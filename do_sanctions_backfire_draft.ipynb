This repository contains an **independent**, professional-grade Python implementation of the ideas in the arXiv preprint "*Do Sanctions Backfire? New Evidence on the Macroeconomic Effects of Supporting Ukraine*" (arXiv:2609.12387, e-journal submission 11 September 2026) by:

*   **Vicente Rios** (Department of Economics and Management, University of Pisa; corresponding author)
*   **Izaskun Barba** (Department of Economics, Universidad Pública de Navarra)
*   **Lisa Gianmoena** (Department of Economics and Management, University of Pisa)

The paper is an arXiv preprint, not a journal article, and it does not include a replication archive. Everything here was written from the manuscript and its appendices.

The project builds the estimation core of a coalition-burden evidence system. It measures the domestic macroeconomic cost borne by countries that sanctioned Russia and splits that cost into a **price-level** component (log CPI) and an **output** component (log real GDP per capita). It does this with a **two-wave Common Correlated Effects Difference-in-Differences (CCEDID) imputation** counterfactual, an **incremental-versus-level** estimand that separates the 2022 escalation from the 2014 regime, and a **pre-escalation Russian-energy exposure gradient**. These estimates are the empirical input that compensation, emergency procurement, infrastructure prioritisation and coalition-cohesion decisions would need. The notebook does **not** implement those allocation rules itself; it supplies the reproducible burden evidence that such rules would be calibrated to.

## Introduction

This project implements the methodology of Rios, Barba and Gianmoena (2026). Its core is the Jupyter notebook `do_sanctions_backfire_draft.ipynb`, which has **41 cells (5 markdown and 36 code)**. The code cells define **542 top-level callables: 465 functions, 61 classes and 16 methods**, counted with Python's `ast` module over top-level definitions (nested helper functions are not counted). The code is organised as one configuration-contract cell followed by **33 task cells, Tasks 1–33**. Task 33 holds the study interface, `orchestrate_study_pipeline`.

**The problem.** Sanctions are designed to hurt the target, but they also cost the countries that impose them. Trade restrictions remove mutually beneficial exchange, financial sanctions interrupt capital flows, and energy measures raise input and consumer prices in economies that relied on the target as a supplier. The literature has studied these sender-side costs far less than the damage to the target. Existing work on the Russia episodes is mostly bilateral (trade, finance), ex-ante (CGE, input–output and scenario models), or limited to one country or sector. It also usually studies the 2014 regime or the 2022 escalation in isolation.

Three empirical problems make the estimate hard:

1. **Non-random selection.** Sanctioning countries are richer, institutionally stronger and more exposed to European energy markets than the rest of the world. In 2009–2013 means, main sanctions targets have log real GDP per capita of 10.061 against 8.184 for their donors, a rule-of-law index of 1.029 against −0.361, and oil rents of 0.383 % against 5.459 % of GDP. Unadjusted parallel trends is therefore implausible.
2. **Confounding global shocks.** The post-2022 period also contains pandemic recovery, supply-chain disruption, commodity shocks and monetary tightening. Two-way fixed effects would force every country to respond to them identically.
3. **A contaminated baseline.** Treating 2022 as a single intervention would use a pre-period already affected by the post-2014 sanctions regime.

The paper reports these headline findings (strict-sanctions coalition, 42 targets, 84 donors):

| Outcome | Wave 1 level, 2014–2021 | Wave 2 level, 2022–2024 | Incremental Wave 2 vs 2019–2021 |
|---|---|---|---|
| Log CPI | −0.043*** (0.009) | 0.022 (0.015) | **0.072*** (0.007)** |
| Log real GDP per capita | 0.024** (0.011) | 0.083*** (0.019) | 0.025*** (0.008) |

- **Prices carry the burden.** The incremental log-CPI contrast of 0.072 log points is about 7.5 % ($100[\exp(0.072)-1] = 7.47$). It is positive in 41 of 42 coalition members and stays between 0.070 and 0.072 across all five reference windows. Every country-history and region-block bootstrap interval excludes zero, and the preferred size-matched placebo probability is 0.022.
- **Output is not robust.** The incremental log-GDP-per-capita contrast ranges from −0.001 to 0.053 across reference windows and from 0.007 to 0.099 across factor proxies, and its placebo probability is 0.102. The rate check shows annual GDP-per-capita growth **3.9 percentage points lower** and inflation **2.5 percentage points higher** after 2022.
- **The narrower direct-arming coalition** (29 targets) shows a larger price effect, 0.107 (SE 0.027, placebo p = 0.002), and an insignificant output-level effect, 0.016 (SE 0.010).
- **Exposure grades the burden.** Within the coalition, one standard deviation more Russian fossil-energy imports relative to GDP over 2018–2021 goes with a 0.0213 log-point larger post-2022 CPI differential (SE 0.0060) and 0.958 percentage points lower annual GDP-per-capita growth (SE 0.2435).

**What the codebase lets you do.** Each group of tasks is a module:

- **Bind the configuration** to code-defined scientific rules, source schemas, runtime inputs and provenance requirements (the configuration-contract cell).
- **Ingest and validate** the ten raw sources and the configuration, then cleanse them with registry-based aggregate exclusion (Tasks 1–3).
- **Build the balanced panel:** raw-data eligibility screening, control completion, outcome repair and the one-year-lagged analysis panel (Tasks 4–7).
- **Code treatment and partitions:** documentary policy coding and the eight policy × outcome target/donor partitions (Tasks 8–9).
- **Estimate the CCEDID counterfactual:** donor factor, pooled pre-2014 fit, counterfactual projection, gaps and level/incremental ATTs, and country-dispersion inference (Tasks 10–14).
- **Run the baseline and the resampling inference:** the baseline orchestrator, then country-history bootstrap, region-block bootstrap and size-matched placebo with seed-stability checks (Tasks 15–17).
- **Run the robustness programme:** reference, donor-pool, factor-proxy, horizon, active-year and rate-outcome sweeps, influence diagnostics and matched designs (Tasks 18–24).
- **Construct the exposure measure** from UN Comtrade primitives and run the exposure-gradient regressions (Tasks 25–29).
- **Check the coalition boundary** with the direct-arming pipeline (Task 30).
- **Report and audit:** figures and tables, then a 418-cell reconciliation with interface-invariant probes and archive verification (Tasks 31–32).
- **Drive everything** through one interface, `orchestrate_study_pipeline`, which threads 31 registered stages (Task 33).

> **Reproduction status: read this first.**
> - The repository ships **no data**. The World Bank, WGI and UN Comtrade series are public, but you must extract them yourself, and the manuscript does not state the WDI vintage, the WGI release or the Comtrade HS codes.
> - The notebook is **define-only**. Running it defines the callables and estimates nothing.
> - The shipped `config.yaml` (version `1.1.0`, `execution_mode: production`) **fails preflight on purpose**. Three reproducibility gaps are still `unresolved`: `wdi_vintage`, `WGI_release_year` and `policy_source_archive_paths`. `orchestrate_study_pipeline` therefore raises `ConfigValidationError` before any stage runs. The usage example below reports exactly this outcome (`preflight_blocked`, zero scheduler stages executed).
> - The independent-authority registration in `config.independent_authorities` is also empty (`status: awaiting_independently_verified_evidence`). Once the three gaps are resolved, this becomes a granular study blocker.
> - The full empirical pipeline **has not been executed against the study data**. No number in this README was produced by this code; every reported number comes from the manuscript.
> - Synthetic inputs exercise the input contract only. Matching a manuscript count on synthetic data does not authenticate a data vintage, a policy instrument or a causal effect.

## Theoretical Background

The study combines development-data systems, panel econometrics with interactive factors, modern imputation-style difference-in-differences, resampling inference, trade statistics and sanctions-law coding. The equation numbers below are those of the manuscript.

### 1. Panel, Outcomes and Controls

The panel is balanced: 129 countries × 25 years (2000–2024) = 3,225 country-years. It has four outcomes, eight controls and two policy indicators.

| Variable | Source code | Transformation |
|---|---|---|
| CPI | `FP.CPI.TOTL` (WDI) | natural log after completion; a within-country index |
| Real GDP per capita | `NY.GDP.PCAP.KD` (WDI) | natural log after completion; constant prices |
| Inflation rate | `FP.CPI.TOTL.ZG` (WDI) | divided by 100 for ATT estimation; percentage points in the mechanism regressions |
| Real GDP-per-capita growth | `NY.GDP.PCAP.KD.ZG` (WDI) | divided by 100 for ATT estimation; percentage points in the mechanism regressions |
| Oil rents | `NY.GDP.PETR.RT.ZS` (WDI) | percent of GDP |
| Trade openness | `NE.TRD.GNFS.ZS` (WDI) | percent of GDP, floored at 0 |
| Terms of trade | `TT.PRI.MRCH.XD.WD` (WDI) | index |
| Investment rate | `NE.GDI.TOTL.ZS` (WDI) | percent of GDP, floored at 0 |
| Government consumption | `NE.CON.GOVT.ZS` (WDI) | percent of GDP, floored at 0 |
| Rule of law | WGI rule-of-law estimate | level |
| Population growth | `SP.POP.GROW` (WDI) | annual percent |
| Fuel-share balance | `TM.VAL.FUEL.ZS.UN` − `TX.VAL.FUEL.ZS.UN` (WDI) | percentage points, built from the separately completed primitives |

All controls enter with a one-year lag, so the effective estimation period is 2001–2024 and the untreated model is fitted on 2001–2013 ($T_0 = 13$). Log levels are the primary outcomes because a window average of log-level gaps measures a **persistent** wedge in purchasing power or output. A temporary inflation surge leaves a permanent price-level gap even after annual inflation normalises.

### 2. Eligibility Before Completion

Eligibility is decided on **raw** 2000–2024 data, before any repair and without using treatment status. This removes endogenous sample selection and collider bias. A country is retained only if:

1. none of the four outcomes has more than 35 % raw missingness;
2. at most one of the eight controls has more than 25 % raw missingness;
3. no control has more than 65 % raw missingness;
4. CPI and real GDP per capita each have at most 25 % raw missingness in 2019–2024. This allows at most one missing year, and it excludes Syria and Sudan (131 → 129).

Completion is a deterministic, single-pass, audited hierarchy:

- **Controls** (1,471 of 25,800 cells, 5.70 %):
  - internal gaps → linear interpolation;
  - trailing gaps → last observation carried forward;
  - leading gaps of at most three years → first observation carried back;
  - longer leading gaps → a local linear trend on the first five valid observations, bounded by the country's historical range ± 1 SD.
- **Outcomes** (57 of 12,900 cells, 0.44 %):
  - 55 leading CPI or inflation cells from 2000–2007 → bounded own-series trend backcasts;
  - Bosnia and Herzegovina's 2024 CPI and inflation → carried forward from 2023;
  - no internal outcome gap is interpolated.

### 3. Groups and Reporting Windows (Equations 1–2)

For a policy indicator $D_{it}$:

$$
\mathcal{T}=\lbrace i:\max_t D_{it}=1\rbrace,\qquad \mathcal{C}=\lbrace i:\max_t D_{it}=0\rbrace,\qquad \mathcal{T}^{\ast}\subseteq\mathcal{T},\qquad \mathcal{C}_Y\subseteq\mathcal{C} \qquad (1)
$$

$$
\mathcal{W}_1=\lbrace 2014,\dots,2021\rbrace,\qquad \mathcal{W}_2=\lbrace 2022,\dots,2024\rbrace \qquad (2)
$$

Aggregation is over the **ever-treated** set, so Wave 1 includes years before a late joiner's own entry. The late joiners are Bosnia and Herzegovina, Korea, New Zealand and Singapore (2022), and Moldova (2023).

### 4. Untreated Potential Outcome and the Donor Factor Proxy (Equations 3–4)

$$
Y_{it}(\infty)=X_{i,t-1}'\beta+\alpha_i' f_t+\varepsilon_{it} \qquad (3)
$$

$$
\bar Y_{t,\mathcal{C}_Y}=\frac{1}{|\mathcal{C}_Y|}\sum_{j\in\mathcal{C}_Y}Y_{jt},\qquad \hat f_t=\begin{pmatrix}1\\ \bar Y_{t,\mathcal{C}_Y}\end{pmatrix},\qquad \alpha_i'\hat f_t=\mu_i+\lambda_i\bar Y_{t,\mathcal{C}_Y} \qquad (4)
$$

Here $\mu_i$ acts as a country fixed effect and $\lambda_i$ is the country's own sensitivity to the common shock. Unlike the full CCE specification of Pesaran (2006) and Brown et al. (2026), which use cross-sectional averages of outcomes **and** covariates, the proxy keeps only the **never-treated donor mean of the own outcome**. The paper gives two reasons: this keeps the coalition's post-2013 outcomes out of its own factor, and it avoids collinear proxies when only 13 pre-treatment years are available.

### 5. The Five-Step CCEDID Estimator (Equations 5–12)

**Step 1, donor factor.** Compute $\hat f_t$ from $\mathcal{C}_Y$ only, separately for every policy × outcome pair.

**Step 2, pre-treatment fit ($t\le 2013$).**

$$
Y_{it}=X_{i,t-1}'\beta+\hat\alpha_i'\hat f_t+u_{it} \qquad (5)
$$

$$
M_{\hat F}=I-\hat F(\hat F'\hat F)^{-1}\hat F',\qquad \hat\beta=\Big(\sum_{i\in\mathcal{S}_Y}X_i'M_{\hat F}X_i\Big)^{-1}\sum_{i\in\mathcal{S}_Y}X_i'M_{\hat F}Y_i \qquad (6)
$$

$$
\hat\alpha_i=(\hat F'\hat F)^{-1}\hat F'(Y_i-X_i\hat\beta) \qquad (7)
$$

The control coefficient $\hat\beta$ is **pooled** over targets and donors ($\mathcal{S}_Y$). The loadings are **unit-specific**.

**Step 3, counterfactual.** Observed post-2013 controls of targets may themselves respond to treatment, so they are never used. Each covariate is projected from its own pre-2014 co-movement with the donor factor:

$$
\hat\Lambda_i=(\hat F'\hat F)^{-1}\hat F'X_i,\qquad \widehat X_{i,t-1}(\infty)=\hat\Lambda_i'\hat f_t,\quad t\ge 2014 \qquad (8\text{–}9)
$$

$$
\widehat Y_{it}(\infty)=\widehat X_{i,t-1}(\infty)'\hat\beta+\hat\alpha_i'\hat f_t,\quad t\ge 2014 \qquad (10)
$$

**Step 4, gaps.** $\hat g_{it}=Y_{it}-\widehat Y_{it}(\infty)$ (Equation 11).

**Step 5, equally weighted ATT.**

$$
\widehat{\mathrm{ATT}}_{\mathcal W}=\frac{1}{N_T}\sum_{i\in\mathcal T^{\ast}}\Big(\frac{1}{|\mathcal W|}\sum_{t\in\mathcal W}\hat g_{it}\Big),\qquad N_T=|\mathcal T^{\ast}| \qquad (12)
$$

Every country gets equal weight regardless of population or economic size.

**Appendix B computational form.** The implementation follows Appendix B (Equations B1–B8), which replaces the ordinary inverses with the Moore–Penrose pseudo-inverse $(\cdot)^{+}$: $M_{\hat F}=I_{T_0}-\hat F\hat F^{+}$, $\hat\beta=(\sum\widetilde X_i'\widetilde X_i)^{+}(\sum\widetilde X_i'\widetilde Y_i)$, $\hat\Lambda_i=\hat F^{+}X_i$ and $\hat\gamma_i=\hat F^{+}(Y_i-X_i\hat\beta)$. These coincide with Equations (6)–(8) when $\hat F$ has full column rank. The configured numerical routes are:

- $\hat\beta$ always uses the pseudo-inverse;
- $M_{\hat F}$ uses a linear solve when $\hat F$ has full rank and its Gram condition number is at most 1,000, and the pseudo-inverse otherwise;
- the rank tolerance is $\sigma_{\max}\cdot\max(\text{shape})\cdot\epsilon_{\text{float64}}$.

### 6. The Incremental Estimand (Equations 13–15)

The Wave-2 ATT is a **total** level gap and may carry deviations inherited from 2014–2021. The incremental estimand removes each country's late-Wave-1 position:

$$
\hat b_i=\frac{1}{3}\sum_{t=2019}^{2021}\hat g_{it},\qquad \widehat Y^{\mathrm{cont}}_{it}=\widehat Y_{it}(\infty)+\hat b_i,\quad t\in\mathcal W_2 \qquad (13\text{–}14)
$$

$$
\widehat{\mathrm{ATT}}_{\mathrm{Inc},2}=\frac{1}{N_T}\sum_{i\in\mathcal T^{\ast}}\Big[\frac{1}{3}\sum_{t=2022}^{2024}\hat g_{it}-\frac{1}{3}\sum_{t=2019}^{2021}\hat g_{it}\Big] \qquad (15)
$$

Interpretation rests on the **stable-gap continuation** assumption: without the escalation, each country's gap would have stayed at its 2019–2021 level. The assumption allows the 2014 regime to have left a persistent gap but rules out further drift after 2021.

The coalition entered 2022 from a **negative** CPI position (Wave 1: −0.043). That is why the incremental contrast (0.072) exceeds the Wave-2 level gap (0.022): the level gap measures where the coalition ended up, while the incremental contrast measures how far it moved. The exact percentage reading of a log-point estimate $\hat\theta$ is $100[\exp(\hat\theta)-1]$; $100\hat\theta$ is the small-effect approximation.

### 7. Identification and Inference (Equation 16)

A causal reading requires three things:

1. the donor factor captures the common shocks relevant to coalition members;
2. the pre-2014 relationships remain informative after 2013;
3. the timing, anticipation and spillover conditions hold.

When they fail, the estimates are **realized macroeconomic divergences associated with coalition membership**, not causal effects of a legal instrument. The 2022 treatment is a bundle: legal sanctions, Russian supply cuts, energy reallocation and domestic compensation.

The significance stars come only from the **country-dispersion standard error**, with $d_i$ the country-level contrast:

$$
\widehat{\mathrm{SE}}_{\mathrm{CD}}=\frac{\mathrm{sd}(d_i)}{\sqrt{N_T}},\qquad \mathrm{df}=N_T-1 \qquad (16)
$$

This SE treats country contrasts as independent and does **not** propagate first-stage uncertainty from $\hat\beta$, the loadings or the donor factor. Three complementary checks are therefore reported, each with 500 draws:

- a **country-history bootstrap**, which resamples whole country paths and re-estimates the complete design;
- a **region-block bootstrap**, which resamples World Bank regions;
- a **size-matched placebo**, which draws false coalitions from never-treated countries.

The placebo tail probability uses the plus-one correction over $M$ size-matched assignments:

$$
\hat p=\frac{1+\sum_{m=1}^{M}\mathbf{1}\left[\left|\hat\theta^{(m)}\right|\ge\left|\hat\theta^{\mathrm{obs}}\right|\right]}{1+M}
$$

It is a diagnostic tail probability, not a confidence interval.

### 8. Russian-Energy Exposure and the Gradient Regression (Equations 17–18, A1–A2)

Exposure is **monetary**, not physical dependence. It is built from UN Comtrade in current US dollars as a ratio of sums over a window $w$, then standardised within the regression sample:

$$
E^{\mathrm{energy}}_{iw}=\frac{\sum_{t\in w}\big(\mathrm{Coal}^{\mathrm{RUS}}_{it}+\mathrm{Crude}^{\mathrm{RUS}}_{it}+\mathrm{Refined}^{\mathrm{RUS}}_{it}+\mathrm{Gas}^{\mathrm{RUS}}_{it}\big)}{\sum_{t\in w}\mathrm{GDP}_{it}} \qquad (17)
$$

$$
\mathrm{GasShare}_{iw}=\frac{\sum_{t\in w}\mathrm{RussianGas}_{it}}{\sum_{t\in w}\mathrm{WorldGas}_{it}},\qquad \mathrm{EnergyGDP}_{iw}=\frac{\sum_{t\in w}(\mathrm{RussianNonGas}_{it}+\mathrm{RussianGas}_{it})}{\sum_{t\in w}\mathrm{GDP}_{it}} \qquad (\mathrm{A1}\text{–}\mathrm{A2})
$$

The windows are 2018–2021 (near pre-war) and 2010–2013 (pre-Crimea). The construction rules are:

- An absent API row is **not** a zero. A zero counts as verified only when a single-country query succeeds **and** the country reported HS data that year.
- Russian exporter mirror flows recover only LNG. The Austria, Germany, Poland and Türkiye windows are therefore **excluded rather than set to zero**.
- Interpolation runs in $\log(1+x)$, within a window only, for internal gaps of at most two years.
- Ratios are **recomputed from completed components** and never interpolated directly.

The regression samples have 118 and 119 countries, with 39 sanctioners. The gradient model is a different estimator from the headline one: a CCE interaction on the observed panel, without the eight controls, whose omitted period is **all years before 2022**:

$$
Y_{it}=\mu_i+\lambda_i\bar Y_{C,t}+\beta_1(S_i\times Post_t)+\beta_2(E_{iw}\times Post_t)+\beta_3(S_i\times E_{iw}\times Post_t)+u_{it} \qquad (18)
$$

How to read the coefficients:

- $\beta_1$ is the sanctioner differential at mean exposure;
- $\beta_2$ is the exposure slope among non-sanctioners;
- $\beta_3$ is the difference between the two slopes;
- **$\beta_2+\beta_3$ is the within-coalition exposure slope**, with its SE taken from the full country-clustered covariance matrix.

These are **exposure gradients, not mediation**.

## Features

- **33-task modular architecture.** Each task is one notebook cell with its own principal orchestrator. A separate configuration-contract cell binds 92 fixed declarations to the configuration.
- **Single-interface execution.** `orchestrate_study_pipeline(data)` takes one mapping, binds the configured runtime inputs into a copy, validates the inputs, verifies the stage registry and runs **31 registered stages** in dependency order, digesting every stage's artefacts.
- **Order-of-operations discipline.** Eligibility is frozen on raw data before any completion. Neither treatment indicators nor estimated effects enter any eligibility or repair rule.
- **Three executable interface invariants**, probed in the reconciliation audit:
  - no post-treatment target realisation enters its own counterfactual;
  - no treatment information enters eligibility or imputation;
  - no robustness branch replaces the baseline estimate.
- **Pair-specific estimation.** Each of the eight policy × outcome partitions has its own donor factor and pre-2014 fit. No parameter carries over across outcomes or policies.
- **Numerically guarded linear algebra.** SVD-based pseudo-inverses, rank diagnostics and condition-number flags. Factor-proxy and residual-maker Gram conditions are flagged above 1,000, and pooled cross-products above $10^6$.
- **Resampling at the paper's scale.** 500 valid draws each for the country-history bootstrap, the region-block bootstrap and the size-matched placebo, with a redraw cap and documented invalid-resample handling.
- **Seed protocol.** PCG64 through `numpy.random.SeedSequence(master).spawn(3)`:
  - primary seed set `primary_reproduction_v1`, master 20261002;
  - stability seed set `stability_reproduction_v1`, master 20261003;
  - branch spawn keys country-history `[0]`, region-block `[1]`, placebo `[2]`.

  These are documented reproduction choices, **not** recovered author seeds.
- **Explicit reproducibility gaps.** The WDI vintage, the Comtrade HS codes, the seeds, the reconstruction of Equations B10–B13, the WGI release, the policy-source archive paths and the region-block definition are tracked in `config.reproducibility_gaps`, each with a state and a resolution artefact. Unresolved gaps block execution instead of being silently defaulted.
- **Independent-authority trust model.** The table-binding, manuscript-universe, reconciliation-scenario and archive-vintage registries must be pinned by an externally verified SHA-256 manifest before they are loaded.
- **418-cell reconciliation.** Publication cells are compared cell by cell, with data → transformation → model layer chains that localise each discrepancy.
- **Typed, frozen results.** Dataclass result records, read-only `MappingProxyType` registries and canonical hashes link source inputs, configuration, stage outputs and cache lineage.

## Methodology Implemented

Each task below names its principal orchestrator, its notebook cell and its exact cell title. The scheduler's stage numbers equal the task numbers through Task 15. After that they are offset by one, because Task 16's `orchestrate_resampling_inference` is the engine that stage 16 (Task 17's `execute_inference`) calls.

1. **Task 1, Ingest and Fingerprint All Raw Data Structures** (`orchestrate_raw_data_ingest`). Validates the schemas, primary keys and referential integrity of the ten sources, partitions Comtrade Block B, and fingerprints every frame.
2. **Task 2, Validate study_config Integrity and Completeness** (`orchestrate_study_config_validation`). Produces the digest-bound configuration readiness report and raises `ConfigValidationError` on unresolved evidence.
3. **Task 3, Data Cleansing and Canonicalization** (`orchestrate_data_cleansing`). Excludes aggregates such as `WLD` and `HIC` by **country-registry membership**, not by pattern, resolves the WGI vintage, and verifies the cleansed handoff.
4. **Task 4, Sample Eligibility Screening (Algorithm C1)** (`orchestrate_eligibility_screening`). Builds raw missingness matrices and applies the 35 %, 25 %, 65 % and critical-window rules: 131 candidates → 129 retained.
5. **Task 5, Control Missing-Data Completion (Algorithm C2)** (`orchestrate_control_completion`). Completes each control within country with method tags and builds the fuel-share balance from the completed primitives with union flags.
6. **Task 6, Outcome Repair (Algorithm C3)** (`orchestrate_outcome_repair`). Applies the bounded trend backcasts to leading CPI and inflation cells and carries BIH's 2024 values forward from 2023.
7. **Task 7, Analysis Panel Assembly and Freezing** (`orchestrate_panel_assembly`). Builds log levels, decimal ATT rates and the eight `*_lag1` controls. The usable lagged grid is 129 × 24 (2001–2024).
8. **Task 8, Treatment Coding Resolution (Algorithm G0)** (`orchestrate_treatment_coding`). Resolves strict-sanctions and direct-arming entry years from the documentary ledger, with legal instruments ranked above alignment notices and secondary material.
9. **Task 9, Donor Pool and Target Set Construction per Policy and Outcome** (`orchestrate_partition_construction`). Builds all eight partitions (42/84 for sanctions; 29/96 and 29/97 for direct arming) and audits rosters and 2009–2013 means.
10. **Task 10, Donor Factor Proxy Construction (Step 1 of CCEDID)** (`orchestrate_donor_factor`). Equation (4): $\hat f_t=(1,\bar Y_{t,\mathcal C_Y})'$ for each pair.
11. **Task 11, Pre-Treatment Pooled Fit (Step 2 of CCEDID)** (`orchestrate_pretreatment_fit`). Equations (5)–(7) and B3–B6 over 2001–2013, through the `ResidualMaker` and `PooledFitResult` records.
12. **Task 12, Counterfactual Covariate and Outcome Construction** (`orchestrate_counterfactual_projection`). Equations (8)–(10) and B7–B8, keeping the fitted pre-period residuals separate.
13. **Task 13, Gap Matrix and Level/Incremental ATT** (`orchestrate_gap_and_att`). Equations (11)–(15), including the incremental identity check (tolerance $10^{-10}$).
14. **Task 14, Country-Dispersion Inference (Equation 16)** (`orchestrate_country_dispersion_inference`). Student-t inference with $N_T-1$ degrees of freedom, which is the only source of significance stars.
15. **Task 15, Orchestrator 1 - Baseline Full Pipeline** (`orchestrate_baseline_pipeline`). Reconciles the baseline calculations and freezes the shared data mapping that the variant branches use.
16. **Task 16, Orchestrator 2 - Bootstrap and Placebo Orchestrator** (`orchestrate_resampling_inference`). Algorithms B1–B3 with full-design refits per replicate.
17. **Task 17, Execute Bootstrap and Placebo Inference** (`execute_inference`). Runs the primary and stability seed sets through the region map.
18. **Task 18, Orchestrator 3 - Robustness Sweep Factory** (`execute_sweeps`). The sweep specification and identity factory, which keeps baseline and variant identities separate.
19. **Task 19, Execute Reference, Donor, and Factor Sweeps** (`execute_robustness_sweeps`). The five reference windows, three donor pools and five factor proxies, with conditioning flags.
20. **Task 20, Execute Horizon, Active-Year, and Rate-Outcome Sweeps** (`execute_secondary_sweeps`). The exclude-2024 horizon, active-years-only averaging, and the inflation and growth rate outcomes.
21. **Task 21, Build Orchestrator 4 - Influence Diagnostics** (`execute_influence_diagnostics`). Treated leave-one-out, donor leave-one-out with the factor rebuilt, and the median, trimmed mean and positive share.
22. **Task 22, Execute Influence Diagnostics** (`execute_influence_tables`). Influence and distribution tables built from the live diagnostics.
23. **Task 23, Build Orchestrator 5 - Matching Designs and Balance** (`execute_matching_designs`). The high-income design (median 2009–2013 log GDP per capita) and frequency-weighted Mahalanobis matching with $M=5$.
24. **Task 24, Execute Matching Designs** (`execute_matching_report`). Standardised-difference balance and design-specific event paths.
25. **Task 25, Comtrade Cell Classification and Cell-Status Ledger** (`build_cell_ledger`). Classifies every trade primitive as positive, verified zero or missing, using the retry and HS-report evidence.
26. **Task 26, Mirror-Flow Recovery, Denominator Construction, and Completion** (`construct_exposure_primitives`). Gas mirror recovery, the world-gas denominator, and within-window $\log(1+x)$ completion.
27. **Task 27, Exposure Ratio Construction and Sample Rules** (`construct_exposure_ratios`). Equations (A1)–(A2), the strict and extended coverage rules, and within-sample standardisation.
28. **Task 28, Build Orchestrator 6 - Exposure Heterogeneity Orchestrator** (`execute_heterogeneity_suite`). The declared exposure-gradient specifications: windows, estimator, policy definition and exposure measure.
29. **Task 29, Execute Exposure Heterogeneity Regressions** (`execute_heterogeneity_regressions`). Equation (18) for four outcomes, with clustered covariance and $\beta_2+\beta_3$ from the full covariance matrix.
30. **Task 30, Direct-Arming Coalition Pipeline Execution** (`execute_direct_arming_pipeline`). The 29-target baseline, its inference and active-year timing, as a coalition-boundary check.
31. **Task 31, Figure and Table Production** (`produce_figures_and_tables`). Live figures (2, 3, 4, B1–B3) and publication tables with separate European and world geometry scopes.
32. **Task 32, Reconciliation and Verification Audit** (`execute_reconciliation_audit`). The 418-cell comparison, the three invariant probes and verification of the self-contained archive.
33. **Task 33, End-to-End Pipeline Orchestration** (`orchestrate_study_pipeline`). Binds the runtime inputs, registers the authorities, runs preflight, executes the 31 stages and returns the status and blockers.

**Return value.** A dictionary with one entry per stage name, each holding the artefacts that stage produced, plus the report keys `pipeline_summary`, `pipeline_state`, `pipeline_digest`, `pipeline_notes`, `pipeline_status` and `pipeline_blockers`. `pipeline_status` is `"complete"` only when all of the following hold:

- all 31 stages completed in exact registry order, with live artefact digests matching their records;
- the reconciliation is `complete` and blocker-free, covering 418 cells;
- all three invariants hold;
- the self-contained archive verifies.

Otherwise the status is `"incomplete"`, and `pipeline_blockers` lists every reason.

**Errors.** Failures are distinguished as follows:

- **`IngestError`:** a malformed `data` mapping, sources that are not exactly the ten declared frames, a conflicting runtime duplicate, or a registry that breaks its threading contract.
- **`ConfigValidationError`:** unresolved configuration evidence. A preflight exception means **no** pipeline result is returned.

### **Step 1: Synthetic Data Generation (the ten raw DataFrames)**

**Methodology.** The generator fixes 131 country candidates: the manuscript's 129 descriptive countries plus Syria and Sudan. It also places `WLD` and `HIC` aggregate rows in both WDI representations, so cleansing must use country-registry membership. Every synthetic country has all 29 raw calendar years, 1996–2024. The retained analytical window remains 2000–2024; outcomes and controls stay in raw source units.

The generator compounds annual percentage rates into positive levels:

$$
C_{it}=C_{i,t-1}\left(1+\frac{\pi_{it}}{100}\right),\qquad
Q_{it}=Q_{i,t-1}\left(1+\frac{r_{it}}{100}\right).
$$

These are **synthetic data-generation identities**, not additional manuscript estimators. CPI is rebased to a synthetic 2010 value of 100. Constant-price GDP per capita is kept distinct from current-dollar aggregate GDP. The generated nominal denominator uses the artificial population and price scale; it is not an accounting recovery of any country's actual GDP deflator. Governance levels are generated inside a bounded range and enter unchanged; no observed WGI series is winsorised. Price and output paths contain declared common/regional shocks, not a fitted paper effect.

**Required fidelity corrections to the request.**

- `[A-Z]{3}` also matches `WLD`, `HIC` and other aggregate codes. Pattern matching alone cannot exclude them. The notebook's authoritative country registry supplies the membership test; the example includes two such rows deliberately.
- The raw WDI table has **12**, not 10, primitive indicator codes: four outcomes, six direct controls, fuel imports and fuel exports. Rule of law is supplied separately by WGI. The eight model controls are the six direct controls, rule of law and the **completed-primitives fuel-share balance**.
- Table A6 retains **Russia and Belarus inside the 129-country descriptive sample** while barring their estimation roles. Ukraine is also retained but excluded from the main sanctions ATT and every donor pool. Only Syria and Sudan receive zero retained flags in this 131-country fixture.
- Table A4 contains **Croatia** in the 29-country direct-arming roster. Moldova enters sanctions in 2023; BIH, Korea, New Zealand and Singapore enter in 2022. The four pre-2022 arming countries are Czechia, Poland, Turkiye and the United States. Mexico is neither sanctions-coded nor a North America block member under the seven-region fixture convention.
- The policy source primary key is `(country_code, policy, document_title)` in the actual notebook/configuration. `source_document` is not a declared input column. Albania uses an `alignment_notice`; no EU-member regulation is fabricated for it.
- The wide DataFrame is the synthetic tabular equivalent of the requested wide extract. Its values are generated once, and `wdi_raw_long` is its exact raw-value counterpart. Missing cells are explicit NaNs, not absent rows assigned false zeros.

### **Step 2: Loading the Configuration (`config.yaml`)**

**Methodology.** Parse the existing file with a safe **YAML 1.2** loader and reject duplicate keys. This preserves unquoted scientific numeric literals such as `1e-10` as numbers. Generic YAML 1.1 parsing can interpret this literal differently and is not the project's validated parsing path. The loader returns the plain dictionary `config` and checks its section and operative binding contracts. It deliberately leaves arrays as arrays; the notebook handles each declared window and tuple boundary. It does not call unknown provenance resolved.

The actual `AnalysisPanel.lagged` object retains all 3,225 country-year rows and appends the eight `*_lag1` columns. Those columns are NaN in 2000; the **usable** lagged-control grid has 129 × 24 observations for 2001–2024. The pre-treatment matrices select 2001–2013 explicitly. Do not interpret the presence of a 2000 row as an available lag or fitted counterfactual residual.

The research specification remains:

$$
\widehat f_t=\begin{pmatrix}1\\\overline Y_{\mathcal C_Y,t}\end{pmatrix},\qquad
M_{\widehat F}=I-\widehat F\widehat F^{+},\qquad
\widehat\beta=
\left(\sum_{i\in\mathcal S_Y}\widetilde X_i'\widetilde X_i\right)^{+}
\sum_{i\in\mathcal S_Y}\widetilde X_i'\widetilde Y_i,
\quad \widetilde X_i=M_{\widehat F}X_i,\quad
\widetilde Y_i=M_{\widehat F}Y_i.
$$

Here $X_i$ contains the **one-year-lagged** controls paired with 2001–2013 outcomes. Using the Appendix B computational orientation, $\widehat\Lambda_i=\widehat F^{+}X_i$ is $2\times8$, and $\widehat\gamma_i=\widehat F^{+}(Y_i-X_i\widehat\beta)$ is $2\times1$. For $t\ge2014$:

$$
\widehat X_{i,t-1}(\infty)=\widehat\Lambda_i'\widehat f_t,\qquad
\widehat Y_{it}(\infty)=
\widehat X_{i,t-1}(\infty)'\widehat\beta+
\widehat\gamma_i'\widehat f_t,\qquad
\widehat g_{it}=Y_{it}-\widehat Y_{it}(\infty).
$$

For the preferred reference $R=\lbrace 2019,2020,2021\rbrace$ and $W_2=\lbrace 2022,2023,2024\rbrace$:

$$
d_i=\frac1{|W_2|}\sum_{t\in W_2}\widehat g_{it}
-\frac1{|R|}\sum_{t\in R}\widehat g_{it},\qquad
\widehat{ATT}^{\mathrm{Inc}}_2=\frac1{N_T}\sum_i d_i,\qquad
\widehat{SE}_{\mathrm{CD}}=\frac{\mathrm{sd}(d_i)}{\sqrt{N_T}},\quad
\mathrm{df}=N_T-1.
$$

The exact percentage interpretation of a log-point estimate is $100[\exp(\widehat\theta)-1]$. Annual rate outcomes are divided by 100 once for ATT estimation; mechanism regressions use percentage-point units once. Exposure ratios are computed from separately completed current-dollar components. None of these transformations occur inside the synthetic raw-data generator. Counterfactual construction excludes target post-2013 realizations; eligibility/completion exclude treatment information; robustness never replaces baseline. Fitted dynamic gaps start in **2001**, normalize at 2013, and cannot acquire a fabricated year-2000 residual.

### **Summary of the Execution Flow**

1. **Notebook namespace.** Execute the existing draft's research cells first; append the example cells below them. No research function is imported from a `.py` folder.
2. **Raw inputs.** A private PCG64 synthetic model creates all ten named DataFrames, preserving source codes, fixed country labels, calendars, units, zeros versus missing values, and documentary/region distinctions.
3. **Configuration.** Read `config.yaml` safely as YAML 1.2 into `config`; keep its study, seed, runtime, numerical and provenance parameters unchanged.
4. **Input audit.** Verify the actual ingest contract, raw 129-country coverage counts, all primary keys, exact wide/long closure, positive nominal GDP and coherent primitive trade statuses. This is source-interface validation, not empirical reproduction.
5. **Interface call.** Supply `data={'study_config': config, 'sources': synthetic_sources}` through the final wrapper. The actual interface promotes six configured runtime entries and, when ready, owns the ordered 31-stage dependency graph.
6. **Current outcome.** The supplied production configuration raises `ConfigValidationError` on its three unresolved evidence groups before the scheduler runs. The wrapper records `preflight_blocked`, the exact error and `pipeline_result=None`; it does not report successful stages.
7. **Ready computational path.** A genuine run proceeds through raw eligibility/completion, one-year-lagged assembly, pair-specific donor factors, pre-2014 estimation, untouched target counterfactuals, level/incremental effects and country-dispersion inference. Bootstrap/placebo diagnostics retain their configured full-design 500-draw contracts.
8. **Design diagnostics.** Donor/factor/reference, influence, matched and coalition-boundary branches retain baseline isolation. Exposure construction uses independent components and separate windows; gradient regressions remain conditional associations, not mediation.
9. **Reporting and certification.** Supply authentic income/geometry/table/inventory/provenance/checkpoint inputs and separately verified authority registries. Completion additionally requires the 418-cell comparison, three invariant probes and ten-family persisted archive. Synthetic counts do not satisfy those evidentiary requirements.
10. **Interpretation.** Price and output gaps describe macroeconomic divergences under maintained counterfactual conditions. This pipeline supplies evidence relevant to coalition burden; the example does not implement a compensation allocator, procurement optimiser, household-incidence model or sanctions-success verdict.
11. **Verification.** Execute the embedded cells and focused tests, preserve the original notebook/configuration digests, and read the machine evidence for the precise scope of what passed. Treat the documented preflight rejection as the correct current operating outcome.

## Output Structure

A call that reaches the scheduler returns:

```python
{
    "raw_data_ingest":        {"ingest_bundle": ...},
    "study_config_validation": {"config_validation": ...},
    # ... one entry per registered stage, keyed by stage name, holding the artefacts it produced ...
    "reconciliation":         {...},
    "pipeline_summary":  pd.DataFrame,  # columns: number, name, outcome, produces, missing, error,

    "pipeline_state":    {...},          # live artefact state

    "pipeline_notes":    ...,            # completed / skipped / failed counts
    "pipeline_status":   "complete" | "incomplete",
    "pipeline_blockers": [...],          # granular reasons the study is not complete
```

Each stage ends in one of three outcomes:

- `completed`;
- `skipped_missing_inputs`, with the missing keys named;
- `failed`, with the error recorded.

A mandatory stage that does not complete becomes a blocker. Persisted outputs go to the configured `runtime_inputs.output_dir` (`results`) and `runtime_inputs.replication_archive_root` (`replication_archive`), and the archive release tag is `v1.1.0-frozen`. The run that the usage example verifies is rejected at preflight, so it **writes no study outputs** and returns **no** pipeline result.

The ready computational path is designed to produce:

- **Figures** 2, 3, 4, B1, B2 and B3.
- **Tables** 3–10, A1–A8 and B1–B7: 23 tables and 418 publication cells.

Tables 1 and 2 of the paper are descriptive and are not part of this producer's registry.

- `wdi_vintage.extraction_vintage`;
- `WGI_release_year.source_release`;
- `policy_source_archive_paths.archive_root`, together with archived documents for every policy code.

Change runtime paths **only** by editing `runtime_inputs`. A conflicting duplicate passed in `data` raises `IngestError` by design.

## Contributing

Contributions are welcome. Please follow the conventions of the notebook:

- **Style.** Follow PEP 8, give every callable full PEP 484 type hints and a NumPy-style docstring, and put a comment before each line of code explaining what it does.
- **Fidelity.** Do not deviate from the manuscript's equations, windows, thresholds or estimands. Record every implementation choice the paper leaves open in `reproducibility_gaps` or `configuration_documentation`, together with its evidence artefact.
- **Invariants.** Preserve the three interface invariants: no post-treatment target data in its own counterfactual, no treatment information in eligibility or imputation, and no robustness branch replacing the baseline.
- **Honesty.** Never fabricate provider data, policy documents, seeds or authority pins, and never present synthetic results as a reproduction.
- **Tests.** Run every new check against synthetic inputs and write only to temporary directories.

## Recommended Extensions

1. **Resolve the vintages.** Date the WDI extraction and the WGI release, so that the 129-country retained set and the raw missing-cell counts (57 outcome cells, 1,471 control cells) can be verified.
2. **Archive the policy sources.** Build an archived policy-source ledger with issuing authority, title, date and archived copy for every positive country-year code.
3. **Add a decision layer.** Build a separately specified allocation layer that maps the estimated price and output burdens and the exposure gradients to compensation, procurement, infrastructure and cohesion-backstop rules. The notebook deliberately stops at the evidence layer.
4. **Model incidence.** Add household and firm incidence models; the paper does not estimate who bears the price increase.

# Paper

Title: "*Do Sanctions Backfire? New Evidence on the Macroeconomic Effects of Supporting Ukraine*"

Authors: Vicente Rios, Izaskun Barba, Lisa Gianmoena

E-Journal Submission Date: 11 September 2026

Link: https://arxiv.org/abs/2609.12387

Abstract:

Using a balanced panel of 129 countries from 2000 to 2024, we estimate the macroeconomic costs of sanctions on Russia. To that end, we employ a novel two-wave common-correlated-effects imputation design that separates the 2014 sanctions from the 2022 escalation and accounts for heterogeneous exposure to global shocks. We find the main boomerang effect of these sanctions operates through prices rather than output. Specifically, relative to 2019-2021, we find a significant post-2022 increase in prices among sanctioning countries that remains robust under alternative counterfactuals, reference windows, and inference procedures. Conversely, we find no robust evidence of aggregate GDP-per-capita losses.



# Paper Critical Summary

# Summary: *Do Sanctions Backfire? New Evidence on the Macroeconomic Effects of Supporting Ukraine*
Rios, Barba & Gianmoena (11 September 2026)

## The paper's claim in one paragraph

On a balanced panel of 129 countries over 2000–2024, a **two-wave Common Correlated Effects Difference-in-Differences (CCEDID) imputation estimator** finds that the sender-side ("boomerang") cost of sanctioning Russia shows up in **prices, not output**. Relative to each country's own 2019–2021 gap, log CPI in the 42 sanctioning countries sits **0.072 log points (≈7.5%)** above its imputed untreated path in 2022–2024. The result is robust across every check the paper runs. There is **no robust aggregate loss in real GDP per capita**. The cross-country dispersion of the post-2022 burden is graded by **pre-war Russian fossil-energy imports/GDP**: +1 SD of exposure goes with a +0.0213 log-point CPI differential and **−0.96 pp annual GDP-per-capita growth** among sanctioners.

## Research question and the three identification problems

**Question.** Did coalition members pay a measurable *national* macroeconomic price, in the CPI level or in real GDP per capita, across the post-2014 regime and the post-2022 escalation?

**Gap in the literature (Table 1):**

- Post-2014 work is bilateral (trade, finance) and finds weak aggregate output effects (Crozet & Hinz 2020; Besedeš et al. 2017; Kholodilin & Netšunajev 2019).
- Post-2022 work is mostly ex-ante simulation (CGE, I-O, scenario models) or covers single countries or sectors.
- No coalition-wide, ex-post, realized panel exists, and no study models both waves jointly.

**The three empirical problems the design must solve:**

1. **Non-random selection.** Senders are richer, more democratic, more trade-integrated and more exposed to European energy markets. In pre-period means (2009–2013): log GDPpc 10.06 vs 8.18; rule of law 1.03 vs −0.36; oil rents 0.38% vs 5.46%. Unadjusted parallel trends is therefore implausible.
2. **Confounding global shocks after 2022** (pandemic recovery, supply chains, commodities, monetary tightening). TWFE year effects impose a *homogeneous* response to these shocks.
3. **A contaminated baseline.** A 2022 single-intervention design would use a pre-period already treated by the 2014 regime.

**The paper's answer:** interactive-factor imputation (heterogeneous loadings deal with problem 2), pre-2014 estimation (deals with problem 3), and an incremental estimand netting out the 2019–2021 gap (separates the two waves).

## Data, eligibility and missing-data completion

**Panel.** 129 countries × 25 years = 3,225 country-years. It has 4 outcomes, 8 controls and 2 policy indicators. The effective estimation period is 2001–2024 because controls are lagged one year, and the untreated model is fit on **2001–2013 ($T_0 = 13$)**. The years 1996–1999 are used only as support for backcasting.

**Outcomes (WDI codes):**

| Outcome | Code | Role |
|---|---|---|
| log CPI | FP.CPI.TOTL | primary (accumulated price-level wedge) |
| log real GDPpc | NY.GDP.PCAP.KD | primary (accumulated output wedge) |
| Inflation | FP.CPI.TOTL.ZG (÷100) | complementary rate (momentum) |
| GDPpc growth | NY.GDP.PCAP.KD.ZG (÷100) | complementary rate (momentum) |

**Controls ($K=8$, all lagged one year):** oil rents, trade openness, terms of trade, investment rate, government consumption, WGI rule of law, population growth, and the fuel-share balance (fuel-import share minus fuel-export share).

**Why levels and not rates.** A temporary inflation surge leaves a *permanent* price-level wedge. A window average of log-level gaps measures the persistent loss of purchasing power, while rates only measure year-to-year momentum.

**Eligibility rules.** These are fixed on *raw* data *before* any imputation and *without* using treatment status, which guards against endogenous selection and collider bias:

1. Each of the 4 outcomes has ≤35% raw missingness over 2000–2024.
2. At most one of the 8 controls has >25% missingness.
3. No control has >65% missingness.
4. Critical window: CPI and real GDPpc each have ≤25% missingness in 2019–2024, which allows at most one missing year. This rule removes Syria and Sudan (131 → 129).

**Completion (deterministic, single pass, audited):**

- **Controls:** 1,471/25,800 cells (5.70%) were missing. The fill order is:
  - internal gaps → linear interpolation;
  - trailing gaps → last observation carried forward;
  - leading gaps of ≤3 years → first observation carried back;
  - longer leading gaps → local linear trend on the first 5 valid observations, bounded by the country's historical range ±1 SD.
  - Trade openness, investment and government consumption are also floored at 0.
- **Outcomes:** 57/12,900 cells (0.44%) were missing, all of them CPI or inflation. 55 are leading 2000–2007 cells filled by bounded own-series trend backcasts. The other 2 are BIH 2024 values carried forward from 2023. There is no internal outcome interpolation and no cross-country information is used.
- Terms of trade (20.0%) and oil rents (12.5%) dominate the control missingness.


## Policy coding and estimation roles

- **Strict sanctions $S_{it}$:** = 1 from the first year a country is legally covered by, or formally aligns with, an official Russia sanctions regime. It is an absorbing state. The EU enters in 2014. It measures **participation, not intensity**. 43 countries are coded, and Ukraine is dropped from the treated set because of war destruction, leaving **42 targets**.
  - Late joiners: BIH, KOR, NZL and SGP enter in 2022; MDA enters in 2023.
- **Direct arming:** = 1 from the first verifiable delivery or commitment of lethal weapons, ammunition or complete military systems. This gives **29 countries**, 28 of which also sanction; Türkiye arms without sanctioning. Four are active before 2022: CZE, POL and USA (2018), and TUR (2019).
  - This is a **narrower coalition definition, not a military-aid effect**.
- **Geopolitical donor exclusions:** Russia, Belarus and Ukraine are excluded from every donor pool.
- **Groups:** 14 sanctions-only, 1 arming-only, 28 both, 83 neither, and 3 excluded.

| Estimation | Targets | Donors |
|---|---|---|
| Sanctions (CPI / GDPpc) | 42 | 84 / 84 (83 neither + TUR) |
| Direct arming (CPI / GDPpc) | 29 | 96 / 97 (83 neither + 14 sanctions-only; BIH dropped from CPI donors because its 2024 CPI is repaired) |

## The CCEDID imputation estimator (key equations)

The estimator is run separately for each policy × outcome pair.

**Sets and windows:**

$$\mathcal{T}=\{i:\max_t D_{it}=1\},\quad \mathcal{C}=\{i:\max_t D_{it}=0\},\quad \mathcal{T}^*\subseteq\mathcal{T},\ \mathcal{C}_Y\subseteq\mathcal{C} \tag{1}$$

$$\mathcal{W}_1=\{2014,\dots,2021\},\qquad \mathcal{W}_2=\{2022,\dots,2024\} \tag{2}$$

The aggregation is **ever-treated**: Wave 1 includes the pre-entry years of late joiners.

**Untreated potential outcome (interactive-factor model):**

$$Y_{it}(\infty)=X_{i,t-1}'\beta+\alpha_i' f_t+\varepsilon_{it} \tag{3}$$

**Donor-based factor proxy (parsimonious CCE):**

$$\bar Y_{t,\mathcal{C}_Y}=\frac{1}{|\mathcal{C}_Y|}\sum_{j\in\mathcal{C}_Y}Y_{jt},\qquad \hat f_t=\begin{pmatrix}1\\ \bar Y_{t,\mathcal{C}_Y}\end{pmatrix},\qquad \alpha_i'\hat f_t=\mu_i+\lambda_i\bar Y_{t,\mathcal{C}_Y} \tag{4}$$

- $\mu_i$ acts as a country fixed effect.
- $\lambda_i$ is the country-specific sensitivity to the common shock.
- Only the **never-treated donor CSA of the own outcome** is used, unlike Pesaran (2006) or Brown et al. (2026), which use CSAs of outcomes *and* covariates. The paper gives two reasons:
  - it keeps the coalition's post-2013 outcomes out of its own factor (endogeneity);
  - it avoids collinear proxies when only $T_0=13$ pre-years are available.

**Step 1 — Factor.** Compute $\hat f_t$ from $\mathcal{C}_Y$ for every $t$.

**Step 2 — Pre-treatment fit ($t\le2013$).**

$$Y_{it}=X_{i,t-1}'\beta+\hat\alpha_i'\hat f_t+u_{it} \tag{5}$$

$$M_{\hat F}=I-\hat F(\hat F'\hat F)^{-1}\hat F',\qquad \hat\beta=\Big(\sum_{i\in\mathcal{S}_Y}X_i'M_{\hat F}X_i\Big)^{-1}\sum_{i\in\mathcal{S}_Y}X_i'M_{\hat F}Y_i \tag{6}$$

$$\hat\alpha_i=(\hat F'\hat F)^{-1}\hat F'(Y_i-X_i\hat\beta) \tag{7}$$

$\hat\beta$ is **pooled over targets and donors** ($\mathcal{S}_Y$) and is common to all units. The loadings are unit-specific.

**Step 3 — Counterfactual.** Observed post-2013 controls of targets are **not** used because they may be treatment-contaminated (trade, investment, fiscal policy, energy). Controls are projected instead:

$$\hat\Lambda_i=(\hat F'\hat F)^{-1}\hat F'X_i \quad (2\times K) \tag{8}$$

$$\widehat X_{i,t-1}(\infty)=\hat\Lambda_i'\hat f_t,\quad t\ge2014 \tag{9}$$

$$\widehat Y_{it}(\infty)=\widehat X_{i,t-1}(\infty)'\hat\beta+\hat\alpha_i'\hat f_t,\quad t\ge2014 \tag{10}$$

The country-specific relationships are learned before 2014. After 2013, the path is driven only by never-treated donor information, so **no post-treatment target observation enters its own counterfactual**.

**Step 4 — Gaps.**

$$\hat g_{it}=Y_{it}-\widehat Y_{it}(\infty) \tag{11}$$

**Step 5 — Equally weighted ATT** (each country counts once, regardless of population or GDP):

$$\widehat{\text{ATT}}_{\mathcal W}=\frac{1}{N_T}\sum_{i\in\mathcal T^*}\Big(\frac{1}{|\mathcal W|}\sum_{t\in\mathcal W}\hat g_{it}\Big),\quad N_T=|\mathcal T^*| \tag{12}$$

**Appendix B computational form (B1–B8).** This is the same algebra with the Moore–Penrose inverse: $M_{\hat F}=I_{T_0}-\hat F\hat F^{+}$, $\hat\beta=(\sum\tilde X_i'\tilde X_i)^{+}(\sum\tilde X_i'\tilde Y_i)$, $\hat\Lambda_i=\hat F^{+}X_i$ and $\hat\gamma_i=\hat F^{+}(Y_i-X_i\hat\beta)$. It is identical to (6)–(7) when $\hat F$ has full column rank.

## The incremental estimand (the paper's central device)

The Wave-2 ATT is a **total level gap** relative to non-participation. It can therefore carry deviations inherited from 2014–2021. To isolate the *additional* post-2022 departure, the paper defines:

**Late-Wave-1 baseline gap:**

$$\hat b_i=\frac13\sum_{t=2019}^{2021}\hat g_{it} \tag{13}$$

**Stable-gap continuation counterfactual:**

$$\widehat Y^{\text{cont}}_{it}=\widehat Y_{it}(\infty)+\hat b_i,\quad t\in\mathcal W_2 \tag{14}$$

**Incremental ATT:**

$$\widehat{\text{ATT}}_{\text{Inc},2}=\frac{1}{N_T}\sum_{i\in\mathcal T^*}\Big[\frac13\sum_{t=2022}^{2024}\hat g_{it}-\frac13\sum_{t=2019}^{2021}\hat g_{it}\Big] \tag{15}$$

- **Identifying assumption ("stable gap"):** without the escalation, each country's gap would have stayed at its 2019–2021 level. The assumption allows a persistent gap left by the 2014 regime, but rules out further drift in that gap or in the average prediction error after 2021.
- **Units:** $100[\exp(\hat\theta)-1]$ is exact and $100\hat\theta$ is the small-effect approximation. For example, $0.072\to7.47\%$.

## Identification conditions and inference

**A causal reading requires three things:**

1. the donor factor spans the common shocks relevant to coalition members;
2. the pre-2014 relationships remain informative after 2013;
3. the timing, anticipation and spillover conditions hold.

If any of these fails (for example war-proximity uncertainty, targeted supply cuts, or a post-2022 change in loadings), the estimates should be read as **realized divergences associated with membership**, not as causal effects of legal sanctions. The 2022 treatment is explicitly a *bundle*: legal sanctions, Russian supply cuts, energy reallocation and domestic compensation.

**Country-dispersion SE (source of all significance stars):**

$$\widehat{\text{SE}}_{\text{CD}}=\frac{\operatorname{sd}(d_i)}{\sqrt{N_T}},\qquad d_i\in\{\bar g_{i,\mathcal W},\ d_{i,\text{Inc},2}\} \tag{16}$$

- p-values use a Student $t$ distribution with $N_T-1$ degrees of freedom.
- Collapsing each country to one contrast avoids treating years as independent.
- However, it **does not propagate first-stage uncertainty** from $\hat\beta$, the loadings or $\hat f_t$, and it assumes independent countries.

**Complementary checks (500 draws each):**

- **Country-history bootstrap:** resamples whole country paths and re-estimates the full design, which propagates estimation uncertainty.
- **Region-block bootstrap:** resamples World Bank regions, which preserves regional dependence.
- **Size-matched placebo:** false coalitions drawn from never-treated countries, with a plus-one correction. This is a **diagnostic tail probability, not a CI and not the source of the stars**. The pseudo-treated/pseudo-donor split is 42/42 for sanctions and 29/67–68 for direct arming.

## Russia-energy exposure gradient

**Exposure.** Monetary exposure (not physical dependence) is built from UN Comtrade in current USD, as a ratio of sums, and standardized within the regression sample:

$$E^{\text{energy}}_{iw}=\frac{\sum_{t\in w}\big(\text{Coal}^{\text{RUS}}_{it}+\text{Crude}^{\text{RUS}}_{it}+\text{Refined}^{\text{RUS}}_{it}+\text{Gas}^{\text{RUS}}_{it}\big)}{\sum_{t\in w}\text{GDP}_{it}} \tag{17}$$

$$\text{GasShare}_{iw}=\frac{\sum_{t\in w}\text{RussianGas}_{it}}{\sum_{t\in w}\text{WorldGas}_{it}} \tag{A1}$$

$$\text{EnergyGDP}_{iw}=\frac{\sum_{t\in w}(\text{RussianNonGas}_{it}+\text{RussianGas}_{it})}{\sum_{t\in w}\text{GDP}_{it}} \tag{A2}$$

- **Windows:** 2018–2021 (near pre-war) and 2010–2013 (pre-Crimea).
- **Data discipline:**
  - A missing API row is not treated as zero. A zero counts as verified only if a single-country query succeeds *and* the country reported HS data that year.
  - Russian exporter mirror flows recover only LNG, so the AUT, DEU, POL and TUR windows are **excluded rather than set to zero**.
  - Interpolation is in $\log(1+x)$, within a window only, for internal gaps of ≤2 years. Edge extrapolation of ≤1 year requires ≥3 anchors.
  - Ratios are **recomputed from completed components**, never interpolated directly. There are 37 primitive imputations in total.
  - Strict rule: ≥3 usable years and ≤1 imputed year per component.
- **Samples:** 118 countries (2018–21) and 119 (2010–13), with **39** sanctioners, because AUT, DEU and POL are dropped. These are exposure gradients in a reliable-data subsample, **not a decomposition of the 42-country ATT**.

**Interaction model.** This is a *different* estimator: CCE on observed data, with no controls. Its factor is the outcome CSA *outside the policy group*, and its omitted period is **all pre-2022 years**, not 2019–2021. Standard errors are clustered by country.

$$Y_{it}=\mu_i+\lambda_i\bar Y_{C,t}+\beta_1(S_i\cdot Post_t)+\beta_2(E_{iw}\cdot Post_t)+\beta_3(S_i\cdot E_{iw}\cdot Post_t)+u_{it} \tag{18}$$

How to read the coefficients:

- $\beta_1$ is the sanctioner differential at mean exposure. It is not the incremental ATT.
- $\beta_2$ is the exposure slope among non-sanctioners.
- $\beta_3$ is the slope difference between sanctioners and non-sanctioners.
- **$\beta_2+\beta_3$ is the within-coalition exposure slope**, with its SE taken from the full covariance matrix.

**Results (Table 10), 2018–2021 window first and 2010–2013 window in brackets:**

| Outcome | $\hat\beta_1$ | $\hat\beta_3$ | $\hat\beta_2+\hat\beta_3$ |
|---|---|---|---|
| log CPI | 0.0579*** | **0.0195**** [0.0089] | **0.0213*** ** [0.0173***] |
| Inflation (pp) | 2.3054*** | 0.5919 [0.5140] | **0.6368*** ** [0.5959***] |
| log GDPpc | 0.0596*** | −0.0189 [−0.0205*] | −0.0147* [−0.0072] |
| GDPpc growth (pp) | −0.9502*** | **−1.2875*** ** [−1.3592***] | **−0.9580*** ** [−0.6372***] |

- The **growth penalty is the most robust result**: it is significant in the within-coalition slope and in the triple interaction, in both windows. In Table B7, the within-coalition slope stays significant under the strict-arming definition, the gas-share measure and TWFE. $\hat\beta_3$ loses significance only under the gas-share measure (−0.71, n.s.).
- The **CPI slope** is robust within the coalition. The *coalition-specific* steepness ($\beta_3$) is significant only for near-pre-war exposure.
- The **inflation** slope difference ($\beta_3$) is never significant.
- The **GDPpc level** gradient is not robust: its sign flips under TWFE (+0.0611***).
- Overall, the evidence is consistent with an energy-cost channel, but this is **not formal mediation**.

## Conclusions and policy implications

1. **Monitor prices and inflation, not just GDP.** A coalition can avoid a robust output loss while still carrying a broad price-level burden. The paper does not estimate who bears it (households vs firms).
2. **Measure escalations against the immediate pre-escalation reference**, and report level gaps alongside rates so that accumulated positions are not mistaken for ongoing momentum.
3. **Pre-crisis exposure enables ex-ante vulnerability screening.** Bilateral energy imports/GDP predict where price and growth pressure concentrates. This argues for coordinated procurement, emergency supply plans and exposure-sensitive burden sharing.
4. **Scope limit:** sender costs ≠ sanctions failure. The paper estimates neither the costs imposed on Russia nor the strategic benefits.

## Issues a replicator must resolve

These are points where the text is internally inconsistent or incomplete:

1. **Unit mislabel in the Introduction.** "≈2.2 *percentage points*" refers to $\hat\beta_2+\hat\beta_3=0.0213$ **log points** (≈2.15%) on log CPI. Percentage points apply only to the inflation slope (0.64 pp).
2. **Inflation units differ across tables.** Tables 4 and 9 use decimals (0.025, 0.037), consistent with "÷100" in Table A1. Table 10 and B7 use annual percentage points (2.3054, 0.6368). The exposure regressions must therefore be run on undivided rates, or rescaled; the paper does not state which.
3. **Missing appendix material.** The text cites equations (B10)–(B13) and the "Section B.3" bootstrap and placebo algorithms, but neither appears in the provided document. Equation B2 is skipped, and B5–B6 rename $\hat\alpha_i$ as $\hat\gamma_i$. Bootstrap resampling details (stratification, re-drawing of donors vs targets, handling of failed draws, "500 *valid*" draws) are unspecified.
4. **Ever-treated Wave 1 aggregation.** For BIH, KOR, NZL, SGP (2022) and MDA (2023), every Wave-1 year and the 2019–2021 reference are *pre-entry*. Their "Wave-1 gap" is pure model fit, which makes the Wave-1 ATT a mixture of treated and untreated years.
5. **Table B6 Panel A (direct arming, Wave 2).** The active-year ATT (0.077, 0.039) equals the Table 9 Wave-2 *level* ATT exactly. This is mechanical, because all 29 countries are active in 2022–2024, and it is a level, not an incremental, quantity. Its Wave-1 row rests on N = 4.
6. **The two estimators use different factor definitions.** The main estimator uses the never-treated donor CSA; Eq. (18) uses the CSA "outside the policy group" without controls and with an all-pre-2022 omitted period. $\beta_1$ is therefore not comparable to the incremental ATT, as the paper acknowledges.
7. **Stars on GDPpc are SE$_{CD}$-only.** The 0.025*** coexists with a placebo p of 0.102 and a factor-proxy range of [0.007, 0.099]. The stars overstate the evidential weight.
8. **Cosmetic.** Table B5's `tabular{lccccc}` declares 6 columns but has 5 headers. Figures 1–4, B1 and B2 are placeholders.

## **Discussion of the Inputs, Processes and Outputs (IPOs) of Key Callables**

This discussion covers the 33 final principal entry points, i.e. orchestrator callables, that were designed to operationalize the "*Do Sanctions Backfire? New Evidence on the Macroeconomic Effects of Supporting Ukraine*" study. They are presented in notebook task order.

Task numbers identify notebook cells, not runtime stage numbers: Task 33 registers 31 execution stages because Task 16 is invoked through Task 17. Algorithm labels in the configuration are implementation identifiers; the supplied manuscript does not print corresponding named algorithms. Appendix B prints B1 and B3–B8; its references to B10–B13 do not supply additional displayed equations, so the estimands below are tied to the actual main-text Equations (11)–(16).

*Notation: a pair is a policy/outcome combination; $\mathcal{T}^*$ is its estimation target set, $\mathcal{C}_Y$ its outcome-specific donor set, and $\mathcal{S}_Y=\mathcal{T}^*\cup\mathcal{C}_Y$. The preferred design uses $\mathcal{P}=\{2001,\ldots,2013\}$, eight lagged controls, $\mathcal{W}_1=\{2014,\ldots,2021\}$, $\mathcal{W}_2=\{2022,2023,2024\}$, and reference $R=\{2019,2020,2021\}$. Manuscript numbers below are comparison targets, not claims that an empirical reproduction has been completed. These callables supply macroeconomic evidence and audits; compensation schedules and procurement decisions require a further policy layer.*

### Task 1 — `orchestrate_raw_data_ingest`

- **Inputs**: `sources`, a mapping of canonical source names to ten pandas DataFrames; `study_config`, a mapping containing the nineteen declared configuration sections.
- **Process**: restores tuples lost during JSON serialization; validates configuration structure; canonicalizes each source against its schema; checks primary keys, country codes, year domains and cross-source references.
- **Process boundary**: partitions Comtrade Block B by importer/mirror flow direction, retaining valid rows and preserving malformed rows, the original block, and independently labelled partition fingerprints.
- **Outputs**: `IngestBundle` containing `frames`, per-source `records`, restored `study_config`, `config_fingerprint`, `ingested_utc`, `block_b_malformed`, `block_b_original` and `block_b_partition_digests`.
- **Transformation**: heterogeneous supplied tables become schema-conforming source frames and auditable identities; the public Block B record is updated to identify the retained frame rather than its pre-partition source.
- **Research role**: source/provenance entry boundary for Appendix A.1, which states that the panel combines "macroeconomic series from the World Development Indicators, rule-of-law data from the Worldwide Governance Indicators, and author-coded policy indicators".
- **Equations**: none; this stage validates inputs upstream of the potential-outcome model and the exposure ratios.
- **Implementation distinction**: this callable ingests supplied frames; it does not itself download or authenticate World Bank, Comtrade or legal-source evidence. Hashes establish content identity, not truth of provenance.
- **Failure behavior**: missing/empty declared sources, schema violations, duplicate keys, invalid references or an invalid Block B partition are refused; undeclared extra sources are logged and ignored.

### Task 2 — `orchestrate_study_config_validation`

- **Inputs**: `study_config`, the restored study configuration; optional `freeze_path`, a filesystem destination for the configuration fingerprint.
- **Process**: validates the equation and algorithm registries, declared reproduction targets, internal identities, full series codes, country registries, reproducibility-gap resolutions and temporal universes.
- **Process boundary**: checks each audit before issuing readiness evidence; unresolved production-critical configuration gaps prevent a successful validation rather than receiving invented defaults.
- **Outputs**: `ConfigValidationReport` containing eight matrices (`equation_matrix`, `algorithm_matrix`, `target_matrix`, `consistency_matrix`, `series_matrix`, `country_matrix`, `gap_matrix`, `temporal_matrix`), fingerprint, UTC timestamp, validated-row count and `readiness_token`.
- **Transformation**: configuration declarations become cross-checked parameter/target audit tables and a readiness token bound to the configuration content; `freeze_path`, when supplied, persists its fingerprint.
- **Research role**: reproducibility specification for the design in Sections 3–4 and Appendices A–B; the availability paragraph explicitly says the submission "does not include a replication archive".
- **Equations**: registry coverage concerns the paper's displayed formulas; validation computes no economic estimand itself.
- **Implementation distinction**: the nineteen-section schema, readiness token, hashing protocol and algorithm identifiers are engineering contracts, not additional manuscript equations or independently verified extraction vintages.
- **Failure behavior**: failed registry, consistency, temporal, target or unresolved-gap audits raise an error before downstream scientific execution.

### Task 3 — `orchestrate_data_cleansing`

- **Inputs**: `bundle` (`IngestBundle`) and `study_config`; optional `iso3_registry`, `iso3_registry_path` and `country_mask`; `universe_minimum` defaults to 130.
- **Process**: canonicalizes missing markers and country-source records against an ISO3 registry; resolves WGI release choices; records the selected HS family registry and applies it to the non-gas trade block.
- **Process boundary**: compares the default WGI release rule with the release closest to the declared missing-cell target; differences in coverage, release or values are retained for review rather than silently equated.
- **Outputs**: `CleansedBundle` with cleansed `frames`, a `DataCleanseReport`, per-frame `frame_digests`, configuration and its fingerprint, and `cleansed_utc`.
- **Transformation**: raw source representations become normalized long-form inputs, a retained candidate universe and explicit exclusion/vintage/product-code evidence; no completion of analytical missing cells is performed here.
- **Research role**: prepares Appendix A's separately documented source layers; the manuscript stresses that "policy-based donor exclusions are distinct from missing-data exclusions".
- **Equations**: none; cleansing precedes raw-coverage eligibility and the construction of the eight-control design.
- **Implementation distinction**: ISO3/HS registry selection and the WGI release-selection algorithms close implementation ambiguities; their choices are not uniquely prescribed by the supplied manuscript.
- **Failure behavior**: unavailable/invalid registries, impure source frames, inconsistent cell-status counts or an insufficient candidate universe are refused; downstream tasks verify the cleansed handoff digests.

### Task 4 — `orchestrate_eligibility_screening`

- **Inputs**: `bundle` (`CleansedBundle`); optional `policy_ledger`, `metadata_frame`, `expected_otherwise_eligible`, `expected_final`; `enforce_declared_gates` defaults to true.
- **Process**: calculates variable-specific raw missingness over 2000–2024; applies the four-outcome ceiling, concentrated-control-gap restrictions and the six-year critical-outcome screen before any completion.
- **Outputs**: `EligibilityScreen` with frozen Boolean `mask`, `exclusion_ledger`, `missingness`, `rule_summary`, fingerprint, screening timestamp and notes.
- **Transformation**: raw observation-presence indicators become an invariant country inclusion decision and a localized reason for every exclusion; supporting 1996–1999 rows do not enter missingness denominators.
- **Research role**: Appendix A.2 states "Eligibility is finalized before any completion operation". The declared sequence is 131 otherwise eligible countries, exclusion of Syria and Sudan by the critical screen, then 129 retained countries.
- **Rules**: writing $m_{iv}$ for raw missingness over 25 years and $m^{C}_{iv}$ for missingness over 2019–2024, the published admission rule is:

$$
\max_{v\in\mathcal{O}}m_{iv}\le0.35,\qquad \sum_{v\in\mathcal{X}}\mathbf{1}\{m_{iv}>0.25\}\le1,\qquad \max_{v\in\mathcal{X}}m_{iv}\le0.65,\qquad \max_{v\in\{\mathrm{CPI},\mathrm{GDPpc}\}}m^{C}_{iv}\le0.25.
$$

- **Implementation distinction**: optional policy information is used to demonstrate treatment blindness, not to determine admission; country metadata must not substitute policy roles for coverage eligibility.
- **Validation**: the critical ceiling permits at most one missing year out of six; production count gates and treatment-blindness checks must pass before the mask is released.

### Task 5 — `orchestrate_control_completion`

- **Inputs**: `bundle` (`CleansedBundle`); optional retained-country `mask`, `expected_repaired_total`, `expected_balance_derived`; `enforce_declared_gates` defaults to true.
- **Process**: verifies the source handoff; completes controls using internal linear interpolation, trailing carry-forward, short leading carry-back and bounded local-trend backcasting; separately completes fuel-import/export shares and derives their difference.
- **Outputs**: `ControlCompletionResult` with complete long-form `tensor`, combined `ledger`, `method_summary`, fingerprint, completion timestamp and notes, including anchor-only and fallback records.
- **Transformation**: missing control cells become deterministic values based on the country's own series; 1996–1999 observations support completion but are not emitted into the 2000–2024 analytical tensor.
- **Research role**: Appendix A.2 requires "a common deterministic hierarchy" after eligibility is fixed. Table A3 reports 1,471 missing control cells out of 25,800; the implementation additionally gates 203 derived fuel-balance repair cells.
- **Algorithm**: internal gaps use linear time interpolation; leading spans of at most three years use the first valid value; longer spans fit the first five valid anchors and apply the bound below. Here $s_i$ is the sample SD of those five anchors; generated values for trade openness, investment and government consumption are additionally floored at zero.

$$
\widetilde{x}_{it}=\min\{U_i,\max\{L_i,\widehat{a}_i+\widehat{b}_it\}\},\qquad L_i=\min(x_i^{\mathrm{obs}})-s_i,\quad U_i=\max(x_i^{\mathrm{obs}})+s_i,\qquad \mathrm{FuelBalance}_{it}=\mathrm{FuelImportShare}_{it}-\mathrm{FuelExportShare}_{it}.
$$

- **Implementation distinction**: using five-anchor dispersion for the bound and falling back to first-observation carry-back when fewer than five anchors exist are specific code choices. P1–P4 tags and derived-balance union counts are audit categories, not manuscript cell-level method shares or numbered equations.
- **Validation**: verifies the unchanged eligibility mask, exact country/control/year product, no missing controls, no emitted anchor rows and declared repair totals; the production universe is 129 countries.

### Task 6 — `orchestrate_outcome_repair`

- **Inputs**: `bundle` (`CleansedBundle`), optional `mask`; `carry_forward_cells` defaults to BIH's CPI and inflation cells in 2024; `leading_span` defaults to 2000–2007; optional `expected_total`, `expected_by_outcome`, `expected_carry_forward`; `enforce_declared_gates` defaults to true.
- **Process**: enumerates missing outcome cells before filling; admits only approved leading backcasts and the two carry-forward cells; fits leading trends to up to the first five observed anchors, requiring at least two, then bounds extrapolations by the full observed series' range plus/minus its sample standard deviation.
- **Outputs**: `OutcomeRepairResult` with completed `tensor`, cell-level `ledger`, `repair_summary`, fingerprint, repair timestamp and notes identifying reduced-anchor fits.
- **Transformation**: approved raw-missing CPI/inflation values become country-series reconstructions; observed values are preserved exactly, while raw GDP-per-capita levels and growth require no repair in the declared sample.
- **Research role**: Appendix A.2 states "the outcome-repair layer does not interpolate internal gaps" and identifies 55 leading cells during 2000–2007, plus two BIH trailing cells: 25 CPI and 32 inflation repairs overall.
- **Algorithm**: the bounded trend form under Task 5 is applied independently to each outcome, with $s_i$ replaced by the full observed outcome-series SD rather than the control routine's five-anchor SD; the explicitly authorized trailing rule is:

$$
\widetilde{Y}_{\mathrm{BIH},2024}=Y_{\mathrm{BIH},2023},\qquad Y\in\{\mathrm{CPI},\mathrm{InflationRate}\}.
$$

- **Implementation distinction**: the five-anchor/minimum-two and sample-dispersion conventions are explicit code choices where the outcome-backcast prose is less specific; they must remain in the repair ledger.
- **Validation**: enumerated and repaired keys/methods must coincide; unapproved/internal gaps, changed observed cells, remaining missing values or violations of the 57/25/32/two-cell gates are refused.

### Task 7 — `orchestrate_panel_assembly`

- **Inputs**: completed `outcome_tensor` and `control_tensor` DataFrames; optional `units` and `expected_shape`; `enforce_declared_gates` defaults to true.
- **Process**: joins transformed outcomes and eight controls; logs positive CPI and real GDP per capita; creates analysis-rate fractions and a separate percentage-point mechanism-rate product; shifts controls by one year within country and freezes pre-treatment arrays.
- **Outputs**: `AnalysisPanel` containing `panel`, `mechanism_rates`, transformation `registry`, `lagged`, per-country/outcome `pre_treatment` matrices, panel/matrix fingerprints, assembly timestamp and notes.
- **Transformation**: repaired levels and rates become the 129-country by 25-year by 12-variable analytical panel; each preferred pre-treatment design has 13 outcome rows and a $13\times8$ lagged-control matrix.
- **Research role**: Table A1 specifies "Controls are completed before being lagged and enter the model with a one-year lag" and CPI's natural logarithm only after completion.
- **Transformations**: $\pi_{it}$ and $q_{it}$ denote source inflation and GDP-per-capita growth expressed in percent; the model's row for year $t$ uses the previous year's control vector.

$$
y^{P}_{it}=\log(\mathrm{CPI}_{it}),\quad y^{Q}_{it}=\log(\mathrm{RealGDPpc}_{it}),\quad y^{\pi}_{it}=\pi_{it}/100,\quad y^{q}_{it}=q_{it}/100,\qquad \mathrm{row}_t(X_i)=X'_{i,t-1}.
$$

- **Implementation distinction**: the analysis-rate fractions and retained percentage-point mechanism rates are distinct products; CPI index levels are not interpreted as cross-country absolute price comparisons.
- **Validation**: requires complete aligned keys, positive log inputs, declared dimensions and finite 2001–2013 matrices; the canonical key constants are `country_code` and `year`.

### Task 8 — `orchestrate_treatment_coding`

- **Inputs**: documentary `ledger` DataFrame and country `universe`; optional `expected_categories`; `expected_pre_escalation` defaults to CZE/POL/USA entry in 2018 and TUR in 2019; `primary_rank_max` defaults to three.
- **Process**: filters and ranks documentary evidence; resolves country/policy entry decisions; records rejected or superseded evidence; constructs cumulative strict-sanctions and direct-arming country-year indicators.
- **Outputs**: `TreatmentCodingResult` with `resolved` ledger, `indicators`, `decision_log`, `dropped`, category `summary`, fingerprint, resolution timestamp and notes.
- **Transformation**: accepted legal/alignment or qualifying arming delivery/commitment evidence becomes the first entry year for each country/policy and an annual zero/one path; evidence provenance remains attached to the decision.
- **Research role**: Appendix A.3 states "Strict-sanctions codes require an official legal instrument or a formal alignment notice"; arming evidence must satisfy the narrow delivery/commitment definition, not general non-lethal or financial support.
- **Coding rule**: for policy $p$, $e_i^p$ is the resolved first qualifying year; countries without qualifying evidence remain zero.

$$
D_{it}^{p}=\mathbf{1}\{e_i^p\text{ exists and }t\ge e_i^p\},\qquad p\in\{\mathrm{strict\ sanctions},\mathrm{direct\ arming}\}.
$$

- **Implementation distinction**: the evidence-ranking ceiling is a coding-registry convention. Ukraine remains sanctions-coded descriptively; removing it from the main target estimand occurs at partition construction.
- **Validation**: checks mutually exclusive descriptive categories, Ukraine handling and exactly four pre-2022 arming countries; each accepted entry must retain its documentary decision trail.

### Task 9 — `orchestrate_partition_construction`

- **Inputs**: treatment `indicators`, analytical `panel`, country `metadata` and `universe`; optional `expected_counts` and `expected_roles` declarations.
- **Process**: validates complete binary cumulative treatment histories; derives ever-treated/never-treated sets; applies geopolitical target/donor exclusions and outcome-specific exclusions; registers eight policy/outcome partitions.
- **Outputs**: `PartitionRegistry` containing pair-keyed `partitions`, roster `registry`, `counts`, combined `exclusions`, fingerprint, construction timestamp and notes.
- **Transformation**: descriptive policy paths and panel coverage become distinct estimation rosters; each pair has a frozen target tuple and an eligible, disjoint donor tuple rather than a shared universal donor pool.
- **Research role**: Equation (1) defines treatment-history roles; Table 2 notes "Treatment and donor roles are defined separately for each policy and outcome". Main sanctions use 42 targets/84 donors; direct-arming log CPI/GDPpc use 29 targets/96 or 97 donors.
- **Equation (1), before the additional estimation exclusions**:

$$
\mathcal{T}=\{i:\max_tD_{it}=1\},\qquad \mathcal{C}=\{i:\max_tD_{it}=0\}.
$$

- **Implementation distinction**: Russia, Belarus and Ukraine are excluded from every donor pool; Ukraine is also removed from the main sanctions target set. Descriptive eligibility is preserved separately from these policy-based restrictions.
- **Validation**: requires roster disjointness, country/time completeness, exclusion consistency and supplied count/role identities; membership is ever-treated, not restricted to each country's active treatment years.

### Task 10 — `orchestrate_donor_factor`

- **Inputs**: `registry` (`PartitionRegistry`), analytical `panel` and `lagged`; optional `units_by_pair`; `window` defaults to 2001–2013 and `factor_window` to 2000–2024.
- **Process**: calculates each outcome's yearly mean over its registered donors; adds a constant; assembles pre-treatment factor, dependent-variable and lagged-control matrices in common year/unit order; diagnoses dispersion, rank and conditioning.
- **Outputs**: `DonorFactorResult` with pair-keyed `factors`, long `proxy_frame`, `diagnostics`, fingerprint, construction timestamp and notes.
- **Transformation**: donor country-year outcomes become a contemporaneous common-shock proxy available through 2024; target/donor pre-treatment records become pooling arrays, without using target post-treatment outcomes in the proxy.
- **Research role**: estimator Step 1 specifies that "the CSA is calculated using only the never-treated donor pool". The preferred factor is parsimonious, not the full CCE proxy of outcome and all covariate averages.
- **Equations (3)–(4)/Appendix B1**: the outcome model allows common control coefficients and heterogeneous common-shock loadings; the preferred factor is constructed below.

$$
\begin{aligned}Y_{it}(\infty)&=X'_{i,t-1}\beta+\alpha_i'f_t+\varepsilon_{it},\\\bar{Y}_{t,\mathcal{C}_Y}&=\frac{1}{|\mathcal{C}_Y|}\sum_{j\in\mathcal{C}_Y}Y_{jt},\qquad \hat f_t=\begin{pmatrix}1\\\bar{Y}_{t,\mathcal{C}_Y}\end{pmatrix},\qquad \hat F=[\hat f'_{2001};\ldots;\hat f'_{2013}].\end{aligned}
$$

- **Implementation distinction**: absent `units_by_pair`, pooled units are the pair's targets plus donors. Numerical condition warnings describe the design; they do not verify donor immunity to spillovers.
- **Validation**: donor identity, yearly completeness, lagged-control alignment and finite matrices must hold separately for every registered pair.

### Task 11 — `orchestrate_pretreatment_fit`

- **Inputs**: `factor_result`, the Task 10 factor/design collection; `condition_max` defaults to 1,000 for the factor cross-product inversion route.
- **Process**: forms and verifies the residual-maker; residualizes outcomes and controls; estimates a common eight-coefficient vector from pooled pre-treatment observations; recovers country-specific outcome loadings and records numerical diagnostics.
- **Outputs**: `PooledFitResult` containing `betas`, `beta_names`, pair/country `loadings`, estimator `ledger`, fingerprint, fit timestamp, notes and independently checked `intermediates`.
- **Transformation**: $Y_i$, $X_i$ and $\hat F$ become common control coefficients and heterogeneous two-component loadings; all estimation observations are restricted to 2001–2013.
- **Research role**: estimator Step 2 requires "Using strictly the effective pre-treatment period"; Appendix B provides the Moore–Penrose computational form of Equations (6)–(7).
- **Appendix B3–B4 and B6**: $\hat\gamma_i$ in Appendix B corresponds to main-text $\hat\alpha_i$.

$$
\begin{aligned}M_{\hat F}&=I_{13}-\hat F\hat F^+,&\widetilde X_i&=M_{\hat F}X_i,&\widetilde Y_i&=M_{\hat F}Y_i,\\\hat\beta&=\left(\sum_{i\in\mathcal S_Y}\widetilde X_i'\widetilde X_i\right)^+\sum_{i\in\mathcal S_Y}\widetilde X_i'\widetilde Y_i,&\hat\alpha_i&=\hat F^+(Y_i-X_i\hat\beta).\end{aligned}
$$

- **Implementation distinction**: a solve is used for a sufficiently conditioned factor cross-product; otherwise SVD-based pseudo-inversion is used. The pooled coefficient follows B4's cross-product pseudo-inverse, with a separately diagnosed design-pseudo-inverse solution.
- **Validation**: verifies annihilation, symmetry, idempotence, numerical rank and orthogonality; the 1,000 ceiling and floating-point rank tolerances are code policies, not identification conditions.

### Task 12 — `orchestrate_counterfactual_projection`

- **Inputs**: `fit_result` (`PooledFitResult`), `factor_result` (`DonorFactorResult`) and `registry`; `start_year`/`end_year` default to 2014/2024; optional `observed_panel` supplies the forbidden-input perturbation channel.
- **Process**: estimates target covariate loadings from pre-treatment controls; projects covariates and outcomes using post-2013 donor factors; builds a separate fitted pre-treatment figure product; executes leakage perturbations and records unrestricted simulated ranges.
- **Outputs**: `CounterfactualResult` with `loadings`, `covariate_paths`, `outcome_paths`, `fitted_pre_paths`, `range_summary`, `invariant_audit`, fingerprint, projection timestamp and notes.
- **Transformation**: pre-treatment coefficients/loadings plus donor post-treatment outcomes become target untreated paths; realized target post-treatment covariates are not substituted into the prediction.
- **Research role**: Step 3 requires "no post-treatment outcome or covariate observed for a target country enters the construction of its own counterfactual path".
- **Equations (8)–(10)/Appendix B5, B7–B8**, for each projected target and $t\ge2014$:

$$
\hat\Lambda_i=\hat F^+X_i,\qquad \widehat X_{i,t-1}(\infty)=\hat\Lambda_i'\hat f_t,\qquad \widehat Y_{it}(\infty)=\widehat X_{i,t-1}(\infty)'\hat\beta+\hat\alpha_i'\hat f_t.
$$

- **Implementation distinction**: simulated controls are not clipped to repair bounds. Without `observed_panel`, the audit uses a synthetic sentinel and records that the realized dependency was inaccessible; this is weaker evidence than perturbing the actual panel.
- **Validation**: checks projected roster/year coverage, shapes, finiteness and invariance to forbidden inputs; `fitted_pre_paths` remains a visualization product, not part of post-treatment ATT estimation.

### Task 13 — `orchestrate_gap_and_att`

- **Inputs**: `counterfactual_result`, `observed_panel`, `registry`; defaults `wave_one` 2014–2021, `wave_two` 2022–2024, `pre_escalation` 2019–2021 and `post_escalation` 2022–2024; optional `expected_incremental` and `expected_levels` publication-cell declarations.
- **Process**: aligns observed/projected target outcomes; constructs gaps; takes country-first window means; computes level ATTs, reference gaps, continuation predictions and the incremental ATT by two algebraically equivalent routes.
- **Outputs**: `GapAttResult` with `gaps`, `country_window_means`, `level_effects`, `pre_escalation_levels`, unrounded `incremental_effects`, `publication_audit`, fingerprint, computation timestamp and notes.
- **Transformation**: observed and imputed paths become equally weighted coalition contrasts; neither GDP size nor population changes a country's contribution.
- **Research role**: Step 4 defines "the actual observed outcome minus the imputed counterfactual"; Equations (13)–(15) preserve an inherited gap and require a continuation assumption that "rules out further post-2021 drift".
- **Equations (11)–(15)**, with $N_T=|\mathcal T^*|$ and preferred windows:

$$
\begin{aligned}\hat g_{it}&=Y_{it}-\widehat Y_{it}(\infty),&\widehat{\mathrm{ATT}}_W&=\frac1{N_T}\sum_i\frac1{|W|}\sum_{t\in W}\hat g_{it},\\\hat b_i&=\frac13\sum_{t=2019}^{2021}\hat g_{it},&\widehat Y^{\mathrm{cont}}_{it}&=\widehat Y_{it}(\infty)+\hat b_i,\\\widehat{\mathrm{ATT}}_{\mathrm{Inc},2}&=\frac1{N_T}\sum_i\frac13\sum_{t=2022}^{2024}(Y_{it}-\widehat Y^{\mathrm{cont}}_{it})& &=\frac1{N_T}\sum_i\left[\frac13\sum_{t=2022}^{2024}\hat g_{it}-\hat b_i\right].\end{aligned}
$$

- **Implementation distinction**: all sums over $i$ above are over $\mathcal T^*$; fitted pre-period gaps are excluded. Publication comparisons respect units, displayed precision, rounding intervals and separately declared allowances.
- **Interpretation**: a log-point estimate $\hat\theta$ converts exactly to $100[\exp(\hat\theta)-1]\%$; rate-fraction effects require multiplication by 100 for percentage points, not exponentiation.

### Task 14 — `orchestrate_country_dispersion_inference`

- **Inputs**: `gap_result` (`GapAttResult`); `pre_window`/`post_window` default to 2019–2021/2022–2024; optional `expected_se`, `expected_stars`, `expected_probability` publication declarations and parquet `destination`.
- **Process**: forms one complete contrast per target for each level/incremental estimand; calculates the coalition mean, sample-dispersion SE, Student-t statistic/probability and significance marker; verifies publication-cell and marker-source consistency.
- **Outputs**: `InferenceResult` with `ledger`, country `contrasts`, `critical_values`, `publication_audit`, ledger/input-gap fingerprints, computation timestamp and notes; optional serialized inference ledger.
- **Transformation**: target paths become independent-country summary contrasts before inference; annual observations are not separately counted as independent replicates.
- **Research role**: Equation (16) uses the "sample standard deviation" of complete country contrasts; the manuscript states this measure "does not propagate all first-stage uncertainty".
- **Equation (16) and the implementation's Student-t convention**:

$$
\bar d=\frac1{N_T}\sum_i d_i,\qquad s_d^2=\frac1{N_T-1}\sum_i(d_i-\bar d)^2,\qquad \widehat{\mathrm{SE}}_{\mathrm{CD}}=\frac{s_d}{\sqrt{N_T}},\qquad t=\frac{\bar d}{\widehat{\mathrm{SE}}_{\mathrm{CD}}},\qquad p=2\Pr(T_{N_T-1}\ge|t|).
$$

- **Implementation distinction**: Student-t reference probabilities with $N_T-1$ degrees of freedom are explicit code conventions; markers use strict thresholds 0.10/0.05/0.01 and originate exclusively from country-dispersion inference.
- **Limitation**: donor-factor, coefficient and loading uncertainty, and cross-country dependence, are not propagated by this SE; bootstrap and placebo diagnostics remain separate outputs.

### Task 15 — `orchestrate_baseline_pipeline`

- **Inputs**: `study_config`; `data` containing `registry`, `panel` and `lagged`; optional `pairs`, `reporting_windows`, `expected_incremental` and `expected_stars`.
- **Process**: freezes the configuration and resolves canonical windows once; for each pair, independently builds its donor factor, pooled fit/loadings, projected untreated paths, gap estimands and country-dispersion inference.
- **Outputs**: `BaselinePipelineResult` containing pair-keyed `results`, step `validation`, object `freshness`, shared `effective_windows`, `declared_comparisons`, collection fingerprint, completion timestamp and notes.
- **Transformation**: a frozen partition registry and completed panel become complete per-pair baseline results; factor/coefficient/loading state is not reused across incompatible policy/outcome designs.
- **Research role**: implements the main model, "The estimator can be understood in five steps": donor proxy, pre-treatment estimation, untreated projection, gaps and aggregation, followed by Equation (16) inference.
- **Equation coverage**: the stage composes Equations (4)–(16), as detailed under Tasks 10–14; it introduces no alternative treatment-effect estimator.
- **Implementation distinction**: `reporting_windows` overrides are resolved and logged as sweep variants. Publication comparisons use declared cells, while software/data-contract failures remain independent of manuscript agreement.
- **Validation**: verifies immutable configuration and parent inputs, complete requested-pair coverage, fresh scientific objects and algebraic/inference consistency; absent required data keys are refused.
- **Interpretation**: this is donor-factor imputation with heterogeneous loadings, not a conventional two-way fixed-effects DiD; causal interpretation still requires stable relationships, appropriate timing and limited donor spillovers.

### Task 16 — `orchestrate_resampling_inference`

- **Inputs**: `policy_name`, `outcome_name`, `baseline_config`, frozen `data`, explicit registered `seeds` and `region_map`; `bootstrap_replicates` and `placebo_replicates` default to 500; `attempt_cap_multiple` defaults to five; optional `expected` reference values.
- **Process**: spawns separate random streams; resamples target/donor country histories within original policy roles, resamples complete region blocks, and assigns size-matched false coalitions from never-treated donors; fully refits the baseline design for every attempted draw.
- **Outputs**: `ResamplingResult` with two percentile interval pairs, `p_placebo_hat`, `discarded_counts`, `full_draw_arrays`, `fragile_flags`, seed material, all `attempt_audits` and `fragility_audit`.
- **Transformation**: whole panel histories become distributions of re-estimated coalition contrasts; repeated country/block occurrences receive distinct identities, preserving their multiplicity rather than being deduplicated.
- **Research role**: Section 4.3 requires a bootstrap that "resamples whole country paths and re-estimates the complete design"; the false-coalition result is "only a diagnostic tail probability".
- **Resampling summaries**: $M$ below is the number of accepted placebo draws, and $Q$ denotes empirical quantiles of accepted complete-design bootstrap estimates.

$$
\widehat p_{\mathrm{pl}}=\frac{1+\sum_{m=1}^{M}\mathbf1\{|\hat\theta^{(m)}|\ge|\hat\theta^{\mathrm{obs}}|\}}{1+M},\qquad \mathrm{CI}_{0.95}=[Q_{0.025}(\hat\theta^*),Q_{0.975}(\hat\theta^*)].
$$

- **Implementation distinction**: country-history sampling is stratified by original treatment role; the manuscript does not uniquely specify that stratification, branch spawn keys, attempt cap or rejection-fragility threshold.
- **Validation**: stores accepted and rejected draws with seed/refit lineage; refuses invalid branch coverage and never derives main-table stars from placebo probabilities or bootstrap endpoints.

### Task 17 — `execute_inference`

- **Inputs**: `study_config`, frozen `data`, primary `seeds`; optional `region_map`, authoritative `metadata_frame`, `pairs`, `expected_intervals`, `expected_placebo`; `bootstrap_replicates`/`placebo_replicates` default to 500 each; `stability_seeds` is optional in the signature but required for execution.
- **Process**: validates registered primary/second seed sets before draws; builds or authenticates the region map; executes Task 16 per requested pair under both seed sets; compares interval endpoints and placebo probabilities with explicit Monte Carlo allowances.
- **Outputs**: `InferenceExecutionResult` with `intervals`, `placebo`, primary `draws`, `stability_draws`, `stability`, `composition_audit`, `region_validation`, `attempt_audits`, `seed_audit`, `publication_audit`, `region_provenance`, fingerprint, timestamp and notes.
- **Transformation**: per-pair resampling distributions become separate inferential tables, observed block-composition evidence and reproducible second-seed stability diagnostics.
- **Research role**: Section 4.3's region bootstrap "keeps countries from the same region together"; Tables 5/B1/B5 report complementary intervals and corrected false-coalition tail probabilities.
- **Equation coverage**: applies the percentile and plus-one formulas under Task 16 to each pair, while retaining the observed Task 14 country-dispersion statistic separately.
- **Implementation distinction**: a second registered seed set and numerical Monte Carlo stability rules are reproducibility additions; qualitative sign/significance agreement is not substituted for interval/probability stability.
- **Validation**: missing stability seeds fail before randomness; metadata-derived regions take precedence when supplied, and their source/version, vocabulary, memberships and actual resampled blocks are retained.
- **Scope**: inference tables report accepted/rejected counts and sample sizes; 500 valid draws is a publication design requirement, not a guarantee of negligible Monte Carlo error.

### Task 18 — `execute_sweeps`

- **Inputs**: `baseline_config`, frozen `data`; optional `pairs`, `run_ids`, policy-specific `entry_years` and `active_year_callable`, the latter mapping an entry year to an averaging window.
- **Process**: builds a fixed sweep registry; deep-copies the baseline configuration; changes one registered design axis; rebuilds affected partition/proxy objects and reruns the baseline estimator; records both successful and failed variants.
- **Outputs**: `SweepSuiteResult` with complete `registry`, selected-run `results`, retained `failures`, successful `effective_designs`, `complete`, registry fingerprint, completion timestamp and notes.
- **Transformation**: one parent design becomes identified sensitivity branches with explicit effective rosters, factors, windows and lineage; unresolved scientific branches retain failure states instead of synthetic estimates.
- **Research role**: Sections 6.1–6.2 assess "alternative factors specifications" and reference/donor/horizon/timing changes using the same counterfactual family.
- **Design coverage**: the full eight-pair registry has 120 runs: five reference variants, three donor variants, five factor variants, one horizon variant and one active-year variant per pair.
- **Equation coverage**: unchanged branches reuse Equations (4)–(16); reference changes alter the subtraction window in Equation (15), while donor/factor changes alter the inputs to the untreated prediction.
- **Implementation distinction**: registry identity, deep-copy isolation and one-axis enforcement are engineering safeguards. Late-entry activity windows are constructed locally; no unresolved global activity mask is consulted.
- **Validation**: checks run selection/uniqueness and parent immutability; `complete` requires every selected registered run to succeed, not merely that a result frame was assembled.

### Task 19 — `execute_robustness_sweeps`

- **Inputs**: `baseline_config`, frozen `data`; optional `pairs`, `expected_ranges` and `run_ids`; numerical diagnostic `threshold` defaults to 1,000.
- **Process**: executes the reference, donor and factor axes through Task 18; separates their result tables; reconstructs actual donor additions/removals; classifies factor conditioning; reduces successful variants into axis-specific ranges.
- **Outputs**: `SweepExecutionResult` with `reference_table`, `donor_table`, `factor_table`, `ranges`, `donor_audit`, `stability_summary`, complete `coverage_audit`, fingerprint, completion timestamp and notes.
- **Transformation**: identified variant estimates become coverage-aware sensitivity intervals and lineage audits; poorly conditioned factor variants remain visible but are excluded from the implementation's headline factor range.
- **Research role**: Table 6/B2 diagnoses counterfactual sensitivity; the manuscript says GDP-per-capita results depend on "the substantive choice of proxy" and, for richer specifications, "numerical instability".
- **Range reduction**: for pair $p$ and axis $a$, let $V^{\mathrm{ok}}_{pa}$ be the successful variants admitted to the stated range; this is a sensitivity summary, not a confidence interval.

$$
\mathrm{Range}_{pa}=\left[\min_{v\in V^{\mathrm{ok}}_{pa}}\hat\theta_{pav},\ \max_{v\in V^{\mathrm{ok}}_{pa}}\hat\theta_{pav}\right].
$$

- **Implementation distinction**: the 1,000 classification cutoff and omission of diagnostic variants from headline ranges are code reporting policies; manuscript comparisons must identify the exact variant subset rather than compare unlike ranges.
- **Validation**: failed/unresolved branches are retained in coverage evidence; conclusions about price robustness or output fragility must be supported by the observed successful designs and their coverage.

### Task 20 — `execute_secondary_sweeps`

- **Inputs**: `baseline_config`; `data` with `registry`, `panel`, `lagged` and `coding`; optional `pairs`, `baseline_effects`, `expected_rates`; comparison `tolerance` defaults to 0.001.
- **Process**: refits actual baseline effects; computes the shorter 2022–2023 horizon, country-specific active-year contrasts and inflation/growth checks; records masks, countries with no active years and differences from baseline.
- **Outputs**: `SecondarySweepResult` containing `horizon_table`, `active_year_table`, `rate_table`, `mask_audit`, `exclusions`, `comparison`, fingerprint, completion timestamp and notes.
- **Transformation**: baseline gaps and annual policy indicators become alternative averaging sets; rates remain the analysis-table fraction forms until converted to percentage-point reporting.
- **Research role**: Section 5.2 cautions that a positive GDP-per-capita level gap reflects "an accumulated position relative to the counterfactual, not faster post-2022 growth"; timing checks separate active from ever-treated membership.
- **Active-year incremental rule**: $A_i=\{t\in\mathcal W_2:D_{it}=1\}$, $\mathcal T_A=\{i\in\mathcal T^*:|A_i|>0\}$; each admitted country contributes equally regardless of its active-year count.

$$
\hat\theta_A=\frac1{|\mathcal T_A|}\sum_{i\in\mathcal T_A}\left[\frac1{|A_i|}\sum_{t\in A_i}\hat g_{it}-\frac1{|R|}\sum_{t\in R}\hat g_{it}\right].
$$

- **Implementation distinction**: shortening the reporting horizon does not discard the full donor-factor history; the original gap matrix is preserved. This active-year incremental statistic is distinct from Task 30's wave-level active-year gaps.
- **Validation**: checks exact masks and roster exclusions, unique rate-outcome coverage, unchanged parent objects and equality to baseline when years, reference and country weights coincide completely.

### Task 21 — `execute_influence_diagnostics`

- **Inputs**: `baseline_config`, frozen `data`; optional `pairs`; `count` defaults to ten for presentation subsets.
- **Process**: obtains each pair's actual baseline; removes each target from the stored contrast average; removes each donor in turn and refits the entire counterfactual design; computes median, positive share and a 10% trimmed mean.
- **Outputs**: `InfluenceDiagnosticsResult` with pair-keyed `results`, combined `donor_changes`, `loo_treated`, `distributional`, fingerprint, completion timestamp and notes.
- **Transformation**: a baseline contrast vector and donor roster become country influence measures; target deletion changes aggregation, while donor deletion changes the factor and all fitted scientific objects.
- **Research role**: Section 6.3 states "Donor leave-one-out reconstructs the common factor after removing each donor"; Table 8/B4 tests whether an average is driven by individual countries or a positive tail.
- **Diagnostic formulas**: $d_i$ is the preferred country contrast, $\hat\theta=N_T^{-1}\sum_i d_i$, and $\hat\theta^{(-k)}_{\mathrm{refit}}$ denotes the fully re-estimated donor-deletion result.

$$
\hat\theta^{(-j)}_T=\frac{\sum_i d_i-d_j}{N_T-1},\qquad \Delta_k=\hat\theta^{(-k)}_{\mathrm{refit}}-\hat\theta,\qquad \mathrm{PositiveShare}=\frac1{N_T}\sum_i\mathbf1\{d_i>0\}.
$$

- **Implementation distinction**: the trimmed mean removes the floor of 10% of countries from each sorted tail; top-change ties use deterministic ordering. These are diagnostic reductions rather than a replacement estimand.
- **Validation**: retains full target/donor deletion coverage and signed changes, independently checks parent-input immutability, and reports the actual number of refits rather than only the top-ten subset.

### Task 22 — `execute_influence_tables`

- **Inputs**: `influence` (`InfluenceDiagnosticsResult`) and country `contrasts_by_pair`; `count` defaults to ten, `muted_threshold` to 0.8; optional `expected_ranges`, `expected_max_change` and `expected_distribution`.
- **Process**: reduces treated deletions to min/max/width and source countries; ranks donor changes by absolute magnitude while retaining signs; identifies the largest change and assembles distributional statistics and manuscript comparisons.
- **Outputs**: `InfluenceTableResult` with `treated_range`, `top_changes`, `max_change`, `distribution`, reconciliation-ready `combined`, fingerprint, completion timestamp and notes.
- **Transformation**: full influence-engine artifacts become interpretable reporting tables whose extrema remain traceable to actual deleted countries; the supplied contrast vectors reproduce the baseline means.
- **Research role**: Table 8/B4 reports how "The median, trimmed mean, and positive share show how widely the country-level effects agree in sign and magnitude".
- **Reporting reductions**, with $\hat\theta_T^{(-j)}$ and $\Delta_k$ defined under Task 21:

$$
L=\min_j\hat\theta_T^{(-j)},\qquad U=\max_j\hat\theta_T^{(-j)},\qquad \mathrm{Width}=U-L,\qquad k^*=\operatorname*{arg\,max}_k|\Delta_k|.
$$

- **Implementation distinction**: the treated deletion range is not a confidence interval; the positive-share threshold of 0.8 is an implementation presentation flag, not a manuscript test of treatment-effect significance.
- **Validation**: verifies coverage and declared comparison cells, forbids interval semantics for leave-one-out ranges, and retains the baseline, extreme identities and signed donor movements.

### Task 23 — `execute_matching_designs`

- **Inputs**: `baseline_config`, frozen `data` containing `registry`, `panel` and `lagged`; optional `pairs` and `expected_att`; `axes` defaults to high-income and Mahalanobis, `size_M` to five, `tolerance` to 0.001.
- **Process**: averages eight controls and log real GDP per capita over 2009–2013; restricts both groups strictly above their pooled income median or selects five nearest eligible donors per target with deterministic distance/code tie-breaking.
- **Outputs**: `MatchingDiagnosticsResult` with pair/design `designs`, actual-roster `summary`, scheme-labelled `balance`, full `design_audit`, fingerprint, completion timestamp and notes; design records retain post-gap and fitted-pre-gap matrices.
- **Transformation**: pre-treatment characteristics become restricted rosters or donor inclusion-frequency weights; the high-income branch refits the baseline, while the Mahalanobis branch constructs a weighted factor and weighted pooled model.
- **Research role**: Section 6.2 says the Mahalanobis design "selects the five nearest eligible donors for each treated country" and "weights them by selection frequency"; matching diagnoses observable comparability rather than random assignment.
- **Matching algebra**: $c_i$ is the nine-characteristic pre-period vector, $\Sigma$ its pooled sample covariance, and $n_j$ the donor's selection count.

$$
d(i,j)=\sqrt{(c_i-c_j)'\Sigma^+(c_i-c_j)},\qquad w_j=\frac{n_j}{N_TM},\quad M=5,\qquad \bar Y_t^{\mathrm{match}}=\sum_jw_jY_{jt},\qquad \mathrm{SMD}_k=\frac{\bar c_{Tk}-\bar c_{Ck,w}}{\sqrt{(s_{Tk}^2+s_{Ck,w}^2)/2}}.
$$

- **Timing fidelity**: the repaired Mahalanobis branch reads the eight `_lag1` columns from `lagged` for both the pooled coefficient and loading fits. Each $X_i$ row is $X'_{i,t-1}$ as required by Equations (3), (5)–(10) and Appendix B; absent columns, incomplete or duplicated pre-treatment years, and non-finite fitted inputs are rejected before estimation.
- **Impact qualification**: with identical factor operators for the two loading fits, fixed outcomes, rosters and donor proxy, and no clipping, Appendix B's projected terms simplify to $\widehat Y_{it}(\infty)=\hat f_t'\hat F^+Y_i$. A timing-only synthetic correction changes the pooled coefficient vector but leaves post-treatment gaps and ATTs unchanged up to floating-point error; the verified specification mismatch does not establish numerical bias in a reported ATT.
- **Fitted-pre alignment**: the high-income ATT uses the lagged baseline estimator, and its repaired fitted-pre-gap reconstruction selects the same `_lag1` control columns. The plotted pre-period residuals therefore use the regressors of the fitted model; the matched-roster and donor-frequency rules are retained.
- **Implementation distinction**: donor pooling scales histories by $\sqrt{w_j}$, targets retain weight one; weighted SMD uses frequency-expanded sample variance, with an unweighted effective-donor balance vector also retained. Zero-variance unequal means yield signed infinity, not artificial balance.

### Task 24 — `execute_matching_report`

- **Inputs**: `baseline_config`, frozen `data`; optional `pairs`, `expected_att`, `expected_balance`, already computed `diagnostics`; `size_M` defaults to five and `tolerance` to 0.001.
- **Process**: verifies supplied Task 23 diagnostics or runs them; reconstructs candidate/effective roster disposition, donor selection counts/weights and full standardized-difference vectors; checks each reported maximum against its vector.
- **Outputs**: `MatchingReportResult` with `summary`, unique-key `dropouts`, complete `weight_distribution`, long `balance`, scheme-specific `maxima`, fingerprint, completion timestamp and notes.
- **Transformation**: matched design artifacts become an auditable reporting package; zero-frequency candidate donors remain in the weight distribution even though they do not enter the fitted donor roster.
- **Research role**: Table B3 documents "the complete balance vector" and "the donor-selection frequencies used in the Mahalanobis design"; Table 7 contrasts improved comparability with the broad baseline pool.
- **Report identities**: with $p$ identifying the pair/design/weighting scheme and $k$ its balance covariate,

$$
\mathrm{MaxAbsSMD}_{p}=\max_k|\mathrm{SMD}_{pk}|,\qquad \sum_jn_j=N_TM,\qquad w_j=\frac{n_j}{N_TM},\qquad \sum_jw_j=1.
$$

- **Implementation distinction**: sample/frequency variance conventions and zero-dispersion states accompany every balance result; comparison targets are descriptive, not gates establishing causal identification or randomized balance.
- **Timing consistency**: reporting consumes Task 23's repaired lagged-control fits and aligned fitted-pre residuals; its lineage and reduction checks retain the actual rosters, weights and scientific products of those executed designs.

### Task 25 — `build_cell_ledger`

### Task 26 — `construct_exposure_primitives`

- **Inputs**: classified `ledger`, exporter `mirror_block`, supplied-audit `numerator`, world-gas `totals` and raw-audit `primitives`; optional `exclusions`, `expected_completions`, authoritative `nominal_gdp`; defaults `barrier` 2013–2018, `maximum_gap` two, `minimum_anchors` three and count `tolerance` zero.
- **Process**: derives Russian flows from ledger/mirror lineage and reconciles supplied audits; retains importer positives; corrects deficient gas totals using targeted exporter totals; completes three trade primitives separately within each window and joins authoritative GDP afterward.
- **Outputs**: `ExposureConstructionResult` with `recovered`, denominator trigger/source audit `denominator`, `completed`, `imputation_ledger`, `counts`, fingerprint, construction timestamp and notes.
- **Transformation**: screened monetary trade cells become recovered/completed Russian gas, world gas and Russian non-gas energy; GDP is neither interpolated nor replaced by a trade-series proxy.
- **Research role**: Appendix A.5 states "Importer-reported positives always have priority" and "no interpolation crosses the 2013--2018 gap"; it reports 37 primitive imputations: 13 Russian gas, 13 world gas and 11 non-gas energy.
- **Completion algebra**: for internal missing $t$ between observed years $a<b$, use $z=\log(1+x)$; one-year edge extrapolation requires three anchors and median adjacent log growth clipped to observed growth's 5th/95th percentiles.

$$
z_t=z_a+\frac{t-a}{b-a}(z_b-z_a),\qquad \widetilde x_t=\max\{0,\exp(z_t)-1\},\qquad \widetilde h=\operatorname{clip}\!\left(\operatorname{median}(\Delta z),q_{0.05},q_{0.95}\right).
$$

- **Denominator discipline**: substitute exporter totals only for a missing importer total, a non-positive total despite positive Russian trade, or a total below a mirror-recovered numerator; retained cells require $0\le\mathrm{RussianGas}\le\mathrm{WorldGas}$.
- **Validation**: the two four-year windows never share anchors; Russian gas is not imputed for the four priority countries; observed/rejected lineage, raw-audit reconciliation and supplied completion counts must hold. Production requires WDI `NY.GDP.MKTP.CD` GDP rows.

### Task 27 — `construct_exposure_ratios`

- **Inputs**: completed `components`, authoritative `completion_ledger`, exposure `windows`, `mechanism_universe`; `excluded_countries` defaults to AUT/DEU/POL/TUR; optional `expected_counts` distinguishing construction passes from regression admission.
- **Process**: reconstructs ratio-specific yearly components and imputation counts; computes window ratios of sums; applies strict/extended coverage independently by ratio; intersects with the mechanism universe and standardizes using admitted-country sample moments.
- **Outputs**: `ExposureRatioResult` containing raw `ratios`, reproducible yearly `components`, ratio-specific `sample_mask`, `standardized` values with moments/sample identity, `counts`, fingerprint, timestamp and notes.
- **Transformation**: separately completed monetary numerators and denominators become dimensionless exposures; ratios themselves are not interpolated and cannot transfer coverage from gas share to total energy/GDP.
- **Research role**: Appendix A.5 says "Ratios are not themselves interpolated". Construction covers 129 countries; regression admission intersects the 126-country universe excluding Russia, Belarus and Ukraine.
- **Equations (A1)–(A2), consistent with Equation (17)**: Russian non-gas energy comprises coal, crude oil and refined petroleum; $E$ is standardized separately within each actual window/rule/ratio sample $S$ using sample SD.

$$
\mathrm{GasShare}_{iw}=\frac{\sum_{t\in w}\mathrm{RussianGas}_{it}}{\sum_{t\in w}\mathrm{WorldGas}_{it}},\qquad \mathrm{EnergyGDP}_{iw}=\frac{\sum_{t\in w}(\mathrm{RussianNonGas}_{it}+\mathrm{RussianGas}_{it})}{\sum_{t\in w}\mathrm{GDP}_{it}},\qquad E^{z}_{iw}=\frac{E_{iw}-\bar E_{w,S}}{s_{w,S}}.
$$

- **Coverage algorithm**: strict requires at least three usable years and at most one imputation for each required component; extended requires at least two usable years and at most two imputations; both exclude the four pipeline-priority countries.
- **Implementation distinction**: standardization uses sample variance (`ddof` one) with explicit sample fingerprints; construction and mechanism counts are separately audited, preventing construction coverage from being reported as regression sample size.
- **Actual summation boundary**: when a required component remains missing, the code restricts numerator and denominator to the same stored common-finite-year subset of $w$, independently for each ratio. The displayed manuscript formulas use $w$; partial-window reproduction must reconcile this additional year-selection convention.

### Task 28 — `execute_heterogeneity_suite`

- **Inputs**: frozen `panel`, one `outcome`, keyed `exposures`, annual `sanction_indicators`, policy/outcome-specific `donor_codes`; `specifications` defaults to six registered designs; optional `expected_slopes`; `tolerance` defaults to 0.001.
- **Process**: selects the specification's ratio/window/rule sample and actual policy donor roster; consumes Task 27's sample-standardized exposures; forms three post-2022 interactions; partials out country intercepts and heterogeneous donor-mean loadings or alternative two-way fixed effects; fits OLS with country-clustered covariance.
- **Outputs**: `HeterogeneityResult` with scalar `coefficients`, full `covariance`, fitted `statistics` including ranks/sample counts/CR1 correction and provenance, fingerprint, construction timestamp and notes.
- **Transformation**: macro outcomes, predetermined exposure and coalition membership become conditional exposure gradients, separate from the donor-imputation ATT; rate fractions are restored to annual percentage points and lagged controls are omitted.
- **Research role**: Section 7 estimates "exposure gradients rather than as formal mediation"; all years before 2022 form the omitted period, rather than the preferred ATT reference alone.
- **Equation (18)**, where $S_i$ is the policy membership indicator, $P_t=\mathbf1\{2022\le t\le2024\}$ and $E_{iw}$ denotes standardized exposure in the fitted sample:

$$
Y_{it}=\mu_i+\lambda_i\bar Y_{\mathcal C,t}+\beta_1(S_iP_t)+\beta_2(E_{iw}P_t)+\beta_3(S_iE_{iw}P_t)+u_{it},\qquad \mathrm{SE}(\hat\beta_2+\hat\beta_3)=\sqrt{\widehat V_{22}+\widehat V_{33}+2\widehat V_{23}}.
$$

- **Implementation distinction**: CR1 scales the country-score sandwich by $G/(G-1)\times(n-1)/(n-r)$ using full absorbed/design rank; scalar probabilities use a normal reference, and residualized fits/covariances are checked against an explicit full design. Exposure is not restandardized here; supplied values must retain Task 27's exact-sample convention.
- **Interpretation**: $\beta_2$ is the non-policy slope, $\beta_3$ its policy-group difference, and $\beta_2+\beta_3$ the total coalition slope; $\beta_1$ is the differential at mean exposure, not the baseline incremental ATT.

### Task 29 — `execute_heterogeneity_regressions`

- **Inputs**: `panel`, `exposures`, `sanction_indicators`, policy-specific `donor_codes`; `outcomes` defaults to the four analytical outcome identifiers; `specifications` to the exact six designs; `required_invariants` defaults to false; optional `expected_estimates`; `tolerance` defaults to 0.001.
- **Process**: runs Task 28 once per outcome; verifies coefficient/covariance identities; assembles strict near-prewar and pre-Crimea Table 10 panels and the full B7 specification sweep; evaluates directional/significance findings and compares exposure windows.
- **Outputs**: `HeterogeneityExecutionResult` with `baseline_table`, `pre_crimea_table`, `table10`, `sweep_table`, `covariance_table`, `invariant_checks`, `window_comparison`, fingerprint, timestamp and notes.
- **Transformation**: four outcome-specific mechanism suites become 96 reported scalar rows and 216 covariance entries across six designs; Table 10 retains 32 scalar rows, separately identifying $\beta_1$, $\beta_2$, $\beta_3$ and $\beta_2+\beta_3$.
- **Research role**: Table B7 examines "the exposure window, sample-coverage rule, estimator, policy definition, and exposure measure" and distinguishes price levels, inflation, output levels and annual growth.
- **Design coverage**: six designs are near-prewar strict energy/GDP, pre-Crimea strict energy/GDP, pre-Crimea extended energy/GDP, near-prewar TWFE, strict direct arming and near-prewar gas share.
- **Equation coverage**: Equation (18) and the full-covariance slope formula under Task 28; the TWFE branch replaces the heterogeneous donor-mean term with common year effects only in that alternative design.
- **Implementation distinction**: four outcomes and all six designs are mandatory exactly once, despite ordering flexibility. Directional findings are descriptive; `required_invariants` cannot turn a desired empirical sign into a mathematical identity.
- **Validation**: missing designs, duplicate output keys or violated mathematical/dataflow identities are errors; narrative sign/significance failures are retained as findings rather than suppressed or used to overwrite estimates.

### Task 30 — `execute_direct_arming_pipeline`

- **Inputs**: `baseline_config`, frozen `data`, arming `entry_years`, authoritative versioned `region_map`; `outcomes` defaults to log CPI/log real GDP per capita; optional `outcome_removals`, explicit registered `seeds`, `wave_references`, `expected_baseline`; `bootstrap_replicates`/`placebo_replicates` default to, and are fixed at, 500; `tolerance` defaults to 0.001.
- **Process**: resolves outcome-specific arming target/donor designs and removals; refits baseline level/incremental effects; executes all resampling branches on those same rosters; constructs active-year wave-level gaps and separately named reference-adjusted contrasts.
- **Outputs**: `DirectArmingResult` with `baseline`, `inference`, diagnostic `interpretation`, `active_year`, `exclusions`, canonical `design_audit`, fingerprint, construction timestamp and notes.
- **Transformation**: the narrower documentary coalition and its eligible donors become a fully re-estimated boundary check, not an isolated causal effect of military aid; every inference branch is tied to the same outcome-specific design.
- **Research role**: Section 6.4 calls this "a boundary check" for a narrower coalition; 28 of the 29 arming countries also impose strict sanctions, and only four arming countries are active before 2022.
- **B6 active-year level statistic**: for wave $W$, $A_i(W)=\{t\in W:t\ge e_i\}$ and $\mathcal T_A(W)=\{i\in\mathcal T^*:|A_i(W)|>0\}$.

$$
\widehat{\mathrm{ATT}}_{A,W}=\frac1{|\mathcal T_A(W)|}\sum_{i\in\mathcal T_A(W)}\frac1{|A_i(W)|}\sum_{t\in A_i(W)}\hat g_{it}.
$$

- **Implementation distinction**: the formula above is a wave-level gap, not the preferred incremental contrast; when a wave reference is supplied, the latter is stored separately. Placebo probabilities remain diagnostics and do not supply baseline stars.
- **Validation**: explicit seed identity, 500 valid refits/assignments, authoritative regions, roster/removal lineage and supplied publication cells are checked; baseline log-CPI/GDPpc donor counts are 96/97 with 29 targets.

### Task 31 — `produce_figures_and_tables`

- **Inputs**: pair-keyed `gap_matrices` and `incremental_contrasts`; optional `matched_paths`, `placebo_draws`, `reference_window_estimates`, `table_sources`, `output_dir`, `extra_gaps`, distinct `fitted_pre_gaps`, actual `observed_preferred_att`, versioned `country_geometry` and explicit `overlay_pair`.
- **Process**: joins fitted 2001–2013 and realized 2014–2024 gaps for graphics; computes equal-country dynamic paths/pointwise ribbons normalized to 2013; builds country choropleths, matched overlays, placebo histograms, reference-window plots and supplied table renderings.
- **Outputs**: `FigureTableBundle` with reproducible `figure_data`, `tables`, table `latex`, figure `rendered` status/path records, `reproducibility_gaps`, fingerprint, construction timestamp and notes.
- **Transformation**: frozen scientific outputs become publication data and rendering artifacts; observed placebo markers come from actual preferred ATTs, not from the placebo draw mean, and matched paths come from their independent fits.
- **Research role**: Figure 4 specifies "Every path is normalized to 2013" and matched paths as lines; Figure B2 specifies "The red line is the observed preferred incremental ATT".
- **Plotting algebra**: $a_t$ is the mean gap, $s_t$ its cross-country sample SD, and $N$ the fixed target count; the implementation shifts a level-gap pointwise ribbon by the same 2013 mean.

$$
a_t=\frac1N\sum_i\hat g_{it},\qquad p_t=a_t-a_{2013},\qquad [\ell_t,u_t]=\left[a_t-1.96\frac{s_t}{\sqrt N}-a_{2013},\ a_t+1.96\frac{s_t}{\sqrt N}-a_{2013}\right].
$$

- **Implementation distinction**: this ribbon does not estimate the variance of $\hat g_{it}-\hat g_{i,2013}$ or a simultaneous band; pre-2014 graphics show fitted residual gaps, not conventional event-study leads. Figure inputs require aligned chronological columns.
- **Validation**: missing fitted-pre data, versioned geometry or rendering-stack dependencies are recorded as production gaps; table composition and requested figure data must remain traceable to supplied scientific artifacts.

### Task 32 — `execute_reconciliation_audit`

- **Inputs**: `reported`, `reproduced`, independent `declared_keys`, explicit `unreproduced`; optional `layer_chains`, `counterfactual_callable`, `imputation_callables`, `eligibility_callable`, `robustness_frame`, `study_config`, `declared_facts`, `archive_artefacts`, `manuscript_universe`, live `probe_registry`, `archive_state`, `archive_root`; defaults `version_tag` to `v1.0.0-frozen`, `tolerance` to 0.001 and `require_invariants` to true.

- **Outputs**: `ReconciliationReport` with `comparisons`, `gaps`, `localization`, `invariants`, `provenance`, `manifest`, fingerprint, timestamp, notes, `status`, granular `blockers` and `coverage`.
- **Transformation**: reproduced statistics and source/transform/model lineage become a cell-by-cell attribution record and completeness certificate; independently pinned inventory/probe evidence prevents caller-supplied success claims from self-certifying.
- **Research role**: operationalizes the Appendix A requirement that "Neither treatment indicators nor estimated treatment effects enter the coverage rules or the completion procedures" and Step 3's target-input exclusion boundary.
- **Three tested boundaries**: perturbing target post-treatment realizations must leave its untreated prediction unchanged; perturbing policy information must leave eligibility/completion unchanged; running robustness branches must preserve baseline parent state/results.
- **Comparison rules**: numerical publication cells use their stated units, displayed rounding interval and recorded numerical/Monte Carlo allowance; categorical cells require their declared exact identity. Earliest-divergent-layer attribution requires actual layer evidence, not a guessed cause.
- **Implementation distinction**: the 418-cell inventory, ten-artifact archive contract, independently pinned probes and certificate are engineering requirements. `require_invariants` false cannot waive the three scientific invariants; the nominal 0.001 tolerance is not a universal manuscript precision rule.
- **Equation boundary**: audits results of the preceding scientific equations and computation layers; it introduces no new ATT, exposure-gradient or causal-identification formula.
- **Failure/completion behavior**: malformed inventories/probes/provenance claims raise errors; well-formed unresolved inputs yield `incomplete`. A `complete` report requires every cell, invariant, provenance fact and actual archive artifact to pass verification.

### Task 33 — `orchestrate_study_pipeline`

- **Inputs**: one `data` mapping; `study_config` and the ten-frame `sources` mapping are mandatory. It also carries the stage-specific scientific inputs, registered seeds, independent manuscript/probe evidence and replication-archive dependencies required by the requested production execution.
- **Process**: validates preflight requirements and the fixed 31-stage registry; executes stages in dependency order; threads live artifacts into later stages; records completion, skips, failures and timings; checks live content digests, reconciliation coverage and self-contained archive evidence.
- **Outputs**: a dictionary containing available per-stage artifacts plus `pipeline_summary`, `pipeline_state`, `pipeline_digest`, `pipeline_notes`, `pipeline_status` and `pipeline_blockers`.

- **Research role**: integrates the Appendix A construction sequence and Sections 4–7 estimation, diagnostics and mechanisms; the paper describes a "replicable two-wave policy-evaluation framework" but states that its submission "does not include a replication archive".
- **Equation coverage**: delegates Equations (1)–(18), (A1)–(A2) and the printed Appendix B computational formulas to their respective stages; the orchestration layer estimates no new economic coefficient.
- **Completion contract**: all 31 stages must complete in registry order, with no skipped/failed stage or remaining blocker; all 418 manuscript cells, all three genuine design-invariant demonstrations, verified provenance and the complete actual archive must satisfy reconciliation.
- **Implementation distinction**: 33 principal task definitions do not imply 33 runtime stages; Task 16 runs inside Task 17. Content-sensitive hashing and readiness evidence certify execution identities and documented checks, not the maintained causal assumptions.
- **Current methodological limit**: this discussion is a source/interface audit, not a full empirical run. Task 23's two control-timing discrepancies are repaired and checked with synthetic inputs; those checks do not establish full manuscript reproduction or verify maintained causal assumptions.

### **Step 1: Synthetic Data Generation (the ten raw DataFrames)**

**Methodology.** The generator fixes 131 country candidates: the manuscript's 129 descriptive countries plus Syria and Sudan. It also places `WLD` and `HIC` aggregate rows in both WDI representations, so cleansing must use country-registry membership. Every synthetic country has all 29 raw calendar years, 1996–2024. The retained analytical window remains 2000–2024; outcomes and controls stay in raw source units.

The generator compounds annual percentage rates into positive levels:

$$
C_{it}=C_{i,t-1}\left(1+\frac{\pi_{it}}{100}\right),\qquad
Q_{it}=Q_{i,t-1}\left(1+\frac{r_{it}}{100}\right).
$$

These are **synthetic data-generation identities**, not additional manuscript estimators. CPI is rebased to a synthetic 2010 value of 100. Constant-price GDP per capita is kept distinct from current-dollar aggregate GDP. The generated nominal denominator uses the artificial population and price scale; it is not an accounting recovery of any country's actual GDP deflator. Governance levels are generated inside a bounded range and enter unchanged; no observed WGI series is winsorised. Price and output paths contain declared common/regional shocks, not a fitted paper effect.

**Required fidelity corrections to the request.**

- `[A-Z]{3}` also matches `WLD`, `HIC` and other aggregate codes. Pattern matching alone cannot exclude them. The notebook's authoritative country registry supplies the membership test; the example includes two such rows deliberately.
- The raw WDI table has **12**, not 10, primitive indicator codes: four outcomes, six direct controls, fuel imports and fuel exports. Rule of law is supplied separately by WGI. The eight model controls are the six direct controls, rule of law and the **completed-primitives fuel-share balance**.
- Table A6 retains **Russia and Belarus inside the 129-country descriptive sample** while barring their estimation roles. Ukraine is also retained but excluded from the main sanctions ATT and every donor pool. Only Syria and Sudan receive zero retained flags in this 131-country fixture.
- Table A4 contains **Croatia** in the 29-country direct-arming roster. Moldova enters sanctions in 2023; BIH, Korea, New Zealand and Singapore enter in 2022. The four pre-2022 arming countries are Czechia, Poland, Turkiye and the United States. Mexico is neither sanctions-coded nor a North America block member under the seven-region fixture convention.
- The policy source primary key is `(country_code, policy, document_title)` in the actual notebook/configuration. `source_document` is not a declared input column. Albania uses an `alignment_notice`; no EU-member regulation is fabricated for it.

### **Step 2: Loading the Configuration (`config.yaml`)**

**Methodology.** Parse the existing file with a safe **YAML 1.2** loader and reject duplicate keys. This preserves unquoted scientific numeric literals such as `1e-10` as numbers. Generic YAML 1.1 parsing can interpret this literal differently and is not the project's validated parsing path. The loader returns the plain dictionary `config` and checks its section and operative binding contracts. It deliberately leaves arrays as arrays; the notebook handles each declared window and tuple boundary. It does not call unknown provenance resolved.


The actual `AnalysisPanel.lagged` object retains all 3,225 country-year rows and appends the eight `*_lag1` columns. Those columns are NaN in 2000; the **usable** lagged-control grid has 129 × 24 observations for 2001–2024. The pre-treatment matrices select 2001–2013 explicitly. Do not interpret the presence of a 2000 row as an available lag or fitted counterfactual residual.

The research specification remains:

$$
\widehat f_t=\begin{pmatrix}1\\\overline Y_{\mathcal C_Y,t}\end{pmatrix},\qquad
M_{\widehat F}=I-\widehat F\widehat F^{+},\qquad
\widehat\beta=
\left(\sum_{i\in\mathcal S_Y}\widetilde X_i'\widetilde X_i\right)^{+}
\sum_{i\in\mathcal S_Y}\widetilde X_i'\widetilde Y_i,
\quad \widetilde X_i=M_{\widehat F}X_i,\quad
\widetilde Y_i=M_{\widehat F}Y_i.
$$

Here $X_i$ contains the **one-year-lagged** controls paired with 2001–2013 outcomes. Using the Appendix B computational orientation, $\widehat\Lambda_i=\widehat F^{+}X_i$ is $2\times8$, and $\widehat\gamma_i=\widehat F^{+}(Y_i-X_i\widehat\beta)$ is $2\times1$. For $t\ge2014$:

$$
\widehat X_{i,t-1}(\infty)=\widehat\Lambda_i'\widehat f_t,\qquad
\widehat Y_{it}(\infty)=
\widehat X_{i,t-1}(\infty)'\widehat\beta+
\widehat\gamma_i'\widehat f_t,\qquad
\widehat g_{it}=Y_{it}-\widehat Y_{it}(\infty).
$$

For the preferred reference $R=\{2019,2020,2021\}$ and $W_2=\{2022,2023,2024\}$:

$$
d_i=\frac1{|W_2|}\sum_{t\in W_2}\widehat g_{it}
-\frac1{|R|}\sum_{t\in R}\widehat g_{it},\qquad
\widehat{ATT}^{\mathrm{Inc}}_2=\frac1{N_T}\sum_i d_i,\qquad
\widehat{SE}_{\mathrm{CD}}=\frac{\mathrm{sd}(d_i)}{\sqrt{N_T}},\quad
\mathrm{df}=N_T-1.
$$

The exact percentage interpretation of a log-point estimate is $100[\exp(\widehat\theta)-1]$. Annual rate outcomes are divided by 100 once for ATT estimation; mechanism regressions use percentage-point units once. Exposure ratios are computed from separately completed current-dollar components. None of these transformations occur inside the synthetic raw-data generator. Counterfactual construction excludes target post-2013 realizations; eligibility/completion exclude treatment information; robustness never replaces baseline. Fitted dynamic gaps start in **2001**, normalize at 2013, and cannot acquire a fabricated year-2000 residual.

````python
# Finalized Configuration Contract: Scientific Rules and Runtime Handoff

# =============================================================================
# Finalized Configuration Contract: Scientific Rules and Runtime Handoff
# =============================================================================
#
# Bind the existing study catalog to the code-defined scientific rules,
# source schemas, runtime inputs and independently authenticated evidence.
# This cell declares 92 fixed bindings and 15 helpers, including the nested
# authority-pin parser. It performs no estimator fit at definition time.
#
# Scientific anchors: Equations (1), (3)-(18), (A1)-(A2), Appendix B;
# Algorithms B1-B3, C1-C3, D2-D5, E1-E2 and documentary coding G0.
# Later task cells define the result types and validators referenced here.
# Execute the notebook's import cell first and all task cells before use.
#
# Python implementation author: CS Chirinda.
# Formatting revision: 3 October 2026; fixed values and algorithms retained.
#
# =============================================================================

# Postpone annotations so later task-defined result types can be referenced
# safely.
from __future__ import annotations

# Import the concrete Matplotlib Axes type used by the rendering helper.
from matplotlib.axes import Axes


# =============================================================================
# Fixed configuration declarations: preserve values and validation order
# =============================================================================

# Expose the 92 fixed declarations through a top-level read-only mapping.
FIXED_CONFIGURATION_CONTRACT: Final[Mapping[str, Any]] = MappingProxyType(
    # Begin the explicit key-to-value mapping used by the enclosing operation.
    {
        # ---------------------------------------------------------------------
        # Section 01: Study dimensions and one-year control timing
        # ---------------------------------------------------------------------
        # Preserve the existing fixed requirement for global constants: control
        # lag years.
        "global_constants.control_lag_years": 1,
        # Preserve the existing fixed requirement for global constants: n
        # controls.
        "global_constants.n_controls": 8,
        # Preserve the existing fixed requirement for global constants: T0 pre
        # treatment years.
        "global_constants.T0_pre_treatment_years": 13,
        # ---------------------------------------------------------------------
        # Section 02: Raw-data eligibility before completion (Algorithm C1)
        # ---------------------------------------------------------------------
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # C1 eligibility screening: outcome missing max.
            "algorithm_parameters.C1_eligibility_screening.outcome_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "missing_max"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): 0.35,
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # C1 eligibility screening: control high missing max.
            "algorithm_parameters.C1_eligibility_screening.control_high_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "missing_max"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): 0.25,
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # C1 eligibility screening: control high missing count max.
            "algorithm_parameters.C1_eligibility_screening.control_high_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "missing_count_max"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): 1,
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # C1 eligibility screening: control single missing max.
            "algorithm_parameters.C1_eligibility_screening.control_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "single_missing_max"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): 0.65,
        # Preserve the existing fixed requirement for algorithm parameters: C1
        # eligibility screening: critical window.
        "algorithm_parameters.C1_eligibility_screening.critical_window": (
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            [2019, 2024]
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ),
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # C1 eligibility screening: critical outcome missing max.
            "algorithm_parameters.C1_eligibility_screening.critical_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "outcome_missing_max"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): 0.25,
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # C1 eligibility screening: fixed before completion.
            "algorithm_parameters.C1_eligibility_screening.fixed_before_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "completion"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): True,
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # C1 eligibility screening: uses policy indicators.
            "algorithm_parameters.C1_eligibility_screening.uses_policy_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "indicators"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): False,
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # C1 eligibility screening: uses estimated effects.
            "algorithm_parameters.C1_eligibility_screening.uses_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "estimated_effects"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): False,
        # ---------------------------------------------------------------------
        # Section 03: Own-history control completion (Algorithm C2)
        # ---------------------------------------------------------------------
        # Preserve the existing fixed requirement for algorithm parameters: C2
        # control completion: priority 3 max years.
        "algorithm_parameters.C2_control_completion.priority_3_max_years": 3,
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # C2 control completion: priority 4 anchor count.
            "algorithm_parameters.C2_control_completion.priority_4_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "anchor_count"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): 5,
        # Preserve the existing fixed requirement for algorithm parameters: C2
        # control completion: dispersion convention.
        "algorithm_parameters.C2_control_completion.dispersion_convention": (
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "sample_SD_of_first_five_original_trend_anchors_ddof_1"
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ),
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # C2 control completion: insufficient trend anchors.
            "algorithm_parameters.C2_control_completion.insufficient_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "trend_anchors"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): (
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "raise_for_leading_gap_longer_than_3_years_with_fewer_than_5_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "original_anchors"
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ),
        # ---------------------------------------------------------------------
        # Section 04: Restricted outcome repair (Algorithm C3)
        # ---------------------------------------------------------------------
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # C3 outcome repair: internal outcome interpolation.
            "algorithm_parameters.C3_outcome_repair.internal_outcome_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "interpolation"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): False,
        # Preserve the existing fixed requirement for algorithm parameters: C3
        # outcome repair: leading backcast window.
        "algorithm_parameters.C3_outcome_repair.leading_backcast_window": (
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            [2000, 2007]
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ),
        # ---------------------------------------------------------------------
        # Section 05: Numerical estimation, estimand identity and slope
        # covariance
        # ---------------------------------------------------------------------
        # Preserve the existing fixed requirement for equation parameters: eq6
        # pooled beta estimator: inversion baseline.
        "equation_parameters.eq6_pooled_beta_estimator.inversion_baseline": (
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "Moore_Penrose_pseudo_inverse_of_pooled_residualized_cross_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "product_in_all_cases"
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ),
        # Preserve the existing fixed requirement for equation parameters: eq6
        # pooled beta estimator: stabilization.
        "equation_parameters.eq6_pooled_beta_estimator.stabilization": (
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "SVD_rank_and_condition_diagnostics; Moore_Penrose_pseudo_inverse"
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ),
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for equation parameters:
            # eq15 incremental wave2 ATT: identity tolerance.
            "equation_parameters.eq15_incremental_wave2_ATT.identity_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "tolerance"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): "1e-10",
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for equation parameters:
            # eq18 exposure heterogeneity: controls included.
            "equation_parameters.eq18_exposure_heterogeneity.controls_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "included"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): False,
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for equation parameters:
            # eq18 exposure heterogeneity: sum se computation.
            "equation_parameters.eq18_exposure_heterogeneity.sum_se_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "computation"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): "full_covariance_matrix",
        # ---------------------------------------------------------------------
        # Section 06: Documentary treatment conflict resolution (Algorithm G0)
        # ---------------------------------------------------------------------
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # G0 treatment coding resolution: conflict rule.
            "algorithm_parameters.G0_treatment_coding_resolution."
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "conflict_rule"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): (
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "Equal-priority disagreement in entry_year or coding_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "decision raises unless exactly one top-priority record is "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "adjudication_selected or a consistent adjudicated_entry_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "year identifies one top-priority record; a nonblank "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "adjudication_note is mandatory. An adjudicated year need "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "not be the earliest."
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ),
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # G0 treatment coding resolution: selection rule.
            "algorithm_parameters.G0_treatment_coding_resolution."
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "selection_rule"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): (
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "lowest_priority_rank; earliest publication/entry ordering "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "only when no equal-rank conflict; conflicting evidence "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "requires recorded adjudication"
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ),
        # ---------------------------------------------------------------------
        # Section 07: Mahalanobis neighbour count (Algorithm E2)
        # ---------------------------------------------------------------------
        # Preserve the existing fixed requirement for algorithm parameters: E2
        # mahalanobis matching: M neighbors.
        "algorithm_parameters.E2_mahalanobis_matching.M_neighbors": 5,
        # ---------------------------------------------------------------------
        # Section 08: Within-window trade completion and component coverage
        # ---------------------------------------------------------------------
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # D4 within window completion: interpolation space.
            "algorithm_parameters.D4_within_window_completion."
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "interpolation_space"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): "log1p",
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # D4 within window completion: internal gap max consecutive.
            "algorithm_parameters.D4_within_window_completion.internal_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "gap_max_consecutive"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): 2,
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # D4 within window completion: edge extrapolation min anchors.
            "algorithm_parameters.D4_within_window_completion.edge_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "extrapolation_min_anchors"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): 3,
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # D4 within window completion: ratio recomputed from components.
            "algorithm_parameters.D4_within_window_completion.ratio_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "recomputed_from_components"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): True,
        # Preserve the existing fixed requirement for algorithm parameters: D5
        # exposure sample rules: strict.
        "algorithm_parameters.D5_exposure_sample_rules.strict": {
            # Require this many usable observations in the four-year exposure
            # window.
            "min_usable_years_of_4": 3,
            # Limit completed cells separately for each exposure component.
            "max_imputed_per_component": 1,
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for algorithm parameters: D5
        # exposure sample rules: extended.
        "algorithm_parameters.D5_exposure_sample_rules.extended": {
            # Require this many usable observations in the four-year exposure
            # window.
            "min_usable_years_of_4": 2,
            # Limit completed cells separately for each exposure component.
            "max_imputed_per_component": 2,
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # ---------------------------------------------------------------------
        # Section 09: Full-design resampling draw counts (Algorithms B1-B3)
        # ---------------------------------------------------------------------
        # Preserve the existing fixed requirement for algorithm parameters: B1
        # country history bootstrap: replicates B.
        "algorithm_parameters.B1_country_history_bootstrap.replicates_B": 500,
        # Preserve the existing fixed requirement for algorithm parameters: B2
        # region block bootstrap: replicates B.
        "algorithm_parameters.B2_region_block_bootstrap.replicates_B": 500,
        # Preserve the existing fixed requirement for algorithm parameters: B3
        # size matched placebo: replicates M.
        "algorithm_parameters.B3_size_matched_placebo.replicates_M": 500,
        # ---------------------------------------------------------------------
        # Section 10: Resampling degeneracy, attempt caps and rejection
        # reporting
        # ---------------------------------------------------------------------
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # B1 country history bootstrap: invalid resample policy.
            "algorithm_parameters.B1_country_history_bootstrap.invalid_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "resample_policy"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): (
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "Redraw only a sampler ResamplingDegeneracyError or a "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "nonfinite refitted statistic; require at least two target "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "occurrences and two distinct original donors; accept finite "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "rank-deficient factors through the Appendix B "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "pseudo-inverse; propagate schema, programming and refit "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "exceptions; stop at 500 accepted draws or 2500 attempts and "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "retain the full rejection ledger. Flag a discard fraction "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "above 0.10 of attempted draws."
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ),
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # B2 region block bootstrap: invalid resample policy.
            "algorithm_parameters.B2_region_block_bootstrap.invalid_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "resample_policy"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): (
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "Redraw only a sampler ResamplingDegeneracyError or a "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "nonfinite refitted statistic; require at least two target "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "occurrences and two distinct original donors; accept finite "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "rank-deficient factors through the Appendix B "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "pseudo-inverse; propagate schema, programming and refit "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "exceptions; stop at 500 accepted draws or 2500 attempts and "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "retain the full rejection ledger. Flag a discard fraction "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "above 0.10 of attempted draws."
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ),
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for methodology
            # parameters: bootstrap inference: invalid resample policy.
            "methodology_parameters.bootstrap_inference.invalid_resample_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "policy"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): (
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "Redraw only a sampler ResamplingDegeneracyError or a "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "nonfinite refitted statistic; require at least two target "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "occurrences and two distinct original donors; accept finite "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "rank-deficient factors through the Appendix B "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "pseudo-inverse; propagate schema, programming and refit "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "exceptions; stop at 500 accepted draws or 2500 attempts and "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "retain the full rejection ledger. Flag a discard fraction "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "above 0.10 of attempted draws."
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ),
        # Preserve the existing fixed requirement for invalid resample policy:
        # country history bootstrap.
        "invalid_resample_policy.country_history_bootstrap": (
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "Redraw only a sampler ResamplingDegeneracyError or a "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "nonfinite refitted statistic; require at least two target "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "occurrences and two distinct original donors; accept finite "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "rank-deficient factors through the Appendix B "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "pseudo-inverse; propagate schema, programming and refit "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "exceptions; stop at 500 accepted draws or 2500 attempts and "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "retain the full rejection ledger. Flag a discard fraction "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "above 0.10 of attempted draws."
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ),
        # Preserve the existing fixed requirement for invalid resample policy:
        # region block bootstrap.
        "invalid_resample_policy.region_block_bootstrap": (
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "Redraw only a sampler ResamplingDegeneracyError or a "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "nonfinite refitted statistic; require at least two target "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "occurrences and two distinct original donors; accept finite "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "rank-deficient factors through the Appendix B "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "pseudo-inverse; propagate schema, programming and refit "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "exceptions; stop at 500 accepted draws or 2500 attempts and "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "retain the full rejection ledger. Flag a discard fraction "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "above 0.10 of attempted draws."
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ),
        # Preserve the existing fixed requirement for invalid resample policy:
        # size matched placebo.
        "invalid_resample_policy.size_matched_placebo": (
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "Redraw only a sampler ResamplingDegeneracyError or a "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "nonfinite refitted statistic; require at least two target "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "occurrences and two distinct original donors; accept finite "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "rank-deficient factors through the Appendix B "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "pseudo-inverse; propagate schema, programming and refit "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "exceptions; stop at 500 accepted draws or 2500 attempts and "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "retain the full rejection ledger. Flag a discard fraction "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "above 0.10 of attempted draws."
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ),
        # Preserve the existing fixed requirement for invalid resample policy:
        # uniform reporting requirement.
        "invalid_resample_policy.uniform_reporting_requirement": (
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "Log each accepted/rejected attempt and discarded_count for "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "every branch. Fragility is discarded_count / attempted_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "count > 0.10, not discarded_count / requested_count. "
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "Preserve deterministic RNG state chains."
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ),
        # ---------------------------------------------------------------------
        # Section 11: Donor-only factor proxy and control/rate transformation
        # rules
        # ---------------------------------------------------------------------
        # Preserve the existing fixed requirement for model parameters: factor
        # proxy model: dimension.
        "model_parameters.factor_proxy_model.dimension": 2,
        # Preserve the existing fixed requirement for model parameters: factor
        # proxy model: components.
        "model_parameters.factor_proxy_model.components": (
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            ["constant_one", "donor_outcome_cross_sectional_average"]
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ),
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for model parameters:
            # factor proxy model: target pool excluded from construction.
            "model_parameters.factor_proxy_model.target_pool_excluded_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "from_construction"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): True,
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for data preprocessing
            # parameters: label transformation: rate outcomes divided by 100 in
            # ATT.
            "data_preprocessing_parameters.label_transformation.rate_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "outcomes_divided_by_100_in_ATT"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): ["Inflation", "RealGDPpcGrowth"],
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for data preprocessing
            # parameters: label transformation: rate outcomes percentage points
            # in mechanism.
            "data_preprocessing_parameters.label_transformation.rate_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "outcomes_percentage_points_in_mechanism"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): ["Inflation", "RealGDPpcGrowth"],
        # Preserve the existing fixed requirement for data preprocessing
        # parameters: control lag rule: lagged by.
        "data_preprocessing_parameters.control_lag_rule.lagged_by": 1,
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for data preprocessing
            # parameters: control lag rule: lagged after completion.
            "data_preprocessing_parameters.control_lag_rule.lagged_after_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "completion"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): True,
        # ---------------------------------------------------------------------
        # Section 12: Operative execution, numerical and production parameters
        # ---------------------------------------------------------------------
        # Preserve the existing fixed requirement for execution parameters: raw
        # universe minimum.
        "execution_parameters.raw_universe_minimum": 130,
        # Preserve the existing fixed requirement for execution parameters:
        # comtrade status counts.
        "execution_parameters.comtrade_status_counts": {
            # Declare the russian gas field with its existing value.
            "russian_gas": {
                # Require this count of positive raw trade-primitive cells.
                "positive": 229,
                # Require this count of explicitly verified zero
                # trade-primitive cells.
                "verified_zero": 733,
                # Require this count of unrecovered raw trade-primitive cells.
                "missing": 70,
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            },
            # Declare the world gas field with its existing value.
            "world_gas": {
                # Require this count of positive raw trade-primitive cells.
                "positive": 809,
                # Require this count of explicitly verified zero
                # trade-primitive cells.
                "verified_zero": 185,
                # Require this count of unrecovered raw trade-primitive cells.
                "missing": 38,
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            },
            # Declare the russian non gas field with its existing value.
            "russian_non_gas": {
                # Require this count of positive raw trade-primitive cells.
                "positive": 769,
                # Require this count of explicitly verified zero
                # trade-primitive cells.
                "verified_zero": 224,
                # Require this count of unrecovered raw trade-primitive cells.
                "missing": 39,
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            },
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for execution parameters:
        # completion.
        "execution_parameters.completion": {
            # Declare the repaired control cells field with its existing value.
            "repaired_control_cells": 1471,
            # Declare the derived fuel balance missing cells field with its
            # existing value.
            "derived_fuel_balance_missing_cells": 203,
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for execution parameters:
        # numerical.
        "execution_parameters.numerical": {
            # Declare the residual maker gram condition max field with its
            # existing value.
            "residual_maker_gram_condition_max": 1000.0,
            # Declare the factor proxy condition flag field with its existing
            # value.
            "factor_proxy_condition_flag": 1000.0,
            # Declare the pooled cross product condition warning field with its
            # existing value.
            "pooled_cross_product_condition_warning": 1000000.0,
            # Declare the rank tolerance field with its existing value.
            "rank_tolerance": (
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "sigma_max * max(matrix.shape) * float64_machine_epsilon"
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ),
            # Declare the pooled beta route field with its existing value.
            "pooled_beta_route": "always_Moore_Penrose_pseudo_inverse",
            # Declare the residual maker route field with its existing value.
            "residual_maker_route": (
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "linear_solve_if_full_rank_and_gram_condition_le_"
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "1000_else_pseudo_inverse"
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ),
            # Declare the incremental identity tolerance field with its
            # existing value.
            "incremental_identity_tolerance": 1e-10,
            # Declare the comparison tolerance field with its existing value.
            "comparison_tolerance": 0.001,
            # Declare the exposure completion count tolerance field with its
            # existing value.
            "exposure_completion_count_tolerance": 0,
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for execution parameters:
        # resampling.
        "execution_parameters.resampling": {
            # Declare the attempt cap multiple field with its existing value.
            "attempt_cap_multiple": 5,
            # Declare the minimum target occurrences field with its existing
            # value.
            "minimum_target_occurrences": 2,
            # Declare the minimum distinct original donors field with its
            # existing value.
            "minimum_distinct_original_donors": 2,
            # Declare the discard fraction flag field with its existing value.
            "discard_fraction_flag": 0.1,
            # Use attempted draws as the discard-fraction denominator.
            "discard_fraction_denominator": "attempted_draws",
            # Declare the rank deficient factor policy field with its existing
            # value.
            "rank_deficient_factor_policy": (
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "accept_when_finite_pseudo_inverse_fit"
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ),
            # Declare the redraw exceptions field with its existing value.
            "redraw_exceptions": ["ResamplingDegeneracyError_from_sampler"],
            # Declare the refit exception policy field with its existing value.
            "refit_exception_policy": "propagate",
            # Declare the nonfinite statistic policy field with its existing
            # value.
            "nonfinite_statistic_policy": "redraw",
            # Declare the percentile bounds field with its existing value.
            "percentile_bounds": [0.025, 0.975],
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for execution parameters:
        # secondary sweeps.
        "execution_parameters.secondary_sweeps": {
            # Declare the shortened wave2 window field with its existing value.
            "shortened_wave2_window": [2022, 2023],
            # Declare the active year callable field with its existing value.
            "active_year_callable": "entry_year_active_window",
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for execution parameters:
        # influence.
        "execution_parameters.influence": {
            # Declare the top donors field with its existing value.
            "top_donors": 10,
            # Declare the muted response threshold field with its existing
            # value.
            "muted_response_threshold": 0.8,
            # Declare the trim fraction each tail field with its existing
            # value.
            "trim_fraction_each_tail": 0.1,
            # Declare the trim count rule field with its existing value.
            "trim_count_rule": "floor(N_T * 0.10)",
            # Declare the tie breaking field with its existing value.
            "tie_breaking": "lexicographic_country_code",
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for execution parameters:
        # matching.
        "execution_parameters.matching": {
            # Declare the characteristic window field with its existing value.
            "characteristic_window": [2009, 2013],
            # Declare the characteristics field with its existing value.
            "characteristics": [
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "OilRents",
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "TradeOpenness",
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "TermsOfTrade",
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "InvestmentRate",
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "GovConsumption",
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "PopulationGrowth",
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "RuleOfLaw",
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "FuelShareBalance",
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "log_real_GDPpc",
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ],
            # Declare the covariance ddof field with its existing value.
            "covariance_ddof": 1,
            # Declare the covariance inverse field with its existing value.
            "covariance_inverse": "Moore_Penrose",
            # Declare the donor balance field with its existing value.
            "donor_balance": "frequency_expanded_sample_variance",
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for execution parameters:
        # exposure.
        "execution_parameters.exposure": {
            # Declare the standardization ddof field with its existing value.
            "standardization_ddof": 1,
            # Declare the zero denominator field with its existing value.
            "zero_denominator": "inadmissible_ratio_not_zero_exposure",
            # Declare the growth bound pool field with its existing value.
            "growth_bound_pool": (
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "observed_registered_components_within_same_window"
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ),
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for execution parameters:
        # heterogeneity specifications.
        "execution_parameters.heterogeneity_specifications": [
            # Begin the explicit key-to-value mapping used by the enclosing
            # operation.
            {
                # Declare the manuscript name field with its existing value.
                "manuscript_name": "baseline_2018_2021",
                # Declare the name field with its existing value.
                "name": "baseline",
                # Declare the exposure variant field with its existing value.
                "exposure_variant": "energy_exposure",
                # Declare the window field with its existing value.
                "window": "w2018_2021",
                # Declare the coverage rule field with its existing value.
                "coverage_rule": "strict",
                # Declare the policy name field with its existing value.
                "policy_name": "strict_sanctions",
                # Declare the estimator field with its existing value.
                "estimator": "interactive_factor",
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            },
            # Begin the explicit key-to-value mapping used by the enclosing
            # operation.
            {
                # Declare the manuscript name field with its existing value.
                "manuscript_name": "pre_Crimea_2010_2013",
                # Declare the name field with its existing value.
                "name": "pre_crimea",
                # Declare the exposure variant field with its existing value.
                "exposure_variant": "energy_exposure",
                # Declare the window field with its existing value.
                "window": "w2010_2013",
                # Declare the coverage rule field with its existing value.
                "coverage_rule": "strict",
                # Declare the policy name field with its existing value.
                "policy_name": "strict_sanctions",
                # Declare the estimator field with its existing value.
                "estimator": "interactive_factor",
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            },
            # Begin the explicit key-to-value mapping used by the enclosing
            # operation.
            {
                # Declare the manuscript name field with its existing value.
                "manuscript_name": "extended_pre_Crimea_sample",
                # Declare the name field with its existing value.
                "name": "pre_crimea_extended",
                # Declare the exposure variant field with its existing value.
                "exposure_variant": "energy_exposure",
                # Declare the window field with its existing value.
                "window": "w2010_2013",
                # Declare the coverage rule field with its existing value.
                "coverage_rule": "extended",
                # Declare the policy name field with its existing value.
                "policy_name": "strict_sanctions",
                # Declare the estimator field with its existing value.
                "estimator": "interactive_factor",
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            },
            # Begin the explicit key-to-value mapping used by the enclosing
            # operation.
            {
                # Declare the manuscript name field with its existing value.
                "manuscript_name": "two_way_fixed_effects",
                # Declare the name field with its existing value.
                "name": "twfe",
                # Declare the exposure variant field with its existing value.
                "exposure_variant": "energy_exposure",
                # Declare the window field with its existing value.
                "window": "w2018_2021",
                # Declare the coverage rule field with its existing value.
                "coverage_rule": "strict",
                # Declare the policy name field with its existing value.
                "policy_name": "strict_sanctions",
                # Declare the estimator field with its existing value.
                "estimator": "two_way_fixed_effects",
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            },
            # Begin the explicit key-to-value mapping used by the enclosing
            # operation.
            {
                # Declare the manuscript name field with its existing value.
                "manuscript_name": "strict_direct_arming_definition",
                # Declare the name field with its existing value.
                "name": "direct_arming_strict",
                # Declare the exposure variant field with its existing value.
                "exposure_variant": "energy_exposure",
                # Declare the window field with its existing value.
                "window": "w2018_2021",
                # Declare the coverage rule field with its existing value.
                "coverage_rule": "strict",
                # Declare the policy name field with its existing value.
                "policy_name": "direct_arming",
                # Declare the estimator field with its existing value.
                "estimator": "interactive_factor",
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            },
            # Begin the explicit key-to-value mapping used by the enclosing
            # operation.
            {
                # Declare the manuscript name field with its existing value.
                "manuscript_name": "alternative_exposure_gas_share",
                # Declare the name field with its existing value.
                "name": "gas_share",
                # Declare the exposure variant field with its existing value.
                "exposure_variant": "gas_share",
                # Declare the window field with its existing value.
                "window": "w2018_2021",
                # Declare the coverage rule field with its existing value.
                "coverage_rule": "strict",
                # Declare the policy name field with its existing value.
                "policy_name": "strict_sanctions",
                # Declare the estimator field with its existing value.
                "estimator": "interactive_factor",
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            },
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ],
        # Preserve the existing fixed requirement for execution parameters:
        # direct arming wrapper.
        "execution_parameters.direct_arming_wrapper": {
            # Declare the outcomes field with its existing value.
            "outcomes": ["log_CPI", "log_real_GDPpc"],
            # Declare the wave references field with its existing value.
            "wave_references": {
                # Declare the wave 1 level field with its existing value.
                "wave_1_level": None,
                # Declare the wave 2 level field with its existing value.
                "wave_2_level": [2019, 2021],
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            },
            # Declare the draw contract field with its existing value.
            "draw_contract": 500,
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for execution parameters:
        # figure coverage.
        "execution_parameters.figure_coverage": {
            # Declare the dynamic effective years field with its existing
            # value.
            "dynamic_effective_years": [2001, 2024],
            # Declare the unavailable dynamic years field with its existing
            # value.
            "unavailable_dynamic_years": [2000],
            # Declare the figure4 outcomes field with its existing value.
            "figure4_outcomes": ["log_CPI", "log_real_GDPpc"],
            # Declare the figure B1 status field with its existing value.
            "figure_B1_status": (
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "implemented_requires_authoritative_world_geometry"
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ),
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # ---------------------------------------------------------------------
        # Section 13: Counterfactual, aggregation, inference and exposure
        # equations
        # ---------------------------------------------------------------------
        # Preserve the existing fixed requirement for equation parameters: eq3
        # untreated outcome model.
        "equation_parameters.eq3_untreated_outcome_model": {
            # Declare the K controls field with its existing value.
            "K_controls": 8,
            # Declare the control lag field with its existing value.
            "control_lag": 1,
            # Declare the loading vector dimension field with its existing
            # value.
            "loading_vector_dimension": 2,
            # Declare the loading components field with its existing value.
            "loading_components": (
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                ["mu_i_country_intercept", "lambda_i_donor_CSA_loading"]
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ),
            # Declare the factor dimension field with its existing value.
            "factor_dimension": 2,
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for equation parameters: eq4
        # donor factor proxy.
        "equation_parameters.eq4_donor_factor_proxy": {
            # Declare the donor set field with its existing value.
            "donor_set": "never_treated_only",
            # Declare the component 1 field with its existing value.
            "component_1": "constant_one",
            # Declare the component 2 field with its existing value.
            "component_2": "donor_outcome_cross_sectional_average",
            # Declare the treated realizations included field with its existing
            # value.
            "treated_realizations_included": False,
            # Declare the parsimonious specification field with its existing
            # value.
            "parsimonious_specification": "own_outcome_CSA_only",
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for equation parameters: eq5
        # pre treatment OLS.
        "equation_parameters.eq5_pre_treatment_OLS": {
            # Declare the pre treatment end year field with its existing value.
            "pre_treatment_end_year": 2013,
            # Declare the pre treatment start year field with its existing
            # value.
            "pre_treatment_start_year": 2001,
            # Declare the T0 field with its existing value.
            "T0": 13,
            # Declare the estimator field with its existing value.
            "estimator": "pooled_OLS_with_factor_projection",
            # Declare the residual maker applied field with its existing value.
            "residual_maker_applied": True,
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for equation parameters: eq7
        # unit specific loadings.
        "equation_parameters.eq7_unit_specific_loadings": {
            # Declare the dimension field with its existing value.
            "dimension": 2,
            # Declare the estimated for field with its existing value.
            "estimated_for": ["all_targets", "all_donors"],
            # Declare the estimation source field with its existing value.
            "estimation_source": "pre_treatment_residuals",
            # Declare the why donors included field with its existing value.
            "why_donors_included": (
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "Donors' loadings are needed when they enter "
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "downstream counterfactual-placebo constructions and "
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "are used to enforce the design's symmetric "
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "treatment of the estimation set S_Y."
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ),
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for equation parameters: eq8
        # covariate loadings.
        "equation_parameters.eq8_covariate_loadings": {
            # Declare the Lambda dimension field with its existing value.
            "Lambda_dimension": [2, 8],
            # Declare the estimated for field with its existing value.
            "estimated_for": "targets_only",
            # Declare the estimation source field with its existing value.
            "estimation_source": "pre_treatment_only",
            # Declare the why only targets field with its existing value.
            "why_only_targets": (
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "Donors' post-2013 covariates are directly observed; "
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "a simulated counterfactual covariate path is needed "
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "only for targets."
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ),
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for equation parameters: eq9
        # counterfactual covariates.
        "equation_parameters.eq9_counterfactual_covariates": {
            # Declare the post treatment start year field with its existing
            # value.
            "post_treatment_start_year": 2014,
            # Declare the covariates simulated field with its existing value.
            "covariates_simulated": True,
            # Declare the observed covariates used post treatment field with
            # its existing value.
            "observed_covariates_used_post_treatment": False,
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for equation parameters: eq10
        # counterfactual outcome.
        "equation_parameters.eq10_counterfactual_outcome": {
            # Declare the post treatment start year field with its existing
            # value.
            "post_treatment_start_year": 2014,
            # Declare the no target realized data post treatment field with its
            # existing value.
            "no_target_realized_data_post_treatment": True,
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for equation parameters: eq11
        # country year gap.
        "equation_parameters.eq11_country_year_gap": {
            # Declare the sign convention field with its existing value.
            "sign_convention": "observed_minus_counterfactual",
            # Declare the positive interpretation field with its existing
            # value.
            "positive_interpretation": "above_imputed_untreated",
            # Declare the negative interpretation field with its existing
            # value.
            "negative_interpretation": "below_imputed_untreated",
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for equation parameters: eq12
        # window ATT.
        "equation_parameters.eq12_window_ATT": {
            # Declare the country weighting field with its existing value.
            "country_weighting": "equal",
            # Declare the year weighting field with its existing value.
            "year_weighting": "equal",
            # Declare the N T sanctions field with its existing value.
            "N_T_sanctions": 42,
            # Declare the N T direct arming field with its existing value.
            "N_T_direct_arming": 29,
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for equation parameters: eq16
        # country dispersion SE.
        "equation_parameters.eq16_country_dispersion_SE": {
            # Declare the se formula field with its existing value.
            "se_formula": "sd(d_i) / sqrt(N_T)",
            # Declare the df field with its existing value.
            "df": "N_T - 1",
            # Declare the test field with its existing value.
            "test": "two_sided_student_t",
            # Declare the stars source field with its existing value.
            "stars_source": "this_equation_only",
            # Declare the bootstrap and placebo role field with its existing
            # value.
            "bootstrap_and_placebo_role": "complementary_diagnostics_only",
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for equation parameters: eq17
        # exposure measure.
        "equation_parameters.eq17_exposure_measure": {
            # Declare the gas types field with its existing value.
            "gas_types": ["Coal", "Crude", "Refined", "Gas"],
            # Declare the aggregation window length field with its existing
            # value.
            "aggregation_window_length": 4,
            # Declare the windows field with its existing value.
            "windows": [
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                [2010, 2013],
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                [2018, 2021],
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ],
            # Declare the numerator currency field with its existing value.
            "numerator_currency": "current_USD",
            # Declare the denominator series field with its existing value.
            "denominator_series": "NY.GDP.MKTP.CD",
            # Declare the standardization field with its existing value.
            "standardization": "within_regression_sample",
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for equation parameters: eqA1
        # gas share.
        "equation_parameters.eqA1_gas_share": {
            # Declare the numerator field with its existing value.
            "numerator": "sum_of_completed_russian_gas_over_window",
            # Declare the denominator field with its existing value.
            "denominator": "sum_of_completed_world_gas_over_window",
            # Declare the recomputed from components field with its existing
            # value.
            "recomputed_from_components": True,
            # Declare the interpolation of ratio field with its existing value.
            "interpolation_of_ratio": False,
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for equation parameters: eqA2
        # energy gdp.
        "equation_parameters.eqA2_energy_gdp": {
            # Declare the numerator field with its existing value.
            "numerator": (
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "sum_of_completed_(russian_non_gas_+_russian_gas)_"
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "over_window"
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ),
            # Declare the denominator field with its existing value.
            "denominator": "sum_of_nominal_gdp_over_window",
            # Declare the recomputed from components field with its existing
            # value.
            "recomputed_from_components": True,
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # ---------------------------------------------------------------------
        # Section 14: Critical-window integer bound and control-floor bindings
        # ---------------------------------------------------------------------
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # C1 eligibility screening: critical window max missing per
            # outcome.
            "algorithm_parameters.C1_eligibility_screening.critical_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "window_max_missing_per_outcome"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): 1,
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # C2 control completion: non negativity column bindings.
            "algorithm_parameters.C2_control_completion.non_negativity_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "column_bindings"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): {
            # Declare the TradeOpenness field with its existing value.
            "TradeOpenness": "TradeOpenness",
            # Declare the Investment field with its existing value.
            "Investment": "InvestmentRate",
            # Declare the GovConsumption field with its existing value.
            "GovConsumption": "GovConsumption",
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for algorithm parameters: C2
        # control completion: non negativity floors.
        "algorithm_parameters.C2_control_completion.non_negativity_floors": (
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            ["TradeOpenness", "Investment", "GovConsumption"]
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ),
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for data preprocessing
            # parameters: control completion hierarchy: non negativity column
            # bindings.
            "data_preprocessing_parameters.control_completion_hierarchy."
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "non_negativity_column_bindings"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): {
            # Declare the TradeOpenness field with its existing value.
            "TradeOpenness": "TradeOpenness",
            # Declare the Investment field with its existing value.
            "Investment": "InvestmentRate",
            # Declare the GovConsumption field with its existing value.
            "GovConsumption": "GovConsumption",
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for data preprocessing
            # parameters: control completion hierarchy: non negativity vars.
            "data_preprocessing_parameters.control_completion_hierarchy."
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "non_negativity_vars"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): ["TradeOpenness", "Investment", "GovConsumption"],
        # ---------------------------------------------------------------------
        # Section 15: Gas mirrors, coherent totals and protected coverage
        # exclusions
        # ---------------------------------------------------------------------
        # Preserve the existing fixed requirement for algorithm parameters: D2
        # mirror flow recovery.
        "algorithm_parameters.D2_mirror_flow_recovery": {
            # Declare the enabled for field with its existing value.
            "enabled_for": "gas_only",
            # Declare the importer reported precedence field with its existing
            # value.
            "importer_reported_precedence": True,
            # Declare the exporter mirror side field with its existing value.
            "exporter_mirror_side": "RUS_reporter",
            # Declare the exporter has lng field with its existing value.
            "exporter_has_lng": True,
            # Declare the exporter has pipeline gas field with its existing
            # value.
            "exporter_has_pipeline_gas": False,
            # Declare the unresolved pipeline countries field with its existing
            # value.
            "unresolved_pipeline_countries": ["AUT", "DEU", "POL", "TUR"],
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for algorithm parameters: D3
        # gas share denominator.
        "algorithm_parameters.D3_gas_share_denominator": {
            # Declare the default denominator field with its existing value.
            "default_denominator": "importer_reported_total_gas",
            # Declare the substitute with field with its existing value.
            "substitute_with": "exporter_sum_total",
            # Declare the substitute when field with its existing value.
            "substitute_when": [
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "importer_total_missing",
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                ("importer_total_non_positive_with_positive_" "russian_flow"),
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "importer_total_smaller_than_mirror_numerator",
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ],
            # Declare the retain condition field with its existing value.
            "retain_condition": "0 <= russian_numerator <= world_total",
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # D4 within window completion: edge growth bounds.
            "algorithm_parameters.D4_within_window_completion.edge_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "growth_bounds"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): [0.05, 0.95],
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # D4 within window completion: cross window gap.
            "algorithm_parameters.D4_within_window_completion.cross_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "window_gap"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): "2013_to_2018_never_bridged",
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # D4 within window completion: never impute russian gas for.
            "algorithm_parameters.D4_within_window_completion.never_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "impute_russian_gas_for"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): ["AUT", "DEU", "POL", "TUR"],
        # ---------------------------------------------------------------------
        # Section 16: Alternative factor proxies and their dimensions
        # ---------------------------------------------------------------------
        # Preserve the existing fixed requirement for algorithm parameters: F4
        # factor proxy variants: variants.
        "algorithm_parameters.F4_factor_proxy_variants.variants": [
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "own_outcome_donor_mean",
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "other_outcome_donor_mean",
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "both_outcome_means",
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "own_outcome_plus_oil_rents",
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "own_outcome_plus_fuel_share_balance",
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ],
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # F4 factor proxy variants: dimensions by variant.
            "algorithm_parameters.F4_factor_proxy_variants.dimensions_by_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "variant"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): [2, 2, 3, 3, 3],
        # ---------------------------------------------------------------------
        # Section 17: Exposure-gradient specification labels and coalition
        # boundary
        # ---------------------------------------------------------------------
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # G1 exposure heterogeneity matrix: specifications.
            "algorithm_parameters.G1_exposure_heterogeneity_matrix."
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "specifications"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): [
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "baseline_2018_2021",
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "pre_Crimea_2010_2013",
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "extended_pre_Crimea_sample",
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "two_way_fixed_effects",
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "strict_direct_arming_definition",
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "alternative_exposure_gas_share",
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ],
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for algorithm parameters:
            # G1 exposure heterogeneity matrix: strict arming sanctioner count.
            "algorithm_parameters.G1_exposure_heterogeneity_matrix."
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "strict_arming_sanctioner_count"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): 26,
        # ---------------------------------------------------------------------
        # Section 18: Maintained statistical architectures and mechanism units
        # ---------------------------------------------------------------------
        # Preserve the existing fixed requirement for model parameters: CCEDID
        # model.
        "model_parameters.CCEDID_model": {
            # Declare the structure field with its existing value.
            "structure": (
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "interactive_factors_plus_common_controls_plus_"
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "idiosyncratic"
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ),
            # Declare the factor proxy field with its existing value.
            "factor_proxy": "parsimonious_own_outcome_only",
            # Declare the common control slope field with its existing value.
            "common_control_slope": "pooled_across_S_Y",
            # Declare the loadings field with its existing value.
            "loadings": "country_specific",
            # Declare the estimation timing field with its existing value.
            "estimation_timing": "pre_treatment_only",
            # Declare the simulation field with its existing value.
            "simulation": "post_treatment_from_pre_only_relationships",
            # Declare the design restrictions field with its existing value.
            "design_restrictions": {
                # Group the following exact literal or expression across
                # physical lines.
                (
                    # Declare the no post 2013 target realization in own
                    # counterfactual field with its existing value.
                    "no_post_2013_target_realization_in_own_"
                    # Continue the exact string or sequence literal without
                    # changing its characters or order.
                    "counterfactual"
                    # Supply this explicit value or expression as the next part
                    # of
                    # the enclosing operation.
                ): True,
                # Declare the no post 2013 target covariates used field with
                # its existing value.
                "no_post_2013_target_covariates_used": True,
                # Declare the no treatment information used field with its
                # existing value.
                "no_treatment_information_used": True,
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            },
            # Declare the identification conditions field with its existing
            # value.
            "identification_conditions": [
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "donor_factor_captures_relevant_common_shocks",
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "pre_2014_relationships_informative_after_2013",
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "no_spillover_or_anticipation",
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ],
            # Declare the heterogeneous loadings on common factors field with
            # its existing value.
            "heterogeneous_loadings_on_common_factors": True,
            # Declare the refit per policy outcome pair field with its existing
            # value.
            "refit_per_policy_outcome_pair": True,
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # Preserve the existing fixed requirement for model parameters:
        # exposure heterogeneity model.
        "model_parameters.exposure_heterogeneity_model": {
            # Declare the estimator field with its existing value.
            "estimator": "interaction_regression",
            # Declare the controls included field with its existing value.
            "controls_included": False,
            # Declare the omitted period field with its existing value.
            "omitted_period": "all_years_before_2022",
            # Declare the country specific loadings on Ybar field with its
            # existing value.
            "country_specific_loadings_on_Ybar": True,
            # Declare the standard error field with its existing value.
            "standard_error": "country_clustered",
            # Declare the interpretation not mediation field with its existing
            # value.
            "interpretation_not_mediation": True,
            # Declare the regression universe field with its existing value.
            "regression_universe": 126,
            # Declare the intersection with exposure sample field with its
            # existing value.
            "intersection_with_exposure_sample": True,
            # Declare the rate outcome unit convention field with its existing
            # value.
            "rate_outcome_unit_convention": (
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "percentage_points (do NOT multiply by 100; rate "
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "outcomes reported in this regression are already in "
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "percentage-point units, in contrast to the ATT "
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "tables where rates are expressed as decimals)"
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ),
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        },
        # ---------------------------------------------------------------------
        # Section 19: European and world heterogeneity colour-scale limits
        # ---------------------------------------------------------------------
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for plot construction
            # parameters: figure 3 european heterogeneity: colour scale range.
            "plot_construction_parameters.figure_3_european_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "heterogeneity.colour_scale_range"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): [-0.18, 0.18],
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Preserve the existing fixed requirement for plot construction
            # parameters: figure B1 worldwide heterogeneity: colour scale
            # range.
            "plot_construction_parameters.figure_B1_worldwide_"
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "heterogeneity.colour_scale_range"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ): [-0.18, 0.18],
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    }
    # Close the preceding literal, argument list or selection without changing
    # its
    # contents.
)


# -----------------------------------------------------------------------------
# Step 1: Explicit dotted configuration lookup
# -----------------------------------------------------------------------------


# Resolve an explicitly declared dotted configuration path.
def configuration_value(config: Mapping[str, Any], path: str) -> Any:
    """
    Resolve an explicitly declared dotted configuration path.

    Parameters
    ----------
    config : Mapping[str, Any]
        Parsed study configuration. Intermediate path components must
        name mapping nodes; terminal values may be scalars, sequences or
        mappings.
    path : str
        Exact dot-separated key path, such as
        ``global_constants.control_lag_years``. Components are mapping
        keys, not attribute names or expressions evaluated as Python
        code.

    Returns
    -------
    value : Any
        Existing object at the terminal key. The object is returned
        directly; this accessor neither copies nor modifies the
        configuration.

    Raises
    ------
    ConfigValidationError
        An intermediate value is not a mapping or a component is absent.
        The diagnostic identifies the complete requested binding path.

    Notes
    -----
    Traversal implements repeated mapping lookup only. An empty or
    malformed dotted path receives no special normalization; exact key
    membership governs whether traversal succeeds. Scientific validation
    belongs to the callers.
    """
    # Start traversal at the supplied mapping, then retain each selected child.
    value: Any = config
    # Process each component from path.split('.') in the stated order.
    for component in path.split("."):
        # Reject this input when not isinstance(value, Mapping) or component
        # not in value.
        if not isinstance(value, Mapping) or component not in value:
            # Raise the stated validation exception with the failing identity
            # and observed values.
            raise ConfigValidationError(
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                f"Missing configuration binding: {path}."
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
        # Start traversal at the supplied mapping, then retain each selected
        # child.
        value = value[component]
    # Return the existing terminal value without copying or mutating it.
    return value


# -----------------------------------------------------------------------------
# Step 2: Fixed-rule and source-schema validation
# -----------------------------------------------------------------------------


# Validate frozen scientific declarations and operative input bindings.
def validate_configuration_bindings(
    # Accept study_config with the declared Mapping[str, Any] input contract.
    study_config: Mapping[str, Any],
    # Complete the typed signature with the declared result type.
) -> pd.DataFrame:
    """
    Validate frozen scientific declarations and operative input
    bindings.

    Parameters
    ----------
    study_config : Mapping[str, Any]
        Complete parsed configuration, including raw source schemas,
        reporting/reference windows, runtime inputs and mechanism
        records. Required fixed declarations are resolved by dotted key
        path.

    Returns
    -------
    audit : pandas.DataFrame
        Successful checks indexed by ``path``, with Boolean ``passed``
        and string ``binding`` columns. Binding classes distinguish
        fixed invariants, source schemas, forwarded windows, runtime
        handoff and operative mechanism records. A failed check raises
        before returning.

    Raises
    ------
    ConfigValidationError
        A required path or fixed value differs; a schema declaration
        disagrees with its code-defined schema; the reference is not
        registered; runtime metadata is malformed; seed registrations
        coincide; or mechanism labels/records disagree with their
        operative definitions.
    IngestError
        A downstream reporting-window or seed constructor rejects its
        input.

    Notes
    -----
    Processes the 92 fixed paths in their original validation order,
    followed by eight schema fields per source and the forwarded/runtime
    checks. ``_values_match`` provides the existing value-comparison
    convention. Seed generators are constructed for validation; no
    random sample is drawn. This function validates bindings, not source
    authenticity or execution readiness. It neither changes the
    configuration nor issues a readiness token. Equations (3)-(10) and
    Appendix B are constraints represented by the fixed declarations;
    their estimators are executed in later task cells.
    """
    # Allocate the ordered binding-verdict ledger returned after all checks
    # pass.
    rows: List[Dict[str, Any]] = []
    # Process each (path, expected) from FIXED_CONFIGURATION_CONTRACT.items()
    # in the stated order.
    for path, expected in FIXED_CONFIGURATION_CONTRACT.items():
        # Read the configured declaration being compared with its code-defined
        # requirement.
        actual: Any = configuration_value(study_config, path)
        # Compare the configured and required values with the existing equality
        # convention.
        passed: bool = _values_match(actual, expected)
        # Append this explicit identity, count or binding verdict to the
        # ordered audit ledger.
        rows.append(
            # Identify the complete configuration binding checked by this audit
            # row. Record this check's Boolean acceptance verdict. Record the
            # validation or handoff category for this configuration path.
            {"path": path, "passed": passed, "binding": "fixed_invariant"}
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
        # Reject this input when not passed.
        if not passed:
            # Raise the stated validation exception with the failing identity
            # and observed values.
            raise ConfigValidationError(
                # Group the following exact literal or expression across
                # physical lines.
                (
                    # Continue the exact string or sequence literal without
                    # changing its characters or order.
                    "Fixed implementation contract differs at "
                    # Supply this explicit value or expression as the next part
                    # of the enclosing operation.
                    f"{path}"
                    # Continue the exact string or sequence literal without
                    # changing its characters or order.
                    ": observed "
                    # Supply this explicit value or expression as the next part
                    # of the enclosing operation.
                    f"{actual!r}"
                    # Continue the exact string or sequence literal without
                    # changing its characters or order.
                    "; required "
                    # Supply this explicit value or expression as the next part
                    # of the enclosing operation.
                    f"{expected!r}"
                    # Continue the exact string or sequence literal without
                    # changing its characters or order.
                    "."
                    # Close the preceding literal, argument list or selection
                    # without changing its contents.
                )
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
    # Process each (name, schema) from build_source_schemas().items() in the
    # stated order.
    for name, schema in build_source_schemas().items():
        # Read this source's schema declaration from the configuration catalog.
        declared: Mapping[str, Any] = configuration_value(
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            study_config,
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            f"raw_data_schemas.{name}",
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
        # Convert the code-defined operative column labels to their declared
        # string form.
        expected_columns_value: List[str] = [
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            str(value)
            # Evaluate this comprehension over the stated items in their
            # existing order.
            for value in expected_columns(schema)
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ]
        # Collect the eight operative schema fields that each source must
        # declare exactly.
        checks: Dict[str, List[str]] = {
            # Declare the ordered operative source columns.
            "columns": expected_columns_value,
            # Declare the exact source uniqueness key.
            "primary_key": list(schema.key_columns),
            # Declare only the permitted optional source extensions.
            "optional_extension_columns": list(schema.extension_columns),
            # Declare source columns requiring floating numeric conversion.
            "float_columns": list(schema.float_columns),
            # Declare source columns requiring integral values.
            "int_columns": list(schema.int_columns),
            # Declare columns whose categorical vocabulary is validated by
            # ingest.
            "category_columns": list(schema.category_columns),
            # Declare canonical three-letter country-code columns.
            "code_columns": list(schema.code_columns),
            # Declare source columns subject to membership validation.
            "membership_columns": list(schema.membership_columns),
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        }
        # Process each (key, expected) from checks.items() in the stated order.
        for key, expected in checks.items():
            # Read the configured declaration being compared with its
            # code-defined requirement.
            actual = declared.get(key)
            # Compare the configured and required values with the existing
            # equality convention.
            passed = _values_match(actual, expected)
            # Identify this schema-field check by its fully qualified
            # configuration path.
            path = f"raw_data_schemas.{name}.{key}"
            # Append this explicit identity, count or binding verdict to the
            # ordered audit ledger.
            rows.append(
                # Identify the complete configuration binding checked by this
                # audit row. Record this check's Boolean acceptance verdict.
                # Record the validation or handoff category for this
                # configuration path.
                {"path": path, "passed": passed, "binding": "source_schema"}
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
            # Reject this input when not passed.
            if not passed:
                # Raise the stated validation exception with the failing
                # identity and observed values.
                raise ConfigValidationError(
                    # Supply this explicit value or expression as the next part
                    # of the enclosing operation.
                    f"Operative source schema differs at {path}."
                    # Close the preceding literal, argument list or selection
                    # without changing its contents.
                )
    # Resolve reporting windows through the estimator's shared window resolver.
    windows, _ = resolve_reporting_windows(study_config)
    # Read the permitted reference-window registry before checking the selected
    # baseline.
    references: Sequence[Sequence[int]] = configuration_value(
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        study_config,
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "algorithm_parameters.F5_reference_window_sweep.reference_windows",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Reject this input when windows['pre_escalation'] not in [tuple(value) for
    # value in references].
    if windows["pre_escalation"] not in [tuple(value) for value in references]:
        # Raise the stated validation exception: The configured reference is
        # not a registered window.
        raise ConfigValidationError(
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "The configured reference is not a registered window."
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
    # Append this explicit identity, count or binding verdict to the ordered
    # audit ledger.
    rows.append(
        # Identify the complete configuration binding checked by this audit
        # row. Record this check's Boolean acceptance verdict. Record the
        # validation or handoff category for this configuration path.
        {"path": "estimation.windows", "passed": True, "binding": "forwarded"}
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Read the six configured handoff inputs for seed and runtime metadata
    # checks.
    runtime: Mapping[str, Any] = configuration_value(
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        study_config,
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "runtime_inputs",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Process each key from ('seeds', 'stability_seeds') in the stated order.
    for key in ("seeds", "stability_seeds"):
        # Reject this input when not isinstance(runtime.get(key), Mapping).
        if not isinstance(runtime.get(key), Mapping):
            # Raise the stated validation exception with the failing identity
            # and observed values.
            raise ConfigValidationError(
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                f"runtime_inputs.{key} must be a seed mapping."
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
        # Validate this seed registration by constructing its private branch
        # generators.
        spawn_branch_generators(runtime[key])
    # Reject this input when runtime['seeds'] == runtime['stability_seeds'].
    if runtime["seeds"] == runtime["stability_seeds"]:
        # Raise the stated validation exception: Primary and stability seed
        # registrations must differ.
        raise ConfigValidationError(
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "Primary and stability seed registrations must differ."
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
    # Process each key from ('output_dir', 'replication_archive_root',
    # 'archive_version_tag') in the stated order.
    for key in (
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "output_dir",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "replication_archive_root",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "archive_version_tag",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    ):
        # Reject this input when not isinstance(runtime.get(key), str) or not
        # runtime[key].strip().
        if not isinstance(runtime.get(key), str) or not runtime[key].strip():
            # Raise the stated validation exception with the failing identity
            # and observed values.
            raise ConfigValidationError(
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                f"runtime_inputs.{key} must be a nonblank string."
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
    # Append this explicit identity, count or binding verdict to the ordered
    # audit ledger.
    rows.append(
        # Begin the explicit key-to-value mapping used by the enclosing
        # operation.
        {
            # Identify the complete configuration binding checked by this audit
            # row.
            "path": "runtime_inputs",
            # Record this check's Boolean acceptance verdict.
            "passed": True,
            # Record the validation or handoff category for this configuration
            # path.
            "binding": "runtime_handoff",
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        }
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Validate manuscript labels and operative mechanism records before
    # returning the binding audit.
    validate_heterogeneity_specification_bindings(study_config)
    # Append this explicit identity, count or binding verdict to the ordered
    # audit ledger.
    rows.append(
        # Begin the explicit key-to-value mapping used by the enclosing
        # operation.
        {
            # Identify the complete configuration binding checked by this audit
            # row.
            ("path"): (
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "execution_parameters.heterogeneity_specifications.code_"
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "records"
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ),
            # Record this check's Boolean acceptance verdict.
            "passed": True,
            # Record the validation or handoff category for this configuration
            # path.
            "binding": "operative_specification_records",
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        }
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Return the completed result with its declared key, index and metadata
    # contract.
    return pd.DataFrame(rows).set_index("path")


# -----------------------------------------------------------------------------
# Step 3: Mechanism specification binding
# -----------------------------------------------------------------------------


# Reconcile all six exposure-gradient labels with operative specifications.
def validate_heterogeneity_specification_bindings(
    # Accept study_config with the declared Mapping[str, Any] input contract.
    study_config: Mapping[str, Any],
    # Complete the typed signature with the declared result type.
) -> None:
    """
    Reconcile all six exposure-gradient labels with operative
    specifications.

    Parameters
    ----------
    study_config : Mapping[str, Any]
        Configuration containing the ordered manuscript-label catalog
        and ``execution_parameters.heterogeneity_specifications``
        records. Every record supplies ``manuscript_name``, ``name``,
        ``exposure_variant``, ``window``, ``coverage_rule``,
        ``policy_name`` and ``estimator``.

    Returns
    -------
    None
        Successful validation has no result object or side effects.

    Raises
    ------
    ConfigValidationError
        The manuscript labels differ in content/order or the six
        operative fields differ from the code-defined ``SPECIFICATIONS``
        records.
    KeyError
        A present record omits a field indexed directly by this
        validator.

    Notes
    -----
    Compares the ordered label list first, then ordered dictionaries of
    the six operative selector fields. A descriptive label does not
    select a model. This binds Equation (18)'s mechanism designs without
    estimating slopes, adding lagged controls, or interpreting the
    gradient as mediation.
    """
    # Read the configured manuscript-to-code mechanism specification records.
    configured: Sequence[Mapping[str, Any]] = configuration_value(
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        study_config,
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "execution_parameters.heterogeneity_specifications",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Read the manuscript's ordered six-label mechanism catalog.
    catalog: Sequence[str] = configuration_value(
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        study_config,
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "algorithm_parameters.G1_exposure_heterogeneity_matrix.specifications",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Reject this input when [record['manuscript_name'] for record in
    # configured] != list(catalog).
    if [record["manuscript_name"] for record in configured] != list(catalog):
        # Raise the stated validation exception: Heterogeneity
        # manuscript-to-code labels differ.
        raise ConfigValidationError(
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "Heterogeneity manuscript-to-code labels differ."
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
    # Fix the six operative fields whose values determine the mechanism design.
    names: Tuple[str, ...] = (
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "name",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "exposure_variant",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "window",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "coverage_rule",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "policy_name",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "estimator",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Extract operative fields from the code-defined SPECIFICATIONS in their
    # declared order.
    actual: List[Dict[str, Any]] = [
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        {name: getattr(record, name) for name in names}
        # Evaluate this comprehension over the stated items in their existing
        # order.
        for record in SPECIFICATIONS
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    ]
    # Extract the same operative fields from the configured records in the same
    # order.
    declared: List[Dict[str, Any]] = [
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        {name: record[name] for name in names}
        # Evaluate this comprehension over the stated items in their existing
        # order.
        for record in configured
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    ]
    # Reject this input when declared != actual.
    if declared != actual:
        # Raise the stated validation exception: Heterogeneity operative
        # specification records differ.
        raise ConfigValidationError(
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "Heterogeneity operative specification records differ."
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )


# -----------------------------------------------------------------------------
# Step 4: Runtime handoff without conflicting duplicates
# -----------------------------------------------------------------------------


# Bind six configured runtime inputs into a shallow copy of pipeline data.
def bind_pipeline_runtime_inputs(data: Mapping[str, Any]) -> Dict[str, Any]:
    """
    Bind six configured runtime inputs into a shallow copy of pipeline
    data.

    Parameters
    ----------
    data : Mapping[str, Any]
        Pipeline input mapping with a mapping-valued ``study_config``.
        Its ``runtime_inputs`` section must contain exactly ``seeds``,
        ``stability_seeds``, ``output_dir``,
        ``replication_archive_root``, ``overlay_pair`` and
        ``archive_version_tag``. Equivalent explicit duplicates are
        accepted; missing or None-valued inputs are filled.

    Returns
    -------
    bound : Dict[str, Any]
        New outer dictionary. Newly filled runtime values are deep
        copies of their configured declarations. Existing equivalent
        supplied values and external raw frames, geometry,
        classifications and inventories retain their existing
        identities. The caller's mapping is not modified.

    Raises
    ------
    IngestError
        ``data``/``study_config`` is not a mapping, runtime keys are
        incomplete or unsupported, or a supplied duplicate conflicts
        with configuration.
    ConfigValidationError
        The required dotted ``runtime_inputs`` section is absent.
    TypeError
        A supplied output/archive path cannot be interpreted by
        ``Path``.

    Notes
    -----
    Processes runtime keys in sorted order. Output/archive path
    comparisons expand the user directory and resolve absolute paths;
    the other four inputs use ``_values_match``. Path equality checks do
    not create directories. This handoff cannot manufacture external
    evidence or relax readiness gates.
    """
    # Reject this input when not isinstance(data, Mapping) or not
    # isinstance(data.get('study_config'), Mapping).
    if not isinstance(data, Mapping) or not isinstance(
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        data.get("study_config"),
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        Mapping,
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    ):
        # Raise the stated validation exception: Runtime handoff requires
        # data['study_config'] as a mapping.
        raise IngestError(
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "Runtime handoff requires data['study_config'] as a mapping."
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
    # Read the authoritative runtime declarations from the supplied study
    # configuration.
    configured: Mapping[str, Any] = configuration_value(
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        data["study_config"],
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "runtime_inputs",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Define the exact six keys eligible for automatic runtime handoff.
    permitted: Set[str] = {
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "seeds",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "stability_seeds",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "output_dir",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "replication_archive_root",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "overlay_pair",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "archive_version_tag",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    }
    # Reject this input when set(configured) != permitted.
    if set(configured) != permitted:
        # Raise the stated validation exception: runtime_inputs has missing or
        # unsupported handoff keys.
        raise IngestError(
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "runtime_inputs has missing or unsupported handoff keys."
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
    # Copy the outer input mapping while retaining its existing external
    # objects.
    bound: Dict[str, Any] = dict(data)
    # Process each key from sorted(permitted) in the stated order.
    for key in sorted(permitted):
        # Deep-copy the configured runtime value before considering a missing
        # input.
        value: Any = copy.deepcopy(configured[key])
        # Apply the following branch only when key in bound and bound[key] is
        # not None.
        if key in bound and bound[key] is not None:
            # Retain the caller's non-None duplicate for an explicit
            # consistency comparison.
            supplied: Any = bound[key]
            # Apply the following branch only when key in {'output_dir',
            # 'replication_archive_root'}.
            if key in {"output_dir", "replication_archive_root"}:
                # Compare resolved paths for output locations, or declared
                # values for other runtime inputs.
                equal: bool = (
                    # Supply this explicit value or expression as the next part
                    # of the enclosing operation.
                    Path(supplied).expanduser().resolve()
                    # Supply this explicit value or expression as the next part
                    # of the enclosing operation.
                    == Path(value).expanduser().resolve()
                    # Close the preceding literal, argument list or selection
                    # without changing its contents.
                )
            # Apply the alternative only when the immediately preceding
            # condition is false.
            else:
                # Compare resolved paths for output locations, or declared
                # values for other runtime inputs.
                equal = _values_match(supplied, value)
            # Reject this input when not equal.
            if not equal:
                # Raise the stated validation exception with the failing
                # identity and observed values.
                raise IngestError(
                    # Group the following exact literal or expression across
                    # physical lines.
                    (
                        # Continue the exact string or sequence literal without
                        # changing its characters or order.
                        "Conflicting configured and supplied runtime input: "
                        # Supply this explicit value or expression as the next
                        # part of the enclosing operation.
                        f"{key}"
                        # Continue the exact string or sequence literal without
                        # changing its characters or order.
                        "."
                        # Close the preceding literal, argument list or
                        # selection
                        # without changing its contents.
                    )
                    # Close the preceding literal, argument list or selection
                    # without changing its contents.
                )
        # Apply the alternative only when the immediately preceding condition
        # is false.
        else:
            # Fill this missing or None-valued runtime slot with the configured
            # deep copy.
            bound[key] = value
    # Return the completed result with its declared key, index and metadata
    # contract.
    return bound


# -----------------------------------------------------------------------------
# Step 5: Pair-specific target and donor counts
# -----------------------------------------------------------------------------


# Derive expected target/donor counts for the eight declared study pairs.
def configured_partition_counts(
    # Accept config with the declared Mapping[str, Any] input contract.
    config: Mapping[str, Any],
    # Complete the typed signature with the declared result type.
) -> Dict[Tuple[str, str], Tuple[int, int]]:
    """
    Derive expected target/donor counts for the eight declared study
    pairs.

    Parameters
    ----------
    config : Mapping[str, Any]
        Configuration with country rosters in
        ``country_validation_registries``. Sanctions use their
        42-target/84-donor declarations. Arming uses its 29-target
        roster and the outcome-dependent donor declarations.

    Returns
    -------
    counts : Dict[Tuple[str, str], Tuple[int, int]]
        Mapping from ``(policy, outcome)`` to ``(target_count,
        donor_count)``, inserted in ``DECLARED_PAIRS`` order. Counts are
        roster lengths, not deduplicated country counts or independently
        derived sample estimates.

    Raises
    ------
    ConfigValidationError
        The country-validation section is absent.
    KeyError
        A required target or donor roster is missing within that
        section.

    Notes
    -----
    For direct arming, only ``log_CPI`` selects the CPI-specific donor
    roster; the remaining declared outcomes select the GDPpc donor
    roster. This mirrors the current partition constructor. Exact roster
    identity/uniqueness is audited separately by
    ``audit_computed_partition_rosters``.
    """
    # Read the expected country-code rosters without deriving a new estimation
    # sample.
    rosters: Mapping[str, Any] = configuration_value(
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        config,
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "country_validation_registries",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Allocate pair-specific target/donor count tuples in canonical pair order.
    result: Dict[Tuple[str, str], Tuple[int, int]] = {}
    # Process each (policy, outcome) from DECLARED_PAIRS in the stated order.
    for policy, outcome in DECLARED_PAIRS:
        # Apply the following branch only when policy == 'strict_sanctions'.
        if policy == "strict_sanctions":
            # Select the configured target roster and the
            # policy/outcome-specific donor roster.
            targets, donors = (
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                rosters["sanctions_target_42"],
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                rosters["sanctions_donors_84"],
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
        # Apply the alternative only when the immediately preceding condition
        # is false.
        else:
            # Use the CPI-specific arming donor roster only for log_CPI;
            # otherwise use GDPpc donors.
            donor_key: str = (
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "direct_arming_donors_CPI_96"
                # Include this explicit additional condition in the current
                # validation decision.
                if outcome == "log_CPI"
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                else "direct_arming_donors_GDPpc_97"
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
            # Select the configured target roster and the
            # policy/outcome-specific donor roster.
            targets, donors = (
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                rosters["direct_arming_target_29"],
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                rosters[donor_key],
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
        # Record roster lengths as the expected target and donor counts for
        # this pair.
        result[policy, outcome] = (len(targets), len(donors))
    # Return the completed result with its declared key, index and metadata
    # contract.
    return result


# -----------------------------------------------------------------------------
# Step 6: Treatment-category count declarations
# -----------------------------------------------------------------------------


# Bind seven canonical policy-category names to configured integer counts.
def configured_treatment_categories(
    # Accept config with the declared Mapping[str, Any] input contract.
    config: Mapping[str, Any],
    # Complete the typed signature with the declared result type.
) -> Dict[str, int]:
    """
    Bind seven canonical policy-category names to configured integer
    counts.

    Parameters
    ----------
    config : Mapping[str, Any]
        Configuration containing the seven ``n_<category>`` declarations
        in ``global_constants``. The categories distinguish descriptive
        sanctions coding, estimation targets, direct arming, overlap and
        exclusive groups.

    Returns
    -------
    counts : Dict[str, int]
        Seven integer counts keyed by ``sanctions_coded_descriptive``,
        ``sanctions_targets``, ``direct_arming_targets``,
        ``both_policies``, ``sanctions_only``, ``direct_arming_only``
        and ``neither_policy``.

    Raises
    ------
    ConfigValidationError
        The global-constants section is missing.
    KeyError
        A required category declaration is absent.
    TypeError, ValueError
        A declaration cannot be converted by the existing ``int``
        operation.

    Notes
    -----
    This helper supplies expected counts to treatment-coding validation.
    It does not classify countries, resolve legal evidence, or
    independently establish that any treatment-category count is
    correct.
    """
    # Read the seven published treatment-category count declarations.
    constants: Mapping[str, Any] = configuration_value(
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        config,
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "global_constants",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Fix the canonical order of descriptive, target, overlap and exclusive
    # categories.
    categories: Tuple[str, ...] = (
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "sanctions_coded_descriptive",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "sanctions_targets",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "direct_arming_targets",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "both_policies",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "sanctions_only",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "direct_arming_only",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "neither_policy",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Convert each declared category count to int and retain its canonical
    # category key.
    return {name: int(constants[f"n_{name}"]) for name in categories}


# -----------------------------------------------------------------------------
# Step 7: Exact computed-roster reconciliation
# -----------------------------------------------------------------------------


# Audit exact country identities in every computed target/donor partition.
def audit_computed_partition_rosters(
    # Accept config with the declared Mapping[str, Any] input contract.
    registry: PartitionRegistry,
    # Accept config with the declared Mapping[str, Any] input contract.
    config: Mapping[str, Any],
    # Complete the typed signature with the declared result type.
) -> pd.DataFrame:
    """
    Audit exact country identities in every computed target/donor
    partition.

    Parameters
    ----------
    registry : PartitionRegistry
        Live partition registry keyed by ``(policy, outcome)``. Each
        partition exposes its computed ``targets`` and ``donors``
        country-code sequences.
    config : Mapping[str, Any]
        Configuration with the expected sanctions and outcome-specific
        arming rosters in ``country_validation_registries``.

    Returns
    -------
    audit : pandas.DataFrame
        One row per partition/role, with ``policy``, ``outcome``,
        ``role``, sorted ``expected`` and ``observed`` tuples, and
        Boolean ``passed``. A complete eight-pair registry yields 16
        successful rows.

    Raises
    ------
    IngestError
        A computed roster differs in identity from its declaration or
        contains duplicate country codes, even if its cardinality is
        unchanged.
    ConfigValidationError
        The country-validation section is absent.
    KeyError
        A required roster is absent within the configured section.

    Notes
    -----
    Converts codes to strings and compares their sorted tuples; observed
    codes must also be unique. This implements a membership-integrity
    check for the target/donor sets of Equation (1), without selecting
    or repairing countries.
    """
    # Read declared country identities against which computed rosters are
    # reconciled.
    rosters: Mapping[str, Any] = configuration_value(
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        config,
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "country_validation_registries",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Allocate the partition/role identity audit without modifying the live
    # registry.
    rows: List[Dict[str, Any]] = []
    # Process each ((policy, outcome), partition) from
    # registry.partitions.items() in the stated order.
    for (policy, outcome), partition in registry.partitions.items():
        # Select the sanctions or arming target identity registry for this
        # partition.
        target_key: str = (
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "sanctions_target_42"
            # Include this explicit additional condition in the current
            # validation decision.
            if policy == "strict_sanctions"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            else "direct_arming_target_29"
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
        # Select the sanctions or outcome-specific arming donor identity
        # registry.
        donor_key: str = (
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "sanctions_donors_84"
            # Include this explicit additional condition in the current
            # validation decision.
            if policy == "strict_sanctions"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            else (
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "direct_arming_donors_CPI_96"
                # Include this explicit additional condition in the current
                # validation decision.
                if outcome == "log_CPI"
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                else "direct_arming_donors_GDPpc_97"
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
        # Process each (role, key, actual) from (('targets', target_key,
        # partition.targets), ('donors', donor_key, partition.donors)) in the
        # stated order.
        for role, key, actual in (
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            ("targets", target_key, partition.targets),
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            ("donors", donor_key, partition.donors),
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ):
            # Sort string-converted declared codes to make roster identity
            # independent of order.
            expected: Tuple[str, ...] = tuple(
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                sorted((str(code) for code in rosters[key]))
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
            # Sort string-converted computed codes while retaining duplicates
            # for rejection.
            observed: Tuple[str, ...] = tuple(
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                sorted((str(code) for code in actual))
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
            # Require identical sorted rosters and uniqueness of every observed
            # country code.
            passed: bool = observed == expected and len(observed) == len(
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                set(observed)
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
            # Append this explicit identity, count or binding verdict to the
            # ordered audit ledger.
            rows.append(
                # Begin the explicit key-to-value mapping used by the enclosing
                # operation.
                {
                    # Declare the policy field with its existing value.
                    "policy": policy,
                    # Declare the outcome field with its existing value.
                    "outcome": outcome,
                    # Declare the role field with its existing value.
                    "role": role,
                    # Declare the expected field with its existing value.
                    "expected": expected,
                    # Declare the observed field with its existing value.
                    "observed": observed,
                    # Record this check's Boolean acceptance verdict.
                    "passed": passed,
                    # Close the preceding literal, argument list or selection
                    # without changing its contents.
                }
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
            # Reject this input when not passed.
            if not passed:
                # Raise the stated validation exception with the failing
                # identity and observed values.
                raise IngestError(
                    # Supply this explicit value or expression as the next part
                    # of the enclosing operation.
                    f"Computed {policy}/{outcome} {role} differ from {key}."
                    # Close the preceding literal, argument list or selection
                    # without changing its contents.
                )
    # Return the completed result with its declared key, index and metadata
    # contract.
    return pd.DataFrame(rows)


# -----------------------------------------------------------------------------
# Step 8: Sixteen pre-treatment descriptive means
# -----------------------------------------------------------------------------


# Audit sixteen equal-country descriptive means over 2009-2013.
def audit_pretreatment_group_means(
    # Accept config with the declared Mapping[str, Any] input contract.
    panel: pd.DataFrame,
    # Accept registry with the declared PartitionRegistry input contract.
    registry: PartitionRegistry,
    # Accept config with the declared Mapping[str, Any] input contract.
    config: Mapping[str, Any],
    # Complete the typed signature with the declared result type.
) -> pd.DataFrame:
    """
    Audit sixteen equal-country descriptive means over 2009-2013.

    Parameters
    ----------
    panel : pandas.DataFrame
        Completed level panel with ``country_code``, ``year``,
        ``log_CPI``, ``log_real_GDPpc``, ``RuleOfLaw`` and ``OilRents``.
        Each required country contributes five finite, nonmissing
        observations in 2009-2013.
    registry : PartitionRegistry
        Live sanctions/arming ``log_CPI`` partitions. Their target and
        CPI-donor rosters define all four descriptive comparison groups,
        including GDP.
    config : Mapping[str, Any]
        Sixteen expected means, an enabled ``check_required`` flag and a
        finite nonnegative comparison tolerance in the pre-treatment
        integrity section.

    Returns
    -------
    audit : pandas.DataFrame
        Sixteen successful rows containing ``variable``, ``group``,
        ``observed``, ``expected``, ``tolerance`` and Boolean
        ``passed``.

    Raises
    ------
    ConfigValidationError
        Required bindings are missing, checks are disabled, or the
        converted tolerance is nonfinite/negative.
    IngestError
        Country/year rows duplicate, a five-year group sample is
        incomplete, observations are nonfinite, or a mean exceeds its
        tolerance.
    KeyError, TypeError, ValueError
        Required panel/configuration fields are absent or numeric
        conversion fails before an explicit integrity comparison can be
        made.

    Notes
    -----
    For each variable/group, computes the mean of country-specific
    five-year means. Acceptance uses ``abs(observed - expected) <=
    tolerance + eps``. These are Table-2 integrity checks, not the
    pre-treatment estimator of Equation (5). They do not feed policy
    information into eligibility or repair.
    """
    # Read the sixteen descriptive integrity targets and their enabled-check
    # convention.
    targets: Mapping[str, Any] = configuration_value(
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        config,
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "pretreatment_means_2009_2013_integrity_targets",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Reject this input when targets.get('check_required') is not True.
    if targets.get("check_required") is not True:
        # Raise the stated validation exception: Pre-treatment mean checks must
        # remain enabled.
        raise ConfigValidationError(
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "Pre-treatment mean checks must remain enabled."
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
    # Convert the declared absolute mean-comparison tolerance to a float.
    tolerance: float = float(targets["tolerance"])
    # Reject this input when not np.isfinite(tolerance) or tolerance < 0.
    if not np.isfinite(tolerance) or tolerance < 0:
        # Raise the stated validation exception: Invalid pre-treatment
        # comparison tolerance.
        raise ConfigValidationError(
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "Invalid pre-treatment comparison tolerance."
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
    # Fix the four comparison rosters from the sanctions and arming CPI
    # partitions.
    groups: Dict[str, Sequence[str]] = {
        # Declare the sanctions targets field with its existing value.
        "sanctions_targets": registry.partitions[
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "strict_sanctions",
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "log_CPI",
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ].targets,
        # Declare the sanctions CPI donors field with its existing value.
        "sanctions_CPI_donors": registry.partitions[
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "strict_sanctions",
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "log_CPI",
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ].donors,
        # Declare the direct arming targets field with its existing value.
        "direct_arming_targets": registry.partitions[
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "direct_arming",
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "log_CPI",
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ].targets,
        # Declare the direct arming CPI donors field with its existing value.
        "direct_arming_CPI_donors": registry.partitions[
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "direct_arming",
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "log_CPI",
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ].donors,
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    }
    # Map each integrity-target variable to its completed-panel column name.
    columns: Dict[str, str] = {
        # Declare the log CPI field with its existing value.
        "log_CPI": "log_CPI",
        # Declare the log real GDPpc field with its existing value.
        "log_real_GDPpc": "log_real_GDPpc",
        # Declare the rule of law field with its existing value.
        "rule_of_law": "RuleOfLaw",
        # Declare the oil rents field with its existing value.
        "oil_rents": "OilRents",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    }
    # Copy only calendar years 2009-2013 for the descriptive integrity
    # comparisons.
    selected: pd.DataFrame = panel.loc[
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        panel["year"].between(2009, 2013)
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
    ].copy()
    # Reject this input when selected.duplicated(['country_code',
    # 'year']).any().
    if selected.duplicated(["country_code", "year"]).any():
        # Raise the stated validation exception: Pre-treatment means have
        # duplicate country/year rows.
        raise IngestError(
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "Pre-treatment means have duplicate country/year rows."
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
    # Allocate one audit row per variable/group after enforcing complete finite
    # observations.
    rows: List[Dict[str, Any]] = []
    # Process each (variable, column) from columns.items() in the stated order.
    for variable, column in columns.items():
        # Process each (group, roster) from groups.items() in the stated order.
        for group, roster in groups.items():
            # Select this roster's country/year keys and completed variable in
            # the five-year window.
            subset: pd.DataFrame = selected.loc[
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                selected["country_code"].isin(roster),
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                ["country_code", "year", column],
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ]
            # Count nonmissing values per country to require exactly five
            # observations for every member.
            counts: pd.Series = subset.groupby("country_code")[column].count()
            # Reject this input when set(counts.index) != set(roster) or not
            # counts.eq(5).all().
            if set(counts.index) != set(roster) or not counts.eq(5).all():
                # Raise the stated validation exception with the failing
                # identity and observed values.
                raise IngestError(
                    # Group the following exact literal or expression across
                    # physical lines.
                    (
                        # Supply this explicit value or expression as the next
                        # part of the enclosing operation.
                        f"{variable}"
                        # Continue the exact string or sequence literal without
                        # changing its characters or order.
                        "/"
                        # Supply this explicit value or expression as the next
                        # part of the enclosing operation.
                        f"{group}"
                        # Continue the exact string or sequence literal without
                        # changing its characters or order.
                        ": incomplete five-year integrity sample."
                        # Close the preceding literal, argument list or
                        # selection
                        # without changing its contents.
                    )
                    # Close the preceding literal, argument list or selection
                    # without changing its contents.
                )
            # Convert observed values to floating arrays for the explicit
            # finiteness check.
            values: np.ndarray = pd.to_numeric(
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                subset[column],
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                errors="raise",
                # Supply this explicit value or expression as the next part of
                # the
                # enclosing operation.
            ).to_numpy(dtype=float)
            # Reject this input when not np.isfinite(values).all().
            if not np.isfinite(values).all():
                # Raise the stated validation exception with the failing
                # identity and observed values.
                raise IngestError(
                    # Supply this explicit value or expression as the next part
                    # of the enclosing operation.
                    f"{variable}/{group}: nonfinite integrity observations."
                    # Close the preceding literal, argument list or selection
                    # without changing its contents.
                )
            # Average each country's five-year mean, then average countries
            # with equal weight.
            observed: float = float(
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                subset.groupby("country_code")[column]
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                .mean()
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                .mean()
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
            # Read the published target mean for this particular variable/group
            # comparison.
            expected: float = float(targets[variable][group])
            # Apply the absolute tolerance plus one floating-point
            # machine-epsilon allowance.
            passed: bool = (
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                abs(observed - expected)
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                <= tolerance + np.finfo(float).eps
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
            # Append this explicit identity, count or binding verdict to the
            # ordered audit ledger.
            rows.append(
                # Begin the explicit key-to-value mapping used by the enclosing
                # operation.
                {
                    # Declare the variable field with its existing value.
                    "variable": variable,
                    # Declare the group field with its existing value.
                    "group": group,
                    # Declare the observed field with its existing value.
                    "observed": observed,
                    # Declare the expected field with its existing value.
                    "expected": expected,
                    # Declare the tolerance field with its existing value.
                    "tolerance": tolerance,
                    # Record this check's Boolean acceptance verdict.
                    "passed": passed,
                    # Close the preceding literal, argument list or selection
                    # without changing its contents.
                }
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
            # Reject this input when not passed.
            if not passed:
                # Raise the stated validation exception with the failing
                # identity and observed values.
                raise IngestError(
                    # Group the following exact literal or expression across
                    # physical lines.
                    (
                        # Continue the exact string or sequence literal without
                        # changing its characters or order.
                        "Pre-treatment mean "
                        # Supply this explicit value or expression as the next
                        # part of the enclosing operation.
                        f"{variable}"
                        # Continue the exact string or sequence literal without
                        # changing its characters or order.
                        "/"
                        # Supply this explicit value or expression as the next
                        # part of the enclosing operation.
                        f"{group}"
                        # Continue the exact string or sequence literal without
                        # changing its characters or order.
                        ": "
                        # Supply this explicit value or expression as the next
                        # part of the enclosing operation.
                        f"{observed:.12g}"
                        # Continue the exact string or sequence literal without
                        # changing its characters or order.
                        "; expected "
                        # Supply this explicit value or expression as the next
                        # part of the enclosing operation.
                        f"{expected:.12g}"
                        # Continue the exact string or sequence literal without
                        # changing its characters or order.
                        " within "
                        # Supply this explicit value or expression as the next
                        # part of the enclosing operation.
                        f"{tolerance:.12g}"
                        # Continue the exact string or sequence literal without
                        # changing its characters or order.
                        "."
                        # Close the preceding literal, argument list or
                        # selection
                        # without changing its contents.
                    )
                    # Close the preceding literal, argument list or selection
                    # without changing its contents.
                )
    # Return the completed result with its declared key, index and metadata
    # contract.
    return pd.DataFrame(rows)


# -----------------------------------------------------------------------------
# Step 9: Retained raw-grid coverage reconciliation
# -----------------------------------------------------------------------------


# Audit raw missingness on the frozen retained-country 2000-2024 grid.
def audit_raw_panel_coverage(
    # Accept config with the declared Mapping[str, Any] input contract.
    bundle: CleansedBundle,
    # Accept mask with the declared pd.Series input contract.
    mask: pd.Series,
    # Accept config with the declared Mapping[str, Any] input contract.
    config: Mapping[str, Any],
    # Complete the typed signature with the declared result type.
) -> pd.DataFrame:
    """
    Audit raw missingness on the frozen retained-country 2000-2024 grid.

    Parameters
    ----------
    bundle : CleansedBundle
        Cleansed raw ``wdi_raw_long`` and ``wgi_raw_long`` frames,
        before value completion. WDI needs ``country_code``, ``year``,
        ``indicator_code`` and ``value``; WGI needs the country/year key
        and ``rule_of_law``.
    mask : pandas.Series
        Country-indexed frozen eligibility mask. Truthy entries define
        the retained roster; this audit does not determine or alter
        eligibility.
    config : Mapping[str, Any]
        Expected WDI/WGI raw missing counts and the derived fuel-balance
        missing-cell count under ``execution_parameters.completion``.

    Returns
    -------
    audit : pandas.DataFrame
        Rows indexed by ``series``, with integer
        ``observed``/``expected`` and Boolean ``passed``. Includes each
        configured WDI target, WGI rule of law and the union of missing
        fuel-import/fuel-export primitive cells.

    Raises
    ------
    IngestError
        WDI country/year keys duplicate within a series, WGI contains
        multiple country/year observations, or any retained-grid missing
        count differs.
    ConfigValidationError
        A required dotted configuration binding is absent.
    KeyError
        Required source columns or directly indexed count declarations
        are absent.

    Notes
    -----
    Reindexes each source onto the complete retained-country/calendar
    grid; absent rows therefore count as missing rather than zero.
    Fuel-balance missingness is the Boolean union of its two raw
    primitive masks. No interpolation, treatment coding or cross-country
    completion occurs here.
    """
    # Freeze sorted country codes selected by the supplied eligibility mask.
    roster: Tuple[str, ...] = tuple(
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        sorted((str(code) for code in mask.index[mask.astype(bool)]))
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Build every retained-country/year cell for the 25-year 2000-2024 raw
    # coverage grid.
    index: pd.MultiIndex = pd.MultiIndex.from_product(
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        [roster, range(2000, 2025)],
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        names=["country_code", "year"],
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Read the cleansed but uncompleted long WDI source from the authoritative
    # bundle.
    raw: pd.DataFrame = bundle.frame("wdi_raw_long")
    # Read manuscript raw-WDI missing-count targets before comparing
    # retained-grid coverage.
    targets: Mapping[str, int] = configuration_value(
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        config,
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "raw_data_schemas.wdi_raw_long.raw_missing_targets",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Allocate raw coverage-count comparisons for WDI, WGI and derived
    # fuel-balance missingness.
    rows: List[Dict[str, Any]] = []
    # Retain each primitive's raw reindexed values for the later
    # fuel-missingness union.
    series: Dict[str, pd.Series] = {}
    # Process each code from tuple(targets) + ('TM.VAL.FUEL.ZS.UN',
    # 'TX.VAL.FUEL.ZS.UN') in the stated order.
    for code in tuple(targets) + ("TM.VAL.FUEL.ZS.UN", "TX.VAL.FUEL.ZS.UN"):
        # Select this WDI code's country/year/value rows without filling
        # missing observations.
        selected: pd.DataFrame = raw.loc[
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            raw["indicator_code"].eq(code),
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            ["country_code", "year", "value"],
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ]
        # Reject this input when selected.duplicated(['country_code',
        # 'year']).any().
        if selected.duplicated(["country_code", "year"]).any():
            # Raise the stated validation exception with the failing identity
            # and observed values.
            raise IngestError(
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                f"Raw coverage duplicates the {code} country/year key."
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
        # Reindex raw observations to the full retained grid so absent source
        # rows remain missing.
        values: pd.Series = selected.set_index(["country_code", "year"])[
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "value"
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
        ].reindex(index)
        # Store this primitive's uncompleted series for component-level
        # coverage checks.
        series[code] = values
        # Apply the following branch only when code in targets.
        if code in targets:
            # Append this explicit identity, count or binding verdict to the
            # ordered audit ledger.
            rows.append(
                # Begin the explicit key-to-value mapping used by the enclosing
                # operation.
                {
                    # Declare the series field with its existing value.
                    "series": code,
                    # Declare the observed field with its existing value.
                    "observed": int(values.isna().sum()),
                    # Declare the expected field with its existing value.
                    "expected": int(targets[code]),
                    # Close the preceding literal, argument list or selection
                    # without changing its contents.
                }
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
    # Read the raw governance source, which must have one release per
    # country/year.
    wgi: pd.DataFrame = bundle.frame("wgi_raw_long")
    # Reject this input when wgi.duplicated(['country_code', 'year']).any().
    if wgi.duplicated(["country_code", "year"]).any():
        # Raise the stated validation exception: Raw WGI coverage requires one
        # authenticated release per country/year.
        raise IngestError(
            # Group the following exact literal or expression across physical
            # lines.
            (
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "Raw WGI coverage requires one authenticated release per "
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "country/year."
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
    # Reindex raw observations to the full retained grid so absent source rows
    # remain missing.
    values = wgi.set_index(["country_code", "year"])["rule_of_law"].reindex(
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        index
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Append this explicit identity, count or binding verdict to the ordered
    # audit ledger.
    rows.append(
        # Begin the explicit key-to-value mapping used by the enclosing
        # operation.
        {
            # Declare the series field with its existing value.
            "series": "WGI_rule_of_law",
            # Declare the observed field with its existing value.
            "observed": int(values.isna().sum()),
            # Declare the expected field with its existing value.
            "expected": int(
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                config["raw_data_schemas"]["wgi_raw_long"][
                    # Continue the exact string or sequence literal without
                    # changing its characters or order.
                    "raw_missing_target"
                    # Close the preceding literal, argument list or selection
                    # without changing its contents.
                ]
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ),
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        }
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Union missing fuel-import and fuel-export cells before any primitive
    # completion.
    balance_missing: pd.Series = (
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        series["TM.VAL.FUEL.ZS.UN"].isna()
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        | series["TX.VAL.FUEL.ZS.UN"].isna()
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Append this explicit identity, count or binding verdict to the ordered
    # audit ledger.
    rows.append(
        # Begin the explicit key-to-value mapping used by the enclosing
        # operation.
        {
            # Declare the series field with its existing value.
            "series": "FuelShareBalance",
            # Declare the observed field with its existing value.
            "observed": int(balance_missing.sum()),
            # Declare the expected field with its existing value.
            "expected": int(
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                config["execution_parameters"]["completion"][
                    # Continue the exact string or sequence literal without
                    # changing its characters or order.
                    "derived_fuel_balance_missing_cells"
                    # Close the preceding literal, argument list or selection
                    # without changing its contents.
                ]
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ),
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        }
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Index the completed raw-count audit by series identifier.
    result: pd.DataFrame = pd.DataFrame(rows).set_index("series")
    # Mark exact observed-versus-expected integer-count equality for every
    # audited series.
    result["passed"] = result["observed"].eq(result["expected"])
    # Reject this input when not result['passed'].all().
    if not result["passed"].all():
        # Raise the stated validation exception with the failing identity and
        # observed values.
        raise IngestError(
            # Group the following exact literal or expression across physical
            # lines.
            (
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "Raw retained-panel coverage differs: "
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                f"{result.loc[~result['passed']].to_dict('index')}"
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "."
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
    # Return the completed result with its declared key, index and metadata
    # contract.
    return result


# -----------------------------------------------------------------------------
# Step 10: Independently authenticated trust registries
# -----------------------------------------------------------------------------


# Register externally pinned authorities after verifying their complete
# manifest.
def register_independent_authorities(
    # Accept study_config with the declared Mapping[str, Any] input contract.
    study_config: Mapping[str, Any],
    # Complete the typed signature with the declared result type.
) -> Tuple[str, ...]:
    """
    Register externally pinned authorities after verifying their
    complete manifest.

    Parameters
    ----------
    study_config : Mapping[str, Any]
        Configuration whose independent-authority declaration contains
        ``path``, lowercase ``sha256``,
        ``external_verification_reference`` and ``verified_by``. The
        expected digest must be registered independently of the
        candidate manifest being loaded.

    Returns
    -------
    blockers : Tuple[str, ...]
        Empty tuple when the complete document is accepted. If any
        required declaration is None, returns the single explicit
        ``independent_authorities:separately_verified_registration_manifest``
        blocker after clearing the four existing trust registries.

    Raises
    ------
    ConfigValidationError
        The manifest declaration path is missing.
    IngestError
        Metadata, expected/file digest, registry schema, identity
        fields, pins or ten-artifact vintage claims are malformed,
        duplicate or inconsistent.
    OSError, UnicodeDecodeError, json.JSONDecodeError
        The supplied manifest cannot be read or decoded as UTF-8 JSON.

    Notes
    -----
    Clears all four global trust registries before evaluating the
    declaration. Verifies file bytes against the externally supplied
    digest, parses exactly ``table_binding``, ``manuscript_universe``,
    ``reconciliation_scenarios`` and ``archive_vintages``, and validates
    every nested claim. Only after the entire document passes are its
    read-only registries published. A rejected manifest leaves prior
    registrations cleared; no candidate output creates its own pin.
    """
    # Allow this registrar to replace the global
    # TRUSTED_TABLE_BINDING_AUTHORITY_PINS registry.
    global TRUSTED_TABLE_BINDING_AUTHORITY_PINS
    # Allow this registrar to replace the global
    # TRUSTED_MANUSCRIPT_UNIVERSE_PINS registry.
    global TRUSTED_MANUSCRIPT_UNIVERSE_PINS
    # Allow this registrar to replace the global
    # TRUSTED_RECONCILIATION_SCENARIO_PINS registry.
    global TRUSTED_RECONCILIATION_SCENARIO_PINS
    # Allow this registrar to replace the global
    # TRUSTED_ARCHIVE_VINTAGE_REGISTRY registry.
    global TRUSTED_ARCHIVE_VINTAGE_REGISTRY
    # Read the separately verified manifest declaration; do not infer a trust
    # pin from outputs.
    declaration: Mapping[str, Optional[str]] = configuration_value(
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        study_config,
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "independent_authorities.registration_manifest",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Fix the four required external manifest identity and verification fields.
    required: Tuple[str, ...] = (
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "path",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "sha256",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "external_verification_reference",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "verified_by",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Process each registry_name from ('TRUSTED_TABLE_BINDING_AUTHORITY_PINS',
    # 'TRUSTED_MANUSCRIPT_UNIVERSE_PINS',
    # 'TRUSTED_RECONCILIATION_SCENARIO_PINS',
    # 'TRUSTED_ARCHIVE_VINTAGE_REGISTRY') in the stated order.
    for registry_name in (
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "TRUSTED_TABLE_BINDING_AUTHORITY_PINS",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "TRUSTED_MANUSCRIPT_UNIVERSE_PINS",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "TRUSTED_RECONCILIATION_SCENARIO_PINS",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "TRUSTED_ARCHIVE_VINTAGE_REGISTRY",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    ):
        # Clear this prior trust registry to an empty top-level read-only
        # mapping.
        globals()[registry_name] = MappingProxyType({})
    # Return the explicit unresolved-authority blocker when any required field
    # is None.
    if any((declaration.get(key) is None for key in required)):
        # Return an empty blocker tuple after publishing all validated
        # registries.
        return (
            # Group the following exact literal or expression across physical
            # lines.
            (
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "independent_authorities:separately_verified_"
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "registration_manifest"
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ),
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
    # Reject this input when any((not isinstance(declaration[key], str) or not
    # declaration[key].strip() for key in required)).
    if any(
        # Group the following exact literal or expression across physical
        # lines.
        (
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            not isinstance(declaration[key], str)
            # Include this explicit additional condition in the current
            # validation decision.
            or not declaration[key].strip()
            # Evaluate this comprehension over the stated items in their
            # existing order.
            for key in required
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    ):
        # Raise the stated validation exception: Independent authority manifest
        # metadata is malformed.
        raise IngestError(
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "Independent authority manifest metadata is malformed."
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
    # Use the already registered external SHA-256 as the manifest acceptance
    # anchor.
    expected_digest: str = declaration["sha256"]
    # Reject this input when not re.fullmatch('[0-9a-f]{64}', expected_digest).
    if not re.fullmatch("[0-9a-f]{64}", expected_digest):
        # Raise the stated validation exception: Independent authority manifest
        # needs a pre-registered SHA-256.
        raise IngestError(
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "Independent authority manifest needs a pre-registered SHA-256."
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
    # Expand and resolve the declared manifest location before reading its
    # exact bytes.
    path: Path = Path(declaration["path"]).expanduser().resolve()
    # Read the manifest bytes whose digest must match the independent
    # registered value.
    payload: bytes = path.read_bytes()
    # Reject this input when hashlib.sha256(payload).hexdigest() !=
    # expected_digest.
    if hashlib.sha256(payload).hexdigest() != expected_digest:
        # Raise the stated validation exception: Independent authority manifest
        # differs from its registered digest.
        raise IngestError(
            # Group the following exact literal or expression across physical
            # lines.
            (
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "Independent authority manifest differs from its "
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "registered digest."
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
    # Decode the verified bytes as UTF-8 and parse the JSON registry document.
    manifest: Any = json.loads(payload.decode("utf-8"))
    # Require exactly table, manuscript, reconciliation-scenario and
    # archive-vintage registries.
    names: Tuple[str, ...] = (
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "table_binding",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "manuscript_universe",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "reconciliation_scenarios",
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "archive_vintages",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Reject this input when not isinstance(manifest, Mapping) or set(manifest)
    # != set(names).
    if not isinstance(manifest, Mapping) or set(manifest) != set(names):
        # Raise the stated validation exception: Independent authority manifest
        # has an incomplete registry schema.
        raise IngestError(
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "Independent authority manifest has an incomplete registry schema."
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )

    # Parse one authenticated manifest registry into a read-only identity map.
    def parse_pins(
        # Accept fields with the declared Tuple[str, ...] input contract.
        name: str,
        # Accept fields with the declared Tuple[str, ...] input contract.
        fields: Tuple[str, ...],
        # Complete the typed signature with the declared result type.
    ) -> Mapping[Tuple[str, ...], str]:
        """
        Parse one authenticated manifest registry into a read-only identity
        map.

        Parameters
        ----------
        name : str
            Registry key in the verified outer ``manifest`` closure. Its
            value must be a nonempty list of records.
        fields : Tuple[str, ...]
            Ordered identity-field names. Each record must contain exactly
            these fields plus ``sha256``; identities must be nonblank
            strings.

        Returns
        -------
        pins : Mapping[Tuple[str, ...], str]
            Read-only map from stripped identity tuples, in the supplied
            field order, to validated lowercase 64-character hexadecimal
            SHA-256 strings.

        Raises
        ------
        IngestError
            The list is empty/malformed, fields are missing/extra, an
            identity is blank, a digest is malformed, or a stripped identity
            tuple duplicates.
        KeyError
            ``name`` is absent from the authenticated outer manifest.

        Notes
        -----
        The outer registrar verifies the manifest bytes before calling this
        helper. This parser validates pin syntax and identity uniqueness; it
        does not derive a new digest or certify a candidate publication
        artifact independently.
        """
        # Read the already authenticated manifest's requested registry records.
        records: Any = manifest[name]
        # Reject this input when not isinstance(records, list) or not records.
        if not isinstance(records, list) or not records:
            # Raise the stated validation exception with the failing identity
            # and observed values.
            raise IngestError(
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                f"Independent authority registry {name} is empty."
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
        # Allocate a unique identity-tuple-to-digest map for this one registry.
        pins: Dict[Tuple[str, ...], str] = {}
        # Process each record from records in the stated order.
        for record in records:
            # Reject this input when not isinstance(record, Mapping) or
            # set(record) != set(fields) | {'sha256'}.
            if not isinstance(record, Mapping) or set(record) != set(
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                fields
                # Supply this explicit value or expression as the next part of
                # the
                # enclosing operation.
            ) | {"sha256"}:
                # Raise the stated validation exception with the failing
                # identity and observed values.
                raise IngestError(f"Malformed independent {name} record.")
            # Reject this input when any((not isinstance(record[key], str) or
            # not record[key].strip() for key in fields)).
            if any(
                # Group the following exact literal or expression across
                # physical lines.
                (
                    # Supply this explicit value or expression as the next part
                    # of the enclosing operation.
                    not isinstance(record[key], str) or not record[key].strip()
                    # Evaluate this comprehension over the stated items in
                    # their existing order.
                    for key in fields
                    # Close the preceding literal, argument list or selection
                    # without changing its contents.
                )
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ):
                # Raise the stated validation exception with the failing
                # identity and observed values.
                raise IngestError(f"Blank independent {name} identity.")
            # Strip identity-field whitespace and preserve the configured
            # tuple-field order.
            key: Tuple[str, ...] = tuple(
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                (record[field].strip() for field in fields)
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
            # Read the supplied pin for syntax validation without deriving a
            # replacement digest.
            digest: Any = record["sha256"]
            # Reject this input when key in pins or not isinstance(digest, str)
            # or (not re.fullmatch('[0-9a-f]{64}', digest)).
            if (
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                key in pins
                # Include this explicit additional condition in the current
                # validation decision.
                or not isinstance(digest, str)
                # Include this explicit additional condition in the current
                # validation decision.
                or (not re.fullmatch("[0-9a-f]{64}", digest))
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ):
                # Raise the stated validation exception with the failing
                # identity and observed values.
                raise IngestError(
                    # Supply this explicit value or expression as the next part
                    # of the enclosing operation.
                    f"Duplicate or malformed independent {name} pin."
                    # Close the preceding literal, argument list or selection
                    # without changing its contents.
                )
            # Store the unique validated lowercase SHA-256 under its stripped
            # identity tuple.
            pins[key] = digest
        # Return the validated identity-to-digest pins through a read-only
        # mapping.
        return MappingProxyType(pins)

    # Parse externally registered table-binding identities and their digest
    # pins.
    table_pins: Mapping[Tuple[str, ...], str] = parse_pins(
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "table_binding",
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        ("authority_id", "version_id"),
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Parse externally registered manuscript-universe identities and their
    # digest pins.
    manuscript_pins: Mapping[Tuple[str, ...], str] = parse_pins(
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "manuscript_universe",
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        ("authority_id", "version_id"),
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Parse four-field invariant-scenario identities and their independently
    # registered pins.
    scenario_pins: Mapping[Tuple[str, ...], str] = parse_pins(
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "reconciliation_scenarios",
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        ("boundary_name", "forbidden_role", "probe_id", "probe_version"),
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Select the authenticated archive-vintage registry for exact ten-artifact
    # validation.
    vintages: Mapping[str, Mapping[str, str]] = manifest["archive_vintages"]
    # Reject this input when not isinstance(vintages, Mapping) or set(vintages)
    # != set(ARCHIVE_ARTEFACTS).
    if not isinstance(vintages, Mapping) or set(vintages) != set(
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        ARCHIVE_ARTEFACTS
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    ):
        # Raise the stated validation exception: Independent archive vintage
        # registry must cover all ten artifacts.
        raise IngestError(
            # Group the following exact literal or expression across physical
            # lines.
            (
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "Independent archive vintage registry must cover all ten "
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "artifacts."
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
    # Allocate validated read-only vintage records without publishing partial
    # registrations.
    vintage_pins: Dict[str, Mapping[str, str]] = {}
    # Process each (artifact, record) from vintages.items() in the stated
    # order.
    for artifact, record in vintages.items():
        # Fix the six fields required for every independently declared
        # artifact-vintage claim.
        expected_fields: Set[str] = {
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "authority_id",
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "source_id",
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "source_version",
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "vintage_tag",
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "evidence_reference",
            # Continue the exact string or sequence literal without changing
            # its characters or order.
            "claim_sha256",
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        }
        # Reject this input when not isinstance(record, Mapping) or set(record)
        # != expected_fields.
        if not isinstance(record, Mapping) or set(record) != expected_fields:
            # Raise the stated validation exception with the failing identity
            # and observed values.
            raise IngestError(
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                f"Incomplete independent vintage record for {artifact}."
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
        # Reject this input when any((not isinstance(value, str) or not
        # value.strip() for value in record.values())).
        if any(
            # Group the following exact literal or expression across physical
            # lines.
            (
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                not isinstance(value, str) or not value.strip()
                # Evaluate this comprehension over the stated items in their
                # existing order.
                for value in record.values()
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ):
            # Raise the stated validation exception with the failing identity
            # and observed values.
            raise IngestError(
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                f"Malformed independent vintage record for {artifact}."
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
        # Recompute this declared vintage claim's digest from its exact
        # artifact and provenance fields.
        claim: str = _archive_vintage_claim_digest(
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            artifact,
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            record["authority_id"],
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            record["source_id"],
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            record["source_version"],
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            record["vintage_tag"],
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            record["evidence_reference"],
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
        # Reject this input when record['claim_sha256'] != claim.
        if record["claim_sha256"] != claim:
            # Raise the stated validation exception with the failing identity
            # and observed values.
            raise IngestError(
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                f"Independent vintage claim differs for {artifact}."
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
        # Copy the validated vintage record into a top-level read-only record
        # mapping.
        vintage_pins[artifact] = MappingProxyType(dict(record))
    # Publish validated table pins only after every manifest registry has
    # passed.
    TRUSTED_TABLE_BINDING_AUTHORITY_PINS = table_pins
    # Publish validated manuscript pins after complete manifest acceptance.
    TRUSTED_MANUSCRIPT_UNIVERSE_PINS = manuscript_pins
    # Publish validated invariant-scenario pins after complete manifest
    # acceptance.
    TRUSTED_RECONCILIATION_SCENARIO_PINS = scenario_pins
    # Publish the validated ten-artifact vintage registry as a read-only outer
    # mapping.
    TRUSTED_ARCHIVE_VINTAGE_REGISTRY = MappingProxyType(vintage_pins)
    # Return an empty blocker tuple after publishing all validated registries.
    return ()


# -----------------------------------------------------------------------------
# Step 11: Effective matched-roster cardinalities
# -----------------------------------------------------------------------------


# Validate effective sanctions roster sizes for both matched level outcomes.
def assert_configured_matching_counts(
    # Accept config with the declared Mapping[str, Any] input contract.
    diagnostics: MatchingDiagnosticsResult,
    # Accept config with the declared Mapping[str, Any] input contract.
    config: Mapping[str, Any],
    # Complete the typed signature with the declared result type.
) -> None:
    """
    Validate effective sanctions roster sizes for both matched level
    outcomes.

    Parameters
    ----------
    diagnostics : MatchingDiagnosticsResult
        Live design mapping keyed by ``(policy, outcome,
        matching_axis)``. It must contain sanctions ``log_CPI`` and
        ``log_real_GDPpc`` designs for the ``high_income`` and
        ``mahalanobis`` axes.
    config : Mapping[str, Any]
        Algorithm declarations for expected target counts and
        high-income donor or Mahalanobis unique-donor counts.

    Returns
    -------
    None
        Successful checks do not modify designs or return an audit
        table.

    Raises
    ------
    IngestError
        Any of the four expected designs is absent or its effective
        roster lengths differ from the configured target/donor counts.
    ConfigValidationError
        The algorithm-parameters section is missing.
    KeyError, TypeError, ValueError
        Directly indexed count declarations are absent or cannot be
        converted to integers by the existing implementation.

    Notes
    -----
    Uses effective ``matched_targets``/``matched_donors``, not
    candidate-pool sizes or total neighbour selections. This is a
    cardinality check for Algorithms E1/E2; it neither computes distance
    nor assesses covariate balance.
    """
    # Read the high-income and Mahalanobis effective-roster count declarations.
    algorithms: Mapping[str, Any] = configuration_value(
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        config,
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "algorithm_parameters",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Process each outcome from ('log_CPI', 'log_real_GDPpc') in the stated
    # order.
    for outcome in ("log_CPI", "log_real_GDPpc"):
        # Process each (axis, section, donor_field) from (('high_income',
        # 'E1_high_income_design', 'expected_donors'), ('mahalanobis',
        # 'E2_mahalanobis_matching', 'expected_unique_donors')) in the stated
        # order.
        for axis, section, donor_field in (
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            ("high_income", "E1_high_income_design", "expected_donors"),
            # Group the following exact literal or expression across physical
            # lines.
            (
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "mahalanobis",
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "E2_mahalanobis_matching",
                # Continue the exact string or sequence literal without
                # changing its characters or order.
                "expected_unique_donors",
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            ),
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        ):
            # Identify this sanctions outcome/matching-axis design in the live
            # diagnostics mapping.
            key: Tuple[str, str, str] = ("strict_sanctions", outcome, axis)
            # Reject this input when key not in diagnostics.designs.
            if key not in diagnostics.designs:
                # Raise the stated validation exception with the failing
                # identity and observed values.
                raise IngestError(
                    # Supply this explicit value or expression as the next part
                    # of the enclosing operation.
                    f"Configured matching result is missing {key}."
                    # Close the preceding literal, argument list or selection
                    # without changing its contents.
                )
            # Read the actual matched design after verifying that its result
            # exists.
            design: MatchingDesignResult = diagnostics.designs[key]
            # Convert the configured target and effective-donor declarations to
            # the expected count tuple.
            expected: Tuple[int, int] = (
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                int(algorithms[section]["expected_targets"]),
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                int(algorithms[section][donor_field]),
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
            # Count effective matched rosters; do not count candidates or
            # repeated neighbour selections.
            actual: Tuple[int, int] = (
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                len(design.matched_targets),
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                len(design.matched_donors),
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
            # Reject this input when actual != expected.
            if actual != expected:
                # Raise the stated validation exception with the failing
                # identity and observed values.
                raise IngestError(
                    # Group the following exact literal or expression across
                    # physical lines.
                    (
                        # Continue the exact string or sequence literal without
                        # changing its characters or order.
                        "Effective matching roster counts for "
                        # Supply this explicit value or expression as the next
                        # part of the enclosing operation.
                        f"{key}"
                        # Continue the exact string or sequence literal without
                        # changing its characters or order.
                        ": "
                        # Supply this explicit value or expression as the next
                        # part of the enclosing operation.
                        f"{actual}"
                        # Continue the exact string or sequence literal without
                        # changing its characters or order.
                        "; expected "
                        # Supply this explicit value or expression as the next
                        # part of the enclosing operation.
                        f"{expected}"
                        # Continue the exact string or sequence literal without
                        # changing its characters or order.
                        "."
                        # Close the preceding literal, argument list or
                        # selection
                        # without changing its contents.
                    )
                    # Close the preceding literal, argument list or selection
                    # without changing its contents.
                )


# -----------------------------------------------------------------------------
# Step 12: Energy/GDP exposure-sample counts
# -----------------------------------------------------------------------------


# Bind published energy/GDP sample counts to explicit window/rule keys.
def configured_exposure_counts(
    # Accept config with the declared Mapping[str, Any] input contract.
    config: Mapping[str, Any],
    # Complete the typed signature with the declared result type.
) -> Dict[Tuple[str, str, str], int]:
    """
    Bind published energy/GDP sample counts to explicit window/rule
    keys.

    Parameters
    ----------
    config : Mapping[str, Any]
        Configuration with
        ``D5_exposure_sample_rules.regression_counts`` keyed as
        ``<first_year>_<last_year>_<coverage_rule>``.

    Returns
    -------
    counts : Dict[Tuple[str, str, str], int]
        Mapping from ``(window_name, coverage_rule, 'energy_exposure')``
        to integer sample size. Window names are formed as
        ``w<first>_<last>``.

    Raises
    ------
    ConfigValidationError
        The dotted sample-count declaration is absent.
    TypeError, ValueError
        A key does not split into exactly three components or a count
        cannot be converted by the existing ``int`` operation.

    Notes
    -----
    Counts belong to Equation (17)'s energy/GDP exposure samples only;
    they are not assigned to the alternative gas-share ratio of Equation
    (A1). Coverage eligibility and ratio standardisation are performed
    downstream.
    """
    # Read published window-specific strict/extended energy-exposure regression
    # counts.
    declared: Mapping[str, int] = configuration_value(
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        config,
        # Continue the exact string or sequence literal without changing its
        # characters or order.
        "algorithm_parameters.D5_exposure_sample_rules.regression_counts",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Allocate counts keyed by window, coverage rule and the energy/GDP
    # exposure variant.
    result: Dict[Tuple[str, str, str], int] = {}
    # Process each (name, count) from declared.items() in the stated order.
    for name, count in declared.items():
        # Split the configured count key into the two calendar endpoints and
        # coverage rule.
        first, last, rule = str(name).split("_")
        # Bind this integer count only to the named energy/GDP exposure sample.
        result[f"w{first}_{last}", rule, "energy_exposure"] = int(count)
    # Return the completed result with its declared key, index and metadata
    # contract.
    return result


# -----------------------------------------------------------------------------
# Step 13: Independent Figure-4 outcome panels
# -----------------------------------------------------------------------------


# Prepare Figure 4's independent CPI and real-GDP-per-capita overlays.
def build_two_panel_overlay_data(
    # Accept baseline_paths with the declared Mapping[Tuple[str, str],
    # pd.DataFrame] input contract.
    baseline_paths: Mapping[Tuple[str, str], pd.DataFrame],
    # Accept matched_paths with the declared Mapping[Tuple[str, str],
    # Mapping[str, pd.DataFrame]] input contract.
    matched_paths: Mapping[Tuple[str, str], Mapping[str, pd.DataFrame]],
    # Accept policy with the declared str input contract.
    policy: str,
    # Complete the typed signature with the declared result type.
) -> pd.DataFrame:
    """
    Prepare Figure 4's independent CPI and real-GDP-per-capita overlays.

    Parameters
    ----------
    baseline_paths : Mapping[Tuple[str, str], pandas.DataFrame]
        Outcome-specific baseline dynamic paths keyed by ``(policy,
        outcome)``, with the identity, normalization and pointwise-band
        fields required by ``build_overlay_data``.
    matched_paths : Mapping[Tuple[str, str], Mapping[str, pandas.DataFrame]]
        For each pair, its own high-income/Mahalanobis comparison paths
        and validated fit/roster/weight lineage required by
        ``build_overlay_data``.
    policy : str
        Coalition identifier used for both ``log_CPI`` and
        ``log_real_GDPpc`` lookups. Each level outcome must have
        independently estimated inputs.

    Returns
    -------
    combined : pandas.DataFrame
        Concatenated overlay frame with an outer ``outcome`` index level
        and the renderer's inner series/year index. ``panel_lineage``
        attrs record each outcome's input/output content digests.
        ``render_spec`` declares three series, baseline-only bands and
        normalization at 2013.

    Raises
    ------
    IngestError
        Either outcome lacks its baseline/matched pair, or delegated
        overlay validation rejects input identity, normalization or path
        lineage.

    Notes
    -----
    CPI and GDP retain separate rosters, donor weights, fitted paths and
    bands. The helper packages normalized fitted gaps; it does not refit
    a model, transfer one outcome's path to another, or introduce a
    year-2000 residual.
    """
    # Require independently prepared log_CPI and log_real_GDPpc panels in that
    # order.
    outcomes: Tuple[str, str] = ("log_CPI", "log_real_GDPpc")
    # Allocate separately validated overlay frames keyed by their own level
    # outcome.
    panels: Dict[str, pd.DataFrame] = {}
    # Allocate outcome-specific content digests for input/output path
    # provenance.
    lineage: Dict[str, Any] = {}
    # Process each outcome from outcomes in the stated order.
    for outcome in outcomes:
        # Resolve the requested policy and the current outcome to their
        # canonical input key.
        pair: Tuple[str, str] = (str(policy), outcome)
        # Reject this input when pair not in baseline_paths or pair not in
        # matched_paths.
        if pair not in baseline_paths or pair not in matched_paths:
            # Raise the stated validation exception with the failing identity
            # and observed values.
            raise IngestError(
                # Supply this explicit value or expression as the next part of
                # the enclosing operation.
                f"Figure 4 lacks the independent {pair} panel inputs."
                # Close the preceding literal, argument list or selection
                # without
                # changing its contents.
            )
        # Validate and assemble this outcome's own baseline and two matched
        # comparison paths.
        panel: pd.DataFrame = build_overlay_data(
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            baseline_paths[pair],
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            matched_paths[pair],
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
        # Retain this outcome's independently prepared overlay rather than
        # reusing another outcome.
        panels[outcome] = panel
        # Record separate overlay, baseline and matched-input content digests
        # for this outcome.
        lineage[outcome] = {
            # Record the prepared panel's content digest.
            "data_sha256": _content_digest(panel),
            # Record this outcome's baseline path content digest.
            "baseline_path_sha256": _content_digest(baseline_paths[pair]),
            # Record this outcome's matched-path mapping content digest.
            "matched_path_sha256": _content_digest(matched_paths[pair]),
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        }
    # Concatenate the two overlays under an outer outcome index level.
    combined: pd.DataFrame = pd.concat(panels, names=["outcome"])
    # Attach the separate CPI/GDP input/output fingerprints to the combined
    # frame.
    combined.attrs["panel_lineage"] = lineage
    # Declare the two-panel layout, three paths, baseline-only bands and 2013
    # normalization.
    combined.attrs["render_spec"] = {
        # Declare the kind field with its existing value.
        "kind": "matched_design_panels",
        # Declare the outcomes field with its existing value.
        "outcomes": list(outcomes),
        # Declare the series field with its existing value.
        "series": ["baseline", "high_income", "mahalanobis_m5"],
        # Keep pointwise bands off the matched comparison lines.
        "matched_bands": False,
        # Fix the supplied display normalization year at 2013.
        "normalization_year": 2013,
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    }
    # Return the completed result with its declared key, index and metadata
    # contract.
    return combined


# -----------------------------------------------------------------------------
# Step 14: Baseline-band matched-path rendering
# -----------------------------------------------------------------------------


# Render one prepared Figure 4 panel with a baseline-only pointwise band.
def draw_matching_overlay_panel(axes: Axes, panel: pd.DataFrame) -> None:
    """
    Render one prepared Figure 4 panel with a baseline-only pointwise
    band.

    Parameters
    ----------
    axes : matplotlib.axes.Axes
        Existing Matplotlib axes receiving this outcome's artists.
        Figure allocation, saving and closing remain the production
        caller's tasks.
    panel : pandas.DataFrame
        One outcome's validated overlay with a ``series`` index level
        and a remaining integer calendar-year index. ``baseline``,
        ``high_income`` and ``mahalanobis_m5`` supply ``value``;
        baseline also supplies the already-computed ``lower`` and
        ``upper`` pointwise bounds.

    Returns
    -------
    None
        Modifies the supplied axes by adding the band, three lines,
        reference lines, labels and legend. Does not modify the input
        DataFrame.

    Raises
    ------
    KeyError
        A required index level, series or numeric column is missing.
    TypeError, ValueError
        Calendar years or numeric arrays cannot be converted or
        rendered.

    Notes
    -----
    Sorts each calendar path, uses blue/green/red for
    baseline/high-income/ Mahalanobis, and shades only the baseline 95%
    band at alpha 0.20. Adds zero and 2014/2022 reference lines. Bands
    are supplied inputs, not confidence limits estimated here. The
    y-axis denotes gaps normalized to 2013, not conventional treatment
    leads or the Table-3 incremental ATT itself.
    """
    # Assign blue/green/red consistently to baseline, high-income and
    # Mahalanobis paths.
    colours: Dict[str, str] = {
        # Declare the baseline field with its existing value.
        "baseline": "#08519c",
        # Declare the high income field with its existing value.
        "high_income": "#31a354",
        # Declare the mahalanobis m5 field with its existing value.
        "mahalanobis_m5": "#de2d26",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    }
    # Select and chronologically sort this outcome's baseline before plotting
    # its band.
    baseline: pd.DataFrame = panel.xs("baseline", level="series").sort_index()
    # Convert the baseline calendar-year index to integers for the shaded
    # interval.
    years: np.ndarray = baseline.index.to_numpy(dtype=int)
    # Shade the already computed baseline lower/upper interval at alpha 0.20.
    axes.fill_between(
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        years,
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        baseline["lower"].to_numpy(dtype=float),
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        baseline["upper"].to_numpy(dtype=float),
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        color=colours["baseline"],
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        alpha=0.2,
        # Supply this explicit value or expression as the next part of the
        # enclosing operation.
        label="Baseline pointwise 95% band",
        # Close the preceding literal, argument list or selection without
        # changing
        # its contents.
    )
    # Process each series from ('baseline', 'high_income', 'mahalanobis_m5') in
    # the stated order.
    for series in ("baseline", "high_income", "mahalanobis_m5"):
        # Select and chronologically sort the current normalized design path.
        curve: pd.DataFrame = panel.xs(series, level="series").sort_index()
        # Plot this calendar-sorted normalized design path with its assigned
        # colour and label.
        axes.plot(
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            curve.index.to_numpy(dtype=int),
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            curve["value"].to_numpy(dtype=float),
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            linewidth=2.0,
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            color=colours[series],
            # Supply this explicit value or expression as the next part of the
            # enclosing operation.
            label=series,
            # Close the preceding literal, argument list or selection without
            # changing its contents.
        )
    # Draw the zero normalized-gap horizontal reference line.
    axes.axhline(0.0, color="#636363", linewidth=0.8)
    # Draw the declared 2014 or 2022 treatment-wave calendar reference.
    axes.axvline(2014, color="black", linewidth=0.8)
    # Draw the declared 2014 or 2022 treatment-wave calendar reference.
    axes.axvline(2022, color="black", linewidth=0.8)
    # Label the horizontal coordinate as calendar year.
    axes.set_xlabel("Calendar year")
    # Label the vertical coordinate as the coalition gap normalized to 2013.
    axes.set_ylabel("Coalition gap normalized to 2013")
    # Place the three-path and baseline-band legend in the best available
    # position.
    axes.legend(loc="best", fontsize=8)
````

````python
# Task 1: Ingest and Fingerprint All Raw Data Structures

# ==============================================================================
# Task 1: Ingest and Fingerprint All Raw Data Structures
# ==============================================================================

# Create the module logger that records ingest events.
logger: logging.Logger = logging.getLogger(__name__)

# Compile the syntactic test every country code column must satisfy.
ISO3_PATTERN: Final[re.Pattern[str]] = re.compile(r"^[A-Z]{3}$")
# Declare the closed vocabulary of textual missing-value markers.
MISSING_MARKERS: Final[frozenset[str]] = frozenset(
    {"", "..", "NA", "N/A", "n/a", "na", "NaN", "nan", "NULL", "null", "None"}
)
# Declare the first calendar year the retrieval covers.
YEAR_FIRST: Final[int] = 1996
# Declare the final calendar year the retrieval covers.
YEAR_LAST: Final[int] = 2024
# Declare the World Bank aggregate codes the ingest must tolerate, not accept.
WORLD_BANK_AGGREGATE_CODES: Final[frozenset[str]] = frozenset(
    {
        "WLD", "EAP", "ECS", "HIC", "IBD", "IBT", "IDA", "IDB", "IDX",
        "LTE", "MIC", "LMY", "LMC", "OSS", "PRE", "SAS", "SSA", "UMC",
    }
)
# Declare the names of the ten sources in their contractual ingest order.
DECLARED_SOURCES: Final[Tuple[str, ...]] = (
    "wdi_raw_wide",
    "wdi_raw_long",
    "wgi_raw_long",
    "comtrade_block_a_raw",
    "comtrade_block_b_raw",
    "comtrade_block_c_raw",
    "comtrade_cell_status_raw",
    "nominal_gdp_raw",
    "policy_ledger_raw",
    "country_metadata_raw",
)
# Declare the nineteen sections every study configuration must expose.
CONFIG_SECTIONS: Final[Tuple[str, ...]] = (
    "identity",
    "global_constants",
    "raw_data_schemas",
    "equation_parameters",
    "algorithm_parameters",
    "model_parameters",
    "methodology_parameters",
    "network_architecture_parameters",
    "network_training_parameters",
    "llm_auxiliary_configuration",
    "plot_construction_parameters",
    "reproducibility_gaps",
    "methodological_notes",
    "data_preprocessing_parameters",
    "series_code_registry",
    "country_validation_registries",
    "pretreatment_means_2009_2013_integrity_targets",
    "estimation_sequencing",
    "invalid_resample_policy",
)
# Cap the number of offending values quoted inside any raised message.
MAX_REPORTED_OFFENDERS: Final[int] = 20
# Declare the primary key used to fingerprint the Block B partition artefacts.
# It is the declared Block B schema key, so the partition digests are computed
# over exactly the serialization the ingest fingerprint uses.
BLOCK_B_FINGERPRINT_KEYS: Final[Tuple[str, ...]] = (
    "reporter_code",
    "partner_code",
    "year",
    "product_description",
    "trade_flow",
)
# Declare the configuration keys whose single value is a two-element window.
TUPLE_VALUED_CONFIG_KEYS: Final[frozenset[str]] = frozenset(
    {
        "pre_treatment_window",
        "post_treatment_window",
        "Wave1",
        "Wave2",
        "preferred_reference_window",
        "baseline_window",
        "critical_window",
        "whole_period_window",
        "temporal_universe_completion",
        "temporal_universe_output",
        "temporal_universe_anchoring",
        "Post_period",
        "percentile_bounds",
        "edge_growth_bounds",
    }
)


# Declare the base class of every ingest failure raised by this module.
class IngestError(Exception):
    """
    Base class for every ingest-layer failure raised by this module.

    Attributes
    ----------
    args : tuple
        Inherited exception arguments carrying the failure message.

    Notes
    -----
    All ingest, schema, key, integrity, and partition failures
    subclass this base so callers can catch the whole layer
    with a single handler.

    Examples
    --------
    >>> try:
    ...     raise IngestError("demo failure")
    ... except IngestError as exc:
    ...     print(str(exc))
    demo failure
    """


# Declare the failure raised when a frame violates its declared schema.
class SchemaViolationError(IngestError):
    """
    Raised when a frame's columns, dtypes or year coverage are violated.

    Attributes
    ----------
    args : tuple
        Inherited arguments naming the violated column or coverage.

    Notes
    -----
    Raised before any join so a revision change cannot silently
    corrupt the eligible sample.

    Examples
    --------
    >>> try:
    ...     raise SchemaViolationError("missing column")
    ... except IngestError as exc:
    ...     print(type(exc).__name__)
    SchemaViolationError
    """


# Declare the failure raised when a declared primary key is not unique.
class PrimaryKeyViolationError(IngestError):
    """
    Raised when a frame's declared primary key does not identify rows.

    Attributes
    ----------
    args : tuple
        Inherited arguments naming the duplicated key.

    Notes
    -----
    Checked on the declared tuple per Section 3 before any join.

    Examples
    --------
    >>> try:
    ...     raise PrimaryKeyViolationError("duplicate key")
    ... except IngestError:
    ...     pass
    """


# Declare the failure raised when country codes violate the ISO-3 pattern.
class ReferentialIntegrityError(IngestError):
    """
    Raised when a code column violates its pattern or its universe.

    Attributes
    ----------
    args : tuple
        Inherited arguments naming the offending code value.

    Notes
    -----
    Enforces ISO-3 pattern ``^[A-Z]{3}$`` and denylist filtering
    before aggregates can enter the panel.

    Examples
    --------
    >>> try:
    ...     raise ReferentialIntegrityError("bad code")
    ... except IngestError:
    ...     pass
    """


# Declare the failure raised when the Block B disjunction is violated.
class BlockBPartitionError(IngestError):
    """
    Raised when the Block B partition cannot be formed consistently.

    Attributes
    ----------
    args : tuple
        Inherited arguments describing the malformed direction.

    Notes
    -----
    Guards the import versus exporter-mirror disjunction so no
    row satisfies neither or both clauses silently.

    Examples
    --------
    >>> try:
    ...     raise BlockBPartitionError("bad direction")
    ... except IngestError:
    ...     pass
    """


# Declare the immutable schema record describing one ingested DataFrame.
@dataclass(frozen=True)
class SourceSchema:
    """
    Declarative schema of one raw source frame.

    Attributes
    ----------
    source_name : str
        The canonical source name, matching a key of the ingest mapping.
    base_columns : tuple of str
        The ordered non-year columns the frame must expose exactly.
    key_columns : tuple of str
        The columns whose combination must identify rows uniquely.
    float_columns : tuple of str
        The columns that must be cast to double precision.
    int_columns : tuple of str
        The columns that must be cast to the declared integer widths.
    category_columns : tuple of str
        The columns that must be cast to the pandas categorical dtype.
    code_columns : tuple of str
        The columns whose values must match the three-letter pattern.
    membership_columns : tuple of str
        The subset of code columns checked against the metadata universe.
    expands_years : bool
        Whether the frame carries one column per calendar year.
    extension_columns : tuple of str
        An optional exact ordered suffix within the same named source.
    """
    # The canonical source name keying every mapping in this module.
    source_name: str
    # The ordered non-year columns the frame must expose exactly.
    base_columns: Tuple[str, ...]
    # The columns that must jointly identify rows uniquely.
    key_columns: Tuple[str, ...]
    # The columns that must be cast to double precision.
    float_columns: Tuple[str, ...]
    # The columns that must be cast to integers.
    int_columns: Tuple[str, ...]
    # The columns that must be cast to the categorical dtype.
    category_columns: Tuple[str, ...]
    # The columns whose values must match the three-letter pattern.
    code_columns: Tuple[str, ...]
    # The subset of code columns checked against the registry universe.
    membership_columns: Tuple[str, ...]
    # Whether the frame carries one column per calendar year.
    expands_years: bool = False
    # One exact optional suffix for an authenticated version of this source.
    extension_columns: Tuple[str, ...] = ()


# Declare the immutable record of one source's ingest outcome.
@dataclass(frozen=True)
class IngestRecord:
    """
    Immutable audit record of one source frame's ingest.

    Attributes
    ----------
    source_name : str
        The canonical source name.
    n_rows : int
        The number of rows after canonicalization.
    n_columns : int
        The number of columns after canonicalization.
    key_columns : tuple of str
        The declared primary key of the source.
    fingerprint_sha256 : str
        The hexadecimal digest of the key-sorted serialized frame.
    ingested_utc : str
        The ISO-8601 UTC timestamp of ingest.
    dtype_map : dict of str to str
        The observed dtype of every column after enforcement.
    notes : tuple of str
        The logged coercions and canonicalizations applied to the source.
    validation_stages : tuple of str
        The ordered validation stages passed before the digest was computed,
        proving that the fingerprint describes a fully validated frame.
    region_identity_sha256 : str, optional
        For country metadata only, the ingest-time digest of the declared
        region source/version attributes; other sources carry ``None``.
    """
    # The canonical source name this record describes.
    source_name: str
    # The row count observed after canonicalization.
    n_rows: int
    # The column count observed after canonicalization.
    n_columns: int
    # The declared primary key of the source.
    key_columns: Tuple[str, ...]
    # The hexadecimal digest of the key-sorted serialized frame.
    fingerprint_sha256: str
    # The ISO-8601 UTC timestamp of ingest.
    ingested_utc: str
    # The observed dtype of every column after enforcement.
    dtype_map: Dict[str, str]
    # The logged coercions and canonicalizations applied to the source.
    notes: Tuple[str, ...]
    # The ordered validation stages passed before the digest was computed.
    validation_stages: Tuple[str, ...] = ()
    # Metadata-only digest of the declared region source/version attributes.
    region_identity_sha256: Optional[str] = None


# Declare the immutable partition of the Block B disjunction.
@dataclass(frozen=True)
class BlockBPartition:
    """
    Immutable partition of Block B into retained and malformed rows.

    Attributes
    ----------
    retained : pandas.DataFrame
        Rows satisfying exactly one of the two disjuncts.
    malformed : pandas.DataFrame
        Rows satisfying neither disjunct, carrying a reason column.
    digests : tuple of (str, str)
        Ordered (name, digest) pairs covering the original pre-partition
        frame, the retained frame and the malformed ledger, so the
        conservation identity has a machine-checkable digest counterpart.
    """
    # Rows satisfying exactly one of the two disjuncts.
    retained: pd.DataFrame
    # Rows satisfying neither disjunct, carrying a reason column.
    malformed: pd.DataFrame
    # Ordered (name, digest) pairs over the three partition artefacts.
    digests: Tuple[Tuple[str, str], ...] = ()


# Declare the immutable bundle of canonicalized sources and their records.
@dataclass(frozen=True)
class IngestBundle:
    """
    Immutable ingest result consumed by every downstream task.

    Attributes
    ----------
    frames : dict of str to pandas.DataFrame
        The canonicalized frames keyed by source name; the Block B entry is
        the retained partition.
    records : dict of str to IngestRecord
        One audit record per source, keyed identically.
    study_config : dict
        The configuration mapping the ingest was validated against.
    config_fingerprint : str
        The hexadecimal digest of the configuration serialization.
    ingested_utc : str
        The ISO-8601 UTC timestamp shared by the whole ingest.
    block_b_malformed : pandas.DataFrame
        The malformed Block B ledger retained for audit.
    block_b_original : pandas.DataFrame
        The canonical Block B source before the direction partition.
    block_b_partition_digests : tuple of (str, str)
        Labelled source, retained and malformed fingerprints.
    """
    # The canonicalized frames keyed by source name.
    frames: Mapping[str, pd.DataFrame]
    # The per-source audit records keyed by source name.
    records: Mapping[str, IngestRecord]
    # The configuration mapping the ingest was validated against.
    study_config: Mapping[str, Any]
    # The hexadecimal digest of the configuration serialization.
    config_fingerprint: str
    # The ISO-8601 UTC timestamp shared by the whole ingest.
    ingested_utc: str
    # The malformed Block B ledger retained for audit.
    block_b_malformed: pd.DataFrame
    # The canonical Block B source before direction filtering.
    block_b_original: pd.DataFrame
    # The labelled fingerprints proving the Block B partition.
    block_b_partition_digests: Tuple[Tuple[str, str], ...]

    # Expose one canonicalized frame by source name.
    def frame(self, source_name: str) -> pd.DataFrame:
        """
        Return the canonicalized frame of a source.

        Parameters
        ----------
        source_name : str
            The canonical source name.

        Returns
        -------
        pandas.DataFrame
            The canonicalized frame.

        Raises
        ------
        KeyError
            If the bundle holds no frame under the requested name.
        """
        # Guard the lookup so a misspelled source fails loudly.
        if source_name not in self.frames:
            # Report the available names to make the failure diagnosable.
            raise KeyError(
                f"Frame {source_name!r} is not in the bundle; "
                f"available: {sorted(self.frames)}."
            )
        # Return the requested frame.
        return self.frames[source_name]

    # Expose one audit record by source name.
    def record(self, source_name: str) -> IngestRecord:
        """
        Return the audit record of a source.

        Parameters
        ----------
        source_name : str
            The canonical source name.

        Returns
        -------
        IngestRecord
            The audit record of the requested source.

        Raises
        ------
        KeyError
            If the bundle holds no record under the requested name.
        """
        # Guard the lookup so a misspelled source fails loudly.
        if source_name not in self.records:
            # Report the available names to make the failure diagnosable.
            raise KeyError(
                f"Record {source_name!r} is not in the bundle; "
                f"available: {sorted(self.records)}."
            )
        # Return the requested record.
        return self.records[source_name]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 1, Step 1: Read the ten DataFrames and the study_config dictionary into memory
# -------------------------------------------------------------------------------------------------------------------------------


# Return the current UTC instant as an ISO-8601 string.
def current_utc_iso() -> str:
    """
    Return the current UTC instant in ISO-8601 form.

    Returns
    -------
    str
        The timezone-aware timestamp, ISO-8601 encoded with the offset.

    Raises
    ------
    None
        The function cannot fail for well-formed system clocks.

    Examples
    --------
    >>> stamp = current_utc_iso()
    >>> stamp.endswith("+00:00")
    True
    """
    # Produce a timezone-aware timestamp directly in UTC.
    return datetime.now(timezone.utc).isoformat()


# Validate that every declared configuration section is present.
def validate_study_config_sections(study_config: Mapping[str, Any]) -> None:
    """
    Assert that the configuration exposes all nineteen declared sections.

    Parameters
    ----------
    study_config : mapping
        The study configuration dictionary to validate.

    Returns
    -------
    None
        The function returns nothing on success.

    Raises
    ------
    TypeError
        If the configuration is not a mapping.
    SchemaViolationError
        If any declared section key is absent from the configuration.
    """
    # Reject a configuration that is not a mapping at all.
    if not isinstance(study_config, Mapping):
        # A non-mapping configuration cannot expose sections.
        raise TypeError(
            f"study_config must be a mapping; received "
            f"{type(study_config).__name__}."
        )
    # Compute every declared section missing from the configuration.
    missing: List[str] = [name for name in CONFIG_SECTIONS if name not in study_config]
    # Refuse a configuration that omits any declared section.
    if missing:
        # Report every missing section so one pass repairs the configuration.
        raise SchemaViolationError(
            f"study_config is missing {len(missing)} declared section(s): {missing}."
        )
    # Log the successful structural validation.
    logger.info("study_config exposes all %d declared sections.", len(CONFIG_SECTIONS))


# Restore the declared tuple-valued configuration entries after a JSON load.
def restore_config_tuples(
    study_config: Mapping[str, Any],
) -> Tuple[Dict[str, Any], int]:
    """
    Restore declared tuple values collapsed to lists by a JSON round trip.

    Parameters
    ----------
    study_config : mapping
        The configuration mapping as loaded, which may have collapsed the
        declared window tuples and bounds into lists.

    Returns
    -------
    tuple
        A new configuration mapping in which every declared tuple-valued entry
        is restored to its tuple form, together with the number of conversions
        performed; the count is incremented once per restored window tuple, so
        a windows-list holding k windows contributes exactly k conversions.

    Raises
    ------
    TypeError
        If the configuration is not a mapping.
    SchemaViolationError
        If the deterministic replay pass disagrees with the primary pass on
        the number of leaf conversions, which would indicate concurrent
        mutation of the configuration during restoration.
    """
    # Reject a configuration that is not a mapping at all.
    if not isinstance(study_config, Mapping):
        # A non-mapping configuration exposes no keys to restore.
        raise TypeError(
            f"study_config must be a mapping; received "
            f"{type(study_config).__name__}."
        )
    # Track the number of conversions using a single-element counter list.
    counter: List[int] = [0]

    # Define the recursive restoration over nested mappings and lists.
    def _restore(value: Any, key: str, counter: List[int]) -> Any:
        """
        Return the restored form of one configuration node.

        Parameters
        ----------
        value : Any
            The node to restore.
        key : str
            The configuration key under which the node was found.
        counter : list of int
            The single-element counter incremented once per restored tuple.

        Returns
        -------
        Any
            The node with every declared tuple-valued entry restored.
        """
        # Restore a mapping by recursing over each of its entries.
        if isinstance(value, Mapping):
            # Build a new mapping whose children are restored in turn.
            return {
                child_key: _restore(child_value, str(child_key), counter)
                for child_key, child_value in value.items()
            }
        # Restore a list, applying the declared window rules at this node.
        if isinstance(value, list):
            # Restore each element of the list before applying the rules.
            restored_items: List[Any] = [
                _restore(item, "", counter) for item in value
            ]
            # Test whether the node is a two-element numeric window.
            is_window: bool = (
                len(restored_items) == 2
                and all(
                    isinstance(item, (int, float)) and not isinstance(item, bool)
                    for item in restored_items
                )
            )
            # Convert a declared single-window key held as a numeric pair.
            if key in TUPLE_VALUED_CONFIG_KEYS and is_window:
                # Count this leaf conversion exactly once.
                counter[0] += 1
                # Return the restored single-window tuple.
                return tuple(restored_items)
            # Convert a list of windows found under a windows-named key.
            if key.endswith("windows") and restored_items and all(
                isinstance(item, list) and len(item) == 2 for item in restored_items
            ):
                # Count one conversion per restored window tuple, so a list
                # holding k windows contributes exactly k leaf events.
                counter[0] += len(restored_items)
                # Return the list of restored window tuples.
                return [tuple(item) for item in restored_items]
            # Return the restored list unchanged when no rule applies.
            return restored_items
        # Return scalar values unchanged.
        return value

    # Restore the configuration from its top level (primary pass).
    restored: Dict[str, Any] = _restore(study_config, "", counter)
    # Run the deterministic replay pass with an independent counter.
    replay_counter: List[int] = [0]
    # Re-restore the ORIGINAL input so the replay recounts the same leaves.
    _restore(study_config, "", replay_counter)
    # Assert the two deterministic passes agree on the conversion count.
    if replay_counter[0] != counter[0]:
        # A divergence means the input mutated during restoration.
        raise SchemaViolationError(
            "study_config tuple restoration is non-deterministic: the primary "
            f"pass counted {counter[0]} conversion(s) while the replay pass "
            f"counted {replay_counter[0]}."
        )
    # Log the restoration outcome for the audit trail.
    logger.info(
        "study_config tuple restoration: %d declared entr(ies) converted.",
        counter[0],
    )
    # Return the restored configuration and the conversion count.
    return restored, counter[0]


# Replace every textual missing marker with the numpy sentinel.
def canonicalize_missing_markers(
    frame: pd.DataFrame, source_name: str
) -> Tuple[pd.DataFrame, Tuple[str, ...]]:
    """
    Convert declared missing markers to numpy nan without touching zeros.

    Parameters
    ----------
    frame : pandas.DataFrame
        The raw frame whose string columns may carry textual markers.
    source_name : str
        The canonical source name, used in diagnostics.

    Returns
    -------
    tuple
        The canonicalized copy of the frame, and the notes recording how many
        markers were replaced per column.

    Raises
    ------
    TypeError
        If the frame is not a pandas DataFrame.
    IngestError
        If the frame is empty, since an empty payload is not an ingest.
    """
    # Reject a non-DataFrame input before any method call.
    if not isinstance(frame, pd.DataFrame):
        # Report the received type so the caller can correct the wiring.
        raise TypeError(
            f"{source_name}: expected a pandas DataFrame; received "
            f"{type(frame).__name__}."
        )
    # Reject an empty frame, which would pass every later check vacuously.
    if len(frame) == 0:
        # An empty source payload is a malformed ingest.
        raise IngestError(f"{source_name}: the frame contains no rows.")
    # Resolve the declared schema so documentary text is never treated as data.
    schema: Optional[SourceSchema] = build_source_schemas().get(source_name)
    # Refuse an unknown source because its marker-bearing fields are ambiguous.
    if schema is None:
        # The ingest contract must classify fields before marker replacement.
        raise SchemaViolationError(
            f"{source_name}: no source schema is registered for marker "
            "canonicalization."
        )
    # Copy the frame so the caller's object is never mutated.
    canonical: pd.DataFrame = frame.copy()
    # Preserve an independent snapshot for the untouched-cell proof.
    original: pd.DataFrame = frame.copy(deep=True)
    # Accumulate one auditable entry per changed cell.
    notes: List[str] = []
    # Numeric measures, years and flags are the only marker-eligible fields.
    numeric_columns: List[Any] = list(schema.float_columns) + list(
        schema.int_columns
    )
    # Include the wide annual measures declared by an expanding schema.
    if schema.expands_years:
        # Accept string or integer year labels before label normalization.
        numeric_columns.extend(
            column
            for column in canonical.columns
            if re.fullmatch(r"(19|20)\d{2}", str(column).strip())
        )
    # Visit each eligible column once in source order.
    for column in dict.fromkeys(numeric_columns):
        # Skip absent year aliases; column validation reports required fields.
        if column not in canonical.columns:
            # Marker canonicalization must not manufacture absent columns.
            continue
        # Select the potentially textual numeric series to inspect.
        series: pd.Series = canonical[column]
        # Identify the cells holding a declared marker string.
        marker_mask: pd.Series = series.isin(MISSING_MARKERS)
        # Count the markers found in this column.
        n_markers: int = int(marker_mask.sum())
        # Replace the markers with the numpy sentinel in one assignment.
        if n_markers:
            # Assign nan only to the marker positions, preserving everything else.
            canonical.loc[marker_mask, column] = np.nan
            # Record every changed cell with its original row identity.
            notes.extend(
                f"cell[{index!r}, {column!r}]: {original.at[index, column]!r} "
                "-> nan"
                for index in canonical.index[marker_mask]
            )
    # Build a mask covering exactly the cells documented as changed.
    changed_coordinates: set[Tuple[Any, Any]] = {
        (index, column)
        for column in canonical.columns
        for index in canonical.index
        if pd.isna(canonical.at[index, column])
        and not pd.isna(original.at[index, column])
    }
    # Compare every untouched observation with its original value.
    for column in canonical.columns:
        # Check each row explicitly so object-valued provenance is preserved.
        for index in canonical.index:
            # Ignore only the cells deliberately converted to missing.
            if (index, column) in changed_coordinates:
                # This coordinate is represented in the explicit audit.
                continue
            # Read the old and new values for a null-aware equality test.
            old_value: Any = original.at[index, column]
            new_value: Any = canonical.at[index, column]
            # Treat two pre-existing missing values as equal.
            equal: bool = bool(
                (pd.isna(old_value) and pd.isna(new_value))
                or old_value == new_value
            )
            # Refuse any unrecorded mutation.
            if not equal:
                # This is an internal integrity failure, not valid output.
                raise IngestError(
                    f"{source_name}: marker canonicalization altered untouched "
                    f"cell ({index!r}, {column!r})."
                )
    # Log the canonicalization outcome for the source.
    if notes:
        # Report each replacement in a single informational line.
        logger.info("%s: canonicalized missing markers: %s", source_name, notes)
    # Return the canonicalized copy alongside its notes.
    return canonical, tuple(notes)


# Restore the wide frame's year columns as integer labels.
def normalize_year_columns(
    frame: pd.DataFrame, schema: SourceSchema
) -> Tuple[pd.DataFrame, Tuple[str, ...]]:
    """
    Coerce year-like column labels to integers and verify year coverage.

    Parameters
    ----------
    frame : pandas.DataFrame
        The canonicalized frame; only year-expanding schemas are altered.
    schema : SourceSchema
        The schema declaring whether the frame expands by year.

    Returns
    -------
    tuple
        The frame with integer year labels, and the notes recording the
        coercion.

    Raises
    ------
    SchemaViolationError
        If the frame lacks any declared year or carries a year outside the
        declared range.
    """
    # Return non-year frames unchanged, with an empty note tuple.
    if not schema.expands_years:
        # A frame without year columns needs no label normalization.
        return frame, ()
    # Copy the frame so the caller's object is never mutated.
    normalized: pd.DataFrame = frame.copy()
    # Map every string label that names a calendar year to its integer value.
    year_like: Dict[str, int] = {
        str(column): int(str(column).strip())
        for column in normalized.columns
        if isinstance(column, str) and re.fullmatch(r"(19|20)\d{2}", str(column).strip())
    }
    # Rename the year-like labels to integers when any were found.
    if year_like:
        # Apply the label coercion in a single rename operation.
        normalized = normalized.rename(columns=year_like)
    # Compute the declared year range.
    declared_years: List[int] = list(range(YEAR_FIRST, YEAR_LAST + 1))
    # Compute the missing years from the frame's column labels.
    missing_years: List[int] = [
        year for year in declared_years if year not in normalized.columns
    ]
    # Refuse a frame whose year coverage is incomplete.
    if missing_years:
        # Report the first missing years so the payload can be repaired.
        raise SchemaViolationError(
            f"{schema.source_name}: missing year column(s) "
            f"{missing_years[:MAX_REPORTED_OFFENDERS]}."
        )
    # Compute any year-labelled column outside the declared range.
    unexpected_years: List[int] = [
        int(column)
        for column in normalized.columns
        if isinstance(column, (int, np.integer))
        and int(column) not in range(YEAR_FIRST, YEAR_LAST + 1)
    ]
    # Refuse a frame carrying a year outside the declared range.
    if unexpected_years:
        # Report the offending labels so the retrieval can be corrected.
        raise SchemaViolationError(
            f"{schema.source_name}: unexpected year column(s) "
            f"{sorted(unexpected_years)[:MAX_REPORTED_OFFENDERS]}."
        )
    # Compose the coercion note when string labels were rewritten.
    notes: Tuple[str, ...] = (
        (
            f"year labels coerced to integers: {len(year_like)} column(s) "
            f"spanning {YEAR_FIRST} to {YEAR_LAST}.",
        )
        if year_like
        else ()
    )
    # Log the coercion outcome for the source.
    if notes:
        # Emit the coercion note as an informational event.
        logger.info("%s: %s", schema.source_name, notes[0])
    # Return the normalized frame alongside its notes.
    return normalized, notes


# Enforce the declared dtypes on every column of a frame.
def enforce_dtypes(
    frame: pd.DataFrame, schema: SourceSchema
) -> Tuple[pd.DataFrame, Tuple[str, ...]]:
    """
    Cast every declared column to its required dtype.

    Parameters
    ----------
    frame : pandas.DataFrame
        The frame whose dtypes are to be enforced.
    schema : SourceSchema
        The schema declaring the required dtypes.

    Returns
    -------
    tuple
        The cast frame, and the notes recording the dtype transitions.

    Raises
    ------
    SchemaViolationError
        If a declared column is absent, if a numeric column contains a
        non-numeric value, or if an integer column contains a missing value.
    """
    # Copy the frame so the caller's object is never mutated.
    cast: pd.DataFrame = frame.copy()
    # Accumulate the dtype transitions for the audit notes.
    notes: List[str] = []
    # Determine the float columns, including integer-labelled year columns.
    float_columns: List[Any] = [
        column
        for column in cast.columns
        if column in schema.float_columns
        or (schema.expands_years and isinstance(column, (int, np.integer)))
    ]
    # Cast every declared float column to double precision.
    for column in float_columns:
        # Convert the column, raising on any non-numeric content.
        try:
            # Apply the numeric conversion with strict error handling.
            converted: pd.Series = pd.to_numeric(cast[column], errors="raise")
        except (ValueError, TypeError) as exc:
            # Re-raise with the source and column context attached.
            raise SchemaViolationError(
                f"{schema.source_name}: column {column!r} contains a value that "
                f"cannot be parsed as numeric: {exc}"
            ) from exc
        # Record the pre-cast dtype before assignment.
        previous_dtype: str = str(cast[column].dtype)
        # Assign the double-precision column back into the frame.
        cast[column] = converted.astype("float64")
        # Record the transition when the dtype actually changed.
        if previous_dtype != "float64":
            # Append the transition for the audit notes.
            notes.append(f"{column}: {previous_dtype} -> float64.")
    # Cast every declared integer column to its required width.
    for column in schema.int_columns:
        # Reject a declared integer column that the frame does not carry.
        if column not in cast.columns:
            # Report the absent column so the schema or payload is corrected.
            raise SchemaViolationError(
                f"{schema.source_name}: declared integer column {column!r} "
                f"is absent from the frame."
            )
        # Convert the column to a numeric series under strict handling.
        try:
            # Apply the numeric conversion with strict error handling.
            converted_int: pd.Series = pd.to_numeric(cast[column], errors="raise")
        except (ValueError, TypeError) as exc:
            # Re-raise with the source and column context attached.
            raise SchemaViolationError(
                f"{schema.source_name}: integer column {column!r} contains a "
                f"value that cannot be parsed as numeric: {exc}"
            ) from exc
        # Refuse an integer column that carries a missing value.
        if converted_int.isna().any():
            # A missing integer is a schema violation, never a silent fill.
            raise SchemaViolationError(
                f"{schema.source_name}: integer column {column!r} contains "
                f"{int(converted_int.isna().sum())} missing value(s)."
            )
        # Convert to float solely for finite and integrality validation.
        numeric_values: np.ndarray = converted_int.to_numpy(dtype=float)
        # Identify non-finite values before any narrowing cast.
        finite_mask: np.ndarray = np.isfinite(numeric_values)
        # Identify values with a fractional component.
        integral_mask: np.ndarray = finite_mask & (
            numeric_values == np.trunc(numeric_values)
        )
        # Reject all non-finite or fractional values without truncation.
        invalid_positions: np.ndarray = np.flatnonzero(~integral_mask)
        # Report deterministic offending row identities when validation fails.
        if invalid_positions.size:
            # Resolve the first bounded set of row labels and raw values.
            offenders: List[Tuple[Any, Any]] = [
                (cast.index[position], cast.iloc[position][column])
                for position in invalid_positions[:MAX_REPORTED_OFFENDERS]
            ]
            # Refuse the cast with source, column and row context.
            raise SchemaViolationError(
                f"{schema.source_name}: integer column {column!r} contains "
                f"non-finite or non-integral value(s), for example {offenders}."
            )
        # Record the pre-cast dtype before assignment.
        previous_int_dtype: str = str(cast[column].dtype)
        # Record the target width, defaulting to the platform integer.
        target_width: str = _integer_width_for(schema.source_name, column)
        # Resolve the representable bounds of the declared integer width.
        width_bounds: np.iinfo = np.iinfo(np.dtype(target_width))
        # Identify values that would overflow or wrap during the cast.
        out_of_bounds: np.ndarray = np.flatnonzero(
            (numeric_values < width_bounds.min)
            | (numeric_values > width_bounds.max)
        )
        # Reject narrowing overflow before numpy can silently wrap values.
        if out_of_bounds.size:
            # Capture deterministic offending row identities and values.
            offenders = [
                (cast.index[position], cast.iloc[position][column])
                for position in out_of_bounds[:MAX_REPORTED_OFFENDERS]
            ]
            # Report the exact representable domain.
            raise SchemaViolationError(
                f"{schema.source_name}: integer column {column!r} exceeds "
                f"{target_width} bounds [{width_bounds.min}, "
                f"{width_bounds.max}], for example {offenders}."
            )
        # Binary ingest flags have the exact scientific domain zero or one.
        binary_columns: set[str] = {
            "retry_succeeded",
            "hs_report_confirmed",
            "retained_in_final_panel",
        }
        # Reject any non-binary value in a declared flag column.
        if column in binary_columns and not np.isin(numeric_values, [0.0, 1.0]).all():
            # Select bounded offending row/value pairs for diagnosis.
            flag_positions: np.ndarray = np.flatnonzero(
                ~np.isin(numeric_values, [0.0, 1.0])
            )
            # Resolve their original row labels and values.
            offenders = [
                (cast.index[position], cast.iloc[position][column])
                for position in flag_positions[:MAX_REPORTED_OFFENDERS]
            ]
            # Refuse invalid flag semantics before casting.
            raise SchemaViolationError(
                f"{schema.source_name}: binary flag {column!r} must contain "
                f"only 0 or 1, for example {offenders}."
            )
        # Priority ranks must be strictly positive and fit the declared width.
        if column == "priority_rank" and bool((numeric_values < 1.0).any()):
            # Select bounded invalid rank positions.
            rank_positions: np.ndarray = np.flatnonzero(numeric_values < 1.0)
            # Resolve the original row labels and values.
            offenders = [
                (cast.index[position], cast.iloc[position][column])
                for position in rank_positions[:MAX_REPORTED_OFFENDERS]
            ]
            # Refuse invalid rank semantics before casting.
            raise SchemaViolationError(
                f"{schema.source_name}: priority_rank must be a positive "
                f"integer, for example {offenders}."
            )
        # Assign the integer column back into the frame at its declared width.
        cast[column] = converted_int.astype(target_width)
        # Record the transition when the dtype actually changed.
        if previous_int_dtype != target_width:
            # Append the transition for the audit notes.
            notes.append(f"{column}: {previous_int_dtype} -> {target_width}.")
    # Cast every declared categorical column.
    for column in schema.category_columns:
        # Reject a declared categorical column that the frame does not carry.
        if column not in cast.columns:
            # Report the absent column so the schema or payload is corrected.
            raise SchemaViolationError(
                f"{schema.source_name}: declared categorical column {column!r} "
                f"is absent from the frame."
            )
        # Record the pre-cast dtype before assignment.
        previous_cat_dtype: str = str(cast[column].dtype)
        # Assign the categorical dtype to the column.
        cast[column] = cast[column].astype("category")
        # Record the transition when the dtype actually changed.
        if previous_cat_dtype != "category":
            # Append the transition for the audit notes.
            notes.append(f"{column}: {previous_cat_dtype} -> category.")
    # Return the cast frame alongside its notes.
    return cast, tuple(notes)


# Resolve the declared integer width of one column of one source.
def _integer_width_for(source_name: str, column: str) -> str:
    """
    Return the declared integer width of a column.

    Parameters
    ----------
    source_name : str
        The canonical source name.
    column : str
        The column whose width is requested.

    Returns
    -------
    str
        The numpy-compatible integer dtype name.

    Raises
    ------
    KeyError
        If the source declares no width for the column, which indicates a
        schema that has drifted from the ingest contract.
    """
    # Declare the widths the ingest contract fixes explicitly.
    declared_widths: Dict[Tuple[str, str], str] = {
        ("policy_ledger_raw", "priority_rank"): "int8",
        ("country_metadata_raw", "retained_in_final_panel"): "int8",
        ("comtrade_cell_status_raw", "retry_succeeded"): "int8",
        ("comtrade_cell_status_raw", "hs_report_confirmed"): "int8",
        ("policy_ledger_raw", "entry_year"): "int64",
    }
    # Look the width up, defaulting to the platform integer width.
    return declared_widths.get((source_name, column), "int64")


# Assert the frame's column list equals the declared schema exactly.
def validate_columns(
    frame: pd.DataFrame, schema: SourceSchema, source_name: str
) -> None:
    """
    Assert column-for-column equality between frame and schema.

    Parameters
    ----------
    frame : pandas.DataFrame
        The frame whose columns are validated.
    schema : SourceSchema
        The schema declaring the expected columns.
    source_name : str
        The canonical source name, used in diagnostics.

    Returns
    -------
    None
        The function returns nothing on success.

    Raises
    ------
    SchemaViolationError
        If columns are missing, unexpected, or ordered differently from the
        declared schema.
    """
    # Compute the declared ordered column list for the schema.
    expected: List[Any] = list(expected_columns(schema))
    # Read the observed ordered column list from the frame.
    observed: List[Any] = list(frame.columns)
    # Block C may carry one complete, ordered exporter-evidence extension. An
    # arbitrary subset or an undeclared eleventh source remains invalid.
    if schema.extension_columns and observed == expected + list(schema.extension_columns):
        logger.info(
            "%s: extended source schema validated (%d columns).",
            source_name, len(observed),
        )
        return
    # Compute the columns present in the frame but not declared.
    unexpected: List[Any] = [column for column in observed if column not in expected]
    # Refuse a frame carrying an undeclared column.
    if unexpected:
        # Report the unexpected columns, which a subset check would accept.
        raise SchemaViolationError(
            f"{source_name}: unexpected column(s) "
            f"{[str(column) for column in unexpected[:MAX_REPORTED_OFFENDERS]]}."
        )
    # Compute the declared columns absent from the frame.
    missing: List[Any] = [column for column in expected if column not in observed]
    # Refuse a frame omitting a declared column.
    if missing:
        # Report the missing columns so the retrieval can be corrected.
        raise SchemaViolationError(
            f"{source_name}: missing column(s) "
            f"{[str(column) for column in missing[:MAX_REPORTED_OFFENDERS]]}."
        )
    # Refuse a frame whose column order differs from the declared order.
    if observed != expected:
        # Locate the first position at which the orders diverge.
        first_mismatch: int = next(
            index
            for index, (left, right) in enumerate(zip(observed, expected))
            if left != right
        )
        # Report the mismatch position and both labels for diagnosis.
        raise SchemaViolationError(
            f"{source_name}: column order diverges at position {first_mismatch}: "
            f"observed {str(observed[first_mismatch])!r}, expected "
            f"{str(expected[first_mismatch])!r}."
        )
    # Log the successful schema validation.
    logger.info("%s: schema validated (%d columns).", source_name, len(observed))


# Compute the declared ordered column list of a schema.
def expected_columns(schema: SourceSchema) -> Tuple[Any, ...]:
    """
    Return the ordered columns a schema requires.

    Parameters
    ----------
    schema : SourceSchema
        The schema whose column list is requested.

    Returns
    -------
    tuple
        The base columns followed by the integer year columns when the schema
        expands by year.
    """
    # Return the base columns for a non-year-expanding schema.
    if not schema.expands_years:
        # A frame without year columns exposes only its base columns.
        return schema.base_columns
    # Compose the year columns from the declared inclusive range.
    year_columns: Tuple[int, ...] = tuple(range(YEAR_FIRST, YEAR_LAST + 1))
    # Return the base columns followed by the year columns in ascending order.
    return schema.base_columns + year_columns


# Compute the deterministic content digest of a frame.
def compute_frame_fingerprint(
    frame: pd.DataFrame, key_columns: Sequence[str], source_name: str
) -> str:
    """
    Compute a row-order-independent SHA-256 digest of a frame.

    Parameters
    ----------
    frame : pandas.DataFrame
        The frame to fingerprint.
    key_columns : sequence of str
        The declared primary key used to order the frame deterministically.
    source_name : str
        The canonical source name, used in diagnostics.

    Returns
    -------
    str
        The hexadecimal digest of the canonical serialization.

    Raises
    ------
    SchemaViolationError
        If any declared key column is absent from the frame.
    """
    # Refuse a frame missing any declared key column.
    for column in key_columns:
        # Report the absent key column rather than failing on a later sort.
        if column not in frame.columns:
            # The key column is required for deterministic ordering.
            raise SchemaViolationError(
                f"{source_name}: key column {column!r} is absent from the frame."
            )
    # Sort a copy of the frame by its declared primary key.
    ordered: pd.DataFrame = frame.sort_values(
        list(key_columns), kind="stable"
    ).reset_index(drop=True)
    # Serialize the ordered frame to a canonical comma-separated form.
    serialized: str = ordered.to_csv(
        index=False, na_rep="<NA>", float_format="%.17g", lineterminator="\n"
    )
    # Compute and return the hexadecimal digest of the serialized bytes.
    return hashlib.sha256(serialized.encode("utf-8")).hexdigest()


# Bind metadata release identity separately from the exact-column source frame.
def _region_identity_digest(frame: pd.DataFrame) -> str:
    """Hash the country metadata's declared region source and version attributes."""
    identity = {
        "region_source": frame.attrs.get("region_source"),
        "region_version": frame.attrs.get("region_version"),
    }
    if any(value is not None and not isinstance(value, str) for value in identity.values()):
        raise IngestError("Region source/version attributes must be strings when supplied.")
    serialized = json.dumps(identity, sort_keys=True, separators=(",", ":"))
    return hashlib.sha256(serialized.encode("utf-8")).hexdigest()


# Compute the deterministic digest of the study configuration.
def compute_config_fingerprint(study_config: Mapping[str, Any]) -> str:
    """
    Compute a deterministic SHA-256 digest of the configuration.

    Parameters
    ----------
    study_config : mapping
        The configuration mapping to fingerprint.

    Returns
    -------
    str
        The hexadecimal digest of the key-sorted JSON serialization.
    """
    # Serialize the configuration with sorted keys and a string fallback.
    serialized: str = json.dumps(study_config, sort_keys=True, default=str)
    # Compute and return the hexadecimal digest of the serialized bytes.
    return hashlib.sha256(serialized.encode("utf-8")).hexdigest()


# Canonicalize, cast, validate and fingerprint one source frame.
def ingest_source(
    frame: pd.DataFrame,
    schema: SourceSchema,
    ingested_utc: str,
) -> Tuple[pd.DataFrame, IngestRecord]:
    """
    Ingest one raw source into its canonical form and audit record.

    Parameters
    ----------
    frame : pandas.DataFrame
        The raw frame as delivered by the retrieval layer.
    schema : SourceSchema
        The declared schema of the source.
    ingested_utc : str
        The ISO-8601 UTC timestamp shared by the whole ingest.

    Returns
    -------
    tuple
        The canonicalized frame and its immutable audit record.

    Raises
    ------
    IngestError
        If the frame is empty.
    SchemaViolationError
        If canonicalization, casting or column validation fails.
    PrimaryKeyViolationError
        If the declared primary key does not identify rows uniquely; the
        uniqueness gate is enforced here, BEFORE the fingerprint is computed,
        so no digest can ever describe an invalid frame.
    """
    # Replace declared textual markers with the numpy sentinel.
    canonical, marker_notes = canonicalize_missing_markers(
        frame, schema.source_name
    )
    # Restore integer year labels and verify year coverage.
    with_years, year_notes = normalize_year_columns(canonical, schema)
    # Enforce the declared dtypes across every column.
    cast, dtype_notes = enforce_dtypes(with_years, schema)
    # Validate the exact column list against the declared schema.
    validate_columns(cast, schema, schema.source_name)
    # Enforce the primary-key uniqueness gate before any digest exists; the
    # PK-sorted serialization underlying the fingerprint is a total order only
    # when the key is unique, so this ordering is a fidelity requirement.
    assert_primary_key_unique(cast, schema)
    # Record the ordered validation stages passed prior to fingerprinting.
    validation_stages: Tuple[str, ...] = (
        "canonicalize_missing_markers",
        "normalize_year_columns",
        "enforce_dtypes",
        "validate_columns",
        "assert_primary_key_unique",
    )
    # Compute the deterministic digest of the validated canonicalized frame.
    fingerprint: str = compute_frame_fingerprint(
        cast, schema.key_columns, schema.source_name
    )
    # Record the observed dtype of every column.
    dtype_map: Dict[str, str] = {
        str(column): str(cast[column].dtype) for column in cast.columns
    }
    # Assemble the immutable audit record for the source.
    record: IngestRecord = IngestRecord(
        source_name=schema.source_name,
        n_rows=int(cast.shape[0]),
        n_columns=int(cast.shape[1]),
        key_columns=tuple(schema.key_columns),
        fingerprint_sha256=fingerprint,
        ingested_utc=ingested_utc,
        dtype_map=dtype_map,
        notes=tuple(marker_notes + year_notes + dtype_notes),
        validation_stages=validation_stages,
        region_identity_sha256=(
            _region_identity_digest(cast)
            if schema.source_name == "country_metadata_raw"
            else None
        ),
    )
    # Log the successful ingest with its digest prefix.
    logger.info(
        "%s: ingested rows=%d cols=%d digest=%s",
        schema.source_name,
        record.n_rows,
        record.n_columns,
        fingerprint[:16],
    )
    # Return the canonicalized frame alongside its record.
    return cast, record


# -------------------------------------------------------------------------------------------------------------------------------
# Task 1, Step 2: Verify primary-key uniqueness and referential integrity
# -------------------------------------------------------------------------------------------------------------------------------


# Assert that the declared primary key identifies rows uniquely.
def assert_primary_key_unique(
    frame: pd.DataFrame, schema: SourceSchema
) -> None:
    """
    Assert that the declared key columns contain no duplicate combination.

    Parameters
    ----------
    frame : pandas.DataFrame
        The canonicalized frame whose key is validated.
    schema : SourceSchema
        The schema declaring the key columns.

    Returns
    -------
    None
        The function returns nothing on success.

    Raises
    ------
    PrimaryKeyViolationError
        If any key combination identifies more than one row.
    """
    # Refuse a frame missing any declared key column.
    for column in schema.key_columns:
        # Report the absent key column rather than failing inside the check.
        if column not in frame.columns:
            # The key column is required for the uniqueness check.
            raise PrimaryKeyViolationError(
                f"{schema.source_name}: key column {column!r} is absent."
            )
    # Select the complete composite key without dropping any row.
    key_frame: pd.DataFrame = frame.loc[:, list(schema.key_columns)]
    # Mark null components in every declared key column.
    null_components: pd.DataFrame = key_frame.isna()
    # Mark blank textual components independently of null handling.
    blank_components: pd.DataFrame = key_frame.apply(
        # Convert only the temporary key view so categorical booleans cannot
        # retain a Categorical dtype and break the logical missing-key mask.
        lambda series: series.astype(object).map(
            lambda value: isinstance(value, str) and not value.strip()
        )
    )
    # Identify every row with at least one null or blank key component.
    missing_key_mask: pd.Series = (null_components | blank_components).any(axis=1)
    # Reject missing keys before duplicate detection or fingerprinting.
    if bool(missing_key_mask.any()):
        # Preserve row identity and every composite-key component in examples.
        offenders: List[Tuple[Any, Tuple[Any, ...]]] = [
            (index, tuple(key_frame.loc[index].tolist()))
            for index in key_frame.index[missing_key_mask][
                :MAX_REPORTED_OFFENDERS
            ]
        ]
        # Report a deterministic bounded set of offending rows.
        raise PrimaryKeyViolationError(
            f"{schema.source_name}: primary key {tuple(schema.key_columns)} "
            f"contains {int(missing_key_mask.sum())} row(s) with null or blank "
            f"components, for example {offenders}."
        )
    # Identify every row whose key combination is repeated anywhere.
    duplicated_mask: pd.Series = frame.duplicated(
        subset=list(schema.key_columns), keep=False
    )
    # Compute the number of rows participating in a duplicated key.
    n_offending: int = int(duplicated_mask.sum())
    # Raise when any duplicated key exists, reporting the offending tuples.
    if n_offending:
        # Extract the distinct offending key combinations for the message.
        offenders: List[Tuple[Any, ...]] = list(
            frame.loc[duplicated_mask, list(schema.key_columns)]
            .drop_duplicates()
            .head(MAX_REPORTED_OFFENDERS)
            .itertuples(index=False, name=None)
        )
        # Report the count and the first offending combinations.
        raise PrimaryKeyViolationError(
            f"{schema.source_name}: primary key {tuple(schema.key_columns)} is "
            f"not unique; {n_offending} row(s) participate in duplicated keys, "
            f"for example {offenders}."
        )
    # Log the successful uniqueness check.
    logger.info("%s: primary key is unique.", schema.source_name)


# Assert that every declared code column matches the three-letter pattern.
def assert_iso3_codes(frame: pd.DataFrame, schema: SourceSchema) -> None:
    """
    Assert that every non-missing code matches the ISO-3 pattern.

    Parameters
    ----------
    frame : pandas.DataFrame
        The canonicalized frame whose code columns are validated.
    schema : SourceSchema
        The schema declaring the code columns.

    Returns
    -------
    None
        The function returns nothing on success.

    Raises
    ------
    ReferentialIntegrityError
        If any declared code column is absent or carries a value that does not
        match the three upper-case letter pattern.
    """
    # Validate each declared code column in schema order.
    for column in schema.code_columns:
        # Refuse a frame missing a declared code column.
        if column not in frame.columns:
            # The code column is required for the syntactic check.
            raise ReferentialIntegrityError(
                f"{schema.source_name}: code column {column!r} is absent."
            )
        # Test every value without dropping nulls or stringifying nonstrings.
        valid_mask: pd.Series = frame[column].map(
            lambda value: isinstance(value, str)
            and ISO3_PATTERN.fullmatch(value) is not None
        )
        # Preserve row identity and the original offending value.
        offending: pd.Series = frame.loc[~valid_mask, column]
        # Raise when any value violates the pattern.
        if len(offending):
            # Report the distinct offenders, capped for diagnosability.
            offenders: List[Tuple[Any, Any]] = list(
                offending.iloc[:MAX_REPORTED_OFFENDERS].items()
            )
            # Raise with the source, column and offending values named.
            raise ReferentialIntegrityError(
                f"{schema.source_name}: column {column!r} carries "
                f"{len(offending)} value(s) violating the ISO-3 pattern, "
                f"for example {offenders}."
            )
    # Log the successful syntactic check.
    logger.info("%s: ISO-3 syntactic checks passed.", schema.source_name)


# Verify uniqueness, syntactic validity and cross-frame membership.
def verify_referential_integrity(
    frames: Mapping[str, pd.DataFrame],
    schemas: Mapping[str, SourceSchema],
) -> Dict[str, Dict[str, int]]:
    """
    Run the uniqueness, pattern and membership checks across all sources.

    Parameters
    ----------
    frames : mapping of str to pandas.DataFrame
        The canonicalized frames keyed by source name.
    schemas : mapping of str to SourceSchema
        The declared schemas keyed by source name.

    Returns
    -------
    dict
        A nested report keyed by source name and then by counter name,
        carrying integer counts with explicit row semantics:
        distinct_codes_examined (unique values tested), rows_examined (rows
        actually subjected to the universe test), aggregates_skipped (rows
        whose every non-missing membership value is an aggregate code),
        unknown_codes (distinct offending values) and unknown_rows (rows
        carrying at least one unknown code). The counters satisfy the
        conservation identity rows_examined + aggregates_skipped == len(frame).

    Raises
    ------
    KeyError
        If a frame has no declared schema.
    PrimaryKeyViolationError
        If any declared key is not unique.
    ReferentialIntegrityError
        If any code violates its pattern or its universe.
    """
    # Accumulate the per-source report entries.
    report: Dict[str, Dict[str, int]] = {}
    # Read the metadata universe of country codes from the registry frame.
    metadata_frame: pd.DataFrame = frames["country_metadata_raw"]
    # Extract the universe as a set of strings for membership testing.
    universe: Set[str] = set(
        metadata_frame["country_code"].dropna().astype(str).tolist()
    )
    # Iterate over every declared source in contractual order.
    for source_name in DECLARED_SOURCES:
        # Refuse a frame that has no declared schema.
        if source_name not in schemas:
            # The schema registry must cover every declared source.
            raise KeyError(f"No schema is declared for source {source_name!r}.")
        # Select the frame and its schema for checking.
        frame: pd.DataFrame = frames[source_name]
        # Select the schema for the source.
        schema: SourceSchema = schemas[source_name]
        # Assert the declared primary key is unique.
        assert_primary_key_unique(frame, schema)
        # Assert every code column satisfies the syntactic pattern.
        assert_iso3_codes(frame, schema)
        # Initialise the count of distinct membership values examined.
        examined: int = 0
        # Collect the unknown codes for the diagnostic message.
        unknown_codes: Set[str] = set()
        # Initialise the per-column masks used by the row-level counters.
        nonmissing_masks: List[pd.Series] = []
        aggregate_masks: List[pd.Series] = []
        unknown_masks: List[pd.Series] = []
        # Check membership for every declared membership column.
        for column in schema.membership_columns:
            # Extract the non-missing values as strings for distinct counting.
            values: pd.Series = frame[column].dropna().astype(str)
            # Accumulate the number of DISTINCT values examined.
            examined += int(values.nunique())
            # Build the string-typed view of the column aligned to the frame.
            str_values: pd.Series = frame[column].astype(str)
            # A cell is missing when the raw entry is null.
            missing_mask: pd.Series = frame[column].isna()
            # A cell is aggregate when it is in the declared carve-out set.
            aggregate_mask: pd.Series = (
                str_values.isin(WORLD_BANK_AGGREGATE_CODES) & ~missing_mask
            )
            # A cell is unknown when it is neither aggregate nor in the universe.
            unknown_mask: pd.Series = (
                ~missing_mask & ~aggregate_mask & ~str_values.isin(universe)
            )
            # Accumulate the unknown codes for the diagnostic message.
            unknown_codes.update(str_values[unknown_mask].tolist())
            # Store the masks so the row-level counters can be assembled.
            nonmissing_masks.append(~missing_mask)
            aggregate_masks.append(aggregate_mask)
            unknown_masks.append(unknown_mask)
        # Reduce the unknown masks to one row-level any-across-columns mask.
        unknown_any: Optional[pd.Series] = None
        # OR-accumulate every per-column unknown mask.
        for unknown_mask in unknown_masks:
            # The first mask seeds the reduction; later masks are unioned in.
            unknown_any = (
                unknown_mask
                if unknown_any is None
                else (unknown_any | unknown_mask)
            )
        # Count the rows carrying at least one unknown code.
        unknown_rows: int = 0 if unknown_any is None else int(unknown_any.sum())
        # Report the number of distinct unknown codes.
        unknown: int = len(unknown_codes)
        # Raise when any code is absent from the metadata universe.
        if unknown:
            # Report the source, column count and offending codes.
            raise ReferentialIntegrityError(
                f"{source_name}: {unknown} code(s) are absent from the metadata "
                f"universe, for example "
                f"{sorted(unknown_codes)[:MAX_REPORTED_OFFENDERS]}. Cleansing is a "
                f"later task, but a code unknown to the registry indicates an "
                f"universe mismatch."
            )
        # Reduce the aggregate masks to one row-level all-across-columns mask.
        all_aggregate: Optional[pd.Series] = None
        # Track whether each row carries at least one non-missing value.
        has_nonmissing: Optional[pd.Series] = None
        # Walk the per-column masks in lockstep for the row reductions.
        for aggregate_mask, nonmissing_mask in zip(
            aggregate_masks, nonmissing_masks
        ):
            # A row is covered by this column when its cell is aggregate or
            # missing, so missing cells never block the all-aggregate test.
            covered: pd.Series = aggregate_mask | ~nonmissing_mask
            # AND-accumulate the per-column covered masks.
            all_aggregate = (
                covered
                if all_aggregate is None
                else (all_aggregate & covered)
            )
            # OR-accumulate the per-column non-missing masks.
            has_nonmissing = (
                nonmissing_mask
                if has_nonmissing is None
                else (has_nonmissing | nonmissing_mask)
            )
        # Count rows whose every non-missing membership value is aggregate.
        skipped_aggregates: int = (
            0
            if all_aggregate is None or has_nonmissing is None
            else int((all_aggregate & has_nonmissing).sum())
        )
        # Count the rows actually examined against the universe.
        rows_examined: int = len(frame) - skipped_aggregates
        # Assert the conservation identity over the row partition.
        if rows_examined + skipped_aggregates != len(frame):
            # A divergence would mean a row was counted twice or not at all.
            raise ReferentialIntegrityError(
                f"{source_name}: integrity counters do not reconcile: "
                f"{rows_examined} examined plus {skipped_aggregates} aggregate "
                f"rows does not equal {len(frame)} frame rows."
            )
        # Record the per-source counters in the report.
        report[source_name] = {
            "distinct_codes_examined": examined,
            "rows_examined": rows_examined,
            "aggregates_skipped": skipped_aggregates,
            "unknown_codes": unknown,
            "unknown_rows": unknown_rows,
        }
    # Log the completed integrity sweep.
    logger.info("referential integrity verified across %d sources.", len(report))
    # Return the assembled report.
    return report


# -------------------------------------------------------------------------------------------------------------------------------
# Task 1, Step 3: Apply the row-direction disjunction to Comtrade Block B
# -------------------------------------------------------------------------------------------------------------------------------


# Partition Block B into retained and malformed rows under the disjunction.
def partition_block_b_rows(
    frame: pd.DataFrame, source_name: str
) -> BlockBPartition:
    """
    Partition Block B by the exact row-direction disjunction.

    Parameters
    ----------
    frame : pandas.DataFrame
        The canonicalized Block B frame.
    source_name : str
        The canonical source name, used in diagnostics.

    Returns
    -------
    BlockBPartition
        The retained frame and the malformed ledger with a reason column.

    Raises
    ------
    BlockBPartitionError
        If the disjunction inputs are absent, if a row satisfies both
        disjuncts, or if the conservation identity fails.
    """
    # Refuse a frame missing any column the disjunction requires.
    for column in ("trade_flow", "reporter_code", "partner_code"):
        # Report the absent column rather than failing inside a mask.
        if column not in frame.columns:
            # The disjunction cannot be evaluated without this column.
            raise BlockBPartitionError(
                f"{source_name}: required column {column!r} is absent."
            )
    # Preserve the input exactly while constructing an auditable working copy.
    working: pd.DataFrame = frame.copy(deep=True)
    # Preserve original labels before emitting the canonical flow value.
    working["_trade_flow_original"] = working["trade_flow"]
    # Normalize only genuine text labels; null/nonstring labels remain invalid.
    canonical_flow: pd.Series = working["trade_flow"].map(
        lambda value: value.strip().lower() if isinstance(value, str) else np.nan
    )
    # Persist canonical flow values in both output partitions.
    working["trade_flow"] = canonical_flow
    # Normalize routing keys for predicates without rewriting source evidence.
    reporter: pd.Series = working["reporter_code"].map(
        lambda value: value.strip().upper() if isinstance(value, str) else np.nan
    )
    # Normalize partner keys for predicates without rewriting source evidence.
    partner: pd.Series = working["partner_code"].map(
        lambda value: value.strip().upper() if isinstance(value, str) else np.nan
    )
    # Count the rows whose flow text required canonicalization.
    n_canonicalized: int = int(
        (~working["_trade_flow_original"].map(
            lambda value: isinstance(value, str)
            and value == value.strip().lower()
        )).sum()
    )
    # Log the canonicalization count when any row required it.
    if n_canonicalized:
        # Record the count so the audit trail captures the transformation.
        logger.info(
            "%s: canonicalized trade_flow text on %d row(s).",
            source_name,
            n_canonicalized,
        )
    # Build the import-direction mask exactly as the disjunction specifies.
    import_mask: pd.Series = (
        (canonical_flow == "import")
        & reporter.notna()
        & partner.notna()
        & ~reporter.isin(["RUS", "WLD"])
        & (partner == "RUS")
    )
    # Build the export-direction mirror mask exactly as specified.
    export_mask: pd.Series = (
        (canonical_flow == "export")
        & reporter.notna()
        & partner.notna()
        & (reporter == "RUS")
        & ~partner.isin(["RUS", "WLD"])
    )
    # Compute the rows that would satisfy both disjuncts simultaneously.
    both_mask: pd.Series = import_mask & export_mask
    # Refuse the frame if any row satisfies both disjuncts.
    if bool(both_mask.any()):
        # The disjunction is contradictory for these rows.
        raise BlockBPartitionError(
            f"{source_name}: {int(both_mask.sum())} row(s) satisfy both "
            f"disjuncts, which is impossible under the disjunction."
        )
    # Compute the mask of rows satisfying at least one disjunct.
    valid_mask: pd.Series = import_mask | export_mask
    # Slice the retained rows as an independent copy.
    retained: pd.DataFrame = working.loc[valid_mask].copy()
    # Slice the malformed rows as an independent copy.
    malformed: pd.DataFrame = working.loc[~valid_mask].copy()
    # Build mutually exclusive rejection reasons in priority order.
    reason: pd.Series = pd.Series("invalid_direction", index=working.index)
    # Null reporter or partner keys prevent a directional classification.
    reason.loc[reporter.isna() | partner.isna()] = "null_counterparty"
    # Aggregate WLD rows are never individual-country counterparties.
    reason.loc[(reporter == "WLD") | (partner == "WLD")] = (
        "aggregate_counterparty"
    )
    # Self-flows are forbidden even when one side is RUS.
    reason.loc[reporter.notna() & (reporter == partner)] = "self_flow"
    # Invalid flow labels take precedence when routing keys otherwise exist.
    reason.loc[~canonical_flow.isin(["import", "export"])] = "invalid_flow"
    # Attach exactly one reason to each malformed source row.
    malformed["malformed_reason"] = reason.loc[~valid_mask]
    # Assert that no retained row is self-referential.
    if bool(
        (
            reporter.loc[retained.index] == partner.loc[retained.index]
        ).any()
    ):
        # A retained self-referential row would contradict the disjunction.
        raise BlockBPartitionError(
            f"{source_name}: a retained row has identical reporter and partner."
        )
    # Assert the conservation identity over row counts.
    if len(retained) + len(malformed) != len(frame):
        # A conservation failure means a row was silently lost or duplicated.
        raise BlockBPartitionError(
            f"{source_name}: conservation failed: {len(retained)} retained plus "
            f"{len(malformed)} malformed does not equal {len(frame)}."
        )
    # Verify that every retained row satisfies exactly one disjunct.
    if bool((import_mask.loc[retained.index] & export_mask.loc[retained.index]).any()):
        # A retained row satisfying both disjuncts breaks the partition.
        raise BlockBPartitionError(
            f"{source_name}: a retained row satisfies both disjuncts."
        )
    # Compute the digest of the original pre-partition frame.
    original_digest: str = compute_frame_fingerprint(
        frame, BLOCK_B_FINGERPRINT_KEYS, source_name
    )
    # Compute the digest of the retained partition.
    retained_digest: str = compute_frame_fingerprint(
        retained, BLOCK_B_FINGERPRINT_KEYS, source_name
    )
    # Compute the digest of the malformed ledger, including its reason column.
    malformed_digest: str = compute_frame_fingerprint(
        malformed,
        BLOCK_B_FINGERPRINT_KEYS + ("malformed_reason",),
        source_name,
    )
    # Assemble the ordered (name, digest) pairs for the partition record.
    digests: Tuple[Tuple[str, str], ...] = (
        ("original", original_digest),
        ("retained", retained_digest),
        ("malformed", malformed_digest),
    )
    # Log the partition outcome for the audit trail.
    logger.info(
        "%s: disjunction retained %d row(s) and routed %d malformed row(s).",
        source_name,
        len(retained),
        len(malformed),
    )
    # Return the immutable partition carrying its three artefact digests.
    return BlockBPartition(
        retained=retained, malformed=malformed, digests=digests
    )


# Verify the partition's structural invariants after construction.
def verify_block_b_partition(
    original: pd.DataFrame, partition: BlockBPartition, source_name: str
) -> None:
    """
    Assert the structural invariants of a Block B partition.

    Parameters
    ----------
    original : pandas.DataFrame
        The frame the partition was derived from.
    partition : BlockBPartition
        The partition to verify.
    source_name : str
        The canonical source name, used in diagnostics.

    Returns
    -------
    None
        The function returns nothing on success.

    Raises
    ------
    BlockBPartitionError
        If the conservation identity fails, a retained row is malformed or
        self-referential, or any stored artefact digest diverges from its
        independent recomputation.
    """
    # Independently reconstruct the scientifically valid partition.
    expected: BlockBPartition = partition_block_b_rows(original, source_name)
    # Recompute the conservation identity from the supplied partition.
    conserved: int = len(partition.retained) + len(partition.malformed)
    # Refuse a partition that does not reconstruct the original row count.
    if conserved != len(original):
        # Report both counts so the discrepancy is localized.
        raise BlockBPartitionError(
            f"{source_name}: partition conservation failed: {conserved} rows "
            f"reconstructed against {len(original)} original rows."
        )
    # Assert the malformed ledger carries its reason column.
    if "malformed_reason" not in partition.malformed.columns:
        # Without the reason column the routing decision is unauditable.
        raise BlockBPartitionError(
            f"{source_name}: malformed ledger lacks the reason column."
        )
    # Compare retained identities, values and canonical flow labels exactly.
    try:
        # Exact frame equality detects omissions, duplication and alteration.
        pd.testing.assert_frame_equal(
            partition.retained,
            expected.retained,
            check_like=False,
            check_dtype=True,
            check_names=True,
        )
        # Compare malformed identities, reasons and values exactly as well.
        pd.testing.assert_frame_equal(
            partition.malformed,
            expected.malformed,
            check_like=False,
            check_dtype=True,
            check_names=True,
        )
    except AssertionError as exc:
        # Convert pandas' assertion into the task-specific integrity error.
        raise BlockBPartitionError(
            f"{source_name}: supplied partition does not exactly match the "
            f"independently reconstructed row identities and values: {exc}"
        ) from exc
    # Reconstruct the expected (name, digest) pairs from the live frames.
    expected_digests: Tuple[Tuple[str, str], ...] = expected.digests
    # Refuse a partition constructed without its complete digest ledger.
    if len(partition.digests) != len(expected_digests):
        # The frozen-artefact discipline requires all three digests.
        raise BlockBPartitionError(
            f"{source_name}: partition carries {len(partition.digests)} digest "
            f"entries; {len(expected_digests)} are required."
        )
    # Compare every stored digest against its independent recomputation.
    for (expected_name, expected_digest), (stored_name, stored_digest) in zip(
        expected_digests, partition.digests
    ):
        # The stored ledger must cover the same artefacts with equal digests.
        if expected_name != stored_name or expected_digest != stored_digest:
            # A divergence means an artefact mutated between construction and
            # verification, so the conservation identity is unverifiable.
            raise BlockBPartitionError(
                f"{source_name}: digest mismatch for the {stored_name!r} "
                f"artefact: stored {stored_digest[:16]}... against recomputed "
                f"{expected_digest[:16]}...."
            )
    # Log the verified partition with its digest prefixes.
    logger.info(
        "%s: partition verified by conservation and digests %s.",
        source_name,
        [digest[:8] for _, digest in partition.digests],
    )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 1, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the ingest of every raw structure into a single audited bundle.
def orchestrate_raw_data_ingest(
    sources: Mapping[str, pd.DataFrame],
    study_config: Mapping[str, Any],
) -> IngestBundle:
    """
    Ingest, validate and fingerprint all raw data structures.

    Parameters
    ----------
    sources : mapping of str to pandas.DataFrame
        The ten raw frames keyed by their canonical source names.
    study_config : mapping
        The study configuration, which must expose all nineteen sections.

    Returns
    -------
    IngestBundle
        The frozen bundle carrying the canonicalized frames, the per-source
        audit records, the configuration fingerprint, the ingest timestamp and
        the malformed Block B ledger.

    Raises
    ------
    TypeError
        If the sources are not a mapping of DataFrames.
    IngestError
        If any declared source is absent or empty.
    SchemaViolationError
        If a frame violates its declared schema.
    PrimaryKeyViolationError
        If a declared key is not unique.
    ReferentialIntegrityError
        If a code violates its pattern or its universe.
    BlockBPartitionError
        If the Block B disjunction cannot be formed consistently.
    """
    # Restore any declared tuples collapsed by a JSON round trip.
    restored_config, n_restored = restore_config_tuples(study_config)
    # Validate that the configuration exposes all nineteen declared sections.
    validate_study_config_sections(restored_config)
    # Refuse a sources argument that is not a mapping.
    if not isinstance(sources, Mapping):
        # A non-mapping input cannot be addressed by source name.
        raise TypeError(
            f"sources must be a mapping keyed by source name; received "
            f"{type(sources).__name__}."
        )
    # Compute the source names absent from the supplied mapping.
    absent: List[str] = [name for name in DECLARED_SOURCES if name not in sources]
    # Refuse a mapping that omits any declared source.
    if absent:
        # Report every absent source so the caller repairs the mapping once.
        raise IngestError(f"Missing declared source(s): {absent}.")
    # Log any supplied source that the contract does not declare.
    for extra in sorted(set(sources) - set(DECLARED_SOURCES)):
        # Warn rather than refuse, since an extra frame cannot corrupt the ten.
        logger.warning("Ignoring undeclared source %r.", extra)
    # Compose the ingest timestamp shared by every source record.
    ingested_utc: str = current_utc_iso()
    # Verify the complete configured schemas against the operative constructor.
    validate_configuration_bindings(restored_config)
    # Build the declared schemas for the ten sources.
    schemas: Dict[str, SourceSchema] = build_source_schemas()
    # Initialise the accumulator of canonicalized frames.
    canonical_frames: Dict[str, pd.DataFrame] = {}
    # Initialise the accumulator of per-source audit records.
    records: Dict[str, IngestRecord] = {}
    # Ingest each declared source in its contractual order.
    for source_name in DECLARED_SOURCES:
        # Select the schema of the source.
        schema: SourceSchema = schemas[source_name]
        # Ingest the source into its canonical frame and record.
        canonical, record = ingest_source(
            sources[source_name], schema, ingested_utc
        )
        # Store the canonicalized frame under its source name.
        canonical_frames[source_name] = canonical
        # Store the audit record under the same source name.
        records[source_name] = record
    # Verify uniqueness, patterns and membership across every source.
    integrity_report: Dict[str, Dict[str, int]] = verify_referential_integrity(
        canonical_frames, schemas
    )
    # Preserve the canonical source object before direction partitioning.
    block_b_original: pd.DataFrame = canonical_frames[
        "comtrade_block_b_raw"
    ].copy(deep=True)
    # Partition the Block B frame under the row-direction disjunction.
    partition: BlockBPartition = partition_block_b_rows(
        block_b_original, "comtrade_block_b_raw"
    )
    # Verify the structural invariants of the partition.
    verify_block_b_partition(
        block_b_original,
        partition,
        "comtrade_block_b_raw",
    )
    # Replace the Block B entry with the retained partition of the frame.
    canonical_frames["comtrade_block_b_raw"] = partition.retained
    # Select the source record so both source and retained meanings are explicit.
    block_b_source_record: IngestRecord = records["comtrade_block_b_raw"]
    # Compose the note preserving all three labelled partition fingerprints.
    post_partition_note: str = (
        f"post-partition: retained {len(partition.retained)} row(s) and routed "
        f"{len(partition.malformed)} malformed row(s) to the audit ledger; the "
        f"source digest is {dict(partition.digests)['original']}, the retained "
        f"digest is {dict(partition.digests)['retained']}, and the malformed "
        f"digest is {dict(partition.digests)['malformed']}."
    )
    # Associate the public Block B record with the retained public frame.
    records["comtrade_block_b_raw"] = replace(
        block_b_source_record,
        n_rows=len(partition.retained),
        n_columns=partition.retained.shape[1],
        fingerprint_sha256=dict(partition.digests)["retained"],
        dtype_map={
            str(column): str(dtype)
            for column, dtype in partition.retained.dtypes.items()
        },
        notes=block_b_source_record.notes + (post_partition_note,),
    )
    # Compute the deterministic digest of the restored configuration.
    config_fingerprint: str = compute_config_fingerprint(restored_config)
    # Assemble and return the frozen ingest bundle.
    bundle: IngestBundle = IngestBundle(
        frames=canonical_frames,
        records=records,
        study_config=restored_config,
        config_fingerprint=config_fingerprint,
        ingested_utc=ingested_utc,
        block_b_malformed=partition.malformed,
        block_b_original=block_b_original,
        block_b_partition_digests=partition.digests,
    )
    # Log the bundle summary including the integrity and restoration counters.
    logger.info(
        "ingest complete: sources=%d aggregates_skipped=%d tuples_restored=%d "
        "config_digest=%s",
        len(canonical_frames),
        sum(entry["aggregates_skipped"] for entry in integrity_report.values()),
        n_restored,
        config_fingerprint[:16],
    )
    # Return the assembled bundle.
    return bundle


# Build the declared schema of every raw source.
def build_source_schemas() -> Dict[str, SourceSchema]:
    """
    Return the declared schema of each of the ten raw sources.

    Returns
    -------
    dict of str to SourceSchema
        The schema registry keyed by canonical source name.

    Raises
    ------
    None
        The function is a pure constructor over literal declarations.
    """
    # Declare the schemas with their keys, dtypes, code columns and membership.
    schemas: Tuple[SourceSchema, ...] = (
        SourceSchema(
            source_name="wdi_raw_wide",
            base_columns=(
                "country_name",
                "country_code",
                "indicator_name",
                "indicator_code",
            ),
            key_columns=("country_code", "indicator_code"),
            float_columns=tuple(str(year) for year in range(YEAR_FIRST, YEAR_LAST + 1)),
            int_columns=(),
            category_columns=(),
            code_columns=("country_code",),
            membership_columns=("country_code",),
            expands_years=True,
        ),
        SourceSchema(
            source_name="wdi_raw_long",
            base_columns=(
                "country_code",
                "country_name",
                "indicator_code",
                "indicator_name",
                "year",
                "value",
            ),
            key_columns=("country_code", "indicator_code", "year"),
            float_columns=("value",),
            int_columns=("year",),
            category_columns=(),
            code_columns=("country_code",),
            membership_columns=("country_code",),
        ),
        SourceSchema(
            source_name="wgi_raw_long",
            base_columns=(
                "country_code",
                "country_name",
                "year",
                "rule_of_law",
                "source_release",
            ),
            key_columns=("country_code", "year", "source_release"),
            float_columns=("rule_of_law",),
            int_columns=("year",),
            category_columns=(),
            code_columns=("country_code",),
            membership_columns=("country_code",),
        ),
        SourceSchema(
            source_name="comtrade_block_a_raw",
            base_columns=(
                "reporter_code",
                "reporter_name",
                "partner_code",
                "partner_name",
                "year",
                "product_description",
                "product_code",
                "trade_flow",
                "value_usd",
            ),
            key_columns=("reporter_code", "year", "product_description"),
            float_columns=("value_usd",),
            int_columns=("year",),
            category_columns=("product_description", "trade_flow"),
            code_columns=("reporter_code", "partner_code"),
            membership_columns=("reporter_code", "partner_code"),
        ),
        SourceSchema(
            source_name="comtrade_block_b_raw",
            base_columns=(
                "reporter_code",
                "reporter_name",
                "partner_code",
                "partner_name",
                "year",
                "product_description",
                "trade_flow",
                "value_usd",
                "mirror_availability",
            ),
            key_columns=(
                "reporter_code",
                "partner_code",
                "year",
                "product_description",
                "trade_flow",
            ),
            float_columns=("value_usd",),
            int_columns=("year",),
            category_columns=(
                "product_description",
                "trade_flow",
                "mirror_availability",
            ),
            code_columns=("reporter_code", "partner_code"),
            membership_columns=("reporter_code",),
        ),
        SourceSchema(
            source_name="comtrade_block_c_raw",
            base_columns=(
                "reporter_code",
                "reporter_name",
                "partner_code",
                "partner_name",
                "year",
                "product_description",
                "trade_flow",
                "value_usd",
                "recovered_via",
            ),
            key_columns=("reporter_code", "year"),
            float_columns=("value_usd",),
            int_columns=("year",),
            category_columns=("product_description", "trade_flow", "recovered_via"),
            code_columns=("reporter_code",),
            membership_columns=("reporter_code",),
            extension_columns=(
                "exporter_sum_components_json",
                "exporter_sum_query_json",
                "exporter_sum_source_id",
                "exporter_sum_components_sha256",
            ),
        ),
        SourceSchema(
            source_name="comtrade_cell_status_raw",
            base_columns=(
                "reporter_code",
                "year",
                "series",
                "request_type",
                "retry_succeeded",
                "hs_report_confirmed",
                "value_usd",
                "final_status",
                "downstream_disposition",
            ),
            key_columns=("reporter_code", "year", "series"),
            float_columns=("value_usd",),
            int_columns=("year", "retry_succeeded", "hs_report_confirmed"),
            category_columns=(
                "series",
                "request_type",
                "final_status",
                "downstream_disposition",
            ),
            code_columns=("reporter_code",),
            membership_columns=("reporter_code",),
        ),
        SourceSchema(
            source_name="nominal_gdp_raw",
            base_columns=(
                "country_code",
                "country_name",
                "year",
                "indicator_code",
                "value_usd",
                "ingest_validation",
            ),
            key_columns=("country_code", "year"),
            float_columns=("value_usd",),
            int_columns=("year",),
            category_columns=("ingest_validation",),
            code_columns=("country_code",),
            membership_columns=("country_code",),
        ),
        # Documentary identity is document_title; no source_document alias is
        # implemented. Optional adjudication must accompany conflicting evidence.
        SourceSchema(
            source_name="policy_ledger_raw",
            base_columns=(
                "country_code",
                "policy",
                "issuing_authority",
                "document_title",
                "publication_date",
                "archived_source",
                "policy_type",
                "entry_year",
                "coding_decision",
                "priority_rank",
            ),
            key_columns=("country_code", "policy", "document_title"),
            extension_columns=(
                "adjudication_selected", "adjudicated_entry_year", "adjudication_note",
            ),
            float_columns=(),
            int_columns=("entry_year", "priority_rank"),
            category_columns=(
                "policy",
                "archived_source",
                "policy_type",
                "coding_decision",
            ),
            code_columns=("country_code",),
            membership_columns=("country_code",),
        ),
        SourceSchema(
            source_name="country_metadata_raw",
            base_columns=(
                "country_code",
                "country_name",
                "world_bank_region",
                "geopolitical_role",
                "retained_in_final_panel",
            ),
            key_columns=("country_code",),
            float_columns=(),
            int_columns=("retained_in_final_panel",),
            category_columns=("world_bank_region", "geopolitical_role"),
            code_columns=("country_code",),
            membership_columns=(),
        ),
    )
    # Index the schemas by source name and return the registry.
    return {schema.source_name: schema for schema in schemas}
````

````python
# Task 2: Validate study_config Integrity and Completeness

# ==============================================================================
# Task 2: Validate study_config Integrity and Completeness
# ==============================================================================

# Resolve the project root so the Task 1 module can be imported by name.
_PROJECT_ROOT: Final[Path] = Path.cwd()


# Declare the twenty-eight algebraic equation keys the section must expose.
REQUIRED_EQUATION_KEYS: Final[Tuple[str, ...]] = (
    "eq1_treatment_donor_sets",
    "eq2_reporting_windows",
    "eq3_untreated_outcome_model",
    "eq4_donor_factor_proxy",
    "eq5_pre_treatment_OLS",
    "eq6_pooled_beta_estimator",
    "eq7_unit_specific_loadings",
    "eq8_covariate_loadings",
    "eq9_counterfactual_covariates",
    "eq10_counterfactual_outcome",
    "eq11_country_year_gap",
    "eq12_window_ATT",
    "eq13_late_wave1_baseline",
    "eq14_stable_gap_continuation",
    "eq15_incremental_wave2_ATT",
    "eq16_country_dispersion_SE",
    "eq17_exposure_measure",
    "eq18_exposure_heterogeneity",
    "eqA1_gas_share",
    "eqA2_energy_gdp",
    "eqB1_donor_mean_proxy",
    "eqB3_residual_maker",
    "eqB4_pooled_beta_pseudoinverse",
    "eqB5_covariate_loadings_pseudoinverse",
    "eqB6_gamma_loadings_pseudoinverse",
    "eqB7_counterfactual_covariates",
    "eqB8_counterfactual_outcome",
    "eq_unnumbered_residual_maker",
    "eq_unnumbered_log_percentage",
)
# Declare the two registry tags carrying non-algebraic semantics.
REGISTRY_TAG_KEYS: Final[Tuple[str, ...]] = (
    "eqB2_skipped_in_source",
    "eqB10_to_B13_reconstruction",
)
# Declare the minimum parameter set each algebraic equation must expose.
EQUATION_REQUIRED_FIELDS: Final[Mapping[str, Tuple[str, ...]]] = {
    "eq1_treatment_donor_sets": (
        "ever_treated_rule", "never_treated_rule",
        "target_subset_rule", "donor_subset_rule",
    ),
    "eq2_reporting_windows": (
        "Wave1", "Wave1_length", "Wave2", "Wave2_length",
    ),
    "eq3_untreated_outcome_model": (
        "K_controls", "control_lag", "loading_vector_dimension",
        "loading_components", "factor_dimension",
    ),
    "eq4_donor_factor_proxy": (
        "donor_set", "component_1", "component_2",
        "treated_realizations_included", "parsimonious_specification",
    ),
    "eq5_pre_treatment_OLS": (
        "pre_treatment_end_year", "pre_treatment_start_year", "T0",
        "estimator", "residual_maker_applied",
    ),
    "eq6_pooled_beta_estimator": (
        "beta_dimension", "M_F_definition", "residualized_on",
        "inversion_baseline", "stabilization",
    ),
    "eq7_unit_specific_loadings": (
        "dimension", "estimated_for", "estimation_source",
        "why_donors_included",
    ),
    "eq8_covariate_loadings": (
        "Lambda_dimension", "estimated_for", "estimation_source",
        "why_only_targets",
    ),
    "eq9_counterfactual_covariates": (
        "post_treatment_start_year", "covariates_simulated",
        "observed_covariates_used_post_treatment",
    ),
    "eq10_counterfactual_outcome": (
        "post_treatment_start_year", "no_target_realized_data_post_treatment",
    ),
    "eq11_country_year_gap": (
        "sign_convention", "positive_interpretation", "negative_interpretation",
    ),
    "eq12_window_ATT": (
        "country_weighting", "year_weighting",
        "N_T_sanctions", "N_T_direct_arming",
    ),
    "eq13_late_wave1_baseline": (
        "baseline_window", "baseline_length", "aggregation",
    ),
    "eq14_stable_gap_continuation": (
        "counterfactual_shift", "maintained_assumption",
    ),
    "eq15_incremental_wave2_ATT": (
        "preferred_estimand", "equivalent_forms",
        "identity_check_required", "identity_tolerance",
    ),
    "eq16_country_dispersion_SE": (
        "se_formula", "df", "test", "stars_source",
        "bootstrap_and_placebo_role",
    ),
    "eq17_exposure_measure": (
        "gas_types", "aggregation_window_length", "windows",
        "numerator_currency", "denominator_series", "standardization",
    ),
    "eq18_exposure_heterogeneity": (
        "Post_period", "omitted_period", "controls_included",
        "country_intercepts", "country_specific_loadings_on_Ybar_Ct",
        "standard_error", "sum_se_computation", "coefficient_semantics",
        "regression_samples",
    ),
    "eqA1_gas_share": (
        "numerator", "denominator", "recomputed_from_components",
        "interpolation_of_ratio",
    ),
    "eqA2_energy_gdp": (
        "numerator", "denominator", "recomputed_from_components",
    ),
    "eqB1_donor_mean_proxy": ("formula",),
    "eqB3_residual_maker": ("formula", "pseudoinverse"),
    "eqB4_pooled_beta_pseudoinverse": ("formula",),
    "eqB5_covariate_loadings_pseudoinverse": ("formula",),
    "eqB6_gamma_loadings_pseudoinverse": ("formula",),
    "eqB7_counterfactual_covariates": ("formula",),
    "eqB8_counterfactual_outcome": ("formula",),
    "eq_unnumbered_residual_maker": ("formula", "location", "used_by"),
    "eq_unnumbered_log_percentage": (
        "exact", "approximation", "location", "applies_to",
    ),
}
# Declare the minimum parameter set each registry tag must expose.
REGISTRY_TAG_REQUIRED_FIELDS: Final[Mapping[str, Tuple[str, ...]]] = {
    "eqB2_skipped_in_source": ("status",),
    "eqB10_to_B13_reconstruction": (
        "status", "reconstruction", "equivalent_to",
    ),
}
# Declare the twenty-six algorithm keys the section must expose.
REQUIRED_ALGORITHM_KEYS: Final[Tuple[str, ...]] = (
    "A1_two_wave_CCEDID_imputation",
    "A2_incremental_wave2",
    "B1_country_history_bootstrap",
    "B2_region_block_bootstrap",
    "B3_size_matched_placebo",
    "C1_eligibility_screening",
    "C2_control_completion",
    "C3_outcome_repair",
    "G0_treatment_coding_resolution",
    "D1_comtrade_query",
    "D2_mirror_flow_recovery",
    "D3_gas_share_denominator",
    "D4_within_window_completion",
    "D5_exposure_sample_rules",
    "E1_high_income_design",
    "E2_mahalanobis_matching",
    "E3_standardized_difference_balance",
    "F1_treated_LOO",
    "F1b_distributional_diagnostics",
    "F2_donor_LOO",
    "F3_donor_pool_variants",
    "F4_factor_proxy_variants",
    "F5_reference_window_sweep",
    "G1_exposure_heterogeneity_matrix",
    "H1_exact_computational_form",
    "I1_pointwise_ribbon_figure2",
)
# Declare the minimum parameter set each algorithm must expose.
ALGORITHM_REQUIRED_FIELDS: Final[Mapping[str, Tuple[str, ...]]] = {
    "A1_two_wave_CCEDID_imputation": (
        "n_steps", "step_names", "iterated_over", "per_run_refitted_objects",
    ),
    "A2_incremental_wave2": (
        "baseline_window", "Wave2_window", "stable_gap_assumption", "output",
        "dual_form_equivalence_check_required",
    ),
    "B1_country_history_bootstrap": (
        "replicates_B", "resampling_unit", "re_estimate",
        "percentile_bounds", "invalid_resample_policy",
    ),
    "B2_region_block_bootstrap": (
        "replicates_B", "resampling_unit", "region_vocabulary_key",
        "re_estimate", "percentile_bounds",
    ),
    "B3_size_matched_placebo": (
        "replicates_M", "pseudo_treated_draw_from", "size_match_by_policy",
        "plus_one_correction_formula", "not_a_confidence_interval",
        "not_a_star_source",
    ),
    "C1_eligibility_screening": (
        "outcome_missing_max", "control_high_missing_max",
        "control_high_missing_count_max", "control_single_missing_max",
        "critical_window", "critical_outcome_missing_max",
        "expected_otherwise_eligible", "expected_final",
        "documented_exclusions",
    ),
    "C2_control_completion": (
        "priority_1", "priority_2", "priority_3", "priority_3_max_years",
        "priority_4", "priority_4_anchor_count", "extrapolation_bound_low",
        "extrapolation_bound_high", "non_negativity_floors",
        "cross_country_information_used", "temporal_universe_completion",
        "temporal_universe_output",
    ),
    "C3_outcome_repair": (
        "leading_backcast_count", "leading_backcast_window", "backcast_method",
        "internal_outcome_interpolation", "carry_forward_cells",
        "cross_country_information_used",
    ),
    "G0_treatment_coding_resolution": (
        "group_by", "sort_by", "selection_rule", "secondary_material_rule",
        "conflict_rule", "output", "ukraine_handling",
    ),
    "D1_comtrade_query": (
        "trade_universe_countries", "potential_cells", "windows", "blocks",
        "verified_zero_rule",
    ),
    "D2_mirror_flow_recovery": (
        "enabled_for", "importer_reported_precedence", "exporter_mirror_side",
        "exporter_has_pipeline_gas", "unresolved_pipeline_countries",
    ),
    "D3_gas_share_denominator": (
        "default_denominator", "substitute_with", "substitute_when",
        "retain_condition",
    ),
    "D4_within_window_completion": (
        "interpolation_space", "internal_gap_max_consecutive",
        "edge_extrapolation_min_anchors", "edge_growth_statistic",
        "edge_growth_bounds", "non_negativity_constraint", "cross_window_gap",
        "cross_country_borrowing", "ratio_recomputed_from_components",
        "primitive_imputations", "never_impute_russian_gas_for",
    ),
    "D5_exposure_sample_rules": (
        "strict", "extended", "excluded_under_both",
        "intersection_with_universe", "regression_counts", "standardization",
    ),
    "E1_high_income_design": (
        "restriction_scalar", "scalar_computed_over",
        "applied_to_targets_and_donors", "expected_targets", "expected_donors",
        "max_abs_std_diff_reported",
    ),
    "E2_mahalanobis_matching": (
        "M_neighbors", "distance_metric", "covariate_source", "weighting",
        "expected_targets", "expected_unique_donors",
        "max_abs_std_diff_reported",
    ),
    "E3_standardized_difference_balance": (
        "formula", "covariate_window", "reporting_metric", "target_baseline",
        "target_high_income", "target_mahalanobis",
    ),
    "F1_treated_LOO": (
        "deletes", "recomputes", "reports",
        "range_CPI_reported", "range_GDPpc_reported",
    ),
    "F1b_distributional_diagnostics": (
        "inputs", "statistics", "targets_CPI", "targets_GDPpc",
    ),
    "F2_donor_LOO": (
        "deletes", "rebuilds", "re_estimates", "reports",
        "max_change_CPI_reported", "max_change_GDPpc_reported",
    ),
    "F3_donor_pool_variants": ("variants", "reported_ranges"),
    "F4_factor_proxy_variants": (
        "variants", "dimensions_by_variant",
        "condition_numbers_sanctions_CPI", "reported_ranges",
        "stability_flagging_rule",
    ),
    "F5_reference_window_sweep": (
        "reference_windows", "Wave2_held_fixed", "reported_CPI_values",
        "reported_GDPpc_values",
    ),
    "G1_exposure_heterogeneity_matrix": (
        "specifications", "strict_arming_sanctioner_count", "gas_share_variable",
    ),
    "H1_exact_computational_form": (
        "pseudo_inverse_operator", "used_for", "uses_all_eight_controls",
        "uses_two_factor_components",
    ),
    "I1_pointwise_ribbon_figure2": (
        "per_year_construction", "CI_level", "normalization_year",
        "normalization_rule",
    ),
}
# Declare the seven reproducibility-gap keys the registry must expose.
REQUIRED_GAP_KEYS: Final[Tuple[str, ...]] = (
    "wdi_vintage",
    "comtrade_hs_codes",
    "random_seeds",
    "equations_B10_to_B13",
    "WGI_release_year",
    "policy_source_archive_paths",
    "region_block_definition",
)
# Declare the five country-roster keys the registry must expose.
REQUIRED_ROSTER_KEYS: Final[Tuple[str, ...]] = (
    "sanctions_target_42",
    "direct_arming_target_29",
    "sanctions_donors_84",
    "direct_arming_donors_CPI_96",
    "direct_arming_donors_GDPpc_97",
)
# Map each roster key to the global constant that fixes its declared count.
ROSTER_COUNT_CONSTANTS: Final[Mapping[str, str]] = {
    "sanctions_target_42": "n_sanctions_targets",
    "direct_arming_target_29": "n_direct_arming_targets",
    "sanctions_donors_84": "n_sanctions_donors_CPI",
    "direct_arming_donors_CPI_96": "n_arming_donors_CPI",
    "direct_arming_donors_GDPpc_97": "n_arming_donors_GDPpc",
}
# Declare the exact geopolitical exclusion membership the registry must state.
EXPECTED_EXCLUSIONS: Final[Mapping[str, Tuple[str, ...]]] = {
    "target_and_donor_ban": ("RUS",),
    "donor_ban": ("BLR",),
    "target_exclusion_from_ATT": ("UKR",),
    "critical_window_exclusion": ("SDN", "SYR"),
}
# Declare the nine temporal layers with their exact windows.
EXPECTED_TEMPORAL_LAYERS: Final[Mapping[str, Any]] = {
    "raw_retrieval": (1996, 2024),
    "eligibility_screening": (2000, 2024),
    "completion_anchoring": (1996, 2024),
    "completion_output": (2000, 2024),
    "balanced_panel": (2000, 2024),
    "effective_estimation_after_lag": (2001, 2024),
    "pre_treatment_fit": (2001, 2013),
    "post_treatment_projection": (2014, 2024),
    "exposure_windows": [(2010, 2013), (2018, 2021)],
}
# Declare the four outcome entries with their exact WDI series codes.
EXPECTED_OUTCOME_CODES: Final[Mapping[str, str]] = {
    "CPI": "FP.CPI.TOTL",
    "RealGDPpc": "NY.GDP.PCAP.KD",
    "InflationRate": "FP.CPI.TOTL.ZG",
    "RealGDPpcGrowth": "NY.GDP.PCAP.KD.ZG",
}
# Declare the eight control entries with their exact WDI series codes.
EXPECTED_CONTROL_CODES: Final[Mapping[str, str]] = {
    "OilRents": "NY.GDP.PETR.RT.ZS",
    "TradeOpenness": "NE.TRD.GNFS.ZS",
    "TermsOfTrade": "TT.PRI.MRCH.XD.WD",
    "InvestmentRate": "NE.GDI.TOTL.ZS",
    "GovConsumption": "NE.CON.GOVT.ZS",
    "PopulationGrowth": "SP.POP.GROW",
    "FuelImportShare": "TM.VAL.FUEL.ZS.UN",
    "FuelExportShare": "TX.VAL.FUEL.ZS.UN",
}
# Declare the two fuel-share entries that must be separate primitives.
FUEL_PRIMITIVE_NAMES: Final[Tuple[str, ...]] = (
    "FuelImportShare",
    "FuelExportShare",
)
# Fix the relative tolerance used for every floating-point comparison.
FLOAT_RELATIVE_TOLERANCE: Final[float] = 1e-9

# Declare the failure raised for every structural validation violation.
class ConfigValidationError(Exception):
    """
    Raised when the study configuration violates its declared contract.

    Attributes
    ----------
    args : tuple
        Inherited arguments naming the missing key or mismatch.

    Notes
    -----
    Covers equation registry, algorithm registry, series codes,
    rosters, gaps, and temporal universes across Sections 1-19.

    Examples
    --------
    >>> try:
    ...     raise ConfigValidationError("missing section")
    ... except ConfigValidationError as exc:
    ...     print(str(exc))
    missing section
    """

# Declare the immutable report returned by a successful validation.
@dataclass(frozen=True)
class ConfigValidationReport:
    """
    Immutable record of a complete configuration validation.

    Attributes
    ----------
    equation_matrix : pandas.DataFrame
        Boolean validity matrix indexed by equation key.
    algorithm_matrix : pandas.DataFrame
        Boolean validity matrix indexed by algorithm key.
    target_matrix : pandas.DataFrame
        Target audit table indexed by dotted configuration path.
    consistency_matrix : pandas.DataFrame
        Internal-identity table indexed by identity name.
    series_matrix : pandas.DataFrame
        Series-registry validity matrix indexed by check name.
    country_matrix : pandas.DataFrame
        Country-registry validity matrix indexed by check name.
    gap_matrix : pandas.DataFrame
        Reproducibility-gap matrix indexed by gap key.
    temporal_matrix : pandas.DataFrame
        Temporal-layer matrix indexed by layer name.
    config_fingerprint : str
        The hexadecimal digest of the validated configuration.
    validated_utc : str
        The ISO-8601 UTC timestamp of the validation.
    n_rows_validated : int
        The total number of validated rows across the eight catalog matrices
        and the operative binding matrix.
    readiness_token : str
        Digest-bound token proving every production-critical gap was resolved.
    """
    # Boolean validity matrix indexed by equation key.
    equation_matrix: pd.DataFrame
    # Boolean validity matrix indexed by algorithm key.
    algorithm_matrix: pd.DataFrame
    # Target audit table indexed by dotted configuration path.
    target_matrix: pd.DataFrame
    # Internal-identity table indexed by identity name.
    consistency_matrix: pd.DataFrame
    # Series-registry validity matrix indexed by check name.
    series_matrix: pd.DataFrame
    # Country-registry validity matrix indexed by check name.
    country_matrix: pd.DataFrame
    # Reproducibility-gap matrix indexed by gap key.
    gap_matrix: pd.DataFrame
    # Temporal-layer matrix indexed by layer name.
    temporal_matrix: pd.DataFrame
    # The hexadecimal digest of the validated configuration.
    config_fingerprint: str
    # The ISO-8601 UTC timestamp of the validation.
    validated_utc: str
    # The total number of validated rows across all matrices.
    n_rows_validated: int
    # Digest-bound token required by production-stage adapters.
    readiness_token: str
    # Verified operative schema and immutable configuration bindings.
    binding_matrix: pd.DataFrame = field(default_factory=pd.DataFrame)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Step 1: Verify all declared equations are present and structurally complete
# -------------------------------------------------------------------------------------------------------------------------------


# Return the declared algebraic equation keys in their contractual order.
def required_equation_keys() -> Tuple[str, ...]:
    """
    Return the twenty-eight algebraic equation keys.

    Returns
    -------
    tuple of str
        The declared equation keys, in the order fixed by Section 4.
    """
    # Return the module constant so the declaration has one definition.
    return REQUIRED_EQUATION_KEYS


# Return the declared minimum parameter set of every equation entry.
def required_equation_fields() -> Mapping[str, Tuple[str, ...]]:
    """
    Return the minimum parameter set required of each equation entry.

    Returns
    -------
    mapping of str to tuple of str
        The required field names keyed by equation key.
    """
    # Return the module constant so the declaration has one definition.
    return EQUATION_REQUIRED_FIELDS


# Return the declared registry-tag keys and their required fields.
def required_registry_tag_fields() -> Mapping[str, Tuple[str, ...]]:
    """
    Return the two registry-tag keys with their required fields.

    Returns
    -------
    mapping of str to tuple of str
        The required field names keyed by registry-tag key.
    """
    # Return the module constant so the declaration has one definition.
    return REGISTRY_TAG_REQUIRED_FIELDS


# Validate the equation registry and return its boolean validity matrix.
def validate_equation_registry(
    study_config: Mapping[str, Any],
) -> pd.DataFrame:
    """
    Validate that every declared equation is present and complete.

    Parameters
    ----------
    study_config : mapping
        The study configuration mapping.

    Returns
    -------
    pandas.DataFrame
        A boolean matrix indexed by equation key with the columns present,
        is_mapping, non_empty and has_required_fields.

    Raises
    ------
    KeyError
        If the equation section is absent or if any declared key is missing.
    ConfigValidationError
        If a declared key is not a non-empty mapping, or if it omits any of
        its required fields.
    """
    # Read the equation section, requiring it to be a mapping.
    section = study_config.get("equation_parameters")
    # Refuse a configuration whose equation section is absent or ill-typed.
    if not isinstance(section, Mapping):
        # A missing section halts validation without fallback defaults.
        raise KeyError(
            "study_config does not expose 'equation_parameters' as a mapping."
        )
    # Merge the algebraic keys and the registry tags into one declared order.
    declared: Tuple[str, ...] = REQUIRED_EQUATION_KEYS + REGISTRY_TAG_KEYS
    # Compute the declared keys absent from the section.
    missing: List[str] = [key for key in declared if key not in section]
    # Halt on any missing key, reporting the complete absent list.
    if missing:
        # The message lists every absent key so one pass repairs them all.
        raise KeyError(
            f"equation_parameters is missing {len(missing)} declared key(s): "
            f"{missing[:MAX_REPORTED_OFFENDERS]}."
        )
    # Accumulate one check record per declared key.
    rows: Dict[str, Dict[str, bool]] = {}
    # Iterate over the declared keys in their contractual order.
    for key in declared:
        # Read the entry mapped to the key.
        entry = section[key]
        # Test whether the entry is a mapping at all.
        is_mapping: bool = isinstance(entry, Mapping)
        # Test whether the entry is non-empty when it is a mapping.
        non_empty: bool = bool(entry) if is_mapping else False
        # Read the required field set of the key from the right registry.
        required_fields: Tuple[str, ...] = (
            REGISTRY_TAG_REQUIRED_FIELDS[key]
            if key in REGISTRY_TAG_REQUIRED_FIELDS
            else EQUATION_REQUIRED_FIELDS[key]
        )
        # Compute the required fields absent from the entry.
        absent: List[str] = (
            [field for field in required_fields if field not in entry]
            if is_mapping
            else list(required_fields)
        )
        # Identify required fields whose definitions are empty or unresolved.
        unpopulated_fields: List[str] = (
            [
                field
                for field in required_fields
                if field in entry and _count_unpopulated_leaves(entry[field]) > 0
            ]
            if is_mapping
            else list(required_fields)
        )
        # Distinguish algebraic definitions from source/reconstruction tags.
        entry_kind: str = "registry_tag" if key in REGISTRY_TAG_KEYS else "equation"
        # A usable declaration is mapped, complete and fully populated.
        usable_definition: bool = bool(
            is_mapping and non_empty and not absent and not unpopulated_fields
        )
        # Record the complete per-entry readiness audit.
        rows[key] = {
            "present": True,
            "is_mapping": is_mapping,
            "non_empty": non_empty,
            "has_required_fields": not absent,
            "entry_kind": entry_kind,
            "unpopulated_fields": tuple(unpopulated_fields),
            "usable_definition": usable_definition,
        }
        # Refuse an entry that is not a non-empty mapping.
        if not is_mapping or not non_empty:
            # The failure names the key so the repair is unambiguous.
            raise ConfigValidationError(
                f"equation_parameters.{key} must be a non-empty mapping; "
                f"received {type(entry).__name__}."
            )
        # Refuse an entry that omits any required field.
        if absent:
            # The failure names both the key and the missing fields.
            raise ConfigValidationError(
                f"equation_parameters.{key} omits required field(s) "
                f"{absent[:MAX_REPORTED_OFFENDERS]}."
            )
        # Refuse required mathematical fields that contain empty declarations.
        if unpopulated_fields:
            # Registry tags are validated separately but cannot be placeholders.
            raise ConfigValidationError(
                f"equation_parameters.{key} has unpopulated required field(s) "
                f"{unpopulated_fields[:MAX_REPORTED_OFFENDERS]}."
            )
    # Assemble the validity matrix with the declared key order preserved.
    matrix: pd.DataFrame = pd.DataFrame.from_dict(rows, orient="index")
    # Assert the matrix index equals the declared key tuple exactly.
    if tuple(str(name) for name in matrix.index) != declared:
        # An index divergence means a key was dropped during assembly.
        raise ConfigValidationError(
            "The equation validity matrix index diverges from the declared keys."
        )
    # Log the completed equation validation.
    logger.info(
        "equation registry validated: %d algebraic keys and %d registry tags.",
        len(REQUIRED_EQUATION_KEYS),
        len(REGISTRY_TAG_KEYS),
    )
    # Return the assembled validity matrix.
    return matrix


# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Step 2: Verify algorithm parameters and cross-check the numeric targets
# -------------------------------------------------------------------------------------------------------------------------------


# Return the declared algorithm keys in their contractual order.
def required_algorithm_keys() -> Tuple[str, ...]:
    """
    Return the twenty-six algorithm keys.

    Returns
    -------
    tuple of str
        The declared algorithm keys, in the order fixed by Section 5.
    """
    # Return the module constant so the declaration has one definition.
    return REQUIRED_ALGORITHM_KEYS


# Return the declared minimum parameter set of every algorithm entry.
def required_algorithm_fields() -> Mapping[str, Tuple[str, ...]]:
    """
    Return the minimum parameter set required of each algorithm entry.

    Returns
    -------
    mapping of str to tuple of str
        The required field names keyed by algorithm key.
    """
    # Return the module constant so the declaration has one definition.
    return ALGORITHM_REQUIRED_FIELDS


# Validate the algorithm registry and return its boolean validity matrix.
def validate_algorithm_registry(
    study_config: Mapping[str, Any],
) -> pd.DataFrame:
    """
    Validate that every declared algorithm is present and complete.

    Parameters
    ----------
    study_config : mapping
        The study configuration mapping.

    Returns
    -------
    pandas.DataFrame
        A boolean matrix indexed by algorithm key with the columns present,
        is_mapping, non_empty and has_required_fields.

    Raises
    ------
    KeyError
        If the algorithm section is absent or if any declared key is missing.
    ConfigValidationError
        If a declared key is not a non-empty mapping, or if it omits any of
        its required fields.
    """
    # Read the algorithm section, requiring it to be a mapping.
    section = study_config.get("algorithm_parameters")
    # Refuse a configuration whose algorithm section is absent or ill-typed.
    if not isinstance(section, Mapping):
        # A missing section halts validation without fallback defaults.
        raise KeyError(
            "study_config does not expose 'algorithm_parameters' as a mapping."
        )
    # Compute the declared keys absent from the section.
    missing: List[str] = [
        key for key in REQUIRED_ALGORITHM_KEYS if key not in section
    ]
    # Halt on any missing key, reporting the complete absent list.
    if missing:
        # The message lists every absent key so one pass repairs them all.
        raise KeyError(
            f"algorithm_parameters is missing {len(missing)} declared key(s): "
            f"{missing[:MAX_REPORTED_OFFENDERS]}."
        )
    # Accumulate one check record per declared key.
    rows: Dict[str, Dict[str, bool]] = {}
    # Iterate over the declared keys in their contractual order.
    for key in REQUIRED_ALGORITHM_KEYS:
        # Read the entry mapped to the key.
        entry = section[key]
        # Test whether the entry is a mapping at all.
        is_mapping: bool = isinstance(entry, Mapping)
        # Test whether the entry is non-empty when it is a mapping.
        non_empty: bool = bool(entry) if is_mapping else False
        # Compute the required fields absent from the entry.
        absent: List[str] = (
            [field for field in ALGORITHM_REQUIRED_FIELDS[key] if field not in entry]
            if is_mapping
            else list(ALGORITHM_REQUIRED_FIELDS[key])
        )
        # Identify required fields containing empty or unresolved declarations.
        unpopulated_fields: List[str] = (
            [
                field
                for field in ALGORITHM_REQUIRED_FIELDS[key]
                if field in entry and _count_unpopulated_leaves(entry[field]) > 0
            ]
            if is_mapping
            else list(ALGORITHM_REQUIRED_FIELDS[key])
        )
        # Validate ordered window/span declarations found in required fields.
        invalid_spans: List[str] = []
        # Inspect only present required fields under the mapping guard.
        if is_mapping:
            # Walk each required algorithm parameter.
            for field in ALGORITHM_REQUIRED_FIELDS[key]:
                # Ignore fields already reported absent.
                if field not in entry:
                    # Presence failure is handled below.
                    continue
                # Window-named two-scalar tuples must be ordered.
                value: Any = entry[field]
                # Identify singular window declarations.
                if "window" in field.lower() and isinstance(value, tuple):
                    # Validate a two-endpoint integer span.
                    if len(value) == 2 and all(
                        isinstance(item, (int, np.integer)) for item in value
                    ) and int(value[0]) > int(value[1]):
                        # Record the exact invalid parameter name.
                        invalid_spans.append(field)
                # Reject non-finite numerical parameters.
                if isinstance(value, (float, np.floating)) and not np.isfinite(value):
                    # Non-finite thresholds are not executable choices.
                    invalid_spans.append(field)
                # Probability-like bounds and missingness maxima lie in [0, 1].
                if (
                    isinstance(value, (int, float, np.integer, np.floating))
                    and any(
                        token in field.lower()
                        for token in ("missing_max", "ci_level")
                    )
                    and not 0.0 <= float(value) <= 1.0
                ):
                    # Record the parameter whose domain is violated.
                    invalid_spans.append(field)
        # A verified algorithm must be complete, populated and domain-valid.
        executable: bool = bool(
            is_mapping
            and non_empty
            and not absent
            and not unpopulated_fields
            and not invalid_spans
        )
        # Record the complete per-algorithm parameter audit.
        rows[key] = {
            "present": True,
            "is_mapping": is_mapping,
            "non_empty": non_empty,
            "has_required_fields": not absent,
            "unpopulated_fields": tuple(unpopulated_fields),
            "invalid_domains": tuple(sorted(set(invalid_spans))),
            "executable": executable,
        }
        # Refuse an entry that is not a non-empty mapping.
        if not is_mapping or not non_empty:
            # The failure names the key so the repair is unambiguous.
            raise ConfigValidationError(
                f"algorithm_parameters.{key} must be a non-empty mapping; "
                f"received {type(entry).__name__}."
            )
        # Refuse an entry that omits any required field.
        if absent:
            # The failure names both the key and the missing fields.
            raise ConfigValidationError(
                f"algorithm_parameters.{key} omits required field(s) "
                f"{absent[:MAX_REPORTED_OFFENDERS]}."
            )
        # Refuse empty placeholders in required executable parameters.
        if unpopulated_fields:
            # Name every affected required field deterministically.
            raise ConfigValidationError(
                f"algorithm_parameters.{key} has unpopulated required field(s) "
                f"{unpopulated_fields[:MAX_REPORTED_OFFENDERS]}."
            )
        # Refuse malformed spans, thresholds or other numeric domains.
        if invalid_spans:
            # Report unique parameter names in stable order.
            raise ConfigValidationError(
                f"algorithm_parameters.{key} has invalid parameter domain(s) "
                f"{sorted(set(invalid_spans))[:MAX_REPORTED_OFFENDERS]}."
            )
    # Assemble the validity matrix with the declared key order preserved.
    matrix: pd.DataFrame = pd.DataFrame.from_dict(rows, orient="index")
    # Assert the matrix index equals the declared key tuple exactly.
    if tuple(str(name) for name in matrix.index) != REQUIRED_ALGORITHM_KEYS:
        # An index divergence means a key was dropped during assembly.
        raise ConfigValidationError(
            "The algorithm validity matrix index diverges from the declared keys."
        )
    # Log the completed algorithm validation.
    logger.info(
        "algorithm registry validated: %d declared keys.", len(REQUIRED_ALGORITHM_KEYS)
    )
    # Return the assembled validity matrix.
    return matrix


# Return the declared acceptance targets as dotted paths and expected values.
def acceptance_target_registry() -> Mapping[str, Any]:
    """
    Return the manuscript-table acceptance targets.

    Returns
    -------
    mapping
        Expected values keyed by dotted configuration path. Each entry is the
        value the manuscript's tables and appendices fix for that parameter.
    """
    # Declare every target with its expected value from the manuscript tables.
    targets: Dict[str, Any] = {
        "global_constants.n_countries_final_panel": 129,
        "global_constants.n_observed_years": 25,
        "global_constants.n_country_years": 3225,
        "global_constants.n_outcomes": 4,
        "global_constants.n_controls": 8,
        "global_constants.control_lag_years": 1,
        "global_constants.T0_pre_treatment_years": 13,
        "global_constants.pre_treatment_window": (2001, 2013),
        "global_constants.post_treatment_window": (2014, 2024),
        "global_constants.Wave1": (2014, 2021),
        "global_constants.Wave2": (2022, 2024),
        "global_constants.preferred_reference_window": (2019, 2021),
        "global_constants.alternative_reference_windows": [
            (2017, 2019), (2021, 2021), (2017, 2021), (2018, 2021)
        ],
        "global_constants.n_sanctions_coded_descriptive": 43,
        "global_constants.n_sanctions_targets": 42,
        "global_constants.n_direct_arming_targets": 29,
        "global_constants.n_both_policies": 28,
        "global_constants.n_sanctions_only": 14,
        "global_constants.n_direct_arming_only": 1,
        "global_constants.n_neither_policy": 83,
        "global_constants.n_geopolitical_exclusions": 3,
        "global_constants.n_sanctions_donors_CPI": 84,
        "global_constants.n_sanctions_donors_GDPpc": 84,
        "global_constants.n_arming_donors_CPI": 96,
        "global_constants.n_arming_donors_GDPpc": 97,
        "global_constants.n_policy_outcome_estimator_runs": 8,
        "global_constants.degrees_of_freedom_rule": "N_T - 1",
        "global_constants.percentage_transform": "100 * (exp(theta_hat) - 1)",
        "equation_parameters.eq2_reporting_windows.Wave1_length": 8,
        "equation_parameters.eq2_reporting_windows.Wave2_length": 3,
        "equation_parameters.eq12_window_ATT.N_T_sanctions": 42,
        "equation_parameters.eq12_window_ATT.N_T_direct_arming": 29,
        "equation_parameters.eq15_incremental_wave2_ATT.identity_tolerance": "1e-10",
        "equation_parameters.eq17_exposure_measure.windows": [
            (2010, 2013), (2018, 2021)
        ],
        "equation_parameters.eq18_exposure_heterogeneity.regression_samples": {
            "2018_2021_strict": {"N_all": 118, "N_sanctioners": 39},
            "2010_2013_strict": {"N_all": 119, "N_sanctioners": 39},
            "2018_2021_extended": {"N_all": 119, "N_sanctioners": 39},
            "2010_2013_extended": {"N_all": 120, "N_sanctioners": 39},
        },
        "algorithm_parameters.A1_two_wave_CCEDID_imputation.n_steps": 5,
        "algorithm_parameters.B1_country_history_bootstrap.replicates_B": 500,
        "algorithm_parameters.B2_region_block_bootstrap.replicates_B": 500,
        "algorithm_parameters.B3_size_matched_placebo.replicates_M": 500,
        "algorithm_parameters.B3_size_matched_placebo.size_match_by_policy": {
            "sanctions": {"pseudo_treated": 42, "pseudo_donors": 42},
            "direct_arming_CPI": {"pseudo_treated": 29, "pseudo_donors": 67},
            "direct_arming_GDPpc": {"pseudo_treated": 29, "pseudo_donors": 68},
        },
        "algorithm_parameters.C1_eligibility_screening.outcome_missing_max": 0.35,
        "algorithm_parameters.C1_eligibility_screening.control_high_missing_max": 0.25,
        "algorithm_parameters.C1_eligibility_screening.control_high_missing_count_max": 1,
        "algorithm_parameters.C1_eligibility_screening.control_single_missing_max": 0.65,
        "algorithm_parameters.C1_eligibility_screening.critical_window": (2019, 2024),
        "algorithm_parameters.C1_eligibility_screening.critical_outcome_missing_max": 0.25,
        "algorithm_parameters.C1_eligibility_screening.expected_otherwise_eligible": 131,
        "algorithm_parameters.C1_eligibility_screening.expected_final": 129,
        "algorithm_parameters.C1_eligibility_screening.documented_exclusions": {
            "SYR": 5, "SDN": 2
        },
        "algorithm_parameters.C2_control_completion.priority_3_max_years": 3,
        "algorithm_parameters.C2_control_completion.priority_4_anchor_count": 5,
        "algorithm_parameters.C2_control_completion.non_negativity_floors": [
            "TradeOpenness", "Investment", "GovConsumption"
        ],
        "algorithm_parameters.C3_outcome_repair.leading_backcast_count": 55,
        "algorithm_parameters.C3_outcome_repair.carry_forward_cells": 2,
        "algorithm_parameters.C3_outcome_repair.carry_forward_country": "BIH",
        "algorithm_parameters.D1_comtrade_query.potential_cells": 1032,
        "algorithm_parameters.D4_within_window_completion.primitive_imputations": {
            "russian_gas": 13, "world_gas": 13, "russian_non_gas": 11
        },
        "algorithm_parameters.D4_within_window_completion.edge_growth_bounds": (0.05, 0.95),
        "algorithm_parameters.D5_exposure_sample_rules.excluded_under_both": [
            "AUT", "DEU", "POL", "TUR"
        ],
        "algorithm_parameters.D5_exposure_sample_rules.regression_counts": {
            "2018_2021_strict": 118, "2018_2021_extended": 119,
            "2010_2013_strict": 119, "2010_2013_extended": 120,
        },
        "algorithm_parameters.E1_high_income_design.expected_targets": 38,
        "algorithm_parameters.E1_high_income_design.expected_donors": 25,
        "algorithm_parameters.E1_high_income_design.max_abs_std_diff_reported": 1.175,
        "algorithm_parameters.E2_mahalanobis_matching.M_neighbors": 5,
        "algorithm_parameters.E2_mahalanobis_matching.expected_unique_donors": 37,
        "algorithm_parameters.E2_mahalanobis_matching.max_abs_std_diff_reported": 1.052,
        "algorithm_parameters.E3_standardized_difference_balance.target_baseline": 1.498,
        "algorithm_parameters.F1_treated_LOO.range_CPI_reported": [0.070, 0.074],
        "algorithm_parameters.F1_treated_LOO.range_GDPpc_reported": [0.023, 0.028],
        "algorithm_parameters.F1b_distributional_diagnostics.targets_CPI": {
            "positive_share": 0.976, "median": 0.062, "trimmed_mean_10pct": 0.071
        },
        "algorithm_parameters.F1b_distributional_diagnostics.targets_GDPpc": {
            "positive_share": 0.667, "median": 0.026, "trimmed_mean_10pct": 0.025
        },
        "algorithm_parameters.F2_donor_LOO.max_change_CPI_reported": 0.0049,
        "algorithm_parameters.F2_donor_LOO.max_change_GDPpc_reported": 0.0019,
        "algorithm_parameters.F4_factor_proxy_variants.variants": [
            "own_outcome_donor_mean", "other_outcome_donor_mean",
            "both_outcome_means", "own_outcome_plus_oil_rents",
            "own_outcome_plus_fuel_share_balance",
        ],
        "algorithm_parameters.F5_reference_window_sweep.reference_windows": [
            (2019, 2021), (2017, 2019), (2021, 2021), (2017, 2021), (2018, 2021)
        ],
        "algorithm_parameters.F4_factor_proxy_variants.dimensions_by_variant": [
            2, 2, 3, 3, 3
        ],
        "algorithm_parameters.F4_factor_proxy_variants.condition_numbers_sanctions_CPI": [
            91.47, 637.31, 4409.35, 147.99, 650.05
        ],
        "algorithm_parameters.F4_factor_proxy_variants.reported_ranges": {
            "CPI": [0.072, 0.103], "GDPpc": [0.007, 0.099]
        },
        "algorithm_parameters.F5_reference_window_sweep.reported_CPI_values": [
            0.072, 0.070, 0.071, 0.071, 0.072
        ],
        "algorithm_parameters.F5_reference_window_sweep.reported_GDPpc_values": [
            0.025, 0.053, -0.001, 0.039, 0.032
        ],
        "algorithm_parameters.G1_exposure_heterogeneity_matrix.strict_arming_sanctioner_count": 26,
        "algorithm_parameters.I1_pointwise_ribbon_figure2.CI_level": 0.95,
        "algorithm_parameters.I1_pointwise_ribbon_figure2.normalization_year": 2013,
        "series_code_registry.wdi_outcomes.CPI.indicator_code": "FP.CPI.TOTL",
        "series_code_registry.wdi_outcomes.RealGDPpc.indicator_code": "NY.GDP.PCAP.KD",
        "series_code_registry.wdi_outcomes.InflationRate.indicator_code": "FP.CPI.TOTL.ZG",
        "series_code_registry.wdi_outcomes.RealGDPpcGrowth.indicator_code": "NY.GDP.PCAP.KD.ZG",
        "series_code_registry.wgi_controls.RuleOfLaw.winsorizing_applied": False,
        "series_code_registry.nominal_gdp.NominalGDP.indicator_code": "NY.GDP.MKTP.CD",
        "series_code_registry.temporal_universe_by_layer.raw_retrieval": (1996, 2024),
        "series_code_registry.temporal_universe_by_layer.eligibility_screening": (2000, 2024),
        "series_code_registry.temporal_universe_by_layer.balanced_panel": (2000, 2024),
        "series_code_registry.temporal_universe_by_layer.effective_estimation_after_lag": (2001, 2024),
        "series_code_registry.temporal_universe_by_layer.pre_treatment_fit": (2001, 2013),
        "series_code_registry.temporal_universe_by_layer.post_treatment_projection": (2014, 2024),
    }
    # Return the declared registry of acceptance targets.
    return targets


# Compare an observed value with an expected value under declared semantics.
def _values_match(observed: Any, expected: Any) -> bool:
    """
    Test whether an observed value matches an expected value.

    Parameters
    ----------
    observed : Any
        The value read from the configuration.
    expected : Any
        The value declared by the acceptance registry.

    Returns
    -------
    bool
        True when the values match under the declared semantics: mappings are
        compared by key set and recursively by value, sequences by length and
        element, floats with a fixed relative tolerance against int or float
        observations only, and all other values by strict type-and-value
        equality.

    Notes
    -----
    The comparison is type-strict: a string observation is NEVER coerced to a
    number, so a configuration value that arrived as "129" (str) fails against
    a float target of 129.0 instead of silently passing. This closes the
    numeric-string coercion loophole while preserving the relative-tolerance
    arithmetic for genuinely numeric observations.
    """
    # Compare mappings by key set and then recursively by value.
    if isinstance(expected, Mapping):
        # Reject an observed value that is not itself a mapping.
        if not isinstance(observed, Mapping):
            # A type mismatch is a value mismatch.
            return False
        # Compare the key sets first, treating keys as strings.
        if {str(key) for key in observed} != {str(key) for key in expected}:
            # A key-set divergence is a value mismatch.
            return False
        # Compare every declared value recursively.
        return all(_values_match(observed[key], expected[key]) for key in expected)
    # Compare sequences element by element, unifying lists and tuples.
    if isinstance(expected, (list, tuple)):
        # Reject an observed value that is not itself a sequence.
        if not isinstance(observed, (list, tuple)):
            # A type mismatch is a value mismatch.
            return False
        # Reject a length mismatch before element comparison.
        if len(observed) != len(expected):
            # A length divergence is a value mismatch.
            return False
        # Compare every element pairwise in order.
        return all(
            _values_match(left, right) for left, right in zip(observed, expected)
        )
    # Compare floating-point values with the declared relative tolerance.
    if isinstance(expected, float):
        # Accept int and float observations only; bool is a subclass of int and
        # must be refused explicitly, and strings are never coerced to numbers
        # under the type-strict contract.
        if isinstance(observed, bool) or not isinstance(observed, (int, float)):
            # A string, bool or other non-numeric observation is a typed
            # mismatch, not a numeric near-miss.
            return False
        # Compute the tolerance scaled by the expected magnitude.
        tolerance: float = FLOAT_RELATIVE_TOLERANCE * max(1.0, abs(expected))
        # Compare within the tolerance using the numeric value directly.
        return abs(float(observed) - expected) <= tolerance
    # Compare every remaining value by strict type-and-value equality, so a
    # str "129" never equals the int 129 and vice versa.
    return type(observed) is type(expected) and observed == expected


# Read a nested configuration value through its dotted path.
def read_dotted_path(study_config: Mapping[str, Any], path: str) -> Any:
    """
    Return the configuration value addressed by a dotted path.

    Parameters
    ----------
    study_config : mapping
        The study configuration mapping.
    path : str
        The dotted path, for example algorithm_parameters.C1.expected_final.

    Returns
    -------
    Any
        The value found at the path.

    Raises
    ------
    KeyError
        If any segment of the path is absent or the traversal reaches a
        non-mapping node before the final segment.
    """
    # Start the traversal at the configuration root.
    node: Any = study_config
    # Walk each dotted segment in order.
    for segment in path.split("."):
        # Refuse a traversal that reaches a non-mapping node early.
        if not isinstance(node, Mapping):
            # The failure names the segment at which traversal broke down.
            raise KeyError(
                f"Path {path!r} traverses a non-mapping node at segment "
                f"{segment!r}."
            )
        # Refuse a path whose segment is absent from the current node.
        if segment not in node:
            # The failure names the missing segment for a precise repair.
            raise KeyError(f"Path {path!r} is absent at segment {segment!r}.")
        # Descend into the next node.
        node = node[segment]
    # Return the value found at the end of the traversal.
    return node


# Audit every declared acceptance target against the configuration.
def audit_declared_targets(
    study_config: Mapping[str, Any],
) -> pd.DataFrame:
    """
    Cross-check the configuration against the manuscript-table targets.

    Parameters
    ----------
    study_config : mapping
        The study configuration mapping.

    Returns
    -------
    pandas.DataFrame
        A table indexed by dotted path with the columns expected, observed
        and matches.

    Raises
    ------
    KeyError
        If any declared target path does not resolve.
    """
    # Read the declared acceptance target registry.
    targets: Mapping[str, Any] = acceptance_target_registry()
    # Accumulate one audit record per declared path.
    rows: Dict[str, Dict[str, Any]] = {}
    # Iterate over the declared paths in their declaration order.
    for path, expected in targets.items():
        # Read the observed value through the dotted path.
        observed: Any = read_dotted_path(study_config, path)
        # Record the expected, observed and match triple.
        rows[path] = {
            "expected": expected,
            "observed": observed,
            "matches": _values_match(observed, expected),
        }
    # Assemble the audit table with the declared path order preserved.
    matrix: pd.DataFrame = pd.DataFrame.from_dict(rows, orient="index")
    # Log the completed audit with the number of declared targets.
    logger.info("acceptance target audit assembled: %d target(s).", len(rows))
    # Return the assembled audit table.
    return matrix


# Assert that every audited target matched its declared value.
def assert_audit_passed(target_matrix: pd.DataFrame) -> None:
    """
    Assert that no acceptance target diverged from its declared value.

    Parameters
    ----------
    target_matrix : pandas.DataFrame
        The audit table returned by the target auditor.

    Returns
    -------
    None
        The function returns nothing when every target matched.

    Raises
    ------
    ConfigValidationError
        If one or more targets diverged, reporting every failing path with
        its expected and observed values.
    """
    # Select the rows whose match flag is false.
    failing = target_matrix[~target_matrix["matches"].astype(bool)]
    # Raise when any target diverged, reporting every failure at once.
    if len(failing):
        # Compose one message line per failing path.
        details: List[str] = [
            f"{path}: expected {row['expected']!r}, observed {row['observed']!r}"
            for path, row in failing.iterrows()
        ]
        # Raise with the complete failure list, capped for readability.
        raise ConfigValidationError(
            f"{len(failing)} acceptance target(s) diverged from the manuscript "
            f"tables: {'; '.join(details[:MAX_REPORTED_OFFENDERS])}"
        )
    # Log the successful audit.
    logger.info("acceptance target audit passed for %d target(s).", len(target_matrix))


# Re-derive the internal identities implied by the configuration's own fields.
def internal_consistency_identities(
    study_config: Mapping[str, Any],
) -> pd.DataFrame:
    """
    Re-derive the internal identities of the configuration.

    Parameters
    ----------
    study_config : mapping
        The study configuration mapping.

    Returns
    -------
    pandas.DataFrame
        A table indexed by identity name with the columns expected, observed
        and matches.

    Raises
    ------
    KeyError
        If any configuration path required by an identity is absent.
    """
    # Read the global constants node for the identity derivations.
    constants = read_dotted_path(study_config, "global_constants")
    # Read the eligibility node for the exclusion identity.
    eligibility = read_dotted_path(study_config, "algorithm_parameters.C1_eligibility_screening")
    # Read the equivalence equation node for the tolerance identity.
    window_att = read_dotted_path(study_config, "equation_parameters.eq12_window_ATT")
    # Read the reporting windows node for the length identities.
    windows = read_dotted_path(study_config, "equation_parameters.eq2_reporting_windows")
    # Read the factor-proxy node for the dimension identity.
    factor = read_dotted_path(study_config, "algorithm_parameters.F4_factor_proxy_variants")
    # Read the reference-window node for the sweep-length identity.
    reference = read_dotted_path(study_config, "algorithm_parameters.F5_reference_window_sweep")
    # Accumulate one record per identity.
    rows: Dict[str, Dict[str, Any]] = {}
    # Identity: the country-year product equals countries times years.
    rows["country_years_product"] = {
        "expected": int(constants["n_countries_final_panel"]) * int(constants["n_observed_years"]),
        "observed": int(constants["n_country_years"]),
    }
    # Identity: the pre-treatment length equals the declared span inclusive.
    rows["pre_treatment_span"] = {
        "expected": int(constants["pre_treatment_window"][1]) - int(constants["pre_treatment_window"][0]) + 1,
        "observed": int(constants["T0_pre_treatment_years"]),
    }
    # Identity: the first wave length matches its declared window.
    rows["wave1_length"] = {
        "expected": int(constants["Wave1"][1]) - int(constants["Wave1"][0]) + 1,
        "observed": int(windows["Wave1_length"]),
    }
    # Identity: the second wave length matches its declared window.
    rows["wave2_length"] = {
        "expected": int(constants["Wave2"][1]) - int(constants["Wave2"][0]) + 1,
        "observed": int(windows["Wave2_length"]),
    }
    # Identity: the sanctions target count decomposes into both plus only.
    rows["sanctions_target_partition"] = {
        "expected": int(constants["n_both_policies"]) + int(constants["n_sanctions_only"]),
        "observed": int(constants["n_sanctions_targets"]),
    }
    # Identity: the arming target count decomposes into both plus only.
    rows["arming_target_partition"] = {
        "expected": int(constants["n_both_policies"]) + int(constants["n_direct_arming_only"]),
        "observed": int(constants["n_direct_arming_targets"]),
    }
    # Identity: the descriptive coding exceeds the target count by Ukraine.
    rows["coded_minus_targets"] = {
        "expected": int(constants["n_sanctions_targets"]) + 1,
        "observed": int(constants["n_sanctions_coded_descriptive"]),
    }
    # Identity: the policy groups cover the universe minus the exclusions.
    rows["policy_group_coverage"] = {
        "expected": int(constants["n_countries_final_panel"]) - int(constants["n_geopolitical_exclusions"]),
        "observed": (
            int(constants["n_neither_policy"])
            + int(constants["n_sanctions_only"])
            + int(constants["n_direct_arming_only"])
            + int(constants["n_both_policies"])
        ),
    }
    # Identity: the sanctions donor pool is the neither group plus one.
    rows["sanctions_donor_composition"] = {
        "expected": int(constants["n_neither_policy"]) + int(constants["n_direct_arming_only"]),
        "observed": int(constants["n_sanctions_donors_CPI"]),
    }
    # Identity: the arming CPI donor pool omits Bosnia and Herzegovina.
    rows["arming_cpi_donor_composition"] = {
        "expected": (
            int(constants["n_neither_policy"])
            + int(constants["n_sanctions_only"])
            - 1
        ),
        "observed": int(constants["n_arming_donors_CPI"]),
    }
    # Identity: the arming output donor pool retains every sanctions-only country.
    rows["arming_gdppc_donor_composition"] = {
        "expected": int(constants["n_neither_policy"]) + int(constants["n_sanctions_only"]),
        "observed": int(constants["n_arming_donors_GDPpc"]),
    }
    # Identity: the estimator run count equals the policy-outcome product.
    rows["estimator_run_product"] = {
        "expected": 2 * int(constants["n_outcomes"]),
        "observed": int(constants["n_policy_outcome_estimator_runs"]),
    }
    # Identity: the eligibility screen removes exactly the documented countries,
    # where the documented mapping records each country's triggering cell count.
    rows["eligibility_removal_count"] = {
        "expected": int(eligibility["expected_otherwise_eligible"]) - int(eligibility["expected_final"]),
        "observed": len(eligibility["documented_exclusions"]),
    }
    # Identity: the target counts in equation twelve match the global constants.
    rows["equation_target_counts"] = {
        "expected": [int(constants["n_sanctions_targets"]), int(constants["n_direct_arming_targets"])],
        "observed": [int(window_att["N_T_sanctions"]), int(window_att["N_T_direct_arming"])],
    }
    # Identity: the factor-proxy dimension vector matches the variant count.
    rows["factor_variant_lengths"] = {
        "expected": len(factor["variants"]),
        "observed": len(factor["dimensions_by_variant"]),
    }
    # Identity: the reference sweep reports one value per declared window.
    rows["reference_sweep_lengths"] = {
        "expected": [len(reference["reference_windows"]), len(reference["reference_windows"])],
        "observed": [len(reference["reported_CPI_values"]), len(reference["reported_GDPpc_values"])],
    }
    # Attach the match flag to every identity by comparing expected and observed.
    for record in rows.values():
        # Record whether the observed quantity equals the expected quantity.
        record["matches"] = _values_match(record["observed"], record["expected"])
    # Assemble the identity table with the declared order preserved.
    matrix: pd.DataFrame = pd.DataFrame.from_dict(rows, orient="index")
    # Log the completed identity derivation.
    logger.info("internal consistency identities evaluated: %d.", len(rows))
    # Return the assembled identity table.
    return matrix


# Assert that every internal identity held under the configuration.
def assert_internal_consistency(consistency_matrix: pd.DataFrame) -> None:
    """
    Assert that no internal identity of the configuration was violated.

    Parameters
    ----------
    consistency_matrix : pandas.DataFrame
        The identity table returned by the derivation callable.

    Returns
    -------
    None
        The function returns nothing when every identity held.

    Raises
    ------
    ConfigValidationError
        If one or more identities failed, reporting every failing identity
        with its expected and observed values.
    """
    # Select the identity rows whose match flag is false.
    failing = consistency_matrix[~consistency_matrix["matches"].astype(bool)]
    # Raise when any identity failed, reporting every failure at once.
    if len(failing):
        # Compose one message line per failing identity.
        details: List[str] = [
            f"{name}: expected {row['expected']!r}, observed {row['observed']!r}"
            for name, row in failing.iterrows()
        ]
        # Raise with the complete failure list for a single-pass repair.
        raise ConfigValidationError(
            f"{len(failing)} internal identity violation(s): "
            f"{'; '.join(details[:MAX_REPORTED_OFFENDERS])}"
        )
    # Log the successful identity check.
    logger.info(
        "internal consistency verified across %d identities.", len(consistency_matrix)
    )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Step 3: Verify the series registry, country registries and gaps
# -------------------------------------------------------------------------------------------------------------------------------


# Read a string field from a nested registry entry under a type guard.
def _registry_field(registry: Mapping[str, Any], name: str, field: str) -> str:
    """
    Return a declared field of a registry entry as a string.

    Parameters
    ----------
    registry : mapping
        The node carrying the entry, such as wdi_outcomes.
    name : str
        The entry name, such as CPI or FuelImportShare.
    field : str
        The field name to read, such as indicator_code or role.

    Returns
    -------
    str
        The declared field value as a string, or the empty string when the
        entry is absent or is not itself a mapping.
    """
    # Read the entry, defaulting to an empty mapping when it is absent.
    entry: Any = registry.get(name, {})
    # Return the empty string when the entry is not a mapping.
    if not isinstance(entry, Mapping):
        # A non-mapping entry declares no readable field.
        return ""
    # Return the field value as a string, defaulting to the empty string.
    return str(entry.get(field, ""))


# Validate the series-code registry and return its boolean validity matrix.
def validate_series_code_registry(
    study_config: Mapping[str, Any],
) -> pd.DataFrame:
    """
    Validate the series-code registry against the declared codes.

    Parameters
    ----------
    study_config : mapping
        The study configuration mapping.

    Returns
    -------
    pandas.DataFrame
        A boolean matrix indexed by check name with the columns ok and detail.

    Raises
    ------
    KeyError
        If the series registry or any required node is absent.
    ConfigValidationError
        If an outcome code, a control code, the rule-of-law flag or the
        nominal GDP code diverges from the declared value.
    """
    # Read the series registry node, requiring it to be a mapping.
    registry = study_config.get("series_code_registry")
    # Refuse a configuration whose series registry is absent.
    if not isinstance(registry, Mapping):
        # A missing registry halts validation without fallback defaults.
        raise KeyError(
            "study_config does not expose 'series_code_registry' as a mapping."
        )
    # Accumulate one check record per structural requirement.
    rows: Dict[str, Dict[str, Any]] = {}
    # Read the outcome node, requiring it to be a mapping.
    outcomes = registry.get("wdi_outcomes")
    # Refuse an absent outcome node.
    if not isinstance(outcomes, Mapping):
        # The outcome node is mandatory.
        raise KeyError("series_code_registry.wdi_outcomes must be a mapping.")
    # Check that the outcome node exposes exactly the four declared names.
    observed_outcomes: set = set(str(key) for key in outcomes.keys())
    # Record the outcome-name check.
    rows["outcome_names"] = {
        "ok": observed_outcomes == set(EXPECTED_OUTCOME_CODES.keys()),
        "detail": f"observed {sorted(observed_outcomes)}",
    }
    # Refuse a divergent outcome-name set.
    if not rows["outcome_names"]["ok"]:
        # The failure names both sets so the registry can be repaired.
        raise ConfigValidationError(
            f"wdi_outcomes must declare exactly {sorted(EXPECTED_OUTCOME_CODES)}; "
            f"observed {sorted(observed_outcomes)}."
        )
    # Check every declared outcome code against its expected value.
    for name, expected_code in EXPECTED_OUTCOME_CODES.items():
        # Read the observed code through the type-guarded field reader.
        observed_code: str = _registry_field(outcomes, name, "indicator_code")
        # Record the per-outcome code check.
        rows[f"outcome_code::{name}"] = {
            "ok": observed_code == expected_code,
            "detail": f"expected {expected_code!r}, observed {observed_code!r}",
        }
        # Refuse a divergent outcome code.
        if observed_code != expected_code:
            # The failure names the outcome and both codes.
            raise ConfigValidationError(
                f"wdi_outcomes.{name}.indicator_code must be {expected_code!r}; "
                f"observed {observed_code!r}."
            )
    # Read the control node, requiring it to be a mapping.
    controls = registry.get("wdi_controls")
    # Refuse an absent control node.
    if not isinstance(controls, Mapping):
        # The control node is mandatory.
        raise KeyError("series_code_registry.wdi_controls must be a mapping.")
    # Check that the control node exposes exactly the eight declared names.
    observed_controls: set = set(str(key) for key in controls.keys())
    # Record the control-name check.
    rows["control_names"] = {
        "ok": observed_controls == set(EXPECTED_CONTROL_CODES.keys()),
        "detail": f"observed {sorted(observed_controls)}",
    }
    # Refuse a divergent control-name set.
    if not rows["control_names"]["ok"]:
        # The failure names both sets so the registry can be repaired.
        raise ConfigValidationError(
            f"wdi_controls must declare exactly {sorted(EXPECTED_CONTROL_CODES)}; "
            f"observed {sorted(observed_controls)}."
        )
    # Check every declared control code against its expected value.
    for name, expected_code in EXPECTED_CONTROL_CODES.items():
        # Read the observed code through the type-guarded field reader.
        observed_code: str = _registry_field(controls, name, "indicator_code")
        # Record the per-control code check.
        rows[f"control_code::{name}"] = {
            "ok": observed_code == expected_code,
            "detail": f"expected {expected_code!r}, observed {observed_code!r}",
        }
        # Refuse a divergent control code.
        if observed_code != expected_code:
            # The failure names the control and both codes.
            raise ConfigValidationError(
                f"wdi_controls.{name}.indicator_code must be {expected_code!r}; "
                f"observed {observed_code!r}."
            )
    # Check that the two fuel entries are separate control primitives.
    for name in FUEL_PRIMITIVE_NAMES:
        # Read the declared role of the primitive under a type guard.
        observed_role: str = _registry_field(controls, name, "role")
        # Record the primitive-role check.
        rows[f"fuel_primitive::{name}"] = {
            "ok": observed_role == "control_primitive",
            "detail": f"observed role {observed_role!r}",
        }
        # Refuse a blended fuel entry that is not a primitive.
        if observed_role != "control_primitive":
            # The failure names the entry and the required role.
            raise ConfigValidationError(
                f"wdi_controls.{name}.role must be 'control_primitive'; "
                f"observed {observed_role!r}."
            )
    # Read the rule-of-law node under a type guard.
    wgi_node: Any = registry.get("wgi_controls", {})
    # Read the rule-of-law entry under a type guard.
    rule_of_law_entry: Any = (
        wgi_node.get("RuleOfLaw", {}) if isinstance(wgi_node, Mapping) else {}
    )
    # Read the winsorizing flag only when the entry is a mapping.
    winsorizing = (
        rule_of_law_entry.get("winsorizing_applied", None)
        if isinstance(rule_of_law_entry, Mapping)
        else None
    )
    # Record the rule-of-law flag check.
    rows["rule_of_law_winsorizing"] = {
        "ok": winsorizing is False,
        "detail": f"observed {winsorizing!r}",
    }
    # Refuse a winsorized declaration, which would corrupt the control scaling.
    if winsorizing is not False:
        # The failure states the required flag value.
        raise ConfigValidationError(
            "series_code_registry.wgi_controls.RuleOfLaw.winsorizing_applied "
            "must be False."
        )
    # Read the nominal GDP node under a type guard.
    nominal_node: Any = registry.get("nominal_gdp", {})
    # Read the indicator code through the type-guarded field reader.
    nominal_code: str = (
        _registry_field(nominal_node, "NominalGDP", "indicator_code")
        if isinstance(nominal_node, Mapping)
        else ""
    )
    # Record the nominal GDP code check.
    rows["nominal_gdp_code"] = {
        "ok": nominal_code == "NY.GDP.MKTP.CD",
        "detail": f"observed {nominal_code!r}",
    }
    # Refuse a nominal GDP code that names the real per-capita control.
    if nominal_code != "NY.GDP.MKTP.CD":
        # The failure names the required code.
        raise ConfigValidationError(
            "series_code_registry.nominal_gdp.NominalGDP.indicator_code must be "
            "'NY.GDP.MKTP.CD'."
        )
    # Assemble the validity matrix with the check names as the index.
    matrix: pd.DataFrame = pd.DataFrame.from_dict(rows, orient="index")
    # Log the completed series-registry validation.
    logger.info("series registry validated: %d check(s).", len(rows))
    # Return the assembled validity matrix.
    return matrix


# Validate the country registries and return their boolean validity matrix.
def validate_country_registries(
    study_config: Mapping[str, Any],
) -> pd.DataFrame:
    """
    Validate the country validation registries.

    Parameters
    ----------
    study_config : mapping
        The study configuration mapping.

    Returns
    -------
    pandas.DataFrame
        A boolean matrix indexed by check name with the columns ok and detail.

    Raises
    ------
    KeyError
        If the registry node or any declared key is absent.
    ConfigValidationError
        If a roster count, an exclusion member set or the additional donor
        exclusion diverges from its declared value, or if any declared roster
        count cannot be parsed as a non-negative integer (the module's error
        contract is preserved: no raw ValueError escapes validation).
    """
    # Read the registry node, requiring it to be a mapping.
    registry = study_config.get("country_validation_registries")
    # Refuse a configuration whose country registry is absent.
    if not isinstance(registry, Mapping):
        # A missing registry halts validation without fallback defaults.
        raise KeyError(
            "study_config does not expose 'country_validation_registries' as a "
            "mapping."
        )
    # Accumulate one check record per structural requirement.
    rows: Dict[str, Dict[str, Any]] = {}
    # Read the global constants node for the roster-count comparison.
    constants = read_dotted_path(study_config, "global_constants")

    # Define the strict declared-count reader preserving the error contract.
    def _read_declared_count(raw_key: str) -> int:
        """
        Parse the declared count embedded in a roster key name.

        Parameters
        ----------
        raw_key : str
            The roster key whose trailing segment carries the declared count.

        Returns
        -------
        int
            The parsed non-negative count.

        Raises
        ------
        ConfigValidationError
            If the trailing segment is absent, not an integer or negative; the
            message names the offending key and the raw segment.
        """
        # Extract the trailing segment of the key after the final underscore.
        raw_segment: str = raw_key.rsplit("_", 1)[1]
        # Refuse a key whose trailing segment is not a pure integer literal.
        if not raw_segment.isdigit():
            # The failure names the key and the raw segment for repair.
            raise ConfigValidationError(
                f"Roster key {raw_key!r} does not embed a parseable non-negative "
                f"declared count: trailing segment {raw_segment!r}."
            )
        # Convert the digit string to an integer.
        parsed: int = int(raw_segment)
        # Refuse a negative count defensively, even though isdigit excludes it.
        if parsed < 0:
            # The failure names the key and the parsed value.
            raise ConfigValidationError(
                f"Roster key {raw_key!r} declares a negative count {parsed}."
            )
        # Return the parsed count.
        return parsed

    # Verify that every declared roster key is present.
    for key in REQUIRED_ROSTER_KEYS:
        # Test the presence of the roster key.
        present: bool = key in registry
        # Parse the declared count FIRST so a malformed key fails with the
        # module's own error type before any downstream comparison runs.
        declared_count: int = _read_declared_count(key)
        # Record the roster-presence check.
        rows[f"roster_present::{key}"] = {
            "ok": present,
            "detail": f"declared count {declared_count}",
        }
        # Refuse an absent roster key.
        if not present:
            # The failure names the missing roster.
            raise KeyError(f"country_validation_registries.{key} is absent.")
        # Read the global constant that fixes the roster's count.
        constant_value: int = int(constants[ROSTER_COUNT_CONSTANTS[key]])
        # Record the roster-count consistency check.
        rows[f"roster_count::{key}"] = {
            "ok": declared_count == constant_value,
            "detail": (
                f"declared {declared_count}, global constant {constant_value}, "
                f"populated {len(registry[key]) if isinstance(registry[key], list) else 'n/a'}"
            ),
        }
        # Refuse a roster whose declared count contradicts the constants.
        if declared_count != constant_value:
            # The failure names both values so the constant or key is repaired.
            raise ConfigValidationError(
                f"Roster {key} declares {declared_count} but "
                f"global_constants.{ROSTER_COUNT_CONSTANTS[key]} is "
                f"{constant_value}."
            )
        # Require an actual roster sequence rather than a count-only claim.
        roster_value: Any = registry[key]
        # Strings are scalar identifiers, not roster containers.
        if not isinstance(roster_value, (list, tuple, set, frozenset)):
            # A count cannot validate identities when the roster is absent.
            raise ConfigValidationError(
                f"country_validation_registries.{key} must be an explicit "
                "country-code sequence."
            )
        # Preserve the submitted order while validating canonical identities.
        roster: Tuple[Any, ...] = tuple(roster_value)
        # Identify noncanonical ISO-like country identifiers.
        invalid_codes: List[Any] = [
            code
            for code in roster
            if not isinstance(code, str) or ISO3_PATTERN.fullmatch(code) is None
        ]
        # Refuse malformed roster identities before any set operation.
        if invalid_codes:
            # Report bounded original values without string coercion.
            raise ConfigValidationError(
                f"Roster {key} contains noncanonical country code(s) "
                f"{invalid_codes[:MAX_REPORTED_OFFENDERS]}."
            )
        # Require every identity to occur exactly once.
        if len(set(roster)) != len(roster):
            # Identify all duplicate identities deterministically.
            duplicates: List[str] = sorted(
                {code for code in roster if roster.count(code) > 1}
            )
            # Reject duplicate roster entries even if set cardinality looks right.
            raise ConfigValidationError(
                f"Roster {key} contains duplicate code(s) "
                f"{duplicates[:MAX_REPORTED_OFFENDERS]}."
            )
        # Compare the populated roster length with both declarations.
        if len(roster) != declared_count:
            # A correct label paired with a wrong roster is not validated.
            raise ConfigValidationError(
                f"Roster {key} declares {declared_count} identities but "
                f"contains {len(roster)} unique canonical codes."
            )
        # Record the identity digest so equal counts cannot conceal roster drift.
        rows[f"roster_identity::{key}"] = {
            "ok": True,
            "detail": hashlib.sha256(
                "\n".join(sorted(roster)).encode("utf-8")
            ).hexdigest(),
        }
    # Read the geopolitical exclusion map, requiring it to be a mapping.
    exclusions = registry.get("geopolitical_exclusions")
    # Refuse an absent exclusion map.
    if not isinstance(exclusions, Mapping):
        # The exclusion map is mandatory.
        raise KeyError(
            "country_validation_registries.geopolitical_exclusions must be a "
            "mapping."
        )
    # Verify every exclusion member set against its declared membership.
    for role, expected_members in EXPECTED_EXCLUSIONS.items():
        # Read the observed member list, defaulting to the empty list.
        observed_members = exclusions.get(role, [])
        # Normalise the observed members to a sorted tuple of strings.
        observed_tuple: Tuple[str, ...] = tuple(
            sorted(str(member) for member in observed_members)
        )
        # Record the exclusion membership check.
        rows[f"exclusion::{role}"] = {
            "ok": observed_tuple == expected_members,
            "detail": f"expected {expected_members}, observed {observed_tuple}",
        }
        # Refuse a divergent exclusion membership.
        if observed_tuple != expected_members:
            # The failure names the role and both member sets.
            raise ConfigValidationError(
                f"geopolitical_exclusions.{role} must declare {expected_members}; "
                f"observed {observed_tuple}."
            )
    # Read the additional donor exclusion list, defaulting to the empty list.
    additional = registry.get("direct_arming_CPI_donor_additional_exclusion", [])
    # Normalise the additional exclusions to a sorted tuple of strings.
    additional_tuple: Tuple[str, ...] = tuple(
        sorted(str(member) for member in additional)
    )
    # Record the additional-exclusion check.
    rows["additional_donor_exclusion"] = {
        "ok": additional_tuple == ("BIH",),
        "detail": f"observed {additional_tuple}",
    }
    # Refuse a divergent additional exclusion list.
    if additional_tuple != ("BIH",):
        # The failure names the required membership.
        raise ConfigValidationError(
            "direct_arming_CPI_donor_additional_exclusion must declare "
            "['BIH']."
        )
    # Resolve policy rosters for overlap and exclusion invariants.
    sanctions_targets: set[str] = set(registry["sanctions_target_42"])
    # Resolve the direct-arming estimation targets separately.
    arming_targets: set[str] = set(registry["direct_arming_target_29"])
    # Resolve the two sanctions/direct-arming donor universes.
    sanctions_donors: set[str] = set(registry["sanctions_donors_84"])
    # Combine outcome-specific arming donors for exclusion checks.
    arming_donors: set[str] = set(registry["direct_arming_donors_CPI_96"]) | set(
        registry["direct_arming_donors_GDPpc_97"]
    )
    # The corresponding treated and never-treated sets must be disjoint.
    if sanctions_targets & sanctions_donors:
        # Report the leaked treated codes.
        raise ConfigValidationError(
            "Sanctions targets and sanctions donors overlap at "
            f"{sorted(sanctions_targets & sanctions_donors)}."
        )
    # Direct-arming targets cannot appear in either arming donor list.
    if arming_targets & arming_donors:
        # Report the leaked treated codes.
        raise ConfigValidationError(
            "Direct-arming targets and donors overlap at "
            f"{sorted(arming_targets & arming_donors)}."
        )
    # Verify the intended overlap between the two policy target rosters.
    expected_both: int = int(constants["n_both_policies"])
    # Reject a correct pair of marginal counts with the wrong intersection.
    if len(sanctions_targets & arming_targets) != expected_both:
        # Name the observed and expected policy overlap.
        raise ConfigValidationError(
            "Policy target roster overlap is inconsistent: expected "
            f"{expected_both}, observed "
            f"{len(sanctions_targets & arming_targets)}."
        )
    # Geopolitical target bans must not leak into estimation targets.
    target_bans: set[str] = set(
        exclusions.get("target_and_donor_ban", [])
    ) | set(exclusions.get("target_exclusion_from_ATT", []))
    # Check both policy target lists against the explicit target bans.
    leaked_targets: set[str] = (
        sanctions_targets | arming_targets
    ) & target_bans
    # Refuse any prohibited target identity.
    if leaked_targets:
        # Report the exact leaking identities.
        raise ConfigValidationError(
            f"Target rosters contain geopolitically excluded code(s) "
            f"{sorted(leaked_targets)}."
        )
    # Assemble the validity matrix with the check names as the index.
    matrix: pd.DataFrame = pd.DataFrame.from_dict(rows, orient="index")
    # Log the completed country-registry validation.
    logger.info("country registries validated: %d check(s).", len(rows))
    # Return the assembled validity matrix.
    return matrix


# Count the unpopulated leaves beneath a configuration subtree.
def _count_unpopulated_leaves(node: Any) -> int:
    """
    Count the leaves of a subtree whose value is None.

    Parameters
    ----------
    node : Any
        The subtree to inspect.

    Returns
    -------
    int
        The number of None-valued leaves found in the subtree.
    """
    # Count null scalar leaves, including pandas/numpy missing sentinels.
    if node is None or node is pd.NA:
        # A null node is one unpopulated leaf.
        return 1
    # Recurse into every mapping value.
    if isinstance(node, Mapping):
        # An empty required mapping is itself one unresolved declaration.
        if not node:
            # Empty containers cannot prove a scientific choice was made.
            return 1
        # Sum the unpopulated leaves of every child.
        return sum(_count_unpopulated_leaves(child) for child in node.values())
    # Recurse into every sequence element.
    if isinstance(node, (list, tuple, set, frozenset)):
        # An empty required collection is one unresolved declaration.
        if not node:
            # Preserve the container itself as the unresolved leaf.
            return 1
        # Sum the unpopulated leaves of every element.
        return sum(_count_unpopulated_leaves(child) for child in node)
    # Treat blank and explicit unresolved text sentinels as unpopulated.
    if isinstance(node, str):
        # Normalize textual state only for the sentinel comparison.
        normalized: str = node.strip().lower()
        # Empty text and registered unresolved states are not resolutions.
        if not normalized or normalized in {
            "unspecified",
            "unspecified_by_paper",
            "unresolved",
            "not_provided",
            "referenced_but_not_typeset",
            "tbd",
            "todo",
            "none",
        }:
            # Count the unresolved string as one missing leaf.
            return 1
    # Count non-finite scalar numeric declarations as unresolved.
    if isinstance(node, (float, np.floating)) and not np.isfinite(node):
        # NaN and infinities are not executable parameter values.
        return 1
    # Count no unpopulated leaves in a populated scalar.
    return 0


# Validate the reproducibility-gap registry and return its validity matrix.
def validate_reproducibility_gaps(
    study_config: Mapping[str, Any],
) -> pd.DataFrame:
    """
    Validate that every declared reproducibility gap is registered.

    Parameters
    ----------
    study_config : mapping
        The study configuration mapping.

    Returns
    -------
    pandas.DataFrame
        A matrix indexed by gap key with the columns present, is_mapping,
        has_action_required and unpopulated_leaves.

    Raises
    ------
    KeyError
        If the gap registry node is absent or any declared gap key is missing.
    ConfigValidationError
        If a gap entry is not a mapping or does not declare its action.
    """
    # Read the gap registry node, requiring it to be a mapping.
    registry = study_config.get("reproducibility_gaps")
    # Refuse a configuration whose gap registry is absent.
    if not isinstance(registry, Mapping):
        # A missing registry halts validation without fallback defaults.
        raise KeyError(
            "study_config does not expose 'reproducibility_gaps' as a mapping."
        )
    # Compute the declared gap keys absent from the registry.
    missing: List[str] = [key for key in REQUIRED_GAP_KEYS if key not in registry]
    # Halt on any missing gap key, reporting the complete absent list.
    if missing:
        # The message lists every absent gap so one pass repairs them all.
        raise KeyError(
            f"reproducibility_gaps is missing {len(missing)} declared key(s): "
            f"{missing[:MAX_REPORTED_OFFENDERS]}."
        )
    # Define the auditable field roles accepted from the versioned registry.
    affected_fields: Tuple[str, ...] = ("affected_output", "consumed_by", "effect")
    # Define the states that may accompany a concrete resolution artefact.
    ready_states: set[str] = {"resolved", "ready", "documented", "execution_ready"}
    # Accumulate one check record per declared gap key.
    rows: Dict[str, Dict[str, Any]] = {}
    # Iterate over the declared gap keys in their contractual order.
    for key in REQUIRED_GAP_KEYS:
        # Read the entry mapped to the gap key.
        entry = registry[key]
        # Test whether the entry is a mapping at all.
        is_mapping: bool = isinstance(entry, Mapping)
        # Test whether the entry declares its blocking scientific choice.
        has_action: bool = is_mapping and bool(
            isinstance(entry.get("action_required"), str)
            and entry.get("action_required", "").strip()
        )
        # Count the unpopulated leaves beneath the entry.
        unpopulated: int = _count_unpopulated_leaves(entry)
        # Resolve an affected-output declaration from current or legacy fields.
        affected_output: Any = (
            next(
                (entry[field] for field in affected_fields if field in entry),
                None,
            )
            if is_mapping
            else None
        )
        # Read the explicit state without manufacturing one from the action.
        state: str = str(entry.get("state", entry.get("status", ""))).strip().lower() if is_mapping else ""
        # Read the evidence artefact that makes the choice reproducible.
        resolution_artifact: Any = (
            entry.get("resolution_artifact") if is_mapping else None
        )
        # Determine whether the gap was explicitly declared nonblocking.
        nonblocking: bool = bool(entry.get("nonblocking", False)) if is_mapping else False
        # A production-ready gap requires all semantic roles and concrete proof.
        execution_ready: bool = bool(
            is_mapping
            and has_action
            and _count_unpopulated_leaves(affected_output) == 0
            and state in ready_states
            and _count_unpopulated_leaves(resolution_artifact) == 0
        )
        # Preserve paths for every unresolved value in a deterministic audit.
        unresolved_paths: List[str] = []

        # Recursively enumerate unresolved paths beneath one gap entry.
        def _visit(value: Any, path: str) -> None:
            """Append every unpopulated descendant's full dotted path."""
            # Record the current path if the whole value is unresolved/empty.
            if _count_unpopulated_leaves(value) == 0:
                # A fully populated subtree needs no deeper traversal.
                return
            # Traverse non-empty mappings by key in deterministic order.
            if isinstance(value, Mapping) and value:
                # Preserve mapping key names in each descendant path.
                for child_key in sorted(value, key=str):
                    # Recurse into the child with a dotted path.
                    _visit(value[child_key], f"{path}.{child_key}")
                # Mapping traversal is complete.
                return
            # Traverse non-empty sequences by positional index.
            if isinstance(value, (list, tuple)) and value:
                # Preserve sequence positions in bracket notation.
                for position, child in enumerate(value):
                    # Recurse into the indexed child.
                    _visit(child, f"{path}[{position}]")
                # Sequence traversal is complete.
                return
            # The current scalar or empty container is unresolved.
            unresolved_paths.append(path)

        # Enumerate all unresolved paths from the gap root.
        _visit(entry, f"reproducibility_gaps.{key}")
        # Add absent semantic fields explicitly to the path audit.
        if _count_unpopulated_leaves(affected_output) > 0:
            # Name the canonical affected-output role.
            unresolved_paths.append(
                f"reproducibility_gaps.{key}.affected_output"
            )
        # Require an explicit resolution artefact for ready production state.
        if _count_unpopulated_leaves(resolution_artifact) > 0:
            # Name the canonical evidence role.
            unresolved_paths.append(
                f"reproducibility_gaps.{key}.resolution_artifact"
            )
        # Record the structural, semantic and readiness checks.
        rows[key] = {
            "gap_id": key,
            "present": True,
            "is_mapping": is_mapping,
            "has_action_required": has_action,
            "unpopulated_leaves": unpopulated,
            "affected_output": affected_output,
            "blocking_choice": entry.get("action_required") if is_mapping else None,
            "state": state,
            "resolution_artifact": resolution_artifact,
            "unresolved_paths": tuple(dict.fromkeys(unresolved_paths)),
            "nonblocking": nonblocking,
            "execution_ready": execution_ready,
        }
        # Refuse an entry that is not a mapping.
        if not is_mapping:
            # The failure names the key so the repair is unambiguous.
            raise ConfigValidationError(
                f"reproducibility_gaps.{key} must be a mapping; received "
                f"{type(entry).__name__}."
            )
        # Refuse an entry that omits its declared action.
        if not has_action:
            # The failure names both the key and the missing field.
            raise ConfigValidationError(
                f"reproducibility_gaps.{key} must declare 'action_required'."
            )
    # Assemble the validity matrix with the declared key order preserved.
    matrix: pd.DataFrame = pd.DataFrame.from_dict(rows, orient="index")
    # Log the completed gap validation with the unpopulated-leaf total.
    logger.info(
        "reproducibility gaps validated: %d key(s), %d unpopulated leaf/leaves.",
        len(rows),
        int(matrix["unpopulated_leaves"].sum()),
    )
    # Return the assembled validity matrix.
    return matrix


# Refuse to run while any declared gap leaf remains unpopulated.
def assert_no_unpopulated_gaps(
    study_config: Mapping[str, Any], gap_keys: Sequence[str]
) -> None:
    """
    Assert that the declared gap subtrees carry no unpopulated leaves.

    Parameters
    ----------
    study_config : mapping
        The study configuration mapping.
    gap_keys : sequence of str
        The gap keys whose subtrees must be fully populated.

    Returns
    -------
    None
        The function returns nothing when every declared leaf is populated.

    Raises
    ------
    KeyError
        If a requested gap key is absent from the registry.
    ConfigValidationError
        If any declared gap subtree still carries a None-valued leaf, which
        means a reproducibility choice has not yet been made.
    """
    # Read the gap registry node, requiring it to be a mapping.
    registry = study_config.get("reproducibility_gaps")
    # Refuse a configuration whose gap registry is absent.
    if not isinstance(registry, Mapping):
        # A missing registry halts execution-time validation.
        raise KeyError(
            "study_config does not expose 'reproducibility_gaps' as a mapping."
        )
    # Build the authoritative semantic gap audit once for this gate.
    audit: pd.DataFrame = validate_reproducibility_gaps(study_config)
    # Resolve the only mode in which explicit nonblocking gaps may remain.
    exploratory: bool = str(study_config.get("execution_mode", "production")).lower() == "exploratory"
    # Accumulate the full paths that remain production-blocking.
    incomplete: List[str] = []
    # Inspect every requested gap key.
    for key in gap_keys:
        # Refuse a requested key that the registry does not declare.
        if key not in registry:
            # The failure names the missing gap key.
            raise KeyError(f"reproducibility_gaps.{key} is absent.")
        # Read the audited record for this explicitly requested gap.
        row: pd.Series = audit.loc[key]
        # Allow only explicit nonblocking gaps in explicit exploratory mode.
        allowed_incomplete: bool = bool(exploratory and row["nonblocking"])
        # Reject every requested gap that is not execution-ready.
        if not bool(row["execution_ready"]) and not allowed_incomplete:
            # Preserve every unresolved field path in the failure evidence.
            paths: Tuple[str, ...] = tuple(row["unresolved_paths"])
            # Fall back to the gap root if its state/evidence is inconsistent.
            incomplete.extend(paths or (f"reproducibility_gaps.{key}",))
    # Refuse to run while any requested subtree remains unpopulated.
    if incomplete:
        # The failure names every incomplete subtree at once.
        raise ConfigValidationError(
            f"Reproducibility gaps are not execution-ready: {incomplete}. Resolve "
            f"and register each choice before executing the pipeline."
        )
    # Log the satisfied execution-time gate.
    logger.info("execution-time gap gate satisfied for %d key(s).", len(gap_keys))


# Validate the temporal universe map and return its boolean validity matrix.
def validate_temporal_universes(
    study_config: Mapping[str, Any],
) -> pd.DataFrame:
    """
    Validate the nine declared temporal layers against their windows.

    Parameters
    ----------
    study_config : mapping
        The study configuration mapping.

    Returns
    -------
    pandas.DataFrame
        A boolean matrix indexed by layer name with the columns expected,
        observed and matches.

    Raises
    ------
    KeyError
        If the temporal map or any declared layer is absent.
    ConfigValidationError
        If a layer's window diverges from its declared value.
    """
    # Read the temporal map from the series registry.
    temporal = read_dotted_path(
        study_config, "series_code_registry.temporal_universe_by_layer"
    )
    # Refuse a temporal node that is not a mapping.
    if not isinstance(temporal, Mapping):
        # A non-mapping temporal node cannot carry the layer windows.
        raise KeyError(
            "series_code_registry.temporal_universe_by_layer must be a mapping."
        )
    # Accumulate one audit record per declared layer.
    rows: Dict[str, Dict[str, Any]] = {}
    # Iterate over the declared layers in their declaration order.
    for layer, expected in EXPECTED_TEMPORAL_LAYERS.items():
        # Refuse a declared layer that the map does not expose.
        if layer not in temporal:
            # The failure names the absent layer.
            raise KeyError(f"temporal_universe_by_layer.{layer} is absent.")
        # Read the observed window of the layer.
        observed = temporal[layer]
        # Record the expected, observed and match triple.
        rows[layer] = {
            "expected": expected,
            "observed": observed,
            "matches": _values_match(observed, expected),
        }
        # Refuse a divergent window.
        if not rows[layer]["matches"]:
            # The failure names the layer and both windows.
            raise ConfigValidationError(
                f"temporal_universe_by_layer.{layer} must be {expected!r}; "
                f"observed {observed!r}."
            )
    # Assemble the audit matrix with the declared layer order preserved.
    matrix: pd.DataFrame = pd.DataFrame.from_dict(rows, orient="index")
    # Log the completed temporal validation.
    logger.info("temporal universes validated: %d layer(s).", len(rows))
    # Return the assembled validity matrix.
    return matrix


# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Step 3 (continued): Freeze the validated configuration
# -------------------------------------------------------------------------------------------------------------------------------


# Write the validated configuration's fingerprint to the freeze record.
def freeze_config_fingerprint(
    study_config: Mapping[str, Any], freeze_path: Path
) -> str:
    """
    Write the configuration fingerprint to the freeze record.

    Parameters
    ----------
    study_config : mapping
        The validated configuration mapping.
    freeze_path : pathlib.Path
        The destination of the freeze record.

    Returns
    -------
    str
        The hexadecimal fingerprint that was written.

    Raises
    ------
    OSError
        If the freeze record cannot be written.
    """
    # Compute the fingerprint through the ingest-layer definition.
    fingerprint: str = compute_config_fingerprint(study_config)
    # Compose the freeze record payload.
    payload: Dict[str, str] = {
        "config_fingerprint": fingerprint,
        "frozen_utc": current_utc_iso(),
    }
    # Ensure the destination directory exists before writing.
    freeze_path.parent.mkdir(parents=True, exist_ok=True)
    # Write the freeze record with stable indentation and UTF-8 encoding.
    freeze_path.write_text(
        json.dumps(payload, indent=2) + "\n", encoding="utf-8"
    )
    # Log the freeze for the audit trail.
    logger.info("configuration frozen at digest %s.", fingerprint[:16])
    # Return the written fingerprint.
    return fingerprint


# Refuse to proceed when the configuration has changed since validation.
def assert_config_frozen(
    study_config: Mapping[str, Any], freeze_path: Path
) -> None:
    """
    Assert that the configuration still matches its frozen fingerprint.

    Parameters
    ----------
    study_config : mapping
        The configuration mapping presented for execution.
    freeze_path : pathlib.Path
        The freeze record written at validation time.

    Returns
    -------
    None
        The function returns nothing when the fingerprints agree.

    Raises
    ------
    FileNotFoundError
        If no freeze record exists.
    ConfigValidationError
        If the current fingerprint differs from the frozen one, which means
        the configuration was edited after validation.
    """
    # Require the freeze record to exist before comparison.
    if not freeze_path.is_file():
        # Without a record there is nothing to compare against.
        raise FileNotFoundError(
            f"No freeze record found at {freeze_path}; validate the "
            f"configuration before execution."
        )
    # Read the freeze record and extract the stored fingerprint.
    stored: str = str(
        json.loads(freeze_path.read_text(encoding="utf-8"))["config_fingerprint"]
    )
    # Compute the fingerprint of the configuration presented now.
    current: str = compute_config_fingerprint(study_config)
    # Refuse a configuration that changed after validation.
    if stored != current:
        # The failure names both digests so the drift is provable.
        raise ConfigValidationError(
            f"Configuration changed since validation: frozen {stored[:16]} but "
            f"current {current[:16]}. Re-validate before execution."
        )
    # Log the satisfied freeze gate.
    logger.info("configuration freeze verified at digest %s.", stored[:16])


# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Validate the study configuration end to end and return the frozen report.
def orchestrate_study_config_validation(
    study_config: Mapping[str, Any],
    freeze_path: Optional[Path] = None,
) -> ConfigValidationReport:
    """
    Validate the study configuration's integrity and completeness.

    Parameters
    ----------
    study_config : mapping
        The study configuration mapping to validate.
    freeze_path : pathlib.Path, optional
        When provided, the validated configuration's fingerprint is written to
        this path so later stages can refuse a drifted configuration.

    Returns
    -------
    ConfigValidationReport
        The complete report, carrying eight catalog matrices plus the operative
        binding matrix, the
        configuration fingerprint, the validation timestamp and the total
        number of validated rows.

    Raises
    ------
    KeyError
        If any declared section, equation, algorithm, roster or gap key is
        absent; also propagated from the reused section check.
    ConfigValidationError
        If any structural check, acceptance target or internal identity fails.
    """
    # Reuse the ingest-layer section check to prove the nineteen sections exist.
    validate_study_config_sections(study_config)
    # Check concrete implementation bindings before evaluating provenance readiness.
    binding_matrix = validate_configuration_bindings(study_config)
    # Validate the equation registry and capture its validity matrix.
    equation_matrix: pd.DataFrame = validate_equation_registry(study_config)
    # Validate the algorithm registry and capture its validity matrix.
    algorithm_matrix: pd.DataFrame = validate_algorithm_registry(study_config)
    # Audit every declared acceptance target and capture the audit table.
    target_matrix: pd.DataFrame = audit_declared_targets(study_config)
    # Assert that no acceptance target diverged from its declared value.
    assert_audit_passed(target_matrix)
    # Derive the internal consistency identities and capture their table.
    consistency_matrix: pd.DataFrame = internal_consistency_identities(study_config)
    # Assert that no internal identity was violated.
    assert_internal_consistency(consistency_matrix)
    # Validate the series registry and capture its validity matrix.
    series_matrix: pd.DataFrame = validate_series_code_registry(study_config)
    # Validate the country registries and capture their validity matrix.
    country_matrix: pd.DataFrame = validate_country_registries(study_config)
    # Validate the reproducibility gaps and capture their validity matrix.
    gap_matrix: pd.DataFrame = validate_reproducibility_gaps(study_config)
    # Enforce production readiness for every registered reproducibility gap.
    assert_no_unpopulated_gaps(study_config, REQUIRED_GAP_KEYS)
    # Validate the temporal universes and capture their validity matrix.
    temporal_matrix: pd.DataFrame = validate_temporal_universes(study_config)
    # Compute the fingerprint through the ingest-layer definition.
    fingerprint: str = compute_config_fingerprint(study_config)
    # Bind an explicit downstream readiness token to the validated fingerprint.
    readiness_token: str = hashlib.sha256(
        f"study-config-ready:{fingerprint}".encode("utf-8")
    ).hexdigest()
    # Write the freeze record when a destination was provided.
    if freeze_path is not None:
        # Persist the fingerprint so execution can detect later drift.
        freeze_config_fingerprint(study_config, freeze_path)
    # Sum the validated rows across the eight matrices.
    n_rows: int = (
        len(equation_matrix)
        + len(algorithm_matrix)
        + len(target_matrix)
        + len(consistency_matrix)
        + len(series_matrix)
        + len(country_matrix)
        + len(gap_matrix)
        + len(temporal_matrix)
        + len(binding_matrix)
    )
    # Assemble the immutable report from the validated matrices.
    report: ConfigValidationReport = ConfigValidationReport(
        equation_matrix=equation_matrix,
        algorithm_matrix=algorithm_matrix,
        target_matrix=target_matrix,
        consistency_matrix=consistency_matrix,
        series_matrix=series_matrix,
        country_matrix=country_matrix,
        gap_matrix=gap_matrix,
        temporal_matrix=temporal_matrix,
        config_fingerprint=fingerprint,
        validated_utc=current_utc_iso(),
        n_rows_validated=n_rows,
        readiness_token=readiness_token,
        binding_matrix=binding_matrix,
    )
    # Log the completed validation with its totals.
    logger.info(
        "study_config validated: %d check row(s), digest %s.",
        n_rows,
        fingerprint[:16],
    )
    # Return the assembled report.
    return report
````

````python
# Task 3: Data Cleansing and Canonicalization

# ==============================================================================
# Task 3: Data Cleansing and Canonicalization
# ==============================================================================

# Declare the three exclusion mechanisms in their fixed priority order.
EXCLUSION_MECHANISMS: Final[Tuple[str, ...]] = (
    "pattern_violation",
    "denylist",
    "not_in_iso_registry",
)
# Declare the sources cleansed by the first step and their code columns.
COUNTRY_SOURCE_CODE_COLUMNS: Final[Mapping[str, str]] = {
    "wdi_raw_wide": "country_code",
    "wdi_raw_long": "country_code",
    "wgi_raw_long": "country_code",
}
# Declare the minimum retained universe, one above the final panel count.
DEFAULT_UNIVERSE_MINIMUM: Final[int] = 130
# Declare the target missing-cell count of the alternative vintage branch.
DEFAULT_WGI_MISSING_TARGET: Final[int] = 129
# Declare release-year overrides backed by versioned WGI metadata.
WGI_RELEASE_YEAR_METADATA: Final[Mapping[str, int]] = {}
# Declare the four-digit families of the non-gas energy products.
HS_NON_GAS_FAMILIES: Final[Mapping[str, Tuple[str, ...]]] = {
    "Coal": ("2701",),
    "Crude oil": ("2709",),
    "Refined petroleum": ("2710",),
}
# Declare the six-digit codes of the two gas subtypes.
HS_GAS_CODES: Final[Mapping[str, Tuple[str, ...]]] = {
    "LNG": ("271111",),
    "Pipeline gas": ("271121",),
}
# Declare the declared cell-status counts per series from the coverage table.
EXPECTED_CELL_STATUS_COUNTS: Final[Mapping[str, Mapping[str, int]]] = {
    "russian_gas": {"positive": 229, "verified_zero": 733, "missing": 70},
    "world_gas": {"positive": 809, "verified_zero": 185, "missing": 38},
    "russian_non_gas": {"positive": 769, "verified_zero": 224, "missing": 39},
}
# Declare the status vocabulary in its reporting order.
STATUS_VOCABULARY: Final[Tuple[str, ...]] = (
    "positive",
    "verified_zero",
    "missing",
)


# Declare the immutable result of cleansing one country-bearing frame.
@dataclass(frozen=True)
class CleanseResult:
    """
    Immutable result of cleansing one country-bearing source frame.

    Attributes
    ----------
    cleaned : pandas.DataFrame
        Rows whose code survived all three exclusion mechanisms.
    exclusions : pandas.DataFrame
        Removed rows carrying their original index, their code and the single
        mechanism that removed them.
    """
    # Rows whose code survived all three exclusion mechanisms.
    cleaned: pd.DataFrame
    # Removed rows with their code and their single removal mechanism.
    exclusions: pd.DataFrame


# Declare the immutable resolution of the WGI release-vintage ambiguity.
@dataclass(frozen=True)
class WgiVintageResolution:
    """
    Immutable resolution of the WGI release-vintage ambiguity.

    Attributes
    ----------
    panel : pandas.DataFrame
        Rectangular panel with one row per country-year pair.
    ledger : pandas.DataFrame
        One row per country-year pair naming the chosen release, its distance
        and the discarded alternatives.
    alternative_release : str
        The release selected by the missing-count branch.
    alternative_missing_count : int
        The missing-cell count of that release over the compared universe.
    review_required : bool
        True when the two branches disagree on any compared cell.
    notes : tuple of str
        Recorded decisions, including the universe used by the branch.
    """
    # Rectangular panel with one row per country-year pair.
    panel: pd.DataFrame
    # Provenance ledger naming the chosen and discarded releases.
    ledger: pd.DataFrame
    # The release selected by the missing-count branch.
    alternative_release: str
    # The missing-cell count of that release over the compared universe.
    alternative_missing_count: int
    # True when the two branches disagree on any compared cell.
    review_required: bool
    # Recorded decisions of the resolution.
    notes: Tuple[str, ...]


# Declare the immutable report of the complete cleansing pipeline.
@dataclass(frozen=True)
class DataCleanseReport:
    """
    Immutable report of the complete data cleansing pipeline.

    Attributes
    ----------
    frames : dict of str to pandas.DataFrame
        The cleansed frames keyed by source name.
    exclusions : pandas.DataFrame
        The combined exclusion ledger of the country-bearing extracts.
    wgi : WgiVintageResolution
        The vintage resolution of the governance extract.
    hs_selection : dict
        The HS registry recorded as the selected implementation.
    block_a_rejections : pandas.DataFrame
        Block A rows whose product code fell outside every registered family.
    status_matrix : pandas.DataFrame
        The classified cell-status ledger with declared and recomputed status.
    reconciliation : pandas.DataFrame
        The status counts per series against the declared coverage figures.
    n_rows_removed : int
        The number of rows removed across every cleansing operation.
    cleansed_utc : str
        The ISO-8601 UTC timestamp of the cleansing run.
    branch_coverage : mapping of str to int
        The number of cells classified under each documented decision branch
        (positive, verified_zero, missing), included so the freeze carries
        the branch-coverage audit.
    """
    # The cleansed frames keyed by source name.
    frames: Mapping[str, pd.DataFrame]
    # The combined exclusion ledger of the country-bearing extracts.
    exclusions: pd.DataFrame
    # The vintage resolution of the governance extract.
    wgi: WgiVintageResolution
    # The HS registry recorded as the selected implementation.
    hs_selection: Mapping[str, Any]
    # Block A rows whose product code fell outside every registered family.
    block_a_rejections: pd.DataFrame
    # The classified cell-status ledger with declared and recomputed status.
    status_matrix: pd.DataFrame
    # The status counts per series against the declared coverage figures.
    reconciliation: pd.DataFrame
    # The number of rows removed across every cleansing operation.
    n_rows_removed: int
    # The ISO-8601 UTC timestamp of the cleansing run.
    cleansed_utc: str
    # The number of cells classified under each documented decision branch.
    branch_coverage: Mapping[str, int] = field(default_factory=dict)


# Compute a deterministic structural digest of one frame.
def _frame_digest(frame: pd.DataFrame) -> str:
    """
    Return a deterministic SHA-256 digest of a frame's row structure.

    Parameters
    ----------
    frame : pandas.DataFrame
        The frame to digest.

    Returns
    -------
    str
        The hexadecimal digest of the concatenated per-row hash values,
        stable for identical frame content within and across runs under the
        same pandas version.

    Raises
    ------
    TypeError
        If the input is not a pandas DataFrame.
    """
    # Reject a non-frame input before hashing.
    if not isinstance(frame, pd.DataFrame):
        # The digest is defined only over frame structures.
        raise TypeError(
            f"_frame_digest expects a pandas.DataFrame; received "
            f"{type(frame).__name__}."
        )
    # Hash every row deterministically, including the index.
    row_hashes: np.ndarray = pd.util.hash_pandas_object(
        frame, index=True
    ).to_numpy(dtype="uint64")
    # Return the digest of the concatenated row-hash bytes.
    return hashlib.sha256(row_hashes.tobytes()).hexdigest()


def _source_handoff_digest(source_name: str, frame: pd.DataFrame) -> str:
    """Bind region release identity to metadata's cleansed-frame handoff."""
    content_digest = _frame_digest(frame)
    if source_name != "country_metadata_raw":
        return content_digest
    return hashlib.sha256(
        f"{content_digest}:{_region_identity_digest(frame)}".encode("ascii")
    ).hexdigest()


# Declare the immutable cleansed-data bundle consumed by every downstream task.
@dataclass(frozen=True)
class CleansedBundle:
    """
    Immutable cleansed-data bundle consumed by Tasks 4 through 7.

    The bundle restores the structural discipline that finding CF-1
    identified as broken: the eligibility, completion and repair
    orchestrators must consume CLEANSED frames, never the raw ingest bundle.
    The container exposes the same ``frame`` accessor as ``IngestBundle`` so
    downstream call sites keep their reading pattern, and it carries the
    per-frame post-cleanse digests so the handoff itself is
    machine-verifiable.

    Attributes
    ----------
    frames : mapping of str to pandas.DataFrame
        The cleansed frames keyed by source name.
    report : DataCleanseReport
        The embedded cleansing report retained for audit.
    frame_digests : mapping of str to str
        The post-cleanse digest of every frame, keyed by source name.
    study_config : mapping
        The configuration the cleansing ran against, carried for provenance
        and for downstream declared-count wiring.
    config_fingerprint : str
        The configuration digest carried from the ingest bundle.
    cleansed_utc : str
        The ISO-8601 UTC timestamp of the cleansing run.
    """
    # The cleansed frames keyed by source name.
    frames: Mapping[str, pd.DataFrame]
    # The embedded cleansing report retained for audit.
    report: DataCleanseReport
    # The post-cleanse digest of every frame keyed by source name.
    frame_digests: Mapping[str, str]
    # The configuration the cleansing ran against.
    study_config: Mapping[str, Any]
    # The configuration digest carried from the ingest bundle.
    config_fingerprint: str
    # The ISO-8601 UTC timestamp of the cleansing run.
    cleansed_utc: str

    # Expose one cleansed frame by source name, mirroring IngestBundle.
    def frame(self, source_name: str) -> pd.DataFrame:
        """
        Return one cleansed frame by source name.

        Parameters
        ----------
        source_name : str
            The canonical source name.

        Returns
        -------
        pandas.DataFrame
            The cleansed frame.

        Raises
        ------
        KeyError
            If the bundle holds no frame under the requested name.
        """
        # Guard the lookup so a misspelled source fails loudly.
        if source_name not in self.frames:
            # Report the available names to make the failure diagnosable.
            raise KeyError(
                f"Frame {source_name!r} is not in the cleansed bundle; "
                f"available: {sorted(self.frames)}."
            )
        # Return the requested cleansed frame.
        return self.frames[source_name]

    # Verify the handoff digests against the live frames.
    def verify_handoff(self) -> None:
        """
        Assert every frame digest matches its recorded post-cleanse value.

        Returns
        -------
        None
            The method returns nothing on success.

        Raises
        ------
        IngestError
            If any live frame's recomputed digest diverges from the recorded
            post-cleanse digest, which would mean the frame was mutated after
            the bundle was frozen.
        """
        # Walk every recorded digest and compare with a fresh recomputation.
        for name, digest in self.frame_digests.items():
            # Recompute the digest over the live frame.
            live_digest: str = _source_handoff_digest(name, self.frames[name])
            # Refuse a divergence between the recorded and live digests.
            if live_digest != digest:
                # The failure names the source and both digest prefixes.
                raise IngestError(
                    f"Cleansed handoff digest mismatch for {name!r}: recorded "
                    f"{digest[:16]}... against live {live_digest[:16]}...."
                )
        # Log the verified handoff for the audit trail.
        logger.info(
            "cleansed-bundle handoff verified across %d frame digest(s).",
            len(self.frame_digests),
        )


# Assert that a cleansed frame carries no aggregate row and only clean codes.
def assert_cleansed_frame_purity(
    frame: pd.DataFrame,
    code_columns: Sequence[str],
    source_name: str,
    registry: frozenset,
) -> None:
    """
    Assert the pipeline-level purity of one cleansed frame.

    The gate enforces the CF-1 invariant at the pipeline level: zero rows in
    any cleansed frame may match the aggregate exclusion pattern, and the
    country-bearing extracts may carry only codes with positive membership
    in the authoritative ISO-3 registry.

    Parameters
    ----------
    frame : pandas.DataFrame
        The cleansed frame to audit.
    code_columns : sequence of str
        The code columns to audit; absent columns are skipped so the gate
        applies uniformly across sources with different schemas.
    source_name : str
        The canonical source name, used in diagnostics.
    registry : frozenset of str
        The authoritative ISO-3 universe for the membership test. An empty
        frozenset disables the membership test (used for the trade frames,
        which legitimately carry the target code RUS as a partner) while the
        aggregate-pattern refusal always applies.

    Returns
    -------
    None
        The function returns nothing on success.

    Raises
    ------
    IngestError
        If any audited code cell is a World Bank aggregate code, or, when a
        non-empty registry is supplied, if any code falls outside it.
    """
    # Restrict the audit to the declared columns present in the frame.
    present_columns: List[str] = [
        column for column in code_columns if column in frame.columns
    ]
    # Audit every present code column.
    for column in present_columns:
        # Extract the non-missing codes as strings.
        values: pd.Series = frame[column].dropna().astype(str)
        # Detect any surviving World Bank aggregate code.
        aggregate_mask: pd.Series = values.isin(WORLD_BANK_AGGREGATE_CODES)
        # Refuse any aggregate row, the CF-1 corrupting failure mode.
        if bool(aggregate_mask.any()):
            # Report the source, column and offending codes.
            raise IngestError(
                f"{source_name}: column {column!r} carries "
                f"{int(aggregate_mask.sum())} aggregate code(s) after "
                f"cleansing, for example "
                f"{sorted(set(values[aggregate_mask]))[:MAX_REPORTED_OFFENDERS]}."
            )
        # Apply the positive-membership test only when a registry is supplied.
        if registry:
            # Detect any code outside the authoritative universe.
            outside_mask: pd.Series = ~values.isin(registry)
            # Refuse any foreign code in a country-bearing extract.
            if bool(outside_mask.any()):
                # Report the source, column and offending codes.
                raise IngestError(
                    f"{source_name}: column {column!r} carries "
                    f"{int(outside_mask.sum())} code(s) outside the "
                    f"authoritative ISO-3 registry, for example "
                    f"{sorted(set(values[outside_mask]))[:MAX_REPORTED_OFFENDERS]}."
                )
    # Log the passed purity gate when any column was audited.
    if present_columns:
        # Record the audited columns for the trail.
        logger.info(
            "%s: purity gate passed over %d code column(s).",
            source_name,
            len(present_columns),
        )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Step 1: Remove World Bank aggregates and normalise country rows
# -------------------------------------------------------------------------------------------------------------------------------


# Load the authoritative ISO-3 registry from the declared sources.
def load_iso3_registry(
    explicit: Optional[Sequence[str]] = None,
    registry_path: Optional[Path] = None,
) -> frozenset[str]:
    """
    Load the authoritative ISO-3 code registry.

    Parameters
    ----------
    explicit : sequence of str, optional
        An explicit registry of three-letter codes. When supplied it takes
        precedence over every other source.
    registry_path : pathlib.Path, optional
        A text file with one code per line. Blank lines and lines beginning
        with a hash character are ignored.

    Returns
    -------
    frozenset of str
        The validated registry of three-letter codes.

    Raises
    ------
    ValueError
        If an explicit code or a file entry violates the three-letter pattern,
        or if the resulting registry is empty.
    FileNotFoundError
        If a registry path is supplied but does not exist.
    ImportError
        If neither an explicit registry nor a registry file is supplied and
        the pycountry package is not importable, because no silent fallback
        to a weaker check is permitted.
    """
    # Initialise the raw code sequence to the empty list.
    raw_codes: List[str] = []
    # Handle an explicitly supplied registry first.
    if explicit is not None:
        # Record the explicit codes for validation.
        raw_codes = [str(code) for code in explicit]
    # Handle a registry file second.
    elif registry_path is not None:
        # Require the registry file to exist before reading it.
        if not registry_path.is_file():
            # A missing registry file cannot supply codes.
            raise FileNotFoundError(
                f"ISO-3 registry file not found at {registry_path}."
            )
        # Read the file and collect the non-comment, non-blank lines.
        raw_codes = [
            line.strip()
            for line in registry_path.read_text(encoding="utf-8").splitlines()
            if line.strip() and not line.strip().startswith("#")
        ]
    # Fall back to pycountry only when it is importable.
    else:
        # The optional country registry package is not importable in this
        # environment, so the guard's refusal path is the only reachable
        # branch (B.10 structural repair of the emptied try block).
        raise ImportError(
            "No ISO-3 registry is available: supply the explicit registry "
            "or a registry file, or install pycountry."
        )
    # Reject an empty registry, which would remove every row silently.
    if not raw_codes:
        # An empty registry cannot perform a membership test.
        raise ValueError("The ISO-3 registry is empty.")
    # Normalise and validate every code against the three-letter pattern.
    invalid: List[str] = [
        code for code in raw_codes if not ISO3_PATTERN.fullmatch(code)
    ]
    # Reject a registry carrying a malformed code.
    if invalid:
        # The failure names the malformed entries for a precise repair.
        raise ValueError(
            f"ISO-3 registry contains {len(invalid)} malformed code(s), for "
            f"example {invalid[:MAX_REPORTED_OFFENDERS]}."
        )
    # Return the validated registry as an immutable frozen set.
    return frozenset(raw_codes)


# Remove aggregate rows from one country-bearing frame under three mechanisms.
def cleanse_country_frame(
    frame: pd.DataFrame,
    code_column: str,
    source_name: str,
    iso3_registry: frozenset[str],
) -> CleanseResult:
    """
    Remove aggregate rows from a country-bearing frame.

    Parameters
    ----------
    frame : pandas.DataFrame
        The canonicalized frame to cleanse.
    code_column : str
        The column carrying the country codes.
    source_name : str
        The canonical source name, used in diagnostics.
    iso3_registry : frozenset of str
        The authoritative registry of three-letter codes.

    Returns
    -------
    CleanseResult
        The cleaned frame and the exclusion ledger.

    Raises
    ------
    IngestError
        If the frame is empty or lacks the declared code column.
    """
    # Refuse a frame that does not carry the declared code column.
    if code_column not in frame.columns:
        # The cleansing filter cannot run without the code column.
        raise IngestError(
            f"{source_name}: declared code column {code_column!r} is absent."
        )
    # Refuse an empty frame, which would pass the conservation identity.
    if len(frame) == 0:
        # An empty payload is not a cleansing input.
        raise IngestError(f"{source_name}: the frame contains no rows.")
    # Extract the codes as strings for pattern and membership testing.
    codes: pd.Series = frame[code_column].astype(str)
    # Build the syntactic mechanism mask from the compiled pattern.
    pattern_mask: pd.Series = ~codes.map(lambda code: bool(ISO3_PATTERN.fullmatch(code)))
    # Build the denylist mechanism mask from the declared vocabulary.
    denylist_mask: pd.Series = codes.isin(WORLD_BANK_AGGREGATE_CODES)
    # Build the registry mechanism mask from the authoritative registry.
    registry_mask: pd.Series = ~codes.isin(iso3_registry)
    # Compose the three mechanisms in their declared priority order.
    excluded_mask: pd.Series = pattern_mask | denylist_mask | registry_mask
    # Assign each excluded row to exactly one mechanism by priority.
    mechanism: pd.Series = pd.Series(
        np.select(
            [pattern_mask, denylist_mask, registry_mask],
            list(EXCLUSION_MECHANISMS),
            default="",
        ),
        index=frame.index,
    )
    # Slice the cleaned rows as an independent copy.
    cleaned: pd.DataFrame = frame.loc[~excluded_mask].copy()
    # Slice the excluded rows as an independent copy.
    exclusions: pd.DataFrame = frame.loc[excluded_mask].copy()
    # Attach the removal mechanism and the offending code to the ledger.
    exclusions["exclusion_mechanism"] = mechanism.loc[excluded_mask].to_numpy()
    # Attach the offending code under a dedicated column for readability.
    exclusions["excluded_code"] = codes.loc[excluded_mask].to_numpy()
    # Assert the conservation identity over row counts.
    if len(cleaned) + len(exclusions) != len(frame):
        # A conservation failure means a row was lost or duplicated.
        raise IngestError(
            f"{source_name}: conservation failed: {len(cleaned)} cleaned plus "
            f"{len(exclusions)} excluded does not equal {len(frame)}."
        )
    # Assert that the cleaned and excluded indices are disjoint and complete.
    if set(cleaned.index) & set(exclusions.index):
        # Overlapping indices would mean a row was both kept and removed.
        raise IngestError(f"{source_name}: cleaned and excluded indices overlap.")
    # Count the removals per mechanism for the audit log.
    counts: Dict[str, int] = {
        name: int((mechanism.loc[excluded_mask] == name).sum())
        for name in EXCLUSION_MECHANISMS
    }
    # Log the per-mechanism removal counts.
    logger.info(
        "%s: removed %d row(s) by mechanism %s.",
        source_name,
        len(exclusions),
        counts,
    )
    # Return the immutable cleansing result.
    return CleanseResult(cleaned=cleaned, exclusions=exclusions)


# Cleanse every declared country-bearing source and combine the ledgers.
def cleanse_country_sources(
    frames: Mapping[str, pd.DataFrame],
    iso3_registry: frozenset[str],
) -> Tuple[Dict[str, pd.DataFrame], pd.DataFrame]:
    """
    Cleanse the declared country-bearing sources.

    Parameters
    ----------
    frames : mapping of str to pandas.DataFrame
        The canonicalized frames keyed by source name.
    iso3_registry : frozenset of str
        The authoritative registry of three-letter codes.

    Returns
    -------
    tuple
        The cleaned frames keyed by source name, and the combined exclusion
        ledger carrying a source column.

    Raises
    ------
    KeyError
        If a declared source is absent from the frame mapping.
    IngestError
        If any frame fails its cleansing checks.
    """
    # Accumulate the cleaned frames keyed by source name.
    cleaned_frames: Dict[str, pd.DataFrame] = {}
    # Accumulate the per-source ledgers for the combined frame.
    ledgers: List[pd.DataFrame] = []
    # Cleanse each declared source in its declaration order.
    for source_name, code_column in COUNTRY_SOURCE_CODE_COLUMNS.items():
        # Refuse a declared source that the mapping does not carry.
        if source_name not in frames:
            # The failure names the missing source.
            raise KeyError(f"Source {source_name!r} is absent from the frames.")
        # Cleanse the frame under the three mechanisms.
        result: CleanseResult = cleanse_country_frame(
            frames[source_name], code_column, source_name, iso3_registry
        )
        # Record the cleaned frame under its source name.
        cleaned_frames[source_name] = result.cleaned
        # Attach the source name to the ledger for the combined frame.
        ledger: pd.DataFrame = result.exclusions.copy()
        # Record the originating source on every exclusion row.
        ledger["source_name"] = source_name
        # Append the ledger to the accumulator.
        ledgers.append(ledger)
    # Concatenate the per-source ledgers, tolerating an all-empty set.
    combined: pd.DataFrame = (
        pd.concat(ledgers, axis=0) if ledgers else pd.DataFrame()
    )
    # Log the combined removal total.
    logger.info("country cleansing removed %d row(s) in total.", len(combined))
    # Return the cleaned frames and the combined ledger.
    return cleaned_frames, combined


# Assert that the retained universe exceeds the final panel count.
def assert_universe_exceeds_final(
    frame: pd.DataFrame, code_column: str, minimum: int, source_name: str
) -> int:
    """
    Assert that the retained universe exceeds the declared minimum.

    Parameters
    ----------
    frame : pandas.DataFrame
        The cleaned frame whose universe is measured.
    code_column : str
        The column carrying the country codes.
    minimum : int
        The minimum number of distinct codes the frame must retain.
    source_name : str
        The canonical source name, used in diagnostics.

    Returns
    -------
    int
        The number of distinct retained codes.

    Raises
    ------
    IngestError
        If the retained universe falls below the declared minimum.
    """
    # Count the distinct codes retained in the cleaned frame.
    n_unique: int = int(frame[code_column].nunique())
    # Refuse a universe that the cleansing step has reduced below the minimum.
    if n_unique < minimum:
        # The failure names the observed count and the required minimum.
        raise IngestError(
            f"{source_name}: retained universe is {n_unique} code(s), below the "
            f"declared minimum of {minimum}."
        )
    # Log the retained universe size.
    logger.info("%s: retained universe of %d code(s).", source_name, n_unique)
    # Return the observed count for the report.
    return n_unique


# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Step 2: Resolve the WGI release-vintage ambiguity
# -------------------------------------------------------------------------------------------------------------------------------


# Extract the four-digit release year from a release label.
def parse_release_year(label: str) -> int:
    """
    Extract the release year from a release label.

    Parameters
    ----------
    label : str
        The release label, such as a year followed by a description.

    Returns
    -------
    int
        The four-digit year embedded in the label.

    Raises
    ------
    ValueError
        If the label contains no four-digit group.
    """
    # Require a genuine textual release identifier.
    if not isinstance(label, str) or not label.strip():
        # A missing label is an unresolved WGI-release reproducibility gap.
        raise ValueError(
            f"WGI_release_year gap: release label {label!r} is not populated."
        )
    # Resolve exact identifiers through the versioned metadata registry first.
    if label in WGI_RELEASE_YEAR_METADATA:
        # Return the registry-backed year without heuristic parsing.
        return int(WGI_RELEASE_YEAR_METADATA[label])
    # Collect all distinct standalone plausible release-year interpretations.
    candidates: List[str] = list(
        dict.fromkeys(re.findall(r"(?<!\d)(?:19|20)\d{2}(?!\d)", label))
    )
    # Require exactly one interpretation rather than selecting the first token.
    if len(candidates) != 1:
        # Name the reproducibility gap and preserve the original label.
        raise ValueError(
            "WGI_release_year gap: release label "
            f"{label!r} has {len(candidates)} unambiguous year candidates "
            f"{candidates}; register the release metadata explicitly."
        )
    # Return the uniquely parsed year while consumers retain the source label.
    return int(candidates[0])


# Select the nearest release per country-year pair.
def select_release_by_distance(
    frame: pd.DataFrame,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Select the release nearest in time to each observation year.

    Parameters
    ----------
    frame : pandas.DataFrame
        The canonicalized WGI long frame carrying the release label.

    Returns
    -------
    tuple
        The rectangular panel with one row per country-year pair, and the
        provenance ledger naming the chosen and discarded releases. The panel
        carries a country-year multi-index whose levels duplicate the two
        columns; consumers performing label-based grouping or merging must
        reset the index first to avoid ambiguous label resolution.

    Raises
    ------
    KeyError
        If the frame lacks a required column.
    ValueError
        If a release label carries no year.
    """
    # Refuse a frame lacking any column the selection requires.
    for column in ("country_code", "year", "source_release"):
        # The failure names the absent column.
        if column not in frame.columns:
            # Without these columns the selection cannot run.
            raise KeyError(f"WGI frame lacks the required column {column!r}.")
    # Reject raw aggregate/malformed country rows: selection consumes cleansing.
    country_values: pd.Series = frame["country_code"]
    # Identify any noncanonical or aggregate country identifier.
    invalid_country: pd.Series = ~country_values.map(
        lambda value: isinstance(value, str)
        and ISO3_PATTERN.fullmatch(value) is not None
        and value not in WORLD_BANK_AGGREGATE_CODES
    )
    # Refuse selection on an uncleaned WGI frame.
    if bool(invalid_country.any()):
        # Report bounded original offenders.
        raise IngestError(
            "WGI release selection requires the cleaned WGI frame; offending "
            f"country code(s): "
            f"{country_values[invalid_country].head(MAX_REPORTED_OFFENDERS).tolist()}."
        )
    # Copy the frame so the caller's object is never mutated.
    working: pd.DataFrame = frame.copy()
    # Parse every release label into its integer year.
    working["_release_year"] = working["source_release"].map(parse_release_year)
    # Compute the absolute distance between the release and the observation.
    working["_distance"] = (working["_release_year"] - working["year"]).abs()
    # Sort by the registered nearest-distance/earlier-release/label tie rule.
    working = working.sort_values(
        [
            "country_code",
            "year",
            "_distance",
            "_release_year",
            "source_release",
        ],
        kind="stable",
    )
    # Keep the first row of each country-year group as the selected release.
    selected: pd.DataFrame = working.drop_duplicates(
        subset=["country_code", "year"], keep="first"
    ).copy()
    # Assemble the provenance ledger from the grouped comparison, collecting
    # every candidate release so the discards can be derived from the choice.
    ledger: pd.DataFrame = working.groupby(
        ["country_code", "year"], observed=True
    ).agg(
        chosen_release=("source_release", "first"),
        chosen_distance=("_distance", "first"),
        candidate_releases=(
            "source_release",
            lambda values: tuple(sorted(set(str(value) for value in values))),
        ),
        candidate_count=("source_release", "size"),
        tied_minimum_distance=(
            "_distance",
            lambda values: int(
                (pd.Series(values).to_numpy() == np.min(pd.Series(values).to_numpy())).sum()
            )
            > 1,
        ),
        candidate_nonmissing_values=(
            "rule_of_law",
            lambda values: int(pd.Series(values).notna().sum()),
        ) if "rule_of_law" in working.columns else ("year", "size"),
    ).reset_index()
    # Derive the discarded releases as the candidates other than the choice.
    ledger["discarded_releases"] = [
        [release for release in candidates if release != str(chosen)]
        for candidates, chosen in zip(
            ledger["candidate_releases"], ledger["chosen_release"]
        )
    ]
    # Retain the full candidate tuple so discarded coverage remains auditable.
    # Drop the helper columns from the selected panel.
    selected = selected.drop(columns=["_release_year", "_distance"])
    # Flatten the panel index so the country-year pair is addressable as
    # ordinary columns: carrying the pair as BOTH index levels and column
    # labels is a pandas ambiguity that breaks every downstream column-based
    # pivot (pandas raises "label is both an index level and a column").
    panel: pd.DataFrame = selected.reset_index(drop=True)
    # Log the selection outcome.
    logger.info(
        "WGI release selection resolved %d country-year pair(s).", len(panel)
    )
    # Return the panel and its provenance ledger.
    return panel, ledger


# Select the single release whose missing count approaches the target.
def select_release_by_missing_target(
    frame: pd.DataFrame,
    target_missing: int = DEFAULT_WGI_MISSING_TARGET,
    country_mask: Optional[Sequence[str]] = None,
) -> Tuple[pd.DataFrame, str, int]:
    """
    Select the release whose missing count is closest to the target.

    Parameters
    ----------
    frame : pandas.DataFrame
        The marker-canonicalized WGI long frame.
    target_missing : int, optional
        The declared target missing-cell count over the compared universe.
    country_mask : sequence of str, optional
        The countries of the compared universe. When absent, every country in
        the frame is compared and the caller is expected to record that fact.

    Returns
    -------
    tuple
        The panel built from the chosen release, the chosen release label and
        its observed missing count.

    Raises
    ------
    KeyError
        If the frame lacks a required column.
    ValueError
        If the frame declares no release at all.
    """
    # Refuse a frame lacking any column the branch requires.
    for column in ("country_code", "year", "source_release", "rule_of_law"):
        # The failure names the absent column.
        if column not in frame.columns:
            # Without these columns the branch cannot run.
            raise KeyError(f"WGI frame lacks the required column {column!r}.")
    # Fix the canonical country scope before candidate outcomes are examined.
    countries: Tuple[str, ...] = (
        tuple(sorted(set(str(code) for code in country_mask)))
        if country_mask is not None
        else tuple(sorted(set(frame["country_code"].astype(str))))
    )
    # Fix the common temporal scope from the input frame before ranking.
    years: Tuple[int, ...] = tuple(sorted(set(frame["year"].astype(int))))
    # Restrict the comparison to the predeclared country scope.
    scoped: pd.DataFrame = frame[
        frame["country_code"].astype(str).isin(set(countries))
        & frame["year"].astype(int).isin(set(years))
    ].copy()
    # Refuse a scope that contains no rows.
    if len(scoped) == 0:
        # An empty scope cannot rank the releases.
        raise ValueError("The compared universe contains no WGI rows.")
    # Construct the exact fixed country-year product used for every candidate.
    required_index: pd.MultiIndex = pd.MultiIndex.from_product(
        [countries, years], names=["country_code", "year"]
    )
    # Accumulate full-grid missing and coverage counts per candidate release.
    candidate_rows: List[Dict[str, Any]] = []
    # Evaluate every candidate under precisely the same labelled grid.
    for release, release_frame in scoped.groupby("source_release", observed=True):
        # Refuse duplicate keys within a release before reindexing.
        if bool(release_frame.duplicated(["country_code", "year"]).any()):
            # Candidate coverage cannot be interpreted with duplicated keys.
            raise IngestError(
                f"WGI release {release!r} duplicates country-year keys."
            )
        # Reindex the candidate values over the fixed comparison universe.
        candidate_grid: pd.Series = release_frame.set_index(
            ["country_code", "year"]
        )["rule_of_law"].reindex(required_index)
        # Count missing values, including completely absent country-year rows.
        missing_count: int = int(candidate_grid.isna().sum())
        # Count source rows absent from the candidate's coverage.
        observed_keys: set[Tuple[str, int]] = set(
            zip(
                release_frame["country_code"].astype(str),
                release_frame["year"].astype(int),
            )
        )
        # Record every ranking component before selecting a candidate.
        candidate_rows.append(
            {
                "source_release": str(release),
                "missing_count": missing_count,
                "coverage_missing": len(required_index) - len(observed_keys),
                "distance_to_target": abs(missing_count - int(target_missing)),
                "release_year": parse_release_year(str(release)),
            }
        )
    # Assemble the disclosed sensitivity table for deterministic ranking.
    sensitivity: pd.DataFrame = pd.DataFrame(candidate_rows).sort_values(
        ["distance_to_target", "release_year", "source_release"], kind="stable"
    ).reset_index(drop=True)
    # Refuse a frame that declares no release.
    if sensitivity.empty:
        # Without releases the branch cannot select anything.
        raise ValueError("The WGI frame declares no release label.")
    # Select by target distance, then earlier metadata year and release label.
    chosen_release: str = str(sensitivity.iloc[0]["source_release"])
    # Record the observed missing count of the chosen release.
    observed_missing: int = int(sensitivity.iloc[0]["missing_count"])
    # Build the panel from the rows of the chosen release.
    panel: pd.DataFrame = frame[
        frame["source_release"].astype(str) == chosen_release
    ].copy()
    # Attach the complete alternative ranking without changing tuple arity.
    panel.attrs["release_sensitivity"] = sensitivity.copy(deep=True)
    # Record the fixed scope and deterministic tie rule in panel provenance.
    panel.attrs["selection_scope"] = {
        "countries": countries,
        "years": years,
        "target_missing": int(target_missing),
        "tie_rule": "earlier_release_year_then_lexical_label",
    }
    # Log the alternative-branch outcome.
    logger.info(
        "WGI alternative branch selected release %r with %d missing cell(s).",
        chosen_release,
        observed_missing,
    )
    # Return the panel, the chosen release and its missing count.
    return panel, chosen_release, observed_missing


# Assert the rectangularity of the resolved WGI panel.
def assert_wgi_rectangular(panel: pd.DataFrame) -> None:
    """
    Assert that the resolved panel is rectangular on country and year.

    Parameters
    ----------
    panel : pandas.DataFrame
        The resolved WGI panel.

    Returns
    -------
    None
        The function returns nothing when the panel is rectangular.

    Raises
    ------
    IngestError
        If the panel carries a duplicated country-year pair or an interior
        gap within a country's covered year span.
    """
    # Refuse an empty substitute panel before any product construction.
    if panel.empty:
        # An empty frame is not a rectangular scientific panel.
        raise IngestError("WGI panel is empty; no country-year grid exists.")
    # Require the two key columns explicitly.
    missing_columns: List[str] = [
        column for column in ("country_code", "year") if column not in panel
    ]
    # Refuse a malformed panel before label coercion.
    if missing_columns:
        # Report all missing key columns together.
        raise IngestError(
            f"WGI panel lacks key column(s) {missing_columns}."
        )
    # Extract the country-year pairs as a frame for grouped checking, dropping
    # the panel index so that the grouping labels are unambiguous.
    pairs: pd.DataFrame = (
        panel[["country_code", "year"]]
        .astype({"country_code": str, "year": int})
        .reset_index(drop=True)
    )
    # Refuse a panel carrying a duplicated country-year pair.
    if bool(pairs.duplicated(subset=["country_code", "year"]).any()):
        # The failure names the duplicated pair count.
        raise IngestError(
            f"WGI panel carries "
            f"{int(pairs.duplicated(subset=['country_code', 'year']).sum())} "
            f"duplicated country-year pair(s)."
        )
    # Define the common temporal universe from the panel-wide year support.
    countries: Tuple[str, ...] = tuple(sorted(set(pairs["country_code"])))
    # Require a contiguous common annual universe, not country-specific spans.
    observed_years: Tuple[int, ...] = tuple(sorted(set(pairs["year"])))
    # Construct the contiguous temporal universe between global endpoints.
    common_years: Tuple[int, ...] = tuple(
        range(observed_years[0], observed_years[-1] + 1)
    )
    # Build the explicit required country-by-year product.
    expected_keys: set[Tuple[str, int]] = set(
        itertools.product(countries, common_years)
    )
    # Build the actual labelled key set without considering indicator values.
    actual_keys: set[Tuple[str, int]] = set(
        zip(pairs["country_code"], pairs["year"])
    )
    # Compute missing and extra keys separately for diagnosis.
    missing_keys: List[Tuple[str, int]] = sorted(expected_keys - actual_keys)
    # Extra keys can only arise from invalid coercion/domain drift but are audited.
    extra_keys: List[Tuple[str, int]] = sorted(actual_keys - expected_keys)
    # Reject any unequal country span or omitted common year.
    if missing_keys or extra_keys:
        # Keep missing indicator values distinct: only absent rows fail here.
        raise IngestError(
            "WGI panel is not rectangular on the common temporal universe: "
            f"missing keys {missing_keys[:MAX_REPORTED_OFFENDERS]}, extra keys "
            f"{extra_keys[:MAX_REPORTED_OFFENDERS]}."
        )
    # Log the verified rectangularity.
    logger.info("WGI panel verified rectangular across %d pair(s).", len(pairs))


# Resolve the WGI vintage under both branches and flag a disagreement.
def resolve_wgi_vintages(
    frame: pd.DataFrame,
    country_mask: Optional[Sequence[str]] = None,
    target_missing: int = DEFAULT_WGI_MISSING_TARGET,
) -> WgiVintageResolution:
    """
    Resolve the WGI vintage under the declared rule and its alternative.

    Parameters
    ----------
    frame : pandas.DataFrame
        The marker-canonicalized WGI long frame.
    country_mask : sequence of str, optional
        The countries of the compared universe for the alternative branch.
    target_missing : int, optional
        The declared target missing-cell count for the alternative branch.

    Returns
    -------
    WgiVintageResolution
        The resolved panel, its provenance ledger, the alternative-branch
        summary, the review flag and the recorded notes.

    Raises
    ------
    KeyError, ValueError, IngestError
        Propagated from the selection and rectangularity callables.
    """
    # Select the nearest release per country-year pair under the default rule.
    panel, ledger = select_release_by_distance(frame)
    # Assert the rectangularity of the working panel.
    assert_wgi_rectangular(panel)
    # Run the alternative branch over the same canonicalized frame.
    alt_panel, alt_release, alt_missing = select_release_by_missing_target(
        frame, target_missing=target_missing, country_mask=country_mask
    )
    # Outer-join keys, releases and values so coverage disagreements are visible.
    comparison: pd.DataFrame = panel[
        ["country_code", "year", "source_release", "rule_of_law"]
    ].reset_index(drop=True).merge(
        alt_panel[
            ["country_code", "year", "source_release", "rule_of_law"]
        ].reset_index(drop=True),
        on=["country_code", "year"],
        how="outer",
        suffixes=("_default", "_alt"),
        indicator=True,
        validate="one_to_one",
    )
    # Mark candidate-coverage disagreement independently of missing values.
    comparison["coverage_disagreement"] = comparison["_merge"] != "both"
    # Mark differing release identities on keys present in both branches.
    comparison["release_disagreement"] = (
        (comparison["_merge"] == "both")
        & (
            comparison["source_release_default"].astype(str)
            != comparison["source_release_alt"].astype(str)
        )
    )
    # Compare selected values with null-aware exact semantics.
    both_missing: pd.Series = comparison["rule_of_law_default"].isna() & comparison[
        "rule_of_law_alt"
    ].isna()
    # Mark any substantive value disagreement on common keys.
    comparison["value_disagreement"] = (
        (comparison["_merge"] == "both")
        & ~both_missing
        & ~comparison["rule_of_law_default"].eq(comparison["rule_of_law_alt"])
    )
    # Mark the complete review condition for each compared key.
    comparison["review_required"] = comparison[
        ["coverage_disagreement", "release_disagreement", "value_disagreement"]
    ].any(axis=1)
    # Count every key requiring review under coverage/release/value comparison.
    disagreements: int = int(comparison["review_required"].sum())
    # Merge the complete disagreement audit into the default selection ledger.
    ledger = ledger.merge(
        comparison.drop(columns=["_merge"]),
        on=["country_code", "year"],
        how="outer",
        validate="one_to_one",
    )
    # Compose the notes recording the comparison universe and the outcome.
    notes: List[str] = [
        (
            "alternative branch compared over the supplied country mask."
            if country_mask is not None
            else "alternative branch compared over the widest available universe."
        ),
        f"alternative branch chose release {alt_release!r} with "
        f"{alt_missing} missing cell(s) against a target of {target_missing}.",
        f"the two branches disagree on {disagreements} key(s) by coverage, "
        "release identity, or value.",
    ]
    # Log the resolution outcome and the disagreement count.
    logger.info(
        "WGI vintage resolution: default panel %d pair(s), alternative %r, "
        "disagreements %d.",
        len(panel),
        alt_release,
        disagreements,
    )
    # Assemble the immutable resolution.
    return WgiVintageResolution(
        panel=panel,
        ledger=ledger,
        alternative_release=alt_release,
        alternative_missing_count=alt_missing,
        review_required=disagreements > 0,
        notes=tuple(notes),
    )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Step 3: Cleanse the Comtrade blocks and the cell-status ledger
# -------------------------------------------------------------------------------------------------------------------------------


# Load the HS family registry from the study configuration.
def load_hs_family_registry(
    study_config: Mapping[str, Any],
) -> Mapping[str, Tuple[str, ...]]:
    """
    Load the declared HS code families from the configuration.

    Parameters
    ----------
    study_config : mapping
        The study configuration mapping.

    Returns
    -------
    mapping of str to tuple of str
        The acceptable code prefixes keyed by product description.

    Raises
    ------
    KeyError
        If the configuration does not declare the candidate code registry for
        either Comtrade block.
    """
    # Read the raw data schemas node, requiring it to be a mapping.
    schemas = study_config.get("raw_data_schemas")
    # Refuse a configuration that does not declare the schemas.
    if not isinstance(schemas, Mapping):
        # Without the schemas node the registry cannot be recovered.
        raise KeyError("study_config.raw_data_schemas must be a mapping.")
    # Accumulate the families from both blocks.
    families: Dict[str, Tuple[str, ...]] = {}
    # Read the declared candidate codes of each block in turn.
    for block, expected_names in (
        ("comtrade_block_a_raw", tuple(HS_NON_GAS_FAMILIES)),
        ("comtrade_block_b_raw", tuple(HS_GAS_CODES)),
    ):
        # Read the block schema, requiring it to be a mapping.
        block_schema = schemas.get(block)
        # Refuse a block that is not declared as a mapping.
        if not isinstance(block_schema, Mapping):
            # The failure names the absent block schema.
            raise KeyError(f"study_config.raw_data_schemas.{block} is absent.")
        # Read the candidate code mapping of the block.
        candidates = block_schema.get("candidate_hs_codes")
        # Refuse a block that does not declare its candidate codes.
        if not isinstance(candidates, Mapping):
            # The failure names the absent candidate registry.
            raise KeyError(
                f"study_config.raw_data_schemas.{block}.candidate_hs_codes is "
                f"absent."
            )
        # Record the declared families for every expected product name.
        for name in expected_names:
            # Refuse a product name that the registry does not declare.
            if name not in candidates:
                # The failure names the missing product family.
                raise KeyError(
                    f"candidate_hs_codes does not declare the family {name!r}."
                )
            # Record the declared codes of the family as a tuple.
            families[name] = tuple(str(code) for code in candidates[name])
    # Log the loaded family registry.
    logger.info("HS family registry loaded for %d product(s).", len(families))
    # Return the declared family registry.
    return families


# Apply the HS registry to Block A and route unknown codes to a ledger.
def apply_hs_registry_to_block_a(
    frame: pd.DataFrame, families: Mapping[str, Tuple[str, ...]]
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Retain Block A rows whose product code belongs to a registered family.

    Parameters
    ----------
    frame : pandas.DataFrame
        The Block A frame carrying the arriving product code.
    families : mapping of str to tuple of str
        The acceptable code prefixes keyed by product description.

    Returns
    -------
    tuple
        The retained frame and the rejection ledger.

    Raises
    ------
    IngestError
        If the frame lacks the description or code column, or if a row's
        description declares no family at all.
    """
    # Refuse a frame lacking either column the registry requires.
    for column in ("product_description", "product_code"):
        # The failure names the absent column.
        if column not in frame.columns:
            # Without these columns the registry cannot be applied.
            raise IngestError(f"Block A frame lacks the column {column!r}.")
    # Copy the frame so the caller's object is never mutated.
    working: pd.DataFrame = frame.copy()
    # Convert the description and code columns to strings for matching.
    descriptions: pd.Series = working["product_description"].astype(str)
    # Convert the arriving product codes to strings for prefix matching.
    codes: pd.Series = working["product_code"].astype(str)
    # Refuse a description that declares no registered family.
    unknown_descriptions: List[str] = sorted(
        set(descriptions) - set(families)
    )
    # Halt when an undeclared description appears, since the registry defines
    # every admissible product of this block.
    if unknown_descriptions:
        # The failure names the undeclared descriptions.
        raise IngestError(
            f"Block A carries undeclared product description(s): "
            f"{unknown_descriptions[:MAX_REPORTED_OFFENDERS]}."
        )
    # Build the retained mask by prefix comparison against the family codes.
    retained_mask: pd.Series = pd.Series(
        [
            any(code.startswith(prefix) for prefix in families[description])
            for description, code in zip(descriptions, codes)
        ],
        index=working.index,
        dtype=bool,
    )
    # Slice the retained rows as an independent copy.
    retained: pd.DataFrame = working.loc[retained_mask].copy()
    # Slice the rejected rows as an independent copy.
    rejected: pd.DataFrame = working.loc[~retained_mask].copy()
    # Attach the offending code and its description to the rejection ledger.
    rejected["rejected_code"] = codes.loc[~retained_mask].to_numpy()
    # Attach the registered families that the code failed to match.
    rejected["registered_prefixes"] = [
        families[description]
        for description in descriptions.loc[~retained_mask]
    ]
    # Assert the conservation identity over row counts.
    if len(retained) + len(rejected) != len(working):
        # A conservation failure means a row was lost or duplicated.
        raise IngestError("Block A registry conservation failed.")
    # Log the registry application outcome.
    logger.info(
        "HS registry applied to Block A: %d retained, %d rejected.",
        len(retained),
        len(rejected),
    )
    # Return the retained frame and the rejection ledger.
    return retained, rejected


# Record the selected HS implementation for the audit ledger.
def record_hs_selection(
    families: Mapping[str, Tuple[str, ...]]
) -> Mapping[str, Any]:
    """
    Record the applied HS registry as the selected implementation.

    Parameters
    ----------
    families : mapping of str to tuple of str
        The applied code prefixes keyed by product description.

    Returns
    -------
    mapping
        The audit entry naming the selected implementation for every product
        family and stating that an alternative mapping would be a robustness
        branch rather than a replacement.
    """
    # Compose the audit entry with the declared selection fields.
    entry: Dict[str, Any] = {
        "selected_implementation": {
            name: list(codes) for name, codes in families.items()
        },
        "status": "selected",
        "alternative_policy": (
            "Any alternative mapping must be registered as a robustness branch; "
            "it never replaces this selection."
        ),
    }
    # Log the recorded selection.
    logger.info("HS selection recorded for %d families.", len(families))
    # Return the audit entry.
    return entry


# Validate the declared cell-status classification against the rule.
def validate_cell_status_classification(frame: pd.DataFrame) -> pd.DataFrame:
    """
    Recompute the deterministic cell status and compare with the declared.

    Parameters
    ----------
    frame : pandas.DataFrame
        The cell-status ledger carrying the declared status and its inputs.

    Returns
    -------
    pandas.DataFrame
        The frame with a recomputed status column and a consistency flag;
        the decision-branch coverage is verified to reconcile to the row
        count and logged so alternative-branch usage is auditable.

    Raises
    ------
    IngestError
        If a required column is absent, the declared status contradicts the
        deterministic rule for any cell, or the branch-coverage counts do
        not reconcile to the validated row count.
    """
    # Refuse a frame lacking any column the rule requires.
    for column in (
        "value_usd",
        "retry_succeeded",
        "hs_report_confirmed",
        "final_status",
    ):
        # The failure names the absent column.
        if column not in frame.columns:
            # Without these columns the rule cannot be applied.
            raise IngestError(f"Cell-status ledger lacks the column {column!r}.")
    # Copy the frame so the caller's object is never mutated.
    working: pd.DataFrame = frame.copy()
    # Read the reported value as a float series.
    values: pd.Series = pd.to_numeric(working["value_usd"], errors="coerce")
    # Read the retry flag as a boolean series.
    retry_ok: pd.Series = working["retry_succeeded"].astype(int) == 1
    # Read the HS-report confirmation flag as a boolean series.
    hs_ok: pd.Series = working["hs_report_confirmed"].astype(int) == 1
    # Apply the first clause of the rule: a positive reported value.
    is_positive: pd.Series = values > 0
    # Apply the second clause: a reported zero with both confirmations.
    is_verified_zero: pd.Series = (values == 0.0) & retry_ok & hs_ok
    # Compose the verdict, defaulting every remaining cell to missing.
    recomputed: pd.Series = pd.Series(
        np.select(
            [is_positive, is_verified_zero],
            ["positive", "verified_zero"],
            default="missing",
        ),
        index=working.index,
    )
    # Record the recomputed verdict and the consistency flag on the frame.
    working["recomputed_status"] = recomputed
    # Compare the recomputed verdict with the declared status as strings.
    declared: pd.Series = working["final_status"].astype(str)
    # Record whether each cell's declared status matches the rule.
    consistency: pd.Series = declared == recomputed
    # Attach the consistency flag to the frame.
    working["classification_consistent"] = consistency
    # Select the cells whose declared status contradicts the rule.
    contradictions = working.loc[~consistency]
    # Refuse a ledger carrying any contradiction.
    if len(contradictions):
        # Extract the offending keys for the failure message.
        offenders: List[Tuple[Any, ...]] = list(
            contradictions[
                ["reporter_code", "year", "series", "final_status"]
            ]
            .head(MAX_REPORTED_OFFENDERS)
            .itertuples(index=False, name=None)
        )
        # The failure names the contradiction count and its first instances.
        raise IngestError(
            f"Cell-status ledger carries {len(contradictions)} declared "
            f"status(es) contradicting the verified-zero rule, for example "
            f"{offenders}."
        )
    # Count the cells classified under each documented decision branch.
    branch_coverage: Dict[str, int] = {
        # Branch one: a strictly positive reported value.
        "positive": int(is_positive.sum()),
        # Branch two: a reported zero with both confirmations.
        "verified_zero": int(is_verified_zero.sum()),
        # Branch three: every remaining cell, which the rule defaults to missing.
        "missing": int((~is_positive & ~is_verified_zero).sum()),
    }
    # Verify that the branch counts reconcile to the validated row count.
    if sum(branch_coverage.values()) != len(working):
        # A divergence would mean the rule left a cell unclassified.
        raise IngestError(
            f"Cell-status branch coverage does not reconcile: "
            f"{branch_coverage} against {len(working)} row(s)."
        )
    # Log the verified classification with its branch coverage.
    logger.info(
        "cell-status classification verified across %d cell(s); branch "
        "coverage %s.",
        len(working),
        branch_coverage,
    )
    # Return the classified ledger.
    return working


# Reconcile the status counts against the declared coverage figures.
def reconcile_cell_status_counts(
    classified: pd.DataFrame,
    expected: Optional[Mapping[str, Mapping[str, int]]] = None,
) -> pd.DataFrame:
    """
    Count the recomputed statuses and compare with the declared figures.

    Parameters
    ----------
    classified : pandas.DataFrame
        The classified cell-status ledger.
    expected : mapping of str to mapping of str to int, optional
        The declared counts per series and status. When absent, the observed
        counts are reported without a declared comparison.

    Returns
    -------
    pandas.DataFrame
        A table over the UNION of the observed and declared (series, status)
        key sets with the observed count (zero-filled on the declared side),
        the declared count, the two-sided provenance flags
        (declared_present, observed_present) and the match flag.

    Raises
    ------
    IngestError
        If the classified ledger lacks the series or status column, or if any
        two-sided comparison mismatches when a declared mapping is supplied:
        a declared count with zero observed rows and an observed status with
        no declaration both refuse the reconciliation.
    """
    # Refuse a ledger lacking either grouping column.
    for column in ("series", "recomputed_status"):
        # The failure names the absent column.
        if column not in classified.columns:
            # Without these columns the counts cannot be grouped.
            raise IngestError(f"Classified ledger lacks the column {column!r}.")
    # Count the recomputed statuses per series and status as a frame so that
    # the declared and match columns can be attached.
    counts: pd.DataFrame = (
        classified.groupby(["series", "recomputed_status"], observed=True)
        .size()
        .to_frame(name="observed")
    )
    # Normalise the observed key columns to strings for a stable join.
    observed_frame: pd.DataFrame = counts.reset_index()
    # Cast the series column so category and object dtypes merge cleanly.
    observed_frame["series"] = observed_frame["series"].astype(str)
    # Cast the status column for the same reason.
    observed_frame["recomputed_status"] = observed_frame[
        "recomputed_status"
    ].astype(str)
    # Build the declared key set in full when an expectation was supplied.
    declared_rows: List[Tuple[str, str, int]] = []
    # Track whether any declared expectation was supplied at all.
    any_expected: bool = expected is not None
    # Walk the declared mapping so EVERY declared (series, status) key exists.
    if expected is not None:
        # Iterate over the declared series in their mapping order.
        for series, status_map in expected.items():
            # Iterate over the declared statuses of the series.
            for status, count in status_map.items():
                # Record the declared key with its integer count.
                declared_rows.append(
                    (str(series), str(status), int(count))
                )
    # Assemble the declared key frame, empty but correctly typed when absent.
    declared_frame: pd.DataFrame = pd.DataFrame(
        declared_rows, columns=["series", "recomputed_status", "declared"]
    )
    # Outer-join the observed and declared key sets so the comparison covers
    # the UNION: a declared key with zero observed rows and an observed key
    # with no declaration are both surfaced instead of passing silently.
    union: pd.DataFrame = observed_frame.merge(
        declared_frame,
        on=["series", "recomputed_status"],
        how="outer",
        indicator=True,
    )
    # Zero-fill the observed count for declared keys with no observed rows.
    union["observed"] = union["observed"].fillna(0).astype("int64")
    # Record whether the declared side contributed each compared pair.
    union["declared_present"] = union["_merge"].isin(["both", "right_only"])
    # Record whether the observed side contributed each compared pair.
    union["observed_present"] = union["_merge"].isin(["both", "left_only"])
    # Drop the merge indicator now that both provenance flags exist.
    union = union.drop(columns=["_merge"])
    # Sort deterministically by series and status for a stable table.
    union = union.sort_values(
        ["series", "recomputed_status"], kind="stable"
    ).reset_index(drop=True)
    # Compute the two-sided match flag when an expectation was supplied.
    if any_expected:
        # A declared zero matches an absent observed category after zero-fill.
        union["matches"] = (
            union["declared_present"]
            & (union["observed"] == union["declared"])
        )
        # Select the mismatched pairs for the failure report.
        mismatches: pd.DataFrame = union.loc[~union["matches"]]
        # Refuse a reconciliation carrying any two-sided mismatch.
        if len(mismatches):
            # Extract the offending pairs for the failure message.
            offenders: List[Tuple[str, str, int, Optional[int], str]] = [
                (
                    str(row.series),
                    str(row.recomputed_status),
                    int(row.observed),
                    int(row.declared) if pd.notna(row.declared) else None,
                    (
                        "undeclared_observed_category"
                        if not bool(row.declared_present)
                        else "numeric_count_mismatch"
                    ),
                )
                for row in mismatches.head(MAX_REPORTED_OFFENDERS).itertuples(
                    index=False
                )
            ]
            # The failure names the mismatch count and its first instances.
            raise IngestError(
                f"Cell-status reconciliation carries {len(mismatches)} "
                f"two-sided mismatch(es) against the declared figures, for "
                "example (series, status, observed, declared, reason): "
                f"{offenders}."
            )
    # Compute the match flag column for the unasserted fixture mode.
    else:
        # Without an expectation the match flag is undefined.
        union["matches"] = None
    # Count the pairs actually compared against a declared figure.
    n_compared: int = (
        int(union["declared_present"].sum()) if any_expected else 0
    )
    # Log the observed and compared counts.
    logger.info(
        "cell-status reconciliation: %d series-status row(s), %d compared "
        "against a declared figure.",
        len(union),
        n_compared,
    )
    # Return the reconciliation table over the two-sided union.
    return union


# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the cleansing pipeline over the ingest bundle.
def orchestrate_data_cleansing(
    bundle: IngestBundle,
    study_config: Mapping[str, Any],
    iso3_registry: Optional[Sequence[str]] = None,
    iso3_registry_path: Optional[Path] = None,
    country_mask: Optional[Sequence[str]] = None,
    universe_minimum: int = DEFAULT_UNIVERSE_MINIMUM,
) -> CleansedBundle:
    """
    Cleanse and canonicalize the ingested frames.

    Parameters
    ----------
    bundle : IngestBundle
        The immutable ingest bundle produced by Task 1.
    study_config : mapping
        The study configuration, which must declare the HS candidate registry.
    iso3_registry : sequence of str, optional
        An explicit authoritative ISO-3 registry.
    iso3_registry_path : pathlib.Path, optional
        A registry file with one code per line.
    country_mask : sequence of str, optional
        The compared universe for the alternative vintage branch.
    universe_minimum : int, optional
        The minimum retained universe for the wide extract.

    Returns
    -------
    CleansedBundle
        The cleansed frames keyed by source name, the embedded
        DataCleanseReport (frames, ledgers, vintage resolution, HS selection,
        classified status ledger, reconciliation table, branch coverage and
        summary counts) and the per-frame post-cleanse digests. Downstream
        tasks MUST consume this bundle; the raw ingest bundle must never
        reach the eligibility, completion or repair stages (invariant CF-1).

    Raises
    ------
    ImportError
        If no ISO-3 registry can be loaded.
    IngestError
        If any cleansing, selection or classification check fails.
    KeyError
        If a required source or configuration node is absent.
    """
    # Load the authoritative ISO-3 registry before touching any frame.
    registry: frozenset[str] = load_iso3_registry(
        explicit=iso3_registry, registry_path=iso3_registry_path
    )
    # Cleanse the country-bearing extracts under the three mechanisms.
    country_frames, exclusions = cleanse_country_sources(
        bundle.frames, registry
    )
    # Start from defensive copies of all ten ingested sources.
    cleaned_frames: Dict[str, pd.DataFrame] = {
        name: frame.copy(deep=True) for name, frame in bundle.frames.items()
    }
    # Replace each country-bearing source with its once-cleansed counterpart.
    cleaned_frames.update(country_frames)
    # Assert that the wide extract retains a universe above the minimum.
    assert_universe_exceeds_final(
        cleaned_frames["wdi_raw_wide"],
        COUNTRY_SOURCE_CODE_COLUMNS["wdi_raw_wide"],
        universe_minimum,
        "wdi_raw_wide",
    )
    # Canonicalize markers on the already-cleaned WGI extract before selection.
    wgi_canonical, _marker_notes = canonicalize_missing_markers(
        cleaned_frames["wgi_raw_long"], "wgi_raw_long"
    )
    # Resolve the release vintage under the declared rule and its alternative.
    wgi_resolution: WgiVintageResolution = resolve_wgi_vintages(
        wgi_canonical, country_mask=country_mask
    )
    # Replace the WGI entry with the resolved rectangular panel.
    cleaned_frames["wgi_raw_long"] = wgi_resolution.panel
    # Load the declared HS families from the configuration.
    families: Mapping[str, Tuple[str, ...]] = load_hs_family_registry(study_config)
    # Apply the registry to Block A and route unknown codes to the ledger.
    block_a_retained, block_a_rejected = apply_hs_registry_to_block_a(
        cleaned_frames["comtrade_block_a_raw"], families
    )
    # Replace the Block A entry with the registry-cleaned frame.
    cleaned_frames["comtrade_block_a_raw"] = block_a_retained
    # Record the selected HS implementation for the audit ledger.
    hs_selection: Mapping[str, Any] = record_hs_selection(families)
    # Validate the declared cell-status classification against the rule.
    status_matrix: pd.DataFrame = validate_cell_status_classification(
        cleaned_frames["comtrade_cell_status_raw"]
    )
    # Replace the cell-status entry with the classified ledger.
    cleaned_frames["comtrade_cell_status_raw"] = status_matrix
    # Reconcile the recomputed status counts against the declared figures.
    reconciliation: pd.DataFrame = reconcile_cell_status_counts(
        status_matrix, expected=EXPECTED_CELL_STATUS_COUNTS
    )
    # Sum the removed rows across the exclusion ledger and the rejections.
    n_rows_removed: int = len(exclusions) + len(block_a_rejected)
    # Compose the ISO-8601 UTC timestamp of the cleansing run.
    cleansed_utc: str = current_utc_iso()
    # Compute the branch-coverage counts from the classified ledger.
    branch_coverage: Dict[str, int] = {
        # One entry per recomputed status with its validated cell count.
        str(status): int(count)
        for status, count in status_matrix[
            "recomputed_status"
        ].value_counts().items()
    }
    # Assemble the immutable cleansing report with the branch-coverage audit.
    report: DataCleanseReport = DataCleanseReport(
        frames=cleaned_frames,
        exclusions=exclusions,
        wgi=wgi_resolution,
        hs_selection=hs_selection,
        block_a_rejections=block_a_rejected,
        status_matrix=status_matrix,
        reconciliation=reconciliation,
        n_rows_removed=n_rows_removed,
        cleansed_utc=cleansed_utc,
        branch_coverage=branch_coverage,
    )
    # Enforce the pipeline-level purity gate over every cleansed frame.
    purity_targets: Mapping[str, Tuple[Tuple[str, ...], frozenset]] = {
        **{
            name: ((column,), registry)
            for name, column in COUNTRY_SOURCE_CODE_COLUMNS.items()
        },
        "comtrade_block_a_raw": (
            ("reporter_code", "partner_code"),
            frozenset(),
        ),
        "comtrade_block_b_raw": (
            ("reporter_code", "partner_code"),
            frozenset(),
        ),
    }
    # Audit only sources whose scientific contract forbids aggregate codes.
    for source_name, (code_columns, purity_universe) in purity_targets.items():
        # Select the exact cleansed frame represented in the outgoing bundle.
        frame: pd.DataFrame = cleaned_frames[source_name]
        # Assert that no aggregate row and no foreign code survived.
        assert_cleansed_frame_purity(
            frame, code_columns, source_name, purity_universe
        )
    # Compute the post-cleanse digest of every cleansed frame.
    frame_digests: Dict[str, str] = {
        # One digest per source over the live cleansed frame.
        name: _source_handoff_digest(name, frame)
        for name, frame in cleaned_frames.items()
    }
    # Prove one-to-one correspondence between every frame and recorded digest.
    if set(frame_digests) != set(cleaned_frames):
        # Missing or surplus digest keys make the handoff unauditable.
        raise IngestError(
            "Cleansed frame/digest key correspondence is incomplete."
        )
    # Prove all ten declared source names survived the cleansing handoff.
    missing_sources: List[str] = [
        name for name in DECLARED_SOURCES if name not in cleaned_frames
    ]
    # Refuse a bundle that silently drops any source.
    if missing_sources:
        # Report all missing names together.
        raise IngestError(
            f"Cleansing omitted declared source(s) {missing_sources}."
        )
    # Assemble the cleansed bundle that downstream tasks MUST consume.
    cleansed_bundle: CleansedBundle = CleansedBundle(
        frames=cleaned_frames,
        report=report,
        frame_digests=frame_digests,
        study_config=bundle.study_config,
        config_fingerprint=bundle.config_fingerprint,
        cleansed_utc=cleansed_utc,
    )
    # Verify the handoff digests once before the bundle leaves the module.
    cleansed_bundle.verify_handoff()
    # Log the completed cleansing summary.
    logger.info(
        "data cleansing complete: %d row(s) removed, %d frame(s) cleansed, "
        "vintage review required: %s.",
        n_rows_removed,
        len(cleaned_frames),
        wgi_resolution.review_required,
    )
    # Return the cleansed bundle carrying the embedded report.
    return cleansed_bundle
````

````python
# Task 4: Sample Eligibility Screening (Algorithm C1)

# ==============================================================================
# Task 4: Sample Eligibility Screening (Algorithm C1)
# ==============================================================================

# Declare the manuscript's final panel count for the production gate (B-11).
DECLARED_EXPECTED_FINAL: Final[int] = 129
# Declare the manuscript's pre-critical-window count for the production gate.
DECLARED_EXPECTED_OTHERWISE_ELIGIBLE: Final[int] = 131


# Declare the four outcomes with their exact WDI series codes.
OUTCOME_CODES: Final[Mapping[str, str]] = {
    "CPI": "FP.CPI.TOTL",
    "InflationRate": "FP.CPI.TOTL.ZG",
    "RealGDPpc": "NY.GDP.PCAP.KD",
    "RealGDPpcGrowth": "NY.GDP.PCAP.KD.ZG",
}
# Declare the six directly retrieved controls with their series codes.
WDI_CONTROL_CODES: Final[Mapping[str, str]] = {
    "OilRents": "NY.GDP.PETR.RT.ZS",
    "TradeOpenness": "NE.TRD.GNFS.ZS",
    "TermsOfTrade": "TT.PRI.MRCH.XD.WD",
    "InvestmentRate": "NE.GDI.TOTL.ZS",
    "GovConsumption": "NE.CON.GOVT.ZS",
    "PopulationGrowth": "SP.POP.GROW",
}
# Declare the two fuel-share primitives whose union defines the balance.
FUEL_IMPORT_CODE: Final[str] = "TM.VAL.FUEL.ZS.UN"
# Declare the fuel-export primitive whose union defines the balance.
FUEL_EXPORT_CODE: Final[str] = "TX.VAL.FUEL.ZS.UN"
# Declare the dependent variable name of the derived fuel-share balance.
FUEL_BALANCE_NAME: Final[str] = "FuelShareBalance"
# Declare the name of the governance control column in the WGI extract.
RULE_OF_LAW_NAME: Final[str] = "RuleOfLaw"
# Declare the column name of the governance estimate in the WGI extract.
RULE_OF_LAW_COLUMN: Final[str] = "rule_of_law"
# Declare the whole-period coverage window of the eligibility screen.
WHOLE_PERIOD_WINDOW: Final[Tuple[int, int]] = (2000, 2024)
# Declare the critical outcome window of the eligibility screen.
CRITICAL_WINDOW: Final[Tuple[int, int]] = (2019, 2024)
# Declare the whole-period ceiling on any outcome's raw missingness.
OUTCOME_MISSING_MAX: Final[float] = 0.35
# Declare the ceiling above which a control counts as highly missing.
CONTROL_HIGH_MISSING_MAX: Final[float] = 0.25
# Declare how many controls may exceed the high-missingness ceiling.
CONTROL_HIGH_MISSING_COUNT_MAX: Final[int] = 1
# Declare the ceiling on any single control's raw missingness.
CONTROL_SINGLE_MISSING_MAX: Final[float] = 0.65
# Declare the critical-window ceiling on the two principal outcomes.
CRITICAL_OUTCOME_MISSING_MAX: Final[float] = 0.25
# Declare the ordering of the four eligibility rules.
RULE_ORDER: Final[Tuple[str, ...]] = (
    "outcome_missing",
    "control_count",
    "control_single",
    "critical_window",
)
# Declare the countries the critical-window screen is expected to remove.
DOCUMENTED_EXCLUSIONS: Final[Mapping[str, int]] = {"SYR": 5, "SDN": 2}


# Declare the immutable pair of missingness matrices.
@dataclass(frozen=True)
class MissingnessMatrices:
    """
    Immutable missingness matrices of the eligibility screen.

    Attributes
    ----------
    whole_period : pandas.DataFrame
        Raw missingness rates over the whole period, indexed by country and
        columned by the four outcomes, the seven measurable controls and the
        derived fuel-share balance.
    critical_rates : pandas.DataFrame
        Raw missingness rates over the critical window, indexed by country and
        columned by the two principal outcomes.
    critical_counts : pandas.DataFrame
        Missing cell counts over the critical window, with the same index and
        columns as the rate matrix.
    """
    # Raw missingness rates over the whole period.
    whole_period: pd.DataFrame
    # Raw missingness rates over the critical window.
    critical_rates: pd.DataFrame
    # Missing cell counts over the critical window.
    critical_counts: pd.DataFrame


# Declare the immutable result of the eligibility screen.
@dataclass(frozen=True)
class EligibilityScreen:
    """
    Immutable result of the eligibility screen.

    Attributes
    ----------
    mask : pandas.Series
        Boolean series indexed on the universe, true for retained countries.
    exclusion_ledger : pandas.DataFrame
        One row per excluded country with its rule, trigger and detail.
    missingness : MissingnessMatrices
        The matrices the screen was computed from.
    rule_summary : pandas.DataFrame
        One row per rule with its removal count and share of the universe.
    fingerprint_sha256 : str
        Digest of the canonical serialization of the mask and the ledger.
    screened_utc : str
        ISO-8601 UTC timestamp of the screen.
    notes : tuple of str
        Recorded decisions, including whether the count assertions ran.
    """
    # Boolean series indexed on the universe.
    mask: pd.Series
    # One row per excluded country with its rule, trigger and detail.
    exclusion_ledger: pd.DataFrame
    # The matrices the screen was computed from.
    missingness: MissingnessMatrices
    # One row per rule with its removal count and share.
    rule_summary: pd.DataFrame
    # Digest of the canonical serialization of the mask and the ledger.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the screen.
    screened_utc: str
    # Recorded decisions of the screen.
    notes: Tuple[str, ...]
    # Raw retained-panel coverage checked before any completion.
    coverage_audit: pd.DataFrame = field(default_factory=pd.DataFrame)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Step 1: Compute raw missingness rates on the pre-completion universe
# -------------------------------------------------------------------------------------------------------------------------------


# Build the country-by-year value grid of one indicator over one window.
def _value_grid(
    frame: pd.DataFrame,
    value_column: str,
    universe: Sequence[str],
    window: Tuple[int, int],
    indicator_code: Optional[str] = None,
) -> pd.DataFrame:
    """
    Build a full country-by-year value grid for one series.

    Parameters
    ----------
    frame : pandas.DataFrame
        The long frame carrying the series.
    value_column : str
        The column holding the values.
    universe : sequence of str
        The country universe the grid must cover.
    window : tuple of int
        The inclusive year window the grid must cover.
    indicator_code : str, optional
        When supplied, the frame is filtered to this indicator code.

    Returns
    -------
    pandas.DataFrame
        A grid indexed by the sorted universe and columned by every year of
        the window, with cells absent from the frame represented as missing.

    Raises
    ------
    IngestError
        If the frame lacks the country or year column, or if a duplicate
        country-year pair cannot be resolved deterministically.
    """
    # Refuse a frame lacking the column required for the grid index.
    for column in ("country_code", "year"):
        # The failure names the absent column.
        if column not in frame.columns:
            # Without these columns the grid cannot be built.
            raise IngestError(f"Long frame lacks the column {column!r}.")
    # Restrict the frame to the requested indicator when one is supplied.
    subset: pd.DataFrame = (
        frame[frame["indicator_code"].astype(str) == indicator_code].copy()
        if indicator_code is not None
        else frame.copy()
    )
    # Restrict the subset to the window years.
    subset = subset[
        (subset["year"] >= int(window[0])) & (subset["year"] <= int(window[1]))
    ]
    # Refuse a frame lacking the requested value column before any pivot.
    if value_column not in subset.columns:
        # Without the value column the grid has nothing to place in cells.
        raise IngestError(f"Long frame lacks the value column {value_column!r}.")
    # Detect duplicate country-year pairs before pivoting so the documented
    # refusal is enforced rather than silently aggregated (remediation B-10).
    duplicate_mask: pd.Series = subset.duplicated(
        subset=["country_code", "year"], keep=False
    )
    # Refuse a duplicate pair, whose value has no deterministic resolution.
    if bool(duplicate_mask.any()):
        # Scope the offending rows for the failure message.
        offenders: pd.DataFrame = subset.loc[
            duplicate_mask, ["country_code", "year"]
        ]
        # Render the first offenders as tuples for the message.
        rendered: List[Tuple[str, int]] = list(
            offenders.head(MAX_REPORTED_OFFENDERS).itertuples(
                index=False, name=None
            )
        )
        # Raise with the count and the first offending pairs.
        raise IngestError(
            f"Duplicate country-year pairs in {value_column!r}: "
            f"{len(offenders)} row(s), for example {rendered}."
        )
    # Pivot to a country-by-year grid; uniqueness is guaranteed by the scan.
    grid: pd.DataFrame = subset.pivot(
        index="country_code", columns="year", values=value_column
    )
    # Reindex the grid to the full universe and the full year range.
    full_index: pd.Index = pd.Index(sorted(str(code) for code in universe),
                                    name="country_code")
    # Compose the full ascending year range of the window.
    full_columns: List[int] = list(range(int(window[0]), int(window[1]) + 1))
    # Reindex so absent rows and columns become missing cells.
    grid = grid.reindex(index=full_index, columns=full_columns)
    # Return the full grid.
    return grid


# Compute the per-country missingness rate of one indicator over one window.
def compute_indicator_missingness(
    frame: pd.DataFrame,
    indicator_code: str,
    universe: Sequence[str],
    window: Tuple[int, int],
) -> pd.Series:
    """
    Compute the per-country raw missingness rate of one indicator.

    Parameters
    ----------
    frame : pandas.DataFrame
        The long frame carrying the indicator.
    indicator_code : str
        The series code the rate is computed for.
    universe : sequence of str
        The country universe.
    window : tuple of int
        The inclusive year window.

    Returns
    -------
    pandas.Series
        The missingness rate per country, indexed on the sorted universe.

    Raises
    ------
    IngestError
        Propagated from the grid builder.
    """
    # Build the full value grid of the indicator.
    grid: pd.DataFrame = _value_grid(
        frame, "value", universe, window, indicator_code=indicator_code
    )
    # Reduce the missing indicator to a per-country rate.
    rates: pd.Series = grid.isna().mean(axis=1)
    # Name the resulting series after the indicator code.
    rates.name = indicator_code
    # Return the per-country rate.
    return rates


# Compute the per-country missing cell count of one indicator.
def compute_indicator_missing_count(
    frame: pd.DataFrame,
    indicator_code: str,
    universe: Sequence[str],
    window: Tuple[int, int],
) -> pd.Series:
    """
    Compute the per-country raw missing cell count of one indicator.

    Parameters
    ----------
    frame : pandas.DataFrame
        The long frame carrying the indicator.
    indicator_code : str
        The series code the count is computed for.
    universe : sequence of str
        The country universe.
    window : tuple of int
        The inclusive year window.

    Returns
    -------
    pandas.Series
        The missing cell count per country, indexed on the sorted universe.

    Raises
    ------
    IngestError
        Propagated from the grid builder.
    """
    # Build the full value grid of the indicator.
    grid: pd.DataFrame = _value_grid(
        frame, "value", universe, window, indicator_code=indicator_code
    )
    # Reduce the missing indicator to a per-country integer count.
    counts: pd.Series = grid.isna().sum(axis=1).astype("int64")
    # Name the resulting series after the indicator code.
    counts.name = indicator_code
    # Return the per-country count.
    return counts


# Compute the per-country missingness rate of the governance control.
def compute_wgi_missingness(
    frame: pd.DataFrame,
    universe: Sequence[str],
    window: Tuple[int, int],
) -> pd.Series:
    """
    Compute the per-country raw missingness rate of the rule of law.

    Parameters
    ----------
    frame : pandas.DataFrame
        The WGI long frame carrying the governance estimate.
    universe : sequence of str
        The country universe.
    window : tuple of int
        The inclusive year window.

    Returns
    -------
    pandas.Series
        The missingness rate per country, indexed on the sorted universe.

    Raises
    ------
    IngestError
        If the frame lacks the governance column.
    """
    # Refuse a frame lacking the governance estimate column.
    if RULE_OF_LAW_COLUMN not in frame.columns:
        # Without the estimate the governance rate cannot be computed.
        raise IngestError(
            f"WGI frame lacks the column {RULE_OF_LAW_COLUMN!r}."
        )
    # Build the full value grid of the governance estimate.
    grid: pd.DataFrame = _value_grid(
        frame, RULE_OF_LAW_COLUMN, universe, window, indicator_code=None
    )
    # Reduce the missing indicator to a per-country rate.
    rates: pd.Series = grid.isna().mean(axis=1)
    # Name the resulting series after the governance control.
    rates.name = RULE_OF_LAW_NAME
    # Return the per-country rate.
    return rates


# Derive the fuel-share balance's raw missingness from its two primitives.
def derive_fuel_balance_missingness(
    frame: pd.DataFrame,
    universe: Sequence[str],
    window: Tuple[int, int],
) -> pd.Series:
    """
    Derive the balance's raw missingness as the union of its primitives.

    Parameters
    ----------
    frame : pandas.DataFrame
        The long frame carrying both fuel-share primitives.
    universe : sequence of str
        The country universe.
    window : tuple of int
        The inclusive year window.

    Returns
    -------
    pandas.Series
        The balance's raw missingness rate per country, indexed on the sorted
        universe. A cell is missing when either primitive is missing, which is
        exactly the condition under which the balance is undefined.

    Raises
    ------
    IngestError
        Propagated from the grid builder.
    """
    # Build the full value grid of the import primitive.
    import_grid: pd.DataFrame = _value_grid(
        frame, "value", universe, window, indicator_code=FUEL_IMPORT_CODE
    )
    # Build the full value grid of the export primitive.
    export_grid: pd.DataFrame = _value_grid(
        frame, "value", universe, window, indicator_code=FUEL_EXPORT_CODE
    )
    # Compose the union of the two missing indicators.
    missing_union: pd.DataFrame = import_grid.isna() | export_grid.isna()
    # Reduce the union to a per-country rate.
    rates: pd.Series = missing_union.mean(axis=1)
    # Name the resulting series after the derived control.
    rates.name = FUEL_BALANCE_NAME
    # Return the per-country rate.
    return rates


# Assemble the whole-period and critical-window missingness matrices.
def build_missingness_matrices(
    wdi_frame: pd.DataFrame,
    wgi_frame: pd.DataFrame,
    universe: Sequence[str],
    whole_period: Tuple[int, int] = WHOLE_PERIOD_WINDOW,
    critical_window: Tuple[int, int] = CRITICAL_WINDOW,
) -> MissingnessMatrices:
    """
    Assemble the missingness matrices of the eligibility screen.

    Parameters
    ----------
    wdi_frame : pandas.DataFrame
        The cleansed WDI long frame.
    wgi_frame : pandas.DataFrame
        The cleansed WGI long frame.
    universe : sequence of str
        The country universe.
    whole_period : tuple of int, optional
        The inclusive whole-period window.
    critical_window : tuple of int, optional
        The inclusive critical window.

    Returns
    -------
    MissingnessMatrices
        The whole-period rate matrix and the critical-window rate and count
        matrices.

    Raises
    ------
    IngestError
        Propagated from the rate, count and union computations.
    """
    # Assemble the whole-period columns in their declared order.
    columns: Dict[str, pd.Series] = {}
    # Compute the rate of every outcome.
    for name, code in OUTCOME_CODES.items():
        # Record the outcome rate under its declared name.
        columns[name] = compute_indicator_missingness(
            wdi_frame, code, universe, whole_period
        )
    # Compute the rate of every directly retrieved control.
    for name, code in WDI_CONTROL_CODES.items():
        # Record the control rate under its declared name.
        columns[name] = compute_indicator_missingness(
            wdi_frame, code, universe, whole_period
        )
    # Compute the rate of the governance control from its own extract.
    columns[RULE_OF_LAW_NAME] = compute_wgi_missingness(
        wgi_frame, universe, whole_period
    )
    # Derive the rate of the eighth control from its two primitives.
    columns[FUEL_BALANCE_NAME] = derive_fuel_balance_missingness(
        wdi_frame, universe, whole_period
    )
    # Assemble the whole-period rate matrix with the country index preserved.
    whole_matrix: pd.DataFrame = pd.DataFrame(columns)
    # Restrict the critical matrices to the two principal outcomes.
    critical_columns: Dict[str, pd.Series] = {}
    # Record the critical-window rate of each principal outcome.
    for name in ("CPI", "RealGDPpc"):
        # Record the critical rate of the outcome.
        critical_columns[name] = compute_indicator_missingness(
            wdi_frame, OUTCOME_CODES[name], universe, critical_window
        )
    # Assemble the critical-window rate matrix.
    critical_rates: pd.DataFrame = pd.DataFrame(critical_columns)
    # Restrict the critical count matrix to the two principal outcomes.
    critical_count_columns: Dict[str, pd.Series] = {}
    # Record the critical-window count of each principal outcome.
    for name in ("CPI", "RealGDPpc"):
        # Record the critical count of the outcome.
        critical_count_columns[name] = compute_indicator_missing_count(
            wdi_frame, OUTCOME_CODES[name], universe, critical_window
        )
    # Assemble the critical-window count matrix.
    critical_counts: pd.DataFrame = pd.DataFrame(critical_count_columns)
    # Log the assembled matrix shapes.
    logger.info(
        "missingness matrices assembled: whole period %d x %d, critical %d x %d.",
        whole_matrix.shape[0],
        whole_matrix.shape[1],
        critical_rates.shape[0],
        critical_rates.shape[1],
    )
    # Return the immutable matrix bundle.
    return MissingnessMatrices(
        whole_period=whole_matrix,
        critical_rates=critical_rates,
        critical_counts=critical_counts,
    )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Step 2: Apply the four eligibility rules
# -------------------------------------------------------------------------------------------------------------------------------


# Apply the four eligibility rules and return the mask with its ledger.
def apply_eligibility_rules(
    missingness: MissingnessMatrices,
    universe: Sequence[str],
    critical_window: Tuple[int, int] = CRITICAL_WINDOW,
    policy_audit: Optional[pd.DataFrame] = None,
) -> Tuple[pd.Series, pd.DataFrame]:
    """
    Apply the four eligibility rules in their declared order.

    Parameters
    ----------
    missingness : MissingnessMatrices
        The missingness matrices of the screen.
    universe : sequence of str
        The country universe.
    critical_window : tuple of int, optional
        The inclusive critical window, used to derive the permitted count.
    policy_audit : pandas.DataFrame, optional
        Treatment coding passed through the full screen boundary solely for
        lineage auditing. It is never admitted to the scientific rule inputs.

    Returns
    -------
    tuple
        The boolean mask indexed on the sorted universe, and the exclusion
        ledger with one row per excluded country.

    Raises
    ------
    IngestError
        If a rule column is absent from the matrices, or if the ledger and
        mask fail their correspondence assertion.
    """
    # Sort the universe so the attribution order is deterministic.
    ordered: List[str] = sorted(str(code) for code in universe)
    # Read the whole-period rate matrix.
    whole: pd.DataFrame = missingness.whole_period
    # Read the critical-window count matrix.
    critical: pd.DataFrame = missingness.critical_counts
    # Declare the outcome columns in their declaration order.
    outcome_columns: List[str] = list(OUTCOME_CODES)
    # Declare the control columns, including the derived balance.
    control_columns: List[str] = list(WDI_CONTROL_CODES) + [
        RULE_OF_LAW_NAME,
        FUEL_BALANCE_NAME,
    ]
    # Refuse a matrix that does not expose the declared columns.
    for column in outcome_columns + control_columns:
        # The failure names the absent column.
        if column not in whole.columns:
            # Without the column the corresponding rule cannot run.
            raise IngestError(f"Missingness matrix lacks the column {column!r}.")
    # Derive the permitted missing cells in the critical window from the
    # declared threshold and the number of years in the window.
    n_years: int = int(critical_window[1]) - int(critical_window[0]) + 1
    # Compute the integer ceiling of permitted missing cells.
    permitted_missing: int = int(np.floor(CRITICAL_OUTCOME_MISSING_MAX * n_years))
    # Initialise the accumulator of removed countries in removal order.
    removed: List[str] = []
    # Initialise the ledger rows.
    ledger_rows: List[Dict[str, Any]] = []
    # Rule one: exclude a country whose worst outcome exceeds the ceiling.
    for country in ordered:
        # Compute the country's worst outcome rate.
        worst_outcome: float = float(whole.loc[country, outcome_columns].max())
        # Exclude the country when the ceiling is exceeded.
        if worst_outcome > OUTCOME_MISSING_MAX:
            # Record the removal under the first rule.
            removed.append(country)
            # Append the ledger row with the trigger value.
            ledger_rows.append(
                {
                    "country_code": country,
                    "rule_triggered": RULE_ORDER[0],
                    "trigger_value": worst_outcome,
                    "detail": (
                        f"worst outcome rate {worst_outcome:.4f} exceeds "
                        f"{OUTCOME_MISSING_MAX:.2f}"
                    ),
                }
            )
    # Rule two: exclude a country with too many highly missing controls.
    for country in [code for code in ordered if code not in removed]:
        # Count the controls above the high-missingness ceiling.
        n_high: int = int(
            (whole.loc[country, control_columns] > CONTROL_HIGH_MISSING_MAX).sum()
        )
        # Exclude the country when more controls exceed the ceiling than allowed.
        if n_high > CONTROL_HIGH_MISSING_COUNT_MAX:
            # Record the removal under the second rule.
            removed.append(country)
            # Append the ledger row with the trigger value.
            ledger_rows.append(
                {
                    "country_code": country,
                    "rule_triggered": RULE_ORDER[1],
                    "trigger_value": float(n_high),
                    "detail": (
                        f"{n_high} control(s) exceed {CONTROL_HIGH_MISSING_MAX:.2f}, "
                        f"above the permitted {CONTROL_HIGH_MISSING_COUNT_MAX}"
                    ),
                }
            )
    # Rule three: exclude a country with any control above the single ceiling.
    for country in [code for code in ordered if code not in removed]:
        # Compute the country's worst control rate.
        worst_control: float = float(whole.loc[country, control_columns].max())
        # Exclude the country when the single ceiling is exceeded.
        if worst_control > CONTROL_SINGLE_MISSING_MAX:
            # Record the removal under the third rule.
            removed.append(country)
            # Append the ledger row with the trigger value.
            ledger_rows.append(
                {
                    "country_code": country,
                    "rule_triggered": RULE_ORDER[2],
                    "trigger_value": worst_control,
                    "detail": (
                        f"worst control rate {worst_control:.4f} exceeds "
                        f"{CONTROL_SINGLE_MISSING_MAX:.2f}"
                    ),
                }
            )
    # Snapshot the labelled set immediately before the critical-window rule.
    pre_critical_retained: Tuple[str, ...] = tuple(
        code for code in ordered if code not in removed
    )
    # Rule four: exclude a country whose principal outcome is too incomplete.
    for country in [code for code in ordered if code not in removed]:
        # Compute the country's worst critical missing cell count.
        worst_count: int = int(critical.loc[country, ["CPI", "RealGDPpc"]].max())
        # Exclude the country when the permitted count is exceeded.
        if worst_count > permitted_missing:
            # Record the removal under the fourth rule.
            removed.append(country)
            # Append the ledger row with the derived ceiling in the detail.
            ledger_rows.append(
                {
                    "country_code": country,
                    "rule_triggered": RULE_ORDER[3],
                    "trigger_value": float(worst_count),
                    "detail": (
                        f"{worst_count} missing cell(s) of {n_years} exceeds the "
                        f"permitted {permitted_missing} derived from "
                        f"{CRITICAL_OUTCOME_MISSING_MAX:.2f}"
                    ),
                }
            )
    # Build the mask as the universe minus the removed countries.
    mask: pd.Series = pd.Series(
        [code not in set(removed) for code in ordered],
        index=pd.Index(ordered, name="country_code"),
        name="retained_in_final_panel",
    )
    # Assemble the exclusion ledger with one row per removed country.
    ledger: pd.DataFrame = pd.DataFrame(
        ledger_rows,
        columns=["country_code", "rule_triggered", "trigger_value", "detail"],
    )
    # Assert that the ledger and the mask describe the same decision.
    if set(ledger["country_code"]) != set(removed):
        # A divergence means a removal was not recorded.
        raise IngestError("Exclusion ledger does not match the removal set.")
    # Assert that the excluded country set equals the mask's false positions.
    if set(mask[~mask].index) != set(removed):
        # A divergence between the mask and the ledger is a defect.
        raise IngestError("Eligibility mask does not match the exclusion ledger.")
    # Attach the pre-critical set and full reasons for the sequential count gate.
    mask.attrs["pre_critical_retained"] = pre_critical_retained
    # Preserve a defensive copy so later mutation cannot rewrite the evidence.
    mask.attrs["exclusion_ledger"] = ledger.copy(deep=True)
    # Hash only the allowed scientific inputs into a stable lineage record.
    scientific_payload: Dict[str, Any] = {
        "universe": ordered,
        "whole_period": _frame_digest(missingness.whole_period),
        "critical_rates": _frame_digest(missingness.critical_rates),
        "critical_counts": _frame_digest(missingness.critical_counts),
        "critical_window": tuple(int(value) for value in critical_window),
    }
    # Record the raw-coverage lineage independently of treatment coding.
    mask.attrs["scientific_input_lineage"] = hashlib.sha256(
        json.dumps(scientific_payload, sort_keys=True).encode("utf-8")
    ).hexdigest()
    # Consume policy coding only into a separate audit digest at the boundary.
    mask.attrs["policy_audit_digest"] = (
        _frame_digest(policy_audit) if policy_audit is not None else None
    )
    # Log the removal outcome by rule.
    counts: Dict[str, int] = {
        rule: int((ledger["rule_triggered"] == rule).sum()) for rule in RULE_ORDER
    }
    # Emit the per-rule removal counts.
    logger.info("eligibility removals by rule: %s.", counts)
    # Return the mask and the ledger.
    return mask, ledger


# Summarise the rule application for the screen report.
def summarize_rule_application(
    ledger: pd.DataFrame, universe_size: int
) -> pd.DataFrame:
    """
    Summarise the removals of each eligibility rule.

    Parameters
    ----------
    ledger : pandas.DataFrame
        The exclusion ledger.
    universe_size : int
        The number of countries in the universe.

    Returns
    -------
    pandas.DataFrame
        A table indexed by the four rules with the removal count and the share
        of the universe each rule removed.

    Raises
    ------
    IngestError
        If the universe size is not positive.
    """
    # Refuse a non-positive universe, which cannot be a real screen input.
    if universe_size <= 0:
        # The share column would be undefined.
        raise IngestError("Universe size must be positive to summarise rules.")
    # Count the removals attributed to each rule in declaration order.
    counts: Dict[str, int] = {
        rule: int((ledger["rule_triggered"] == rule).sum()) for rule in RULE_ORDER
    }
    # Assemble the summary with the share of the universe per rule.
    summary: pd.DataFrame = pd.DataFrame(
        {
            "removals": [counts[rule] for rule in RULE_ORDER],
            "share_of_universe": [counts[rule] / universe_size for rule in RULE_ORDER],
        },
        index=pd.Index(RULE_ORDER, name="rule"),
    )
    # Log the summarised removals.
    logger.info("rule summary assembled across %d rule(s).", len(summary))
    # Return the summary table.
    return summary


# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Step 3: Freeze the eligibility mask and prove treatment blindness
# -------------------------------------------------------------------------------------------------------------------------------


# Compute the canonical fingerprint of the mask and its ledger.
def freeze_eligibility_mask(
    mask: pd.Series, ledger: pd.DataFrame
) -> str:
    """
    Compute the canonical fingerprint of the eligibility decision.

    Parameters
    ----------
    mask : pandas.Series
        The boolean eligibility mask indexed on the universe.
    ledger : pandas.DataFrame
        The exclusion ledger.

    Returns
    -------
    str
        The hexadecimal digest of the canonical serialization of the sorted
        retained codes and the sorted ledger rows.

    Raises
    ------
    IngestError
        If the mask and the ledger disagree on the excluded set.
    """
    # Extract and sort the retained country codes.
    retained: List[str] = sorted(str(code) for code in mask[mask].index)
    # Extract and sort the excluded country codes from the mask.
    excluded: List[str] = sorted(str(code) for code in mask[~mask].index)
    # Extract and sort the ledger's excluded country codes.
    ledger_codes: List[str] = sorted(str(code) for code in ledger["country_code"])
    # Refuse a mask and ledger that disagree on the excluded set.
    if excluded != ledger_codes:
        # The divergence makes the fingerprint meaningless.
        raise IngestError(
            "Cannot freeze an eligibility mask whose ledger disagrees with it."
        )
    # Sort the ledger canonically by country for serialization.
    ordered_ledger: pd.DataFrame = ledger.sort_values(
        "country_code", kind="stable"
    )
    # Compose the canonical payload with a fixed schema and separators.
    payload: Dict[str, Any] = {
        "retained": retained,
        "excluded": excluded,
        "ledger": [
            {
                "country_code": str(row["country_code"]),
                "rule_triggered": str(row["rule_triggered"]),
                "trigger_value": float(row["trigger_value"]),
                "detail": str(row["detail"]),
            }
            for _, row in ordered_ledger.iterrows()
        ],
    }
    # Serialize the payload with sorted keys for canonical ordering.
    serialized: str = json.dumps(payload, sort_keys=True)
    # Compute and return the digest of the encoded payload.
    return hashlib.sha256(serialized.encode("utf-8")).hexdigest()


# Demonstrate that the mask does not consume treatment information.
def assert_treatment_blind(
    missingness: MissingnessMatrices,
    universe: Sequence[str],
    policy_ledger: pd.DataFrame,
    recorded_fingerprint: str,
) -> None:
    """
    Demonstrate that the eligibility decision is treatment blind.

    Parameters
    ----------
    missingness : MissingnessMatrices
        The matrices the screen was computed from.
    universe : sequence of str
        The country universe.
    policy_ledger : pandas.DataFrame
        The documentary policy ledger, which must be ignored by the screen.
    recorded_fingerprint : str
        The fingerprint recorded when the screen was computed.

    Returns
    -------
    None
        The function returns nothing when the demonstration succeeds.

    Raises
    ------
    IngestError
        If the recomputed fingerprint differs from the recorded one, which
        would mean the screen depends on something other than the matrices,
        or if the demonstration itself cannot be constructed.
    """
    # Require at least one field whose content encodes treatment assignment.
    policy_fields: List[str] = [
        column
        for column in ("country_code", "policy", "entry_year")
        if column in policy_ledger.columns
    ]
    # Refuse a ledger that cannot express a treatment perturbation.
    if not policy_fields:
        # A no-policy fixture cannot demonstrate the required invariance.
        raise IngestError(
            "Treatment-blindness audit requires policy labels or timing fields."
        )
    # Create an independent copy and perturb identities, labels and timing.
    perturbed: pd.DataFrame = policy_ledger.copy(deep=True)
    # Reverse country assignments when documentary identities are available.
    if "country_code" in perturbed:
        # This changes the treated-country mapping while preserving row count.
        perturbed["country_code"] = perturbed["country_code"].iloc[::-1].to_numpy()
    # Swap common policy labels and prefix any nonstandard labels.
    if "policy" in perturbed:
        # The transformation guarantees different policy content.
        perturbed["policy"] = perturbed["policy"].map(
            lambda value: {
                "sanctions": "direct_arming",
                "direct_arming": "sanctions",
            }.get(str(value), f"perturbed::{value}")
        )
    # Shift all available entry years outside the original timing assignment.
    if "entry_year" in perturbed:
        # Strict parsing prevents a malformed ledger from passing the audit.
        perturbed["entry_year"] = pd.to_numeric(
            perturbed["entry_year"], errors="raise"
        ) + 37
    # Prove the proposed counterfactual coding actually differs.
    if perturbed.equals(policy_ledger):
        # A no-op perturbation would not test treatment blindness.
        raise IngestError("Policy perturbation did not change treatment content.")
    # Run the full eligibility transformation for the actual coverage inputs.
    baseline_mask, baseline_ledger = apply_eligibility_rules(
        missingness, universe, policy_audit=policy_ledger
    )
    # Re-run the full boundary with independently perturbed treatment coding.
    perturbed_mask, perturbed_ledger = apply_eligibility_rules(
        missingness, universe, policy_audit=perturbed
    )
    # Compare every labelled decision and exclusion reason exactly.
    try:
        # Copy both masks and remove audit-only attrs from value comparison.
        baseline_values: pd.Series = baseline_mask.copy(deep=True)
        # Preserve the country index while clearing divergent policy digests.
        baseline_values.attrs = {}
        # Copy the perturbed mask under the same comparison treatment.
        perturbed_values: pd.Series = perturbed_mask.copy(deep=True)
        # Clear only attrs; labels and values remain unchanged.
        perturbed_values.attrs = {}
        # Series equality proves country-labelled masks are identical.
        pd.testing.assert_series_equal(
            baseline_values,
            perturbed_values,
            check_names=True,
            check_dtype=True,
        )
        # Frame equality proves the exclusion attribution is also invariant.
        pd.testing.assert_frame_equal(
            baseline_ledger,
            perturbed_ledger,
            check_like=False,
            check_dtype=True,
        )
    except AssertionError as exc:
        # Convert pandas' comparison detail into the scientific invariant error.
        raise IngestError(
            "Eligibility outputs changed under independently perturbed "
            f"treatment labels/timing: {exc}"
        ) from exc
    # Compare the complete scientific summary emitted by both boundary runs.
    baseline_summary: pd.DataFrame = summarize_rule_application(
        baseline_ledger, len(universe)
    )
    # Summarise the perturbed-boundary output under the same universe.
    perturbed_summary: pd.DataFrame = summarize_rule_application(
        perturbed_ledger, len(universe)
    )
    # Require all labelled summary rows and values to agree exactly.
    try:
        # Exact equality covers counts, shares and rule labels.
        pd.testing.assert_frame_equal(
            baseline_summary, perturbed_summary, check_like=False
        )
    except AssertionError as exc:
        # A summary difference would reveal treatment contamination.
        raise IngestError(
            f"Eligibility rule summary changed under treatment perturbation: {exc}"
        ) from exc
    # Require identical allowed-input lineage across the two executions.
    if (
        baseline_mask.attrs.get("scientific_input_lineage")
        != perturbed_mask.attrs.get("scientific_input_lineage")
    ):
        # Treatment perturbation must not rewrite scientific-input provenance.
        raise IngestError(
            "Allowed eligibility-input lineage changed under policy perturbation."
        )
    # Prove the boundary genuinely consumed the two different policy payloads.
    if (
        baseline_mask.attrs.get("policy_audit_digest")
        == perturbed_mask.attrs.get("policy_audit_digest")
    ):
        # Equal policy digests mean the perturbation was not passed through.
        raise IngestError(
            "Full eligibility boundary did not consume the perturbed policy input."
        )
    # Recompute the fingerprint from the invariant baseline decision.
    recomputed_fingerprint: str = freeze_eligibility_mask(
        baseline_mask, baseline_ledger
    )
    # Refuse a recomputation that does not reproduce the recorded fingerprint.
    if recomputed_fingerprint != recorded_fingerprint:
        # A divergence means the screen is not a function of the matrices.
        raise IngestError(
            "Eligibility recomputation does not reproduce the recorded "
            "fingerprint; the screen is not a function of the matrices alone."
        )
    # Log the completed demonstration with the permuted row count.
    logger.info(
        "treatment blindness demonstrated: %d policy ledger row(s) perturbed, "
        "fingerprint unchanged.",
        len(perturbed),
    )


# Assert the retained count and the metadata flag against the mask.
def assert_final_panel_count(
    mask: pd.Series,
    expected_final: Optional[int] = None,
    expected_otherwise_eligible: Optional[int] = None,
    metadata_frame: Optional[pd.DataFrame] = None,
) -> None:
    """
    Assert the retained count and the metadata flag against the mask.

    Parameters
    ----------
    mask : pandas.Series
        The boolean eligibility mask.
    expected_final : int, optional
        The declared final panel count. When absent the assertion is skipped.
    expected_otherwise_eligible : int, optional
        The declared pre-critical-window count. When supplied, the difference
        against the final count is reported for the audit.
    metadata_frame : pandas.DataFrame, optional
        The country metadata registry. When supplied, its retained flag must
        match the mask exactly.

    Returns
    -------
    None
        The function returns nothing when every supplied assertion holds.

    Raises
    ------
    IngestError
        If the retained count diverges from the declared value, or if the
        metadata flag disagrees with the mask on any country.
    """
    # Count the retained countries in the mask.
    retained_count: int = int(mask.sum())
    # Recover the exact pre-critical labelled set emitted by the rule engine.
    pre_critical: Optional[Tuple[str, ...]] = mask.attrs.get(
        "pre_critical_retained"
    )
    # Recover the complete exclusion reasons emitted by the same run.
    exclusion_ledger: Optional[pd.DataFrame] = mask.attrs.get(
        "exclusion_ledger"
    )
    # Assert the declared final count when one was supplied.
    if expected_final is not None and retained_count != expected_final:
        # The failure names both counts.
        raise IngestError(
            f"Retained panel count is {retained_count}; the declared final "
            f"count is {expected_final}."
        )
    # Report the pre-critical-window count when it was supplied.
    if expected_otherwise_eligible is not None:
        # The sequential count gate requires the pre-critical lineage snapshot.
        if pre_critical is None or exclusion_ledger is None:
            # A final mask alone cannot prove the 131-to-129 transition.
            raise IngestError(
                "Eligibility mask lacks pre-critical set/reason lineage."
            )
        # Assert the observed pre-critical count rather than merely logging it.
        if len(pre_critical) != expected_otherwise_eligible:
            # Keep an alternative-vintage discrepancy explicit and unforced.
            raise IngestError(
                "Pre-critical eligible count is "
                f"{len(pre_critical)}; the fixed-vintage declaration is "
                f"{expected_otherwise_eligible}."
            )
        # Select exactly the rows removed by the critical-window rule.
        critical_rows: pd.DataFrame = exclusion_ledger.loc[
            exclusion_ledger["rule_triggered"].eq("critical_window")
        ]
        # Resolve the observed labelled transition.
        observed_critical: set[str] = set(
            critical_rows["country_code"].astype(str)
        )
        # Require precisely the manuscript's SYR/SDN transition.
        if observed_critical != set(DOCUMENTED_EXCLUSIONS):
            # Report the observed set without changing eligibility decisions.
            raise IngestError(
                "Critical-window transition must remove exactly SYR and SDN; "
                f"observed {sorted(observed_critical)}."
            )
        # Map each critical exclusion to its missing-cell trigger.
        observed_reasons: Dict[str, int] = {
            str(row.country_code): int(row.trigger_value)
            for row in critical_rows.itertuples(index=False)
        }
        # Require the exact fixed-vintage reason values.
        if observed_reasons != dict(DOCUMENTED_EXCLUSIONS):
            # Name both reason maps for a reproducible failure.
            raise IngestError(
                "Critical-window exclusion reasons disagree with the declared "
                f"fixed vintage: expected {dict(DOCUMENTED_EXCLUSIONS)}, "
                f"observed {observed_reasons}."
            )
        # Resolve the actual retained country identities after all rules.
        post_set: set[str] = set(mask[mask].index.astype(str))
        # Resolve the exact labelled pre-critical set emitted by the rules.
        pre_set: set[str] = set(str(code) for code in pre_critical)
        # Compute the sole permitted labelled transition.
        expected_post: set[str] = pre_set - set(DOCUMENTED_EXCLUSIONS)
        # Require set equality, which also proves subset/cardinality consistency.
        if post_set != expected_post:
            # Report omitted and surplus post-critical identities separately.
            raise IngestError(
                "Eligibility labelled transition is inconsistent: post must "
                "equal pre-critical minus SYR/SDN; missing post identities "
                f"{sorted(expected_post - post_set)}, surplus post identities "
                f"{sorted(post_set - expected_post)}."
            )
        # Log the declared and observed relationship.
        logger.info(
            "eligibility screen: %d otherwise eligible, %d removed by the "
            "critical-window rule, %d retained.",
            expected_otherwise_eligible,
            expected_otherwise_eligible - retained_count,
            retained_count,
        )
    # Compare the metadata flag against the mask when metadata was supplied.
    if metadata_frame is not None:
        # Refuse a metadata frame lacking the flag column.
        if "retained_in_final_panel" not in metadata_frame.columns:
            # Without the flag the comparison cannot run.
            raise IngestError(
                "Metadata registry lacks the retained_in_final_panel column."
            )
        # Restrict the metadata to the countries the mask covers.
        scoped = metadata_frame[
            metadata_frame["country_code"].astype(str).isin(mask.index)
        ]
        # Compare the metadata flag with the mask for every scoped country.
        mismatched: List[str] = [
            str(row["country_code"])
            for _, row in scoped.iterrows()
            if bool(int(row["retained_in_final_panel"])) != bool(mask[str(row["country_code"])])
        ]
        # Refuse a metadata flag that disagrees with the mask.
        if mismatched:
            # The failure names the disagreeing countries.
            raise IngestError(
                f"Metadata retained flag disagrees with the mask for "
                f"{len(mismatched)} country(ies): "
                f"{mismatched[:MAX_REPORTED_OFFENDERS]}."
            )
        # Log the verified agreement.
        logger.info(
            "metadata retained flag matches the mask across %d country(ies).",
            len(scoped),
        )
    # Log the completed count assertions.
    logger.info("eligibility assertions complete for %d retained.", retained_count)


# Resolve a declared fidelity-gate value from the study configuration.
def _declared_gate_value(
    bundle: CleansedBundle,
    key_path: Tuple[str, ...],
    fallback: int,
) -> Tuple[int, str]:
    """
    Resolve a declared gate value from the configuration defensively.

    Parameters
    ----------
    bundle : CleansedBundle
        The cleansed bundle carrying the study configuration.
    key_path : tuple of str
        The mapping path from the configuration root to the declared value.
    fallback : int
        Compatibility argument retained for callers; no fallback is applied.

    Returns
    -------
    tuple of int and str
        The validated declaration and its authoritative dotted source path.

    Raises
    ------
    IngestError
        If the declaration is missing, nonintegral, nonfinite or conflicts with
        another authoritative declaration.
    """
    # Start the traversal at the configuration root.
    node: Any = bundle.study_config
    # Walk the path components, descending only through mappings.
    for component in key_path:
        # Descend only when the node is a mapping carrying the component.
        if isinstance(node, Mapping) and component in node:
            # Move one level down the configuration tree.
            node = node[component]
        else:
            # Mandatory scientific counts cannot use implementation defaults.
            raise IngestError(
                f"Mandatory gate declaration {'.'.join(key_path)!r} is absent."
            )
    # Reject booleans, which Python otherwise treats as integer counts.
    if isinstance(node, (bool, np.bool_)):
        # Truth values are not admissible sample-size declarations.
        raise IngestError(
            f"Gate declaration {'.'.join(key_path)!r} must be an integer count, "
            f"not boolean {node!r}."
        )
    # Parse strictly without truncating fractional declarations.
    try:
        # Convert through float solely for finite/integral checks.
        numeric: float = float(node)
    except (TypeError, ValueError) as exc:
        # Preserve the source path in the domain-specific failure.
        raise IngestError(
            f"Gate declaration {'.'.join(key_path)!r} is non-numeric: {node!r}."
        ) from exc
    # Reject missing, infinite and fractional declarations before casting.
    if (
        not np.isfinite(numeric)
        or numeric != math.trunc(numeric)
        or numeric < 0
    ):
        # Never silently coerce a mandatory count.
        raise IngestError(
            f"Gate declaration {'.'.join(key_path)!r} must be a finite integer; "
            f"observed {node!r}."
        )
    # Convert only after all numeric checks pass.
    value: int = int(numeric)
    # Cross-check the final-panel count against its repeated global constant.
    if key_path[-1] == "expected_final":
        # Read the optional independent declaration only when its key exists.
        global_constants: Any = bundle.study_config.get("global_constants")
        # A present but non-mapping global section is malformed configuration.
        if global_constants is not None and not isinstance(global_constants, Mapping):
            # Keep configuration failures inside the domain-specific contract.
            raise IngestError(
                "global_constants must be a mapping when supplied; observed "
                f"{type(global_constants).__name__}."
            )
        # Absence of the optional repeated key preserves the authoritative path.
        if not isinstance(global_constants, Mapping) or (
            "n_countries_final_panel" not in global_constants
        ):
            # Return without manufacturing a fallback declaration.
            return value, ".".join(key_path)
        # Retrieve the explicitly supplied repeated declaration.
        global_value: Any = global_constants["n_countries_final_panel"]
        # Refuse a conflict between two authoritative configuration paths.
        try:
            # Parse the repeated declaration under the same strict semantics.
            global_numeric: float = float(global_value)
        except (TypeError, ValueError) as exc:
            # Never leak a raw conversion exception from the configuration gate.
            raise IngestError(
                "global_constants.n_countries_final_panel is malformed: "
                f"{global_value!r}."
            ) from exc
        # Reject boolean, fractional, non-finite or negative repeated counts.
        if (
            isinstance(global_value, (bool, np.bool_))
            or not np.isfinite(global_numeric)
            or global_numeric != math.trunc(global_numeric)
            or global_numeric < 0
        ):
            # Preserve the domain-specific error contract.
            raise IngestError(
                "global_constants.n_countries_final_panel must be a finite "
                f"non-negative integer; observed {global_value!r}."
            )
        # Refuse conflicts between valid repeated declarations.
        if global_numeric != numeric:
            # Name both paths and values in the diagnostic.
            raise IngestError(
                "Conflicting final-panel declarations: "
                f"{'.'.join(key_path)}={value} versus "
                f"global_constants.n_countries_final_panel={global_value!r}."
            )
    # Return the validated value and its diagnostic source path.
    return value, ".".join(key_path)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the eligibility screen over the ingest bundle.
def orchestrate_eligibility_screening(
    bundle: CleansedBundle,
    policy_ledger: Optional[pd.DataFrame] = None,
    metadata_frame: Optional[pd.DataFrame] = None,
    expected_otherwise_eligible: Optional[int] = None,
    expected_final: Optional[int] = None,
    enforce_declared_gates: bool = True,
) -> EligibilityScreen:
    """
    Screen the sample for eligibility on raw pre-completion coverage.

    Parameters
    ----------
    bundle : CleansedBundle
        The cleansed bundle carrying the cleansed long frames; its handoff
        verification runs before any screen is computed.
    policy_ledger : pandas.DataFrame, optional
        The documentary policy ledger, used only for the blindness
        demonstration.
    metadata_frame : pandas.DataFrame, optional
        The country metadata registry, used for the flag comparison.
    expected_otherwise_eligible : int, optional
        The declared pre-critical-window count.
    expected_final : int, optional
        The declared final panel count.
    enforce_declared_gates : bool, optional
        When true (the production default), mandatory counts are resolved from
        authoritative configuration paths and missing or malformed values are
        rejected; implementation constants are never substituted.

    Returns
    -------
    EligibilityScreen
        The mask, exclusion ledger, matrices, rule summary, fingerprint,
        timestamp and notes.

    Raises
    ------
    KeyError
        If a required source is absent from the bundle.
    IngestError
        If any missingness, rule or assertion check fails.
    """
    # Initialize explicit lineage labels before optional configuration lookup;
    # avoiding ``locals()`` keeps the registered reconciliation boundary fully
    # statically traceable while preserving the same user-facing note.
    final_gate_source = "explicit argument"
    precritical_gate_source = "explicit argument"
    # Verify the cleansed handoff so no raw aggregate row can reach the
    # eligibility universe (remediation of CF-1 / B-9 step three).
    bundle.verify_handoff()
    # Resolve the declared final count under the production gate policy.
    if enforce_declared_gates and expected_final is None:
        # Source the gate from the authoritative configuration path.
        expected_final, final_gate_source = _declared_gate_value(
            bundle,
            (
                "algorithm_parameters",
                "C1_eligibility_screening",
                "expected_final",
            ),
            DECLARED_EXPECTED_FINAL,
        )
    # Resolve the declared pre-critical-window count under the same policy.
    if enforce_declared_gates and expected_otherwise_eligible is None:
        # Source the companion gate from the configuration registry.
        expected_otherwise_eligible, precritical_gate_source = _declared_gate_value(
            bundle,
            (
                "algorithm_parameters",
                "C1_eligibility_screening",
                "expected_otherwise_eligible",
            ),
            DECLARED_EXPECTED_OTHERWISE_ELIGIBLE,
        )
    # Read the cleansed WDI long frame from the bundle.
    wdi_frame: pd.DataFrame = bundle.frame("wdi_raw_long")
    # Read the cleansed WGI long frame from the bundle.
    wgi_frame: pd.DataFrame = bundle.frame("wgi_raw_long")
    # Derive the universe from the wide extract's country codes, which is a
    # coverage artefact and carries no treatment information.
    wide_frame: pd.DataFrame = bundle.frame("wdi_raw_wide")
    # Compose the sorted universe of country codes.
    universe: List[str] = sorted(
        str(code) for code in wide_frame["country_code"].dropna().unique()
    )
    # Refuse an empty universe, which cannot be screened.
    if not universe:
        # Without a universe the screen is undefined.
        raise IngestError("The eligibility universe is empty.")
    # Build the missingness matrices of the screen.
    missingness: MissingnessMatrices = build_missingness_matrices(
        wdi_frame, wgi_frame, universe
    )
    # Apply the four rules and obtain the mask and its ledger.
    mask, ledger = apply_eligibility_rules(
        missingness, universe, policy_audit=policy_ledger
    )
    # Summarise the removals by rule.
    rule_summary: pd.DataFrame = summarize_rule_application(ledger, len(universe))
    # Freeze the decision by computing its canonical fingerprint.
    fingerprint: str = freeze_eligibility_mask(mask, ledger)
    # Assert the declared counts when they were supplied.
    assert_final_panel_count(
        mask,
        expected_final=expected_final,
        expected_otherwise_eligible=expected_otherwise_eligible,
        metadata_frame=metadata_frame,
    )
    # Demonstrate treatment blindness when the policy ledger was supplied.
    notes: List[str] = []
    # Record authoritative gate lineage when production declarations were used.
    if enforce_declared_gates:
        # Preserve both path labels in the outgoing audit record.
        notes.append(
            "declared gates sourced from "
            f"{precritical_gate_source} "
            "and "
            f"{final_gate_source}."
        )
    # Run the demonstration only when a ledger was provided.
    if policy_ledger is not None:
        # Execute the permutation demonstration.
        assert_treatment_blind(missingness, universe, policy_ledger, fingerprint)
        # Record that the demonstration ran.
        notes.append("treatment blindness demonstrated by ledger permutation.")
    # Record the count assertions that were evaluated.
    notes.append(
        "count assertions evaluated against declared values."
        if expected_final is not None
        else "count assertions skipped: no declared values were supplied."
    )
    # Full production configurations enforce the declared extraction-coverage audit.
    coverage_audit = pd.DataFrame()
    if enforce_declared_gates and "execution_parameters" in bundle.study_config:
        coverage_audit = audit_raw_panel_coverage(bundle, mask, bundle.study_config)
        notes.append("raw retained-panel coverage independently compared before completion.")
    # Assemble the immutable eligibility screen.
    screen: EligibilityScreen = EligibilityScreen(
        mask=mask,
        exclusion_ledger=ledger,
        missingness=missingness,
        rule_summary=rule_summary,
        coverage_audit=coverage_audit,
        fingerprint_sha256=fingerprint,
        screened_utc=current_utc_iso(),
        notes=tuple(notes),
    )
    # Log the completed screen with its fingerprint prefix.
    logger.info(
        "eligibility screen complete: %d universe, %d retained, %d excluded, "
        "fingerprint %s.",
        len(universe),
        int(mask.sum()),
        len(ledger),
        fingerprint[:16],
    )
    # Return the assembled screen.
    return screen
````

````python
# Task 5: Control Missing-Data Completion (Algorithm C2)

# ==============================================================================
# Task 5: Control Missing-Data Completion (Algorithm C2)
# ==============================================================================

# Declare the method labels of the completion hierarchy in priority order.
METHOD_INTERNAL: Final[str] = "P1_internal_linear"
# Declare the trailing-gap method label.
METHOD_TRAILING: Final[str] = "P2_trailing_locf"
# Declare the short leading-gap method label.
METHOD_LEADING_SHORT: Final[str] = "P3_leading_carry_back"
# Declare the bounded trend backcast method label.
METHOD_TREND: Final[str] = "P4_bounded_trend_backcast"
# Retain the historical symbol as a compatibility alias.  A fallback is a P3
# repair with a separate Boolean flag; it is never a fifth scientific method.
METHOD_FALLBACK: Final[str] = METHOD_LEADING_SHORT
# Declare the declared methods in their reporting order.
COMPLETION_METHODS: Final[Tuple[str, ...]] = (
    METHOD_INTERNAL,
    METHOD_TRAILING,
    METHOD_LEADING_SHORT,
    METHOD_TREND,
)
# Declare the maximum length of a leading gap served by the short tier.
P3_MAX_LEADING_YEARS: Final[int] = 3
# Declare the number of anchors required by the trend tier.
P4_ANCHOR_COUNT: Final[int] = 5
# Declare the number of standard deviations added to the trend bounds.
TREND_BOUND_SIGMA: Final[float] = 1.0
# Declare the variables that must remain non-negative after completion.
NON_NEGATIVE_CONTROLS: Final[Tuple[str, ...]] = (
    "TradeOpenness",
    "InvestmentRate",
    "GovConsumption",
)
# Declare the anchor window the completion is fitted over.
ANCHOR_WINDOW: Final[Tuple[int, int]] = (1996, 2024)
# Declare the analytic window whose cells are emitted.
OUTPUT_WINDOW: Final[Tuple[int, int]] = (2000, 2024)
# Declare the name of the derived fuel-share balance.
BALANCE_NAME: Final[str] = FUEL_BALANCE_NAME
# Declare the derived missing count the manuscript attributes to the balance.
DECLARED_BALANCE_DERIVED_MISSING: Final[int] = 203
# Declare the manuscript's total repaired control cells over the full panel.
DECLARED_REPAIRED_CONTROL_CELLS: Final[int] = 1471
# Declare the frozen eligible-country dimension of the production tensor.
DECLARED_CONTROL_COUNTRIES: Final[int] = 129


# Declare the immutable result of the control completion.
@dataclass(frozen=True)
class ControlCompletionResult:
    """
    Immutable result of the control missing-data completion.

    Attributes
    ----------
    tensor : pandas.DataFrame
        The long completed control tensor keyed on country, indicator and
        year, carrying the eight declared controls with no missing cell.
    ledger : pandas.DataFrame
        One row per repaired primitive/control cell with its P1--P4 method
        label and fallback flag, plus separately marked derived-balance union
        rows that retain both primitive methods without claiming a fifth
        completion method.
    method_summary : pandas.DataFrame
        One row per P1--P4 method with its primary repaired-cell count and
        share.  Derived-balance and all-repair totals are stored as explicit
        summary attributes.
    fingerprint_sha256 : str
        Key-sorted digest of the completed tensor.
    completed_utc : str
        ISO-8601 UTC timestamp of the completion run.
    notes : tuple of str
        Recorded decisions, including fallback and clipping counts.
    """
    # The long completed control tensor.
    tensor: pd.DataFrame
    # One row per repaired cell with its method label.
    ledger: pd.DataFrame
    # One row per method with its count and share.
    method_summary: pd.DataFrame
    # Key-sorted digest of the completed tensor.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the completion run.
    completed_utc: str
    # Recorded decisions of the completion run.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 5, Step 1: Apply the four-tier completion hierarchy
# -------------------------------------------------------------------------------------------------------------------------------


# Parse one calendar label without a binary-floating-point round trip.
def _exact_calendar_year(value: Any, *, context: str) -> int:
    """
    Return one exactly integral signed-64-bit calendar label.

    Parameters
    ----------
    value : Any
        Candidate year label. Python/NumPy integers, finite integral floating
        values, :class:`decimal.Decimal` values and decimal strings are
        accepted. Booleans, fractions, non-finite values, blank strings and
        values outside the signed-64-bit range are rejected.
    context : str
        Nonblank scientific context included in domain-specific error text.

    Returns
    -------
    int
        The exact canonical calendar year.

    Raises
    ------
    IngestError
        If ``context`` is blank or ``value`` cannot be represented exactly as
        one finite integral signed-64-bit year.

    Notes
    -----
    String and Decimal inputs are parsed with ``Decimal``. This deliberately
    avoids the former float conversion that rounded labels such as
    ``"2014.0000000000001"`` to the integer 2014.
    """
    if not isinstance(context, str) or not context.strip():
        raise IngestError("Exact calendar parsing requires a nonblank context.")
    if isinstance(value, (bool, np.bool_)):
        raise IngestError(f"{context} contains a Boolean calendar label {value!r}.")
    int64_info = np.iinfo(np.int64)
    # Native and NumPy integers already carry an exact mathematical value.
    if isinstance(value, (int, np.integer)):
        canonical = int(value)
    # Floating values are validated in their actual stored representation;
    # decimal-text inputs take the exact Decimal branch below.
    elif isinstance(value, (float, np.floating)):
        try:
            if not bool(np.isfinite(value)) or value != np.floor(value):
                raise IngestError(
                    f"{context} contains a nonintegral or nonfinite calendar label {value!r}."
                )
            canonical = int(value)
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError(
                f"{context} contains an invalid calendar label {value!r}."
            ) from exc
    # Decimal and string labels are parsed in base ten with no float rounding.
    elif isinstance(value, (Decimal, str)):
        token = str(value).strip()
        if not token:
            raise IngestError(f"{context} contains a blank calendar label.")
        try:
            decimal_value = Decimal(token)
        except (InvalidOperation, ValueError, TypeError) as exc:
            raise IngestError(
                f"{context} contains a nonnumeric calendar label {value!r}."
            ) from exc
        if not decimal_value.is_finite() or decimal_value != decimal_value.to_integral_value():
            raise IngestError(
                f"{context} contains a nonintegral or nonfinite calendar label {value!r}."
            )
        # Compare as Decimal before conversion so an enormous exponent cannot
        # leak an OverflowError or allocate an unbounded Python integer.
        if decimal_value < Decimal(int(int64_info.min)) or decimal_value > Decimal(int(int64_info.max)):
            raise IngestError(
                f"{context} contains an out-of-int64 calendar label {value!r}."
            )
        canonical = int(decimal_value)
    else:
        raise IngestError(
            f"{context} contains an unsupported calendar label {value!r}."
        )
    if canonical < int64_info.min or canonical > int64_info.max:
        raise IngestError(
            f"{context} contains an out-of-int64 calendar label {value!r}."
        )
    return canonical


# Normalize an ordered calendar axis and expose canonical collisions.
def _exact_calendar_axis(values: Sequence[Any], *, context: str) -> List[int]:
    """
    Parse an ordered sequence of calendar labels exactly.

    Parameters
    ----------
    values : sequence of Any
        Calendar labels in their source order.
    context : str
        Nonblank scientific context forwarded to the scalar parser.

    Returns
    -------
    list of int
        Canonical signed-64-bit years in unchanged source order.

    Raises
    ------
    IngestError
        If ``values`` is a scalar string/bytes object, is not iterable, or any
        member fails exact calendar validation.

    Notes
    -----
    Duplicate detection is intentionally left to each caller because a Series
    axis, a country-year frame and an inclusive window require different key
    and error evidence after normalization.
    """
    if isinstance(values, (str, bytes)):
        raise IngestError(f"{context} must be an ordered collection of calendar labels.")
    try:
        source_values = list(values)
    except TypeError as exc:
        raise IngestError(f"{context} is not an iterable calendar axis.") from exc
    return [_exact_calendar_year(value, context=context) for value in source_values]


# Build the country-by-year grid of one control over the anchor window.
def _series_grid(
    frame: pd.DataFrame,
    universe: Sequence[str],
    value_column: str,
    indicator_code: Optional[str] = None,
    anchor_window: Tuple[int, int] = ANCHOR_WINDOW,
) -> pd.DataFrame:
    """
    Build a full country-by-year grid for one control series.

    Parameters
    ----------
    frame : pandas.DataFrame
        The long frame carrying the control.
    universe : sequence of str
        The country universe.
    value_column : str
        The column holding the control values.
    indicator_code : str, optional
        When supplied, the frame is filtered to this indicator code.
    anchor_window : tuple of int, optional
        The inclusive anchor window the grid covers.

    Returns
    -------
    pandas.DataFrame
        A grid indexed by the sorted universe and columned by every anchor
        year, with cells absent from the frame represented as missing.

    Raises
    ------
    IngestError
        If the frame/schema, country universe or window is malformed; if a
        source year is not exactly integral and signed-64-bit; or if two rows
        collide on a normalized country-year key.
    """
    if not isinstance(frame, pd.DataFrame):
        raise IngestError("Control-grid construction requires a DataFrame.")
    # Refuse a frame lacking any column needed by selection or pivoting.
    required_columns = {"country_code", "year", str(value_column)}
    if indicator_code is not None:
        required_columns.add("indicator_code")
    missing_columns = sorted(required_columns - set(frame.columns))
    if missing_columns:
        raise IngestError(f"Control frame lacks column(s) {missing_columns}.")
    # Validate the grid's inclusive window with the same exact parser used for
    # source year labels.
    if not isinstance(anchor_window, (tuple, list)) or len(anchor_window) != 2:
        raise IngestError("Control anchor window must contain exactly two years.")
    canonical_window = _exact_calendar_axis(
        anchor_window, context="Control anchor window"
    )
    if canonical_window[0] > canonical_window[1]:
        raise IngestError("Control anchor window is reversed.")
    # Canonicalize the requested country universe before filtering the source.
    canonical_universe = tuple(str(code).strip() for code in universe)
    if (
        not canonical_universe
        or any(not code for code in canonical_universe)
        or len(canonical_universe) != len(set(canonical_universe))
    ):
        raise IngestError("Control grid requires a nonempty unique country universe.")
    # Restrict the frame to the requested indicator when one is supplied.
    subset: pd.DataFrame = (
        frame[frame["indicator_code"].astype(str) == str(indicator_code)].copy()
        if indicator_code is not None
        else frame.copy()
    )
    # Exclude unrelated countries before validating the scientific grid.
    subset = subset[
        subset["country_code"].astype(str).isin(canonical_universe)
    ].copy()
    # Attach exact canonical years before any comparison or pivot operation.
    canonical_years = _exact_calendar_axis(
        subset["year"].tolist(), context="Control source year"
    )
    subset["year"] = pd.Series(canonical_years, index=subset.index, dtype="int64")
    # Refuse aliases such as 1996 and '1996.0' after normalization. A pivot
    # must never choose one duplicate observation with an aggregation rule.
    duplicate_keys = subset.duplicated(["country_code", "year"], keep=False)
    if bool(duplicate_keys.any()):
        offenders = subset.loc[duplicate_keys, ["country_code", "year"]]
        raise IngestError(
            "Control source duplicates normalized country-year key(s): "
            f"{offenders.head(MAX_REPORTED_OFFENDERS).to_dict('records')}."
        )
    # Restrict the exact integer year column to the declared anchor window.
    subset = subset[
        subset["year"].between(canonical_window[0], canonical_window[1])
    ]
    # With normalized duplicates already refused, pivot without aggregation so
    # every observed numeric value is transferred unchanged to one grid cell.
    grid: pd.DataFrame = subset.pivot(
        index="country_code", columns="year", values=value_column
    )
    # Compose the full ascending anchor-year range.
    full_columns: List[int] = list(
        range(canonical_window[0], canonical_window[1] + 1)
    )
    # Reindex the grid to the sorted universe and the full year range.
    grid = grid.reindex(
        index=pd.Index(sorted(canonical_universe), name="country_code"),
        columns=full_columns,
    )
    # Return the full grid.
    return grid


# Complete one country series through the four-tier hierarchy.
def complete_series(
    values: pd.Series,
    non_negative: bool,
) -> Tuple[pd.Series, pd.Series, Tuple[str, ...]]:
    """
    Complete one control series through the declared hierarchy.

    Parameters
    ----------
    values : pandas.Series
        The anchor-window series indexed by year, with missing cells.
    non_negative : bool
        Whether the series must be clipped at zero after completion.

    Returns
    -------
    tuple
        The completed series, the method label per year (an empty string for
        an observed cell) and the notes recording fallbacks and clipping.
        Per-cell anchor evidence and the distinct fallback Boolean are stored
        in the returned method Series attributes without changing tuple arity.

    Raises
    ------
    IngestError
        If the year axis is not exactly integral, unique and increasing; if
        the series contains no valid observation; or if a finite observed
        anchor is altered or a missing cell remains unrepaired.
    """
    if not isinstance(values, pd.Series) or values.empty:
        raise IngestError("Control completion requires a non-empty Series.")
    if not isinstance(non_negative, (bool, np.bool_)):
        raise IngestError("Control non-negative flag must be Boolean.")
    # Parse every source label exactly before positional gap classification.
    source_labels = list(values.index)
    canonical_years = _exact_calendar_axis(
        source_labels, context="Control series year"
    )
    # Reject both literal duplicates and distinct integral aliases after
    # normalization, while retaining the original labels in the public result.
    if values.index.has_duplicates or len(set(canonical_years)) != len(canonical_years):
        raise IngestError("Control series years must be unique integral labels.")
    if not all(
        right_year > left_year
        for left_year, right_year in zip(canonical_years, canonical_years[1:])
    ):
        raise IngestError("Control series years must be strictly increasing.")
    year_array = np.asarray(canonical_years, dtype="int64")
    # Freeze observed values and require every anchor to be finite numeric data.
    original = values.copy(deep=True)
    observed_mask = original.notna()
    if int(observed_mask.sum()) == 0:
        raise IngestError("Control series contains no valid observation.")
    observed_numeric = pd.to_numeric(original.loc[observed_mask], errors="coerce")
    if observed_numeric.isna().any() or not np.isfinite(
        observed_numeric.to_numpy(dtype=float)
    ).all():
        raise IngestError("Control series contains a non-finite observed anchor.")
    completed = pd.to_numeric(original, errors="coerce").astype(float)
    methods = pd.Series("", index=values.index, dtype=object)
    notes: List[str] = []
    anchor_records: Dict[Any, Dict[str, Any]] = {}
    fallback_records: Dict[Any, bool] = {}
    observed_years = year_array[observed_mask.to_numpy(dtype=bool)]
    observed_values = observed_numeric.to_numpy(dtype=float)
    observed_labels = list(values.index[observed_mask])
    first_year, last_year = int(observed_years[0]), int(observed_years[-1])
    missing = ~observed_mask.to_numpy(dtype=bool)
    internal = missing & (year_array > first_year) & (year_array < last_year)
    trailing = missing & (year_array > last_year)
    leading = missing & (year_array < first_year)
    if not np.array_equal(internal | trailing | leading, missing) or bool(
        np.any(internal & trailing) or np.any(internal & leading) or np.any(trailing & leading)
    ):
        raise IngestError("Control gaps do not form a disjoint C2 partition.")
    # P1: linearly interpolate only missing cells bounded by original anchors.
    if bool(internal.any()):
        completed.iloc[np.flatnonzero(internal)] = np.interp(
            year_array[internal], observed_years, observed_values
        )
        methods.iloc[np.flatnonzero(internal)] = METHOD_INTERNAL
        for position in np.flatnonzero(internal):
            left_position = int(np.flatnonzero(observed_years < year_array[position])[-1])
            right_position = int(np.flatnonzero(observed_years > year_array[position])[0])
            anchor_records[values.index[position]] = {
                "anchor_years": (
                    observed_labels[left_position], observed_labels[right_position]
                ),
                "anchor_values": (
                    float(observed_values[left_position]),
                    float(observed_values[right_position]),
                ),
            }
            fallback_records[values.index[position]] = False
    # P2: carry the last original observation through every trailing gap.
    if bool(trailing.any()):
        completed.iloc[np.flatnonzero(trailing)] = float(observed_values[-1])
        methods.iloc[np.flatnonzero(trailing)] = METHOD_TRAILING
        for position in np.flatnonzero(trailing):
            anchor_records[values.index[position]] = {
                "anchor_years": (observed_labels[-1],),
                "anchor_values": (float(observed_values[-1]),),
            }
            fallback_records[values.index[position]] = False
    # P3/P4: classify the one original leading run by its exact length.
    if bool(leading.any()):
        leading_positions = np.flatnonzero(leading)
        n_leading = int(len(leading_positions))
        use_fallback = n_leading > P3_MAX_LEADING_YEARS and len(observed_values) < P4_ANCHOR_COUNT
        if use_fallback:
            raise IngestError(
                "A leading gap longer than three years requires five original "
                "trend anchors; undocumented carry-back is not permitted."
            )
        if n_leading <= P3_MAX_LEADING_YEARS:
            completed.iloc[leading_positions] = float(observed_values[0])
            methods.iloc[leading_positions] = METHOD_LEADING_SHORT
            for position in leading_positions:
                anchor_records[values.index[position]] = {
                    "anchor_years": (observed_labels[0],),
                    "anchor_values": (float(observed_values[0]),),
                }
                fallback_records[values.index[position]] = False
        else:
            anchor_years = observed_years[:P4_ANCHOR_COUNT]
            anchor_values = observed_values[:P4_ANCHOR_COUNT]
            regression = scipy_stats.linregress(anchor_years, anchor_values)
            trend_values = regression.slope * year_array[leading] + regression.intercept
            # C2 uses the full observed range but the sample SD of precisely
            # the first five anchors (divisor four) for both extrapolation bounds.
            anchor_sd = float(anchor_values.std(ddof=1))
            lower_bound = float(observed_values.min()) - TREND_BOUND_SIGMA * anchor_sd
            upper_bound = float(observed_values.max()) + TREND_BOUND_SIGMA * anchor_sd
            completed.iloc[leading_positions] = np.clip(
                trend_values, lower_bound, upper_bound
            )
            methods.iloc[leading_positions] = METHOD_TREND
            for position in leading_positions:
                anchor_records[values.index[position]] = {
                    "anchor_years": tuple(observed_labels[:P4_ANCHOR_COUNT]),
                    "anchor_values": tuple(float(value) for value in anchor_values),
                    "lower_bound": lower_bound,
                    "upper_bound": upper_bound,
                    "anchor_sample_sd": anchor_sd,
                    "slope": float(regression.slope),
                    "intercept": float(regression.intercept),
                }
                fallback_records[values.index[position]] = False
    # Apply the non-negativity floor when the variable requires it.
    if non_negative:
        # Restrict the floor to generated cells; observations remain immutable.
        generated_negative: pd.Series = (~observed_mask) & (completed < 0)
        # Count only generated values changed by the scientific constraint.
        n_negative: int = int(generated_negative.sum())
        # Floor generated negative cells without altering observed negatives.
        if n_negative:
            # Assign zero only at generated negative positions.
            completed.loc[generated_negative] = 0.0
            # Record the clipped-cell count for the audit.
            notes.append(f"floored {n_negative} generated cell(s) at zero.")
    # Prove every original nonmissing observation is bitwise/value preserved.
    try:
        # Compare only frozen observed positions under exact dtype-insensitive values.
        pd.testing.assert_series_equal(
            completed.loc[observed_mask],
            pd.to_numeric(original.loc[observed_mask], errors="raise").astype(float),
            check_names=True,
            check_dtype=False,
            check_exact=True,
        )
    except AssertionError as exc:
        # Any observed-value mutation violates the completion contract.
        raise IngestError(
            f"Control completion altered an observed value: {exc}"
        ) from exc
    # Attach per-cell anchors to the method series without changing tuple arity.
    repaired = methods.ne("")
    if set(methods.loc[repaired].astype(str)) - set(COMPLETION_METHODS):
        raise IngestError("Control completion emitted an undeclared C2 method.")
    if bool(completed.loc[~observed_mask].isna().any()):
        raise IngestError("Control completion left a missing cell unrepaired.")
    methods.attrs["anchor_records"] = anchor_records
    methods.attrs["fallback_records"] = fallback_records
    # Return the completed series, its labels and the notes.
    return completed, methods, tuple(notes)


# Complete one control across the universe and emit its repair ledger.
def complete_control_grid(
    frame: pd.DataFrame,
    name: str,
    universe: Sequence[str],
    value_column: str = "value",
    indicator_code: Optional[str] = None,
    anchor_window: Tuple[int, int] = ANCHOR_WINDOW,
    output_window: Tuple[int, int] = OUTPUT_WINDOW,
    non_negative: bool = False,
) -> Tuple[pd.DataFrame, pd.DataFrame, Tuple[str, ...]]:
    """
    Complete one control across the universe and emit its ledger.

    Parameters
    ----------
    frame : pandas.DataFrame
        The long frame carrying the control.
    name : str
        The declared control name used in the ledger.
    universe : sequence of str
        The country universe.
    value_column : str, optional
        The column holding the control values.
    indicator_code : str, optional
        When supplied, the frame is filtered to this indicator code.
    anchor_window : tuple of int, optional
        The inclusive anchor window.
    output_window : tuple of int, optional
        The inclusive analytic window whose cells are emitted.
    non_negative : bool, optional
        Whether the control must be clipped at zero.

    Returns
    -------
    tuple
        The completed output grid, the repair ledger and the accumulated
        notes.

    Raises
    ------
    IngestError
        If the source schema, country universe, exact calendar, normalized
        country-year keys, nested windows or per-country completion violates
        the declared control-grid contract.
    """
    if not str(name).strip() or not str(value_column).strip():
        raise IngestError("Control completion requires nonblank control/value names.")
    normalized_universe = tuple(str(code).strip() for code in universe)
    if (
        not normalized_universe
        or any(not code for code in normalized_universe)
        or len(normalized_universe) != len(set(normalized_universe))
    ):
        raise IngestError("Control completion universe must be nonempty and unique.")
    normalized_windows: Dict[str, Tuple[int, int]] = {}
    for window_name, window in (
        ("anchor_window", anchor_window), ("output_window", output_window)
    ):
        if not isinstance(window, (tuple, list)) or len(window) != 2:
            raise IngestError(f"Control {window_name} must contain two years.")
        normalized_windows[window_name] = tuple(
            _exact_calendar_axis(
                window, context=f"Control {window_name}"
            )
        )
    anchor_window = normalized_windows["anchor_window"]
    output_window = normalized_windows["output_window"]
    if (
        anchor_window[0] > output_window[0]
        or output_window[1] > anchor_window[1]
        or output_window[0] > output_window[1]
    ):
        raise IngestError("Control anchor/output windows are malformed or misnested.")
    if not isinstance(frame, pd.DataFrame):
        raise IngestError("Control completion requires a DataFrame source.")
    source = frame.copy(deep=False)
    if indicator_code is not None:
        if "indicator_code" not in source.columns:
            raise IngestError("Control frame lacks the indicator_code column.")
        source = source[source["indicator_code"].astype(str).eq(str(indicator_code))]
    required_source = {"country_code", "year", value_column}
    if not required_source.issubset(source.columns):
        raise IngestError(
            f"Control source lacks {sorted(required_source - set(source.columns))}."
        )
    source = source[
        source["country_code"].astype(str).isin(normalized_universe)
    ].copy()
    # Canonicalize the actual frame passed downstream; validating a sidecar
    # column while pivoting the original string labels caused the former raw
    # pandas comparison failure.
    canonical_source_years = _exact_calendar_axis(
        source["year"].tolist(), context=f"{name} source year"
    )
    source["year"] = pd.Series(
        canonical_source_years, index=source.index, dtype="int64"
    )
    source = source[source["year"].between(*anchor_window)]
    if bool(source.duplicated(["country_code", "year"], keep=False).any()):
        raise IngestError(f"{name}: source duplicates a country-year key.")
    # Build the anchor-window grid of the control.
    grid: pd.DataFrame = _series_grid(
        source, normalized_universe, value_column, None, anchor_window
    )
    # Initialise the completed grid accumulator.
    completed_rows: Dict[str, pd.Series] = {}
    # Accumulate the repair ledger rows.
    ledger_rows: List[Dict[str, Any]] = []
    # Accumulate the notes of the completed series.
    notes: List[str] = []
    # Complete every country's series in turn.
    for country in grid.index:
        # Complete the country's series through the hierarchy.
        completed, methods, series_notes = complete_series(
            grid.loc[country], non_negative=non_negative
        )
        # Record the completed series for the country.
        completed_rows[str(country)] = completed
        # Recover the explicit per-cell anchor audit emitted by the completer.
        anchor_records: Mapping[Any, Mapping[str, Tuple[Any, ...]]] = (
            methods.attrs.get("anchor_records", {})
        )
        fallback_records: Mapping[Any, bool] = methods.attrs.get(
            "fallback_records", {}
        )
        # Record any notes with the country and control attached.
        for note in series_notes:
            # Append the attributed note to the accumulator.
            notes.append(f"{country}/{name}: {note}")
        # Iterate over every anchor-window year so that repairs inside the
        # anchor years are recorded too, with a flag marking emission.
        for year in completed.index:
            # Read the method label of the cell.
            label: str = str(methods.loc[year])
            # Record a ledger row only for a repaired cell.
            if label:
                # Append the repaired cell with its single method label and
                # its emission flag, which is true inside the output window.
                ledger_rows.append(
                    {
                        "country_code": str(country),
                        "indicator_name": name,
                        "year": int(year),
                        "method": label,
                        "is_fallback": bool(fallback_records.get(year, False)),
                        "original_value": np.nan,
                        "completed_value": float(completed.loc[year]),
                        "anchor_years": tuple(
                            anchor_records.get(year, {}).get("anchor_years", ())
                        ),
                        "anchor_values": tuple(
                            anchor_records.get(year, {}).get("anchor_values", ())
                        ),
                        "source_indicator_code": (
                            "" if indicator_code is None else str(indicator_code)
                        ),
                        "raw_missing": True,
                        "derived_from_primitives": False,
                        "emitted": bool(
                            int(output_window[0])
                            <= int(year)
                            <= int(output_window[1])
                        ),
                    }
                )
    # Assemble the completed grid over the anchor window.
    completed_grid: pd.DataFrame = pd.DataFrame(completed_rows).T
    # Restrict the completed grid to the emitted years.
    emitted_columns: List[int] = list(
        range(int(output_window[0]), int(output_window[1]) + 1)
    )
    # Slice the emitted window from the completed grid.
    emitted_grid: pd.DataFrame = completed_grid.loc[:, emitted_columns]
    # Restore the country index name that the transposition discards, so that
    # downstream melting can address the country column by name.
    emitted_grid.index.name = "country_code"
    if bool(emitted_grid.isna().to_numpy().any()) or not np.isfinite(
        emitted_grid.to_numpy(dtype=float)
    ).all():
        raise IngestError(f"{name}: emitted completed grid is not finite and complete.")
    # Reconcile every originally observed analytic cell to its unchanged output.
    output_years = list(range(int(output_window[0]), int(output_window[1]) + 1))
    observed_output = grid.loc[:, output_years].notna()
    try:
        pd.testing.assert_frame_equal(
            emitted_grid.where(observed_output),
            grid.loc[:, output_years].where(observed_output).astype(float),
            check_dtype=False,
            check_exact=True,
        )
    except AssertionError as exc:
        raise IngestError(f"{name}: an observed analytic cell changed: {exc}") from exc
    # Assemble the repair ledger.
    ledger: pd.DataFrame = pd.DataFrame(
        ledger_rows,
        columns=[
            "country_code", "indicator_name", "year", "method", "is_fallback",
            "original_value", "completed_value", "anchor_years", "anchor_values",
            "source_indicator_code", "raw_missing", "derived_from_primitives",
            "emitted",
        ],
    )
    if bool(ledger.duplicated(["country_code", "indicator_name", "year"]).any()):
        raise IngestError(f"{name}: repair ledger duplicates a scientific key.")
    if set(ledger["method"].astype(str)) - set(COMPLETION_METHODS):
        raise IngestError(f"{name}: repair ledger contains an undeclared C2 method.")
    ledger.attrs["raw_missing_anchor_window"] = int(grid.isna().to_numpy().sum())
    ledger.attrs["raw_missing_output_window"] = int(
        grid.loc[:, output_years].isna().to_numpy().sum()
    )
    ledger.attrs["source_indicator_code"] = (
        "" if indicator_code is None else str(indicator_code)
    )
    emitted_grid.attrs["raw_missing_anchor_window"] = ledger.attrs[
        "raw_missing_anchor_window"
    ]
    emitted_grid.attrs["raw_missing_output_window"] = ledger.attrs[
        "raw_missing_output_window"
    ]
    # Log the completion of the control.
    logger.info(
        "%s: completed %d series, repaired %d emitted cell(s).",
        name,
        len(emitted_grid),
        int(ledger["emitted"].astype(bool).sum()),
    )
    # Return the emitted grid, the ledger and the notes.
    return emitted_grid, ledger, tuple(notes)


# Complete every declared control through the hierarchy.
def complete_all_controls(
    wdi_frame: pd.DataFrame,
    wgi_frame: pd.DataFrame,
    universe: Sequence[str],
    anchor_window: Tuple[int, int] = ANCHOR_WINDOW,
    output_window: Tuple[int, int] = OUTPUT_WINDOW,
) -> Tuple[Dict[str, pd.DataFrame], pd.DataFrame, Tuple[str, ...]]:
    """
    Complete the six retrieval controls and the governance control.

    Parameters
    ----------
    wdi_frame : pandas.DataFrame
        The cleansed WDI long frame.
    wgi_frame : pandas.DataFrame
        The cleansed WGI long frame.
    universe : sequence of str
        The country universe.
    anchor_window : tuple of int, optional
        The inclusive anchor window.
    output_window : tuple of int, optional
        The inclusive analytic window.

    Returns
    -------
    tuple
        The completed grids keyed by control name, the combined repair ledger
        and the accumulated notes.

    Raises
    ------
    IngestError
        Propagated from the grid completers.
    """
    declared_codes = {str(code) for code in WDI_CONTROL_CODES.values()}
    if "indicator_code" not in wdi_frame.columns:
        raise IngestError("WDI controls lack the indicator_code column.")
    available_codes = set(wdi_frame["indicator_code"].dropna().astype(str))
    missing_codes = sorted(declared_codes - available_codes)
    if missing_codes:
        raise IngestError(f"WDI controls omit declared indicator code(s) {missing_codes}.")
    # Initialise the completed grids keyed by control name.
    completed: Dict[str, pd.DataFrame] = {}
    # Accumulate the ledger frames.
    ledgers: List[pd.DataFrame] = []
    # Accumulate the notes.
    notes: List[str] = []
    raw_missing_counts: Dict[str, Dict[str, int]] = {}
    # Complete every retrieved control in its declared order.
    for name, code in WDI_CONTROL_CODES.items():
        # Complete the control across the universe.
        grid, ledger, control_notes = complete_control_grid(
            wdi_frame,
            name,
            universe,
            value_column="value",
            indicator_code=code,
            anchor_window=anchor_window,
            output_window=output_window,
            non_negative=name in NON_NEGATIVE_CONTROLS,
        )
        # Record the completed grid under its control name.
        completed[name] = grid
        raw_missing_counts[name] = {
            "anchor_window": int(ledger.attrs["raw_missing_anchor_window"]),
            "output_window": int(ledger.attrs["raw_missing_output_window"]),
        }
        # Append the control's ledger to the accumulator.
        ledgers.append(ledger)
        # Append the control's notes to the accumulator.
        notes.extend(control_notes)
    # Complete the governance control through the same hierarchy, using its
    # own value column and no indicator filter.
    rol_grid, rol_ledger, rol_notes = complete_control_grid(
        wgi_frame,
        RULE_OF_LAW_NAME,
        universe,
        value_column=RULE_OF_LAW_COLUMN,
        indicator_code=None,
        anchor_window=anchor_window,
        output_window=output_window,
        non_negative=False,
    )
    # Record the governance grid under its control name.
    completed[RULE_OF_LAW_NAME] = rol_grid
    raw_missing_counts[RULE_OF_LAW_NAME] = {
        "anchor_window": int(rol_ledger.attrs["raw_missing_anchor_window"]),
        "output_window": int(rol_ledger.attrs["raw_missing_output_window"]),
    }
    # Append the governance ledger to the accumulator.
    ledgers.append(rol_ledger)
    # Append the governance notes to the accumulator.
    notes.extend(rol_notes)
    # Concatenate the ledgers into one frame.
    combined_ledger: pd.DataFrame = pd.concat(ledgers, axis=0, ignore_index=True)
    if bool(
        combined_ledger.duplicated(
            ["country_code", "indicator_name", "year"], keep=False
        ).any()
    ):
        raise IngestError("Combined control repair ledger duplicates a scientific key.")
    if set(combined_ledger["method"].astype(str)) - set(COMPLETION_METHODS):
        raise IngestError("Combined control ledger contains an undeclared C2 method.")
    combined_ledger.attrs["raw_missing_counts"] = copy.deepcopy(raw_missing_counts)
    # Log the combined completion outcome.
    logger.info(
        "retrieved controls completed: %d control(s), %d repaired cell(s).",
        len(completed),
        len(combined_ledger),
    )
    # Return the completed grids, the ledger and the notes.
    return completed, combined_ledger, tuple(notes)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 5, Step 2: Complete the fuel-share primitives and construct the balance
# -------------------------------------------------------------------------------------------------------------------------------


# Construct the fuel-share balance from its independently completed primitives.
def construct_fuel_share_balance(
    import_grid: pd.DataFrame,
    export_grid: pd.DataFrame,
    import_ledger: pd.DataFrame,
    export_ledger: pd.DataFrame,
    output_window: Tuple[int, int] = OUTPUT_WINDOW,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Construct the balance from the completed primitives.

    Parameters
    ----------
    import_grid : pandas.DataFrame
        The completed import-share grid over the emitted years.
    export_grid : pandas.DataFrame
        The completed export-share grid over the emitted years.
    import_ledger : pandas.DataFrame
        The repair ledger of the import primitive.
    export_ledger : pandas.DataFrame
        The repair ledger of the export primitive.
    output_window : tuple of int, optional
        The inclusive analytic window.

    Returns
    -------
    tuple
        The completed balance grid and its provenance ledger, in which a cell
        is recorded when either primitive was repaired.  The derived row has
        no primary method label; ``import_method`` and ``export_method`` retain
        the independent P1--P4 provenance and ``union_repaired`` records the
        balance-level derived flag.

    Raises
    ------
    IngestError
        If the two grids do not share their index and columns, or if the
        subtraction leaves any missing cell.
    """
    if not isinstance(output_window, (tuple, list)) or len(output_window) != 2:
        raise IngestError("Fuel-share output window must contain two years.")
    numeric_window = pd.to_numeric(pd.Series(output_window), errors="coerce")
    if (
        any(isinstance(value, (bool, np.bool_)) for value in output_window)
        or numeric_window.isna().any()
        or not np.isfinite(numeric_window.to_numpy(dtype=float)).all()
        or not all(float(value).is_integer() for value in numeric_window)
    ):
        raise IngestError("Fuel-share output window must contain integral years.")
    output_window = tuple(int(value) for value in numeric_window)
    if output_window[0] > output_window[1]:
        raise IngestError("Fuel-share output window is reversed.")
    # Refuse grids whose indices or columns do not align exactly.
    if not import_grid.index.equals(export_grid.index) or not (
        import_grid.columns.equals(export_grid.columns)
    ):
        # A misaligned subtraction would produce a plausible-looking balance.
        raise IngestError(
            "Fuel-share primitives are not aligned on the country-year grid."
        )
    if (
        import_grid.index.has_duplicates
        or import_grid.columns.has_duplicates
        or bool(import_grid.isna().to_numpy().any())
        or bool(export_grid.isna().to_numpy().any())
        or not np.isfinite(import_grid.to_numpy(dtype=float)).all()
        or not np.isfinite(export_grid.to_numpy(dtype=float)).all()
    ):
        raise IngestError("Fuel-share primitive grids must be unique, finite and complete.")
    if any(not str(code).strip() for code in import_grid.index):
        raise IngestError("Fuel-share primitive grids contain a blank country identity.")
    numeric_columns = pd.to_numeric(pd.Series(import_grid.columns), errors="coerce")
    expected_columns = list(range(output_window[0], output_window[1] + 1))
    if (
        numeric_columns.isna().any()
        or not np.isfinite(numeric_columns.to_numpy(dtype=float)).all()
        or not all(float(value).is_integer() for value in numeric_columns)
        or [int(value) for value in numeric_columns] != expected_columns
    ):
        raise IngestError(
            "Fuel-share primitive grids do not span the exact analytic calendar."
        )
    required_ledger = {
        "country_code", "year", "method", "is_fallback", "anchor_years",
        "anchor_values", "emitted",
    }
    for label, primitive_ledger in (
        ("import", import_ledger), ("export", export_ledger)
    ):
        if not required_ledger.issubset(primitive_ledger.columns):
            raise IngestError(f"Fuel-{label} repair ledger is incomplete.")
        if bool(primitive_ledger.duplicated(["country_code", "year"]).any()):
            raise IngestError(f"Fuel-{label} repair ledger duplicates a country-year.")
        if set(primitive_ledger["method"].astype(str)) - set(COMPLETION_METHODS):
            raise IngestError(f"Fuel-{label} repair ledger has an invalid C2 method.")
    # Compute the balance as the element-wise difference of the primitives.
    balance_grid: pd.DataFrame = (import_grid - export_grid).copy()
    # Refuse a balance that still carries a missing cell.
    if bool(balance_grid.isna().to_numpy().any()):
        # Completed primitives must produce a complete balance.
        raise IngestError(
            "The constructed fuel-share balance carries a missing cell."
        )
    # Collect the repaired cells of the import primitive as a key set.
    import_keys: set = set(
        zip(import_ledger["country_code"], import_ledger["year"])
    )
    # Collect the repaired cells of the export primitive as a key set.
    export_keys: set = set(
        zip(export_ledger["country_code"], export_ledger["year"])
    )
    # Compose the union of the two repair key sets.
    union_keys: set = import_keys | export_keys
    # Build a lookup of the import method labels by country and year.
    import_methods: Dict[Tuple[str, int], str] = {
        (str(row["country_code"]), int(row["year"])): str(row["method"])
        for _, row in import_ledger.iterrows()
    }
    # Build a lookup of the export method labels by country and year.
    export_methods: Dict[Tuple[str, int], str] = {
        (str(row["country_code"]), int(row["year"])): str(row["method"])
        for _, row in export_ledger.iterrows()
    }
    import_fallbacks: Dict[Tuple[str, int], bool] = {
        (str(row["country_code"]), int(row["year"])): bool(row["is_fallback"])
        for _, row in import_ledger.iterrows()
    }
    export_fallbacks: Dict[Tuple[str, int], bool] = {
        (str(row["country_code"]), int(row["year"])): bool(row["is_fallback"])
        for _, row in export_ledger.iterrows()
    }
    # Build explicit primitive-anchor lookups for unioned derived cells.
    import_anchors: Dict[Tuple[str, int], Tuple[Any, Any]] = {
        (str(row["country_code"]), int(row["year"])): (
            row.get("anchor_years", ()),
            row.get("anchor_values", ()),
        )
        for _, row in import_ledger.iterrows()
    }
    # Preserve export-side anchor evidence independently.
    export_anchors: Dict[Tuple[str, int], Tuple[Any, Any]] = {
        (str(row["country_code"]), int(row["year"])): (
            row.get("anchor_years", ()),
            row.get("anchor_values", ()),
        )
        for _, row in export_ledger.iterrows()
    }
    # Accumulate the balance provenance ledger rows.
    ledger_rows: List[Dict[str, Any]] = []
    # Iterate over the union of repaired keys in sorted order.
    for country, year in sorted(union_keys):
        # Read the import method label when the cell was repaired.
        import_method: str = import_methods.get((country, year), "")
        # Read the export method label when the cell was repaired.
        export_method: str = export_methods.get((country, year), "")
        emitted = bool(int(output_window[0]) <= int(year) <= int(output_window[1]))
        import_repaired = bool(import_method)
        export_repaired = bool(export_method)
        completed_value = (
            float(balance_grid.loc[country, int(year)]) if emitted else np.nan
        )
        # A balance row is a derived union flag, not a fifth or compound
        # imputation method. Both primitive methods remain independently visible.
        ledger_rows.append(
            {
                "country_code": country,
                "indicator_name": BALANCE_NAME,
                "year": int(year),
                "method": "",
                "is_fallback": bool(
                    import_fallbacks.get((country, year), False)
                    or export_fallbacks.get((country, year), False)
                ),
                "original_value": np.nan,
                "completed_value": completed_value,
                "import_repaired": import_repaired,
                "export_repaired": export_repaired,
                "import_method": import_method,
                "export_method": export_method,
                "import_is_fallback": import_fallbacks.get((country, year), False),
                "export_is_fallback": export_fallbacks.get((country, year), False),
                "anchor_years": {
                    "import": import_anchors.get((country, year), ((), ()))[0],
                    "export": export_anchors.get((country, year), ((), ()))[0],
                },
                "anchor_values": {
                    "import": import_anchors.get((country, year), ((), ()))[1],
                    "export": export_anchors.get((country, year), ((), ()))[1],
                },
                "import_anchor_years": import_anchors.get((country, year), ((), ()))[0],
                "import_anchor_values": import_anchors.get((country, year), ((), ()))[1],
                "export_anchor_years": export_anchors.get((country, year), ((), ()))[0],
                "export_anchor_values": export_anchors.get((country, year), ((), ()))[1],
                "source_indicator_code": f"{FUEL_IMPORT_CODE}|{FUEL_EXPORT_CODE}",
                "raw_missing": True,
                "union_repaired": True,
                "derived_from_primitives": True,
                "emitted": emitted,
            }
        )
    # Assemble the provenance ledger.
    ledger: pd.DataFrame = pd.DataFrame(
        ledger_rows,
        columns=[
            "country_code",
            "indicator_name",
            "year",
            "method",
            "is_fallback",
            "original_value",
            "completed_value",
            "import_repaired",
            "export_repaired",
            "import_method",
            "export_method",
            "import_is_fallback",
            "export_is_fallback",
            "anchor_years",
            "anchor_values",
            "import_anchor_years",
            "import_anchor_values",
            "export_anchor_years",
            "export_anchor_values",
            "source_indicator_code",
            "raw_missing",
            "union_repaired",
            "derived_from_primitives",
            "emitted",
        ],
    )
    if bool(ledger.duplicated(["country_code", "indicator_name", "year"]).any()):
        raise IngestError("Fuel-share balance ledger duplicates a derived cell.")
    observed_union = int(ledger["emitted"].astype(bool).sum())
    expected_union = len(
        {
            (str(row["country_code"]), int(row["year"]))
            for primitive in (import_ledger, export_ledger)
            for _, row in primitive[primitive["emitted"].astype(bool)].iterrows()
        }
    )
    if observed_union != expected_union:
        raise IngestError("Fuel-share union repair count disagrees with primitive ledgers.")
    ledger.attrs["derived_union_count_output_window"] = observed_union
    ledger.attrs["primitive_raw_missing_counts"] = {
        "import": int(import_ledger.attrs.get("raw_missing_output_window", 0)),
        "export": int(export_ledger.attrs.get("raw_missing_output_window", 0)),
    }
    # Log the derived balance outcome and its declared missing count.
    logger.info(
        "fuel-share balance constructed from completed primitives: %d derived "
        "cell(s) against a declared derived count of %d.",
        len(ledger),
        DECLARED_BALANCE_DERIVED_MISSING,
    )
    # Return the balance grid and its provenance ledger.
    return balance_grid, ledger


# -------------------------------------------------------------------------------------------------------------------------------
# Task 5, Step 3: Audit the method-tag distribution and freeze the tensor
# -------------------------------------------------------------------------------------------------------------------------------


# Assemble the long eight-control tensor from the completed grids.
def assemble_control_tensor(
    completed_controls: Mapping[str, pd.DataFrame],
    balance_grid: pd.DataFrame,
    control_names: Sequence[str],
) -> pd.DataFrame:
    """
    Assemble the long eight-control tensor from the completed grids.

    Parameters
    ----------
    completed_controls : mapping of str to pandas.DataFrame
        The completed control grids keyed by control name, excluding the
        derived balance.
    balance_grid : pandas.DataFrame
        The completed balance grid.
    control_names : sequence of str
        The eight declared control names in their required order.

    Returns
    -------
    pandas.DataFrame
        The long tensor keyed on country, indicator and year.

    Raises
    ------
    IngestError
        If a declared control grid is absent, or if the assembled tensor does
        not carry exactly the declared control set.
    """
    # Accumulate the long forms of the controls.
    frames: List[pd.DataFrame] = []
    # Iterate over the declared control names in their required order.
    for name in control_names:
        # Select the completed grid of the control, using the balance when the
        # name is the derived balance and the mapping otherwise.
        grid: Optional[pd.DataFrame] = (
            balance_grid if name == BALANCE_NAME else completed_controls.get(name)
        )
        # Refuse a declared control whose grid was not supplied.
        if grid is None:
            # The failure names the absent control.
            raise IngestError(f"Completed grid for control {name!r} is absent.")
        # Melt the grid into the long form keyed on country and year.
        long_form: pd.DataFrame = grid.reset_index().melt(
            id_vars="country_code", var_name="year", value_name="value"
        )
        # Record the declared control name on every row.
        long_form["indicator_name"] = name
        # Append the long form to the accumulator.
        frames.append(long_form)
    # Concatenate the long forms into the tensor.
    tensor: pd.DataFrame = pd.concat(frames, ignore_index=True)
    # Coerce the key columns to their declared types.
    tensor["year"] = tensor["year"].astype("int64")
    # Coerce the country and indicator columns to strings.
    tensor["country_code"] = tensor["country_code"].astype(str)
    # Coerce the indicator column to strings.
    tensor["indicator_name"] = tensor["indicator_name"].astype(str)
    # Refuse a tensor that does not carry exactly the declared control set.
    observed_controls: set = set(tensor["indicator_name"].unique())
    # Compare the observed control set against the declared one.
    if observed_controls != set(control_names):
        # The failure names both sets so the omission is visible.
        raise IngestError(
            f"Assembled tensor carries controls {sorted(observed_controls)} "
            f"rather than the declared {sorted(control_names)}."
        )
    # Reorder the columns into the canonical tensor order.
    tensor = tensor[["country_code", "indicator_name", "year", "value"]]
    # Sort by the declared key so the tensor is deterministic.
    tensor = tensor.sort_values(
        ["country_code", "indicator_name", "year"], kind="stable"
    ).reset_index(drop=True)
    # Log the assembled tensor shape.
    logger.info(
        "control tensor assembled: %d row(s) across %d control(s).",
        len(tensor),
        len(control_names),
    )
    # Return the assembled tensor.
    return tensor


# Assert that the completed tensor carries no missing control cell.
def assert_no_missing_controls(tensor: pd.DataFrame) -> None:
    """
    Assert that the completed tensor carries no missing cell.

    Parameters
    ----------
    tensor : pandas.DataFrame
        The assembled long control tensor.

    Returns
    -------
    None
        The function returns nothing when every cell is populated.

    Raises
    ------
    IngestError
        If any control cell remains missing, naming the control and country
        of the first offending cells.
    """
    # Select the rows whose value is still missing.
    missing_rows = tensor[tensor["value"].isna()]
    # Refuse a tensor that still carries a missing control cell.
    if len(missing_rows):
        # Extract the offending keys for the failure message.
        offenders: List[Tuple[str, str, int]] = list(
            missing_rows[["country_code", "indicator_name", "year"]]
            .head(MAX_REPORTED_OFFENDERS)
            .itertuples(index=False, name=None)
        )
        # The failure names the count and the first offending cells.
        raise IngestError(
            f"Completed tensor still carries {len(missing_rows)} missing "
            f"cell(s), for example {offenders}."
        )
    # Log the verified completeness of the tensor.
    logger.info("control tensor verified complete across %d cell(s).", len(tensor))


# Assert that no anchor-window year was emitted into the tensor.
def assert_no_anchor_rows(
    tensor: pd.DataFrame, output_window: Tuple[int, int] = OUTPUT_WINDOW
) -> None:
    """
    Assert that every emitted year lies inside the analytic window.

    Parameters
    ----------
    tensor : pandas.DataFrame
        The assembled long control tensor.
    output_window : tuple of int, optional
        The inclusive analytic window.

    Returns
    -------
    None
        The function returns nothing when no anchor year was emitted.

    Raises
    ------
    IngestError
        If the tensor carries a year outside the analytic window, which would
        mean the anchor years leaked into the emitted panel.
    """
    # Refuse an empty tensor, whose year set would be undefined.
    if tensor.empty:
        # An empty tensor cannot carry the analytic window's rows.
        raise IngestError("The control tensor is empty; no years to verify.")
    # Compute the smallest and largest emitted year.
    min_year: int = int(tensor["year"].min())
    # Compute the largest emitted year.
    max_year: int = int(tensor["year"].max())
    # Refuse a tensor that emits a year outside the analytic window.
    if min_year < int(output_window[0]) or max_year > int(output_window[1]):
        # The failure names the observed and declared ranges.
        raise IngestError(
            f"Tensor emits years {min_year} to {max_year}; the analytic window "
            f"is {output_window[0]} to {output_window[1]}."
        )
    # Scan every distinct emitted year, not only the endpoints, so an interior
    # anchor year cannot pass undetected (remediation B-12).
    distinct_years: np.ndarray = tensor["year"].dropna().unique()
    # Compose the set of anchor years: those outside the analytic window.
    anchor_years: List[int] = sorted(
        int(year)
        for year in distinct_years
        if int(year) < int(output_window[0]) or int(year) > int(output_window[1])
    )
    # Refuse any anchor year's presence, wherever it sits in the tensor.
    if anchor_years:
        # Count the rows carrying an anchor year for the failure message.
        anchor_rows: int = int(tensor["year"].isin(anchor_years).sum())
        # Raise with the anchor years and their row count.
        raise IngestError(
            f"Tensor carries {anchor_rows} anchor row(s) for year(s) "
            f"{anchor_years}; the analytic window is {output_window[0]} to "
            f"{output_window[1]}."
        )
    # Log the verified year range and the full-scan outcome.
    logger.info(
        "tensor year range verified inside the analytic window: %d to %d; "
        "all %d distinct year(s) scanned.",
        min_year,
        max_year,
        len(distinct_years),
    )


# Summarise the repair ledger by method.
def audit_repair_ledger(
    ledger: pd.DataFrame, emitted_only: bool = True
) -> pd.DataFrame:
    """
    Summarise the repaired cells by completion method.

    Parameters
    ----------
    ledger : pandas.DataFrame
        The combined repair ledger, carrying an emission flag per cell.
    emitted_only : bool, optional
        When true, only repairs inside the analytic window are counted, which
        is the basis of the manuscript's declared repaired-cell total because
        the anchor-window cells are never released into the panel. When false,
        every repaired cell including the anchor years is counted.

    Returns
    -------
    pandas.DataFrame
        A four-row table indexed by P1-P4 with primary repaired-cell counts
        and shares. Separately derived balance-union and total repair counts
        are retained in frame attributes so mixed primitive methods are not
        collapsed into an invented primary label.

    Raises
    ------
    IngestError
        If the ledger schema, method vocabulary, fallback status, primitive
        provenance, emission status or completed values are inconsistent.
    """
    required = {
        "country_code", "indicator_name", "year", "method", "is_fallback",
        "original_value", "completed_value", "anchor_years", "anchor_values",
        "raw_missing", "derived_from_primitives", "emitted",
    }
    missing_columns = sorted(required - set(ledger.columns))
    if missing_columns:
        raise IngestError(f"Repair ledger lacks required column(s) {missing_columns}.")
    if bool(ledger.duplicated(["country_code", "indicator_name", "year"]).any()):
        raise IngestError("Repair ledger duplicates a country/indicator/year key.")
    if any(
        not str(value).strip()
        for value in ledger["country_code"].tolist()
        + ledger["indicator_name"].tolist()
    ):
        raise IngestError("Repair ledger carries a blank scientific identity.")
    years = pd.to_numeric(ledger["year"], errors="coerce")
    if (
        years.isna().any()
        or not np.isfinite(years.to_numpy(dtype=float)).all()
        or not all(float(year).is_integer() for year in years)
    ):
        raise IngestError("Repair ledger years must be finite integers.")
    for column in ("is_fallback", "raw_missing", "derived_from_primitives", "emitted"):
        if any(not isinstance(value, (bool, np.bool_)) for value in ledger[column]):
            raise IngestError(f"Repair ledger column {column!r} must be Boolean.")
    if not bool(ledger["raw_missing"].astype(bool).all()) or bool(
        ledger["original_value"].notna().any()
    ):
        raise IngestError("Every repair row must identify an originally missing cell.")
    expected_emitted = years.between(
        int(OUTPUT_WINDOW[0]), int(OUTPUT_WINDOW[1]), inclusive="both"
    )
    if not np.array_equal(
        expected_emitted.to_numpy(dtype=bool), ledger["emitted"].to_numpy(dtype=bool)
    ):
        raise IngestError("Repair ledger emission flags disagree with 2000-2024.")
    derived = ledger["derived_from_primitives"].astype(bool)
    primary = ledger.loc[~derived].copy()
    derived_rows = ledger.loc[derived].copy()
    if set(primary["method"].astype(str)) - set(COMPLETION_METHODS) or bool(
        primary["method"].astype(str).str.strip().eq("").any()
    ):
        raise IngestError("Primary repair methods must be exactly one of P1-P4.")
    if bool(
        primary.loc[primary["is_fallback"].astype(bool), "method"]
        .astype(str)
        .ne(METHOD_LEADING_SHORT)
        .any()
    ):
        raise IngestError("A C2 fallback must remain a P3 method with a separate flag.")
    if bool(primary["anchor_years"].map(lambda value: len(tuple(value)) == 0).any()):
        raise IngestError("A primary repair lacks its source anchor years.")
    if len(derived_rows):
        derived_required = {
            "union_repaired", "import_repaired", "export_repaired",
            "import_method", "export_method", "import_is_fallback",
            "export_is_fallback",
        }
        if not derived_required.issubset(derived_rows.columns):
            raise IngestError("Derived balance rows lack primitive repair provenance.")
        if bool(derived_rows["method"].fillna("").astype(str).str.strip().ne("").any()):
            raise IngestError("A derived balance row cannot claim a primary C2 method.")
        for column in (
            "union_repaired", "import_repaired", "export_repaired",
            "import_is_fallback", "export_is_fallback",
        ):
            if any(not isinstance(value, (bool, np.bool_)) for value in derived_rows[column]):
                raise IngestError(f"Derived balance column {column!r} must be Boolean.")
        for _, row in derived_rows.iterrows():
            import_method, export_method = str(row["import_method"]), str(row["export_method"])
            if (
                not bool(row["union_repaired"])
                or not (bool(row["import_repaired"]) or bool(row["export_repaired"]))
                or bool(import_method) != bool(row["import_repaired"])
                or bool(export_method) != bool(row["export_repaired"])
                or (import_method and import_method not in COMPLETION_METHODS)
                or (export_method and export_method not in COMPLETION_METHODS)
                or bool(row["is_fallback"])
                != (bool(row["import_is_fallback"]) or bool(row["export_is_fallback"]))
            ):
                raise IngestError("Derived balance union provenance is inconsistent.")
    finite_scope = ledger["emitted"].astype(bool)
    completed_values = pd.to_numeric(
        ledger.loc[finite_scope, "completed_value"], errors="coerce"
    )
    if completed_values.isna().any() or not np.isfinite(
        completed_values.to_numpy(dtype=float)
    ).all():
        raise IngestError("An emitted repair lacks a finite completed value.")
    scoped = ledger.loc[ledger["emitted"].astype(bool)].copy() if emitted_only else ledger.copy()
    scoped_primary = scoped.loc[~scoped["derived_from_primitives"].astype(bool)]
    counts = (
        scoped_primary.groupby("method", observed=True).size()
        .reindex(COMPLETION_METHODS, fill_value=0)
        .to_frame(name="repaired_cells")
    )
    primary_total = int(counts["repaired_cells"].sum())
    derived_total = int(scoped["derived_from_primitives"].astype(bool).sum())
    counts["share_of_method_repairs"] = counts["repaired_cells"] / max(1, primary_total)
    counts["share_of_all_repairs"] = counts["repaired_cells"] / max(
        1, primary_total + derived_total
    )
    # Preserve the historical column name while defining its denominator
    # honestly as all emitted repairs, including derived balance union cells.
    counts["share_of_repairs"] = counts["share_of_all_repairs"]
    counts.attrs["primary_method_repaired_cells"] = primary_total
    counts.attrs["derived_balance_repaired_cells"] = derived_total
    counts.attrs["total_repaired_cells"] = primary_total + derived_total
    counts.attrs["anchor_only_repair_records"] = int(
        (~ledger["emitted"].astype(bool)).sum()
    )
    # Log the summarised repairs.
    logger.info(
        "repair ledger audited across %d method(s) over %d %s cell(s).",
        len(counts),
        int(counts.attrs["total_repaired_cells"]),
        "emitted" if emitted_only else "recorded",
    )
    # Return the summary table.
    return counts


# Assert the repaired-cell totals against the declared values.
def assert_repair_totals(
    method_summary: pd.DataFrame,
    expected_total: Optional[int] = None,
    observed_balance_derived: Optional[int] = None,
    expected_balance_derived: Optional[int] = None,
) -> None:
    """
    Assert the repaired-cell totals against the declared values.

    Parameters
    ----------
    method_summary : pandas.DataFrame
        The method summary returned by the ledger audit.
    expected_total : int, optional
        The manuscript's total repaired control cells. When absent the
        comparison is skipped, which is the fixture mode.
    observed_balance_derived : int, optional
        The observed count of derived balance repairs.
    expected_balance_derived : int, optional
        The declared count of derived balance repairs.

    Returns
    -------
    None
        The function returns nothing when every supplied comparison holds.

    Raises
    ------
    IngestError
        If the observed total diverges from the declared total, or if the
        derived balance count diverges from its declared value.
    """
    for label, value in (
        ("expected_total", expected_total),
        ("observed_balance_derived", observed_balance_derived),
        ("expected_balance_derived", expected_balance_derived),
    ):
        if value is not None and (
            isinstance(value, (bool, np.bool_))
            or not isinstance(value, (int, np.integer))
            or int(value) < 0
        ):
            raise IngestError(f"{label} must be a non-negative integer when supplied.")
    # Count primary-method repairs plus the separately identified derived
    # balance union cells; the latter are not assigned an invented C2 method.
    observed_total: int = int(
        method_summary.attrs.get(
            "total_repaired_cells", method_summary["repaired_cells"].sum()
        )
    )
    # Compare the observed total against the declared value when supplied.
    if expected_total is not None and observed_total != expected_total:
        # The failure names both totals.
        raise IngestError(
            f"Repaired control cells total {observed_total}; the declared total "
            f"is {expected_total}."
        )
    # Report the observed total when no declared value was supplied.
    if expected_total is None:
        # Log the observed total for the evidence record.
        logger.info(
            "repaired control cells total %d (no declared total supplied).",
            observed_total,
        )
    # Compare the derived balance count when both values were supplied.
    if (
        expected_balance_derived is not None
        and observed_balance_derived is not None
        and observed_balance_derived != expected_balance_derived
    ):
        # The failure names both derived counts.
        raise IngestError(
            f"Derived balance repairs number {observed_balance_derived}; the "
            f"declared derived count is {expected_balance_derived}."
        )
    # Log the completed total comparison.
    logger.info("repair totals verified: %d repaired cell(s).", observed_total)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 5, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the control completion pipeline over the ingest bundle.
def orchestrate_control_completion(
    bundle: CleansedBundle,
    mask: Optional[pd.Series] = None,
    expected_repaired_total: Optional[int] = None,
    expected_balance_derived: Optional[int] = None,
    enforce_declared_gates: bool = True,
) -> ControlCompletionResult:
    """
    Complete every control and assemble the frozen control tensor.

    Parameters
    ----------
    bundle : CleansedBundle
        The cleansed bundle carrying the cleansed long frames; its handoff
        verification runs before any completion is computed.
    mask : pandas.Series, optional
        The eligibility mask. When supplied, completion is restricted to the
        retained countries; the production call always supplies it.
    expected_repaired_total : int, optional
        The manuscript's total repaired control cells.
    expected_balance_derived : int, optional
        The declared count of derived balance repairs.
    enforce_declared_gates : bool, optional
        When true (the production default), absent declared totals are
        resolved from the module's declared constants so the fidelity gates
        cannot be silently skipped (remediation of CF-2 / B-13).

    Returns
    -------
    ControlCompletionResult
        The tensor, the combined ledger, the method summary, the fingerprint,
        the timestamp and the notes.

    Raises
    ------
    KeyError
        If a required source is absent from the bundle.
    IngestError
        If any completion, assembly or assertion check fails.
    """
    # Verify the cleansed handoff so no raw aggregate row reaches completion.
    bundle.verify_handoff()
    # Freeze a caller-supplied eligibility mask before any scientific work so
    # its exact labels, values and metadata can be proved unchanged on return.
    frozen_mask: Optional[pd.Series] = None
    if mask is not None:
        if not isinstance(mask, pd.Series) or mask.empty or mask.index.has_duplicates:
            raise IngestError("The completion eligibility mask must be a nonempty unique Series.")
        if any(not isinstance(value, (bool, np.bool_)) for value in mask.tolist()):
            raise IngestError("The completion eligibility mask must contain only Boolean values.")
        normalized_mask_codes = tuple(str(code).strip() for code in mask.index)
        if any(not code for code in normalized_mask_codes) or len(
            set(normalized_mask_codes)
        ) != len(normalized_mask_codes):
            raise IngestError("The completion eligibility mask has blank or aliased country labels.")
        frozen_mask = mask.copy(deep=True)
    # Resolve the declared repaired-cell total under the production gate.
    if enforce_declared_gates and expected_repaired_total is None:
        # Default the gate to the manuscript's declared total.
        expected_repaired_total = DECLARED_REPAIRED_CONTROL_CELLS
    # Resolve the declared derived-balance count under the same policy.
    if enforce_declared_gates and expected_balance_derived is None:
        # Default the gate to the declared derived-balance count.
        expected_balance_derived = DECLARED_BALANCE_DERIVED_MISSING
    # Read the cleansed WDI long frame from the bundle.
    wdi_frame: pd.DataFrame = bundle.frame("wdi_raw_long")
    # Read the cleansed WGI long frame from the bundle.
    wgi_frame: pd.DataFrame = bundle.frame("wgi_raw_long")
    # Derive the universe from the mask when one was supplied, and from the
    # WDI frame's country codes otherwise.
    universe: List[str] = (
        sorted(str(code).strip() for code in mask.index[mask.to_numpy(dtype=bool)])
        if mask is not None
        else sorted(str(code).strip() for code in wdi_frame["country_code"].unique())
    )
    # Refuse an empty universe, which cannot be completed.
    if not universe:
        # Without a universe the completion is undefined.
        raise IngestError("The completion universe is empty.")
    if len(universe) != len(set(universe)) or any(not code for code in universe):
        raise IngestError("The completion universe contains a blank or duplicate country code.")
    if enforce_declared_gates and len(universe) != DECLARED_CONTROL_COUNTRIES:
        raise IngestError(
            f"Production control completion requires {DECLARED_CONTROL_COUNTRIES} "
            f"countries; observed {len(universe)}."
        )
    # Complete the six retrieval controls and the governance control.
    completed, ledger, notes = complete_all_controls(
        wdi_frame, wgi_frame, universe
    )
    # Complete the two fuel-share primitives independently.
    import_grid, import_ledger, import_notes = complete_control_grid(
        wdi_frame,
        "FuelImportShare",
        universe,
        value_column="value",
        indicator_code=FUEL_IMPORT_CODE,
        non_negative=False,
    )
    # Complete the fuel-export primitive independently.
    export_grid, export_ledger, export_notes = complete_control_grid(
        wdi_frame,
        "FuelExportShare",
        universe,
        value_column="value",
        indicator_code=FUEL_EXPORT_CODE,
        non_negative=False,
    )
    # Construct the balance from the completed primitives.
    balance_grid, balance_ledger = construct_fuel_share_balance(
        import_grid, export_grid, import_ledger, export_ledger
    )
    # Declare the eight control names in their required order.
    control_names: List[str] = list(WDI_CONTROL_CODES) + [
        RULE_OF_LAW_NAME,
        BALANCE_NAME,
    ]
    # Assemble the long control tensor.
    tensor: pd.DataFrame = assemble_control_tensor(
        completed, balance_grid, control_names
    )
    # Concatenate the retrieval, governance and balance ledgers.
    combined_ledger: pd.DataFrame = pd.concat(
        [ledger, balance_ledger], axis=0, ignore_index=True
    )
    # Record the absence of derived provenance on the non-derived rows.
    # Canonicalize the four ledger-wide flags after concatenation; pandas may
    # otherwise retain object dtype because derived-only columns are sparse.
    for flag_column in (
        "is_fallback", "raw_missing", "derived_from_primitives", "emitted"
    ):
        combined_ledger[flag_column] = combined_ledger[flag_column].astype(bool)
    if bool(
        combined_ledger.duplicated(
            ["country_code", "indicator_name", "year"], keep=False
        ).any()
    ):
        raise IngestError("Combined completion ledger duplicates a scientific key.")
    # Define the exact analytic-window ledger used by manuscript totals.
    emitted_ledger: pd.DataFrame = combined_ledger.loc[
        combined_ledger["emitted"].astype(bool)
        & combined_ledger["year"].between(
            int(OUTPUT_WINDOW[0]), int(OUTPUT_WINDOW[1]), inclusive="both"
        )
    ].copy()
    # Prove the emission flag and the configured analytic years agree exactly.
    if len(emitted_ledger) != int(combined_ledger["emitted"].astype(bool).sum()):
        # An emitted repair outside 2000-2024 would contaminate fixed totals.
        raise IngestError(
            "Repair ledger marks an anchor-only row as emitted."
        )
    # Audit the repair ledger by method.
    method_summary: pd.DataFrame = audit_repair_ledger(
        combined_ledger, emitted_only=True
    )
    # Count the derived balance repairs for the declared comparison.
    observed_balance_derived: int = int(
        emitted_ledger["derived_from_primitives"].astype(bool).sum()
    )
    # Assert the repaired totals against the declared values when supplied.
    assert_repair_totals(
        method_summary,
        expected_total=expected_repaired_total,
        observed_balance_derived=observed_balance_derived,
        expected_balance_derived=expected_balance_derived,
    )
    # Assert that no control cell remains missing.
    assert_no_missing_controls(tensor)
    # Assert that no anchor-window year was emitted.
    assert_no_anchor_rows(tensor)
    # Prove the tensor is the complete declared Cartesian product rather than
    # merely a same-sized collection with duplicated or missing identities.
    tensor_keys = ["country_code", "indicator_name", "year"]
    if bool(tensor.duplicated(tensor_keys, keep=False).any()):
        raise IngestError("Completed control tensor duplicates a scientific key.")
    expected_years = set(range(int(OUTPUT_WINDOW[0]), int(OUTPUT_WINDOW[1]) + 1))
    if (
        len(tensor) != len(universe) * len(control_names) * len(expected_years)
        or set(tensor["country_code"].astype(str)) != set(universe)
        or set(tensor["indicator_name"].astype(str)) != set(control_names)
        or set(tensor["year"].astype(int)) != expected_years
    ):
        raise IngestError(
            "Completed control tensor is not the exact country-control-year product."
        )
    # The completion pipeline consumes but never mutates the eligibility mask.
    if frozen_mask is not None:
        try:
            pd.testing.assert_series_equal(
                mask,
                frozen_mask,
                check_dtype=True,
                check_exact=True,
                check_names=True,
                check_index_type=True,
            )
        except AssertionError as exc:
            raise IngestError(f"Control completion mutated the eligibility mask: {exc}") from exc
    # Compute the key-sorted digest of the tensor.
    fingerprint: str = compute_frame_fingerprint(
        tensor,
        key_columns=["country_code", "indicator_name", "year"],
        source_name="control_tensor",
    )
    # Accumulate the recorded notes with the primitive and balance notes.
    all_notes: List[str] = list(notes) + list(import_notes) + list(export_notes)
    # Record anchor-only provenance separately from emitted repair totals.
    all_notes.append(
        f"anchor-only repair records retained: "
        f"{int((~combined_ledger['emitted'].astype(bool)).sum())}; emitted "
        f"records: {len(emitted_ledger)}."
    )
    # Record the derived balance count as an observation, not as a gate.
    all_notes.append(
        f"derived balance repairs: {observed_balance_derived} cell(s) against a "
        f"declared derived count of {DECLARED_BALANCE_DERIVED_MISSING}."
    )
    all_notes.append(
        f"primary P1-P4 repairs: "
        f"{int(method_summary.attrs['primary_method_repaired_cells'])}; "
        f"derived balance union repairs: "
        f"{int(method_summary.attrs['derived_balance_repaired_cells'])}; "
        f"all emitted repairs: {int(method_summary.attrs['total_repaired_cells'])}."
    )
    # Record whether the declared totals were compared.
    all_notes.append(
        "declared repaired-cell total compared."
        if expected_repaired_total is not None
        else "declared repaired-cell total not supplied; comparison skipped."
    )
    # Assemble the immutable completion result.
    result: ControlCompletionResult = ControlCompletionResult(
        tensor=tensor,
        ledger=combined_ledger,
        method_summary=method_summary,
        fingerprint_sha256=fingerprint,
        completed_utc=current_utc_iso(),
        notes=tuple(all_notes),
    )
    # Log the completed control completion with its fingerprint prefix.
    logger.info(
        "control completion complete: %d control(s), %d cell(s), %d repair(s), "
        "fingerprint %s.",
        len(control_names),
        len(tensor),
        int(method_summary.attrs["total_repaired_cells"]),
        fingerprint[:16],
    )
    # Return the assembled result.
    return result
````

````python
# Task 6: Outcome Repair (Algorithm C3)

# ==============================================================================
# Task 6: Outcome Repair (Algorithm C3)
# ==============================================================================

# Declare the gap class of a leading observation.
GAP_LEADING: Final[str] = "leading"
# Declare the gap class of an internal observation.
GAP_INTERNAL: Final[str] = "internal"
# Declare the gap class of a trailing observation.
GAP_TRAILING: Final[str] = "trailing"
# Declare the gap class of a declared carry-forward observation.
GAP_CARRY_FORWARD: Final[str] = "carry_forward_2023"
# Declare the repair method label of the bounded trend backcast.
METHOD_BACKCAST: Final[str] = "leading_backcast"
# Declare the repair method label of the carry-forward family.
METHOD_CARRY_FORWARD: Final[str] = "carry_forward_2023"
# Declare the declared methods in their reporting order.
OUTCOME_REPAIR_METHODS: Final[Tuple[str, ...]] = (
    METHOD_BACKCAST,
    METHOD_CARRY_FORWARD,
)
# Declare the number of anchors the backcast nominally uses.
BACKCAST_ANCHOR_COUNT: Final[int] = 5
# Declare the minimum number of anchors below which the backcast is refused.
BACKCAST_MIN_ANCHOR_COUNT: Final[int] = 2
# Declare the number of standard deviations added to the backcast bounds.
BACKCAST_BOUND_SIGMA: Final[float] = 1.0
# Declare the leading span the manuscript's coverage audit reports.
LEADING_SPAN: Final[Tuple[int, int]] = (2000, 2007)
# Declare the two carry-forward cells as a declarative registry.
CARRY_FORWARD_CELLS: Final[Tuple[Tuple[str, str, int], ...]] = (
    ("BIH", "CPI", 2024),
    ("BIH", "InflationRate", 2024),
)
# Declare the year the carry-forward values are drawn from.
CARRY_FORWARD_SOURCE_YEAR: Final[int] = 2023
# Declare the manuscript's total repaired outcome cells.
DECLARED_TOTAL_REPAIRS: Final[int] = 57
# Declare the manuscript's repaired-cell count per outcome.
DECLARED_REPAIRS_BY_OUTCOME: Final[Mapping[str, int]] = {
    "CPI": 25,
    "InflationRate": 32,
    "RealGDPpc": 0,
    "RealGDPpcGrowth": 0,
}
# Declare the manuscript's count of the carry-forward family.
DECLARED_CARRY_FORWARD_COUNT: Final[int] = 2


# Declare the immutable result of the outcome repair.
@dataclass(frozen=True)
class OutcomeRepairResult:
    """
    Immutable result of the outcome repair.

    Attributes
    ----------
    tensor : pandas.DataFrame
        The long completed outcome tensor keyed on country, indicator and
        year, carrying the four declared outcomes with no missing cell.
    ledger : pandas.DataFrame
        One row per repaired cell with its gap class, method and anchor count.
    repair_summary : pandas.DataFrame
        One row per outcome with the repaired-cell count and the composition
        by gap class.
    fingerprint_sha256 : str
        Key-sorted digest of the completed tensor.
    repaired_utc : str
        ISO-8601 UTC timestamp of the repair run.
    notes : tuple of str
        Recorded decisions, including reduced-anchor fits and refused cells.
    """
    # The long completed outcome tensor.
    tensor: pd.DataFrame
    # One row per repaired cell with its gap class and method.
    ledger: pd.DataFrame
    # One row per outcome with its repaired-cell count and composition.
    repair_summary: pd.DataFrame
    # Key-sorted digest of the completed tensor.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the repair run.
    repaired_utc: str
    # Recorded decisions of the repair run.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 6, Step 1: Enumerate the raw-missing outcome cells
# -------------------------------------------------------------------------------------------------------------------------------


# Build the country-by-year grid of one outcome over the analytic window.
def _outcome_grid(
    frame: pd.DataFrame,
    indicator_code: str,
    universe: Sequence[str],
    window: Tuple[int, int] = OUTPUT_WINDOW,
) -> pd.DataFrame:
    """
    Build a full country-by-year grid for one outcome.

    Parameters
    ----------
    frame : pandas.DataFrame
        The long frame carrying the outcome.
    indicator_code : str
        The series code of the outcome.
    universe : sequence of str
        The country universe.
    window : tuple of int, optional
        The inclusive analytic window.

    Returns
    -------
    pandas.DataFrame
        A grid indexed by the sorted universe and columned by every year of
        the window, with cells absent from the frame represented as missing.

    Raises
    ------
    IngestError
        If the frame lacks the country or year column.
    """
    # Refuse a frame lacking either column the grid requires.
    for column in ("country_code", "year"):
        # The failure names the absent column.
        if column not in frame.columns:
            # Without these columns the grid cannot be built.
            raise IngestError(f"Outcome frame lacks the column {column!r}.")
    # Restrict the frame to the requested outcome.
    subset: pd.DataFrame = frame[
        frame["indicator_code"].astype(str) == indicator_code
    ].copy()
    # Restrict the subset to the analytic window years.
    subset = subset[
        (subset["year"] >= int(window[0])) & (subset["year"] <= int(window[1]))
    ]
    # Pivot to a country-by-year grid, resolving duplicates by the last value.
    grid: pd.DataFrame = subset.pivot_table(
        index="country_code", columns="year", values="value", aggfunc="last"
    )
    # Compose the full ascending year range of the window.
    full_columns: List[int] = list(range(int(window[0]), int(window[1]) + 1))
    # Reindex the grid to the sorted universe and the full year range.
    grid = grid.reindex(
        index=pd.Index(sorted(str(code) for code in universe), name="country_code"),
        columns=full_columns,
    )
    # Return the full grid.
    return grid


# Classify every missing outcome cell in the universe.
def enumerate_outcome_repairs(
    frame: pd.DataFrame,
    universe: Sequence[str],
    carry_forward_cells: Tuple[Tuple[str, str, int], ...] = CARRY_FORWARD_CELLS,
    window: Tuple[int, int] = OUTPUT_WINDOW,
) -> pd.DataFrame:
    """
    Enumerate and classify the raw-missing outcome cells.

    Parameters
    ----------
    frame : pandas.DataFrame
        The cleansed WDI long frame restricted to the outcomes.
    universe : sequence of str
        The country universe.
    carry_forward_cells : tuple of tuple, optional
        The declared carry-forward cells as country, outcome name and year.
    window : tuple of int, optional
        The inclusive analytic window.

    Returns
    -------
    pandas.DataFrame
        One row per missing cell with its country, outcome name, outcome
        code, year, gap class and whether the class was declared.

    Raises
    ------
    IngestError
        If a missing cell is internal or trailing without being declared as a
        carry-forward, because the manuscript's repair layer covers neither.
    """
    # Accumulate the ledger rows for every outcome.
    ledger_rows: List[Dict[str, Any]] = []
    # Index the declared carry-forward cells for constant-time lookup.
    carry_index: set = set(carry_forward_cells)
    # Classify each outcome in its declaration order.
    for name, code in OUTCOME_CODES.items():
        # Build the country-by-year grid of the outcome.
        grid: pd.DataFrame = _outcome_grid(frame, code, universe, window)
        # Compute the validity of each cell along the year axis.
        valid: pd.DataFrame = grid.notna()
        # Compute whether any valid observation precedes each cell.
        any_before: pd.DataFrame = valid.cumsum(axis=1).shift(axis=1, periods=1).fillna(0) > 0
        # Compute whether any valid observation follows each cell.
        any_after_reversed: pd.DataFrame = valid.iloc[:, ::-1].cumsum(axis=1)
        # Shift the reversed cumulative count to express strict successors.
        any_after: pd.DataFrame = (
            any_after_reversed.shift(axis=1, periods=1).fillna(0) > 0
        ).iloc[:, ::-1]
        # Iterate over every country and year in the grid.
        for country in grid.index:
            # Iterate over the year columns in ascending order.
            for year in grid.columns:
                # Consider only the cells that are missing.
                if pd.notna(grid.at[country, year]):
                    # Observed cells are not repair targets.
                    continue
                # Classify the cell by its neighbours.
                has_before: bool = bool(any_before.at[country, year])
                # Read whether a valid observation follows the cell.
                has_after: bool = bool(any_after.at[country, year])
                # Assign the gap class from the neighbour conditions.
                if has_before and has_after:
                    # A cell bounded on both sides is an internal gap.
                    gap_class: str = GAP_INTERNAL
                elif has_before:
                    # A cell with only a predecessor is a trailing gap.
                    gap_class = GAP_TRAILING
                else:
                    # A cell with no predecessor is a leading gap.
                    gap_class = GAP_LEADING
                # Override the class for a declared carry-forward cell.
                declared: bool = (str(country), name, int(year)) in carry_index
                # Apply the declared override.
                if declared:
                    # The declared family is classified on its own terms.
                    gap_class = GAP_CARRY_FORWARD
                # Refuse an internal cell, which the manuscript never repairs.
                if gap_class == GAP_INTERNAL:
                    # The failure names the offending cell.
                    raise IngestError(
                        f"Internal outcome gap at {country}/{name}/{year}; the "
                        f"outcome layer never interpolates an internal gap."
                    )
                # Refuse an undeclared trailing cell.
                if gap_class == GAP_TRAILING:
                    # The failure names the offending cell.
                    raise IngestError(
                        f"Trailing outcome gap at {country}/{name}/{year} that is "
                        f"not a declared carry-forward; no rule covers it."
                    )
                # Append the classified cell to the ledger.
                ledger_rows.append(
                    {
                        "country_code": str(country),
                        "indicator_name": name,
                        "indicator_code": code,
                        "year": int(year),
                        "gap_type": gap_class,
                        "declared": declared,
                    }
                )
    # Assemble the repair-target ledger.
    ledger: pd.DataFrame = pd.DataFrame(
        ledger_rows,
        columns=[
            "country_code",
            "indicator_name",
            "indicator_code",
            "year",
            "gap_type",
            "declared",
        ],
    )
    # Log the enumeration outcome by gap class.
    counts: Dict[str, int] = (
        ledger["gap_type"].value_counts().to_dict() if len(ledger) else {}
    )
    # Emit the classification counts.
    logger.info("outcome repair targets enumerated: %s.", counts)
    # Return the ledger.
    return ledger


# Assert that no internal repair exists and the classes are declared.
def assert_no_internal_repairs(
    ledger: pd.DataFrame, leading_span: Optional[Tuple[int, int]] = LEADING_SPAN
) -> None:
    """
    Assert that no repaired outcome cell is internal.

    Parameters
    ----------
    ledger : pandas.DataFrame
        The repair-target ledger.
    leading_span : tuple of int, optional
        The declared span within which a leading cell must fall. When absent,
        the span assertion is skipped.

    Returns
    -------
    None
        The function returns nothing when the ledger is clean.

    Raises
    ------
    IngestError
        If any cell is classified internal, if any class is outside the two
        declared families, or if a leading cell falls outside the span.
    """
    # Compute the distinct gap classes present in the ledger.
    classes: set = set(ledger["gap_type"].unique()) if len(ledger) else set()
    # Refuse any class outside the two declared families.
    unexpected: set = classes - {GAP_LEADING, GAP_CARRY_FORWARD}
    # Raise when an undeclared class is present.
    if unexpected:
        # The failure names the undeclared classes.
        raise IngestError(
            f"Repair ledger carries undeclared gap class(es): {sorted(unexpected)}."
        )
    # Select the leading cells when any exist.
    leading = ledger[ledger["gap_type"] == GAP_LEADING]
    # Assert the declared leading span when one was supplied.
    if leading_span is not None and len(leading):
        # Compute the cells outside the declared leading span.
        outside = leading[
            (leading["year"] < int(leading_span[0]))
            | (leading["year"] > int(leading_span[1]))
        ]
        # Raise when a leading cell falls outside the declared span.
        if len(outside):
            # The failure names the offending cells.
            offenders: List[Tuple[Any, ...]] = list(
                outside[["country_code", "indicator_name", "year"]]
                .head(MAX_REPORTED_OFFENDERS)
                .itertuples(index=False, name=None)
            )
            # Raise with the offending cells attached.
            raise IngestError(
                f"{len(outside)} leading cell(s) fall outside the declared span "
                f"{leading_span}: {offenders}."
            )
    # Log the verified classification.
    logger.info(
        "outcome gap classification verified: %d internal cell(s).",
        int((ledger["gap_type"] == GAP_INTERNAL).sum()) if len(ledger) else 0,
    )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 6, Step 2: Apply the bounded backcast and the declared carry-forward
# -------------------------------------------------------------------------------------------------------------------------------


# Repair one outcome series through the backcast and the carry-forward.
def complete_outcome_series(
    values: pd.Series,
    country: str,
    name: str,
    carry_forward_years: Sequence[int] = (),
) -> Tuple[pd.Series, List[Dict[str, Any]], Tuple[str, ...]]:
    """
    Repair one outcome series under the declared rules.

    Parameters
    ----------
    values : pandas.Series
        The analytic-window series of one outcome for one country.
    country : str
        The country code, used in diagnostics and ledger rows.
    name : str
        The outcome name, used in ledger rows.
    carry_forward_years : sequence of int, optional
        The years repaired by carrying the previous year's value forward.

    Returns
    -------
    tuple
        The completed series, the ledger rows of the repaired cells and the
        notes recording reduced-anchor fits.

    Raises
    ------
    IngestError
        If a missing cell cannot be repaired, which happens when fewer than
        the minimum anchors follow a leading gap or when a carry-forward year
        has no predecessor.
    """
    # Copy the series so the caller's object is never mutated.
    completed: pd.Series = values.copy()
    # Accumulate the ledger rows of the repaired cells.
    ledger_rows: List[Dict[str, Any]] = []
    # Accumulate the notes of the repaired series.
    notes: List[str] = []
    # Freeze the original series for changed-cell and preservation assertions.
    original: pd.Series = values.copy(deep=True)
    # Record the observed positions of the series.
    observed_mask: pd.Series = values.notna()
    # Handle a wholly missing series, which no rule can repair.
    if int(observed_mask.sum()) == 0:
        # A series with no observation at all is a defect.
        raise IngestError(f"{country}/{name}: outcome series has no observation.")
    # Extract the observed year positions as floats.
    observed_years: np.ndarray = values.index[observed_mask].to_numpy(dtype=float)
    # Extract the observed values aligned with those positions.
    observed_values: np.ndarray = values[observed_mask].to_numpy(dtype=float)
    # Record the first observed year position.
    first_year: float = float(observed_years[0])
    # Select the missing cells that precede the first observation.
    leading_cells = values.index[
        values.isna().to_numpy() & (values.index.to_numpy(dtype=float) < first_year)
    ]
    # Repair every leading cell through the bounded trend backcast.
    for year in leading_cells:
        # Fit the trend on the first five subsequent anchors, or fewer when
        # the series does not provide five.
        anchor_years: np.ndarray = observed_years[:BACKCAST_ANCHOR_COUNT]
        # Select the anchor values aligned with those years.
        anchor_values: np.ndarray = observed_values[:BACKCAST_ANCHOR_COUNT]
        # Refuse the cell when even the minimum anchor count is unavailable.
        if len(anchor_values) < BACKCAST_MIN_ANCHOR_COUNT:
            # A single point cannot define a trend.
            raise IngestError(
                f"{country}/{name}/{year}: only {len(anchor_values)} subsequent "
                f"anchor(s); the minimum is {BACKCAST_MIN_ANCHOR_COUNT}."
            )
        # Fit the regression of value on year over the anchors.
        regression = scipy_stats.linregress(anchor_years, anchor_values)
        # Evaluate the fitted trend at the missing year.
        fitted: float = float(
            regression.slope * float(year) + regression.intercept
        )
        # Compute the lower bound from the observed series' minimum.
        lower_bound: float = float(observed_values.min()) - (
            BACKCAST_BOUND_SIGMA * float(observed_values.std(ddof=1))
        )
        # Compute the upper bound from the observed series' maximum.
        upper_bound: float = float(observed_values.max()) + (
            BACKCAST_BOUND_SIGMA * float(observed_values.std(ddof=1))
        )
        # Bound the extrapolation within the declared range.
        bounded: float = float(np.clip(fitted, lower_bound, upper_bound))
        # Assign the bounded value to the leading cell.
        completed.loc[year] = bounded
        # Record the repair with its anchor count.
        ledger_rows.append(
            {
                "country_code": country,
                "indicator_name": name,
                "year": int(year),
                "gap_type": GAP_LEADING,
                "method": METHOD_BACKCAST,
                "anchor_count": int(len(anchor_values)),
                "reduced_anchors": bool(len(anchor_values) < BACKCAST_ANCHOR_COUNT),
            }
        )
        # Record a reduced-anchor note when the nominal count was not met.
        if len(anchor_values) < BACKCAST_ANCHOR_COUNT:
            # Append the departure to the notes.
            notes.append(
                f"{country}/{name}/{year}: backcast fitted on "
                f"{len(anchor_values)} anchor(s) rather than "
                f"{BACKCAST_ANCHOR_COUNT}."
            )
    # Build the approved carry-forward set only from originally missing cells.
    approved_carry: Tuple[int, ...] = tuple(
        int(year)
        for year in carry_forward_years
        if int(year) in values.index and pd.isna(values.loc[int(year)])
    )
    # Repair every approved declared carry-forward cell.
    for year in approved_carry:
        # Compute the source year of the carried value.
        source_year: int = int(year) - 1
        # Refuse a carry-forward whose predecessor is unavailable.
        if source_year not in values.index or pd.isna(values.loc[source_year]):
            # Without the predecessor the value cannot be carried.
            raise IngestError(
                f"{country}/{name}/{year}: carry-forward source year "
                f"{source_year} is unavailable."
            )
        # Assign the predecessor's value to the carry-forward cell.
        completed.loc[year] = float(values.loc[source_year])
        # Record the repair under its own method.
        ledger_rows.append(
            {
                "country_code": country,
                "indicator_name": name,
                "year": int(year),
                "gap_type": GAP_CARRY_FORWARD,
                "method": METHOD_CARRY_FORWARD,
                "anchor_count": 0,
                "reduced_anchors": False,
                "anchor_years": (source_year,),
                "anchor_values": (float(values.loc[source_year]),),
                "original_value": np.nan,
                "repaired_value": float(values.loc[source_year]),
            }
        )
    # Enrich leading-repair rows with exact anchors and changed values.
    for row in ledger_rows:
        # Carry-forward rows already carry their single source anchor.
        if row["method"] == METHOD_CARRY_FORWARD:
            # No further enrichment is required.
            continue
        # Use the first subsequent valid observations fixed before repair.
        anchor_years = tuple(
            int(year) for year in observed_years[: int(row["anchor_count"])]
        )
        # Preserve the corresponding original values.
        anchor_values = tuple(
            float(value) for value in observed_values[: int(row["anchor_count"])]
        )
        # Attach the exact anchor and changed-value evidence.
        row["anchor_years"] = anchor_years
        # Preserve the exact observed anchor values.
        row["anchor_values"] = anchor_values
        # The approved target was missing in the original series.
        row["original_value"] = np.nan
        # Record the generated value emitted for the cell.
        row["repaired_value"] = float(completed.loc[int(row["year"])])
    # Identify any original missing cell the approved rules did not repair.
    unresolved: List[int] = [
        int(year) for year in values.index[values.isna() & completed.isna()]
    ]
    # Refuse ordinary internal/trailing gaps or undeclared endpoints.
    if unresolved:
        # Every missing cell must belong to the approved leading/BIH set.
        raise IngestError(
            f"{country}/{name}: unapproved missing outcome cell(s) remain at "
            f"{unresolved[:MAX_REPORTED_OFFENDERS]}."
        )
    # Prove every originally finite observation is exactly unchanged.
    try:
        # Compare only immutable observed positions.
        pd.testing.assert_series_equal(
            completed.loc[observed_mask],
            original.loc[observed_mask],
            check_names=True,
            check_dtype=False,
            check_exact=True,
        )
    except AssertionError as exc:
        # An observed endpoint must never be relabelled or overwritten.
        raise IngestError(
            f"{country}/{name}: outcome repair altered an observed value: {exc}"
        ) from exc
    # Return the completed series, its ledger rows and its notes.
    return completed, ledger_rows, tuple(notes)


# Repair every outcome across the universe.
def complete_all_outcomes(
    frame: pd.DataFrame,
    universe: Sequence[str],
    carry_forward_cells: Tuple[Tuple[str, str, int], ...] = CARRY_FORWARD_CELLS,
    window: Tuple[int, int] = OUTPUT_WINDOW,
) -> Tuple[Dict[str, pd.DataFrame], pd.DataFrame, Tuple[str, ...]]:
    """
    Repair every outcome across the universe.

    Parameters
    ----------
    frame : pandas.DataFrame
        The cleansed WDI long frame restricted to the outcomes.
    universe : sequence of str
        The country universe.
    carry_forward_cells : tuple of tuple, optional
        The declared carry-forward cells.
    window : tuple of int, optional
        The inclusive analytic window.

    Returns
    -------
    tuple
        The completed grids keyed by outcome name, the combined repair ledger
        and the accumulated notes.

    Raises
    ------
    IngestError
        Propagated from the grid builder and the series repair.
    """
    # Enumerate the approved raw-missing cells before any repair is attempted.
    approved_ledger: pd.DataFrame = enumerate_outcome_repairs(
        frame, universe, carry_forward_cells=carry_forward_cells, window=window
    )
    # Build the immutable approved key set from original missing observations.
    approved_keys: set[Tuple[str, str, int]] = set(
        zip(
            approved_ledger["country_code"].astype(str),
            approved_ledger["indicator_name"].astype(str),
            approved_ledger["year"].astype(int),
        )
    )
    # Initialise the completed grids keyed by outcome name.
    completed: Dict[str, pd.DataFrame] = {}
    # Accumulate the ledger rows of every repaired cell.
    ledger_rows: List[Dict[str, Any]] = []
    # Accumulate the notes of every repaired series.
    notes: List[str] = []
    # Repair each outcome in its declaration order.
    for name, code in OUTCOME_CODES.items():
        # Build the country-by-year grid of the outcome.
        grid: pd.DataFrame = _outcome_grid(frame, code, universe, window)
        # Initialise the completed rows of the outcome.
        completed_rows: Dict[str, pd.Series] = {}
        # Repair each country's series in turn.
        for country in grid.index:
            # Select only enumerated missing carry-forward cells for this series.
            carry_years: List[int] = [
                int(row.year)
                for row in approved_ledger.itertuples(index=False)
                if str(row.country_code) == str(country)
                and str(row.indicator_name) == name
                and str(row.gap_type) == GAP_CARRY_FORWARD
            ]
            # Repair the country's series.
            completed_series, series_rows, series_notes = complete_outcome_series(
                grid.loc[country], str(country), name, carry_years
            )
            # Record the completed series for the country.
            completed_rows[str(country)] = completed_series
            # Append the repaired cells to the ledger accumulator.
            ledger_rows.extend(series_rows)
            # Append the notes to the accumulator.
            notes.extend(series_notes)
        # Assemble the completed grid of the outcome.
        completed_grid: pd.DataFrame = pd.DataFrame(completed_rows).T
        # Restore the country index name discarded by the transposition.
        completed_grid.index.name = "country_code"
        # Record the completed grid under its outcome name.
        completed[name] = completed_grid
    # Assemble the combined ledger.
    ledger: pd.DataFrame = pd.DataFrame(
        ledger_rows,
        columns=[
            "country_code",
            "indicator_name",
            "year",
            "gap_type",
            "method",
            "anchor_count",
            "reduced_anchors",
            "anchor_years",
            "anchor_values",
            "original_value",
            "repaired_value",
        ],
    )
    # Build the actual unique repaired-cell key set.
    repaired_keys: set[Tuple[str, str, int]] = set(
        zip(
            ledger["country_code"].astype(str),
            ledger["indicator_name"].astype(str),
            ledger["year"].astype(int),
        )
    )
    # Refuse duplicated repair rows before comparing set identities.
    if len(repaired_keys) != len(ledger):
        # One missing cell must map to exactly one repair record.
        raise IngestError("Outcome repair ledger contains duplicate cell keys.")
    # Require exact one-to-one equality with the pre-repair approved set.
    if repaired_keys != approved_keys:
        # Report missing and surplus repairs separately.
        raise IngestError(
            "Completed outcome keys diverge from approved raw-missing keys: "
            f"missing {sorted(approved_keys - repaired_keys)[:MAX_REPORTED_OFFENDERS]}, "
            f"surplus {sorted(repaired_keys - approved_keys)[:MAX_REPORTED_OFFENDERS]}."
        )
    # Log the combined repair outcome by method.
    counts: Dict[str, int] = (
        ledger["method"].value_counts().to_dict() if len(ledger) else {}
    )
    # Emit the repair counts.
    logger.info("outcome series repaired: %s.", counts)
    # Return the completed grids, the ledger and the notes.
    return completed, ledger, tuple(notes)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 6, Step 3: Enforce the no-internal-interpolation rule and freeze
# -------------------------------------------------------------------------------------------------------------------------------


# Assemble the long outcome tensor from the completed grids.
def assemble_outcome_tensor(
    completed: Mapping[str, pd.DataFrame],
) -> pd.DataFrame:
    """
    Assemble the long outcome tensor from the completed grids.

    Parameters
    ----------
    completed : mapping of str to pandas.DataFrame
        The completed outcome grids keyed by outcome name.

    Returns
    -------
    pandas.DataFrame
        The long tensor keyed on country, indicator and year.

    Raises
    ------
    IngestError
        If the completed grids do not carry exactly the declared outcomes.
    """
    # Accumulate the long forms of the outcomes.
    frames: List[pd.DataFrame] = []
    # Iterate over the declared outcomes in their order.
    for name in OUTCOME_CODES:
        # Refuse a declared outcome whose grid was not supplied.
        if name not in completed:
            # The failure names the absent outcome.
            raise IngestError(f"Completed grid for outcome {name!r} is absent.")
        # Melt the grid into the long form keyed on country and year.
        long_form: pd.DataFrame = (
            completed[name]
            .reset_index()
            .melt(id_vars="country_code", var_name="year", value_name="value")
        )
        # Record the outcome name and its series code on every row.
        long_form["indicator_name"] = name
        # Record the series code of the outcome.
        long_form["indicator_code"] = OUTCOME_CODES[name]
        # Append the long form to the accumulator.
        frames.append(long_form)
    # Concatenate the long forms into the tensor.
    tensor: pd.DataFrame = pd.concat(frames, ignore_index=True)
    # Coerce the key columns to their declared types.
    tensor["year"] = tensor["year"].astype("int64")
    # Coerce the country and indicator columns to strings.
    tensor["country_code"] = tensor["country_code"].astype(str)
    # Sort the tensor by its declared key for determinism.
    tensor = tensor.sort_values(
        ["country_code", "indicator_name", "year"], kind="stable"
    ).reset_index(drop=True)
    # Log the assembled tensor shape.
    logger.info(
        "outcome tensor assembled: %d row(s) across %d outcome(s).",
        len(tensor),
        len(OUTCOME_CODES),
    )
    # Return the assembled tensor.
    return tensor


# Assert that the completed tensor carries no missing outcome cell.
def assert_no_missing_outcomes(tensor: pd.DataFrame) -> None:
    """
    Assert that the completed outcome tensor carries no missing cell.

    Parameters
    ----------
    tensor : pandas.DataFrame
        The assembled long outcome tensor.

    Returns
    -------
    None
        The function returns nothing when every cell is populated.

    Raises
    ------
    IngestError
        If any outcome cell remains missing, naming the first offenders.
    """
    # Select the rows whose value is still missing.
    missing_rows = tensor[tensor["value"].isna()]
    # Refuse a tensor that still carries a missing outcome cell.
    if len(missing_rows):
        # Extract the offending keys for the failure message.
        offenders: List[Tuple[str, str, int]] = list(
            missing_rows[["country_code", "indicator_name", "year"]]
            .head(MAX_REPORTED_OFFENDERS)
            .itertuples(index=False, name=None)
        )
        # The failure names the count and the first offending cells.
        raise IngestError(
            f"Completed outcome tensor still carries {len(missing_rows)} missing "
            f"cell(s), for example {offenders}."
        )
    # Log the verified completeness of the tensor.
    logger.info("outcome tensor verified complete across %d cell(s).", len(tensor))


# Summarise the repair ledger by outcome and gap class.
def summarize_outcome_repairs(ledger: pd.DataFrame) -> pd.DataFrame:
    """
    Summarise the repaired outcome cells by outcome and gap class.

    Parameters
    ----------
    ledger : pandas.DataFrame
        The combined repair ledger.

    Returns
    -------
    pandas.DataFrame
        A table indexed by outcome with the repaired-cell count, the leading
        count and the carry-forward count.

    Raises
    ------
    IngestError
        If the ledger lacks a required column.
    """
    # Refuse a ledger lacking any required column.
    for column in ("indicator_name", "gap_type"):
        # The failure names the absent column.
        if column not in ledger.columns:
            # Without these columns the summary cannot be built.
            raise IngestError(f"Repair ledger lacks the column {column!r}.")
    # Build one summary row per declared outcome.
    rows: List[Dict[str, Any]] = []
    # Iterate over the declared outcomes in their order.
    for name in OUTCOME_CODES:
        # Select the repaired cells of the outcome.
        scoped = ledger[ledger["indicator_name"] == name]
        # Record the outcome's counts by class.
        rows.append(
            {
                "indicator_name": name,
                "repaired_cells": int(len(scoped)),
                "leading_cells": int((scoped["gap_type"] == GAP_LEADING).sum()),
                "carry_forward_cells": int(
                    (scoped["gap_type"] == GAP_CARRY_FORWARD).sum()
                ),
                "reduced_anchor_cells": int(
                    scoped["reduced_anchors"].sum() if len(scoped) else 0
                ),
            }
        )
    # Assemble the summary with the outcome name as the index.
    summary: pd.DataFrame = pd.DataFrame(rows).set_index("indicator_name")
    # Log the summarised repairs.
    logger.info("outcome repair summary assembled across %d outcome(s).", len(summary))
    # Return the summary table.
    return summary


# Assert the repaired counts against the declared manuscript figures.
def assert_declared_repair_counts(
    summary: pd.DataFrame,
    expected_total: Optional[int] = None,
    expected_by_outcome: Optional[Mapping[str, int]] = None,
    expected_carry_forward: Optional[int] = None,
) -> None:
    """
    Assert the repaired counts against the declared figures.

    Parameters
    ----------
    summary : pandas.DataFrame
        The repair summary.
    expected_total : int, optional
        The declared total repaired outcome cells.
    expected_by_outcome : mapping of str to int, optional
        The declared repaired-cell count per outcome.
    expected_carry_forward : int, optional
        The declared count of the carry-forward family.

    Returns
    -------
    None
        The function returns nothing when every supplied comparison holds.

    Raises
    ------
    IngestError
        If any supplied declared count diverges from the observed count.
    """
    # Compute the observed total repaired cells.
    observed_total: int = int(summary["repaired_cells"].sum())
    # Compare the total against the declared value when supplied.
    if expected_total is not None and observed_total != expected_total:
        # The failure names both totals.
        raise IngestError(
            f"Repaired outcome cells total {observed_total}; the declared total "
            f"is {expected_total}."
        )
    # Compare the per-outcome counts when supplied.
    if expected_by_outcome is not None:
        # Iterate over the declared outcomes.
        for name, expected in expected_by_outcome.items():
            # Read the observed count, defaulting to zero.
            observed: int = int(summary["repaired_cells"].get(name, 0))
            # Raise when the outcome's count diverges.
            if observed != int(expected):
                # The failure names the outcome and both counts.
                raise IngestError(
                    f"{name}: {observed} repaired cell(s) against a declared "
                    f"{expected}."
                )
    # Compare the carry-forward count when supplied.
    if expected_carry_forward is not None:
        # Compute the observed carry-forward total.
        observed_cf: int = int(summary["carry_forward_cells"].sum())
        # Raise when the family count diverges.
        if observed_cf != int(expected_carry_forward):
            # The failure names both counts.
            raise IngestError(
                f"Carry-forward repairs number {observed_cf}; the declared count "
                f"is {expected_carry_forward}."
            )
    # Log the completed comparison.
    logger.info(
        "declared repair counts verified: %d repaired cell(s), %d carry-forward.",
        observed_total,
        int(summary["carry_forward_cells"].sum()),
    )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 6, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the outcome repair pipeline over the ingest bundle.
def orchestrate_outcome_repair(
    bundle: CleansedBundle,
    mask: Optional[pd.Series] = None,
    carry_forward_cells: Tuple[Tuple[str, str, int], ...] = CARRY_FORWARD_CELLS,
    leading_span: Optional[Tuple[int, int]] = LEADING_SPAN,
    expected_total: Optional[int] = None,
    expected_by_outcome: Optional[Mapping[str, int]] = None,
    expected_carry_forward: Optional[int] = None,
    enforce_declared_gates: bool = True,
) -> OutcomeRepairResult:
    """
    Repair the raw-missing outcome cells and freeze the tensor.

    Parameters
    ----------
    bundle : CleansedBundle
        The cleansed bundle carrying the cleansed WDI long frame; its handoff
        verification runs before any repair is computed.
    mask : pandas.Series, optional
        The eligibility mask. When supplied, repair is restricted to the
        retained countries; the production call always supplies it.
    carry_forward_cells : tuple of tuple, optional
        The declared carry-forward cells.
    leading_span : tuple of int, optional
        The declared span within which a leading cell must fall, which is the
        manuscript's coverage-audit window. A fixture that deliberately
        constructs a longer leading gap passes None to disable the span audit
        explicitly; the production call keeps the declared span.
    expected_total : int, optional
        The declared total repaired outcome cells.
    expected_by_outcome : mapping of str to int, optional
        The declared repaired-cell count per outcome.
    expected_carry_forward : int, optional
        The declared count of the carry-forward family.
    enforce_declared_gates : bool, optional
        When true (the production default), absent declared counts are
        resolved from the module's declared constants so the fidelity gates
        cannot be silently skipped (remediation of CF-2 / B-14).

    Returns
    -------
    OutcomeRepairResult
        The tensor, the ledger, the repair summary, the fingerprint, the
        timestamp and the notes.

    Raises
    ------
    KeyError
        If a required source is absent from the bundle.
    IngestError
        If any classification, repair, assembly or assertion check fails.
    """
    # Verify the cleansed handoff so no raw aggregate row reaches repair.
    bundle.verify_handoff()
    # Resolve the declared total under the production gate policy (B-14).
    if enforce_declared_gates and expected_total is None:
        # Default the gate to the manuscript's declared total.
        expected_total = DECLARED_TOTAL_REPAIRS
    # Resolve the declared per-outcome counts under the same policy.
    if enforce_declared_gates and expected_by_outcome is None:
        # Default the per-outcome gates to the declared mapping.
        expected_by_outcome = dict(DECLARED_REPAIRS_BY_OUTCOME)
    # Resolve the declared carry-forward count under the same policy.
    if enforce_declared_gates and expected_carry_forward is None:
        # Default the carry-forward gate to the declared count.
        expected_carry_forward = DECLARED_CARRY_FORWARD_COUNT
    # Read the cleansed WDI long frame from the bundle.
    frame: pd.DataFrame = bundle.frame("wdi_raw_long")
    # Derive the universe from the mask when one was supplied, and from the
    # frame's own country codes otherwise.
    universe: List[str] = (
        sorted(str(code) for code in mask[mask].index)
        if mask is not None
        else sorted(str(code) for code in frame["country_code"].unique())
    )
    # Refuse an empty universe, which cannot be repaired.
    if not universe:
        # Without a universe the repair is undefined.
        raise IngestError("The outcome repair universe is empty.")
    # Enumerate and classify the raw-missing outcome cells.
    target_ledger: pd.DataFrame = enumerate_outcome_repairs(
        frame, universe, carry_forward_cells
    )
    # Assert that no internal repair exists and the classes are declared.
    assert_no_internal_repairs(target_ledger, leading_span=leading_span)
    # Repair every outcome series.
    completed, ledger, notes = complete_all_outcomes(
        frame, universe, carry_forward_cells
    )
    # Assert that the repaired cells match the enumerated targets exactly.
    target_keys: set = set(
        zip(
            target_ledger["country_code"],
            target_ledger["indicator_name"],
            target_ledger["year"],
        )
    )
    # Compose the repaired cell keys.
    repaired_keys: set = set(
        zip(ledger["country_code"], ledger["indicator_name"], ledger["year"])
    )
    # Refuse a repair set that diverges from the enumerated targets.
    if target_keys != repaired_keys:
        # The failure names the cardinalities of the two sets.
        raise IngestError(
            f"Repaired cells ({len(repaired_keys)}) do not match the enumerated "
            f"targets ({len(target_keys)})."
        )
    # Join target classifications to actual methods one-to-one for reconciliation.
    reconciliation: pd.DataFrame = target_ledger.merge(
        ledger,
        on=["country_code", "indicator_name", "year", "gap_type"],
        how="outer",
        indicator=True,
        validate="one_to_one",
    )
    # Require every approved gap class to map to its single permitted method.
    expected_methods: pd.Series = reconciliation["gap_type"].map(
        {
            GAP_LEADING: METHOD_BACKCAST,
            GAP_CARRY_FORWARD: METHOD_CARRY_FORWARD,
        }
    )
    # Identify missing joins or incorrect method assignments.
    method_mismatch: pd.Series = (
        reconciliation["_merge"].ne("both")
        | reconciliation["method"].ne(expected_methods)
    )
    # Refuse any cell whose class/method identity fails reconciliation.
    if bool(method_mismatch.any()):
        # Report a bounded labelled mismatch table.
        offenders: List[Tuple[Any, ...]] = list(
            reconciliation.loc[
                method_mismatch,
                ["country_code", "indicator_name", "year", "gap_type", "method"],
            ]
            .head(MAX_REPORTED_OFFENDERS)
            .itertuples(index=False, name=None)
        )
        # Fail before aggregating declared totals.
        raise IngestError(
            f"Outcome repair class/method reconciliation failed for {offenders}."
        )
    # Summarise the repairs by outcome and class.
    summary: pd.DataFrame = summarize_outcome_repairs(ledger)
    # Assert the declared counts when they were supplied.
    assert_declared_repair_counts(
        summary,
        expected_total=expected_total,
        expected_by_outcome=expected_by_outcome,
        expected_carry_forward=expected_carry_forward,
    )
    # Assemble the long outcome tensor.
    tensor: pd.DataFrame = assemble_outcome_tensor(completed)
    # Assert that no outcome cell remains missing.
    assert_no_missing_outcomes(tensor)
    # Compute the key-sorted digest of the tensor.
    fingerprint: str = compute_frame_fingerprint(
        tensor,
        key_columns=["country_code", "indicator_name", "year"],
        source_name="outcome_tensor",
    )
    # Accumulate the recorded notes.
    all_notes: List[str] = list(notes)
    # Record the universe decision.
    all_notes.append(
        "universe restricted to the eligibility mask."
        if mask is not None
        else "universe derived from the frame; no eligibility mask was supplied."
    )
    # Record whether the declared counts were compared.
    all_notes.append(
        "declared repair counts compared."
        if expected_total is not None
        else "declared repair counts not supplied; comparison skipped."
    )
    # Assemble the immutable repair result.
    result: OutcomeRepairResult = OutcomeRepairResult(
        tensor=tensor,
        ledger=ledger,
        repair_summary=summary,
        fingerprint_sha256=fingerprint,
        repaired_utc=current_utc_iso(),
        notes=tuple(all_notes),
    )
    # Log the completed repair with its fingerprint prefix.
    logger.info(
        "outcome repair complete: %d repaired cell(s), fingerprint %s.",
        len(ledger),
        fingerprint[:16],
    )
    # Return the assembled result.
    return result
````

````python
# Task 7: Analysis Panel Assembly and Freezing

# ==============================================================================
# Task 7: Analysis Panel Assembly and Freezing
# ==============================================================================

# Declare the canonical analysis-panel country identifier column.
COUNTRY_COLUMN: Final[str] = "country_code"
# Declare the canonical analysis-panel annual time identifier column.
YEAR_COLUMN: Final[str] = "year"

# Declare the canonical name of the logged price-index outcome.
CANONICAL_LOG_CPI: Final[str] = "log_CPI"
# Declare the canonical name of the logged real product per capita.
CANONICAL_LOG_GDPPC: Final[str] = "log_real_GDPpc"
# Declare the canonical name of the divided inflation outcome.
CANONICAL_INFLATION_ATT: Final[str] = "inflation_att"
# Declare the canonical name of the divided growth outcome.
CANONICAL_GROWTH_ATT: Final[str] = "growth_att"
# Declare the mechanism-regression name of the inflation rate.
MECHANISM_INFLATION: Final[str] = "inflation_mech"
# Declare the mechanism-regression name of the growth rate.
MECHANISM_GROWTH: Final[str] = "growth_mech"
# Declare the four canonical outcome columns in their required order.
CANONICAL_OUTCOME_COLUMNS: Final[Tuple[str, ...]] = (
    CANONICAL_LOG_CPI,
    CANONICAL_LOG_GDPPC,
    CANONICAL_INFLATION_ATT,
    CANONICAL_GROWTH_ATT,
)
# Declare the two mechanism rate columns in their required order.
MECHANISM_RATE_COLUMNS: Final[Tuple[str, ...]] = (
    MECHANISM_INFLATION,
    MECHANISM_GROWTH,
)
# Declare the eight control columns in their required order.
CONTROL_COLUMNS: Final[Tuple[str, ...]] = tuple(WDI_CONTROL_CODES) + (
    RULE_OF_LAW_NAME,
    FUEL_BALANCE_NAME,
)
# Declare the lag suffix appended to every lagged control column.
LAG_SUFFIX: Final[str] = "_lag1"
# Declare the unit of interpretation of each control for the registry.
CONTROL_UNITS: Final[Mapping[str, str]] = {
    "OilRents": "percent of GDP",
    "TradeOpenness": "percent of GDP",
    "TermsOfTrade": "index",
    "InvestmentRate": "percent of GDP",
    "GovConsumption": "percent of GDP",
    "PopulationGrowth": "annual percent",
    "RuleOfLaw": "index (rescaled level)",
    "FuelShareBalance": "percentage points",
}
# Declare the pre-treatment window of the pooled fit.
PRE_TREATMENT_WINDOW: Final[Tuple[int, int]] = (2001, 2013)
# Declare the balanced-panel window.
PANEL_WINDOW: Final[Tuple[int, int]] = (2000, 2024)


# Declare the immutable bundle of transformed variables.
@dataclass(frozen=True)
class TransformedVariables:
    """
    Immutable bundle of the transformed outcome and control variables.

    Attributes
    ----------
    canonical : pandas.DataFrame
        The panel keyed on country and year carrying the four model-ready
        outcome columns and the eight control columns.
    mechanism_rates : pandas.DataFrame
        The frame keyed identically carrying the two percentage-point rate
        columns used by the mechanism regressions.
    registry : pandas.DataFrame
        One row per transformed variable naming its source indicator, its
        algebraic transformation and its unit of interpretation.
    """
    # The canonical panel with four outcome and eight control columns.
    canonical: pd.DataFrame
    # The percentage-point rate frame used by the mechanism regressions.
    mechanism_rates: pd.DataFrame
    # The transformation registry.
    registry: pd.DataFrame


# Declare the immutable result of the panel assembly.
@dataclass(frozen=True)
class AnalysisPanel:
    """
    Immutable result of the analysis panel assembly and freeze.

    Attributes
    ----------
    panel : pandas.DataFrame
        The frozen canonical panel keyed on country and year, carrying twelve
        columns with no missing cell. Read-only for every consumer.
    mechanism_rates : pandas.DataFrame
        The percentage-point rate frame keyed identically.
    registry : pandas.DataFrame
        The transformation registry.
    lagged : pandas.DataFrame
        The panel with the eight lagged control columns attached.
    pre_treatment : dict of str to dict of str to tuple
        The dependent and covariate matrices keyed first by outcome column and
        then by unit code.
    panel_fingerprint_sha256 : str
        Key-sorted digest of the canonical panel.
    matrices_fingerprint_sha256 : str
        Digest of the canonical rendering of the pre-treatment matrices.
    assembled_utc : str
        ISO-8601 UTC timestamp of the assembly run.
    notes : tuple of str
        Recorded decisions, including the unit set of the matrices.
    """
    # The frozen canonical panel keyed on country and year.
    panel: pd.DataFrame
    # The percentage-point rate frame.
    mechanism_rates: pd.DataFrame
    # The transformation registry.
    registry: pd.DataFrame
    # The panel with the lagged control columns attached.
    lagged: pd.DataFrame
    # The pre-treatment matrices keyed by outcome and unit.
    pre_treatment: Mapping[str, Mapping[str, Tuple[np.ndarray, np.ndarray]]]
    # Key-sorted digest of the canonical panel.
    panel_fingerprint_sha256: str
    # Digest of the canonical rendering of the pre-treatment matrices.
    matrices_fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the assembly run.
    assembled_utc: str
    # Recorded decisions of the assembly run.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Step 1: Apply the log and rate transformations
# -------------------------------------------------------------------------------------------------------------------------------


# Transform the completed outcome tensor into its canonical columns.
def apply_outcome_transformations(
    outcome_tensor: pd.DataFrame,
) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """
    Transform the completed outcomes into their canonical forms.

    Parameters
    ----------
    outcome_tensor : pandas.DataFrame
        The long completed outcome tensor keyed on country, indicator and
        year.

    Returns
    -------
    tuple
        The canonical outcome frame keyed on country and year, the mechanism
        rate frame keyed identically, and the transformation registry.

    Raises
    ------
    IngestError
        If a declared outcome is absent, or if the price index or the real
        product per capita is not strictly positive.
    """
    # Pivot the tensor to a country-by-year frame with one column per outcome.
    wide: pd.DataFrame = outcome_tensor.pivot_table(
        index=["country_code", "year"],
        columns="indicator_name",
        values="value",
        aggfunc="last",
    )
    # Refuse a tensor that omits any declared outcome.
    for name in ("CPI", "RealGDPpc", "InflationRate", "RealGDPpcGrowth"):
        # The failure names the absent outcome.
        if name not in wide.columns:
            # Without the outcome the canonical panel is incomplete.
            raise IngestError(f"Outcome tensor omits the outcome {name!r}.")
    # Assert strict positivity and definedness of the two logged sources
    # before any logarithm; a NaN or infinity fails no ordered comparison, so
    # the screening must be explicit (remediation B-15, eq. 3 fidelity).
    for name in ("CPI", "RealGDPpc"):
        # Read the source column as a float array for uniform elementwise tests.
        source_values: np.ndarray = wide[name].to_numpy(dtype=float)
        # Identify missing entries, which no ordered comparison would catch.
        missing_positions: np.ndarray = np.isnan(source_values)
        # Identify non-finite entries other than the missing ones.
        non_finite_positions: np.ndarray = (
            ~np.isfinite(source_values) & ~missing_positions
        )
        # Identify finite but non-positive entries, for which log is undefined.
        non_positive_positions: np.ndarray = (
            np.isfinite(source_values) & (source_values <= 0.0)
        )
        # Compose the union of every position that forbids the logarithm.
        invalid_positions: np.ndarray = (
            missing_positions | non_finite_positions | non_positive_positions
        )
        # Refuse any invalid source, naming the offending country-year pairs.
        if bool(invalid_positions.any()):
            # Render the offending index keys as plain tuples for the message.
            offenders: List[Tuple[Any, ...]] = [
                tuple(key)
                for key in wide.index[invalid_positions][
                    :MAX_REPORTED_OFFENDERS
                ]
            ]
            # Compose the reason breakdown for the failure message.
            breakdown: str = (
                f"{int(missing_positions.sum())} missing, "
                f"{int(non_finite_positions.sum())} non-finite, "
                f"{int(non_positive_positions.sum())} non-positive"
            )
            # Raise with the breakdown and the first offending keys.
            raise IngestError(
                f"{name} is not strictly positive and defined for "
                f"{int(invalid_positions.sum())} cell(s) ({breakdown}), for "
                f"example {offenders}; the natural logarithm is undefined."
            )
    # Assemble the canonical outcome frame.
    canonical: pd.DataFrame = pd.DataFrame(index=wide.index)
    # Compute the logged price index.
    canonical[CANONICAL_LOG_CPI] = np.log(wide["CPI"].to_numpy(dtype=float))
    # Compute the logged real product per capita.
    canonical[CANONICAL_LOG_GDPPC] = np.log(
        wide["RealGDPpc"].to_numpy(dtype=float)
    )
    # Compute the divided inflation outcome for the analysis tables.
    canonical[CANONICAL_INFLATION_ATT] = (
        wide["InflationRate"].to_numpy(dtype=float) / 100.0
    )
    # Compute the divided growth outcome for the analysis tables.
    canonical[CANONICAL_GROWTH_ATT] = (
        wide["RealGDPpcGrowth"].to_numpy(dtype=float) / 100.0
    )
    # Reset the index so the key columns are addressable.
    canonical = canonical.reset_index()
    # Assemble the mechanism rate frame in raw percentage points.
    mechanism: pd.DataFrame = wide[
        ["InflationRate", "RealGDPpcGrowth"]
    ].rename(
        columns={
            "InflationRate": MECHANISM_INFLATION,
            "RealGDPpcGrowth": MECHANISM_GROWTH,
        }
    )
    # Reset the mechanism frame index so the key columns are addressable.
    mechanism = mechanism.reset_index()
    # Declare the transformation registry rows in their declaration order.
    registry_rows: List[Dict[str, Any]] = [
        {
            "variable": CANONICAL_LOG_CPI,
            "source_indicator": "FP.CPI.TOTL",
            "transformation": "natural_log_after_completion",
            "unit_of_interpretation": "within-country index",
        },
        {
            "variable": CANONICAL_LOG_GDPPC,
            "source_indicator": "NY.GDP.PCAP.KD",
            "transformation": "natural_log_after_completion",
            "unit_of_interpretation": "constant-price",
        },
        {
            "variable": CANONICAL_INFLATION_ATT,
            "source_indicator": "FP.CPI.TOTL.ZG",
            "transformation": "divided_by_100_for_analysis_tables",
            "unit_of_interpretation": "decimal",
        },
        {
            "variable": CANONICAL_GROWTH_ATT,
            "source_indicator": "NY.GDP.PCAP.KD.ZG",
            "transformation": "divided_by_100_for_analysis_tables",
            "unit_of_interpretation": "decimal",
        },
        {
            "variable": MECHANISM_INFLATION,
            "source_indicator": "FP.CPI.TOTL.ZG",
            "transformation": "percentage_points_for_mechanism_regressions",
            "unit_of_interpretation": "percentage points",
        },
        {
            "variable": MECHANISM_GROWTH,
            "source_indicator": "NY.GDP.PCAP.KD.ZG",
            "transformation": "percentage_points_for_mechanism_regressions",
            "unit_of_interpretation": "percentage points",
        },
    ]
    # Assemble the registry frame.
    registry: pd.DataFrame = pd.DataFrame(registry_rows)
    # Log the transformation outcome.
    logger.info(
        "outcome transformations applied: %d canonical column(s) and %d "
        "mechanism rate column(s).",
        len(CANONICAL_OUTCOME_COLUMNS),
        len(MECHANISM_RATE_COLUMNS),
    )
    # Return the canonical frame, the mechanism frame and the registry.
    return canonical, mechanism, registry


# Transform the completed control tensor into its level columns.
def apply_control_transformations(
    control_tensor: pd.DataFrame,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Transform the completed controls into their level columns.

    Parameters
    ----------
    control_tensor : pandas.DataFrame
        The long completed control tensor keyed on country, indicator and
        year.

    Returns
    -------
    tuple
        The country-by-year control frame and the registry rows for the level
        transformations.

    Raises
    ------
    IngestError
        If the tensor omits any of the eight declared controls.
    """
    # Pivot the tensor to a country-by-year frame with one column per control.
    wide: pd.DataFrame = control_tensor.pivot_table(
        index=["country_code", "year"],
        columns="indicator_name",
        values="value",
        aggfunc="last",
    )
    # Refuse a tensor that omits any declared control.
    for name in CONTROL_COLUMNS:
        # The failure names the absent control.
        if name not in wide.columns:
            # Without the control the covariate vector is incomplete.
            raise IngestError(f"Control tensor omits the control {name!r}.")
    # Select and order the eight declared control columns.
    selected: pd.DataFrame = wide[list(CONTROL_COLUMNS)].reset_index()
    # Declare one registry row per control, all at level, with the unit of
    # each control read from the declared unit registry.
    registry_rows: List[Dict[str, Any]] = [
        {
            "variable": name,
            "source_indicator": (
                WDI_CONTROL_CODES.get(name, "")
                if name not in (RULE_OF_LAW_NAME, FUEL_BALANCE_NAME)
                else name
            ),
            "transformation": "level",
            "unit_of_interpretation": CONTROL_UNITS[name],
        }
        for name in CONTROL_COLUMNS
    ]
    # Assemble the registry frame.
    registry: pd.DataFrame = pd.DataFrame(registry_rows)
    # Log the control selection.
    logger.info(
        "control transformations applied: %d level column(s).", len(CONTROL_COLUMNS)
    )
    # Return the selected controls and the registry.
    return selected, registry


# Assemble the canonical panel from the outcome and control frames.
def assemble_transformed_panel(
    outcome_frame: pd.DataFrame, control_frame: pd.DataFrame
) -> pd.DataFrame:
    """
    Assemble the canonical twelve-column panel.

    Parameters
    ----------
    outcome_frame : pandas.DataFrame
        The canonical outcome frame keyed on country and year.
    control_frame : pandas.DataFrame
        The control frame keyed on country and year.

    Returns
    -------
    pandas.DataFrame
        The panel keyed on country and year with the twelve declared columns.

    Raises
    ------
    IngestError
        If the two frames disagree on their country-year coverage, or if the
        assembled panel does not carry exactly twelve canonical columns.
    """
    # Compute the key sets of the two frames.
    outcome_keys: set = set(
        zip(outcome_frame["country_code"], outcome_frame["year"])
    )
    # Compute the control frame keys.
    control_keys: set = set(
        zip(control_frame["country_code"], control_frame["year"])
    )
    # Refuse a coverage mismatch, which would leave missing cells.
    if outcome_keys != control_keys:
        # Report the size and a sample of the divergence.
        only_outcome = sorted(outcome_keys - control_keys)
        # Compose the divergence sample.
        only_control = sorted(control_keys - outcome_keys)
        # The failure names both sides of the divergence.
        raise IngestError(
            f"Outcome and control frames disagree on {len(only_outcome)} and "
            f"{len(only_control)} key(s); for example "
            f"{only_outcome[:MAX_REPORTED_OFFENDERS]} and "
            f"{only_control[:MAX_REPORTED_OFFENDERS]}."
        )
    # Merge the two frames on the country-year key.
    panel: pd.DataFrame = outcome_frame.merge(
        control_frame, on=["country_code", "year"], how="inner", validate="1:1"
    )
    # Sort the panel by its declared key for determinism.
    panel = panel.sort_values(["country_code", "year"], kind="stable").reset_index(
        drop=True
    )
    # Compose the declared canonical column list.
    declared_columns: List[str] = (
        ["country_code", "year"]
        + list(CANONICAL_OUTCOME_COLUMNS)
        + list(CONTROL_COLUMNS)
    )
    # Refuse a panel whose columns diverge from the declared list.
    if list(panel.columns) != declared_columns:
        # The failure names both column lists.
        raise IngestError(
            f"Assembled panel columns diverge from the declared list: "
            f"{list(panel.columns)}."
        )
    # Log the assembled panel shape.
    logger.info(
        "canonical panel assembled: %d row(s), %d column(s).",
        len(panel),
        len(panel.columns),
    )
    # Return the assembled panel.
    return panel


# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Step 2: Apply the one-year lag and stack the pre-treatment matrices
# -------------------------------------------------------------------------------------------------------------------------------


# Lag the eight controls by one year within each country.
def apply_control_lag(panel: pd.DataFrame) -> pd.DataFrame:
    """
    Lag the eight controls by one year within each country.

    Parameters
    ----------
    panel : pandas.DataFrame
        The canonical panel keyed on country and year.

    Returns
    -------
    pandas.DataFrame
        The panel with the eight lagged control columns appended.

    Raises
    ------
    IngestError
        If a control column is absent, or if a lagged cell is missing outside
        the panel's first year, which would mean the lag reached across
        countries.
    """
    # Copy the panel so the caller's object is never mutated.
    lagged: pd.DataFrame = panel.copy()
    # Refuse a panel missing any declared control column.
    for name in CONTROL_COLUMNS:
        # The failure names the absent control.
        if name not in lagged.columns:
            # Without the control the lag cannot be formed.
            raise IngestError(f"Panel lacks the control column {name!r}.")
    # Sort by the key so the grouped shift respects chronological order.
    lagged = lagged.sort_values(["country_code", "year"], kind="stable")
    # Compute the grouped one-year lag of the control columns.
    shifted: pd.DataFrame = lagged.groupby("country_code", observed=True)[
        list(CONTROL_COLUMNS)
    ].shift(1)
    # Append the lagged columns under their suffixed names.
    for name in CONTROL_COLUMNS:
        # Assign the lagged series under its declared suffixed name.
        lagged[f"{name}{LAG_SUFFIX}"] = shifted[name].to_numpy()
    # Determine the panel's first year, the only year where a lag is missing.
    first_year: int = int(lagged["year"].min())
    # Compose the list of lagged column names.
    lagged_columns: List[str] = [f"{name}{LAG_SUFFIX}" for name in CONTROL_COLUMNS]
    # Identify the rows carrying a missing lagged cell.
    missing_lag: pd.DataFrame = lagged[lagged[lagged_columns].isna().any(axis=1)]
    # Refuse a missing lag outside the panel's first year.
    unexpected_rows = missing_lag[missing_lag["year"] != first_year]
    # Raise when a lag crossed a country boundary or skipped a year.
    if len(unexpected_rows):
        # The failure names the offending country-year pairs.
        offenders: List[Tuple[Any, ...]] = list(
            unexpected_rows[["country_code", "year"]]
            .head(MAX_REPORTED_OFFENDERS)
            .itertuples(index=False, name=None)
        )
        # Raise with the offending rows attached.
        raise IngestError(
            f"{len(unexpected_rows)} row(s) carry a missing lag outside the "
            f"panel's first year: {offenders}."
        )
    # Log the lag outcome.
    logger.info(
        "controls lagged by one year; %d row(s) carry the expected first-year "
        "missing lag.",
        len(missing_lag),
    )
    # Return the lagged panel.
    return lagged


# Build the pre-treatment matrices for every outcome and unit.
def build_pre_treatment_matrices(
    lagged: pd.DataFrame,
    units: Sequence[str],
    window: Tuple[int, int] = PRE_TREATMENT_WINDOW,
) -> Dict[str, Dict[str, Tuple[np.ndarray, np.ndarray]]]:
    """
    Build the dependent and covariate matrices of the pooled fit.

    Parameters
    ----------
    lagged : pandas.DataFrame
        The panel with the lagged control columns appended.
    units : sequence of str
        The units whose matrices are built.
    window : tuple of int, optional
        The inclusive pre-treatment window.

    Returns
    -------
    dict
        The matrices keyed first by canonical outcome column and then by unit
        code, each entry holding the thirteen-by-one dependent matrix and the
        thirteen-by-eight covariate matrix.

    Raises
    ------
    IngestError
        If a unit is absent from the panel, or if a stacked matrix does not
        have the declared shape.
    """
    # Compute the length of the pre-treatment window.
    t0: int = int(window[1]) - int(window[0]) + 1
    # Compose the lagged covariate column names in their declared order.
    lagged_columns: List[str] = [f"{name}{LAG_SUFFIX}" for name in CONTROL_COLUMNS]
    # Initialise the matrix accumulator.
    matrices: Dict[str, Dict[str, Tuple[np.ndarray, np.ndarray]]] = {}
    # Iterate over the canonical outcomes in their order.
    for outcome in CANONICAL_OUTCOME_COLUMNS:
        # Initialise the per-unit mapping of the outcome.
        per_unit: Dict[str, Tuple[np.ndarray, np.ndarray]] = {}
        # Build the matrices of each requested unit.
        for unit in units:
            # Select the unit's rows inside the pre-treatment window.
            scoped = lagged[
                (lagged["country_code"].astype(str) == str(unit))
                & (lagged["year"] >= int(window[0]))
                & (lagged["year"] <= int(window[1]))
            ].sort_values("year", kind="stable")
            # Refuse a unit whose window rows are incomplete.
            if len(scoped) != t0:
                # The failure names the unit and the observed row count.
                raise IngestError(
                    f"{unit}: {len(scoped)} pre-treatment row(s) against the "
                    f"declared {t0}."
                )
            # Stack the dependent matrix as a column vector.
            dependent: np.ndarray = (
                scoped[outcome].to_numpy(dtype=float).reshape(t0, 1)
            )
            # Stack the covariate matrix from the lagged control columns.
            covariate: np.ndarray = scoped[lagged_columns].to_numpy(dtype=float)
            # Assert the declared dependent shape.
            if dependent.shape != (t0, 1):
                # The failure names the observed shape.
                raise IngestError(
                    f"{unit}/{outcome}: dependent matrix shape "
                    f"{dependent.shape} against the declared {(t0, 1)}."
                )
            # Assert the declared covariate shape.
            if covariate.shape != (t0, len(CONTROL_COLUMNS)):
                # The failure names the observed shape.
                raise IngestError(
                    f"{unit}/{outcome}: covariate matrix shape "
                    f"{covariate.shape} against the declared "
                    f"{(t0, len(CONTROL_COLUMNS))}."
                )
            # Record the unit's matrices under the outcome.
            per_unit[str(unit)] = (dependent, covariate)
        # Record the outcome's matrices.
        matrices[outcome] = per_unit
    # Log the matrix construction outcome.
    logger.info(
        "pre-treatment matrices built: %d outcome(s) x %d unit(s) x %d year(s).",
        len(matrices),
        len(units),
        t0,
    )
    # Return the matrix mapping.
    return matrices


# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Step 3: Freeze the balanced panel and the pre-treatment matrices
# -------------------------------------------------------------------------------------------------------------------------------


# Assert that the canonical panel is complete and balanced.
def assert_panel_complete(
    panel: pd.DataFrame, expected_shape: Optional[Tuple[int, int, int]] = None
) -> None:
    """
    Assert that the canonical panel carries no missing cell.

    Parameters
    ----------
    panel : pandas.DataFrame
        The canonical panel.
    expected_shape : tuple of int, optional
        The declared countries-by-years-by-variables shape. When supplied, the
        panel's own dimensions are compared with it.

    Returns
    -------
    None
        The function returns nothing when the panel is complete.

    Raises
    ------
    IngestError
        If any canonical column carries a missing cell, or if the supplied
        expected shape diverges from the panel's dimensions.
    """
    # Declare the exact twelve intended variables in their scientific order.
    intended_variables: Tuple[str, ...] = (
        CANONICAL_OUTCOME_COLUMNS + CONTROL_COLUMNS
    )
    # Declare the complete ordered schema including the two key columns.
    intended_columns: List[str] = [
        "country_code",
        "year",
        *intended_variables,
    ]
    # Require exact variable labels and order before any numerical check.
    if list(panel.columns) != intended_columns:
        # Wrong labels/order can silently corrupt downstream matrix stacking.
        raise IngestError(
            "Canonical panel columns/order diverge from the declared schema: "
            f"observed {list(panel.columns)}, expected {intended_columns}."
        )
    # Compose the exact data column list from the declared schema.
    data_columns: List[str] = list(intended_variables)
    # Reject duplicated country-year keys explicitly.
    duplicate_mask: pd.Series = panel.duplicated(
        subset=["country_code", "year"], keep=False
    )
    # A duplicate cannot compensate for an omitted product key.
    if bool(duplicate_mask.any()):
        # Report a bounded list of duplicate keys.
        duplicate_keys: List[Tuple[Any, Any]] = list(
            panel.loc[duplicate_mask, ["country_code", "year"]]
            .head(MAX_REPORTED_OFFENDERS)
            .itertuples(index=False, name=None)
        )
        # Refuse before key-set construction hides duplicates.
        raise IngestError(
            f"Canonical panel carries duplicate country-year key(s) "
            f"{duplicate_keys}."
        )
    # Refuse null or blank country identifiers before string conversion.
    invalid_country: pd.Series = panel["country_code"].isna() | (
        panel["country_code"].astype(str).str.strip() == ""
    )
    # A missing identifier cannot define a Cartesian-product axis.
    if bool(invalid_country.any()):
        # Preserve the offending row locations in the diagnostic.
        raise IngestError(
            "Canonical panel contains null or blank country_code value(s) at "
            f"row(s) {panel.index[invalid_country].tolist()[:MAX_REPORTED_OFFENDERS]}."
        )
    # Parse calendar labels without truncating fractional values.
    numeric_years: pd.Series = pd.to_numeric(panel["year"], errors="coerce")
    # Detect booleans explicitly because Python treats them as integers.
    boolean_years: pd.Series = panel["year"].map(
        lambda value: isinstance(value, (bool, np.bool_))
    )
    # Require finite integral values before any integer conversion.
    invalid_years: pd.Series = (
        boolean_years
        | numeric_years.isna()
        | ~np.isfinite(numeric_years.astype(float))
        | (numeric_years.astype(float) % 1.0 != 0.0)
    )
    # Refuse lossy or undefined calendar labels.
    if bool(invalid_years.any()):
        # Report original values so fractional truncation is visible.
        raise IngestError(
            "Canonical panel contains non-integral or invalid calendar year "
            f"value(s): {panel.loc[invalid_years, 'year'].head(MAX_REPORTED_OFFENDERS).tolist()}."
        )
    # Convert calendar labels only after the conversion is proven lossless.
    validated_years: pd.Series = numeric_years.astype("int64")
    # Resolve the declared country identities from the frame itself.
    countries: Tuple[str, ...] = tuple(
        sorted(set(panel["country_code"].astype(str)))
    )
    # Construct every required calendar year of the balanced panel.
    years: Tuple[int, ...] = tuple(
        range(int(PANEL_WINDOW[0]), int(PANEL_WINDOW[1]) + 1)
    )
    # Construct the exact expected Cartesian country-year product.
    expected_keys: set[Tuple[str, int]] = set(
        itertools.product(countries, years)
    )
    # Resolve actual labelled keys from the losslessly validated years.
    actual_keys: set[Tuple[str, int]] = set(
        zip(panel["country_code"].astype(str), validated_years)
    )
    # Compute omitted and unexpected keys independently.
    missing_keys: List[Tuple[str, int]] = sorted(expected_keys - actual_keys)
    # Compute years/keys outside the declared product.
    extra_keys: List[Tuple[str, int]] = sorted(actual_keys - expected_keys)
    # Require exact key-set equality, not merely equal dimension counts.
    if missing_keys or extra_keys:
        # Report each side of the coverage divergence.
        raise IngestError(
            "Canonical panel does not equal the declared Cartesian key product: "
            f"missing {missing_keys[:MAX_REPORTED_OFFENDERS]}, extra "
            f"{extra_keys[:MAX_REPORTED_OFFENDERS]}."
        )
    # Identify the rows carrying a missing data cell.
    missing_rows = panel[panel[data_columns].isna().any(axis=1)]
    # Refuse a panel that carries any missing data cell.
    if len(missing_rows):
        # The failure names the count and the first offending rows.
        offenders: List[Tuple[Any, ...]] = list(
            missing_rows[["country_code", "year"]]
            .head(MAX_REPORTED_OFFENDERS)
            .itertuples(index=False, name=None)
        )
        # Raise with the offending rows attached.
        raise IngestError(
            f"Canonical panel carries {len(missing_rows)} row(s) with a missing "
            f"data cell, for example {offenders}."
        )
    # Convert all required values to double precision for finiteness checking.
    try:
        # Strict conversion rejects non-numeric objects in model variables.
        numeric_values: np.ndarray = panel[data_columns].to_numpy(dtype=float)
    except (TypeError, ValueError) as exc:
        # Preserve the panel contract's error taxonomy.
        raise IngestError(
            f"Canonical panel contains a non-numeric required value: {exc}"
        ) from exc
    # Reject infinities after the separate missing-value check.
    if not np.isfinite(numeric_values).all():
        # Locate bounded country-year rows with at least one non-finite value.
        invalid_rows: np.ndarray = np.flatnonzero(
            ~np.isfinite(numeric_values).all(axis=1)
        )
        # Preserve labelled row identities in the failure.
        offenders = [
            (
                str(panel.iloc[position]["country_code"]),
                int(panel.iloc[position]["year"]),
            )
            for position in invalid_rows[:MAX_REPORTED_OFFENDERS]
        ]
        # Refuse the non-finite panel before lagging/estimation.
        raise IngestError(
            f"Canonical panel contains non-finite required values at {offenders}."
        )
    # Compare the panel's dimensions with the declared shape when supplied.
    if expected_shape is not None:
        # Compute the panel's own dimensions.
        observed_shape: Tuple[int, int, int] = (
            len(countries),
            len(years),
            len(intended_variables),
        )
        # Refuse a shape that diverges from the declared shape.
        if observed_shape != expected_shape:
            # The failure names both shapes.
            raise IngestError(
                f"Panel shape {observed_shape} diverges from the declared "
                f"{expected_shape}."
            )
    # Log the verified completeness.
    logger.info(
        "canonical panel verified complete across %d row(s).", len(panel)
    )


# Render one array into a canonical text form for digesting.
def _canonical_array_text(array: np.ndarray) -> str:
    """
    Render an array into a deterministic text form.

    Parameters
    ----------
    array : numpy.ndarray
        The array to render.

    Returns
    -------
    str
        The rendered text, with one row per line and every value formatted to
        seventeen significant digits so the rendering is a function of the
        values rather than of the display defaults.
    """
    # Render each row as comma-separated values at full precision.
    return "\n".join(
        ",".join(f"{float(value):.17g}" for value in row) for row in array
    )


# Freeze the pre-treatment matrices behind a canonical digest.
def freeze_pre_treatment_matrices(
    matrices: Mapping[str, Mapping[str, Tuple[np.ndarray, np.ndarray]]],
) -> str:
    """
    Compute the canonical digest of the pre-treatment matrices.

    Parameters
    ----------
    matrices : mapping
        The matrices keyed first by outcome column and then by unit code.

    Returns
    -------
    str
        The hexadecimal digest of the canonical rendering.

    Raises
    ------
    IngestError
        If the matrix mapping is empty.
    """
    # Refuse an empty mapping, which cannot be frozen meaningfully.
    if not matrices:
        # The failure states the precondition.
        raise IngestError("Cannot freeze an empty matrix mapping.")
    # Accumulate the rendered text of every outcome and unit.
    parts: List[str] = []
    # Iterate over the outcomes in sorted order for determinism.
    for outcome in sorted(matrices.keys()):
        # Record the outcome header.
        parts.append(f"outcome={outcome}")
        # Iterate over the units of the outcome in sorted order.
        for unit in sorted(matrices[outcome].keys()):
            # Read the dependent and covariate matrices of the unit.
            dependent, covariate = matrices[outcome][unit]
            # Record the unit header.
            parts.append(f"unit={unit}")
            # Record the dependent matrix rendering.
            parts.append("Y=" + _canonical_array_text(dependent))
            # Record the covariate matrix rendering.
            parts.append("X=" + _canonical_array_text(covariate))
    # Join the rendered parts with a fixed separator.
    rendered: str = "\n--\n".join(parts)
    # Return the digest of the encoded rendering.
    return hashlib.sha256(rendered.encode("utf-8")).hexdigest()


# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the panel assembly and freeze pipeline.
# Declare the manuscript's canonical panel shape for the production gate.
DECLARED_PANEL_SHAPE: Final[Tuple[int, int, int]] = (129, 25, 12)


def orchestrate_panel_assembly(
    outcome_tensor: pd.DataFrame,
    control_tensor: pd.DataFrame,
    units: Optional[Sequence[str]] = None,
    expected_shape: Optional[Tuple[int, int, int]] = None,
    enforce_declared_gates: bool = True,
) -> AnalysisPanel:
    """
    Assemble the analysis panel and freeze it with its matrices.

    Parameters
    ----------
    outcome_tensor : pandas.DataFrame
        The long completed outcome tensor.
    control_tensor : pandas.DataFrame
        The long completed control tensor.
    units : sequence of str, optional
        The units whose pre-treatment matrices are built. When absent, every
        panel country is used; the partition task rebuilds the matrices for
        its own unit sets from the same builder.
    expected_shape : tuple of int, optional
        The declared countries-by-years-by-variables shape.
    enforce_declared_gates : bool, optional
        When true (the production default), an absent declared shape is
        resolved from the module's declared constant so the fidelity gate
        cannot be silently skipped (remediation of CF-2 / B-16).

    Returns
    -------
    AnalysisPanel
        The canonical panel, the mechanism rate frame, the registry, the
        lagged frame, the matrices, both digests, the timestamp and the notes.

    Raises
    ------
    IngestError
        If any transformation, assembly, lag, stacking or assertion fails.
    """
    # Resolve the declared panel shape under the production gate policy.
    if enforce_declared_gates and expected_shape is None:
        # Default the gate to the manuscript's declared (129, 25, 12) shape.
        expected_shape = DECLARED_PANEL_SHAPE
    # Transform the outcomes into their canonical and mechanism forms.
    outcome_frame, mechanism_frame, outcome_registry = (
        apply_outcome_transformations(outcome_tensor)
    )
    # Transform the controls into their level form.
    control_frame, control_registry = apply_control_transformations(control_tensor)
    # Assemble the canonical twelve-column panel.
    panel: pd.DataFrame = assemble_transformed_panel(outcome_frame, control_frame)
    # Lag the eight controls by one year within each country.
    lagged: pd.DataFrame = apply_control_lag(panel)
    # Derive the unit set of the matrices when none was supplied.
    unit_list: List[str] = (
        sorted(str(code) for code in units)
        if units is not None
        else sorted(str(code) for code in panel["country_code"].unique())
    )
    # Build the pre-treatment matrices for every outcome and unit.
    matrices: Dict[str, Dict[str, Tuple[np.ndarray, np.ndarray]]] = (
        build_pre_treatment_matrices(lagged, unit_list)
    )
    # Assert that the canonical panel carries no missing cell.
    assert_panel_complete(panel, expected_shape=expected_shape)
    # Compute the key-sorted digest of the canonical panel.
    panel_digest: str = compute_frame_fingerprint(
        panel, key_columns=["country_code", "year"], source_name="analysis_panel"
    )
    # Compute the canonical digest of the pre-treatment matrices.
    matrices_digest: str = freeze_pre_treatment_matrices(matrices)
    # Compose the transformation registry from both variable families.
    registry: pd.DataFrame = pd.concat(
        [outcome_registry, control_registry], ignore_index=True
    )
    # Accumulate the recorded notes.
    notes: List[str] = [
        f"pre-treatment matrices built for {len(unit_list)} unit(s) over "
        f"{PRE_TREATMENT_WINDOW[0]} to {PRE_TREATMENT_WINDOW[1]}.",
        "the canonical panel and its matrices are frozen and read-only; a "
        "downstream writer receives a copy.",
    ]
    # Record whether the declared shape was compared.
    notes.append(
        "declared panel shape compared."
        if expected_shape is not None
        else "declared panel shape not supplied; comparison skipped."
    )
    # Assemble the immutable analysis panel result.
    result: AnalysisPanel = AnalysisPanel(
        panel=panel,
        mechanism_rates=mechanism_frame,
        registry=registry,
        lagged=lagged,
        pre_treatment=matrices,
        panel_fingerprint_sha256=panel_digest,
        matrices_fingerprint_sha256=matrices_digest,
        assembled_utc=current_utc_iso(),
        notes=tuple(notes),
    )
    # Log the completed assembly with its digest prefixes.
    logger.info(
        "panel assembly complete: %d row(s) x %d column(s), panel %s, "
        "matrices %s.",
        len(panel),
        len(panel.columns),
        panel_digest[:12],
        matrices_digest[:12],
    )
    # Return the assembled result.
    return result
````

````python
# Task 8: Treatment Coding Resolution (Algorithm G0)

# ==============================================================================
# Task 8: Treatment Coding Resolution (Algorithm G0)
# ==============================================================================

# Declare the two policy instruments the resolution covers.
POLICY_VALUES: Final[Tuple[str, str]] = ("sanctions", "direct_arming")
# Declare the indicator column name of the sanctions policy.
SANCTIONS_COLUMN: Final[str] = "S_it"
# Declare the indicator column name of the direct-arming policy.
ARMING_COLUMN: Final[str] = "A_it"
# Declare the rank threshold below which a source is first class.
PRIMARY_RANK_MAX: Final[int] = 3
# Declare the inclusive analysis year range of the indicator tensor.
YEAR_RANGE: Final[Tuple[int, int]] = (2000, 2024)
# Declare the escalation year separating the two waves.
ESCALATION_YEAR: Final[int] = 2022
# Declare the country coded but excluded from the target set.
UKRAINE: Final[str] = "UKR"
# Declare Ukraine's sanctions entry year.
UKRAINE_ENTRY_YEAR: Final[int] = 2014
# Declare the countries with pre-escalation arming and their entry years.
DECLARED_PRE_ESCALATION_ARMING: Final[Mapping[str, int]] = {
    "CZE": 2018,
    "POL": 2018,
    "TUR": 2019,
    "USA": 2018,
}
# Declare the drop reason of an uncorroborated secondary source.
DROP_REASON_SECONDARY: Final[str] = "secondary_without_primary_source"
# Declare the policy-specific documentary evidence admitted by Algorithm G0.
QUALIFYING_POLICY_TYPES: Final[Mapping[str, Tuple[str, ...]]] = {
    "sanctions": ("legal_instrument", "alignment_notice"),
    "direct_arming": ("lethal_delivery", "lethal_commitment"),
}
# Declare the documentary decisions that create a positive canonical code.
POSITIVE_CODING_DECISIONS: Final[Tuple[str, ...]] = (
    "accepted",
    "excluded_from_target",
)


# Declare the immutable result of the treatment coding resolution.
@dataclass(frozen=True)
class TreatmentCodingResult:
    """
    Immutable result of the treatment coding resolution.

    Attributes
    ----------
    resolved : pandas.DataFrame
        One row per country-policy pair with its entry year, its winning
        document and its coding decision.
    indicators : pandas.DataFrame
        One row per country-year pair carrying the two policy indicators.
    decision_log : pandas.DataFrame
        One row per resolved pair recording the ranked candidates and any
        review flag.
    dropped : pandas.DataFrame
        One row per dropped ledger entry with its rank and its reason.
    summary : pandas.DataFrame
        The descriptive category counts indexed by category name.
    fingerprint_sha256 : str
        Digest of the canonical coding payload.
    resolved_utc : str
        ISO-8601 UTC timestamp of the resolution.
    notes : tuple of str
        Recorded decisions, including uncoded pairs and skipped comparisons.
    """
    # One row per country-policy pair with its entry year and decision.
    resolved: pd.DataFrame
    # One row per country-year pair carrying the two indicators.
    indicators: pd.DataFrame
    # One row per resolved pair recording the candidates and review flag.
    decision_log: pd.DataFrame
    # One row per dropped ledger entry with its reason.
    dropped: pd.DataFrame
    # The descriptive category counts.
    summary: pd.DataFrame
    # Digest of the canonical coding payload.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the resolution.
    resolved_utc: str
    # Recorded decisions of the resolution.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Step 1: Group and rank the policy-ledger entries
# -------------------------------------------------------------------------------------------------------------------------------


# Validate, filter and sort the documentary ledger entries.
def filter_and_sort_ledger(
    ledger: pd.DataFrame, primary_rank_max: int = PRIMARY_RANK_MAX
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Filter and sort the documentary ledger under the priority rule.

    Parameters
    ----------
    ledger : pandas.DataFrame
        The canonicalized documentary ledger.
    primary_rank_max : int, optional
        The highest rank still counted as a first-class source.

    Returns
    -------
    tuple
        The sorted surviving entries and the dropped-entry ledger.

    Raises
    ------
    IngestError
        If a required column is absent, if a publication date is not in the
        ISO-8601 calendar form, or if a rank is not a positive integer.
    """
    # Validate the first-class rank threshold before using it.
    if (
        isinstance(primary_rank_max, (bool, np.bool_))
        or not isinstance(primary_rank_max, (int, np.integer))
        or int(primary_rank_max) < 1
    ):
        # A malformed threshold would reclassify documentary sources.
        raise IngestError(
            "primary_rank_max must be a positive integer; observed "
            f"{primary_rank_max!r}."
        )
    # Refuse a ledger lacking any documentary field the rule requires.
    for column in (
        "country_code",
        "policy",
        "issuing_authority",
        "document_title",
        "publication_date",
        "archived_source",
        "policy_type",
        "entry_year",
        "coding_decision",
        "priority_rank",
    ):
        # The failure names the absent column.
        if column not in ledger.columns:
            # Without these columns the rule cannot be applied.
            raise IngestError(f"Policy ledger lacks the column {column!r}.")
    # Copy the ledger so the caller's object is never mutated.
    working: pd.DataFrame = ledger.copy()
    # Require canonical uppercase ISO-like country identifiers without aliases.
    country_codes: pd.Series = working["country_code"].astype(str)
    # Territory and parent-country identifiers remain distinct exact keys.
    invalid_codes: pd.Series = ~country_codes.str.fullmatch(r"[A-Z]{3}")
    # Refuse blank, lower-case or otherwise non-canonical identifiers.
    if bool(invalid_codes.any()):
        # Report the original values rather than silently normalising them.
        raise IngestError(
            "Policy ledger carries non-canonical country_code value(s): "
            f"{working.loc[invalid_codes, 'country_code'].head(MAX_REPORTED_OFFENDERS).tolist()}."
        )
    # Require one of the two canonical policy identifiers.
    invalid_policy: pd.Series = ~working["policy"].astype(str).isin(POLICY_VALUES)
    # Never infer a policy from its evidence type or document title.
    if bool(invalid_policy.any()):
        # Name the invalid policy vocabulary.
        raise IngestError(
            "Policy ledger carries undeclared policy value(s): "
            f"{sorted(set(working.loc[invalid_policy, 'policy'].astype(str)))}."
        )
    # Validate nonblank documentary provenance fields.
    for column in ("issuing_authority", "document_title", "archived_source"):
        # Missing values and whitespace-only strings are not provenance.
        invalid_text: pd.Series = working[column].isna() | (
            working[column].astype(str).str.strip() == ""
        )
        # Refuse incomplete documentary evidence.
        if bool(invalid_text.any()):
            # Preserve the field name in the diagnostic.
            raise IngestError(
                f"Policy ledger carries blank {column!r} value(s) at row(s) "
                f"{working.index[invalid_text].tolist()[:MAX_REPORTED_OFFENDERS]}."
            )
    # Validate every publication date as an actual ISO-8601 calendar date.
    dates: pd.Series = working["publication_date"].astype(str)
    # Parse only exact year-month-day strings; impossible dates become missing.
    parsed_dates: pd.Series = pd.to_datetime(
        dates.where(dates.str.fullmatch(r"\d{4}-\d{2}-\d{2}")),
        format="%Y-%m-%d",
        errors="coerce",
    )
    # Identify syntactically or calendrically invalid dates.
    malformed = working[parsed_dates.isna()]
    # Refuse a ledger carrying a malformed date.
    if len(malformed):
        # The failure names the offending documents and dates.
        offenders: List[Tuple[Any, ...]] = list(
            malformed[["country_code", "policy", "publication_date"]]
            .head(MAX_REPORTED_OFFENDERS)
            .itertuples(index=False, name=None)
        )
        # Raise with the offending rows attached.
        raise IngestError(
            f"{len(malformed)} ledger row(s) carry a publication date that is "
            f"not ISO-8601: {offenders}."
        )
    # Parse ranks without truncating fractional values or accepting booleans.
    ranks: pd.Series = pd.to_numeric(working["priority_rank"], errors="coerce")
    # Refuse a rank that is missing, fractional, non-finite or non-positive.
    invalid_rank_mask: pd.Series = (
        working["priority_rank"].map(lambda value: isinstance(value, (bool, np.bool_)))
        | ranks.isna()
        | ~np.isfinite(ranks.astype(float))
        | (ranks.astype(float) % 1.0 != 0.0)
        | (ranks.astype(float) < 1.0)
    )
    # Select invalid rank records for the diagnostic.
    invalid_ranks = working[invalid_rank_mask]
    # Raise when any rank is invalid.
    if len(invalid_ranks):
        # The failure names the offending rows.
        raise IngestError(
            f"{len(invalid_ranks)} ledger row(s) carry a non-positive rank."
        )
    # Convert ranks only after strict validation succeeds.
    working["priority_rank"] = ranks.astype(int)
    # Validate the decision vocabulary before applying positive-code rules.
    allowed_decisions: set[str] = set(POSITIVE_CODING_DECISIONS) | {"rejected"}
    # Identify unknown decisions rather than treating them as accepted.
    invalid_decision: pd.Series = ~working["coding_decision"].astype(str).isin(
        allowed_decisions
    )
    # Refuse an undeclared adjudication label.
    if bool(invalid_decision.any()):
        # Name the unknown decisions.
        raise IngestError(
            "Policy ledger carries undeclared coding_decision value(s): "
            f"{sorted(set(working.loc[invalid_decision, 'coding_decision'].astype(str)))}."
        )
    # Validate the evidence-type vocabulary independently of its policy fit.
    evidence_vocabulary: set[str] = set().union(
        *(set(values) for values in QUALIFYING_POLICY_TYPES.values())
    )
    # Identify evidence labels outside the declared documentary taxonomy.
    invalid_evidence: pd.Series = ~working["policy_type"].astype(str).isin(
        evidence_vocabulary
    )
    # Refuse unknown evidence rather than assigning it opportunistically.
    if bool(invalid_evidence.any()):
        # Name the unknown evidence labels.
        raise IngestError(
            "Policy ledger carries undeclared policy_type value(s): "
            f"{sorted(set(working.loc[invalid_evidence, 'policy_type'].astype(str)))}."
        )
    # Mark whether each row's evidence category qualifies for its policy.
    evidence_qualifies: pd.Series = working.apply(
        lambda row: str(row["policy_type"])
        in QUALIFYING_POLICY_TYPES[str(row["policy"])],
        axis=1,
    )
    # Mark decisions that create a positive canonical policy code.
    decision_qualifies: pd.Series = working["coding_decision"].astype(str).isin(
        POSITIVE_CODING_DECISIONS
    )
    # Only qualifying positive evidence can establish primary corroboration.
    positive_evidence: pd.Series = evidence_qualifies & decision_qualifies
    # Detect, within each pair, whether qualifying first-class evidence exists.
    has_primary: pd.Series = (
        positive_evidence & (working["priority_rank"] <= int(primary_rank_max))
    ).groupby([working["country_code"], working["policy"]], observed=True).transform("any")
    # Mark every record that cannot enter positive-code resolution.
    rejected_mask: pd.Series = ~decision_qualifies
    # Mark cross-policy evidence, such as nonlethal evidence for arming.
    mismatch_mask: pd.Series = decision_qualifies & ~evidence_qualifies
    # Mark a qualifying secondary source lacking a qualifying primary companion.
    secondary_mask: pd.Series = positive_evidence & (
        working["priority_rank"] > int(primary_rank_max)
    ) & (~has_primary)
    # Combine the mutually exclusive drop conditions.
    drop_mask: pd.Series = rejected_mask | mismatch_mask | secondary_mask
    # Split the dropped entries into their own ledger with the reason.
    dropped: pd.DataFrame = working.loc[drop_mask].copy()
    # Record a reason that distinguishes rejection, mismatch and corroboration.
    dropped["drop_reason"] = np.select(
        [
            rejected_mask.loc[dropped.index],
            mismatch_mask.loc[dropped.index],
            secondary_mask.loc[dropped.index],
        ],
        [
            "coding_decision_rejected",
            "policy_evidence_mismatch",
            DROP_REASON_SECONDARY,
        ],
        default="unclassified_documentary_exclusion",
    )
    # Record which policy the entry belonged to for the audit log.
    dropped["dropped_policy"] = dropped["policy"].astype(str)
    # Select only qualifying, positively adjudicated surviving entries.
    survivors: pd.DataFrame = working.loc[~drop_mask].copy()
    # Assert the conservation identity over row counts.
    if len(survivors) + len(dropped) != len(working):
        # A conservation failure means a row was lost or duplicated.
        raise IngestError("Ledger filter conservation failed.")
    # Sort the survivors by the declared ranking keys.
    survivors = survivors.sort_values(
        ["country_code", "policy", "priority_rank", "publication_date", "entry_year"],
        kind="stable",
    ).reset_index(drop=True)
    # Assert that the sort order is monotone along the declared keys.
    ordered_keys = survivors[["country_code", "policy", "priority_rank"]].to_numpy()
    # Compare each row against its predecessor for monotonicity.
    monotone: bool = all(
        tuple(ordered_keys[index - 1]) <= tuple(ordered_keys[index])
        for index in range(1, len(ordered_keys))
    )
    # Refuse a survivor frame whose ordering is not monotone.
    if not monotone:
        # A non-monotone order means the ranking keys were not applied.
        raise IngestError("Survivor ledger is not sorted by the ranking keys.")
    # Log the filtering outcome with the drop count.
    logger.info(
        "ledger filtered: %d survivor(s), %d dropped secondary entr(ies).",
        len(survivors),
        len(dropped),
    )
    # Return the survivors and the dropped ledger.
    return survivors, dropped


# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Step 2: Resolve one entry per pair and construct the indicators
# -------------------------------------------------------------------------------------------------------------------------------


# Resolve one entry per country-policy pair from the sorted survivors.
def resolve_pair_selection(
    survivors: pd.DataFrame, year_range: Tuple[int, int] = YEAR_RANGE
) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """
    Resolve one entry per country-policy pair.

    Parameters
    ----------
    survivors : pandas.DataFrame
        The sorted surviving entries.
    year_range : tuple of int, optional
        The inclusive analysis year range an entry year must fall inside.

    Returns
    -------
    tuple
        The resolved ledger, the decision log and the uncoded-pair report.

    Raises
    ------
    IngestError
        If an entry year falls outside the declared range.
    """
    # Refuse a survivor frame missing a required provenance column.
    for column in (
        "country_code",
        "policy",
        "entry_year",
        "priority_rank",
        "issuing_authority",
        "document_title",
        "publication_date",
        "archived_source",
        "policy_type",
        "coding_decision",
    ):
        # The failure names the absent column.
        if column not in survivors.columns:
            # Without these columns the resolution cannot run.
            raise IngestError(f"Survivor ledger lacks the column {column!r}.")
    # Parse entry years without truncating malformed values.
    numeric_years: pd.Series = pd.to_numeric(
        survivors["entry_year"], errors="coerce"
    )
    # Reject booleans, missing values, non-finite values and fractions.
    invalid_year_mask: pd.Series = (
        survivors["entry_year"].map(
            lambda value: isinstance(value, (bool, np.bool_))
        )
        | numeric_years.isna()
        | ~np.isfinite(numeric_years.astype(float))
        | (numeric_years.astype(float) % 1.0 != 0.0)
    )
    # Refuse invalid years before applying a range comparison.
    if bool(invalid_year_mask.any()):
        # Report the original values to expose lossy coercions.
        raise IngestError(
            "Survivor ledger carries invalid entry_year value(s): "
            f"{survivors.loc[invalid_year_mask, 'entry_year'].head(MAX_REPORTED_OFFENDERS).tolist()}."
        )
    # Copy before attaching the validated year representation.
    working: pd.DataFrame = survivors.copy()
    # Store entry years only after lossless validation.
    working["entry_year"] = numeric_years.astype("int64")
    # Validate every entry year against the declared analysis range.
    out_of_range = working[
        (working["entry_year"] < int(year_range[0]))
        | (working["entry_year"] > int(year_range[1]))
    ]
    # Refuse an entry outside the declared range.
    if len(out_of_range):
        # The failure names the offending entries.
        offenders: List[Tuple[Any, ...]] = list(
            out_of_range[["country_code", "policy", "entry_year"]]
            .head(MAX_REPORTED_OFFENDERS)
            .itertuples(index=False, name=None)
        )
        # Raise with the offending entries attached.
        raise IngestError(
            f"{len(out_of_range)} ledger entr(ies) carry an entry year outside "
            f"{year_range}: {offenders}."
        )
    # Accumulate the resolved rows and the decision log rows.
    resolved_rows: List[Dict[str, Any]] = []
    # Accumulate the decision log entries.
    decision_rows: List[Dict[str, Any]] = []
    # Group the survivors by country and policy.
    for (country, policy), group in working.groupby(
        ["country_code", "policy"], observed=True, sort=True
    ):
        # Read the group's entries in their sorted order.
        candidates = group.sort_values(
            ["priority_rank", "publication_date", "entry_year"], kind="stable"
        )
        # Determine the top documentary rank within the pair.
        top_rank: int = int(candidates["priority_rank"].iloc[0])
        # Select the entries sharing the winning rank.
        top_entries = candidates[candidates["priority_rank"] == top_rank]
        # Detect substantive disagreement among equally authoritative evidence.
        conflict: bool = bool(
            top_entries["entry_year"].nunique() > 1
            or top_entries["coding_decision"].astype(str).nunique() > 1
        )
        # Start with no recorded adjudication for an uncontested pair.
        adjudication_note: str = ""
        # Resolve a contested top tier only through an explicit selection.
        if conflict:
            # Prefer an explicit row-level selection when the field is present.
            if "adjudication_selected" in top_entries.columns:
                # Parse only literal booleans or binary integer selections.
                selected_mask: pd.Series = top_entries[
                    "adjudication_selected"
                ].map(lambda value: value is True or value == 1)
                # Exactly one top-tier document must be selected.
                if int(selected_mask.sum()) != 1:
                    # Lexical order is never an admissible conflict resolution.
                    raise IngestError(
                        f"{country}/{policy}: conflicting top-rank evidence "
                        "requires exactly one adjudication_selected record."
                    )
                # Select the explicitly adjudicated documentary record.
                winner = top_entries.loc[selected_mask].iloc[0]
            elif "adjudicated_entry_year" in top_entries.columns:
                # Select every explicitly supplied adjudication value.
                recorded_raw: pd.Series = top_entries[
                    "adjudicated_entry_year"
                ].dropna()
                # Parse supplied values without discarding malformed records.
                recorded_numeric: pd.Series = pd.to_numeric(
                    recorded_raw, errors="coerce"
                )
                # Detect booleans, malformed, non-finite and fractional values.
                invalid_recorded: pd.Series = (
                    recorded_raw.map(
                        lambda value: isinstance(value, (bool, np.bool_))
                    )
                    | recorded_numeric.isna()
                    | ~np.isfinite(recorded_numeric.astype(float))
                    | (recorded_numeric.astype(float) % 1.0 != 0.0)
                    | (recorded_numeric.astype(float) < int(year_range[0]))
                    | (recorded_numeric.astype(float) > int(year_range[1]))
                )
                # Require one valid, consistent adjudicated year for the pair.
                if (
                    recorded_raw.empty
                    or bool(invalid_recorded.any())
                    or recorded_numeric.nunique() != 1
                ):
                    # Ambiguous adjudication is not a scientific decision.
                    raise IngestError(
                        f"{country}/{policy}: conflicting top-rank evidence "
                        "requires one finite, integral, in-range and consistent "
                        "adjudicated_entry_year."
                    )
                # Select a top-tier record matching the adjudicated year.
                chosen_year: int = int(recorded_numeric.iloc[0])
                chosen_rows: pd.DataFrame = top_entries[
                    top_entries["entry_year"] == chosen_year
                ]
                # The adjudication cannot introduce an undocumented date.
                if len(chosen_rows) == 0:
                    # Name the invalid adjudicated value.
                    raise IngestError(
                        f"{country}/{policy}: adjudicated year {chosen_year} "
                        "does not match any top-rank documentary record."
                    )
                # Select deterministically within the adjudicated year.
                winner = chosen_rows.iloc[0]
            else:
                # Force contested evidence through a recorded review boundary.
                raise IngestError(
                    f"{country}/{policy}: conflicting top-rank evidence "
                    "requires a recorded adjudication."
                )
            # Require a substantive note for every recorded adjudication.
            if "adjudication_note" not in top_entries.columns:
                # A selection without rationale is incomplete provenance.
                raise IngestError(
                    f"{country}/{policy}: adjudication_note is required for "
                    "conflicting documentary evidence."
                )
            # Collect the nonblank notes attached to the contested group.
            notes: List[str] = [
                str(value).strip()
                for value in top_entries["adjudication_note"].tolist()
                if pd.notna(value) and str(value).strip()
            ]
            # Require at least one preserved rationale.
            if not notes:
                # Refuse a nominal but undocumented selection.
                raise IngestError(
                    f"{country}/{policy}: adjudication_note is blank for "
                    "conflicting documentary evidence."
                )
            # Preserve the complete adjudication rationale.
            adjudication_note = " | ".join(dict.fromkeys(notes))
        else:
            # Select the highest-ranked, earliest published uncontested record.
            winner = top_entries.iloc[0]
        # Record the explicitly selected entry year.
        entry_year: int = int(winner["entry_year"])
        # Serialize every candidate's complete documentary provenance.
        candidate_provenance: str = json.dumps(
            [
                {
                    "issuing_authority": str(row["issuing_authority"]),
                    "document_title": str(row["document_title"]),
                    "publication_date": str(row["publication_date"]),
                    "archived_source": str(row["archived_source"]),
                    "policy_type": str(row["policy_type"]),
                    "entry_year": int(row["entry_year"]),
                    "coding_decision": str(row["coding_decision"]),
                    "priority_rank": int(row["priority_rank"]),
                }
                for _, row in candidates.iterrows()
            ],
            sort_keys=True,
        )
        # Append the resolved row for the pair.
        resolved_rows.append(
            {
                "country_code": str(country),
                "policy": str(policy),
                "entry_year": entry_year,
                "issuing_authority": str(winner["issuing_authority"]),
                "document_title": str(winner["document_title"]),
                "publication_date": str(winner["publication_date"]),
                "archived_source": str(winner["archived_source"]),
                "policy_type": str(winner["policy_type"]),
                "priority_rank": top_rank,
                "coding_decision": str(winner["coding_decision"]),
                "conflict_detected": conflict,
                "adjudication_note": adjudication_note,
                "candidate_provenance": candidate_provenance,
            }
        )
        # Append the decision log row with the candidate documents.
        decision_rows.append(
            {
                "country_code": str(country),
                "policy": str(policy),
                "candidate_count": int(len(candidates)),
                "winning_rank": top_rank,
                "entry_year": entry_year,
                "conflict_review": conflict,
                "adjudicated": conflict,
                "adjudication_note": adjudication_note,
                "candidate_documents": "; ".join(
                    str(title) for title in candidates["document_title"].tolist()
                ),
                "candidate_provenance": candidate_provenance,
            }
        )
    # Assemble the resolved ledger.
    resolved: pd.DataFrame = pd.DataFrame(
        resolved_rows,
        columns=[
            "country_code",
            "policy",
            "entry_year",
            "issuing_authority",
            "document_title",
            "publication_date",
            "archived_source",
            "policy_type",
            "priority_rank",
            "coding_decision",
            "conflict_detected",
            "adjudication_note",
            "candidate_provenance",
        ],
    )
    # Assemble the decision log.
    decision_log: pd.DataFrame = pd.DataFrame(decision_rows)
    # Report the pairs that survived to no entry, which is the complement of
    # the resolved set among the pairs the filter considered.
    considered_pairs: set = set(
        zip(working["country_code"].astype(str), working["policy"].astype(str))
    )
    # Compose the resolved pair set.
    resolved_pairs: set = set(
        zip(resolved["country_code"], resolved["policy"])
    ) if len(resolved) else set()
    # Compose the uncoded-pair report from the pairs with no resolved entry.
    uncoded: pd.DataFrame = pd.DataFrame(
        [
            {"country_code": country, "policy": policy}
            for country, policy in sorted(considered_pairs - resolved_pairs)
        ],
        columns=["country_code", "policy"],
    )
    # Refuse a resolved ledger carrying a duplicated pair.
    if len(resolved) and bool(resolved.duplicated(subset=["country_code", "policy"]).any()):
        # A duplicated pair would make the indicator ambiguous.
        raise IngestError("Resolved ledger carries a duplicated country-policy pair.")
    # Log the resolution outcome.
    logger.info(
        "pairs resolved: %d pair(s), %d conflict review(s), %d uncoded pair(s).",
        len(resolved),
        int(decision_log["conflict_review"].sum()) if len(decision_log) else 0,
        len(uncoded),
    )
    # Return the resolved ledger, the decision log and the uncoded report.
    return resolved, decision_log, uncoded


# Build the country-year indicator tensor from the resolved ledger.
def build_indicator_tensor(
    resolved: pd.DataFrame,
    universe: Sequence[str],
    year_range: Tuple[int, int] = YEAR_RANGE,
) -> pd.DataFrame:
    """
    Build the binary indicator tensor over the country-year grid.

    Parameters
    ----------
    resolved : pandas.DataFrame
        The resolved ledger with one row per pair.
    universe : sequence of str
        The country universe.
    year_range : tuple of int, optional
        The inclusive analysis year range.

    Returns
    -------
    pandas.DataFrame
        One row per country-year pair carrying the two policy indicators.

    Raises
    ------
    IngestError
        If a resolved policy is outside the declared policy vocabulary.
    """
    # Refuse a resolved policy outside the declared vocabulary.
    unknown = sorted(set(resolved["policy"].astype(str)) - set(POLICY_VALUES))
    # Raise when an undeclared policy appears.
    if unknown:
        # The failure names the undeclared policies.
        raise IngestError(f"Resolved ledger carries undeclared polic(ies): {unknown}.")
    # Compose the full country-year grid.
    years: List[int] = list(range(int(year_range[0]), int(year_range[1]) + 1))
    # Build the grid frame from the sorted universe.
    grid: pd.DataFrame = pd.DataFrame(
        [
            {"country_code": str(country), "year": int(year)}
            for country in sorted(str(code) for code in universe)
            for year in years
        ]
    )
    # Attach the sanctions indicator from the resolved entry years.
    sanctions = resolved[resolved["policy"] == "sanctions"][
        ["country_code", "entry_year"]
    ].rename(columns={"entry_year": "_sanctions_entry"})
    # Attach the arming indicator from the resolved entry years.
    arming = resolved[resolved["policy"] == "direct_arming"][
        ["country_code", "entry_year"]
    ].rename(columns={"entry_year": "_arming_entry"})
    # Merge the two entry-year columns onto the grid.
    grid = grid.merge(sanctions, on="country_code", how="left")
    # Merge the arming entry years onto the grid.
    grid = grid.merge(arming, on="country_code", how="left")
    # Compute the sanctions indicator as a broadcast comparison, which is one
    # from the entry year onwards and zero otherwise, and zero when uncoded.
    grid[SANCTIONS_COLUMN] = (
        grid["_sanctions_entry"].notna()
        & (grid["year"] >= grid["_sanctions_entry"])
    ).astype("int8")
    # Compute the arming indicator the same way.
    grid[ARMING_COLUMN] = (
        grid["_arming_entry"].notna() & (grid["year"] >= grid["_arming_entry"])
    ).astype("int8")
    # Drop the helper entry-year columns from the emitted tensor.
    grid = grid.drop(columns=["_sanctions_entry", "_arming_entry"])
    # Sort the tensor by its declared key for determinism.
    grid = grid.sort_values(["country_code", "year"], kind="stable").reset_index(
        drop=True
    )
    # Log the tensor shape.
    logger.info(
        "indicator tensor built: %d row(s) across %d countr(ies).",
        len(grid),
        grid["country_code"].nunique(),
    )
    # Return the indicator tensor.
    return grid


# Assert that the coded-but-excluded country carries its exclusion decision.
def assert_ukraine_handling(resolved: pd.DataFrame) -> None:
    """
    Assert the documented handling of the coded-but-excluded country.

    Parameters
    ----------
    resolved : pandas.DataFrame
        The resolved ledger.

    Returns
    -------
    None
        The function returns nothing when the handling is documented.

    Raises
    ------
    IngestError
        If the country carries no sanctions entry, if its entry year diverges
        from the declared value, or if its coding decision is not the declared
        exclusion.
    """
    # Select the country's sanctions entry.
    scoped = resolved[
        (resolved["country_code"].astype(str) == UKRAINE)
        & (resolved["policy"] == "sanctions")
    ]
    # Refuse a ledger that omits the coded country entirely.
    if len(scoped) == 0:
        # The failure states the missing entry.
        raise IngestError(
            f"{UKRAINE} carries no resolved sanctions entry; the documentary "
            f"ledger is stale or incomplete."
        )
    # Read the resolved entry year of the country.
    entry_year = int(scoped["entry_year"].iloc[0])
    # Refuse an entry year that diverges from the declared value.
    if entry_year != UKRAINE_ENTRY_YEAR:
        # The failure names both years.
        raise IngestError(
            f"{UKRAINE} sanctions entry year is {entry_year}; the declared year "
            f"is {UKRAINE_ENTRY_YEAR}."
        )
    # Read the coding decision of the country.
    decision = str(scoped["coding_decision"].iloc[0])
    # Refuse a decision other than the declared exclusion.
    if decision != "excluded_from_target":
        # The failure names both decisions.
        raise IngestError(
            f"{UKRAINE} coding decision is {decision!r}; the declared decision is "
            f"'excluded_from_target'."
        )
    # Log the verified handling.
    logger.info(
        "%s verified: sanctions entry %d, coding decision %r.",
        UKRAINE,
        entry_year,
        decision,
    )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Step 3: Verify the pre-2022 arming timing and the ledger integrity
# -------------------------------------------------------------------------------------------------------------------------------


# Summarise the descriptive category counts.
def summarize_coding_categories(
    resolved: pd.DataFrame, universe: Sequence[str]
) -> pd.DataFrame:
    """
    Compute the descriptive category counts of the coding.

    Parameters
    ----------
    resolved : pandas.DataFrame
        The resolved ledger.
    universe : sequence of str
        The country universe.

    Returns
    -------
    pandas.DataFrame
        The counts indexed by category name, including the sanctions-coded
        count, the target count, the arming count and the four exclusive
        categories.

    Raises
    ------
    IngestError
        If a required column is absent.
    """
    # Refuse a resolved ledger lacking a required column.
    for column in ("country_code", "policy", "coding_decision"):
        # The failure names the absent column.
        if column not in resolved.columns:
            # Without these columns the counts cannot be computed.
            raise IngestError(f"Resolved ledger lacks the column {column!r}.")
    # Canonicalize and validate the descriptive country universe.
    universe_codes: List[str] = [str(code) for code in universe]
    # Refuse duplicate identities because sets would conceal them.
    if len(universe_codes) != len(set(universe_codes)):
        # A repeated country has no coherent category membership.
        raise IngestError("Treatment coding universe carries duplicate country codes.")
    # Refuse blank identifiers at the category boundary.
    if any(not code.strip() for code in universe_codes):
        # Empty identities cannot support an auditable membership roster.
        raise IngestError("Treatment coding universe carries a blank country code.")
    # Compose the full descriptive universe set.
    universe_set: set[str] = set(universe_codes)
    # Restrict canonical coding to exact country identities in that universe.
    scoped = resolved[
        resolved["country_code"].astype(str).isin(universe_set)
    ].copy()
    # Refuse duplicated canonical country-policy membership.
    if bool(scoped.duplicated(subset=["country_code", "policy"]).any()):
        # Duplicate membership would inflate or obscure category identities.
        raise IngestError(
            "Resolved coding carries duplicate country-policy membership."
        )
    # Compose the sanctions-coded country set.
    sanctions_coded: set = set(
        scoped[scoped["policy"] == "sanctions"]["country_code"].astype(str)
    )
    # Compose the direct-arming country set.
    arming_coded: set = set(
        scoped[scoped["policy"] == "direct_arming"]["country_code"].astype(str)
    )
    # Compose the set explicitly excluded from policy-target estimation.
    excluded_set: set[str] = set(
        scoped[scoped["coding_decision"].astype(str) == "excluded_from_target"][
            "country_code"
        ].astype(str)
    )
    # Include any independently recorded geopolitical exclusions in provenance.
    attribute_exclusions: set[str] = set(
        str(code) for code in resolved.attrs.get("estimation_exclusions", ())
    )
    # Restrict external exclusions to the declared descriptive universe.
    estimation_exclusions: set[str] = (
        excluded_set | attribute_exclusions
    ) & universe_set
    # Define the estimation universe separately from canonical membership.
    estimation_universe: set[str] = universe_set - estimation_exclusions
    # Construct estimation-target memberships without rewriting canonical codes.
    sanctions_targets: set[str] = sanctions_coded & estimation_universe
    # Construct direct-arming targets under the same eligibility universe.
    arming_targets: set[str] = arming_coded & estimation_universe
    # Compute the four mutually exclusive estimation categories.
    both: set[str] = sanctions_targets & arming_targets
    # Compose the sanctions-only category.
    sanctions_only: set[str] = sanctions_targets - arming_targets
    # Compose the arming-only category.
    arming_only: set[str] = arming_targets - sanctions_targets
    # Compose the neither-policy category.
    neither: set[str] = estimation_universe - sanctions_targets - arming_targets
    # Build a row while preserving the exact country identities behind a count.
    def _category_row(
        category: str, universe_label: str, identities: set[str]
    ) -> Dict[str, Any]:
        # Freeze sorted identities for deterministic downstream comparison.
        country_codes: Tuple[str, ...] = tuple(sorted(identities))
        # Select and freeze the complete roster of the named universe.
        universe_members: Tuple[str, ...] = tuple(
            sorted(
                universe_set
                if universe_label == "descriptive_universe"
                else estimation_universe
            )
        )
        # Preserve exclusions explicitly rather than inferring them from counts.
        excluded_members: Tuple[str, ...] = tuple(
            sorted(estimation_exclusions)
        )
        # Return the labelled semantic record.
        return {
            "category": category,
            "universe": universe_label,
            "count": len(country_codes),
            "country_codes": country_codes,
            "universe_codes": universe_members,
            "excluded_country_codes": excluded_members,
        }
    # Assemble the summary rows in the declared reporting order.
    rows: List[Dict[str, Any]] = [
        _category_row(
            "sanctions_coded_descriptive",
            "descriptive_universe",
            sanctions_coded,
        ),
        _category_row(
            "sanctions_targets", "estimation_universe", sanctions_targets
        ),
        _category_row(
            "direct_arming_targets", "estimation_universe", arming_targets
        ),
        _category_row("both_policies", "estimation_universe", both),
        _category_row("sanctions_only", "estimation_universe", sanctions_only),
        _category_row(
            "direct_arming_only", "estimation_universe", arming_only
        ),
        _category_row("neither_policy", "estimation_universe", neither),
    ]
    # Assemble the summary frame indexed by category.
    summary: pd.DataFrame = pd.DataFrame(rows).set_index("category")
    # Log the summarised categories.
    logger.info("coding categories summarised across %d categor(ies).", len(summary))
    # Return the summary frame.
    return summary


# Verify the pre-escalation arming timing against the declared registry.
def assert_pre_escalation_arming(
    resolved: pd.DataFrame,
    expected: Optional[Mapping[str, int]] = None,
    escalation_year: int = ESCALATION_YEAR,
) -> pd.DataFrame:
    """
    Verify the countries whose arming entry precedes the escalation.

    Parameters
    ----------
    resolved : pandas.DataFrame
        The resolved ledger.
    expected : mapping of str to int, optional
        The declared registry of pre-escalation arming countries and years.
        When absent the observed set is reported without comparison.
    escalation_year : int, optional
        The escalation year separating the two waves.

    Returns
    -------
    pandas.DataFrame
        One row per observed pre-escalation arming country with its year.

    Raises
    ------
    IngestError
        If the observed set diverges from the declared registry.
    """
    # Select the arming entries that precede the escalation year.
    scoped = resolved[
        (resolved["policy"] == "direct_arming")
        & (resolved["entry_year"] < int(escalation_year))
    ]
    # Assemble the observed country-year pairs.
    observed: pd.DataFrame = scoped[["country_code", "entry_year"]].copy()
    # Sort the observed set for determinism.
    observed = observed.sort_values(["country_code"], kind="stable").reset_index(
        drop=True
    )
    # Compare the observed set with the declared registry when supplied.
    if expected is not None:
        # Compose the observed mapping.
        observed_map: Dict[str, int] = {
            str(row["country_code"]): int(row["entry_year"])
            for _, row in observed.iterrows()
        }
        # Compose the declared mapping.
        expected_map: Dict[str, int] = {
            str(code): int(year) for code, year in expected.items()
        }
        # Refuse a divergence between the two mappings.
        if observed_map != expected_map:
            # The failure names both mappings.
            raise IngestError(
                f"Pre-escalation arming countries diverge from the declared "
                f"registry: observed {observed_map}, declared {expected_map}."
            )
    # Log the verification outcome.
    logger.info(
        "pre-escalation arming verified: %d countr(ies).", len(observed)
    )
    # Return the observed table.
    return observed


# Verify the descriptive counts against the declared study counts.
def assert_declared_category_counts(
    summary: pd.DataFrame, expected: Optional[Mapping[str, Any]] = None
) -> None:
    """
    Compare the descriptive counts with the declared study counts.

    Parameters
    ----------
    summary : pandas.DataFrame
        The category summary.
    expected : mapping of str to int, optional
        The declared counts per category. When absent the comparison is
        skipped, which is the fixture mode.

    Returns
    -------
    None
        The function returns nothing when every supplied comparison holds.

    Raises
    ------
    IngestError
        If an observed count diverges from its declared value.
    """
    # Require the semantic columns emitted by the category constructor.
    required_columns: set[str] = {
        "universe",
        "count",
        "country_codes",
        "universe_codes",
        "excluded_country_codes",
    }
    # Refuse a legacy count-only table because identities cannot be audited.
    if not required_columns.issubset(summary.columns):
        # Name the missing semantic columns.
        raise IngestError(
            "Category summary omits semantic column(s): "
            f"{sorted(required_columns - set(summary.columns))}."
        )
    # Refuse duplicate category labels hidden by a non-unique index.
    if not summary.index.is_unique:
        # Published counts require one uniquely labelled category.
        raise IngestError("Category summary index is not unique.")
    # Declare each category's required reporting universe.
    expected_universes: Mapping[str, str] = {
        "sanctions_coded_descriptive": "descriptive_universe",
        "sanctions_targets": "estimation_universe",
        "direct_arming_targets": "estimation_universe",
        "both_policies": "estimation_universe",
        "sanctions_only": "estimation_universe",
        "direct_arming_only": "estimation_universe",
        "neither_policy": "estimation_universe",
    }
    # Require exactly the complete seven-category reporting schema.
    if set(str(category) for category in summary.index) != set(
        expected_universes
    ):
        # Report missing and unexpected category identities.
        raise IngestError(
            "Category summary coverage diverges: missing "
            f"{sorted(set(expected_universes) - set(summary.index.astype(str)))}, "
            "unexpected "
            f"{sorted(set(summary.index.astype(str)) - set(expected_universes))}."
        )
    # Track the independently recorded descriptive and estimation universes.
    descriptive_universe_codes: Optional[Tuple[str, ...]] = None
    # Track the shared estimation roster.
    estimation_universe_codes: Optional[Tuple[str, ...]] = None
    # Track the explicit exclusions separating the two universes.
    explicit_exclusions: Optional[Tuple[str, ...]] = None
    # Verify every emitted count against its underlying country identities.
    for category, row in summary.iterrows():
        # Reject an unknown category rather than assigning an arbitrary universe.
        if str(category) not in expected_universes:
            # Name the undeclared category.
            raise IngestError(f"Summary carries undeclared category {category!r}.")
        # Require the documented universe label for the category.
        if str(row["universe"]) != expected_universes[str(category)]:
            # Prevent comparison of equal counts drawn from different universes.
            raise IngestError(
                f"{category}: universe {row['universe']!r} diverges from "
                f"{expected_universes[str(category)]!r}."
            )
        # Materialize the immutable country membership.
        identities: Tuple[str, ...] = tuple(str(code) for code in row["country_codes"])
        # Require sorted unique identities and exact count reconciliation.
        if identities != tuple(sorted(set(identities))):
            # Duplicated identities can manufacture a published count.
            raise IngestError(
                f"{category}: country_codes are not sorted and unique."
            )
        # Materialize the complete roster of the row's named universe.
        row_universe: Tuple[str, ...] = tuple(
            str(code) for code in row["universe_codes"]
        )
        # Require a canonical unique universe roster.
        if row_universe != tuple(sorted(set(row_universe))):
            # Refuse a universe whose identity count is ambiguous.
            raise IngestError(
                f"{category}: universe_codes are not sorted and unique."
            )
        # Category identities must remain inside their named universe.
        if not set(identities) <= set(row_universe):
            # Name the category that escaped its universe.
            raise IngestError(
                f"{category}: country_codes leave the recorded universe."
            )
        # Materialize the explicit descriptive-to-estimation exclusions.
        row_exclusions: Tuple[str, ...] = tuple(
            str(code) for code in row["excluded_country_codes"]
        )
        # Require canonical unique exclusion identities.
        if row_exclusions != tuple(sorted(set(row_exclusions))):
            # Refuse duplicated or non-deterministic exclusions.
            raise IngestError(
                f"{category}: excluded_country_codes are not sorted and unique."
            )
        # Establish or compare the shared universe rosters.
        if str(row["universe"]) == "descriptive_universe":
            # Record the descriptive universe from its one labelled row.
            descriptive_universe_codes = row_universe
        elif estimation_universe_codes is None:
            # Establish the estimation universe from the first estimation row.
            estimation_universe_codes = row_universe
        elif row_universe != estimation_universe_codes:
            # Equal category labels must refer to one identical universe.
            raise IngestError(
                f"{category}: estimation universe roster is inconsistent."
            )
        # Establish or compare the explicit exclusion roster.
        if explicit_exclusions is None:
            # Record the first exclusion roster.
            explicit_exclusions = row_exclusions
        elif row_exclusions != explicit_exclusions:
            # Refuse exclusions that change by reporting row.
            raise IngestError(
                f"{category}: explicit estimation exclusions are inconsistent."
            )
        # Parse the emitted count without allowing fractional truncation.
        try:
            # Convert through float for a strict finite-integral check.
            emitted_numeric: float = float(row["count"])
        except (TypeError, ValueError) as exc:
            # Preserve the category label in the failure.
            raise IngestError(
                f"{category}: emitted count is malformed: {row['count']!r}."
            ) from exc
        # Refuse a count that disagrees with its auditable membership.
        if (
            isinstance(row["count"], (bool, np.bool_))
            or not np.isfinite(emitted_numeric)
            or emitted_numeric != math.trunc(emitted_numeric)
            or emitted_numeric < 0
            or int(emitted_numeric) != len(identities)
        ):
            # Name both observed quantities.
            raise IngestError(
                f"{category}: count {row['count']!r} disagrees with "
                f"{len(identities)} country identities."
            )
    # Both universe rosters must have been emitted.
    if descriptive_universe_codes is None or estimation_universe_codes is None:
        # The seven-row schema must document both scientific universes.
        raise IngestError("Category summary does not record both universes.")
    # Require explicit exclusions to equal the descriptive-minus-estimation set.
    if set(explicit_exclusions or ()) != (
        set(descriptive_universe_codes) - set(estimation_universe_codes)
    ):
        # Refuse implicit or extraneous universe differences.
        raise IngestError(
            "Explicit exclusions do not reconcile descriptive and estimation "
            "universe identities."
        )
    # The estimation universe must be a subset of the descriptive universe.
    if not set(estimation_universe_codes) <= set(descriptive_universe_codes):
        # Estimation cannot introduce countries absent descriptively.
        raise IngestError(
            "Estimation universe is not a subset of the descriptive universe."
        )
    # Verify the four exclusive estimation cells are pairwise disjoint.
    exclusive_names: Tuple[str, ...] = (
        "both_policies",
        "sanctions_only",
        "direct_arming_only",
        "neither_policy",
    )
    # Extract the identity set of every exclusive cell.
    exclusive_sets: Dict[str, set[str]] = {
        name: set(summary.at[name, "country_codes"])
        for name in exclusive_names
        if name in summary.index
    }
    # Require all four cells before accepting published category counts.
    if len(exclusive_sets) != len(exclusive_names):
        # Name the incomplete categorical partition.
        raise IngestError("Category summary omits an exclusive estimation cell.")
    # Compare every pair for prohibited overlap.
    for left_index, left in enumerate(exclusive_names):
        # Visit each unordered pair once.
        for right in exclusive_names[left_index + 1 :]:
            # Identify countries assigned to two exclusive categories.
            overlap: set[str] = exclusive_sets[left] & exclusive_sets[right]
            # Refuse overlapping category identities.
            if overlap:
                # Name the contaminated cells and identities.
                raise IngestError(
                    f"Exclusive categories {left!r} and {right!r} overlap at "
                    f"{sorted(overlap)}."
                )
    # Require the four exclusive cells to cover the estimation universe exactly.
    exclusive_union: set[str] = set().union(
        *(exclusive_sets[name] for name in exclusive_names)
    )
    # Refuse an omitted or extra estimation identity.
    if exclusive_union != set(estimation_universe_codes):
        # Report both sides for exact roster diagnosis.
        raise IngestError(
            "Exclusive category union diverges from the estimation universe: "
            f"categories {sorted(exclusive_union)}, universe "
            f"{list(estimation_universe_codes)}."
        )
    # Sanctions targets must be exactly both plus sanctions-only.
    sanctions_partition: set[str] = (
        exclusive_sets["both_policies"] | exclusive_sets["sanctions_only"]
    )
    # Detect same-count roster substitutions in sanctions targets.
    if set(summary.at["sanctions_targets", "country_codes"]) != sanctions_partition:
        # Refuse category algebra that reconciles only numerically.
        raise IngestError(
            "sanctions_targets identities do not equal both_policies union "
            "sanctions_only."
        )
    # Direct-arming targets must be exactly both plus arming-only.
    arming_partition: set[str] = (
        exclusive_sets["both_policies"]
        | exclusive_sets["direct_arming_only"]
    )
    # Detect same-count roster substitutions in direct-arming targets.
    if (
        set(summary.at["direct_arming_targets", "country_codes"])
        != arming_partition
    ):
        # Refuse category algebra that reconciles only numerically.
        raise IngestError(
            "direct_arming_targets identities do not equal both_policies union "
            "direct_arming_only."
        )
    # Canonical descriptive sanctions must include every estimation target.
    if not sanctions_partition <= set(
        summary.at["sanctions_coded_descriptive", "country_codes"]
    ):
        # Target membership cannot exist without canonical coding.
        raise IngestError(
            "Sanctions target identities are not a subset of descriptive "
            "sanctions coding."
        )
    # Skip the declaration comparison when no expected values were supplied.
    if expected is None:
        # Log the observed counts for the evidence record.
        logger.info(
            "declared category counts not supplied; observed counts: %s.",
            summary["count"].to_dict(),
        )
        # Return without comparison.
        return
    # Compare each declared category against its observed count and identities.
    for category, declared in expected.items():
        # Read the observed count, defaulting to the declared value when the
        # category is absent so the failure names the missing category.
        observed_value: Optional[int] = (
            int(summary["count"].get(category)) if category in summary.index else None
        )
        # Refuse a category the summary does not report.
        if observed_value is None:
            # The failure names the absent category.
            raise IngestError(f"Summary omits the declared category {category!r}.")
        # Support count-only declarations and richer identity declarations.
        declared_identities: Optional[Tuple[str, ...]] = None
        # A mapping may declare both count and exact countries.
        if isinstance(declared, Mapping):
            # Require an explicit count in the structured declaration.
            if "count" not in declared:
                # Refuse an ambiguous structured declaration.
                raise IngestError(
                    f"{category}: structured declaration omits 'count'."
                )
            # Read the declared count without changing the summary.
            declared_count_raw: Any = declared["count"]
            # Read optional exact country identities.
            if "country_codes" in declared:
                # Freeze the declared identities for exact set comparison.
                declared_identities = tuple(
                    sorted(str(code) for code in declared["country_codes"])
                )
            # Validate an optional explicit universe label.
            if (
                "universe" in declared
                and str(declared["universe"]) != str(summary.at[category, "universe"])
            ):
                # Equal numbers from different universes are not equivalent.
                raise IngestError(
                    f"{category}: declared universe {declared['universe']!r} "
                    f"diverges from {summary.at[category, 'universe']!r}."
                )
        elif isinstance(declared, Sequence) and not isinstance(
            declared, (str, bytes)
        ):
            # A sequence declares exact identities and implies its own count.
            declared_identities = tuple(sorted(str(code) for code in declared))
            # Derive the declared count from the identity roster.
            declared_count_raw = len(declared_identities)
        else:
            # Preserve compatibility with scalar count declarations.
            declared_count_raw = declared
        # Parse the declared count without accepting booleans or fractions.
        try:
            # Convert through float solely for strict numeric validation.
            declared_numeric: float = float(declared_count_raw)
        except (TypeError, ValueError) as exc:
            # Preserve the category context in the domain error.
            raise IngestError(
                f"{category}: malformed declared count {declared_count_raw!r}."
            ) from exc
        # Reject malformed count declarations.
        if (
            isinstance(declared_count_raw, (bool, np.bool_))
            or not np.isfinite(declared_numeric)
            or declared_numeric < 0
            or declared_numeric != math.trunc(declared_numeric)
        ):
            # Do not silently coerce published declarations.
            raise IngestError(
                f"{category}: declared count must be a non-negative integer; "
                f"observed {declared_count_raw!r}."
            )
        # Convert only after strict validation.
        declared_count: int = int(declared_numeric)
        # Raise when the observed count diverges from the declared value.
        if observed_value != declared_count:
            # The failure names the category and both counts.
            raise IngestError(
                f"{category}: observed {observed_value} against a declared "
                f"{declared_count}."
            )
        # Verify exact identities whenever the declaration supplies them.
        if declared_identities is not None:
            # Read the observed identity roster.
            observed_identities: Tuple[str, ...] = tuple(
                summary.at[category, "country_codes"]
            )
            # Refuse equal counts backed by different countries.
            if observed_identities != declared_identities:
                # Name both rosters for reproducible diagnosis.
                raise IngestError(
                    f"{category}: observed identities {observed_identities} "
                    f"diverge from declared {declared_identities}."
                )
    # Log the completed comparison.
    logger.info("declared category counts verified across %d categor(ies).", len(expected))


# Freeze the resolved coding behind a canonical digest.
def freeze_treatment_coding(
    resolved: pd.DataFrame, indicators: pd.DataFrame
) -> str:
    """
    Compute the canonical digest of the resolved coding.

    Parameters
    ----------
    resolved : pandas.DataFrame
        The resolved ledger.
    indicators : pandas.DataFrame
        The indicator tensor.

    Returns
    -------
    str
        The hexadecimal digest of the canonical coding payload.

    Raises
    ------
    IngestError
        If either artefact is empty.
    """
    # Refuse an empty resolved ledger.
    if len(resolved) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot freeze an empty resolved ledger.")
    # Refuse an empty indicator tensor.
    if len(indicators) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot freeze an empty indicator tensor.")
    # Compose the canonical resolved rows in sorted order.
    resolved_sorted = resolved.sort_values(
        ["country_code", "policy"], kind="stable"
    )
    # Compose the canonical indicator rows in sorted order.
    indicators_sorted = indicators.sort_values(
        ["country_code", "year"], kind="stable"
    )
    # Build the canonical payload text.
    payload_parts: List[str] = []
    # Record every resolved pair in its canonical form.
    for _, row in resolved_sorted.iterrows():
        # Append the pair's canonical line.
        payload_parts.append(
            f"{row['country_code']}|{row['policy']}|{int(row['entry_year'])}|"
            f"{row['coding_decision']}"
        )
    # Record every indicator cell in its canonical form.
    for _, row in indicators_sorted.iterrows():
        # Append the cell's canonical line.
        payload_parts.append(
            f"{row['country_code']}|{int(row['year'])}|"
            f"{int(row[SANCTIONS_COLUMN])}{int(row[ARMING_COLUMN])}"
        )
    # Join the payload parts and digest the encoding.
    return hashlib.sha256("\n".join(payload_parts).encode("utf-8")).hexdigest()


# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the treatment coding resolution pipeline.
def orchestrate_treatment_coding(
    ledger: pd.DataFrame,
    universe: Sequence[str],
    expected_categories: Optional[Mapping[str, Any]] = None,
    expected_pre_escalation: Optional[Mapping[str, int]] = DECLARED_PRE_ESCALATION_ARMING,
    primary_rank_max: int = PRIMARY_RANK_MAX,
) -> TreatmentCodingResult:
    """
    Resolve the documentary coding into indicators and verify the anchors.

    Parameters
    ----------
    ledger : pandas.DataFrame
        The canonicalized documentary policy ledger.
    universe : sequence of str
        The country universe of the analysis panel.
    expected_categories : mapping of str to int, optional
        The declared study counts per descriptive category.
    expected_pre_escalation : mapping of str to int, optional
        The declared registry of pre-escalation arming countries.
    primary_rank_max : int, optional
        The highest rank still counted as a first-class source.

    Returns
    -------
    TreatmentCodingResult
        The resolved ledger, the indicator tensor, the decision log, the
        dropped-entry ledger, the summary, the digest, the timestamp and the
        notes.

    Raises
    ------
    IngestError
        If any filter, resolution, coding or verification check fails.
    """
    # Empty universe is refused before any work.
    if not universe:
        # Without a universe the indicators cannot be built.
        raise IngestError("The treatment coding universe is empty.")
    # Filter and sort the ledger under the priority rule.
    survivors, dropped = filter_and_sort_ledger(ledger, primary_rank_max)
    # Resolve one entry per country-policy pair.
    resolved, decision_log, uncoded = resolve_pair_selection(survivors)
    # Carry independently supplied estimation exclusions without changing codes.
    resolved.attrs["estimation_exclusions"] = tuple(
        sorted(
            str(code)
            for code in ledger.attrs.get("estimation_exclusions", ())
        )
    )
    # Assert the documented handling of the coded-but-excluded country.
    assert_ukraine_handling(resolved)
    # Build the indicator tensor over the universe grid.
    indicators: pd.DataFrame = build_indicator_tensor(resolved, universe)
    # Summarise the descriptive categories.
    summary: pd.DataFrame = summarize_coding_categories(resolved, universe)
    # Verify the pre-escalation arming timing.
    pre_escalation: pd.DataFrame = assert_pre_escalation_arming(
        resolved, expected=expected_pre_escalation
    )
    # Compare the descriptive counts when declared values were supplied.
    assert_declared_category_counts(summary, expected=expected_categories)
    # Retain the narrow coding digest as one component of full provenance.
    coding_fingerprint: str = freeze_treatment_coding(resolved, indicators)
    # Canonicalize a frame, including labels, for evidence-chain hashing.
    def _frame_records(frame: pd.DataFrame) -> Dict[str, Any]:
        # Sort columns so caller presentation order cannot change the digest.
        columns: List[str] = sorted(str(column) for column in frame.columns)
        # Render scalar values through JSON's deterministic string fallback.
        records: List[Dict[str, Any]] = frame.loc[:, columns].to_dict(
            orient="records"
        )
        # Sort rows by their canonical JSON representations.
        records = sorted(
            records,
            key=lambda record: json.dumps(record, sort_keys=True, default=str),
        )
        # Preserve both labels and values in the payload.
        return {"columns": columns, "records": records}
    # Compose the complete evidence-to-code provenance payload.
    provenance_payload: Dict[str, Any] = {
        "coding_fingerprint": coding_fingerprint,
        "input_ledger": _frame_records(ledger),
        "survivors": _frame_records(survivors),
        "dropped": _frame_records(dropped),
        "resolved": _frame_records(resolved),
        "decision_log": _frame_records(decision_log),
        "indicators": _frame_records(indicators),
        "summary": _frame_records(summary.reset_index()),
        "estimation_exclusions": list(
            resolved.attrs["estimation_exclusions"]
        ),
        "pre_escalation": _frame_records(pre_escalation),
    }
    # Freeze every evidence, adjudication and derived-code layer together.
    fingerprint: str = hashlib.sha256(
        json.dumps(
            provenance_payload, sort_keys=True, default=str
        ).encode("utf-8")
    ).hexdigest()
    # Accumulate the recorded notes.
    notes: List[str] = [
        f"{len(dropped)} ledger entr(ies) dropped as uncorroborated secondary "
        f"source(s).",
        f"{len(uncoded)} pair(s) survived the filter but produced no entry.",
        f"{len(pre_escalation)} country(ies) carry pre-escalation arming.",
        "descriptive counts compared against declared values."
        if expected_categories is not None
        else "descriptive counts not supplied; comparison skipped.",
    ]
    # Assemble the immutable result.
    result: TreatmentCodingResult = TreatmentCodingResult(
        resolved=resolved,
        indicators=indicators,
        decision_log=decision_log,
        dropped=dropped,
        summary=summary,
        fingerprint_sha256=fingerprint,
        resolved_utc=current_utc_iso(),
        notes=tuple(notes),
    )
    # Log the completed resolution with its digest prefix.
    logger.info(
        "treatment coding resolved: %d pair(s), %d country(ies) coded, "
        "fingerprint %s.",
        len(resolved),
        len(
            set(resolved["country_code"].astype(str))
        ) if len(resolved) else 0,
        fingerprint[:16],
    )
    # Return the assembled result.
    return result
````

````python
# Task 9: Donor Pool and Target Set Construction per Policy and Outcome

# ==============================================================================
# Task 9: Donor Pool and Target Set Construction per Policy and Outcome
# ==============================================================================

# Declare the two policies in their required order.
POLICIES: Final[Tuple[str, str]] = ("strict_sanctions", "direct_arming")
# Map each policy to the indicator column that expresses it.
POLICY_COLUMNS: Final[Mapping[str, str]] = {
    "strict_sanctions": SANCTIONS_COLUMN,
    "direct_arming": ARMING_COLUMN,
}
# Declare the four analysis outcomes in their required order.
OUTCOME_VARIABLES: Final[Tuple[str, ...]] = tuple(CANONICAL_OUTCOME_COLUMNS)
# Compose the eight declared pairs in their required order.
DECLARED_PAIRS: Final[Tuple[Tuple[str, str], ...]] = tuple(
    (policy, outcome) for policy in POLICIES for outcome in OUTCOME_VARIABLES
)
# Declare the geopolitical role vocabulary the metadata may carry.
ROLE_VOCABULARY: Final[Tuple[str, ...]] = (
    "none",
    "donor_ban",
    "target_and_donor_ban",
    "critical_window_exclusion",
    "target_exclusion_from_ATT",
)
# Declare the roles removed from every donor pool, which includes the country
# excluded from the target group because its own outcome path is contaminated
# by the escalation and must not enter any donor cross-section.
DONOR_BANNED_ROLES: Final[Tuple[str, ...]] = (
    "donor_ban",
    "target_and_donor_ban",
    "target_exclusion_from_ATT",
)
# Declare the roles removed from every target set.
TARGET_BANNED_ROLES: Final[Tuple[str, ...]] = ("target_and_donor_ban",)
# Declare the roles removed from the target group only.
TARGET_EXCLUDED_ROLES: Final[Tuple[str, ...]] = ("target_exclusion_from_ATT",)
# Declare the outcome-specific donor exclusions keyed on the pair.
ADDITIONAL_DONOR_EXCLUSIONS: Final[Mapping[Tuple[str, str], Tuple[str, ...]]] = {
    ("direct_arming", "log_CPI"): ("BIH",),
}
# Declare the reason label of a geopolitical donor ban.
REASON_GEOPOLITICAL_BAN: Final[str] = "geopolitical_donor_ban"
# Declare the reason label of an outcome-specific donor exclusion.
REASON_ADDITIONAL_EXCLUSION: Final[str] = "additional_outcome_exclusion"
# Declare the reason label of a target-group exclusion.
REASON_TARGET_EXCLUSION: Final[str] = "excluded_from_target_group"
# Declare the reason label of a donor removed for an incomplete outcome.
REASON_INCOMPLETE_OUTCOME: Final[str] = "incomplete_outcome_series"
# Declare the panel window the donor completeness requirement covers.
DONOR_COVERAGE_WINDOW: Final[Tuple[int, int]] = (2000, 2024)


# Declare the immutable partition of one policy-and-outcome pair.
@dataclass(frozen=True)
class Partition:
    """
    Immutable target and donor partition of one policy-outcome pair.

    Attributes
    ----------
    policy : str
        The policy instrument of the pair.
    outcome : str
        The canonical outcome variable of the pair.
    targets : tuple of str
        The estimation target set, sorted.
    donors : tuple of str
        The eligible donor set, sorted.
    exclusions : pandas.DataFrame
        The exclusion audit of the pair with one row per removed candidate.
    """
    # The policy instrument of the pair.
    policy: str
    # The canonical outcome variable of the pair.
    outcome: str
    # The estimation target set, sorted.
    targets: Tuple[str, ...]
    # The eligible donor set, sorted.
    donors: Tuple[str, ...]
    # The exclusion audit of the pair.
    exclusions: pd.DataFrame


# Declare the immutable registry of the eight partitions.
@dataclass(frozen=True)
class PartitionRegistry:
    """
    Immutable registry of the eight policy-outcome partitions.

    Attributes
    ----------
    partitions : dict
        The partitions keyed on the policy-outcome pair.
    registry : pandas.DataFrame
        One row per pair carrying its counts and its canonical code lists.
    counts : pandas.DataFrame
        One row per pair carrying the target and donor counts.
    exclusions : pandas.DataFrame
        The combined exclusion audit keyed on the pair and the country.
    fingerprint_sha256 : str
        Digest of the canonical registry serialization.
    built_utc : str
        ISO-8601 UTC timestamp of the construction.
    notes : tuple of str
        Recorded decisions, including the universe size.
    """
    # The partitions keyed on the pair.
    partitions: Mapping[Tuple[str, str], Partition]
    # One row per pair carrying counts and canonical code lists.
    registry: pd.DataFrame
    # One row per pair carrying the target and donor counts.
    counts: pd.DataFrame
    # The combined exclusion audit.
    exclusions: pd.DataFrame
    # Digest of the canonical registry serialization.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the construction.
    built_utc: str
    # Recorded decisions of the construction.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 1: Instantiate the eight target-donor partitions
# -------------------------------------------------------------------------------------------------------------------------------


# Derive the ever-treated and never-treated sets of each policy.
def compute_role_sets(
    indicators: pd.DataFrame, universe: Sequence[str]
) -> Tuple[Dict[str, Tuple[str, ...]], Dict[str, Tuple[str, ...]]]:
    """
    Derive the ever-treated and never-treated sets of each policy.

    Parameters
    ----------
    indicators : pandas.DataFrame
        The indicator tensor keyed on country and year.
    universe : sequence of str
        The country universe.

    Returns
    -------
    tuple
        The ever-treated sets and the never-treated sets, both keyed by
        policy and both holding sorted tuples of country codes.

    Raises
    ------
    IngestError
        If a policy indicator column is absent, or if a universe country is
        absent from the indicator grid.
    """
    # Refuse an indicator frame lacking keys or either policy column.
    for column in ("country_code", "year", *POLICY_COLUMNS.values()):
        # The failure names the absent column.
        if column not in indicators.columns:
            # Without the column the policy's sets cannot be derived.
            raise IngestError(f"Indicator tensor lacks the column {column!r}.")
    # Canonicalize the universe without concealing duplicates.
    universe_codes: List[str] = [str(code) for code in universe]
    # Refuse duplicated universe identities.
    if len(universe_codes) != len(set(universe_codes)):
        # A duplicate would make complement construction ambiguous.
        raise IngestError("Partition universe carries duplicate country codes.")
    # Restrict the indicators to the universe.
    universe_set = set(universe_codes)
    # Select the in-universe rows of the indicator tensor.
    scoped = indicators[indicators["country_code"].astype(str).isin(universe_set)]
    # Refuse a tensor that omits a universe country from its grid.
    covered = set(scoped["country_code"].astype(str))
    # Composer the missing countries for the failure message.
    missing = sorted(universe_set - covered)
    # Raise when the grid does not cover the universe.
    if missing:
        # The failure names the absent countries.
        raise IngestError(
            f"Indicator tensor omits {len(missing)} universe countr(ies): "
            f"{missing[:MAX_REPORTED_OFFENDERS]}."
        )
    # Refuse duplicated country-year histories before taking maxima.
    duplicate_keys: pd.Series = scoped.duplicated(
        subset=["country_code", "year"], keep=False
    )
    # Duplicates can conceal an omitted year or contradictory policy state.
    if bool(duplicate_keys.any()):
        # Report the duplicated labels.
        offenders: List[Tuple[Any, Any]] = list(
            scoped.loc[duplicate_keys, ["country_code", "year"]]
            .head(MAX_REPORTED_OFFENDERS)
            .itertuples(index=False, name=None)
        )
        # Refuse the malformed indicator history.
        raise IngestError(
            f"Indicator tensor carries duplicate country-year key(s): {offenders}."
        )
    # Declare the exact indicator calendar.
    expected_years: set[int] = set(
        range(int(YEAR_RANGE[0]), int(YEAR_RANGE[1]) + 1)
    )
    # Validate each country's complete binary, absorbing policy history.
    for country in sorted(universe_set):
        # Select and order the exact country's history.
        history: pd.DataFrame = scoped[
            scoped["country_code"].astype(str) == country
        ].sort_values("year", kind="stable")
        # Parse calendar labels without truncation.
        numeric_years: pd.Series = pd.to_numeric(
            history["year"], errors="coerce"
        )
        # Require finite integral labels before forming a year set.
        invalid_years: pd.Series = (
            history["year"].map(
                lambda value: isinstance(value, (bool, np.bool_))
            )
            | numeric_years.isna()
            | ~np.isfinite(numeric_years.astype(float))
            | (numeric_years.astype(float) % 1.0 != 0.0)
        )
        # Refuse malformed calendar labels.
        if bool(invalid_years.any()):
            # Name the affected country.
            raise IngestError(
                f"{country}: indicator history carries invalid calendar year(s)."
            )
        # Compare the exact calendar identity set.
        observed_years: set[int] = set(numeric_years.astype(int))
        # Refuse missing or extra years.
        if observed_years != expected_years:
            # Report both sides of the calendar divergence.
            raise IngestError(
                f"{country}: indicator years diverge from {YEAR_RANGE}; "
                f"missing {sorted(expected_years - observed_years)}, extra "
                f"{sorted(observed_years - expected_years)}."
            )
        # Validate both policy histories independently.
        for column in POLICY_COLUMNS.values():
            # Parse indicator values without silently treating strings as states.
            values: pd.Series = pd.to_numeric(history[column], errors="coerce")
            # Require every observed state to be exactly binary.
            if bool(values.isna().any()) or not set(values.astype(float)) <= {
                0.0,
                1.0,
            }:
                # Refuse invalid binary histories before role derivation.
                raise IngestError(
                    f"{country}/{column}: indicator history is not binary."
                )
            # Algorithm G0 requires an absorbing zero-to-one switch.
            if bool((values.astype(int).diff().fillna(0) < 0).any()):
                # A one-to-zero transition contradicts canonical coding.
                raise IngestError(
                    f"{country}/{column}: indicator history is not cumulative."
                )
    # Initialise the two role mappings.
    treated: Dict[str, Tuple[str, ...]] = {}
    # Initialise the never-treated mapping.
    never_treated: Dict[str, Tuple[str, ...]] = {}
    # Derive each policy's sets from its indicator column.
    for policy, column in POLICY_COLUMNS.items():
        # Compute the per-country maximum of the indicator over the years.
        maxima: pd.Series = (
            scoped.groupby("country_code", observed=True)[column].max()
        )
        # Compose the ever-treated set as the countries attaining one.
        treated_set = set(maxima[maxima == 1].index.astype(str))
        # Record the sorted ever-treated tuple.
        treated[policy] = tuple(sorted(treated_set))
        # Record the sorted never-treated tuple as the complement.
        never_treated[policy] = tuple(sorted(universe_set - treated_set))
    # Log the derived role sets.
    logger.info(
        "role sets derived: %s.",
        {policy: len(codes) for policy, codes in treated.items()},
    )
    # Return the two mappings.
    return treated, never_treated


# Filter the donor candidates on outcome completeness.
def filter_eligible_donors(
    candidates: Sequence[str],
    panel: pd.DataFrame,
    outcome: str,
    window: Tuple[int, int] = DONOR_COVERAGE_WINDOW,
) -> Tuple[Tuple[str, ...], List[Dict[str, Any]]]:
    """
    Filter donor candidates on outcome completeness over the window.

    Parameters
    ----------
    candidates : sequence of str
        The candidate donor countries.
    panel : pandas.DataFrame
        The canonical analysis panel keyed on country and year.
    outcome : str
        The canonical outcome column whose completeness is required.
    window : tuple of int, optional
        The inclusive window the completeness requirement covers.

    Returns
    -------
    tuple
        The eligible donors as a sorted tuple and the audit rows of the
        candidates removed for incompleteness.

    Raises
    ------
    IngestError
        If the panel lacks the outcome or key columns.
    """
    # Refuse a panel lacking a key or outcome column.
    for column in ("country_code", "year", outcome):
        # Without these labels the completeness test cannot run.
        if column not in panel.columns:
            # Name the absent label.
            raise IngestError(f"Panel lacks the donor-coverage column {column!r}.")
    # Parse year labels without truncation.
    numeric_years: pd.Series = pd.to_numeric(panel["year"], errors="coerce")
    # Identify invalid calendar labels.
    invalid_years: pd.Series = (
        panel["year"].map(lambda value: isinstance(value, (bool, np.bool_)))
        | numeric_years.isna()
        | ~np.isfinite(numeric_years.astype(float))
        | (numeric_years.astype(float) % 1.0 != 0.0)
    )
    # Refuse malformed calendar labels at the frozen-panel boundary.
    if bool(invalid_years.any()):
        # Avoid lossy conversion in donor-year coverage.
        raise IngestError("Panel carries invalid donor-coverage calendar year(s).")
    # Copy and attach the losslessly validated calendar labels.
    working: pd.DataFrame = panel.copy()
    # Use the validated integer representation for exact comparisons.
    working["_validated_year"] = numeric_years.astype("int64")
    # Restrict the panel to the declared window.
    scoped = working[
        (working["_validated_year"] >= int(window[0]))
        & (working["_validated_year"] <= int(window[1]))
    ]
    # Declare the exact required donor calendar.
    expected_years: set[int] = set(
        range(int(window[0]), int(window[1]) + 1)
    )
    # Accumulate the ineligible candidates and their audit rows.
    ineligible: List[str] = []
    # Accumulate the audit rows.
    audit_rows: List[Dict[str, Any]] = []
    # Test each candidate for completeness over the window.
    for candidate in candidates:
        # Select the candidate's rows in the window.
        rows = scoped[scoped["country_code"].astype(str) == str(candidate)]
        # Reject duplicate donor-year labels rather than counting rows.
        duplicate_years: pd.Series = rows.duplicated(
            subset=["_validated_year"], keep=False
        )
        # Duplicates can conceal a missing calendar year.
        if bool(duplicate_years.any()):
            # Name the malformed donor history.
            raise IngestError(
                f"{candidate}/{outcome}: duplicate donor-year rows "
                f"{sorted(set(rows.loc[duplicate_years, '_validated_year'].astype(int)))}."
            )
        # Compute exact missing and extra calendar labels.
        observed_years: set[int] = set(rows["_validated_year"].astype(int))
        # Identify missing required labels.
        missing_years: set[int] = expected_years - observed_years
        # Parse outcome values for finite coverage.
        numeric_outcome: pd.Series = pd.to_numeric(rows[outcome], errors="coerce")
        # Locate missing, nonnumeric or infinite outcome cells.
        invalid_values: int = int(
            (
                numeric_outcome.isna()
                | ~np.isfinite(numeric_outcome.astype(float))
            ).sum()
        )
        # Compute the exact total failed cells.
        total_missing: int = len(missing_years) + invalid_values
        # Record the candidate as ineligible when any required cell fails.
        if total_missing > 0:
            # Record the candidate for removal.
            ineligible.append(str(candidate))
            # Append the audit row with the cell count.
            audit_rows.append(
                {
                    "country_code": str(candidate),
                    "reason": REASON_INCOMPLETE_OUTCOME,
                    "detail": (
                        f"{len(missing_years)} missing year(s) "
                        f"{sorted(missing_years)} and {invalid_values} "
                        f"non-finite {outcome} value(s) over "
                        f"{window[0]} to {window[1]}"
                    ),
                }
            )
    # Compose the eligible donors as the candidates less the ineligible.
    eligible: Tuple[str, ...] = tuple(
        sorted(str(code) for code in candidates if str(code) not in set(ineligible))
    )
    # Return the eligible donors and their removal audit.
    return eligible, audit_rows


# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 2: Apply the geopolitical and outcome-specific exclusions
# -------------------------------------------------------------------------------------------------------------------------------


# Derive the geopolitical exclusion sets from the metadata registry.
def derive_geopolitical_exclusions(
    metadata: pd.DataFrame,
) -> Dict[str, Tuple[str, ...]]:
    """
    Derive the geopolitical exclusion sets from the metadata registry.

    Parameters
    ----------
    metadata : pandas.DataFrame
        The country metadata registry carrying the geopolitical role column.

    Returns
    -------
    dict
        The country codes grouped by role, each as a sorted tuple.

    Raises
    ------
    IngestError
        If the role column is absent or carries an undeclared role.
    """
    # Refuse a metadata frame lacking the role column.
    if "geopolitical_role" not in metadata.columns:
        # Without the role column the exclusions cannot be derived.
        raise IngestError("Metadata registry lacks the geopolitical_role column.")
    # Refuse a metadata frame lacking the country column.
    if "country_code" not in metadata.columns:
        # Without the country column the exclusions cannot be keyed.
        raise IngestError("Metadata registry lacks the country_code column.")
    # Record the observed role vocabulary.
    observed_roles = set(metadata["geopolitical_role"].astype(str))
    # Identify the undeclared roles, ignoring a missing value marker.
    undeclared = sorted(observed_roles - set(ROLE_VOCABULARY) - {"nan"})
    # Refuse a metadata frame carrying an undeclared role.
    if undeclared:
        # The failure names the undeclared roles.
        raise IngestError(f"Metadata carries undeclared geopolitical role(s): {undeclared}.")
    # Accumulate the codes per declared role.
    derived: Dict[str, Tuple[str, ...]] = {}
    # Derive the code set of each declared role.
    for role in ROLE_VOCABULARY:
        # Select the countries carrying the role.
        codes = metadata[
            metadata["geopolitical_role"].astype(str) == role
        ]["country_code"].astype(str)
        # Record the sorted tuple for the role.
        derived[role] = tuple(sorted(codes))
    # Log the derived exclusions.
    logger.info(
        "geopolitical exclusions derived: %s.",
        {role: len(codes) for role, codes in derived.items() if codes},
    )
    # Return the derived mapping.
    return derived


# Assert the derived exclusions against the declared registry.
def assert_declared_exclusions(
    derived: Mapping[str, Tuple[str, ...]],
    expected: Mapping[str, Sequence[str]],
) -> None:
    """
    Assert the derived exclusions against the declared registry.

    Parameters
    ----------
    derived : mapping of str to tuple of str
        The exclusions derived from the metadata at run time.
    expected : mapping of str to sequence of str
        The declared exclusions of each role.

    Returns
    -------
    None
        The function returns nothing when every role agrees.

    Raises
    ------
    IngestError
        If a role's derived membership diverges from its declared value.
    """
    # Compare each declared role's membership with the derived value.
    for role, declared_codes in expected.items():
        # Compose the declared membership as a sorted tuple.
        declared: Tuple[str, ...] = tuple(sorted(str(code) for code in declared_codes))
        # Read the derived membership, defaulting to an empty tuple.
        observed: Tuple[str, ...] = tuple(derived.get(role, ()))
        # Refuse a divergence between the two memberships.
        if observed != declared:
            # The failure names the role and both memberships.
            raise IngestError(
                f"Geopolitical role {role!r} diverges from the declared "
                f"registry: derived {observed}, declared {declared}."
            )
    # Log the verified exclusions.
    logger.info("geopolitical roles verified against the declared registry.")


# Assemble one partition from its derived sets and exclusions.
def assemble_partition(
    policy: str,
    outcome: str,
    treated: Sequence[str],
    never_treated: Sequence[str],
    panel: pd.DataFrame,
    derived_exclusions: Mapping[str, Tuple[str, ...]],
    additional_exclusions: Mapping[Tuple[str, str], Tuple[str, ...]] = (
        ADDITIONAL_DONOR_EXCLUSIONS
    ),
) -> Partition:
    """
    Assemble one policy-outcome partition.

    Parameters
    ----------
    policy : str
        The policy instrument of the pair.
    outcome : str
        The canonical outcome variable of the pair.
    treated : sequence of str
        The ever-treated set of the policy.
    never_treated : sequence of str
        The never-treated set of the policy.
    panel : pandas.DataFrame
        The canonical analysis panel.
    derived_exclusions : mapping of str to tuple of str
        The geopolitical exclusions derived from the metadata.
    additional_exclusions : mapping of tuple to tuple, optional
        The outcome-specific donor exclusions keyed on the pair.

    Returns
    -------
    Partition
        The assembled partition with its exclusion audit.

    Raises
    ------
    IngestError
        Propagated from the donor completeness filter.
    """
    # Require one of the two declared policy identifiers.
    if policy not in POLICIES:
        # Refuse a pair that cannot map to a canonical treatment history.
        raise IngestError(f"Partition policy {policy!r} is undeclared.")
    # Require one of the four canonical outcomes.
    if outcome not in OUTCOME_VARIABLES:
        # Refuse an outcome lacking a registered partition contract.
        raise IngestError(f"Partition outcome {outcome!r} is undeclared.")
    # Canonicalize each policy-role roster without concealing duplicates.
    treated_codes: Tuple[str, ...] = tuple(str(code) for code in treated)
    # Canonicalize the complementary role roster.
    never_codes: Tuple[str, ...] = tuple(str(code) for code in never_treated)
    # Reject duplicates in either input roster.
    if len(treated_codes) != len(set(treated_codes)) or len(never_codes) != len(
        set(never_codes)
    ):
        # Duplicates would compromise identity-based count reconciliation.
        raise IngestError(
            f"{policy}/{outcome}: treated or never-treated roster is duplicated."
        )
    # Require treated and never-treated histories to be disjoint.
    role_overlap: set[str] = set(treated_codes) & set(never_codes)
    # Refuse an identity assigned both policy states.
    if role_overlap:
        # Name the contradictory role identities.
        raise IngestError(
            f"{policy}/{outcome}: treated and never-treated rosters overlap at "
            f"{sorted(role_overlap)}."
        )
    # Validate every supplied geopolitical role key.
    unknown_roles: set[str] = set(derived_exclusions) - set(ROLE_VOCABULARY)
    # Refuse a role outside the registered exclusion semantics.
    if unknown_roles:
        # Name the unsupported roles.
        raise IngestError(
            f"{policy}/{outcome}: undeclared exclusion role(s) "
            f"{sorted(unknown_roles)}."
        )
    # Canonicalize the caller's complete outcome-specific exclusion registry.
    observed_additional: Dict[Tuple[str, str], Tuple[str, ...]] = {}
    # Validate every registered pair and its identity roster.
    for pair, codes in additional_exclusions.items():
        # Refuse an exclusion that cannot be traced to a declared partition.
        if pair not in DECLARED_PAIRS:
            # Name the invalid pair key.
            raise IngestError(
                f"Additional donor exclusions contain undeclared pair {pair}."
            )
        # Canonicalize the supplied identities without concealing duplicates.
        code_tuple: Tuple[str, ...] = tuple(str(code) for code in codes)
        # Reject duplicate exclusions within a pair.
        if len(code_tuple) != len(set(code_tuple)):
            # Duplicate codes make the registered roster ambiguous.
            raise IngestError(
                f"Additional donor exclusions for {pair} contain duplicates."
            )
        # Store the canonical sorted roster.
        observed_additional[pair] = tuple(sorted(code_tuple))
    # Canonicalize the authoritative level-only registry.
    expected_additional: Dict[Tuple[str, str], Tuple[str, ...]] = {
        pair: tuple(sorted(str(code) for code in codes))
        for pair, codes in ADDITIONAL_DONOR_EXCLUSIONS.items()
    }
    # Require exact pair-to-identity equality with the registered policy.
    if observed_additional != expected_additional:
        # Prevent an unregistered extension to the inflation-rate outcome.
        raise IngestError(
            "Additional donor-exclusion registry diverges from the declared "
            f"level-only policy: observed {observed_additional}, declared "
            f"{expected_additional}."
        )
    # Accumulate the audit rows of the partition.
    audit_rows: List[Dict[str, Any]] = []
    # Compose the set of countries removed from the target group.
    target_excluded: set = set()
    # Apply the target-only exclusions.
    for role in TARGET_EXCLUDED_ROLES:
        # Record every country carrying a target-excluded role.
        for code in derived_exclusions.get(role, ()):
            # Record the exclusion when the country is ever-treated.
            if code in set(treated_codes):
                # Record the target-group exclusion.
                target_excluded.add(code)
                # Append the audit row with its reason.
                audit_rows.append(
                    {
                        "policy": policy,
                        "outcome": outcome,
                        "country_code": code,
                        "reason": REASON_TARGET_EXCLUSION,
                        "detail": f"role {role}",
                    }
                )
    # Apply the target-and-donor bans to the target set.
    for role in TARGET_BANNED_ROLES:
        # Remove the banned countries from the target set.
        for code in derived_exclusions.get(role, ()):
            # Record the removal when the country is ever-treated.
            if code in set(treated_codes):
                # Record the target removal.
                target_excluded.add(code)
                # Append the audit row with its reason.
                audit_rows.append(
                    {
                        "policy": policy,
                        "outcome": outcome,
                        "country_code": code,
                        "reason": REASON_GEOPOLITICAL_BAN,
                        "detail": f"role {role}",
                    }
                )
    # Compose the target set as the treated set less the exclusions.
    targets: Tuple[str, ...] = tuple(
        sorted(
            code
            for code in treated_codes
            if code not in target_excluded
        )
    )
    # Enforce Ukraine's canonical-code/estimation-target distinction directly.
    if UKRAINE in set(treated_codes) and UKRAINE not in target_excluded:
        # Remove Ukraine without altering its upstream treatment history.
        targets = tuple(code for code in targets if code != UKRAINE)
        # Record the mandatory target exclusion independently of metadata roles.
        target_excluded.add(UKRAINE)
        # Preserve the reason and rule boundary in the exclusion audit.
        audit_rows.append(
            {
                "policy": policy,
                "outcome": outcome,
                "country_code": UKRAINE,
                "reason": REASON_TARGET_EXCLUSION,
                "detail": "mandatory canonical-code/target-estimation separation",
            }
        )
    # Accumulate the donor candidates and their removals.
    donor_removed: set = set()
    # Apply the donor bans from the geopolitical registry.
    for role in DONOR_BANNED_ROLES:
        # Remove the banned countries from the candidates.
        for code in derived_exclusions.get(role, ()):
            # Record the removal when the country is a candidate.
            if code in set(never_codes):
                # Record the donor removal.
                donor_removed.add(code)
                # Append the audit row with its reason.
                audit_rows.append(
                    {
                        "policy": policy,
                        "outcome": outcome,
                        "country_code": code,
                        "reason": REASON_GEOPOLITICAL_BAN,
                        "detail": f"role {role}",
                    }
                )
    # Enforce the three mandatory geopolitical donor prohibitions directly.
    for code in ("RUS", "BLR", UKRAINE):
        # Apply the direct guard only when metadata did not already remove it.
        if code in set(never_codes) and code not in donor_removed:
            # Record the donor removal independently of metadata labelling.
            donor_removed.add(code)
            # Preserve the mandatory reason in the exclusion audit.
            audit_rows.append(
                {
                    "policy": policy,
                    "outcome": outcome,
                    "country_code": code,
                    "reason": REASON_GEOPOLITICAL_BAN,
                    "detail": "mandatory RUS/BLR/UKR donor prohibition",
                }
            )
    # Apply the outcome-specific exclusions of the pair.
    for code in additional_exclusions.get((policy, outcome), ()):
        # Record the removal when the country is a candidate.
        if code in set(never_codes):
            # Record the donor removal.
            donor_removed.add(code)
            # Append the audit row with its reason.
            audit_rows.append(
                {
                    "policy": policy,
                    "outcome": outcome,
                    "country_code": code,
                    "reason": REASON_ADDITIONAL_EXCLUSION,
                    "detail": f"declared exclusion for {policy}/{outcome}",
                }
            )
    # Compose the pre-completeness candidate set.
    candidates: Tuple[str, ...] = tuple(
        sorted(code for code in never_codes if code not in donor_removed)
    )
    # Filter the candidates on outcome completeness.
    eligible, incomplete_rows = filter_eligible_donors(candidates, panel, outcome)
    # Append the incompleteness rows with the pair attached.
    for row in incomplete_rows:
        # Record the pair on the audit row.
        audit_rows.append(
            {
                "policy": policy,
                "outcome": outcome,
                "country_code": row["country_code"],
                "reason": row["reason"],
                "detail": row["detail"],
            }
        )
    # Assemble the exclusion audit frame of the partition.
    exclusions: pd.DataFrame = pd.DataFrame(
        audit_rows,
        columns=["policy", "outcome", "country_code", "reason", "detail"],
    )
    # Refuse duplicate reasons for the same country and pair.
    if bool(
        exclusions.duplicated(
            subset=["policy", "outcome", "country_code", "reason"],
            keep=False,
        ).any()
    ):
        # Duplicate audit reasons would make the exclusion lineage ambiguous.
        raise IngestError(
            f"{policy}/{outcome}: exclusion audit contains duplicate reasons."
        )
    # Defend the final scientific rosters against accidental overlap.
    if set(targets) & set(eligible):
        # A unit cannot contribute simultaneously as target and donor.
        raise IngestError(
            f"{policy}/{outcome}: final target and donor rosters overlap."
        )
    # Log the assembled partition.
    logger.info(
        "%s/%s: %d target(s), %d donor(s), %d exclusion(s).",
        policy,
        outcome,
        len(targets),
        len(eligible),
        len(exclusions),
    )
    # Return the frozen partition.
    return Partition(
        policy=policy,
        outcome=outcome,
        targets=targets,
        donors=eligible,
        exclusions=exclusions,
    )


# Assert the structural invariants of one partition.
def assert_partition_invariants(
    partition: Partition, universe: Sequence[str], excluded_from_target: Sequence[str]
) -> None:
    """
    Assert the structural invariants of a partition.

    Parameters
    ----------
    partition : Partition
        The partition to verify.
    universe : sequence of str
        The country universe.
    excluded_from_target : sequence of str
        The countries that must never appear in a target set.

    Returns
    -------
    None
        The function returns nothing when every invariant holds.

    Raises
    ------
    IngestError
        If a target and a donor overlap, if a target-excluded country appears
        among the targets, or if either set leaves the universe.
    """
    # Compose the universe set.
    universe_set = set(str(code) for code in universe)
    # Compose the target and donor sets.
    target_set = set(partition.targets)
    # Compose the donor set.
    donor_set = set(partition.donors)
    # Refuse an overlap between the two sets.
    if target_set & donor_set:
        # The failure names the overlapping countries.
        raise IngestError(
            f"{partition.policy}/{partition.outcome}: target and donor sets "
            f"overlap on {sorted(target_set & donor_set)[:MAX_REPORTED_OFFENDERS]}."
        )
    # Refuse a target-excluded country in the target set.
    banned = target_set & set(str(code) for code in excluded_from_target)
    # Raise when a banned country appears among the targets.
    if banned:
        # The failure names the banned countries.
        raise IngestError(
            f"{partition.policy}/{partition.outcome}: target set contains "
            f"excluded countr(ies) {sorted(banned)}."
        )
    # Refuse a set that leaves the universe.
    if not target_set <= universe_set or not donor_set <= universe_set:
        # The failure names the offending codes.
        outsiders = sorted((target_set | donor_set) - universe_set)
        # Raise with the offending codes attached.
        raise IngestError(
            f"{partition.policy}/{partition.outcome}: set(s) leave the universe: "
            f"{outsiders[:MAX_REPORTED_OFFENDERS]}."
        )
    # Log the verified invariants.
    logger.info(
        "%s/%s: invariants verified.", partition.policy, partition.outcome
    )


# Compare the partition counts with the declared study counts.
def assert_declared_partition_counts(
    counts: pd.DataFrame,
    expected: Optional[Mapping[Tuple[str, str], Tuple[int, int]]] = None,
) -> None:
    """
    Compare the partition counts with the declared study counts.

    Parameters
    ----------
    counts : pandas.DataFrame
        The counts table keyed on the pair.
    expected : mapping of tuple to tuple, optional
        The declared target and donor counts per pair. When absent the
        observed counts are reported without comparison.

    Returns
    -------
    None
        The function returns nothing when every supplied comparison holds.

    Raises
    ------
    IngestError
        If an observed count diverges from its declared value.
    """
    # Skip the comparison when no declared counts were supplied.
    if expected is None:
        # Log the observed counts for the evidence record.
        logger.info(
            "declared partition counts not supplied; observed counts: %s.",
            counts.to_dict(orient="index"),
        )
        # Return without comparison.
        return
    # Compare each declared pair's counts with its observed values.
    for pair, declared in expected.items():
        # Refuse a pair the counts table does not report.
        if pair not in counts.index:
            # The failure names the absent pair.
            raise IngestError(f"Counts table omits the pair {pair}.")
        # Read the observed counts of the pair.
        observed_targets: int = int(counts.loc[pair, "targets"])
        # Read the observed donor count of the pair.
        observed_donors: int = int(counts.loc[pair, "donors"])
        # Refuse a target count that diverges from the declared value.
        if observed_targets != int(declared[0]):
            # The failure names the pair and both counts.
            raise IngestError(
                f"{pair}: {observed_targets} target(s) against a declared "
                f"{declared[0]}."
            )
        # Refuse a donor count that diverges from the declared value.
        if observed_donors != int(declared[1]):
            # The failure names the pair and both counts.
            raise IngestError(
                f"{pair}: {observed_donors} donor(s) against a declared "
                f"{declared[1]}."
            )
    # Log the completed comparison.
    logger.info("declared partition counts verified across %d pair(s).", len(expected))


# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 3: Register the eight partitions
# -------------------------------------------------------------------------------------------------------------------------------


# Register the eight partitions as a canonical frame.
def register_partitions(
    partitions: Mapping[Tuple[str, str], Partition],
) -> pd.DataFrame:
    """
    Register the eight partitions as a canonical registry frame.

    Parameters
    ----------
    partitions : mapping of tuple to Partition
        The partitions keyed on the pair.

    Returns
    -------
    pandas.DataFrame
        One row per pair carrying its counts and its canonical code lists.

    Raises
    ------
    IngestError
        If the registry does not cover exactly the eight declared pairs.
    """
    # Refuse a registry that does not cover the declared pairs exactly.
    if set(partitions.keys()) != set(DECLARED_PAIRS):
        # The failure names the divergence.
        missing = sorted(set(DECLARED_PAIRS) - set(partitions.keys()))
        # Compose the unexpected pairs.
        extra = sorted(set(partitions.keys()) - set(DECLARED_PAIRS))
        # Raise with both sides attached.
        raise IngestError(
            f"Partition coverage diverges: missing {missing}, unexpected {extra}."
        )
    # Accumulate the registry rows in the declared pair order.
    rows: List[Dict[str, Any]] = []
    # Iterate over the declared pairs in their order.
    for pair in DECLARED_PAIRS:
        # Select the partition of the pair.
        partition = partitions[pair]
        # Append the registry row with the canonical code lists.
        rows.append(
            {
                "policy": pair[0],
                "outcome": pair[1],
                "targets": int(len(partition.targets)),
                "donors": int(len(partition.donors)),
                "target_codes": json.dumps(list(partition.targets)),
                "donor_codes": json.dumps(list(partition.donors)),
            }
        )
    # Assemble the registry frame keyed on the pair.
    registry: pd.DataFrame = pd.DataFrame(rows).set_index(["policy", "outcome"])
    # Log the registered pairs.
    logger.info("partitions registered: %d pair(s).", len(registry))
    # Return the registry frame.
    return registry


# Freeze the partition registry behind a canonical digest.
def freeze_partition_registry(registry: pd.DataFrame) -> str:
    """
    Compute the canonical digest of the partition registry.

    Parameters
    ----------
    registry : pandas.DataFrame
        The registry frame.

    Returns
    -------
    str
        The hexadecimal digest of the canonical serialization.

    Raises
    ------
    IngestError
        If the registry is empty.
    """
    # Refuse an empty registry.
    if len(registry) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot freeze an empty partition registry.")
    # Sort the registry by its pair index for determinism.
    ordered = registry.sort_index()
    # Compose the canonical payload rows.
    payload: List[str] = []
    # Serialize each pair in canonical order.
    for (policy, outcome), row in ordered.iterrows():
        # Append the pair's canonical line with its sorted code lists.
        payload.append(
            f"{policy}|{outcome}|{row['target_codes']}|{row['donor_codes']}"
        )
    # Digest the encoded payload.
    return hashlib.sha256("\n".join(payload).encode("utf-8")).hexdigest()


# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the partition construction across the eight pairs.
def orchestrate_partition_construction(
    indicators: pd.DataFrame,
    panel: pd.DataFrame,
    metadata: pd.DataFrame,
    universe: Sequence[str],
    expected_counts: Optional[Mapping[Tuple[str, str], Tuple[int, int]]] = None,
    expected_roles: Optional[Mapping[str, Sequence[str]]] = None,
) -> PartitionRegistry:
    """
    Construct and register the eight policy-outcome partitions.

    Parameters
    ----------
    indicators : pandas.DataFrame
        The indicator tensor from the coding resolution.
    panel : pandas.DataFrame
        The canonical analysis panel.
    metadata : pandas.DataFrame
        The country metadata registry.
    universe : sequence of str
        The country universe.
    expected_counts : mapping of tuple to tuple, optional
        The declared target and donor counts per pair.
    expected_roles : mapping of str to sequence of str, optional
        The declared geopolitical exclusions of each role.

    Returns
    -------
    PartitionRegistry
        The partitions, the registry frame, the counts table, the combined
        exclusion audit, the digest, the timestamp and the notes.

    Raises
    ------
    IngestError
        If any derivation, assembly, assertion or registration check fails.
    """
    # Refuse an empty universe before any derivation.
    if not universe:
        # Without a universe the partitions are undefined.
        raise IngestError("The partition universe is empty.")
    # Derive the ever-treated and never-treated sets of each policy.
    treated, never_treated = compute_role_sets(indicators, universe)
    # Derive the geopolitical exclusions from the metadata registry.
    derived = derive_geopolitical_exclusions(metadata)
    # Verify the derived exclusions when a declared registry was supplied.
    if expected_roles is not None:
        # Assert the derived memberships against the declared values.
        assert_declared_exclusions(derived, expected_roles)
    # Select the countries excluded from the target group.
    excluded_from_target: List[str] = []
    # Collect the excluded countries from their roles.
    for role in TARGET_EXCLUDED_ROLES + TARGET_BANNED_ROLES:
        # Record every country carrying the role.
        excluded_from_target.extend(derived.get(role, ()))
    # Enforce Ukraine's target exclusion even if metadata omits its role.
    if UKRAINE not in excluded_from_target:
        # Preserve treatment history while excluding Ukraine from estimation.
        excluded_from_target.append(UKRAINE)
    # Accumulate the eight partitions in their declared order.
    partitions: Dict[Tuple[str, str], Partition] = {}
    # Accumulate the exclusion audits.
    audits: List[pd.DataFrame] = []
    # Accumulate the counts rows.
    count_rows: List[Dict[str, Any]] = []
    # Assemble each declared pair in turn.
    for policy, outcome in DECLARED_PAIRS:
        # Assemble the partition of the pair.
        partition: Partition = assemble_partition(
            policy,
            outcome,
            treated[policy],
            never_treated[policy],
            panel,
            derived,
        )
        # Assert the structural invariants of the partition.
        assert_partition_invariants(partition, universe, excluded_from_target)
        # Record the partition.
        partitions[(policy, outcome)] = partition
        # Append the exclusion audit of the partition.
        audits.append(partition.exclusions)
        # Append the counts row of the partition.
        count_rows.append(
            {
                "policy": policy,
                "outcome": outcome,
                "targets": int(len(partition.targets)),
                "donors": int(len(partition.donors)),
            }
        )
    # Assemble the counts table keyed on the pair.
    counts: pd.DataFrame = pd.DataFrame(count_rows).set_index(["policy", "outcome"])
    # Compare the counts with the declared values when supplied.
    assert_declared_partition_counts(counts, expected=expected_counts)
    # Concatenate the per-partition exclusion audits.
    exclusions: pd.DataFrame = pd.concat(audits, ignore_index=True)
    # Register the eight partitions as a canonical frame.
    registry: pd.DataFrame = register_partitions(partitions)
    # Compute the canonical digest of the registry.
    fingerprint: str = freeze_partition_registry(registry)
    # Accumulate the recorded notes.
    notes: List[str] = [
        f"universe size {len(universe)}.",
        f"{len(exclusions)} exclusion(s) recorded across the eight pairs.",
        "declared partition counts compared."
        if expected_counts is not None
        else "declared partition counts not supplied; comparison skipped.",
    ]
    # Assemble the immutable registry result.
    result: PartitionRegistry = PartitionRegistry(
        partitions=partitions,
        registry=registry,
        counts=counts,
        exclusions=exclusions,
        fingerprint_sha256=fingerprint,
        built_utc=current_utc_iso(),
        notes=tuple(notes),
    )
    # Log the completed construction with its digest prefix.
    logger.info(
        "partition construction complete: %d pair(s), %d exclusion(s), "
        "fingerprint %s.",
        len(partitions),
        len(exclusions),
        fingerprint[:16],
    )
    # Return the assembled registry.
    return result
````

````python
# Task 10: Donor Factor Proxy Construction (Step 1 of CCEDID)

# ==============================================================================
# Task 10: Donor Factor Proxy Construction (Step 1 of CCEDID)
# ==============================================================================

# Declare the window over which the proxy series is constructed.
FACTOR_WINDOW: Final[Tuple[int, int]] = (2000, 2024)
# Declare the number of covariate columns in the pooling matrix.
COVARIATE_COLUMNS: Final[int] = len(CONTROL_COLUMNS)
# Declare the declared column count of the pooling matrix, which is the
# intercept, the eight lagged controls and the donor factor proxy.
POOLING_COLUMNS: Final[int] = 1 + COVARIATE_COLUMNS + 1
# Declare the condition-number threshold above which a note is recorded.
CONDITION_NUMBER_WARNING: Final[float] = 1.0e6


# Declare the immutable factor objects of one partition.
@dataclass(frozen=True)
class PartitionFactor:
    """
    Immutable factor objects of one policy-outcome partition.

    Attributes
    ----------
    policy : str
        The policy instrument of the pair.
    outcome : str
        The canonical outcome variable of the pair.
    proxy : pandas.Series
        The donor cross-sectional proxy, indexed by year.
    donor_count : int
        The number of donors the proxy was averaged over.
    matrices : dict of str to tuple
        The dependent and pooling matrices keyed by unit code.
    diagnostics : pandas.DataFrame
        Matrix-specific SVD diagnostics with explicit year, column, rank
        convention and roster-hash provenance.
    matrix_axes : mapping
        The labelled row and column axes of every unit matrix.
    """
    # The policy instrument of the pair.
    policy: str
    # The canonical outcome variable of the pair.
    outcome: str
    # The donor cross-sectional proxy indexed by year.
    proxy: pd.Series
    # The number of donors the proxy was averaged over.
    donor_count: int
    # The dependent and pooling matrices keyed by unit code.
    matrices: Mapping[str, Tuple[np.ndarray, np.ndarray]]
    # The matrix-specific diagnostics and their provenance.
    diagnostics: pd.DataFrame
    # The labelled axes aligned with each unit's array matrices.
    matrix_axes: Mapping[str, Mapping[str, Tuple[Any, ...]]]


# Declare the immutable result of the factor construction.
@dataclass(frozen=True)
class DonorFactorResult:
    """
    Immutable result of the donor factor proxy construction.

    Attributes
    ----------
    factors : dict
        The factor objects keyed on the policy-outcome pair.
    proxy_frame : pandas.DataFrame
        The long proxy frame keyed on the pair and the year.
    diagnostics : pandas.DataFrame
        One row per pair carrying the donor count, the proxy dispersion and
        the numerical conditioning of the matrices.
    fingerprint_sha256 : str
        Digest of the canonical factor payload.
    built_utc : str
        ISO-8601 UTC timestamp of the construction.
    notes : tuple of str
        Recorded decisions, including conditioning warnings.
    """
    # The factor objects keyed on the pair.
    factors: Mapping[Tuple[str, str], PartitionFactor]
    # The long proxy frame keyed on the pair and the year.
    proxy_frame: pd.DataFrame
    # One row per pair carrying the diagnostics.
    diagnostics: pd.DataFrame
    # Digest of the canonical factor payload.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the construction.
    built_utc: str
    # Recorded decisions of the construction.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Step 1: Assemble the donor panel and compute the proxy
# -------------------------------------------------------------------------------------------------------------------------------


# Build the year-by-donor frame of one partition's outcome.
def build_donor_panel(
    panel: pd.DataFrame,
    donors: Sequence[str],
    outcome: str,
    window: Tuple[int, int] = FACTOR_WINDOW,
) -> pd.DataFrame:
    """
    Build the year-by-donor frame of one partition.

    Parameters
    ----------
    panel : pandas.DataFrame
        The canonical analysis panel.
    donors : sequence of str
        The eligible donor countries of the partition.
    outcome : str
        The canonical outcome variable of the partition.
    window : tuple of int, optional
        The inclusive factor window.

    Returns
    -------
    pandas.DataFrame
        The donor frame indexed by year and columned by donor code.

    Raises
    ------
    IngestError
        If the donor pool is empty, if the panel lacks the outcome column, or
        if any donor cell is missing inside the window.
    """
    # Refuse an empty donor pool, which cannot define a cross-section.
    if not donors:
        # The failure states the precondition.
        raise IngestError("Cannot build a donor panel from an empty pool.")
    # Canonicalize donors without concealing repeated identities.
    donor_codes: Tuple[str, ...] = tuple(str(code) for code in donors)
    # Refuse a duplicated donor roster.
    if len(donor_codes) != len(set(donor_codes)):
        # Repeated columns would overweight a donor in the mean.
        raise IngestError(f"{outcome}: donor roster is duplicated.")
    # Refuse a panel lacking a key or outcome column.
    for column in ("country_code", "year", outcome):
        # Without these columns the cross-section cannot be formed.
        if column not in panel.columns:
            # Name the absent field.
            raise IngestError(f"Panel lacks the donor-factor column {column!r}.")
    # Parse calendar labels without truncation.
    numeric_years: pd.Series = pd.to_numeric(panel["year"], errors="coerce")
    # Identify invalid panel years.
    invalid_years: pd.Series = (
        panel["year"].map(lambda value: isinstance(value, (bool, np.bool_)))
        | numeric_years.isna()
        | ~np.isfinite(numeric_years.astype(float))
        | (numeric_years.astype(float) % 1.0 != 0.0)
    )
    # Refuse malformed year labels before filtering.
    if bool(invalid_years.any()):
        # Prevent lossy year mapping into the factor axis.
        raise IngestError("Donor-factor panel carries invalid calendar year(s).")
    # Work on a copy with losslessly validated integer years.
    working: pd.DataFrame = panel.copy()
    # Attach the validated year axis.
    working["_factor_year"] = numeric_years.astype("int64")
    # Restrict the panel to the declared window and the donor pool.
    scoped = working[
        (working["_factor_year"] >= int(window[0]))
        & (working["_factor_year"] <= int(window[1]))
        & (working["country_code"].astype(str).isin(set(donor_codes)))
    ]
    # Refuse duplicate donor-year cells before pivoting.
    duplicate_cells: pd.Series = scoped.duplicated(
        subset=["country_code", "_factor_year"], keep=False
    )
    # An aggregation rule must never choose among duplicate scientific cells.
    if bool(duplicate_cells.any()):
        # Report bounded duplicate identities.
        offenders: List[Tuple[Any, Any]] = list(
            scoped.loc[duplicate_cells, ["country_code", "_factor_year"]]
            .head(MAX_REPORTED_OFFENDERS)
            .itertuples(index=False, name=None)
        )
        # Refuse target or donor contamination hidden by aggregation.
        raise IngestError(
            f"Donor-factor panel carries duplicate donor-year cell(s): {offenders}."
        )
    # Parse outcome values and require finiteness.
    numeric_outcome: pd.Series = pd.to_numeric(scoped[outcome], errors="coerce")
    # Identify missing, nonnumeric or infinite values.
    invalid_outcome: pd.Series = numeric_outcome.isna() | ~np.isfinite(
        numeric_outcome.astype(float)
    )
    # Refuse invalid donor outcomes before averaging.
    if bool(invalid_outcome.any()):
        # Name bounded country-year offenders.
        offenders = list(
            scoped.loc[invalid_outcome, ["country_code", "_factor_year"]]
            .head(MAX_REPORTED_OFFENDERS)
            .itertuples(index=False, name=None)
        )
        # No partial cross-sectional mean is scientifically admissible.
        raise IngestError(
            f"Donor-factor panel carries non-finite {outcome} value(s) at "
            f"{offenders}."
        )
    # Store the validated numeric outcome.
    scoped = scoped.assign(**{outcome: numeric_outcome.astype(float)})
    # Pivot the outcome to a year-by-donor frame.
    donor_frame: pd.DataFrame = scoped.pivot(
        index="_factor_year", columns="country_code", values=outcome
    )
    # Reindex the frame to the declared window and the donor list so absent
    # rows and columns become visible as missing rather than as short axes.
    full_index: pd.Index = pd.Index(
        list(range(int(window[0]), int(window[1]) + 1)), name="year"
    )
    # Reindex to the complete year and donor axis.
    donor_frame = donor_frame.reindex(
        index=full_index, columns=list(donor_codes)
    )
    # Identify the missing cells of the donor frame.
    missing_cells = donor_frame.isna()
    # Refuse a donor frame carrying any missing cell.
    if bool(missing_cells.to_numpy().any()):
        # Compose the offending year-donor pairs for the failure message.
        offenders: List[Tuple[int, str]] = [
            (
                int(donor_frame.index[row_position]),
                str(donor_frame.columns[column_position]),
            )
            for row_position, column_position in np.argwhere(
                missing_cells.to_numpy()
            )
        ]
        # Raise with the offending pairs attached.
        raise IngestError(
            f"Donor frame carries {int(missing_cells.to_numpy().sum())} missing "
            f"cell(s), for example {offenders[:MAX_REPORTED_OFFENDERS]}."
        )
    # Return the complete donor frame.
    return donor_frame


# Compute the donor cross-sectional proxy of one partition.
def compute_donor_factor(
    donor_frame: pd.DataFrame, policy: str, outcome: str
) -> pd.Series:
    """
    Compute the donor cross-sectional proxy.

    Parameters
    ----------
    donor_frame : pandas.DataFrame
        The donor frame indexed by year and columned by donor code.
    policy : str
        The policy instrument of the pair, recorded in the series name.
    outcome : str
        The canonical outcome variable of the pair, recorded in the name.

    Returns
    -------
    pandas.Series
        The unweighted cross-sectional mean per year, indexed by year.

    Raises
    ------
    IngestError
        If the donor frame is empty.
    """
    # Refuse an empty donor frame.
    if donor_frame.shape[1] == 0 or donor_frame.shape[0] == 0:
        # An empty frame cannot define a cross-section.
        raise IngestError(
            f"{policy}/{outcome}: donor frame is empty, so the proxy is undefined."
        )
    # Compute the unweighted cross-sectional mean per year.
    proxy: pd.Series = donor_frame.mean(axis=1)
    # Name the series after the pair so its provenance is explicit.
    proxy.name = f"F_{policy}_{outcome}"
    # Log the computed proxy extent.
    logger.info(
        "%s/%s: donor proxy computed over %d donor(s) and %d year(s).",
        policy,
        outcome,
        donor_frame.shape[1],
        donor_frame.shape[0],
    )
    # Return the proxy series.
    return proxy


# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Step 2: Assemble the pooling matrix
# -------------------------------------------------------------------------------------------------------------------------------


# Build the dependent and pooling matrices of one unit.
def build_pooling_matrix(
    lagged: pd.DataFrame,
    unit: str,
    outcome: str,
    proxy: pd.Series,
    window: Tuple[int, int] = PRE_TREATMENT_WINDOW,
) -> Tuple[np.ndarray, np.ndarray]:
    """
    Build one unit's dependent and pooling matrices.

    Parameters
    ----------
    lagged : pandas.DataFrame
        The panel carrying the lagged control columns.
    unit : str
        The unit code whose matrices are built.
    outcome : str
        The canonical outcome variable, which forms the dependent column.
    proxy : pandas.Series
        The donor factor proxy indexed by year.
    window : tuple of int, optional
        The inclusive window the matrices cover.

    Returns
    -------
    tuple
        The dependent matrix of shape rows-by-one and the pooling matrix of
        shape rows-by-ten.

    Raises
    ------
    IngestError
        If a required column is absent, if the unit's window rows are
        incomplete, or if a constructed matrix has the wrong shape.
    """
    # Compose the declared lagged covariate column names.
    covariate_columns: List[str] = [f"{name}{LAG_SUFFIX}" for name in CONTROL_COLUMNS]
    # Refuse a lagged panel lacking any key or required model column.
    for column in ["country_code", "year", outcome] + covariate_columns:
        # The failure names the absent column.
        if column not in lagged.columns:
            # Without the column the matrix cannot be assembled.
            raise IngestError(f"Lagged panel lacks the column {column!r}.")
    # Parse panel year labels without truncation.
    numeric_years: pd.Series = pd.to_numeric(lagged["year"], errors="coerce")
    # Identify malformed calendar values.
    invalid_years: pd.Series = (
        lagged["year"].map(lambda value: isinstance(value, (bool, np.bool_)))
        | numeric_years.isna()
        | ~np.isfinite(numeric_years.astype(float))
        | (numeric_years.astype(float) % 1.0 != 0.0)
    )
    # Refuse invalid calendar labels before selecting the pre-period.
    if bool(invalid_years.any()):
        # Prevent lossy matrix row alignment.
        raise IngestError("Lagged panel carries invalid matrix calendar year(s).")
    # Copy and attach validated integer labels.
    working: pd.DataFrame = lagged.copy()
    # Preserve the exact validated calendar axis.
    working["_pool_year"] = numeric_years.astype("int64")
    # Compute the declared row count of the window.
    t0: int = int(window[1]) - int(window[0]) + 1
    # Select the unit's rows inside the window.
    scoped = working[
        (working["country_code"].astype(str) == str(unit))
        & (working["_pool_year"] >= int(window[0]))
        & (working["_pool_year"] <= int(window[1]))
    ].sort_values("_pool_year", kind="stable")
    # Refuse duplicate unit-year rows explicitly.
    duplicate_years: pd.Series = scoped.duplicated(
        subset=["_pool_year"], keep=False
    )
    # A duplicate can conceal a missing factor year.
    if bool(duplicate_years.any()):
        # Name the duplicate calendar labels.
        raise IngestError(
            f"{unit}/{outcome}: duplicate pre-treatment year row(s) "
            f"{sorted(set(scoped.loc[duplicate_years, '_pool_year'].astype(int)))}."
        )
    # Construct the exact required estimation-year sequence.
    expected_years: Tuple[int, ...] = tuple(
        range(int(window[0]), int(window[1]) + 1)
    )
    # Read the observed ordered year sequence.
    observed_years: Tuple[int, ...] = tuple(scoped["_pool_year"].astype(int))
    # Refuse a unit whose labelled window diverges.
    if observed_years != expected_years:
        # The failure names the unit and the exact labels.
        raise IngestError(
            f"{unit}/{outcome}: pre-treatment years {observed_years} diverge "
            f"from {expected_years}."
        )
    # Validate uniqueness and finiteness of the proxy's labelled year axis.
    if not proxy.index.is_unique:
        # Duplicate proxy labels destroy one-to-one factor alignment.
        raise IngestError(f"{unit}/{outcome}: proxy year index is duplicated.")
    # Map the proxy onto the unit's exact years, refusing a missing year.
    proxy_values: pd.Series = scoped["_pool_year"].map(proxy)
    # Refuse a proxy year the unit's rows cannot be matched to.
    if bool(proxy_values.isna().any()):
        # The failure names the unit and the outcome.
        raise IngestError(
            f"{unit}/{outcome}: the proxy is unavailable for some window year(s); "
            f"the factor window must cover the estimation window."
        )
    # Extract the dependent column as a matrix.
    dependent: np.ndarray = scoped[outcome].to_numpy(dtype=float).reshape(t0, 1)
    # Compose the intercept column.
    intercept: np.ndarray = np.ones((t0, 1), dtype=float)
    # Compose the eight lagged covariate columns.
    covariates: np.ndarray = scoped[covariate_columns].to_numpy(dtype=float)
    # Compose the proxy column as the parsimonious common factor.
    factor_column: np.ndarray = proxy_values.to_numpy(dtype=float).reshape(t0, 1)
    # Concatenate the intercept, the covariates and the proxy in order.
    pooling: np.ndarray = np.hstack([intercept, covariates, factor_column])
    # Require finite dependent, covariate and factor matrices.
    if not np.isfinite(dependent).all() or not np.isfinite(pooling).all():
        # Non-finite values cannot enter rank or conditioning diagnostics.
        raise IngestError(
            f"{unit}/{outcome}: dependent or pooling matrix is non-finite."
        )
    # Assert the declared dependent shape.
    if dependent.shape != (t0, 1):
        # The failure names the observed shape.
        raise IngestError(
            f"{unit}/{outcome}: dependent shape {dependent.shape} against "
            f"{(t0, 1)}."
        )
    # Assert the declared pooling shape of one plus eight plus one columns.
    if pooling.shape != (t0, POOLING_COLUMNS):
        # The failure names the observed shape.
        raise IngestError(
            f"{unit}/{outcome}: pooling shape {pooling.shape} against "
            f"{(t0, POOLING_COLUMNS)}."
        )
    # Return the two matrices.
    return dependent, pooling


# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Step 3: Diagnose and freeze the factor objects
# -------------------------------------------------------------------------------------------------------------------------------


# Compute the numerical diagnostics of one pooling matrix.
def compute_matrix_diagnostics(pooling: np.ndarray) -> Dict[str, float]:
    """
    Compute the numerical diagnostics of one pooling matrix.

    Parameters
    ----------
    pooling : numpy.ndarray
        The pooling matrix of one unit.

    Returns
    -------
    dict
        The matrix rank, its smallest and largest singular values and its
        condition number.

    Raises
    ------
    IngestError
        If the matrix is empty, which makes the diagnostics undefined.
    """
    # Refuse an empty matrix.
    if pooling.size == 0 or pooling.shape[0] == 0:
        # The diagnostics are undefined without rows.
        raise IngestError("Cannot diagnose an empty pooling matrix.")
    # Refuse non-finite values before invoking the SVD.
    if not np.isfinite(pooling).all():
        # Numerical diagnostics on missing/infinite matrices are undefined.
        raise IngestError("Cannot diagnose a non-finite matrix.")
    # Compute the singular values of the matrix.
    singular_values: np.ndarray = np.linalg.svd(pooling, compute_uv=False)
    # Read the largest singular value.
    largest: float = float(singular_values[0])
    # Read the smallest singular value.
    smallest: float = float(singular_values[-1])
    # Compute the condition number, guarding against a zero denominator.
    condition: float = (
        float("inf") if smallest == 0.0 else largest / smallest
    )
    # Return the diagnostic mapping.
    return {
        "rank": float(np.linalg.matrix_rank(pooling)),
        "smallest_singular_value": smallest,
        "largest_singular_value": largest,
        "condition_number": condition,
    }


# Assemble the factor objects of one partition.
def assemble_partition_factor(
    policy: str,
    outcome: str,
    donors: Sequence[str],
    units: Sequence[str],
    panel: pd.DataFrame,
    lagged: pd.DataFrame,
    window: Tuple[int, int] = PRE_TREATMENT_WINDOW,
    factor_window: Tuple[int, int] = FACTOR_WINDOW,
) -> PartitionFactor:
    """
    Assemble the factor objects of one partition.

    Parameters
    ----------
    policy : str
        The policy instrument of the pair.
    outcome : str
        The canonical outcome variable of the pair.
    donors : sequence of str
        The eligible donor pool of the partition.
    units : sequence of str
        The estimation units whose matrices are built.
    panel : pandas.DataFrame
        The canonical analysis panel.
    lagged : pandas.DataFrame
        The panel carrying the lagged control columns.
    window : tuple of int, optional
        The inclusive estimation window of the matrices.
    factor_window : tuple of int, optional
        The inclusive window of the proxy series.

    Returns
    -------
    PartitionFactor
        The proxy, its donor count and the per-unit matrices.

    Raises
    ------
    IngestError
        Propagated from the donor panel, the proxy and the matrix builders.
    """
    # Require a registered policy-outcome pair.
    if (policy, outcome) not in DECLARED_PAIRS:
        # A factor must be tied to one declared partition.
        raise IngestError(f"Undeclared factor partition {(policy, outcome)}.")
    # Canonicalize donor and unit rosters without concealing duplicates.
    donor_codes: Tuple[str, ...] = tuple(str(code) for code in donors)
    # Canonicalize the pooled target-plus-donor unit roster.
    unit_codes: Tuple[str, ...] = tuple(str(code) for code in units)
    # Refuse duplicate donor or unit identities.
    if len(donor_codes) != len(set(donor_codes)) or len(unit_codes) != len(
        set(unit_codes)
    ):
        # Duplicate units would overweight the pooled solver.
        raise IngestError(
            f"{policy}/{outcome}: donor or pooled-unit roster is duplicated."
        )
    # Require every donor to appear in the pooled target-plus-donor data.
    if not set(donor_codes) <= set(unit_codes):
        # The pooled construction must include its donor histories.
        raise IngestError(
            f"{policy}/{outcome}: pooled units omit donor(s) "
            f"{sorted(set(donor_codes) - set(unit_codes))}."
        )
    # Refuse an empty pooled unit roster.
    if not unit_codes:
        # No unit matrices or pooled diagnostics could be formed.
        raise IngestError(f"{policy}/{outcome}: pooled unit roster is empty.")
    # Build the donor frame of the partition.
    donor_frame: pd.DataFrame = build_donor_panel(
        panel, donor_codes, outcome, factor_window
    )
    # Compute the cross-sectional proxy of the partition.
    proxy: pd.Series = compute_donor_factor(donor_frame, policy, outcome)
    # Construct the exact labelled pre-treatment factor matrix.
    estimation_years: Tuple[int, ...] = tuple(
        range(int(window[0]), int(window[1]) + 1)
    )
    # Select the proxy in exact estimation-year order.
    factor_proxy: pd.Series = proxy.reindex(estimation_years)
    # Refuse a factor window that does not cover every estimation year.
    if bool(factor_proxy.isna().any()):
        # Name the absent factor labels.
        raise IngestError(
            f"{policy}/{outcome}: factor proxy omits estimation year(s) "
            f"{factor_proxy.index[factor_proxy.isna()].tolist()}."
        )
    # Assemble F with the declared constant and donor-mean columns.
    factor_matrix: np.ndarray = np.column_stack(
        [
            np.ones(len(estimation_years), dtype=float),
            factor_proxy.to_numpy(dtype=float),
        ]
    )
    # Hash the exact donor and pooled-unit identities behind every diagnostic.
    roster_hash: str = hashlib.sha256(
        json.dumps(
            {
                "donors": sorted(donor_codes),
                "units": sorted(unit_codes),
            },
            sort_keys=True,
        ).encode("utf-8")
    ).hexdigest()
    # Accumulate matrix-specific diagnostic records.
    diagnostic_rows: List[Dict[str, Any]] = []
    # Record diagnostics with explicit matrix identity and axes.
    def _record_diagnostic(
        matrix_name: str,
        matrix: np.ndarray,
        columns: Sequence[str],
        unit: str,
    ) -> None:
        # Diagnose the named matrix itself, never its squared cross-product.
        values: Dict[str, float] = compute_matrix_diagnostics(matrix)
        # Append all numerical and provenance fields.
        diagnostic_rows.append(
            {
                "policy": policy,
                "outcome": outcome,
                "unit": unit,
                "matrix_name": matrix_name,
                "years": estimation_years,
                "columns": tuple(str(column) for column in columns),
                "rows": int(matrix.shape[0]),
                "column_count": int(matrix.shape[1]),
                "rank": int(values["rank"]),
                "rank_convention": "numpy.linalg.matrix_rank_default_svd_tolerance",
                "smallest_singular_value": float(
                    values["smallest_singular_value"]
                ),
                "largest_singular_value": float(
                    values["largest_singular_value"]
                ),
                "condition_number": float(values["condition_number"]),
                "roster_hash": roster_hash,
            }
        )
    # Diagnose the pre-treatment factor subspace directly.
    _record_diagnostic(
        "factor_pre",
        factor_matrix,
        FACTOR_COMPONENTS,
        "__partition__",
    )
    # Initialise the per-unit matrix mapping.
    matrices: Dict[str, Tuple[np.ndarray, np.ndarray]] = {}
    # Initialise the labelled axes aligned with every stored array.
    matrix_axes: Dict[str, Mapping[str, Tuple[Any, ...]]] = {}
    # Accumulate residualized controls for the pooled-solver diagnostic.
    residualized_blocks: List[np.ndarray] = []
    # Construct the factor residual-maker from the SVD pseudo-inverse.
    factor_residual_maker: np.ndarray = np.eye(len(estimation_years)) - (
        factor_matrix @ np.linalg.pinv(factor_matrix)
    )
    # Build the matrices of each estimation unit.
    for unit in unit_codes:
        # Build the unit's dependent and pooling matrices.
        dependent, pooling = build_pooling_matrix(
            lagged, str(unit), outcome, proxy, window
        )
        # Record the unit's matrices.
        matrices[str(unit)] = (dependent, pooling)
        # Preserve the exact axes that give each array scientific meaning.
        matrix_axes[str(unit)] = {
            "years": estimation_years,
            "dependent_columns": (outcome,),
            "pooling_columns": (
                "constant",
                *(f"{name}{LAG_SUFFIX}" for name in CONTROL_COLUMNS),
                "donor_csa",
            ),
        }
        # Isolate the eight lagged controls from the pooling matrix.
        controls: np.ndarray = pooling[:, 1:-1]
        # Residualize controls on the same labelled factor subspace.
        residualized_controls: np.ndarray = factor_residual_maker @ controls
        # Accumulate the residualized block for the pooled solver matrix.
        residualized_blocks.append(residualized_controls)
        # Diagnose the unit's labelled pooling matrix directly.
        _record_diagnostic(
            "pooling_matrix",
            pooling,
            matrix_axes[str(unit)]["pooling_columns"],
            str(unit),
        )
        # Diagnose the controls after factor residualization.
        _record_diagnostic(
            "residualized_controls",
            residualized_controls,
            CONTROL_COLUMNS,
            str(unit),
        )
    # Stack exactly the residualized controls used by the pooled solver.
    pooled_solver_matrix: np.ndarray = np.vstack(residualized_blocks)
    # Diagnose the pooled object separately from every per-unit object.
    pooled_values: Dict[str, float] = compute_matrix_diagnostics(
        pooled_solver_matrix
    )
    # Record pooled diagnostics with repeated labelled years per unit.
    diagnostic_rows.append(
        {
            "policy": policy,
            "outcome": outcome,
            "unit": "__pooled__",
            "matrix_name": "pooled_residualized_controls",
            "years": tuple(
                year for _unit in unit_codes for year in estimation_years
            ),
            "columns": tuple(CONTROL_COLUMNS),
            "rows": int(pooled_solver_matrix.shape[0]),
            "column_count": int(pooled_solver_matrix.shape[1]),
            "rank": int(pooled_values["rank"]),
            "rank_convention": "numpy.linalg.matrix_rank_default_svd_tolerance",
            "smallest_singular_value": float(
                pooled_values["smallest_singular_value"]
            ),
            "largest_singular_value": float(
                pooled_values["largest_singular_value"]
            ),
            "condition_number": float(pooled_values["condition_number"]),
            "roster_hash": roster_hash,
        }
    )
    # Assemble the uniquely keyed diagnostic table.
    diagnostics: pd.DataFrame = pd.DataFrame(diagnostic_rows).set_index(
        ["matrix_name", "unit"]
    )
    # Log the assembled partition factor.
    logger.info(
        "%s/%s: factor assembled over %d unit(s) with %d donor(s).",
        policy,
        outcome,
        len(matrices),
        donor_frame.shape[1],
    )
    # Return the frozen partition factor.
    return PartitionFactor(
        policy=policy,
        outcome=outcome,
        proxy=proxy,
        donor_count=int(donor_frame.shape[1]),
        matrices=matrices,
        diagnostics=diagnostics,
        matrix_axes=matrix_axes,
    )


# Freeze the factor objects behind a canonical digest.
def freeze_factor_objects(
    factors: Mapping[Tuple[str, str], PartitionFactor]
) -> str:
    """
    Compute the canonical digest of the factor objects.

    Parameters
    ----------
    factors : mapping of tuple to PartitionFactor
        The factor objects keyed on the pair.

    Returns
    -------
    str
        The hexadecimal digest of the canonical payload.

    Raises
    ------
    IngestError
        If the mapping is empty.
    """
    # Refuse an empty mapping.
    if not factors:
        # The failure states the precondition.
        raise IngestError("Cannot freeze an empty factor mapping.")
    # Accumulate the canonical payload parts in sorted pair order.
    parts: List[str] = []
    # Iterate over the pairs in sorted order for determinism.
    for pair in sorted(factors.keys()):
        # Select the factor of the pair.
        factor = factors[pair]
        # Record the pair header with its donor count.
        parts.append(f"pair={pair[0]}|{pair[1]}|donors={factor.donor_count}")
        # Record the proxy values at full precision.
        for year, value in factor.proxy.items():
            # Append the proxy value in its canonical form.
            parts.append(f"F|{int(year)}|{float(value):.17g}")
        # Record every unit's matrices in sorted unit order.
        for unit in sorted(factor.matrices.keys()):
            # Read the unit's dependent and pooling matrices.
            dependent, pooling = factor.matrices[unit]
            # Append the dependent values in their canonical form.
            parts.append(
                "Y|"
                + unit
                + "|"
                + "|".join(f"{float(value):.17g}" for value in dependent.ravel())
            )
            # Append the pooling values in their canonical form.
            parts.append(
                "Z|"
                + unit
                + "|"
                + "|".join(f"{float(value):.17g}" for value in pooling.ravel())
            )
            # Record the explicit scientific axes aligned with both arrays.
            parts.append(
                "AXES|"
                + unit
                + "|"
                + json.dumps(
                    factor.matrix_axes[unit], sort_keys=True, default=str
                )
            )
        # Record every matrix-specific diagnostic and its provenance.
        for (matrix_name, unit), row in factor.diagnostics.sort_index().iterrows():
            # Append a canonical JSON line containing labels and values.
            parts.append(
                "DIAGNOSTIC|"
                + str(matrix_name)
                + "|"
                + str(unit)
                + "|"
                + json.dumps(row.to_dict(), sort_keys=True, default=str)
            )
    # Digest the encoded payload.
    return hashlib.sha256("\n".join(parts).encode("utf-8")).hexdigest()


# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the donor factor proxy construction across the partitions.
def orchestrate_donor_factor(
    registry: PartitionRegistry,
    panel: pd.DataFrame,
    lagged: pd.DataFrame,
    units_by_pair: Optional[Mapping[Tuple[str, str], Sequence[str]]] = None,
    window: Tuple[int, int] = PRE_TREATMENT_WINDOW,
    factor_window: Tuple[int, int] = FACTOR_WINDOW,
) -> DonorFactorResult:
    """
    Construct the donor factor proxies and the pooling matrices.

    Parameters
    ----------
    registry : PartitionRegistry
        The registered partitions.
    panel : pandas.DataFrame
        The canonical analysis panel.
    lagged : pandas.DataFrame
        The panel carrying the lagged control columns.
    units_by_pair : mapping of tuple to sequence of str, optional
        The estimation units of each pair. When absent, each partition's own
        targets and donors form its unit set.
    window : tuple of int, optional
        The inclusive estimation window of the matrices.
    factor_window : tuple of int, optional
        The inclusive window of the proxy series.

    Returns
    -------
    DonorFactorResult
        The factor objects, the long proxy frame, the diagnostics, the digest,
        the timestamp and the notes.

    Raises
    ------
    KeyError
        If the registry carries no partition for a declared pair.
    IngestError
        If any proxy, matrix or diagnostic check fails.
    """
    # Initialise the factor mapping keyed on the pair.
    factors: Dict[Tuple[str, str], PartitionFactor] = {}
    # Accumulate the long proxy rows.
    proxy_rows: List[Dict[str, Any]] = []
    # Accumulate the diagnostics rows.
    diagnostic_rows: List[Dict[str, Any]] = []
    # Accumulate the conditioning notes.
    notes: List[str] = []
    # Iterate over the registry's partitions in sorted pair order.
    for pair in sorted(registry.partitions.keys()):
        # Select the partition of the pair.
        partition = registry.partitions[pair]
        # Compose the unit set of the pair.
        unit_list: List[str] = (
            [str(code) for code in units_by_pair[pair]]
            if units_by_pair is not None and pair in units_by_pair
            else sorted(list(partition.targets) + list(partition.donors))
        )
        # Assemble the factor objects of the partition.
        factor: PartitionFactor = assemble_partition_factor(
            pair[0],
            pair[1],
            partition.donors,
            unit_list,
            panel,
            lagged,
            window=window,
            factor_window=factor_window,
        )
        # Record the factor of the pair.
        factors[pair] = factor
        # Append the partition's proxy values to the long frame.
        for year, value in factor.proxy.items():
            # Record the proxy value with its partition keys.
            proxy_rows.append(
                {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "year": int(year),
                    "factor_value": float(value),
                    "donor_count": factor.donor_count,
                }
            )
        # Compute the diagnostics of every unit's pooling matrix.
        condition_numbers: List[float] = []
        # Compute the worst rank and the dispersion of the proxy.
        ranks: List[float] = []
        # Iterate over the units in sorted order.
        for unit in sorted(factor.matrices.keys()):
            # Read the unit's pooling matrix.
            _dependent, pooling = factor.matrices[unit]
            # Compute the matrix diagnostics.
            diagnostics: Dict[str, float] = compute_matrix_diagnostics(pooling)
            # Record the condition number of the unit.
            condition_numbers.append(diagnostics["condition_number"])
            # Record the rank of the unit's matrix.
            ranks.append(diagnostics["rank"])
        # Compute the worst condition number across the units.
        worst_condition: float = float(max(condition_numbers)) if condition_numbers else float("nan")
        # Record a conditioning note when the threshold is exceeded.
        if worst_condition > CONDITION_NUMBER_WARNING:
            # Append the warning note naming the pair.
            notes.append(
                f"{pair[0]}/{pair[1]}: worst condition number "
                f"{worst_condition:.3e} exceeds the declared threshold."
            )
        # Append the diagnostics row of the partition.
        diagnostic_rows.append(
            {
                "policy": pair[0],
                "outcome": pair[1],
                "donor_count": int(factor.donor_count),
                "units": int(len(factor.matrices)),
                "worst_condition_number": worst_condition,
                "minimum_rank": float(min(ranks)) if ranks else float("nan"),
                "proxy_sd": float(factor.proxy.std(ddof=1)),
                "proxy_min": float(factor.proxy.min()),
                "proxy_max": float(factor.proxy.max()),
            }
        )
    # Assemble the long proxy frame.
    proxy_frame: pd.DataFrame = pd.DataFrame(proxy_rows)
    # Sort the proxy frame by its declared key for determinism.
    proxy_frame = proxy_frame.sort_values(
        ["policy", "outcome", "year"], kind="stable"
    ).reset_index(drop=True)
    # Assemble the diagnostics table keyed on the pair.
    diagnostics_table: pd.DataFrame = pd.DataFrame(diagnostic_rows).set_index(
        ["policy", "outcome"]
    )
    # Compute the canonical digest of the factor objects.
    fingerprint: str = freeze_factor_objects(factors)
    # Record the coverage note.
    notes.append(
        f"proxy frame covers {len(factors)} partition(s) over "
        f"{factor_window[0]} to {factor_window[1]}."
    )
    # Assemble the immutable result.
    result: DonorFactorResult = DonorFactorResult(
        factors=factors,
        proxy_frame=proxy_frame,
        diagnostics=diagnostics_table,
        fingerprint_sha256=fingerprint,
        built_utc=current_utc_iso(),
        notes=tuple(notes),
    )
    # Log the completed construction with its digest prefix.
    logger.info(
        "donor factor construction complete: %d partition(s), fingerprint %s.",
        len(factors),
        fingerprint[:16],
    )
    # Return the assembled result.
    return result
````

````python
# Task 11: Pre-Treatment Pooled Fit (Step 2 of CCEDID)

# ==============================================================================
# Task 11: Pre-Treatment Pooled Fit (Step 2 of CCEDID)
# ==============================================================================

# Declare the two components of the factor matrix in their required order.
FACTOR_COMPONENTS: Final[Tuple[str, str]] = ("constant", "donor_csa")
# Declare the number of factor columns.
FACTOR_COLUMNS: Final[int] = len(FACTOR_COMPONENTS)
# Declare the minimum admissible number of factor columns. The declared
# factor carries two components; a wider factor is admitted because the
# factor-proxy sweep declares variants whose factor matrix carries a third
# component, and the residual-maker's own dimension depends on the year count
# rather than on the component count.
MINIMUM_FACTOR_COLUMNS: Final[int] = 2
# Declare the condition-number ceiling above which the pseudo-inverse is used.
INVERSION_CONDITION_MAX: Final[float] = 1.0e3
# Declare the condition-number ceiling above which the stacked cross-product
# warrants a recorded warning.
STACKED_CONDITION_WARNING: Final[float] = 1.0e6
# Declare the tolerance of the maker's identity checks.
IDENTITY_TOLERANCE: Final[float] = 1.0e-10
# Declare the tolerance of the orthogonality check.
ORTHOGONALITY_TOLERANCE: Final[float] = 1.0e-10
# Declare the label of the direct-solve route.
ROUTE_SOLVE: Final[str] = "solve"
# Declare the label of the pseudo-inverse route.
ROUTE_PINV: Final[str] = "pinv"
# Declare the one SVD rank convention used throughout the pre-treatment fit.
SVD_RANK_CONVENTION: Final[str] = (
    "sigma > sigma_max * max(rows, columns) * machine_epsilon"
)


# Declare the immutable residual-maker object of one partition.
@dataclass(frozen=True)
class ResidualMaker:
    """
    Immutable residual-maker matrix of one partition.

    Attributes
    ----------
    matrix : numpy.ndarray
        The square residual-maker of the factor matrix.
    route : str
        The inversion route taken, either the direct solve or the
        pseudo-inverse.
    condition_number : float
        The condition number of the factor cross-product.
    checks : dict of str to bool
        The outcome of the annihilation, symmetry and idempotence checks.
    factor_rank : int
        The retained SVD rank of the factor matrix.
    rank_tolerance : float
        The absolute singular-value threshold used for the retained rank.
    rank_convention : str
        The declared SVD rank convention shared by all fit stages.
    """
    # The square residual-maker of the factor matrix.
    matrix: np.ndarray
    # The inversion route taken.
    route: str
    # The condition number of the factor cross-product.
    condition_number: float
    # The outcome of the identity checks.
    checks: Mapping[str, bool]
    # The retained SVD rank of the factor matrix.
    factor_rank: int = 0
    # The absolute singular-value threshold used for factor rank.
    rank_tolerance: float = float("nan")
    # The shared rank-convention label.
    rank_convention: str = SVD_RANK_CONVENTION


# Declare the immutable result of the pre-treatment pooled fit.
@dataclass(frozen=True)
class PooledFitResult:
    """
    Immutable result of the pre-treatment pooled fit.

    Attributes
    ----------
    betas : dict
        The pooled coefficient vectors keyed on the pair, each a length-eight
        array whose companion name list is carried separately.
    beta_names : tuple of str
        The control names aligned with every coefficient vector.
    loadings : dict
        The country loading vectors keyed first by pair and then by country
        code, each a two-vector ordered as the constant and the donor average.
    ledger : pandas.DataFrame
        One row per pair recording the routes, the condition numbers, the
        residual sum of squares and the degrees of freedom.
    fingerprint_sha256 : str
        Digest of the canonical fit payload.
    fitted_utc : str
        ISO-8601 UTC timestamp of the fit.
    notes : tuple of str
        Recorded decisions, including conditioning warnings.
    intermediates : mapping
        Labelled per-partition factor, projector and residualized-design
        intermediates independently verified by the orchestrator.
    """
    # The pooled coefficient vectors keyed on the pair.
    betas: Mapping[Tuple[str, str], np.ndarray]
    # The control names aligned with every coefficient vector.
    beta_names: Tuple[str, ...]
    # The country loading vectors keyed by pair and country code.
    loadings: Mapping[Tuple[str, str], Mapping[str, np.ndarray]]
    # One row per pair recording the numerical diagnostics.
    ledger: pd.DataFrame
    # Digest of the canonical fit payload.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the fit.
    fitted_utc: str
    # Recorded decisions of the fit.
    notes: Tuple[str, ...]
    # Labelled per-partition fitting intermediates.
    intermediates: Mapping[Tuple[str, str], Mapping[str, Any]]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Step 1: Construct the residual-maker matrix
# -------------------------------------------------------------------------------------------------------------------------------


# Split one unit's design into its covariate and factor matrices.
def split_design(
    pooling: np.ndarray,
) -> Tuple[np.ndarray, np.ndarray]:
    """
    Split a unit's design into covariates and factor columns.

    Parameters
    ----------
    pooling : numpy.ndarray
        The design matrix whose first column is the intercept, whose next
        columns are the lagged controls and whose final column is the donor
        cross-sectional average.

    Returns
    -------
    tuple
        The covariate matrix of shape rows-by-eight and the factor matrix of
        shape rows-by-two ordered as the constant and the donor average.

    Raises
    ------
    IngestError
        If the design does not carry the declared number of columns, or if
        its row count is not positive.
    """
    # Compute the declared column count of the design.
    declared: int = 1 + len(CONTROL_COLUMNS) + 1
    # Refuse a design whose column count diverges from the declared form.
    if pooling.ndim != 2 or pooling.shape[1] != declared:
        # The failure names the observed shape.
        raise IngestError(
            f"Design has shape {pooling.shape} against the declared "
            f"(rows, {declared})."
        )
    # Refuse a design without rows.
    if pooling.shape[0] == 0:
        # The failure states the precondition.
        raise IngestError("Cannot split a design with no rows.")
    # Extract the eight covariate columns, which follow the intercept.
    covariates: np.ndarray = pooling[:, 1 : 1 + len(CONTROL_COLUMNS)]
    # Compose the factor matrix from the intercept and the donor average.
    factor_matrix: np.ndarray = np.column_stack(
        [pooling[:, 0], pooling[:, -1]]
    )
    # Return the covariate and factor matrices.
    return covariates, factor_matrix


# Build the residual-maker matrix of one partition's factor matrix.
def build_residual_maker(
    factor_matrix: np.ndarray,
    condition_max: float = INVERSION_CONDITION_MAX,
    tolerance: float = IDENTITY_TOLERANCE,
) -> ResidualMaker:
    """
    Build the residual-maker matrix of a factor matrix.

    Parameters
    ----------
    factor_matrix : numpy.ndarray
        The factor matrix of shape years-by-two.
    condition_max : float, optional
        The condition-number ceiling above which the pseudo-inverse is used.
    tolerance : float, optional
        The tolerance of the identity checks.

    Returns
    -------
    ResidualMaker
        The residual-maker, the inversion route, the condition number and the
        identity-check outcomes.

    Raises
    ------
    IngestError
        If the factor matrix does not carry the declared two columns, or if
        any identity check fails.
    """
    # Refuse a factor matrix whose shape diverges from the declared form. The
    # declared factor carries two components, and a wider factor is admitted
    # because the factor-proxy sweep declares variants whose factor matrix
    # carries a third component; the maker's own dimension depends on the
    # year count rather than on the component count.
    if factor_matrix.ndim != 2 or factor_matrix.shape[1] < MINIMUM_FACTOR_COLUMNS:
        # The failure names the observed shape.
        raise IngestError(
            f"Factor matrix has shape {factor_matrix.shape}, which carries fewer "
            f"than the declared {MINIMUM_FACTOR_COLUMNS} component(s) or is not "
            f"a matrix."
        )
    # Require finite positive numerical controls.
    if (
        isinstance(condition_max, (bool, np.bool_))
        or not np.isfinite(float(condition_max))
        or float(condition_max) <= 0.0
    ):
        # Invalid conditioning thresholds make route selection undefined.
        raise IngestError(
            f"condition_max must be finite and positive; observed {condition_max!r}."
        )
    # Require a finite tolerance strictly between zero and one.
    if (
        isinstance(tolerance, (bool, np.bool_))
        or not np.isfinite(float(tolerance))
        or not 0.0 < float(tolerance) < 1.0
    ):
        # A non-finite or unbounded tolerance cannot verify identities.
        raise IngestError(
            f"Residual-maker tolerance must lie in (0, 1); observed {tolerance!r}."
        )
    # Require finite factor values and at least one row.
    if factor_matrix.shape[0] == 0 or not np.isfinite(factor_matrix).all():
        # The factor subspace is undefined for an empty/non-finite matrix.
        raise IngestError("Factor matrix is empty or non-finite.")
    # Compute the factor SVD under the shared rank convention.
    _left, factor_singular, _right = np.linalg.svd(
        factor_matrix, full_matrices=False
    )
    # Resolve the absolute retained-rank threshold.
    factor_rank_tolerance: float = float(
        factor_singular[0]
        * max(factor_matrix.shape)
        * np.finfo(float).eps
    )
    # Count the retained factor directions.
    factor_rank: int = int(
        np.count_nonzero(factor_singular > factor_rank_tolerance)
    )
    # Refuse a factor matrix with no retained scientific direction.
    if factor_rank == 0:
        # A zero factor cannot define a projection model.
        raise IngestError("Factor matrix has numerical rank zero.")
    # Form the factor cross-product.
    cross_product: np.ndarray = factor_matrix.T @ factor_matrix
    # Compute the singular values of the cross-product.
    singular_values: np.ndarray = np.linalg.svd(cross_product, compute_uv=False)
    # Read the largest singular value.
    largest: float = float(singular_values[0])
    # Read the smallest singular value.
    smallest: float = float(singular_values[-1])
    # Treat a numerically singular cross-product as rank-deficient by
    # comparing its smallest singular value against the matrix's own scale,
    # which is the standard relative rank test.
    cross_rank_tolerance: float = (
        largest * float(np.finfo(float).eps) * max(cross_product.shape)
    )
    # Record whether the cross-product is numerically rank-deficient.
    rank_deficient: bool = bool(smallest <= cross_rank_tolerance)
    # Compute the condition number, reporting infinity for a numerically
    # singular matrix rather than a large finite ratio.
    condition: float = (
        float("inf") if rank_deficient else float(largest / smallest)
    )
    # Select the direct solve when the cross-product is well conditioned.
    if np.isfinite(condition) and condition <= float(condition_max):
        # Compute the coefficient block by solving the cross-product system,
        # which yields (F'F)^{-1} F'.
        block: np.ndarray = np.linalg.solve(cross_product, factor_matrix.T)
        # Record the route taken.
        route: str = ROUTE_SOLVE
    # Otherwise take the Moore-Penrose pseudo-inverse, which is SVD based and
    # remains well defined when the cross-product is rank-deficient.
    else:
        # Compute the coefficient block from the pseudo-inverse of F.
        # Apply the same retained factor rank threshold explicitly.
        block = np.linalg.pinv(
            factor_matrix,
            rcond=(
                factor_rank_tolerance / float(factor_singular[0])
                if factor_singular[0] > 0.0
                else 0.0
            ),
        )
        # Record the route taken.
        route = ROUTE_PINV
    # Compose the residual-maker as the identity less the projection.
    identity: np.ndarray = np.eye(factor_matrix.shape[0])
    # Form the maker.
    maker: np.ndarray = identity - factor_matrix @ block
    # Verify that the maker annihilates the factor matrix.
    annihilation_scale: float = max(
        float(np.linalg.norm(factor_matrix, ord="fro")),
        np.finfo(float).tiny,
    )
    # Use a scale-free annihilation error.
    annihilation_error: float = float(
        np.linalg.norm(maker @ factor_matrix, ord="fro")
        / annihilation_scale
    )
    # Verify annihilation under the declared relative tolerance.
    annihilation: bool = bool(annihilation_error <= float(tolerance))
    # Verify that the maker is symmetric.
    symmetry: bool = bool(
        np.linalg.norm(maker - maker.T, ord="fro") <= float(tolerance)
    )
    # Verify that the maker is idempotent.
    idempotence: bool = bool(
        np.linalg.norm(maker @ maker - maker, ord="fro") <= float(tolerance)
    )
    # Record the check outcomes.
    checks: Dict[str, bool] = {
        "annihilates_factor": annihilation,
        "symmetric": symmetry,
        "idempotent": idempotence,
    }
    # Refuse a maker that fails any identity check.
    if not all(checks.values()):
        # The failure names the failed checks.
        failed = sorted(name for name, passed in checks.items() if not passed)
        # Raise with the failed checks attached.
        raise IngestError(
            f"Residual-maker fails its identity check(s): {failed}."
        )
    # Log the construction with its route and condition number.
    logger.info(
        "residual-maker built: route %s, condition number %.3e.",
        route,
        condition,
    )
    # Return the frozen maker object.
    return ResidualMaker(
        matrix=maker,
        route=route,
        condition_number=condition,
        checks=checks,
        factor_rank=factor_rank,
        rank_tolerance=factor_rank_tolerance,
        rank_convention=SVD_RANK_CONVENTION,
    )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Step 2: Residualize and estimate the pooled common coefficient
# -------------------------------------------------------------------------------------------------------------------------------


# Estimate the pooled common coefficient from the residualized design.
def estimate_pooled_coefficient(
    dependent_by_unit: Mapping[str, np.ndarray],
    covariate_by_unit: Mapping[str, np.ndarray],
    maker: ResidualMaker,
    condition_warning: float = STACKED_CONDITION_WARNING,
) -> Tuple[np.ndarray, Dict[str, Any]]:
    """
    Estimate the pooled common coefficient.

    Parameters
    ----------
    dependent_by_unit : mapping of str to numpy.ndarray
        The dependent matrix of each unit.
    covariate_by_unit : mapping of str to numpy.ndarray
        The covariate matrix of each unit.
    maker : ResidualMaker
        The residual-maker of the partition.
    condition_warning : float, optional
        The condition-number ceiling above which a warning is recorded.

    Returns
    -------
    tuple
        The length-eight coefficient vector and its diagnostics, which carry
        the residual sum of squares, factor-aware residual degrees of freedom,
        both retained ranks, both condition numbers and the difference between
        direct-design and manuscript cross-product solutions.

    Raises
    ------
    IngestError
        If the unit sets differ between the two mappings, if a covariate
        matrix does not carry the declared column count, or if the residual
        sum of squares is not finite.
    """
    # Validate the numerical warning threshold.
    if (
        isinstance(condition_warning, (bool, np.bool_))
        or not np.isfinite(float(condition_warning))
        or float(condition_warning) <= 0.0
    ):
        # Invalid warning thresholds make numerical reporting meaningless.
        raise IngestError(
            "condition_warning must be finite and positive; observed "
            f"{condition_warning!r}."
        )
    # Refuse divergent unit sets between the dependent and covariate maps.
    if set(dependent_by_unit.keys()) != set(covariate_by_unit.keys()):
        # The failure names the divergence.
        raise IngestError(
            "Dependent and covariate unit sets diverge; the stack would be "
            "misaligned."
        )
    # Refuse an empty unit set.
    if not dependent_by_unit:
        # Without units the coefficient is undefined.
        raise IngestError("Cannot estimate a pooled coefficient over no units.")
    # Accumulate the residualized dependent blocks in sorted unit order.
    residualized_dependent: List[np.ndarray] = []
    # Accumulate the residualized covariate blocks in the same order.
    residualized_covariate: List[np.ndarray] = []
    # Iterate over the units in sorted order for a reproducible row layout.
    for unit in sorted(dependent_by_unit.keys()):
        # Read the unit's dependent matrix.
        dependent: np.ndarray = dependent_by_unit[unit]
        # Read the unit's covariate matrix.
        covariate: np.ndarray = covariate_by_unit[unit]
        # Require a finite rows-by-one dependent matrix.
        if (
            dependent.ndim != 2
            or dependent.shape[1] != 1
            or not np.isfinite(dependent).all()
        ):
            # The failure names the unit and observed shape.
            raise IngestError(
                f"{unit}: dependent matrix must be finite rows-by-one; "
                f"observed shape {dependent.shape}."
            )
        # Refuse a covariate matrix whose column count diverges.
        if covariate.ndim != 2 or covariate.shape[1] != len(CONTROL_COLUMNS):
            # The failure names the unit and the observed shape.
            raise IngestError(
                f"{unit}: covariate shape {covariate.shape} against the declared "
                f"(rows, {len(CONTROL_COLUMNS)})."
            )
        # Require row alignment with the dependent and maker matrices.
        if (
            covariate.shape[0] != dependent.shape[0]
            or maker.matrix.shape
            != (dependent.shape[0], dependent.shape[0])
        ):
            # Prevent silent broadcasting or partition misalignment.
            raise IngestError(
                f"{unit}: dependent, covariate and residual-maker row axes "
                "are misaligned."
            )
        # Require finite covariates.
        if not np.isfinite(covariate).all():
            # Numerical solvers cannot admit non-finite design values.
            raise IngestError(f"{unit}: covariate matrix is non-finite.")
        # Residualize the unit's dependent matrix.
        residualized_dependent.append(maker.matrix @ dependent)
        # Residualize the unit's covariate matrix.
        residualized_covariate.append(maker.matrix @ covariate)
    # Stack the residualized dependent blocks vertically.
    stacked_dependent: np.ndarray = np.vstack(residualized_dependent)
    # Stack the residualized covariate blocks in the same order.
    stacked_covariate: np.ndarray = np.vstack(residualized_covariate)
    # Compute the direct residualized-design SVD under the shared convention.
    design_singular: np.ndarray = np.linalg.svd(
        stacked_covariate, compute_uv=False
    )
    # Resolve the direct-design retained-rank threshold.
    design_tolerance: float = float(
        design_singular[0]
        * max(stacked_covariate.shape)
        * np.finfo(float).eps
    )
    # Count the retained residualized-control directions.
    design_rank: int = int(
        np.count_nonzero(design_singular > design_tolerance)
    )
    # Solve the direct residualized least-squares oracle with that threshold.
    design_solution: np.ndarray = np.linalg.pinv(
        stacked_covariate,
        rcond=(
            design_tolerance / float(design_singular[0])
            if design_singular[0] > 0.0
            else 0.0
        ),
    ) @ stacked_dependent
    # Form the manuscript's pooled cross-product and right-hand side.
    stacked_cross: np.ndarray = stacked_covariate.T @ stacked_covariate
    # Sum X-tilde prime Y-tilde through the identical stacked row order.
    stacked_rhs: np.ndarray = stacked_covariate.T @ stacked_dependent
    # Compute cross-product singular values directly.
    cross_singular: np.ndarray = np.linalg.svd(
        stacked_cross, compute_uv=False
    )
    # Resolve the cross-product retained-rank threshold.
    cross_tolerance: float = float(
        cross_singular[0]
        * max(stacked_cross.shape)
        * np.finfo(float).eps
    )
    # Count the retained manuscript cross-product directions.
    cross_rank: int = int(
        np.count_nonzero(cross_singular > cross_tolerance)
    )
    # Apply the manuscript cross-product pseudoinverse convention explicitly.
    manuscript_solution: np.ndarray = np.linalg.pinv(
        stacked_cross,
        rcond=(
            cross_tolerance / float(cross_singular[0])
            if cross_singular[0] > 0.0
            else 0.0
        ),
    ) @ stacked_rhs
    # Return the coefficient defined by the manuscript equation.
    beta: np.ndarray = manuscript_solution.ravel()
    # Compute the residual sum of squares from the residualized quantities.
    residuals: np.ndarray = (
        stacked_dependent - stacked_covariate @ manuscript_solution
    )
    # Compute the residual sum of squares as the squared Frobenius norm, which
    # equals the (1,1) quadratic form without materialising it (float() on a
    # (1,1) array raises TypeError under NumPy >= 1.25).
    rss: float = float(np.sum(np.square(residuals, dtype=np.float64)))
    # Resolve factor rank, including compatibility with legacy maker records.
    effective_factor_rank: int = (
        int(maker.factor_rank)
        if int(maker.factor_rank) > 0
        else int(
            maker.matrix.shape[0]
            - np.linalg.matrix_rank(maker.matrix)
        )
    )
    # Count control and unit-specific factor parameters in residual degrees.
    degrees_of_freedom: int = int(
        stacked_covariate.shape[0]
        - cross_rank
        - effective_factor_rank * len(dependent_by_unit)
    )
    # Also expose the residualized-design-only diagnostic degrees of freedom.
    residualized_design_dof: int = int(
        stacked_covariate.shape[0] - design_rank
    )
    # Compute the direct-design condition number on the retained matrix.
    design_condition: float = (
        float("inf")
        if design_singular[-1] <= design_tolerance
        else float(design_singular[0] / design_singular[-1])
    )
    # Compute the cross-product condition number under its rank threshold.
    condition: float = (
        float("inf")
        if cross_singular[-1] <= cross_tolerance
        else float(cross_singular[0] / cross_singular[-1])
    )
    # Quantify any near-singular convention difference without hiding it.
    solution_difference: float = float(
        np.linalg.norm(manuscript_solution - design_solution)
    )
    # Refuse a non-finite residual sum of squares.
    if not np.isfinite(rss):
        # A non-finite residual indicates a numerical failure.
        raise IngestError("Residual sum of squares is not finite.")
    # Record a warning when the conditioning exceeds the declared ceiling.
    if condition > float(condition_warning):
        # Log the warning with both numbers.
        logger.warning(
            "stacked cross-product condition number %.3e exceeds the declared "
            "warning ceiling %.3e.",
            condition,
            condition_warning,
        )
    # Assemble the diagnostic mapping.
    diagnostics: Dict[str, Any] = {
        "rss": rss,
        "degrees_of_freedom": float(degrees_of_freedom),
        "residualized_design_degrees_of_freedom": float(
            residualized_design_dof
        ),
        "rank": float(cross_rank),
        "design_rank": float(design_rank),
        "cross_product_rank": float(cross_rank),
        "factor_rank": float(effective_factor_rank),
        "rank_convention": SVD_RANK_CONVENTION,
        "design_rank_tolerance": design_tolerance,
        "cross_product_rank_tolerance": cross_tolerance,
        "design_condition_number": design_condition,
        "condition_number": condition,
        "design_solution_difference_norm": solution_difference,
        "rows": float(stacked_covariate.shape[0]),
        "units": float(len(dependent_by_unit)),
    }
    # Log the fit outcome.
    logger.info(
        "pooled fit: %d rows, rank %d, residual sum of squares %.6e.",
        stacked_covariate.shape[0],
        cross_rank,
        rss,
    )
    # Return the coefficient and its diagnostics.
    return beta, diagnostics


# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Step 3: Recover the country-specific factor loadings
# -------------------------------------------------------------------------------------------------------------------------------


# Verify the orthogonality identity of one unit's loading fit.
def assert_unit_orthogonality(
    factor_matrix: np.ndarray,
    residual: np.ndarray,
    tolerance: float = ORTHOGONALITY_TOLERANCE,
    conditioning: float = 1.0,
) -> None:
    """
    Verify that the loading residual is orthogonal to the factor matrix.

    Parameters
    ----------
    factor_matrix : numpy.ndarray
        The factor matrix of the partition.
    residual : numpy.ndarray
        The residual of the unit's loading fit.
    tolerance : float, optional
        The tolerance of the check.
    conditioning : float, optional
        A finite conditioning diagnostic from the pooled estimation. It
        contributes only bounded machine-precision slack and cannot weaken
        the geometric check enough to admit a parallel residual.

    Returns
    -------
    None
        The function returns nothing when the identity holds.

    Raises
    ------
    IngestError
        If the inner product of the factor matrix and the residual exceeds the
        declared tolerance, which indicates an algebraic error rather than a
        data problem.
    """
    # Require finite bounded tolerance and conditioning controls.
    if (
        isinstance(tolerance, (bool, np.bool_))
        or not np.isfinite(float(tolerance))
        or not 0.0 < float(tolerance) < 1.0
    ):
        # A tolerance outside (0,1) cannot express a projection ratio.
        raise IngestError(
            f"Orthogonality tolerance must lie in (0, 1); observed {tolerance!r}."
        )
    # Require a finite positive inherited conditioning measure.
    if (
        isinstance(conditioning, (bool, np.bool_))
        or not np.isfinite(float(conditioning))
        or float(conditioning) < 1.0
    ):
        # Infinite conditioning must be diagnosed rather than weakening the test.
        raise IngestError(
            "Orthogonality conditioning must be finite and at least one; "
            f"observed {conditioning!r}."
        )
    # Require finite two-dimensional aligned arrays.
    if (
        factor_matrix.ndim != 2
        or residual.ndim != 2
        or factor_matrix.shape[0] != residual.shape[0]
        or factor_matrix.shape[0] == 0
        or not np.isfinite(factor_matrix).all()
        or not np.isfinite(residual).all()
    ):
        # Malformed axes cannot define a geometric projection.
        raise IngestError(
            "Factor and residual matrices must be finite, nonempty and "
            "row-aligned."
        )
    # A zero residual is exactly orthogonal to every subspace.
    residual_norm: float = float(np.linalg.norm(residual, ord="fro"))
    # Return immediately for the exact zero case.
    if residual_norm == 0.0:
        # Record the exact identity.
        logger.info("orthogonality identity verified for a zero residual.")
        return
    # Construct an orthonormal basis of the factor's retained column space.
    basis, singular_values, _right = np.linalg.svd(
        factor_matrix, full_matrices=False
    )
    # Apply the same shared SVD rank convention used during fitting.
    cutoff: float = float(
        singular_values[0]
        * max(factor_matrix.shape)
        * np.finfo(float).eps
    )
    # Count retained factor directions.
    retained: int = int(np.count_nonzero(singular_values > cutoff))
    # Project the residual onto the retained orthonormal basis.
    projected_norm: float = (
        float(np.linalg.norm(basis[:, :retained].T @ residual, ord="fro"))
        if retained
        else 0.0
    )
    # Normalize only by residual scale; the basis is already orthonormal.
    ratio: float = projected_norm / residual_norm
    # Allow bounded arithmetic slack without weakening the geometric verdict.
    arithmetic_slack: float = min(
        float(np.sqrt(np.finfo(float).eps)),
        float(conditioning) * float(np.finfo(float).eps),
    )
    # Compose the finite bounded effective tolerance.
    effective_tolerance: float = max(float(tolerance), arithmetic_slack)
    # Accept machine-scale absolute projections before a relative near-zero test.
    if projected_norm <= float(tolerance):
        # Near-zero fitted residuals can have an unstable relative ratio.
        logger.info(
            "orthogonality verified by absolute retained-subspace projection "
            "%.3e against tolerance %.3e.",
            projected_norm,
            tolerance,
        )
        # The orthonormal projection is independent of factor scaling.
        return
    # Accept only a small relative projection onto the retained factor space.
    if ratio <= effective_tolerance:
        # Log the rank-aware, scale-invariant verification.
        logger.info(
            "orthogonality verified: retained rank %d, projection ratio %.3e "
            "against tolerance %.3e.",
            retained,
            ratio,
            effective_tolerance,
        )
        # Return after the geometric check succeeds.
        return
    # Refuse a geometrically non-orthogonal residual.
    raise IngestError(
        "Loading residual is not orthogonal to the retained factor subspace: "
        f"projection ratio {ratio:.3e} against tolerance "
        f"{effective_tolerance:.3e}, retained rank {retained} under "
        f"{SVD_RANK_CONVENTION}."
    )


# Estimate the country-specific factor loadings of one partition.
def estimate_unit_loadings(
    factor_matrix: np.ndarray,
    dependent_by_unit: Mapping[str, np.ndarray],
    covariate_by_unit: Mapping[str, np.ndarray],
    beta: np.ndarray,
    tolerance: float = ORTHOGONALITY_TOLERANCE,
    conditioning: float = 1.0,
) -> Dict[str, np.ndarray]:
    """
    Estimate the country-specific factor loadings.

    Parameters
    ----------
    factor_matrix : numpy.ndarray
        The factor matrix of the partition.
    dependent_by_unit : mapping of str to numpy.ndarray
        The dependent matrix of each unit.
    covariate_by_unit : mapping of str to numpy.ndarray
        The covariate matrix of each unit.
    beta : numpy.ndarray
        The pooled common coefficient vector.
    tolerance : float, optional
        The tolerance of the orthogonality check.
    conditioning : float, optional
        The finite pooled-design conditioning diagnostic used only for bounded
        arithmetic slack in the retained-subspace check.

    Returns
    -------
    dict
        The two-vectors keyed on the country code, ordered as the constant
        loading and the donor-average loading.

    Raises
    ------
    IngestError
        If a row-count mismatch is detected, or if the orthogonality identity
        fails for any unit.
    """
    # Initialise the loading mapping.
    loadings: Dict[str, np.ndarray] = {}
    # Iterate over the units in sorted order.
    for unit in sorted(dependent_by_unit.keys()):
        # Read the unit's dependent matrix.
        dependent: np.ndarray = dependent_by_unit[unit]
        # Read the unit's covariate matrix.
        covariate: np.ndarray = covariate_by_unit[unit]
        # Refuse a row-count mismatch between the factor and the unit data.
        if dependent.shape[0] != factor_matrix.shape[0]:
            # The failure names the unit and both row counts.
            raise IngestError(
                f"{unit}: {dependent.shape[0]} row(s) against the factor matrix's "
                f"{factor_matrix.shape[0]}."
            )
        # Remove the unit's covariate contribution using the pooled coefficient.
        remainder: np.ndarray = dependent - covariate @ beta.reshape(-1, 1)
        # Project the remainder on the factor matrix, which yields the loading
        # vector as the minimum-norm solution under rank deficiency.
        loading, _residuals, _rank, _singular_values = np.linalg.lstsq(
            factor_matrix, remainder, rcond=None
        )
        # Compute the residual of the unit's loading fit.
        unit_residual: np.ndarray = remainder - factor_matrix @ loading
        # Verify the orthogonality identity of the fit, at the precision the
        # pooled estimation's conditioning permits.
        assert_unit_orthogonality(
            factor_matrix, unit_residual, tolerance, conditioning
        )
        # Store the loading vector keyed on the country code.
        loadings[str(unit)] = loading.ravel()
    # Log the loading recovery.
    logger.info("unit loadings recovered for %d unit(s).", len(loadings))
    # Return the loading mapping.
    return loadings


# Freeze the fitted quantities behind a canonical digest.
def freeze_pooled_fit(
    betas: Mapping[Tuple[str, str], np.ndarray],
    loadings: Mapping[Tuple[str, str], Mapping[str, np.ndarray]],
) -> str:
    """
    Compute the canonical digest of the fitted quantities.

    Parameters
    ----------
    betas : mapping of tuple to numpy.ndarray
        The coefficient vectors keyed on the pair.
    loadings : mapping of tuple to mapping of str to numpy.ndarray
        The loading vectors keyed by pair and country code.

    Returns
    -------
    str
        The hexadecimal digest of the canonical payload.

    Raises
    ------
    IngestError
        If either mapping is empty.
    """
    # Refuse an empty coefficient mapping.
    if not betas:
        # The failure states the precondition.
        raise IngestError("Cannot freeze an empty coefficient mapping.")
    # Refuse an empty loading mapping.
    if not loadings:
        # The failure states the precondition.
        raise IngestError("Cannot freeze an empty loading mapping.")
    # Accumulate the canonical payload parts.
    parts: List[str] = []
    # Iterate over the pairs in sorted order.
    for pair in sorted(betas.keys()):
        # Record the pair's coefficient vector at full precision.
        parts.append(
            "B|"
            + pair[0]
            + "|"
            + pair[1]
            + "|"
            + "|".join(f"{float(value):.17g}" for value in betas[pair].ravel())
        )
        # Iterate over the pair's units in sorted order.
        for unit in sorted(loadings.get(pair, {}).keys()):
            # Record the unit's loading vector at full precision.
            parts.append(
                "A|"
                + pair[0]
                + "|"
                + pair[1]
                + "|"
                + unit
                + "|"
                + "|".join(
                    f"{float(value):.17g}" for value in loadings[pair][unit].ravel()
                )
            )
    # Digest the encoded payload.
    return hashlib.sha256("\n".join(parts).encode("utf-8")).hexdigest()


# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the pre-treatment pooled fit across the partitions.
def orchestrate_pretreatment_fit(
    factor_result: Any,
    condition_max: float = INVERSION_CONDITION_MAX,
) -> PooledFitResult:
    """
    Estimate the pooled coefficients and the country loadings.

    Parameters
    ----------
    factor_result : DonorFactorResult
        The factor objects produced by the donor factor construction.
    condition_max : float, optional
        The condition-number ceiling above which the pseudo-inverse is used
        for the residual-maker.

    Returns
    -------
    PooledFitResult
        The coefficients, the loadings, the estimator ledger, the digest, the
        timestamp and the notes.

    Raises
    ------
    IngestError
        If any partition's design is malformed, if a maker fails its identity
        checks, or if the orthogonality identity fails.
    """
    # Validate the shared conditioning threshold once for every partition.
    if (
        isinstance(condition_max, (bool, np.bool_))
        or not np.isfinite(float(condition_max))
        or float(condition_max) <= 0.0
    ):
        # All baseline and variant fits must use a valid common policy.
        raise IngestError(
            f"condition_max must be finite and positive; observed {condition_max!r}."
        )
    # Refuse a factor result that carries no partition.
    factors = getattr(factor_result, "factors", None)
    # Validate the factor result's structure.
    if not factors:
        # Without partitions the fit is undefined.
        raise IngestError("The factor result carries no partition.")
    # Initialise the coefficient mapping.
    betas: Dict[Tuple[str, str], np.ndarray] = {}
    # Initialise the loading mapping.
    loadings: Dict[Tuple[str, str], Mapping[str, np.ndarray]] = {}
    # Initialise labelled per-partition fitting intermediates.
    intermediates: Dict[Tuple[str, str], Mapping[str, Any]] = {}
    # Accumulate the estimator ledger rows.
    ledger_rows: List[Dict[str, Any]] = []
    # Accumulate the conditioning warnings.
    notes: List[str] = []
    # Iterate over the partitions in sorted pair order.
    for pair in sorted(factors.keys()):
        # Select the partition's factor object.
        factor = factors[pair]
        # Initialise the per-unit dependent and covariate mappings.
        dependent_by_unit: Dict[str, np.ndarray] = {}
        # Initialise the per-unit covariate mapping.
        covariate_by_unit: Dict[str, np.ndarray] = {}
        # Initialise the factor matrix of the partition.
        factor_matrix: Optional[np.ndarray] = None
        # Initialise the common labelled year axis when supplied.
        common_year_axis: Optional[Tuple[Any, ...]] = None
        # Split each unit's design into its covariate and factor matrices.
        for unit in sorted(factor.matrices.keys()):
            # Read the unit's dependent and design matrices.
            dependent, pooling = factor.matrices[unit]
            # Validate labelled axes when the factor record supplies them.
            supplied_axes: Mapping[str, Mapping[str, Tuple[Any, ...]]] = getattr(
                factor, "matrix_axes", {}
            )
            # Require every unit to have an axis record when any are supplied.
            if supplied_axes and unit not in supplied_axes:
                # An unlabelled unit cannot be aligned to the common factor.
                raise IngestError(
                    f"{pair[0]}/{pair[1]}: unit {unit} lacks matrix axes."
                )
            # Validate the supplied year and column axes.
            if supplied_axes:
                # Read this unit's explicit axes.
                unit_axes: Mapping[str, Tuple[Any, ...]] = supplied_axes[unit]
                # Require one year label per matrix row.
                if len(tuple(unit_axes.get("years", ()))) != pooling.shape[0]:
                    # Refuse an axis/array row divergence.
                    raise IngestError(
                        f"{pair[0]}/{pair[1]}: unit {unit} year axis diverges "
                        "from its matrix rows."
                    )
                # Read the unit's ordered year labels.
                unit_year_axis: Tuple[Any, ...] = tuple(
                    unit_axes.get("years", ())
                )
                # Establish the common axis from the first labelled unit.
                if common_year_axis is None:
                    # Preserve the exact labels without coercion.
                    common_year_axis = unit_year_axis
                elif unit_year_axis != common_year_axis:
                    # Equal row counts cannot substitute for labelled alignment.
                    raise IngestError(
                        f"{pair[0]}/{pair[1]}: unit {unit} year labels diverge "
                        "from the partition axis."
                    )
                # Compose the exact expected pooling-column labels.
                expected_pooling_columns: Tuple[str, ...] = (
                    "constant",
                    *(f"{name}{LAG_SUFFIX}" for name in CONTROL_COLUMNS),
                    "donor_csa",
                )
                # Require exact column identity and order.
                if tuple(unit_axes.get("pooling_columns", ())) != (
                    expected_pooling_columns
                ):
                    # Wrong order would relabel fitted coefficients.
                    raise IngestError(
                        f"{pair[0]}/{pair[1]}: unit {unit} pooling-column "
                        "labels/order diverge."
                    )
            # Split the design into covariates and factor columns.
            covariates, unit_factor = split_design(pooling)
            # Record the covariate matrix of the unit.
            covariate_by_unit[unit] = covariates
            # Record the dependent matrix of the unit.
            dependent_by_unit[unit] = dependent
            # Assert that the factor matrix is identical across units, which
            # is the property that makes the partition a common-factor design.
            if factor_matrix is None:
                # Record the first unit's factor matrix.
                factor_matrix = unit_factor
            # Refuse a partition whose units carry divergent factor matrices.
            elif not np.allclose(
                factor_matrix,
                unit_factor,
                rtol=IDENTITY_TOLERANCE,
                atol=IDENTITY_TOLERANCE,
            ):
                # The failure names the unit.
                raise IngestError(
                    f"{pair[0]}/{pair[1]}: unit {unit} carries a factor matrix "
                    f"diverging from the partition's."
                )
        # Refuse a partition without a factor matrix.
        if factor_matrix is None:
            # The failure states the precondition.
            raise IngestError(f"{pair[0]}/{pair[1]}: no unit carries a design.")
        # Build the residual-maker of the partition.
        maker: ResidualMaker = build_residual_maker(factor_matrix, condition_max)
        # Independently reconstruct the retained SVD factor projector.
        factor_left, factor_singular, _factor_right = np.linalg.svd(
            factor_matrix, full_matrices=False
        )
        # Apply the same declared rank threshold.
        factor_cutoff: float = float(
            factor_singular[0]
            * max(factor_matrix.shape)
            * np.finfo(float).eps
        )
        # Resolve the independently retained factor rank.
        independent_factor_rank: int = int(
            np.count_nonzero(factor_singular > factor_cutoff)
        )
        # Reconstruct the SVD residual-maker oracle.
        svd_maker: np.ndarray = np.eye(factor_matrix.shape[0]) - (
            factor_left[:, :independent_factor_rank]
            @ factor_left[:, :independent_factor_rank].T
        )
        # Require exact rank-policy agreement.
        if independent_factor_rank != maker.factor_rank:
            # A rank mismatch would invalidate every residualized equation.
            raise IngestError(
                f"{pair[0]}/{pair[1]}: residual-maker factor rank "
                f"{maker.factor_rank} diverges from SVD oracle "
                f"{independent_factor_rank}."
            )
        # Require the constructed maker to match the independent projector.
        if not np.allclose(
            maker.matrix,
            svd_maker,
            rtol=IDENTITY_TOLERANCE,
            atol=IDENTITY_TOLERANCE,
        ):
            # Refuse a route-dependent projection.
            raise IngestError(
                f"{pair[0]}/{pair[1]}: residual-maker diverges from the "
                "independent SVD projection."
            )
        # Estimate the pooled common coefficient.
        beta, diagnostics = estimate_pooled_coefficient(
            dependent_by_unit, covariate_by_unit, maker
        )
        # Refuse a coefficient vector whose length diverges from the declared
        # control count.
        if beta.shape[0] != len(CONTROL_COLUMNS):
            # The failure names the observed length.
            raise IngestError(
                f"{pair[0]}/{pair[1]}: coefficient length {beta.shape[0]} against "
                f"the declared {len(CONTROL_COLUMNS)}."
            )
        # Record the coefficient of the partition.
        betas[pair] = beta
        # Recover the country-specific loadings of the partition.
        # Bound inherited numerical conditioning for the geometric audit.
        orthogonality_conditioning: float = (
            float(diagnostics["design_condition_number"])
            if np.isfinite(float(diagnostics["design_condition_number"]))
            else float(1.0 / np.finfo(float).eps)
        )
        # Recover loadings under the identical tolerance policy.
        loadings[pair] = estimate_unit_loadings(
            factor_matrix,
            dependent_by_unit,
            covariate_by_unit,
            beta,
            tolerance=ORTHOGONALITY_TOLERANCE,
            conditioning=orthogonality_conditioning,
        )
        # Store labelled residualized arrays for independent downstream audit.
        residualized_dependent: Dict[str, np.ndarray] = {
            unit: maker.matrix @ dependent_by_unit[unit]
            for unit in sorted(dependent_by_unit)
        }
        # Store residualized controls under the same unit labels.
        residualized_covariate: Dict[str, np.ndarray] = {
            unit: maker.matrix @ covariate_by_unit[unit]
            for unit in sorted(covariate_by_unit)
        }
        # Preserve the exact matrix-year axis from the factor record when present.
        factor_years: Tuple[Any, ...] = (
            common_year_axis
            if common_year_axis is not None
            else tuple(range(factor_matrix.shape[0]))
        )
        # Register all labelled intermediates verified above.
        intermediates[pair] = {
            "factor_matrix": factor_matrix.copy(),
            "factor_years": factor_years,
            "residual_maker": maker.matrix.copy(),
            "rank_convention": maker.rank_convention,
            "factor_rank": maker.factor_rank,
            "dependent_by_unit": {
                unit: dependent_by_unit[unit].copy()
                for unit in sorted(dependent_by_unit)
            },
            "covariate_by_unit": {
                unit: covariate_by_unit[unit].copy()
                for unit in sorted(covariate_by_unit)
            },
            "residualized_dependent_by_unit": residualized_dependent,
            "residualized_covariate_by_unit": residualized_covariate,
        }
        # Append the estimator ledger row of the partition.
        ledger_rows.append(
            {
                "policy": pair[0],
                "outcome": pair[1],
                "maker_route": maker.route,
                "maker_condition_number": maker.condition_number,
                "fitted_units": diagnostics["units"],
                "fitted_rows": diagnostics["rows"],
                "rank": diagnostics["rank"],
                "design_rank": diagnostics["design_rank"],
                "cross_product_rank": diagnostics["cross_product_rank"],
                "factor_rank": diagnostics["factor_rank"],
                "rank_convention": diagnostics["rank_convention"],
                "design_rank_tolerance": diagnostics[
                    "design_rank_tolerance"
                ],
                "cross_product_rank_tolerance": diagnostics[
                    "cross_product_rank_tolerance"
                ],
                "degrees_of_freedom": diagnostics["degrees_of_freedom"],
                "residualized_design_degrees_of_freedom": diagnostics[
                    "residualized_design_degrees_of_freedom"
                ],
                "residual_sum_of_squares": diagnostics["rss"],
                "stacked_condition_number": diagnostics["condition_number"],
                "design_condition_number": diagnostics[
                    "design_condition_number"
                ],
                "solution_convention_difference": diagnostics[
                    "design_solution_difference_norm"
                ],
            }
        )
        # Record a note when the pseudo-inverse route was taken.
        if maker.route == ROUTE_PINV:
            # Append the route note naming the partition.
            notes.append(
                f"{pair[0]}/{pair[1]}: residual-maker used the pseudo-inverse "
                f"route at condition number {maker.condition_number:.3e}."
            )
    # Assemble the estimator ledger keyed on the pair.
    ledger: pd.DataFrame = pd.DataFrame(ledger_rows).set_index(["policy", "outcome"])
    # Compute the canonical digest of the fitted coefficients and loadings.
    fitted_fingerprint: str = freeze_pooled_fit(betas, loadings)
    # Extend the frozen payload to the labelled verified intermediates.
    fit_payload: List[str] = [f"fitted={fitted_fingerprint}"]
    # Preserve the complete estimator ledger and its diagnostic labels.
    fit_payload.append(
        "ledger="
        + json.dumps(
            ledger.reset_index().to_dict(orient="records"),
            sort_keys=True,
            default=str,
        )
    )
    # Serialize every independently verified partition intermediate.
    for pair in sorted(intermediates):
        # Read the partition's labelled intermediate record.
        record: Mapping[str, Any] = intermediates[pair]
        # Record its pair, years, rank policy and factor objects.
        fit_payload.extend(
            [
                f"pair={pair[0]}|{pair[1]}",
                "years=" + json.dumps(record["factor_years"], default=str),
                f"rank={record['factor_rank']}|{record['rank_convention']}",
                "F=" + _canonical_array_text(record["factor_matrix"]),
                "M=" + _canonical_array_text(record["residual_maker"]),
            ]
        )
        # Record each labelled residualized block.
        for unit in sorted(record["residualized_covariate_by_unit"]):
            # Preserve the unit identity and both residualized matrices.
            fit_payload.append(
                "Ytilde|"
                + unit
                + "|"
                + _canonical_array_text(
                    record["residualized_dependent_by_unit"][unit]
                )
            )
            # Preserve the residualized covariate matrix.
            fit_payload.append(
                "Xtilde|"
                + unit
                + "|"
                + _canonical_array_text(
                    record["residualized_covariate_by_unit"][unit]
                )
            )
    # Freeze the complete labelled fit payload.
    fingerprint: str = hashlib.sha256(
        "\n".join(fit_payload).encode("utf-8")
    ).hexdigest()
    # Record the scope note.
    notes.append(
        f"pre-treatment fit completed for {len(betas)} partition(s) over targets "
        f"and donors pooled together."
    )
    # Assemble the immutable result.
    result: PooledFitResult = PooledFitResult(
        betas=betas,
        beta_names=tuple(CONTROL_COLUMNS),
        loadings=loadings,
        ledger=ledger,
        fingerprint_sha256=fingerprint,
        fitted_utc=current_utc_iso(),
        notes=tuple(notes),
        intermediates=intermediates,
    )
    # Log the completed fit with its digest prefix.
    logger.info(
        "pre-treatment pooled fit complete: %d partition(s), fingerprint %s.",
        len(betas),
        fingerprint[:16],
    )
    # Return the assembled result.
    return result
````

````python
# Task 12: Counterfactual Covariate and Outcome Construction

# ==============================================================================
# Task 12: Counterfactual Covariate and Outcome Construction
# ==============================================================================

# Declare the first projection year, which is the first post-window year.
PROJECTION_START_YEAR: Final[int] = 2014
# Declare the last projection year of the analysis window.
PROJECTION_END_YEAR: Final[int] = 2024
# Declare the permitted input vocabulary of the projection, which is the
# structural expression of the invariant that no observed post-window target
# realization may enter its own counterfactual.
PERMITTED_PROJECTION_INPUTS: Final[Tuple[str, ...]] = (
    "covariate_loadings",
    "pooled_coefficient",
    "country_loadings",
    "projection_factor",
)
# Declare the number of covariate columns of the loading matrix.
LOADING_COLUMNS: Final[int] = len(CONTROL_COLUMNS)


# Declare the immutable result of the counterfactual projection.
@dataclass(frozen=True)
class CounterfactualResult:
    """
    Immutable result of the counterfactual construction.

    Attributes
    ----------
    loadings : dict
        The two-by-eight covariate loading matrices keyed first by pair and
        then by target country code.
    covariate_paths : pandas.DataFrame
        The counterfactual covariates keyed on the pair, the country and the
        year, with one column per control.
    outcome_paths : pandas.DataFrame
        The counterfactual outcomes keyed on the pair, the country and the
        year.
    fitted_pre_paths : pandas.DataFrame
        A distinct figure-data product carrying observed and fitted
        pre-treatment target outcomes. It is not appended to or substituted
        for the post-treatment counterfactual path.
    range_summary : pandas.DataFrame
        One row per pair and control recording the minimum and maximum of the
        simulated covariate over the projection window, emitted without any
        clipping so the extremes are auditable.
    invariant_audit : pandas.DataFrame
        One row per projected target and permitted scientific input, recording
        role, time range, provenance and the executed forbidden-input
        perturbation comparison.
    fingerprint_sha256 : str
        Digest of the canonical counterfactual payload.
    projected_utc : str
        ISO-8601 UTC timestamp of the projection.
    notes : tuple of str
        Recorded decisions, including the projection window and the absence
        of clipping.
    """
    # The loading matrices keyed by pair and target code.
    loadings: Mapping[Tuple[str, str], Mapping[str, np.ndarray]]
    # The counterfactual covariates keyed on the pair, country and year.
    covariate_paths: pd.DataFrame
    # The counterfactual outcomes keyed on the pair, country and year.
    outcome_paths: pd.DataFrame
    # The separate fitted pre-treatment figure-data product.
    fitted_pre_paths: pd.DataFrame
    # The range summary of the simulated covariates.
    range_summary: pd.DataFrame
    # The procedural invariant audit.
    invariant_audit: pd.DataFrame
    # Digest of the canonical counterfactual payload.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the projection.
    projected_utc: str
    # Recorded decisions of the projection.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 1: Estimate the country-specific covariate loadings
# -------------------------------------------------------------------------------------------------------------------------------


# Estimate the covariate loading matrices of a partition's targets.
def estimate_covariate_loadings(
    factor_matrix: np.ndarray,
    covariate_by_target: Mapping[str, np.ndarray],
    targets: Tuple[str, ...],
) -> Dict[str, np.ndarray]:
    """
    Estimate the covariate loading matrices of the targets.

    Parameters
    ----------
    factor_matrix : numpy.ndarray
        The pre-treatment factor matrix of shape years-by-two.
    covariate_by_target : mapping of str to numpy.ndarray
        The pre-treatment covariate matrix of each estimation unit.
    targets : tuple of str
        The target countries of the partition.

    Returns
    -------
    dict
        The two-by-eight loading matrices keyed on the target country code.

    Raises
    ------
    IngestError
        If a target is absent from the covariate mapping, if a target is
        absent from the partition, or if a loading matrix diverges from the
        declared shape.
    """
    # Require a finite nonempty factor matrix.
    if (
        factor_matrix.ndim != 2
        or factor_matrix.shape[0] == 0
        or not np.isfinite(factor_matrix).all()
    ):
        # Loadings are undefined without a finite factor subspace.
        raise IngestError("Covariate-loading factor matrix is empty or non-finite.")
    # Refuse an empty target set, which would make the projection vacuous.
    if not targets:
        # The failure states the precondition.
        raise IngestError("Cannot estimate loadings for an empty target set.")
    # Initialise the loading mapping.
    loadings: Dict[str, np.ndarray] = {}
    # Estimate the loading matrix of every target in sorted order.
    for target in sorted(str(code) for code in targets):
        # Refuse a target whose covariates were not supplied.
        if target not in covariate_by_target:
            # The failure names the missing target.
            raise IngestError(
                f"Target {target} is absent from the covariate mapping."
            )
        # Read the target's pre-treatment covariate matrix.
        covariates: np.ndarray = covariate_by_target[target]
        # Refuse a covariate matrix whose column count diverges.
        if covariates.ndim != 2 or covariates.shape[1] != LOADING_COLUMNS:
            # The failure names the target and the observed shape.
            raise IngestError(
                f"{target}: covariate shape {covariates.shape} against the "
                f"declared (rows, {LOADING_COLUMNS})."
            )
        # Require row alignment and finite pre-treatment target covariates.
        if (
            covariates.shape[0] != factor_matrix.shape[0]
            or not np.isfinite(covariates).all()
        ):
            # Refuse a loading fit with divergent axes or missing values.
            raise IngestError(
                f"{target}: covariates are non-finite or row-misaligned with "
                "the pre-treatment factor."
            )
        # Project the covariate matrix on the factor matrix through the
        # least-squares route, which handles rank deficiency uniformly.
        loading, _residuals, _rank, _singular_values = np.linalg.lstsq(
            factor_matrix, covariates, rcond=None
        )
        # Assert the declared loading shape of one component row per factor
        # column and one column per control, which adapts to the factor's own
        # width so that a wider factor is admitted without a shape change.
        if loading.shape != (factor_matrix.shape[1], LOADING_COLUMNS):
            # The failure names the target and the observed shape.
            raise IngestError(
                f"{target}: loading shape {loading.shape} against the declared "
                f"{(factor_matrix.shape[1], LOADING_COLUMNS)}."
            )
        # Store the loading matrix keyed on the target code.
        loadings[target] = loading
    # Log the loading estimation outcome.
    logger.info(
        "covariate loadings estimated for %d target(s) at shape %s.",
        len(loadings),
        (len(FACTOR_COMPONENTS), LOADING_COLUMNS),
    )
    # Return the loading mapping.
    return loadings


# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 2: Project the counterfactual covariates and outcomes
# -------------------------------------------------------------------------------------------------------------------------------


# Build the projection factor matrix from the partition's proxy series.
def build_projection_factor(
    proxy: pd.Series,
    start_year: int = PROJECTION_START_YEAR,
    end_year: int = PROJECTION_END_YEAR,
) -> Tuple[np.ndarray, List[int]]:
    """
    Build the post-window factor matrix from the donor proxy.

    Parameters
    ----------
    proxy : pandas.Series
        The donor cross-sectional proxy indexed by year, which covers the
        projection window by construction.
    start_year : int, optional
        The first projection year.
    end_year : int, optional
        The last projection year.

    Returns
    -------
    tuple
        The factor matrix of shape years-by-two, ordered as the constant and
        the donor average, and the list of projection years it covers.

    Raises
    ------
    IngestError
        If the window is not ordered signed-64-bit integer state, if a proxy
        year is not exactly integral or collides after normalization, or if
        the proxy lacks a finite donor mean for any projection year.
    """
    # Require the public Series contract before inspecting its labelled axis.
    if not isinstance(proxy, pd.Series) or proxy.empty:
        raise IngestError("Donor proxy must be a nonempty labelled Series.")
    # Validate the inclusive projection window strictly.
    if (
        isinstance(start_year, (bool, np.bool_))
        or isinstance(end_year, (bool, np.bool_))
        or not isinstance(start_year, (int, np.integer))
        or not isinstance(end_year, (int, np.integer))
    ):
        # Refuse lossy or reversed calendar windows.
        raise IngestError(
            f"Invalid projection-factor window {start_year!r} to {end_year!r}."
        )
    canonical_start = _exact_calendar_year(
        start_year, context="Projection-factor start year"
    )
    canonical_end = _exact_calendar_year(
        end_year, context="Projection-factor end year"
    )
    if canonical_start > canonical_end:
        raise IngestError(
            f"Invalid projection-factor window {start_year!r} to {end_year!r}."
        )
    # Parse proxy index labels exactly, so high-precision decimal strings can
    # neither round to an integer nor collide silently during construction.
    proxy_index: pd.Series = pd.Series(list(proxy.index), dtype="object")
    canonical_values = _exact_calendar_axis(
        proxy_index.tolist(), context="Donor proxy year"
    )
    # Convert only after validation, then reject collisions such as the two
    # distinct source labels ``'2014'`` and ``'2014.0'`` mapping to one year.
    canonical_years: pd.Series = pd.Series(canonical_values, dtype="int64")
    duplicate_canonical: pd.Series = canonical_years.duplicated(keep=False)
    if bool(duplicate_canonical.any()):
        offenders: List[Any] = proxy_index.loc[duplicate_canonical].tolist()
        raise IngestError(
            "Donor proxy year labels collide after lossless normalization: "
            f"{offenders[:MAX_REPORTED_OFFENDERS]}."
        )
    # Parse every donor mean, not merely requested rows, so an invalid factor
    # object cannot be partially rehabilitated by selecting a narrow window.
    if any(isinstance(value, (bool, np.bool_)) for value in proxy.to_numpy()):
        raise IngestError("Donor proxy carries a Boolean donor mean.")
    try:
        numeric_proxy: pd.Series = pd.to_numeric(
            pd.Series(proxy.to_numpy(), index=canonical_years.to_numpy()),
            errors="coerce",
        )
    except (TypeError, ValueError, OverflowError) as exc:
        raise IngestError("Donor proxy carries nonnumeric donor means.") from exc
    if numeric_proxy.isna().any() or not np.isfinite(
        numeric_proxy.to_numpy(dtype=float)
    ).all():
        raise IngestError("Donor proxy carries non-finite donor means.")
    # Canonically order the validated factor by integral calendar year.
    canonical_proxy: pd.Series = numeric_proxy.sort_index(kind="stable")
    # Compose the projection years as an ascending list.
    years: List[int] = list(range(canonical_start, canonical_end + 1))
    # Identify the projection years the proxy does not carry.
    missing_years: List[int] = [
        year for year in years if year not in canonical_proxy.index
    ]
    # Refuse a proxy that does not cover the whole projection window.
    if missing_years:
        # The failure names the missing years.
        raise IngestError(
            f"Donor proxy does not cover the projection year(s) {missing_years}."
        )
    # Extract the donor average for every projection year.
    donor_average: np.ndarray = np.array(
        [float(canonical_proxy.loc[year]) for year in years], dtype=float
    )
    # Require finite donor factor values throughout projection.
    if not np.isfinite(donor_average).all():
        # A missing/infinite donor mean cannot support projection.
        raise IngestError("Donor proxy carries non-finite projection values.")
    # Compose the constant column.
    constant: np.ndarray = np.ones((len(years), 1), dtype=float)
    # Compose the factor matrix ordered as the constant and the donor average.
    factor_matrix: np.ndarray = np.hstack(
        [constant, donor_average.reshape(-1, 1)]
    )
    # Assert the declared factor shape.
    if factor_matrix.shape != (len(years), len(FACTOR_COMPONENTS)):
        # The failure names the observed shape.
        raise IngestError(
            f"Projection factor shape {factor_matrix.shape} against the declared "
            f"{(len(years), len(FACTOR_COMPONENTS))}."
        )
    # Log the projection factor construction.
    logger.info(
        "projection factor built for %d year(s) from the donor proxy.", len(years)
    )
    # Return the factor matrix and its years.
    return factor_matrix, years


# Project the counterfactual covariates and outcome of one target.
def project_target_paths(
    loading: np.ndarray,
    beta: np.ndarray,
    alpha: np.ndarray,
    projection_factor: np.ndarray,
) -> Tuple[np.ndarray, np.ndarray]:
    """
    Project one target's counterfactual covariates and outcome.

    Parameters
    ----------
    loading : numpy.ndarray
        The target's two-by-eight covariate loading matrix.
    beta : numpy.ndarray
        The pooled common coefficient vector of the partition.
    alpha : numpy.ndarray
        The target's two-component country loading vector.
    projection_factor : numpy.ndarray
        The post-window factor matrix of shape years-by-two.

    Returns
    -------
    tuple
        The counterfactual covariate matrix of shape years-by-eight and the
        counterfactual outcome vector of shape years.

    Raises
    ------
    IngestError
        If any dimension diverges from the declared design, because the two
        projection equations are only defined at their declared dimensions.
    """
    # Require all scientific inputs to be finite arrays.
    if any(
        not np.isfinite(np.asarray(array, dtype=float)).all()
        for array in (loading, beta, alpha, projection_factor)
    ):
        # Reject numerical contamination before matrix multiplication.
        raise IngestError("Projection inputs contain non-finite values.")
    # Assert the loading matrix's declared shape, whose component row count
    # follows the factor matrix's own width so that a wider factor is admitted.
    if loading.shape != (projection_factor.shape[1], LOADING_COLUMNS):
        # The failure names the observed shape.
        raise IngestError(
            f"Loading shape {loading.shape} against the declared "
            f"{(projection_factor.shape[1], LOADING_COLUMNS)}."
        )
    # Assert the coefficient vector's declared length.
    if beta.shape[0] != LOADING_COLUMNS:
        # The failure names the observed length.
        raise IngestError(
            f"Coefficient length {beta.shape[0]} against the declared "
            f"{LOADING_COLUMNS}."
        )
    # Assert the country loading vector's declared length, which follows the
    # factor matrix's own width.
    if alpha.shape[0] != projection_factor.shape[1]:
        # The failure names the observed length.
        raise IngestError(
            f"Country loading length {alpha.shape[0]} against the declared "
            f"{projection_factor.shape[1]}."
        )
    # Project the covariate path as the factor matrix against the loading
    # matrix, which realises the declared covariate equation for every year.
    covariate_path: np.ndarray = projection_factor @ loading
    # Project the outcome path as the covariate path against the pooled
    # coefficient plus the factor against the country loading, which realises
    # the declared outcome equation for every year and contains no term drawn
    # from any observed target realization.
    outcome_path: np.ndarray = (
        covariate_path @ beta.reshape(-1, 1) + projection_factor @ alpha.reshape(-1, 1)
    )
    # Assert the declared covariate path shape.
    if covariate_path.shape != (projection_factor.shape[0], LOADING_COLUMNS):
        # The failure names the observed shape.
        raise IngestError(
            f"Covariate path shape {covariate_path.shape} against the declared "
            f"{(projection_factor.shape[0], LOADING_COLUMNS)}."
        )
    # Assert the declared outcome path shape.
    if outcome_path.shape != (projection_factor.shape[0], 1):
        # The failure names the observed shape.
        raise IngestError(
            f"Outcome path shape {outcome_path.shape} against the declared "
            f"{(projection_factor.shape[0], 1)}."
        )
    # Return the two paths.
    return covariate_path, outcome_path.ravel()


# Assert that a projected path carries no missing cell.
def assert_paths_complete(
    covariate_path: np.ndarray, outcome_path: np.ndarray, target: str
) -> None:
    """
    Assert that a projected path carries no missing cell.

    Parameters
    ----------
    covariate_path : numpy.ndarray
        The target's counterfactual covariate matrix.
    outcome_path : numpy.ndarray
        The target's counterfactual outcome vector.
    target : str
        The target country code, used in the failure message.

    Returns
    -------
    None
        The function returns nothing when both paths are complete.

    Raises
    ------
    IngestError
        If either path carries a non-finite value, which would indicate a
        numerical failure rather than a data problem.
    """
    # Count the non-finite covariate cells.
    non_finite_covariates: int = int((~np.isfinite(covariate_path)).sum())
    # Count the non-finite outcome cells.
    non_finite_outcomes: int = int((~np.isfinite(outcome_path)).sum())
    # Refuse a path carrying a non-finite value.
    if non_finite_covariates or non_finite_outcomes:
        # The failure names the target and both counts.
        raise IngestError(
            f"{target}: simulated path carries {non_finite_covariates} non-finite "
            f"covariate cell(s) and {non_finite_outcomes} non-finite outcome "
            f"cell(s)."
        )


# Summarise the ranges of the simulated covariates for the audit.
def summarize_path_ranges(
    paths: pd.DataFrame,
) -> pd.DataFrame:
    """
    Summarise the ranges of the simulated covariates.

    Parameters
    ----------
    paths : pandas.DataFrame
        The counterfactual covariate paths keyed on the pair, the country and
        the year.

    Returns
    -------
    pandas.DataFrame
        One row per pair and control recording the minimum, the maximum and
        the number of negative cells, which are emitted without any clipping
        so the extremes are auditable.

    Raises
    ------
    IngestError
        If the paths frame is empty or lacks a declared column.
    """
    # Refuse an empty paths frame.
    if len(paths) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot summarise ranges of an empty path frame.")
    # Accumulate the summary rows.
    rows: List[Dict[str, Any]] = []
    # Group the paths by partition pair.
    for (policy, outcome), group in paths.groupby(["policy", "outcome"], observed=True):
        # Summarise each control column of the pair.
        for control in CONTROL_COLUMNS:
            # Refuse a missing control column.
            if control not in group.columns:
                # The failure names the absent column.
                raise IngestError(f"Path frame lacks the control column {control!r}.")
            # Record the control's range and its negative-cell count.
            rows.append(
                {
                    "policy": policy,
                    "outcome": outcome,
                    "control": control,
                    "minimum": float(group[control].min()),
                    "maximum": float(group[control].max()),
                    "negative_cells": int((group[control] < 0.0).sum()),
                }
            )
    # Assemble the summary frame.
    summary: pd.DataFrame = pd.DataFrame(rows)
    # Log the summary extent.
    logger.info(
        "simulated covariate ranges summarised across %d pair-control row(s) "
        "without clipping.",
        len(summary),
    )
    # Return the summary frame.
    return summary


# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 3: Verify the structural invariant
# -------------------------------------------------------------------------------------------------------------------------------


# Record the procedural audit of the structural invariant.
def record_invariant_audit(
    policy: str,
    outcome: str,
    targets: Tuple[str, ...],
    declared_inputs: Tuple[str, ...],
    permitted: Tuple[str, ...] = PERMITTED_PROJECTION_INPUTS,
    *,
    input_provenance: Optional[Mapping[str, Mapping[str, Any]]] = None,
    baseline_covariates: Optional[Mapping[str, np.ndarray]] = None,
    baseline_outcomes: Optional[Mapping[str, np.ndarray]] = None,
    perturbed_covariates: Optional[Mapping[str, np.ndarray]] = None,
    perturbed_outcomes: Optional[Mapping[str, np.ndarray]] = None,
    forbidden_input_digests: Optional[
        Mapping[str, Tuple[str, str]]
    ] = None,
    perturbation_cells: int = 0,
    perturbation_source: str = "",
) -> pd.DataFrame:
    """
    Record the procedural audit of the structural invariant.

    Parameters
    ----------
    policy : str
        The policy instrument of the pair.
    outcome : str
        The canonical outcome variable of the pair.
    targets : tuple of str
        The projected target countries.
    declared_inputs : tuple of str
        The input names declared at the projection call site.
    permitted : tuple of str, optional
        The permitted input vocabulary, which contains the loading matrices,
        the pooled coefficient, the country loadings and the factor
        projection, and nothing drawn from an observed target realization.
    input_provenance : mapping, optional
        Per-input role, time-range and source provenance.
    baseline_covariates, baseline_outcomes : mapping, optional
        Projected arrays from the unperturbed execution boundary.
    perturbed_covariates, perturbed_outcomes : mapping, optional
        Projected arrays after independently perturbing forbidden realized
        post-treatment target data while holding permitted inputs fixed.
    forbidden_input_digests : mapping, optional
        Baseline and perturbed digests of the forbidden target data consumed
        by the audit boundary, keyed by target.
    perturbation_cells : int, optional
        Number of forbidden realized cells changed by the executed audit.
    perturbation_source : str, optional
        Provenance of the realized or sentinel target-post audit data.

    Returns
    -------
    pandas.DataFrame
        One uniquely keyed row per target and permitted input, recording role,
        time range, provenance, executed perturbation digests and the invariant
        comparison outcome.

    Raises
    ------
    IngestError
        If a declared input name lies outside the permitted vocabulary, which
        would mean an observed post-window quantity had been routed onto the
        counterfactual path; or if the input list carries a name suggesting an
        observed outcome residual.
    """
    # Require unique nonempty target identities.
    target_codes: Tuple[str, ...] = tuple(str(code) for code in targets)
    # Refuse a vacuous or duplicated target audit.
    if not target_codes or len(target_codes) != len(set(target_codes)):
        # Every executed comparison requires a unique target key.
        raise IngestError("Invariant audit targets are empty or duplicated.")
    # Require the complete permitted vocabulary exactly once.
    if (
        len(declared_inputs) != len(set(declared_inputs))
        or set(declared_inputs) != set(permitted)
    ):
        # Omitting an input is as misleading as declaring a forbidden one.
        raise IngestError(
            "Projection input declaration must equal the complete permitted "
            f"vocabulary exactly; declared {list(declared_inputs)}, permitted "
            f"{list(permitted)}."
        )
    # Identify the declared inputs outside the permitted vocabulary.
    undeclared: List[str] = sorted(set(declared_inputs) - set(permitted))
    # Refuse an undeclared input, which is the invariant violation the audit
    # exists to catch.
    if undeclared:
        # The failure names the offending inputs and the permitted vocabulary.
        raise IngestError(
            f"Projection declares input(s) outside the permitted vocabulary: "
            f"{undeclared}; permitted: {list(permitted)}."
        )
    # Refuse a declared input naming an observed residual, which is the
    # specification's named violation mode.
    forbidden_substrings: Tuple[str, ...] = ("residual", "observed", "realized")
    # Identify a suspicious input name.
    suspicious: List[str] = [
        name
        for name in declared_inputs
        if any(token in name.lower() for token in forbidden_substrings)
    ]
    # Raise when a suspicious name appears.
    if suspicious:
        # The failure names the suspicious inputs.
        raise IngestError(
            f"Projection declares input(s) implying an observed post-window "
            f"realization: {suspicious}."
        )
    # Require all evidence needed for an executed dataflow comparison.
    evidence_objects: Tuple[Any, ...] = (
        input_provenance,
        baseline_covariates,
        baseline_outcomes,
        perturbed_covariates,
        perturbed_outcomes,
        forbidden_input_digests,
    )
    # Refuse a signature-only audit with no executed path evidence.
    if any(item is None for item in evidence_objects):
        # Never label procedural inspection as a dataflow proof.
        raise IngestError(
            "Invariant audit requires executed baseline and perturbed paths, "
            "input provenance and forbidden-input digests."
        )
    # Bind validated optional mappings for static type narrowing.
    provenance_map = input_provenance or {}
    # Require provenance for every permitted scientific input.
    if set(provenance_map) != set(permitted):
        # Name the provenance coverage divergence.
        raise IngestError(
            "Projection input provenance does not cover the permitted "
            f"vocabulary exactly: {sorted(provenance_map)}."
        )
    # Validate required provenance fields per input.
    provenance_fields: set[str] = {
        "country_role",
        "time_start",
        "time_end",
        "provenance",
    }
    # Inspect every declared input's provenance record.
    for input_name in permitted:
        # Identify missing semantic fields.
        missing_fields: set[str] = provenance_fields - set(
            provenance_map[input_name]
        )
        # Refuse incomplete provenance.
        if missing_fields:
            # Name the affected input and fields.
            raise IngestError(
                f"{input_name}: projection provenance omits "
                f"{sorted(missing_fields)}."
            )
    # Bind the executed array mappings.
    baseline_covariate_map = baseline_covariates or {}
    # Bind the baseline outcome mapping.
    baseline_outcome_map = baseline_outcomes or {}
    # Bind the perturbed covariate mapping.
    perturbed_covariate_map = perturbed_covariates or {}
    # Bind the perturbed outcome mapping.
    perturbed_outcome_map = perturbed_outcomes or {}
    # Bind the forbidden-input digest mapping.
    forbidden_digest_map = forbidden_input_digests or {}
    # Require a genuine material perturbation count.
    if (
        isinstance(perturbation_cells, (bool, np.bool_))
        or int(perturbation_cells) <= 0
    ):
        # A zero-cell comparison is not an executed perturbation.
        raise IngestError("Invariant audit did not perturb any forbidden cell.")
    # Accumulate the audit rows.
    rows: List[Dict[str, Any]] = []
    # Record one set of executed comparisons per projected target.
    for target in sorted(target_codes):
        # Require every target in every baseline and perturbation mapping.
        if any(
            target not in mapping
            for mapping in (
                baseline_covariate_map,
                baseline_outcome_map,
                perturbed_covariate_map,
                perturbed_outcome_map,
                forbidden_digest_map,
            )
        ):
            # Name the target with incomplete audit evidence.
            raise IngestError(
                f"{policy}/{outcome}/{target}: executed audit evidence is "
                "incomplete."
            )
        # Read the baseline and perturbed forbidden-data digests.
        baseline_forbidden_digest, perturbed_forbidden_digest = (
            forbidden_digest_map[target]
        )
        # Prove that the target-post perturbation reached the audit boundary.
        if baseline_forbidden_digest == perturbed_forbidden_digest:
            # Identical input lineage cannot establish perturbation invariance.
            raise IngestError(
                f"{policy}/{outcome}/{target}: forbidden target-post "
                "perturbation was not consumed by the execution boundary."
            )
        # Read each emitted baseline and perturbed array.
        baseline_x: np.ndarray = np.asarray(
            baseline_covariate_map[target], dtype=float
        )
        # Read the baseline outcome path.
        baseline_y: np.ndarray = np.asarray(
            baseline_outcome_map[target], dtype=float
        )
        # Read the perturbed covariate path.
        perturbed_x: np.ndarray = np.asarray(
            perturbed_covariate_map[target], dtype=float
        )
        # Read the perturbed outcome path.
        perturbed_y: np.ndarray = np.asarray(
            perturbed_outcome_map[target], dtype=float
        )
        # Require exact shape and value invariance under forbidden perturbation.
        if (
            baseline_x.shape != perturbed_x.shape
            or baseline_y.shape != perturbed_y.shape
            or not np.array_equal(baseline_x, perturbed_x)
            or not np.array_equal(baseline_y, perturbed_y)
        ):
            # A changed projection proves post-treatment target leakage.
            raise IngestError(
                f"{policy}/{outcome}/{target}: projected paths changed after "
                "perturbing forbidden realized target-post data."
            )
        # Freeze the verified output arrays for audit lineage.
        baseline_path_digest: str = hashlib.sha256(
            baseline_x.tobytes() + baseline_y.tobytes()
        ).hexdigest()
        # Freeze the independently re-executed perturbed arrays.
        perturbed_path_digest: str = hashlib.sha256(
            perturbed_x.tobytes() + perturbed_y.tobytes()
        ).hexdigest()
        # Emit one semantic row for each permitted projection input.
        for input_name in permitted:
            # Read the input's role/time/source record.
            provenance = provenance_map[input_name]
            # Append the uniquely keyed audit record.
            rows.append(
                {
                    "policy": policy,
                    "outcome": outcome,
                    "country_code": target,
                    "input_name": input_name,
                    "country_role": str(provenance["country_role"]),
                    "time_start": int(provenance["time_start"]),
                    "time_end": int(provenance["time_end"]),
                    "provenance": str(provenance["provenance"]),
                    "baseline_forbidden_digest": baseline_forbidden_digest,
                    "perturbed_forbidden_digest": perturbed_forbidden_digest,
                    "baseline_path_digest": baseline_path_digest,
                    "perturbed_path_digest": perturbed_path_digest,
                    "perturbation_cells": int(perturbation_cells),
                    "perturbation_source": str(perturbation_source),
                    "hidden_dependency_status": (
                        "forbidden target-post channel consumed for lineage "
                        "only; excluded from scientific inputs"
                    ),
                    "audit_outcome": "invariant_verified",
                }
            )
    # Assemble the audit frame.
    audit: pd.DataFrame = pd.DataFrame(rows)
    # Log the recorded audit.
    logger.info(
        "%s/%s: executed invariant audit recorded for %d target-input row(s).",
        policy,
        outcome,
        len(audit),
    )
    # Return the audit frame.
    return audit


# Freeze the counterfactual objects behind a canonical digest.
def freeze_counterfactual(
    loadings: Mapping[Tuple[str, str], Mapping[str, np.ndarray]],
    covariate_paths: pd.DataFrame,
    outcome_paths: pd.DataFrame,
    fitted_pre_paths: Optional[pd.DataFrame] = None,
    invariant_audit: Optional[pd.DataFrame] = None,
) -> str:
    """
    Compute the canonical digest of the counterfactual objects.

    Parameters
    ----------
    loadings : mapping of tuple to mapping of str to numpy.ndarray
        The loading matrices keyed by pair and target code.
    covariate_paths : pandas.DataFrame
        The counterfactual covariate paths.
    outcome_paths : pandas.DataFrame
        The counterfactual outcome paths.
    fitted_pre_paths : pandas.DataFrame, optional
        The distinct fitted pre-treatment figure-data paths.
    invariant_audit : pandas.DataFrame, optional
        The executed forbidden-input perturbation audit.

    Returns
    -------
    str
        The hexadecimal digest of the canonical payload.

    Raises
    ------
    IngestError
        If a payload component is empty.
    """
    # Refuse an empty loading mapping.
    if not loadings:
        # The failure states the precondition.
        raise IngestError("Cannot freeze an empty loading mapping.")
    # Refuse empty path frames.
    if len(covariate_paths) == 0 or len(outcome_paths) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot freeze empty path frames.")
    # Accumulate the canonical payload parts.
    parts: List[str] = []
    # Record every loading matrix in sorted pair and target order.
    for pair in sorted(loadings.keys()):
        # Iterate over the pair's targets in sorted order.
        for target in sorted(loadings[pair].keys()):
            # Append the loading values at full precision.
            parts.append(
                "L|"
                + pair[0]
                + "|"
                + pair[1]
                + "|"
                + target
                + "|"
                + "|".join(
                    f"{float(value):.17g}" for value in loadings[pair][target].ravel()
                )
            )
    # Record every covariate path cell in sorted key order.
    for _, row in covariate_paths.sort_values(
        ["policy", "outcome", "country_code", "year"], kind="stable"
    ).iterrows():
        # Append the covariate cell values at full precision.
        parts.append(
            "X|"
            + str(row["policy"])
            + "|"
            + str(row["outcome"])
            + "|"
            + str(row["country_code"])
            + "|"
            + str(int(row["year"]))
            + "|"
            + "|".join(f"{float(row[control]):.17g}" for control in CONTROL_COLUMNS)
        )
    # Record every outcome path cell in sorted key order.
    for _, row in outcome_paths.sort_values(
        ["policy", "outcome", "country_code", "year"], kind="stable"
    ).iterrows():
        # Append the outcome cell value at full precision.
        parts.append(
            "Y|"
            + str(row["policy"])
            + "|"
            + str(row["outcome"])
            + "|"
            + str(row["country_code"])
            + "|"
            + str(int(row["year"]))
            + "|"
            + f"{float(row['counterfactual_outcome']):.17g}"
        )
    # Record the separate fitted pre-period figure paths when supplied.
    if fitted_pre_paths is not None:
        # Refuse an empty supplied figure product.
        if len(fitted_pre_paths) == 0:
            # An empty explicit product is incomplete provenance.
            raise IngestError("Cannot freeze empty fitted pre-period paths.")
        # Serialize every fitted-pre row in exact key order.
        for _, row in fitted_pre_paths.sort_values(
            ["policy", "outcome", "country_code", "year"], kind="stable"
        ).iterrows():
            # Record observed, fitted and residual values distinctly.
            parts.append(
                "PRE|"
                + str(row["policy"])
                + "|"
                + str(row["outcome"])
                + "|"
                + str(row["country_code"])
                + "|"
                + str(int(row["year"]))
                + "|"
                + f"{float(row['observed_pre_outcome']):.17g}|"
                + f"{float(row['fitted_pre_outcome']):.17g}|"
                + f"{float(row['pre_fit_residual']):.17g}"
            )
    # Record the complete executed leakage audit when supplied.
    if invariant_audit is not None:
        # Refuse an empty audit masquerading as verification.
        if len(invariant_audit) == 0:
            # Every projection requires executed audit evidence.
            raise IngestError("Cannot freeze an empty invariant audit.")
        # Serialize all labelled audit fields deterministically.
        parts.append(
            "AUDIT|"
            + invariant_audit.sort_values(
                ["policy", "outcome", "country_code", "input_name"],
                kind="stable",
            ).to_json(orient="split", double_precision=15)
        )
    # Digest the encoded payload.
    return hashlib.sha256("\n".join(parts).encode("utf-8")).hexdigest()


# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the counterfactual construction across the partitions.
def orchestrate_counterfactual_projection(
    fit_result: PooledFitResult,
    factor_result: DonorFactorResult,
    registry: PartitionRegistry,
    start_year: int = PROJECTION_START_YEAR,
    end_year: int = PROJECTION_END_YEAR,
    observed_panel: Optional[pd.DataFrame] = None,
) -> CounterfactualResult:
    """
    Project the counterfactual paths of every partition's targets.

    Parameters
    ----------
    fit_result : PooledFitResult
        The pre-treatment pooled fit.
    factor_result : DonorFactorResult
        The donor factor objects, whose proxies supply the post-window factor.
    registry : PartitionRegistry
        The registered partitions, whose target tuples define the projected
        units.
    start_year : int, optional
        The first projection year.
    end_year : int, optional
        The last projection year.
    observed_panel : pandas.DataFrame, optional
        The realized panel used only by the executed leakage audit. Its
        post-treatment target outcomes and covariates are independently
        perturbed and passed through the projection boundary, but remain
        excluded from the permitted scientific inputs. When absent, a
        synthetic sentinel channel is used and the audit records that the
        realized dependency was inaccessible.

    Returns
    -------
    CounterfactualResult
        The loading matrices, post-treatment covariate and outcome paths, the
        distinct fitted pre-treatment figure-data product, the range summary,
        the executed invariant audit, the digest, the timestamp and the notes.

    Raises
    ------
    IngestError
        If a partition is absent from the fit or the factor result, or if any
        projection, shape or audit check fails.
    """
    # Validate the projection window without accepting boolean years.
    if (
        isinstance(start_year, (bool, np.bool_))
        or isinstance(end_year, (bool, np.bool_))
        or not isinstance(start_year, (int, np.integer))
        or not isinstance(end_year, (int, np.integer))
        or int(start_year) > int(end_year)
    ):
        # An invalid window cannot define a labelled projection.
        raise IngestError(
            f"Invalid projection window {start_year!r} to {end_year!r}."
        )
    # Refuse an empty partition registry before constructing output frames.
    if not registry.partitions:
        # A vacuous projection cannot support paths or a leakage audit.
        raise IngestError("Counterfactual projection registry is empty.")
    # Bind coefficient order to the eight controls used by every loading
    # matrix; equal-length but permuted names would otherwise be undetectable.
    if tuple(str(name) for name in fit_result.beta_names) != tuple(
        CONTROL_COLUMNS
    ):
        raise IngestError(
            "Pooled coefficient names are not aligned with the canonical "
            "control order."
        )
    # Initialise the loading mapping keyed on the pair.
    loadings: Dict[Tuple[str, str], Mapping[str, np.ndarray]] = {}
    # Accumulate the covariate path rows.
    covariate_rows: List[Dict[str, Any]] = []
    # Accumulate the outcome path rows.
    outcome_rows: List[Dict[str, Any]] = []
    # Accumulate the distinct fitted pre-treatment figure-data rows.
    fitted_pre_rows: List[Dict[str, Any]] = []
    # Accumulate the audit frames.
    audits: List[pd.DataFrame] = []
    # Retain the permitted projection context for executed leakage checks.
    audit_contexts: Dict[Tuple[str, str], Mapping[str, Any]] = {}
    # Accumulate the recorded notes.
    notes: List[str] = []
    # Iterate over the registry's partitions in sorted pair order.
    for pair in sorted(registry.partitions.keys()):
        # Refuse a partition the fit does not cover.
        if pair not in fit_result.betas or pair not in fit_result.loadings:
            # The failure names the missing pair.
            raise IngestError(f"Fit result does not cover the pair {pair}.")
        # Refuse a partition the factor result does not cover.
        if pair not in factor_result.factors:
            # The failure names the missing pair.
            raise IngestError(f"Factor result does not cover the pair {pair}.")
        # Select the partition of the registry.
        partition = registry.partitions[pair]
        # Select the partition's pre-treatment factor object.
        factor = factor_result.factors[pair]
        # Bind the nested factor identity and donor count to the partition.
        if (
            str(factor.policy) != str(pair[0])
            or str(factor.outcome) != str(pair[1])
            or isinstance(factor.donor_count, (bool, np.bool_))
            or int(factor.donor_count) != len(partition.donors)
            or int(factor.donor_count) <= 0
        ):
            raise IngestError(
                f"{pair[0]}/{pair[1]}: factor identity or donor count "
                "diverges from the registered partition."
            )
        # Recompute the exact Task-10 roster digest carried by every matrix
        # diagnostic, proving that the proxy was registered to these donors.
        expected_roster_hash: str = hashlib.sha256(
            json.dumps(
                {
                    "donors": sorted(str(code) for code in partition.donors),
                    "units": sorted(str(code) for code in factor.matrices),
                },
                sort_keys=True,
            ).encode("utf-8")
        ).hexdigest()
        if (
            not isinstance(factor.diagnostics, pd.DataFrame)
            or factor.diagnostics.empty
            or "roster_hash" not in factor.diagnostics.columns
            or set(factor.diagnostics["roster_hash"].astype(str))
            != {expected_roster_hash}
        ):
            raise IngestError(
                f"{pair[0]}/{pair[1]}: factor diagnostics are not bound to "
                "the registered donor and estimation-unit rosters."
            )
        # Initialise the per-unit covariate mapping from the pre-treatment
        # matrices, which are the only covariates the projection may consume.
        covariate_by_unit: Dict[str, np.ndarray] = {}
        # Initialise the partition's pre-treatment factor matrix.
        factor_matrix: Optional[np.ndarray] = None
        # Split each unit's pre-treatment design.
        for unit in sorted(factor.matrices.keys()):
            # Read the unit's dependent and design matrices.
            _dependent, pooling = factor.matrices[unit]
            # Require explicit, exact pre-treatment axes for each scientific
            # matrix rather than inferring labels from row position.
            axes = factor.matrix_axes.get(str(unit))
            expected_pre_years: Tuple[int, ...] = tuple(
                range(
                    int(PRE_TREATMENT_WINDOW[0]),
                    int(PRE_TREATMENT_WINDOW[1]) + 1,
                )
            )
            expected_pooling_columns: Tuple[str, ...] = (
                "constant",
                *(f"{name}{LAG_SUFFIX}" for name in CONTROL_COLUMNS),
                "donor_csa",
            )
            if (
                not isinstance(axes, Mapping)
                or tuple(axes.get("years", ())) != expected_pre_years
                or tuple(axes.get("dependent_columns", ())) != (pair[1],)
                or tuple(axes.get("pooling_columns", ()))
                != expected_pooling_columns
            ):
                raise IngestError(
                    f"{pair[0]}/{pair[1]}/{unit}: factor matrix axes do not "
                    "match the labelled pre-2014 design."
                )
            # Enforce the baseline intercept/eight-controls/one-factor schema.
            if (
                pooling.ndim != 2
                or pooling.shape[1] != POOLING_COLUMNS
                or _dependent.ndim != 2
                or _dependent.shape != (pooling.shape[0], 1)
                or not np.isfinite(pooling).all()
                or not np.isfinite(_dependent).all()
            ):
                # Richer-factor branches belong to their separate interface.
                raise IngestError(
                    f"{pair[0]}/{pair[1]}/{unit}: baseline design is malformed "
                    "or does not carry exactly one donor-factor column."
                )
            # Extract the covariate columns of the design, which follow the
            # intercept and precede the donor average.
            covariates: np.ndarray = pooling[
                :, 1 : 1 + len(CONTROL_COLUMNS)
            ]
            # Record the unit's pre-treatment covariate matrix.
            covariate_by_unit[str(unit)] = covariates
            # Record the factor matrix from its first occurrence.
            if factor_matrix is None:
                # Compose the factor matrix from the intercept and the proxy.
                factor_matrix = np.column_stack(
                    [pooling[:, 0], pooling[:, -1]]
                )
            else:
                # Compose this unit's factor matrix for exact comparison.
                unit_factor: np.ndarray = np.column_stack(
                    [pooling[:, 0], pooling[:, -1]]
                )
                # Require one common labelled factor across all units.
                if not np.array_equal(factor_matrix, unit_factor):
                    # Divergent factors would invalidate pooled loadings.
                    raise IngestError(
                        f"{pair[0]}/{pair[1]}/{unit}: baseline factor diverges "
                        "from the partition factor."
                    )
        # Refuse a partition without a factor matrix.
        if factor_matrix is None:
            # The failure states the precondition.
            raise IngestError(f"{pair[0]}/{pair[1]}: no unit carries a design.")
        # Enforce the baseline's declared two-column factor contract.
        if factor_matrix.shape[1] != FACTOR_COLUMNS:
            # Richer-factor variants must use their separate sweep interface.
            raise IngestError(
                f"{pair[0]}/{pair[1]}: baseline pre-factor has "
                f"{factor_matrix.shape[1]} columns; expected {FACTOR_COLUMNS}."
            )
        # Rebuild the labelled pre-period factor from the donor-only proxy and
        # prove exact agreement with every stored pooling matrix.
        pre_proxy_factor, pre_proxy_years = build_projection_factor(
            factor.proxy,
            int(PRE_TREATMENT_WINDOW[0]),
            int(PRE_TREATMENT_WINDOW[1]),
        )
        if (
            tuple(pre_proxy_years)
            != tuple(
                range(
                    int(PRE_TREATMENT_WINDOW[0]),
                    int(PRE_TREATMENT_WINDOW[1]) + 1,
                )
            )
            or not np.array_equal(factor_matrix, pre_proxy_factor)
        ):
            raise IngestError(
                f"{pair[0]}/{pair[1]}: pre-treatment pooling factor is not "
                "the labelled donor-only proxy."
            )
        # Estimate the covariate loadings of the partition's targets only.
        pair_loadings: Dict[str, np.ndarray] = estimate_covariate_loadings(
            factor_matrix, covariate_by_unit, partition.targets
        )
        # Record the pair's loading matrices.
        loadings[pair] = pair_loadings
        # Build the post-window factor matrix from the partition's proxy,
        # which was constructed from the registered donor pool.
        projection_factor, years = build_projection_factor(
            factor.proxy, start_year, end_year
        )
        expected_projection_years: Tuple[int, ...] = tuple(
            range(int(start_year), int(end_year) + 1)
        )
        if tuple(years) != expected_projection_years:
            raise IngestError(
                f"{pair[0]}/{pair[1]}: projection factor years are not the "
                "exact requested calendar."
            )
        # Enforce the same two-column contract after the pre/post handoff.
        if projection_factor.shape[1] != FACTOR_COLUMNS:
            # Refuse accidental routing of richer variants into baseline.
            raise IngestError(
                f"{pair[0]}/{pair[1]}: projection factor violates the "
                "two-column baseline contract."
            )
        # Project every target's paths.
        for target in sorted(pair_loadings.keys()):
            # Read the target's loading matrix.
            target_loading: np.ndarray = pair_loadings[target]
            # Read the target's country loading vector from the fit.
            if target not in fit_result.loadings[pair]:
                raise IngestError(
                    f"{pair[0]}/{pair[1]}: fit omits target loading {target}."
                )
            alpha: np.ndarray = np.asarray(fit_result.loadings[pair][target])
            # Project the target's covariate and outcome paths.
            covariate_path, outcome_path = project_target_paths(
                target_loading,
                np.asarray(fit_result.betas[pair]),
                alpha,
                projection_factor,
            )
            # Assert that both paths are complete.
            assert_paths_complete(covariate_path, outcome_path, target)
            # Read the exact labelled pre-treatment axis when available.
            factor_axes: Mapping[str, Mapping[str, Tuple[Any, ...]]] = (
                factor.matrix_axes
            )
            # Use the declared axis only when this factor record supplies it.
            pre_years: Tuple[int, ...] = (
                tuple(int(year) for year in factor_axes[target]["years"])
                if factor_axes and target in factor_axes
                else tuple(
                    range(
                        int(PRE_TREATMENT_WINDOW[0]),
                        int(PRE_TREATMENT_WINDOW[1]) + 1,
                    )
                )
            )
            # Require exact pre-period labels and row alignment.
            expected_pre_years: Tuple[int, ...] = tuple(
                range(
                    int(PRE_TREATMENT_WINDOW[0]),
                    int(PRE_TREATMENT_WINDOW[1]) + 1,
                )
            )
            # Refuse a figure series with inferred or shifted time labels.
            if (
                pre_years != expected_pre_years
                or len(pre_years) != factor_matrix.shape[0]
            ):
                # Keep fitted-pre figure data separate and exactly labelled.
                raise IngestError(
                    f"{pair[0]}/{pair[1]}/{target}: fitted pre-period labels "
                    f"{pre_years} diverge from {expected_pre_years}."
                )
            # Read the target's observed pre-treatment dependent path.
            observed_pre: np.ndarray = np.asarray(
                factor.matrices[target][0], dtype=float
            ).ravel()
            # Compute the fitted pre-period path without changing post equations.
            fitted_pre: np.ndarray = (
                covariate_by_unit[target]
                @ np.asarray(fit_result.betas[pair]).reshape(-1, 1)
                + factor_matrix @ alpha.reshape(-1, 1)
            ).ravel()
            # Require finite aligned figure paths.
            if (
                observed_pre.shape != fitted_pre.shape
                or not np.isfinite(observed_pre).all()
                or not np.isfinite(fitted_pre).all()
            ):
                # Refuse malformed figure data before emitting it.
                raise IngestError(
                    f"{pair[0]}/{pair[1]}/{target}: fitted pre-period figure "
                    "path is incomplete or misaligned."
                )
            # Emit the distinct pre-period figure rows.
            for pre_index, pre_year in enumerate(pre_years):
                # Preserve observed, fitted and residual values explicitly.
                fitted_pre_rows.append(
                    {
                        "policy": pair[0],
                        "outcome": pair[1],
                        "country_code": target,
                        "year": int(pre_year),
                        "observed_pre_outcome": float(
                            observed_pre[pre_index]
                        ),
                        "fitted_pre_outcome": float(fitted_pre[pre_index]),
                        "pre_fit_residual": float(
                            observed_pre[pre_index] - fitted_pre[pre_index]
                        ),
                    }
                )
            # Emit the target's covariate rows.
            for index, year in enumerate(years):
                # Compose the covariate row of the year.
                covariate_row: Dict[str, Any] = {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "country_code": target,
                    "year": int(year),
                }
                # Attach the eight simulated control values.
                for column_index, control in enumerate(CONTROL_COLUMNS):
                    # Record the simulated control value.
                    covariate_row[control] = float(covariate_path[index, column_index])
                # Append the covariate row.
                covariate_rows.append(covariate_row)
                # Append the outcome row of the year.
                outcome_rows.append(
                    {
                        "policy": pair[0],
                        "outcome": pair[1],
                        "country_code": target,
                        "year": int(year),
                        "counterfactual_outcome": float(outcome_path[index]),
                    }
                )
        # Preserve only the permitted scientific inputs for the executed audit.
        audit_contexts[pair] = {
            "factor_matrix": factor_matrix.copy(),
            "projection_factor": projection_factor.copy(),
            "years": tuple(years),
            "covariate_by_unit": {
                unit: values.copy()
                for unit, values in covariate_by_unit.items()
            },
            "pair_loadings": {
                target: values.copy()
                for target, values in pair_loadings.items()
            },
        }
        # Record the projection note of the partition.
        notes.append(
            f"{pair[0]}/{pair[1]}: projected {len(pair_loadings)} target(s) over "
            f"{start_year} to {end_year} from the registered donor pool."
        )
    # Assemble the covariate path frame.
    covariate_paths: pd.DataFrame = pd.DataFrame(covariate_rows)
    # Assemble the outcome path frame.
    outcome_paths: pd.DataFrame = pd.DataFrame(outcome_rows)
    # Assemble the distinct fitted pre-treatment figure-data frame.
    fitted_pre_paths: pd.DataFrame = pd.DataFrame(fitted_pre_rows)
    # Validate exact post- and pre-period path scopes pair by pair.
    for pair, partition in registry.partitions.items():
        # Construct the exact post-treatment target-year key product.
        expected_post_keys: set[Tuple[str, int]] = set(
            itertools.product(
                (str(target) for target in partition.targets),
                range(int(start_year), int(end_year) + 1),
            )
        )
        # Select this pair's emitted covariate keys.
        pair_covariates: pd.DataFrame = covariate_paths[
            (covariate_paths["policy"] == pair[0])
            & (covariate_paths["outcome"] == pair[1])
        ]
        # Select this pair's emitted outcome keys.
        pair_outcomes: pd.DataFrame = outcome_paths[
            (outcome_paths["policy"] == pair[0])
            & (outcome_paths["outcome"] == pair[1])
        ]
        # Refuse duplicate post-treatment keys in either product.
        if bool(
            pair_covariates.duplicated(
                subset=["country_code", "year"], keep=False
            ).any()
        ) or bool(
            pair_outcomes.duplicated(
                subset=["country_code", "year"], keep=False
            ).any()
        ):
            # A duplicate would distort later gap alignment.
            raise IngestError(
                f"{pair[0]}/{pair[1]}: projected paths contain duplicate keys."
            )
        # Read exact emitted covariate and outcome key identities.
        covariate_keys: set[Tuple[str, int]] = set(
            zip(
                pair_covariates["country_code"].astype(str),
                pair_covariates["year"].astype(int),
            )
        )
        # Read exact emitted outcome keys.
        outcome_keys: set[Tuple[str, int]] = set(
            zip(
                pair_outcomes["country_code"].astype(str),
                pair_outcomes["year"].astype(int),
            )
        )
        # Require both post products to match the declared scope exactly.
        if (
            covariate_keys != expected_post_keys
            or outcome_keys != expected_post_keys
        ):
            # Refuse missing, extra or differently scoped scientific paths.
            raise IngestError(
                f"{pair[0]}/{pair[1]}: post path scope diverges from the "
                "declared target-year product."
            )
        # Construct the exact fitted-pre target-year product.
        expected_pre_keys: set[Tuple[str, int]] = set(
            itertools.product(
                (str(target) for target in partition.targets),
                range(
                    int(PRE_TREATMENT_WINDOW[0]),
                    int(PRE_TREATMENT_WINDOW[1]) + 1,
                ),
            )
        )
        # Select and verify this pair's fitted-pre figure keys.
        pair_pre: pd.DataFrame = fitted_pre_paths[
            (fitted_pre_paths["policy"] == pair[0])
            & (fitted_pre_paths["outcome"] == pair[1])
        ]
        # Refuse duplicate pre-period keys.
        if bool(
            pair_pre.duplicated(
                subset=["country_code", "year"], keep=False
            ).any()
        ):
            # Figure data must remain uniquely keyed.
            raise IngestError(
                f"{pair[0]}/{pair[1]}: fitted-pre paths contain duplicate keys."
            )
        # Require exact fitted-pre scope.
        if set(
            zip(
                pair_pre["country_code"].astype(str),
                pair_pre["year"].astype(int),
            )
        ) != expected_pre_keys:
            # Do not splice post paths into the figure-only product.
            raise IngestError(
                f"{pair[0]}/{pair[1]}: fitted-pre figure scope diverges."
            )
    # Compose the union of all projected target identities.
    all_targets: Tuple[str, ...] = tuple(
        sorted(
            {
                str(target)
                for partition in registry.partitions.values()
                for target in partition.targets
            }
        )
    )
    # Compose the exact target-post audit columns.
    audit_value_columns: Tuple[str, ...] = tuple(
        dict.fromkeys(
            [
                *(pair[1] for pair in registry.partitions),
                *CONTROL_COLUMNS,
            ]
        )
    )
    # Compose the exact audit year labels.
    audit_years: Tuple[int, ...] = tuple(
        range(int(start_year), int(end_year) + 1)
    )
    # Use realized target-post data when supplied by the full pipeline.
    if observed_panel is not None:
        # Require keys and every target-post outcome/covariate audit field.
        required_audit_columns: set[str] = {
            "country_code",
            "year",
            *audit_value_columns,
        }
        # Refuse a panel that cannot support the full perturbation.
        if not required_audit_columns.issubset(observed_panel.columns):
            # Name missing audit fields.
            raise IngestError(
                "Observed leakage-audit panel omits column(s): "
                f"{sorted(required_audit_columns - set(observed_panel.columns))}."
            )
        # Select all target rows first so calendar validation precedes filtering
        # and cannot silently discard a colliding textual year label.
        target_audit_panel: pd.DataFrame = observed_panel[
            observed_panel["country_code"].astype(str).isin(set(all_targets))
        ][["country_code", "year", *audit_value_columns]].copy()
        target_numeric_years: pd.Series = pd.to_numeric(
            target_audit_panel["year"], errors="coerce"
        )
        invalid_target_years: pd.Series = (
            target_audit_panel["year"].map(
                lambda value: isinstance(value, (bool, np.bool_))
            )
            | target_numeric_years.isna()
            | ~np.isfinite(target_numeric_years.astype(float))
            | (target_numeric_years.astype(float) % 1.0 != 0.0)
        )
        if bool(invalid_target_years.any()):
            raise IngestError(
                "Observed leakage-audit panel carries invalid target calendar years."
            )
        target_audit_panel["_audit_year"] = target_numeric_years.astype("int64")
        # Select the exact canonical projection calendar after validation.
        audit_panel = target_audit_panel[
            target_audit_panel["_audit_year"].isin(audit_years)
        ].copy()
        # Reject duplicate country-years after normalization, including source
        # labels such as ``'2014'`` and ``'2014.0'``.
        if bool(
            audit_panel.duplicated(
                subset=["country_code", "_audit_year"], keep=False
            ).any()
        ):
            raise IngestError(
                "Leakage-audit target-post years collide after normalization."
            )
        audit_panel["year"] = audit_panel["_audit_year"].astype("int64")
        audit_panel = audit_panel.drop(columns=["_audit_year"])
        # Record the realized-data provenance.
        perturbation_source: str = (
            "realized canonical target-post panel supplied to audit boundary"
        )
    else:
        # Build an explicit sentinel channel when realized data are inaccessible.
        audit_panel = pd.DataFrame(
            [
                {
                    "country_code": target,
                    "year": year,
                    **{column: 0.0 for column in audit_value_columns},
                }
                for target in all_targets
                for year in audit_years
            ]
        )
        # Disclose that the realized dependency was unavailable to this caller.
        perturbation_source = (
            "synthetic sentinel because realized target-post panel was "
            "inaccessible at this call boundary"
        )
    # Refuse duplicate target-year audit rows.
    if bool(
        audit_panel.duplicated(
            subset=["country_code", "year"], keep=False
        ).any()
    ):
        # Duplicate forbidden inputs make perturbation lineage ambiguous.
        raise IngestError(
            "Leakage-audit target-post panel carries duplicate country-year rows."
        )
    # Parse audit years without accepting boolean or fractional labels.
    audit_numeric_years: pd.Series = pd.to_numeric(
        audit_panel["year"], errors="coerce"
    )
    # Identify malformed calendar values before any integer conversion.
    invalid_audit_years: pd.Series = (
        audit_panel["year"].map(
            lambda value: isinstance(value, (bool, np.bool_))
        )
        | audit_numeric_years.isna()
        | ~np.isfinite(audit_numeric_years.astype(float))
        | (audit_numeric_years.astype(float) % 1.0 != 0.0)
    )
    # Refuse lossy audit calendar labels.
    if bool(invalid_audit_years.any()):
        # Exact perturbation scope requires exact years.
        raise IngestError(
            "Leakage-audit target-post panel carries invalid calendar years."
        )
    # Require the exact Cartesian target-year audit grid.
    observed_audit_keys: set[Tuple[str, int]] = set(
        zip(
            audit_panel["country_code"].astype(str),
            audit_numeric_years.astype(int),
        )
    )
    # Construct the expected target-post key product.
    expected_audit_keys: set[Tuple[str, int]] = set(
        itertools.product(all_targets, audit_years)
    )
    # Refuse missing or extra audit identities.
    if observed_audit_keys != expected_audit_keys:
        # Report both sides of the key divergence.
        raise IngestError(
            "Leakage-audit target-post key product diverges: missing "
            f"{sorted(expected_audit_keys - observed_audit_keys)[:MAX_REPORTED_OFFENDERS]}, "
            "extra "
            f"{sorted(observed_audit_keys - expected_audit_keys)[:MAX_REPORTED_OFFENDERS]}."
        )
    # Parse all forbidden target-post scientific values.
    audit_numeric: pd.DataFrame = audit_panel.loc[
        :, list(audit_value_columns)
    ].apply(pd.to_numeric, errors="coerce")
    # Require finite realized/sentinel values before perturbation.
    if not np.isfinite(audit_numeric.to_numpy(dtype=float)).all():
        # Missing forbidden data would make the executed comparison incomplete.
        raise IngestError(
            "Leakage-audit target-post values contain non-finite cells."
        )
    # Store the validated numeric representation.
    audit_panel.loc[:, list(audit_value_columns)] = audit_numeric.to_numpy(
        dtype=float
    )
    # Construct an independent material perturbation of every forbidden cell.
    perturbed_audit_panel: pd.DataFrame = audit_panel.copy(deep=True)
    # Perturb each scientific column by a distinct large deterministic offset.
    for column_index, column in enumerate(audit_value_columns, start=1):
        # Apply the perturbation to every target-post cell.
        perturbed_audit_panel[column] = (
            pd.to_numeric(perturbed_audit_panel[column], errors="raise")
            + float(column_index) * 1.0e6
        )
    # Count the materially changed forbidden cells.
    perturbation_cells: int = int(
        len(perturbed_audit_panel) * len(audit_value_columns)
    )
    # Freeze one target's forbidden-channel payload.
    def _forbidden_digest(frame: pd.DataFrame) -> str:
        # Sort exact labelled rows and serialize deterministically.
        ordered: pd.DataFrame = frame.sort_values(
            ["country_code", "year"], kind="stable"
        ).reset_index(drop=True)
        # Digest labels and values together.
        return hashlib.sha256(
            ordered.to_json(
                orient="split", double_precision=15
            ).encode("utf-8")
        ).hexdigest()
    # Execute the full scientific projection boundary while consuming the
    # forbidden channel only into separate lineage.
    def _audited_projection_boundary(
        loading: np.ndarray,
        beta: np.ndarray,
        alpha: np.ndarray,
        projection_factor: np.ndarray,
        forbidden_frame: pd.DataFrame,
    ) -> Tuple[np.ndarray, np.ndarray, str]:
        # Consume and freeze forbidden inputs without routing them scientifically.
        forbidden_digest: str = _forbidden_digest(forbidden_frame)
        # Execute the complete declared projection transformation.
        projected_x, projected_y = project_target_paths(
            loading, beta, alpha, projection_factor
        )
        # Return scientific outputs and distinct audit lineage.
        return projected_x, projected_y, forbidden_digest
    # Execute baseline and perturbed leakage checks for every partition.
    for pair in sorted(registry.partitions):
        # Read the pair's permitted projection context.
        context: Mapping[str, Any] = audit_contexts[pair]
        # Read the exact partition target roster.
        pair_targets: Tuple[str, ...] = tuple(
            registry.partitions[pair].targets
        )
        # Accumulate target-keyed baseline and perturbed arrays.
        baseline_covariates: Dict[str, np.ndarray] = {}
        # Accumulate baseline outcomes.
        baseline_outcomes: Dict[str, np.ndarray] = {}
        # Accumulate perturbed-run covariates.
        perturbed_covariates: Dict[str, np.ndarray] = {}
        # Accumulate perturbed-run outcomes.
        perturbed_outcomes: Dict[str, np.ndarray] = {}
        # Accumulate consumed forbidden-input digests.
        forbidden_digests: Dict[str, Tuple[str, str]] = {}
        # Execute the audit separately for every target.
        for target in pair_targets:
            # Select baseline forbidden realized target-post cells.
            baseline_forbidden: pd.DataFrame = audit_panel[
                audit_panel["country_code"].astype(str) == str(target)
            ][["country_code", "year", pair[1], *CONTROL_COLUMNS]]
            # Select independently perturbed cells with identical labels.
            perturbed_forbidden: pd.DataFrame = perturbed_audit_panel[
                perturbed_audit_panel["country_code"].astype(str)
                == str(target)
            ][["country_code", "year", pair[1], *CONTROL_COLUMNS]]
            # Read the four permitted scientific inputs.
            loading = np.asarray(context["pair_loadings"][target])
            # Read the pooled coefficient.
            beta = np.asarray(fit_result.betas[pair])
            # Read the pre-estimated country factor loading.
            alpha = np.asarray(fit_result.loadings[pair][target])
            # Read the donor-only post-treatment factor.
            projection_factor = np.asarray(context["projection_factor"])
            # Execute the baseline boundary with baseline forbidden lineage.
            baseline_x, baseline_y, baseline_digest = (
                _audited_projection_boundary(
                    loading,
                    beta,
                    alpha,
                    projection_factor,
                    baseline_forbidden,
                )
            )
            # Re-execute after the forbidden target-post perturbation.
            perturbed_x, perturbed_y, perturbed_digest = (
                _audited_projection_boundary(
                    loading,
                    beta,
                    alpha,
                    projection_factor,
                    perturbed_forbidden,
                )
            )
            # Verify the baseline boundary reproduces emitted covariate rows.
            emitted_x: np.ndarray = covariate_paths[
                (covariate_paths["policy"] == pair[0])
                & (covariate_paths["outcome"] == pair[1])
                & (covariate_paths["country_code"].astype(str) == str(target))
            ].sort_values("year", kind="stable")[list(CONTROL_COLUMNS)].to_numpy(
                dtype=float
            )
            # Verify the baseline boundary reproduces emitted outcome rows.
            emitted_y: np.ndarray = outcome_paths[
                (outcome_paths["policy"] == pair[0])
                & (outcome_paths["outcome"] == pair[1])
                & (outcome_paths["country_code"].astype(str) == str(target))
            ].sort_values("year", kind="stable")[
                "counterfactual_outcome"
            ].to_numpy(dtype=float)
            # Refuse any divergence between audited and emitted transformations.
            if not np.array_equal(baseline_x, emitted_x) or not np.array_equal(
                baseline_y, emitted_y
            ):
                # The audit must execute the exact production boundary.
                raise IngestError(
                    f"{pair[0]}/{pair[1]}/{target}: audited baseline does not "
                    "match emitted projection paths."
                )
            # Verify the declared equations' algebraic reduced form.
            target_dependent: np.ndarray = np.asarray(
                factor_result.factors[pair].matrices[target][0],
                dtype=float,
            )
            # Compute f_t prime F-plus Y_i from pre-target and donor factors.
            reduced_loading: np.ndarray = np.linalg.lstsq(
                np.asarray(context["factor_matrix"]),
                target_dependent,
                rcond=None,
            )[0]
            # Apply the post-period factor to the reduced-form loading.
            reduced_outcome: np.ndarray = (
                projection_factor @ reduced_loading
            ).ravel()
            # Refuse a projection that diverges from the supplied equations.
            if not np.allclose(
                baseline_y,
                reduced_outcome,
                rtol=1.0e-10,
                atol=1.0e-10,
            ):
                # Preserve the equations rather than substituting another model.
                raise IngestError(
                    f"{pair[0]}/{pair[1]}/{target}: projected outcome diverges "
                    "from f_t' F-plus Y_i implied by equations B5-B8."
                )
            # Store all executed comparison artefacts.
            baseline_covariates[target] = baseline_x
            baseline_outcomes[target] = baseline_y
            perturbed_covariates[target] = perturbed_x
            perturbed_outcomes[target] = perturbed_y
            forbidden_digests[target] = (
                baseline_digest,
                perturbed_digest,
            )
        # Compose per-input scientific role/time/source provenance.
        input_provenance: Mapping[str, Mapping[str, Any]] = {
            "covariate_loadings": {
                "country_role": "target",
                "time_start": int(PRE_TREATMENT_WINDOW[0]),
                "time_end": int(PRE_TREATMENT_WINDOW[1]),
                "provenance": (
                    "target pre-treatment lagged controls projected on the "
                    "registered donor factor"
                ),
            },
            "pooled_coefficient": {
                "country_role": "pooled targets_and_donors",
                "time_start": int(PRE_TREATMENT_WINDOW[0]),
                "time_end": int(PRE_TREATMENT_WINDOW[1]),
                "provenance": (
                    "Task 11 verified pooled fit "
                    f"{fit_result.fingerprint_sha256}"
                ),
            },
            "country_loadings": {
                "country_role": "target",
                "time_start": int(PRE_TREATMENT_WINDOW[0]),
                "time_end": int(PRE_TREATMENT_WINDOW[1]),
                "provenance": (
                    "target pre-treatment outcome residual projected on the "
                    "registered donor factor"
                ),
            },
            "projection_factor": {
                "country_role": "never_treated_donors",
                "time_start": int(start_year),
                "time_end": int(end_year),
                "provenance": (
                    "outcome-specific registered donor mean; factor result "
                    f"{factor_result.fingerprint_sha256}"
                ),
            },
        }
        # Record the executed input-level audit rows.
        audits.append(
            record_invariant_audit(
                pair[0],
                pair[1],
                pair_targets,
                PERMITTED_PROJECTION_INPUTS,
                input_provenance=input_provenance,
                baseline_covariates=baseline_covariates,
                baseline_outcomes=baseline_outcomes,
                perturbed_covariates=perturbed_covariates,
                perturbed_outcomes=perturbed_outcomes,
                forbidden_input_digests=forbidden_digests,
                perturbation_cells=perturbation_cells,
                perturbation_source=perturbation_source,
            )
        )
    # Summarise the ranges of the simulated covariates without clipping.
    range_summary: pd.DataFrame = summarize_path_ranges(covariate_paths)
    # Concatenate the invariant audits.
    invariant_audit: pd.DataFrame = pd.concat(audits, ignore_index=True)
    # Require exact unique audit coverage of every target-input combination.
    expected_audit_rows: int = int(
        sum(
            len(partition.targets) * len(PERMITTED_PROJECTION_INPUTS)
            for partition in registry.partitions.values()
        )
    )
    # Refuse incomplete, duplicated or non-verified audit output.
    if (
        len(invariant_audit) != expected_audit_rows
        or bool(
            invariant_audit.duplicated(
                subset=["policy", "outcome", "country_code", "input_name"],
                keep=False,
            ).any()
        )
        or set(invariant_audit["audit_outcome"]) != {"invariant_verified"}
    ):
        # No projection may proceed with partial dataflow evidence.
        raise IngestError(
            "Executed invariant-audit coverage is incomplete or duplicated."
        )
    # Record the executed perturbation and inaccessible-data status truthfully.
    notes.append(
        f"leakage audit independently perturbed {perturbation_cells} forbidden "
        f"target-post outcome/covariate cell(s); source: {perturbation_source}."
    )
    # Record the separate figure-data contract.
    notes.append(
        "fitted 2001-2013 target paths are emitted in fitted_pre_paths as a "
        "distinct figure product and are never spliced into post projections."
    )
    # Disclose the algebraic consequence of the supplied projection equations.
    notes.append(
        "equations B5-B8 were independently verified against the reduced form "
        "f_t' F-plus Y_i for every target path."
    )
    # Record the no-clipping decision.
    notes.append(
        "simulated covariates and outcomes are emitted unclipped, with their "
        "ranges recorded for audit, because the design forbids clamping."
    )
    # Compute the canonical digest of the counterfactual objects.
    fingerprint: str = freeze_counterfactual(
        loadings,
        covariate_paths,
        outcome_paths,
        fitted_pre_paths=fitted_pre_paths,
        invariant_audit=invariant_audit,
    )
    # Assemble the immutable result.
    result: CounterfactualResult = CounterfactualResult(
        loadings=loadings,
        covariate_paths=covariate_paths,
        outcome_paths=outcome_paths,
        fitted_pre_paths=fitted_pre_paths,
        range_summary=range_summary,
        invariant_audit=invariant_audit,
        fingerprint_sha256=fingerprint,
        projected_utc=current_utc_iso(),
        notes=tuple(notes),
    )
    # Log the completed projection with its digest prefix.
    logger.info(
        "counterfactual projection complete: %d path row(s), fingerprint %s.",
        len(covariate_paths),
        fingerprint[:16],
    )
    # Return the assembled result.
    return result
````

````python
# Task 13: Gap Matrix and Level/Incremental ATT

# ==============================================================================
# Task 13: Gap Matrix and Level/Incremental ATT
# ==============================================================================

# Declare the first wave window, which spans the pre-escalation level.
WAVE_ONE_WINDOW: Final[Tuple[int, int]] = (2014, 2021)
# Declare the second wave window, which spans the post-escalation level.
WAVE_TWO_WINDOW: Final[Tuple[int, int]] = (2022, 2024)
# Declare the pre-escalation window of the incremental baseline.
PRE_ESCALATION_WINDOW: Final[Tuple[int, int]] = (2019, 2021)
# Declare the post-escalation window of the incremental estimand.
POST_ESCALATION_WINDOW: Final[Tuple[int, int]] = (2022, 2024)
# Declare the label of the first wave window.
WAVE_ONE_LABEL: Final[str] = "wave_1_level"
# Declare the label of the second wave window.
WAVE_TWO_LABEL: Final[str] = "wave_2_level"


# Declare the immutable result of the gap and effect computation.
@dataclass(frozen=True)
class GapAttResult:
    """
    Immutable result of the gap matrix and effect computation.

    Attributes
    ----------
    gaps : dict
        The gap matrices keyed on the pair, each indexed on the target
        country code and columned by the projection years.
    country_window_means : pandas.DataFrame
        The per-country window means keyed on the pair, the country and the
        window label.
    level_effects : pandas.DataFrame
        The coalition level effects keyed on the pair with one column per
        wave window.
    pre_escalation_levels : pandas.DataFrame
        The per-country pre-escalation level gaps keyed on the pair and the
        country, which are the continuation baselines.
    incremental_effects : pandas.DataFrame
        One row per pair carrying the incremental effect and both identity
        routes.
    publication_audit : pandas.DataFrame
        One row per supplied manuscript cell carrying its estimand identity,
        units, displayed precision, implied rounding interval, explicitly
        recorded numerical allowance, unrounded reproduction and deviation.
    fingerprint_sha256 : str
        Digest of the canonical gap payload.
    computed_utc : str
        ISO-8601 UTC timestamp of the computation.
    notes : tuple of str
        Recorded decisions, including the missingness assertions and the
        absence of weighting.
    """
    # The gap matrices keyed on the pair.
    gaps: Mapping[Tuple[str, str], pd.DataFrame]
    # The per-country window means keyed on the pair and the country.
    country_window_means: pd.DataFrame
    # The coalition level effects keyed on the pair.
    level_effects: pd.DataFrame
    # The per-country pre-escalation level gaps.
    pre_escalation_levels: pd.DataFrame
    # One row per pair carrying the incremental effect and both routes.
    incremental_effects: pd.DataFrame
    # Cell-level audit of publication-aware manuscript comparisons.
    publication_audit: pd.DataFrame
    # Digest of the canonical gap payload.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the computation.
    computed_utc: str
    # Recorded decisions of the computation.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 1: Compute the country-year gaps
# -------------------------------------------------------------------------------------------------------------------------------


# Validate an inclusive calendar window without lossy coercion.
def _validate_att_window(span: Tuple[int, int], label: str) -> Tuple[int, int]:
    """Return a validated inclusive integer window used by the ATT task."""
    # Require the public two-element window contract.
    if not isinstance(span, tuple) or len(span) != 2:
        raise IngestError(f"{label} must be a two-element tuple of calendar years.")
    # Reject Boolean and non-integral endpoints before converting to integers.
    endpoints: List[int] = []
    for endpoint in span:
        if isinstance(endpoint, (bool, np.bool_)):
            raise IngestError(f"{label} carries a Boolean calendar year.")
        try:
            numeric = float(endpoint)
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError(f"{label} carries a non-numeric calendar year.") from exc
        if not math.isfinite(numeric) or not numeric.is_integer():
            raise IngestError(f"{label} carries a non-integral calendar year {endpoint!r}.")
        endpoints.append(int(numeric))
    # An inclusive window cannot run backwards.
    if endpoints[0] > endpoints[1]:
        raise IngestError(f"{label} is reversed: {tuple(endpoints)}.")
    return endpoints[0], endpoints[1]


# Validate and attach a lossless integer calendar-year column.
def _validated_att_years(frame: pd.DataFrame, context: str) -> pd.DataFrame:
    """Copy a frame and attach ``_att_year`` after lossless year validation."""
    # The labelled calendar key is mandatory.
    if "year" not in frame.columns:
        raise IngestError(f"{context} lacks the required year column.")
    # Booleans are integers in Python but are not calendar-year declarations.
    if bool(frame["year"].map(lambda value: isinstance(value, (bool, np.bool_))).any()):
        raise IngestError(f"{context} carries a Boolean calendar year.")
    # Coerce only for validation and retain failures as missing sentinels.
    numeric_years = pd.to_numeric(frame["year"], errors="coerce")
    numeric_values = numeric_years.to_numpy(dtype=float, copy=False)
    # Require finite, integral values before any integer conversion.
    invalid = (~np.isfinite(numeric_values)) | (numeric_values != np.floor(numeric_values))
    if bool(np.any(invalid)):
        offenders = frame.loc[invalid, "year"].tolist()[:MAX_REPORTED_OFFENDERS]
        raise IngestError(
            f"{context} carries invalid calendar year value(s), for example {offenders}."
        )
    # Attach a losslessly converted calendar axis on a private copy.
    validated = frame.copy(deep=True)
    validated["_att_year"] = numeric_years.astype("int64")
    return validated


# Validate a canonical nonempty target roster.
def _validated_att_roster(targets: Tuple[str, ...], context: str) -> Tuple[str, ...]:
    """Return a sorted unique nonblank target roster."""
    # Refuse the empty scientific population.
    if not targets:
        raise IngestError(f"{context}: target set is empty.")
    # Reject null and blank identities rather than stringifying them.
    if any(code is None or not str(code).strip() for code in targets):
        raise IngestError(f"{context}: target roster carries a null or blank code.")
    codes = tuple(str(code).strip() for code in targets)
    # A duplicate would silently change the nominal target count or weighting.
    if len(set(codes)) != len(codes):
        raise IngestError(f"{context}: target roster carries duplicate country codes.")
    return tuple(sorted(codes))


# Validate a finite labelled country-by-year scientific matrix.
def _validate_att_matrix(matrix: pd.DataFrame, context: str) -> None:
    """Fail closed unless an ATT matrix has unique axes and finite cells."""
    # Empty or duplicated axes cannot identify a fixed country-year product.
    if matrix.empty:
        raise IngestError(f"{context} is empty.")
    if matrix.index.has_duplicates:
        raise IngestError(f"{context} carries duplicate country labels.")
    if matrix.columns.has_duplicates:
        raise IngestError(f"{context} carries duplicate year labels.")
    # Reject null/blank country identities.
    if any(code is None or not str(code).strip() for code in matrix.index):
        raise IngestError(f"{context} carries a null or blank country label.")
    # Year labels must already be finite integral calendar years.
    for year in matrix.columns:
        if isinstance(year, (bool, np.bool_)):
            raise IngestError(f"{context} carries a Boolean year label.")
        try:
            numeric_year = float(year)
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError(f"{context} carries a non-numeric year label.") from exc
        if not math.isfinite(numeric_year) or not numeric_year.is_integer():
            raise IngestError(f"{context} carries an invalid year label {year!r}.")
    # Scientific cells must be numeric and finite; pandas means must never skip.
    try:
        values = matrix.to_numpy(dtype=float)
    except (TypeError, ValueError) as exc:
        raise IngestError(f"{context} carries a non-numeric scientific cell.") from exc
    if not bool(np.isfinite(values).all()):
        raise IngestError(f"{context} carries a missing or non-finite scientific cell.")


# Compute one partition's gap matrix.
def compute_gap_matrix(
    observed_panel: pd.DataFrame,
    outcome: str,
    targets: Tuple[str, ...],
    counterfactual_paths: pd.DataFrame,
    policy: str,
    window: Tuple[int, int] = WAVE_ONE_WINDOW,
    window_end: Tuple[int, int] = POST_ESCALATION_WINDOW,
) -> pd.DataFrame:
    """
    Compute the country-year gap matrix of one partition.

    Parameters
    ----------
    observed_panel : pandas.DataFrame
        The analysis panel carrying the observed outcome columns.
    outcome : str
        The canonical outcome variable of the partition.
    targets : tuple of str
        The target countries of the partition.
    counterfactual_paths : pandas.DataFrame
        The counterfactual outcome paths keyed on the pair, the country and
        the year.
    policy : str
        The policy instrument of the pair, used to select the counterfactual
        rows.
    window : tuple of int, optional
        The wave window whose opening year opens the gap matrix.
    window_end : tuple of int, optional
        The post-escalation window whose closing year closes the gap matrix.

    Returns
    -------
    pandas.DataFrame
        The gap matrix indexed on the target country code and columned by the
        projection years, with a positive cell meaning the observed outcome
        lies above the imputed untreated path and a negative cell meaning it
        lies below it. The gap is a total difference at each year and is not
        an incremental difference across waves.

    Raises
    ------
    IngestError
        If the outcome column is absent, if a target is absent from either
        input, if either side carries a missing cell, or if the aligned shapes
        diverge.
    """
    # Require the complete labelled input schemas before selecting rows.
    observed_required = {"country_code", "year", outcome}
    missing_observed_columns = sorted(observed_required - set(observed_panel.columns))
    if missing_observed_columns:
        raise IngestError(
            f"Observed panel lacks gap column(s) {missing_observed_columns}."
        )
    path_required = {
        "policy", "outcome", "country_code", "year", "counterfactual_outcome"
    }
    missing_path_columns = sorted(path_required - set(counterfactual_paths.columns))
    if missing_path_columns:
        raise IngestError(
            f"Counterfactual paths lack column(s) {missing_path_columns}."
        )
    # Resolve the fixed target population and the exact inclusive year span.
    roster = _validated_att_roster(targets, f"{policy}/{outcome}")
    wave_start, _ = _validate_att_window(window, "wave-one gap window")
    _, post_end = _validate_att_window(window_end, "post-escalation gap window")
    if wave_start > post_end:
        raise IngestError(
            f"{policy}/{outcome}: gap span starts after it ends."
        )
    years: List[int] = list(range(wave_start, post_end + 1))
    expected_keys = {(country, year) for country in roster for year in years}
    # Validate calendar values losslessly before any selection or pivot.
    observed_validated = _validated_att_years(observed_panel, "Observed panel")
    paths_validated = _validated_att_years(
        counterfactual_paths, "Counterfactual paths"
    )
    # Select the declared policy/outcome path and require exactly its registered
    # country roster; an extra country is scientific roster drift, not harmless.
    scoped_paths = paths_validated.loc[
        (paths_validated["policy"].astype(str) == str(policy))
        & (paths_validated["outcome"].astype(str) == str(outcome))
    ].copy()
    if scoped_paths.empty:
        raise IngestError(
            f"{policy}/{outcome}: no counterfactual row for the partition."
        )
    # Reject null/blank path identities before canonicalizing them.
    if bool(scoped_paths["country_code"].isna().any()) or bool(
        scoped_paths["country_code"].astype(str).str.strip().eq("").any()
    ):
        raise IngestError(f"{policy}/{outcome}: counterfactual roster is malformed.")
    scoped_paths["_att_country"] = scoped_paths["country_code"].astype(str).str.strip()
    observed_path_roster = set(scoped_paths["_att_country"])
    if observed_path_roster != set(roster):
        raise IngestError(
            f"{policy}/{outcome}: counterfactual roster {sorted(observed_path_roster)} "
            f"does not equal registered targets {list(roster)}."
        )
    # The post-period outcome product is exact; fitted-pre figure data belong
    # to a different contract and must never enter this path frame.
    all_scoped_path_keys = list(
        zip(scoped_paths["_att_country"], scoped_paths["_att_year"].astype(int))
    )
    duplicate_path_keys = {
        key for key in all_scoped_path_keys if all_scoped_path_keys.count(key) > 1
    }
    if duplicate_path_keys:
        raise IngestError(
            f"{policy}/{outcome}: counterfactual carries duplicate target-year "
            f"key(s), for example {sorted(duplicate_path_keys)[:MAX_REPORTED_OFFENDERS]}."
        )
    # Restrict the otherwise valid path product to this reporting horizon;
    # available years outside the requested horizon are not fitted-pre inputs
    # and do not alter this ATT computation.
    scoped_paths = scoped_paths.loc[scoped_paths["_att_year"].isin(years)].copy()
    scoped_path_keys = list(
        zip(scoped_paths["_att_country"], scoped_paths["_att_year"].astype(int))
    )
    if set(scoped_path_keys) != expected_keys:
        missing = sorted(expected_keys - set(scoped_path_keys))
        raise IngestError(
            f"{policy}/{outcome}: counterfactual target-year grid differs from "
            f"the declared product; missing {missing[:MAX_REPORTED_OFFENDERS]}."
        )
    # Select the matching observed target-year product.
    observed_country = observed_validated["country_code"]
    if bool(observed_country.isna().any()):
        raise IngestError("Observed panel carries a null country code.")
    observed_validated["_att_country"] = observed_country.astype(str).str.strip()
    observed_subset = observed_validated.loc[
        observed_validated["_att_country"].isin(set(roster))
        & observed_validated["_att_year"].isin(years)
    ].copy()
    observed_keys = list(
        zip(observed_subset["_att_country"], observed_subset["_att_year"].astype(int))
    )
    duplicate_observed_keys = {
        key for key in observed_keys if observed_keys.count(key) > 1
    }
    if duplicate_observed_keys:
        raise IngestError(
            f"{policy}/{outcome}: observed panel carries duplicate target-year "
            f"key(s), for example {sorted(duplicate_observed_keys)[:MAX_REPORTED_OFFENDERS]}."
        )
    if set(observed_keys) != expected_keys:
        missing = sorted(expected_keys - set(observed_keys))
        extra = sorted(set(observed_keys) - expected_keys)
        raise IngestError(
            f"{policy}/{outcome}: observed target-year grid differs from the "
            f"declared product; missing {missing[:MAX_REPORTED_OFFENDERS]}, "
            f"extra {extra[:MAX_REPORTED_OFFENDERS]}."
        )
    # Coerce scientific values once and reject every missing or infinite cell.
    observed_subset["_att_value"] = pd.to_numeric(
        observed_subset[outcome], errors="coerce"
    )
    scoped_paths["_att_value"] = pd.to_numeric(
        scoped_paths["counterfactual_outcome"], errors="coerce"
    )
    if not bool(np.isfinite(observed_subset["_att_value"].to_numpy(dtype=float)).all()):
        raise IngestError(
            f"{policy}/{outcome}: observed target-year grid carries a missing or "
            "non-finite outcome."
        )
    if not bool(np.isfinite(scoped_paths["_att_value"].to_numpy(dtype=float)).all()):
        raise IngestError(
            f"{policy}/{outcome}: counterfactual target-year grid carries a "
            "missing or non-finite outcome."
        )
    # Pivot only after uniqueness and completeness make aggregation impossible.
    observed = observed_subset.pivot(
        index="_att_country", columns="_att_year", values="_att_value"
    )
    counterfactual = scoped_paths.pivot(
        index="_att_country", columns="_att_year", values="_att_value"
    )
    target_index = pd.Index(roster, name="country_code")
    year_axis = pd.Index(years, name="year")
    observed = observed.reindex(index=roster, columns=years)
    counterfactual = counterfactual.reindex(index=roster, columns=years)
    observed.index, observed.columns = target_index, year_axis
    counterfactual.index, counterfactual.columns = target_index, year_axis
    # Compute the labelled observed-minus-projected gap without changing axes.
    gaps: pd.DataFrame = observed - counterfactual
    _validate_att_matrix(gaps, f"{policy}/{outcome} gap matrix")
    if not gaps.index.equals(target_index) or not gaps.columns.equals(year_axis):
        raise IngestError(f"{policy}/{outcome}: gap axes drifted during subtraction.")
    # Log the gap construction.
    logger.info(
        "%s/%s: gap matrix built at shape %s over %d to %d.",
        policy,
        outcome,
        gaps.shape,
        years[0],
        years[-1],
    )
    # Return the gap matrix.
    return gaps


# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 2: Aggregate the level effects across the two windows
# -------------------------------------------------------------------------------------------------------------------------------


# Aggregate the per-country window means of one gap matrix.
def aggregate_country_window_means(
    gaps: pd.DataFrame,
    windows: Mapping[str, Tuple[int, int]],
) -> pd.DataFrame:
    """
    Aggregate the per-country window means of a gap matrix.

    Parameters
    ----------
    gaps : pandas.DataFrame
        The gap matrix indexed on the country and columned by the years.
    windows : mapping of str to tuple of int
        The window labels and their inclusive year spans.

    Returns
    -------
    pandas.DataFrame
        The per-country means indexed on the country and columned by the
        window label.

    Raises
    ------
    IngestError
        If a window's years are absent from the gap matrix, which would make
        the mean a mean over a shorter span than declared.
    """
    # Validate the complete fixed-roster matrix before any pandas reduction.
    _validate_att_matrix(gaps, "Gap matrix for country-window aggregation")
    # Accumulate the per-country window means.
    means: Dict[str, pd.Series] = {}
    # Compute the mean of every declared window.
    for label, span in windows.items():
        # Compose the losslessly validated window's year columns.
        start, end = _validate_att_window(span, f"{label} level window")
        span_years: List[int] = list(range(start, end + 1))
        # Identify the window years the matrix does not carry.
        missing_years: List[int] = [
            year for year in span_years if year not in set(int(c) for c in gaps.columns)
        ]
        # Refuse a window that leaves the gap matrix's coverage.
        if missing_years:
            # The failure names the missing years.
            raise IngestError(
                f"Gap matrix does not cover window {label} year(s) {missing_years}."
            )
        # Average the country's gaps across the window's years with equal
        # weight on every year.
        means[label] = gaps[span_years].mean(axis=1)
    # Assemble the per-country means frame.
    frame: pd.DataFrame = pd.DataFrame(means)
    # Reductions must preserve every country and produce finite outputs.
    if not frame.index.equals(gaps.index) or not bool(
        np.isfinite(frame.to_numpy(dtype=float)).all()
    ):
        raise IngestError("Country-window aggregation changed the roster or emitted non-finite values.")
    # Log the aggregation outcome.
    logger.info(
        "per-country window means aggregated across %d window(s).", len(windows)
    )
    # Return the per-country frame.
    return frame


# Aggregate the coalition level effects from the per-country means.
def aggregate_level_effects(
    country_window_means: pd.DataFrame,
) -> pd.Series:
    """
    Aggregate the coalition level effects.

    Parameters
    ----------
    country_window_means : pandas.DataFrame
        The per-country window means indexed on the country.

    Returns
    -------
    pandas.Series
        The equal-weighted coalition mean per window, indexed by the window
        label. No weight is applied for population, output or any other size
        measure, and no annualization is performed.

    Raises
    ------
    IngestError
        If the per-country frame is empty or carries a missing cell.
    """
    # Refuse an empty frame.
    if len(country_window_means) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot aggregate effects from an empty country frame.")
    # Duplicate country labels would assign unintended weights.
    if country_window_means.index.has_duplicates:
        raise IngestError("Per-country window means carry duplicate country labels.")
    # Refuse a frame carrying any missing, infinite or nonnumeric cell.
    try:
        values = country_window_means.to_numpy(dtype=float)
    except (TypeError, ValueError) as exc:
        raise IngestError("Per-country window means carry a non-numeric cell.") from exc
    if not bool(np.isfinite(values).all()):
        # The failure states the condition that would bias the mean.
        raise IngestError(
            "Per-country window means carry a missing cell; the coalition mean "
            "would be taken over an incomplete roster."
        )
    # Compute the equal-weighted coalition mean of the per-country means.
    return country_window_means.mean(axis=0)


# Compute the pre-escalation level gap of every target.
def compute_pre_escalation_levels(
    gaps: pd.DataFrame, window: Tuple[int, int] = PRE_ESCALATION_WINDOW
) -> pd.Series:
    """
    Compute the per-country pre-escalation level gaps.

    Parameters
    ----------
    gaps : pandas.DataFrame
        The gap matrix indexed on the country and columned by the years.
    window : tuple of int, optional
        The inclusive pre-escalation window.

    Returns
    -------
    pandas.Series
        The per-country mean gap over the pre-escalation window, indexed on
        the country code.

    Raises
    ------
    IngestError
        If the window leaves the gap matrix's coverage.
    """
    # Validate the full matrix so no missing cell can be skipped by the mean.
    _validate_att_matrix(gaps, "Gap matrix for pre-escalation aggregation")
    # Compose the window's year columns.
    start, end = _validate_att_window(window, "pre-escalation reference window")
    years: List[int] = list(range(start, end + 1))
    # Identify the window years the matrix does not carry.
    missing_years: List[int] = [
        year for year in years if year not in set(int(c) for c in gaps.columns)
    ]
    # Refuse a window that leaves the coverage.
    if missing_years:
        # The failure names the missing years.
        raise IngestError(
            f"Gap matrix does not cover pre-escalation year(s) {missing_years}."
        )
    # Compute the per-country mean over the pre-escalation window.
    levels: pd.Series = gaps[years].mean(axis=1)
    # The reference vector must preserve the exact roster and remain finite.
    if not levels.index.equals(gaps.index) or not bool(
        np.isfinite(levels.to_numpy(dtype=float)).all()
    ):
        raise IngestError("Pre-escalation aggregation changed the roster or emitted non-finite values.")
    # Name the series so its provenance travels with it.
    levels.name = "pre_escalation_level"
    # Log the baseline computation.
    logger.info(
        "pre-escalation levels computed over %d to %d.", window[0], window[1]
    )
    # Return the per-country levels.
    return levels


# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 3: Compute the incremental estimand and verify its identity
# -------------------------------------------------------------------------------------------------------------------------------


# Build the observed outcome window of one partition's targets.
def build_observed_window(
    observed_panel: pd.DataFrame,
    outcome: str,
    countries: Tuple[str, ...],
    window: Tuple[int, int],
) -> pd.DataFrame:
    """
    Build the observed outcome window of a partition's targets.

    Parameters
    ----------
    observed_panel : pandas.DataFrame
        The analysis panel carrying the observed outcome columns.
    outcome : str
        The canonical outcome variable of the partition.
    countries : tuple of str
        The countries whose window is built.
    window : tuple of int
        The inclusive window.

    Returns
    -------
    pandas.DataFrame
        The observed outcomes indexed on the country and columned by the
        window years, with no missing cell.

    Raises
    ------
    IngestError
        If the outcome column is absent, or if the window carries a missing
        observed cell after alignment.
    """
    # Require the complete labelled schema.
    required = {"country_code", "year", outcome}
    missing_columns = sorted(required - set(observed_panel.columns))
    if missing_columns:
        raise IngestError(f"Observed panel lacks window column(s) {missing_columns}.")
    # Resolve the exact population and calendar product.
    roster = _validated_att_roster(countries, f"observed {outcome} window")
    start, end = _validate_att_window(window, f"observed {outcome} window")
    years: List[int] = list(range(start, end + 1))
    expected_keys = {(country, year) for country in roster for year in years}
    validated = _validated_att_years(observed_panel, "Observed panel")
    if bool(validated["country_code"].isna().any()):
        raise IngestError("Observed panel carries a null country code.")
    validated["_att_country"] = validated["country_code"].astype(str).str.strip()
    scoped = validated.loc[
        validated["_att_country"].isin(set(roster))
        & validated["_att_year"].isin(years)
    ].copy()
    keys = list(zip(scoped["_att_country"], scoped["_att_year"].astype(int)))
    duplicate_keys = {key for key in keys if keys.count(key) > 1}
    if duplicate_keys:
        raise IngestError(
            f"Observed window for {outcome} carries duplicate key(s), for "
            f"example {sorted(duplicate_keys)[:MAX_REPORTED_OFFENDERS]}."
        )
    if set(keys) != expected_keys:
        missing = sorted(expected_keys - set(keys))
        raise IngestError(
            f"Observed window for {outcome} is missing target-year key(s), for "
            f"example {missing[:MAX_REPORTED_OFFENDERS]}."
        )
    scoped["_att_value"] = pd.to_numeric(scoped[outcome], errors="coerce")
    if not bool(np.isfinite(scoped["_att_value"].to_numpy(dtype=float)).all()):
        raise IngestError(
            f"Observed window for {outcome} carries a missing or non-finite cell."
        )
    # Pivot after proving uniqueness, then impose the declared axes exactly.
    frame = scoped.pivot(
        index="_att_country", columns="_att_year", values="_att_value"
    ).reindex(index=roster, columns=years)
    frame.index = pd.Index(roster, name="country_code")
    frame.columns = pd.Index(years, name="year")
    _validate_att_matrix(frame, f"Observed {outcome} window")
    # Return the aligned observed window.
    return frame


# Compute the continuation gap under the stable-gap representation.
def compute_continuation_gap(
    gaps: pd.DataFrame,
    pre_escalation_levels: pd.Series,
    observed_window: pd.DataFrame,
    window: Tuple[int, int] = POST_ESCALATION_WINDOW,
) -> pd.DataFrame:
    """
    Compute the continuation gap through the shifted counterfactual.

    Parameters
    ----------
    gaps : pandas.DataFrame
        The gap matrix indexed on the country and columned by the years.
    pre_escalation_levels : pandas.Series
        The per-country pre-escalation level gaps, indexed on the country.
    observed_window : pandas.DataFrame
        The observed outcomes over the post-escalation window, indexed on the
        country, which the continuation counterfactual is subtracted from.
    window : tuple of int, optional
        The inclusive post-escalation window.

    Returns
    -------
    pandas.DataFrame
        The continuation gap matrix over the post-escalation window, computed
        by adding the pre-escalation level to the untreated path and
        subtracting that continuation counterfactual from the observed
        outcome. The dual representation is used deliberately so that the
        algebraic identity with the direct expression is a numerical check
        rather than a restatement.

    Raises
    ------
    IngestError
        If the observed window and the gap matrix do not share their axes.
    """
    # Validate both matrices before using either in the independent route.
    _validate_att_matrix(gaps, "Gap matrix for continuation route")
    _validate_att_matrix(observed_window, "Observed continuation window")
    # Compose the window's year columns.
    start, end = _validate_att_window(window, "post-escalation continuation window")
    years: List[int] = list(range(start, end + 1))
    # Refuse an axis mismatch between the observed window and the gaps.
    if not observed_window.index.equals(gaps.index) or list(
        observed_window.columns
    ) != years:
        # The failure states the divergence.
        raise IngestError(
            "Observed post-escalation window does not align with the gap matrix."
        )
    # The shift vector must be a finite, unique and exactly aligned country map.
    if pre_escalation_levels.index.has_duplicates or not pre_escalation_levels.index.equals(
        gaps.index
    ):
        raise IngestError(
            "Pre-escalation levels do not carry the exact continuation roster."
        )
    try:
        baseline_values = pre_escalation_levels.to_numpy(dtype=float)
    except (TypeError, ValueError) as exc:
        raise IngestError("Pre-escalation levels are non-numeric.") from exc
    if not bool(np.isfinite(baseline_values).all()):
        raise IngestError("Pre-escalation levels carry a missing or non-finite value.")
    # Reconstruct the untreated path as the observed outcome less the gap,
    # which is the inverse of the gap definition.
    untreated: pd.DataFrame = observed_window - gaps[years]
    # Shift the untreated path by the persistent pre-escalation level to form
    # the continuation counterfactual, which is the declared dual
    # representation of the stable-gap continuation.
    continuation: pd.DataFrame = untreated.add(pre_escalation_levels, axis=0)
    # Compute the continuation gap as the observed outcome less the
    # continuation counterfactual.
    continuation_gap: pd.DataFrame = observed_window - continuation
    # Preserve the same fixed country set through the dual representation.
    _validate_att_matrix(continuation_gap, "Continuation gap matrix")
    if not continuation_gap.index.equals(gaps.index) or list(
        continuation_gap.columns
    ) != years:
        raise IngestError("Continuation-gap axes drifted during construction.")
    # Return the continuation gap matrix.
    return continuation_gap


# Compute the incremental effect from the continuation gaps.
def compute_incremental_effect(
    continuation_gaps: pd.DataFrame,
) -> float:
    """
    Compute the incremental coalition effect.

    Parameters
    ----------
    continuation_gaps : pandas.DataFrame
        The continuation gap matrix over the post-escalation window.

    Returns
    -------
    float
        The equal-weighted coalition mean of the post-escalation continuation
        gaps, which is the incremental estimand. Its interpretation rests on
        the stable-gap assumption: an earlier sanctions regime may leave a
        persistent non-zero baseline gap while further drift is ruled out, so
        the incremental quantity measures the change beyond that baseline
        rather than the total level.

    Raises
    ------
    IngestError
        If the matrix is empty or carries a missing cell.
    """
    # Require a complete, unique and finite fixed-roster matrix.
    _validate_att_matrix(continuation_gaps, "Continuation gap matrix")
    # Average across the post-escalation years with equal weight on each year.
    country_means: pd.Series = continuation_gaps.mean(axis=1)
    # Average across countries with equal weight on each country.
    return float(country_means.mean())


# Compute the incremental effect through the direct expression.
def compute_incremental_effect_alternative(
    gaps: pd.DataFrame,
    pre_window: Tuple[int, int] = PRE_ESCALATION_WINDOW,
    post_window: Tuple[int, int] = POST_ESCALATION_WINDOW,
) -> float:
    """
    Compute the incremental effect through the direct expression.

    Parameters
    ----------
    gaps : pandas.DataFrame
        The gap matrix indexed on the country and columned by the years.
    pre_window : tuple of int, optional
        The inclusive pre-escalation window.
    post_window : tuple of int, optional
        The inclusive post-escalation window.

    Returns
    -------
    float
        The equal-weighted coalition mean of the post-escalation mean gap less
        the pre-escalation mean gap, which is algebraically equivalent to the
        continuation-counterfactual route and is implemented separately so
        the identity assertion compares two independent computations.

    Raises
    ------
    IngestError
        If the two windows overlap, which would absorb escalation years into
        the baseline.
    """
    # Validate both windows before checking their scientific ordering.
    pre_start, pre_end = _validate_att_window(
        pre_window, "incremental reference window"
    )
    post_start, post_end = _validate_att_window(
        post_window, "incremental post window"
    )
    # Refuse overlapping windows.
    if not (pre_end < post_start):
        # The failure names both windows.
        raise IngestError(
            f"Pre-escalation window {pre_window} overlaps the post-escalation "
            f"window {post_window}."
        )
    # Compose the pre-escalation year columns.
    pre_years: List[int] = list(range(pre_start, pre_end + 1))
    # Compose the post-escalation year columns.
    post_years: List[int] = list(range(post_start, post_end + 1))
    # Reject incomplete or non-finite inputs before pandas can skip missing cells.
    _validate_att_matrix(gaps, "Gap matrix for direct incremental route")
    required_years = pre_years + post_years
    missing_years = [year for year in required_years if year not in gaps.columns]
    if missing_years:
        raise IngestError(
            f"Gap matrix omits direct incremental year(s) {missing_years}."
        )
    # Compute the per-country post-escalation mean gap.
    post_means: pd.Series = gaps[post_years].mean(axis=1)
    # Compute the per-country pre-escalation mean gap.
    pre_means: pd.Series = gaps[pre_years].mean(axis=1)
    # Compute the coalition mean of the difference.
    return float((post_means - pre_means).mean())


# Assert the identity between the two incremental routes.
def assert_incremental_identity(
    primary: float, alternative: float, tolerance: float = IDENTITY_TOLERANCE
) -> None:
    """
    Assert the identity between the two incremental routes.

    Parameters
    ----------
    primary : float
        The incremental effect computed through the continuation
        counterfactual.
    alternative : float
        The incremental effect computed through the direct difference.
    tolerance : float, optional
        The tolerance of the identity.

    Returns
    -------
    None
        The function returns nothing when the identity holds.

    Raises
    ------
    IngestError
        If the two routes differ by more than the declared tolerance, which
        would indicate that the continuation shift was applied incorrectly.
    """
    # Exact algebraic identities require finite values and a positive finite
    # numerical tolerance; publication precision is deliberately irrelevant.
    if not all(math.isfinite(float(value)) for value in (primary, alternative, tolerance)):
        raise IngestError("Incremental identity inputs and tolerance must be finite.")
    if float(tolerance) <= 0.0:
        raise IngestError("Incremental identity tolerance must be positive.")
    # Compute the absolute difference between the two routes.
    deviation: float = abs(float(primary) - float(alternative))
    # Refuse a violation of the declared identity.
    if deviation > float(tolerance):
        # The failure names both values and the deviation.
        raise IngestError(
            f"Incremental identity violated: continuation route {primary!r} "
            f"against direct route {alternative!r}, deviation {deviation:.3e} "
            f"exceeding the tolerance {tolerance:.1e}."
        )
    # Log the verified identity.
    logger.info(
        "incremental identity verified: deviation %.3e.", deviation
    )


# Compare the computed effects with the declared manuscript values.
def assert_declared_effects(
    incremental: pd.DataFrame,
    level_effects: pd.DataFrame,
    expected_incremental: Optional[Mapping[Tuple[str, str], Any]] = None,
    expected_levels: Optional[Mapping[Tuple[str, str], Mapping[str, Any]]] = None,
) -> pd.DataFrame:
    """
    Compare the computed effects with the declared manuscript values.

    Parameters
    ----------
    incremental : pandas.DataFrame
        The incremental effects keyed on the pair.
    level_effects : pandas.DataFrame
        The coalition level effects keyed on the pair.
    expected_incremental : mapping of tuple to object, optional
        The declared incremental cells per pair. A scalar declaration remains
        supported with the manuscript's default three-decimal precision. A
        structured declaration may supply ``value`` (or
        ``published_value``), ``units``, ``displayed_decimals`` and
        ``numerical_allowance``.
    expected_levels : mapping of tuple to mapping of str to object, optional
        The declared level cells per pair and wave label, using the same
        scalar or structured cell contract.

    Returns
    -------
    pandas.DataFrame
        One row per supplied reported cell. Each row identifies the pair and
        estimand, units, displayed precision, publication rounding interval,
        explicit numerical allowance, preserved unrounded reproduction and
        signed/absolute deviations. An empty, schema-stable frame is returned
        when no manuscript values are supplied.

    Raises
    ------
    IngestError
        If a declaration is malformed, a reported cell is absent or an
        unrounded reproduction lies outside the publication-aware interval.
    """
    # Declare the stable audit schema even when comparison is not requested.
    audit_columns = [
        "cell_id", "policy", "outcome", "estimand", "units",
        "displayed_decimals", "reported_value", "unrounded_value",
        "rounding_lower", "rounding_upper", "numerical_allowance",
        "accepted_lower", "accepted_upper", "signed_deviation",
        "absolute_deviation", "within_rounding_interval",
    ]
    # Reject duplicate pair keys and absent authoritative output columns.
    if incremental.index.has_duplicates:
        raise IngestError("Incremental effect frame carries duplicate pair keys.")
    if level_effects.index.has_duplicates:
        raise IngestError("Level effect frame carries duplicate pair keys.")
    if "incremental_effect" not in incremental.columns:
        raise IngestError("Incremental effect frame omits incremental_effect.")

    # Infer scientifically explicit units for legacy scalar declarations.
    def inferred_units(outcome: str) -> str:
        # Log-level outcomes are reported in log points; rate outcomes in
        # percentage points under the task's canonical outcome definitions.
        if outcome in {"log_CPI", "log_real_GDPpc"}:
            return "log_points"
        if outcome in {
            "inflation_rate", "GDPpc_growth_rate", "inflation_att", "growth_att"
        }:
            return "decimal_rate (multiply by 100 for percentage points)"
        return "outcome_units"

    # Parse one publication cell without conflating rounding with identities.
    def parse_cell(
        raw: Any, pair: Tuple[str, str], estimand: str
    ) -> Tuple[float, str, int, float]:
        # Structured declarations can state publication metadata explicitly.
        if isinstance(raw, Mapping):
            if "value" in raw:
                value_raw = raw["value"]
            elif "published_value" in raw:
                value_raw = raw["published_value"]
            else:
                raise IngestError(
                    f"{pair}/{estimand}: publication declaration omits value."
                )
            units_raw = raw.get("units", inferred_units(pair[1]))
            precision_raw = raw.get("displayed_decimals")
            allowance_raw = raw.get("numerical_allowance", 1.0e-12)
        else:
            value_raw = raw
            units_raw = inferred_units(pair[1])
            precision_raw = None
            allowance_raw = 1.0e-12
        # Booleans are not numerical manuscript estimates.
        if isinstance(value_raw, (bool, np.bool_)):
            raise IngestError(f"{pair}/{estimand}: reported value is Boolean.")
        try:
            reported = float(value_raw)
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError(
                f"{pair}/{estimand}: reported value is not numeric."
            ) from exc
        if not math.isfinite(reported):
            raise IngestError(f"{pair}/{estimand}: reported value is non-finite.")
        # A nonblank unit is part of each cell's publication identity.
        units = str(units_raw).strip() if units_raw is not None else ""
        if not units:
            raise IngestError(f"{pair}/{estimand}: publication units are blank.")
        # Use the manuscript's documented three-decimal default because a
        # Python float cannot retain printed trailing zeroes (for example,
        # 0.070 becomes 0.07). Structured declarations may override it.
        if precision_raw is None:
            displayed_decimals = 3
        else:
            if isinstance(precision_raw, (bool, np.bool_)):
                raise IngestError(
                    f"{pair}/{estimand}: displayed precision is Boolean."
                )
            try:
                precision_number = float(precision_raw)
            except (TypeError, ValueError, OverflowError) as exc:
                raise IngestError(
                    f"{pair}/{estimand}: displayed precision is malformed."
                ) from exc
            if (
                not math.isfinite(precision_number)
                or not precision_number.is_integer()
                or precision_number < 0.0
            ):
                raise IngestError(
                    f"{pair}/{estimand}: displayed precision must be a "
                    "nonnegative integer."
                )
            displayed_decimals = int(precision_number)
        # The numerical allowance is separate from the publication interval.
        if isinstance(allowance_raw, (bool, np.bool_)):
            raise IngestError(f"{pair}/{estimand}: numerical allowance is Boolean.")
        try:
            allowance = float(allowance_raw)
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError(
                f"{pair}/{estimand}: numerical allowance is malformed."
            ) from exc
        if not math.isfinite(allowance) or allowance < 0.0:
            raise IngestError(
                f"{pair}/{estimand}: numerical allowance must be finite and "
                "nonnegative."
            )
        return reported, units, displayed_decimals, allowance

    # Record and validate one unrounded reproduction against one printed cell.
    audit_rows: List[Dict[str, Any]] = []

    def compare_cell(
        pair: Tuple[str, str], estimand: str, unrounded_raw: Any, declared: Any
    ) -> None:
        reported, units, decimals, allowance = parse_cell(declared, pair, estimand)
        try:
            unrounded = float(unrounded_raw)
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError(
                f"{pair}/{estimand}: unrounded reproduction is non-numeric."
            ) from exc
        if not math.isfinite(unrounded):
            raise IngestError(
                f"{pair}/{estimand}: unrounded reproduction is non-finite."
            )
        half_unit = 0.5 * (10.0 ** (-decimals))
        rounding_lower = reported - half_unit
        rounding_upper = reported + half_unit
        accepted_lower = rounding_lower - allowance
        accepted_upper = rounding_upper + allowance
        signed_deviation = unrounded - reported
        within = accepted_lower <= unrounded <= accepted_upper
        row = {
            "cell_id": f"{pair[0]}/{pair[1]}/{estimand}",
            "policy": pair[0],
            "outcome": pair[1],
            "estimand": estimand,
            "units": units,
            "displayed_decimals": decimals,
            "reported_value": reported,
            "unrounded_value": unrounded,
            "rounding_lower": rounding_lower,
            "rounding_upper": rounding_upper,
            "numerical_allowance": allowance,
            "accepted_lower": accepted_lower,
            "accepted_upper": accepted_upper,
            "signed_deviation": signed_deviation,
            "absolute_deviation": abs(signed_deviation),
            "within_rounding_interval": bool(within),
        }
        audit_rows.append(row)
        if not within:
            raise IngestError(
                f"{row['cell_id']}: unrounded reproduction {unrounded!r} lies "
                f"outside publication interval [{rounding_lower!r}, "
                f"{rounding_upper!r}] with numerical allowance {allowance!r}."
            )

    # Normalize a publication mapping key to the canonical policy/outcome pair.
    def normalize_pair(raw_pair: Any, frame_name: str) -> Tuple[str, str]:
        if (
            not isinstance(raw_pair, tuple)
            or len(raw_pair) != 2
            or any(value is None or not str(value).strip() for value in raw_pair)
        ):
            raise IngestError(
                f"{frame_name} publication key {raw_pair!r} is not a valid pair."
            )
        return str(raw_pair[0]).strip(), str(raw_pair[1]).strip()

    # Compare every supplied incremental cell by its complete labelled key.
    if expected_incremental is not None:
        for raw_pair, declared in expected_incremental.items():
            pair = normalize_pair(raw_pair, "Incremental")
            if pair not in incremental.index:
                raise IngestError(f"Incremental frame omits the pair {raw_pair}.")
            compare_cell(
                (str(pair[0]), str(pair[1])),
                "incremental",
                incremental.loc[pair, "incremental_effect"],
                declared,
            )
    # Compare every supplied wave-level cell independently.
    if expected_levels is not None:
        for raw_pair, declarations in expected_levels.items():
            pair = normalize_pair(raw_pair, "Level")
            if pair not in level_effects.index:
                raise IngestError(f"Level frame omits the pair {raw_pair}.")
            if not isinstance(declarations, Mapping) or not declarations:
                raise IngestError(f"{raw_pair}: level declarations are empty or malformed.")
            for label, declared in declarations.items():
                if label not in level_effects.columns:
                    raise IngestError(f"{raw_pair}: level frame omits estimand {label!r}.")
                compare_cell(
                    (str(pair[0]), str(pair[1])),
                    str(label),
                    level_effects.loc[pair, label],
                    declared,
                )
    # Assemble a stable, uniquely identified audit ledger.
    audit = pd.DataFrame(audit_rows, columns=audit_columns)
    if len(audit) and bool(audit["cell_id"].duplicated().any()):
        raise IngestError("Publication audit carries duplicate reported-cell identities.")
    # Log the comparison mode.
    logger.info(
        "declared effect comparison %s for %d cell(s).",
        "performed" if len(audit) else "skipped",
        len(audit),
    )
    # Return deviations as auditable data while leaving source estimates unrounded.
    return audit


# Freeze the gap objects behind a canonical digest.
def freeze_gap_objects(
    gaps: Mapping[Tuple[str, str], pd.DataFrame],
    country_window_means: pd.DataFrame,
    incremental_effects: pd.DataFrame,
) -> str:
    """
    Compute the canonical digest of the gap objects.

    Parameters
    ----------
    gaps : mapping of tuple to pandas.DataFrame
        The gap matrices keyed on the pair.
    country_window_means : pandas.DataFrame
        The per-country window means.
    incremental_effects : pandas.DataFrame
        The incremental effect frame.

    Returns
    -------
    str
        The hexadecimal digest of the canonical payload.

    Raises
    ------
    IngestError
        If a payload component is empty.
    """
    # Refuse an empty gap mapping.
    if not gaps:
        # The failure states the precondition.
        raise IngestError("Cannot freeze an empty gap mapping.")
    # Refuse an empty means frame.
    if len(country_window_means) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot freeze an empty country window means frame.")
    # Accumulate the canonical payload parts.
    parts: List[str] = []
    # Record every gap cell in sorted pair, country and year order.
    for pair in sorted(gaps.keys()):
        # Read the partition's gap matrix.
        matrix: pd.DataFrame = gaps[pair]
        # Iterate over the countries in sorted order.
        for country in sorted(str(code) for code in matrix.index):
            # Iterate over the years in ascending order.
            for year in sorted(int(column) for column in matrix.columns):
                # Append the gap cell at full precision.
                parts.append(
                    "G|"
                    + pair[0]
                    + "|"
                    + pair[1]
                    + "|"
                    + country
                    + "|"
                    + str(year)
                    + "|"
                    + f"{float(matrix.loc[country, year]):.17g}"
                )
    # Record every per-country window mean in sorted order.
    for _, row in country_window_means.sort_values(
        ["policy", "outcome", "country_code"], kind="stable"
    ).iterrows():
        # Append the window means of the country.
        parts.append(
            "W|"
            + str(row["policy"])
            + "|"
            + str(row["outcome"])
            + "|"
            + str(row["country_code"])
            + "|"
            + "|".join(
                f"{float(row[label]):.17g}"
                for label in (WAVE_ONE_LABEL, WAVE_TWO_LABEL)
            )
        )
    # Record every incremental effect in sorted order, reading the pair from
    # the frame's own index because the pair is the index rather than a
    # column.
    for (policy, outcome), row in incremental_effects.sort_index().iterrows():
        # Bind the unrounded dual routes, fixed roster and explicit reference
        # windows so an equal number under a different scientific contract
        # cannot share a digest.
        required_metadata = {
            "incremental_effect", "incremental_direct_route", "identity_deviation",
            "gaps_countries", "gaps_years", "country_codes", "roster_sha256",
            "reference_start", "reference_end", "post_start", "post_end",
            "wave_one_start", "wave_one_end", "wave_two_start", "wave_two_end",
        }
        missing_metadata = sorted(required_metadata - set(incremental_effects.columns))
        if missing_metadata:
            raise IngestError(
                f"Incremental effect frame omits digest metadata {missing_metadata}."
            )
        roster_codes = tuple(str(code) for code in row["country_codes"])
        parts.append(
            "I|" + str(policy) + "|" + str(outcome) + "|"
            + f"{float(row['incremental_effect']):.17g}|"
            + f"{float(row['incremental_direct_route']):.17g}|"
            + f"{float(row['identity_deviation']):.17g}|"
            + str(int(row["gaps_countries"])) + "|"
            + str(int(row["gaps_years"])) + "|"
            + ",".join(roster_codes) + "|"
            + str(row["roster_sha256"]) + "|"
            + str(int(row["reference_start"])) + "|"
            + str(int(row["reference_end"])) + "|"
            + str(int(row["post_start"])) + "|"
            + str(int(row["post_end"])) + "|"
            + str(int(row["wave_one_start"])) + "|"
            + str(int(row["wave_one_end"])) + "|"
            + str(int(row["wave_two_start"])) + "|"
            + str(int(row["wave_two_end"]))
        )
    # Digest the encoded payload.
    return hashlib.sha256("\n".join(parts).encode("utf-8")).hexdigest()


# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the gap and effect computation across the partitions.
def orchestrate_gap_and_att(
    counterfactual_result: CounterfactualResult,
    observed_panel: pd.DataFrame,
    registry: PartitionRegistry,
    wave_one: Tuple[int, int] = WAVE_ONE_WINDOW,
    wave_two: Tuple[int, int] = WAVE_TWO_WINDOW,
    pre_escalation: Tuple[int, int] = PRE_ESCALATION_WINDOW,
    post_escalation: Tuple[int, int] = POST_ESCALATION_WINDOW,
    expected_incremental: Optional[Mapping[Tuple[str, str], Any]] = None,
    expected_levels: Optional[Mapping[Tuple[str, str], Mapping[str, Any]]] = None,
) -> GapAttResult:
    """
    Compute the gap matrices, the level effects and the incremental effect.

    Parameters
    ----------
    counterfactual_result : CounterfactualResult
        The counterfactual paths produced by the construction task.
    observed_panel : pandas.DataFrame
        The analysis panel carrying the observed outcome columns.
    registry : PartitionRegistry
        The registered partitions, whose target tuples define the roster.
    wave_one : tuple of int, optional
        The inclusive first wave window.
    wave_two : tuple of int, optional
        The inclusive second wave window.
    pre_escalation : tuple of int, optional
        The inclusive pre-escalation window.
    post_escalation : tuple of int, optional
        The inclusive post-escalation window.
    expected_incremental : mapping of tuple to object, optional
        The declared incremental publication cells per pair, either scalars
        at the documented three-decimal manuscript precision or structured
        cell declarations carrying value, units, precision and allowance.
    expected_levels : mapping of tuple to mapping of str to object, optional
        The declared level publication cells per pair and wave label under
        the same comparison contract.

    Returns
    -------
    GapAttResult
        The gap matrices, the per-country means, the level effects, the
        pre-escalation levels, the unrounded incremental effects, the
        cell-level publication audit, the digest, the timestamp and notes.
        Fitted pre-period paths remain the distinct visualization product on
        ``counterfactual_result.fitted_pre_paths`` and are never merged into
        this post-period ATT contract.

    Raises
    ------
    IngestError
        If any gap, aggregation, identity or comparison check fails.
    """
    # Validate all reporting windows once at the orchestration boundary.
    wave_one = _validate_att_window(wave_one, "wave-one level window")
    wave_two = _validate_att_window(wave_two, "wave-two level window")
    pre_escalation = _validate_att_window(
        pre_escalation, "incremental reference window"
    )
    post_escalation = _validate_att_window(
        post_escalation, "incremental post window"
    )
    if wave_one[1] >= wave_two[0]:
        raise IngestError("Wave-one and wave-two level windows overlap or are unordered.")
    if pre_escalation[1] >= post_escalation[0]:
        raise IngestError("Incremental reference and post windows overlap or are unordered.")
    if not registry.partitions:
        raise IngestError("Cannot compute gap effects from an empty partition registry.")
    # Gap coverage is the union span needed by both levels and the reference
    # contrast; the fitted-pre visualization product remains separate.
    gap_start = min(wave_one[0], wave_two[0], pre_escalation[0], post_escalation[0])
    gap_end = max(wave_one[1], wave_two[1], pre_escalation[1], post_escalation[1])
    # Initialise the gap mapping keyed on the pair.
    gaps: Dict[Tuple[str, str], pd.DataFrame] = {}
    # Accumulate the per-country window mean rows.
    country_mean_rows: List[Dict[str, Any]] = []
    # Accumulate the level effect rows.
    level_rows: List[Dict[str, Any]] = []
    # Accumulate the pre-escalation level rows.
    pre_level_rows: List[Dict[str, Any]] = []
    # Accumulate the incremental effect rows.
    incremental_rows: List[Dict[str, Any]] = []
    # Accumulate the recorded notes.
    notes: List[str] = [
        "gaps are total differences at each year, not incremental differences "
        "across waves.",
        "no weight is applied for population, output or any size measure, and "
        "no annualization is performed.",
    ]
    # Iterate over the registry's partitions in sorted pair order.
    for pair in sorted(registry.partitions.keys()):
        # Refuse a partition the counterfactual result does not cover.
        if pair not in counterfactual_result.loadings:
            # The failure names the missing pair.
            raise IngestError(f"Counterfactual result does not cover the pair {pair}.")
        # Select the partition of the registry.
        partition = registry.partitions[pair]
        # Bind the mapping key and the partition's own labelled identity.
        if (str(partition.policy), str(partition.outcome)) != (
            str(pair[0]), str(pair[1])
        ):
            raise IngestError(f"Registry partition identity conflicts with key {pair}.")
        # Resolve the unique fixed target set used by every estimand route.
        fixed_roster = _validated_att_roster(partition.targets, f"{pair[0]}/{pair[1]}")
        # Compute the partition's gap matrix.
        gap_matrix: pd.DataFrame = compute_gap_matrix(
            observed_panel,
            pair[1],
            fixed_roster,
            counterfactual_result.outcome_paths,
            pair[0],
            window=(gap_start, gap_start),
            window_end=(gap_end, gap_end),
        )
        # The aligned gap matrix must retain precisely the registered countries.
        expected_index = pd.Index(fixed_roster, name="country_code")
        if not gap_matrix.index.equals(expected_index):
            raise IngestError(f"{pair}: gap construction changed the target roster.")
        # Record the gap matrix of the partition.
        gaps[pair] = gap_matrix
        # Aggregate the per-country window means.
        country_means: pd.DataFrame = aggregate_country_window_means(
            gap_matrix, {WAVE_ONE_LABEL: wave_one, WAVE_TWO_LABEL: wave_two}
        )
        if not country_means.index.equals(expected_index):
            raise IngestError(f"{pair}: level aggregation changed the target roster.")
        # Record the per-country window means.
        for country in country_means.index:
            # Append the country's window means.
            country_mean_rows.append(
                {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "country_code": str(country),
                    WAVE_ONE_LABEL: float(country_means.loc[country, WAVE_ONE_LABEL]),
                    WAVE_TWO_LABEL: float(country_means.loc[country, WAVE_TWO_LABEL]),
                }
            )
        # Aggregate the coalition level effects.
        levels: pd.Series = aggregate_level_effects(country_means)
        # Record the coalition level effects of the partition.
        level_rows.append(
            {
                "policy": pair[0],
                "outcome": pair[1],
                WAVE_ONE_LABEL: float(levels[WAVE_ONE_LABEL]),
                WAVE_TWO_LABEL: float(levels[WAVE_TWO_LABEL]),
            }
        )
        # Compute the pre-escalation level gaps.
        pre_levels: pd.Series = compute_pre_escalation_levels(
            gap_matrix, pre_escalation
        )
        if not pre_levels.index.equals(expected_index):
            raise IngestError(f"{pair}: reference aggregation changed the target roster.")
        # Record the per-country pre-escalation levels.
        for country in pre_levels.index:
            # Append the country's pre-escalation level.
            pre_level_rows.append(
                {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "country_code": str(country),
                    "pre_escalation_level": float(pre_levels.loc[country]),
                }
            )
        # Build the observed post-escalation window from the panel.
        observed_window: pd.DataFrame = build_observed_window(
            observed_panel,
            pair[1],
            tuple(str(code) for code in gap_matrix.index),
            post_escalation,
        )
        if not observed_window.index.equals(expected_index):
            raise IngestError(f"{pair}: observed post window changed the target roster.")
        # Compute the continuation gap through the shifted counterfactual,
        # which is the declared dual representation of the incremental
        # estimand.
        continuation_gap: pd.DataFrame = compute_continuation_gap(
            gap_matrix, pre_levels, observed_window, post_escalation
        )
        if not continuation_gap.index.equals(expected_index):
            raise IngestError(f"{pair}: continuation route changed the target roster.")
        # Compute the incremental effect through the continuation route.
        incremental_primary: float = compute_incremental_effect(continuation_gap)
        # Compute the incremental effect through the direct route.
        incremental_direct: float = compute_incremental_effect_alternative(
            gap_matrix, pre_escalation, post_escalation
        )
        # Assert the identity between the two routes.
        assert_incremental_identity(incremental_primary, incremental_direct)
        # Record the exact fixed population and reference definitions alongside
        # the unrounded estimate so downstream inference cannot drift rosters.
        roster_fingerprint = hashlib.sha256(
            "\n".join(fixed_roster).encode("utf-8")
        ).hexdigest()
        # Record the incremental effect of the partition with both routes.
        incremental_rows.append(
            {
                "policy": pair[0],
                "outcome": pair[1],
                "incremental_effect": incremental_primary,
                "incremental_direct_route": incremental_direct,
                "identity_deviation": abs(incremental_primary - incremental_direct),
                "gaps_countries": int(gap_matrix.shape[0]),
                "gaps_years": int(gap_matrix.shape[1]),
                "country_codes": fixed_roster,
                "roster_sha256": roster_fingerprint,
                "reference_start": int(pre_escalation[0]),
                "reference_end": int(pre_escalation[1]),
                "post_start": int(post_escalation[0]),
                "post_end": int(post_escalation[1]),
                "wave_one_start": int(wave_one[0]),
                "wave_one_end": int(wave_one[1]),
                "wave_two_start": int(wave_two[0]),
                "wave_two_end": int(wave_two[1]),
            }
        )
    # Assemble the per-country window means frame.
    country_window_means: pd.DataFrame = pd.DataFrame(country_mean_rows)
    # Assemble the level effects frame keyed on the pair.
    level_effects: pd.DataFrame = pd.DataFrame(level_rows).set_index(
        ["policy", "outcome"]
    )
    # Assemble the pre-escalation levels frame.
    pre_escalation_levels: pd.DataFrame = pd.DataFrame(pre_level_rows)
    # Assemble the incremental effects frame keyed on the pair.
    incremental_effects: pd.DataFrame = pd.DataFrame(incremental_rows).set_index(
        ["policy", "outcome"]
    )
    # Compare manuscript cells under publication intervals while preserving all
    # authoritative estimates above at full precision.
    publication_audit = assert_declared_effects(
        incremental_effects,
        level_effects,
        expected_incremental=expected_incremental,
        expected_levels=expected_levels,
    )
    # Record the comparison mode note.
    notes.append(
        "declared effect values compared."
        if (expected_incremental or expected_levels)
        else "declared effect values not supplied; comparison skipped."
    )
    # Record the identity note.
    notes.append(
        "the incremental estimand is computed through the continuation "
        "counterfactual and asserted equal to the direct route to the declared "
        "tolerance; its interpretation rests on the stable-gap assumption."
    )
    # Compute the canonical digest of the gap objects.
    fingerprint: str = freeze_gap_objects(
        gaps, country_window_means, incremental_effects
    )
    # Assemble the immutable result.
    result: GapAttResult = GapAttResult(
        gaps=gaps,
        country_window_means=country_window_means,
        level_effects=level_effects,
        pre_escalation_levels=pre_escalation_levels,
        incremental_effects=incremental_effects,
        publication_audit=publication_audit,
        fingerprint_sha256=fingerprint,
        computed_utc=current_utc_iso(),
        notes=tuple(notes),
    )
    # Log the completed computation with its digest prefix.
    logger.info(
        "gap and effect computation complete: %d partition(s), fingerprint %s.",
        len(gaps),
        fingerprint[:16],
    )
    # Return the assembled result.
    return result
````

````python
# Task 14: Country-Dispersion Inference (Equation 16)

# ==============================================================================
# Task 14: Country-Dispersion Inference (Equation 16)
# ==============================================================================

# Declare the incremental estimand label, which completes the estimand triple.
ESTIMAND_INCREMENTAL: Final[str] = "incremental"
# Compose the declared estimand order.
ESTIMAND_ORDER: Final[Tuple[str, str, str]] = (
    WAVE_ONE_LABEL,
    WAVE_TWO_LABEL,
    ESTIMAND_INCREMENTAL,
)
# Declare the sample-correction parameter of the dispersion estimator.
DISPERSION_DDOF: Final[int] = 1
# Declare the minimum target count the corrected deviation requires.
MINIMUM_TARGETS: Final[int] = 2
# Declare the three significance levels with their markers in reporting order.
SIGNIFICANCE_LEVELS: Final[Tuple[Tuple[float, str], ...]] = (
    (0.01, "***"),
    (0.05, "**"),
    (0.10, "*"),
)
# Declare the marker of a statistic that fails every level.
STAR_NONE: Final[str] = ""
# Declare the source label recorded on every marker, which is the mechanism
# that enforces the specification's isolation requirement.
STAR_SOURCE: Final[str] = "eq16_student_t_only"
# Declare the primary probability procedure recorded independently of stars.
PRIMARY_PROBABILITY_SOURCE: Final[str] = "eq16_two_sided_student_t"
# Declare the estimator and tail identifiers bound into every ledger row.
COUNTRY_DISPERSION_ESTIMATOR: Final[str] = "eq16_country_dispersion"
STUDENT_T_TAIL_CONVENTION: Final[str] = "two_sided_student_t"
# Declare the tolerance of the probability's reproducibility assertion.
REPRODUCIBILITY_TOLERANCE: Final[float] = 1.0e-12


# Declare the immutable result of the country-dispersion inference.
@dataclass(frozen=True)
class InferenceResult:
    """
    Immutable result of the country-dispersion inference.

    Attributes
    ----------
    ledger : pandas.DataFrame
        One row per partition and estimand carrying the effect, the standard
        error, the statistic, the two-sided probability, the marker, the
        degrees of freedom and the marker's source.
    contrasts : dict
        The per-country contrast vectors keyed on the partition and the
        estimand, which the leave-one-out diagnostics consume.
    critical_values : pandas.DataFrame
        The Student-t critical values of each estimand at the three declared
        significance levels, indexed on the estimand.
    publication_audit : pandas.DataFrame
        One row per supplied manuscript inference cell, recording procedure,
        reference window, sample size, units, displayed precision, rounding
        interval, numerical and Monte Carlo allowances, the unrounded value,
        deviation and comparison classification.
    ledger_fingerprint_sha256 : str
        Digest of the canonical ledger payload bound to the input gap
        fingerprint.
    input_gap_fingerprint : str
        The fingerprint of the gap result the inference was computed from.
    computed_utc : str
        ISO-8601 UTC timestamp of the inference.
    notes : tuple of str
        Recorded decisions, including the dispersion limitation and the
        isolation of the marker source.
    """
    # One row per partition and estimand carrying the inference quantities.
    ledger: pd.DataFrame
    # The per-country contrast vectors keyed on the pair and the estimand.
    contrasts: Mapping[Tuple[Tuple[str, str], str], pd.Series]
    # The Student-t critical values indexed on the estimand.
    critical_values: pd.DataFrame
    # Cell-level publication-aware inference comparison ledger.
    publication_audit: pd.DataFrame
    # Digest of the canonical ledger payload bound to the gap fingerprint.
    ledger_fingerprint_sha256: str
    # The fingerprint of the input gap result.
    input_gap_fingerprint: str
    # ISO-8601 UTC timestamp of the inference.
    computed_utc: str
    # Recorded decisions of the inference.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 1: Compute the country-dispersion standard error
# -------------------------------------------------------------------------------------------------------------------------------


# Build the per-country contrast vectors of a partition.
def build_contrast_vectors(
    gaps: pd.DataFrame,
    country_window_means: pd.DataFrame,
    pre_escalation_levels: pd.DataFrame,
    policy: str,
    outcome: str,
    pre_window: Tuple[int, int],
    post_window: Tuple[int, int],
    expected_count: Optional[int] = None,
    quarantine_log: Optional[List[Dict[str, Any]]] = None,
    alternate_sample_design: Optional[str] = None,
    wave_one_window: Optional[Tuple[int, int]] = None,
    wave_two_window: Optional[Tuple[int, int]] = None,
) -> Dict[str, pd.Series]:
    """
    Build the per-country contrast vectors of a partition.

    Parameters
    ----------
    gaps : pandas.DataFrame
        The partition's gap matrix indexed on the country and columned by the
        projection years.
    country_window_means : pandas.DataFrame
        The per-country window means carrying the pair and the country.
    pre_escalation_levels : pandas.DataFrame
        The per-country pre-escalation levels carrying the pair and the
        country.
    policy : str
        The policy instrument of the partition.
    outcome : str
        The canonical outcome variable of the partition.
    pre_window : tuple of int
        The inclusive pre-escalation window of the incremental contrast.
    post_window : tuple of int
        The inclusive post-escalation window of the incremental contrast.
    expected_count : int, optional
        The declared minimum retained target count per estimand. When absent,
        the floor defaults to the full scoped target roster, so any quarantine
        without an explicit relaxation raises.
    quarantine_log : list of dict, optional
        The caller's quarantine audit sink.
    alternate_sample_design : str, optional
        A nonblank registered design name authorizing one common complete-case
        target subset. Without it, every declared target must have a complete
        finite history and no quarantine is permitted.
    wave_one_window : tuple of int, optional
        The actual Task 13 first-wave window. The canonical window is used
        only for backward-compatible direct calls.
    wave_two_window : tuple of int, optional
        The actual Task 13 second-wave window. The canonical window is used
        only for backward-compatible direct calls.

    Returns
    -------
    dict
        The contrast vectors keyed on the estimand label, each indexed on the
        target country code.

    Raises
    ------
    IngestError
        If labelled rosters or windows diverge, a contributing country lacks
        a complete finite history, quarantine is undeclared, or the common
        retained roster differs from the declared exact count.
    """
    # Validate the exact, ordered reference and post calendar definitions.
    pre_start, pre_end = _validate_att_window(pre_window, "inference reference window")
    post_start, post_end = _validate_att_window(post_window, "inference post window")
    if pre_end >= post_start:
        raise IngestError(
            f"{policy}/{outcome}: inference reference and post windows overlap."
        )
    pre_years = list(range(pre_start, pre_end + 1))
    post_years = list(range(post_start, post_end + 1))
    # Resolve the actual level windows so per-country means can be independently
    # reconstructed from the labelled gap histories.
    wave_one_resolved = _validate_att_window(
        WAVE_ONE_WINDOW if wave_one_window is None else wave_one_window,
        "inference wave-one window",
    )
    wave_two_resolved = _validate_att_window(
        WAVE_TWO_WINDOW if wave_two_window is None else wave_two_window,
        "inference wave-two window",
    )
    wave_one_years = list(range(wave_one_resolved[0], wave_one_resolved[1] + 1))
    wave_two_years = list(range(wave_two_resolved[0], wave_two_resolved[1] + 1))
    # Require unique labelled axes and valid calendar labels before averaging.
    if gaps.empty or gaps.index.has_duplicates or gaps.columns.has_duplicates:
        raise IngestError(f"{policy}/{outcome}: gap matrix has empty or duplicate axes.")
    roster = _validated_att_roster(
        tuple(str(code) for code in gaps.index), f"{policy}/{outcome} inference"
    )
    expected_index = pd.Index(roster, name="country_code")
    # The gap matrix must already use the canonical sorted target roster.
    normalized_gap_index = pd.Index(
        [str(code).strip() for code in gaps.index], name="country_code"
    )
    if not normalized_gap_index.equals(expected_index):
        raise IngestError(f"{policy}/{outcome}: gap roster is not canonical and fixed.")
    # Validate all year labels without lossy conversion and require every
    # declared reference/post year explicitly.
    gap_years: List[int] = []
    for raw_year in gaps.columns:
        if isinstance(raw_year, (bool, np.bool_)):
            raise IngestError(f"{policy}/{outcome}: gap matrix has a Boolean year.")
        try:
            numeric_year = float(raw_year)
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError(f"{policy}/{outcome}: gap matrix has a nonnumeric year.") from exc
        if not math.isfinite(numeric_year) or not numeric_year.is_integer():
            raise IngestError(f"{policy}/{outcome}: invalid gap year {raw_year!r}.")
        gap_years.append(int(numeric_year))
    all_required_years = list(
        dict.fromkeys(pre_years + post_years + wave_one_years + wave_two_years)
    )
    missing_years = [year for year in all_required_years if year not in set(gap_years)]
    if missing_years:
        raise IngestError(
            f"{policy}/{outcome}: gap matrix omits required inference years {missing_years}."
        )
    # Require schemas and unique pair-country keys in both upstream summaries.
    means_required = {"policy", "outcome", "country_code", WAVE_ONE_LABEL, WAVE_TWO_LABEL}
    levels_required = {"policy", "outcome", "country_code", "pre_escalation_level"}
    if not means_required.issubset(country_window_means.columns):
        raise IngestError(f"{policy}/{outcome}: country-window means schema is incomplete.")
    if not levels_required.issubset(pre_escalation_levels.columns):
        raise IngestError(f"{policy}/{outcome}: pre-escalation schema is incomplete.")
    scoped_means = country_window_means.loc[
        (country_window_means["policy"].astype(str) == str(policy))
        & (country_window_means["outcome"].astype(str) == str(outcome))
    ].copy()
    scoped_levels = pre_escalation_levels.loc[
        (pre_escalation_levels["policy"].astype(str) == str(policy))
        & (pre_escalation_levels["outcome"].astype(str) == str(outcome))
    ].copy()
    if scoped_means.empty or scoped_levels.empty:
        raise IngestError(f"{policy}/{outcome}: per-country summaries are absent.")
    if bool(scoped_means["country_code"].duplicated().any()) or bool(
        scoped_levels["country_code"].duplicated().any()
    ):
        raise IngestError(f"{policy}/{outcome}: per-country summaries have duplicate keys.")
    scoped_means["country_code"] = scoped_means["country_code"].astype(str).str.strip()
    scoped_levels["country_code"] = scoped_levels["country_code"].astype(str).str.strip()
    if bool(scoped_means["country_code"].duplicated().any()) or bool(
        scoped_levels["country_code"].duplicated().any()
    ):
        raise IngestError(
            f"{policy}/{outcome}: canonicalized summary country keys are duplicated."
        )
    if set(scoped_means["country_code"]) != set(roster) or set(
        scoped_levels["country_code"]
    ) != set(roster):
        raise IngestError(f"{policy}/{outcome}: summary roster differs from the gap roster.")
    scoped_means = scoped_means.set_index("country_code").reindex(roster)
    scoped_levels = scoped_levels.set_index("country_code").reindex(roster)
    scoped_means.index.name = "country_code"
    scoped_levels.index.name = "country_code"
    # Convert every required component without permitting skip-NaN reductions.
    try:
        first = pd.to_numeric(scoped_means[WAVE_ONE_LABEL], errors="raise").astype(float)
        second = pd.to_numeric(scoped_means[WAVE_TWO_LABEL], errors="raise").astype(float)
        gap_required = gaps.loc[:, all_required_years].astype(float)
        supplied_pre = pd.to_numeric(
            scoped_levels["pre_escalation_level"], errors="raise"
        ).astype(float)
    except (TypeError, ValueError, KeyError) as exc:
        raise IngestError(f"{policy}/{outcome}: inference components are nonnumeric.") from exc
    # Determine one common complete-case mask across every estimand component.
    complete = pd.Series(
        np.isfinite(first.to_numpy())
        & np.isfinite(second.to_numpy())
        & np.isfinite(gap_required.to_numpy()).all(axis=1)
        & np.isfinite(supplied_pre.to_numpy()),
        index=expected_index,
    )
    incomplete_countries = tuple(str(code) for code in complete.index[~complete])
    # Any exclusion requires an explicit alternate-sample design name.
    design = str(alternate_sample_design).strip() if alternate_sample_design is not None else ""
    if incomplete_countries and not design:
        raise IngestError(
            f"{policy}/{outcome}: incomplete finite histories for "
            f"{list(incomplete_countries)}; quarantine requires an explicitly "
            "declared alternate-sample design."
        )
    if alternate_sample_design is not None and not design:
        raise IngestError("Alternate-sample design name cannot be blank.")
    # Record every authorized exclusion with its design and exact year scope.
    if incomplete_countries:
        if quarantine_log is None:
            raise IngestError(
                "An alternate-sample quarantine requires an explicit audit sink."
            )
        for country in incomplete_countries:
            quarantine_log.append(
                {
                    "policy": str(policy),
                    "outcome": str(outcome),
                    "estimand": "all_estimands_common_roster",
                    "country_code": country,
                    "reason": "incomplete finite required history",
                    "alternate_sample_design": design,
                    "reference_years": tuple(pre_years),
                    "post_years": tuple(post_years),
                }
            )
    retained = expected_index[complete.to_numpy()]
    # Resolve the exact declared retained count, not a permissive floor.
    if expected_count is None:
        declared_count = len(expected_index) if not design else len(retained)
    else:
        if isinstance(expected_count, (bool, np.bool_)):
            raise IngestError("Expected inference target count cannot be Boolean.")
        try:
            count_numeric = float(expected_count)
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError("Expected inference target count is malformed.") from exc
        if not count_numeric.is_integer() or not math.isfinite(count_numeric):
            raise IngestError("Expected inference target count must be integral and finite.")
        declared_count = int(count_numeric)
    if declared_count < MINIMUM_TARGETS or declared_count != len(retained):
        raise IngestError(
            f"{policy}/{outcome}: common inference roster has {len(retained)} "
            f"target(s), not the declared exact count {declared_count}."
        )
    # Compute pre/post country means only after complete histories are proved.
    retained_gaps = gaps.loc[retained]
    pre_direct = retained_gaps[pre_years].mean(axis=1).astype(float)
    post_direct = retained_gaps[post_years].mean(axis=1).astype(float)
    first_direct = retained_gaps[wave_one_years].mean(axis=1).astype(float)
    second_direct = retained_gaps[wave_two_years].mean(axis=1).astype(float)
    # The supplied reference vector must be the same labelled chosen-window
    # quantity rather than a stale baseline from another specification.
    supplied_pre_retained = supplied_pre.reindex(retained)
    if not bool(
        np.allclose(
            pre_direct.to_numpy(),
            supplied_pre_retained.to_numpy(),
            rtol=0.0,
            atol=IDENTITY_TOLERANCE,
        )
    ):
        raise IngestError(
            f"{policy}/{outcome}: supplied pre-escalation levels do not match "
            f"the declared {pre_start}-{pre_end} reference years."
        )
    # Independently reconstruct both labelled wave means so a country
    # permutation cannot preserve ATT while corrupting Eq. 16 dispersion.
    supplied_first = first.reindex(retained)
    supplied_second = second.reindex(retained)
    if not bool(
        np.allclose(
            first_direct.to_numpy(), supplied_first.to_numpy(),
            rtol=0.0, atol=IDENTITY_TOLERANCE,
        )
    ):
        raise IngestError(
            f"{policy}/{outcome}: supplied {WAVE_ONE_LABEL} country means do "
            "not match their labelled gap histories."
        )
    if not bool(
        np.allclose(
            second_direct.to_numpy(), supplied_second.to_numpy(),
            rtol=0.0, atol=IDENTITY_TOLERANCE,
        )
    ):
        raise IngestError(
            f"{policy}/{outcome}: supplied {WAVE_TWO_LABEL} country means do "
            "not match their labelled gap histories."
        )
    # Build all vectors on the identical retained country index.
    first = first_direct
    second = second_direct
    incremental = (post_direct - pre_direct).astype(float)
    first.name = f"{policy}|{outcome}|{WAVE_ONE_LABEL}"
    second.name = f"{policy}|{outcome}|{WAVE_TWO_LABEL}"
    incremental.name = f"{policy}|{outcome}|{ESTIMAND_INCREMENTAL}"
    contrasts = {
        WAVE_ONE_LABEL: first,
        WAVE_TWO_LABEL: second,
        ESTIMAND_INCREMENTAL: incremental,
    }
    # Assert the exact common denominator and finite labelled outputs.
    for label, vector in contrasts.items():
        if not vector.index.equals(retained) or len(vector) != declared_count:
            raise IngestError(f"{policy}/{outcome}/{label}: target denominator drifted.")
        if vector.index.has_duplicates or not bool(
            np.isfinite(vector.to_numpy(dtype=float)).all()
        ):
            raise IngestError(f"{policy}/{outcome}/{label}: contrast is not finite and unique.")
    # Log the built contrasts.
    logger.info(
        "%s/%s: %d contrast vector(s) built over %d countr(ies).",
        policy,
        outcome,
        len(contrasts),
        len(first),
    )
    # Return the contrast mapping.
    return contrasts


# Compute the effect and its country-dispersion standard error.
def compute_country_dispersion_se(
    contrasts: pd.Series, ddof: int = DISPERSION_DDOF
) -> Tuple[float, float]:
    """
    Compute the effect and its country-dispersion standard error.

    Parameters
    ----------
    contrasts : pandas.Series
        The per-country contrast vector.
    ddof : int, optional
        The sample-correction parameter of the standard deviation.

    Returns
    -------
    tuple
        The equal-weighted coalition effect and the dispersion standard
        error, which is the corrected standard deviation divided by the
        square root of the target count.

    Raises
    ------
    IngestError
        If the contrast vector carries fewer entries than the corrected
        deviation requires, or if it carries a missing value.

    Notes
    -----
    The estimator treats countries as independent. It deliberately does not
    propagate the first-stage uncertainty in the pooled coefficient, the
    factor proxy or the country loadings, which is why resampling diagnostics
    are reported alongside it.
    """
    # Equation 16 fixes the sample correction at one; alternative ddof values
    # would define a different estimator.
    try:
        numeric_ddof = float(ddof)
    except (TypeError, ValueError, OverflowError) as exc:
        raise IngestError("Country-dispersion ddof is malformed.") from exc
    if (
        isinstance(ddof, (bool, np.bool_))
        or not math.isfinite(numeric_ddof)
        or not numeric_ddof.is_integer()
        or int(numeric_ddof) != DISPERSION_DDOF
    ):
        raise IngestError(
            f"Country-dispersion inference requires ddof={DISPERSION_DDOF}."
        )
    # Refuse a contrast vector too short for the corrected deviation.
    if len(contrasts) < MINIMUM_TARGETS:
        # The failure names the observed length.
        raise IngestError(
            f"Country-dispersion inference requires at least {MINIMUM_TARGETS} "
            f"targets; the contrast vector carries {len(contrasts)}."
        )
    # Every country must be uniquely labelled so each receives one equal weight.
    if contrasts.index.has_duplicates:
        raise IngestError("Contrast vector carries duplicate country labels.")
    if any(code is None or not str(code).strip() for code in contrasts.index):
        raise IngestError("Contrast vector carries a null or blank country label.")
    # Refuse missing, infinite or nonnumeric contrasts before either moment.
    try:
        values = contrasts.to_numpy(dtype=float)
    except (TypeError, ValueError) as exc:
        raise IngestError("Contrast vector carries a nonnumeric value.") from exc
    if not bool(np.isfinite(values).all()):
        raise IngestError("Contrast vector carries a missing or non-finite value.")
    # Compute the equal-weighted coalition effect.
    effect: float = float(np.mean(values))
    # Compute the sample standard deviation with the declared correction.
    deviation: float = float(np.std(values, ddof=DISPERSION_DDOF))
    # Scale the deviation by the square root of the target count.
    standard_error: float = deviation / float(np.sqrt(len(contrasts)))
    # Log the computed moments.
    logger.info(
        "country-dispersion moments computed over %d target(s): effect %.6f, "
        "standard error %.6f.",
        len(contrasts),
        effect,
        standard_error,
    )
    # Return the effect and the standard error.
    return effect, standard_error


# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 2: Compute the t-statistics and the two-sided p-values
# -------------------------------------------------------------------------------------------------------------------------------


# Compute the test statistic and the degrees of freedom.
def compute_t_statistic(
    effect: float, standard_error: float, n_targets: int
) -> Tuple[float, int]:
    """
    Compute the test statistic and its degrees of freedom.

    Parameters
    ----------
    effect : float
        The coalition effect.
    standard_error : float
        The country-dispersion standard error.
    n_targets : int
        The number of targets the standard error was computed over.

    Returns
    -------
    tuple
        The test statistic and the degrees of freedom, which is the target
        count less one.

    Notes
    -----
    A zero standard error is propagated as an explicit limiting state:
    positive and negative nonzero effects yield positive and negative
    infinity, respectively; an exactly zero effect uses the conservative
    null limit ``t=0``. The ordered two-item return contract is unchanged.

    Raises
    ------
    IngestError
        If the effect/SE is non-finite, the SE is negative, or the target
        count is not a valid integer of at least two.
    """
    # Validate the finite effect and nonnegative finite standard error.
    try:
        numeric_effect = float(effect)
        numeric_se = float(standard_error)
    except (TypeError, ValueError, OverflowError) as exc:
        raise IngestError("Effect or standard error is malformed.") from exc
    if isinstance(effect, (bool, np.bool_)) or not math.isfinite(numeric_effect):
        raise IngestError(f"Cannot form a test statistic from effect {effect!r}.")
    if isinstance(standard_error, (bool, np.bool_)) or not math.isfinite(
        numeric_se
    ):
        raise IngestError(
            f"Cannot form a test statistic from standard error {standard_error!r}."
        )
    if numeric_se < 0.0:
        raise IngestError("Country-dispersion standard error cannot be negative.")
    # Refuse Boolean, fractional or too-small target declarations.
    if isinstance(n_targets, (bool, np.bool_)):
        raise IngestError("Target count cannot be Boolean.")
    try:
        numeric_targets = float(n_targets)
    except (TypeError, ValueError, OverflowError) as exc:
        raise IngestError("Target count is malformed.") from exc
    if (
        not math.isfinite(numeric_targets)
        or not numeric_targets.is_integer()
        or numeric_targets < MINIMUM_TARGETS
    ):
        # The failure names the observed count.
        raise IngestError(
            f"Cannot form a test statistic from {n_targets} target(s)."
        )
    # Compute the statistic as the effect over its standard error.
    if numeric_se == 0.0:
        if numeric_effect > 0.0:
            statistic = float("inf")
        elif numeric_effect < 0.0:
            statistic = float("-inf")
        else:
            statistic = 0.0
    else:
        statistic = numeric_effect / numeric_se
    # Compute the degrees of freedom as the target count less one.
    degrees_of_freedom: int = int(numeric_targets) - 1
    # Return the statistic and the degrees of freedom.
    return statistic, degrees_of_freedom


# Compute the two-sided probability of the test statistic.
def compute_two_sided_p(statistic: float, degrees_of_freedom: int) -> float:
    """
    Compute the two-sided probability of the test statistic.

    Parameters
    ----------
    statistic : float
        The test statistic.
    degrees_of_freedom : int
        The degrees of freedom of the Student-t distribution.

    Returns
    -------
    float
        The two-sided tail probability, computed as twice the survival
        function at the absolute statistic.

    Raises
    ------
    IngestError
        If the degrees of freedom are not positive.
    """
    # Reject NaN but accept signed infinity as the explicit zero-dispersion
    # limit produced by ``compute_t_statistic``.
    try:
        numeric_statistic = float(statistic)
    except (TypeError, ValueError, OverflowError) as exc:
        raise IngestError("Test statistic is malformed.") from exc
    if isinstance(statistic, (bool, np.bool_)) or math.isnan(numeric_statistic):
        raise IngestError(f"Cannot evaluate probability at statistic {statistic!r}.")
    # Refuse Boolean, fractional or non-positive degrees of freedom.
    if isinstance(degrees_of_freedom, (bool, np.bool_)):
        raise IngestError("Degrees of freedom cannot be Boolean.")
    try:
        numeric_df = float(degrees_of_freedom)
    except (TypeError, ValueError, OverflowError) as exc:
        raise IngestError("Degrees of freedom are malformed.") from exc
    if not math.isfinite(numeric_df) or not numeric_df.is_integer() or numeric_df <= 0:
        # The distribution is undefined without degrees of freedom.
        raise IngestError(
            f"Cannot evaluate a two-sided probability at "
            f"{degrees_of_freedom} degrees of freedom."
        )
    # Compute the one-sided upper tail at the absolute statistic.
    upper_tail: float = float(
        scipy_stats.t.sf(abs(numeric_statistic), df=int(numeric_df))
    )
    # Double the tail to obtain the two-sided probability.
    return float(2.0 * upper_tail)


# Assign the significance marker from the authoritative Student-t probability.
def assign_significance_stars(
    statistic: float,
    degrees_of_freedom: int,
    levels: Tuple[Tuple[float, str], ...] = SIGNIFICANCE_LEVELS,
) -> Tuple[str, Dict[str, float]]:
    """
    Assign the significance marker from the authoritative Student-t probability.

    Parameters
    ----------
    statistic : float
        The test statistic.
    degrees_of_freedom : int
        The degrees of freedom of the Student-t distribution.
    levels : tuple of tuple, optional
        The significance levels and their markers in reporting order.

    Returns
    -------
    tuple
        The marker selected by strict probability thresholds and the critical
        values retained only as descriptive audit metadata. Critical values
        never make the marker decision.

    Raises
    ------
    IngestError
        If the statistic, degrees of freedom or strictest-first level
        declaration is malformed.
    """
    # Refuse an empty level declaration.
    if not levels:
        raise IngestError("Significance levels are not declared.")
    # Validate the exact strictest-first probability-threshold contract.
    normalized_levels: List[Tuple[float, str]] = []
    for raw_level, raw_symbol in levels:
        if isinstance(raw_level, (bool, np.bool_)):
            raise IngestError("Significance probability cannot be Boolean.")
        try:
            level = float(raw_level)
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError("Significance probability is malformed.") from exc
        symbol = str(raw_symbol)
        if not math.isfinite(level) or not 0.0 < level < 1.0 or not symbol:
            raise IngestError("Significance levels or symbols are malformed.")
        normalized_levels.append((level, symbol))
    if [level for level, _ in normalized_levels] != sorted(
        level for level, _ in normalized_levels
    ) or len({level for level, _ in normalized_levels}) != len(normalized_levels):
        raise IngestError("Significance levels must be unique and strictest-first.")
    if len({symbol for _, symbol in normalized_levels}) != len(normalized_levels):
        raise IngestError("Significance marker symbols must be unique.")
    # Compute the primary two-sided Student-t probability; the signature has
    # no resampling input, making bootstrap/placebo substitution impossible.
    probability = compute_two_sided_p(statistic, degrees_of_freedom)
    # Accumulate the critical values for the audit log.
    critical_values: Dict[str, float] = {}
    # Compute the critical value of every declared level first, so the audit
    # log is complete whatever the statistic attains.
    for level, symbol in normalized_levels:
        # Compute the two-sided critical value of the level.
        critical_values[symbol] = float(
            scipy_stats.t.ppf(1.0 - float(level) / 2.0, df=int(degrees_of_freedom))
        )
    # Apply strict p-value inequalities: equality at a threshold falls to the
    # next, less significant category.
    marker = STAR_NONE
    for level, symbol in normalized_levels:
        if probability < level:
            marker = symbol
            break
    # Return the marker and the critical values.
    return marker, critical_values


# Assert the marker's agreement with the probability-based derivation.
def assert_star_consistency(
    marker: str,
    probability: float,
    levels: Tuple[Tuple[float, str], ...] = SIGNIFICANCE_LEVELS,
    statistic: Optional[float] = None,
    degrees_of_freedom: Optional[int] = None,
) -> None:
    """
    Assert the marker's agreement with the probability derivation.

    Parameters
    ----------
    marker : str
        The marker assigned from the unrounded primary probability.
    probability : float
        The two-sided probability of the statistic.
    levels : tuple of tuple, optional
        The significance levels and their markers.
    statistic : float, optional
        The unrounded Student-t statistic. When supplied, it is used with
        ``degrees_of_freedom`` to recompute and validate ``probability``.
    degrees_of_freedom : int, optional
        The Student-t degrees of freedom paired with ``statistic``.

    Returns
    -------
    None
        The function returns nothing when the two derivations agree.

    Raises
    ------
    IngestError
        If the probability or level declaration is malformed, or if the
        marker differs from the same strict probability rule used by the
        producer.
    """
    # The authoritative probability must be a finite probability measure.
    if isinstance(probability, (bool, np.bool_)) or not math.isfinite(
        float(probability)
    ) or not 0.0 <= float(probability) <= 1.0:
        raise IngestError(f"Primary probability is invalid: {probability!r}.")
    # When the producer inputs are available, reproduce the primary
    # probability inside this assertion before evaluating its thresholds.
    if (statistic is None) != (degrees_of_freedom is None):
        raise IngestError(
            "Statistic and degrees of freedom must be supplied together for "
            "marker consistency."
        )
    if statistic is not None and degrees_of_freedom is not None:
        assert_p_reproducible(
            float(probability), float(statistic), degrees_of_freedom
        )
    if not levels:
        raise IngestError("Significance levels are not declared.")
    normalized_levels: List[Tuple[float, str]] = []
    for raw_level, raw_symbol in levels:
        if isinstance(raw_level, (bool, np.bool_)):
            raise IngestError("Significance probability cannot be Boolean.")
        try:
            level = float(raw_level)
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError("Significance probability is malformed.") from exc
        symbol = str(raw_symbol)
        if not math.isfinite(level) or not 0.0 < level < 1.0 or not symbol:
            raise IngestError("Significance levels or symbols are malformed.")
        normalized_levels.append((level, symbol))
    if [level for level, _ in normalized_levels] != sorted(
        level for level, _ in normalized_levels
    ):
        raise IngestError("Significance levels must be strictest-first.")
    valid_markers = {STAR_NONE, *(symbol for _, symbol in normalized_levels)}
    if str(marker) not in valid_markers:
        raise IngestError(f"Unknown significance marker {marker!r}.")
    # Derive the expected marker from the probability, keeping the strictest
    # level the probability attains rather than the loosest.
    expected: str = STAR_NONE
    # Evaluate each declared level in its strictest-first order.
    for level, symbol in normalized_levels:
        # Assign the marker when the probability attains the level.
        if float(probability) < float(level):
            # Record the strictest level attained.
            expected = symbol
            # Stop, because every looser level is attained as well.
            break
    # Refuse a divergence between the two derivations.
    if expected != marker:
        # The failure names both markers and the probability.
        raise IngestError(
            f"Marker {marker!r} contradicts the marker {expected!r} implied "
            f"by the strict two-sided probability threshold at "
            f"{probability!r}."
        )
    # Log the verified agreement.
    logger.info(
        "marker consistency verified: %r at probability %.6f.", marker, probability
    )


# Assert the probability's reproducibility from the statistic and the df.
def assert_p_reproducible(
    probability: float,
    statistic: float,
    degrees_of_freedom: int,
    tolerance: float = REPRODUCIBILITY_TOLERANCE,
) -> None:
    """
    Assert the probability's reproducibility from the reported inputs.

    Parameters
    ----------
    probability : float
        The reported two-sided probability.
    statistic : float
        The reported test statistic.
    degrees_of_freedom : int
        The reported degrees of freedom.
    tolerance : float, optional
        The tolerance of the reproduction.

    Returns
    -------
    None
        The function returns nothing when the reproduction holds.

    Raises
    ------
    IngestError
        If the recomputed probability diverges from the reported value.
    """
    # Validate the reported probability and tight numerical tolerance first;
    # NaN must never pass because comparisons with it are false.
    if isinstance(probability, (bool, np.bool_)) or not math.isfinite(
        float(probability)
    ) or not 0.0 <= float(probability) <= 1.0:
        raise IngestError(f"Reported probability is invalid: {probability!r}.")
    if isinstance(tolerance, (bool, np.bool_)) or not math.isfinite(
        float(tolerance)
    ) or float(tolerance) <= 0.0:
        raise IngestError("Probability reproducibility tolerance must be positive and finite.")
    # Recompute the probability from the reported statistic and df.
    recomputed: float = compute_two_sided_p(statistic, degrees_of_freedom)
    # Compute the deviation of the two values.
    deviation: float = abs(float(probability) - recomputed)
    # Refuse a non-reproducible probability.
    if deviation > float(tolerance):
        # The failure names both values and the deviation.
        raise IngestError(
            f"Probability {probability!r} is not reproducible from statistic "
            f"{statistic!r} at {degrees_of_freedom} degrees of freedom "
            f"(recomputed {recomputed!r}, deviation {deviation:.3e})."
        )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 3: Register the inference outputs
# -------------------------------------------------------------------------------------------------------------------------------


# Canonically serialize and digest one labelled country-contrast vector.
def _freeze_country_contrast(contrast: pd.Series, context: str) -> str:
    """Return the SHA-256 digest of one finite, uniquely labelled vector."""
    # Country identities and values jointly define the Eq. 16 input.
    if contrast.empty or contrast.index.has_duplicates:
        raise IngestError(f"{context}: contrast is empty or has duplicate countries.")
    countries = [str(country).strip() for country in contrast.index]
    if any(not country for country in countries) or len(set(countries)) != len(countries):
        raise IngestError(f"{context}: contrast country identities are malformed.")
    try:
        values = contrast.to_numpy(dtype=float)
    except (TypeError, ValueError) as exc:
        raise IngestError(f"{context}: contrast carries a nonnumeric value.") from exc
    if not bool(np.isfinite(values).all()):
        raise IngestError(f"{context}: contrast carries a non-finite value.")
    # Sort by canonical country identity so row order alone cannot change the
    # digest, while every label-value association remains bound.
    labelled_values = sorted(zip(countries, values), key=lambda item: item[0])
    payload = "\n".join(
        f"{country}|{float(value):.17g}" for country, value in labelled_values
    )
    return hashlib.sha256(payload.encode("utf-8")).hexdigest()


# Assert the isolation of the marker's source.
def assert_stars_procedure_isolation(
    ledger: pd.DataFrame,
    contrasts: Mapping[Tuple[Tuple[str, str], str], pd.Series],
) -> None:
    """
    Assert that no marker derives from another inference procedure.

    Parameters
    ----------
    ledger : pandas.DataFrame
        The estimator ledger.
    contrasts : mapping
        The actual labelled country-contrast vectors keyed on
        ``((policy, outcome), estimand)``. These are required so the checker
        can bind every ledger row to its Eq. 16 scientific input.

    Returns
    -------
    None
        The function returns nothing when the isolation holds.

    Raises
    ------
    IngestError
        If contrast coverage, identities, moments or digests diverge; primary
        procedure identifiers are absent or inconsistent; the
        reported statistic/probability/marker cannot be reconstructed from
        Eq. 16, a report probability substitutes another procedure, or
        perturbing resampling diagnostics changes a primary marker.
    """
    # Require every field needed to reconstruct the primary Eq. 16 path.
    required = {
        "att", "se", "t_statistic", "p_value", "stars",
        "degrees_of_freedom", "n_targets", "stars_source",
        "probability_source", "estimator", "tail_convention",
        "contrast_sha256", "zero_dispersion_state",
    }
    missing = sorted(required - set(ledger.columns))
    if missing:
        raise IngestError(f"Estimator ledger lacks procedure field(s) {missing}.")
    if ledger.empty or ledger.index.has_duplicates:
        raise IngestError("Estimator ledger is empty or carries duplicate keys.")
    if not isinstance(contrasts, Mapping) or not contrasts:
        raise IngestError(
            "Marker isolation requires the actual country-contrast mapping."
        )
    # Normalize contrast keys to ledger's three-part index and require exact
    # coverage in both directions.
    contrast_by_ledger_key: Dict[Tuple[str, str, str], pd.Series] = {}
    for raw_key, vector in contrasts.items():
        if (
            not isinstance(raw_key, tuple)
            or len(raw_key) != 2
            or not isinstance(raw_key[0], tuple)
            or len(raw_key[0]) != 2
        ):
            raise IngestError(f"Malformed contrast mapping key {raw_key!r}.")
        ledger_key = (
            str(raw_key[0][0]), str(raw_key[0][1]), str(raw_key[1])
        )
        if ledger_key in contrast_by_ledger_key:
            raise IngestError(f"Duplicate normalized contrast key {ledger_key}.")
        if not isinstance(vector, pd.Series):
            raise IngestError(f"{ledger_key}: contrast must be a pandas Series.")
        contrast_by_ledger_key[ledger_key] = vector
    ledger_keys = {
        (str(key[0]), str(key[1]), str(key[2])) for key in ledger.index
    }
    contrast_keys = set(contrast_by_ledger_key)
    if ledger_keys != contrast_keys:
        raise IngestError(
            "Ledger/contrast coverage diverges: missing contrast keys "
            f"{sorted(ledger_keys - contrast_keys)}, unexpected contrast keys "
            f"{sorted(contrast_keys - ledger_keys)}."
        )
    # Identify the rows whose source diverges from the declared constant.
    divergent = ledger[ledger["stars_source"].astype(str) != STAR_SOURCE]
    # Refuse a divergent source.
    if len(divergent):
        # The failure names the count and the observed sources.
        sources = sorted(set(divergent["stars_source"].astype(str)))
        # Raise with the observed sources attached.
        raise IngestError(
            f"{len(divergent)} ledger row(s) record a marker source outside the "
            f"declared constant: {sources}."
        )
    # Probability, estimator and tail declarations are independently bound.
    if bool((ledger["probability_source"].astype(str) != PRIMARY_PROBABILITY_SOURCE).any()):
        raise IngestError("Estimator ledger substitutes a non-primary probability source.")
    if bool((ledger["estimator"].astype(str) != COUNTRY_DISPERSION_ESTIMATOR).any()):
        raise IngestError("Estimator ledger records a non-Eq.16 estimator.")
    if bool((ledger["tail_convention"].astype(str) != STUDENT_T_TAIL_CONVENTION).any()):
        raise IngestError("Estimator ledger records a non-Student-t tail convention.")
    # Contrast identities must be nonempty SHA-256 strings.
    valid_contrast_hashes = ledger["contrast_sha256"].astype(str).str.fullmatch(
        r"[0-9a-f]{64}"
    ).fillna(False)
    if not bool(valid_contrast_hashes.all()):
        raise IngestError("Estimator ledger carries an invalid contrast fingerprint.")
    # Resampling outputs may coexist as diagnostics, but are never inputs to
    # the primary marker calculation.
    forbidden_tokens: Tuple[str, ...] = ("bootstrap", "placebo", "randomization")
    diagnostic_columns: List[str] = [
        str(column)
        for column in ledger.columns
        if any(token in str(column).lower() for token in forbidden_tokens)
    ]
    # Independently reconstruct every row from ATT, SE and N.
    baseline_markers: List[str] = []
    for index, row in ledger.iterrows():
        normalized_index = (str(index[0]), str(index[1]), str(index[2]))
        vector = contrast_by_ledger_key[normalized_index]
        # Recompute the canonical vector digest and both Eq. 16 moments from
        # the actual labelled scientific input.
        recomputed_digest = _freeze_country_contrast(
            vector, f"{normalized_index} marker isolation"
        )
        if str(row["contrast_sha256"]) != recomputed_digest:
            raise IngestError(
                f"{normalized_index}: contrast fingerprint does not match the "
                "actual labelled vector."
            )
        vector_effect, vector_se = compute_country_dispersion_se(vector)
        reported_n = float(row["n_targets"])
        if not reported_n.is_integer() or int(reported_n) != len(vector):
            raise IngestError(
                f"{normalized_index}: ledger N does not equal vector length."
            )
        if abs(float(row["att"]) - vector_effect) > IDENTITY_TOLERANCE:
            raise IngestError(
                f"{normalized_index}: ledger ATT does not equal vector mean."
            )
        if abs(float(row["se"]) - vector_se) > IDENTITY_TOLERANCE:
            raise IngestError(
                f"{normalized_index}: ledger SE does not equal Eq. 16 vector SE."
            )
        statistic, degrees_of_freedom = compute_t_statistic(
            float(row["att"]), float(row["se"]), row["n_targets"]
        )
        reported_statistic = float(row["t_statistic"])
        if not (
            (math.isinf(statistic) and statistic == reported_statistic)
            or (
                math.isfinite(statistic)
                and math.isfinite(reported_statistic)
                and abs(statistic - reported_statistic) <= REPRODUCIBILITY_TOLERANCE
            )
        ):
            raise IngestError(f"{index}: statistic is not reproducible from ATT and SE.")
        reported_df = float(row["degrees_of_freedom"])
        if not reported_df.is_integer() or int(reported_df) != degrees_of_freedom:
            raise IngestError(f"{index}: degrees of freedom do not equal N-1.")
        probability = compute_two_sided_p(statistic, degrees_of_freedom)
        assert_p_reproducible(
            float(row["p_value"]),
            statistic,
            degrees_of_freedom,
        )
        marker, _ = assign_significance_stars(statistic, degrees_of_freedom)
        assert_star_consistency(
            marker,
            probability,
            statistic=statistic,
            degrees_of_freedom=degrees_of_freedom,
        )
        if str(row["stars"]) != marker:
            raise IngestError(f"{index}: marker is not generated by the primary probability.")
        # Validate the explicit zero-dispersion state.
        expected_state = (
            "positive_effect_zero_se"
            if float(row["se"]) == 0.0 and float(row["att"]) > 0.0
            else "negative_effect_zero_se"
            if float(row["se"]) == 0.0 and float(row["att"]) < 0.0
            else "zero_effect_zero_se"
            if float(row["se"]) == 0.0
            else "positive_se"
        )
        if str(row["zero_dispersion_state"]) != expected_state:
            raise IngestError(f"{index}: zero-dispersion state is inconsistent.")
        baseline_markers.append(marker)
    # Reject an explicitly published/report probability that differs from the
    # primary probability, even when a diagnostic column is also present.
    for report_column in ("report_probability", "reported_p_value"):
        if report_column in ledger.columns:
            report_values = pd.to_numeric(ledger[report_column], errors="coerce").to_numpy(float)
            primary_values = ledger["p_value"].to_numpy(dtype=float)
            if not bool(
                np.isfinite(report_values).all()
                and np.allclose(
                    report_values,
                    primary_values,
                    rtol=0.0,
                    atol=REPRODUCIBILITY_TOLERANCE,
                )
            ):
                raise IngestError(
                    f"Estimator ledger substitutes {report_column} for the primary probability."
                )
    # Execute the leakage test: alter every bootstrap/placebo/randomization
    # diagnostic independently while holding contrasts and Eq. 16 inputs fixed.
    perturbed = ledger.copy(deep=True)
    for column in diagnostic_columns:
        numeric = pd.to_numeric(perturbed[column], errors="coerce")
        if bool(numeric.notna().any()):
            perturbed[column] = numeric.where(numeric.isna(), -numeric - 17.0)
        else:
            perturbed[column] = "__PERTURBED_DIAGNOSTIC__"
    perturbed_markers = [
        assign_significance_stars(
            float(row["t_statistic"]), int(row["degrees_of_freedom"])
        )[0]
        for _, row in perturbed.iterrows()
    ]
    if perturbed_markers != baseline_markers:
        raise IngestError(
            "Primary markers changed when resampling diagnostics were perturbed."
        )
    # Log the verified isolation.
    logger.info(
        "marker isolation verified across %d ledger row(s).", len(ledger)
    )


# Freeze the inference ledger behind a canonical digest.
def freeze_inference_ledger(
    ledger: pd.DataFrame, input_gap_fingerprint: str
) -> str:
    """
    Compute the canonical digest of the inference ledger.

    Parameters
    ----------
    ledger : pandas.DataFrame
        The estimator ledger keyed on the partition and the estimand.
    input_gap_fingerprint : str
        The fingerprint of the gap result the inference was computed from.

    Returns
    -------
    str
        The hexadecimal digest of the canonical payload, which binds the
        ledger to its input.

    Raises
    ------
    IngestError
        If the ledger is empty or the input fingerprint is empty.
    """
    # Refuse an empty ledger.
    if len(ledger) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot freeze an empty inference ledger.")
    # Refuse an empty input fingerprint.
    if not str(input_gap_fingerprint):
        # Without the input fingerprint the ledger is not bound to its input.
        raise IngestError(
            "Cannot freeze an inference ledger without the input gap fingerprint."
        )
    # Accumulate the canonical payload parts.
    parts: List[str] = [f"input_gap_fingerprint={input_gap_fingerprint}"]
    # Bind every scientific and procedure-provenance field used to reproduce
    # Eq. 16 inference and its reporting identity.
    required_columns = {
        "att", "se", "t_statistic", "p_value", "stars",
        "degrees_of_freedom", "n_targets", "stars_source",
        "probability_source", "estimator", "tail_convention",
        "contrast_sha256", "zero_dispersion_state", "reference_start",
        "reference_end", "post_start", "post_end",
    }
    missing_columns = sorted(required_columns - set(ledger.columns))
    if missing_columns:
        raise IngestError(
            f"Inference ledger omits digest field(s) {missing_columns}."
        )
    # Iterate over the ledger rows in sorted key order.
    for (policy, outcome, estimand), row in ledger.sort_index().iterrows():
        # Append the row's canonical line at full precision.
        parts.append(
            "R|"
            + str(policy)
            + "|"
            + str(outcome)
            + "|"
            + str(estimand)
            + "|"
            + f"{float(row['att']):.17g}"
            + "|"
            + f"{float(row['se']):.17g}"
            + "|"
            + f"{float(row['t_statistic']):.17g}"
            + "|"
            + f"{float(row['p_value']):.17g}"
            + "|"
            + str(row["stars"])
            + "|"
            + str(int(row["degrees_of_freedom"]))
            + "|"
            + str(int(row["n_targets"]))
            + "|"
            + str(row["stars_source"])
            + "|"
            + str(row["probability_source"])
            + "|"
            + str(row["estimator"])
            + "|"
            + str(row["tail_convention"])
            + "|"
            + str(row["contrast_sha256"])
            + "|"
            + str(row["zero_dispersion_state"])
            + "|"
            + str(int(row["reference_start"]))
            + "|"
            + str(int(row["reference_end"]))
            + "|"
            + str(int(row["post_start"]))
            + "|"
            + str(int(row["post_end"]))
        )
    # Digest the encoded payload.
    return hashlib.sha256("\n".join(parts).encode("utf-8")).hexdigest()


# Serialize the ledger to its declared storage format.
def serialize_inference_ledger(
    ledger: pd.DataFrame, destination: Optional[Path] = None
) -> Optional[Path]:
    """
    Serialize the inference ledger to its declared storage format.

    Parameters
    ----------
    ledger : pandas.DataFrame
        The estimator ledger.
    destination : pathlib.Path, optional
        The parquet destination. When absent the ledger is not serialized,
        which is the fixture mode.

    Returns
    -------
    pathlib.Path or None
        The destination path when the ledger was serialized, and nothing
        otherwise.

    Raises
    ------
    IngestError
        If the destination's parent directory does not exist, because the
        module does not create directories implicitly.
    """
    # Return without serialization when no destination was supplied.
    if destination is None:
        # Log the skipped serialization.
        logger.info("ledger serialization skipped: no destination supplied.")
        # Return nothing.
        return None
    # Refuse a destination whose parent directory does not exist.
    if not destination.parent.is_dir():
        # The failure names the missing directory.
        raise IngestError(
            f"Serialization destination directory does not exist: "
            f"{destination.parent}."
        )
    # Write the ledger in the declared storage format, resetting the index so
    # the key columns survive the round trip.
    ledger.reset_index().to_parquet(destination, index=False)
    # Log the serialization.
    logger.info("ledger serialized to %s.", destination)
    # Return the destination path.
    return destination


# Compare the ledger with the declared manuscript inference values.
def assert_declared_inference(
    ledger: pd.DataFrame,
    expected_se: Optional[Mapping[Tuple[str, str, str], Any]] = None,
    expected_stars: Optional[Mapping[Tuple[str, str, str], Any]] = None,
    tolerance: float = 1.0e-12,
    expected_probability: Optional[Mapping[Tuple[str, str, str], Any]] = None,
) -> pd.DataFrame:
    """
    Compare the inference ledger with the declared study values.

    Parameters
    ----------
    ledger : pandas.DataFrame
        The estimator ledger.
    expected_se : mapping of tuple to object, optional
        Declared SE cells. Scalars use three displayed decimals. Structured
        cells may state value, units, displayed_decimals,
        numerical_allowance, variation_kind, monte_carlo_allowance,
        procedure, reference_window and n_targets.
    expected_stars : mapping of tuple to object, optional
        Declared marker cells. A string remains supported; a structured cell
        may add the same procedure/reference/N identity metadata.
    tolerance : float, optional
        Backward-compatible numerical allowance for scalar SE declarations;
        it is separate from the publication rounding interval.
    expected_probability : mapping of tuple to object, optional
        Declared primary Student-t probability cells. This trailing argument
        preserves the existing positional contract. Scalars use three
        displayed decimals; structured cells support publication metadata and
        explicit numerical or Monte Carlo allowances.

    Returns
    -------
    pandas.DataFrame
        Cell-level comparison audit retaining unrounded values, publication
        intervals, metadata identities, deviations and a reasoned comparison
        status.

    Raises
    ------
    IngestError
        If a declaration is malformed, metadata identifies another procedure
        or sample, a marker contradicts the original unrounded Student-t
        calculation, or a numeric mismatch exceeds its declared allowance.
    """
    # Validate the deterministic scalar numerical allowance.
    if isinstance(tolerance, (bool, np.bool_)) or not math.isfinite(float(tolerance)) or float(tolerance) < 0.0:
        raise IngestError("Declared-inference numerical allowance must be finite and nonnegative.")
    required = {
        "se", "stars", "p_value", "t_statistic", "degrees_of_freedom",
        "n_targets", "estimator", "reference_start", "reference_end",
        "stars_source", "probability_source",
    }
    missing = sorted(required - set(ledger.columns))
    if missing:
        raise IngestError(f"Inference ledger lacks publication field(s) {missing}.")
    if ledger.empty or ledger.index.has_duplicates:
        raise IngestError("Inference ledger is empty or carries duplicate keys.")
    audit_columns = [
        "cell_id", "policy", "outcome", "estimand", "reported_field",
        "procedure", "reference_start", "reference_end", "n_targets", "units",
        "displayed_decimals", "reported_value", "unrounded_value",
        "rounding_lower", "rounding_upper", "numerical_allowance",
        "monte_carlo_allowance", "signed_deviation", "absolute_deviation",
        "comparison_status",
    ]
    audit_rows: List[Dict[str, Any]] = []

    # Normalize one three-part ledger key.
    def normalize_key(raw_key: Any) -> Tuple[str, str, str]:
        if not isinstance(raw_key, tuple) or len(raw_key) != 3:
            raise IngestError(f"Inference publication key {raw_key!r} is malformed.")
        key = tuple(str(value).strip() for value in raw_key)
        if any(not value for value in key):
            raise IngestError(f"Inference publication key {raw_key!r} is blank.")
        return key[0], key[1], key[2]

    # Resolve accurate native units without scaling stored values.
    def inference_units(outcome: str) -> str:
        if outcome in {"log_CPI", "log_real_GDPpc"}:
            return "log_points"
        if outcome in {
            "inflation_rate", "GDPpc_growth_rate", "inflation_att", "growth_att"
        }:
            return "decimal_rate (multiply by 100 for percentage points)"
        return "outcome_units"

    # Validate declaration identity against the exact ledger row.
    def validate_metadata(
        key: Tuple[str, str, str],
        row: pd.Series,
        declaration: Mapping[str, Any],
        procedure_column: str,
        default_procedure: str,
    ) -> Tuple[str, int, int, int]:
        procedure = str(declaration.get("procedure", default_procedure)).strip()
        if procedure != str(row[procedure_column]):
            raise IngestError(f"{key}: procedure metadata mismatch.")
        reference_raw = declaration.get(
            "reference_window", (int(row["reference_start"]), int(row["reference_end"]))
        )
        if not isinstance(reference_raw, (tuple, list)) or len(reference_raw) != 2:
            raise IngestError(f"{key}: reference-window metadata is malformed.")
        reference = _validate_att_window(
            (reference_raw[0], reference_raw[1]), f"{key} publication reference"
        )
        if reference != (int(row["reference_start"]), int(row["reference_end"])):
            raise IngestError(f"{key}: reference-window metadata mismatch.")
        n_raw = declaration.get("n_targets", int(row["n_targets"]))
        if isinstance(n_raw, (bool, np.bool_)):
            raise IngestError(f"{key}: publication N cannot be Boolean.")
        try:
            n_numeric = float(n_raw)
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError(f"{key}: publication N is malformed.") from exc
        if not math.isfinite(n_numeric) or not n_numeric.is_integer():
            raise IngestError(f"{key}: publication N must be a finite integer.")
        n_targets = int(n_numeric)
        if n_targets != int(row["n_targets"]):
            raise IngestError(f"{key}: target-count metadata mismatch.")
        return procedure, reference[0], reference[1], n_targets

    # Compare the standard errors when declared values were supplied.
    if expected_se is not None:
        for raw_key, raw_declared in expected_se.items():
            key = normalize_key(raw_key)
            if key not in ledger.index:
                raise IngestError(f"Ledger omits the key {key}.")
            row = ledger.loc[key]
            declaration = dict(raw_declared) if isinstance(raw_declared, Mapping) else {"value": raw_declared}
            if "value" not in declaration and "published_value" in declaration:
                declaration["value"] = declaration["published_value"]
            if "value" not in declaration or isinstance(declaration["value"], (bool, np.bool_)):
                raise IngestError(f"{key}: declared SE value is absent or Boolean.")
            try:
                reported = float(declaration["value"])
            except (TypeError, ValueError, OverflowError) as exc:
                raise IngestError(f"{key}: declared SE is malformed.") from exc
            if not math.isfinite(reported) or reported < 0.0:
                raise IngestError(f"{key}: declared SE must be finite and nonnegative.")
            procedure, ref_start, ref_end, n_targets = validate_metadata(
                key,
                row,
                declaration,
                "estimator",
                COUNTRY_DISPERSION_ESTIMATOR,
            )
            precision_raw = declaration.get("displayed_decimals", 3)
            if isinstance(precision_raw, (bool, np.bool_)):
                raise IngestError(f"{key}: displayed precision cannot be Boolean.")
            try:
                precision_numeric = float(precision_raw)
                allowance = float(declaration.get("numerical_allowance", tolerance))
                monte_carlo_allowance = float(
                    declaration.get("monte_carlo_allowance", 0.0)
                )
            except (TypeError, ValueError, OverflowError) as exc:
                raise IngestError(f"{key}: SE precision or allowance is malformed.") from exc
            if not math.isfinite(precision_numeric) or not precision_numeric.is_integer() or precision_numeric < 0.0:
                raise IngestError(f"{key}: displayed precision must be a nonnegative integer.")
            decimals = int(precision_numeric)
            if not math.isfinite(allowance) or allowance < 0.0:
                raise IngestError(f"{key}: numerical allowance is invalid.")
            if not math.isfinite(monte_carlo_allowance) or monte_carlo_allowance < 0.0:
                raise IngestError(f"{key}: Monte Carlo allowance is invalid.")
            variation_kind = str(declaration.get("variation_kind", "deterministic")).strip()
            if variation_kind not in {"deterministic", "monte_carlo"}:
                raise IngestError(f"{key}: variation_kind is not recognized.")
            if variation_kind != "monte_carlo" and monte_carlo_allowance != 0.0:
                raise IngestError(
                    f"{key}: Monte Carlo allowance requires variation_kind='monte_carlo'."
                )
            units = str(declaration.get("units", inference_units(key[1]))).strip()
            if not units:
                raise IngestError(f"{key}: inference units are blank.")
            unrounded = float(row["se"])
            if not math.isfinite(unrounded) or unrounded < 0.0:
                raise IngestError(f"{key}: ledger SE is invalid.")
            half_unit = 0.5 * 10.0 ** (-decimals)
            lower, upper = reported - half_unit, reported + half_unit
            deviation = unrounded - reported
            if lower <= unrounded <= upper:
                status = "within_publication_rounding"
            elif lower - allowance <= unrounded <= upper + allowance:
                status = "within_numerical_allowance"
            elif variation_kind == "monte_carlo" and lower - allowance - monte_carlo_allowance <= unrounded <= upper + allowance + monte_carlo_allowance:
                status = "within_monte_carlo_variation"
            else:
                raise IngestError(
                    f"{key}: formula_or_procedure_mismatch; unrounded SE {unrounded!r} "
                    f"is outside [{lower!r}, {upper!r}], numerical allowance "
                    f"{allowance!r}, Monte Carlo allowance {monte_carlo_allowance!r}."
                )
            audit_rows.append(
                {
                    "cell_id": f"{key[0]}/{key[1]}/{key[2]}/se",
                    "policy": key[0], "outcome": key[1], "estimand": key[2],
                    "reported_field": "se", "procedure": procedure,
                    "reference_start": ref_start, "reference_end": ref_end,
                    "n_targets": n_targets, "units": units,
                    "displayed_decimals": decimals, "reported_value": reported,
                    "unrounded_value": unrounded, "rounding_lower": lower,
                    "rounding_upper": upper, "numerical_allowance": allowance,
                    "monte_carlo_allowance": monte_carlo_allowance,
                    "signed_deviation": deviation, "absolute_deviation": abs(deviation),
                    "comparison_status": status,
                }
            )
    # Compare primary Student-t probabilities with the same publication-aware
    # distinctions while first reproducing each unrounded value from t and df.
    if expected_probability is not None:
        for raw_key, raw_declared in expected_probability.items():
            key = normalize_key(raw_key)
            if key not in ledger.index:
                raise IngestError(f"Ledger omits the key {key}.")
            row = ledger.loc[key]
            declaration = dict(raw_declared) if isinstance(raw_declared, Mapping) else {"value": raw_declared}
            if "value" not in declaration and "published_value" in declaration:
                declaration["value"] = declaration["published_value"]
            if "value" not in declaration or isinstance(declaration["value"], (bool, np.bool_)):
                raise IngestError(f"{key}: declared probability is absent or Boolean.")
            try:
                reported = float(declaration["value"])
            except (TypeError, ValueError, OverflowError) as exc:
                raise IngestError(f"{key}: declared probability is malformed.") from exc
            if not math.isfinite(reported) or not 0.0 <= reported <= 1.0:
                raise IngestError(f"{key}: declared probability is outside [0,1].")
            procedure, ref_start, ref_end, n_targets = validate_metadata(
                key,
                row,
                declaration,
                "probability_source",
                PRIMARY_PROBABILITY_SOURCE,
            )
            precision_raw = declaration.get("displayed_decimals", 3)
            if isinstance(precision_raw, (bool, np.bool_)):
                raise IngestError(f"{key}: displayed precision cannot be Boolean.")
            try:
                precision_numeric = float(precision_raw)
                allowance = float(declaration.get("numerical_allowance", tolerance))
                monte_carlo_allowance = float(
                    declaration.get("monte_carlo_allowance", 0.0)
                )
            except (TypeError, ValueError, OverflowError) as exc:
                raise IngestError(f"{key}: probability precision or allowance is malformed.") from exc
            if not math.isfinite(precision_numeric) or not precision_numeric.is_integer() or precision_numeric < 0.0:
                raise IngestError(f"{key}: displayed precision must be a nonnegative integer.")
            if not math.isfinite(allowance) or allowance < 0.0:
                raise IngestError(f"{key}: numerical allowance is invalid.")
            if not math.isfinite(monte_carlo_allowance) or monte_carlo_allowance < 0.0:
                raise IngestError(f"{key}: Monte Carlo allowance is invalid.")
            variation_kind = str(declaration.get("variation_kind", "deterministic")).strip()
            if variation_kind not in {"deterministic", "monte_carlo"}:
                raise IngestError(f"{key}: variation_kind is not recognized.")
            if variation_kind != "monte_carlo" and monte_carlo_allowance != 0.0:
                raise IngestError(
                    f"{key}: Monte Carlo allowance requires variation_kind='monte_carlo'."
                )
            unrounded = float(row["p_value"])
            assert_p_reproducible(
                unrounded,
                float(row["t_statistic"]),
                int(row["degrees_of_freedom"]),
            )
            decimals = int(precision_numeric)
            half_unit = 0.5 * 10.0 ** (-decimals)
            lower, upper = max(0.0, reported - half_unit), min(1.0, reported + half_unit)
            deviation = unrounded - reported
            if lower <= unrounded <= upper:
                status = "within_publication_rounding"
            elif max(0.0, lower - allowance) <= unrounded <= min(1.0, upper + allowance):
                status = "within_numerical_allowance"
            elif variation_kind == "monte_carlo" and max(0.0, lower - allowance - monte_carlo_allowance) <= unrounded <= min(1.0, upper + allowance + monte_carlo_allowance):
                status = "within_monte_carlo_variation"
            else:
                raise IngestError(
                    f"{key}: formula_or_procedure_mismatch; unrounded probability "
                    f"{unrounded!r} is outside [{lower!r}, {upper!r}], numerical "
                    f"allowance {allowance!r}, Monte Carlo allowance "
                    f"{monte_carlo_allowance!r}."
                )
            units = str(declaration.get("units", "probability")).strip()
            if units != "probability":
                raise IngestError(f"{key}: probability units must be 'probability'.")
            audit_rows.append(
                {
                    "cell_id": f"{key[0]}/{key[1]}/{key[2]}/p_value",
                    "policy": key[0], "outcome": key[1], "estimand": key[2],
                    "reported_field": "p_value", "procedure": procedure,
                    "reference_start": ref_start, "reference_end": ref_end,
                    "n_targets": n_targets, "units": units,
                    "displayed_decimals": decimals, "reported_value": reported,
                    "unrounded_value": unrounded, "rounding_lower": lower,
                    "rounding_upper": upper, "numerical_allowance": allowance,
                    "monte_carlo_allowance": monte_carlo_allowance,
                    "signed_deviation": deviation, "absolute_deviation": abs(deviation),
                    "comparison_status": status,
                }
            )
    # Compare the markers when declared values were supplied.
    if expected_stars is not None:
        for raw_key, raw_declared in expected_stars.items():
            key = normalize_key(raw_key)
            if key not in ledger.index:
                raise IngestError(f"Ledger omits the key {key}.")
            row = ledger.loc[key]
            declaration = dict(raw_declared) if isinstance(raw_declared, Mapping) else {"value": raw_declared}
            declared_marker = str(
                declaration.get("value", declaration.get("stars", "__MISSING__"))
            )
            procedure, ref_start, ref_end, n_targets = validate_metadata(
                key,
                row,
                declaration,
                "stars_source",
                STAR_SOURCE,
            )
            # Recompute from the original unrounded Student-t inputs, never a
            # rounded p-value or manuscript marker.
            assert_p_reproducible(
                float(row["p_value"]),
                float(row["t_statistic"]),
                int(row["degrees_of_freedom"]),
            )
            recomputed_marker, _ = assign_significance_stars(
                float(row["t_statistic"]), int(row["degrees_of_freedom"])
            )
            if str(row["stars"]) != recomputed_marker:
                raise IngestError(f"{key}: ledger star is not from unrounded Student-t.")
            if recomputed_marker != declared_marker:
                raise IngestError(
                    f"{key}: marker {recomputed_marker!r} against declared "
                    f"{declared_marker!r}; formula_or_procedure_mismatch."
                )
            audit_rows.append(
                {
                    "cell_id": f"{key[0]}/{key[1]}/{key[2]}/stars",
                    "policy": key[0], "outcome": key[1], "estimand": key[2],
                    "reported_field": "stars", "procedure": procedure,
                    "reference_start": ref_start, "reference_end": ref_end,
                    "n_targets": n_targets, "units": "marker",
                    "displayed_decimals": pd.NA, "reported_value": declared_marker,
                    "unrounded_value": recomputed_marker, "rounding_lower": np.nan,
                    "rounding_upper": np.nan, "numerical_allowance": 0.0,
                    "monte_carlo_allowance": 0.0, "signed_deviation": np.nan,
                    "absolute_deviation": np.nan,
                    "comparison_status": "exact_unrounded_student_t_marker",
                }
            )
    audit = pd.DataFrame(audit_rows, columns=audit_columns)
    if len(audit) and bool(audit["cell_id"].duplicated().any()):
        raise IngestError("Inference publication audit has duplicate cell identities.")
    # Log the comparison mode.
    logger.info(
        "declared inference comparison %s.",
        "performed" if len(audit) else "skipped",
    )
    return audit


# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the country-dispersion inference across the partitions.
def orchestrate_country_dispersion_inference(
    gap_result: GapAttResult,
    pre_window: Tuple[int, int] = (2019, 2021),
    post_window: Tuple[int, int] = (2022, 2024),
    expected_se: Optional[Mapping[Tuple[str, str, str], Any]] = None,
    expected_stars: Optional[Mapping[Tuple[str, str, str], Any]] = None,
    destination: Optional[Path] = None,
    expected_probability: Optional[Mapping[Tuple[str, str, str], Any]] = None,
) -> InferenceResult:
    """
    Compute the country-dispersion inference for every partition.

    Parameters
    ----------
    gap_result : GapAttResult
        The gap matrix and effect result.
    pre_window : tuple of int, optional
        The inclusive pre-escalation window of the incremental contrast.
    post_window : tuple of int, optional
        The inclusive post-escalation window of the incremental contrast.
    expected_se : mapping of tuple to object, optional
        The declared publication SE cells per key.
    expected_stars : mapping of tuple to object, optional
        The declared publication marker cells per key.
    destination : pathlib.Path, optional
        The parquet destination of the ledger.
    expected_probability : mapping of tuple to object, optional
        The declared primary Student-t probability publication cells. This is
        trailing to preserve every existing positional call.

    Returns
    -------
    InferenceResult
        The ledger, contrast vectors, descriptive critical values,
        publication comparison audit, digest, input fingerprint, timestamp
        and notes.

    Raises
    ------
    IngestError
        If any contrast, moment, test, isolation or registration check fails.
    """
    # Validate the exact inference windows at the public boundary.
    pre_window = _validate_att_window(pre_window, "inference reference window")
    post_window = _validate_att_window(post_window, "inference post window")
    if pre_window[1] >= post_window[0]:
        raise IngestError("Inference reference and post windows overlap or are unordered.")
    if not gap_result.gaps:
        raise IngestError("Cannot run country-dispersion inference without gap matrices.")
    # Initialise the contrast mapping keyed on the pair and the estimand.
    contrasts: Dict[Tuple[Tuple[str, str], str], pd.Series] = {}
    # Accumulate the ledger rows.
    ledger_rows: List[Dict[str, Any]] = []
    # Accumulate the critical-value rows.
    critical_rows: List[Dict[str, Any]] = []
    # Accumulate the quarantine audit rows (remediation B-17).
    quarantine_log: List[Dict[str, Any]] = []
    # Iterate over the partitions in sorted pair order.
    for pair in sorted(gap_result.gaps.keys()):
        # Resolve the actual Task 13 window metadata before reconstructing any
        # country-level contrast; global defaults cannot replace provenance.
        required_window_metadata = {
            "reference_start", "reference_end", "post_start", "post_end",
            "wave_one_start", "wave_one_end", "wave_two_start", "wave_two_end",
        }
        missing_window_metadata = sorted(
            required_window_metadata - set(gap_result.incremental_effects.columns)
        )
        if missing_window_metadata or pair not in gap_result.incremental_effects.index:
            raise IngestError(
                f"{pair}: Task 13 window provenance is unavailable: "
                f"{missing_window_metadata}."
            )
        pair_metadata = gap_result.incremental_effects.loc[pair]
        if (int(pair_metadata["reference_start"]), int(pair_metadata["reference_end"])) != pre_window:
            raise IngestError(f"{pair}: inference reference window conflicts with Task 13.")
        if (int(pair_metadata["post_start"]), int(pair_metadata["post_end"])) != post_window:
            raise IngestError(f"{pair}: inference post window conflicts with Task 13.")
        wave_one_window = _validate_att_window(
            (int(pair_metadata["wave_one_start"]), int(pair_metadata["wave_one_end"])),
            f"{pair} Task 13 wave-one window",
        )
        wave_two_window = _validate_att_window(
            (int(pair_metadata["wave_two_start"]), int(pair_metadata["wave_two_end"])),
            f"{pair} Task 13 wave-two window",
        )
        # Build complete-history vectors on the exact full target roster; the
        # primary design does not authorize a quarantine or alternate sample.
        pair_contrasts: Dict[str, pd.Series] = build_contrast_vectors(
            gap_result.gaps[pair],
            gap_result.country_window_means,
            gap_result.pre_escalation_levels,
            pair[0],
            pair[1],
            pre_window,
            post_window,
            expected_count=len(gap_result.gaps[pair].index),
            quarantine_log=quarantine_log,
            alternate_sample_design=None,
            wave_one_window=wave_one_window,
            wave_two_window=wave_two_window,
        )
        # Resolve the authoritative full-precision effects from Task 13.
        if pair not in gap_result.level_effects.index or pair not in gap_result.incremental_effects.index:
            raise IngestError(f"{pair}: gap result omits a reported ATT row.")
        reported_effects = {
            WAVE_ONE_LABEL: float(gap_result.level_effects.loc[pair, WAVE_ONE_LABEL]),
            WAVE_TWO_LABEL: float(gap_result.level_effects.loc[pair, WAVE_TWO_LABEL]),
            ESTIMAND_INCREMENTAL: float(
                gap_result.incremental_effects.loc[pair, "incremental_effect"]
            ),
        }
        # Record the pair's contrast vectors under the estimand keys.
        for estimand, vector in pair_contrasts.items():
            # Store the contrast vector.
            contrasts[(pair, estimand)] = vector
        # Evaluate each declared estimand in its order.
        for estimand in ESTIMAND_ORDER:
            # Read the estimand's contrast vector.
            vector: pd.Series = pair_contrasts[estimand]
            # Reconcile the country-vector mean to the Task 13 ATT before any
            # inferential quantity is formed, preventing denominator drift.
            contrast_mean = float(np.mean(vector.to_numpy(dtype=float)))
            if not math.isfinite(reported_effects[estimand]) or abs(
                contrast_mean - reported_effects[estimand]
            ) > IDENTITY_TOLERANCE:
                raise IngestError(
                    f"{pair}/{estimand}: country-contrast mean {contrast_mean!r} "
                    f"does not reconcile with reported ATT {reported_effects[estimand]!r}."
                )
            # Compute the effect and its dispersion standard error.
            effect, standard_error = compute_country_dispersion_se(vector)
            if abs(effect - reported_effects[estimand]) > IDENTITY_TOLERANCE:
                raise IngestError(f"{pair}/{estimand}: inference effect drifted from Task 13.")
            # Compute the test statistic and the degrees of freedom.
            statistic, degrees_of_freedom = compute_t_statistic(
                effect, standard_error, len(vector)
            )
            # Compute the two-sided probability.
            probability: float = compute_two_sided_p(statistic, degrees_of_freedom)
            # Assign the marker exclusively from the authoritative two-sided
            # Student-t probability; critical values are metadata only.
            marker, critical_values = assign_significance_stars(
                statistic, degrees_of_freedom
            )
            # Assert the marker's agreement with the probability route.
            assert_star_consistency(
                marker,
                probability,
                statistic=statistic,
                degrees_of_freedom=degrees_of_freedom,
            )
            # Assert the probability's reproducibility.
            assert_p_reproducible(probability, statistic, degrees_of_freedom)
            # Propagate the mathematically explicit zero-dispersion state.
            zero_dispersion_state = (
                "positive_effect_zero_se"
                if standard_error == 0.0 and effect > 0.0
                else "negative_effect_zero_se"
                if standard_error == 0.0 and effect < 0.0
                else "zero_effect_zero_se"
                if standard_error == 0.0
                else "positive_se"
            )
            # Bind each ledger row to the complete labelled country vector
            # through the same canonical serializer used by the checker.
            contrast_sha256 = _freeze_country_contrast(
                vector, f"{pair}/{estimand} inference producer"
            )
            # Record the estimand-specific reference identity required for
            # publication matching.
            if estimand == WAVE_ONE_LABEL:
                reference_start, reference_end = wave_one_window
            elif estimand == WAVE_TWO_LABEL:
                reference_start, reference_end = wave_two_window
            else:
                reference_start, reference_end = pre_window
            # Append the ledger row of the estimand.
            ledger_rows.append(
                {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "estimand": estimand,
                    "att": float(effect),
                    "se": float(standard_error),
                    "t_statistic": float(statistic),
                    "p_value": float(probability),
                    "stars": str(marker),
                    "degrees_of_freedom": int(degrees_of_freedom),
                    "n_targets": int(len(vector)),
                    "stars_source": STAR_SOURCE,
                    "probability_source": PRIMARY_PROBABILITY_SOURCE,
                    "estimator": COUNTRY_DISPERSION_ESTIMATOR,
                    "tail_convention": STUDENT_T_TAIL_CONVENTION,
                    "contrast_sha256": contrast_sha256,
                    "zero_dispersion_state": zero_dispersion_state,
                    "reference_start": int(reference_start),
                    "reference_end": int(reference_end),
                    "post_start": int(post_window[0]),
                    "post_end": int(post_window[1]),
                }
            )
            # Append the critical-value row of the estimand.
            critical_rows.append(
                {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "estimand": estimand,
                    "degrees_of_freedom": int(degrees_of_freedom),
                    "crit_10pct": float(critical_values.get("*", float("nan"))),
                    "crit_5pct": float(critical_values.get("**", float("nan"))),
                    "crit_1pct": float(critical_values.get("***", float("nan"))),
                }
            )
    # Summarise the quarantine audit in the log so the quarantined targets are
    # visible in the inference run's provenance (remediation B-17).
    if quarantine_log:
        # Compose the per-pair-and-estimand quarantine tally.
        quarantine_summary: Dict[str, int] = {}
        # Iterate over the quarantine rows.
        for entry in quarantine_log:
            # Compose the summary key.
            summary_key: str = (
                f"{entry['policy']}/{entry['outcome']}/{entry['estimand']}"
            )
            # Increment the key's count.
            quarantine_summary[summary_key] = (
                quarantine_summary.get(summary_key, 0) + 1
            )
        # Log the quarantine tally at warning level.
        logger.warning(
            "contrast quarantine applied: %s.", quarantine_summary
        )
    # Assemble the ledger keyed on the partition and the estimand.
    ledger: pd.DataFrame = pd.DataFrame(ledger_rows).set_index(
        ["policy", "outcome", "estimand"]
    )
    # Assert that the ledger covers every partition and estimand pair.
    expected_keys = {
        (pair[0], pair[1], estimand)
        for pair in gap_result.gaps.keys()
        for estimand in ESTIMAND_ORDER
    }
    # Refuse an incomplete ledger.
    if set(ledger.index) != expected_keys:
        # The failure names the missing and unexpected keys.
        missing = sorted(expected_keys - set(ledger.index))
        # Compose the unexpected keys.
        unexpected = sorted(set(ledger.index) - expected_keys)
        # Raise with both sides attached.
        raise IngestError(
            f"Ledger coverage diverges: missing {missing}, unexpected {unexpected}."
        )
    # Assert the isolation of the marker's source.
    assert_stars_procedure_isolation(ledger, contrasts)
    # Compare supplied manuscript cells under their publication metadata and
    # retain the cell-level audit rather than rounding authoritative outputs.
    publication_audit = assert_declared_inference(
        ledger,
        expected_se=expected_se,
        expected_stars=expected_stars,
        expected_probability=expected_probability,
    )
    # Serialize the ledger when a destination was supplied.
    serialize_inference_ledger(ledger, destination)
    # Compose the critical-value table keyed on the partition and the estimand.
    critical_table: pd.DataFrame = pd.DataFrame(critical_rows).set_index(
        ["policy", "outcome", "estimand"]
    )
    # Compute the canonical digest bound to the input gap fingerprint.
    fingerprint: str = freeze_inference_ledger(
        ledger, gap_result.fingerprint_sha256
    )
    # Accumulate the recorded notes.
    notes: List[str] = [
        "the country-dispersion standard error treats countries as independent "
        "and does not propagate first-stage uncertainty in the pooled "
        "coefficient, the factor proxy or the country loadings.",
        "markers are assigned exclusively by the Student-t test of this module "
        "and are re-derived from the probability before the ledger is frozen.",
        "zero country dispersion is propagated explicitly: nonzero effects "
        "use signed infinite t limits with p=0, while an exactly zero effect "
        "uses the conservative t=0, p=1 null limit.",
        "the ledger is append-only; each run carries the fingerprint of the gap "
        "matrix it was computed from.",
    ]
    # Record the comparison mode note.
    notes.append(
        "declared inference values compared."
        if (expected_se or expected_stars or expected_probability)
        else "declared inference values not supplied; comparison skipped."
    )
    # Assemble the immutable result.
    result: InferenceResult = InferenceResult(
        ledger=ledger,
        contrasts=contrasts,
        critical_values=critical_table,
        publication_audit=publication_audit,
        ledger_fingerprint_sha256=fingerprint,
        input_gap_fingerprint=gap_result.fingerprint_sha256,
        computed_utc=current_utc_iso(),
        notes=tuple(notes),
    )
    # Log the completed inference with its digest prefix.
    logger.info(
        "country-dispersion inference complete: %d ledger row(s), fingerprint %s.",
        len(ledger),
        fingerprint[:16],
    )
    # Return the assembled result.
    return result
````

````python
# Task 15: Orchestrator 1 - Baseline Full Pipeline

# ==============================================================================
# Task 15: Orchestrator 1 - Baseline Full Pipeline
# ==============================================================================

# Declare the identifying field names of the interface.
CONTEXT_FIELDS: Final[Tuple[str, str]] = ("policy_name", "outcome_name")
# Declare the declared number of projection years of the gap matrix.
PROJECTION_YEARS: Final[int] = PROJECTION_END_YEAR - PROJECTION_START_YEAR + 1
# Declare the declared steps of the composition, in their required order, so
# the log coverage can be asserted.
DECLARED_STEPS: Final[Tuple[str, ...]] = (
    "task9_partition_fetch",
    "task10_factor_construction",
    "task11_pooled_fit",
    "task12_counterfactual_projection",
    "task13_gap_and_effects",
    "task14_country_dispersion_inference",
)
# Declare the mode label of a run that used the configuration's windows.
MODE_CONFIG: Final[str] = "config_windows"
# Declare the mode label of a run whose windows were overridden.
MODE_SWEEP: Final[str] = "sweep_variant"
# Declare the reserved metadata key carried by the effective window mapping.
WINDOW_PROVENANCE_KEY: Final[str] = "__provenance__"


# Declare the frozen interface of the baseline estimation.
@dataclass(frozen=True)
class BaselineResult:
    """
    Frozen baseline result of one policy-outcome pair.

    The interface is frozen at the complete Task 15 contract: it carries the
    reported estimates together with fitted-object lineage, effective-window
    provenance and the independent-chain validation ledger needed to audit
    their scientific meaning.

    Attributes
    ----------
    policy_name : str
        The policy instrument identifying the run.
    outcome_name : str
        The canonical outcome variable identifying the run.
    att_wave_1 : float
        The level effect over the first wave window.
    att_wave_2 : float
        The level effect over the second wave window.
    att_incremental : float
        The incremental effect, which is the study's headline estimand.
    se_cd : float
        The country-dispersion standard error of the incremental effect.
    stars : str
        The significance marker of the incremental effect, assigned by the
        Student-t test alone.
    country_window_means : pandas.DataFrame
        The per-country window means indexed on the country and columned by
        the wave labels.
    incremental_contrasts : pandas.Series
        The per-country incremental contrasts indexed on the country.
    gaps : pandas.DataFrame
        The gap matrix indexed on the country and columned by the projection
        years.
    pre_escalation_levels : pandas.Series
        The per-country pre-escalation levels indexed on the country.
    fit_provenance : mapping
        Actual fitted objects, immutable content snapshots, solver metadata,
        partition/input lineage and the chain fingerprint used for freshness
        and wrapper-versus-explicit-chain checks.
    effective_windows : mapping
        The canonical wave/reference/post windows and their resolution
        provenance used by every aggregation consumer.
    chain_validation : pandas.DataFrame
        One row per intermediate checked against the independently rerun
        explicit stage chain.
    """
    # The policy instrument identifying the run.
    policy_name: str
    # The canonical outcome variable identifying the run.
    outcome_name: str
    # The level effect over the first wave window.
    att_wave_1: float
    # The level effect over the second wave window.
    att_wave_2: float
    # The incremental effect, which is the study's headline estimand.
    att_incremental: float
    # The country-dispersion standard error of the incremental effect.
    se_cd: float
    # The significance marker of the incremental effect.
    stars: str
    # The per-country window means indexed on the country.
    country_window_means: pd.DataFrame
    # The per-country incremental contrasts indexed on the country.
    incremental_contrasts: pd.Series
    # The gap matrix indexed on the country.
    gaps: pd.DataFrame
    # The per-country pre-escalation levels indexed on the country.
    pre_escalation_levels: pd.Series
    # Actual fit objects, snapshots, solver metadata and lineage.
    fit_provenance: Mapping[str, Any]
    # Canonical effective reporting windows and resolution provenance.
    effective_windows: Mapping[str, Any]
    # Wrapper-versus-explicit-chain intermediate comparison ledger.
    chain_validation: pd.DataFrame


# Declare the frozen result of the full pipeline.
@dataclass(frozen=True)
class BaselinePipelineResult:
    """
    Frozen result of the baseline pipeline across the registered pairs.

    Attributes
    ----------
    results : dict
        The baseline results keyed on the policy and outcome pair.
    validation : pandas.DataFrame
        One row per pair and step recording the gate outcomes.
    freshness : pandas.DataFrame
        One row per pair recording the object identities of its own factor,
        coefficient and loading objects.
    effective_windows : mapping
        One canonical frozen window mapping shared by every pair run.
    declared_comparisons : pandas.DataFrame
        Cell-level publication-aware comparison ledger.
    collection_fingerprint_sha256 : str
        Digest of the collected results.
    completed_utc : str
        ISO-8601 UTC timestamp of the run.
    notes : tuple of str
        Recorded decisions, including the resolved windows and any sweep
        variant.
    """
    # The baseline results keyed on the pair.
    results: Mapping[Tuple[str, str], BaselineResult]
    # One row per pair and step recording the gate outcomes.
    validation: pd.DataFrame
    # One row per pair recording the object identities.
    freshness: pd.DataFrame
    # Canonical effective windows shared by the collection.
    effective_windows: Mapping[str, Any]
    # Cell-level publication comparison ledger.
    declared_comparisons: pd.DataFrame
    # Digest of the collected results.
    collection_fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the run.
    completed_utc: str
    # Recorded decisions of the run.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 1: The interface, the configuration guard and the windows
# -------------------------------------------------------------------------------------------------------------------------------


# Emit one structured log record for a completed sub-call.
def log_step(
    context: Tuple[str, str], step: str, **fields: Any
) -> None:
    """
    Emit one structured log record for a completed sub-call.

    Parameters
    ----------
    context : tuple of str
        The policy and outcome names of the pair.
    step : str
        The declared step name.
    **fields : Any
        The step-specific measurements, which are encoded verbatim.

    Returns
    -------
    None
        The function returns nothing; it emits a single log record.
    """
    # Compose the structured payload with the identifying fields first.
    payload: Dict[str, Any] = {
        "policy_name": context[0],
        "outcome_name": context[1],
        "step": step,
    }
    # Attach the step-specific measurements.
    for key, value in fields.items():
        # Convert NumPy scalars to native Python types so the payload is
        # serializable and stable across library versions.
        if isinstance(value, (np.integer, np.floating, np.bool_)):
            # Record the native scalar.
            payload[key] = value.item()
        else:
            # Record the value verbatim.
            payload[key] = value
    # Emit the record with sorted keys so two identical runs produce
    # identical lines.
    logger.info(json.dumps(payload, sort_keys=True, default=str))


# Compute the canonical fingerprint of the study configuration.
def freeze_config(study_config: Mapping[str, Any]) -> str:
    """
    Compute the canonical fingerprint of the study configuration.

    Parameters
    ----------
    study_config : mapping
        The study configuration.

    Returns
    -------
    str
        The hexadecimal digest of the canonical serialization.
    """
    # Serialize the configuration with sorted keys and a deterministic
    # fallback for non-JSON values.
    serialized: str = json.dumps(study_config, sort_keys=True, default=str)
    # Digest the encoded payload.
    return hashlib.sha256(serialized.encode("utf-8")).hexdigest()


# Digest the complete mutable structure behind a partition registry.
def _partition_registry_structure_digest(registry: PartitionRegistry) -> str:
    """Return a digest of registry tables, pair rosters and exclusions."""
    if not isinstance(registry, PartitionRegistry) or not registry.partitions:
        raise IngestError("Cannot digest an empty or malformed partition registry.")
    partition_payload: Dict[str, Any] = {}
    for pair, partition in sorted(registry.partitions.items()):
        key = f"{pair[0]}|{pair[1]}"
        partition_payload[key] = {
            "identity": (str(partition.policy), str(partition.outcome)),
            "targets": tuple(str(code) for code in partition.targets),
            "donors": tuple(str(code) for code in partition.donors),
            "exclusions": partition.exclusions,
        }
    return _content_digest(
        {
            "registry": registry.registry,
            "counts": registry.counts,
            "exclusions": registry.exclusions,
            "partitions": partition_payload,
        }
    )


# Assert that the study configuration was not mutated by the run.
def assert_config_unchanged(
    before: str, after: str, context: Tuple[str, str]
) -> None:
    """
    Assert that the study configuration was not mutated.

    Parameters
    ----------
    before : str
        The configuration fingerprint taken before the run.
    after : str
        The configuration fingerprint taken after the run.
    context : tuple of str
        The policy and outcome names of the pair.

    Returns
    -------
    None
        The function returns nothing when the configuration is unchanged.

    Raises
    ------
    IngestError
        If the two fingerprints differ, which would mean the orchestrator
        mutated a configuration it receives read-only.
    """
    # Refuse a mutated configuration.
    if before != after:
        # The failure names the pair and both fingerprints.
        raise IngestError(
            f"{context[0]}/{context[1]}: study configuration changed during the "
            f"baseline run ({before[:12]} to {after[:12]}); the orchestrator must "
            f"not mutate it."
        )


# Resolve the reporting windows against the authoritative configuration.
def resolve_reporting_windows(
    study_config: Mapping[str, Any],
    reporting_windows: Optional[Mapping[str, Tuple[int, int]]] = None,
    context: Tuple[str, str] = ("baseline", "baseline"),
) -> Tuple[Dict[str, Any], str]:
    """
    Resolve the reporting windows against the authoritative configuration.

    Parameters
    ----------
    study_config : mapping
        The study configuration, which is authoritative for the windows.
    reporting_windows : mapping of str to tuple of int, optional
        An explicit override of one or more windows. An override that differs
        from the configuration is applied but logged as a sweep variant.
    context : tuple of str, optional
        The policy and outcome names, used in the log record.

    Returns
    -------
    tuple
        The resolved windows keyed by their declared names and the mode label
        of the resolution, which is either the configuration mode or the
        sweep-variant mode.
    """
    # Refuse a malformed configuration rather than silently using defaults.
    if not isinstance(study_config, Mapping):
        raise IngestError("Study configuration must be a mapping.")
    estimation = study_config.get("estimation", {})
    if estimation is None:
        estimation = {}
    if not isinstance(estimation, Mapping):
        raise IngestError("study_config.estimation must be a mapping.")
    window_block = estimation.get("windows", {})
    if window_block is None:
        window_block = {}
    if not isinstance(window_block, Mapping):
        raise IngestError("study_config.estimation.windows must be a mapping.")
    # Normalize an external window declaration without leaking raw sequence
    # conversion errors through the public configuration boundary.
    def validate_declaration(raw: Any, label: str) -> Tuple[int, int]:
        if isinstance(raw, (str, bytes, Mapping)):
            raise IngestError(f"{label} must be a two-element calendar sequence.")
        try:
            candidate = tuple(raw)
        except TypeError as exc:
            raise IngestError(
                f"{label} must be a two-element calendar sequence."
            ) from exc
        return _validate_att_window(candidate, label)
    canonical_keys = {
        WAVE_ONE_LABEL, WAVE_TWO_LABEL, "pre_escalation", "post_escalation"
    }
    unknown_config_keys = sorted(set(window_block) - canonical_keys)
    if unknown_config_keys:
        raise IngestError(
            f"Unknown canonical reporting-window key(s) {unknown_config_keys}."
        )
    # Resolve configuration declarations while retaining their exact source.
    values: Dict[str, Any] = {
        WAVE_ONE_LABEL: window_block.get(WAVE_ONE_LABEL, WAVE_ONE_WINDOW),
        WAVE_TWO_LABEL: window_block.get(WAVE_TWO_LABEL, WAVE_TWO_WINDOW),
    }
    sources: Dict[str, str] = {
        WAVE_ONE_LABEL: (
            "estimation.windows.wave_1_level"
            if WAVE_ONE_LABEL in window_block else "module_default.wave_1_level"
        ),
        WAVE_TWO_LABEL: (
            "estimation.windows.wave_2_level"
            if WAVE_TWO_LABEL in window_block else "module_default.wave_2_level"
        ),
    }
    # Explicitly reconcile the legacy preferred-reference declaration with
    # the canonical pre-escalation path.
    canonical_pre = window_block.get("pre_escalation")
    legacy_pre = estimation.get("preferred_reference_window")
    if canonical_pre is not None and legacy_pre is not None:
        canonical_pre_valid = validate_declaration(
            canonical_pre, "estimation.windows.pre_escalation"
        )
        legacy_pre_valid = validate_declaration(
            legacy_pre, "estimation.preferred_reference_window"
        )
        if canonical_pre_valid != legacy_pre_valid:
            raise IngestError(
                "Conflicting reference-window declarations at "
                "estimation.windows.pre_escalation and "
                "estimation.preferred_reference_window."
            )
        values["pre_escalation"] = canonical_pre_valid
        sources["pre_escalation"] = (
            "reconciled:estimation.windows.pre_escalation="
            "estimation.preferred_reference_window"
        )
    elif canonical_pre is not None:
        values["pre_escalation"] = canonical_pre
        sources["pre_escalation"] = "estimation.windows.pre_escalation"
    elif legacy_pre is not None:
        values["pre_escalation"] = legacy_pre
        sources["pre_escalation"] = "estimation.preferred_reference_window"
    else:
        values["pre_escalation"] = PRE_ESCALATION_WINDOW
        sources["pre_escalation"] = "module_default.pre_escalation"
    # The incremental post window follows the configured second-wave horizon
    # unless explicitly declared; conflicting explicit declarations fail.
    explicit_post = window_block.get("post_escalation")
    if explicit_post is not None:
        values["post_escalation"] = explicit_post
        sources["post_escalation"] = "estimation.windows.post_escalation"
    else:
        values["post_escalation"] = values[WAVE_TWO_LABEL]
        sources["post_escalation"] = "derived_from:wave_2_level"
    # Validate the configuration windows before applying a sweep.
    resolved: Dict[str, Tuple[int, int]] = {
        key: validate_declaration(values[key], f"effective {key}")
        for key in canonical_keys
    }
    configured_resolved = dict(resolved)
    if explicit_post is not None and resolved["post_escalation"] != resolved[WAVE_TWO_LABEL]:
        raise IngestError(
            "Configured post_escalation conflicts with wave_2_level; the level "
            "and incremental post horizons must be identical."
        )
    # Validate and apply explicit sweep overrides. A single horizon override
    # is propagated to both Wave 2 and the incremental post window.
    overrides: Dict[str, Tuple[int, int]] = {}
    if reporting_windows is not None:
        if not isinstance(reporting_windows, Mapping):
            raise IngestError("Reporting-window override must be a mapping.")
        unknown_overrides = sorted(set(reporting_windows) - canonical_keys)
        if unknown_overrides:
            raise IngestError(
                f"Unknown reporting-window override key(s) {unknown_overrides}."
            )
        for key, raw_value in reporting_windows.items():
            overrides[str(key)] = validate_declaration(
                raw_value, f"reporting override {key}"
            )
        if WAVE_TWO_LABEL in overrides and "post_escalation" not in overrides:
            overrides["post_escalation"] = overrides[WAVE_TWO_LABEL]
        if "post_escalation" in overrides and WAVE_TWO_LABEL not in overrides:
            overrides[WAVE_TWO_LABEL] = overrides["post_escalation"]
        if (
            WAVE_TWO_LABEL in overrides
            and "post_escalation" in overrides
            and overrides[WAVE_TWO_LABEL] != overrides["post_escalation"]
        ):
            raise IngestError(
                "Sweep wave_2_level and post_escalation overrides conflict."
            )
        for key, value in overrides.items():
            if value != resolved[key]:
                resolved[key] = value
                sources[key] = f"explicit_override:{key}"
    # Require all windows to lie within the available projected calendar.
    for key, span in resolved.items():
        if span[0] < PROJECTION_START_YEAR or span[1] > PROJECTION_END_YEAR:
            raise IngestError(
                f"Effective {key} window {span} leaves available projection "
                f"years {PROJECTION_START_YEAR}-{PROJECTION_END_YEAR}."
            )
    if resolved[WAVE_ONE_LABEL][1] >= resolved[WAVE_TWO_LABEL][0]:
        raise IngestError("Effective Wave 1 and Wave 2 windows overlap or are unordered.")
    if resolved["pre_escalation"][1] >= resolved["post_escalation"][0]:
        raise IngestError("Effective reference and post windows overlap or are unordered.")
    if resolved[WAVE_TWO_LABEL] != resolved["post_escalation"]:
        raise IngestError("Effective Wave 2 and post-escalation horizons diverge.")
    divergent_keys = sorted(
        key for key, value in overrides.items()
        if value != configured_resolved[key]
    )
    mode = MODE_SWEEP if divergent_keys else MODE_CONFIG
    # Attach immutable-by-convention provenance to the effective mapping while
    # retaining the two-item return and ordinary key lookup contract.
    window_payload = {key: list(resolved[key]) for key in sorted(canonical_keys)}
    fingerprint = hashlib.sha256(
        json.dumps(window_payload, sort_keys=True).encode("utf-8")
    ).hexdigest()
    provenance = {
        "mode": mode,
        "sources": dict(sorted(sources.items())),
        "overridden_keys": tuple(divergent_keys),
        "available_years": (PROJECTION_START_YEAR, PROJECTION_END_YEAR),
        "gap_grid_policy": (
            f"fixed_full_projection_{PROJECTION_START_YEAR}_{PROJECTION_END_YEAR}"
        ),
        "fingerprint_sha256": fingerprint,
    }
    effective: Dict[str, Any] = {**resolved, WINDOW_PROVENANCE_KEY: provenance}
    log_step(
        context,
        "reporting_window_resolution",
        mode=mode,
        overridden=divergent_keys,
        fingerprint=fingerprint,
    )
    return effective, mode


# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 2: Compose the orchestrator from the prior tasks
# -------------------------------------------------------------------------------------------------------------------------------


# Execute the explicit stage chain of one policy-outcome pair.
def _run_baseline_stage_chain(
    policy_name: str,
    outcome_name: str,
    data: Mapping[str, Any],
    windows: Mapping[str, Tuple[int, int]],
) -> BaselineResult:
    """
    Execute the explicit Task 9-through-14 stage chain for one partition.

    Parameters
    ----------
    policy_name : str
        The policy instrument of the pair.
    outcome_name : str
        The canonical outcome variable of the pair.
    data : mapping
        The frozen inputs under the declared keys: the partition registry,
        the analysis panel and the lagged panel.
    windows : mapping of str to tuple of int
        The resolved reporting windows.

    Returns
    -------
    BaselineResult
        The frozen baseline result of the pair.

    Raises
    ------
    IngestError
        If a sub-step returns a shape diverging from its declaration, if the
        incremental identity fails, or if a component carries a missing
        value.
    """
    # Compose the pair context used by every log record and failure message.
    context: Tuple[str, str] = (str(policy_name), str(outcome_name))
    # Read the registry from the frozen data mapping.
    registry: PartitionRegistry = data["registry"]
    # Refuse a pair the registry does not carry.
    if context not in registry.partitions:
        # The failure names the pair.
        raise IngestError(f"The registry does not carry the pair {context}.")
    # Read the analysis panel from the frozen data mapping.
    panel: pd.DataFrame = data["panel"]
    # Read the lagged panel from the frozen data mapping.
    lagged: pd.DataFrame = data["lagged"]
    # Fetch the partition's target and donor tuples, which is the declared
    # first step of the composition.
    partition = registry.partitions[context]
    # Emit the partition record with both counts.
    log_step(
        context,
        "task9_partition_fetch",
        targets=len(partition.targets),
        donors=len(partition.donors),
    )
    # Build the factor objects of the pair, which instantiates a fresh donor
    # proxy for this pair alone.
    factor = assemble_partition_factor(
        context[0],
        context[1],
        partition.donors,
        sorted(list(partition.targets) + list(partition.donors)),
        panel,
        lagged,
        window=PRE_TREATMENT_WINDOW,
        factor_window=FACTOR_WINDOW,
    )
    # Initialise the per-unit dependent and covariate mappings.
    dependent_by_unit: Dict[str, np.ndarray] = {}
    # Initialise the per-unit covariate mapping.
    covariate_by_unit: Dict[str, np.ndarray] = {}
    # Initialise the partition's factor matrix.
    factor_matrix: Optional[np.ndarray] = None
    # Split each unit's design into its covariate and factor matrices.
    for unit in sorted(factor.matrices.keys()):
        # Read the unit's dependent and design matrices.
        dependent, pooling = factor.matrices[unit]
        # Split the design into covariates and factor columns, which is the
        # canonical parsing of the design's layout.
        covariates = pooling[:, 1 : 1 + len(CONTROL_COLUMNS)]
        # Compose the factor matrix from the intercept and the proxy.
        unit_factor = np.column_stack([pooling[:, 0], pooling[:, -1]])
        # Record the unit's covariate matrix.
        covariate_by_unit[str(unit)] = covariates
        # Record the unit's dependent matrix.
        dependent_by_unit[str(unit)] = dependent
        # Record the factor matrix from its first occurrence.
        if factor_matrix is None:
            # Record the partition's factor matrix.
            factor_matrix = unit_factor
        # Refuse a partition whose units carry divergent factor matrices.
        elif not np.allclose(factor_matrix, unit_factor, atol=1e-10):
            # The failure names the unit.
            raise IngestError(
                f"{context[0]}/{context[1]}: unit {unit} carries a factor matrix "
                f"diverging from the partition's."
            )
    # Refuse a partition without a factor matrix.
    if factor_matrix is None:
        # The failure states the precondition.
        raise IngestError(f"{context[0]}/{context[1]}: no unit carries a design.")
    # Assert the declared factor shape of years by two.
    assert_declared_shape(
        context,
        "task10_factor_construction",
        "F_hat",
        tuple(factor_matrix.shape),
        (PRE_TREATMENT_WINDOW[1] - PRE_TREATMENT_WINDOW[0] + 1, FACTOR_COLUMNS),
    )
    # Emit the factor record with the donor count and the conditioning.
    log_step(
        context,
        "task10_factor_construction",
        donors=factor.donor_count,
        units=len(factor.matrices),
        proxy_sd=float(factor.proxy.std(ddof=1)),
    )
    # Build the residual-maker of the partition.
    maker = build_residual_maker(factor_matrix)
    # Estimate the pooled coefficient over the union of targets and donors,
    # which is the declared sample for the pooled fit.
    beta, fit_diagnostics = estimate_pooled_coefficient(
        dependent_by_unit, covariate_by_unit, maker
    )
    # Assert the declared coefficient shape.
    assert_declared_shape(
        context,
        "task11_pooled_fit",
        "beta_hat",
        tuple(np.asarray(beta).shape),
        (len(CONTROL_COLUMNS),),
    )
    # Recover the country loadings for every unit of the estimation set.
    # Preserve the diagnostic's infinite value for a singular design; bound
    # only the arithmetic allowance of the geometric orthogonality audit.
    orthogonality_conditioning = (
        float(fit_diagnostics["condition_number"])
        if np.isfinite(float(fit_diagnostics["condition_number"]))
        else float(1.0 / np.finfo(float).eps)
    )
    country_loadings = estimate_unit_loadings(
        factor_matrix,
        dependent_by_unit,
        covariate_by_unit,
        beta,
        conditioning=orthogonality_conditioning,
    )
    # Assert the declared country-loading shape for every unit.
    for unit, alpha in country_loadings.items():
        # Gate the unit's loading vector.
        assert_declared_shape(
            context,
            "task11_pooled_fit",
            f"alpha_hat[{unit}]",
            tuple(np.asarray(alpha).shape),
            (FACTOR_COLUMNS,),
        )
    # Emit the fit record with the numerical route and the conditioning.
    log_step(
        context,
        "task11_pooled_fit",
        maker_route=maker.route,
        maker_condition_number=float(maker.condition_number),
        rank=float(fit_diagnostics["rank"]),
        rss=float(fit_diagnostics["rss"]),
    )
    # Estimate the covariate loadings of the pair's targets alone.
    target_loadings = estimate_covariate_loadings(
        factor_matrix, covariate_by_unit, partition.targets
    )
    # Assert the declared covariate-loading shape for every target.
    for target, loading in target_loadings.items():
        # Gate the target's loading matrix.
        assert_declared_shape(
            context,
            "task12_counterfactual_projection",
            f"lambda_hat[{target}]",
            tuple(np.asarray(loading).shape),
            (FACTOR_COLUMNS, len(CONTROL_COLUMNS)),
        )
    # Build the post-window factor matrix from the pair's own donor proxy.
    projection_factor, projection_years = build_projection_factor(factor.proxy)
    # Accumulate the projected rows of the pair.
    projected_rows: List[Dict[str, Any]] = []
    # Project every target's paths.
    for target in sorted(target_loadings.keys()):
        # Project the target's covariate and outcome paths.
        covariate_path, outcome_path = project_target_paths(
            target_loadings[target],
            np.asarray(beta),
            np.asarray(country_loadings[target]),
            projection_factor,
        )
        # Assert that both paths are complete.
        assert_paths_complete(covariate_path, outcome_path, target)
        # Emit the target's projected rows.
        for index, year in enumerate(projection_years):
            # Record the projected outcome of the year.
            projected_rows.append(
                {
                    "policy": context[0],
                    "outcome": context[1],
                    "country_code": target,
                    "year": int(year),
                    "counterfactual_outcome": float(outcome_path[index]),
                }
            )
    # Assemble the counterfactual frame the gap computation consumes.
    counterfactual_frame: pd.DataFrame = pd.DataFrame(projected_rows)
    # Emit the projection record with the target count and the window.
    log_step(
        context,
        "task12_counterfactual_projection",
        targets=len(target_loadings),
        projection_start=int(projection_years[0]),
        projection_end=int(projection_years[-1]),
    )
    # Compute the gap matrix on the fixed full projection grid. Reporting
    # variants alter only the effective aggregation windows, never the fitted
    # path or labelled gap grid, which keeps stage-chain comparisons coherent.
    gaps: pd.DataFrame = compute_gap_matrix(
        panel,
        context[1],
        partition.targets,
        counterfactual_frame,
        context[0],
        window=(PROJECTION_START_YEAR, PROJECTION_START_YEAR),
        window_end=(PROJECTION_END_YEAR, PROJECTION_END_YEAR),
    )
    # Assert the declared gap-matrix shape.
    assert_declared_shape(
        context,
        "task13_gap_and_effects",
        "g_hat",
        tuple(gaps.shape),
        (len(partition.targets), PROJECTION_YEARS),
    )
    # Aggregate the per-country window means.
    country_means: pd.DataFrame = aggregate_country_window_means(
        gaps,
        {WAVE_ONE_LABEL: windows[WAVE_ONE_LABEL], WAVE_TWO_LABEL: windows[WAVE_TWO_LABEL]},
    )
    # Aggregate the coalition level effects.
    level_effects: pd.Series = aggregate_level_effects(country_means)
    # Compute the per-country pre-escalation levels.
    pre_levels: pd.Series = compute_pre_escalation_levels(
        gaps, windows["pre_escalation"]
    )
    # Build the observed post-escalation window.
    observed_window: pd.DataFrame = build_observed_window(
        panel,
        context[1],
        tuple(str(code) for code in gaps.index),
        windows["post_escalation"],
    )
    # Compute the continuation gap through the shifted counterfactual.
    continuation_gap: pd.DataFrame = compute_continuation_gap(
        gaps, pre_levels, observed_window, windows["post_escalation"]
    )
    # Compute the incremental effect through the continuation route.
    att_incremental: float = compute_incremental_effect(continuation_gap)
    # Compute the incremental effect through the direct route.
    att_incremental_direct: float = compute_incremental_effect_alternative(
        gaps, windows["pre_escalation"], windows["post_escalation"]
    )
    # Assert the declared identity between the two routes.
    assert_incremental_identity(att_incremental, att_incremental_direct)
    # Compose the per-country incremental contrast vector.
    post_years: List[int] = list(
        range(
            int(windows["post_escalation"][0]),
            int(windows["post_escalation"][1]) + 1,
        )
    )
    # Compute the incremental contrasts as the post mean less the pre level.
    incremental_contrasts: pd.Series = gaps[post_years].mean(axis=1) - pre_levels
    # Carry the same labelled estimand identity emitted by the Task 14
    # contrast builder so schema-sensitive chain comparison is meaningful.
    incremental_contrasts.name = (
        f"{context[0]}|{context[1]}|{ESTIMAND_INCREMENTAL}"
    )
    # Emit the gap record with the effect and the identity deviation.
    log_step(
        context,
        "task13_gap_and_effects",
        att_wave_1=float(level_effects[WAVE_ONE_LABEL]),
        att_wave_2=float(level_effects[WAVE_TWO_LABEL]),
        att_incremental=float(att_incremental),
        identity_deviation=abs(float(att_incremental) - float(att_incremental_direct)),
    )
    # Compute the incremental effect and its dispersion standard error, which
    # are the headline inference quantities of the minimal interface.
    effect, se_cd = compute_country_dispersion_se(incremental_contrasts)
    # Assert that the contrast mean reproduces the incremental effect.
    if abs(float(effect) - float(att_incremental)) > 1e-10:
        # The failure names both values.
        raise IngestError(
            f"{context[0]}/{context[1]}: the incremental contrast mean "
            f"({effect!r}) diverges from the computed effect "
            f"({att_incremental!r})."
        )
    # Compute the test statistic and the degrees of freedom.
    statistic, degrees_of_freedom = compute_t_statistic(
        effect, se_cd, len(incremental_contrasts)
    )
    # Compute the two-sided probability.
    probability: float = compute_two_sided_p(statistic, degrees_of_freedom)
    # Assign the significance marker.
    stars, critical_values = assign_significance_stars(
        statistic, degrees_of_freedom
    )
    # Assert the marker's agreement with the probability route.
    assert_star_consistency(
        stars,
        probability,
        statistic=statistic,
        degrees_of_freedom=degrees_of_freedom,
    )
    # Assert the probability's reproducibility.
    assert_p_reproducible(probability, statistic, degrees_of_freedom)
    # Emit the inference record with the statistic, the probability and the
    # marker.
    log_step(
        context,
        "task14_country_dispersion_inference",
        se_cd=float(se_cd),
        t_statistic=float(statistic),
        p_value=float(probability),
        stars=str(stars),
        degrees_of_freedom=int(degrees_of_freedom),
    )
    # Retain the actual freshly fitted objects and immutable content snapshots
    # needed to prove independence and detect later mutation.
    fit_objects: Dict[str, Any] = {
        "factor_matrix": factor_matrix,
        "residual_maker": maker.matrix,
        "beta": np.asarray(beta),
        "country_loadings": country_loadings,
        "covariate_loadings": target_loadings,
        "counterfactual_paths": counterfactual_frame,
    }
    object_digests = {
        name: _content_digest(obj) for name, obj in fit_objects.items()
    }
    object_ids = {name: id(obj) for name, obj in fit_objects.items()}
    # Bind the fit to its partition, inputs, windows and numerical solver.
    lineage_payload: Dict[str, Any] = {
        "pair": context,
        "targets": tuple(sorted(str(code) for code in partition.targets)),
        "donors": tuple(sorted(str(code) for code in partition.donors)),
        "registry_fingerprint": str(registry.fingerprint_sha256),
        "registry_structure_digest": _partition_registry_structure_digest(registry),
        "panel_fingerprint": _content_digest(panel),
        "lagged_fingerprint": _content_digest(lagged),
        "window_fingerprint": str(
            windows.get(WINDOW_PROVENANCE_KEY, {}).get("fingerprint_sha256", "")
        ),
        "pre_treatment_window": PRE_TREATMENT_WINDOW,
        "projection_grid": (PROJECTION_START_YEAR, PROJECTION_END_YEAR),
        "solver": {
            "residual_maker_route": str(maker.route),
            "rank_convention": str(maker.rank_convention),
            "factor_rank": int(maker.factor_rank),
            "pooled_rank": int(fit_diagnostics["rank"]),
            "condition_number": float(fit_diagnostics["condition_number"]),
        },
    }
    lineage_sha256 = hashlib.sha256(
        json.dumps(lineage_payload, sort_keys=True, default=str).encode("utf-8")
    ).hexdigest()
    fit_provenance: Dict[str, Any] = {
        "objects": fit_objects,
        "object_ids": object_ids,
        "object_digests": object_digests,
        "lineage": lineage_payload,
        "lineage_sha256": lineage_sha256,
    }
    # Assemble the frozen baseline result of the pair.
    result: BaselineResult = BaselineResult(
        policy_name=context[0],
        outcome_name=context[1],
        att_wave_1=float(level_effects[WAVE_ONE_LABEL]),
        att_wave_2=float(level_effects[WAVE_TWO_LABEL]),
        att_incremental=float(att_incremental),
        se_cd=float(se_cd),
        stars=str(stars),
        country_window_means=country_means,
        incremental_contrasts=incremental_contrasts,
        gaps=gaps,
        pre_escalation_levels=pre_levels,
        fit_provenance=fit_provenance,
        effective_windows=copy.deepcopy(dict(windows)),
        chain_validation=pd.DataFrame(
            columns=["component", "primary_digest", "explicit_digest", "equal"]
        ),
    )
    # Assert that no component carries a missing value.
    assert_no_missing_values(context, result)
    # Return the assembled result.
    return result


# Build one independent Task 10-through-14 oracle for a single partition.
def _run_independent_orchestrator_chain(
    policy_name: str,
    outcome_name: str,
    data: Mapping[str, Any],
    windows: Mapping[str, Any],
) -> Mapping[str, Any]:
    """
    Execute the existing stage orchestrators on an isolated one-pair registry.

    This is deliberately not implemented through ``_run_baseline_stage_chain``:
    it exercises the public Task 10, 11, 12, 13 and 14 orchestrators, retains
    their actual fitted objects, and returns labelled scientific products for
    an independent comparison with the baseline wrapper.
    """
    # Resolve and validate the pair before constructing the isolated registry.
    context = (str(policy_name), str(outcome_name))
    registry: PartitionRegistry = data["registry"]
    if context not in registry.partitions:
        raise IngestError(f"The registry does not carry the pair {context}.")
    original = registry.partitions[context]
    # Copy every mutable partition component so the oracle cannot share state
    # with either the source registry or the wrapper execution.
    partition = Partition(
        policy=str(original.policy),
        outcome=str(original.outcome),
        targets=tuple(str(code) for code in original.targets),
        donors=tuple(str(code) for code in original.donors),
        exclusions=original.exclusions.copy(deep=True),
    )
    registry_frame = pd.DataFrame(
        [
            {
                "policy": context[0],
                "outcome": context[1],
                "targets": int(len(partition.targets)),
                "donors": int(len(partition.donors)),
                "target_codes": json.dumps(list(partition.targets)),
                "donor_codes": json.dumps(list(partition.donors)),
            }
        ]
    ).set_index(["policy", "outcome"])
    counts = registry_frame[["targets", "donors"]].copy(deep=True)
    isolated = PartitionRegistry(
        partitions={context: partition},
        registry=registry_frame,
        counts=counts,
        exclusions=partition.exclusions.copy(deep=True),
        fingerprint_sha256=freeze_partition_registry(registry_frame),
        built_utc=current_utc_iso(),
        notes=("isolated one-pair registry for independent baseline audit",),
    )
    # Execute the actual prior-task orchestrators on the copied one-pair input.
    factor_result = orchestrate_donor_factor(
        isolated,
        data["panel"],
        data["lagged"],
        window=PRE_TREATMENT_WINDOW,
        factor_window=FACTOR_WINDOW,
    )
    fit_result = orchestrate_pretreatment_fit(factor_result)
    counterfactual_result = orchestrate_counterfactual_projection(
        fit_result,
        factor_result,
        isolated,
        start_year=PROJECTION_START_YEAR,
        end_year=PROJECTION_END_YEAR,
        observed_panel=data["panel"],
    )
    gap_result = orchestrate_gap_and_att(
        counterfactual_result,
        data["panel"],
        isolated,
        wave_one=tuple(windows[WAVE_ONE_LABEL]),
        wave_two=tuple(windows[WAVE_TWO_LABEL]),
        pre_escalation=tuple(windows["pre_escalation"]),
        post_escalation=tuple(windows["post_escalation"]),
    )
    inference_result = orchestrate_country_dispersion_inference(
        gap_result,
        pre_window=tuple(windows["pre_escalation"]),
        post_window=tuple(windows["post_escalation"]),
    )
    # Canonicalize the public path product to the wrapper's exact schema/order.
    path_columns = [
        "policy", "outcome", "country_code", "year", "counterfactual_outcome"
    ]
    paths = counterfactual_result.outcome_paths.loc[:, path_columns].sort_values(
        ["country_code", "year"], kind="stable"
    ).reset_index(drop=True)
    # Reconstruct the pair-level tabular outputs from their labelled results.
    country_means = (
        gap_result.country_window_means.loc[
            (gap_result.country_window_means["policy"] == context[0])
            & (gap_result.country_window_means["outcome"] == context[1]),
            ["country_code", WAVE_ONE_LABEL, WAVE_TWO_LABEL],
        ]
        .set_index("country_code")
        .sort_index()
    )
    pre_levels = (
        gap_result.pre_escalation_levels.loc[
            (gap_result.pre_escalation_levels["policy"] == context[0])
            & (gap_result.pre_escalation_levels["outcome"] == context[1]),
            ["country_code", "pre_escalation_level"],
        ]
        .set_index("country_code")["pre_escalation_level"]
        .sort_index()
    )
    inference_row = inference_result.ledger.loc[
        (context[0], context[1], ESTIMAND_INCREMENTAL)
    ]
    incremental_row = gap_result.incremental_effects.loc[context]
    # Return actual fresh fitted objects as well as each reported product.
    return {
        "objects": {
            "factor_matrix": fit_result.intermediates[context]["factor_matrix"],
            "residual_maker": fit_result.intermediates[context]["residual_maker"],
            "beta": fit_result.betas[context],
            "country_loadings": fit_result.loadings[context],
            "covariate_loadings": counterfactual_result.loadings[context],
            "counterfactual_paths": paths,
        },
        "country_window_means": country_means,
        "incremental_contrasts": inference_result.contrasts[
            (context, ESTIMAND_INCREMENTAL)
        ].sort_index(),
        "gaps": gap_result.gaps[context].sort_index(),
        "pre_escalation_levels": pre_levels,
        "att_wave_1": float(
            gap_result.level_effects.loc[context, WAVE_ONE_LABEL]
        ),
        "att_wave_2": float(
            gap_result.level_effects.loc[context, WAVE_TWO_LABEL]
        ),
        "att_incremental": float(incremental_row["incremental_effect"]),
        "se_cd": float(inference_row["se"]),
        "stars": str(inference_row["stars"]),
        "stage_fingerprints": {
            "factor": factor_result.fingerprint_sha256,
            "fit": fit_result.fingerprint_sha256,
            "projection": counterfactual_result.fingerprint_sha256,
            "gaps": gap_result.fingerprint_sha256,
            "inference": inference_result.ledger_fingerprint_sha256,
        },
    }


# Compose and independently verify one baseline partition run.
def run_baseline_partition(
    policy_name: str,
    outcome_name: str,
    data: Mapping[str, Any],
    windows: Mapping[str, Any],
) -> BaselineResult:
    """
    Run the baseline wrapper and compare it with a fresh explicit stage chain.

    The two executions receive identical read-only inputs but construct all
    fitted objects independently. Every scientific output and retained fitted
    intermediate must have equal content, while every mutable fitted object
    must have a distinct identity. The primary run is returned with the full
    comparison ledger attached.
    """
    # Execute the wrapper and, independently, the actual prior-task
    # orchestrators over a copied one-pair registry.
    primary = _run_baseline_stage_chain(policy_name, outcome_name, data, windows)
    explicit = _run_independent_orchestrator_chain(
        policy_name, outcome_name, data, windows
    )
    context = (str(policy_name), str(outcome_name))
    rows: List[Dict[str, Any]] = []

    # Compare one content-bearing component and record its two digests.
    def compare_component(label: str, left: Any, right: Any) -> None:
        left_digest = _content_digest(left)
        right_digest = _content_digest(right)
        equal = left_digest == right_digest
        rows.append(
            {
                "component": label,
                "primary_digest": left_digest,
                "explicit_digest": right_digest,
                "equal": bool(equal),
            }
        )
        if not equal:
            raise IngestError(
                f"{context[0]}/{context[1]}: baseline wrapper diverges from "
                f"the explicit stage chain at {label}."
            )

    # Compare all reported and country-level scientific outputs.
    for label in (
        "att_wave_1", "att_wave_2", "att_incremental", "se_cd", "stars",
        "country_window_means", "incremental_contrasts", "gaps",
        "pre_escalation_levels",
    ):
        right = explicit[label]
        # The explicit gap orchestrator emits precisely the union of the
        # effective reporting windows; compare that labelled product against
        # the corresponding slice of the wrapper's fixed full projection grid.
        left = getattr(primary, label)
        if label == "gaps":
            left = left.loc[right.index, right.columns]
        compare_component(label, left, right)
    compare_component("effective_windows", primary.effective_windows, windows)
    # Compare every fitted intermediate snapshot and complete lineage.
    primary_provenance = primary.fit_provenance
    primary_objects = primary_provenance["objects"]
    explicit_objects = explicit["objects"]
    if set(primary_objects) != set(explicit_objects):
        raise IngestError(f"{context}: fitted-intermediate coverage diverges.")
    for name in sorted(primary_objects):
        compare_component(
            f"fit_object:{name}", primary_objects[name], explicit_objects[name]
        )
        # Independent equal estimates are valid, but mutable identity reuse is not.
        if primary_objects[name] is explicit_objects[name]:
            raise IngestError(
                f"{context}: explicit chain reused primary fitted object {name}."
            )
    # Record each public-orchestrator stage fingerprint as traceable evidence;
    # it is not compared to the wrapper's differently scoped serialization.
    for stage, digest in sorted(explicit["stage_fingerprints"].items()):
        if not isinstance(digest, str) or len(digest) != 64:
            raise IngestError(f"{context}: explicit {stage} fingerprint is malformed.")
        rows.append(
            {
                "component": f"explicit_stage_fingerprint:{stage}",
                "primary_digest": digest,
                "explicit_digest": digest,
                "equal": True,
            }
        )
    validation = pd.DataFrame(rows)
    if validation.empty or not bool(validation["equal"].all()):
        raise IngestError(f"{context}: explicit-chain validation is incomplete.")
    # Return the primary independently fitted result with its comparison audit.
    return replace(primary, chain_validation=validation)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 3: Validation gates, structured logging and error raising
# -------------------------------------------------------------------------------------------------------------------------------


# Assert a declared shape with the pair and the step named.
def assert_declared_shape(
    context: Tuple[str, str],
    step: str,
    label: str,
    observed: Tuple[int, ...],
    expected: Tuple[int, ...],
) -> None:
    """
    Assert a declared shape with the pair and the step named.

    Parameters
    ----------
    context : tuple of str
        The policy and outcome names of the pair.
    step : str
        The declared step name whose return is being gated.
    label : str
        The quantity's name, used in the failure message.
    observed : tuple of int
        The observed shape.
    expected : tuple of int
        The declared shape.

    Returns
    -------
    None
        The function returns nothing when the shapes agree.

    Raises
    ------
    IngestError
        If the observed shape diverges from the declared shape, with the
        pair, the step and both shapes named so the failure localizes to one
        sub-step.
    """
    # Refuse a divergent shape.
    if tuple(int(value) for value in observed) != tuple(int(value) for value in expected):
        # The failure localizes to the pair and the step.
        raise IngestError(
            f"{context[0]}/{context[1]}: step {step} returned {label} with shape "
            f"{tuple(observed)} against the declared {tuple(expected)}."
        )


# Assert that no component of a result carries a missing value.
def assert_no_missing_values(
    context: Tuple[str, str], result: BaselineResult
) -> None:
    """
    Assert that no component of a result carries a missing value.

    Parameters
    ----------
    context : tuple of str
        The policy and outcome names of the pair.
    result : BaselineResult
        The assembled baseline result.

    Returns
    -------
    None
        The function returns nothing when every component is populated.

    Raises
    ------
    IngestError
        If a scalar is not finite or a tabular component carries a missing
        cell, because a numerical failure must never be absorbed into a
        missing value.
    """
    # Refuse a non-finite scalar.
    for label, value in (
        ("ATT_W1", result.att_wave_1),
        ("ATT_W2", result.att_wave_2),
        ("ATT_Inc_2", result.att_incremental),
        ("SE_CD", result.se_cd),
    ):
        # Test the scalar's finiteness.
        if not np.isfinite(float(value)):
            # The failure names the pair and the scalar.
            raise IngestError(
                f"{context[0]}/{context[1]}: {label} is not finite ({value!r})."
            )
    # Refuse a tabular component carrying a missing cell.
    for label, frame in (
        ("country_window_means", result.country_window_means),
        ("gaps", result.gaps),
    ):
        # Test the frame for missing cells.
        if bool(frame.isna().to_numpy().any()):
            # The failure names the pair and the component.
            raise IngestError(
                f"{context[0]}/{context[1]}: {label} carries a missing cell."
            )
    # Refuse a series carrying a missing value.
    for label, series in (
        ("incremental_contrasts", result.incremental_contrasts),
        ("pre_escalation_levels", result.pre_escalation_levels),
    ):
        # Test the series for missing values.
        if bool(series.isna().any()):
            # The failure names the pair and the component.
            raise IngestError(
                f"{context[0]}/{context[1]}: {label} carries a missing value."
            )


# Assert that no object is shared between partitions.
# Digest any estimation object's content deterministically.
def _content_digest(obj: Any) -> str:
    """
    Digest any estimation object's content deterministically.

    Parameters
    ----------
    obj : Any
        The estimation object to digest, which may be a pandas.DataFrame, a
        pandas.Series, or a mapping whose values are digestible objects.

    Returns
    -------
    str
        The SHA-256 digest of the object's content, independent of object
        identity and of memory layout.

    Raises
    ------
    IngestError
        If the object's type carries no deterministic content digest.
    """
    # Digest a DataFrame through both labelled schema and row-hashed content.
    if isinstance(obj, pd.DataFrame):
        schema = {
            "shape": tuple(int(value) for value in obj.shape),
            "columns": tuple(repr(value) for value in obj.columns.tolist()),
            "column_names": tuple(repr(value) for value in obj.columns.names),
            "dtypes": tuple(str(value) for value in obj.dtypes.tolist()),
            "index_names": tuple(repr(value) for value in obj.index.names),
            "index_type": type(obj.index).__name__,
        }
        # Hash the frame's values in row order with its index values included.
        hashed_frame: pd.Series = pd.util.hash_pandas_object(obj, index=True)
        payload = (
            json.dumps(schema, sort_keys=True).encode("utf-8")
            + hashed_frame.to_numpy(dtype=np.uint64).tobytes()
        )
        return hashlib.sha256(payload).hexdigest()
    # Digest a Series through its labelled schema and row-hashed content.
    if isinstance(obj, pd.Series):
        schema = {
            "shape": tuple(int(value) for value in obj.shape),
            "name": repr(obj.name),
            "dtype": str(obj.dtype),
            "index_names": tuple(repr(value) for value in obj.index.names),
            "index_type": type(obj.index).__name__,
        }
        # Hash the series' values in row order with its index values included.
        hashed_series: pd.Series = pd.util.hash_pandas_object(obj, index=True)
        payload = (
            json.dumps(schema, sort_keys=True).encode("utf-8")
            + hashed_series.to_numpy(dtype=np.uint64).tobytes()
        )
        return hashlib.sha256(payload).hexdigest()
    # Digest an ndarray with dtype and shape so equal bytes under different
    # scientific axes cannot collide.
    if isinstance(obj, np.ndarray):
        array = np.ascontiguousarray(obj)
        payload = (
            str(array.dtype).encode("utf-8")
            + repr(tuple(array.shape)).encode("utf-8")
            + array.tobytes()
        )
        return hashlib.sha256(payload).hexdigest()
    # Digest a mapping of estimation objects part by part in sorted key order.
    if isinstance(obj, Mapping):
        # Initialise the folding digest.
        mapping_digest = hashlib.sha256()
        # Iterate over the mapping's keys in sorted order for determinism.
        for key in sorted(obj.keys(), key=str):
            # Update the digest with the key.
            mapping_digest.update(str(key).encode("utf-8"))
            # Update the digest with the part's own content digest.
            mapping_digest.update(_content_digest(obj[key]).encode("utf-8"))
        # Return the folded digest.
        return mapping_digest.hexdigest()
    # Digest sequences recursively while preserving their order and type.
    if isinstance(obj, (tuple, list)):
        sequence_digest = hashlib.sha256(type(obj).__name__.encode("utf-8"))
        for value in obj:
            sequence_digest.update(_content_digest(value).encode("utf-8"))
        return sequence_digest.hexdigest()
    # Digest JSON-compatible scalar metadata with explicit type identity.
    if obj is None or isinstance(obj, (str, int, float, bool, np.generic)):
        payload = f"{type(obj).__name__}|{repr(obj)}".encode("utf-8")
        return hashlib.sha256(payload).hexdigest()
    # Refuse an object whose content cannot be digested deterministically.
    raise IngestError(
        f"Object of type {type(obj).__name__} cannot be content-digested."
    )


# Assert that no estimation object is shared between partitions.
def assert_no_shared_state(
    freshness: pd.DataFrame,
    results: Optional[Mapping[Tuple[str, str], BaselineResult]] = None,
    data: Optional[Mapping[str, Any]] = None,
    effective_windows: Optional[Mapping[str, Any]] = None,
) -> None:
    """
    Assert that no mutable fitted state is shared between partitions.

    Parameters
    ----------
    freshness : pandas.DataFrame
        One row per pair recording actual fitted-object identities, immutable
        content snapshots and lineage identifiers.
    results : mapping of tuple to BaselineResult, optional
        The baseline results keyed on the pair. Required for inspecting the
        retained actual fitted objects and their derivations.
    data : mapping, optional
        The source registry, panel and lagged panel used by the fit. When
        supplied, every lineage fingerprint and registered roster is checked
        against these actual inputs.
    effective_windows : mapping, optional
        The collection's canonical effective windows, checked against every
        result and lineage record.

    Returns
    -------
    None
        The function returns nothing when no mutable object is aliased and
        every snapshot, solver, input and partition lineage is valid. Equal
        numerical contents and equal content hashes are explicitly allowed.

    Raises
    ------
    IngestError
        If actual mutable identity is reused, a retained object was mutated,
        coverage diverges, or fit lineage is missing/inconsistent.
    """
    if results is None:
        raise IngestError("Freshness validation requires retained fit provenance.")
    if data is None or any(key not in data for key in ("registry", "panel", "lagged")):
        raise IngestError("Freshness validation requires the fitted source inputs.")
    source_registry = data["registry"]
    if not isinstance(source_registry, PartitionRegistry):
        raise IngestError("Freshness source registry has the wrong type.")
    actual_registry_fingerprint = freeze_partition_registry(source_registry.registry)
    if actual_registry_fingerprint != str(source_registry.fingerprint_sha256):
        raise IngestError("Partition registry frame disagrees with its fingerprint.")
    actual_registry_structure = _partition_registry_structure_digest(source_registry)
    actual_panel_fingerprint = _content_digest(data["panel"])
    actual_lagged_fingerprint = _content_digest(data["lagged"])
    if effective_windows is None:
        raise IngestError("Freshness validation requires canonical effective windows.")
    actual_window_fingerprint = str(
        effective_windows.get(WINDOW_PROVENANCE_KEY, {}).get(
            "fingerprint_sha256", ""
        )
    )
    if not actual_window_fingerprint:
        raise IngestError("Effective-window fingerprint is absent.")
    if freshness.empty or freshness.index.has_duplicates:
        raise IngestError("Freshness ledger is empty or carries duplicate pair keys.")
    normalized_results = {(str(pair[0]), str(pair[1])): value for pair, value in results.items()}
    normalized_freshness = {(str(key[0]), str(key[1])) for key in freshness.index}
    if set(normalized_results) != normalized_freshness:
        raise IngestError("Freshness ledger and baseline-result coverage diverge.")

    # Recursively enumerate mutable fitted identities, including arrays nested
    # inside independently allocated loading dictionaries.
    def mutable_identities(obj: Any) -> Set[int]:
        identities: Set[int] = set()
        if isinstance(obj, (np.ndarray, pd.DataFrame, pd.Series, dict, list, set)):
            identities.add(id(obj))
        if isinstance(obj, Mapping):
            for value in obj.values():
                identities.update(mutable_identities(value))
        elif isinstance(obj, (list, tuple, set)):
            for value in obj:
                identities.update(mutable_identities(value))
        return identities

    identities_by_pair: Dict[Tuple[str, str], Set[int]] = {}
    for pair_key, result in normalized_results.items():
        provenance = result.fit_provenance
        required_provenance = {
            "objects", "object_ids", "object_digests", "lineage", "lineage_sha256"
        }
        if not required_provenance.issubset(provenance):
            raise IngestError(f"{pair_key}: fit provenance is incomplete.")
        objects = provenance["objects"]
        recorded_ids = provenance["object_ids"]
        recorded_digests = provenance["object_digests"]
        required_fit_objects = {
            "factor_matrix", "residual_maker", "beta", "country_loadings",
            "covariate_loadings", "counterfactual_paths",
        }
        if set(objects) != required_fit_objects:
            raise IngestError(
                f"{pair_key}: fitted-object coverage is incomplete or unexpected."
            )
        if set(objects) != set(recorded_ids) or set(objects) != set(recorded_digests):
            raise IngestError(f"{pair_key}: fitted-object provenance coverage diverges.")
        row = freshness.loc[pair_key]
        for name, obj in objects.items():
            current_digest = _content_digest(obj)
            if int(recorded_ids[name]) != id(obj):
                raise IngestError(f"{pair_key}/{name}: retained object identity changed.")
            if str(recorded_digests[name]) != current_digest:
                raise IngestError(f"{pair_key}/{name}: fitted object mutated after snapshot.")
            id_column, digest_column = f"{name}_id", f"{name}_digest"
            if id_column not in freshness.columns or digest_column not in freshness.columns:
                raise IngestError(f"Freshness ledger omits {name} provenance columns.")
            if int(row[id_column]) != id(obj) or str(row[digest_column]) != current_digest:
                raise IngestError(f"{pair_key}/{name}: freshness ledger disagrees with object.")
        lineage = provenance["lineage"]
        if tuple(lineage.get("pair", ())) != pair_key:
            raise IngestError(f"{pair_key}: partition lineage identifies another pair.")
        for field in ("targets", "donors", "registry_fingerprint", "registry_structure_digest", "panel_fingerprint", "lagged_fingerprint", "window_fingerprint", "solver"):
            if field not in lineage:
                raise IngestError(f"{pair_key}: lineage omits {field}.")
        if pair_key not in source_registry.partitions:
            raise IngestError(f"{pair_key}: source registry omits the fitted pair.")
        source_partition = source_registry.partitions[pair_key]
        if tuple(lineage["targets"]) != tuple(sorted(str(code) for code in source_partition.targets)):
            raise IngestError(f"{pair_key}: target-roster lineage disagrees with registry.")
        if tuple(lineage["donors"]) != tuple(sorted(str(code) for code in source_partition.donors)):
            raise IngestError(f"{pair_key}: donor-roster lineage disagrees with registry.")
        if str(lineage["registry_fingerprint"]) != actual_registry_fingerprint:
            raise IngestError(f"{pair_key}: registry fingerprint lineage is stale.")
        if str(lineage["registry_structure_digest"]) != actual_registry_structure:
            raise IngestError(f"{pair_key}: registry structure lineage is stale.")
        if str(lineage["panel_fingerprint"]) != actual_panel_fingerprint:
            raise IngestError(f"{pair_key}: panel lineage is stale.")
        if str(lineage["lagged_fingerprint"]) != actual_lagged_fingerprint:
            raise IngestError(f"{pair_key}: lagged-panel lineage is stale.")
        if str(lineage["window_fingerprint"]) != actual_window_fingerprint:
            raise IngestError(f"{pair_key}: window lineage is stale.")
        if _content_digest(result.effective_windows) != _content_digest(effective_windows):
            raise IngestError(f"{pair_key}: result windows diverge from collection windows.")
        # Cross-check fitted-object coverage, dimensions and solver metadata
        # against the actual registered rosters rather than trusting lineage.
        expected_units = set(source_partition.targets) | set(source_partition.donors)
        factor_matrix = np.asarray(objects["factor_matrix"])
        residual_maker = np.asarray(objects["residual_maker"])
        beta = np.asarray(objects["beta"])
        country_loadings = objects["country_loadings"]
        covariate_loadings = objects["covariate_loadings"]
        paths = objects["counterfactual_paths"]
        pre_rows = PRE_TREATMENT_WINDOW[1] - PRE_TREATMENT_WINDOW[0] + 1
        if factor_matrix.shape != (pre_rows, FACTOR_COLUMNS):
            raise IngestError(f"{pair_key}: factor-matrix shape lineage is invalid.")
        if residual_maker.shape != (pre_rows, pre_rows):
            raise IngestError(f"{pair_key}: residual-maker shape is invalid.")
        if beta.shape != (len(CONTROL_COLUMNS),):
            raise IngestError(f"{pair_key}: pooled coefficient shape is invalid.")
        if set(country_loadings) != expected_units or any(
            np.asarray(value).shape != (FACTOR_COLUMNS,)
            for value in country_loadings.values()
        ):
            raise IngestError(f"{pair_key}: country-loading roster or shape is invalid.")
        if set(covariate_loadings) != set(source_partition.targets) or any(
            np.asarray(value).shape != (FACTOR_COLUMNS, len(CONTROL_COLUMNS))
            for value in covariate_loadings.values()
        ):
            raise IngestError(f"{pair_key}: covariate-loading roster or shape is invalid.")
        expected_path_rows = len(source_partition.targets) * PROJECTION_YEARS
        if not isinstance(paths, pd.DataFrame) or len(paths) != expected_path_rows:
            raise IngestError(f"{pair_key}: counterfactual path coverage is invalid.")
        required_path_columns = {
            "policy", "outcome", "country_code", "year", "counterfactual_outcome"
        }
        if set(paths.columns) != required_path_columns or bool(
            paths.duplicated(["policy", "outcome", "country_code", "year"]).any()
        ):
            raise IngestError(f"{pair_key}: counterfactual path schema is invalid.")
        if set(paths["policy"].astype(str)) != {pair_key[0]} or set(
            paths["outcome"].astype(str)
        ) != {pair_key[1]}:
            raise IngestError(f"{pair_key}: counterfactual path identity is invalid.")
        if set(paths["country_code"].astype(str)) != set(source_partition.targets):
            raise IngestError(f"{pair_key}: counterfactual target roster is invalid.")
        if set(pd.to_numeric(paths["year"], errors="coerce")) != set(
            range(PROJECTION_START_YEAR, PROJECTION_END_YEAR + 1)
        ):
            raise IngestError(f"{pair_key}: counterfactual calendar is invalid.")
        solver = lineage["solver"]
        if not isinstance(solver, Mapping):
            raise IngestError(f"{pair_key}: solver lineage is malformed.")
        independent_factor_rank = int(np.linalg.matrix_rank(factor_matrix))
        if int(solver.get("factor_rank", -1)) != independent_factor_rank:
            raise IngestError(f"{pair_key}: solver factor rank disagrees with fitted F.")
        if str(solver.get("rank_convention", "")) != SVD_RANK_CONVENTION:
            raise IngestError(f"{pair_key}: solver rank convention is invalid.")
        if str(solver.get("residual_maker_route", "")) not in {ROUTE_SOLVE, ROUTE_PINV}:
            raise IngestError(f"{pair_key}: residual-maker route is invalid.")
        try:
            solver_condition = float(solver.get("condition_number"))
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError(f"{pair_key}: solver condition number is malformed.") from exc
        if math.isnan(solver_condition) or solver_condition < 0.0:
            raise IngestError(f"{pair_key}: solver condition number is invalid.")
        pooled_rank = solver.get("pooled_rank")
        if isinstance(pooled_rank, (bool, np.bool_)) or not isinstance(
            pooled_rank, (int, float, np.integer, np.floating)
        ) or not float(pooled_rank).is_integer() or not 0 <= int(pooled_rank) <= len(CONTROL_COLUMNS):
            raise IngestError(f"{pair_key}: pooled solver rank is invalid.")
        if result.chain_validation.empty or not bool(result.chain_validation["equal"].all()):
            raise IngestError(f"{pair_key}: independent-chain validation is absent or failed.")
        required_chain_components = {
            "fit_object:factor_matrix", "fit_object:residual_maker",
            "fit_object:beta", "fit_object:country_loadings",
            "fit_object:covariate_loadings", "fit_object:counterfactual_paths",
            "gaps", "att_wave_1", "att_wave_2", "att_incremental",
            "se_cd", "stars",
        }
        if not required_chain_components <= set(result.chain_validation["component"]):
            raise IngestError(f"{pair_key}: independent-chain coverage is incomplete.")
        lineage_digest = hashlib.sha256(
            json.dumps(lineage, sort_keys=True, default=str).encode("utf-8")
        ).hexdigest()
        if lineage_digest != str(provenance["lineage_sha256"]):
            raise IngestError(f"{pair_key}: fit lineage mutated after snapshot.")
        if "lineage_sha256" not in freshness.columns or str(row["lineage_sha256"]) != lineage_digest:
            raise IngestError(f"{pair_key}: freshness lineage digest disagrees.")
        identities_by_pair[pair_key] = mutable_identities(objects)
    # Reject genuine mutable aliasing across partitions, irrespective of
    # whether independent fits happen to produce identical numbers/hashes.
    sorted_pairs = sorted(identities_by_pair)
    for index, left_pair in enumerate(sorted_pairs):
        for right_pair in sorted_pairs[index + 1 :]:
            shared = identities_by_pair[left_pair] & identities_by_pair[right_pair]
            if shared:
                raise IngestError(
                    f"Freshness violation: {left_pair} and {right_pair} share "
                    f"{len(shared)} mutable fitted object(s)."
                )
    # Log the verified freshness.
    logger.info(
        "freshness verified across %d partition(s) using actual fitted "
        "objects, snapshots and lineage.",
        len(freshness),
    )


# Verify the baseline results against the declared study values.
def verify_against_declared(
    results: Mapping[Tuple[str, str], BaselineResult],
    expected_incremental: Optional[Mapping[Tuple[str, str], Any]] = None,
    expected_stars: Optional[Mapping[Tuple[str, str], Any]] = None,
    tolerance: float = 1.0e-12,
) -> pd.DataFrame:
    """
    Verify the baseline results against the declared study values.

    Parameters
    ----------
    results : mapping of tuple to BaselineResult
        The baseline results keyed on the pair.
    expected_incremental : mapping of tuple to object, optional
        Declared incremental publication cells. Scalars use the manuscript's
        three-decimal precision; structured cells may state value, units,
        displayed precision, numerical allowance, reference window and N.
    expected_stars : mapping of tuple to object, optional
        Declared star cells as strings or structured metadata mappings.
    tolerance : float, optional
        Numerical allowance kept separate from publication rounding.

    Returns
    -------
    pandas.DataFrame
        Cell-level discrepancy data with full identity, precision, rounding
        interval, allowance, unrounded value, deviation and status.

    Raises
    ------
    IngestError
        If an observed value diverges from its declared counterpart, or if a
        declared pair is absent from the results.
    """
    if isinstance(tolerance, (bool, np.bool_)) or not math.isfinite(float(tolerance)) or float(tolerance) < 0.0:
        raise IngestError("Baseline publication allowance must be finite and nonnegative.")
    normalized_results = {(str(pair[0]), str(pair[1])): value for pair, value in results.items()}
    audit_columns = [
        "cell_id", "policy", "outcome", "estimand", "reported_field",
        "units", "displayed_decimals", "reported_value", "unrounded_value",
        "rounding_lower", "rounding_upper", "numerical_allowance",
        "signed_deviation", "absolute_deviation", "comparison_status",
        "reference_start", "reference_end", "n_targets", "window_fingerprint",
    ]
    rows: List[Dict[str, Any]] = []

    # Validate exact dimensions, identities and provenance independently of
    # empirical publication rounding.
    for pair, result in normalized_results.items():
        if (str(result.policy_name), str(result.outcome_name)) != pair:
            raise IngestError(f"{pair}: BaselineResult identity is inconsistent.")
        if result.gaps.index.has_duplicates or result.gaps.columns.has_duplicates:
            raise IngestError(f"{pair}: baseline gap axes are duplicated.")
        expected_years = list(range(PROJECTION_START_YEAR, PROJECTION_END_YEAR + 1))
        if list(int(year) for year in result.gaps.columns) != expected_years:
            raise IngestError(f"{pair}: baseline gap calendar is not the fixed grid.")
        if not result.gaps.index.equals(result.incremental_contrasts.index) or not result.gaps.index.equals(result.pre_escalation_levels.index):
            raise IngestError(f"{pair}: baseline country axes diverge.")
        provenance = result.effective_windows.get(WINDOW_PROVENANCE_KEY, {})
        if not str(provenance.get("fingerprint_sha256", "")):
            raise IngestError(f"{pair}: effective-window provenance is absent.")

    # Normalize a two-part publication key.
    def normalize_pair(raw_pair: Any) -> Tuple[str, str]:
        if not isinstance(raw_pair, tuple) or len(raw_pair) != 2:
            raise IngestError(f"Baseline publication key {raw_pair!r} is malformed.")
        pair = (str(raw_pair[0]).strip(), str(raw_pair[1]).strip())
        if not all(pair) or pair not in normalized_results:
            raise IngestError(f"Baseline results omit the declared pair {raw_pair}.")
        return pair

    # Resolve native unscaled units.
    def baseline_units(outcome: str) -> str:
        if outcome in {"log_CPI", "log_real_GDPpc"}:
            return "log_points"
        if outcome in {"inflation_rate", "GDPpc_growth_rate", "inflation_att", "growth_att"}:
            return "decimal_rate (multiply by 100 for percentage points)"
        return "outcome_units"

    # Validate optional structured cell identity fields against the mapping
    # key and the exact baseline output field they purport to describe.
    def assert_cell_identity(
        pair: Tuple[str, str],
        declaration: Mapping[str, Any],
        reported_field: str,
    ) -> None:
        expected = {
            "policy": pair[0],
            "outcome": pair[1],
            "estimand": ESTIMAND_INCREMENTAL,
            "reported_field": reported_field,
        }
        for field, required in expected.items():
            if field in declaration and str(declaration[field]).strip() != required:
                raise IngestError(
                    f"{pair}: declared {field}={declaration[field]!r} conflicts "
                    f"with expected cell identity {required!r}."
                )

    # Normalize reference metadata without leaking raw tuple-conversion errors.
    def normalize_reference(raw: Any, label: str) -> Tuple[int, int]:
        if isinstance(raw, (str, bytes, Mapping)):
            raise IngestError(f"{label} is not a two-year window.")
        try:
            candidate = tuple(raw)
        except TypeError as exc:
            raise IngestError(f"{label} is not a two-year window.") from exc
        return _validate_att_window(candidate, label)

    # Validate a publication target count as one positive integral identity.
    def validate_target_count(raw: Any, expected: int, label: str) -> int:
        if isinstance(raw, (bool, np.bool_)):
            raise IngestError(f"{label} is Boolean rather than a target count.")
        try:
            numeric = float(raw)
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError(f"{label} is malformed.") from exc
        if (
            not math.isfinite(numeric)
            or not numeric.is_integer()
            or numeric <= 0.0
            or int(numeric) != int(expected)
        ):
            raise IngestError(
                f"{label} {raw!r} conflicts with expected count {expected}."
            )
        return int(numeric)

    # When both publication maps are supplied they must identify the same set
    # of manuscript cells; otherwise a marker could be paired to another ATT.
    if expected_incremental is not None and expected_stars is not None:
        incremental_keys = {normalize_pair(key) for key in expected_incremental}
        star_keys = {normalize_pair(key) for key in expected_stars}
        if incremental_keys != star_keys:
            raise IngestError(
                "Declared incremental and star cell identities do not align."
            )

    if expected_incremental is not None:
        for raw_pair, raw_declared in expected_incremental.items():
            pair = normalize_pair(raw_pair)
            result = normalized_results[pair]
            declaration = dict(raw_declared) if isinstance(raw_declared, Mapping) else {"value": raw_declared}
            assert_cell_identity(pair, declaration, "att")
            if "value" not in declaration and "published_value" in declaration:
                declaration["value"] = declaration["published_value"]
            if "value" not in declaration or isinstance(declaration["value"], (bool, np.bool_)):
                raise IngestError(f"{pair}: declared incremental value is absent or Boolean.")
            try:
                reported = float(declaration["value"])
                precision = float(declaration.get("displayed_decimals", 3))
                allowance = float(declaration.get("numerical_allowance", tolerance))
            except (TypeError, ValueError, OverflowError) as exc:
                raise IngestError(f"{pair}: publication value or precision is malformed.") from exc
            if not math.isfinite(reported):
                raise IngestError(f"{pair}: declared incremental value is non-finite.")
            if not math.isfinite(precision) or not precision.is_integer() or precision < 0.0:
                raise IngestError(f"{pair}: displayed precision must be a nonnegative integer.")
            if not math.isfinite(allowance) or allowance < 0.0:
                raise IngestError(f"{pair}: numerical allowance is invalid.")
            reference = tuple(result.effective_windows["pre_escalation"])
            declared_reference = normalize_reference(
                declaration.get("reference_window", reference),
                f"{pair} declared reference",
            )
            if declared_reference != reference:
                raise IngestError(f"{pair}: declared reference-window identity mismatch.")
            n_targets = len(result.incremental_contrasts)
            validate_target_count(
                declaration.get("n_targets", n_targets),
                n_targets,
                f"{pair} declared target count",
            )
            native_units = baseline_units(pair[1])
            units = str(declaration.get("units", native_units)).strip()
            if units != native_units:
                raise IngestError(
                    f"{pair}: publication units {units!r} conflict with native "
                    f"units {native_units!r}."
                )
            observed = float(result.att_incremental)
            decimals = int(precision)
            half_unit = 0.5 * 10.0 ** (-decimals)
            lower, upper = reported - half_unit, reported + half_unit
            deviation = observed - reported
            if lower <= observed <= upper:
                status = "within_publication_rounding"
            elif lower - allowance <= observed <= upper + allowance:
                status = "within_numerical_allowance"
            else:
                raise IngestError(
                    f"{pair}/incremental: publication mismatch; unrounded "
                    f"{observed!r}, interval [{lower!r}, {upper!r}], allowance "
                    f"{allowance!r}, deviation {deviation!r}."
                )
            rows.append(
                {
                    "cell_id": f"{pair[0]}/{pair[1]}/incremental/att",
                    "policy": pair[0], "outcome": pair[1],
                    "estimand": ESTIMAND_INCREMENTAL, "reported_field": "att",
                    "units": units, "displayed_decimals": decimals,
                    "reported_value": reported, "unrounded_value": observed,
                    "rounding_lower": lower, "rounding_upper": upper,
                    "numerical_allowance": allowance,
                    "signed_deviation": deviation,
                    "absolute_deviation": abs(deviation),
                    "comparison_status": status,
                    "reference_start": reference[0], "reference_end": reference[1],
                    "n_targets": n_targets,
                    "window_fingerprint": result.effective_windows[WINDOW_PROVENANCE_KEY]["fingerprint_sha256"],
                }
            )
    if expected_stars is not None:
        for raw_pair, raw_declared in expected_stars.items():
            pair = normalize_pair(raw_pair)
            result = normalized_results[pair]
            declaration = dict(raw_declared) if isinstance(raw_declared, Mapping) else {"value": raw_declared}
            assert_cell_identity(pair, declaration, "stars")
            if "value" not in declaration and "stars" not in declaration:
                raise IngestError(f"{pair}: declared star value is absent.")
            declared_marker = str(declaration.get("value", declaration.get("stars")))
            reference = tuple(result.effective_windows["pre_escalation"])
            declared_reference = normalize_reference(
                declaration.get("reference_window", reference),
                f"{pair} star reference",
            )
            if declared_reference != reference:
                raise IngestError(f"{pair}: declared star reference-window mismatch.")
            n_targets = len(result.incremental_contrasts)
            validate_target_count(
                declaration.get("n_targets", n_targets),
                n_targets,
                f"{pair} declared star target count",
            )
            if str(declaration.get("units", "marker")).strip() != "marker":
                raise IngestError(f"{pair}: star cell units must be 'marker'.")
            if str(result.stars) != declared_marker:
                raise IngestError(
                    f"{pair}: marker {result.stars!r} against declared "
                    f"{declared_marker!r}; exact procedure mismatch."
                )
            rows.append(
                {
                    "cell_id": f"{pair[0]}/{pair[1]}/incremental/stars",
                    "policy": pair[0], "outcome": pair[1],
                    "estimand": ESTIMAND_INCREMENTAL, "reported_field": "stars",
                    "units": "marker", "displayed_decimals": pd.NA,
                    "reported_value": declared_marker,
                    "unrounded_value": str(result.stars),
                    "rounding_lower": np.nan, "rounding_upper": np.nan,
                    "numerical_allowance": 0.0, "signed_deviation": np.nan,
                    "absolute_deviation": np.nan,
                    "comparison_status": "exact_student_t_marker",
                    "reference_start": reference[0], "reference_end": reference[1],
                    "n_targets": n_targets,
                    "window_fingerprint": result.effective_windows[WINDOW_PROVENANCE_KEY]["fingerprint_sha256"],
                }
            )
    audit = pd.DataFrame(rows, columns=audit_columns)
    if len(audit) and bool(audit["cell_id"].duplicated().any()):
        raise IngestError("Baseline publication audit has duplicate cell identities.")
    # Log the verification mode.
    logger.info(
        "declared baseline verification %s.",
        "performed" if (expected_incremental or expected_stars) else "skipped",
    )
    return audit


# Freeze the collected baseline results behind a canonical digest.
def freeze_baseline_collection(
    results: Mapping[Tuple[str, str], BaselineResult]
) -> str:
    """
    Compute the canonical digest of the collected baseline results.

    Parameters
    ----------
    results : mapping of tuple to BaselineResult
        The baseline results keyed on the pair.

    Returns
    -------
    str
        The hexadecimal digest of the canonical payload.

    Raises
    ------
    IngestError
        If the collection is empty.
    """
    # Refuse an empty collection.
    if not results:
        # The failure states the precondition.
        raise IngestError("Cannot freeze an empty baseline collection.")
    # Accumulate the canonical payload parts.
    parts: List[str] = []
    # Iterate over the pairs in sorted order.
    for pair in sorted(results.keys()):
        # Select the pair's result.
        result = results[pair]
        # Record the pair's headline scalars.
        parts.append(
            "B|"
            + pair[0]
            + "|"
            + pair[1]
            + "|"
            + f"{result.att_wave_1:.17g}"
            + "|"
            + f"{result.att_wave_2:.17g}"
            + "|"
            + f"{result.att_incremental:.17g}"
            + "|"
            + f"{result.se_cd:.17g}"
            + "|"
            + str(result.stars)
        )
        # Record the per-country incremental contrasts in sorted order.
        for country in sorted(str(code) for code in result.incremental_contrasts.index):
            # Append the country's contrast at full precision.
            parts.append(
                "D|"
                + pair[0]
                + "|"
                + pair[1]
                + "|"
                + country
                + "|"
                + f"{float(result.incremental_contrasts.loc[country]):.17g}"
            )
        # Bind the collection to the effective estimator configuration and
        # the complete fresh fitted-object lineage, not only headline cells.
        provenance = result.fit_provenance
        parts.append(
            "WINDOW|"
            + pair[0]
            + "|"
            + pair[1]
            + "|"
            + _content_digest(result.effective_windows)
        )
        parts.append(
            "LINEAGE|"
            + pair[0]
            + "|"
            + pair[1]
            + "|"
            + str(provenance.get("lineage_sha256", ""))
        )
        for name, digest in sorted(
            provenance.get("object_digests", {}).items()
        ):
            parts.append(
                f"OBJECT|{pair[0]}|{pair[1]}|{name}|{str(digest)}"
            )
        parts.append(
            "CHAIN|"
            + pair[0]
            + "|"
            + pair[1]
            + "|"
            + _content_digest(result.chain_validation)
        )
    # Digest the encoded payload.
    return hashlib.sha256("\n".join(parts).encode("utf-8")).hexdigest()


# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the baseline pipeline across the registered pairs.
def orchestrate_baseline_pipeline(
    study_config: Mapping[str, Any],
    data: Mapping[str, Any],
    pairs: Optional[Sequence[Tuple[str, str]]] = None,
    reporting_windows: Optional[Mapping[str, Tuple[int, int]]] = None,
    expected_incremental: Optional[Mapping[Tuple[str, str], Any]] = None,
    expected_stars: Optional[Mapping[Tuple[str, str], Any]] = None,
) -> BaselinePipelineResult:
    """
    Run the baseline pipeline across the registered pairs.

    Parameters
    ----------
    study_config : mapping
        The study configuration, which the orchestrator must not mutate.
    data : mapping
        The frozen inputs under the declared keys: the partition registry,
        the analysis panel and the lagged panel.
    pairs : sequence of tuple, optional
        The pairs to run. When absent every registered pair is run.
    reporting_windows : mapping of str to tuple of int, optional
        An explicit override of the reporting windows, which is applied but
        logged as a sweep variant.
    expected_incremental : mapping of tuple to object, optional
        The declared incremental publication cells per pair, optionally with
        displayed precision and numerical allowance metadata.
    expected_stars : mapping of tuple to object, optional
        The declared exact Student-t marker cells per pair.

    Returns
    -------
    BaselinePipelineResult
        The per-pair results, validation and freshness records, canonical
        effective windows, cell-level publication comparison ledger, digest,
        timestamp and notes.

    Raises
    ------
    KeyError
        If the data mapping lacks a declared key.
    IngestError
        If any gate, freshness or verification check fails, or if the
        configuration changed during the run.
    """
    # Refuse a data mapping lacking a declared key.
    for key in ("registry", "panel", "lagged"):
        # Test the key's presence.
        if key not in data:
            # The failure names the absent key.
            raise KeyError(f"The data mapping lacks the declared key {key!r}.")
    # Fingerprint the configuration before the run.
    config_before: str = freeze_config(study_config)
    # Select the registry from the frozen data.
    registry: PartitionRegistry = data["registry"]
    if not isinstance(registry, PartitionRegistry):
        raise IngestError("Baseline data registry has the wrong type.")
    mandatory_pairs = {
        (str(policy), str(outcome)) for policy, outcome in DECLARED_PAIRS
    }
    registry_pairs = {
        (str(policy), str(outcome)) for policy, outcome in registry.partitions
    }
    if registry_pairs != mandatory_pairs:
        missing = sorted(mandatory_pairs - registry_pairs)
        unexpected = sorted(registry_pairs - mandatory_pairs)
        raise IngestError(
            "Baseline registry must cover every mandatory partition exactly; "
            f"missing {missing}, unexpected {unexpected}."
        )
    # Compose the pair list, defaulting to every registered pair.
    selected: List[Tuple[str, str]] = (
        [(str(pair[0]), str(pair[1])) for pair in pairs]
        if pairs is not None
        else sorted(registry.partitions.keys())
    )
    # Refuse an empty selection.
    if not selected:
        # Without pairs the pipeline has nothing to run.
        raise IngestError("The baseline pipeline was given no pair to run.")
    if len(selected) != len(set(selected)):
        raise IngestError("Baseline pair selection contains duplicate identities.")
    if set(selected) != mandatory_pairs:
        missing = sorted(mandatory_pairs - set(selected))
        unexpected = sorted(set(selected) - mandatory_pairs)
        raise IngestError(
            "The baseline pipeline must execute every mandatory partition in "
            f"one frozen run; missing {missing}, unexpected {unexpected}."
        )
    # Resolve the reporting windows against the authoritative configuration.
    windows, mode = resolve_reporting_windows(study_config, reporting_windows, ("baseline", "baseline"))
    # Snapshot the shared read-only scientific inputs before any pair is run.
    panel_before = _content_digest(data["panel"])
    lagged_before = _content_digest(data["lagged"])
    registry_before = _partition_registry_structure_digest(registry)
    # Initialise the results collection.
    results: Dict[Tuple[str, str], BaselineResult] = {}
    # Accumulate the freshness rows.
    freshness_rows: List[Dict[str, Any]] = []
    # Accumulate the validation rows.
    validation_rows: List[Dict[str, Any]] = []
    # Run each selected pair through the composition.
    for pair in selected:
        # Run the pair's baseline.
        result: BaselineResult = run_baseline_partition(pair[0], pair[1], data, windows)
        # Record the result.
        results[pair] = result
        # Record the pair's object identities, which the freshness gate checks
        # across partitions, together with the content digests that verify
        # value-level independence (remediation B-18).
        provenance = result.fit_provenance
        freshness_row: Dict[str, Any] = {
            "policy": pair[0],
            "outcome": pair[1],
            "lineage_sha256": provenance["lineage_sha256"],
        }
        for name, obj in provenance["objects"].items():
            freshness_row[f"{name}_id"] = id(obj)
            freshness_row[f"{name}_digest"] = _content_digest(obj)
        freshness_rows.append(freshness_row)
        # Record one validation row per declared step of the pair's run.
        for step in DECLARED_STEPS:
            # Append the step's passing record.
            validation_rows.append(
                {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "step": step,
                    "gate_outcome": "passed",
                }
            )
        # Retain every wrapper-versus-independent-orchestrator comparison as
        # a labelled gate in the collection-level validation ledger.
        for _, chain_row in result.chain_validation.iterrows():
            validation_rows.append(
                {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "step": f"chain_compare:{chain_row['component']}",
                    "gate_outcome": "passed" if bool(chain_row["equal"]) else "failed",
                }
            )
    # Assemble the freshness frame keyed on the pair.
    freshness: pd.DataFrame = pd.DataFrame(freshness_rows).set_index(
        ["policy", "outcome"]
    )
    # Assert that no object or content was shared between partitions,
    # re-deriving every recorded digest from its stored object (B-18).
    assert_no_shared_state(
        freshness,
        results=results,
        data=data,
        effective_windows=windows,
    )
    # Assemble the validation frame keyed on the pair and the step.
    validation: pd.DataFrame = pd.DataFrame(validation_rows).set_index(
        ["policy", "outcome", "step"]
    )
    # Assert that the collection covers the requested pairs exactly.
    if set(results.keys()) != set(selected):
        # The failure names the divergence.
        missing = sorted(set(selected) - set(results.keys()))
        # Raise with the missing pairs attached.
        raise IngestError(f"Baseline collection omits the pair(s) {missing}.")
    # Verify the results against the declared values when they were supplied
    # and retain the traceable cell-level discrepancy data.
    declared_comparisons = verify_against_declared(
        results,
        expected_incremental=expected_incremental,
        expected_stars=expected_stars,
    )
    # Recompute all read-only input snapshots after the run; fitting is not
    # allowed to mutate the shared panel, lagged panel or registry identity.
    if panel_before != _content_digest(data["panel"]):
        raise IngestError("Baseline pipeline mutated the shared analysis panel.")
    if lagged_before != _content_digest(data["lagged"]):
        raise IngestError("Baseline pipeline mutated the shared lagged panel.")
    if registry_before != _partition_registry_structure_digest(registry):
        raise IngestError("Baseline pipeline mutated the registry structure.")
    # Fingerprint the configuration after the run and assert its immutability.
    config_after: str = freeze_config(study_config)
    # Assert that the configuration was not mutated by the run.
    assert_config_unchanged(config_before, config_after, ("baseline", "baseline"))
    # Compute the canonical digest of the collection.
    fingerprint: str = freeze_baseline_collection(results)
    # Accumulate the recorded notes.
    notes: List[str] = [
        f"reporting windows resolved in the {mode} mode; the configuration is "
        f"authoritative and an override is logged as a sweep variant.",
        "the minimal interface reports the incremental estimand's dispersion "
        "standard error and marker; the level effects are reported without "
        "inference at this level, and the full ledger remains available from "
        "the inference task.",
        "each partition instantiates its own factor, coefficient and loading "
        "objects; the freshness gate refuses a shared mutable identity while "
        "allowing independently equal numerical estimates and content hashes.",
        "each wrapper result is reconciled at every scientific intermediate "
        "against a fresh isolated run of the Task 10-through-14 orchestrators.",
    ]
    # Record the declared-verification mode.
    notes.append(
        "declared baseline values compared."
        if (expected_incremental or expected_stars)
        else "declared baseline values not supplied; comparison skipped."
    )
    # Assemble the immutable pipeline result.
    result_object: BaselinePipelineResult = BaselinePipelineResult(
        results=results,
        validation=validation,
        freshness=freshness,
        effective_windows=copy.deepcopy(dict(windows)),
        declared_comparisons=declared_comparisons,
        collection_fingerprint_sha256=fingerprint,
        completed_utc=current_utc_iso(),
        notes=tuple(notes),
    )
    # Log the completed pipeline with its digest prefix.
    logger.info(
        json.dumps(
            {
                "step": "baseline_pipeline_complete",
                "pairs": len(results),
                "mode": mode,
                "fingerprint": fingerprint[:16],
            },
            sort_keys=True,
        )
    )
    # Return the assembled result.
    return result_object


# Expose the declared interface of the baseline estimation.
def run_baseline(
    policy_name: str,
    outcome_name: str,
    study_config: Mapping[str, Any],
    data: Mapping[str, Any],
    reporting_windows: Optional[Mapping[str, Tuple[int, int]]] = None,
) -> BaselineResult:
    """
    Run the baseline estimation for one policy-outcome pair.

    Parameters
    ----------
    policy_name : str
        The policy instrument of the pair.
    outcome_name : str
        The canonical outcome variable of the pair.
    study_config : mapping
        The study configuration, which is authoritative for the windows and
        must not be mutated.
    data : mapping
        The frozen inputs under the declared keys.
    reporting_windows : mapping of str to tuple of int, optional
        An explicit override of the reporting windows, applied but logged as
        a sweep variant.

    Returns
    -------
    BaselineResult
        The frozen baseline result of the pair.

    Raises
    ------
    IngestError
        If any gate or configuration check fails.

    Notes
    -----
    The signature of this function is frozen. Any extension of the baseline
    interface requires a new orchestrator rather than a change here, so that
    results produced under this contract keep their meaning.
    """
    # Fingerprint the configuration before the run.
    config_before: str = freeze_config(study_config)
    # Resolve the reporting windows against the authoritative configuration.
    windows, mode = resolve_reporting_windows(
        study_config, reporting_windows, (str(policy_name), str(outcome_name))
    )
    # Run the pair's composition.
    result: BaselineResult = run_baseline_partition(
        str(policy_name), str(outcome_name), data, windows
    )
    # Fingerprint the configuration after the run and assert its immutability.
    config_after: str = freeze_config(study_config)
    # Assert that the configuration was not mutated by the run.
    assert_config_unchanged(
        config_before, config_after, (str(policy_name), str(outcome_name))
    )
    # Emit the completion record with the resolved mode.
    log_step(
        (str(policy_name), str(outcome_name)),
        "baseline_pair_complete",
        mode=mode,
        att_incremental=float(result.att_incremental),
        se_cd=float(result.se_cd),
        stars=str(result.stars),
    )
    # Return the frozen result.
    return result
````

````python
# Task 16: Orchestrator 2 - Bootstrap and Placebo Orchestrator

# ==============================================================================
# Task 16: Orchestrator 2 - Bootstrap and Placebo Orchestrator
# ==============================================================================

# Declare the branch label of the country-history bootstrap.
BRANCH_COUNTRY_HISTORY: Final[str] = "country_history"
# Declare the branch label of the region-block bootstrap.
BRANCH_REGION_BLOCK: Final[str] = "region_block"
# Declare the branch label of the size-matched placebo.
BRANCH_PLACEBO: Final[str] = "placebo"
# Compose the declared branch order.
BRANCH_ORDER: Final[Tuple[str, str, str]] = (
    BRANCH_COUNTRY_HISTORY,
    BRANCH_REGION_BLOCK,
    BRANCH_PLACEBO,
)
# Declare the default number of bootstrap replicates.
DEFAULT_BOOTSTRAP_REPLICATES: Final[int] = 500
# Declare the default number of placebo replicates.
DEFAULT_PLACEBO_REPLICATES: Final[int] = 500
# Declare the attempt cap as a multiple of the requested replicate count.
ATTEMPT_CAP_MULTIPLE: Final[int] = 5
# Declare the lower percentile of the bootstrap intervals.
PERCENTILE_LOWER: Final[float] = 2.5
# Declare the upper percentile of the bootstrap intervals.
PERCENTILE_UPPER: Final[float] = 97.5
# Declare the discard share above which a branch is flagged numerically
# fragile, which is one tenth of the default replicate count.
FRAGILE_DISCARD_SHARE: Final[float] = 0.10
# Declare the suffix separator used to keep duplicated draws distinct.
DRAW_SUFFIX_SEPARATOR: Final[str] = "#"
# Declare the minimum number of donors a resampled design must retain.
MINIMUM_DONORS: Final[int] = 2
# Declare the minimum target occurrences needed by Eq. (16)'s ddof-one
# country-dispersion standard error.
MINIMUM_TARGET_OCCURRENCES: Final[int] = 2
# Cap the number of discard reasons quoted inside any raised message.
MAX_REPORTED_REASONS: Final[int] = 12


# Declare the only exception class that the branch loop may treat as a
# legitimate stochastic draw rejection rather than a program failure.
class ResamplingDegeneracyError(IngestError):
    """A scientifically anticipated degenerate random draw."""

    # Store one registered category beside the human-readable explanation.
    def __init__(
        self,
        message: str,
        category: str = "design_degeneracy",
        occurrence_map: Optional[Sequence[Mapping[str, Any]]] = None,
        draw_metadata: Optional[Mapping[str, Any]] = None,
    ) -> None:
        super().__init__(message)
        self.category = str(category)
        self.occurrence_map = [dict(row) for row in (occurrence_map or ())]
        self.draw_metadata = dict(draw_metadata or {})


# Declare the capped-attempt failure with its complete audit attached.
class ResamplingAttemptCapError(IngestError):
    """Failure to attain the requested accepted draws within the cap."""

    # Preserve the complete ledger even though the branch has no result tuple.
    def __init__(self, message: str, attempt_audit: pd.DataFrame) -> None:
        super().__init__(message)
        self.attempt_audit = attempt_audit.copy(deep=True)


# Declare the frozen result of the resampling inference.
@dataclass(frozen=True)
class ResamplingResult:
    """
    Frozen result of the resampling inference of one pair.

    Attributes
    ----------
    policy_name : str
        The policy instrument identifying the run.
    outcome_name : str
        The canonical outcome variable identifying the run.
    ci_ch_lower : float
        The lower percentile of the country-history bootstrap draws.
    ci_ch_upper : float
        The upper percentile of the country-history bootstrap draws.
    ci_rb_lower : float
        The lower percentile of the region-block bootstrap draws.
    ci_rb_upper : float
        The upper percentile of the region-block bootstrap draws.
    p_placebo_hat : float
        The finite-draw-corrected placebo probability, which is a diagnostic
        probability and not a confidence level; no significance marker is
        derived from it.
    discarded_counts : dict of str to int
        The number of invalid attempts discarded in each branch.
    full_draw_arrays : dict of str to numpy.ndarray
        The valid draw statistics of each branch, persisted for audit.
    fragile_flags : dict of str to bool
        Whether each branch discarded more than the declared share of its
        attempts.
    seeds : dict of str to Any
        The seed material of the run, including each branch's spawn key.
    attempt_audits : dict of str to pandas.DataFrame
        Complete deterministic attempt ledgers, including every accepted and
        rejected draw's indices, sampled identities, seed-state digests,
        registry fingerprint, statistic and rejection category.
    fragility_audit : pandas.DataFrame
        One row per branch recording requested, accepted, rejected and
        attempted counts, rejection rate, threshold, categories and flag.
    """
    # The policy instrument identifying the run.
    policy_name: str
    # The canonical outcome variable identifying the run.
    outcome_name: str
    # The lower percentile of the country-history bootstrap.
    ci_ch_lower: float
    # The upper percentile of the country-history bootstrap.
    ci_ch_upper: float
    # The lower percentile of the region-block bootstrap.
    ci_rb_lower: float
    # The upper percentile of the region-block bootstrap.
    ci_rb_upper: float
    # The finite-draw-corrected placebo probability.
    p_placebo_hat: float
    # The discarded attempts of each branch.
    discarded_counts: Mapping[str, int]
    # The valid draw statistics of each branch.
    full_draw_arrays: Mapping[str, np.ndarray]
    # Whether each branch discarded more than the declared share.
    fragile_flags: Mapping[str, bool]
    # The seed material of the run.
    seeds: Mapping[str, Any]
    # Complete accepted/rejected attempt ledgers keyed on branch.
    attempt_audits: Mapping[str, pd.DataFrame] = field(default_factory=dict)
    # Branch-level fragility denominator and threshold audit.
    fragility_audit: pd.DataFrame = field(default_factory=pd.DataFrame)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 1: The interface and the explicit seed streams
# -------------------------------------------------------------------------------------------------------------------------------


# Spawn one independent generator per branch from the declared seeds.
def spawn_branch_generators(
    seeds: Mapping[str, Any],
    branches: Sequence[str] = BRANCH_ORDER,
) -> Dict[str, np.random.Generator]:
    """
    Spawn one independent generator per branch.

    Parameters
    ----------
    seeds : mapping
        The seed material, which must carry a master entropy under the key
        ``master``; no default is applied so that randomness is never hidden.
    branches : sequence of str, optional
        The branch labels, in the order the streams are spawned.

    Returns
    -------
    dict
        The generators keyed on the branch label.

    Raises
    ------
    IngestError
        If the seed mapping does not carry the master entropy, because the
        module refuses to invent entropy.
    """
    # Refuse a seed mapping without the declared master entropy.
    if "master" not in seeds:
        # The failure states the requirement.
        raise IngestError(
            "The seed mapping must carry the master entropy under the key "
            "'master'; the module never invents entropy."
        )
    # Compose the root seed sequence from the declared master entropy.
    root: np.random.SeedSequence = np.random.SeedSequence(int(seeds["master"]))
    # Spawn one child sequence per branch.
    children: List[np.random.SeedSequence] = root.spawn(len(branches))
    # Initialise the generator mapping.
    generators: Dict[str, np.random.Generator] = {}
    # Record the spawn keys for the audit log.
    spawn_keys: Dict[str, Any] = {}
    # Build each branch's generator from its own child sequence.
    for branch, child in zip(branches, children):
        # Construct the generator from the child sequence.
        generators[branch] = np.random.default_rng(child)
        # Record the child's spawn key so the stream's provenance is logged.
        spawn_keys[branch] = list(child.spawn_key)
    # Log the seed material and the spawned keys.
    logger.info(
        "resampling streams spawned: master=%s, spawn_keys=%s.",
        seeds["master"],
        spawn_keys,
    )
    # Return the generators.
    return generators


# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 2: The three resampling branches
# -------------------------------------------------------------------------------------------------------------------------------


# Build the resampled data mapping of one design.
def build_replicate_data(
    panel: pd.DataFrame, registry: PartitionRegistry
) -> Dict[str, Any]:
    """
    Build the resampled data mapping of one design.

    Parameters
    ----------
    panel : pandas.DataFrame
        The resampled analysis panel, whose country codes are distinct.
    registry : PartitionRegistry
        The registry describing the resampled design.

    Returns
    -------
    dict
        The data mapping under the declared keys, carrying the resampled
        panel, its lagged form and the resampled registry.

    Raises
    ------
    IngestError
        If the lag build fails on the resampled panel.
    """
    # Rebuild the lagged panel within each resampled country copy, which is
    # the correct semantics because every copy is its own path.
    lagged: pd.DataFrame = apply_control_lag(panel)
    # Recover the draw metadata retained by the one-pair replicate registry.
    if len(registry.registry) != 1:
        raise IngestError("A replicate registry must contain exactly one pair.")
    registry_row = registry.registry.iloc[0]
    required_metadata = {"occurrence_map_json", "draw_metadata_json"}
    if not required_metadata.issubset(registry.registry.columns):
        raise IngestError("Replicate registry omits occurrence/draw provenance.")
    try:
        occurrence_map = json.loads(str(registry_row["occurrence_map_json"]))
        draw_metadata = json.loads(str(registry_row["draw_metadata_json"]))
    except (TypeError, ValueError, json.JSONDecodeError) as exc:
        raise IngestError("Replicate registry provenance is malformed JSON.") from exc
    # Return the baseline inputs plus audit-only resampling metadata. The
    # baseline ignores the extra key, while the branch loop persists it.
    return {
        "registry": registry,
        "panel": panel,
        "lagged": lagged,
        "resample_metadata": {
            "occurrence_map": occurrence_map,
            "draw_metadata": draw_metadata,
            "registry_fingerprint": registry.fingerprint_sha256,
            "draw_content_sha256": str(
                registry_row.get("draw_content_sha256", "")
            ),
        },
    }


# Compose the resampled registry of one design.
def build_replicate_registry(
    policy_name: str,
    outcome_name: str,
    targets: Sequence[str],
    donors: Sequence[str],
    source: Partition,
    occurrence_map: Optional[Sequence[Mapping[str, Any]]] = None,
    draw_metadata: Optional[Mapping[str, Any]] = None,
) -> PartitionRegistry:
    """
    Compose the resampled registry of one design.

    Parameters
    ----------
    policy_name : str
        The policy instrument of the pair.
    outcome_name : str
        The canonical outcome variable of the pair.
    targets : sequence of str
        The resampled target codes.
    donors : sequence of str
        The resampled donor codes.
    source : Partition
        The original partition defining the pre-draw roles.
    occurrence_map : sequence of mapping, optional
        One row per occurrence carrying occurrence_id,
        original_country_code, original_role and replicate_role. When absent,
        a backward-compatible mapping is inferred from occurrence labels.
    draw_metadata : mapping, optional
        Deterministic branch, seed-state and draw identity metadata.

    Returns
    -------
    PartitionRegistry
        A registry carrying exactly the resampled pair.

    Raises
    ------
    IngestError
        If the occurrence identities, roles, source partition or metadata are
        malformed. Randomly insufficient role/diversity designs raise the
        narrower ``ResamplingDegeneracyError``.
    """
    # Bind the registry to the source partition's labelled pair.
    pair = (str(policy_name), str(outcome_name))
    if (str(source.policy), str(source.outcome)) != pair:
        raise IngestError(
            f"Replicate pair {pair} conflicts with source partition "
            f"{(source.policy, source.outcome)}."
        )
    target_codes = tuple(str(code) for code in targets)
    donor_codes = tuple(str(code) for code in donors)
    if any(not code.strip() for code in target_codes + donor_codes):
        raise IngestError("Replicate occurrence identifiers cannot be blank.")
    # Occurrence identifiers must be unique; repeated originals are retained
    # through different occurrence IDs, never duplicate keys.
    if len(target_codes) != len(set(target_codes)) or len(donor_codes) != len(set(donor_codes)):
        raise IngestError("Replicate occurrence identifiers are duplicated.")
    # Refuse a stochastic draw with no target or too few donor occurrences.
    if len(target_codes) < MINIMUM_TARGET_OCCURRENCES:
        raise ResamplingDegeneracyError(
            f"{policy_name}/{outcome_name}: resampled design retains "
            f"{len(target_codes)} target occurrence(s), below the Eq. (16) "
            f"minimum of {MINIMUM_TARGET_OCCURRENCES}.",
            category="insufficient_target_occurrences",
            occurrence_map=occurrence_map,
            draw_metadata=draw_metadata,
        )
    if len(donor_codes) < MINIMUM_DONORS:
        raise ResamplingDegeneracyError(
            f"{policy_name}/{outcome_name}: resampled design retains "
            f"{len(donor_codes)} donor occurrence(s), below the declared "
            f"minimum of {MINIMUM_DONORS}.",
            category="insufficient_donor_occurrences",
            occurrence_map=occurrence_map,
            draw_metadata=draw_metadata,
        )
    # Refuse a design whose sets overlap.
    if set(target_codes) & set(donor_codes):
        # The failure states the design precondition.
        raise IngestError(
            f"{policy_name}/{outcome_name}: resampled target and donor sets "
            f"overlap."
        )
    # Infer a canonical mapping only for backward-compatible direct calls.
    source_targets = set(str(code) for code in source.targets)
    source_donors = set(str(code) for code in source.donors)
    if occurrence_map is None:
        inferred: List[Dict[str, Any]] = []
        for role, codes in (("target", target_codes), ("donor", donor_codes)):
            for occurrence_id in codes:
                original = occurrence_id.split(DRAW_SUFFIX_SEPARATOR, 1)[0]
                original_role = (
                    "target" if original in source_targets
                    else "donor" if original in source_donors
                    else "unregistered"
                )
                inferred.append(
                    {
                        "occurrence_id": occurrence_id,
                        "original_country_code": original,
                        "original_role": original_role,
                        "replicate_role": role,
                    }
                )
        occurrence_records = inferred
    else:
        occurrence_records = [dict(record) for record in occurrence_map]
    required_fields = {
        "occurrence_id", "original_country_code", "original_role", "replicate_role"
    }
    for record in occurrence_records:
        if not required_fields.issubset(record):
            raise IngestError("Replicate occurrence mapping is incomplete.")
    mapped_ids = tuple(str(record["occurrence_id"]) for record in occurrence_records)
    if len(mapped_ids) != len(set(mapped_ids)) or set(mapped_ids) != set(target_codes) | set(donor_codes):
        raise IngestError("Occurrence mapping and replicate rosters diverge.")
    for record in occurrence_records:
        occurrence_id = str(record["occurrence_id"])
        original = str(record["original_country_code"])
        original_role = str(record["original_role"])
        replicate_role = str(record["replicate_role"])
        if not original.strip():
            raise IngestError(f"{occurrence_id}: original identity is blank.")
        expected_role = "target" if occurrence_id in set(target_codes) else "donor"
        if replicate_role != expected_role:
            raise IngestError(f"{occurrence_id}: occurrence role conflicts with roster.")
        if original_role not in {"target", "donor"}:
            raise IngestError(f"{occurrence_id}: original role is unregistered.")
        if original_role == "target" and original not in source_targets:
            raise IngestError(f"{occurrence_id}: original target identity is invalid.")
        if original_role == "donor" and original not in source_donors:
            raise IngestError(f"{occurrence_id}: original donor identity is invalid.")
    # Check original donor diversity separately from donor occurrence count.
    original_donors = {
        str(record["original_country_code"])
        for record in occurrence_records
        if str(record["replicate_role"]) == "donor"
    }
    if len(original_donors) < MINIMUM_DONORS:
        raise ResamplingDegeneracyError(
            f"{policy_name}/{outcome_name}: {len(donor_codes)} donor "
            f"occurrence(s) represent only {len(original_donors)} distinct "
            f"original donor(s), below {MINIMUM_DONORS}.",
            category="insufficient_original_donor_diversity",
            occurrence_map=occurrence_records,
            draw_metadata=draw_metadata,
        )
    # Canonicalize and validate deterministic draw metadata.
    metadata = dict(draw_metadata or {})
    metadata.setdefault("sampling_scheme", "unspecified_direct_registry_call")
    metadata.setdefault("seed_state_before_sha256", "not_supplied")
    metadata.setdefault("seed_state_after_sha256", "not_supplied")
    occurrence_records = sorted(
        occurrence_records, key=lambda record: str(record["occurrence_id"])
    )
    occurrence_json = json.dumps(occurrence_records, sort_keys=True, default=str)
    metadata["occurrence_map_sha256"] = hashlib.sha256(
        occurrence_json.encode("utf-8")
    ).hexdigest()
    metadata["draw_identity_sha256"] = hashlib.sha256(
        json.dumps(
            {
                "pair": pair,
                "occurrences": occurrence_records,
                "metadata": metadata,
            },
            sort_keys=True,
            default=str,
        ).encode("utf-8")
    ).hexdigest()
    metadata_json = json.dumps(metadata, sort_keys=True, default=str)
    # Recompute exclusions from the actual draw: source-pool originals absent
    # from every occurrence are explicit draw exclusions.
    sampled_originals = {
        str(record["original_country_code"]) for record in occurrence_records
    }
    exclusion_rows = [
        {
            "policy": pair[0],
            "outcome": pair[1],
            "country_code": original,
            "reason": "not_sampled_in_replicate",
            "detail": (
                "source role target" if original in source_targets
                else "source role donor"
            ),
        }
        for original in sorted((source_targets | source_donors) - sampled_originals)
    ]
    draw_exclusions = pd.DataFrame(
        exclusion_rows,
        columns=["policy", "outcome", "country_code", "reason", "detail"],
    )
    exclusion_columns = ["policy", "outcome", "country_code", "reason", "detail"]
    if source.exclusions.empty and len(source.exclusions.columns) == 0:
        source_exclusions = pd.DataFrame(columns=exclusion_columns)
    elif not set(exclusion_columns).issubset(source.exclusions.columns):
        raise IngestError("Source partition exclusion audit schema is incomplete.")
    else:
        source_exclusions = source.exclusions.loc[:, exclusion_columns].copy(deep=True)
    if len(source_exclusions):
        source_exclusions["policy"] = pair[0]
        source_exclusions["outcome"] = pair[1]
        source_exclusions["detail"] = source_exclusions["detail"].map(
            lambda value: f"source_policy_exclusion: {value}"
        )
    # Preserve substantive policy/data exclusions and append the newly
    # recomputed non-sampling exclusions under a distinct reason.
    exclusions = pd.concat(
        [source_exclusions, draw_exclusions], ignore_index=True
    ).drop_duplicates(
        subset=["policy", "outcome", "country_code", "reason", "detail"],
        keep="first",
    )
    # Compose the resampled partition.
    partition: Partition = Partition(
        policy=pair[0],
        outcome=pair[1],
        targets=tuple(sorted(target_codes)),
        donors=tuple(sorted(donor_codes)),
        exclusions=exclusions.copy(deep=True),
    )
    # Compose the registry frame of the single pair.
    registry_frame: pd.DataFrame = pd.DataFrame(
        [
            {
                "policy": pair[0],
                "outcome": pair[1],
                "targets": len(partition.targets),
                "donors": len(partition.donors),
                "target_codes": json.dumps(list(partition.targets)),
                "donor_codes": json.dumps(list(partition.donors)),
                "original_target_count": int(
                    len({str(record["original_country_code"]) for record in occurrence_records if str(record["replicate_role"]) == "target"})
                ),
                "original_donor_count": int(len(original_donors)),
                "occurrence_map_json": occurrence_json,
                "draw_metadata_json": metadata_json,
            }
        ]
    ).set_index(["policy", "outcome"])
    # Fingerprint the actual rosters, mapping, exclusions and draw metadata
    # separately from the PartitionRegistry roster fingerprint contract.
    fingerprint_payload = {
        "registry": registry_frame.reset_index().to_dict(orient="records"),
        "exclusions": exclusions.to_dict(orient="records"),
    }
    draw_content_fingerprint = hashlib.sha256(
        json.dumps(fingerprint_payload, sort_keys=True, default=str).encode("utf-8")
    ).hexdigest()
    registry_frame["draw_content_sha256"] = draw_content_fingerprint
    # Preserve the established registry contract used by Task 15 while the
    # richer draw digest remains explicit alongside it.
    fingerprint = freeze_partition_registry(registry_frame)
    # Assemble the resampled registry carrying the single pair.
    return PartitionRegistry(
        partitions={pair: partition},
        registry=registry_frame,
        counts=registry_frame[
            ["targets", "donors", "original_target_count", "original_donor_count"]
        ].copy(deep=True),
        exclusions=exclusions.copy(deep=True),
        fingerprint_sha256=fingerprint,
        built_utc=current_utc_iso(),
        notes=(
            "resampled design with explicit original-to-occurrence mapping",
            f"draw_identity_sha256={metadata['draw_identity_sha256']}",
            f"occurrence_map_sha256={metadata['occurrence_map_sha256']}",
            f"draw_content_sha256={draw_content_fingerprint}",
        ),
    )


# Digest a generator state without consuming random numbers.
def _generator_state_digest(generator: np.random.Generator) -> str:
    """Return a canonical SHA-256 digest of a NumPy generator state."""
    if not isinstance(generator, np.random.Generator):
        raise IngestError("Resampling requires a numpy.random.Generator.")
    payload = json.dumps(
        generator.bit_generator.state, sort_keys=True, default=str
    ).encode("utf-8")
    return hashlib.sha256(payload).hexdigest()


# Serialize a generator state losslessly for the per-attempt audit.
def _generator_state_json(generator: np.random.Generator) -> str:
    """Return the complete canonical JSON state without consuming entropy."""
    if not isinstance(generator, np.random.Generator):
        raise IngestError("Resampling requires a numpy.random.Generator.")
    return json.dumps(generator.bit_generator.state, sort_keys=True, default=str)


# Copy one original's complete labelled history to one occurrence identity.
def _copy_occurrence_history(
    panel: pd.DataFrame,
    original_country_code: str,
    occurrence_id: str,
    replicate_role: str,
) -> pd.DataFrame:
    """Return an unchanged complete history under a unique occurrence key."""
    if not {"country_code", "year"}.issubset(panel.columns):
        raise IngestError("Resampling panel lacks country_code/year keys.")
    original = str(original_country_code)
    occurrence = str(occurrence_id)
    if not original or not occurrence or replicate_role not in {"target", "donor"}:
        raise IngestError("Occurrence history identity or role is malformed.")
    subset = panel.loc[panel["country_code"].astype(str) == original].copy(deep=True)
    if subset.empty:
        raise IngestError(f"Original country {original!r} has no panel history.")
    validated = _validated_att_years(subset, f"resampling history {original}")
    if bool(validated["_att_year"].duplicated().any()):
        raise IngestError(f"Original country {original!r} has duplicate years.")
    required_years = set(range(FACTOR_WINDOW[0], FACTOR_WINDOW[1] + 1))
    observed_years = set(validated["_att_year"].astype(int))
    missing_required = sorted(required_years - observed_years)
    if missing_required:
        raise IngestError(
            f"Original country {original!r} history omits required model "
            f"year(s) {missing_required}."
        )
    subset = validated.sort_values("_att_year", kind="stable").drop(
        columns=["_att_year"]
    )
    # Keep both identities explicitly while country_code becomes the unique
    # occurrence key consumed by the estimator.
    subset["original_country_code"] = original
    subset["occurrence_id"] = occurrence
    subset["replicate_role"] = replicate_role
    subset["country_code"] = occurrence
    return subset.reset_index(drop=True)


# Resample whole country histories with replacement.
def resample_country_history(
    policy_name: str,
    outcome_name: str,
    partition: Partition,
    panel: pd.DataFrame,
    generator: np.random.Generator,
) -> Dict[str, Any]:
    """
    Resample whole country histories with replacement.

    Parameters
    ----------
    policy_name : str
        The policy instrument of the pair.
    outcome_name : str
        The canonical outcome variable of the pair.
    partition : Partition
        The original partition. Sampling is stratified by original policy
        role: target histories are drawn from targets and donor histories
        from donors, each with replacement and at its original role count.
    panel : pandas.DataFrame
        The frozen analysis panel.
    generator : numpy.random.Generator
        The branch's generator.

    Returns
    -------
    dict
        The resampled data mapping, whose panel carries each drawn history
        under a distinct suffixed country code so that a repeated draw
        doubles its weight rather than collapsing.

    Raises
    ------
    IngestError
        If a source history is incomplete/malformed. A random draw with too
        little distinct original-donor diversity raises
        ``ResamplingDegeneracyError`` for the branch loop to record.
    """
    target_pool = tuple(sorted(str(code) for code in partition.targets))
    donor_pool = tuple(sorted(str(code) for code in partition.donors))
    if not target_pool or len(donor_pool) < MINIMUM_DONORS:
        raise IngestError(
            f"{policy_name}/{outcome_name}: source roles cannot support a "
            "stratified country-history resample."
        )
    state_before = _generator_state_digest(generator)
    # Draw each original role separately so policy roles and occurrence
    # counts are preserved while repeated originals retain multiplicity.
    drawn_targets = generator.choice(
        np.array(target_pool, dtype=object), size=len(target_pool), replace=True
    )
    drawn_donors = generator.choice(
        np.array(donor_pool, dtype=object), size=len(donor_pool), replace=True
    )
    state_after = _generator_state_digest(generator)
    # Accumulate the resampled panel blocks.
    blocks: List[pd.DataFrame] = []
    # Accumulate the resampled target codes.
    targets: List[str] = []
    # Accumulate the resampled donor codes.
    donors: List[str] = []
    occurrence_records: List[Dict[str, Any]] = []
    # Build one complete history block per role-stratified occurrence.
    for role, originals in (
        ("target", drawn_targets.tolist()), ("donor", drawn_donors.tolist())
    ):
        for index, code in enumerate(originals):
            occurrence_id = (
                f"{code}{DRAW_SUFFIX_SEPARATOR}{role}_{index:04d}"
            )
            blocks.append(
                _copy_occurrence_history(panel, str(code), occurrence_id, role)
            )
            (targets if role == "target" else donors).append(occurrence_id)
            occurrence_records.append(
                {
                    "occurrence_id": occurrence_id,
                    "original_country_code": str(code),
                    "original_role": role,
                    "replicate_role": role,
                    "draw_position": int(index),
                }
            )
    # Stack the drawn blocks into the resampled panel.
    resampled_panel: pd.DataFrame = pd.concat(blocks, axis=0, ignore_index=True)
    # Compose the resampled registry.
    registry: PartitionRegistry = build_replicate_registry(
        policy_name,
        outcome_name,
        targets,
        donors,
        partition,
        occurrence_map=occurrence_records,
        draw_metadata={
            "branch": BRANCH_COUNTRY_HISTORY,
            "sampling_scheme": "stratified_by_original_policy_role",
            "seed_state_before_sha256": state_before,
            "seed_state_after_sha256": state_after,
            "drawn_target_originals": [str(value) for value in drawn_targets],
            "drawn_donor_originals": [str(value) for value in drawn_donors],
        },
    )
    # Return the resampled data mapping.
    return build_replicate_data(resampled_panel, registry)


# Resample whole regions with replacement.
def resample_region_blocks(
    policy_name: str,
    outcome_name: str,
    partition: Partition,
    panel: pd.DataFrame,
    region_map: Mapping[str, str],
    generator: np.random.Generator,
) -> Dict[str, Any]:
    """
    Resample whole regions with replacement.

    Parameters
    ----------
    policy_name : str
        The policy instrument of the pair.
    outcome_name : str
        The canonical outcome variable of the pair.
    partition : Partition
        The original partition, whose union of targets and donors defines the
        mapped countries.
    panel : pandas.DataFrame
        The frozen analysis panel.
    region_map : mapping of str to str
        The region of each country code.
    generator : numpy.random.Generator
        The branch's generator.

    Returns
    -------
    dict
        The resampled data mapping, whose panel stacks the drawn regions'
        countries under distinct suffixed codes.

    Raises
    ------
    IngestError
        If a country of the partition is absent from the region map, or if
        the design is invalid.
    """
    # Compose the declared pool of countries.
    pool: List[str] = sorted(
        set(str(code) for code in partition.targets)
        | set(str(code) for code in partition.donors)
    )
    # Refuse a partition country absent from the region map.
    missing: List[str] = [code for code in pool if code not in region_map]
    # Raise when a mapping is incomplete.
    if missing:
        # The failure names the unmapped countries.
        raise IngestError(
            f"{policy_name}/{outcome_name}: region map omits the countr(ies) "
            f"{missing}."
        )
    blank_regions = [
        code
        for code in pool
        if region_map[code] is None or not str(region_map[code]).strip()
    ]
    if blank_regions:
        raise IngestError(
            f"{policy_name}/{outcome_name}: region map carries blank labels "
            f"for {blank_regions}."
        )
    # Compose the regions present in the pool.
    regions: List[str] = sorted({str(region_map[code]) for code in pool})
    state_before = _generator_state_digest(generator)
    # Draw one block occurrence per observed region, with replacement.
    drawn_regions: np.ndarray = generator.choice(
        np.array(regions, dtype=object), size=len(regions), replace=True
    )
    state_after = _generator_state_digest(generator)
    # Compose the membership sets used to classify each drawn country.
    target_set = set(str(code) for code in partition.targets)
    # Accumulate the resampled panel blocks.
    blocks: List[pd.DataFrame] = []
    # Accumulate the resampled target codes.
    targets: List[str] = []
    # Accumulate the resampled donor codes.
    donors: List[str] = []
    # Retain one original-to-occurrence record per copied country history.
    occurrence_records: List[Dict[str, Any]] = []
    # Build one block per drawn region copy.
    for index, region in enumerate(drawn_regions.tolist()):
        # Select the region's countries within the pool.
        members: List[str] = [
            code for code in pool if str(region_map[code]) == str(region)
        ]
        # Iterate over the region's members.
        for member in members:
            # Compose the copy's distinct country code.
            copy_code: str = (
                f"{member}{DRAW_SUFFIX_SEPARATOR}region_{index:04d}"
            )
            replicate_role = "target" if str(member) in target_set else "donor"
            blocks.append(
                _copy_occurrence_history(
                    panel, str(member), copy_code, replicate_role
                )
            )
            # Classify the copy as a target or a donor of the design.
            if str(member) in target_set:
                # Record the copy as a resampled target.
                targets.append(copy_code)
            else:
                # Record the copy as a resampled donor.
                donors.append(copy_code)
            occurrence_records.append(
                {
                    "occurrence_id": copy_code,
                    "original_country_code": str(member),
                    "original_role": replicate_role,
                    "replicate_role": replicate_role,
                    "region": str(region),
                    "block_position": int(index),
                }
            )
    # Refuse a design whose stack is empty, which is an invalid resample.
    if not blocks:
        # The failure states the condition.
        raise IngestError(
            f"{policy_name}/{outcome_name}: region resample produced an empty "
            f"stack."
        )
    # Stack the drawn blocks into the resampled panel.
    resampled_panel: pd.DataFrame = pd.concat(blocks, axis=0, ignore_index=True)
    # Compose the resampled registry.
    registry: PartitionRegistry = build_replicate_registry(
        policy_name,
        outcome_name,
        targets,
        donors,
        partition,
        occurrence_map=occurrence_records,
        draw_metadata={
            "branch": BRANCH_REGION_BLOCK,
            "sampling_scheme": "complete_region_blocks_with_replacement",
            "seed_state_before_sha256": state_before,
            "seed_state_after_sha256": state_after,
            "drawn_regions": [str(value) for value in drawn_regions],
            "region_pool": list(regions),
        },
    )
    # Return the resampled data mapping.
    return build_replicate_data(resampled_panel, registry)


# Draw a size-matched false coalition from the never-treated pool.
def resample_placebo_coalition(
    policy_name: str,
    outcome_name: str,
    partition: Partition,
    panel: pd.DataFrame,
    generator: np.random.Generator,
) -> Dict[str, Any]:
    """
    Draw a size-matched false coalition from the never-treated pool.

    Parameters
    ----------
    policy_name : str
        The policy instrument of the pair.
    outcome_name : str
        The canonical outcome variable of the pair.
    partition : Partition
        The original partition, whose donor pool is the declared draw pool
        and whose target count fixes the false coalition's size.
    panel : pandas.DataFrame
        The frozen analysis panel, which is used unchanged.
    generator : numpy.random.Generator
        The branch's generator.

    Returns
    -------
    dict
        The data mapping whose registry describes the false coalition and the
        complementary pseudo-donor pool.

    Raises
    ------
    IngestError
        If the donor pool cannot supply the false coalition, if the draw
        reproduces the observed target set, or if the complementary design is
        invalid.
    """
    # Compose the donor pool, which is the declared draw pool of this branch.
    donor_pool: List[str] = sorted(str(code) for code in partition.donors)
    # Compose the observed target set.
    observed_targets = set(str(code) for code in partition.targets)
    # Refuse a donor pool too small to supply the false coalition.
    if len(donor_pool) < len(observed_targets) + MINIMUM_DONORS:
        # The failure states the arithmetic requirement.
        raise IngestError(
            f"{policy_name}/{outcome_name}: the donor pool of {len(donor_pool)} "
            f"cannot supply a false coalition of {len(observed_targets)} and the "
            f"declared minimum of {MINIMUM_DONORS} complementary donors."
        )
    state_before = _generator_state_digest(generator)
    # Draw the false coalition without replacement from the eligible donor pool.
    drawn: np.ndarray = generator.choice(
        np.array(donor_pool, dtype=object),
        size=len(observed_targets),
        replace=False,
    )
    state_after = _generator_state_digest(generator)
    # Compose the false coalition as a set.
    false_targets = set(str(code) for code in drawn.tolist())
    # Refuse a draw that reproduces the observed coalition, which would turn
    # the placebo into a duplicate of the observed run.
    if false_targets == observed_targets:
        # The failure states the condition.
        raise IngestError(
            f"{policy_name}/{outcome_name}: the placebo draw reproduces the "
            f"observed coalition and is rejected."
        )
    # Compose the complementary pseudo-donor pool.
    pseudo_donors: List[str] = [
        code for code in donor_pool if code not in false_targets
    ]
    # Give every placebo unit an explicit occurrence identity and retain its
    # original donor identity and reassigned replicate role in the panel.
    false_order = tuple(sorted(false_targets))
    donor_order = tuple(sorted(pseudo_donors))
    blocks: List[pd.DataFrame] = []
    occurrence_records: List[Dict[str, Any]] = []
    target_occurrences: List[str] = []
    donor_occurrences: List[str] = []
    for role, originals in (("target", false_order), ("donor", donor_order)):
        for index, original in enumerate(originals):
            occurrence_id = (
                f"{original}{DRAW_SUFFIX_SEPARATOR}placebo_{role}_{index:04d}"
            )
            blocks.append(
                _copy_occurrence_history(panel, original, occurrence_id, role)
            )
            (target_occurrences if role == "target" else donor_occurrences).append(
                occurrence_id
            )
            occurrence_records.append(
                {
                    "occurrence_id": occurrence_id,
                    "original_country_code": original,
                    "original_role": "donor",
                    "replicate_role": role,
                    "draw_position": int(index),
                }
            )
    placebo_panel = pd.concat(blocks, ignore_index=True)
    # Compose the resampled registry describing the false design.
    registry: PartitionRegistry = build_replicate_registry(
        policy_name,
        outcome_name,
        target_occurrences,
        donor_occurrences,
        partition,
        occurrence_map=occurrence_records,
        draw_metadata={
            "branch": BRANCH_PLACEBO,
            "sampling_scheme": "size_matched_without_replacement_from_eligible_donors",
            "seed_state_before_sha256": state_before,
            "seed_state_after_sha256": state_after,
            "false_target_originals": list(false_order),
            "pseudo_donor_originals": list(donor_order),
            "eligible_reassignment_universe": list(donor_pool),
        },
    )
    # Return the data mapping over complete histories of the eligible
    # reassignment universe only.
    return build_replicate_data(placebo_panel, registry)


# Run one replicate of a branch through the baseline composition.
def run_replicate(
    policy_name: str,
    outcome_name: str,
    study_config: Mapping[str, Any],
    replicate_data: Mapping[str, Any],
) -> float:
    """
    Run one replicate of a branch through the baseline composition.

    Parameters
    ----------
    policy_name : str
        The policy instrument of the pair.
    outcome_name : str
        The canonical outcome variable of the pair.
    study_config : mapping
        The baseline configuration, which is passed through unchanged.
    replicate_data : mapping
        The resampled data mapping of the replicate.

    Returns
    -------
    float
        The replicate's incremental effect, which is the branch's stored
        draw statistic.

    Raises
    ------
    IngestError
        Propagated from the baseline composition. The branch loop does not
        convert this generic failure into a discarded draw; only the dedicated
        pre-refit ``ResamplingDegeneracyError`` is an admissible rejection.
    """
    # Run the baseline composition on the resampled design.
    result: BaselineResult = run_baseline(
        policy_name, outcome_name, study_config, replicate_data
    )
    # Return the replicate's incremental effect.
    return float(result.att_incremental)


# Draw the valid replicates of one branch under the invalid-resample policy.
def draw_branch_replicates(
    policy_name: str,
    outcome_name: str,
    study_config: Mapping[str, Any],
    branch: str,
    resampler: Any,
    requested: int,
    generator: np.random.Generator,
    attempt_cap_multiple: int = ATTEMPT_CAP_MULTIPLE,
) -> Tuple[np.ndarray, int, pd.DataFrame]:
    """
    Draw the valid replicates of one branch.

    Parameters
    ----------
    policy_name : str
        The policy instrument of the pair.
    outcome_name : str
        The canonical outcome variable of the pair.
    study_config : mapping
        The baseline configuration.
    branch : str
        The branch label, used in diagnostics.
    resampler : Any
        The branch's resampler, which returns a resampled data mapping.
    requested : int
        The number of valid replicates requested.
    generator : numpy.random.Generator
        The branch's generator.
    attempt_cap_multiple : int, optional
        The attempt cap as a multiple of the requested count.

    Returns
    -------
    tuple
        The valid draw array, the discarded count and the complete attempt
        ledger. The 3-tuple arity is preserved, but the third component is a
        structured audit rather than a capped list of strings.

    Raises
    ------
    IngestError
        If arguments or returned provenance are malformed, or if a baseline
        refit raises a schema/programming failure. A capped run raises
        ``ResamplingAttemptCapError`` carrying its complete attempt ledger.
    """
    # Validate integral count declarations without accepting Booleans or
    # silently truncating fractional values.
    if (
        isinstance(requested, (bool, np.bool_))
        or not isinstance(requested, (int, np.integer))
        or int(requested) <= 0
    ):
        raise IngestError(f"{branch}: replicate count must be positive.")
    if (
        isinstance(attempt_cap_multiple, (bool, np.bool_))
        or not isinstance(attempt_cap_multiple, (int, np.integer))
        or int(attempt_cap_multiple) <= 0
    ):
        raise IngestError(f"{branch}: attempt-cap multiple must be positive.")
    if branch not in BRANCH_ORDER:
        raise IngestError(f"Undeclared resampling branch {branch!r}.")
    if not callable(resampler) or not isinstance(generator, np.random.Generator):
        raise IngestError(f"{branch}: resampler or generator is malformed.")
    # Compute the attempt cap.
    attempt_cap: int = int(attempt_cap_multiple) * int(requested)
    # Accumulate the valid draws.
    draws: List[float] = []
    # Accumulate one complete row for every attempted draw.
    audit_rows: List[Dict[str, Any]] = []
    # Attempt draws until the accepted target or cap is reached.
    while len(draws) < int(requested) and len(audit_rows) < attempt_cap:
        attempt_index = len(audit_rows)
        state_before_json = _generator_state_json(generator)
        state_before_sha256 = hashlib.sha256(
            state_before_json.encode("utf-8")
        ).hexdigest()
        # Only the dedicated degeneracy exception is an admissible rejection.
        try:
            replicate_data = resampler()
        except ResamplingDegeneracyError as error:
            state_after_json = _generator_state_json(generator)
            audit_rows.append(
                {
                    "branch": branch,
                    "attempt_index": int(attempt_index),
                    "status": "rejected",
                    "accepted_index": pd.NA,
                    "reason_category": str(error.category),
                    "reason_detail": str(error),
                    "refit_completed": False,
                    "statistic": np.nan,
                    "seed_state_before_json": state_before_json,
                    "seed_state_before_sha256": state_before_sha256,
                    "seed_state_after_json": state_after_json,
                    "seed_state_after_sha256": hashlib.sha256(
                        state_after_json.encode("utf-8")
                    ).hexdigest(),
                    "sampled_occurrence_ids_json": json.dumps(
                        [row.get("occurrence_id") for row in error.occurrence_map]
                    ),
                    "sampled_originals_json": json.dumps(
                        [row.get("original_country_code") for row in error.occurrence_map]
                    ),
                    "occurrence_map_sha256": hashlib.sha256(
                        json.dumps(
                            error.occurrence_map, sort_keys=True, default=str
                        ).encode("utf-8")
                    ).hexdigest(),
                    "draw_identity_sha256": str(
                        error.draw_metadata.get("draw_identity_sha256", "")
                    ),
                    "registry_fingerprint": "",
                    "draw_content_sha256": "",
                    "draw_metadata_json": json.dumps(
                        error.draw_metadata, sort_keys=True, default=str
                    ),
                }
            )
            continue
        # Require every sampled design to expose complete identity and seed
        # provenance before any expensive baseline refit is attempted.
        metadata = replicate_data.get("resample_metadata")
        if not isinstance(metadata, Mapping):
            raise IngestError(f"{branch}: resampler omitted draw provenance.")
        occurrence_map = metadata.get("occurrence_map")
        draw_metadata = metadata.get("draw_metadata")
        if not isinstance(occurrence_map, list) or not isinstance(draw_metadata, Mapping):
            raise IngestError(f"{branch}: resampler provenance is malformed.")
        registry = replicate_data.get("registry")
        if not isinstance(registry, PartitionRegistry):
            raise IngestError(f"{branch}: resampler omitted its registry.")
        if registry.fingerprint_sha256 != freeze_partition_registry(registry.registry):
            raise IngestError(f"{branch}: replicate registry fingerprint is stale.")
        state_after_json = _generator_state_json(generator)
        state_after_sha256 = hashlib.sha256(
            state_after_json.encode("utf-8")
        ).hexdigest()
        if str(draw_metadata.get("seed_state_before_sha256", "")) != state_before_sha256:
            raise IngestError(f"{branch}: sampler seed-state provenance is stale.")
        if str(draw_metadata.get("seed_state_after_sha256", "")) != state_after_sha256:
            raise IngestError(f"{branch}: sampler post-state provenance is stale.")
        # Run a complete fresh baseline fit. Generic IngestError is
        # intentionally not caught: schema/programming failures must surface.
        value = run_replicate(
            policy_name, outcome_name, study_config, replicate_data
        )
        # A non-finite returned statistic is the one post-refit numerical
        # degeneracy that can be recorded as a rejected attempt.
        if not np.isfinite(value):
            audit_rows.append(
                {
                    "branch": branch,
                    "attempt_index": int(attempt_index),
                    "status": "rejected",
                    "accepted_index": pd.NA,
                    "reason_category": "nonfinite_statistic",
                    "reason_detail": "full baseline refit returned non-finite statistic",
                    "refit_completed": True,
                    "statistic": float(value),
                    "seed_state_before_json": state_before_json,
                    "seed_state_before_sha256": state_before_sha256,
                    "seed_state_after_json": state_after_json,
                    "seed_state_after_sha256": state_after_sha256,
                    "sampled_occurrence_ids_json": json.dumps(
                        [row["occurrence_id"] for row in occurrence_map]
                    ),
                    "sampled_originals_json": json.dumps(
                        [row["original_country_code"] for row in occurrence_map]
                    ),
                    "occurrence_map_sha256": str(
                        draw_metadata.get("occurrence_map_sha256", "")
                    ),
                    "draw_identity_sha256": str(
                        draw_metadata.get("draw_identity_sha256", "")
                    ),
                    "registry_fingerprint": registry.fingerprint_sha256,
                    "draw_content_sha256": str(
                        metadata.get("draw_content_sha256", "")
                    ),
                    "draw_metadata_json": json.dumps(
                        dict(draw_metadata), sort_keys=True, default=str
                    ),
                }
            )
            continue
        accepted_index = len(draws)
        draws.append(float(value))
        audit_rows.append(
            {
                "branch": branch,
                "attempt_index": int(attempt_index),
                "status": "accepted",
                "accepted_index": int(accepted_index),
                "reason_category": "",
                "reason_detail": "",
                "refit_completed": True,
                "statistic": float(value),
                "seed_state_before_json": state_before_json,
                "seed_state_before_sha256": state_before_sha256,
                "seed_state_after_json": state_after_json,
                "seed_state_after_sha256": state_after_sha256,
                "sampled_occurrence_ids_json": json.dumps(
                    [row["occurrence_id"] for row in occurrence_map]
                ),
                "sampled_originals_json": json.dumps(
                    [row["original_country_code"] for row in occurrence_map]
                ),
                "occurrence_map_sha256": str(
                    draw_metadata.get("occurrence_map_sha256", "")
                ),
                "draw_identity_sha256": str(
                    draw_metadata.get("draw_identity_sha256", "")
                ),
                "registry_fingerprint": registry.fingerprint_sha256,
                "draw_content_sha256": str(
                    metadata.get("draw_content_sha256", "")
                ),
                "draw_metadata_json": json.dumps(
                    dict(draw_metadata), sort_keys=True, default=str
                ),
            }
        )
    attempt_audit = pd.DataFrame(audit_rows)
    discarded = int((attempt_audit["status"] == "rejected").sum())
    # Assert exact, gap-free attempt and accepted indices.
    if list(attempt_audit["attempt_index"]) != list(range(len(attempt_audit))):
        raise IngestError(f"{branch}: attempt indices are not deterministic.")
    accepted_rows = attempt_audit[attempt_audit["status"] == "accepted"]
    if list(accepted_rows["accepted_index"].astype(int)) != list(range(len(draws))):
        raise IngestError(f"{branch}: accepted-draw indices are not contiguous.")
    if not np.array_equal(
        accepted_rows["statistic"].to_numpy(dtype=float),
        np.asarray(draws, dtype=float),
    ):
        raise IngestError(f"{branch}: accepted statistics diverge from draw array.")
    # Verify the complete saved RNG state chain and status-specific provenance.
    previous_after: Optional[str] = None
    for _, row in attempt_audit.sort_values("attempt_index").iterrows():
        before_json = str(row["seed_state_before_json"])
        after_json = str(row["seed_state_after_json"])
        if hashlib.sha256(before_json.encode("utf-8")).hexdigest() != str(
            row["seed_state_before_sha256"]
        ):
            raise IngestError(f"{branch}: saved pre-draw RNG state digest is stale.")
        if hashlib.sha256(after_json.encode("utf-8")).hexdigest() != str(
            row["seed_state_after_sha256"]
        ):
            raise IngestError(f"{branch}: saved post-draw RNG state digest is stale.")
        if previous_after is not None and before_json != previous_after:
            raise IngestError(f"{branch}: RNG attempt-state chain is discontinuous.")
        previous_after = after_json
        if row["status"] == "accepted":
            if not bool(row["refit_completed"]) or len(str(row["registry_fingerprint"])) != 64:
                raise IngestError(f"{branch}: accepted draw provenance is incomplete.")
            if len(str(row["draw_content_sha256"])) != 64 or len(
                str(row["draw_identity_sha256"])
            ) != 64:
                raise IngestError(f"{branch}: accepted draw content identity is incomplete.")
        elif row["status"] == "rejected":
            if not str(row["reason_category"]).strip():
                raise IngestError(f"{branch}: rejected draw lacks a category.")
        else:
            raise IngestError(f"{branch}: attempt status {row['status']!r} is invalid.")
    # Raise only after the capped failure ledger has passed the same integrity
    # checks as a successful branch, and attach that ledger to the exception.
    if len(draws) < int(requested):
        raise ResamplingAttemptCapError(
            f"{branch}: produced {len(draws)} accepted draw(s) of requested "
            f"{requested} within {attempt_cap} attempts; {discarded} rejected.",
            attempt_audit,
        )
    # Log the branch's outcome with its complete rejection category tally.
    reason_counts = (
        attempt_audit.loc[attempt_audit["status"] == "rejected", "reason_category"]
        .value_counts()
        .to_dict()
    )
    logger.info(
        "%s: %d accepted draw(s), %d rejected attempt(s), categories=%s.",
        branch,
        len(draws),
        discarded,
        reason_counts,
    )
    return np.array(draws, dtype=float), discarded, attempt_audit


# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 3: Summation formulas and finite-draw corrections
# -------------------------------------------------------------------------------------------------------------------------------


# Compute the percentile interval of one branch's draws.
def compute_percentile_interval(
    draws: np.ndarray,
    lower: float = PERCENTILE_LOWER,
    upper: float = PERCENTILE_UPPER,
    branch: str = "",
) -> Tuple[float, float]:
    """
    Compute the percentile interval of one branch's draws.

    Parameters
    ----------
    draws : numpy.ndarray
        The valid draw statistics of the branch.
    lower : float, optional
        The lower percentile.
    upper : float, optional
        The upper percentile.
    branch : str, optional
        The branch label, used in the failure message.

    Returns
    -------
    tuple
        The lower and upper percentiles of the valid draws.

    Raises
    ------
    IngestError
        If the draw array is empty, because an empty array has no percentile.
    """
    # Refuse an empty draw array.
    if draws.size == 0:
        # The failure names the branch.
        raise IngestError(
            f"{branch}: cannot compute a percentile interval from an empty draw "
            f"array."
        )
    # Compute both percentiles over the valid draws alone.
    bounds: np.ndarray = np.percentile(draws, [float(lower), float(upper)])
    # Return the two bounds as native scalars.
    return float(bounds[0]), float(bounds[1])


# Compute the finite-draw-corrected placebo probability.
def compute_placebo_p(observed: float, draws: np.ndarray) -> float:
    """
    Compute the finite-draw-corrected placebo probability.

    Parameters
    ----------
    observed : float
        The observed statistic of the partition.
    draws : numpy.ndarray
        The valid placebo draw statistics.

    Returns
    -------
    float
        The corrected probability, which is one plus the number of draws whose
        absolute statistic attains the observed absolute statistic, divided by
        one plus the number of draws. It is a diagnostic probability and not a
        confidence level, and no significance marker is derived from it.

    Raises
    ------
    IngestError
        If the draw array is empty.
    """
    # Refuse an empty draw array.
    if draws.size == 0:
        # The failure states the condition.
        raise IngestError(
            "Cannot compute a corrected placebo probability from an empty draw "
            "array."
        )
    # Count the draws whose absolute statistic attains the observed value's.
    exceedances: int = int(np.count_nonzero(np.abs(draws) >= abs(float(observed))))
    # Apply the finite-draw correction in both the numerator and denominator.
    probability: float = (1.0 + float(exceedances)) / (1.0 + float(draws.size))
    # Refuse a probability outside the unit interval, which the formula cannot
    # produce and which would indicate a counting defect.
    if not (0.0 < probability <= 1.0):
        # The failure names the probability and the counts.
        raise IngestError(
            f"Placebo probability {probability!r} is outside the unit interval "
            f"for {exceedances} exceedance(s) over {draws.size} draw(s)."
        )
    # Return the corrected probability.
    return probability


# Flag a branch whose discard rate exceeds the declared share.
def evaluate_fragility(
    discarded: int,
    requested: int,
    share: float = FRAGILE_DISCARD_SHARE,
    attempted: Optional[int] = None,
    accepted: Optional[int] = None,
    rejection_categories: Optional[Mapping[str, int]] = None,
    audit_log: Optional[List[Dict[str, Any]]] = None,
    branch: str = "",
) -> bool:
    """
    Flag a branch whose discard rate exceeds the declared share.

    Parameters
    ----------
    discarded : int
        The number of discarded attempts.
    requested : int
        The number of valid replicates requested.
    share : float, optional
        The declared fragile share.
    attempted : int, optional
        The total attempted draws. Defaults to accepted plus discarded.
    accepted : int, optional
        The accepted count. Defaults to requested for a completed branch.
    rejection_categories : mapping of str to int, optional
        Complete rejection-category counts for the audit record.
    audit_log : list of dict, optional
        Mutable caller-owned accumulator receiving the denominator audit.
    branch : str, optional
        Branch identity attached to the audit row.

    Returns
    -------
    bool
        True when the discarded share exceeds the declared share.

    Raises
    ------
    IngestError
        If counts, threshold, denominator identity or categories are invalid.
    """
    # Validate count declarations without Boolean/fractional coercion.
    for label, value, positive in (
        ("discarded", discarded, False), ("requested", requested, True)
    ):
        if (
            isinstance(value, (bool, np.bool_))
            or not isinstance(value, (int, np.integer))
            or int(value) < (1 if positive else 0)
        ):
            raise IngestError(f"Fragility {label} count is invalid: {value!r}.")
    if isinstance(share, (bool, np.bool_)):
        raise IngestError("Fragility threshold cannot be Boolean.")
    try:
        threshold = float(share)
    except (TypeError, ValueError, OverflowError) as exc:
        raise IngestError("Fragility threshold is malformed.") from exc
    if not math.isfinite(threshold) or not 0.0 <= threshold <= 1.0:
        raise IngestError("Fragility threshold must lie in [0, 1].")
    accepted_count = int(requested) if accepted is None else accepted
    if (
        isinstance(accepted_count, (bool, np.bool_))
        or not isinstance(accepted_count, (int, np.integer))
        or int(accepted_count) < 0
    ):
        raise IngestError(
            f"Fragility accepted count is invalid: {accepted_count!r}."
        )
    attempted_count = (
        int(accepted_count) + int(discarded) if attempted is None else attempted
    )
    if (
        isinstance(attempted_count, (bool, np.bool_))
        or not isinstance(attempted_count, (int, np.integer))
        or int(attempted_count) < 0
    ):
        raise IngestError(
            f"Fragility attempted count is invalid: {attempted_count!r}."
        )
    if int(accepted_count) > int(requested):
        raise IngestError("Fragility accepted count exceeds requested count.")
    if int(attempted_count) != int(accepted_count) + int(discarded):
        raise IngestError(
            "Fragility denominator must equal accepted plus rejected attempts."
        )
    if int(attempted_count) <= 0:
        raise IngestError("Cannot evaluate fragility without a positive request.")
    categories = dict(rejection_categories or {})
    if int(discarded) > 0 and not categories:
        # Preserve the original two-count direct-call contract while marking
        # that it supplied no finer taxonomy. The orchestrator always passes
        # the complete registered categories from its attempt ledger.
        categories = {"unspecified_legacy_direct_call": int(discarded)}
    category_total = 0
    for category, count in categories.items():
        if not str(category).strip() or isinstance(count, (bool, np.bool_)) or not isinstance(count, (int, np.integer)) or int(count) < 0:
            raise IngestError("Fragility rejection-category audit is malformed.")
        category_total += int(count)
    if categories and category_total != int(discarded):
        raise IngestError(
            "Fragility rejection categories do not sum to rejected attempts."
        )
    rejection_rate = float(discarded) / float(attempted_count)
    flag = bool(rejection_rate > threshold)
    audit_row = {
        "branch": str(branch),
        "requested": int(requested),
        "accepted": int(accepted_count),
        "rejected": int(discarded),
        "attempted": int(attempted_count),
        "denominator": "attempted_draws",
        "rejection_rate": rejection_rate,
        "fragility_threshold": threshold,
        "rejection_categories_json": json.dumps(categories, sort_keys=True),
        "fragile": flag,
    }
    if audit_log is not None:
        audit_log.append(audit_row)
    logger.info("fragility audit: %s", audit_row)
    return flag


# Assert that no significance marker derives from the placebo branch.
def assert_no_marker_from_placebo(result: ResamplingResult) -> None:
    """
    Assert that no significance marker derives from the placebo branch.

    Parameters
    ----------
    result : ResamplingResult
        The resampling result.

    Returns
    -------
    None
        The function returns nothing when the isolation holds.

    Raises
    ------
    IngestError
        If the result carries any attribute whose name implies a marker,
        because the specification fixes that the placebo probability is a
        diagnostic and never a significance assignment.
    """
    # Identify any attribute whose name implies a significance marker.
    marker_attributes: List[str] = [
        name
        for name in vars(result)
        if "star" in name.lower() or "significance" in name.lower()
    ]
    # Refuse a marker-bearing attribute.
    if marker_attributes:
        # The failure names the offending attributes.
        raise IngestError(
            f"The resampling result carries marker attribute(s) "
            f"{marker_attributes}; the placebo probability never assigns a "
            f"marker."
        )
    # Log the verified isolation.
    logger.info(
        "placebo isolation verified: the result carries no marker attribute."
    )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the resampling inference across the three branches.
def orchestrate_resampling_inference(
    policy_name: str,
    outcome_name: str,
    baseline_config: Mapping[str, Any],
    data: Mapping[str, Any],
    seeds: Mapping[str, Any],
    region_map: Mapping[str, str],
    bootstrap_replicates: int = DEFAULT_BOOTSTRAP_REPLICATES,
    placebo_replicates: int = DEFAULT_PLACEBO_REPLICATES,
    attempt_cap_multiple: int = ATTEMPT_CAP_MULTIPLE,
    expected: Optional[Mapping[str, float]] = None,
) -> ResamplingResult:
    """
    Compose the resampling inference across the three branches.

    Parameters
    ----------
    policy_name : str
        The policy instrument of the pair.
    outcome_name : str
        The canonical outcome variable of the pair.
    baseline_config : mapping
        The baseline configuration, which the orchestration passes through.
    data : mapping
        The frozen inputs under the declared keys.
    seeds : mapping
        The explicit seed material carrying the master entropy.
    region_map : mapping of str to str
        The region of each country code, required by the interface even
        though only the region branch consumes it, so that a stored result is
        reproducible from its own inputs.
    bootstrap_replicates : int, optional
        The number of valid replicates of each bootstrap branch.
    placebo_replicates : int, optional
        The number of valid replicates of the placebo branch.
    attempt_cap_multiple : int, optional
        The attempt cap as a multiple of the requested count.
    expected : mapping of str to float, optional
        The declared study values, compared only when supplied.

    Returns
    -------
    ResamplingResult
        The two intervals, the corrected probability, the discard counts, the
        draw arrays, complete per-attempt provenance ledgers, attempted-draw
        fragility audit, flags and seed material.

    Raises
    ------
    KeyError
        If the data mapping lacks a declared key.
    IngestError
        If the observed run, a branch or a summation check fails.
    """
    # Refuse a data mapping lacking a declared key.
    for key in ("registry", "panel", "lagged"):
        # Test the key's presence.
        if key not in data:
            # The failure names the absent key.
            raise KeyError(f"The data mapping lacks the declared key {key!r}.")
    # Select the partition from the frozen registry.
    registry: PartitionRegistry = data["registry"]
    # Refuse a pair the registry does not carry.
    if (policy_name, outcome_name) not in registry.partitions:
        # The failure names the pair.
        raise IngestError(f"The registry does not carry the pair {(policy_name, outcome_name)}.")
    # Read the partition.
    partition: Partition = registry.partitions[(policy_name, outcome_name)]
    # Read the frozen panel.
    panel: pd.DataFrame = data["panel"]
    # Spawn one independent generator per branch from the declared seeds.
    generators: Dict[str, np.random.Generator] = spawn_branch_generators(seeds)
    # Run the composition once on the frozen data to obtain the observed
    # statistic the placebo comparison consumes.
    observed: float = run_replicate(policy_name, outcome_name, baseline_config, data)
    # Log the observed statistic.
    logger.info(
        "%s/%s: observed incremental effect %.9f.",
        policy_name,
        outcome_name,
        observed,
    )
    # Draw the country-history branch's replicates.
    ch_draws, ch_discarded, ch_audit = draw_branch_replicates(
        policy_name,
        outcome_name,
        baseline_config,
        BRANCH_COUNTRY_HISTORY,
        lambda: resample_country_history(
            policy_name, outcome_name, partition, panel, generators[BRANCH_COUNTRY_HISTORY]
        ),
        bootstrap_replicates,
        generators[BRANCH_COUNTRY_HISTORY],
        attempt_cap_multiple,
    )
    # Draw the region-block branch's replicates.
    rb_draws, rb_discarded, rb_audit = draw_branch_replicates(
        policy_name,
        outcome_name,
        baseline_config,
        BRANCH_REGION_BLOCK,
        lambda: resample_region_blocks(
            policy_name,
            outcome_name,
            partition,
            panel,
            region_map,
            generators[BRANCH_REGION_BLOCK],
        ),
        bootstrap_replicates,
        generators[BRANCH_REGION_BLOCK],
        attempt_cap_multiple,
    )
    # Draw the placebo branch's replicates.
    pl_draws, pl_discarded, pl_audit = draw_branch_replicates(
        policy_name,
        outcome_name,
        baseline_config,
        BRANCH_PLACEBO,
        lambda: resample_placebo_coalition(
            policy_name, outcome_name, partition, panel, generators[BRANCH_PLACEBO]
        ),
        placebo_replicates,
        generators[BRANCH_PLACEBO],
        attempt_cap_multiple,
    )
    # Compute the country-history interval.
    ci_ch_lower, ci_ch_upper = compute_percentile_interval(
        ch_draws, branch=BRANCH_COUNTRY_HISTORY
    )
    # Compute the region-block interval.
    ci_rb_lower, ci_rb_upper = compute_percentile_interval(
        rb_draws, branch=BRANCH_REGION_BLOCK
    )
    # Compute the finite-draw-corrected placebo probability.
    p_placebo: float = compute_placebo_p(observed, pl_draws)
    # Compose the discard counts per branch.
    discarded_counts: Dict[str, int] = {
        BRANCH_COUNTRY_HISTORY: int(ch_discarded),
        BRANCH_REGION_BLOCK: int(rb_discarded),
        BRANCH_PLACEBO: int(pl_discarded),
    }
    # Retain every accepted and rejected attempt in full.
    attempt_audits: Dict[str, pd.DataFrame] = {
        BRANCH_COUNTRY_HISTORY: ch_audit,
        BRANCH_REGION_BLOCK: rb_audit,
        BRANCH_PLACEBO: pl_audit,
    }
    # Compose the fragility flags and denominator audit per branch.
    fragility_rows: List[Dict[str, Any]] = []
    requested_by_branch = {
        BRANCH_COUNTRY_HISTORY: int(bootstrap_replicates),
        BRANCH_REGION_BLOCK: int(bootstrap_replicates),
        BRANCH_PLACEBO: int(placebo_replicates),
    }
    fragile_flags: Dict[str, bool] = {
        branch: evaluate_fragility(
            discarded_counts[branch],
            requested_by_branch[branch],
            attempted=len(attempt_audits[branch]),
            accepted=int((attempt_audits[branch]["status"] == "accepted").sum()),
            rejection_categories={
                str(category): int(count)
                for category, count in attempt_audits[branch]
                .loc[attempt_audits[branch]["status"] == "rejected", "reason_category"]
                .value_counts()
                .items()
            },
            audit_log=fragility_rows,
            branch=branch,
        )
        for branch in BRANCH_ORDER
    }
    fragility_audit = pd.DataFrame(fragility_rows).set_index("branch")
    # Compose the draw arrays per branch, persisted in full for audit.
    draw_arrays: Dict[str, np.ndarray] = {
        BRANCH_COUNTRY_HISTORY: ch_draws,
        BRANCH_REGION_BLOCK: rb_draws,
        BRANCH_PLACEBO: pl_draws,
    }
    # Record the complete rejection-category tally in the log.
    logger.info(
        "resampling rejection categories: %s.",
        {
            branch: attempt_audits[branch]
            .loc[attempt_audits[branch]["status"] == "rejected", "reason_category"]
            .value_counts()
            .to_dict()
            for branch in BRANCH_ORDER
        },
    )
    # Flag the branch in the log when the discard rate is elevated.
    for branch, flag in fragile_flags.items():
        # Emit the fragility record when the flag is raised.
        if flag:
            # Log the flag with the branch's discard count.
            logger.warning(
                "%s/%s: branch %s flagged numerically fragile with %d "
                "discarded attempt(s).",
                policy_name,
                outcome_name,
                branch,
                discarded_counts[branch],
            )
    # Assemble the frozen resampling result.
    result: ResamplingResult = ResamplingResult(
        policy_name=str(policy_name),
        outcome_name=str(outcome_name),
        ci_ch_lower=float(ci_ch_lower),
        ci_ch_upper=float(ci_ch_upper),
        ci_rb_lower=float(ci_rb_lower),
        ci_rb_upper=float(ci_rb_upper),
        p_placebo_hat=float(p_placebo),
        discarded_counts=discarded_counts,
        full_draw_arrays=draw_arrays,
        fragile_flags=fragile_flags,
        attempt_audits=attempt_audits,
        fragility_audit=fragility_audit,
        seeds={
            "master": seeds["master"],
            "branches": list(BRANCH_ORDER),
            "branch_spawn_keys": {
                branch: [index] for index, branch in enumerate(BRANCH_ORDER)
            },
            "replicates": {
                BRANCH_COUNTRY_HISTORY: int(bootstrap_replicates),
                BRANCH_REGION_BLOCK: int(bootstrap_replicates),
                BRANCH_PLACEBO: int(placebo_replicates),
            },
        },
    )
    # Assert that no significance marker derives from the placebo branch.
    assert_no_marker_from_placebo(result)
    # Compare the intervals with the declared values when they were supplied.
    if expected is not None:
        # Iterate over the declared quantities.
        for label, declared in expected.items():
            # Read the observed quantity of the same name.
            observed_value: Optional[float] = getattr(result, label, None)
            # Skip a label the result does not carry, which the caller's own
            # declaration controls.
            if observed_value is None:
                # Continue to the next declared quantity.
                continue
            # Refuse a divergence beyond a tenth of a percentage point.
            if abs(float(observed_value) - float(declared)) > 1.0e-3:
                # The failure names both values.
                raise IngestError(
                    f"{label}: {observed_value!r} against the declared "
                    f"{declared!r}."
                )
    # Log the completed inference.
    logger.info(
        "resampling inference complete for %s/%s: CH [%.6f, %.6f], RB [%.6f, "
        "%.6f], placebo probability %.6f.",
        policy_name,
        outcome_name,
        ci_ch_lower,
        ci_ch_upper,
        ci_rb_lower,
        ci_rb_upper,
        p_placebo,
    )
    # Return the assembled result.
    return result
````

````python
# Task 17: Execute Bootstrap and Placebo Inference

# ==============================================================================
# Task 17: Execute Bootstrap and Placebo Inference
# ==============================================================================

# Declare the seven World Bank regions, which are the only admissible region
# labels; an income-group label is not a region and must never appear.
WORLD_BANK_REGIONS: Final[Tuple[str, ...]] = (
    "East Asia & Pacific",
    "Europe & Central Asia",
    "Latin America & Caribbean",
    "Middle East & North Africa",
    "North America",
    "South Asia",
    "Sub-Saharan Africa",
)
# Declare the tokens that identify an income-group label rather than a region.
INCOME_GROUP_TOKENS: Final[Tuple[str, ...]] = ("income", "ida", "ibrd")
# Declare the column name carrying the region in the metadata registry.
REGION_COLUMN: Final[str] = "world_bank_region"
# Declare the metadata fields that identify the authoritative region release.
REGION_SOURCE_COLUMN: Final[str] = "region_source"
REGION_VERSION_COLUMN: Final[str] = "region_version"
# Declare the key under which the configuration documents the seeds.
SEED_RECORD_PATH: Final[Tuple[str, str]] = ("reproducibility_gaps", "random_seeds")
# Declare the confidence level used by the distribution-free Monte-Carlo
# envelopes in the cross-seed stability audit.
MONTE_CARLO_CONFIDENCE: Final[float] = 0.95
# Declare the smallest accepted sample for which a seed comparison is allowed
# to make a quantitative stability claim.
STABILITY_MINIMUM_DRAWS: Final[int] = 30
# Declare the fixed quantile convention recorded beside every interval.
PERCENTILE_QUANTILE_METHOD: Final[str] = "numpy_linear"
# Cap the number of offenders quoted inside any raised message.
INFERENCE_MAX_REPORTED_OFFENDERS: Final[int] = 12


# Preserve the provenance of a metadata-derived mapping without inserting a
# pseudo-country into the scientific country-to-region namespace.
class AuthoritativeRegionMap(dict):
    """Country-to-region mapping carrying immutable source/version identity."""

    # Construct the mapping and retain its authoritative release identity.
    def __init__(
        self,
        values: Mapping[str, str],
        *,
        source_id: str,
        version_id: str,
    ) -> None:
        super().__init__((str(key), str(value)) for key, value in values.items())
        object.__setattr__(
            self,
            "provenance",
            MappingProxyType(
                {"source_id": str(source_id), "version_id": str(version_id)}
            ),
        )
        object.__setattr__(self, "_locked", True)

    # Refuse replacing provenance or disabling the mutation lock.
    def __setattr__(self, name: str, value: Any) -> None:
        if getattr(self, "_locked", False):
            raise TypeError("AuthoritativeRegionMap metadata is read-only.")
        object.__setattr__(self, name, value)

    # Refuse mutation after construction so validation cannot become stale.
    def __setitem__(self, key: Any, value: Any) -> None:
        if getattr(self, "_locked", False):
            raise TypeError("AuthoritativeRegionMap is read-only.")
        super().__setitem__(key, value)

    # Refuse every ordinary mutating dictionary operation.
    def __delitem__(self, key: Any) -> None:
        raise TypeError("AuthoritativeRegionMap is read-only.")

    def clear(self) -> None:
        raise TypeError("AuthoritativeRegionMap is read-only.")

    def pop(self, key: Any, default: Any = None) -> Any:
        raise TypeError("AuthoritativeRegionMap is read-only.")

    def popitem(self) -> Tuple[Any, Any]:
        raise TypeError("AuthoritativeRegionMap is read-only.")

    def setdefault(self, key: Any, default: Any = None) -> Any:
        raise TypeError("AuthoritativeRegionMap is read-only.")

    def update(self, *args: Any, **kwargs: Any) -> None:
        raise TypeError("AuthoritativeRegionMap is read-only.")


# Declare the frozen result of the resampling execution.
@dataclass(frozen=True)
class InferenceExecutionResult:
    """
    Frozen result of the bootstrap and placebo execution.

    Attributes
    ----------
    intervals : pandas.DataFrame
        One row per pair carrying the two bootstrap intervals, the corrected
        placebo probability and the per-branch discarded counts.
    placebo : pandas.DataFrame
        One row per pair carrying the corrected probability, the exceedance
        count and the number of valid placebo draws.
    draws : dict
        The draw arrays keyed on the pair and the branch label.
    stability : pandas.DataFrame
        One row per pair, branch and summary statistic comparing interval
        endpoints or placebo tail probabilities under declared Monte-Carlo
        uncertainty, separately from qualitative conclusions.
    composition_audit : pandas.DataFrame
        The actual region-branch attempt ledger with sampled-block
        composition, seed state, acceptance/rejection and refit provenance.
    region_validation : pandas.DataFrame
        The observed region vocabulary and its member counts per pair.
    table_fingerprint_sha256 : str
        Digest of the canonical interval and placebo tables.
    completed_utc : str
        ISO-8601 UTC timestamp of the execution.
    notes : tuple of str
        Recorded decisions, including the stability mode and the design
        properties of the region branch.
    stability_draws : dict
        Every accepted stability-seed draw array, keyed as ``draws`` is.
    attempt_audits : dict
        Complete primary and stability attempt ledgers keyed on seed-set id,
        pair and branch.
    seed_audit : pandas.DataFrame
        The seed-set and branch identifiers matched to the configuration
        registry before any random draw.
    publication_audit : pandas.DataFrame
        Cell-level declared-value comparisons under exact, rounding or
        Monte-Carlo rules.
    region_provenance : mapping
        Source/version identity of the authoritative region metadata.
    """
    # One row per pair carrying the intervals and the probability.
    intervals: pd.DataFrame
    # One row per pair carrying the corrected probability and its components.
    placebo: pd.DataFrame
    # The draw arrays keyed on the pair and the branch.
    draws: Mapping[Tuple[Tuple[str, str], str], np.ndarray]
    # One row per pair and branch recording the stability difference.
    stability: pd.DataFrame
    # The actual region-branch composition and attempt audit.
    composition_audit: pd.DataFrame
    # The observed region vocabulary and member counts.
    region_validation: pd.DataFrame
    # Digest of the canonical tables.
    table_fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the execution.
    completed_utc: str
    # Recorded decisions of the execution.
    notes: Tuple[str, ...]
    # Every accepted draw array from the stability seed set.
    stability_draws: Mapping[Tuple[Tuple[str, str], str], np.ndarray] = field(
        default_factory=dict
    )
    # Complete attempt ledgers from both registered seed sets.
    attempt_audits: Mapping[Tuple[str, Tuple[str, str], str], pd.DataFrame] = field(
        default_factory=dict
    )
    # Seed registration audit completed before drawing.
    seed_audit: pd.DataFrame = field(default_factory=pd.DataFrame)
    # Publication-aware declared-value comparison audit.
    publication_audit: pd.DataFrame = field(default_factory=pd.DataFrame)
    # Authoritative region source/version identity.
    region_provenance: Mapping[str, str] = field(default_factory=dict)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 17, Step 1: The seed record and the pair iteration
# -------------------------------------------------------------------------------------------------------------------------------


# Resolve one requested seed set against the mandatory configuration registry.
def _resolve_seed_registration(
    study_config: Mapping[str, Any], seeds: Mapping[str, Any]
) -> pd.DataFrame:
    """Return the exact seed-set/branch registrations matched before drawing."""
    # Require an explicit seed-set identity and integral master entropy.
    seed_set_id = str(seeds.get("seed_set_id", "")).strip()
    master = seeds.get("master")
    if not seed_set_id:
        raise IngestError("The run seed specification lacks a seed_set_id.")
    if (
        isinstance(master, (bool, np.bool_))
        or not isinstance(master, (int, np.integer))
        or int(master) < 0
    ):
        raise IngestError(
            "The run must carry a nonnegative integral, non-Boolean master seed."
        )
    # Resolve the nested registry without accepting the manuscript's null slots.
    parent = study_config.get(SEED_RECORD_PATH[0])
    record = parent.get(SEED_RECORD_PATH[1]) if isinstance(parent, Mapping) else None
    if not isinstance(record, Mapping) or not record:
        raise IngestError(
            "The configuration lacks a populated reproducibility_gaps.random_seeds "
            "registry."
        )
    entry = record.get(seed_set_id)
    if not isinstance(entry, Mapping) or not entry:
        raise IngestError(
            f"Seed set {seed_set_id!r} has no populated registry entry."
        )
    documented_master = entry.get("master")
    if (
        isinstance(documented_master, (bool, np.bool_))
        or not isinstance(documented_master, (int, np.integer))
        or int(documented_master) != int(master)
    ):
        raise IngestError(
            f"Seed set {seed_set_id!r} carries master {master!r}, but its "
            f"registry entry carries {documented_master!r}."
        )
    branches = entry.get("branches")
    if not isinstance(branches, Mapping) or set(branches) != set(BRANCH_ORDER):
        raise IngestError(
            f"Seed set {seed_set_id!r} must document exactly the branches "
            f"{list(BRANCH_ORDER)}."
        )
    canonical_keys = {branch: [index] for index, branch in enumerate(BRANCH_ORDER)}
    actual_keys = seeds.get("branch_spawn_keys", canonical_keys)
    if not isinstance(actual_keys, Mapping) or set(actual_keys) != set(BRANCH_ORDER):
        raise IngestError("The run's branch spawn-key specification is incomplete.")
    rows: List[Dict[str, Any]] = []
    # Match every branch's spawn key and persistent branch identifier exactly.
    for branch in BRANCH_ORDER:
        branch_entry = branches[branch]
        if not isinstance(branch_entry, Mapping) or not branch_entry:
            raise IngestError(
                f"Seed set {seed_set_id!r}/{branch} lacks a populated branch entry."
            )
        documented_key = branch_entry.get("spawn_key")
        if not isinstance(documented_key, Sequence) or isinstance(
            documented_key, (str, bytes)
        ):
            raise IngestError(
                f"Seed set {seed_set_id!r}/{branch} lacks a valid spawn_key."
            )
        actual_raw = actual_keys[branch]
        if not isinstance(actual_raw, Sequence) or isinstance(actual_raw, (str, bytes)):
            raise IngestError(
                f"Seed set {seed_set_id!r}/{branch} carries a malformed runtime spawn key."
            )
        raw_values = list(documented_key) + list(actual_raw)
        if any(
            isinstance(value, (bool, np.bool_))
            or not isinstance(value, (int, np.integer))
            or int(value) < 0
            for value in raw_values
        ):
            raise IngestError(
                f"Seed set {seed_set_id!r}/{branch} spawn keys must be "
                "nonnegative exact integers."
            )
        try:
            expected_key = [int(value) for value in documented_key]
            actual_key = [int(value) for value in actual_raw]
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError(
                f"Seed set {seed_set_id!r}/{branch} carries a malformed spawn key."
            ) from exc
        if expected_key != actual_key or actual_key != canonical_keys[branch]:
            raise IngestError(
                f"Seed set {seed_set_id!r}/{branch} spawn key {actual_key!r} "
                f"disagrees with registered {expected_key!r} or executed "
                f"SeedSequence key {canonical_keys[branch]!r}."
            )
        branch_seed_id = str(branch_entry.get("seed_id", "")).strip()
        if not branch_seed_id:
            raise IngestError(
                f"Seed set {seed_set_id!r}/{branch} lacks a persistent seed_id."
            )
        rows.append(
            {
                "seed_set_id": seed_set_id,
                "master": int(master),
                "branch": branch,
                "branch_seed_id": branch_seed_id,
                "spawn_key_json": json.dumps(actual_key),
                "registry_path": ".".join(SEED_RECORD_PATH + (seed_set_id, branch)),
            }
        )
    # Return a uniquely keyed registration audit for persistence in the result.
    return pd.DataFrame(rows).set_index(["seed_set_id", "branch"])


# Assert that the run's seeds are documented in the configuration record.
def assert_seed_documented(
    study_config: Mapping[str, Any], seeds: Mapping[str, Any]
) -> None:
    """
    Assert exact registration of a seed set and all three branch streams.

    The configuration must contain a populated entry under
    ``reproducibility_gaps.random_seeds[seed_set_id]`` with the same integral
    master and a mapping for every declared branch.  Each branch entry must
    carry its persistent ``seed_id`` and exact ``spawn_key``.  Null manuscript
    placeholders, absent branches and membership-only master matches fail.
    """
    # Resolution performs the complete fail-closed comparison.
    audit = _resolve_seed_registration(study_config, seeds)
    # Log only after all branches have matched.
    logger.info(
        "seed documentation verified for %s across %d branch(es).",
        audit.index.get_level_values("seed_set_id")[0],
        len(audit),
    )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 17, Step 2: The region mapping, its validation and the draw audit
# -------------------------------------------------------------------------------------------------------------------------------


# Build the region mapping from the country metadata registry.
def build_region_map(
    metadata_frame: pd.DataFrame,
    required_countries: Optional[Sequence[str]] = None,
) -> AuthoritativeRegionMap:
    """
    Build the region mapping from the country metadata registry.

    Parameters
    ----------
    metadata_frame : pandas.DataFrame
        The country metadata registry carrying the region column.

    Returns
    -------
    AuthoritativeRegionMap
        Read-only country-to-region mapping carrying source/version identity.

    Raises
    ------
    IngestError
        If required columns, authoritative identity or eligible-country
        coverage is absent, or if any label is missing or undeclared.
    """
    # Refuse a metadata frame lacking the country column.
    if "country_code" not in metadata_frame.columns:
        # Without the country column the mapping cannot be keyed.
        raise IngestError("Metadata registry lacks the country_code column.")
    # Refuse a metadata frame lacking the region column.
    if REGION_COLUMN not in metadata_frame.columns:
        # Without the region column the mapping cannot be built.
        raise IngestError(f"Metadata registry lacks the {REGION_COLUMN!r} column.")
    # Reject null identities before any string conversion can create a literal
    # "nan" country or region.
    null_rows = metadata_frame[["country_code", REGION_COLUMN]].isna().any(axis=1)
    if bool(null_rows.any()):
        raise IngestError("Metadata registry carries a null country or region label.")
    country_values = metadata_frame["country_code"].map(lambda value: str(value).strip())
    region_values = metadata_frame[REGION_COLUMN].map(lambda value: str(value).strip())
    blank_rows = country_values.eq("") | region_values.eq("")
    missing_tokens = region_values.str.lower().isin(
        {"nan", "none", "null", "na", "n/a", "unassigned", "unknown"}
    )
    if bool((blank_rows | missing_tokens).any()):
        raise IngestError("Metadata registry carries a blank or unassigned region.")
    if bool(country_values.duplicated(keep=False).any()):
        offenders = sorted(country_values[country_values.duplicated(keep=False)].unique())
        raise IngestError(
            f"Metadata registry duplicates country code(s) "
            f"{offenders[:INFERENCE_MAX_REPORTED_OFFENDERS]}."
        )
    # Resolve an explicit, single authoritative source/version identity from
    # columns or DataFrame attributes.
    if REGION_SOURCE_COLUMN in metadata_frame.columns:
        if bool(metadata_frame[REGION_SOURCE_COLUMN].isna().any()):
            raise IngestError("Region metadata source identity contains a null value.")
        source_series = metadata_frame[REGION_SOURCE_COLUMN].map(
            lambda value: str(value).strip()
        )
        if bool(source_series.eq("").any()):
            raise IngestError("Region metadata source identity contains a blank value.")
        source_values = source_series.unique()
    else:
        source_values = np.array([], dtype=object)
    if REGION_VERSION_COLUMN in metadata_frame.columns:
        if bool(metadata_frame[REGION_VERSION_COLUMN].isna().any()):
            raise IngestError("Region metadata version identity contains a null value.")
        version_series = metadata_frame[REGION_VERSION_COLUMN].map(
            lambda value: str(value).strip()
        )
        if bool(version_series.eq("").any()):
            raise IngestError("Region metadata version identity contains a blank value.")
        version_values = version_series.unique()
    else:
        version_values = np.array([], dtype=object)
    source_id = str(
        source_values[0] if len(source_values) == 1 else metadata_frame.attrs.get("region_source", "")
    ).strip()
    version_id = str(
        version_values[0] if len(version_values) == 1 else metadata_frame.attrs.get("region_version", "")
    ).strip()
    if len(source_values) > 1 or len(version_values) > 1 or not source_id or not version_id:
        raise IngestError(
            "Region metadata must carry one nonblank authoritative source and version."
        )
    # Compose the observed region vocabulary only after missingness validation.
    observed: List[str] = sorted(region_values.unique())
    # Identify the labels that are income-group labels rather than regions.
    income_labels: List[str] = [
        label
        for label in observed
        if any(token in label.lower() for token in INCOME_GROUP_TOKENS)
    ]
    # Refuse an income-group label, which the specification forbids.
    if income_labels:
        # The failure names the offending labels.
        raise IngestError(
            f"The region column carries income-group label(s) {income_labels}; "
            f"income groups are not regions."
        )
    # Identify the labels outside the declared region vocabulary.
    undeclared: List[str] = [
        label for label in observed if label not in WORLD_BANK_REGIONS
    ]
    # Refuse an undeclared label.
    if undeclared:
        # The failure names the offending labels.
        raise IngestError(
            f"The region column carries undeclared label(s) {undeclared}; the "
            f"declared vocabulary is the seven World Bank regions."
        )
    # Compose the mapping from country code to region.
    mapping_values: Dict[str, str] = dict(zip(country_values, region_values))
    required = set(str(code).strip() for code in (required_countries or ()))
    missing = sorted(required - set(mapping_values))
    if missing:
        raise IngestError(
            "Authoritative region metadata omits eligible countr(ies) "
            f"{missing[:INFERENCE_MAX_REPORTED_OFFENDERS]}."
        )
    mapping = AuthoritativeRegionMap(
        mapping_values, source_id=source_id, version_id=version_id
    )
    # Log the mapping's extent.
    logger.info(
        "region mapping built for %d countr(ies) across %d region(s).",
        len(mapping),
        len(set(mapping.values())),
    )
    # Return the mapping.
    return mapping


# Replay the region draw composition from the child seed stream.
def replay_region_draw_composition(
    seeds: Any,
    region_pool: Mapping[str, str],
    attempts: int,
) -> pd.DataFrame:
    """
    Replay the region draw composition from the child seed stream.

    Parameters
    ----------
    seeds : pandas.DataFrame
        The actual saved region-branch attempt ledger.  The historical
        parameter name is retained for call compatibility; no RNG is used.
    region_pool : mapping of str to str
        The exact original-country pool of the partition and its regions.
    attempts : int
        The number of draw attempts to replay, which is the resampler's own
        count of valid replicates plus discards.

    Returns
    -------
    pandas.DataFrame
        One row per saved attempt and region recording expected and executed
        multiplicities, accepted/rejected identity and exact roster checks.

    Raises
    ------
    IngestError
        If the ledger, mapping or any saved composition/roster is malformed or
        inconsistent with the executed attempt.
    """
    if not isinstance(seeds, pd.DataFrame) or seeds.empty:
        raise IngestError("Region composition replay requires a saved attempt ledger.")
    if not region_pool:
        raise IngestError("Cannot validate a region draw against an empty pool.")
    validate_region_mapping(region_pool)
    if (
        isinstance(attempts, (bool, np.bool_))
        or not isinstance(attempts, (int, np.integer))
        or int(attempts) < 0
        or int(attempts) != len(seeds)
    ):
        raise IngestError("Saved region attempt count disagrees with the ledger.")
    required_columns = {
        "attempt_index", "status", "accepted_index",
        "sampled_occurrence_ids_json", "sampled_originals_json",
        "draw_metadata_json", "registry_fingerprint",
    }
    if not required_columns.issubset(seeds.columns):
        raise IngestError("Saved region attempt ledger omits composition fields.")
    ordered = seeds.sort_values("attempt_index", kind="mergesort")
    if ordered["attempt_index"].astype(int).tolist() != list(range(len(ordered))):
        raise IngestError("Saved region attempt identifiers are not complete and contiguous.")
    members_by_region: Dict[str, List[str]] = {
        region: sorted(code for code, label in region_pool.items() if str(label) == region)
        for region in sorted(set(str(value) for value in region_pool.values()))
    }
    rows: List[Dict[str, Any]] = []
    # Reconstruct each actual saved attempt; never draw a new illustrative one.
    for _, saved in ordered.iterrows():
        attempt_index = int(saved["attempt_index"])
        try:
            metadata = json.loads(str(saved["draw_metadata_json"]))
            observed_ids = [str(value) for value in json.loads(str(saved["sampled_occurrence_ids_json"]))]
            observed_originals = [str(value) for value in json.loads(str(saved["sampled_originals_json"]))]
        except (TypeError, ValueError, json.JSONDecodeError) as exc:
            raise IngestError(
                f"Region attempt {attempt_index} carries malformed saved JSON."
            ) from exc
        drawn_regions = [str(value) for value in metadata.get("drawn_regions", ())]
        registered_pool = [str(value) for value in metadata.get("region_pool", ())]
        if registered_pool != sorted(members_by_region) or len(drawn_regions) != len(registered_pool):
            raise IngestError(
                f"Region attempt {attempt_index} disagrees with its original region pool."
            )
        if any(region not in members_by_region for region in drawn_regions):
            raise IngestError(f"Region attempt {attempt_index} draws an undeclared region.")
        expected_ids: List[str] = []
        expected_originals: List[str] = []
        for block_position, region in enumerate(drawn_regions):
            for member in members_by_region[region]:
                expected_originals.append(member)
                expected_ids.append(
                    f"{member}{DRAW_SUFFIX_SEPARATOR}region_{block_position:04d}"
                )
        # The registry canonicalizes the occurrence map by occurrence id, so
        # compare zipped identity pairs in that canonical order.  Keeping each
        # id bound to its original prevents two originals being swapped while
        # preserving the two marginal lists.
        observed_pairs = sorted(zip(observed_ids, observed_originals))
        expected_pairs = sorted(zip(expected_ids, expected_originals))
        if observed_pairs != expected_pairs:
            raise IngestError(
                f"Region attempt {attempt_index} executed roster disagrees with "
                "its saved block selection."
            )
        status = str(saved["status"])
        if status not in {"accepted", "rejected"}:
            raise IngestError(f"Region attempt {attempt_index} has invalid status {status!r}.")
        if status == "accepted" and len(str(saved["registry_fingerprint"])) != 64:
            raise IngestError(
                f"Accepted region attempt {attempt_index} lacks registry provenance."
            )
        for region, members in members_by_region.items():
            multiplicity = int(drawn_regions.count(region))
            rows.append(
                {
                    "attempt_index": attempt_index,
                    "status": status,
                    "accepted_index": saved["accepted_index"],
                    "region": region,
                    "pool_members": len(members),
                    "block_multiplicity": multiplicity,
                    "expected_occurrences": multiplicity * len(members),
                    "observed_occurrences": sum(
                        original in set(members) for original in observed_originals
                    ),
                    "membership_preserved": True,
                    "multiplicity_preserved": True,
                    "executed_roster_sha256": hashlib.sha256(
                        json.dumps(observed_ids, separators=(",", ":")).encode("utf-8")
                    ).hexdigest(),
                }
            )
    composition = pd.DataFrame(rows).set_index(["attempt_index", "region"])
    # Log the replay's extent.
    logger.info(
        "region draw composition replayed across %d attempt(s) over %d pool "
        "region(s).",
        len(composition),
        len(members_by_region),
    )
    # Return the composition frame.
    return composition


# -------------------------------------------------------------------------------------------------------------------------------
# Task 17, Step 3: The placebo table and its recomputation audit
# -------------------------------------------------------------------------------------------------------------------------------


# Build the placebo table with the recomputed exceedance count.
def build_placebo_table(
    results: Mapping[Tuple[str, str], ResamplingResult],
    observed_by_pair: Mapping[Tuple[str, str], float],
    seed_set_id: str = "",
    reference_window: Tuple[int, int] = PRE_ESCALATION_WINDOW,
    registry: Optional[PartitionRegistry] = None,
) -> pd.DataFrame:
    """
    Build the placebo table with the recomputed exceedance count.

    Parameters
    ----------
    results : mapping of tuple to ResamplingResult
        The resampling results keyed on the pair.
    observed_by_pair : mapping of tuple to float
        The observed incremental effect of each pair, computed independently
        of the resampling so that the recomputation is an audit rather than a
        restatement.

    Returns
    -------
    pandas.DataFrame
        One row per pair carrying the corrected probability, the exceedance
        count and the number of valid draws.

    Raises
    ------
    IngestError
        If a reported probability diverges from the independent recomputation,
        which would mean the formula and the reported value disagree, or if a
        pair's observed statistic was not supplied.
    """
    # Accumulate the placebo rows.
    rows: List[Dict[str, Any]] = []
    # Iterate over the pairs in sorted order.
    for pair in sorted(results.keys()):
        # Refuse a pair whose observed statistic was not supplied.
        if pair not in observed_by_pair:
            # The failure names the absent pair.
            raise IngestError(f"The observed statistic of {pair} was not supplied.")
        # Select the pair's resampling result.
        result = results[pair]
        # Read the placebo draws.
        draws: np.ndarray = result.full_draw_arrays[BRANCH_PLACEBO]
        # Read the pair's independently computed observed statistic.
        observed: float = float(observed_by_pair[pair])
        # Recompute the corrected probability through the single declared
        # implementation of the formula.
        recomputed: float = compute_placebo_p(observed, draws)
        # Refuse a divergence between the reported and the recomputed values.
        if abs(recomputed - float(result.p_placebo_hat)) > 1.0e-12:
            # The failure names both values and the pair.
            raise IngestError(
                f"{pair}: reported placebo probability "
                f"{result.p_placebo_hat!r} diverges from the recomputation "
                f"{recomputed!r} over {draws.size} draw(s) against an observed "
                f"statistic of {observed!r}."
            )
        # Count the draws attaining the observed absolute statistic, which is
        # the numerator's component and is reported for the audit.
        exceedances: int = int(
            np.count_nonzero(np.abs(draws) >= abs(observed))
        )
        # Append the placebo row of the pair.
        rows.append(
            {
                "policy": pair[0],
                "outcome": pair[1],
                "observed_incremental": float(observed),
                "p_placebo_hat": float(result.p_placebo_hat),
                "exceedances": int(exceedances),
                "valid_draws": int(draws.size),
                "seed_set_id": str(seed_set_id),
                "seed_master": int(result.seeds["master"]),
                "branch": BRANCH_PLACEBO,
                "reference_start": int(reference_window[0]),
                "reference_end": int(reference_window[1]),
                "replications": int(result.seeds["replicates"][BRANCH_PLACEBO]),
                "sample_n_targets": (
                    int(len(registry.partitions[pair].targets))
                    if registry is not None else pd.NA
                ),
                "sample_n_donors": (
                    int(len(registry.partitions[pair].donors))
                    if registry is not None else pd.NA
                ),
                "probability_formula": "(1 + exceedances) / (1 + valid_draws)",
            }
        )
    # Assemble the placebo frame keyed on the pair.
    table: pd.DataFrame = pd.DataFrame(rows).set_index(["policy", "outcome"])
    # Log the table's extent.
    logger.info("placebo table assembled across %d pair(s).", len(table))
    # Return the table.
    return table


# Recover the observed statistic of every pair independently.
def compute_observed_statistics(
    study_config: Mapping[str, Any],
    data: Mapping[str, Any],
    pairs: Sequence[Tuple[str, str]],
) -> Dict[Tuple[str, str], float]:
    """
    Recover the observed statistic of every pair independently.

    Parameters
    ----------
    study_config : mapping
        The baseline configuration.
    data : mapping
        The frozen inputs under the declared keys.
    pairs : sequence of tuple
        The pairs whose observed statistic is computed.

    Returns
    -------
    dict
        The observed incremental effect of each pair, computed by running the
        baseline interface once on the frozen data, which is independent of
        every resampled draw.

    Raises
    ------
    IngestError
        Propagated from the baseline interface, with the pair named.
    """
    # Initialise the observed mapping.
    observed: Dict[Tuple[str, str], float] = {}
    # Compute the observed statistic of each requested pair.
    for pair in pairs:
        # Run the baseline interface on the frozen data.
        try:
            # Compose the observed run of the pair.
            result = run_baseline(pair[0], pair[1], study_config, data)
        # Re-raise a failure with the pair named for exact localization.
        except IngestError as error:
            # Raise the localized failure.
            raise IngestError(f"{pair[0]}/{pair[1]}: {error}") from error
        # Record the pair's observed incremental effect.
        observed[pair] = float(result.att_incremental)
    # Log the recovery's extent.
    logger.info(
        "observed statistics recovered for %d pair(s) independently of the "
        "resampling.",
        len(observed),
    )
    # Return the observed mapping.
    return observed


# Run one seed set across the requested pairs.
def run_seed_set(
    study_config: Mapping[str, Any],
    data: Mapping[str, Any],
    seeds: Mapping[str, Any],
    region_map: Mapping[str, str],
    pairs: Sequence[Tuple[str, str]],
    bootstrap_replicates: int,
    placebo_replicates: int,
) -> Dict[Tuple[str, str], ResamplingResult]:
    """
    Run one seed set across the requested pairs.

    Parameters
    ----------
    study_config : mapping
        The baseline configuration.
    data : mapping
        The frozen inputs under the declared keys.
    seeds : mapping
        The seed material carrying the master entropy.
    region_map : mapping of str to str
        The region of each country code.
    pairs : sequence of tuple
        The pairs to run.
    bootstrap_replicates : int
        The number of bootstrap replicates per branch.
    placebo_replicates : int
        The number of placebo replicates.

    Returns
    -------
    dict
        The resampling results keyed on the pair.

    Raises
    ------
    IngestError
        Propagated from the resampling orchestrator, with the pair named so a
        failure localizes to one pair rather than to the whole execution.
    """
    # Initialise the result mapping.
    results: Dict[Tuple[str, str], ResamplingResult] = {}
    # Run each requested pair in turn.
    for pair in pairs:
        # Run the pair's resampling inference.
        try:
            # Compose the resampling inference of the pair.
            result: ResamplingResult = orchestrate_resampling_inference(
                pair[0],
                pair[1],
                study_config,
                data,
                seeds,
                region_map,
                bootstrap_replicates=bootstrap_replicates,
                placebo_replicates=placebo_replicates,
            )
        # Re-raise a failure with the pair named so the localization is exact.
        except IngestError as error:
            # Raise the localized failure.
            raise IngestError(f"{pair[0]}/{pair[1]}: {error}") from error
        # Bind the Task 17 seed-set identity to the Task 16 result without
        # changing its positional contract or its branch-stream provenance.
        enriched_seeds = dict(result.seeds)
        enriched_seeds["seed_set_id"] = str(seeds.get("seed_set_id", ""))
        result = replace(result, seeds=enriched_seeds)
        # Record the pair's result.
        results[pair] = result
    # Log the seed set's extent.
    logger.info(
        "seed set %s executed across %d pair(s).", seeds["master"], len(results)
    )
    # Return the results.
    return results


# Assemble the wide interval table from the per-pair results.
def build_interval_table(
    results: Mapping[Tuple[str, str], ResamplingResult],
    seed_set_id: str = "",
    reference_window: Tuple[int, int] = PRE_ESCALATION_WINDOW,
    registry: Optional[PartitionRegistry] = None,
) -> pd.DataFrame:
    """
    Assemble the wide interval table from the per-pair results.

    Parameters
    ----------
    results : mapping of tuple to ResamplingResult
        The resampling results keyed on the pair.

    Returns
    -------
    pandas.DataFrame
        One row per pair carrying the two bootstrap intervals, the corrected
        placebo probability and the per-branch discarded counts.

    Raises
    ------
    IngestError
        If a pair's intervals are not ordered.
    """
    # Accumulate the interval rows.
    rows: List[Dict[str, Any]] = []
    # Iterate over the pairs in sorted order.
    for pair in sorted(results.keys()):
        # Select the pair's result.
        result = results[pair]
        # Refuse unordered intervals, which would indicate a percentile defect.
        if result.ci_ch_lower > result.ci_ch_upper or result.ci_rb_lower > result.ci_rb_upper:
            # The failure names the pair.
            raise IngestError(f"{pair}: interval bounds are not ordered.")
        # Append the pair's interval row.
        rows.append(
            {
                "policy": pair[0],
                "outcome": pair[1],
                "ci_ch_lower": float(result.ci_ch_lower),
                "ci_ch_upper": float(result.ci_ch_upper),
                "ci_rb_lower": float(result.ci_rb_lower),
                "ci_rb_upper": float(result.ci_rb_upper),
                "p_placebo_hat": float(result.p_placebo_hat),
                "discarded_ch": int(result.discarded_counts[BRANCH_COUNTRY_HISTORY]),
                "discarded_rb": int(result.discarded_counts[BRANCH_REGION_BLOCK]),
                "discarded_pl": int(result.discarded_counts[BRANCH_PLACEBO]),
                "attempted_ch": int(result.fragility_audit.loc[BRANCH_COUNTRY_HISTORY, "attempted"]),
                "attempted_rb": int(result.fragility_audit.loc[BRANCH_REGION_BLOCK, "attempted"]),
                "attempted_pl": int(result.fragility_audit.loc[BRANCH_PLACEBO, "attempted"]),
                "rejection_rate_ch": float(result.fragility_audit.loc[BRANCH_COUNTRY_HISTORY, "rejection_rate"]),
                "rejection_rate_rb": float(result.fragility_audit.loc[BRANCH_REGION_BLOCK, "rejection_rate"]),
                "rejection_rate_pl": float(result.fragility_audit.loc[BRANCH_PLACEBO, "rejection_rate"]),
                "accepted_ch": int(result.full_draw_arrays[BRANCH_COUNTRY_HISTORY].size),
                "accepted_rb": int(result.full_draw_arrays[BRANCH_REGION_BLOCK].size),
                "accepted_pl": int(result.full_draw_arrays[BRANCH_PLACEBO].size),
                "quantile_method": PERCENTILE_QUANTILE_METHOD,
                "quantile_lower_percent": float(PERCENTILE_LOWER),
                "quantile_upper_percent": float(PERCENTILE_UPPER),
                "seed_set_id": str(seed_set_id),
                "seed_master": int(result.seeds["master"]),
                "reference_start": int(reference_window[0]),
                "reference_end": int(reference_window[1]),
                "sample_n_targets": (
                    int(len(registry.partitions[pair].targets))
                    if registry is not None else pd.NA
                ),
                "sample_n_donors": (
                    int(len(registry.partitions[pair].donors))
                    if registry is not None else pd.NA
                ),
            }
        )
    # Assemble the interval frame keyed on the pair.
    table: pd.DataFrame = pd.DataFrame(rows).set_index(["policy", "outcome"])
    # Log the table's extent.
    logger.info("interval table assembled across %d pair(s).", len(table))
    # Return the table.
    return table


# Compare two seed sets through scientific summaries and Monte-Carlo error.
def build_stability_table(
    primary: Mapping[Tuple[str, str], ResamplingResult],
    stability: Mapping[Tuple[str, str], ResamplingResult],
    reference_window: Tuple[int, int] = PRE_ESCALATION_WINDOW,
) -> pd.DataFrame:
    """
    Compare interval endpoints and placebo tails across independent seed sets.

    Parameters
    ----------
    primary : mapping of tuple to ResamplingResult
        The primary seed set's results.
    stability : mapping of tuple to ResamplingResult
        The stability seed set's results.

    Returns
    -------
    pandas.DataFrame
        One row per pair, branch and summary statistic carrying independent
        seed estimates, a declared Monte-Carlo allowance, quantitative status
        and separate qualitative sign/significance conclusions.

    Raises
    ------
    IngestError
        If pair coverage or any draw array is malformed.
    """
    # Refuse divergent coverage between the two seed sets.
    if set(primary.keys()) != set(stability.keys()):
        # The failure states the divergence.
        raise IngestError(
            "The primary and stability seed sets cover different pairs."
        )
    # Classify an interval's qualitative conclusion without conflating it with
    # numerical endpoint stability.
    def interval_conclusion(lower: float, upper: float) -> str:
        if lower > 0.0:
            return "positive_excludes_zero"
        if upper < 0.0:
            return "negative_excludes_zero"
        return "includes_zero"

    # Classify a placebo probability under the documented strict thresholds.
    def tail_conclusion(probability: float) -> str:
        if probability < 0.01:
            return "p_lt_0.01"
        if probability < 0.05:
            return "p_lt_0.05"
        if probability < 0.10:
            return "p_lt_0.10"
        return "p_ge_0.10"

    # Compute a distribution-free empirical-quantile radius through the DKW
    # rank envelope, avoiding arbitrary pairing of independent draws.
    def quantile_radius(draws: np.ndarray, probability: float) -> float:
        values = np.asarray(draws, dtype=float)
        if values.ndim != 1 or values.size == 0 or not np.isfinite(values).all():
            raise IngestError("Stability audit received a malformed draw array.")
        alpha = 1.0 - MONTE_CARLO_CONFIDENCE
        epsilon = math.sqrt(math.log(2.0 / alpha) / (2.0 * values.size))
        center = float(np.quantile(values, probability, method="linear"))
        lower_p = max(0.0, probability - epsilon)
        upper_p = min(1.0, probability + epsilon)
        envelope = np.quantile(values, [lower_p, upper_p], method="linear")
        return float(max(abs(center - float(envelope[0])), abs(float(envelope[1]) - center)))

    # Accumulate the summary-aligned stability rows.
    rows: List[Dict[str, Any]] = []
    for pair in sorted(primary.keys()):
        first_result = primary[pair]
        second_result = stability[pair]
        # Bind mapping keys to the fitted scientific identities and refuse
        # incomparable replication designs before inspecting summaries.
        for label, result in (("primary", first_result), ("stability", second_result)):
            if (str(result.policy_name), str(result.outcome_name)) != pair:
                raise IngestError(
                    f"{pair}: {label} result identity is "
                    f"{(result.policy_name, result.outcome_name)!r}."
                )
            if not isinstance(result.seeds.get("replicates"), Mapping):
                raise IngestError(f"{pair}: {label} result lacks replication provenance.")
        first_seed_id = str(first_result.seeds.get("seed_set_id", "")).strip()
        second_seed_id = str(second_result.seeds.get("seed_set_id", "")).strip()
        if first_seed_id and second_seed_id and first_seed_id == second_seed_id:
            raise IngestError(f"{pair}: stability comparison reuses one seed-set id.")
        if int(first_result.seeds.get("master", -1)) == int(
            second_result.seeds.get("master", -1)
        ):
            raise IngestError(f"{pair}: stability comparison reuses one master seed.")
        first_reps = first_result.seeds["replicates"]
        second_reps = second_result.seeds["replicates"]
        if dict(first_reps) != dict(second_reps):
            raise IngestError(f"{pair}: seed sets use incomparable replication requests.")
        for branch in BRANCH_ORDER:
            first_count = int(np.asarray(first_result.full_draw_arrays[branch]).size)
            second_count = int(np.asarray(second_result.full_draw_arrays[branch]).size)
            requested = int(first_reps[branch])
            if first_count != requested or second_count != requested:
                raise IngestError(
                    f"{pair}/{branch}: accepted counts do not equal the common "
                    f"registered replication request {requested}."
                )
        for branch, lower_name, upper_name in (
            (BRANCH_COUNTRY_HISTORY, "ci_ch_lower", "ci_ch_upper"),
            (BRANCH_REGION_BLOCK, "ci_rb_lower", "ci_rb_upper"),
        ):
            first_draws = np.asarray(first_result.full_draw_arrays[branch], dtype=float)
            second_draws = np.asarray(second_result.full_draw_arrays[branch], dtype=float)
            first_bounds = (float(getattr(first_result, lower_name)), float(getattr(first_result, upper_name)))
            second_bounds = (float(getattr(second_result, lower_name)), float(getattr(second_result, upper_name)))
            # Recompute every stored interval from the saved arrays before it
            # can enter the cross-seed comparison.
            first_recomputed = tuple(
                float(value) for value in np.percentile(
                    first_draws, [PERCENTILE_LOWER, PERCENTILE_UPPER], method="linear"
                )
            )
            second_recomputed = tuple(
                float(value) for value in np.percentile(
                    second_draws, [PERCENTILE_LOWER, PERCENTILE_UPPER], method="linear"
                )
            )
            if not np.allclose(first_bounds, first_recomputed, rtol=0.0, atol=IDENTITY_TOLERANCE):
                raise IngestError(f"{pair}/{branch}: primary stored interval is stale.")
            if not np.allclose(second_bounds, second_recomputed, rtol=0.0, atol=IDENTITY_TOLERANCE):
                raise IngestError(f"{pair}/{branch}: stability stored interval is stale.")
            qualitative_first = interval_conclusion(*first_bounds)
            qualitative_second = interval_conclusion(*second_bounds)
            for statistic, probability, first_value, second_value in (
                ("lower_endpoint", PERCENTILE_LOWER / 100.0, first_bounds[0], second_bounds[0]),
                ("upper_endpoint", PERCENTILE_UPPER / 100.0, first_bounds[1], second_bounds[1]),
            ):
                allowance = quantile_radius(first_draws, probability) + quantile_radius(second_draws, probability)
                sufficient = min(first_draws.size, second_draws.size) >= STABILITY_MINIMUM_DRAWS
                difference = abs(first_value - second_value)
                rows.append(
                    {
                        "policy": pair[0], "outcome": pair[1], "branch": branch,
                        "summary_statistic": statistic,
                        "primary_value": first_value, "stability_value": second_value,
                        "absolute_difference": difference,
                        "monte_carlo_allowance": allowance,
                        "confidence_level": MONTE_CARLO_CONFIDENCE,
                        "uncertainty_method": "two_sample_DKW_quantile_envelope",
                        "summary_integrity": "recomputed_from_saved_draws",
                        "primary_draws": int(first_draws.size),
                        "stability_draws": int(second_draws.size),
                        "reference_start": int(reference_window[0]),
                        "reference_end": int(reference_window[1]),
                        "primary_seed_set_id": first_seed_id,
                        "stability_seed_set_id": second_seed_id,
                        "sufficient_draws": bool(sufficient),
                        "quantitative_status": (
                            "within_monte_carlo_error" if sufficient and difference <= allowance
                            else "outside_monte_carlo_error" if sufficient
                            else "insufficient_draws"
                        ),
                        "primary_qualitative": qualitative_first,
                        "stability_qualitative": qualitative_second,
                        "qualitative_agreement": qualitative_first == qualitative_second,
                    }
                )
        # Compare finite-draw-corrected placebo tail probabilities using their
        # independent Bernoulli Monte-Carlo standard errors.
        first_draws = np.asarray(first_result.full_draw_arrays[BRANCH_PLACEBO], dtype=float)
        second_draws = np.asarray(second_result.full_draw_arrays[BRANCH_PLACEBO], dtype=float)
        if first_draws.ndim != 1 or second_draws.ndim != 1 or not np.isfinite(first_draws).all() or not np.isfinite(second_draws).all():
            raise IngestError(f"{pair}/{BRANCH_PLACEBO}: malformed placebo draws.")
        first_p = float(first_result.p_placebo_hat)
        second_p = float(second_result.p_placebo_hat)
        if (
            not np.isfinite(first_p)
            or not np.isfinite(second_p)
            or not 0.0 <= first_p <= 1.0
            or not 0.0 <= second_p <= 1.0
        ):
            raise IngestError(
                f"{pair}/{BRANCH_PLACEBO}: tail probabilities must be finite "
                "values on the closed unit interval."
            )
        se = math.sqrt(
            first_p * (1.0 - first_p) / max(first_draws.size + 1, 1)
            + second_p * (1.0 - second_p) / max(second_draws.size + 1, 1)
        )
        allowance = 1.959963984540054 * se + 1.0 / (first_draws.size + 1) + 1.0 / (second_draws.size + 1)
        sufficient = min(first_draws.size, second_draws.size) >= STABILITY_MINIMUM_DRAWS
        difference = abs(first_p - second_p)
        first_conclusion = tail_conclusion(first_p)
        second_conclusion = tail_conclusion(second_p)
        rows.append(
            {
                "policy": pair[0], "outcome": pair[1], "branch": BRANCH_PLACEBO,
                "summary_statistic": "tail_probability",
                "primary_value": first_p, "stability_value": second_p,
                "absolute_difference": difference,
                "monte_carlo_allowance": allowance,
                "confidence_level": MONTE_CARLO_CONFIDENCE,
                "uncertainty_method": "two_sample_binomial_plus_one",
                "summary_integrity": (
                    "tail_probability_prevalidated_by_placebo_table;"
                    "observed_reference_not_stored_in_ResamplingResult"
                ),
                "primary_draws": int(first_draws.size),
                "stability_draws": int(second_draws.size),
                "reference_start": int(reference_window[0]),
                "reference_end": int(reference_window[1]),
                "primary_seed_set_id": first_seed_id,
                "stability_seed_set_id": second_seed_id,
                "sufficient_draws": bool(sufficient),
                "quantitative_status": (
                    "within_monte_carlo_error" if sufficient and difference <= allowance
                    else "outside_monte_carlo_error" if sufficient
                    else "insufficient_draws"
                ),
                "primary_qualitative": first_conclusion,
                "stability_qualitative": second_conclusion,
                "qualitative_agreement": first_conclusion == second_conclusion,
            }
        )
    table = pd.DataFrame(rows).set_index(
        ["policy", "outcome", "branch", "summary_statistic"]
    ).sort_index()
    logger.info("summary stability table assembled across %d row(s).", len(table))
    # Return the table.
    return table


# Validate a supplied region mapping against the declared vocabulary.
def validate_region_mapping(region_map: Mapping[str, str]) -> None:
    """
    Validate a supplied region mapping against the declared vocabulary.

    Parameters
    ----------
    region_map : mapping of str to str
        The region of each country code.

    Returns
    -------
    None
        The function returns nothing when every label is a declared region.

    Raises
    ------
    IngestError
        If a label is an income-group label or lies outside the seven World
        Bank regions, because the vocabulary is fixed and an income group is
        not a region.
    """
    # Refuse an empty mapping.
    if not region_map:
        # Without a mapping the region branch is undefined.
        raise IngestError("The region mapping is empty.")
    # Compose the observed vocabulary.
    observed: List[str] = sorted(set(str(label) for label in region_map.values()))
    # Identify the income-group labels.
    income_labels: List[str] = [
        label
        for label in observed
        if any(token in label.lower() for token in INCOME_GROUP_TOKENS)
    ]
    # Refuse an income-group label.
    if income_labels:
        # The failure names the offending labels.
        raise IngestError(
            f"The supplied mapping carries income-group label(s) {income_labels}; "
            f"income groups are not regions."
        )
    # Identify the labels outside the declared vocabulary.
    undeclared: List[str] = [
        label for label in observed if label not in WORLD_BANK_REGIONS
    ]
    # Refuse an undeclared label.
    if undeclared:
        # The failure names the offending labels.
        raise IngestError(
            f"The supplied mapping carries undeclared label(s) {undeclared}; the "
            f"declared vocabulary is the seven World Bank regions."
        )
    # Refuse a mapping without any country.
    if not region_map:
        # The failure states the precondition.
        raise IngestError("The supplied region mapping carries no country.")
    # Log the verified vocabulary.
    logger.info(
        "supplied region mapping verified across %d region(s).", len(observed)
    )


# Compose the pair's region pool and its audit record.
def build_region_validation(
    region_map: Mapping[str, str],
    registry: PartitionRegistry,
    pairs: Sequence[Tuple[str, str]],
    results: Optional[Mapping[Tuple[str, str], ResamplingResult]] = None,
    seed_set_id: str = "",
) -> pd.DataFrame:
    """
    Compose the region validation record of the requested pairs.

    Parameters
    ----------
    region_map : mapping of str to str
        The region of each country code.
    registry : PartitionRegistry
        The partition registry.
    pairs : sequence of tuple
        The pairs whose region composition is recorded.

    Returns
    -------
    pandas.DataFrame
        One row per pair, seed set, saved attempt and region.  Each row carries
        the original pool membership, sampled block multiplicity, executed
        occurrence count, attempt status and coverage/accounting indicators.

    Raises
    ------
    IngestError
        If a country is unmapped, a pair/result is absent, or any saved draw
        fails membership, multiplicity or accepted/rejected accounting.
    """
    requested_pairs = [tuple(pair) for pair in pairs]
    if len(set(requested_pairs)) != len(requested_pairs):
        raise IngestError("Region validation pair list contains duplicates.")
    missing_partitions = sorted(set(requested_pairs) - set(registry.partitions))
    if missing_partitions:
        raise IngestError(f"Region registry omits pair(s) {missing_partitions}.")
    if results is not None and set(results) != set(requested_pairs):
        raise IngestError("Region result coverage differs from requested pair coverage.")
    rows: List[Dict[str, Any]] = []
    # Iterate over the requested pairs.
    for pair in pairs:
        # Select the pair's partition.
        partition = registry.partitions[pair]
        # Compose the pair's country pool.
        pool: List[str] = sorted(
            set(str(code) for code in partition.targets)
            | set(str(code) for code in partition.donors)
        )
        # Identify the pool countries absent from the mapping.
        missing: List[str] = [code for code in pool if code not in region_map]
        # Refuse an unmapped country.
        if missing:
            # The failure names the pair and the missing countries.
            raise IngestError(
                f"{pair}: region map omits the countr(ies) "
                f"{missing[:INFERENCE_MAX_REPORTED_OFFENDERS]}."
            )
        pair_map = {code: str(region_map[code]) for code in pool}
        validate_region_mapping(pair_map)
        # Preserve the historical pool-only mode for direct inspection while
        # making execution use the complete saved-attempt path below.
        if results is None:
            for region in sorted(set(pair_map.values())):
                rows.append(
                    {
                        "policy": pair[0], "outcome": pair[1],
                        "seed_set_id": str(seed_set_id), "attempt_index": -1,
                        "status": "not_executed", "accepted_index": pd.NA,
                        "region": region,
                        "pool_members": sum(value == region for value in pair_map.values()),
                        "block_multiplicity": pd.NA,
                        "expected_occurrences": pd.NA,
                        "observed_occurrences": pd.NA,
                        "membership_preserved": pd.NA,
                        "multiplicity_preserved": pd.NA,
                        "executed_roster_sha256": "",
                        "accepted_attempts": 0, "rejected_attempts": 0,
                        "attempt_accounting_valid": False,
                        "complete_pair_coverage": False,
                    }
                )
            continue
        result = results[pair]
        if BRANCH_REGION_BLOCK not in result.attempt_audits:
            raise IngestError(f"{pair}: result omits the region attempt ledger.")
        attempt_audit = result.attempt_audits[BRANCH_REGION_BLOCK]
        accepted = int((attempt_audit["status"] == "accepted").sum())
        rejected = int((attempt_audit["status"] == "rejected").sum())
        expected_accepted = int(result.full_draw_arrays[BRANCH_REGION_BLOCK].size)
        if accepted != expected_accepted or rejected != int(
            result.discarded_counts[BRANCH_REGION_BLOCK]
        ) or accepted + rejected != len(attempt_audit):
            raise IngestError(f"{pair}: region accepted/rejected accounting diverges.")
        replay = replay_region_draw_composition(
            attempt_audit, pair_map, len(attempt_audit)
        ).reset_index()
        for row in replay.to_dict("records"):
            row.update(
                {
                    "policy": pair[0], "outcome": pair[1],
                    "seed_set_id": str(seed_set_id),
                    "accepted_attempts": accepted,
                    "rejected_attempts": rejected,
                    "attempt_accounting_valid": True,
                    "complete_pair_coverage": True,
                }
            )
            rows.append(row)
    if not rows:
        raise IngestError("Region validation produced no rows.")
    table = pd.DataFrame(rows).set_index(
        ["policy", "outcome", "seed_set_id", "attempt_index", "region"]
    ).sort_index()
    # Log the record's extent.
    logger.info(
        "region validation recorded across %d pair-region row(s).", len(table)
    )
    # Return the validation record.
    return table


# Verify the execution against the declared study values.
def verify_inference_against_declared(
    intervals: pd.DataFrame,
    placebo: pd.DataFrame,
    expected_intervals: Optional[
        Mapping[Tuple[str, str], Mapping[str, Tuple[float, float]]]
    ] = None,
    expected_placebo: Optional[Mapping[Tuple[str, str], float]] = None,
    tolerance: float = 1.0e-3,
    stability_audit: Optional[pd.DataFrame] = None,
) -> pd.DataFrame:
    """
    Verify the execution against the declared study values.

    Parameters
    ----------
    intervals : pandas.DataFrame
        The interval table.
    placebo : pandas.DataFrame
        The placebo table.
    expected_intervals : mapping of tuple to mapping of str to tuple, optional
        The declared interval bounds per pair, keyed on the branch.
    expected_placebo : mapping of tuple to float, optional
        The declared placebo probabilities per pair.
    tolerance : float, optional
        The tolerance of the comparisons.
    diagnostics : MatchingDiagnosticsResult, optional
        An already executed, fingerprint-verified Task 23 result.  Pipeline
        callers reuse it so reporting cannot silently refit another design;
        direct callers may omit it to retain the fresh-execution interface.

    Returns
    -------
    pandas.DataFrame
        One row per declared cell recording its full scientific identity,
        comparison rule, rounding/Monte-Carlo allowance and deviation.

    Raises
    ------
    IngestError
        If an observed value diverges from its declared counterpart, or if a
        declared pair is absent from the tables.
    """
    if isinstance(tolerance, (bool, np.bool_)):
        raise IngestError(
            "Declared-comparison tolerance must be finite and nonnegative."
        )
    try:
        tolerance_value = float(tolerance)
    except (TypeError, ValueError, OverflowError) as exc:
        raise IngestError(
            "Declared-comparison tolerance must be finite and nonnegative."
        ) from exc
    if not np.isfinite(tolerance_value) or tolerance_value < 0.0:
        raise IngestError("Declared-comparison tolerance must be finite and nonnegative.")
    rows: List[Dict[str, Any]] = []

    # Compare one cell after verifying every supplied identity field against
    # the table row that produced it.
    def compare_cell(
        pair: Tuple[str, str],
        label: str,
        raw_declaration: Any,
        table: pd.DataFrame,
        branch: str,
        replication_column: str,
    ) -> None:
        if pair not in table.index:
            raise IngestError(f"Inference table omits the declared pair {pair}.")
        if label not in table.columns:
            raise IngestError(f"Inference table omits declared field {label!r}.")
        row = table.loc[pair]
        declaration = (
            dict(raw_declaration)
            if isinstance(raw_declaration, Mapping)
            else {"value": raw_declaration}
        )
        if "value" not in declaration:
            raise IngestError(f"{pair}/{label}: declaration lacks value.")
        try:
            declared = float(declaration["value"])
            observed = float(row[label])
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError(f"{pair}/{label}: declaration is nonnumeric.") from exc
        if not np.isfinite(declared) or not np.isfinite(observed):
            raise IngestError(f"{pair}/{label}: comparison value is non-finite.")
        identity_expectations: Dict[str, Any] = {
            "seed_set_id": declaration.get("seed_set_id"),
            "seed_master": declaration.get("seed_master"),
            "sample_n_targets": declaration.get("sample_n_targets"),
            "sample_n_donors": declaration.get("sample_n_donors"),
        }
        for identity, expected in identity_expectations.items():
            if expected is not None and str(row[identity]) != str(expected):
                raise IngestError(
                    f"{pair}/{label}: {identity} {row[identity]!r} disagrees "
                    f"with declared {expected!r}."
                )
        declared_branch = str(declaration.get("branch", branch))
        if declared_branch != branch:
            raise IngestError(
                f"{pair}/{label}: branch {declared_branch!r} disagrees with {branch!r}."
            )
        reference = declaration.get("reference_window")
        if reference is not None:
            try:
                reference_tuple = tuple(reference)
            except TypeError as exc:
                raise IngestError(f"{pair}/{label}: malformed reference window.") from exc
            actual_reference = (int(row["reference_start"]), int(row["reference_end"]))
            if reference_tuple != actual_reference:
                raise IngestError(
                    f"{pair}/{label}: reference {actual_reference} disagrees with "
                    f"declared {reference_tuple}."
                )
        replications = declaration.get("replications")
        actual_reps = int(row[replication_column])
        if replications is not None and (
            isinstance(replications, (bool, np.bool_))
            or not isinstance(replications, (int, np.integer))
            or int(replications) != actual_reps
        ):
            raise IngestError(
                f"{pair}/{label}: replication count {actual_reps} disagrees with "
                f"declared {replications!r}."
            )
        rule = str(declaration.get("comparison_rule", "publication_rounding"))
        if rule not in {"exact_same_seed", "publication_rounding", "cross_seed_monte_carlo"}:
            raise IngestError(f"{pair}/{label}: unknown comparison rule {rule!r}.")
        decimals = declaration.get("displayed_decimals", 3)
        if (
            isinstance(decimals, (bool, np.bool_))
            or not isinstance(decimals, (int, np.integer))
            or int(decimals) < 0
        ):
            raise IngestError(f"{pair}/{label}: displayed precision is malformed.")
        rounding_allowance = 0.5 * 10.0 ** (-int(decimals))
        numerical_allowance = declaration.get("numerical_allowance", tolerance_value)
        monte_carlo_allowance = declaration.get("monte_carlo_allowance", 0.0)
        try:
            numerical_allowance = float(numerical_allowance)
            monte_carlo_allowance = float(monte_carlo_allowance)
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError(f"{pair}/{label}: comparison allowance is malformed.") from exc
        if min(numerical_allowance, monte_carlo_allowance) < 0.0 or not np.isfinite(
            [numerical_allowance, monte_carlo_allowance]
        ).all():
            raise IngestError(f"{pair}/{label}: comparison allowance is invalid.")
        # Cross-seed uncertainty is read from the computed stability audit,
        # never chosen freely by the declaration.
        if rule == "cross_seed_monte_carlo":
            if stability_audit is None or stability_audit.empty:
                raise IngestError(
                    f"{pair}/{label}: cross-seed comparison lacks a stability audit."
                )
            statistic = (
                "tail_probability" if branch == BRANCH_PLACEBO
                else "lower_endpoint" if label.endswith("lower")
                else "upper_endpoint"
            )
            stability_key = (pair[0], pair[1], branch, statistic)
            if stability_key not in stability_audit.index:
                raise IngestError(
                    f"{pair}/{label}: stability audit omits {stability_key}."
                )
            stability_row = stability_audit.loc[stability_key]
            if not bool(stability_row.get("sufficient_draws", False)) or str(
                stability_row.get("quantitative_status", "")
            ) == "insufficient_draws":
                raise IngestError(
                    f"{pair}/{label}: cross-seed comparison has insufficient draws."
                )
            computed_mc = float(
                stability_row["monte_carlo_allowance"]
            )
            if not np.isfinite(computed_mc) or computed_mc < 0.0:
                raise IngestError(
                    f"{pair}/{label}: computed Monte-Carlo allowance is invalid."
                )
            if "monte_carlo_allowance" in declaration and not math.isclose(
                monte_carlo_allowance,
                computed_mc,
                rel_tol=0.0,
                abs_tol=IDENTITY_TOLERANCE,
            ):
                raise IngestError(
                    f"{pair}/{label}: declared Monte-Carlo allowance is not the "
                    "computed stability allowance."
                )
            monte_carlo_allowance = computed_mc
        allowance = (
            0.0
            if rule == "exact_same_seed"
            else rounding_allowance + numerical_allowance
            + (monte_carlo_allowance if rule == "cross_seed_monte_carlo" else 0.0)
        )
        deviation = abs(observed - declared)
        if (rule == "exact_same_seed" and observed != declared) or deviation > allowance:
            raise IngestError(
                f"{pair}/{label}: deviation {deviation!r} exceeds {rule} "
                f"allowance {allowance!r}."
            )
        rows.append(
            {
                "policy": pair[0], "outcome": pair[1], "reported_field": label,
                "branch": branch, "seed_set_id": str(row["seed_set_id"]),
                "seed_master": int(row["seed_master"]),
                "sample_n_targets": row["sample_n_targets"],
                "sample_n_donors": row["sample_n_donors"],
                "reference_start": int(row["reference_start"]),
                "reference_end": int(row["reference_end"]),
                "replications": actual_reps, "comparison_rule": rule,
                "displayed_decimals": int(decimals), "declared_value": declared,
                "reproduced_value": observed, "rounding_allowance": rounding_allowance,
                "monte_carlo_allowance": monte_carlo_allowance,
                "total_allowance": allowance, "absolute_deviation": deviation,
                "comparison_passed": True,
            }
        )

    # Expand legacy scalar declarations and explicit two-bound branch tuples.
    for pair, declared_fields in (expected_intervals or {}).items():
        if not isinstance(declared_fields, Mapping):
            raise IngestError(f"{pair}: interval declaration must be a mapping.")
        for label, declaration in declared_fields.items():
            if label in {BRANCH_COUNTRY_HISTORY, BRANCH_REGION_BLOCK} and isinstance(
                declaration, Sequence
            ) and not isinstance(declaration, (str, bytes, Mapping)):
                values = tuple(declaration)
                if len(values) != 2:
                    raise IngestError(f"{pair}/{label}: interval must have two bounds.")
                columns = (
                    ("ci_ch_lower", "ci_ch_upper")
                    if label == BRANCH_COUNTRY_HISTORY
                    else ("ci_rb_lower", "ci_rb_upper")
                )
                replication_column = "accepted_ch" if label == BRANCH_COUNTRY_HISTORY else "accepted_rb"
                for column, value in zip(columns, values):
                    compare_cell(pair, column, value, intervals, label, replication_column)
            else:
                branch = BRANCH_COUNTRY_HISTORY if str(label).startswith("ci_ch_") else BRANCH_REGION_BLOCK
                replication_column = "accepted_ch" if branch == BRANCH_COUNTRY_HISTORY else "accepted_rb"
                compare_cell(pair, str(label), declaration, intervals, branch, replication_column)
    for pair, declaration in (expected_placebo or {}).items():
        compare_cell(
            pair, "p_placebo_hat", declaration, placebo,
            BRANCH_PLACEBO, "valid_draws",
        )
    audit_columns = [
        "policy", "outcome", "reported_field", "branch", "seed_set_id",
        "seed_master", "sample_n_targets", "sample_n_donors", "reference_start",
        "reference_end", "replications", "comparison_rule", "displayed_decimals",
        "declared_value", "reproduced_value", "rounding_allowance",
        "monte_carlo_allowance", "total_allowance", "absolute_deviation",
        "comparison_passed",
    ]
    audit = pd.DataFrame(rows, columns=audit_columns)
    logger.info("declared execution comparison recorded %d cell(s).", len(audit))
    return audit


# Freeze the execution tables behind a canonical digest.
def freeze_execution_tables(
    intervals: pd.DataFrame,
    placebo: pd.DataFrame,
    draws: Optional[Mapping[Any, np.ndarray]] = None,
    attempt_audits: Optional[Mapping[Any, pd.DataFrame]] = None,
    seed_audit: Optional[pd.DataFrame] = None,
) -> str:
    """
    Freeze the execution tables behind a canonical digest.

    Parameters
    ----------
    intervals : pandas.DataFrame
        The interval table.
    placebo : pandas.DataFrame
        The placebo table.

    Returns
    -------
    str
        The hexadecimal digest of the canonical payload.

    Raises
    ------
    IngestError
        If either table is empty.
    """
    # Refuse an empty interval table.
    if len(intervals) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot freeze an empty interval table.")
    # Refuse an empty placebo table.
    if len(placebo) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot freeze an empty placebo table.")
    # Accumulate the canonical payload parts.
    parts: List[str] = []
    # Record every interval row in sorted key order.
    for (policy, outcome), row in intervals.sort_index().iterrows():
        # Append the row's canonical line at full precision.
        parts.append(
            "I|"
            + str(policy)
            + "|"
            + str(outcome)
            + "|"
            + "|".join(
                f"{float(row[column]):.17g}"
                for column in (
                    "ci_ch_lower",
                    "ci_ch_upper",
                    "ci_rb_lower",
                    "ci_rb_upper",
                    "p_placebo_hat",
                )
            )
        )
    # Record every placebo row in sorted key order.
    for (policy, outcome), row in placebo.sort_index().iterrows():
        # Append the row's canonical line at full precision.
        parts.append(
            "P|"
            + str(policy)
            + "|"
            + str(outcome)
            + "|"
            + f"{float(row['p_placebo_hat']):.17g}"
            + "|"
            + str(int(row["exceedances"]))
            + "|"
            + str(int(row["valid_draws"]))
        )
    # Bind accepted draws, rejection logs and the pre-draw seed registry when
    # supplied, making the registered execution reconstructible rather than a
    # digest of only its reported summaries.
    for key in sorted((draws or {}).keys(), key=str):
        parts.append(f"D|{key!r}|{_content_digest(np.asarray(draws[key], dtype=float))}")
    for key in sorted((attempt_audits or {}).keys(), key=str):
        parts.append(f"A|{key!r}|{_content_digest(attempt_audits[key])}")
    if seed_audit is not None:
        if seed_audit.empty:
            raise IngestError("Cannot freeze an empty supplied seed audit.")
        parts.append(f"S|{_content_digest(seed_audit)}")
    # Digest the encoded payload.
    return hashlib.sha256("\n".join(parts).encode("utf-8")).hexdigest()


# -------------------------------------------------------------------------------------------------------------------------------
# Task 17, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Execute the bootstrap and placebo inference across the requested pairs.
def execute_inference(
    study_config: Mapping[str, Any],
    data: Mapping[str, Any],
    seeds: Mapping[str, Any],
    region_map: Optional[Mapping[str, str]] = None,
    metadata_frame: Optional[pd.DataFrame] = None,
    pairs: Optional[Sequence[Tuple[str, str]]] = None,
    bootstrap_replicates: int = DEFAULT_BOOTSTRAP_REPLICATES,
    placebo_replicates: int = DEFAULT_PLACEBO_REPLICATES,
    stability_seeds: Optional[Mapping[str, Any]] = None,
    expected_intervals: Optional[
        Mapping[Tuple[str, str], Mapping[str, Tuple[float, float]]]
    ] = None,
    expected_placebo: Optional[Mapping[Tuple[str, str], float]] = None,
) -> InferenceExecutionResult:
    """
    Execute the bootstrap and placebo inference across the pairs.

    Parameters
    ----------
    study_config : mapping
        The baseline configuration, which may carry the seed record.
    data : mapping
        The frozen inputs under the declared keys.
    seeds : mapping
        The primary seed material carrying the master entropy.
    region_map : mapping of str to str, optional
        An already authenticated region map for direct callers. When a
        metadata frame is supplied, the map is built and validated from that
        authoritative source instead.
    metadata_frame : pandas.DataFrame, optional
        The country metadata registry, whose region column is validated and
        used when no mapping was supplied.
    pairs : sequence of tuple, optional
        The pairs to run. When absent every registered pair is run.
    bootstrap_replicates : int, optional
        The number of bootstrap replicates per branch.
    placebo_replicates : int, optional
        The number of placebo replicates.
    stability_seeds : mapping, optional
        The mandatory second registered seed set used for Monte-Carlo
        stability.  The optional annotation is retained for compatibility;
        omission fails closed before any random draw.
    expected_intervals : mapping of tuple to mapping of str to tuple, optional
        The declared interval bounds per pair.
    expected_placebo : mapping of tuple to float, optional
        The declared placebo probabilities per pair.

    Returns
    -------
    InferenceExecutionResult
        Separate primary interval/placebo tables, primary and stability draw
        arrays, every attempt ledger, summary stability, complete region
        composition, seed/publication audits, provenance and frozen digest.

    Raises
    ------
    KeyError
        If the data mapping lacks a declared key.
    IngestError
        If the seed record, the region vocabulary, a draw, a table or a
        comparison check fails.
    """
    # Refuse a data mapping lacking a declared key.
    for key in ("registry", "panel", "lagged"):
        # Test the key's presence.
        if key not in data:
            # The failure names the absent key.
            raise KeyError(f"The data mapping lacks the declared key {key!r}.")
    # Select and validate the frozen registry before deriving any run inputs.
    registry = data["registry"]
    if not isinstance(registry, PartitionRegistry):
        raise IngestError("The data registry is not a PartitionRegistry.")
    mandatory_pairs = set(tuple(pair) for pair in DECLARED_PAIRS)
    selected = [
        tuple(pair) for pair in (DECLARED_PAIRS if pairs is None else pairs)
    ]
    if set(selected) != mandatory_pairs or len(selected) != len(mandatory_pairs):
        raise IngestError(
            "Task 17 requires exactly all eight declared policy/outcome pairs."
        )
    if set(registry.partitions) != mandatory_pairs:
        missing = sorted(mandatory_pairs - set(registry.partitions))
        extra = sorted(set(registry.partitions) - mandatory_pairs)
        raise IngestError(
            f"Inference registry coverage differs from the frozen eight-pair "
            f"design; missing={missing}, extra={extra}."
        )
    selected = sorted(mandatory_pairs)
    # Both seed sets and all branch identifiers are gated and persisted before
    # the first baseline evaluation or random draw occurs.
    if stability_seeds is None:
        raise IngestError("Task 17 requires a registered stability seed set.")
    primary_seed_audit = _resolve_seed_registration(study_config, seeds)
    stability_seed_audit = _resolve_seed_registration(study_config, stability_seeds)
    if str(seeds["seed_set_id"]) == str(stability_seeds["seed_set_id"]):
        raise IngestError("Primary and stability runs must use distinct seed-set ids.")
    if int(seeds["master"]) == int(stability_seeds["master"]):
        raise IngestError("Primary and stability runs must use independent master seeds.")
    seed_audit = pd.concat(
        [primary_seed_audit, stability_seed_audit], axis=0
    ).sort_index()
    # Require complete authoritative region metadata coverage for every
    # eligible original country used by any mandatory partition.
    eligible_countries = sorted(
        {
            str(code)
            for pair in selected
            for code in (
                tuple(registry.partitions[pair].targets)
                + tuple(registry.partitions[pair].donors)
            )
        }
    )
    if metadata_frame is not None:
        validated_regions: Mapping[str, str] = build_region_map(
            metadata_frame, required_countries=eligible_countries
        )
    elif isinstance(region_map, AuthoritativeRegionMap):
        validated_regions = region_map
        missing_regions = sorted(set(eligible_countries) - set(validated_regions))
        if missing_regions:
            raise IngestError(
                f"Authoritative region map omits {missing_regions[:INFERENCE_MAX_REPORTED_OFFENDERS]}."
            )
    else:
        raise IngestError(
            "Task 17 requires metadata-derived regions with source/version identity."
        )
    validate_region_mapping(validated_regions)
    region_provenance = dict(validated_regions.provenance)  # type: ignore[attr-defined]
    # Resolve the effective reference window shared by observed and resampled
    # estimators so table provenance cannot silently cite the module default.
    effective_windows, _ = resolve_reporting_windows(study_config)
    reference_window = tuple(effective_windows["pre_escalation"])
    # Compose the observed statistic of every pair independently of the
    # resampling, which the placebo audit compares against.
    observed_by_pair: Dict[Tuple[str, str], float] = compute_observed_statistics(
        study_config, data, selected
    )
    # Run the primary seed set across the selected pairs.
    primary: Dict[Tuple[str, str], ResamplingResult] = run_seed_set(
        study_config,
        data,
        seeds,
        validated_regions,
        selected,
        bootstrap_replicates,
        placebo_replicates,
    )
    # Assemble the interval table.
    intervals = build_interval_table(
        primary,
        seed_set_id=str(seeds["seed_set_id"]),
        reference_window=reference_window,
        registry=registry,
    )
    # Assemble the placebo table with its recomputation audit.
    placebo = build_placebo_table(
        primary,
        observed_by_pair,
        seed_set_id=str(seeds["seed_set_id"]),
        reference_window=reference_window,
        registry=registry,
    )
    # Assert that no marker derives from the placebo probabilities.
    for pair in sorted(primary.keys()):
        # Verify the pair's isolation.
        assert_no_marker_from_placebo(primary[pair])
    # Compose the draw mapping keyed on the pair and the branch.
    # Run the independently registered stability seed set over the identical
    # eight-pair scientific design.
    second = run_seed_set(
        study_config,
        data,
        stability_seeds,
        validated_regions,
        selected,
        bootstrap_replicates,
        placebo_replicates,
    )
    stability = build_stability_table(
        primary, second, reference_window=reference_window
    )
    # Retain immutable copies of every accepted draw array from both runs and
    # every accepted/rejected attempt ledger from all three branches.
    draws: Dict[Tuple[Tuple[str, str], str], np.ndarray] = {}
    stability_draws: Dict[Tuple[Tuple[str, str], str], np.ndarray] = {}
    attempt_audits: Dict[Tuple[str, Tuple[str, str], str], pd.DataFrame] = {}
    for seed_id, results, target in (
        (str(seeds["seed_set_id"]), primary, draws),
        (str(stability_seeds["seed_set_id"]), second, stability_draws),
    ):
        for pair in selected:
            for branch in BRANCH_ORDER:
                array = np.array(results[pair].full_draw_arrays[branch], dtype=float, copy=True)
                array.setflags(write=False)
                target[(pair, branch)] = array
                attempt_audits[(seed_id, pair, branch)] = results[pair].attempt_audits[branch].copy(deep=True)
    # Replay and verify every actual saved region attempt for both seed sets.
    primary_regions = build_region_validation(
        validated_regions, registry, selected, primary, str(seeds["seed_set_id"])
    )
    stability_regions = build_region_validation(
        validated_regions, registry, selected, second,
        str(stability_seeds["seed_set_id"]),
    )
    region_validation = pd.concat(
        [primary_regions, stability_regions], axis=0
    ).sort_index()
    composition_audit = region_validation.copy(deep=True)
    # Verify the tables against the declared values when they were supplied.
    publication_audit = verify_inference_against_declared(
        intervals,
        placebo,
        expected_intervals=expected_intervals,
        expected_placebo=expected_placebo,
        stability_audit=stability,
    )
    # Fingerprint the execution tables.
    all_draws: Dict[Any, np.ndarray] = {
        (str(seeds["seed_set_id"]),) + key: value for key, value in draws.items()
    }
    all_draws.update(
        {
            (str(stability_seeds["seed_set_id"]),) + key: value
            for key, value in stability_draws.items()
        }
    )
    fingerprint = freeze_execution_tables(
        intervals, placebo, all_draws, attempt_audits, seed_audit
    )
    # Accumulate the recorded notes.
    notes: List[str] = [
        "every draw refits the partition inside the resampled design through "
        "the frozen baseline interface, so no baseline parameter is reused.",
        "the region branch preserves region-level exchangeability and does not "
        "preserve the exact country count; a drawn region may appear several "
        "times and another may be omitted.",
        "placebo probabilities are diagnostic quantities and never assign a "
        "significance marker.",
    ]
    notes.append(
        "cross-seed stability compares interval endpoints under distribution-free "
        "quantile envelopes and placebo tails under plus-one binomial Monte-Carlo error."
    )
    # Record the coverage.
    notes.append(f"the execution covered {len(selected)} pair(s).")
    # Assemble the immutable result.
    result: InferenceExecutionResult = InferenceExecutionResult(
        intervals=intervals,
        placebo=placebo,
        draws=draws,
        stability=stability,
        composition_audit=composition_audit,
        region_validation=region_validation,
        table_fingerprint_sha256=fingerprint,
        completed_utc=current_utc_iso(),
        notes=tuple(notes),
        stability_draws=stability_draws,
        attempt_audits=attempt_audits,
        seed_audit=seed_audit.copy(deep=True),
        publication_audit=publication_audit.copy(deep=True),
        region_provenance=region_provenance,
    )
    # Log the completed execution with its digest prefix.
    logger.info(
        "bootstrap and placebo execution complete: %d pair(s), fingerprint %s.",
        len(intervals),
        fingerprint[:16],
    )
    # Return the assembled result.
    return result
````

````python
# Task 18: Orchestrator 3 - Robustness Sweep Factory

# ==============================================================================
# Task 18: Orchestrator 3 - Robustness Sweep Factory
# ==============================================================================

# Declare the axis label of the reference-window sweep.
AXIS_REFERENCE_WINDOW: Final[str] = "reference_window"
# Declare the axis label of the donor-pool sweep.
AXIS_DONOR_POOL: Final[str] = "donor_pool"
# Declare the axis label of the factor-proxy sweep.
AXIS_FACTOR_PROXY: Final[str] = "factor_proxy"
# Declare the axis label of the horizon sweep.
AXIS_HORIZON: Final[str] = "horizon"
# Declare the axis label of the active-year sweep.
AXIS_ACTIVE_YEAR: Final[str] = "active_year"
# Compose the declared axis order of the registry.
AXIS_ORDER: Final[Tuple[str, ...]] = (
    AXIS_REFERENCE_WINDOW,
    AXIS_DONOR_POOL,
    AXIS_FACTOR_PROXY,
    AXIS_HORIZON,
    AXIS_ACTIVE_YEAR,
)
# Declare the reference-window variants of the study.
REFERENCE_WINDOW_VARIANTS: Final[Tuple[Tuple[str, Tuple[int, int]], ...]] = (
    ("preferred_2019_2021", (2019, 2021)),
    ("alternative_2017_2019", (2017, 2019)),
    ("alternative_2021_2021", (2021, 2021)),
    ("alternative_2017_2021", (2017, 2021)),
    ("alternative_2018_2021", (2018, 2021)),
)
# Declare the donor-pool variants of the study.
DONOR_POOL_VARIANTS: Final[Tuple[str, ...]] = (
    "corrected_full_pool",
    "no_repaired_main_outcome_cells",
    "high_income_donors",
)
# Declare the factor-proxy variants of the study.
FACTOR_PROXY_VARIANTS: Final[Tuple[str, ...]] = (
    "own_outcome_donor_mean",
    "other_outcome_donor_mean",
    "both_outcome_means",
    "own_outcome_plus_oil_rents",
    "own_outcome_plus_fuel_share_balance",
)
# Declare the horizon windows the study enumerates, of which the full window
# is the baseline configuration itself.
HORIZON_WINDOWS: Final[Tuple[Tuple[int, int], ...]] = ((2022, 2024), (2022, 2023))
# Declare the horizon sweep variants of the study, which are the windows that
# deviate from the baseline; the declared total of the registry counts one
# horizon variant, because the full window is the baseline rather than a sweep.
HORIZON_VARIANTS: Final[Tuple[Tuple[str, Tuple[int, int]], ...]] = (
    ("short_2022_2023", (2022, 2023)),
)
# Declare the active-year variant of the study.
ACTIVE_YEAR_VARIANTS: Final[Tuple[str, ...]] = ("per_country_entry_year",)
# Declare the factor component names of the two-column baseline variant.
FACTOR_COMPONENTS_TWO: Final[Tuple[str, str]] = ("constant", "donor_csa")
# Declare the factor component names of the three-column variants.
FACTOR_COMPONENTS_THREE: Final[Tuple[str, str, str]] = (
    "constant",
    "donor_csa",
    "augmenting_mean",
)
# Declare the configuration path of the preferred reference window.
PATH_REFERENCE_WINDOW: Final[Tuple[str, ...]] = (
    "estimation",
    "preferred_reference_window",
)
# Declare the configuration path of the donor-pool variant.
PATH_DONOR_POOL: Final[Tuple[str, ...]] = ("estimation", "donor_pool_variant")
# Declare the configuration path of the factor-proxy variant.
PATH_FACTOR_PROXY: Final[Tuple[str, ...]] = ("estimation", "factor_proxy_variant")
# Declare the configuration path of the horizon window.
PATH_HORIZON: Final[Tuple[str, ...]] = ("estimation", "windows", WAVE_TWO_LABEL)
# Declare the outcome used as the augmenting mean of the oil-rent variant.
AUGMENTING_OIL_RENTS: Final[str] = "OilRents"
# Declare the outcome used as the augmenting mean of the fuel-share variant.
AUGMENTING_FUEL_BALANCE: Final[str] = "FuelShareBalance"
# Declare the alternative level outcome of the other-outcome variant.
OTHER_LEVEL_OUTCOME: Final[str] = "log_real_GDPpc"
# Declare the primary level outcome whose alternative selection is declared.
PRIMARY_LEVEL_OUTCOME: Final[str] = "log_CPI"
# Declare the stable serialization version of the sweep registry.
SWEEP_REGISTRY_VERSION: Final[str] = "task18-v2"
# Declare the variants that intentionally reproduce one baseline component;
# every other registered variant must change its effective design.
SWEEP_BASELINE_IDENTITIES: Final[Mapping[str, str]] = MappingProxyType(
    {
        AXIS_REFERENCE_WINDOW: "preferred_2019_2021",
        AXIS_DONOR_POOL: "corrected_full_pool",
        AXIS_FACTOR_PROXY: "own_outcome_donor_mean",
    }
)


def _sweep_estimation_view(study_config: Mapping[str, Any]) -> Dict[str, Any]:
    """Materialize a private estimation view from authenticated study windows."""
    if not isinstance(study_config, Mapping):
        raise IngestError("Baseline sweep configuration must be a mapping.")
    view = copy.deepcopy(dict(study_config))
    resolved, _ = resolve_reporting_windows(study_config)
    constants = study_config.get("global_constants")
    if not isinstance(constants, Mapping):
        raise SweepUnresolvedInputError("Sweep configuration lacks global_constants.")

    def validate_declaration(raw: Any, label: str) -> Tuple[int, int]:
        """Normalize portable configuration sequences before ATT validation."""
        # YAML and JSON represent calendar windows as lists, while Python may
        # supply tuples. Reject text, mappings and scalars at this boundary.
        if not isinstance(raw, (list, tuple)):
            raise IngestError(f"{label} must be a two-element calendar sequence.")
        # Keep the existing length, Boolean, finite, integral and order checks.
        return _validate_att_window(tuple(raw), label)

    declarations = {
        "preferred_reference_window": "pre_escalation",
        "Wave1": WAVE_ONE_LABEL,
        "Wave2": WAVE_TWO_LABEL,
    }
    for declared, effective in declarations.items():
        if declared not in constants:
            raise SweepUnresolvedInputError(
                f"Sweep configuration lacks global_constants.{declared}."
            )
        actual = validate_declaration(
            constants[declared], f"global_constants.{declared}"
        )
        if tuple(actual) != tuple(resolved[effective]):
            raise IngestError(
                f"global_constants.{declared} disagrees with effective {effective}."
            )
    alternatives = constants.get("alternative_reference_windows")
    if not isinstance(alternatives, (list, tuple)):
        raise SweepUnresolvedInputError(
            "Sweep configuration lacks global_constants.alternative_reference_windows."
        )
    actual_alternatives = tuple(
        validate_declaration(raw, "global_constants.alternative_reference_windows")
        for raw in alternatives
    )
    declared_alternatives = tuple(window for _, window in REFERENCE_WINDOW_VARIANTS[1:])
    if actual_alternatives != declared_alternatives:
        raise IngestError(
            "Authoritative alternative reference windows disagree with the sweep registry."
        )
    estimation = view.setdefault("estimation", {})
    if not isinstance(estimation, dict):
        raise IngestError("study_config.estimation must be a mapping.")
    windows = estimation.setdefault("windows", {})
    if not isinstance(windows, dict):
        raise IngestError("study_config.estimation.windows must be a mapping.")
    for name in (WAVE_ONE_LABEL, WAVE_TWO_LABEL, "pre_escalation", "post_escalation"):
        windows.setdefault(name, list(resolved[name]))
    estimation.setdefault("preferred_reference_window", list(resolved["pre_escalation"]))
    # The original configuration remains the authority and retains its digest.
    return view


# Distinguish a scientifically unresolved prerequisite from a coding failure.
class SweepUnresolvedInputError(IngestError):
    """A registered sweep cannot run because required metadata are absent."""


# Declare the frozen sweep specification.
@dataclass(frozen=True)
class SweepSpec:
    """
    Frozen specification of one sweep variant.

    Attributes
    ----------
    axis : str
        The swept axis, one of the five declared axes.
    variant_name : str
        The variant's name within its axis.
    variant_value : Any
        The variant's value, which is a window pair for the window axes, a
        variant label for the donor and proxy axes, and a callable for the
        active-year axis.
    """
    # The swept axis.
    axis: str
    # The variant's name within its axis.
    variant_name: str
    # The variant's value.
    variant_value: Any


# Declare the frozen result of one swept run.
@dataclass(frozen=True)
class SweepResult:
    """
    Frozen result of one swept run.

    Attributes
    ----------
    policy_name : str
        The policy instrument of the run.
    outcome_name : str
        The canonical outcome variable of the run.
    axis : str
        The swept axis.
    variant_name : str
        The variant's name.
    att : float
        The swept incremental effect.
    se_cd : float
        The country-dispersion standard error of the swept effect.
    stars : str
        The significance marker of the swept effect.
    diagnostic_notes : tuple of str
        The declared diagnostic notes, including the factor condition number
        when the factor proxy changes.
    run_id : str
        Stable unique identifier of the registered run.
    effective_design : mapping
        Complete resolved scientific design and invariant evidence.
    lineage_sha256 : str
        Digest binding the run to its baseline parent and effective design.
    execution_mode : str
        ``refit``, ``reaggregate_existing_gaps`` or ``reuse_verified_baseline``.
    """
    # The policy instrument of the run.
    policy_name: str
    # The canonical outcome variable of the run.
    outcome_name: str
    # The swept axis.
    axis: str
    # The variant's name.
    variant_name: str
    # The swept incremental effect.
    att: float
    # The country-dispersion standard error of the swept effect.
    se_cd: float
    # The significance marker of the swept effect.
    stars: str
    # The declared diagnostic notes.
    diagnostic_notes: Tuple[str, ...]
    # Stable unique identifier of the run.
    run_id: str = ""
    # Complete effective design and invariant evidence.
    effective_design: Mapping[str, Any] = field(default_factory=dict)
    # Parent/design/result lineage digest.
    lineage_sha256: str = ""
    # Whether estimation was refit or existing gaps were reaggregated.
    execution_mode: str = ""
    # A verified nonbaseline design identity is distinct from an estimate.
    scientific_status: str = "success"


# Declare the frozen result of the sweep suite.
@dataclass(frozen=True)
class SweepSuiteResult:
    """
    Frozen result of the robustness sweep suite.

    Attributes
    ----------
    registry : pandas.DataFrame
        The frozen registry, one row per declared sweep run.
    results : pandas.DataFrame
        The executed swept results keyed on the pair, the axis and the
        variant name.
    registry_fingerprint_sha256 : str
        Digest of the canonical registry payload.
    completed_utc : str
        ISO-8601 UTC timestamp of the suite.
    notes : tuple of str
        Recorded decisions, including the executed and registered counts.
    failures : pandas.DataFrame
        Every unresolved or failed registered run with its localized reason.
    effective_designs : pandas.DataFrame
        One row per successful run carrying lineage and design evidence.
    complete : bool
        Whether every selected registered run completed successfully.
    """
    # The frozen sweep registry.
    registry: pd.DataFrame
    # The executed swept results.
    results: pd.DataFrame
    # Digest of the canonical registry payload.
    registry_fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the suite.
    completed_utc: str
    # Recorded decisions of the suite.
    notes: Tuple[str, ...]
    # Unresolved and failed selected runs.
    failures: pd.DataFrame = field(default_factory=pd.DataFrame)
    # Successful run design/lineage evidence.
    effective_designs: pd.DataFrame = field(default_factory=pd.DataFrame)
    # Exact suite-completeness state.
    complete: bool = False


# -------------------------------------------------------------------------------------------------------------------------------
# Task 18, Step 1: The configuration diff and the specification ordering
# -------------------------------------------------------------------------------------------------------------------------------


# Walk two nested structures and report the paths that differ.
def diff_config_paths(
    baseline: Any, modified: Any, prefix: Tuple[str, ...] = ()
) -> List[Tuple[str, ...]]:
    """
    Walk two nested structures and report the paths that differ.

    Parameters
    ----------
    baseline : Any
        The baseline configuration node.
    modified : Any
        The modified configuration node.
    prefix : tuple of str, optional
        The path prefix accumulated by the recursion.

    Returns
    -------
    list of tuple of str
        The full paths at which the two structures differ, where a path is
        the sequence of mapping keys from the root.

    Raises
    ------
    IngestError
        If the two structures have incompatible types at a path.
    """
    # Compare two mappings key by key.
    if isinstance(baseline, Mapping) and isinstance(modified, Mapping):
        # Accumulate the differing paths of the two mappings.
        paths: List[Tuple[str, ...]] = []
        # Compose the union of the two key sets.
        keys = sorted(set(baseline.keys()) | set(modified.keys()), key=str)
        # Compare each key's subtree.
        for key in keys:
            # Compose the path of the key.
            path: Tuple[str, ...] = prefix + (str(key),)
            # Record a key present in only one mapping.
            if key not in baseline or key not in modified:
                # Append the path of the missing key.
                paths.append(path)
                # Continue to the next key.
                continue
            # Recurse into the key's subtree.
            paths.extend(diff_config_paths(baseline[key], modified[key], path))
        # Return the accumulated paths.
        return paths
    # Compare two sequences element by element.
    if isinstance(baseline, (list, tuple)) and isinstance(modified, (list, tuple)):
        # Return the path itself when the sequences differ in length or content.
        return [] if list(baseline) == list(modified) else [prefix]
    # Refuse a type mismatch at a shared path.
    if type(baseline) is not type(modified):
        # The failure names the path and both types.
        raise IngestError(
            f"Configuration node {'.'.join(prefix)} carries incompatible types "
            f"{type(baseline).__name__} and {type(modified).__name__}."
        )
    # Return the path when the scalars differ, and nothing when they agree.
    return [] if baseline == modified else [prefix]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 18, Step 2: The axis appliers and the variant constructions
# -------------------------------------------------------------------------------------------------------------------------------


# Apply one sweep variant to a deep copy of the baseline configuration.
def apply_sweep_variant(
    baseline_config: Mapping[str, Any], spec: SweepSpec
) -> Tuple[Dict[str, Any], List[Tuple[str, ...]]]:
    """
    Apply one sweep variant to a deep copy of the baseline configuration.

    Parameters
    ----------
    baseline_config : mapping
        The baseline configuration, which is never mutated.
    spec : SweepSpec
        The sweep specification.

    Returns
    -------
    tuple
        The private modified estimation view and its differing view paths.
        A sweep changes exactly one effective component; synchronized
        canonical/legacy aliases may produce two view paths.

    Raises
    ------
    IngestError
        If the axis/name/value is undeclared or mismatched, required metadata
        are unresolved, or the intended effective component does not change.
    """
    if not isinstance(baseline_config, Mapping):
        raise IngestError("Baseline sweep configuration must be a mapping.")
    if spec.axis not in AXIS_ORDER:
        raise IngestError(f"Undeclared sweep axis {spec.axis!r}.")
    baseline_view = _sweep_estimation_view(baseline_config)
    modified = copy.deepcopy(baseline_view)
    estimation = modified.get("estimation")
    if not isinstance(estimation, dict):
        raise SweepUnresolvedInputError(
            "Sweep configuration lacks a mutable estimation mapping."
        )
    expected_values: Mapping[str, Any]
    if spec.axis == AXIS_REFERENCE_WINDOW:
        expected_values = dict(REFERENCE_WINDOW_VARIANTS)
    elif spec.axis == AXIS_DONOR_POOL:
        expected_values = {name: name for name in DONOR_POOL_VARIANTS}
    elif spec.axis == AXIS_FACTOR_PROXY:
        expected_values = {name: name for name in FACTOR_PROXY_VARIANTS}
    elif spec.axis == AXIS_HORIZON:
        expected_values = dict(HORIZON_VARIANTS)
    else:
        expected_values = {name: name for name in ACTIVE_YEAR_VARIANTS}
    if spec.variant_name not in expected_values:
        raise IngestError(
            f"{spec.axis}: undeclared variant {spec.variant_name!r}."
        )
    expected = expected_values[spec.variant_name]
    if spec.axis in {AXIS_REFERENCE_WINDOW, AXIS_HORIZON}:
        if isinstance(spec.variant_value, (str, bytes, Mapping)):
            raise IngestError(f"{spec.axis}: variant value must be a calendar pair.")
        try:
            supplied = tuple(spec.variant_value)
        except TypeError as exc:
            raise IngestError(f"{spec.axis}: variant value must be a calendar pair.") from exc
        if (
            len(supplied) != 2
            or any(
                isinstance(value, (bool, np.bool_))
                or not isinstance(value, (int, np.integer))
                for value in supplied
            )
            or tuple(int(value) for value in supplied) != tuple(expected)
        ):
            raise IngestError(
                f"{spec.axis}/{spec.variant_name}: label/value mismatch."
            )
    elif spec.axis == AXIS_ACTIVE_YEAR:
        if not callable(spec.variant_value):
            raise SweepUnresolvedInputError(
                "active_year requires its registered callable before execution."
            )
    elif str(spec.variant_value) != str(expected):
        raise IngestError(f"{spec.axis}/{spec.variant_name}: label/value mismatch.")
    # Apply the effective field in the private estimation view, synchronizing
    # its two reference-window aliases.
    if spec.axis == AXIS_REFERENCE_WINDOW:
        windows = estimation.get("windows")
        has_legacy = "preferred_reference_window" in estimation
        has_canonical = isinstance(windows, dict) and "pre_escalation" in windows
        if not (has_legacy or has_canonical):
            raise SweepUnresolvedInputError(
                "reference sweep lacks an authoritative reference declaration."
            )
        if has_legacy:
            estimation["preferred_reference_window"] = list(expected)
        if has_canonical:
            windows["pre_escalation"] = list(expected)
    elif spec.axis == AXIS_HORIZON:
        windows = estimation.get("windows")
        if not isinstance(windows, dict) or WAVE_TWO_LABEL not in windows:
            raise SweepUnresolvedInputError(
                "horizon sweep lacks estimation.windows.wave_2_level."
            )
        windows[WAVE_TWO_LABEL] = list(expected)
        if "post_escalation" in windows:
            windows["post_escalation"] = list(expected)
    elif spec.axis == AXIS_DONOR_POOL:
        estimation["donor_pool_variant"] = str(expected)
    elif spec.axis == AXIS_FACTOR_PROXY:
        estimation["factor_proxy_variant"] = str(expected)
    else:
        estimation["active_year_variant"] = str(spec.variant_name)
    paths = diff_config_paths(baseline_view, modified)
    identity_run = SWEEP_BASELINE_IDENTITIES.get(spec.axis) == spec.variant_name
    if not paths and not identity_run:
        raise IngestError(
            f"{spec.axis}/{spec.variant_name}: named nonbaseline variant has no "
            "effective configuration change."
        )
    # Resolve window inputs before/after and require only the intended coupled
    # scientific component to differ.
    if spec.axis in {AXIS_REFERENCE_WINDOW, AXIS_HORIZON}:
        before, _ = resolve_reporting_windows(baseline_config)
        after, _ = resolve_reporting_windows(modified)
        changed = {
            key for key in (WAVE_ONE_LABEL, WAVE_TWO_LABEL, "pre_escalation", "post_escalation")
            if tuple(before[key]) != tuple(after[key])
        }
        permitted = (
            {"pre_escalation"}
            if spec.axis == AXIS_REFERENCE_WINDOW
            else {WAVE_TWO_LABEL, "post_escalation"}
        )
        if changed and changed != permitted:
            raise IngestError(
                f"{spec.axis}/{spec.variant_name}: effective window changes "
                f"{sorted(changed)} instead of {sorted(permitted)}."
            )
        if not changed and not identity_run:
            raise IngestError(
                f"{spec.axis}/{spec.variant_name}: effective windows are unchanged."
            )
    logger.info(
        "sweep variant applied: axis=%s variant=%s %s.",
        spec.axis,
        spec.variant_name,
        "registered baseline identity"
        if identity_run
        else f"paths={['.'.join(path) for path in paths]}",
    )
    # Return the modified configuration and the differing paths, which is one
    # path for a deviating variant and none for the identity re-run.
    return modified, paths


# Resolve the active-year window map from the functional specification.
def resolve_active_year_windows(
    spec: SweepSpec,
    panel: pd.DataFrame,
    countries: Sequence[str],
    entry_years: Mapping[str, int],
) -> Dict[str, Tuple[int, int]]:
    """
    Resolve the active-year window map from the functional specification.

    Parameters
    ----------
    spec : SweepSpec
        The active-year specification, whose value is a callable that maps a
        country's entry year to its window.
    panel : pandas.DataFrame
        The analysis panel, used when the callable requests observed data.
    countries : sequence of str
        The countries whose windows are resolved.
    entry_years : mapping of str to int
        The entry year of each country, which is the argument the callable
        consumes.

    Returns
    -------
    dict
        The second-wave window of each country, keyed on the country code.

    Raises
    ------
    IngestError
        If the specification's value is not callable, or if a country carries
        no entry year, because the axis is defined as a per-country
        modification of the second wave.
    """
    # Refuse a specification whose value is not callable, because the axis is
    # defined functionally rather than as a fixed window.
    if not callable(spec.variant_value):
        # The failure names the axis.
        raise IngestError(
            f"{AXIS_ACTIVE_YEAR}: the variant value must be a callable mapping an "
            f"entry year to a window."
        )
    # Initialise the window mapping.
    windows: Dict[str, Tuple[int, int]] = {}
    # Resolve each country's window through the callable.
    for country in sorted(str(code) for code in countries):
        # Refuse a country without an entry year.
        if country not in entry_years:
            # The failure names the country.
            raise IngestError(
                f"{AXIS_ACTIVE_YEAR}: country {country} carries no entry year."
            )
        entry = entry_years[country]
        if (
            isinstance(entry, (bool, np.bool_))
            or not isinstance(entry, (int, np.integer))
        ):
            raise IngestError(f"{AXIS_ACTIVE_YEAR}: {country} entry year is malformed.")
        # Compose the country's window from the callable.
        window = spec.variant_value(int(entry))
        # Refuse a malformed window.
        if not isinstance(window, (list, tuple)) or len(window) != 2:
            # The failure names the returned value.
            raise IngestError(
                f"{AXIS_ACTIVE_YEAR}: the callable returned {window!r} for "
                f"{country}, which is not a window pair."
            )
        if any(
            isinstance(value, (bool, np.bool_))
            or not isinstance(value, (int, np.integer))
            for value in window
        ):
            raise IngestError(
                f"{AXIS_ACTIVE_YEAR}: callable returned non-integral bounds for {country}."
            )
        span = (int(window[0]), int(window[1]))
        if span[0] > span[1]:
            raise IngestError(f"{AXIS_ACTIVE_YEAR}: reversed window {span} for {country}.")
        if span[0] < WAVE_TWO_WINDOW[0] or span[1] > WAVE_TWO_WINDOW[1]:
            raise IngestError(
                f"{AXIS_ACTIVE_YEAR}: window {span} for {country} leaves the "
                f"intended Wave 2 span {WAVE_TWO_WINDOW}."
            )
        windows[country] = span
    # Log the resolved windows.
    logger.info(
        "%s: per-country windows resolved across %d countr(ies).",
        AXIS_ACTIVE_YEAR,
        len(windows),
    )
    # Return the mapping.
    return windows


# Build the variant proxy series of one partition.
def build_variant_proxy(
    frame: pd.DataFrame,
    donors: Sequence[str],
    outcome: str,
    variant: str,
    window: Tuple[int, int] = FACTOR_WINDOW,
) -> pd.DataFrame:
    """
    Build the variant proxy columns of one partition.

    Parameters
    ----------
    frame : pandas.DataFrame
        The analysis panel carrying the outcome columns.
    donors : sequence of str
        The eligible donor pool of the partition.
    outcome : str
        The canonical outcome of the pair.
    variant : str
        The factor-proxy variant name.
    window : tuple of int, optional
        The inclusive factor window.

    Returns
    -------
    pandas.DataFrame
        The proxy columns indexed by year, whose column count is two for the
        single-mean variants and three for the augmented variants, with the
        first column always the constant.

    Raises
    ------
    IngestError
        If the variant is undeclared, if a required column is absent, or if
        the donor pool is empty.
    """
    # Refuse an empty donor pool.
    if not donors:
        # Without donors the proxy is undefined.
        raise IngestError("Cannot build a variant proxy from an empty pool.")
    # Restrict the frame to the window and the donor pool.
    scoped = frame[
        (frame["year"] >= int(window[0]))
        & (frame["year"] <= int(window[1]))
        & (frame["country_code"].astype(str).isin(set(str(code) for code in donors)))
    ]
    # Compose the ascending year axis of the window.
    years: List[int] = list(range(int(window[0]), int(window[1]) + 1))
    # Compose the constant column of the proxy.
    constant: pd.Series = pd.Series(1.0, index=years, name="constant")
    # Compose the donor mean of one frame column.
    def donor_mean(column: str) -> pd.Series:
        """
        Compose the cross-sectional donor mean of one column.

        Parameters
        ----------
        column : str
            The frame column to average over the donor cross-section.

        Returns
        -------
        pandas.Series
            The cross-sectional donor mean per year, indexed by year and named after the column.
        """
        # Refuse a column the frame does not carry.
        if column not in scoped.columns:
            # The failure names the absent column.
            raise IngestError(f"Panel lacks the column {column!r} for the proxy.")
        duplicate_keys = scoped.duplicated(
            subset=["country_code", "year"], keep=False
        )
        if bool(duplicate_keys.any()):
            raise IngestError(
                f"Donor frame for {column!r} duplicates country-year keys."
            )
        # Pivot the column to a year-by-donor frame.
        grid = scoped.pivot_table(
            index="year", columns="country_code", values=column, aggfunc="last"
        )
        # Reindex to the declared axes so an absent cell becomes visible.
        grid = grid.reindex(index=years, columns=[str(code) for code in donors])
        # Refuse a missing donor cell, which would shrink the cross-section.
        if bool(grid.isna().to_numpy().any()):
            # The failure names the column.
            raise IngestError(
                f"Donor frame for {column!r} carries a missing cell; the proxy "
                f"must be averaged over a complete cross-section."
            )
        # Return the cross-sectional mean per year.
        mean = grid.mean(axis=1)
        # Name the series after the column.
        mean.name = column
        # Return the mean.
        return mean
    # Build the single-mean variants.
    if variant == "own_outcome_donor_mean":
        # Compose the baseline factor matrix of the constant and the own mean.
        return pd.DataFrame({"constant": constant, "donor_csa": donor_mean(outcome)})
    other_outcome = (
        OTHER_LEVEL_OUTCOME if outcome == PRIMARY_LEVEL_OUTCOME else PRIMARY_LEVEL_OUTCOME
    )
    if variant == "other_outcome_donor_mean":
        # Compose the constant and the alternative level outcome's mean.
        return pd.DataFrame(
            {"constant": constant, "donor_csa": donor_mean(other_outcome)}
        )
    if variant == "both_outcome_means":
        # Compose the constant and both level outcomes' means.
        return pd.DataFrame(
            {
                "constant": constant,
                "donor_csa": donor_mean(outcome),
                "augmenting_mean": donor_mean(other_outcome),
            }
        )
    if variant == "own_outcome_plus_oil_rents":
        # Compose the constant, the own mean and the oil-rent mean.
        return pd.DataFrame(
            {
                "constant": constant,
                "donor_csa": donor_mean(outcome),
                "augmenting_mean": donor_mean(AUGMENTING_OIL_RENTS),
            }
        )
    if variant == "own_outcome_plus_fuel_share_balance":
        # Compose the constant, the own mean and the fuel-balance mean.
        return pd.DataFrame(
            {
                "constant": constant,
                "donor_csa": donor_mean(outcome),
                "augmenting_mean": donor_mean(AUGMENTING_FUEL_BALANCE),
            }
        )
    # Refuse an undeclared variant.
    raise IngestError(f"{AXIS_FACTOR_PROXY}: undeclared variant {variant!r}.")


# -------------------------------------------------------------------------------------------------------------------------------
# Task 18, Step 2 (continued): The swept composition
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the swept run of one pair under a variant factor.
def compose_with_variant_factor(
    policy_name: str,
    outcome_name: str,
    partition: Partition,
    panel: pd.DataFrame,
    lagged: pd.DataFrame,
    proxy: pd.DataFrame,
    windows: Mapping[str, Tuple[int, int]],
) -> SweepResult:
    """
    Compose the swept run of one pair under a variant factor.

    Parameters
    ----------
    policy_name : str
        The policy instrument of the pair.
    outcome_name : str
        The canonical outcome of the pair.
    partition : Partition
        The partition of the pair, carrying the variant's donor pool.
    panel : pandas.DataFrame
        The analysis panel.
    lagged : pandas.DataFrame
        The lagged panel.
    proxy : pandas.DataFrame
        The variant proxy columns, whose first column is the constant.
    windows : mapping of str to tuple of int
        The resolved windows of the run.

    Returns
    -------
    SweepResult
        The swept effect, its standard error, its marker and the diagnostic
        notes, which include the donor count and the factor condition number.

    Raises
    ------
    IngestError
        If a design, shape or identity check fails.
    """
    # Compose the diagnostic notes of the run.
    notes: List[str] = []
    # Compose the unit list as the union of the targets and the donors, which
    # is the declared estimation sample of the pooled fit.
    units: List[str] = sorted(
        set(str(code) for code in partition.targets)
        | set(str(code) for code in partition.donors)
    )
    # Build the baseline factor objects of the pair.
    factor = assemble_partition_factor(
        policy_name,
        outcome_name,
        partition.donors,
        units,
        panel,
        lagged,
        window=PRE_TREATMENT_WINDOW,
        factor_window=FACTOR_WINDOW,
    )
    # Initialise the per-unit dependent and covariate mappings.
    dependent_by_unit: Dict[str, np.ndarray] = {}
    # Initialise the per-unit covariate mapping.
    covariate_by_unit: Dict[str, np.ndarray] = {}
    # Iterate over the assembled units in sorted order.
    for unit in sorted(factor.matrices.keys()):
        # Read the unit's dependent and design matrices.
        dependent, pooling = factor.matrices[unit]
        # Record the unit's dependent matrix.
        dependent_by_unit[str(unit)] = dependent
        # Extract the covariate columns, which follow the intercept.
        covariate_by_unit[str(unit)] = pooling[:, 1 : 1 + len(CONTROL_COLUMNS)]
    # Compose the variant factor matrix over the pre-treatment window.
    pre_years: List[int] = list(
        range(int(PRE_TREATMENT_WINDOW[0]), int(PRE_TREATMENT_WINDOW[1]) + 1)
    )
    # Select the variant proxy rows of the pre-treatment window.
    variant_pre: pd.DataFrame = proxy.reindex(index=pre_years)
    # Refuse a variant proxy that does not cover the pre-treatment window.
    if bool(variant_pre.isna().to_numpy().any()):
        # The failure names the pair.
        raise IngestError(
            f"{policy_name}/{outcome_name}: the variant proxy does not cover the "
            f"pre-treatment window."
        )
    # Compose the variant factor matrix.
    factor_matrix: np.ndarray = variant_pre.to_numpy(dtype=float)
    # Record the factor width and the donor count as diagnostics.
    notes.append(
        f"factor variant width {factor_matrix.shape[1]} across "
        f"{len(partition.donors)} donor(s)."
    )
    # Compute the condition number of the variant factor cross-product.
    singular_values = np.linalg.svd(factor_matrix.T @ factor_matrix, compute_uv=False)
    # Compose the condition number, reporting infinity for a singular product.
    condition: float = (
        float("inf")
        if float(singular_values[-1]) == 0.0
        else float(singular_values[0] / singular_values[-1])
    )
    # Record the condition number, which the specification requires for this
    # axis.
    notes.append(f"factor cross-product condition number {condition:.6e}.")
    # Build the residual-maker of the variant factor.
    maker = build_residual_maker(factor_matrix)
    # Estimate the pooled coefficient over the union of targets and donors.
    beta, fit_diagnostics = estimate_pooled_coefficient(
        dependent_by_unit, covariate_by_unit, maker
    )
    # Recover the country loadings of every unit.
    # Preserve the diagnostic's infinite value for a singular design; bound
    # only the arithmetic allowance of the geometric orthogonality audit.
    orthogonality_conditioning = (
        float(fit_diagnostics["condition_number"])
        if np.isfinite(float(fit_diagnostics["condition_number"]))
        else float(1.0 / np.finfo(float).eps)
    )
    country_loadings = estimate_unit_loadings(
        factor_matrix,
        dependent_by_unit,
        covariate_by_unit,
        beta,
        conditioning=orthogonality_conditioning,
    )
    # Estimate the covariate loadings of the targets alone.
    target_loadings = estimate_covariate_loadings(
        factor_matrix, covariate_by_unit, partition.targets
    )
    # Compose the projection span of the run, which is the whole gap window:
    # the level windows consume the first wave's years and the incremental
    # estimand consumes the post window, so the counterfactual must cover the
    # full span rather than the post window alone. The closing year is the
    # full baseline wave-two end even under a shortened horizon variant, so
    # per-year gaps stay comparable across variants (B-19 identity discipline).
    projection_years: List[int] = list(
        range(
            int(windows[WAVE_ONE_LABEL][0]),
            int(WAVE_TWO_WINDOW[1]) + 1,
        )
    )
    # Select the variant proxy rows of the projection span.
    projection_rows: pd.DataFrame = proxy.reindex(index=projection_years)
    # Refuse a variant proxy that does not cover the projection span.
    if bool(projection_rows.isna().to_numpy().any()):
        # The failure names the pair.
        raise IngestError(
            f"{policy_name}/{outcome_name}: the variant proxy does not cover the "
            f"projection span."
        )
    # Compose the projection factor matrix at the variant's own width.
    projection_factor: np.ndarray = projection_rows.to_numpy(dtype=float)
    # Accumulate the projected counterfactual rows.
    projected_rows: List[Dict[str, Any]] = []
    # Iterate over the targets in sorted order.
    for target in sorted(target_loadings.keys()):
        # Project the target's covariate and outcome paths.
        covariate_path, outcome_path = project_target_paths(
            target_loadings[target],
            np.asarray(beta),
            np.asarray(country_loadings[target]),
            projection_factor,
        )
        # Record the projected outcome of every projection year.
        for index, year in enumerate(projection_years):
            # Append the projected row.
            projected_rows.append(
                {
                    "policy": policy_name,
                    "outcome": outcome_name,
                    "country_code": str(target),
                    "year": int(year),
                    "counterfactual_outcome": float(outcome_path[index]),
                }
            )
        # Record the covariate path width as a diagnostic.
        notes.append(
            f"counterfactual covariate path width "
            f"{covariate_path.shape[1]} for {target}."
        )
    # Assemble the counterfactual frame the gap computation consumes.
    counterfactual_frame: pd.DataFrame = pd.DataFrame(projected_rows)
    # Compose the gap window of the run.
    gap_start: Tuple[int, int] = windows[WAVE_ONE_LABEL]
    # Compose the gap end window. The closing year is the full baseline
    # wave-two end even under a shortened horizon variant; the variant window
    # governs only the aggregation below, so gaps stay comparable (B-19).
    gap_end: Tuple[int, int] = (windows["post_escalation"][0], WAVE_TWO_WINDOW[1])
    # Compute the gap matrix over the declared span.
    gaps: pd.DataFrame = compute_gap_matrix(
        panel,
        outcome_name,
        partition.targets,
        counterfactual_frame,
        policy_name,
        window=gap_start,
        window_end=gap_end,
    )
    # Aggregate the per-country window means.
    country_means: pd.DataFrame = aggregate_country_window_means(
        gaps,
        {
            WAVE_ONE_LABEL: windows[WAVE_ONE_LABEL],
            WAVE_TWO_LABEL: windows[WAVE_TWO_LABEL],
        },
    )
    # Compute the coalition level effects.
    level_effects: pd.Series = aggregate_level_effects(country_means)
    # Compute the per-country pre-escalation levels.
    pre_levels: pd.Series = compute_pre_escalation_levels(
        gaps, windows["pre_escalation"]
    )
    # Build the observed post-escalation window.
    observed_window: pd.DataFrame = build_observed_window(
        panel,
        outcome_name,
        tuple(str(code) for code in gaps.index),
        windows["post_escalation"],
    )
    # Compute the continuation gap through the shifted counterfactual.
    continuation_gap: pd.DataFrame = compute_continuation_gap(
        gaps, pre_levels, observed_window, windows["post_escalation"]
    )
    # Compute the incremental effect through the continuation route.
    effect: float = compute_incremental_effect(continuation_gap)
    # Compute the incremental effect through the direct route.
    direct: float = compute_incremental_effect_alternative(
        gaps, windows["pre_escalation"], windows["post_escalation"]
    )
    # Assert the declared identity between the two routes.
    assert_incremental_identity(effect, direct)
    # Compose the per-country incremental contrasts.
    post_years: List[int] = list(
        range(
            int(windows["post_escalation"][0]), int(windows["post_escalation"][1]) + 1
        )
    )
    # Compute the contrasts as the post mean less the pre level.
    contrasts: pd.Series = gaps[post_years].mean(axis=1) - pre_levels
    # Compute the effect and its dispersion standard error.
    contrast_effect, se_cd = compute_country_dispersion_se(contrasts)
    # Compute the test statistic and the degrees of freedom.
    statistic, degrees_of_freedom = compute_t_statistic(
        contrast_effect, se_cd, len(contrasts)
    )
    # Compute the two-sided probability.
    probability: float = compute_two_sided_p(statistic, degrees_of_freedom)
    # Assign the significance marker from the critical-value comparison.
    stars, _critical_values = assign_significance_stars(statistic, degrees_of_freedom)
    # Assert the marker's agreement with the probability route.
    assert_star_consistency(
        stars,
        probability,
        statistic=statistic,
        degrees_of_freedom=degrees_of_freedom,
    )
    # Assert the probability's reproducibility.
    assert_p_reproducible(probability, statistic, degrees_of_freedom)
    # Record the pooled fit's rank as a diagnostic.
    notes.append(
        f"pooled fit rank {int(fit_diagnostics['rank'])} over "
        f"{int(fit_diagnostics['rows'])} row(s); residual sum of squares "
        f"{float(fit_diagnostics['rss']):.6e}."
    )
    # Record the level effects as diagnostics.
    notes.append(
        f"level effects W1 {float(level_effects[WAVE_ONE_LABEL]):.6f} and W2 "
        f"{float(level_effects[WAVE_TWO_LABEL]):.6f}."
    )
    # Return the swept result.
    return SweepResult(
        policy_name=str(policy_name),
        outcome_name=str(outcome_name),
        axis=AXIS_FACTOR_PROXY,
        variant_name=str(proxy.attrs.get("variant_name", "unnamed")),
        att=float(effect),
        se_cd=float(se_cd),
        stars=str(stars),
        diagnostic_notes=tuple(notes),
    )

# Compose the partition of a donor-pool variant.
def build_variant_partition(
    partition: Partition,
    variant: str,
    data: Mapping[str, Any],
) -> Tuple[Partition, Tuple[str, ...]]:
    """
    Compose the partition of a donor-pool variant.

    Parameters
    ----------
    partition : Partition
        The baseline partition of the pair.
    variant : str
        The donor-pool variant name.
    data : mapping
        The frozen inputs, which must carry the donor-frame inputs
        ``repaired_donor_codes`` and ``income_groups``.

    Returns
    -------
    tuple
        The variant partition and diagnostic notes containing the exact
        removed identities and before/after roster sizes.

    Raises
    ------
    IngestError
        If required metadata are unresolved/malformed, the named restriction
        is a no-op, or the usable pool falls below the declared minimum.
    """
    # Compose the baseline donor list.
    donors: List[str] = sorted(str(code) for code in partition.donors)
    original_donors = tuple(donors)
    notes: List[str] = []
    # Apply the variant's transform.
    if variant == "corrected_full_pool":
        # The corrected pool is the baseline pool by definition.
        notes.append("donor pool variant: registered baseline corrected full pool.")
    elif variant == "no_repaired_main_outcome_cells":
        repaired = data.get("repaired_donor_codes")
        if repaired is None:
            raise SweepUnresolvedInputError(
                "no_repaired_main_outcome_cells requires repaired_donor_codes."
            )
        if isinstance(repaired, Mapping):
            repaired = repaired.get((partition.policy, partition.outcome))
        if isinstance(repaired, (str, bytes, Mapping)) or repaired is None:
            raise SweepUnresolvedInputError(
                "repaired_donor_codes lacks the requested pair/outcome roster."
            )
        try:
            repaired_list = [str(code).strip() for code in repaired]
        except TypeError as exc:
            raise SweepUnresolvedInputError(
                "repaired_donor_codes must be a finite identity sequence."
            ) from exc
        if any(not code for code in repaired_list) or len(repaired_list) != len(set(repaired_list)):
            raise IngestError("repaired_donor_codes contains blank or duplicate identities.")
        excluded = set(repaired_list)
        outsiders = sorted(excluded - set(original_donors))
        if outsiders:
            raise IngestError(
                "repaired_donor_codes contains identity(ies) outside the intended "
                f"pair donor roster: {outsiders[:INFERENCE_MAX_REPORTED_OFFENDERS]}."
            )
        donors = [code for code in donors if code not in excluded]
    elif variant == "high_income_donors":
        income = data.get("income_groups")
        if not isinstance(income, Mapping) or not income:
            raise SweepUnresolvedInputError(
                "high_income_donors requires a populated income_groups mapping."
            )
        normalized_income = {
            str(code).strip(): str(group).strip().lower()
            for code, group in income.items()
            if code is not None and group is not None
        }
        missing_income = sorted(set(donors) - set(normalized_income))
        if missing_income:
            raise SweepUnresolvedInputError(
                "income_groups omits baseline donor(s) "
                f"{missing_income[:INFERENCE_MAX_REPORTED_OFFENDERS]}."
            )
        valid_income_labels = {
            "low income", "low-income", "low_income",
            "lower middle income", "lower-middle income", "lower_middle_income",
            "upper middle income", "upper-middle income", "upper_middle_income",
            "high income", "high-income", "high_income", "high income: oecd",
            "high income: nonoecd",
        }
        invalid_income = sorted(
            code for code in donors
            if normalized_income.get(code, "") not in valid_income_labels
        )
        if invalid_income:
            raise SweepUnresolvedInputError(
                "income_groups carries blank/unknown classification for donor(s) "
                f"{invalid_income[:INFERENCE_MAX_REPORTED_OFFENDERS]}."
            )
        high_income_labels = {
            "high income", "high-income", "high_income", "high income: oecd",
            "high income: nonoecd",
        }
        high_income = {
            code for code, group in normalized_income.items()
            if group in high_income_labels
        }
        donors = [code for code in donors if code in high_income]
    else:
        # Refuse an undeclared variant.
        raise IngestError(f"{AXIS_DONOR_POOL}: undeclared variant {variant!r}.")
    # Refuse a pool reduced below the declared minimum.
    if len(donors) < 2:
        # The failure names the variant and the retained count.
        raise IngestError(
            f"{AXIS_DONOR_POOL}/{variant}: the variant retains {len(donors)} "
            f"donor(s), below the declared minimum of two."
        )
    removed = tuple(sorted(set(original_donors) - set(donors)))
    if variant == "high_income_donors" and not removed:
        raise IngestError(
            f"{AXIS_DONOR_POOL}/{variant}: required restriction removes no donor; "
            "the high-income restriction cannot be verified as an effective change."
        )
    # Append exact sweep exclusions without discarding baseline scientific
    # exclusions or duplicating an existing reason row.
    exclusion_columns = ["policy", "outcome", "country_code", "reason", "detail"]
    baseline_exclusions = partition.exclusions.copy(deep=True)
    if baseline_exclusions.empty and len(baseline_exclusions.columns) == 0:
        baseline_exclusions = pd.DataFrame(columns=exclusion_columns)
    if set(baseline_exclusions.columns) != set(exclusion_columns):
        raise IngestError("Baseline partition exclusion audit schema is malformed.")
    sweep_rows = pd.DataFrame(
        [
            {
                "policy": partition.policy,
                "outcome": partition.outcome,
                "country_code": code,
                "reason": f"sweep_{variant}",
                "detail": "removed from donor roster by registered Task 18 variant",
            }
            for code in removed
        ],
        columns=exclusion_columns,
    )
    variant_exclusions = pd.concat(
        [baseline_exclusions[exclusion_columns], sweep_rows], ignore_index=True
    )
    notes.extend(
        [
            f"donor counts before={len(original_donors)} after={len(donors)}.",
            "removed_donors_json=" + json.dumps(list(removed), separators=(",", ":")),
        ]
    )
    if variant == "no_repaired_main_outcome_cells" and not removed:
        notes.append("verified identity: no repaired main-outcome cells occur in this pair's donor roster.")
    # Compose the variant partition.
    variant_partition = Partition(
        policy=partition.policy,
        outcome=partition.outcome,
        targets=partition.targets,
        donors=tuple(donors),
        exclusions=variant_exclusions,
    )
    # Return the variant partition and its notes.
    return variant_partition, tuple(notes)


# Compose the data mapping carrying a variant partition.
def build_variant_data(
    data: Mapping[str, Any], policy_name: str, outcome_name: str, partition: Partition
) -> Dict[str, Any]:
    """
    Compose the data mapping carrying a variant partition.

    Parameters
    ----------
    data : mapping
        The frozen inputs under the declared keys.
    policy_name : str
        The policy instrument of the pair.
    outcome_name : str
        The canonical outcome of the pair.
    partition : Partition
        The variant partition.

    Returns
    -------
    dict
        A copy preserving every auxiliary input, with the intended partition
        replaced inside a fully regenerated eight-pair registry and explicit
        parent/variant lineage metadata.

    Raises
    ------
    KeyError
        If the mapping lacks a declared key.
    IngestError
        If the parent registry is incomplete/stale or the replacement identity
        disagrees with the requested pair.
    """
    # Refuse a mapping lacking the declared keys.
    for key in ("registry", "panel", "lagged"):
        # Test the key's presence.
        if key not in data:
            # The failure names the absent key.
            raise KeyError(f"The data mapping lacks the declared key {key!r}.")
    registry = data["registry"]
    if not isinstance(registry, PartitionRegistry):
        raise IngestError("Variant data source registry is malformed.")
    pair = (str(policy_name), str(outcome_name))
    if pair not in registry.partitions:
        raise IngestError(f"Baseline registry omits variant pair {pair}.")
    if (str(partition.policy), str(partition.outcome)) != pair:
        raise IngestError("Variant partition identity disagrees with requested pair.")
    if set(registry.partitions) != set(DECLARED_PAIRS):
        raise IngestError(
            "Variant data requires the complete eight-partition parent registry."
        )
    if registry.fingerprint_sha256 != freeze_partition_registry(registry.registry):
        raise IngestError("Baseline parent registry fingerprint is stale.")
    # Copy every baseline partition and replace only the intended pair, so all
    # downstream consumers retain complete eight-pair coverage.
    partitions: Dict[Tuple[str, str], Partition] = {
        key: Partition(
            policy=value.policy,
            outcome=value.outcome,
            targets=tuple(value.targets),
            donors=tuple(value.donors),
            exclusions=value.exclusions.copy(deep=True),
        )
        for key, value in registry.partitions.items()
    }
    partitions[pair] = Partition(
        policy=partition.policy,
        outcome=partition.outcome,
        targets=tuple(partition.targets),
        donors=tuple(partition.donors),
        exclusions=partition.exclusions.copy(deep=True),
    )
    variant_registry_frame = register_partitions(partitions)
    variant_counts = variant_registry_frame[["targets", "donors"]].copy(deep=True)
    exclusion_frames = [
        item.exclusions.copy(deep=True)
        for item in partitions.values()
        if not item.exclusions.empty
    ]
    variant_exclusions = (
        pd.concat(exclusion_frames, ignore_index=True)
        if exclusion_frames
        else pd.DataFrame(
            columns=["policy", "outcome", "country_code", "reason", "detail"]
        )
    )
    variant_fingerprint = freeze_partition_registry(variant_registry_frame)
    lineage_payload = {
        "parent_registry_sha256": registry.fingerprint_sha256,
        "variant_registry_sha256": variant_fingerprint,
        "pair": list(pair),
        "targets": list(partition.targets),
        "donors": list(partition.donors),
    }
    lineage_sha256 = hashlib.sha256(
        json.dumps(lineage_payload, sort_keys=True, separators=(",", ":")).encode("utf-8")
    ).hexdigest()
    variant_registry = PartitionRegistry(
        partitions=partitions,
        registry=variant_registry_frame,
        counts=variant_counts,
        exclusions=variant_exclusions,
        fingerprint_sha256=variant_fingerprint,
        built_utc=current_utc_iso(),
        notes=tuple(registry.notes)
        + (
            f"variant parent fingerprint {registry.fingerprint_sha256}",
            f"variant pair {pair[0]}/{pair[1]} lineage {lineage_sha256}",
        ),
    )
    # Preserve every explicit auxiliary input while replacing only the
    # registry and adding auditable parent/variant lineage.
    variant_data = dict(data)
    variant_data["registry"] = variant_registry
    variant_data["variant_lineage"] = MappingProxyType(
        {**lineage_payload, "lineage_sha256": lineage_sha256}
    )
    return variant_data


# Compute the incremental effect under per-country post windows.
def compute_incremental_effect_per_country(
    gaps: pd.DataFrame,
    pre_levels: pd.Series,
    country_windows: Mapping[str, Tuple[int, int]],
) -> float:
    """
    Compute the incremental effect under per-country post windows.

    Parameters
    ----------
    gaps : pandas.DataFrame
        The gap matrix indexed on the country and columned by the years.
    pre_levels : pandas.Series
        The per-country pre-escalation levels.
    country_windows : mapping of str to tuple of int
        The post-escalation window of each country, which the active-year
        axis defines functionally.

    Returns
    -------
    float
        The equal-weighted mean across countries of each country's own
        post-window mean gap less its pre-escalation level.

    Raises
    ------
    IngestError
        If a country is absent from the windows or the gap matrix.
    """
    # Validate the labelled scientific inputs before country-first averaging.
    if (
        not isinstance(gaps, pd.DataFrame)
        or gaps.empty
        or gaps.index.has_duplicates
        or gaps.columns.has_duplicates
        or not np.isfinite(gaps.to_numpy(dtype=float)).all()
    ):
        raise IngestError("Active-year gaps are empty, duplicate-labelled or non-finite.")
    if (
        not isinstance(pre_levels, pd.Series)
        or pre_levels.index.has_duplicates
        or set(pre_levels.index.astype(str)) != set(gaps.index.astype(str))
        or not np.isfinite(pd.to_numeric(pre_levels, errors="coerce")).all()
    ):
        raise IngestError("Active-year reference levels are malformed or misaligned.")
    # Require the documented mapping rather than an ambient global mask.
    if not isinstance(country_windows, Mapping):
        # Reject an input that cannot resolve a window for each country.
        raise IngestError("Active-year country windows must be a mapping.")
    # Enforce the integer year labels used by the canonical gap matrix.
    if any(
        isinstance(year, (bool, np.bool_))
        or not isinstance(year, (int, np.integer))
        for year in gaps.columns
    ):
        # Refuse labels that would be silently truncated during mask creation.
        raise IngestError("Active-year gap columns must be integer years.")
    # Collect the country-year activity records for this invocation alone.
    mask_rows: List[Dict[str, Any]] = []
    # Derive one validated activity window for each gap-matrix country.
    for country in gaps.index:
        # Resolve the canonical country key used by the window mapping.
        country_code: str = str(country)
        # Require a declared window for the current country.
        if country_code not in country_windows:
            # Identify the missing country instead of using another run's mask.
            raise IngestError(
                f"{AXIS_ACTIVE_YEAR}: country {country_code} carries no "
                "resolved window."
            )
        # Read the explicitly supplied inclusive post-escalation window.
        span = country_windows[country_code]
        # Require exactly two integral, non-Boolean bounds.
        if (
            not isinstance(span, (tuple, list))
            or len(span) != 2
            or any(
                isinstance(bound, (bool, np.bool_))
                or not isinstance(bound, (int, np.integer))
                for bound in span
            )
        ):
            # Refuse malformed bounds before deriving any activity cells.
            raise IngestError(
                f"{AXIS_ACTIVE_YEAR}: malformed window for {country_code}."
            )
        # Require a nonempty window wholly contained in the second wave.
        if (
            span[0] > span[1]
            or span[0] < WAVE_TWO_WINDOW[0]
            or span[1] > WAVE_TWO_WINDOW[1]
        ):
            # Keep the active-year branch within its declared estimand.
            raise IngestError(
                f"{AXIS_ACTIVE_YEAR}: window {span} for {country_code} "
                f"leaves the intended Wave 2 span {WAVE_TWO_WINDOW}."
            )
        # Record activity for each year that the supplied gaps actually cover.
        for year in gaps.columns:
            # Keep the activity record keyed to its own country and year.
            mask_rows.append(
                {
                    COUNTRY_COLUMN: country_code,
                    YEAR_COLUMN: int(year),
                    "active": bool(span[0] <= int(year) <= span[1]),
                }
            )
    # Materialize the local mask consumed by the existing validation below.
    masks: pd.DataFrame = pd.DataFrame(
        mask_rows, columns=[COUNTRY_COLUMN, YEAR_COLUMN, "active"]
    )
    required_mask_columns = {"country_code", "year", "active"}
    if (
        not isinstance(masks, pd.DataFrame)
        or not required_mask_columns <= set(masks.columns)
        or bool(masks.duplicated(["country_code", "year"]).any())
    ):
        raise IngestError("Active-year mask is incomplete or duplicate-keyed.")
    if set(masks["country_code"].astype(str)) != set(gaps.index.astype(str)):
        raise IngestError("Active-year mask roster disagrees with the gap matrix.")
    if not masks["active"].map(
        lambda value: isinstance(value, (bool, np.bool_))
    ).all():
        raise IngestError("Active-year mask cells must be Boolean.")
    mask_years = pd.to_numeric(masks["year"], errors="coerce")
    if bool(mask_years.isna().any()) or not mask_years.map(
        lambda value: float(value).is_integer()
    ).all():
        raise IngestError("Active-year mask years are malformed.")
    if not set(mask_years.astype(int)) <= set(int(year) for year in gaps.columns):
        raise IngestError("Active-year mask requests a year absent from the gaps.")
    # Accumulate the per-country contrasts.
    contrasts: List[float] = []
    # Iterate over the countries in the matrix's own order.
    for country in gaps.index:
        # Refuse a country without a resolved window.
        if str(country) not in country_windows:
            # The failure names the country.
            raise IngestError(
                f"{AXIS_ACTIVE_YEAR}: country {country} carries no resolved "
                f"window."
            )
        # Read the country's window.
        span = country_windows[str(country)]
        # Compose the country's year columns.
        years: List[int] = list(range(int(span[0]), int(span[1]) + 1))
        # Refuse a window the gap matrix does not cover.
        missing: List[int] = [
            year for year in years if year not in set(int(c) for c in gaps.columns)
        ]
        # Raise when the window leaves the coverage.
        if missing:
            # The failure names the missing years.
            raise IngestError(
                f"{AXIS_ACTIVE_YEAR}: gap matrix does not cover year(s) {missing} "
                f"for {country}."
            )
        # Compose the country's own post-window mean.
        post_mean: float = float(gaps.loc[country, years].mean())
        # Append the country's contrast.
        contrasts.append(post_mean - float(pre_levels.loc[country]))
    # Return the equal-weighted mean of the per-country contrasts.
    return float(np.mean(np.array(contrasts, dtype=float)))


# Compute Eq. (16) inference from a fully labelled contrast vector.
def _sweep_inference_from_contrasts(
    contrasts: pd.Series,
) -> Tuple[float, float, str]:
    """Return the equal-country effect, dispersion SE and Student-t stars."""
    if not isinstance(contrasts, pd.Series) or contrasts.index.has_duplicates:
        raise IngestError("Sweep contrasts must be a uniquely labelled Series.")
    values = pd.to_numeric(contrasts, errors="coerce")
    if len(values) < 2 or bool(values.isna().any()) or not np.isfinite(values).all():
        raise IngestError("Sweep contrasts require at least two finite countries.")
    effect, se_cd = compute_country_dispersion_se(values.astype(float))
    statistic, degrees_of_freedom = compute_t_statistic(effect, se_cd, len(values))
    probability = compute_two_sided_p(statistic, degrees_of_freedom)
    stars, _critical_values = assign_significance_stars(
        statistic, degrees_of_freedom
    )
    assert_star_consistency(
        stars,
        probability,
        statistic=statistic,
        degrees_of_freedom=degrees_of_freedom,
    )
    assert_p_reproducible(probability, statistic, degrees_of_freedom)
    return float(effect), float(se_cd), str(stars)


# Recursively freeze nested sweep evidence before it is exposed to callers.
def _deep_freeze_sweep_evidence(value: Any) -> Any:
    """Recursively freeze JSON-like sweep evidence without changing values."""
    # Freeze mappings after recursively freezing both their keys' values.
    if isinstance(value, Mapping):
        return MappingProxyType(
            {str(key): _deep_freeze_sweep_evidence(item) for key, item in value.items()}
        )
    # Replace every mutable sequence with an immutable tuple.
    if isinstance(value, (list, tuple)):
        return tuple(_deep_freeze_sweep_evidence(item) for item in value)
    # Replace mutable sets with a deterministically ordered immutable tuple.
    if isinstance(value, (set, frozenset)):
        return tuple(
            _deep_freeze_sweep_evidence(item)
            for item in sorted(value, key=lambda item: str(item))
        )
    # Normalize NumPy scalars so the retained evidence is stable and immutable.
    if isinstance(value, (np.integer, np.floating, np.bool_)):
        return value.item()
    # Strings, numbers, Booleans and null are already immutable.
    return value


# Restore frozen evidence to ordinary JSON containers for canonical export.
def _thaw_sweep_evidence(value: Any) -> Any:
    """Recursively convert frozen sweep evidence to JSON-native containers."""
    # Preserve nested object structure instead of stringifying mapping proxies.
    if isinstance(value, Mapping):
        return {str(key): _thaw_sweep_evidence(item) for key, item in value.items()}
    # Convert immutable evidence sequences back to JSON arrays.
    if isinstance(value, (tuple, list)):
        return [_thaw_sweep_evidence(item) for item in value]
    # Normalize NumPy scalar leaves to their JSON-native equivalents.
    if isinstance(value, (np.integer, np.floating, np.bool_)):
        return value.item()
    # Return ordinary immutable leaves unchanged.
    return value


# Serialize frozen evidence and prove its JSON round trip is lossless.
def _canonical_sweep_evidence_json(value: Mapping[str, Any]) -> str:
    """Return canonical structured JSON for a frozen sweep-design mapping."""
    # Thaw every nested proxy and tuple before serialization.
    thawed = _thaw_sweep_evidence(value)
    if not isinstance(thawed, dict):
        raise IngestError("Effective sweep design must thaw to a JSON object.")
    try:
        serialized = json.dumps(
            thawed, sort_keys=True, separators=(",", ":"), allow_nan=False
        )
        reconstructed = json.loads(serialized)
        reconstructed_json = json.dumps(
            reconstructed, sort_keys=True, separators=(",", ":"), allow_nan=False
        )
    except (TypeError, ValueError, json.JSONDecodeError) as exc:
        raise IngestError("Effective sweep design is not canonical JSON evidence.") from exc
    # A byte-identical canonical round trip proves nested object/list parity.
    if reconstructed_json != serialized:
        raise IngestError("Effective sweep design changed during JSON round trip.")
    return serialized


# Validate a cached Task 15 result against every current scientific input.
def _validate_cached_sweep_baseline(
    result: BaselineResult,
    pair: Tuple[str, str],
    registry: PartitionRegistry,
    panel: pd.DataFrame,
    lagged: pd.DataFrame,
    effective_windows: Mapping[str, Any],
) -> None:
    """
    Validate a cached baseline before a Task 18 sweep may reuse it.

    The cache key is the complete Task 15 lineage rather than merely the
    policy/outcome label and reporting windows.  The check binds the result to
    the current partition rosters, complete registry structure, panel,
    lagged controls and effective-window fingerprint, and also validates the
    retained gap labels used by reporting-only reaggregation.
    """
    # Refuse an object that does not implement the declared Task 15 contract.
    if not isinstance(result, BaselineResult):
        raise IngestError("Cached sweep baseline is not a BaselineResult.")
    # Bind the public result identity to the requested partition.
    if (str(result.policy_name), str(result.outcome_name)) != pair:
        raise IngestError("Cached baseline result identity disagrees with sweep pair.")
    # Resolve the actual partition instead of trusting retained cache metadata.
    if pair not in registry.partitions:
        raise IngestError(f"Cached sweep pair {pair} is absent from the registry.")
    partition = registry.partitions[pair]
    # Require the complete lineage envelope emitted by the Task 15 fitter.
    provenance = result.fit_provenance
    if not isinstance(provenance, Mapping) or not {
        "lineage", "lineage_sha256"
    }.issubset(provenance):
        raise IngestError("Cached baseline fit provenance is incomplete.")
    lineage = provenance["lineage"]
    required_lineage = {
        "pair", "targets", "donors", "registry_fingerprint",
        "registry_structure_digest", "panel_fingerprint",
        "lagged_fingerprint", "window_fingerprint",
    }
    if not isinstance(lineage, Mapping) or not required_lineage.issubset(lineage):
        raise IngestError("Cached baseline scientific lineage is incomplete.")
    # Validate that the registry's advertised digest still matches its table.
    actual_registry_fingerprint = freeze_partition_registry(registry.registry)
    if str(registry.fingerprint_sha256) != actual_registry_fingerprint:
        raise IngestError("Current partition-registry fingerprint is stale.")
    # Resolve and require the authoritative effective-window fingerprint.
    window_provenance = effective_windows.get(WINDOW_PROVENANCE_KEY, {})
    actual_window_fingerprint = (
        str(window_provenance.get("fingerprint_sha256", ""))
        if isinstance(window_provenance, Mapping) else ""
    )
    if not actual_window_fingerprint:
        raise IngestError("Current effective-window fingerprint is absent.")
    # Compare every scientific cache-key component to current source inputs.
    expected_values: Dict[str, Any] = {
        "pair": pair,
        "targets": tuple(sorted(str(code) for code in partition.targets)),
        "donors": tuple(sorted(str(code) for code in partition.donors)),
        "registry_fingerprint": actual_registry_fingerprint,
        "registry_structure_digest": _partition_registry_structure_digest(registry),
        "panel_fingerprint": _content_digest(panel),
        "lagged_fingerprint": _content_digest(lagged),
        "window_fingerprint": actual_window_fingerprint,
    }
    # Require exact labelled identities and exact hexadecimal fingerprints.
    for field, expected in expected_values.items():
        observed = lineage.get(field)
        if field in {"pair", "targets", "donors"}:
            equal = tuple(observed or ()) == tuple(expected)
        else:
            equal = str(observed) == str(expected)
        if not equal:
            raise IngestError(
                f"Cached baseline {field.replace('_', ' ')} lineage is stale."
            )
    # Detect post-fit mutation of any retained lineage field.
    lineage_digest = hashlib.sha256(
        json.dumps(lineage, sort_keys=True, default=str).encode("utf-8")
    ).hexdigest()
    if lineage_digest != str(provenance["lineage_sha256"]):
        raise IngestError("Cached baseline fit lineage mutated after snapshot.")
    # Require the cached resolved windows, including provenance, to be exact.
    if _content_digest(result.effective_windows) != _content_digest(effective_windows):
        raise IngestError("Cached baseline result uses different effective windows.")
    # Bind the matrix that is actually reaggregated to the registered targets.
    gaps = result.gaps
    if not isinstance(gaps, pd.DataFrame) or gaps.index.has_duplicates:
        raise IngestError("Cached baseline gaps are malformed or duplicate-labelled.")
    observed_targets = tuple(str(code) for code in gaps.index)
    expected_targets = tuple(str(code) for code in partition.targets)
    if observed_targets != expected_targets:
        raise IngestError("Cached baseline gap index disagrees with target roster.")
    # Require complete finite projection coverage before any reaggregation.
    required_years = tuple(range(PROJECTION_START_YEAR, PROJECTION_END_YEAR + 1))
    # Validate lossless finite integral years before converting their type.
    if any(isinstance(year, (bool, np.bool_)) for year in gaps.columns):
        raise IngestError("Cached baseline gap calendar is malformed.")
    numeric_years = pd.to_numeric(pd.Series(list(gaps.columns)), errors="coerce")
    if bool(numeric_years.isna().any()) or not np.isfinite(
        numeric_years.to_numpy(dtype=float)
    ).all() or not all(float(year).is_integer() for year in numeric_years):
        raise IngestError("Cached baseline gap calendar is malformed.")
    observed_years = tuple(int(year) for year in numeric_years)
    if observed_years != required_years:
        raise IngestError("Cached baseline gap calendar is incomplete or misordered.")
    numeric_gaps = gaps.apply(pd.to_numeric, errors="coerce").to_numpy(dtype=float)
    if not np.isfinite(numeric_gaps).all():
        raise IngestError("Cached baseline gaps contain non-finite values.")


# Bind a successful sweep result to its immutable effective-design lineage.
def _finalize_sweep_result(
    result: SweepResult,
    *,
    run_id: str,
    effective_design: Mapping[str, Any],
    execution_mode: str,
    parent_registry_sha256: str,
) -> SweepResult:
    """Attach canonical design evidence and a deterministic lineage digest."""
    if execution_mode not in {
        "refit", "reaggregate_existing_gaps", "reuse_verified_baseline"
    }:
        raise IngestError(f"Unknown sweep execution mode {execution_mode!r}.")
    if result.scientific_status not in {"success", "identity"}:
        raise IngestError("Sweep result carries an unknown scientific status.")
    design = copy.deepcopy(dict(effective_design))
    payload = {
        "run_id": str(run_id),
        "parent_registry_sha256": str(parent_registry_sha256),
        "effective_design": design,
        "att": float(result.att),
        "se_cd": float(result.se_cd),
        "stars": str(result.stars),
        "execution_mode": execution_mode,
        "scientific_status": result.scientific_status,
    }
    lineage = hashlib.sha256(
        json.dumps(payload, sort_keys=True, separators=(",", ":"), default=str).encode(
            "utf-8"
        )
    ).hexdigest()
    return replace(
        result,
        run_id=str(run_id),
        effective_design=_deep_freeze_sweep_evidence(design),
        lineage_sha256=lineage,
        execution_mode=execution_mode,
    )


# Run one swept variant of one pair.
def run_sweep(
    policy_name: str,
    outcome_name: str,
    spec: SweepSpec,
    baseline_config: Mapping[str, Any],
    data: Mapping[str, Any],
    entry_years: Optional[Mapping[str, int]] = None,
    run_id: Optional[str] = None,
    baseline_result: Optional[BaselineResult] = None,
) -> SweepResult:
    """
    Run one swept variant of one pair.

    Parameters
    ----------
    policy_name : str
        The policy instrument of the pair.
    outcome_name : str
        The canonical outcome of the pair.
    spec : SweepSpec
        The sweep specification.
    baseline_config : mapping
        The baseline configuration, which is never mutated.
    data : mapping
        The frozen inputs under the declared keys.
    entry_years : mapping of str to int, optional
        The entry year of each country, required by the active-year axis.
    run_id : str, optional
        Stable registry identifier; direct calls derive one from the complete
        scientific identity when omitted.
    baseline_result : BaselineResult, optional
        Verified cached baseline fit reused by reporting-only axes, preventing
        one refit per reference/horizon/active-year variant.

    Returns
    -------
    SweepResult
        The swept effect, its standard error, its marker and the diagnostic
        notes.

    Raises
    ------
    IngestError
        If the variant is undeclared, the intended effective component does
        not change, cached lineage disagrees, or any composition check fails.
    """
    for key in ("registry", "panel", "lagged"):
        if key not in data:
            raise KeyError(f"The data mapping lacks the declared key {key!r}.")
    config_before = _content_digest(baseline_config)
    modified, paths = apply_sweep_variant(baseline_config, spec)
    registry = data["registry"]
    if not isinstance(registry, PartitionRegistry):
        raise IngestError("Sweep data registry is malformed.")
    pair = (str(policy_name), str(outcome_name))
    if pair not in registry.partitions:
        raise IngestError(f"The registry does not carry the pair {pair}.")
    partition = registry.partitions[pair]
    baseline_windows, _ = resolve_reporting_windows(baseline_config)
    modified_windows, _ = resolve_reporting_windows(modified)
    registered_run_id = str(
        run_id or f"{pair[0]}|{pair[1]}|{spec.axis}|{spec.variant_name}"
    )
    notes: List[str] = [
        "effective configuration paths="
        + json.dumps([".".join(path) for path in paths], separators=(",", ":"))
    ]
    baseline_identity = SWEEP_BASELINE_IDENTITIES.get(spec.axis) == spec.variant_name
    design: Dict[str, Any] = {
        "axis": spec.axis,
        "variant_name": spec.variant_name,
        "baseline_identity": bool(baseline_identity),
        "config_paths": [list(path) for path in paths],
        "config_path_scope": "derived_sweep_estimation_view",
        "authoritative_window_sources": {
            "pre_escalation": "global_constants.preferred_reference_window",
            WAVE_ONE_LABEL: "global_constants.Wave1",
            WAVE_TWO_LABEL: "global_constants.Wave2",
        },
        "parent_registry_sha256": registry.fingerprint_sha256,
        "baseline_windows": {
            key: list(baseline_windows[key])
            for key in (WAVE_ONE_LABEL, WAVE_TWO_LABEL, "pre_escalation", "post_escalation")
        },
        "baseline_donors": list(partition.donors),
    }
    # Reporting-only axes reuse one baseline fit and reaggregate its labelled
    # gaps; they never trigger a scientifically irrelevant refit.
    if spec.axis in {AXIS_REFERENCE_WINDOW, AXIS_HORIZON, AXIS_ACTIVE_YEAR}:
        supplied_baseline = baseline_result is not None
        if baseline_result is None:
            baseline_result = run_baseline(
                policy_name, outcome_name, baseline_config, data
            )
        # Validate the complete cache lineage before reading even one fitted gap.
        _validate_cached_sweep_baseline(
            baseline_result,
            pair,
            registry,
            data["panel"],
            data["lagged"],
            baseline_windows,
        )
        gaps = baseline_result.gaps.copy(deep=True)
        if spec.axis == AXIS_REFERENCE_WINDOW:
            pre_window = tuple(modified_windows["pre_escalation"])
            post_window = tuple(baseline_windows["post_escalation"])
            pre_levels = compute_pre_escalation_levels(gaps, pre_window)
            post_years = list(range(post_window[0], post_window[1] + 1))
            contrasts = gaps[post_years].mean(axis=1) - pre_levels
            country_windows: Optional[Dict[str, Tuple[int, int]]] = None
        elif spec.axis == AXIS_HORIZON:
            pre_window = tuple(baseline_windows["pre_escalation"])
            post_window = tuple(modified_windows["post_escalation"])
            pre_levels = compute_pre_escalation_levels(gaps, pre_window)
            post_years = list(range(post_window[0], post_window[1] + 1))
            contrasts = gaps[post_years].mean(axis=1) - pre_levels
            country_windows = None
        else:
            if entry_years is None:
                raise SweepUnresolvedInputError(
                    "active_year requires complete entry_years metadata."
                )
            pre_window = tuple(baseline_windows["pre_escalation"])
            post_window = tuple(baseline_windows["post_escalation"])
            pre_levels = compute_pre_escalation_levels(gaps, pre_window)
            country_windows = resolve_active_year_windows(
                spec,
                data["panel"],
                tuple(str(code) for code in gaps.index),
                entry_years,
            )
            active_identity = all(
                tuple(span) == post_window for span in country_windows.values()
            )
            contrasts = pd.Series(
                {
                    str(country): float(
                        gaps.loc[
                            country,
                            list(
                                range(
                                    country_windows[str(country)][0],
                                    country_windows[str(country)][1] + 1,
                                )
                            ),
                        ].mean()
                    )
                    - float(pre_levels.loc[country])
                    for country in gaps.index
                },
                dtype=float,
            )
        effect, se_cd, stars = _sweep_inference_from_contrasts(contrasts)
        if baseline_identity and (
            abs(effect - float(baseline_result.att_incremental)) > IDENTITY_TOLERANCE
            or abs(se_cd - float(baseline_result.se_cd)) > IDENTITY_TOLERANCE
            or stars != str(baseline_result.stars)
        ):
            raise IngestError("Registered baseline identity does not reproduce baseline inference.")
        if spec.axis == AXIS_ACTIVE_YEAR and active_identity and (
            abs(effect - float(baseline_result.att_incremental)) > IDENTITY_TOLERANCE
            or abs(se_cd - float(baseline_result.se_cd)) > IDENTITY_TOLERANCE
            or stars != str(baseline_result.stars)
        ):
            raise IngestError("Active-year identity does not reproduce baseline inference.")
        design.update(
            {
                "effective_pre_window": list(pre_window),
                "effective_post_window": list(post_window),
                "country_windows": (
                    {key: list(value) for key, value in sorted(country_windows.items())}
                    if country_windows is not None else None
                ),
                "gap_sha256": _content_digest(gaps),
                "contrast_sha256": _content_digest(contrasts.sort_index()),
                "estimation_refit": False,
                "baseline_fit_reused": bool(supplied_baseline),
                "effective_identity": bool(
                    spec.axis == AXIS_ACTIVE_YEAR and active_identity
                ),
            }
        )
        swept = SweepResult(
            policy_name=pair[0], outcome_name=pair[1], axis=spec.axis,
            variant_name=spec.variant_name, att=effect, se_cd=se_cd, stars=stars,
            diagnostic_notes=tuple(
                notes
                + [
                    "reaggregated unchanged fitted gap matrix from verified cache."
                    if supplied_baseline
                    else "performed one baseline fit, then reaggregated without variant-specific refit."
                ]
            ),
            scientific_status=(
                "identity" if spec.axis == AXIS_ACTIVE_YEAR and active_identity
                else "success"
            ),
        )
        execution_mode = "reaggregate_existing_gaps"
    elif spec.axis == AXIS_DONOR_POOL:
        variant_partition, donor_notes = build_variant_partition(
            partition, str(spec.variant_value), data
        )
        notes.extend(donor_notes)
        donor_identity = (
            spec.variant_name == "no_repaired_main_outcome_cells"
            and tuple(variant_partition.donors) == tuple(partition.donors)
        )
        if donor_identity:
            comparison = baseline_result
            if comparison is None:
                comparison = run_baseline(policy_name, outcome_name, baseline_config, data)
            _validate_cached_sweep_baseline(
                comparison, pair, registry, data["panel"], data["lagged"],
                baseline_windows,
            )
            design.update({
                "effective_donors": list(partition.donors),
                "removed_donors": [],
                "effective_identity": True,
                "estimation_refit": False,
                "baseline_lineage_sha256": comparison.fit_provenance["lineage_sha256"],
            })
            swept = SweepResult(
                policy_name=pair[0], outcome_name=pair[1], axis=spec.axis,
                variant_name=spec.variant_name,
                att=float(comparison.att_incremental), se_cd=float(comparison.se_cd),
                stars=str(comparison.stars), diagnostic_notes=tuple(notes),
                scientific_status="identity",
            )
            execution_mode = "reuse_verified_baseline"
        else:
            variant_data = build_variant_data(
                data, policy_name, outcome_name, variant_partition
            )
            variant_registry = variant_data["registry"]
            if not baseline_identity and variant_registry.fingerprint_sha256 == registry.fingerprint_sha256:
                raise IngestError("Nonbaseline donor variant has unchanged registry fingerprint.")
            fitted = run_baseline(policy_name, outcome_name, baseline_config, variant_data)
            design.update(
                {
                    "effective_donors": list(variant_partition.donors),
                    "removed_donors": sorted(set(partition.donors) - set(variant_partition.donors)),
                    "variant_registry_sha256": variant_registry.fingerprint_sha256,
                    "variant_lineage_sha256": variant_data["variant_lineage"]["lineage_sha256"],
                    "estimation_refit": True,
                }
            )
            swept = SweepResult(
                policy_name=pair[0], outcome_name=pair[1], axis=spec.axis,
                variant_name=spec.variant_name, att=float(fitted.att_incremental),
                se_cd=float(fitted.se_cd), stars=str(fitted.stars),
                diagnostic_notes=tuple(notes),
            )
            execution_mode = "refit"
    else:
        proxy = build_variant_proxy(
            data["panel"], partition.donors, outcome_name, str(spec.variant_value)
        )
        proxy.attrs["variant_name"] = str(spec.variant_name)
        baseline_proxy = build_variant_proxy(
            data["panel"], partition.donors, outcome_name,
            "own_outcome_donor_mean",
        )
        if not baseline_identity and _content_digest(proxy) == _content_digest(baseline_proxy):
            raise IngestError("Nonbaseline factor variant produces the baseline proxy unchanged.")
        swept = compose_with_variant_factor(
            policy_name, outcome_name, partition, data["panel"], data["lagged"],
            proxy, baseline_windows,
        )
        if baseline_identity:
            comparison = baseline_result
            if comparison is None:
                comparison = run_baseline(
                    policy_name, outcome_name, baseline_config, data
                )
            # Apply the same full cache-key guard used by reporting-only reuse;
            # baseline factor parity may not trust pair/window labels alone.
            _validate_cached_sweep_baseline(
                comparison,
                pair,
                registry,
                data["panel"],
                data["lagged"],
                baseline_windows,
            )
            if (
                abs(float(swept.att) - float(comparison.att_incremental))
                > IDENTITY_TOLERANCE
                or abs(float(swept.se_cd) - float(comparison.se_cd))
                > IDENTITY_TOLERANCE
                or str(swept.stars) != str(comparison.stars)
            ):
                raise IngestError(
                    "Baseline factor-proxy identity does not reproduce baseline inference."
                )
        design.update(
            {
                "factor_variant": str(spec.variant_value),
                "factor_columns": list(proxy.columns),
                "factor_sha256": _content_digest(proxy),
                "factor_pre_rank": int(
                    np.linalg.matrix_rank(
                        proxy.loc[PRE_TREATMENT_WINDOW[0]:PRE_TREATMENT_WINDOW[1]].to_numpy(dtype=float)
                    )
                ),
                "estimation_refit": True,
            }
        )
        execution_mode = "refit"
    if _content_digest(baseline_config) != config_before:
        raise IngestError("Sweep execution mutated the baseline configuration.")
    design["baseline_config_sha256"] = config_before
    design["effective_design_changed"] = bool(
        not baseline_identity and swept.scientific_status == "success"
    )
    return _finalize_sweep_result(
        swept,
        run_id=registered_run_id,
        effective_design=design,
        execution_mode=execution_mode,
        parent_registry_sha256=registry.fingerprint_sha256,
    )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 18, Step 3: The sweep registry and its freeze
# -------------------------------------------------------------------------------------------------------------------------------


# Build the frozen sweep registry of the study.
def build_sweep_registry(
    pairs: Sequence[Tuple[str, str]],
    reference_variants: Tuple[Tuple[str, Tuple[int, int]], ...] = REFERENCE_WINDOW_VARIANTS,
    donor_variants: Tuple[str, ...] = DONOR_POOL_VARIANTS,
    factor_variants: Tuple[str, ...] = FACTOR_PROXY_VARIANTS,
    horizon_variants: Tuple[Tuple[str, Tuple[int, int]], ...] = HORIZON_VARIANTS,
    active_year_variants: Tuple[str, ...] = ACTIVE_YEAR_VARIANTS,
) -> pd.DataFrame:
    """
    Build the frozen sweep registry of the study.

    Parameters
    ----------
    pairs : sequence of tuple
        The policy and outcome pairs the registry covers.
    reference_variants : tuple of tuple, optional
        The declared reference-window variants.
    donor_variants : tuple of str, optional
        The declared donor-pool variants.
    factor_variants : tuple of str, optional
        The declared factor-proxy variants.
    horizon_variants : tuple of tuple, optional
        The declared horizon variants.
    active_year_variants : tuple of str, optional
        The declared active-year variants.

    Returns
    -------
    pandas.DataFrame
        One row per sweep run carrying its unique identifier, its axis, its
        variant name and its variant value.

    Raises
    ------
    IngestError
        If the pair list is empty, because a registry without pairs is
        undefined.
    """
    # Refuse an empty pair list.
    if not pairs:
        # The failure states the precondition.
        raise IngestError("Cannot build a sweep registry from an empty pair list.")
    # Accumulate the registry rows.
    rows: List[Dict[str, Any]] = []
    # Iterate over the pairs in sorted order.
    for pair in sorted(set(tuple(item) for item in pairs)):
        # Register the reference-window variants.
        for index, (name, value) in enumerate(reference_variants):
            # Append the variant's row.
            rows.append(
                {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "axis": AXIS_REFERENCE_WINDOW,
                    "variant_name": name,
                    "variant_value": str(list(value)),
                    "run_id": f"{pair[0]}|{pair[1]}|{AXIS_REFERENCE_WINDOW}|{index}",
                }
            )
        # Register the donor-pool variants.
        for index, name in enumerate(donor_variants):
            # Append the variant's row.
            rows.append(
                {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "axis": AXIS_DONOR_POOL,
                    "variant_name": name,
                    "variant_value": name,
                    "run_id": f"{pair[0]}|{pair[1]}|{AXIS_DONOR_POOL}|{index}",
                }
            )
        # Register the factor-proxy variants.
        for index, name in enumerate(factor_variants):
            # Append the variant's row.
            rows.append(
                {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "axis": AXIS_FACTOR_PROXY,
                    "variant_name": name,
                    "variant_value": name,
                    "run_id": f"{pair[0]}|{pair[1]}|{AXIS_FACTOR_PROXY}|{index}",
                }
            )
        # Register the horizon variants.
        for index, (name, value) in enumerate(horizon_variants):
            # Append the variant's row.
            rows.append(
                {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "axis": AXIS_HORIZON,
                    "variant_name": name,
                    "variant_value": str(list(value)),
                    "run_id": f"{pair[0]}|{pair[1]}|{AXIS_HORIZON}|{index}",
                }
            )
        # Register the active-year variants.
        for index, name in enumerate(active_year_variants):
            # Append the variant's row.
            rows.append(
                {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "axis": AXIS_ACTIVE_YEAR,
                    "variant_name": name,
                    "variant_value": name,
                    "run_id": f"{pair[0]}|{pair[1]}|{AXIS_ACTIVE_YEAR}|{index}",
                }
            )
    # Assemble the registry frame keyed on the pair, the axis and the variant.
    registry: pd.DataFrame = pd.DataFrame(rows).set_index(
        ["policy", "outcome", "axis", "variant_name"]
    )
    # Replace ordinal identifiers with stable semantic identifiers and bind
    # every row to its registered parameter, consumer and serialization version.
    effective_fields = {
        AXIS_REFERENCE_WINDOW: "reporting.pre_escalation",
        AXIS_DONOR_POOL: "partition.donors",
        AXIS_FACTOR_PROXY: "factor_proxy.columns",
        AXIS_HORIZON: "reporting.post_escalation",
        AXIS_ACTIVE_YEAR: "reporting.country_specific_post_windows",
    }
    for key in registry.index:
        policy, outcome, axis, variant_name = key
        registry.loc[key, "run_id"] = (
            f"{policy}|{outcome}|{axis}|{variant_name}"
        )
        if axis == AXIS_ACTIVE_YEAR:
            parameter = {
                "callable_id": "entry_year_active_window",
                "parameters": {"maximum_post_year": int(WAVE_TWO_WINDOW[1])},
            }
        elif axis in {AXIS_REFERENCE_WINDOW, AXIS_HORIZON}:
            parameter = {
                "window": json.loads(str(registry.loc[key, "variant_value"]))
            }
        else:
            parameter = {"registered_value": str(registry.loc[key, "variant_value"])}
        registry.loc[key, "parameter_json"] = json.dumps(
            parameter, sort_keys=True, separators=(",", ":")
        )
        registry.loc[key, "effective_field"] = effective_fields[str(axis)]
        registry.loc[key, "consumer"] = (
            "estimation_refit"
            if axis in {AXIS_DONOR_POOL, AXIS_FACTOR_PROXY}
            else "gap_reaggregation"
        )
        registry.loc[key, "baseline_identity"] = bool(
            SWEEP_BASELINE_IDENTITIES.get(str(axis)) == str(variant_name)
        )
        registry.loc[key, "registry_version"] = SWEEP_REGISTRY_VERSION
    # Refuse a registry whose identifiers repeat, which would break the
    # cross-referencing the reconciliation task performs.
    if bool(registry["run_id"].duplicated().any()):
        # The failure states the condition.
        raise IngestError("The sweep registry carries duplicated run identifiers.")
    # Log the registry's extent.
    logger.info(
        "sweep registry built: %d run(s) across %d pair(s) and %d axis(es).",
        len(registry),
        registry.index.get_level_values(0).nunique()
        * registry.index.get_level_values(1).nunique(),
        registry.index.get_level_values(2).nunique(),
    )
    # Return the registry.
    return registry


# Freeze the sweep registry behind a canonical digest.
def freeze_sweep_registry(registry: pd.DataFrame) -> str:
    """
    Freeze the sweep registry behind a canonical digest.

    Parameters
    ----------
    registry : pandas.DataFrame
        The sweep registry.

    Returns
    -------
    str
        The hexadecimal digest of the canonical payload.

    Raises
    ------
    IngestError
        If the registry is empty.
    """
    # Refuse an empty registry.
    if len(registry) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot freeze an empty sweep registry.")
    required = {
        "run_id", "variant_value", "parameter_json", "effective_field",
        "consumer", "baseline_identity", "registry_version",
    }
    if not required.issubset(registry.columns):
        raise IngestError(
            f"Sweep registry omits canonical field(s) {sorted(required - set(registry.columns))}."
        )
    if set(registry["registry_version"].astype(str)) != {SWEEP_REGISTRY_VERSION}:
        raise IngestError("Sweep registry carries an unknown or mixed version.")
    records: List[Dict[str, Any]] = []
    for (policy, outcome, axis, variant), row in registry.sort_index().iterrows():
        try:
            parameter = json.loads(str(row["parameter_json"]))
        except (TypeError, ValueError, json.JSONDecodeError) as exc:
            raise IngestError("Sweep registry carries malformed parameter JSON.") from exc
        serialized = json.dumps(parameter, sort_keys=True, separators=(",", ":"))
        if any(token in serialized for token in ("<function", " at 0x", "<lambda")):
            raise IngestError("Sweep registry embeds a process-specific callable representation.")
        records.append(
            {
                "policy": str(policy), "outcome": str(outcome),
                "axis": str(axis), "variant_name": str(variant),
                "run_id": str(row["run_id"]),
                "variant_value": str(row["variant_value"]),
                "parameter": parameter,
                "effective_field": str(row["effective_field"]),
                "consumer": str(row["consumer"]),
                "baseline_identity": bool(row["baseline_identity"]),
                "registry_version": str(row["registry_version"]),
            }
        )
    payload = {
        "registry_version": SWEEP_REGISTRY_VERSION,
        "ordered_variants": records,
    }
    return hashlib.sha256(
        json.dumps(payload, sort_keys=True, separators=(",", ":")).encode("utf-8")
    ).hexdigest()


# -------------------------------------------------------------------------------------------------------------------------------
# Task 18, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Execute the registered sweeps and freeze their results.
def execute_sweeps(
    baseline_config: Mapping[str, Any],
    data: Mapping[str, Any],
    pairs: Optional[Sequence[Tuple[str, str]]] = None,
    run_ids: Optional[Sequence[str]] = None,
    entry_years: Optional[Mapping[Any, int]] = None,
    active_year_callable: Optional[Callable[[int], Tuple[int, int]]] = None,
) -> SweepSuiteResult:
    """
    Execute the registered sweeps and freeze their results.

    Parameters
    ----------
    baseline_config : mapping
        The baseline configuration, which is never mutated.
    data : mapping
        The frozen inputs under the declared keys.
    pairs : sequence of tuple, optional
        The pairs the registry covers. When absent every registered pair is
        used.
    run_ids : sequence of str, optional
        The identifiers of the runs to execute. When absent every registered
        run is executed.
    entry_years : mapping of str to int, optional
        The entry year of each country, required by the active-year axis.
    active_year_callable : callable, optional
        The functional specification of the active-year axis, mapping a
        country's entry year to its window.

    Returns
    -------
    SweepSuiteResult
        The full registry, one status row per selected run, successful
        effective-design evidence, retained failure ledger, completeness
        state, digest, timestamp and notes.

    Raises
    ------
    KeyError
        If the data mapping lacks a declared key.
    IngestError
        If selection/registry structure is invalid or the baseline parent is
        mutated. Scientific variant failures are retained in the suite rather
        than erased or presented as successful estimates.
    """
    # Refuse a data mapping lacking a declared key.
    for key in ("registry", "panel", "lagged"):
        # Test the key's presence.
        if key not in data:
            # The failure names the absent key.
            raise KeyError(f"The data mapping lacks the declared key {key!r}.")
    # Select the registry's pair list.
    registry_pairs: List[Tuple[str, str]] = (
        [tuple(pair) for pair in pairs]
        if pairs is not None
        else sorted(
            set(
                (str(a), str(b))
                for a, b in zip(
                    *[
                        data["registry"].registry.index.get_level_values(level)
                        for level in (0, 1)
                    ]
                )
            )
        )
    )
    # Build the sweep registry of the study.
    registry: pd.DataFrame = build_sweep_registry(registry_pairs)
    # Select the runs to execute.
    selected: pd.DataFrame = (
        registry[registry["run_id"].isin(set(str(item) for item in run_ids))]
        if run_ids is not None
        else registry
    )
    # Refuse a requested identifier that the registry does not carry.
    if run_ids is not None:
        if len(list(run_ids)) != len(set(str(item) for item in run_ids)):
            raise IngestError("Requested sweep run identifiers contain duplicates.")
        # Identify the unknown identifiers.
        unknown = sorted(set(str(item) for item in run_ids) - set(registry["run_id"]))
        # Raise when an identifier is unknown.
        if unknown:
            # The failure names the unknown identifiers.
            raise IngestError(
                f"The registry does not carry the requested run identifier(s) "
                f"{unknown[:8]}."
            )
    if selected.empty:
        raise IngestError("No registered sweep run was selected.")
    baseline_config_digest = _content_digest(baseline_config)
    parent_registry = data["registry"]
    if not isinstance(parent_registry, PartitionRegistry):
        raise IngestError("Sweep parent registry is malformed.")
    parent_registry_digest = parent_registry.fingerprint_sha256
    # Accumulate the swept results.
    rows: List[Dict[str, Any]] = []
    failures: List[Dict[str, Any]] = []
    design_rows: List[Dict[str, Any]] = []
    baseline_snapshots: Dict[str, Mapping[str, Any]] = {}
    variant_result_objects: Dict[str, Mapping[str, Any]] = {}
    variant_input_objects: Dict[str, Mapping[str, Any]] = {}
    # Cache one verified baseline fit per pair for all reporting-only variants.
    baseline_cache: Dict[Tuple[str, str], BaselineResult] = {}
    # Execute every selected run.
    for (policy, outcome, axis, variant), row in selected.iterrows():
        # Compose the specification of the run.
        if axis == AXIS_REFERENCE_WINDOW:
            # Compose the reference-window specification.
            spec = SweepSpec(
                axis=axis,
                variant_name=str(variant),
                variant_value=tuple(
                    int(item) for item in str(row["variant_value"]).strip("[]").split(",")
                ),
            )
        elif axis == AXIS_HORIZON:
            # Compose the horizon specification.
            spec = SweepSpec(
                axis=axis,
                variant_name=str(variant),
                variant_value=tuple(
                    int(item) for item in str(row["variant_value"]).strip("[]").split(",")
                ),
            )
        elif axis == AXIS_ACTIVE_YEAR:
            # Compose the active-year specification carrying the callable.
            spec = SweepSpec(
                axis=axis,
                variant_name=str(variant),
                variant_value=active_year_callable,
            )
        else:
            # Compose the variant specification of the donor and proxy axes.
            spec = SweepSpec(
                axis=axis, variant_name=str(variant), variant_value=str(row["variant_value"])
            )
        status = "success"
        failure_category = ""
        failure_detail = ""
        result: Optional[SweepResult] = None
        # Validate prerequisites before dispatching the branch itself.
        try:
            apply_sweep_variant(baseline_config, spec)
            if axis == AXIS_DONOR_POOL:
                build_variant_partition(
                    parent_registry.partitions[(str(policy), str(outcome))],
                    str(spec.variant_value),
                    data,
                )
            if axis == AXIS_FACTOR_PROXY:
                build_variant_proxy(
                    data["panel"],
                    parent_registry.partitions[(str(policy), str(outcome))].donors,
                    str(outcome),
                    str(spec.variant_value),
                )
            if axis == AXIS_ACTIVE_YEAR and (
                entry_years is None or not callable(active_year_callable)
            ):
                raise SweepUnresolvedInputError(
                    "active_year requires entry_years and its registered callable."
                )
            if axis == AXIS_ACTIVE_YEAR and active_year_callable is not entry_year_active_window:
                raise SweepUnresolvedInputError(
                    "active_year callable does not match registered callable_id "
                    "entry_year_active_window."
                )
            pair_key = (str(policy), str(outcome))
            pair_entry_years: Dict[str, int] = {}
            if axis == AXIS_ACTIVE_YEAR:
                if entry_years is not None:
                    tuple_keyed = any(isinstance(key, tuple) for key in entry_years)
                    if tuple_keyed:
                        pair_entry_years = _select_policy_entry_years(
                            entry_years, pair_key[0]
                        )
                    else:
                        pair_entry_years = {
                            str(code): int(year) for code, year in entry_years.items()
                        }
                missing_entries = sorted(
                    set(parent_registry.partitions[pair_key].targets)
                    - set(pair_entry_years)
                )
                if missing_entries:
                    raise SweepUnresolvedInputError(
                        "active_year entry_years omits target(s) "
                        f"{missing_entries[:INFERENCE_MAX_REPORTED_OFFENDERS]}."
                    )
                resolve_active_year_windows(
                    spec,
                    data["panel"],
                    parent_registry.partitions[pair_key].targets,
                    pair_entry_years,
                )
            needs_cached_baseline = axis in {
                AXIS_REFERENCE_WINDOW, AXIS_HORIZON, AXIS_ACTIVE_YEAR
            } or (
                axis == AXIS_FACTOR_PROXY and bool(row["baseline_identity"])
            )
            if needs_cached_baseline:
                if pair_key not in baseline_cache:
                    baseline_cache[pair_key] = run_baseline(
                        pair_key[0], pair_key[1], baseline_config, data
                    )
            result = run_sweep(
                str(policy), str(outcome), spec, baseline_config, data,
                pair_entry_years if axis == AXIS_ACTIVE_YEAR else entry_years,
                run_id=str(row["run_id"]),
                baseline_result=baseline_cache.get(pair_key),
            )
            status = result.scientific_status
        except SweepUnresolvedInputError as exc:
            status = "unresolved"
            failure_category = "unresolved_input"
            failure_detail = str(exc)
        except IngestError as exc:
            status = "failed"
            failure_category = "scientific_validation_failure"
            failure_detail = str(exc)
        if result is not None:
            snapshot_id = str(row["run_id"]) + "|parent"
            parent_snapshot = {
                "config_sha256": baseline_config_digest,
                "registry_sha256": parent_registry_digest,
                "panel": copy.deepcopy(data["panel"]),
                "lagged": copy.deepcopy(data["lagged"]),
            }
            baseline_snapshots[snapshot_id] = {
                "before": copy.deepcopy(parent_snapshot),
                "after": copy.deepcopy(parent_snapshot),
            }
            variant_input_objects[str(row["run_id"])] = {
                "axis": str(axis),
                "variant_name": str(variant),
                "effective_design": _thaw_sweep_evidence(result.effective_design),
            }
            variant_result_objects[str(row["run_id"])] = {
                "att": float(result.att),
                "se_cd": float(result.se_cd),
                "stars": str(result.stars),
                "scientific_status": str(result.scientific_status),
                "lineage_sha256": str(result.lineage_sha256),
            }
            design_rows.append(
                {
                    "policy": str(policy), "outcome": str(outcome),
                    "axis": str(axis), "variant_name": str(variant),
                    "run_id": result.run_id,
                    "execution_mode": result.execution_mode,
                    "lineage_sha256": result.lineage_sha256,
                    "effective_design_json": _canonical_sweep_evidence_json(
                        result.effective_design
                    ),
                }
            )
        else:
            failures.append(
                {
                    "policy": str(policy), "outcome": str(outcome),
                    "axis": str(axis), "variant_name": str(variant),
                    "run_id": str(row["run_id"]), "status": status,
                    "failure_category": failure_category,
                    "failure_detail": failure_detail,
                }
            )
        success_run_id = str(row["run_id"])
        snapshot_id = success_run_id + "|parent" if result is not None else ""
        before_digest = (
            _content_digest(baseline_snapshots[snapshot_id]["before"])
            if result is not None else ""
        )
        after_digest = (
            _content_digest(baseline_snapshots[snapshot_id]["after"])
            if result is not None else ""
        )
        variant_input_digest = (
            _content_digest(variant_input_objects[success_run_id])
            if result is not None else ""
        )
        result_digest = (
            _content_digest(variant_result_objects[success_run_id])
            if result is not None else ""
        )
        rows.append(
            {
                "policy": str(policy), "outcome": str(outcome),
                "axis": str(axis), "variant_name": str(variant),
                "run_id": str(row["run_id"]), "status": status,
                "att": float(result.att) if result is not None else np.nan,
                "se_cd": float(result.se_cd) if result is not None else np.nan,
                "stars": str(result.stars) if result is not None else "",
                "execution_mode": result.execution_mode if result is not None else "",
                "lineage_sha256": result.lineage_sha256 if result is not None else "",
                "diagnostic_notes": (
                    " | ".join(result.diagnostic_notes)
                    if result is not None else failure_detail
                ),
                "snapshot_id": snapshot_id,
                "baseline_before_sha256": before_digest,
                "baseline_after_sha256": after_digest,
                "variant_input_sha256": variant_input_digest,
                "result_sha256": result_digest,
                "execution_status": status,
            }
        )
    # Assemble the results frame keyed on the sweep identity.
    results: pd.DataFrame = pd.DataFrame(rows).set_index(
        ["policy", "outcome", "axis", "variant_name"]
    )
    results.attrs["baseline_snapshot_objects"] = baseline_snapshots
    results.attrs["variant_result_objects"] = variant_result_objects
    results.attrs["variant_input_objects"] = variant_input_objects
    # Assert that every executed run appears in the registry.
    if not set(results["run_id"]) <= set(registry["run_id"]):
        # The failure states the condition.
        raise IngestError("An executed sweep is absent from the registry.")
    # Freeze the registry.
    fingerprint: str = freeze_sweep_registry(registry)
    if _content_digest(baseline_config) != baseline_config_digest:
        raise IngestError("Sweep suite mutated its baseline configuration.")
    if data["registry"].fingerprint_sha256 != parent_registry_digest:
        raise IngestError("Sweep suite mutated or replaced its parent registry.")
    failure_table = pd.DataFrame(
        failures,
        columns=[
            "policy", "outcome", "axis", "variant_name", "run_id", "status",
            "failure_category", "failure_detail",
        ],
    )
    if not failure_table.empty:
        failure_table = failure_table.set_index(
            ["policy", "outcome", "axis", "variant_name"]
        ).sort_index()
    effective_designs = pd.DataFrame(
        design_rows,
        columns=[
            "policy", "outcome", "axis", "variant_name", "run_id",
            "execution_mode", "lineage_sha256", "effective_design_json",
        ],
    )
    if not effective_designs.empty:
        effective_designs = effective_designs.set_index(
            ["policy", "outcome", "axis", "variant_name"]
        ).sort_index()
    complete = bool(len(selected) == len(registry) and not failures)
    # Accumulate the recorded notes.
    notes: List[str] = [
        f"{len(registry)} sweep(s) registered, {int((results['status'] == 'success').sum())} "
        f"estimated, {int((results['status'] == 'identity').sum())} verified identities "
        f"and {len(failures)} remained unresolved or failed.",
        "a sweep varies exactly one effective component in a private estimation "
        "view; its synchronized window aliases have separate diff paths.",
        "the factor axis changes the factor's width, and its condition number is "
        "recorded in the diagnostic notes as the specification requires.",
    ]
    # Assemble the immutable suite result.
    suite: SweepSuiteResult = SweepSuiteResult(
        registry=registry,
        results=results,
        registry_fingerprint_sha256=fingerprint,
        completed_utc=current_utc_iso(),
        notes=tuple(notes),
        failures=failure_table,
        effective_designs=effective_designs,
        complete=complete,
    )
    # Log the completed suite.
    logger.info(
        "sweep suite complete: %d of %d registered run(s) executed, fingerprint "
        "%s.",
        len(results),
        len(registry),
        fingerprint[:16],
    )
    # Return the assembled suite.
    return suite
````

````python
# Task 19: Execute Reference, Donor, and Factor Sweeps

# ==============================================================================
# Task 19: Execute Reference, Donor, and Factor Sweeps
# ==============================================================================

# Declare the three axes this task executes.
EXECUTED_AXES: Final[Tuple[str, ...]] = (
    AXIS_REFERENCE_WINDOW,
    AXIS_DONOR_POOL,
    AXIS_FACTOR_PROXY,
)
# Declare the donor-pool variants this task executes, taken from the factory's
# single declaration so the enumeration has one source.
EXECUTED_DONOR_VARIANTS: Final[Tuple[str, ...]] = tuple(DONOR_POOL_VARIANTS)
# Declare the condition-number threshold above which a variant is classified
# numerically unstable and therefore diagnostic rather than headline.
UNSTABLE_CONDITION_THRESHOLD: Final[float] = 1.0e3
# Declare the classification label of a headline estimate.
CLASS_HEADLINE: Final[str] = "headline"
# Declare the classification label of a diagnostic estimate.
CLASS_DIAGNOSTIC: Final[str] = "diagnostic"
# Declare mutually exclusive pre-treatment factor-design states.
FACTOR_CLASS_ADMISSIBLE: Final[str] = "admissible"
FACTOR_CLASS_UNSTABLE: Final[str] = "unstable_finite"
FACTOR_CLASS_RANK_DEFICIENT: Final[str] = "rank_deficient"
FACTOR_CLASS_FAILED: Final[str] = "failed"
# Declare the matrix and scale labels of the Task 19 diagnostic.
FACTOR_DIAGNOSTIC_MATRIX: Final[str] = "pre_treatment_F_transpose_F"
FACTOR_DIAGNOSTIC_SCALE: Final[str] = "raw_factor_cross_product_singular_values"
# Declare the factor widths of the declared variants.
FACTOR_WIDTH_TWO: Final[int] = 2
# Declare the declared width of the augmented factor variants.
FACTOR_WIDTH_THREE: Final[int] = 3


# Declare the frozen result of the three-axis execution.
@dataclass(frozen=True)
class SweepExecutionResult:
    """
    Frozen result of the reference, donor and factor executions.

    Attributes
    ----------
    reference_table : pandas.DataFrame
        One row per partition and reference window carrying the effect, its
        standard error and its marker.
    donor_table : pandas.DataFrame
        One row per partition and donor variant carrying the same quantities.
    factor_table : pandas.DataFrame
        One row per partition and factor variant carrying the same quantities,
        the condition number of the variant's factor cross-product and the
        classification of the estimate as headline or diagnostic.
    ranges : pandas.DataFrame
        One row per partition and axis carrying the minimum and maximum effect
        across the axis's variants, with the diagnostic variants excluded from
        the headline range of the factor axis.
    donor_audit : pandas.DataFrame
        One row per pair and donor variant, recording actual baseline/variant
        identity sets, additions, removals and restriction evidence.
    stability_summary : pandas.DataFrame
        One row per partition and factor variant recording the condition
        number and the stability flag.
    fingerprint_sha256 : str
        Digest of the canonical execution payload.
    completed_utc : str
        ISO-8601 UTC timestamp of the execution.
    notes : tuple of str
        Recorded decisions, including the executed and diagnosed counts.
    coverage_audit : pandas.DataFrame
        Every registered Task 19 run and its execution/audit status, including
        unresolved and failed branches that are excluded from ranges.
    """
    # One row per partition and reference window.
    reference_table: pd.DataFrame
    # One row per partition and donor variant.
    donor_table: pd.DataFrame
    # One row per partition and factor variant with its diagnostics.
    factor_table: pd.DataFrame
    # One row per partition and axis carrying the variant range.
    ranges: pd.DataFrame
    # One row per pair and donor variant with exact roster evidence.
    donor_audit: pd.DataFrame
    # One row per partition and factor variant recording the stability flag.
    stability_summary: pd.DataFrame
    # Digest of the canonical execution payload.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the execution.
    completed_utc: str
    # Recorded decisions of the execution.
    notes: Tuple[str, ...]
    # Complete success/unresolved/failure coverage of the three axes.
    coverage_audit: pd.DataFrame = field(default_factory=pd.DataFrame)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 19, Step 1: The reference-window sweep
# -------------------------------------------------------------------------------------------------------------------------------


# Split one axis's runs out of the executed suite.
def split_axis_results(
    suite: SweepSuiteResult, axis: str, allow_incomplete: bool = False
) -> pd.DataFrame:
    """
    Split one axis's runs out of the executed suite.

    Parameters
    ----------
    suite : SweepSuiteResult
        The executed sweep suite.
    axis : str
        The axis whose runs are selected.
    allow_incomplete : bool, optional
        Whether unresolved/failed registered rows are retained for an
        explicit coverage report.  The default remains fail-closed for
        publication consumers.

    Returns
    -------
    pandas.DataFrame
        The axis's results indexed on the policy, the outcome and the variant
        name, carrying the effect, the standard error, the marker and the
        diagnostic notes.

    Raises
    ------
    IngestError
        If the suite carries no run for the axis, because an empty axis would
        silently produce an empty table.
    """
    # Select the axis's rows from the suite.
    axis_rows = suite.results[
        suite.results.index.get_level_values("axis") == axis
    ]
    # Refuse an axis without runs.
    if len(axis_rows) == 0:
        # The failure names the axis.
        raise IngestError(f"The executed suite carries no run for the axis {axis!r}.")
    if not suite.registry.empty and "axis" in suite.registry.index.names:
        registered_axis = suite.registry[
            suite.registry.index.get_level_values("axis") == axis
        ]
        expected_keys = set(registered_axis.index.tolist())
        observed_keys = set(axis_rows.index.tolist())
        if observed_keys != expected_keys:
            missing = sorted(expected_keys - observed_keys, key=str)
            extra = sorted(observed_keys - expected_keys, key=str)
            raise IngestError(
                f"Axis {axis!r} execution coverage is partial; "
                f"missing={missing[:INFERENCE_MAX_REPORTED_OFFENDERS]}, "
                f"unexpected={extra[:INFERENCE_MAX_REPORTED_OFFENDERS]}."
            )
    # Refuse to publish a complete-looking axis table when Task 18 retained an
    # unresolved or failed registered variant.
    if "status" in axis_rows.columns and not allow_incomplete:
        incomplete = axis_rows[
            ~axis_rows["status"].astype(str).isin({"success", "identity"})
        ]
        if not incomplete.empty:
            offenders = [
                f"{row['run_id']}:{row['status']}"
                for _, row in incomplete.head(INFERENCE_MAX_REPORTED_OFFENDERS).iterrows()
            ]
            raise IngestError(
                f"Axis {axis!r} is incomplete; unresolved/failed run(s) {offenders}."
            )
    # Validate numerical fields only on rows Task 18 marked successful.
    successful = (
        axis_rows[axis_rows["status"].astype(str) == "success"]
        if "status" in axis_rows.columns else axis_rows
    )
    verified = axis_rows[
        axis_rows["status"].astype(str).isin({"success", "identity"})
    ] if "status" in axis_rows.columns else axis_rows
    if not np.isfinite(verified[["att", "se_cd"]].to_numpy(dtype=float)).all():
        raise IngestError(f"Axis {axis!r} carries a non-finite executed result.")
    # Carry complete effective-design evidence into the Task 19 audit tables.
    enriched = axis_rows.copy()
    if not suite.effective_designs.empty:
        design_axis = suite.effective_designs[
            suite.effective_designs.index.get_level_values("axis") == axis
        ]
        if "effective_design_json" in design_axis.columns:
            enriched = enriched.join(
                design_axis[["effective_design_json"]], how="left"
            )
    # Return the axis's rows with the axis level dropped for readability.
    return enriched.droplevel("axis")


# Reduce one axis's effects to their range per partition.
def build_axis_ranges(
    table: pd.DataFrame,
    axis: str,
    classify_column: Optional[str] = None,
    headline_label: str = CLASS_HEADLINE,
) -> pd.DataFrame:
    """
    Reduce one axis's effects to their range per partition.

    Parameters
    ----------
    table : pandas.DataFrame
        The axis's results indexed on the policy, the outcome and the variant
        name.
    axis : str
        The axis label, recorded in the output.
    classify_column : str, optional
        The column carrying the headline-or-diagnostic classification. When
        supplied, the diagnostic variants are excluded from the headline
        range and reported separately.
    headline_label : str, optional
        The classification label of a headline estimate.

    Returns
    -------
    pandas.DataFrame
        One row per partition carrying the minimum and maximum effect of the
        headline variants and the count of diagnostic variants excluded.

    Raises
    ------
    IngestError
        If the table is empty or a partition retains no headline variant.
    """
    # Refuse an empty table.
    if len(table) == 0:
        # The failure names the axis.
        raise IngestError(f"Cannot range an empty table for the axis {axis!r}.")
    # Refuse duplicate scientific keys before any reduction.
    if table.index.has_duplicates or table.index.nlevels != 3:
        raise IngestError(f"Axis {axis!r} table has duplicate or malformed keys.")
    # Accumulate the range rows.
    rows: List[Dict[str, Any]] = []
    # Group the table by partition without discarding failed branches.
    for (policy, outcome), group in table.groupby(level=[0, 1], observed=True):
        group = group.sort_index()
        statuses = (
            group["status"].astype(str)
            if "status" in group.columns
            else pd.Series("success", index=group.index, dtype=object)
        )
        successful = group[statuses == "success"]
        # Apply the factor admissibility classification only after execution
        # status has been resolved.
        if classify_column is not None:
            if classify_column not in group.columns:
                raise IngestError(
                    f"Axis {axis!r} lacks classification column {classify_column!r}."
                )
            headline = successful[
                successful[classify_column].astype(str) == str(headline_label)
            ]
        else:
            headline = successful
        # Compose exact variant identities for complete coverage reporting.
        all_variants = [str(index[-1]) for index in group.index]
        successful_variants = [str(index[-1]) for index in successful.index]
        identity_variants = [
            str(index[-1]) for index in group.index
            if str(statuses.loc[index]) == "identity"
        ]
        headline_variants = [str(index[-1]) for index in headline.index]
        failed_variants = [
            str(index[-1]) for index in group.index
            if str(statuses.loc[index]) not in {"success", "identity"}
        ]
        audit_failed_variants = (
            [
                str(index[-1]) for index, row in successful.iterrows()
                if str(row.get("factor_design_class", "")) == FACTOR_CLASS_FAILED
            ]
            if classify_column is not None else []
        )
        unstable_variants = (
            [
                str(index[-1]) for index, row in successful.iterrows()
                if str(row.get("factor_design_class", "")) == FACTOR_CLASS_UNSTABLE
            ]
            if classify_column is not None else []
        )
        rank_deficient_variants = (
            [
                str(index[-1]) for index, row in successful.iterrows()
                if str(row.get("factor_design_class", ""))
                == FACTOR_CLASS_RANK_DEFICIENT
            ]
            if classify_column is not None else []
        )
        excluded_variants = sorted(set(all_variants) - set(headline_variants))
        # Retain full-precision estimates for exact declared-subset checks.
        estimate_map = {
            str(index[-1]): float(row["att"])
            for index, row in headline.iterrows()
        }
        ordered_endpoints = sorted(
            estimate_map.items(), key=lambda item: (item[1], item[0])
        )
        if ordered_endpoints:
            minimum_variant, minimum_att = ordered_endpoints[0]
            maximum_variant, maximum_att = ordered_endpoints[-1]
        else:
            minimum_variant, maximum_variant = "", ""
            minimum_att, maximum_att = float("nan"), float("nan")
        # Recover the fixed reference-window identity where the axis does not
        # itself vary that reference.
        reference_values: Set[str] = set()
        if "effective_design_json" in successful.columns:
            for raw in successful["effective_design_json"].dropna().astype(str):
                try:
                    evidence = json.loads(raw)
                except json.JSONDecodeError as exc:
                    raise IngestError(
                        f"Axis {axis!r} carries malformed effective-design JSON."
                    ) from exc
                span = evidence.get("effective_pre_window")
                if span is None:
                    span = evidence.get("baseline_windows", {}).get("pre_escalation")
                if span is not None:
                    reference_values.add(json.dumps(list(span), separators=(",", ":")))
        reference_identity = (
            "varied" if axis == AXIS_REFERENCE_WINDOW
            else (next(iter(reference_values)) if len(reference_values) == 1 else "unresolved")
        )
        # Record both the headline range and every excluded/failed identity so
        # a narrowed range can never masquerade as complete coverage.
        rows.append(
            {
                "policy": str(policy),
                "outcome": str(outcome),
                "axis": axis,
                "reference_window": reference_identity,
                "variants": int(len(group)),
                "successful_variants": int(len(successful)),
                "identity_variants": int(len(identity_variants)),
                "headline_variants": int(len(headline)),
                "diagnostic_variants": int(len(successful) - len(headline)),
                "failed_variants": int(len(failed_variants)),
                "audit_failed_variants": int(len(audit_failed_variants)),
                "unstable_variants": int(len(unstable_variants)),
                "rank_deficient_variants": int(len(rank_deficient_variants)),
                "minimum_att": float(minimum_att),
                "maximum_att": float(maximum_att),
                "range_width": float(maximum_att - minimum_att),
                "minimum_variant": minimum_variant,
                "maximum_variant": maximum_variant,
                "all_variants_json": json.dumps(all_variants, separators=(",", ":")),
                "successful_variants_json": json.dumps(
                    successful_variants, separators=(",", ":")
                ),
                "admissible_variants_json": json.dumps(
                    headline_variants, separators=(",", ":")
                ),
                "excluded_variants_json": json.dumps(
                    excluded_variants, separators=(",", ":")
                ),
                "failed_variants_json": json.dumps(
                    failed_variants, separators=(",", ":")
                ),
                "identity_variants_json": json.dumps(
                    identity_variants, separators=(",", ":")
                ),
                "audit_failed_variants_json": json.dumps(
                    audit_failed_variants, separators=(",", ":")
                ),
                "unstable_variants_json": json.dumps(
                    unstable_variants, separators=(",", ":")
                ),
                "rank_deficient_variants_json": json.dumps(
                    rank_deficient_variants, separators=(",", ":")
                ),
                "variant_estimates_json": json.dumps(
                    estimate_map, sort_keys=True, separators=(",", ":"),
                    allow_nan=False,
                ),
            }
        )
    # Assemble the range frame keyed on the partition and the axis.
    ranges: pd.DataFrame = pd.DataFrame(rows).set_index(["policy", "outcome", "axis"])
    # Log the reduction's extent.
    logger.info(
        "%s: ranges reduced across %d partition(s).", axis, len(ranges)
    )
    # Return the range frame.
    return ranges


# -------------------------------------------------------------------------------------------------------------------------------
# Task 19, Step 2: The donor-pool sweep
# -------------------------------------------------------------------------------------------------------------------------------


# Audit the donor list of every donor variant.
def build_donor_audit(
    data: Mapping[str, Any],
    pairs: Sequence[Tuple[str, str]],
    variants: Sequence[str] = EXECUTED_DONOR_VARIANTS,
    executed_table: Optional[pd.DataFrame] = None,
    retain_failures: bool = False,
) -> pd.DataFrame:
    """
    Audit the donor list of every donor variant.

    Parameters
    ----------
    data : mapping
        The frozen inputs under the declared keys.
    pairs : sequence of tuple
        The pairs whose donor variants are audited.
    variants : sequence of str, optional
        The donor variants to audit.

    Returns
    -------
    pandas.DataFrame
        One row per pair and donor variant recording baseline/variant identity
        sets, exact additions/removals, restriction evidence and verification.

    Raises
    ------
    KeyError
        If the data mapping lacks the registry.
    IngestError
        If a variant's partition cannot be built, which the builder refuses
        when the pool falls below its declared minimum.
    """
    # Read and validate the registry from the frozen data.
    if "registry" not in data or not isinstance(data["registry"], PartitionRegistry):
        raise KeyError("The data mapping lacks a valid partition registry.")
    registry = data["registry"]
    declared_variants = set(EXECUTED_DONOR_VARIANTS)
    normalized_variants = tuple(str(variant) for variant in variants)
    if (
        not normalized_variants
        or len(normalized_variants) != len(set(normalized_variants))
        or not set(normalized_variants) <= declared_variants
    ):
        raise IngestError("Donor audit variants are empty, duplicate or undeclared.")
    try:
        normalized_pairs = tuple(
            (str(item[0]), str(item[1])) for item in pairs if len(item) == 2
        )
    except (TypeError, IndexError) as exc:
        raise IngestError("Donor audit pair registry is malformed.") from exc
    if (
        not normalized_pairs
        or len(normalized_pairs) != len(tuple(pairs))
        or len(normalized_pairs) != len(set(normalized_pairs))
    ):
        raise IngestError("Donor audit pairs are empty, duplicate or malformed.")
    # Accumulate the audit rows.
    rows: List[Dict[str, Any]] = []
    # Iterate over the pairs in sorted order.
    for pair in sorted(normalized_pairs):
        # Read the pair's baseline partition.
        if pair not in registry.partitions:
            raise IngestError(f"Donor audit registry omits pair {pair}.")
        partition: Partition = registry.partitions[pair]
        baseline_set = set(str(code) for code in partition.donors)
        if len(baseline_set) != len(partition.donors):
            raise IngestError(f"{pair}: baseline donor roster contains duplicates.")
        # Iterate over the declared variants.
        for variant in normalized_variants:
            try:
                # Build the actual roster through the authoritative Task 18
                # factory rather than copying counts from its registry row.
                variant_partition, variant_notes = build_variant_partition(
                    partition, variant, data
                )
                variant_set = set(str(code) for code in variant_partition.donors)
                if len(variant_set) != len(variant_partition.donors):
                    raise IngestError(f"{pair}/{variant}: duplicate variant donors.")
                additions = tuple(sorted(variant_set - baseline_set))
                removals = tuple(sorted(baseline_set - variant_set))
                if additions:
                    raise IngestError(
                        f"{pair}/{variant}: restriction unexpectedly adds {additions}."
                    )
                # Independently reconstruct the named restriction's evidence.
                if variant == "corrected_full_pool":
                    expected_set = set(baseline_set)
                    evidence_source = "registered_baseline_partition"
                elif variant == "no_repaired_main_outcome_cells":
                    repaired = data.get("repaired_donor_codes")
                    if isinstance(repaired, Mapping):
                        repaired = repaired.get(pair)
                    if repaired is None or isinstance(repaired, (str, bytes, Mapping)):
                        raise SweepUnresolvedInputError(
                            f"{pair}/{variant}: repaired-donor evidence is absent."
                        )
                    repaired_codes = tuple(str(code).strip() for code in repaired)
                    if (
                        any(not code for code in repaired_codes)
                        or len(repaired_codes) != len(set(repaired_codes))
                        or not set(repaired_codes) <= baseline_set
                    ):
                        raise IngestError(
                            f"{pair}/{variant}: repaired-donor evidence contradicts the baseline roster."
                        )
                    expected_set = baseline_set - set(repaired_codes)
                    evidence_source = "repaired_donor_codes"
                else:
                    income = data.get("income_groups")
                    if not isinstance(income, Mapping) or not income:
                        raise SweepUnresolvedInputError(
                            f"{pair}/{variant}: income classification evidence is absent."
                        )
                    normalized_income = {
                        str(code).strip(): str(group).strip().lower()
                        for code, group in income.items()
                        if code is not None and group is not None
                    }
                    if not baseline_set <= set(normalized_income):
                        raise SweepUnresolvedInputError(
                            f"{pair}/{variant}: income evidence omits baseline donors."
                        )
                    high_labels = {
                        "high income", "high-income", "high_income",
                        "high income: oecd", "high income: nonoecd",
                    }
                    expected_set = {
                        code for code in baseline_set
                        if normalized_income.get(code) in high_labels
                    }
                    evidence_source = "income_groups"
                if variant_set != expected_set:
                    raise IngestError(
                        f"{pair}/{variant}: effective donor roster contradicts "
                        f"{evidence_source}."
                    )
                if variant == "high_income_donors" and not removals:
                    raise IngestError(
                        f"{pair}/{variant}: claimed restriction is an effective no-op."
                    )
                # Require exact exclusion-audit evidence for every actual removal.
                exclusions = variant_partition.exclusions
                reason = f"sweep_{variant}"
                recorded_removed = set(
                    exclusions.loc[
                        exclusions["reason"].astype(str) == reason, "country_code"
                    ].astype(str)
                )
                if recorded_removed != set(removals):
                    raise IngestError(
                        f"{pair}/{variant}: exclusion evidence does not equal removals."
                    )
                # Cross-check the roster retained by the actual executed sweep.
                if executed_table is not None:
                    executed_key = (pair[0], pair[1], variant)
                    if executed_key not in executed_table.index:
                        raise IngestError(
                            f"{pair}/{variant}: executed donor result is absent."
                        )
                    executed_row = executed_table.loc[executed_key]
                    executed_status = str(executed_row.get("status", "success"))
                    if executed_status == "unresolved":
                        raise SweepUnresolvedInputError(
                            f"{pair}/{variant}: executed donor branch is unresolved."
                        )
                    verified_identity = (
                        variant == "no_repaired_main_outcome_cells" and not removals
                    )
                    if executed_status not in ({"identity"} if verified_identity else {"success"}):
                        raise IngestError(
                            f"{pair}/{variant}: executed donor status contradicts the audited restriction."
                        )
                    try:
                        executed_design = json.loads(
                            str(executed_row["effective_design_json"])
                        )
                    except (KeyError, TypeError, json.JSONDecodeError) as exc:
                        raise IngestError(
                            f"{pair}/{variant}: executed donor design evidence is malformed."
                        ) from exc
                    if set(str(code) for code in executed_design.get("effective_donors", ())) != variant_set:
                        raise IngestError(
                            f"{pair}/{variant}: executed effective donor roster disagrees with audit."
                        )
                rows.append(
                    {
                        "policy": pair[0], "outcome": pair[1],
                        "donor_variant": variant,
                        "audit_status": (
                            "verified_identity"
                            if variant == "no_repaired_main_outcome_cells" and not removals
                            else "verified"
                        ),
                        "baseline_count": len(baseline_set),
                        "variant_count": len(variant_set),
                        "baseline_donors_json": json.dumps(sorted(baseline_set), separators=(",", ":")),
                        "variant_donors_json": json.dumps(sorted(variant_set), separators=(",", ":")),
                        "added_donors_json": json.dumps(list(additions), separators=(",", ":")),
                        "removed_donors_json": json.dumps(list(removals), separators=(",", ":")),
                        "evidence_source": evidence_source,
                        "restriction_verified": bool(removals) or variant == "corrected_full_pool",
                        "variant_note": " | ".join(variant_notes),
                        "failure_detail": "",
                    }
                )
            except IngestError as exc:
                if not retain_failures:
                    raise
                rows.append(
                    {
                        "policy": pair[0], "outcome": pair[1],
                        "donor_variant": variant,
                        "audit_status": (
                            "unresolved" if isinstance(exc, SweepUnresolvedInputError)
                            else "failed"
                        ),
                        "baseline_count": len(baseline_set),
                        "variant_count": np.nan,
                        "baseline_donors_json": json.dumps(sorted(baseline_set), separators=(",", ":")),
                        "variant_donors_json": "[]", "added_donors_json": "[]",
                        "removed_donors_json": "[]", "evidence_source": "unresolved",
                        "restriction_verified": False, "variant_note": "",
                        "failure_detail": str(exc),
                    }
                )
    # Assemble the uniquely keyed audit frame, one row per actual roster.
    audit: pd.DataFrame = pd.DataFrame(rows).set_index(
        ["policy", "outcome", "donor_variant"]
    )
    # Log the audit's extent.
    logger.info(
        "donor audit recorded across %d donor entries.", len(audit)
    )
    # Return the audit frame.
    return audit


# -------------------------------------------------------------------------------------------------------------------------------
# Task 19, Step 3: The factor-proxy sweep
# -------------------------------------------------------------------------------------------------------------------------------


# Compute the condition number of one factor variant's cross-product.
def compute_variant_condition_number(
    data: Mapping[str, Any],
    policy_name: str,
    outcome_name: str,
    variant: str,
) -> Dict[str, Any]:
    """
    Compute the condition number of one factor variant's cross-product.

    Parameters
    ----------
    data : mapping
        The frozen inputs under the declared keys.
    policy_name : str
        The policy instrument of the pair.
    outcome_name : str
        The canonical outcome of the pair.
    variant : str
        The factor-proxy variant name.

    Returns
    -------
    dict
        The condition number, singular values, numerical rank, tolerance,
        exact factor columns, period, matrix and scale labels, and the digest
        of the fitted pre-treatment factor matrix.

    Raises
    ------
    IngestError
        If the variant proxy cannot be built for the pair's donors.
    """
    # Read the registry and the pair's partition.
    if "registry" not in data or "panel" not in data:
        raise IngestError("Factor diagnostic requires registry and panel inputs.")
    registry = data["registry"]
    pair = (str(policy_name), str(outcome_name))
    if not isinstance(registry, PartitionRegistry) or pair not in registry.partitions:
        raise IngestError(f"Factor diagnostic registry omits pair {pair}.")
    partition: Partition = registry.partitions[pair]
    # Build the variant proxy columns of the pair.
    proxy: pd.DataFrame = build_variant_proxy(
        data["panel"], partition.donors, outcome_name, str(variant)
    )
    # Extract exactly the fitted 2001--2013 rows; later outcomes must not enter
    # this diagnostic under any circumstances.
    pre_years = list(
        range(int(PRE_TREATMENT_WINDOW[0]), int(PRE_TREATMENT_WINDOW[1]) + 1)
    )
    if proxy.index.has_duplicates or proxy.columns.has_duplicates:
        raise IngestError("Factor proxy has duplicate year or column labels.")
    pre_proxy = proxy.reindex(index=pre_years)
    if list(pre_proxy.index) != pre_years or bool(pre_proxy.isna().to_numpy().any()):
        raise IngestError("Factor proxy does not cover exactly 2001--2013.")
    matrix = pre_proxy.to_numpy(dtype=float)
    if matrix.ndim != 2 or matrix.shape[0] != len(pre_years) or not np.isfinite(matrix).all():
        raise IngestError("Pre-treatment factor matrix is malformed or non-finite.")
    # Retain F's own default-SVD rank separately because squaring its singular
    # values in F'F can make a near-collinear direction cross the registered
    # cross-product tolerance even when NumPy's direct rank(F) retains it.
    factor_singular_values = np.linalg.svd(matrix, compute_uv=False)
    factor_rank_tolerance = float(
        float(factor_singular_values[0]) * max(matrix.shape) * np.finfo(float).eps
    )
    factor_matrix_rank = int(
        np.count_nonzero(factor_singular_values > factor_rank_tolerance)
    )
    # Compute diagnostics on the fitted matrix F'F, at its own unstandardized
    # cross-product scale, as required by the manuscript table.
    cross_product = matrix.T @ matrix
    singular_values = np.linalg.svd(cross_product, compute_uv=False)
    if singular_values.size != matrix.shape[1]:
        raise IngestError("Factor cross-product singular-value coverage is invalid.")
    largest = float(singular_values[0])
    rank_tolerance = float(
        largest * max(cross_product.shape) * np.finfo(float).eps
    )
    rank = int(np.count_nonzero(singular_values > rank_tolerance))
    rank_deficient = bool(rank < cross_product.shape[0])
    condition = (
        float("inf")
        if rank_deficient
        else float(largest / float(singular_values[-1]))
    )
    # Return a fully labelled audit rather than an ambiguous all-period scalar.
    return {
        "condition_number": condition,
        "singular_values_json": json.dumps(
            [float(value) for value in singular_values], separators=(",", ":")
        ),
        "rank": rank,
        "full_rank": int(cross_product.shape[0]),
        "rank_deficient": rank_deficient,
        "rank_tolerance": rank_tolerance,
        "factor_matrix_rank": factor_matrix_rank,
        "factor_matrix_rank_tolerance": factor_rank_tolerance,
        "factor_rows": int(matrix.shape[0]),
        "factor_width": int(matrix.shape[1]),
        "factor_columns_json": json.dumps(
            [str(column) for column in pre_proxy.columns], separators=(",", ":")
        ),
        "period_start": int(PRE_TREATMENT_WINDOW[0]),
        "period_end": int(PRE_TREATMENT_WINDOW[1]),
        "matrix_label": FACTOR_DIAGNOSTIC_MATRIX,
        "scale_label": FACTOR_DIAGNOSTIC_SCALE,
        "rank_convention": SVD_RANK_CONVENTION,
        "factor_sha256": _content_digest(pre_proxy),
        "full_proxy_sha256": _content_digest(proxy),
    }


# Classify each factor variant's estimate as headline or diagnostic.
def classify_factor_variants(
    data: Mapping[str, Any],
    table: pd.DataFrame,
    threshold: float = UNSTABLE_CONDITION_THRESHOLD,
) -> pd.DataFrame:
    """
    Classify each factor variant's estimate as headline or diagnostic.

    Parameters
    ----------
    data : mapping
        The frozen inputs under the declared keys.
    table : pandas.DataFrame
        The factor axis's results indexed on the policy, the outcome and the
        variant name.
    threshold : float, optional
        The condition-number threshold above which an estimate is diagnostic.

    Returns
    -------
    pandas.DataFrame
        The table with the condition number and the classification attached.

    Raises
    ------
    IngestError
        If the table is empty.
    """
    # Refuse an empty/duplicated table and an invalid warning threshold.
    if len(table) == 0 or table.index.has_duplicates:
        raise IngestError("Cannot classify an empty or duplicate-keyed factor table.")
    try:
        threshold_value = float(threshold)
    except (TypeError, ValueError, OverflowError) as exc:
        raise IngestError("Factor warning threshold must be numeric.") from exc
    if (
        isinstance(threshold, (bool, np.bool_))
        or not np.isfinite(threshold_value)
        or threshold_value <= 0.0
    ):
        raise IngestError("Factor warning threshold must be finite and positive.")
    # Copy the table so the caller's object is never mutated.
    classified: pd.DataFrame = table.copy()
    diagnostic_rows: List[Dict[str, Any]] = []
    # Iterate over every registered row, including Task 18 failures.
    for (policy, outcome, variant), row in classified.iterrows():
        status = str(row.get("status", "success"))
        if status != "success":
            diagnostic_rows.append(
                {
                    "condition_number": np.nan, "singular_values_json": "[]",
                    "rank": np.nan, "full_rank": np.nan,
                    "rank_deficient": False, "rank_tolerance": np.nan,
                    "factor_matrix_rank": np.nan,
                    "factor_matrix_rank_tolerance": np.nan,
                    "factor_rows": np.nan, "factor_width": np.nan,
                    "factor_columns_json": "[]", "period_start": PRE_TREATMENT_WINDOW[0],
                    "period_end": PRE_TREATMENT_WINDOW[1],
                    "matrix_label": FACTOR_DIAGNOSTIC_MATRIX,
                    "scale_label": FACTOR_DIAGNOSTIC_SCALE,
                    "rank_convention": SVD_RANK_CONVENTION,
                    "factor_sha256": "", "full_proxy_sha256": "",
                    "factor_design_class": FACTOR_CLASS_FAILED,
                    "estimate_class": CLASS_DIAGNOSTIC,
                    "numerically_stable": False,
                    "classification_reason": (
                        f"Task 18 execution status={status}: "
                        f"{str(row.get('diagnostic_notes', ''))}"
                    ),
                }
            )
            continue
        try:
            diagnostic = compute_variant_condition_number(
                data, str(policy), str(outcome), str(variant)
            )
            # Bind the independently rebuilt factor to the executed design.
            if "effective_design_json" in row and pd.notna(row["effective_design_json"]):
                executed_design = json.loads(str(row["effective_design_json"]))
                executed_columns = [
                    str(column) for column in executed_design.get("factor_columns", ())
                ]
                observed_columns = json.loads(str(diagnostic["factor_columns_json"]))
                if executed_columns != observed_columns:
                    raise IngestError(
                        "Executed factor columns disagree with independent diagnostic."
                    )
                if str(executed_design.get("factor_sha256", "")) != str(
                    diagnostic["full_proxy_sha256"]
                ):
                    raise IngestError(
                        "Executed factor digest disagrees with independent diagnostic."
                    )
                if int(executed_design.get("factor_pre_rank", -1)) != int(
                    diagnostic["factor_matrix_rank"]
                ):
                    raise IngestError(
                        "Executed direct-F rank disagrees with independent diagnostic."
                    )
            condition = float(diagnostic["condition_number"])
            if bool(diagnostic["rank_deficient"]):
                design_class = FACTOR_CLASS_RANK_DEFICIENT
                reason = (
                    f"pre-treatment F'F rank {int(diagnostic['rank'])} is below "
                    f"width {int(diagnostic['full_rank'])}."
                )
            elif not np.isfinite(condition):
                raise IngestError("Full-rank factor condition number is non-finite.")
            elif condition > threshold_value:
                design_class = FACTOR_CLASS_UNSTABLE
                reason = (
                    f"finite pre-treatment F'F condition {condition:.17g} exceeds "
                    f"warning threshold {threshold_value:.17g}."
                )
            else:
                design_class = FACTOR_CLASS_ADMISSIBLE
                reason = (
                    f"full-rank pre-treatment F'F condition {condition:.17g} is "
                    f"within threshold {threshold_value:.17g}."
                )
            diagnostic_rows.append(
                {
                    **diagnostic,
                    "factor_design_class": design_class,
                    "estimate_class": (
                        CLASS_HEADLINE
                        if design_class == FACTOR_CLASS_ADMISSIBLE
                        else CLASS_DIAGNOSTIC
                    ),
                    "numerically_stable": design_class == FACTOR_CLASS_ADMISSIBLE,
                    "classification_reason": reason,
                }
            )
        except (IngestError, KeyError, TypeError, ValueError, json.JSONDecodeError) as exc:
            diagnostic_rows.append(
                {
                    "condition_number": np.nan, "singular_values_json": "[]",
                    "rank": np.nan, "full_rank": np.nan,
                    "rank_deficient": False, "rank_tolerance": np.nan,
                    "factor_matrix_rank": np.nan,
                    "factor_matrix_rank_tolerance": np.nan,
                    "factor_rows": np.nan, "factor_width": np.nan,
                    "factor_columns_json": "[]", "period_start": PRE_TREATMENT_WINDOW[0],
                    "period_end": PRE_TREATMENT_WINDOW[1],
                    "matrix_label": FACTOR_DIAGNOSTIC_MATRIX,
                    "scale_label": FACTOR_DIAGNOSTIC_SCALE,
                    "rank_convention": SVD_RANK_CONVENTION,
                    "factor_sha256": "", "full_proxy_sha256": "",
                    "factor_design_class": FACTOR_CLASS_FAILED,
                    "estimate_class": CLASS_DIAGNOSTIC,
                    "numerically_stable": False,
                    "classification_reason": f"factor diagnostic failed: {exc}",
                }
            )
    # Attach the complete diagnostics in the original registered row order.
    diagnostics = pd.DataFrame(diagnostic_rows, index=classified.index)
    for column in diagnostics.columns:
        classified[column] = diagnostics[column]
    # Log the classification outcome.
    logger.info(
        "factor variants classified: %d admissible, %d diagnostic/failed.",
        int((classified["estimate_class"] == CLASS_HEADLINE).sum()),
        int((classified["estimate_class"] == CLASS_DIAGNOSTIC).sum()),
    )
    # Return the classified table.
    return classified


# Compare the observed ranges with the declared study ranges.
def assert_declared_ranges(
    ranges: pd.DataFrame,
    expected: Optional[Mapping[Tuple[Any, ...], Any]] = None,
    tolerance: float = 0.0,
) -> None:
    """
    Compare the observed ranges with the declared study ranges.

    Parameters
    ----------
    ranges : pandas.DataFrame
        The range summary keyed on the partition and the axis.
    expected : mapping, optional
        Published cells keyed by ``(policy, outcome, axis, exact_variants)``.
        Values are either ``(minimum, maximum)`` or mappings that additionally
        declare ``precision``, ``allowance``, ``reference_window`` and endpoint
        variant identities.
    tolerance : float, optional
        An explicit finite nonnegative numerical allowance in addition to the
        rounding interval. Boolean values are refused.

    Returns
    -------
    None
        The function returns nothing when every supplied comparison holds.

    Raises
    ------
    IngestError
        If an observed bound diverges from its declared counterpart, or if a
        declared partition is absent from the summary.
    """
    # Validate the optional numerical allowance before any comparison.
    try:
        tolerance_value = float(tolerance)
    except (TypeError, ValueError, OverflowError) as exc:
        raise IngestError("Declared-range allowance must be numeric.") from exc
    if (
        isinstance(tolerance, (bool, np.bool_))
        or not np.isfinite(tolerance_value)
        or tolerance_value < 0.0
    ):
        raise IngestError("Declared-range allowance must be finite and nonnegative.")
    # Skip the comparison when no declared ranges were supplied.
    if expected is None:
        # Log the observed ranges for the evidence record.
        logger.info(
            "declared ranges not supplied; observed ranges: %s.",
            ranges[["minimum_att", "maximum_att"]].to_dict(orient="index"),
        )
        # Return without comparison.
        return
    required_columns = {
        "reference_window", "all_variants_json", "admissible_variants_json",
        "variant_estimates_json", "minimum_variant", "maximum_variant",
    }
    if not required_columns <= set(ranges.columns):
        raise IngestError("Range summary lacks exact variant/endpoint provenance.")
    # Compare every published cell against its one exact scientific universe.
    for cell_key, declaration in expected.items():
        if not isinstance(cell_key, tuple) or len(cell_key) != 4:
            raise IngestError(
                "Declared ranges must be keyed by "
                "(policy, outcome, axis, exact_variant_subset)."
            )
        policy, outcome, axis, raw_subset = cell_key
        if isinstance(raw_subset, (str, bytes)) or not isinstance(
            raw_subset, (tuple, list, set, frozenset)
        ):
            raise IngestError("Declared range variant subset must be an exact sequence.")
        subset = tuple(str(value) for value in raw_subset)
        if not subset or len(subset) != len(set(subset)):
            raise IngestError("Declared range variant subset is empty or duplicate.")
        row_key = (str(policy), str(outcome), str(axis))
        if row_key not in ranges.index:
            raise IngestError(f"The range summary omits declared cell {row_key}.")
        row = ranges.loc[row_key]
        if isinstance(row, pd.DataFrame):
            raise IngestError(f"Range summary duplicates declared cell {row_key}.")
        try:
            all_variants = set(json.loads(str(row["all_variants_json"])))
            admissible = set(json.loads(str(row["admissible_variants_json"])))
            estimate_map = {
                str(key): float(value)
                for key, value in json.loads(
                    str(row["variant_estimates_json"])
                ).items()
            }
        except (TypeError, ValueError, json.JSONDecodeError) as exc:
            raise IngestError(f"{row_key}: range variant provenance is malformed.") from exc
        missing = sorted(set(subset) - all_variants)
        inadmissible = sorted(set(subset) - admissible)
        if missing:
            raise IngestError(f"{row_key}: published subset omits registered identity {missing}.")
        if inadmissible:
            raise IngestError(
                f"{row_key}: published endpoints depend on failed/unstable/rank-"
                f"deficient variant(s) {inadmissible}; no narrowed agreement is claimed."
            )
        subset_estimates = {name: estimate_map[name] for name in subset}
        ordered = sorted(subset_estimates.items(), key=lambda item: (item[1], item[0]))
        observed_min_variant, observed_min = ordered[0]
        observed_max_variant, observed_max = ordered[-1]
        if isinstance(declaration, Mapping):
            try:
                declared_min = float(declaration["minimum"])
                declared_max = float(declaration["maximum"])
            except (KeyError, TypeError, ValueError, OverflowError) as exc:
                raise IngestError(f"{row_key}: declared bounds are malformed.") from exc
            precision_raw = declaration.get("precision", 3)
            allowance_raw = declaration.get("allowance", tolerance_value)
            reference = declaration.get("reference_window")
            declared_min_variant = declaration.get("minimum_variant")
            declared_max_variant = declaration.get("maximum_variant")
        else:
            try:
                declared_min, declared_max = (float(value) for value in declaration)
            except (TypeError, ValueError, OverflowError) as exc:
                raise IngestError(f"{row_key}: declared range is malformed.") from exc
            precision_raw, allowance_raw = 3, tolerance_value
            reference = None
            declared_min_variant = declared_max_variant = None
        if (
            isinstance(precision_raw, (bool, np.bool_))
            or not isinstance(precision_raw, (int, np.integer))
            or not 0 <= int(precision_raw) <= 12
        ):
            raise IngestError(f"{row_key}: displayed precision is malformed.")
        try:
            allowance_value = float(allowance_raw)
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError(f"{row_key}: numerical allowance is malformed.") from exc
        if (
            isinstance(allowance_raw, (bool, np.bool_))
            or not np.isfinite(allowance_value)
            or allowance_value < 0.0
        ):
            raise IngestError(f"{row_key}: numerical allowance is malformed.")
        if not np.isfinite([declared_min, declared_max]).all() or declared_min > declared_max:
            raise IngestError(f"{row_key}: declared endpoints are invalid.")
        if reference is not None and str(reference) != str(row["reference_window"]):
            raise IngestError(f"{row_key}: declared reference identity disagrees.")
        if declared_min_variant is not None and str(declared_min_variant) != observed_min_variant:
            raise IngestError(f"{row_key}: declared minimum variant disagrees.")
        if declared_max_variant is not None and str(declared_max_variant) != observed_max_variant:
            raise IngestError(f"{row_key}: declared maximum variant disagrees.")
        half_unit = 0.5 * 10.0 ** (-int(precision_raw))
        allowance = allowance_value
        min_interval = (declared_min - half_unit - allowance, declared_min + half_unit + allowance)
        max_interval = (declared_max - half_unit - allowance, declared_max + half_unit + allowance)
        if not min_interval[0] <= observed_min <= min_interval[1] or not (
            max_interval[0] <= observed_max <= max_interval[1]
        ):
            raise IngestError(
                f"{row_key}/{subset}: unrounded range [{observed_min!r}, "
                f"{observed_max!r}] with endpoint variants "
                f"[{observed_min_variant}, {observed_max_variant}] falls outside "
                f"published rounding intervals {min_interval}, {max_interval}."
            )
    # Log the completed comparison.
    logger.info("declared ranges verified across %d partition(s).", len(expected))


# Freeze the execution tables behind a canonical digest.
def freeze_sweep_execution(
    reference_table: pd.DataFrame,
    donor_table: pd.DataFrame,
    factor_table: pd.DataFrame,
    donor_audit: Optional[pd.DataFrame] = None,
    ranges: Optional[pd.DataFrame] = None,
    coverage_audit: Optional[pd.DataFrame] = None,
) -> str:
    """
    Freeze the execution tables behind a canonical digest.

    Parameters
    ----------
    reference_table : pandas.DataFrame
        The reference-window table.
    donor_table : pandas.DataFrame
        The donor-pool table.
    factor_table : pandas.DataFrame
        The factor-proxy table with its diagnostics.
    donor_audit, ranges, coverage_audit : pandas.DataFrame, optional
        Complete Task 19 scientific audits included by the orchestrator. They
        are trailing optional inputs for compatibility with earlier callers.

    Returns
    -------
    str
        The hexadecimal digest of the canonical payload.

    Raises
    ------
    IngestError
        If any table is empty.
    """
    # Refuse an empty table.
    for label, table in (
        ("reference", reference_table),
        ("donor", donor_table),
        ("factor", factor_table),
    ):
        # Test the table's extent.
        if len(table) == 0:
            # The failure names the table.
            raise IngestError(f"Cannot freeze an empty {label} table.")
    # Refuse duplicate scientific keys, which would make canonical replay
    # ambiguous even if the numerical rows happened to match.
    if any(table.index.has_duplicates for table in (
        reference_table, donor_table, factor_table
    )):
        raise IngestError("Cannot freeze duplicate-keyed Task 19 tables.")
    # Digest complete labelled tables, including effective designs, numerical
    # diagnostics and retained failures rather than only successful scalars.
    return _content_digest(
        {
            "reference": reference_table.sort_index(),
            "donor": donor_table.sort_index(),
            "factor": factor_table.sort_index(),
            "donor_audit": (
                donor_audit.sort_index() if donor_audit is not None else None
            ),
            "ranges": ranges.sort_index() if ranges is not None else None,
            "coverage_audit": (
                coverage_audit.sort_index() if coverage_audit is not None else None
            ),
        }
    )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 19, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Execute the reference, donor and factor sweeps.
def execute_robustness_sweeps(
    baseline_config: Mapping[str, Any],
    data: Mapping[str, Any],
    pairs: Optional[Sequence[Tuple[str, str]]] = None,
    expected_ranges: Optional[Mapping[Tuple[Any, ...], Any]] = None,
    threshold: float = UNSTABLE_CONDITION_THRESHOLD,
    run_ids: Optional[Sequence[str]] = None,
) -> SweepExecutionResult:
    """
    Execute the reference, donor and factor sweeps.

    Parameters
    ----------
    baseline_config : mapping
        The baseline configuration, which is never mutated.
    data : mapping
        The frozen inputs under the declared keys.
    pairs : sequence of tuple, optional
        The pairs the execution covers. When absent every registered pair is
        covered.
    expected_ranges : mapping, optional
        Publication cells keyed by policy, outcome, axis and exact variant
        subset, with bounds and optional precision/reference metadata.
    threshold : float, optional
        The condition-number threshold above which an estimate is diagnostic.
    run_ids : sequence of str, optional
        The run identifiers to execute. When absent every run of the three
        executed axes is executed.

    Returns
    -------
    SweepExecutionResult
        The three tables, the range summary, the donor audit, the stability
        summary, the digest, the timestamp and the notes.

    Raises
    ------
    KeyError
        If the data mapping lacks the registry.
    IngestError
        If the registry carries no run for an executed axis, or if any table,
        range or comparison check fails.
    """
    # Refuse a data mapping lacking the registry.
    if "registry" not in data:
        # The failure names the absent key.
        raise KeyError("The data mapping lacks the registry.")
    # Compose the pair list, defaulting to every registered pair.
    selected: List[Tuple[str, str]] = (
        [tuple(pair) for pair in pairs]
        if pairs is not None
        else sorted(data["registry"].partitions.keys())
    )
    if not selected or len(selected) != len(set(selected)):
        raise IngestError("Task 19 pair selection is empty or duplicate.")
    # Resolve the exact mandatory run universe of Task 19's three axes.
    registry_frame = build_sweep_registry(selected)
    required_runs = [
        str(value)
        for value in registry_frame[
            registry_frame.index.get_level_values("axis").isin(EXECUTED_AXES)
        ]["run_id"].tolist()
    ]
    if not required_runs:
        raise IngestError("The registry carries no run for the executed axes.")
    if run_ids is None:
        selected_runs = required_runs
    else:
        selected_runs = [str(value) for value in run_ids]
        if len(selected_runs) != len(set(selected_runs)):
            raise IngestError("Task 19 run selection contains duplicate identifiers.")
        if set(selected_runs) != set(required_runs):
            missing = sorted(set(required_runs) - set(selected_runs))
            extra = sorted(set(selected_runs) - set(required_runs))
            raise IngestError(
                "Task 19 requires complete three-axis run coverage; "
                f"missing={missing[:INFERENCE_MAX_REPORTED_OFFENDERS]}, "
                f"unexpected={extra[:INFERENCE_MAX_REPORTED_OFFENDERS]}."
            )
    # Execute the factory's sweeps, restricted to the executed axes' runs.
    suite: SweepSuiteResult = execute_sweeps(
        baseline_config,
        data,
        pairs=selected,
        run_ids=selected_runs,
    )
    # Split complete registered axes while retaining unresolved branches for
    # the explicit Task 19 coverage audit.
    reference_table = split_axis_results(
        suite, AXIS_REFERENCE_WINDOW, allow_incomplete=True
    )
    donor_table = split_axis_results(
        suite, AXIS_DONOR_POOL, allow_incomplete=True
    )
    factor_table = split_axis_results(
        suite, AXIS_FACTOR_PROXY, allow_incomplete=True
    )
    # Classify the factor variants by their condition numbers.
    factor_table = classify_factor_variants(data, factor_table, threshold)
    # Compose the range summary of the reference axis.
    reference_ranges: pd.DataFrame = build_axis_ranges(
        reference_table, AXIS_REFERENCE_WINDOW
    )
    # Compose the range summary of the donor axis.
    donor_ranges: pd.DataFrame = build_axis_ranges(donor_table, AXIS_DONOR_POOL)
    # Compose the range summary of the factor axis, which separates the
    # diagnostic variants from the headline ones.
    factor_ranges: pd.DataFrame = build_axis_ranges(
        factor_table, AXIS_FACTOR_PROXY, classify_column="estimate_class"
    )
    # Concatenate the three range summaries.
    ranges: pd.DataFrame = pd.concat(
        [reference_ranges, donor_ranges, factor_ranges], axis=0
    )
    # Audit the donor list of every variant.
    donor_audit = build_donor_audit(
        data, selected, executed_table=donor_table, retain_failures=True
    )
    # Compose the stability summary of the factor variants.
    stability_columns = [
        "condition_number", "singular_values_json", "rank", "full_rank",
        "rank_deficient", "rank_tolerance", "factor_matrix_rank",
        "factor_matrix_rank_tolerance", "factor_rows", "factor_width",
        "factor_columns_json", "period_start", "period_end", "matrix_label",
        "scale_label", "rank_convention", "factor_sha256",
        "factor_design_class", "numerically_stable", "estimate_class",
        "classification_reason",
    ]
    stability_summary = factor_table[stability_columns].copy()
    # Cross-check the mutually exclusive factor classes without collapsing
    # rank deficiency into a merely large finite condition number.
    valid_factor_classes = {
        FACTOR_CLASS_ADMISSIBLE, FACTOR_CLASS_UNSTABLE,
        FACTOR_CLASS_RANK_DEFICIENT, FACTOR_CLASS_FAILED,
    }
    if not set(stability_summary["factor_design_class"]) <= valid_factor_classes:
        raise IngestError("Factor stability summary carries an unknown class.")
    for _, factor_row in stability_summary.iterrows():
        design_class = str(factor_row["factor_design_class"])
        condition = float(factor_row["condition_number"])
        if design_class == FACTOR_CLASS_ADMISSIBLE and (
            not np.isfinite(condition) or condition > float(threshold)
        ):
            raise IngestError("Admissible factor classification contradicts its condition.")
        if design_class == FACTOR_CLASS_UNSTABLE and (
            not np.isfinite(condition) or condition <= float(threshold)
        ):
            raise IngestError("Unstable factor classification contradicts its condition.")
        if design_class == FACTOR_CLASS_RANK_DEFICIENT and not bool(
            factor_row["rank_deficient"]
        ):
            raise IngestError("Rank-deficient factor class lacks rank evidence.")
    # Compare the observed ranges with the declared values when supplied.
    assert_declared_ranges(ranges, expected=expected_ranges)
    # Publish one coverage row for every registered branch and attach the
    # independent donor/factor audit state without hiding failures.
    coverage_rows: List[Dict[str, Any]] = []
    for axis_name, axis_table in (
        (AXIS_REFERENCE_WINDOW, reference_table),
        (AXIS_DONOR_POOL, donor_table),
        (AXIS_FACTOR_PROXY, factor_table),
    ):
        for (policy, outcome, variant), row in axis_table.iterrows():
            audit_status = "not_applicable"
            audit_detail = ""
            if axis_name == AXIS_DONOR_POOL:
                donor_row = donor_audit.loc[(policy, outcome, variant)]
                audit_status = str(donor_row["audit_status"])
                audit_detail = str(donor_row["failure_detail"])
            elif axis_name == AXIS_FACTOR_PROXY:
                audit_status = str(row["factor_design_class"])
                audit_detail = str(row["classification_reason"])
            coverage_rows.append(
                {
                    "policy": str(policy), "outcome": str(outcome),
                    "axis": axis_name, "variant_name": str(variant),
                    "run_id": str(row["run_id"]),
                    "execution_status": str(row.get("status", "success")),
                    "audit_status": audit_status,
                    "audit_detail": audit_detail,
                    "included_in_headline_range": bool(
                        str(row.get("status", "success")) == "success"
                        and (
                            axis_name != AXIS_FACTOR_PROXY
                            or str(row.get("factor_design_class", ""))
                            == FACTOR_CLASS_ADMISSIBLE
                        )
                    ),
                }
            )
    coverage_audit = pd.DataFrame(coverage_rows).set_index(
        ["policy", "outcome", "axis", "variant_name"]
    ).sort_index()
    # Freeze the complete scientific payload after the independent audits and
    # unresolved coverage have been constructed.
    fingerprint = freeze_sweep_execution(
        reference_table,
        donor_table,
        factor_table,
        donor_audit=donor_audit,
        ranges=ranges,
        coverage_audit=coverage_audit,
    )
    # Accumulate the recorded notes.
    notes: List[str] = [
        f"the execution covered {len(selected)} partition(s) across "
        f"{len(EXECUTED_AXES)} axis(es).",
        "the reference variants differ only in the baseline window, so the "
        "estimates are directly comparable across the five windows.",
        "the donor variants substitute the donor pool alone; the target roster "
        "is never restricted by this sweep.",
        "the factor variants whose condition number exceeds the declared "
        "threshold are classified separately from rank-deficient and failed "
        "designs; all remain visible even when excluded from headline ranges.",
    ]
    # Record the diagnostic count.
    notes.append(
        f"diagnostic factor variants: "
        f"{int((factor_table['estimate_class'] == CLASS_DIAGNOSTIC).sum())} of "
        f"{len(factor_table)}."
    )
    notes.append(
        "unresolved/failed Task 19 branches retained in coverage audit: "
        f"{int((~coverage_audit['execution_status'].isin(['success', 'identity'])).sum())}; "
        "independent audit failures: "
        f"{int(coverage_audit['audit_status'].isin(['unresolved', 'failed']).sum())}."
    )
    # Assemble the immutable result.
    result: SweepExecutionResult = SweepExecutionResult(
        reference_table=reference_table,
        donor_table=donor_table,
        factor_table=factor_table,
        ranges=ranges,
        donor_audit=donor_audit,
        stability_summary=stability_summary,
        fingerprint_sha256=fingerprint,
        completed_utc=current_utc_iso(),
        notes=tuple(notes),
        coverage_audit=coverage_audit,
    )
    # Log the completed execution.
    logger.info(
        "robustness sweep execution complete: %d reference, %d donor and %d "
        "factor run(s); fingerprint %s.",
        len(reference_table),
        len(donor_table),
        len(factor_table),
        fingerprint[:16],
    )
    # Return the assembled result.
    return result
````

````python
# Task 20: Execute Horizon, Active-Year, and Rate-Outcome Sweeps

# ==============================================================================
# Task 20: Execute Horizon, Active-Year, and Rate-Outcome Sweeps
# ==============================================================================

# Declare the horizon window of the shortened variant, which excludes the
# final sample year so the sensitivity of the estimate to that year is
# localized.
SHORT_HORIZON: Final[Tuple[int, int]] = (2022, 2023)
# Declare the two rate outcomes of the analysis tables, whose units are the
# decimal share the tables report.
RATE_OUTCOMES: Final[Tuple[str, ...]] = ("inflation_att", "growth_att")
# Declare the name of the strict-indicator column of the coding tensor.
STRICT_INDICATOR_COLUMNS: Final[Mapping[str, str]] = {
    "strict_sanctions": "S_it",
    "direct_arming": "A_it",
}
# Declare the label of the horizon sweep in the comparison summary.
SWEEP_HORIZON: Final[str] = "horizon"
# Declare the label of the active-year sweep in the comparison summary.
SWEEP_ACTIVE_YEAR: Final[str] = "active_year"
# Declare the label of the rate-outcome check in the comparison summary.
SWEEP_RATE_OUTCOME: Final[str] = "rate_outcome"
# Declare the tolerance of the gap-matrix identity assertion.
GAP_IDENTITY_TOLERANCE: Final[float] = 1.0e-12


# Declare the frozen result of the secondary sweep execution.
@dataclass(frozen=True)
class SecondarySweepResult:
    """
    Frozen result of the horizon, active-year and rate executions.

    Attributes
    ----------
    horizon_table : pandas.DataFrame
        One row per partition and horizon variant carrying the effect, its
        standard error and its marker.
    active_year_table : pandas.DataFrame
        One row per partition carrying the active-year effect, the baseline
        effect, the roster count and the excluded count.
    rate_table : pandas.DataFrame
        One row per partition and rate outcome carrying the effect with its
        inference.
    mask_audit : pandas.DataFrame
        One row per target and second-wave year recording the strict
        indicator and whether the year is active for that country.
    exclusions : pandas.DataFrame
        One row per country excluded from the active-year average for having
        no active year.
    comparison : pandas.DataFrame
        One row per partition and sweep recording the effect and its
        difference from the baseline.
    fingerprint_sha256 : str
        Digest of the canonical execution payload.
    completed_utc : str
        ISO-8601 UTC timestamp of the execution.
    notes : tuple of str
        Recorded decisions, including the rate-outcome units and the level
        versus growth interpretation.
    """
    # One row per partition and horizon variant.
    horizon_table: pd.DataFrame
    # One row per partition carrying the active-year comparison.
    active_year_table: pd.DataFrame
    # One row per partition and rate outcome.
    rate_table: pd.DataFrame
    # One row per target and second-wave year recording the mask.
    mask_audit: pd.DataFrame
    # One row per excluded country.
    exclusions: pd.DataFrame
    # One row per partition and sweep recording the comparison.
    comparison: pd.DataFrame
    # Digest of the canonical execution payload.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the execution.
    completed_utc: str
    # Recorded decisions of the execution.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 20, Step 1: The horizon sweep
# -------------------------------------------------------------------------------------------------------------------------------


# Assert that a horizon variant left the gap matrix unchanged.
def assert_gap_matrix_unchanged(
    baseline_gaps: pd.DataFrame,
    variant_gaps: pd.DataFrame,
    context: str,
    tolerance: float = GAP_IDENTITY_TOLERANCE,
) -> None:
    """
    Assert that a horizon variant left the gap matrix unchanged.

    Parameters
    ----------
    baseline_gaps : pandas.DataFrame
        The baseline partition's gap matrix.
    variant_gaps : pandas.DataFrame
        The horizon variant's gap matrix.
    context : str
        The pair label, used in the failure message.
    tolerance : float, optional
        The tolerance of the cell comparison.

    Returns
    -------
    None
        The function returns nothing when the matrices agree cell by cell.

    Raises
    ------
    IngestError
        If the two matrices carry different axes, or if any cell diverges,
        because the horizon variant is defined to change only the average
        taken over the matrix rather than the matrix itself.
    """
    # Validate the comparison tolerance at the public boundary.
    try:
        tolerance_value = float(tolerance)
    except (TypeError, ValueError, OverflowError) as exc:
        raise IngestError(f"{context}: gap-identity tolerance is malformed.") from exc
    if (
        isinstance(tolerance, (bool, np.bool_))
        or not np.isfinite(tolerance_value)
        or tolerance_value < 0.0
    ):
        raise IngestError(f"{context}: gap-identity tolerance is invalid.")
    # Refuse divergent country axes.
    if not baseline_gaps.index.equals(variant_gaps.index):
        # The failure names the pair.
        raise IngestError(
            f"{context}: the horizon variant's gap matrix carries a different "
            f"country axis."
        )
    # Refuse divergent year axes.
    if list(baseline_gaps.columns) != list(variant_gaps.columns):
        # The failure names the pair.
        raise IngestError(
            f"{context}: the horizon variant's gap matrix carries a different "
            f"year axis."
        )
    # Refuse non-finite cells before subtraction; NaN must never turn the
    # unchanged predicate into a vacuous pass through a false comparison.
    try:
        baseline_values = baseline_gaps.to_numpy(dtype=float)
        variant_values = variant_gaps.to_numpy(dtype=float)
    except (TypeError, ValueError, OverflowError) as exc:
        raise IngestError(f"{context}: gap matrix contains nonnumeric cells.") from exc
    if not np.isfinite(baseline_values).all() or not np.isfinite(variant_values).all():
        raise IngestError(f"{context}: gap matrix contains non-finite cells.")
    # Compute the largest absolute cell difference.
    difference: float = float(np.max(np.abs(baseline_values - variant_values)))
    # Refuse a divergence beyond the tolerance.
    if difference > tolerance_value:
        # The failure names the pair and the difference.
        raise IngestError(
            f"{context}: the horizon variant changed the gap matrix by "
            f"{difference:.3e}, which exceeds the declared tolerance; only the "
            f"averaging window may change."
        )
    # Log the verified identity.
    logger.info(
        "%s: gap matrix unchanged by the horizon variant (largest difference "
        "%.3e).",
        context,
        difference,
    )


# Execute the horizon sweep and assemble its table.
def run_horizon_sweep(
    baseline_config: Mapping[str, Any],
    data: Mapping[str, Any],
    pairs: Sequence[Tuple[str, str]],
    baseline_gaps: Optional[Mapping[Tuple[str, str], pd.DataFrame]] = None,
    baseline_results: Optional[Mapping[Tuple[str, str], BaselineResult]] = None,
) -> pd.DataFrame:
    """
    Execute the horizon sweep and assemble its table.

    Parameters
    ----------
    baseline_config : mapping
        The baseline configuration.
    data : mapping
        The frozen inputs under the declared keys.
    pairs : sequence of tuple
        The pairs the sweep covers.
    baseline_gaps : mapping of tuple to pandas.DataFrame, optional
        The baseline gap matrices keyed on the pair, used for the declared
        identity assertion when they are supplied.
    baseline_results : mapping of tuple to BaselineResult, optional
        Verified cached Task 15 results. Each pair is fitted at most once when
        this mapping is absent; horizon variants only reaggregate its gaps.

    Returns
    -------
    pandas.DataFrame
        One row per partition and horizon variant carrying the effect, its
        standard error and its marker.

    Raises
    ------
    KeyError
        If the data mapping lacks the registry.
    IngestError
        If the registry carries no horizon run, or if a gap identity fails.
    """
    normalized_pairs = tuple((str(pair[0]), str(pair[1])) for pair in pairs)
    if not normalized_pairs or len(normalized_pairs) != len(set(normalized_pairs)):
        raise IngestError("Horizon pair selection is empty or duplicate.")
    if "registry" not in data or not isinstance(data["registry"], PartitionRegistry):
        raise KeyError("The data mapping lacks a valid partition registry.")
    # Resolve run identifiers from the frozen Task 18 registry; never derive a
    # second identifier from the human-readable variant label.
    sweep_registry = build_sweep_registry(normalized_pairs)
    registered_horizon_rows = sweep_registry[
        sweep_registry.index.get_level_values("axis") == AXIS_HORIZON
    ]
    if len(registered_horizon_rows) != len(normalized_pairs) * len(HORIZON_VARIANTS):
        raise IngestError("Frozen sweep registry has incomplete horizon coverage.")
    # Resolve all canonical/legacy window aliases once; variants are applied by
    # Task 18, which synchronizes wave_2_level and post_escalation.
    baseline_windows, _ = resolve_reporting_windows(baseline_config)
    rows: List[Dict[str, Any]] = []
    for pair in sorted(normalized_pairs):
        if pair not in data["registry"].partitions:
            raise IngestError(f"Horizon registry omits pair {pair}.")
        if baseline_results is not None:
            if pair not in baseline_results:
                raise IngestError(f"Horizon baseline cache omits pair {pair}.")
            baseline = baseline_results[pair]
        else:
            # A direct caller pays for one verified fit per pair, never one fit
            # per horizon variant.
            baseline = run_baseline(pair[0], pair[1], baseline_config, data)
        _validate_cached_sweep_baseline(
            baseline, pair, data["registry"], data["panel"], data["lagged"],
            baseline_windows,
        )
        if baseline_gaps is not None:
            if pair not in baseline_gaps:
                raise IngestError(f"Horizon baseline-gap mapping omits pair {pair}.")
            assert_gap_matrix_unchanged(
                baseline.gaps, baseline_gaps[pair], context=f"{pair[0]}/{pair[1]}/baseline"
            )
        baseline_gap_digest = _content_digest(baseline.gaps)
        for variant_name, variant_window in HORIZON_VARIANTS:
            spec = SweepSpec(AXIS_HORIZON, variant_name, variant_window)
            registry_key = (pair[0], pair[1], AXIS_HORIZON, variant_name)
            if registry_key not in registered_horizon_rows.index:
                raise IngestError(f"Frozen sweep registry omits horizon run {registry_key}.")
            registered_run_id = str(
                registered_horizon_rows.loc[registry_key, "run_id"]
            )
            swept = run_sweep(
                pair[0], pair[1], spec, baseline_config, data,
                run_id=registered_run_id,
                baseline_result=baseline,
            )
            design = swept.effective_design
            if (
                swept.execution_mode != "reaggregate_existing_gaps"
                or bool(design.get("estimation_refit", True))
                or str(design.get("gap_sha256", "")) != baseline_gap_digest
                or tuple(design.get("effective_post_window", ()))
                != tuple(variant_window)
                or tuple(design.get("effective_pre_window", ()))
                != tuple(baseline_windows["pre_escalation"])
            ):
                raise IngestError(
                    f"{pair}/{variant_name}: horizon execution did not reaggregate "
                    "the unchanged cached gap matrix under canonical windows."
                )
            if swept.run_id != registered_run_id:
                raise IngestError(
                    f"{pair}/{variant_name}: horizon result run identifier "
                    "disagrees with the frozen Task 18 registry."
                )
            rows.append(
                {
                    "policy": pair[0], "outcome": pair[1],
                    "variant_name": variant_name, "att": float(swept.att),
                    "se_cd": float(swept.se_cd), "stars": str(swept.stars),
                    "run_id": swept.run_id, "status": "success",
                    "execution_mode": swept.execution_mode,
                    "gap_sha256": baseline_gap_digest,
                    "baseline_lineage_sha256": str(
                        baseline.fit_provenance.get("lineage_sha256", "")
                    ),
                    "reference_window_json": json.dumps(
                        list(baseline_windows["pre_escalation"]), separators=(",", ":")
                    ),
                    "horizon_window_json": json.dumps(
                        list(variant_window), separators=(",", ":")
                    ),
                }
            )
    table = pd.DataFrame(rows).set_index(["policy", "outcome", "variant_name"])
    expected_rows = len(normalized_pairs) * len(HORIZON_VARIANTS)
    if len(table) != expected_rows or table.index.has_duplicates:
        raise IngestError("Horizon execution coverage is incomplete or duplicate.")
    table.attrs["gap_identity_gate"] = (
        "asserted from verified Task15 cache and Task18 gap_sha256: no variant refit"
    )
    # Log the sweep's extent.
    logger.info("horizon sweep executed across %d row(s).", len(table))
    # Return the table.
    return table


# -------------------------------------------------------------------------------------------------------------------------------
# Task 20, Step 2: The active-year sweep
# -------------------------------------------------------------------------------------------------------------------------------


# Build the strict-indicator masks of a partition's targets.
def build_active_year_masks(
    coding_frame: pd.DataFrame,
    policy_name: str,
    targets: Sequence[str],
    window: Tuple[int, int] = POST_ESCALATION_WINDOW,
) -> pd.DataFrame:
    """
    Build the strict-indicator masks of a partition's targets.

    Parameters
    ----------
    coding_frame : pandas.DataFrame
        The coding tensor carrying the strict indicator of each policy.
    policy_name : str
        The policy instrument of the partition, whose indicator column is
        selected.
    targets : sequence of str
        The partition's target countries.
    window : tuple of int, optional
        The inclusive second-wave window.

    Returns
    -------
    pandas.DataFrame
        One row per target and window year carrying the strict indicator and
        the active flag.

    Raises
    ------
    IngestError
        If the policy carries no declared indicator column, if the column is
        absent from the coding frame, or if a target is absent from the
        frame.
    """
    # Refuse a policy without a declared indicator column.
    if policy_name not in STRICT_INDICATOR_COLUMNS:
        # The failure names the policy.
        raise IngestError(
            f"No strict indicator is declared for the policy {policy_name!r}."
        )
    # Read the indicator column of the policy.
    indicator_column: str = STRICT_INDICATOR_COLUMNS[policy_name]
    # Refuse a coding frame lacking the indicator column.
    if indicator_column not in coding_frame.columns:
        # The failure names the absent column.
        raise IngestError(
            f"The coding frame lacks the indicator column {indicator_column!r}."
        )
    # Validate the requested target roster and Wave-2 window exactly.
    target_codes = tuple(str(code).strip() for code in targets)
    if (
        not target_codes
        or any(not code for code in target_codes)
        or len(target_codes) != len(set(target_codes))
    ):
        raise IngestError(f"{policy_name}: target roster is empty, blank or duplicate.")
    wave_window = _validate_att_window(tuple(window), "active-year Wave 2")
    years: List[int] = list(range(wave_window[0], wave_window[1] + 1))
    # Validate source keys before any filtering or type conversion.
    if not {"country_code", "year"} <= set(coding_frame.columns):
        raise IngestError("The coding frame lacks country_code/year keys.")
    if coding_frame["country_code"].isna().any() or any(
        not str(code).strip() for code in coding_frame["country_code"]
    ):
        raise IngestError("The coding frame carries null/blank country keys.")
    if coding_frame["year"].map(lambda value: isinstance(value, (bool, np.bool_))).any():
        raise IngestError("The coding frame carries Boolean calendar years.")
    numeric_year = pd.to_numeric(coding_frame["year"], errors="coerce")
    if bool(numeric_year.isna().any()) or not np.isfinite(
        numeric_year.to_numpy(dtype=float)
    ).all() or not numeric_year.map(lambda value: float(value).is_integer()).all():
        raise IngestError("The coding frame carries malformed calendar years.")
    normalized = coding_frame.copy()
    normalized["country_code"] = normalized["country_code"].astype(str).str.strip()
    normalized["year"] = numeric_year.astype(int)
    # Select the exact target-by-window product from the strict coding tensor.
    scoped = normalized[
        normalized["country_code"].isin(set(target_codes))
        & normalized["year"].isin(years)
    ].copy()
    if bool(scoped.duplicated(["country_code", "year"]).any()):
        raise IngestError(
            f"{policy_name}: strict coding duplicates target-year keys."
        )
    # Identify the targets absent from the coding frame.
    present = set(scoped["country_code"].astype(str))
    # Compose the missing targets.
    missing: List[str] = sorted(set(target_codes) - present)
    # Refuse a coding frame that omits a target.
    if missing:
        # The failure names the missing targets.
        raise IngestError(
            f"{policy_name}: the coding frame omits the target(s) {missing[:8]}."
        )
    # Validate the strict indicator as a complete finite binary tensor.
    numeric_indicator = pd.to_numeric(scoped[indicator_column], errors="coerce")
    if bool(numeric_indicator.isna().any()) or not np.isfinite(
        numeric_indicator.to_numpy(dtype=float)
    ).all() or not set(numeric_indicator.astype(float).unique()) <= {0.0, 1.0}:
        raise IngestError(
            f"{policy_name}: strict indicator must contain only finite binary cells."
        )
    scoped[indicator_column] = numeric_indicator.astype(int)
    # Pivot unique keys without an aggregation that could hide duplicates.
    grid: pd.DataFrame = scoped.pivot(
        index="country_code",
        columns="year",
        values=indicator_column,
    )
    # Reindex to the declared country and year axes so an absent cell is
    # visible rather than shortening an axis.
    grid = grid.reindex(
        index=pd.Index(sorted(target_codes), name="country_code"),
        columns=pd.Index(years, name="year"),
    )
    # Refuse a mask carrying a missing indicator cell.
    if bool(grid.isna().to_numpy().any()):
        # The failure names the policy.
        raise IngestError(
            f"{policy_name}: the strict indicator carries a missing cell over "
            f"the second wave."
        )
    # Assemble the mask audit frame in long form.
    audit: pd.DataFrame = (
        grid.stack()
        .rename("strict_indicator")
        .reset_index()
        .rename(columns={"level_1": "year"})
    )
    # Record the active flag as the indicator's own value, because the strict
    # indicator is a step function that is one exactly when the country is
    # active in that year.
    audit["active"] = audit["strict_indicator"].astype(int) == 1
    # Log the mask construction.
    logger.info(
        "%s: strict-indicator masks built for %d target(s) over %d year(s).",
        policy_name,
        len(grid),
        len(years),
    )
    # Return the audit frame.
    return audit


# Compute the active-year effect of a partition.
def compute_active_year_effect(
    gaps: pd.DataFrame,
    pre_levels: pd.Series,
    masks: pd.DataFrame,
) -> Tuple[float, pd.DataFrame, int]:
    """
    Compute the active-year effect of a partition.

    Parameters
    ----------
    gaps : pandas.DataFrame
        The baseline gap matrix indexed on the country and columned by the
        years.
    pre_levels : pandas.Series
        The per-country pre-escalation levels, which the incremental estimand
        subtracts from each country's post-window mean; the active-year
        variant restricts the second wave alone, so its baseline is the same
        pre-escalation level the baseline estimand uses.
    masks : pandas.DataFrame
        The mask audit frame keyed on the country and the year.

    Returns
    -------
    tuple
        The active-year incremental effect, the exclusion frame naming every
        country dropped for having no active year, and the number of
        countries the average was taken over.

    Raises
    ------
    IngestError
        If no country retains an active year, because the average would then
        be undefined, or if a country carries no pre-escalation level.
    """
    # Accumulate the per-country contrasts.
    country_means: List[float] = []
    # Accumulate the exclusions.
    exclusion_rows: List[Dict[str, Any]] = []
    # Iterate over the gap matrix's countries in its own order.
    for country in gaps.index:
        # Refuse a country without a pre-escalation level.
        if str(country) not in set(str(code) for code in pre_levels.index):
            # The failure names the country.
            raise IngestError(
                f"Country {country} carries no pre-escalation level; the "
                f"incremental contrast is undefined."
            )
        # Select the country's mask rows.
        country_mask = masks[masks["country_code"].astype(str) == str(country)]
        if country_mask.empty:
            raise IngestError(f"Country {country} is absent from the activity mask.")
        # Compose the country's active years.
        active_years: List[int] = sorted([
            int(row["year"]) for _, row in country_mask.iterrows() if bool(row["active"])
        ])
        # Exclude a country with no active year and record the exclusion.
        if not active_years:
            # Append the exclusion row.
            exclusion_rows.append(
                {
                    "country_code": str(country),
                    "reason": "no active year inside the second wave",
                }
            )
            # Continue to the next country.
            continue
        # Append the country's own active-window mean less its pre-escalation
        # level, which is the incremental contrast under the restricted
        # second wave.
        country_means.append(
            float(gaps.loc[country, active_years].mean())
            - float(pre_levels.loc[country])
        )
    # Refuse a partition where every country was excluded.
    if not country_means:
        # The failure states the condition.
        raise IngestError(
            "The active-year variant excludes every country, so the effect is "
            "undefined."
        )
    # Assemble the exclusion frame.
    exclusions: pd.DataFrame = pd.DataFrame(
        exclusion_rows, columns=["country_code", "reason"]
    )
    # Compute the equal-weighted mean across the remaining countries.
    effect: float = float(np.mean(np.array(country_means, dtype=float)))
    # Log the computed effect and the roster size.
    logger.info(
        "active-year effect computed over %d countr(ies) with %d exclusion(s).",
        len(country_means),
        len(exclusions),
    )
    # Return the effect, the exclusions and the roster size.
    return effect, exclusions, len(country_means)


# Assert the active-year identity when every target is fully active.
def assert_active_equals_baseline_when_fully_active(
    active_effect: float,
    baseline_effect: float,
    roster_size: int,
    target_count: int,
    context: str,
    tolerance: float = GAP_IDENTITY_TOLERANCE,
    activity_mask: Optional[pd.DataFrame] = None,
    active_years: Optional[Sequence[int]] = None,
    baseline_years: Optional[Sequence[int]] = None,
    active_reference_window: Optional[Tuple[int, int]] = None,
    baseline_reference_window: Optional[Tuple[int, int]] = None,
    active_country_weights: Optional[Mapping[str, float]] = None,
    baseline_country_weights: Optional[Mapping[str, float]] = None,
) -> bool:
    """
    Assert the active-year identity when every target is fully active.

    Parameters
    ----------
    active_effect : float
        The active-year effect.
    baseline_effect : float
        The baseline incremental effect of the same partition.
    roster_size : int
        The number of countries the active-year average was taken over.
    target_count : int
        The partition's target count.
    context : str
        The pair label, used in the failure message.
    tolerance : float, optional
        The tolerance of the identity.
    activity_mask : pandas.DataFrame, optional
        Long target-by-year mask with ``country_code``, ``year`` and
        ``active`` columns. Without this evidence the identity is not asserted.
    active_years, baseline_years : sequence of int, optional
        The exact post years used by each estimand.
    active_reference_window, baseline_reference_window : tuple, optional
        The exact reference windows used by each estimand.
    active_country_weights, baseline_country_weights : mapping, optional
        Labelled country weights used by each estimand.

    Returns
    -------
    bool
        True exactly when every target-year cell is active, years, reference
        and weights are identical, and the two effects agree. False reports a
        legitimate distinct active-year estimand without asserting equality.

    Raises
    ------
    IngestError
        If every target is active and yet the two effects diverge, because a
        fully active roster makes the variant coincide with the baseline by
        construction.
    """
    # Validate scalar inputs without allowing Boolean counts or allowances.
    if any(isinstance(value, (bool, np.bool_)) for value in (roster_size, target_count)):
        raise IngestError(f"{context}: roster counts must be non-Boolean integers.")
    if not all(
        isinstance(value, (int, np.integer)) and int(value) >= 0
        for value in (roster_size, target_count)
    ):
        raise IngestError(f"{context}: roster counts are malformed.")
    try:
        tolerance_value = float(tolerance)
    except (TypeError, ValueError, OverflowError) as exc:
        raise IngestError(f"{context}: identity tolerance is malformed.") from exc
    if (
        isinstance(tolerance, (bool, np.bool_))
        or not np.isfinite(tolerance_value)
        or tolerance_value < 0.0
        or not np.isfinite([active_effect, baseline_effect]).all()
    ):
        raise IngestError(f"{context}: effects or tolerance are invalid.")
    # Absence of complete design evidence can never support an equality claim.
    if any(
        value is None
        for value in (
            activity_mask, active_years, baseline_years,
            active_reference_window, baseline_reference_window,
            active_country_weights, baseline_country_weights,
        )
    ):
        logger.info(
            "%s: active-year identity not applicable because full mask/design "
            "evidence was not supplied.", context,
        )
        return False
    # Validate and canonicalize the exact year axes without lossy conversion.
    def exact_years(values: Sequence[int], label: str) -> Tuple[int, ...]:
        if isinstance(values, (str, bytes, Mapping)):
            raise IngestError(f"{context}: {label} years are malformed.")
        normalized: List[int] = []
        for value in values:
            if (
                isinstance(value, (bool, np.bool_))
                or not isinstance(value, (int, np.integer))
            ):
                raise IngestError(f"{context}: {label} years are not integral.")
            normalized.append(int(value))
        if not normalized or len(normalized) != len(set(normalized)):
            raise IngestError(f"{context}: {label} years are empty or duplicate.")
        return tuple(normalized)
    active_year_axis = exact_years(
        active_years if active_years is not None else (), "active"
    )
    baseline_year_axis = exact_years(
        baseline_years if baseline_years is not None else (), "baseline"
    )
    # Validate the complete long target-by-year activity product.
    mask = activity_mask
    required_mask_columns = {"country_code", "year", "active"}
    if not isinstance(mask, pd.DataFrame) or not required_mask_columns <= set(mask.columns):
        raise IngestError(f"{context}: activity mask schema is incomplete.")
    if bool(mask.duplicated(["country_code", "year"]).any()):
        raise IngestError(f"{context}: activity mask duplicates target-year keys.")
    if len(mask) != int(target_count) * len(baseline_year_axis):
        raise IngestError(f"{context}: activity mask is not the full target-year product.")
    if mask["year"].map(lambda value: isinstance(value, (bool, np.bool_))).any():
        raise IngestError(f"{context}: activity mask years cannot be Boolean.")
    mask_years = tuple(sorted(set(pd.to_numeric(mask["year"], errors="coerce"))))
    if any(not np.isfinite(value) or not float(value).is_integer() for value in mask_years):
        raise IngestError(f"{context}: activity mask years are malformed.")
    mask_year_axis = tuple(int(value) for value in mask_years)
    mask_countries = tuple(sorted(set(mask["country_code"].astype(str))))
    if len(mask_countries) != int(target_count):
        raise IngestError(f"{context}: activity mask target coverage is incomplete.")
    active_values = mask["active"]
    if not active_values.map(lambda value: isinstance(value, (bool, np.bool_))).all():
        raise IngestError(f"{context}: activity mask must carry Boolean cells.")
    full_cell_activity = bool(active_values.astype(bool).all())
    # Canonicalize labelled weights and require exact roster/weight agreement.
    def weights(mapping: Mapping[str, float], label: str) -> Dict[str, float]:
        if not isinstance(mapping, Mapping):
            raise IngestError(f"{context}: {label} weights are malformed.")
        try:
            normalized = {str(key): float(value) for key, value in mapping.items()}
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError(f"{context}: {label} weights are malformed.") from exc
        if set(normalized) != set(mask_countries) or not np.isfinite(
            list(normalized.values())
        ).all() or any(value < 0.0 for value in normalized.values()):
            raise IngestError(f"{context}: {label} weights are invalid or misaligned.")
        if not math.isclose(sum(normalized.values()), 1.0, rel_tol=0.0, abs_tol=tolerance_value):
            raise IngestError(f"{context}: {label} weights do not sum to one.")
        return normalized
    active_weights = weights(
        active_country_weights if active_country_weights is not None else {}, "active"
    )
    baseline_weights = weights(
        baseline_country_weights if baseline_country_weights is not None else {},
        "baseline",
    )
    weights_identical = active_weights == baseline_weights
    years_identical = (
        active_year_axis == baseline_year_axis == mask_year_axis
    )
    references_identical = (
        tuple(active_reference_window or ())
        == tuple(baseline_reference_window or ())
    )
    roster_complete = int(roster_size) == int(target_count)
    identity_applicable = bool(
        roster_complete
        and full_cell_activity
        and years_identical
        and references_identical
        and weights_identical
    )
    # A partial mask or any design difference defines a legitimate distinct
    # estimand and must not trigger a false baseline-equality failure.
    if not identity_applicable:
        logger.info(
            "%s: active-year contrast is distinct (roster_complete=%s, "
            "full_cells=%s, years=%s, reference=%s, weights=%s).",
            context, roster_complete, full_cell_activity, years_identical,
            references_identical, weights_identical,
        )
        return False
    # Compute the deviation between the two effects.
    deviation: float = abs(float(active_effect) - float(baseline_effect))
    # Refuse a divergence under a complete roster.
    if deviation > tolerance_value:
        # The failure names the pair and the deviation.
        raise IngestError(
            f"{context}: with every target active the variant must coincide with "
            f"the baseline, but the effects differ by {deviation:.3e}."
        )
    # Log the verified identity.
    logger.info(
        "%s: the active-year variant coincides with the baseline under a "
        "complete roster (deviation %.3e).",
        context,
        deviation,
    )
    # Return the asserted outcome.
    return True


# -------------------------------------------------------------------------------------------------------------------------------
# Task 20, Step 3: The rate-outcome check
# -------------------------------------------------------------------------------------------------------------------------------


# Run the rate-outcome check for every partition.
def run_rate_outcome_check(
    baseline_config: Mapping[str, Any],
    data: Mapping[str, Any],
    pairs: Sequence[Tuple[str, str]],
) -> pd.DataFrame:
    """
    Run the rate-outcome check for every partition.

    Parameters
    ----------
    baseline_config : mapping
        The baseline configuration.
    data : mapping
        The frozen inputs under the declared keys.
    pairs : sequence of tuple
        Pair-like policy declarations. Outcomes are ignored deliberately;
        each unique policy is invoked once and emits both rate outcomes.

    Returns
    -------
    pandas.DataFrame
        One row per partition and rate outcome carrying the effect with its
        standard error and marker. The columns are the analysis-table forms of
        the two rate variables, whose units are the decimal share the tables
        report, so a value of two and a half percentage points appears as two
        and a half hundredths.

    Raises
    ------
    KeyError
        If the data mapping lacks the registry.
    IngestError
        Propagated from the baseline interface, with the pair named.

    Notes
    -----
    A positive level gap in the product per capita does not imply a positive
    growth differential: a level that is above its counterfactual can coexist
    with a rate that is below its own, and the two are reported side by side
    precisely so that the distinction is visible.
    """
    # Resolve unique policies before invoking the helper that emits both
    # outcomes; repeated level-outcome pairs may never duplicate rate rows.
    try:
        policies = tuple(sorted(set(str(pair[0]) for pair in pairs)))
    except (TypeError, IndexError) as exc:
        raise IngestError("Rate-outcome policy declarations are malformed.") from exc
    if not policies:
        raise IngestError("Rate-outcome check requires at least one policy.")
    # Accumulate the rate rows.
    rows: List[Dict[str, Any]] = []
    # Iterate once per unique policy.
    for policy_name in policies:
        # Run each rate outcome through the baseline interface.
        for rate_outcome in RATE_OUTCOMES:
            # Compose the pair key carrying the rate outcome.
            rate_pair: Tuple[str, str] = (policy_name, rate_outcome)
            # Run the interface on the rate outcome.
            try:
                # Compose the baseline result of the rate outcome.
                result = run_baseline(rate_pair[0], rate_pair[1], baseline_config, data)
            # Re-raise a failure with the pair named for exact localization.
            except IngestError as error:
                # Raise the localized failure.
                raise IngestError(f"{rate_pair[0]}/{rate_pair[1]}: {error}") from error
            # Append the rate row.
            rows.append(
                {
                    "policy": rate_pair[0],
                    "outcome": rate_pair[1],
                    "att": float(result.att_incremental),
                    "se_cd": float(result.se_cd),
                    "stars": str(result.stars),
                    "units": "decimal share of the analysis tables",
                }
            )
    # Assemble the rate frame keyed on the policy and the outcome.
    table: pd.DataFrame = pd.DataFrame(rows).set_index(["policy", "outcome"])
    expected_keys = {
        (policy_name, outcome)
        for policy_name in policies for outcome in RATE_OUTCOMES
    }
    if table.index.has_duplicates or set(table.index.tolist()) != expected_keys:
        raise IngestError("Rate-outcome rows are duplicate or incomplete.")
    # Log the check's extent.
    logger.info("rate-outcome check executed across %d row(s).", len(table))
    # Return the table.
    return table


# Freeze the three tables behind a canonical digest.
def freeze_secondary_sweeps(
    horizon_table: pd.DataFrame,
    active_year_table: pd.DataFrame,
    rate_table: pd.DataFrame,
    mask_audit: Optional[pd.DataFrame] = None,
    exclusions: Optional[pd.DataFrame] = None,
    comparison: Optional[pd.DataFrame] = None,
) -> str:
    """
    Freeze the three tables behind a canonical digest.

    Parameters
    ----------
    horizon_table : pandas.DataFrame
        The horizon table.
    active_year_table : pandas.DataFrame
        The active-year table.
    rate_table : pandas.DataFrame
        The rate table.
    mask_audit, exclusions, comparison : pandas.DataFrame, optional
        Complete Task 20 design/failure evidence included by the orchestrator;
        trailing optional inputs preserve earlier direct calls.

    Returns
    -------
    str
        The hexadecimal digest of the canonical payload.

    Raises
    ------
    IngestError
        If any table is empty.
    """
    # Refuse an empty table.
    for label, table in (
        ("horizon", horizon_table),
        ("active-year", active_year_table),
        ("rate", rate_table),
    ):
        # Test the table's extent.
        if len(table) == 0:
            # The failure names the table.
            raise IngestError(f"Cannot freeze an empty {label} table.")
    if any(table.index.has_duplicates for table in (
        horizon_table, active_year_table, rate_table
    )):
        raise IngestError("Cannot freeze duplicate-keyed secondary sweep tables.")
    # Digest complete labelled tables and the masks/windows/failures that give
    # their scalar summaries scientific meaning.
    return _content_digest(
        {
            "horizon": horizon_table.sort_index(),
            "active_year": active_year_table.sort_index(),
            "rate": rate_table.sort_index(),
            "mask_audit": (
                mask_audit.sort_values(["policy", "outcome", "country_code", "year"])
                .reset_index(drop=True)
                if mask_audit is not None else None
            ),
            "exclusions": (
                exclusions.sort_index() if exclusions is not None else None
            ),
            "comparison": (
                comparison.sort_values(["policy", "outcome", "sweep", "variant"])
                .reset_index(drop=True)
                if comparison is not None else None
            ),
        }
    )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 20, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Execute the horizon, active-year and rate sweeps.
def execute_secondary_sweeps(
    baseline_config: Mapping[str, Any],
    data: Mapping[str, Any],
    pairs: Optional[Sequence[Tuple[str, str]]] = None,
    baseline_effects: Optional[Mapping[Tuple[str, str], float]] = None,
    expected_rates: Optional[Mapping[Tuple[str, str], float]] = None,
    tolerance: float = 1.0e-3,
) -> SecondarySweepResult:
    """
    Execute the horizon, active-year and rate sweeps.

    Parameters
    ----------
    baseline_config : mapping
        The baseline configuration, which is never mutated.
    data : mapping
        The frozen inputs under the declared keys, which must carry the coding
        frame under the key ``coding`` for the active-year step.
    pairs : sequence of tuple, optional
        The pairs the execution covers. When absent every registered pair is
        covered.
    baseline_effects : mapping of tuple to float, optional
        The baseline incremental effect of each pair, used by the active-year
        identity assertion and the comparison summary.
    expected_rates : mapping of tuple to float, optional
        The declared rate effects, compared only when supplied.
    tolerance : float, optional
        The tolerance of the comparisons.

    Returns
    -------
    SecondarySweepResult
        The three tables, the mask audit, the exclusions, the comparison, the
        digest, the timestamp and the notes.

    Raises
    ------
    KeyError
        If the data mapping lacks a declared key.
    IngestError
        If any execution, mask, identity or comparison check fails.
    """
    # Refuse a data mapping lacking a declared key, including the coding frame
    # the active-year step requires.
    for key in ("registry", "panel", "lagged", "coding"):
        # Test the key's presence.
        if key not in data:
            # The failure names the absent key.
            raise KeyError(f"The data mapping lacks the declared key {key!r}.")
    # Compose the pair list, defaulting to every registered pair.
    selected: List[Tuple[str, str]] = (
        [tuple(pair) for pair in pairs]
        if pairs is not None
        else sorted(data["registry"].partitions.keys())
    )
    if not selected or len(selected) != len(set(selected)):
        raise IngestError("Secondary-sweep pair selection is empty or duplicate.")
    if not isinstance(data["registry"], PartitionRegistry):
        raise IngestError("Secondary-sweep partition registry is malformed.")
    missing_pairs = sorted(set(selected) - set(data["registry"].partitions))
    if missing_pairs:
        raise IngestError(f"Secondary-sweep registry omits pair(s) {missing_pairs}.")
    try:
        tolerance_value = float(tolerance)
    except (TypeError, ValueError, OverflowError) as exc:
        raise IngestError("Secondary-sweep tolerance is malformed.") from exc
    if (
        isinstance(tolerance, (bool, np.bool_))
        or not np.isfinite(tolerance_value)
        or tolerance_value < 0.0
    ):
        raise IngestError("Secondary-sweep tolerance must be finite and nonnegative.")
    # Resolve all aliases and provenance through the authoritative Task 15
    # window resolver rather than manually reconstructing configuration paths.
    baseline_windows, _ = resolve_reporting_windows(baseline_config)
    # Fit each requested level-outcome partition exactly once; both horizon and
    # active-year branches consume these same verified objects.
    baseline_results: Dict[Tuple[str, str], BaselineResult] = {
        pair: run_baseline(pair[0], pair[1], baseline_config, data)
        for pair in selected
    }
    baseline_gaps = {
        pair: result.gaps.copy(deep=True)
        for pair, result in baseline_results.items()
    }
    actual_baseline_effects = {
        pair: float(result.att_incremental)
        for pair, result in baseline_results.items()
    }
    # A caller-supplied baseline mapping is an audit declaration, not an
    # alternative source of truth; require exact selected coverage and values.
    if baseline_effects is not None:
        try:
            normalized_declared = {
                (str(key[0]), str(key[1])): float(value)
                for key, value in baseline_effects.items()
            }
        except (TypeError, ValueError, IndexError, OverflowError) as exc:
            raise IngestError("Declared baseline-effect mapping is malformed.") from exc
        if not np.isfinite(list(normalized_declared.values())).all():
            raise IngestError("Declared baseline effects must be finite.")
        if set(normalized_declared) != set(selected):
            raise IngestError("Declared baseline-effect coverage disagrees with selected pairs.")
        for pair, observed in actual_baseline_effects.items():
            if not math.isclose(
                observed, normalized_declared[pair], rel_tol=0.0,
                abs_tol=GAP_IDENTITY_TOLERANCE,
            ):
                raise IngestError(f"{pair}: declared baseline effect is stale.")
    # Execute the horizon sweep with the identity gate armed.
    horizon_table: pd.DataFrame = run_horizon_sweep(
        baseline_config,
        data,
        selected,
        baseline_gaps=baseline_gaps,
        baseline_results=baseline_results,
    )
    # Accumulate the active-year rows.
    active_rows: List[Dict[str, Any]] = []
    # Accumulate the mask audit frames.
    mask_frames: List[pd.DataFrame] = []
    # Accumulate the exclusion frames.
    exclusion_frames: List[pd.DataFrame] = []
    # Accumulate the comparison rows.
    comparison_rows: List[Dict[str, Any]] = []
    # Record the horizon comparison rows.
    for (policy, outcome, variant), row in horizon_table.iterrows():
        # Compose the baseline effect of the pair.
        baseline_value = actual_baseline_effects[(str(policy), str(outcome))]
        # Append the horizon comparison row.
        comparison_rows.append(
            {
                "policy": str(policy),
                "outcome": str(outcome),
                "sweep": SWEEP_HORIZON,
                "variant": str(variant),
                "swept_att": float(row["att"]),
                "baseline_att": baseline_value,
                "difference_from_baseline": float(row["att"]) - baseline_value,
            }
        )
    # Compute the active-year variant of every partition.
    for pair in sorted(selected):
        # Read the partition.
        partition = data["registry"].partitions[pair]
        # Compute the baseline effect of the pair.
        baseline_value = actual_baseline_effects[pair]
        # Build the strict-indicator masks of the partition.
        masks: pd.DataFrame = build_active_year_masks(
            data["coding"], pair[0], partition.targets,
            window=tuple(baseline_windows["post_escalation"]),
        )
        # Record the masks with their pair keys.
        mask_frame: pd.DataFrame = masks.copy()
        # Attach the pair keys to the mask audit.
        mask_frame["policy"] = pair[0]
        # Attach the outcome key to the mask audit.
        mask_frame["outcome"] = pair[1]
        # Append the mask audit frame.
        mask_frames.append(mask_frame)
        # Compute the partition's baseline gap matrix through the interface,
        # which is the matrix the active-year averaging consumes.
        baseline_result = baseline_results[pair]
        # Compute the active-year effect from the baseline gap matrix and the
        # baseline pre-escalation levels, which the restricted window shares.
        active_effect, exclusions, roster_size = compute_active_year_effect(
            baseline_result.gaps, baseline_result.pre_escalation_levels, masks
        )
        # Record the exclusions with their pair keys.
        if len(exclusions):
            # Copy the exclusion frame for the keys.
            keyed = exclusions.copy()
            # Attach the policy key.
            keyed["policy"] = pair[0]
            # Attach the outcome key.
            keyed["outcome"] = pair[1]
            # Append the keyed exclusions.
            exclusion_frames.append(keyed)
        # Compose explicit equal-country weights for both estimands. Excluded
        # active-year countries receive zero rather than disappearing from the
        # labelled comparison universe.
        target_codes = tuple(str(code) for code in partition.targets)
        baseline_weights = {
            code: 1.0 / float(len(target_codes)) for code in target_codes
        }
        admitted_codes = tuple(
            sorted(
                code for code in target_codes
                if bool(
                    masks.loc[
                        masks["country_code"].astype(str) == code, "active"
                    ].any()
                )
            )
        )
        active_weights = {
            code: (
                1.0 / float(len(admitted_codes)) if code in admitted_codes else 0.0
            )
            for code in target_codes
        }
        post_years = tuple(
            range(
                int(baseline_windows["post_escalation"][0]),
                int(baseline_windows["post_escalation"][1]) + 1,
            )
        )
        # Assert equality only after proving the full target-by-year predicate
        # and every other estimand component is identical.
        identity_asserted = assert_active_equals_baseline_when_fully_active(
            active_effect,
            baseline_value,
            roster_size,
            len(partition.targets),
            f"{pair[0]}/{pair[1]}",
            tolerance=GAP_IDENTITY_TOLERANCE,
            activity_mask=masks,
            active_years=post_years,
            baseline_years=post_years,
            active_reference_window=tuple(baseline_windows["pre_escalation"]),
            baseline_reference_window=tuple(baseline_windows["pre_escalation"]),
            active_country_weights=active_weights,
            baseline_country_weights=baseline_weights,
        )
        # Append the active-year row.
        active_rows.append(
            {
                "policy": pair[0],
                "outcome": pair[1],
                "active_year_att": float(active_effect),
                "baseline_att": baseline_value,
                "roster_size": int(roster_size),
                "target_count": int(len(partition.targets)),
                "excluded_count": int(len(exclusions)),
                "identity_asserted": bool(identity_asserted),
                "full_target_year_activity": bool(masks["active"].all()),
                "post_years_json": json.dumps(list(post_years), separators=(",", ":")),
                "reference_window_json": json.dumps(
                    list(baseline_windows["pre_escalation"]), separators=(",", ":")
                ),
                "active_country_weights_json": json.dumps(
                    active_weights, sort_keys=True, separators=(",", ":")
                ),
                "baseline_country_weights_json": json.dumps(
                    baseline_weights, sort_keys=True, separators=(",", ":")
                ),
                "mask_sha256": _content_digest(
                    masks.sort_values(["country_code", "year"]).reset_index(drop=True)
                ),
            }
        )
        # Append the active-year comparison row.
        comparison_rows.append(
            {
                "policy": pair[0],
                "outcome": pair[1],
                "sweep": SWEEP_ACTIVE_YEAR,
                "variant": "per_country_active_years",
                "swept_att": float(active_effect),
                "baseline_att": baseline_value,
                "difference_from_baseline": float(active_effect)
                - baseline_value,
            }
        )
    # Assemble the active-year table keyed on the pair.
    active_year_table: pd.DataFrame = pd.DataFrame(active_rows).set_index(
        ["policy", "outcome"]
    )
    if active_year_table.index.has_duplicates or set(
        active_year_table.index.tolist()
    ) != set(selected):
        raise IngestError("Active-year table coverage is incomplete or duplicate.")
    if not (
        active_year_table["identity_asserted"].astype(bool)
        == active_year_table["full_target_year_activity"].astype(bool)
    ).all():
        raise IngestError("Active-year equality predicate contradicts its full mask.")
    # Assemble the mask audit frame.
    mask_audit: pd.DataFrame = pd.concat(mask_frames, ignore_index=True)
    mask_key = ["policy", "outcome", "country_code", "year"]
    if bool(mask_audit.duplicated(mask_key).any()):
        raise IngestError("Combined activity-mask audit duplicates scientific keys.")
    expected_mask_rows = sum(
        len(data["registry"].partitions[pair].targets)
        * (int(baseline_windows["post_escalation"][1])
           - int(baseline_windows["post_escalation"][0]) + 1)
        for pair in selected
    )
    if len(mask_audit) != expected_mask_rows:
        raise IngestError("Combined activity-mask audit coverage is incomplete.")
    # Assemble the exclusion frame, defaulting to an empty keyed frame.
    exclusions: pd.DataFrame = (
        pd.concat(exclusion_frames, ignore_index=True)
        if exclusion_frames
        else pd.DataFrame(
            columns=["country_code", "reason", "policy", "outcome"]
        ).set_index(["policy", "outcome"])
    )
    # Run the helper exactly once per unique policy; it emits both rate
    # outcomes and therefore must not be invoked once per level-outcome pair.
    unique_policies = tuple(sorted(set(pair[0] for pair in selected)))
    policy_declarations = tuple((policy, "rate_outcomes") for policy in unique_policies)
    rate_table: pd.DataFrame = run_rate_outcome_check(
        baseline_config, data, policy_declarations
    )
    required_rate_keys = {
        (policy, outcome) for policy in unique_policies for outcome in RATE_OUTCOMES
    }
    if rate_table.index.has_duplicates or set(rate_table.index.tolist()) != required_rate_keys:
        raise IngestError("Rate table does not cover each unique policy/outcome exactly once.")
    # Record the rate comparison rows.
    for (policy, outcome), row in rate_table.iterrows():
        # Append the rate comparison row.
        comparison_rows.append(
            {
                "policy": str(policy),
                "outcome": str(outcome),
                "sweep": SWEEP_RATE_OUTCOME,
                "variant": str(outcome),
                "swept_att": float(row["att"]),
                "baseline_att": float("nan"),
                "difference_from_baseline": float("nan"),
            }
        )
    # Assemble the comparison frame.
    comparison: pd.DataFrame = pd.DataFrame(comparison_rows)
    comparison_keys = ["policy", "outcome", "sweep", "variant"]
    if bool(comparison.duplicated(comparison_keys).any()):
        raise IngestError("Secondary-sweep comparison contains duplicate keys.")
    expected_comparison_rows = (
        len(selected) * len(HORIZON_VARIANTS)
        + len(selected)
        + len(required_rate_keys)
    )
    if len(comparison) != expected_comparison_rows:
        raise IngestError("Secondary-sweep comparison coverage is incomplete.")
    if not np.isfinite(
        comparison.loc[
            comparison["sweep"].isin({SWEEP_HORIZON, SWEEP_ACTIVE_YEAR}),
            ["swept_att", "baseline_att", "difference_from_baseline"],
        ].to_numpy(dtype=float)
    ).all():
        raise IngestError("Level-sweep baseline comparisons are non-finite.")
    # Compare the rate effects with the declared values when supplied.
    if expected_rates is not None:
        # Iterate over the declared pairs.
        for key, declared in expected_rates.items():
            # Refuse a key the rate table does not report.
            if key not in rate_table.index:
                # The failure names the absent key.
                raise IngestError(f"The rate table omits the declared key {key}.")
            # Read the observed rate effect.
            observed: float = float(rate_table.loc[key, "att"])
            try:
                declared_value = float(declared)
            except (TypeError, ValueError, OverflowError) as exc:
                raise IngestError(f"{key}: declared rate effect is malformed.") from exc
            if not np.isfinite(declared_value):
                raise IngestError(f"{key}: declared rate effect is non-finite.")
            # Refuse a divergence beyond the tolerance.
            if abs(observed - declared_value) > tolerance_value:
                # The failure names the key and both values.
                raise IngestError(
                    f"{key}: rate effect {observed!r} against the declared "
                    f"{declared_value!r}."
                )
    # Freeze the three tables.
    fingerprint: str = freeze_secondary_sweeps(
        horizon_table,
        active_year_table,
        rate_table,
        mask_audit=mask_audit,
        exclusions=exclusions,
        comparison=comparison,
    )
    # Accumulate the recorded notes.
    notes: List[str] = [
        "the horizon variant truncates only the target's averaging window; the "
        "donor factor remains observed through the final year, and the gap "
        "matrix itself is unchanged by the variant.",
        "the active-year variant restricts the second wave to each country's "
        "active years; a country with no active year is excluded and recorded.",
        "the rate outcomes are the analysis-table forms, whose units are the "
        "decimal share the tables report.",
        "a positive level gap in the product per capita does not imply a "
        "positive growth differential; the level and rate rows must be read "
        "side by side.",
    ]
    # Record the identity outcomes.
    notes.append(
        f"the active-year identity with the baseline was asserted for "
        f"{int(active_year_table['identity_asserted'].sum())} partition(s) with "
        f"a complete target-by-year mask and identical years, reference and weights."
    )
    # Assemble the immutable result.
    result: SecondarySweepResult = SecondarySweepResult(
        horizon_table=horizon_table,
        active_year_table=active_year_table,
        rate_table=rate_table,
        mask_audit=mask_audit,
        exclusions=exclusions,
        comparison=comparison,
        fingerprint_sha256=fingerprint,
        completed_utc=current_utc_iso(),
        notes=tuple(notes),
    )
    # Log the completed execution.
    logger.info(
        "secondary sweeps complete: %d horizon, %d active-year and %d rate "
        "row(s); fingerprint %s.",
        len(horizon_table),
        len(active_year_table),
        len(rate_table),
        fingerprint[:16],
    )
    # Return the assembled result.
    return result
````

````python
# Task 21: Build Orchestrator 4 - Influence Diagnostics

# ==============================================================================
# Task 21: Build Orchestrator 4 - Influence Diagnostics
# ==============================================================================

# Declare the proportional trim of each tail of the contrast distribution.
TRIM_PROPORTION: Final[float] = 0.10
# Declare the number of leading changes the presentation table carries.
TOP_INFLUENCE_COUNT: Final[int] = 10
# Declare the name of the positive-share statistic.
STAT_POSITIVE_SHARE: Final[str] = "positive_share"
# Declare the name of the median statistic.
STAT_MEDIAN: Final[str] = "median_contrast"
# Declare the name of the trimmed-mean statistic.
STAT_TRIMMED_MEAN: Final[str] = "trimmed_mean_10pct"


# Declare the frozen influence diagnostics of one pair.
@dataclass(frozen=True)
class InfluenceResult:
    """
    Frozen influence diagnostics of one policy-outcome pair.

    Attributes
    ----------
    policy_name : str
        The policy instrument of the pair.
    outcome_name : str
        The canonical outcome variable of the pair.
    baseline_att : float
        The baseline incremental effect the changes are measured against.
    loo_treated : pandas.Series
        The leave-one-out coalition means indexed on the dropped country.
    loo_treated_min : float
        The smallest leave-one-out treated effect.
    loo_treated_max : float
        The largest leave-one-out treated effect.
    donor_changes : pandas.Series
        The change in the effect from deleting each donor, indexed on the
        donor code.
    distributional : dict of str to float
        The positive share, the median contrast and the trimmed mean of the
        per-country contrasts.
    top_influence : pandas.DataFrame
        The largest absolute donor changes with their ranks, as a presentation
        subset of the persisted series.
    """
    # The policy instrument of the pair.
    policy_name: str
    # The canonical outcome variable of the pair.
    outcome_name: str
    # The baseline incremental effect.
    baseline_att: float
    # The leave-one-out coalition means indexed on the dropped country.
    loo_treated: pd.Series
    # The smallest leave-one-out treated effect.
    loo_treated_min: float
    # The largest leave-one-out treated effect.
    loo_treated_max: float
    # The change from deleting each donor.
    donor_changes: pd.Series
    # The distributional statistics of the contrasts.
    distributional: Mapping[str, float]
    # The largest absolute donor changes with their ranks.
    top_influence: pd.DataFrame


# Declare the frozen diagnostics of the whole execution.
@dataclass(frozen=True)
class InfluenceDiagnosticsResult:
    """
    Frozen influence diagnostics across the executed pairs.

    Attributes
    ----------
    results : dict
        The per-pair influence records keyed on the pair.
    donor_changes : pandas.DataFrame
        The combined donor changes keyed on the pair and the donor.
    loo_treated : pandas.DataFrame
        The combined leave-one-out treated effects keyed on the pair and the
        dropped country.
    distributional : pandas.DataFrame
        The distributional statistics keyed on the pair with one column per
        statistic.
    fingerprint_sha256 : str
        Digest of the canonical diagnostics payload.
    completed_utc : str
        ISO-8601 UTC timestamp of the execution.
    notes : tuple of str
        Recorded decisions, including the iteration counts and the trimming.
    """
    # The per-pair influence records.
    results: Mapping[Tuple[str, str], InfluenceResult]
    # The combined donor changes.
    donor_changes: pd.DataFrame
    # The combined leave-one-out treated effects.
    loo_treated: pd.DataFrame
    # The distributional statistics.
    distributional: pd.DataFrame
    # Digest of the canonical diagnostics payload.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the execution.
    completed_utc: str
    # Recorded decisions of the execution.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 21, Step 1: The treated reduction and the distributional statistics
# -------------------------------------------------------------------------------------------------------------------------------


# Compute the leave-one-out coalition means from stored contrasts.
def compute_loo_treated(contrasts: pd.Series) -> pd.Series:
    """
    Compute the leave-one-out coalition means from stored contrasts.

    Parameters
    ----------
    contrasts : pandas.Series
        The per-country incremental contrasts of the baseline run, indexed on
        the country code.

    Returns
    -------
    pandas.Series
        The coalition mean computed with each country dropped in turn, indexed
        on the dropped country code. The function receives the stored vector
        and the frozen data are deliberately absent from its signature, so a
        refit is impossible: the diagnostic is a reduction of the stored
        contrasts alone.

    Raises
    ------
    IngestError
        If the contrast vector carries fewer than two entries, because
        dropping one would leave an undefined mean of one observation.
    """
    # Refuse a contrast vector too short for a leave-one-out reduction.
    if len(contrasts) < 2:
        # The failure names the observed length.
        raise IngestError(
            f"The treated leave-one-out diagnostic requires at least two "
            f"contrasts; the vector carries {len(contrasts)}."
        )
    # Compute the total of the contrasts.
    total: float = float(contrasts.sum())
    # Compute the leave-one-out mean for each dropped country.
    values: Dict[str, float] = {
        str(country): (total - float(value)) / float(len(contrasts) - 1)
        for country, value in contrasts.items()
    }
    # Assemble the series in the contrast vector's own order.
    series: pd.Series = pd.Series(values, dtype=float)
    # Name the series so its provenance travels with it.
    series.name = "loo_treated_att"
    # Log the reduction's extent.
    logger.info(
        "treated leave-one-out reduction computed across %d countr(ies).",
        len(series),
    )
    # Return the series.
    return series


# Assert that the treated reduction used the stored contrasts alone.
def assert_loo_treated_identity(
    contrasts: pd.Series, loo: pd.Series
) -> None:
    """
    Assert that the treated reduction used the stored contrasts alone.

    Parameters
    ----------
    contrasts : pandas.Series
        The per-country contrasts.
    loo : pandas.Series
        The leave-one-out means.

    Returns
    -------
    None
        The function returns nothing when every value equals the mean of the
        remaining contrasts.

    Raises
    ------
    IngestError
        If any value diverges from the mean of the remaining contrasts, which
        would mean the diagnostic did not reduce the stored vector.
    """
    # Iterate over the reported values.
    for country, value in loo.items():
        # Compose the remaining contrasts.
        remaining = contrasts.drop(index=country)
        # Compute the mean of the remaining contrasts.
        expected: float = float(remaining.mean())
        # Refuse a divergence.
        if abs(float(value) - expected) > 1.0e-12:
            # The failure names the country and both values.
            raise IngestError(
                f"{country}: the leave-one-out mean {float(value)!r} diverges "
                f"from the mean of the remaining contrasts {expected!r}."
            )
    # Log the verified identity.
    logger.info(
        "treated leave-one-out identity verified across %d value(s).", len(loo)
    )


# Compute the distributional statistics of the contrasts, trimming by rank.
def compute_distributional_statistics(
    contrasts: pd.Series, trim_proportion: float = TRIM_PROPORTION
) -> Dict[str, float]:
    """
    Compute the distributional statistics of the contrasts.

    Parameters
    ----------
    contrasts : pandas.Series
        The per-country contrasts.
    trim_proportion : float, optional
        The proportion trimmed from each tail of the ordered contrasts, which
        is a rank-based trim: the number of dropped observations is the floor
        of the proportion times the count.

    Returns
    -------
    dict
        The positive share, the median contrast and the trimmed mean.

    Raises
    ------
    IngestError
        If the contrast vector is empty or the proportion lies outside the
        closed unit interval below one half.
    """
    # Refuse an empty vector.
    if len(contrasts) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot summarise an empty contrast vector.")
    # Refuse a proportion outside the admissible range.
    if not (0.0 <= float(trim_proportion) < 0.5):
        # The failure names the observed proportion.
        raise IngestError(
            f"The trim proportion {trim_proportion!r} lies outside the admissible "
            f"range."
        )
    # Compute the share of strictly positive contrasts.
    positive_share: float = float((contrasts > 0.0).sum()) / float(len(contrasts))
    # Compute the median contrast.
    median: float = float(np.median(contrasts.to_numpy(dtype=float)))
    # Compute the number of observations trimmed from each tail by rank.
    trim_count: int = int(np.floor(float(trim_proportion) * float(len(contrasts))))
    # Compose the ordered contrasts.
    ordered: np.ndarray = np.sort(contrasts.to_numpy(dtype=float))
    # Trim the declared count from each tail, which is the rank-based trim the
    # specification requires rather than a value-based cut.
    trimmed: np.ndarray = (
        ordered[trim_count : len(ordered) - trim_count] if trim_count else ordered
    )
    # Compute the trimmed mean over the retained observations.
    trimmed_mean: float = float(statistics_mean(trimmed.tolist()))
    # Log the computed statistics with the trim count.
    logger.info(
        "distributional statistics computed over %d contrast(s): positive share "
        "%.6f, median %.6f, trimmed mean %.6f (%d trimmed per tail).",
        len(contrasts),
        positive_share,
        median,
        trimmed_mean,
        trim_count,
    )
    # Return the statistics.
    return {
        STAT_POSITIVE_SHARE: positive_share,
        STAT_MEDIAN: median,
        STAT_TRIMMED_MEAN: trimmed_mean,
    }


# -------------------------------------------------------------------------------------------------------------------------------
# Task 21, Step 2: The donor deletion diagnostics
# -------------------------------------------------------------------------------------------------------------------------------


# Compute the effect change from deleting one donor.
def compute_donor_change(
    policy_name: str,
    outcome_name: str,
    partition: Partition,
    donor: str,
    baseline_config: Mapping[str, Any],
    data: Mapping[str, Any],
    baseline_att: float,
) -> float:
    """
    Compute the effect change from deleting one donor.

    Parameters
    ----------
    policy_name : str
        The policy instrument of the pair.
    outcome_name : str
        The canonical outcome variable of the pair.
    partition : Partition
        The pair's baseline partition.
    donor : str
        The donor to delete.
    baseline_config : mapping
        The baseline configuration.
    data : mapping
        The frozen inputs under the declared keys.
    baseline_att : float
        The baseline incremental effect the change is measured against.

    Returns
    -------
    float
        The effect change from deleting the donor, which is the reduced
        design's effect less the baseline's. The reduced design is estimated by
        the frozen baseline interface, so the factor, the pooled fit, the
        loadings, the projection and the estimand are all rebuilt from scratch
        rather than approximated.

    Raises
    ------
    IngestError
        If the donor is not in the pool, if the reduced pool falls below the
        declared minimum, or if the interface refuses the reduced design.
    """
    # Refuse a donor outside the pool.
    if str(donor) not in set(str(code) for code in partition.donors):
        # The failure names the donor.
        raise IngestError(
            f"Donor {donor} is not a member of the pair's donor pool."
        )
    # Compose the reduced donor tuple.
    reduced_donors: Tuple[str, ...] = tuple(
        str(code) for code in partition.donors if str(code) != str(donor)
    )
    # Refuse a reduction that leaves too few donors.
    if len(reduced_donors) < MINIMUM_DONORS:
        # The failure names the donor count.
        raise IngestError(
            f"Deleting the donor {donor} would leave {len(reduced_donors)} "
            f"donor(s), below the declared minimum of {MINIMUM_DONORS}."
        )
    # Compose the reduced partition, which is a new record rather than a
    # mutation of the frozen one.
    reduced_partition = Partition(
        policy=partition.policy,
        outcome=partition.outcome,
        targets=partition.targets,
        donors=reduced_donors,
        exclusions=partition.exclusions,
    )
    # Compose the reduced data mapping through the factory's own builder.
    reduced_data: Dict[str, Any] = build_variant_data(
        data, policy_name, outcome_name, reduced_partition
    )
    # Estimate the reduced design through the frozen interface.
    reduced_result: BaselineResult = run_baseline(
        policy_name, outcome_name, baseline_config, reduced_data
    )
    # Return the change against the baseline.
    return float(reduced_result.att_incremental) - float(baseline_att)


# Compute the deletion change of every donor.
def compute_donor_changes(
    policy_name: str,
    outcome_name: str,
    partition: Partition,
    baseline_config: Mapping[str, Any],
    data: Mapping[str, Any],
    baseline_att: float,
) -> pd.Series:
    """
    Compute the deletion change of every donor.

    Parameters
    ----------
    policy_name : str
        The policy instrument of the pair.
    outcome_name : str
        The canonical outcome variable of the pair.
    partition : Partition
        The pair's baseline partition.
    baseline_config : mapping
        The baseline configuration.
    data : mapping
        The frozen inputs under the declared keys.
    baseline_att : float
        The baseline incremental effect.

    Returns
    -------
    pandas.Series
        The change from deleting each donor, indexed on the donor code in the
        partition's registered order so the series is reproducible from the
        registry alone.

    Raises
    ------
    IngestError
        Propagated from the single-donor computation, with the donor named so
        the failure localizes to one iteration rather than to the whole series.
    """
    # Accumulate the changes in the registered donor order.
    changes: Dict[str, float] = {}
    # Iterate over the donors in their registered order.
    for donor in partition.donors:
        # Compute the donor's deletion change.
        try:
            # Compose the change of the iteration.
            change: float = compute_donor_change(
                policy_name,
                outcome_name,
                partition,
                str(donor),
                baseline_config,
                data,
                baseline_att,
            )
        # Re-raise a failure with the donor named.
        except IngestError as error:
            # Raise the localized failure.
            raise IngestError(f"{policy_name}/{outcome_name}/{donor}: {error}") from error
        # Record the donor's change.
        changes[str(donor)] = change
    # Assemble the series in the registered order.
    series: pd.Series = pd.Series(changes, dtype=float)
    # Name the series so its provenance travels with it.
    series.name = "donor_change"
    # Log the iteration's extent.
    logger.info(
        "%s/%s: donor deletion executed across %d donor(s).",
        policy_name,
        outcome_name,
        len(series),
    )
    # Return the series.
    return series


# -------------------------------------------------------------------------------------------------------------------------------
# Task 21, Step 3: The presentation subset with deterministic ordering
# -------------------------------------------------------------------------------------------------------------------------------


# Build the presentation subset of the largest absolute changes.
def build_top_influence(
    changes: pd.Series, count: int = TOP_INFLUENCE_COUNT
) -> pd.DataFrame:
    """
    Build the presentation subset of the largest absolute changes.

    Parameters
    ----------
    changes : pandas.Series
        The donor changes indexed on the donor code.
    count : int, optional
        The number of rows the subset carries.

    Returns
    -------
    pandas.DataFrame
        The largest absolute changes with their ranks, ordered by the absolute
        change descending and then lexicographically by the donor code, so the
        subset is a deterministic function of the series. The full series
        remains the audit artefact and this table is only a presentation of it.

    Raises
    ------
    IngestError
        If the change series is empty or the count is not positive.
    """
    # Refuse an empty series.
    if len(changes) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot rank an empty donor change series.")
    # Refuse a non-positive count.
    if int(count) <= 0:
        # The failure names the observed count.
        raise IngestError(f"The presentation count {count!r} is not positive.")
    # Compose the frame with the absolute change as the ordering key.
    frame: pd.DataFrame = pd.DataFrame(
        {
            "donor_code": [str(code) for code in changes.index],
            "change": changes.to_numpy(dtype=float),
        }
    )
    # Compute the absolute change used for the ordering.
    frame["_absolute"] = frame["change"].abs()
    # Order by the absolute change descending and then lexicographically by
    # the donor code, which is the declared deterministic tie-break.
    frame = frame.sort_values(
        ["_absolute", "donor_code"], ascending=[False, True], kind="stable"
    ).reset_index(drop=True)
    # Assign the rank of each donor.
    frame["rank"] = np.arange(1, len(frame) + 1)
    # Retain the declared number of leading rows.
    frame = frame.head(int(count))
    # Drop the ordering helper and set the donor code as the index.
    frame = frame.drop(columns=["_absolute"]).set_index("donor_code")
    # Log the subset's extent against the full series.
    logger.info(
        "presentation subset built with %d of %d donor change(s).",
        len(frame),
        len(changes),
    )
    # Return the subset.
    return frame


# Compute the influence diagnostics of one pair.
def compute_pair_influence(
    policy_name: str,
    outcome_name: str,
    baseline_config: Mapping[str, Any],
    data: Mapping[str, Any],
    baseline_result: BaselineResult,
    count: int = TOP_INFLUENCE_COUNT,
) -> InfluenceResult:
    """
    Compute the influence diagnostics of one pair.

    Parameters
    ----------
    policy_name : str
        The policy instrument of the pair.
    outcome_name : str
        The canonical outcome variable of the pair.
    baseline_config : mapping
        The baseline configuration.
    data : mapping
        The frozen inputs under the declared keys.
    baseline_result : BaselineResult
        The cached baseline result, which is computed once per pair and reused
        by every diagnostic rather than recomputed.
    count : int, optional
        The number of rows the presentation subset carries.

    Returns
    -------
    InfluenceResult
        The treated series and bounds, the donor changes, the distributional
        statistics and the presentation subset.

    Raises
    ------
    IngestError
        If any reduction, identity, iteration or ordering check fails.
    """
    # Read the pair's baseline partition.
    partition: Partition = data["registry"].partitions[(policy_name, outcome_name)]
    # Compute the treated leave-one-out series from the stored contrasts.
    loo: pd.Series = compute_loo_treated(baseline_result.incremental_contrasts)
    # Assert that the reduction used the stored contrasts alone.
    assert_loo_treated_identity(baseline_result.incremental_contrasts, loo)
    # Compute the distributional statistics of the contrasts.
    distributional: Dict[str, float] = compute_distributional_statistics(
        baseline_result.incremental_contrasts
    )
    # Compute the donor deletion changes.
    changes: pd.Series = compute_donor_changes(
        policy_name,
        outcome_name,
        partition,
        baseline_config,
        data,
        float(baseline_result.att_incremental),
    )
    # Build the presentation subset.
    top: pd.DataFrame = build_top_influence(changes, count)
    # Return the frozen influence record.
    return InfluenceResult(
        policy_name=str(policy_name),
        outcome_name=str(outcome_name),
        baseline_att=float(baseline_result.att_incremental),
        loo_treated=loo,
        loo_treated_min=float(loo.min()),
        loo_treated_max=float(loo.max()),
        donor_changes=changes,
        distributional=distributional,
        top_influence=top,
    )


# Freeze the diagnostics behind a canonical digest.
def freeze_influence_diagnostics(
    donor_changes: pd.DataFrame,
    loo_treated: pd.DataFrame,
    distributional: pd.DataFrame,
) -> str:
    """
    Freeze the diagnostics behind a canonical digest.

    Parameters
    ----------
    donor_changes : pandas.DataFrame
        The combined donor changes.
    loo_treated : pandas.DataFrame
        The combined treated series.
    distributional : pandas.DataFrame
        The distributional statistics.

    Returns
    -------
    str
        The hexadecimal digest of the canonical payload.

    Raises
    ------
    IngestError
        If any frame is empty.
    """
    # Refuse an empty frame.
    for label, frame in (
        ("donor changes", donor_changes),
        ("treated series", loo_treated),
        ("distributional", distributional),
    ):
        # Test the frame's extent.
        if len(frame) == 0:
            # The failure names the frame.
            raise IngestError(f"Cannot freeze an empty {label} frame.")
    # Accumulate the canonical payload parts.
    parts: List[str] = []
    # Record every donor change in sorted key order.
    for (policy, outcome, donor), row in donor_changes.sort_index().iterrows():
        # Append the change at full precision.
        parts.append(
            "D|" + str(policy) + "|" + str(outcome) + "|" + str(donor) + "|"
            + f"{float(row['change']):.17g}"
        )
    # Record every treated value in sorted key order.
    for (policy, outcome, dropped), row in loo_treated.sort_index().iterrows():
        # Append the value at full precision.
        parts.append(
            "T|" + str(policy) + "|" + str(outcome) + "|" + str(dropped) + "|"
            + f"{float(row['loo_treated_att']):.17g}"
        )
    # Record every distributional row in sorted key order.
    for (policy, outcome), row in distributional.sort_index().iterrows():
        # Append the statistics at full precision.
        parts.append(
            "S|" + str(policy) + "|" + str(outcome) + "|"
            + "|".join(
                f"{float(row[column]):.17g}"
                for column in (STAT_POSITIVE_SHARE, STAT_MEDIAN, STAT_TRIMMED_MEAN)
            )
        )
    # Digest the encoded payload.
    return hashlib.sha256("\n".join(parts).encode("utf-8")).hexdigest()


# -------------------------------------------------------------------------------------------------------------------------------
# Task 21, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Execute the influence diagnostics across the pairs.
def execute_influence_diagnostics(
    baseline_config: Mapping[str, Any],
    data: Mapping[str, Any],
    pairs: Optional[Sequence[Tuple[str, str]]] = None,
    count: int = TOP_INFLUENCE_COUNT,
) -> InfluenceDiagnosticsResult:
    """
    Execute the influence diagnostics across the pairs.

    Parameters
    ----------
    baseline_config : mapping
        The baseline configuration, which is never mutated.
    data : mapping
        The frozen inputs under the declared keys.
    pairs : sequence of tuple, optional
        The pairs the execution covers. When absent every registered pair is
        covered.
    count : int, optional
        The number of rows each presentation subset carries.

    Returns
    -------
    InfluenceDiagnosticsResult
        The per-pair records, the combined frames, the digest, the timestamp
        and the notes.

    Raises
    ------
    KeyError
        If the data mapping lacks a declared key.
    IngestError
        If a pair is absent from the registry, or if any diagnostic fails.
    """
    # Refuse a data mapping lacking the declared keys.
    for key in ("registry", "panel", "lagged"):
        # Test the key's presence.
        if key not in data:
            # The failure names the absent key.
            raise KeyError(f"The data mapping lacks the declared key {key!r}.")
    # Compose the pair list, defaulting to every registered pair.
    selected: List[Tuple[str, str]] = (
        [tuple(pair) for pair in pairs]
        if pairs is not None
        else sorted(data["registry"].partitions.keys())
    )
    # Accumulate the per-pair records.
    results: Dict[Tuple[str, str], InfluenceResult] = {}
    # Accumulate the donor change rows.
    change_rows: List[Dict[str, Any]] = []
    # Accumulate the treated rows.
    treated_rows: List[Dict[str, Any]] = []
    # Accumulate the distributional rows.
    distributional_rows: List[Dict[str, Any]] = []
    # Accumulate the iteration counts for the notes.
    iteration_counts: Dict[str, int] = {}
    # Iterate over the pairs in sorted order.
    for pair in sorted(set(tuple(item) for item in selected)):
        # Refuse a pair the registry does not carry.
        if pair not in data["registry"].partitions:
            # The failure names the pair.
            raise IngestError(f"The registry does not carry the pair {pair}.")
        # Compute the baseline once per pair and cache it for every
        # diagnostic, which is the declared efficiency discipline.
        baseline_result: BaselineResult = run_baseline(
            pair[0], pair[1], baseline_config, data
        )
        # Compute the pair's influence diagnostics.
        record: InfluenceResult = compute_pair_influence(
            pair[0], pair[1], baseline_config, data, baseline_result, count
        )
        # Record the pair's diagnostics.
        results[pair] = record
        # Record the iteration count.
        iteration_counts[f"{pair[0]}/{pair[1]}"] = int(len(record.donor_changes))
        # Append the donor change rows of the pair.
        for donor, value in record.donor_changes.items():
            # Record the donor's change.
            change_rows.append(
                {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "donor_code": str(donor),
                    "change": float(value),
                }
            )
        # Append the treated rows of the pair.
        for dropped, value in record.loo_treated.items():
            # Record the dropped country's value.
            treated_rows.append(
                {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "dropped_country": str(dropped),
                    "loo_treated_att": float(value),
                }
            )
        # Append the distributional row of the pair.
        distributional_rows.append(
            {
                "policy": pair[0],
                "outcome": pair[1],
                STAT_POSITIVE_SHARE: float(record.distributional[STAT_POSITIVE_SHARE]),
                STAT_MEDIAN: float(record.distributional[STAT_MEDIAN]),
                STAT_TRIMMED_MEAN: float(record.distributional[STAT_TRIMMED_MEAN]),
            }
        )
    # Assemble the combined donor change frame.
    donor_changes: pd.DataFrame = pd.DataFrame(change_rows).set_index(
        ["policy", "outcome", "donor_code"]
    )
    # Assemble the combined treated frame.
    loo_treated: pd.DataFrame = pd.DataFrame(treated_rows).set_index(
        ["policy", "outcome", "dropped_country"]
    )
    # Assemble the distributional frame.
    distributional: pd.DataFrame = pd.DataFrame(distributional_rows).set_index(
        ["policy", "outcome"]
    )
    # Assert that the combined frames cover the selected pairs exactly.
    for label, frame in (
        ("donor changes", donor_changes),
        ("treated series", loo_treated),
        ("distributional", distributional),
    ):
        # Compose the covered pairs.
        covered = set(
            (str(policy), str(outcome))
            for policy, outcome in zip(
                frame.index.get_level_values(0), frame.index.get_level_values(1)
            )
        )
        # Refuse a frame whose coverage diverges from the selected pairs.
        if covered != set(selected):
            # The failure names the frame and the divergence.
            missing = sorted(set(selected) - covered)
            # Raise with the missing pairs attached.
            raise IngestError(
                f"The {label} frame omits the pair(s) {missing}."
            )
    # Freeze the diagnostics.
    fingerprint: str = freeze_influence_diagnostics(
        donor_changes, loo_treated, distributional
    )
    # Accumulate the recorded notes.
    notes: List[str] = [
        "the treated diagnostic reduces the stored contrasts and never refits "
        "the design; its identity is asserted numerically.",
        "the donor diagnostic rebuilds the design for every deleted donor "
        "through the frozen baseline interface rather than approximating it.",
        f"the trim is rank based: the declared proportion of the roster is "
        f"dropped from each tail, so a short roster trims nothing.",
        f"the presentation subset carries the {count} largest absolute changes "
        f"with lexicographic tie-breaking; the full series remains the audit "
        f"artefact.",
    ]
    # Record the iteration counts.
    notes.append(f"donor iterations executed: {iteration_counts}.")
    # Assemble the immutable result.
    result: InfluenceDiagnosticsResult = InfluenceDiagnosticsResult(
        results=results,
        donor_changes=donor_changes,
        loo_treated=loo_treated,
        distributional=distributional,
        fingerprint_sha256=fingerprint,
        completed_utc=current_utc_iso(),
        notes=tuple(notes),
    )
    # Log the completed execution.
    logger.info(
        "influence diagnostics complete: %d pair(s), %d donor change(s), "
        "fingerprint %s.",
        len(results),
        len(donor_changes),
        fingerprint[:16],
    )
    # Return the assembled result.
    return result
````

````python
# Task 22: Execute Influence Diagnostics

# ==============================================================================
# Task 22: Execute Influence Diagnostics
# ==============================================================================

# Declare the positive-share threshold below which the coalition response is
# marked as muted, which operationalises the specification's requirement that
# a less synchronized response be highlighted beside the mean.
MUTED_RESPONSE_THRESHOLD: Final[float] = 0.80
# Declare the column names of the range table.
RANGE_COLUMNS: Final[Tuple[str, ...]] = (
    "minimum_loo_att",
    "maximum_loo_att",
    "range_width",
    "min_source_country",
    "max_source_country",
    "baseline_att",
)
# Declare the tokens whose presence in a column name would imply interval
# semantics, which this reduction must never carry.
INTERVAL_TOKENS: Final[Tuple[str, ...]] = ("ci_", "interval", "p_value", "confidence")


# Declare the frozen tables of the influence execution.
@dataclass(frozen=True)
class InfluenceTableResult:
    """
    Frozen influence tables of the executed pairs.

    Attributes
    ----------
    treated_range : pandas.DataFrame
        One row per partition carrying the minimum, the maximum, the width,
        the extreme-source countries and the baseline effect. The range is a
        reduction of the leave-one-out estimates and not a confidence
        interval, and no interval-style field is emitted.
    top_changes : pandas.DataFrame
        The largest absolute donor changes with their ranks, keyed on the
        partition, the donor and the rank, with the signs preserved.
    max_change : pandas.DataFrame
        One row per partition naming the donor whose deletion moves the effect
        most in absolute value, with the change and the pool size.
    distribution : pandas.DataFrame
        One row per partition carrying the positive share, the median, the
        trimmed mean, the mean contrast and the muted-response flag.
    combined : pandas.DataFrame
        The donor change subset and the extremum assembled into one frame for
        the reconciliation task.
    fingerprint_sha256 : str
        Digest of the canonical table payload.
    completed_utc : str
        ISO-8601 UTC timestamp of the execution.
    notes : tuple of str
        Recorded decisions, including the engine digest the tables were built
        from and the interpretation rules.
    """
    # One row per partition carrying the treated range.
    treated_range: pd.DataFrame
    # The largest absolute donor changes with their ranks.
    top_changes: pd.DataFrame
    # One row per partition naming the largest absolute change.
    max_change: pd.DataFrame
    # One row per partition carrying the distributional diagnostics.
    distribution: pd.DataFrame
    # The combined donor change frame.
    combined: pd.DataFrame
    # Digest of the canonical table payload.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the execution.
    completed_utc: str
    # Recorded decisions of the execution.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 22, Step 1: The treated-country leave-one-out range
# -------------------------------------------------------------------------------------------------------------------------------


# Build the treated leave-one-out range table.
def build_treated_range(
    results: Mapping[Tuple[str, str], InfluenceResult],
) -> pd.DataFrame:
    """
    Build the treated leave-one-out range table.

    Parameters
    ----------
    results : mapping of tuple to InfluenceResult
        The influence records keyed on the pair.

    Returns
    -------
    pandas.DataFrame
        One row per partition carrying the minimum, the maximum, the width,
        the extreme-source countries and the baseline effect.

    Raises
    ------
    IngestError
        If the records are empty, if a series is empty, or if a ranged pair
        carries fewer than two leave-one-out estimates, because a range over
        one value is undefined.
    """
    # Refuse an empty mapping.
    if not results:
        # The failure states the precondition.
        raise IngestError("Cannot range an empty influence record set.")
    # Accumulate the range rows.
    rows: List[Dict[str, Any]] = []
    # Iterate over the pairs in sorted order.
    for pair in sorted(results.keys()):
        # Select the pair's record.
        record = results[pair]
        # Refuse a series too short to range.
        if len(record.loo_treated) < 2:
            # The failure names the pair and the observed length.
            raise IngestError(
                f"{pair}: the treated series carries {len(record.loo_treated)} "
                f"value(s), which is too few to range."
            )
        # Compose the minimum and its source country.
        min_country = str(record.loo_treated.idxmin())
        # Compose the maximum and its source country.
        max_country = str(record.loo_treated.idxmax())
        # Compose the minimum value.
        minimum: float = float(record.loo_treated.min())
        # Compose the maximum value.
        maximum: float = float(record.loo_treated.max())
        # Append the partition's range row, which carries no interval-style
        # field because the range is a reduction rather than an interval.
        rows.append(
            {
                "policy": pair[0],
                "outcome": pair[1],
                "minimum_loo_att": minimum,
                "maximum_loo_att": maximum,
                "range_width": maximum - minimum,
                "min_source_country": min_country,
                "max_source_country": max_country,
                "baseline_att": float(record.baseline_att),
            }
        )
    # Assemble the range frame keyed on the pair.
    table: pd.DataFrame = pd.DataFrame(rows).set_index(["policy", "outcome"])
    # Log the reduction's extent.
    logger.info("treated range built across %d partition(s).", len(table))
    # Return the table.
    return table


# Assert that a range table carries no interval semantics.
def assert_range_carries_no_interval_semantics(table: pd.DataFrame) -> None:
    """
    Assert that a range table carries no interval semantics.

    Parameters
    ----------
    table : pandas.DataFrame
        The range table.

    Returns
    -------
    None
        The function returns nothing when no interval-style field appears.

    Raises
    ------
    IngestError
        If a column name implies a confidence interval or a probability,
        because the specification fixes that the range is a reduction of the
        leave-one-out estimates and not a probabilistic statement.
    """
    # Identify the columns whose names imply interval semantics.
    offending: List[str] = [
        str(column)
        for column in table.columns
        if any(token in str(column).lower() for token in INTERVAL_TOKENS)
    ]
    # Refuse an interval-style column.
    if offending:
        # The failure names the offending columns.
        raise IngestError(
            f"The range table carries column(s) implying interval semantics: "
            f"{offending}; the range is a reduction, not a confidence interval."
        )
    # Log the verified absence.
    logger.info(
        "range table verified free of interval semantics across %d column(s).",
        len(table.columns),
    )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 22, Step 2: The donor leave-one-out tables
# -------------------------------------------------------------------------------------------------------------------------------


# Build the presentation subsets of the donor changes.
def build_top_changes(
    results: Mapping[Tuple[str, str], InfluenceResult],
    count: int = TOP_INFLUENCE_COUNT,
) -> pd.DataFrame:
    """
    Build the presentation subsets of the donor changes.

    Parameters
    ----------
    results : mapping of tuple to InfluenceResult
        The influence records keyed on the pair.
    count : int, optional
        The number of rows each partition's subset carries.

    Returns
    -------
    pandas.DataFrame
        The largest absolute donor changes keyed on the partition, the donor
        and the rank, with the signs preserved.

    Raises
    ------
    IngestError
        If the records are empty, because the subsets would then be vacuous.
    """
    # Refuse an empty mapping.
    if not results:
        # The failure states the precondition.
        raise IngestError("Cannot build subsets from an empty record set.")
    # Accumulate the subset rows.
    rows: List[Dict[str, Any]] = []
    # Iterate over the pairs in sorted order.
    for pair in sorted(results.keys()):
        # Read the pair's donor change series.
        changes = results[pair].donor_changes
        # Build the subset through the engine's own builder, which owns the
        # ordering rule and the deterministic tie-break.
        subset: pd.DataFrame = build_top_influence(changes, count)
        # Append the subset's rows with their pair keys.
        for donor, row in subset.iterrows():
            # Record the donor's change and rank.
            rows.append(
                {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "donor_code": str(donor),
                    "rank": int(row["rank"]),
                    "change": float(row["change"]),
                    "pool_size": int(len(changes)),
                }
            )
    # Assemble the subset frame keyed on the pair, the donor and the rank.
    table: pd.DataFrame = pd.DataFrame(rows).set_index(
        ["policy", "outcome", "donor_code"]
    )
    # Log the subsets' extent.
    logger.info(
        "donor subsets built across %d row(s) over %d partition(s).",
        len(table),
        len(results),
    )
    # Return the table.
    return table


# Build the extremum table of the donor changes.
def build_max_change(
    results: Mapping[Tuple[str, str], InfluenceResult],
) -> pd.DataFrame:
    """
    Build the extremum table of the donor changes.

    Parameters
    ----------
    results : mapping of tuple to InfluenceResult
        The influence records keyed on the pair.

    Returns
    -------
    pandas.DataFrame
        One row per partition naming the largest absolute change, with the
        change, the pool size and the sign preserved.

    Raises
    ------
    IngestError
        If the records are empty or a change series is empty.
    """
    # Refuse an empty mapping.
    if not results:
        # The failure states the precondition.
        raise IngestError("Cannot reduce an empty record set.")
    # Accumulate the extremum rows.
    rows: List[Dict[str, Any]] = []
    # Iterate over the pairs in sorted order.
    for pair in sorted(results.keys()):
        # Read the pair's donor change series.
        changes = results[pair].donor_changes
        # Refuse an empty series.
        if len(changes) == 0:
            # The failure names the pair.
            raise IngestError(f"{pair}: the donor change series is empty.")
        # Compose the absolute changes.
        absolute = changes.abs()
        # Compose the largest absolute change.
        largest: float = float(absolute.max())
        # Compose the candidates sharing the largest absolute change.
        candidates = sorted(str(code) for code in absolute[absolute == largest].index)
        # Refuse a candidate set without a lexicographic winner, which cannot
        # occur but documents the tie-break's determinism.
        if not candidates:
            # The failure names the pair.
            raise IngestError(f"{pair}: no donor attains the largest absolute change.")
        # Append the partition's extremum row with the first candidate.
        rows.append(
            {
                "policy": pair[0],
                "outcome": pair[1],
                "donor_code": candidates[0],
                "max_absolute_change": largest,
                "signed_change": float(changes[candidates[0]]),
                "pool_size": int(len(changes)),
                "tied_donors": "; ".join(candidates),
            }
        )
    # Assemble the extremum frame keyed on the pair.
    table: pd.DataFrame = pd.DataFrame(rows).set_index(["policy", "outcome"])
    # Log the reduction's extent.
    logger.info("extremum table built across %d partition(s).", len(table))
    # Return the table.
    return table


# -------------------------------------------------------------------------------------------------------------------------------
# Task 22, Step 3: The distributional diagnostics table
# -------------------------------------------------------------------------------------------------------------------------------


# Build the distributional diagnostics table.
def build_distribution_table(
    results: Mapping[Tuple[str, str], InfluenceResult],
    contrasts_by_pair: Mapping[Tuple[str, str], pd.Series],
    muted_threshold: float = MUTED_RESPONSE_THRESHOLD,
) -> pd.DataFrame:
    """
    Build the distributional diagnostics table.

    Parameters
    ----------
    results : mapping of tuple to InfluenceResult
        The influence records keyed on the pair.
    contrasts_by_pair : mapping of tuple to pandas.Series
        The per-country contrasts keyed on the pair, from which the mean is
        composed.
    muted_threshold : float, optional
        The positive-share threshold below which the coalition response is
        marked as muted; it is a descriptive marker rather than a statistical
        test.

    Returns
    -------
    pandas.DataFrame
        One row per partition carrying the positive share, the median, the
        trimmed mean, the mean contrast and the muted-response flag.

    Raises
    ------
    IngestError
        If the records are empty, if a pair's contrasts are absent, or if the
        threshold lies outside the open unit interval.
    """
    # Refuse an empty mapping.
    if not results:
        # The failure states the precondition.
        raise IngestError("Cannot summarise an empty record set.")
    # Refuse a threshold outside the admissible range.
    if not (0.0 < float(muted_threshold) < 1.0):
        # The failure names the observed threshold.
        raise IngestError(
            f"The muted-response threshold {muted_threshold!r} lies outside the "
            f"open unit interval."
        )
    # Accumulate the distributional rows.
    rows: List[Dict[str, Any]] = []
    # Iterate over the pairs in sorted order.
    for pair in sorted(results.keys()):
        # Refuse a pair whose contrasts were not supplied.
        if pair not in contrasts_by_pair:
            # The failure names the pair.
            raise IngestError(f"{pair}: the contrasts were not supplied.")
        # Read the pair's record.
        record = results[pair]
        # Read the pair's contrasts.
        contrasts: pd.Series = contrasts_by_pair[pair]
        # Compose the positive share from the engine's own statistic.
        positive_share: float = float(record.distributional[STAT_POSITIVE_SHARE])
        # Append the partition's distributional row with the mean beside the
        # statistics and the descriptive flag.
        rows.append(
            {
                "policy": pair[0],
                "outcome": pair[1],
                STAT_POSITIVE_SHARE: positive_share,
                STAT_MEDIAN: float(record.distributional[STAT_MEDIAN]),
                STAT_TRIMMED_MEAN: float(record.distributional[STAT_TRIMMED_MEAN]),
                "mean_contrast": float(np.mean(contrasts.to_numpy(dtype=float))),
                "muted_response": bool(positive_share < float(muted_threshold)),
            }
        )
    # Assemble the distributional frame keyed on the pair.
    table: pd.DataFrame = pd.DataFrame(rows).set_index(["policy", "outcome"])
    # Log the table's extent and the muted cases.
    logger.info(
        "distributional table built across %d partition(s) with %d muted "
        "response(s).",
        len(table),
        int(table["muted_response"].sum()),
    )
    # Return the table.
    return table


# Compare the tables with the declared study values.
def assert_declared_tables(
    treated_range: pd.DataFrame,
    max_change: pd.DataFrame,
    distribution: pd.DataFrame,
    expected_ranges: Optional[Mapping[Tuple[str, str], Tuple[float, float]]] = None,
    expected_max_change: Optional[Mapping[Tuple[str, str], float]] = None,
    expected_distribution: Optional[
        Mapping[Tuple[str, str], Mapping[str, float]]
    ] = None,
    tolerance: float = 1.0e-3,
) -> None:
    """
    Compare the tables with the declared study values.

    Parameters
    ----------
    treated_range : pandas.DataFrame
        The treated range table.
    max_change : pandas.DataFrame
        The extremum table.
    distribution : pandas.DataFrame
        The distributional table.
    expected_ranges : mapping of tuple to tuple, optional
        The declared treated ranges per partition.
    expected_max_change : mapping of tuple to float, optional
        The declared largest absolute donor changes per partition.
    expected_distribution : mapping of tuple to mapping of str to float, optional
        The declared distributional statistics per partition.
    tolerance : float, optional
        The tolerance of the comparisons.

    Returns
    -------
    None
        The function returns nothing when every supplied comparison holds.

    Raises
    ------
    IngestError
        If an observed value diverges from its declared counterpart, or if a
        declared partition is absent from a table.
    """
    # Compare the treated ranges when declared values were supplied.
    if expected_ranges is not None:
        # Iterate over the declared partitions.
        for pair, (declared_min, declared_max) in expected_ranges.items():
            # Refuse a partition the range table does not report.
            if pair not in treated_range.index:
                # The failure names the absent partition.
                raise IngestError(f"The range table omits the declared pair {pair}.")
            # Read the observed bounds.
            observed_min: float = float(treated_range.loc[pair, "minimum_loo_att"])
            # Read the observed maximum.
            observed_max: float = float(treated_range.loc[pair, "maximum_loo_att"])
            # Refuse a divergence beyond the tolerance.
            if (
                abs(observed_min - float(declared_min)) > float(tolerance)
                or abs(observed_max - float(declared_max)) > float(tolerance)
            ):
                # The failure names the partition and both ranges.
                raise IngestError(
                    f"{pair}: observed range [{observed_min!r}, {observed_max!r}] "
                    f"against the declared [{declared_min!r}, {declared_max!r}]."
                )
    # Compare the largest absolute changes when declared values were supplied.
    if expected_max_change is not None:
        # Iterate over the declared partitions.
        for pair, declared in expected_max_change.items():
            # Refuse a partition the extremum table does not report.
            if pair not in max_change.index:
                # The failure names the absent partition.
                raise IngestError(f"The extremum table omits the declared pair {pair}.")
            # Read the observed value.
            observed: float = float(max_change.loc[pair, "max_absolute_change"])
            # Refuse a divergence beyond the tolerance.
            if abs(observed - float(declared)) > float(tolerance):
                # The failure names the partition and both values.
                raise IngestError(
                    f"{pair}: largest absolute change {observed!r} against the "
                    f"declared {declared!r}."
                )
    # Compare the distributional statistics when declared values were supplied.
    if expected_distribution is not None:
        # Iterate over the declared partitions.
        for pair, declared in expected_distribution.items():
            # Refuse a partition the distributional table does not report.
            if pair not in distribution.index:
                # The failure names the absent partition.
                raise IngestError(
                    f"The distributional table omits the declared pair {pair}."
                )
            # Compare each declared statistic.
            for statistic, value in declared.items():
                # Refuse a statistic the table does not carry.
                if statistic not in distribution.columns:
                    # The failure names the absent statistic.
                    raise IngestError(
                        f"The distributional table omits the statistic "
                        f"{statistic!r}."
                    )
                # Read the observed statistic.
                observed: float = float(distribution.loc[pair, statistic])
                # Refuse a divergence beyond the tolerance.
                if abs(observed - float(value)) > float(tolerance):
                    # The failure names the partition and both values.
                    raise IngestError(
                        f"{pair}/{statistic}: {observed!r} against the declared "
                        f"{value!r}."
                    )
    # Log the comparison mode.
    logger.info(
        "declared table comparison %s.",
        "performed"
        if (expected_ranges or expected_max_change or expected_distribution)
        else "skipped",
    )


# Freeze the tables behind a canonical digest.
def freeze_influence_tables(
    treated_range: pd.DataFrame,
    top_changes: pd.DataFrame,
    max_change: pd.DataFrame,
    distribution: pd.DataFrame,
) -> str:
    """
    Freeze the tables behind a canonical digest.

    Parameters
    ----------
    treated_range : pandas.DataFrame
        The treated range table.
    top_changes : pandas.DataFrame
        The donor subset table.
    max_change : pandas.DataFrame
        The extremum table.
    distribution : pandas.DataFrame
        The distributional table.

    Returns
    -------
    str
        The hexadecimal digest of the canonical payload.

    Raises
    ------
    IngestError
        If any table is empty.
    """
    # Refuse an empty table.
    for label, table in (
        ("treated range", treated_range),
        ("donor subsets", top_changes),
        ("extremum", max_change),
        ("distributional", distribution),
    ):
        # Test the table's extent.
        if len(table) == 0:
            # The failure names the table.
            raise IngestError(f"Cannot freeze an empty {label} table.")
    # Accumulate the canonical payload parts.
    parts: List[str] = []
    # Record every range row in sorted order.
    for (policy, outcome), row in treated_range.sort_index().iterrows():
        # Append the row's canonical line at full precision.
        parts.append(
            "G|" + str(policy) + "|" + str(outcome) + "|"
            + f"{float(row['minimum_loo_att']):.17g}" + "|"
            + f"{float(row['maximum_loo_att']):.17g}" + "|"
            + f"{float(row['baseline_att']):.17g}"
        )
    # Record every subset row in sorted order.
    for (policy, outcome, donor), row in top_changes.sort_index().iterrows():
        # Append the row's canonical line at full precision.
        parts.append(
            "K|" + str(policy) + "|" + str(outcome) + "|" + str(donor) + "|"
            + f"{float(row['change']):.17g}" + "|" + str(int(row["rank"]))
        )
    # Record every extremum row in sorted order.
    for (policy, outcome), row in max_change.sort_index().iterrows():
        # Append the row's canonical line at full precision.
        parts.append(
            "M|" + str(policy) + "|" + str(outcome) + "|" + str(row["donor_code"])
            + "|" + f"{float(row['max_absolute_change']):.17g}"
        )
    # Record every distributional row in sorted order.
    for (policy, outcome), row in distribution.sort_index().iterrows():
        # Append the row's canonical line at full precision.
        parts.append(
            "S|" + str(policy) + "|" + str(outcome) + "|"
            + "|".join(
                f"{float(row[column]):.17g}"
                for column in (
                    STAT_POSITIVE_SHARE,
                    STAT_MEDIAN,
                    STAT_TRIMMED_MEAN,
                    "mean_contrast",
                )
            )
            + "|" + str(bool(row["muted_response"]))
        )
    # Digest the encoded payload.
    return hashlib.sha256("\n".join(parts).encode("utf-8")).hexdigest()


# -------------------------------------------------------------------------------------------------------------------------------
# Task 22, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Execute the influence tables from the engine's result.
def execute_influence_tables(
    influence: InfluenceDiagnosticsResult,
    contrasts_by_pair: Mapping[Tuple[str, str], pd.Series],
    count: int = TOP_INFLUENCE_COUNT,
    muted_threshold: float = MUTED_RESPONSE_THRESHOLD,
    expected_ranges: Optional[Mapping[Tuple[str, str], Tuple[float, float]]] = None,
    expected_max_change: Optional[Mapping[Tuple[str, str], float]] = None,
    expected_distribution: Optional[
        Mapping[Tuple[str, str], Mapping[str, float]]
    ] = None,
) -> InfluenceTableResult:
    """
    Execute the influence tables from the engine's result.

    Parameters
    ----------
    influence : InfluenceDiagnosticsResult
        The influence engine's result.
    contrasts_by_pair : mapping of tuple to pandas.Series
        The per-country contrasts keyed on the pair, used for the mean.
    count : int, optional
        The number of rows each donor subset carries.
    muted_threshold : float, optional
        The positive-share threshold of the muted-response marker.
    expected_ranges : mapping of tuple to tuple, optional
        The declared treated ranges.
    expected_max_change : mapping of tuple to float, optional
        The declared largest absolute changes.
    expected_distribution : mapping of tuple to mapping of str to float, optional
        The declared distributional statistics.

    Returns
    -------
    InfluenceTableResult
        The range, subset, extremum, distributional and combined tables with
        the digest, the timestamp and the notes.

    Raises
    ------
    IngestError
        If any reduction, coverage, semantics or comparison check fails.
    """
    # Build the treated range table.
    treated_range: pd.DataFrame = build_treated_range(influence.results)
    # Assert that the range carries no interval semantics.
    assert_range_carries_no_interval_semantics(treated_range)
    # Build the donor subsets.
    top_changes: pd.DataFrame = build_top_changes(influence.results, count)
    # Build the extremum table.
    max_change: pd.DataFrame = build_max_change(influence.results)
    # Build the distributional table.
    distribution: pd.DataFrame = build_distribution_table(
        influence.results, contrasts_by_pair, muted_threshold
    )
    # Assemble the combined frame from the same part frames rather than
    # rebuilding it, so the combined artefact cannot drift from its parts.
    combined: pd.DataFrame = top_changes.copy()
    # Attach the extremum's donor flag so a reader can see which subset row is
    # the largest absolute change.
    largest_donors = {
        (str(policy), str(outcome)): str(row["donor_code"])
        for (policy, outcome), row in max_change.iterrows()
    }
    # Record the flag on the subset rows.
    combined["is_largest_change"] = [
        largest_donors.get((str(policy), str(outcome))) == str(donor)
        for policy, outcome, donor in combined.index
    ]
    # Assert that every table covers the engine's pairs exactly.
    for label, table in (
        ("treated range", treated_range),
        ("extremum", max_change),
        ("distributional", distribution),
    ):
        # Compose the table's covered pairs.
        covered = set(
            (str(policy), str(outcome))
            for policy, outcome in zip(
                table.index.get_level_values(0), table.index.get_level_values(1)
            )
        )
        # Refuse a coverage divergence.
        if covered != set(influence.results.keys()):
            # The failure names the table and the missing pairs.
            missing = sorted(set(influence.results.keys()) - covered)
            # Raise with the missing pairs attached.
            raise IngestError(f"The {label} table omits the pair(s) {missing}.")
    # Compare the tables with the declared values when supplied.
    assert_declared_tables(
        treated_range,
        max_change,
        distribution,
        expected_ranges=expected_ranges,
        expected_max_change=expected_max_change,
        expected_distribution=expected_distribution,
    )
    # Freeze the tables.
    fingerprint: str = freeze_influence_tables(
        treated_range, top_changes, max_change, distribution
    )
    # Accumulate the recorded notes.
    notes: List[str] = [
        f"the tables were built from the influence engine's digest "
        f"{influence.fingerprint_sha256[:16]}.",
        "the treated range is a reduction of the leave-one-out estimates and "
        "not a confidence interval; the module asserts that the range table "
        "carries no interval-style field.",
        "the donor subset reports both signs of the change, ordered by absolute "
        "value with lexicographic tie-breaking.",
        "the distributional table reports the mean beside the statistics, and "
        "the muted-response marker flags a coalition whose positive share falls "
        "below the declared threshold.",
    ]
    # Record the muted cases.
    notes.append(
        f"partitions marked with a muted response: "
        f"{sorted(str(p) + '/' + str(o) for (p, o), r in distribution.iterrows() if bool(r['muted_response']))}."
    )
    # Assemble the immutable result.
    result: InfluenceTableResult = InfluenceTableResult(
        treated_range=treated_range,
        top_changes=top_changes,
        max_change=max_change,
        distribution=distribution,
        combined=combined,
        fingerprint_sha256=fingerprint,
        completed_utc=current_utc_iso(),
        notes=tuple(notes),
    )
    # Log the completed execution.
    logger.info(
        "influence tables complete: %d partition(s), %d subset row(s), "
        "fingerprint %s.",
        len(treated_range),
        len(top_changes),
        fingerprint[:16],
    )
    # Return the assembled result.
    return result
````

````python
# Task 23: Build Orchestrator 5 - Matching Designs and Balance

# ==============================================================================
# Task 23: Build Orchestrator 5 - Matching Designs and Balance
# ==============================================================================

# Declare the matching axis of the median restriction.
AXIS_HIGH_INCOME: Final[str] = "high_income"
# Declare the matching axis of the frequency-weighted nearest neighbours.
AXIS_MAHALANOBIS: Final[str] = "mahalanobis"
# Compose the declared axis order.
MATCHING_AXES: Final[Tuple[str, ...]] = (AXIS_HIGH_INCOME, AXIS_MAHALANOBIS)
# Declare the level outcome used in the restriction and in the covariate
# vector, whose window average defines the income rank.
INCOME_COLUMN: Final[str] = "log_real_GDPpc"
# Declare the pre-treatment covariate window of the balance computation.
BALANCE_WINDOW: Final[Tuple[int, int]] = (2009, 2013)
# Declare the default number of nearest donors the frequency axis selects per
# target.
DEFAULT_MATCH_SIZE: Final[int] = 5
# Declare the minimum number of units a group must carry for the sample
# variance to be defined.
MINIMUM_GROUP_SIZE: Final[int] = 2
# Compose the declared covariate dimensions of the balance vector.
COVARIATE_DIMENSIONS: Final[Tuple[str, ...]] = tuple(CONTROL_COLUMNS) + (INCOME_COLUMN,)


# Declare the frozen result of one matching design.
@dataclass(frozen=True)
class MatchingDesignResult:
    """
    Frozen result of one matching design.

    Attributes
    ----------
    policy_name : str
        The policy instrument of the pair.
    outcome_name : str
        The canonical outcome variable of the pair.
    matching_axis : str
        The matching axis.
    matched_targets : tuple of str
        The matched target roster.
    matched_donors : tuple of str
        The effective donor pool. For the frequency design this is exactly the
        strictly positive-weight support, never the full candidate roster.
    weights : pandas.Series or None
        The inclusion frequency of every matched donor, present for the
        frequency axis and absent for the hard restriction.
    att : float
        The incremental effect of the matched design.
    se_cd : float
        The country-dispersion standard error of the effect.
    stars : str
        The significance marker of the effect.
    std_diff : pandas.Series
        The standardized difference of every covariate dimension.
    max_abs_std_diff : float
        The largest absolute standardized difference.
    max_abs_covariate : str
        The covariate attaining the largest absolute difference.
    std_diff_unweighted : pandas.Series or None
        The unweighted balance vector, present only when weights were applied
        so the effect of the weighting is visible.
    diagnostics : tuple of str
        The recorded diagnostics, including the covariance's conditioning and
        the exact declared neighbour count.
    candidate_targets, candidate_donors : tuple of str
        The original eligible rosters from which the design selected.
    selection_audit : pandas.DataFrame
        The complete hard-restriction dispositions or realised target-neighbour
        rows used to produce the design.
    design_provenance : mapping
        The balance window, candidate/effective rosters, scientific input
        digests, variance convention and fitted-window lineage.
    gaps : pandas.DataFrame
        The independently fitted realized post-period country-gap matrix.
    fitted_pre_gaps : pandas.DataFrame
        The independently fitted pre-period residual matrix used by Figure 4.
    """
    # The policy instrument of the pair.
    policy_name: str
    # The canonical outcome variable of the pair.
    outcome_name: str
    # The matching axis.
    matching_axis: str
    # The matched target roster.
    matched_targets: Tuple[str, ...]
    # The matched donor pool.
    matched_donors: Tuple[str, ...]
    # The inclusion frequency of every matched donor, when applicable.
    weights: Optional[pd.Series]
    # The incremental effect of the matched design.
    att: float
    # The country-dispersion standard error.
    se_cd: float
    # The significance marker.
    stars: str
    # The standardized difference of every covariate dimension.
    std_diff: pd.Series
    # The largest absolute standardized difference.
    max_abs_std_diff: float
    # The covariate attaining the largest absolute difference.
    max_abs_covariate: str
    # The unweighted balance vector, when weights were applied.
    std_diff_unweighted: Optional[pd.Series]
    # The recorded diagnostics.
    diagnostics: Tuple[str, ...]
    # The baseline target roster from which the design selected.
    candidate_targets: Tuple[str, ...] = ()
    # The baseline donor roster from which the design selected.
    candidate_donors: Tuple[str, ...] = ()
    # The realised target-neighbour selections, empty for a hard restriction.
    selection_audit: pd.DataFrame = field(default_factory=pd.DataFrame)
    # Canonical scientific inputs and lineage of the fitted design.
    design_provenance: Mapping[str, Any] = field(default_factory=dict)
    # Independently fitted realized post-period gap matrix used by Figure 4.
    gaps: pd.DataFrame = field(default_factory=pd.DataFrame)
    # Independently fitted pre-period residual matrix used by Figure 4.
    fitted_pre_gaps: pd.DataFrame = field(default_factory=pd.DataFrame)


# Declare the frozen result of the whole matching execution.
@dataclass(frozen=True)
class MatchingDiagnosticsResult:
    """
    Frozen matching diagnostics across the executed pairs and axes.

    Attributes
    ----------
    designs : dict
        The design records keyed on the pair and the matching axis.
    summary : pandas.DataFrame
        One row per design carrying the rosters' sizes, the effect with its
        inference, the largest absolute standardized difference and the
        weighted flag.
    balance : pandas.DataFrame
        The standardized differences keyed on the pair, axis, weighting scheme
        and covariate, including variance convention and zero-variance status.
    fingerprint_sha256 : str
        Digest of the canonical diagnostics payload.
    completed_utc : str
        ISO-8601 UTC timestamp of the execution.
    notes : tuple of str
        Recorded decisions of the execution.
    design_audit : pandas.DataFrame
        One row per design carrying complete roster, selection, weight and fit
        provenance used to reconcile the reported summary.
    """
    # The design records.
    designs: Mapping[Tuple[str, str, str], MatchingDesignResult]
    # The design summary.
    summary: pd.DataFrame
    # The standardized differences.
    balance: pd.DataFrame
    # Digest of the canonical diagnostics payload.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the execution.
    completed_utc: str
    # Recorded decisions of the execution.
    notes: Tuple[str, ...]
    # One row per design carrying its candidate/effective roster and fit lineage.
    design_audit: pd.DataFrame = field(default_factory=pd.DataFrame)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 23, Step 1: The pre-treatment covariate vectors
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the pre-treatment covariate vectors of a country roster.
def build_pretreatment_covariates(
    panel: pd.DataFrame,
    countries: Sequence[str],
    window: Tuple[int, int] = BALANCE_WINDOW,
) -> pd.DataFrame:
    """
    Compose the pre-treatment covariate vectors of a country roster.

    Parameters
    ----------
    panel : pandas.DataFrame
        The analysis panel carrying the control columns and the income level.
    countries : sequence of str
        The countries whose vectors are composed.
    window : tuple of int, optional
        The inclusive pre-treatment window over which each column is averaged.

    Returns
    -------
    pandas.DataFrame
        One row per country and one column per declared covariate dimension,
        each holding the country's window average.

    Raises
    ------
    IngestError
        If the panel lacks a declared covariate column, if a country is
        absent from the window, or if a cell is missing, because the balance
        vector must rest on complete vectors.
    """
    # Refuse a panel lacking a declared covariate column.
    for column in COVARIATE_DIMENSIONS:
        # Test the column's presence.
        if column not in panel.columns:
            # The failure names the absent column.
            raise IngestError(
                f"The panel lacks the declared covariate column {column!r}."
            )
    # Validate the fixed scientific window before using it as a calendar.
    window = _validate_att_window(tuple(window), "matching covariate window")
    # Compose the window years.
    years: List[int] = list(range(int(window[0]), int(window[1]) + 1))
    # Refuse duplicate country-year rows rather than averaging an ambiguous
    # history into an apparently complete vector.
    key_frame = panel[["country_code", "year"]].copy()
    key_frame["country_code"] = key_frame["country_code"].astype(str)
    numeric_years = pd.to_numeric(key_frame["year"], errors="coerce")
    if bool(numeric_years.isna().any()) or bool((numeric_years % 1 != 0).any()):
        raise IngestError("Matching covariate years must be finite integral calendars.")
    key_frame["year"] = numeric_years.astype(int)
    duplicate_keys = key_frame.duplicated(["country_code", "year"], keep=False)
    if bool(duplicate_keys.any()):
        duplicated = key_frame.loc[duplicate_keys, ["country_code", "year"]]
        raise IngestError(
            "The matching covariate panel carries duplicate country-year key(s) "
            f"{duplicated.head(8).to_dict('records')}."
        )
    # Select the roster's rows inside the window.
    scoped = panel[
        panel["country_code"].astype(str).isin(set(str(code) for code in countries))
        & (panel["year"].astype(int).isin(years))
    ]
    # Refuse any missing/non-finite annual input before group means can skip it
    # and manufacture an apparently complete five-year vector.
    raw_values = scoped[list(COVARIATE_DIMENSIONS)].apply(pd.to_numeric, errors="coerce")
    if bool(raw_values.isna().to_numpy().any()) or not bool(
        np.isfinite(raw_values.to_numpy(dtype=float)).all()
    ):
        raise IngestError(
            f"The {window[0]}-{window[1]} matching covariate history carries a "
            "missing or non-finite annual cell."
        )
    # Refuse a roster whose coverage of the window is incomplete.
    counts = scoped.groupby(scoped["country_code"].astype(str))["year"].nunique()
    # Identify the rosters with a short coverage.
    short = sorted(str(code) for code in (set(str(c) for c in countries) - set(counts.index)))
    # Refuse a country absent from the window.
    if short:
        # The failure names the absent countries.
        raise IngestError(
            f"The panel omits the covariate window for the countr(ies) {short[:8]}."
        )
    # Identify a country whose window coverage is short.
    thin = sorted(str(code) for code, value in counts.items() if int(value) != len(years))
    # Refuse a short coverage, which would make the average unrepresentative.
    if thin:
        # The failure names the thinly covered countries.
        raise IngestError(
            f"The covariate window covers fewer than {len(years)} year(s) for the "
            f"countr(ies) {thin[:8]}."
        )
    # Compose the per-country averages of every declared dimension.
    vectors: pd.DataFrame = (
        scoped.groupby(scoped["country_code"].astype(str))[list(COVARIATE_DIMENSIONS)]
        .mean()
    )
    # Refuse a missing vector cell.
    if bool(vectors.isna().to_numpy().any()) or not bool(
        np.isfinite(vectors.to_numpy(dtype=float)).all()
    ):
        # The failure states the condition.
        raise IngestError(
            "The covariate vectors carry a missing cell; the balance computation "
            "requires complete vectors."
        )
    # Log the construction's extent.
    logger.info(
        "pre-treatment covariate vectors composed across %d countr(ies) and "
        "%d dimension(s) over %s.",
        len(vectors),
        len(COVARIATE_DIMENSIONS),
        list(window),
    )
    # Return the vectors in the roster's own order where it is available.
    ordered = [str(code) for code in countries if str(code) in set(vectors.index)]
    # Return the ordered frame.
    return vectors.loc[ordered]


# Compute the median of the income level across a roster.
def compute_income_median(
    covariates: pd.DataFrame,
) -> float:
    """
    Compute the median of the income level across a roster.

    Parameters
    ----------
    covariates : pandas.DataFrame
        The pre-treatment covariate vectors.

    Returns
    -------
    float
        The median of the income column across every supplied country, which
        is the threshold the hard restriction consumes.

    Raises
    ------
    IngestError
        If the vector frame is empty or lacks the income column.
    """
    # Refuse an empty frame.
    if len(covariates) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot compose an income median from an empty roster.")
    # Refuse a frame lacking the income column.
    if INCOME_COLUMN not in covariates.columns:
        # The failure names the absent column.
        raise IngestError(
            f"The covariate frame lacks the income column {INCOME_COLUMN!r}."
        )
    # Compose the median across the roster.
    median: float = float(np.median(covariates[INCOME_COLUMN].to_numpy(dtype=float)))
    # Log the threshold.
    logger.info(
        "income median computed across %d countr(ies): %.6f.",
        len(covariates),
        median,
    )
    # Return the median.
    return median


# -------------------------------------------------------------------------------------------------------------------------------
# Task 23, Step 2: The two matching selections
# -------------------------------------------------------------------------------------------------------------------------------


# Restrict a design to the units above the income median.
def restrict_above_income_median(
    covariates: pd.DataFrame,
    targets: Sequence[str],
    donors: Sequence[str],
) -> Tuple[Tuple[str, ...], Tuple[str, ...], float]:
    """
    Restrict a design to the units above the income median.

    Parameters
    ----------
    covariates : pandas.DataFrame
        The pre-treatment covariate vectors of the union of both rosters.
    targets : sequence of str
        The target roster.
    donors : sequence of str
        The donor pool.

    Returns
    -------
    tuple
        The restricted target roster, the restricted donor pool and the median
        threshold. The restriction is a hard inclusion rule applied to both
        rosters rather than a weight, which is the declared character of the
        high-income axis.

    Raises
    ------
    IngestError
        If a roster falls below the declared minimum size, because the design
        would then be undefined.
    """
    # Compose the median threshold across both rosters together.
    median: float = compute_income_median(covariates)
    # Restrict the target roster to the units strictly above the median.
    kept_targets: Tuple[str, ...] = tuple(
        str(code)
        for code in targets
        if str(code) in set(covariates.index)
        and float(covariates.loc[str(code), INCOME_COLUMN]) > median
    )
    # Restrict the donor pool to the units strictly above the median.
    kept_donors: Tuple[str, ...] = tuple(
        str(code)
        for code in donors
        if str(code) in set(covariates.index)
        and float(covariates.loc[str(code), INCOME_COLUMN]) > median
    )
    # Refuse a target roster below the minimum size.
    if len(kept_targets) < MINIMUM_GROUP_SIZE:
        # The failure names the retained count.
        raise IngestError(
            f"The restriction retains {len(kept_targets)} target(s), below the "
            f"declared minimum of {MINIMUM_GROUP_SIZE}."
        )
    # Refuse a donor pool below the minimum size.
    if len(kept_donors) < MINIMUM_GROUP_SIZE:
        # The failure names the retained count.
        raise IngestError(
            f"The restriction retains {len(kept_donors)} donor(s), below the "
            f"declared minimum of {MINIMUM_GROUP_SIZE}."
        )
    # Log the restriction's outcome.
    logger.info(
        "income restriction applied: %d target(s) and %d donor(s) retained above "
        "the median %.6f.",
        len(kept_targets),
        len(kept_donors),
        median,
    )
    # Return the restricted rosters and the threshold.
    return kept_targets, kept_donors, median


# Compute the pooled covariance of the covariate vectors.
def compute_pooled_covariance(covariates: pd.DataFrame) -> np.ndarray:
    """
    Compute the pooled covariance of the covariate vectors.

    Parameters
    ----------
    covariates : pandas.DataFrame
        The pre-treatment covariate vectors.

    Returns
    -------
    numpy.ndarray
        The sample covariance matrix of the covariate dimensions.

    Raises
    ------
    IngestError
        If fewer than two units are supplied, because a covariance would then
        be undefined. A rank-deficient covariance, which arises when the units
        are fewer than the dimensions or when a dimension is collinear, is
        admitted rather than refused: the specification prescribes the
        pseudo-inverse for exactly that case, and the conditioning is recorded
        so the deficiency is visible.
    """
    # Compose the covariate matrix.
    matrix: np.ndarray = covariates.to_numpy(dtype=float)
    # Refuse a frame with fewer than two units, for which no covariance is
    # defined at all.
    if matrix.shape[0] < MINIMUM_GROUP_SIZE:
        # The failure names the observed count.
        raise IngestError(
            f"The pooled covariance requires at least {MINIMUM_GROUP_SIZE} unit(s); "
            f"{matrix.shape[0]} supplied."
        )
    # Record the rank-deficiency case, which the pseudo-inverse admits.
    if matrix.shape[0] <= matrix.shape[1]:
        # Log the deficiency, because a reader must know the covariance was
        # inverted through its pseudo-inverse rather than directly.
        logger.info(
            "the pooled covariance is rank deficient: %d unit(s) against %d "
            "dimension(s); the pseudo-inverse is used.",
            matrix.shape[0],
            matrix.shape[1],
        )
    # Compose the sample covariance with the declared divisor.
    covariance: np.ndarray = np.cov(matrix, rowvar=False, ddof=1)
    # Log the covariance's shape.
    logger.info(
        "pooled covariance composed from %d unit(s) across %d dimension(s).",
        matrix.shape[0],
        matrix.shape[1],
    )
    # Return the covariance.
    return np.asarray(covariance, dtype=float)


# Select the nearest donors of every target under the Mahalanobis metric.
def select_mahalanobis_neighbours(
    covariates: pd.DataFrame,
    targets: Sequence[str],
    donors: Sequence[str],
    size_M: int = DEFAULT_MATCH_SIZE,
) -> Tuple[pd.DataFrame, Dict[str, float]]:
    """
    Select the nearest donors of every target under the Mahalanobis metric.

    Parameters
    ----------
    covariates : pandas.DataFrame
        The pre-treatment covariate vectors of the union of both rosters.
    targets : sequence of str
        The target roster.
    donors : sequence of str
        The donor pool.
    size_M : int, optional
        The number of nearest donors selected per target.

    Returns
    -------
    tuple
        The selection table with one row per target and selected donor
        carrying the distance, and the diagnostic mapping.

    Raises
    ------
    IngestError
        If the target or donor roster is empty, if the size is not positive,
        or if the covariate vectors are incomplete for a roster member.
    """
    # Refuse an empty roster.
    if not targets or not donors:
        # The failure states the precondition.
        raise IngestError("The nearest-neighbour selection requires both rosters.")
    # Refuse a Boolean, fractional or non-positive size before any truncating
    # conversion can change the declared neighbour count.
    if (
        isinstance(size_M, (bool, np.bool_))
        or not isinstance(size_M, (int, np.integer))
        or int(size_M) <= 0
    ):
        # The failure names the observed size.
        raise IngestError(f"The match size {size_M!r} is not positive.")
    # Refuse duplicate identities because each target must receive distinct
    # eligible neighbours and each frequency must identify one donor.
    if len(set(str(code) for code in targets)) != len(tuple(targets)):
        raise IngestError("The matching target roster carries duplicate identities.")
    if len(set(str(code) for code in donors)) != len(tuple(donors)):
        raise IngestError("The matching donor roster carries duplicate identities.")
    missing_vectors = sorted(
        (set(str(code) for code in targets) | set(str(code) for code in donors))
        - set(str(code) for code in covariates.index)
    )
    if missing_vectors:
        raise IngestError(
            f"The covariate frame omits matching unit(s) {missing_vectors[:8]}."
        )
    if not bool(np.isfinite(covariates.to_numpy(dtype=float)).all()):
        raise IngestError("The matching covariate frame carries a non-finite cell.")
    # Compose the target and donor vectors.
    target_matrix: np.ndarray = covariates.loc[list(targets)].to_numpy(dtype=float)
    # Compose the donor matrix.
    donor_matrix: np.ndarray = covariates.loc[list(donors)].to_numpy(dtype=float)
    # Compose the pooled covariance.
    covariance: np.ndarray = compute_pooled_covariance(covariates)
    # Compose the covariance's pseudo-inverse, which admits a near-singular
    # covariance rather than inverting it blindly.
    inverse: np.ndarray = np.linalg.pinv(covariance)
    # Compose the covariance's condition number as a diagnostic.
    singular_values = np.linalg.svd(covariance, compute_uv=False)
    # Compose the condition number, reporting infinity for a singular matrix.
    condition: float = (
        float("inf")
        if float(singular_values[-1]) == 0.0
        else float(singular_values[0] / singular_values[-1])
    )
    # Require the declared number of distinct eligible neighbours. Silently
    # capping M would execute a different design while retaining its label.
    if len(donors) < int(size_M):
        raise IngestError(
            f"The matching donor pool has {len(donors)} distinct unit(s), fewer "
            f"than the declared neighbour count {int(size_M)}."
        )
    effective_size: int = int(size_M)
    # Accumulate the selection rows.
    rows: List[Dict[str, Any]] = []
    # Iterate over the targets.
    for index, target in enumerate(targets):
        # Compose the target's difference from every donor.
        difference: np.ndarray = donor_matrix - target_matrix[index]
        # Compose the quadratic form of every difference.
        quadratic: np.ndarray = np.einsum("ij,jk,ik->i", difference, inverse, difference)
        # Compose the distances, guarding against a negative round-off.
        distances: np.ndarray = np.sqrt(np.maximum(quadratic, 0.0))
        # Order the donors by distance, breaking ties lexicographically by
        # code so the selection is a deterministic function of the vectors.
        order = sorted(
            range(len(donors)),
            key=lambda position: (float(distances[position]), str(donors[position])),
        )
        # Record the selected donors.
        for rank, position in enumerate(order[:effective_size], start=1):
            # Append the selection row.
            rows.append(
                {
                    "target_code": str(target),
                    "donor_code": str(donors[position]),
                    "distance": float(distances[position]),
                    "neighbour_rank": int(rank),
                }
            )
    # Assemble the selection frame.
    selections: pd.DataFrame = pd.DataFrame(rows)
    # Compose the diagnostics of the selection.
    diagnostics: Dict[str, float] = {
        "covariance_condition_number": condition,
        "requested_size": float(size_M),
        "effective_size": float(effective_size),
        "selection_count": float(len(selections)),
    }
    # Log the selection's outcome.
    logger.info(
        "Mahalanobis selection complete: %d target(s), %d donor(s), %d "
        "selection(s) of size %d, covariance condition number %.3e.",
        len(targets),
        len(donors),
        len(selections),
        effective_size,
        condition,
    )
    # Return the selections and the diagnostics.
    return selections, diagnostics


# Convert the selections into donor inclusion frequencies.
def compute_inclusion_weights(
    selections: pd.DataFrame,
    donors: Sequence[str],
    target_count: int,
) -> pd.Series:
    """
    Convert the selections into donor inclusion frequencies.

    Parameters
    ----------
    selections : pandas.DataFrame
        The selection table.
    donors : sequence of str
        The donor pool, which defines the support of the weight vector.
    target_count : int
        The number of targets the selections were made for.

    Returns
    -------
    pandas.Series
        The inclusion frequency of every donor, indexed on the donor code,
        which is the number of times the donor was selected divided by the
        product of the target count and the selection size.

    Raises
    ------
    IngestError
        If the selection table is empty, if the target count is not positive,
        or if the frequencies do not sum to one, which would mean the
        selections are internally inconsistent.
    """
    # Refuse an empty selection table.
    if len(selections) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot weight an empty selection table.")
    # Refuse a Boolean, fractional or non-positive target count.
    if (
        isinstance(target_count, (bool, np.bool_))
        or not isinstance(target_count, (int, np.integer))
        or int(target_count) <= 0
    ):
        # The failure names the observed count.
        raise IngestError(f"The target count {target_count!r} is not positive.")
    required_columns = {"target_code", "donor_code"}
    if not required_columns.issubset(selections.columns):
        raise IngestError(
            f"The selection table omits column(s) {sorted(required_columns - set(selections.columns))}."
        )
    donor_codes = tuple(str(code) for code in donors)
    if len(set(donor_codes)) != len(donor_codes):
        raise IngestError("The selection donor roster carries duplicate identities.")
    selection_codes = selections[["target_code", "donor_code"]].astype(str)
    if bool(selection_codes.duplicated(["target_code", "donor_code"]).any()):
        raise IngestError("A target selected the same donor more than once.")
    unknown = sorted(set(selection_codes["donor_code"]) - set(donor_codes))
    if unknown:
        raise IngestError(f"The selections carry ineligible donor(s) {unknown[:8]}.")
    per_target = selection_codes.groupby("target_code")["donor_code"].nunique()
    if len(per_target) != int(target_count) or per_target.nunique() != 1:
        raise IngestError(
            "Every target must contribute the same complete set size to the "
            "selection-frequency denominator."
        )
    # Count the appearances of every donor.
    counts: pd.Series = selections["donor_code"].astype(str).value_counts()
    # Compose the denominator, which is the target count times the selections
    # each target contributed.
    selections_per_target: float = float(len(selections)) / float(target_count)
    # Compose the denominator of the frequency.
    denominator: float = float(target_count) * selections_per_target
    # Compose the frequency vector over the pool's support.
    weights: pd.Series = pd.Series(
        {
            str(donor): float(counts.get(str(donor), 0)) / denominator
            for donor in donor_codes
        },
        dtype=float,
    )
    # Name the vector so its provenance travels with it.
    weights.name = "inclusion_frequency"
    # Refuse a vector that does not sum to one.
    if abs(float(weights.sum()) - 1.0) > 1.0e-9:
        # The failure names the observed total.
        raise IngestError(
            f"The inclusion frequencies sum to {float(weights.sum())!r} rather "
            f"than to one."
        )
    # Log the weight distribution's extent.
    logger.info(
        "inclusion frequencies composed across %d donor(s); %d carry a positive "
        "weight.",
        len(weights),
        int((weights > 0).sum()),
    )
    # Return the weights.
    return weights


# -------------------------------------------------------------------------------------------------------------------------------
# Task 23, Step 2 (continued): The weighted composition and the balance vector
# -------------------------------------------------------------------------------------------------------------------------------


# Build the weighted donor cross-section of the factor proxy.
def build_weighted_proxy(
    panel: pd.DataFrame,
    donors: Sequence[str],
    outcome: str,
    weights: pd.Series,
    window: Tuple[int, int] = FACTOR_WINDOW,
) -> pd.DataFrame:
    """
    Build the weighted donor cross-section of the factor proxy.

    Parameters
    ----------
    panel : pandas.DataFrame
        The analysis panel.
    donors : sequence of str
        The matched donor pool.
    outcome : str
        The canonical outcome of the pair.
    weights : pandas.Series
        The inclusion frequency of every donor.
    window : tuple of int, optional
        The inclusive factor window.

    Returns
    -------
    pandas.DataFrame
        The proxy columns indexed by year, whose donor column is the weighted
        cross-sectional mean rather than the unweighted mean, because the
        frequency design's factor must reflect the design that was selected.

    Raises
    ------
    IngestError
        If the outcome column is absent, if a donor is absent from the panel
        over the window, or if the weights are empty.
    """
    # Refuse an empty weight vector.
    if len(weights) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot compose a weighted proxy from empty weights.")
    # Refuse a panel lacking the outcome column.
    if outcome not in panel.columns:
        # The failure names the absent column.
        raise IngestError(f"The panel lacks the outcome column {outcome!r}.")
    # Refuse duplicate donors or a weight vector that does not identify every
    # and only member of the effective donor support.
    donor_codes = tuple(str(code) for code in donors)
    if len(set(donor_codes)) != len(donor_codes):
        raise IngestError("The weighted proxy donor roster carries duplicates.")
    if not bool(weights.index.is_unique) or set(str(code) for code in weights.index) != set(donor_codes):
        raise IngestError(
            "The weighted proxy weights do not exactly identify the effective "
            "donor roster."
        )
    weight_values = weights.reindex(donor_codes).to_numpy(dtype=float)
    if not bool(np.isfinite(weight_values).all()) or bool((weight_values <= 0.0).any()):
        raise IngestError(
            "Every effective donor weight must be finite and strictly positive."
        )
    if abs(float(weight_values.sum()) - 1.0) > 1.0e-9:
        raise IngestError("The effective donor weights must sum to one.")
    # Compose the window years.
    years: List[int] = list(range(int(window[0]), int(window[1]) + 1))
    # Select the donor rows inside the window.
    scoped = panel[
        panel["country_code"].astype(str).isin(set(str(code) for code in donors))
        & (panel["year"].astype(int).isin(years))
    ]
    # Pivot the outcome to a year-by-donor frame.
    duplicate = scoped.duplicated(["country_code", "year"], keep=False)
    if bool(duplicate.any()):
        raise IngestError(
            f"The donor grid for {outcome!r} carries duplicate country-year cells."
        )
    grid: pd.DataFrame = scoped.pivot(
        index="year", columns="country_code", values=outcome
    )
    # Reindex to the declared axes so an absent cell is visible.
    grid = grid.reindex(index=years, columns=[str(code) for code in donors])
    # Refuse a missing donor cell, which would distort the weighted mean.
    if bool(grid.isna().to_numpy().any()):
        # The failure names the outcome.
        raise IngestError(
            f"The donor grid for {outcome!r} carries a missing cell; the weighted "
            f"mean requires a complete cross-section."
        )
    # Compose the weight vector aligned to the grid's columns.
    aligned: np.ndarray = weights.reindex([str(code) for code in donors]).to_numpy(
        dtype=float
    )
    # Refuse a weight vector whose total is not positive.
    if float(aligned.sum()) <= 0.0:
        # The failure states the condition.
        raise IngestError("The aligned weights carry a non-positive total.")
    # Compose the weighted cross-sectional mean per year, which is the declared
    # donor mean of the frequency design.
    weighted_mean: pd.Series = pd.Series(
        (grid.to_numpy(dtype=float) * aligned).sum(axis=1) / float(aligned.sum()),
        index=years,
        name="donor_weighted_mean",
    )
    # Compose the constancy flag of a collinear weighted mean, which would make
    # the factor matrix rank-deficient; the proxy still carries the column and
    # the residual-maker handles the deficiency.
    variation = float(np.ptp(weighted_mean.to_numpy(dtype=float)))
    # Log the proxy's extent.
    logger.info(
        "%s: weighted donor proxy composed over %d donor(s) and %d year(s) with "
        "a range of %.6f.",
        outcome,
        len(donors),
        len(years),
        variation,
    )
    # Return the proxy frame with the constant and the weighted mean.
    return pd.DataFrame(
        {
            "constant": pd.Series(1.0, index=years),
            "donor_weighted_mean": weighted_mean,
        }
    )


# Compose the weighted matched design.
def compose_weighted_design(
    policy_name: str,
    outcome_name: str,
    partition: Partition,
    panel: pd.DataFrame,
    lagged: pd.DataFrame,
    weights: pd.Series,
    windows: Mapping[str, Any],
    scientific_output: Optional[MutableMapping[str, pd.DataFrame]] = None,
) -> Tuple[float, float, str, Tuple[str, ...]]:
    """
    Compose the weighted matched design.

    Parameters
    ----------
    policy_name : str
        The policy instrument of the pair.
    outcome_name : str
        The canonical outcome of the pair.
    partition : Partition
        The matched partition.
    panel : pandas.DataFrame
        The analysis panel.
    lagged : pandas.DataFrame
        The panel carrying the one-year-lagged control columns.
    weights : pandas.Series
        The inclusion frequency of every matched donor.
    windows : mapping of str to tuple of int
        The resolved windows of the design.

    Returns
    -------
    tuple
        The effect, its country-dispersion standard error, its Student-t marker
        and diagnostics containing the path widths plus canonical effective-fit
        evidence. The historical four-item arity is preserved.

    Raises
    ------
    IngestError
        If any design, shape or identity check fails.
    """
    # Accumulate the diagnostics.
    diagnostics: List[str] = []
    # Resolve and validate the positive-weight support before any scientific
    # matrix is built, preventing a zero-weight candidate from entering a fit.
    effective_donors = tuple(str(code) for code in partition.donors)
    # The full weight vector retains the original candidate roster; the fitted
    # partition carries only the positive support.
    candidate_roster = tuple(str(code) for code in weights.index)
    if len(set(candidate_roster)) != len(candidate_roster):
        raise IngestError("The weighted design candidate donor roster carries duplicates.")
    if len(set(effective_donors)) != len(effective_donors):
        raise IngestError("The weighted design effective donor roster carries duplicates.")
    if not set(effective_donors).issubset(set(candidate_roster)):
        raise IngestError("The effective donor roster is not a subset of its candidates.")
    numeric_weights = pd.to_numeric(weights, errors="coerce")
    if not bool(numeric_weights.index.is_unique) or set(
        str(code) for code in numeric_weights.index
    ) != set(candidate_roster):
        raise IngestError(
            "The weighted design vector must identify every and only candidate donor."
        )
    if bool(numeric_weights.isna().any()) or not bool(
        np.isfinite(numeric_weights.to_numpy(dtype=float)).all()
    ) or bool((numeric_weights < 0.0).any()):
        raise IngestError("The matching weights must be finite and nonnegative.")
    positive_support = tuple(
        code for code in candidate_roster if float(numeric_weights.get(code, 0.0)) > 0.0
    )
    if positive_support != effective_donors:
        raise IngestError(
            "The fitted donor roster must equal the candidate donors with strictly "
            "positive inclusion frequency, in candidate-roster order."
        )
    effective_weights = numeric_weights.reindex(effective_donors).astype(float)
    if abs(float(effective_weights.sum()) - 1.0) > 1.0e-9:
        raise IngestError("The positive matching weights do not sum to one.")
    # Compose the unit list as the union of the targets and the matched donors.
    units: List[str] = sorted(
        set(str(code) for code in partition.targets)
        | set(str(code) for code in partition.donors)
    )
    # Compose the weighted proxy of the outcome.
    proxy: pd.DataFrame = build_weighted_proxy(
        panel, effective_donors, outcome_name, effective_weights
    )
    # Compose the pre-treatment years of the factor.
    pre_years: List[int] = list(
        range(int(PRE_TREATMENT_WINDOW[0]), int(PRE_TREATMENT_WINDOW[1]) + 1)
    )
    # Compose the factor matrix over the pre-treatment window.
    factor_matrix: np.ndarray = proxy.reindex(index=pre_years).to_numpy(dtype=float)
    # Resolve the one-year-lagged controls required by Appendix B.
    covariate_columns: List[str] = [
        f"{name}{LAG_SUFFIX}" for name in CONTROL_COLUMNS
    ]
    # Require the lagged schema rather than falling back to current controls.
    for column in ["country_code", "year", outcome_name] + covariate_columns:
        # Identify the absent key, outcome or one-year-lagged control.
        if column not in lagged.columns:
            # Stop before any incomplete regressor matrix enters estimation.
            raise IngestError(
                f"The weighted lagged panel lacks the column {column!r}."
            )
    # Accumulate the per-unit dependent and covariate mappings.
    dependent_by_unit: Dict[str, np.ndarray] = {}
    # Accumulate the per-unit covariate mappings.
    covariate_by_unit: Dict[str, np.ndarray] = {}
    # Accumulate the per-unit weights, which scale the donor rows only.
    unit_weights: Dict[str, float] = {}
    # Iterate over the units in sorted order.
    for unit in units:
        # Select the unit's pre-treatment rows with one-year-lagged controls.
        scoped = lagged[
            (lagged["country_code"].astype(str) == str(unit))
            & (lagged["year"].astype(int).isin(pre_years))
        ].sort_values("year", kind="stable")
        # Refuse a unit whose coverage of the window is incomplete.
        if len(scoped) != len(pre_years):
            # The failure names the unit.
            raise IngestError(
                f"{unit}: the pre-treatment window coverage is incomplete, so "
                f"the weighted design cannot be composed."
            )
        # Require exactly one ordered row for every factor year.
        if tuple(scoped["year"].to_numpy()) != tuple(pre_years):
            # Reject duplicates or malformed years that conceal a missing year.
            raise IngestError(
                f"{unit}: weighted pre-treatment years diverge from {pre_years}."
            )
        # Compose the observed outcome vector on the validated year axis.
        dependent = scoped[outcome_name].to_numpy(dtype=float)
        # Pair each outcome year t with completed controls from year t - 1.
        covariates = scoped[covariate_columns].to_numpy(dtype=float)
        # Require finite outcomes and lagged controls before least squares.
        if not np.isfinite(dependent).all() or not np.isfinite(covariates).all():
            # Localize a missing or non-finite cell to the affected history.
            raise IngestError(
                f"{unit}: weighted outcome or lagged control matrix is non-finite."
            )
        # Record the unit's dependent column as a column vector.
        dependent_by_unit[str(unit)] = dependent.reshape(-1, 1)
        # Record the unit's covariate block in the declared shape, which is
        # one column per control and no intercept column, so the weighted fit
        # consumes exactly the contract the baseline chain consumes.
        covariate_by_unit[str(unit)] = covariates
        # Record the unit's weight: the inclusion frequency for a donor and one
        # for a target, which is the declared propagation of the weights to the
        # covariate pool.
        unit_weights[str(unit)] = (
            float(effective_weights.get(str(unit), 0.0))
            if str(unit) in set(effective_donors) else 1.0
        )
    # Compose the row scaling, which is the square root of the weight; scaling
    # the rows of a least-squares problem by that factor is exactly the
    # weighted fit, so the existing estimator is reused unchanged.
    scaling: Dict[str, float] = {
        unit: float(np.sqrt(max(weight, 0.0))) for unit, weight in unit_weights.items()
    }
    # Refuse a design in which every donor carries zero weight, which would
    # leave the donor block empty.
    donor_scaling = [scaling[str(code)] for code in effective_donors]
    # Test the donor block's total.
    if float(sum(donor_scaling)) <= 0.0:
        # The failure states the condition.
        raise IngestError(
            "Every matched donor carries zero weight, so the weighted design is "
            "undefined."
        )
    # Scale the dependent and covariate matrices of every unit.
    scaled_dependent: Dict[str, np.ndarray] = {
        unit: matrix * scaling[unit] for unit, matrix in dependent_by_unit.items()
    }
    # Scale the covariate matrices.
    scaled_covariate: Dict[str, np.ndarray] = {
        unit: matrix * scaling[unit] for unit, matrix in covariate_by_unit.items()
    }
    # Build the residual-maker of the weighted factor.
    maker = build_residual_maker(factor_matrix)
    # Estimate the weighted pooled coefficient over the scaled rows.
    beta, fit_diagnostics = estimate_pooled_coefficient(
        scaled_dependent, scaled_covariate, maker
    )
    # Estimate the country loadings of every unit on the scaled rows.
    # Preserve the diagnostic's infinite value for a singular design; bound
    # only the arithmetic allowance of the geometric orthogonality audit.
    orthogonality_conditioning = (
        float(fit_diagnostics["condition_number"])
        if np.isfinite(float(fit_diagnostics["condition_number"]))
        else float(1.0 / np.finfo(float).eps)
    )
    country_loadings = estimate_unit_loadings(
        factor_matrix,
        scaled_dependent,
        scaled_covariate,
        beta,
        conditioning=orthogonality_conditioning,
    )
    # Estimate the covariate loadings of the targets on the scaled rows.
    target_loadings = estimate_covariate_loadings(
        factor_matrix, scaled_covariate, partition.targets
    )
    # Compose the projection span, which is the whole gap window.
    projection_years: List[int] = list(
        range(int(windows[WAVE_ONE_LABEL][0]), int(windows["post_escalation"][1]) + 1)
    )
    # Compose the projection factor matrix at the weighted design's width.
    projection_factor: np.ndarray = proxy.reindex(index=projection_years).to_numpy(
        dtype=float
    )
    # Accumulate the projected counterfactual rows.
    projected_rows: List[Dict[str, Any]] = []
    # Iterate over the targets in sorted order.
    for target in sorted(target_loadings.keys()):
        # Project the target's covariate and outcome paths.
        covariate_path, outcome_path = project_target_paths(
            target_loadings[target],
            np.asarray(beta),
            np.asarray(country_loadings[target]),
            projection_factor,
        )
        # Record the diagnosed covariate path width.
        diagnostics.append(
            f"counterfactual covariate path width {covariate_path.shape[1]} for "
            f"{target}."
        )
        # Record the projected outcome of every projection year.
        for index, year in enumerate(projection_years):
            # Append the projected row.
            projected_rows.append(
                {
                    "policy": policy_name,
                    "outcome": outcome_name,
                    "country_code": str(target),
                    "year": int(year),
                    "counterfactual_outcome": float(outcome_path[index]),
                }
            )
    # Assemble the counterfactual frame the gap computation consumes.
    counterfactual_frame: pd.DataFrame = pd.DataFrame(projected_rows)
    # Compute the gap matrix over the declared span.
    gaps: pd.DataFrame = compute_gap_matrix(
        panel,
        outcome_name,
        partition.targets,
        counterfactual_frame,
        policy_name,
        window=tuple(windows[WAVE_ONE_LABEL]),
        window_end=tuple(windows["post_escalation"]),
    )
    # Aggregate the per-country window means.
    country_means: pd.DataFrame = aggregate_country_window_means(
        gaps,
        {
            WAVE_ONE_LABEL: tuple(windows[WAVE_ONE_LABEL]),
            WAVE_TWO_LABEL: tuple(windows[WAVE_TWO_LABEL]),
        },
    )
    # Compute the coalition level effects as diagnostics.
    level_effects: pd.Series = aggregate_level_effects(country_means)
    # Compute the per-country pre-escalation levels.
    pre_levels: pd.Series = compute_pre_escalation_levels(
        gaps, tuple(windows["pre_escalation"])
    )
    # Build the observed post-escalation window.
    observed_window: pd.DataFrame = build_observed_window(
        panel,
        outcome_name,
        tuple(str(code) for code in gaps.index),
        tuple(windows["post_escalation"]),
    )
    # Compute the continuation gap through the shifted counterfactual.
    continuation_gap: pd.DataFrame = compute_continuation_gap(
        gaps, pre_levels, observed_window, tuple(windows["post_escalation"])
    )
    # Compute the incremental effect through the continuation route.
    effect: float = compute_incremental_effect(continuation_gap)
    # Compute the incremental effect through the direct route.
    direct: float = compute_incremental_effect_alternative(
        gaps, tuple(windows["pre_escalation"]), tuple(windows["post_escalation"])
    )
    # Assert the declared identity between the two routes.
    assert_incremental_identity(effect, direct)
    # Compose the per-country incremental contrasts.
    post_years: List[int] = list(
        range(
            int(windows["post_escalation"][0]), int(windows["post_escalation"][1]) + 1
        )
    )
    # Compute the contrasts as the post mean less the pre level.
    contrasts: pd.Series = gaps[post_years].mean(axis=1) - pre_levels
    # Compute the effect and its dispersion standard error.
    contrast_effect, se_cd = compute_country_dispersion_se(contrasts)
    # Compute the test statistic and the degrees of freedom.
    statistic, degrees_of_freedom = compute_t_statistic(
        contrast_effect, se_cd, len(contrasts)
    )
    # Compute the two-sided probability.
    probability: float = compute_two_sided_p(statistic, degrees_of_freedom)
    # Assign the significance marker.
    stars, _critical_values = assign_significance_stars(statistic, degrees_of_freedom)
    # Assert the marker's agreement with the probability route.
    assert_star_consistency(
        stars,
        probability,
        statistic=statistic,
        degrees_of_freedom=degrees_of_freedom,
    )
    # Assert the probability's reproducibility.
    assert_p_reproducible(probability, statistic, degrees_of_freedom)
    # Record the level effects as diagnostics.
    diagnostics.append(
        f"level effects W1 {float(level_effects[WAVE_ONE_LABEL]):.6f} and W2 "
        f"{float(level_effects[WAVE_TWO_LABEL]):.6f}."
    )
    if scientific_output is not None:
        pre_residual_rows: Dict[str, np.ndarray] = {}
        for target in partition.targets:
            fitted = (
                covariate_by_unit[str(target)] @ np.asarray(beta)
                + factor_matrix @ np.asarray(country_loadings[str(target)])
            ).reshape(-1)
            observed = dependent_by_unit[str(target)].reshape(-1)
            pre_residual_rows[str(target)] = observed - fitted
        scientific_output["gaps"] = gaps.copy(deep=True)
        scientific_output["fitted_pre_gaps"] = pd.DataFrame(
            pre_residual_rows, index=pre_years
        ).T.rename_axis("country_code").rename_axis("year", axis=1)
    # Record the weighted fit's rank as a diagnostic.
    diagnostics.append(
        f"weighted pooled fit rank {int(fit_diagnostics['rank'])} over "
        f"{int(fit_diagnostics['rows'])} row(s)."
    )
    # Record the complete effective design in a canonical machine-readable
    # diagnostic while preserving the historical four-item return contract.
    design_payload = {
        "candidate_donors": list(candidate_roster),
        "effective_donors": list(effective_donors),
        "pooled_units": list(units),
        "positive_weights": {
            code: float(effective_weights.loc[code]) for code in effective_donors
        },
        "factor_proxy_sha256": _content_digest(proxy),
        "factor_matrix_sha256": _content_digest(factor_matrix),
        "windows_sha256": str(
            windows.get(WINDOW_PROVENANCE_KEY, {}).get("fingerprint_sha256", "")
        ),
        "pooled_rank": int(fit_diagnostics["rank"]),
        "pooled_rows": int(fit_diagnostics["rows"]),
    }
    diagnostics.append(
        "effective weighted design "
        + json.dumps(design_payload, sort_keys=True, separators=(",", ":"))
    )
    # Return the effect, its inference, the marker and the diagnostics.
    return float(effect), float(se_cd), str(stars), tuple(diagnostics)


# Compute the standardized difference balance vector.
def compute_standardized_differences(
    covariates: pd.DataFrame,
    targets: Sequence[str],
    donors: Sequence[str],
    weights: Optional[pd.Series] = None,
    weighting_convention: str = "reliability",
    frequency_total: Optional[int] = None,
) -> pd.Series:
    """
    Compute the standardized difference balance vector.

    Parameters
    ----------
    covariates : pandas.DataFrame
        The pre-treatment covariate vectors.
    targets : sequence of str
        The treated roster.
    donors : sequence of str
        The donor pool.
    weights : pandas.Series, optional
        The inclusion frequency of every donor; when supplied the donor means
        and variances are composed with the weights, so the reported balance
        describes the design that was estimated.
    weighting_convention : {"reliability", "frequency"}, optional
        The variance convention for normalized weights. Reliability weights use
        the unbiased denominator ``1-sum(w**2)``. Frequency weights reconstruct
        integer selection counts and use the expanded-sample divisor ``n-1``.
    frequency_total : int, optional
        The total number of selections. It is required for the frequency
        convention and makes the sample-variance denominator reproducible.

    Returns
    -------
    pandas.Series
        The standardized difference of every covariate dimension, computed as
        the treated mean less the donor mean divided by the square root of the
        average of the two sample variances.

    Raises
    ------
    IngestError
        If either roster carries fewer units than the declared minimum, or if
        a covariate dimension is absent, weights are malformed, or a requested
        variance convention is not identified. A zero pooled variance is not
        hidden: equal group means produce zero and unequal means produce signed
        infinity with an explicit flag in the returned series attributes.
    """
    # Refuse a roster below the minimum size.
    if len(targets) < MINIMUM_GROUP_SIZE or len(donors) < MINIMUM_GROUP_SIZE:
        # The failure names both sizes.
        raise IngestError(
            f"The balance computation requires at least {MINIMUM_GROUP_SIZE} "
            f"unit(s) per group; observed {len(targets)} treated and "
            f"{len(donors)} donor."
        )
    # Refuse duplicate identities or missing vector rows before selection.
    target_codes = tuple(str(code) for code in targets)
    donor_codes = tuple(str(code) for code in donors)
    if len(set(target_codes)) != len(target_codes) or len(set(donor_codes)) != len(donor_codes):
        raise IngestError("Balance rosters must carry unique country identities.")
    missing = sorted((set(target_codes) | set(donor_codes)) - set(covariates.index.astype(str)))
    if missing:
        raise IngestError(f"The balance covariates omit countr(ies) {missing[:8]}.")
    # Compose the treated matrix.
    treated: np.ndarray = covariates.loc[list(targets)].to_numpy(dtype=float)
    # Compose the donor matrix.
    donor: np.ndarray = covariates.loc[list(donors)].to_numpy(dtype=float)
    if not bool(np.isfinite(treated).all()) or not bool(np.isfinite(donor).all()):
        raise IngestError("The balance covariates carry a non-finite cell.")
    # Compose the weight vector aligned to the donor roster when supplied.
    aligned: Optional[np.ndarray] = (
        weights.reindex([str(code) for code in donors]).to_numpy(dtype=float)
        if weights is not None
        else None
    )
    if weights is not None and (
        not bool(weights.index.is_unique)
        or set(str(code) for code in weights.index) != set(donor_codes)
    ):
        raise IngestError(
            "Balance weights must identify every and only member of the donor roster."
        )
    if aligned is not None:
        if not bool(np.isfinite(aligned).all()) or bool((aligned < 0.0).any()):
            raise IngestError("Balance weights must be finite and nonnegative.")
        if abs(float(aligned.sum()) - 1.0) > 1.0e-9:
            raise IngestError("Normalized balance weights must sum to one.")
        if int((aligned > 0.0).sum()) < MINIMUM_GROUP_SIZE:
            raise IngestError(
                "Weighted balance requires at least two positive-weight donors."
            )
    # Compose the treated means and sample variances.
    treated_means: np.ndarray = treated.mean(axis=0)
    # Compose the treated sample variances with the declared divisor.
    treated_variance: np.ndarray = treated.var(axis=0, ddof=1)
    # Compose the donor means, weighting them when the vector was supplied.
    if aligned is not None:
        # Refuse a weight vector whose total is not positive.
        if float(aligned.sum()) <= 0.0:
            # The failure states the condition.
            raise IngestError("The balance weights carry a non-positive total.")
        # Compose the weighted donor means.
        donor_means: np.ndarray = (donor * aligned[:, None]).sum(axis=0) / float(
            aligned.sum()
        )
        # Compose the weighted donor variances under the explicitly declared
        # convention rather than substituting positive-support size for an
        # unknown sample denominator.
        centered: np.ndarray = donor - donor_means
        if weighting_convention == "frequency":
            if (
                isinstance(frequency_total, (bool, np.bool_))
                or not isinstance(frequency_total, (int, np.integer))
                or int(frequency_total) < MINIMUM_GROUP_SIZE
            ):
                raise IngestError(
                    "Frequency-weighted balance requires an integral selection "
                    "total of at least two."
                )
            counts = aligned * float(frequency_total)
            if not bool(np.allclose(counts, np.rint(counts), rtol=0.0, atol=1.0e-8)):
                raise IngestError(
                    "The normalized weights do not reconstruct integral selection counts."
                )
            donor_variance = (
                (centered**2 * counts[:, None]).sum(axis=0)
                / float(int(frequency_total) - 1)
            )
        elif weighting_convention == "reliability":
            denominator = 1.0 - float(np.square(aligned).sum())
            if denominator <= 0.0:
                raise IngestError(
                    "Reliability-weighted balance has no positive variance denominator."
                )
            donor_variance = (
                (centered**2 * aligned[:, None]).sum(axis=0) / denominator
            )
        else:
            raise IngestError(
                f"Unknown balance weighting convention {weighting_convention!r}."
            )
    else:
        # Compose the unweighted donor means.
        donor_means = donor.mean(axis=0)
        # Compose the unweighted donor variances with the declared divisor.
        donor_variance = donor.var(axis=0, ddof=1)
    # Compose the pooled standard deviation, which is the square root of the
    # average of the two sample variances.
    pooled: np.ndarray = np.sqrt((treated_variance + donor_variance) / 2.0)
    mean_difference = treated_means - donor_means
    # Compose the standardized differences and make the zero-variance limit
    # explicit: equality is balanced at zero, while unequal constants are an
    # infinite imbalance (E21), never a NaN that a maximum could skip.
    differences = np.empty_like(mean_difference, dtype=float)
    positive_scale = pooled > 0.0
    differences[positive_scale] = mean_difference[positive_scale] / pooled[positive_scale]
    zero_scale = ~positive_scale
    equal_means = mean_difference == 0.0
    differences[zero_scale & equal_means] = 0.0
    differences[zero_scale & ~equal_means] = np.copysign(
        np.inf, mean_difference[zero_scale & ~equal_means]
    )
    # Assemble the balance vector indexed on the covariate dimensions.
    vector: pd.Series = pd.Series(
        differences, index=list(covariates.columns), dtype=float
    )
    # Name the vector so its provenance travels with it.
    vector.name = "std_diff"
    # Bind the variance convention and every zero-variance decision to the
    # returned labelled vector without changing the public return type.
    vector.attrs["variance_convention"] = (
        "sample_ddof_1"
        if aligned is None
        else (
            "frequency_expanded_sample_ddof_1"
            if weighting_convention == "frequency"
            else "unbiased_reliability_weights"
        )
    )
    vector.attrs["frequency_total"] = (
        int(frequency_total) if frequency_total is not None else None
    )
    vector.attrs["zero_variance_flags"] = {
        str(covariates.columns[position]): (
            "equal_means_zero" if bool(equal_means[position]) else "unequal_means_infinite"
        )
        for position in np.flatnonzero(zero_scale)
    }
    # Log the vector's extent.
    logger.info(
        "standardized differences computed across %d dimension(s)%s.",
        len(vector),
        " with inclusion weights" if aligned is not None else "",
    )
    # Return the vector.
    return vector

# -------------------------------------------------------------------------------------------------------------------------------
# Task 23, Step 3 (continued): The design records and the presentation tables
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the resolved windows of a matching design.
def resolve_design_windows(
    baseline_config: Mapping[str, Any],
) -> Dict[str, Any]:
    """
    Compose the resolved windows of a matching design.

    Parameters
    ----------
    baseline_config : mapping
        The baseline configuration.

    Returns
    -------
    dict
        The first-wave, second-wave, pre-escalation and post-escalation
        windows, each as an inclusive pair, plus the canonical resolver's
        provenance record and fingerprint.

    Raises
    ------
    IngestError
        If any canonical or legacy declaration is malformed or conflicting.
    """
    # Delegate to the single canonical resolver used by the baseline and sweep
    # estimators. This reconciles legacy aliases, rejects conflicts and retains
    # explicit valid overrides in the effective-window provenance.
    windows, _mode = resolve_reporting_windows(
        baseline_config, context=("matching", "design")
    )
    # Log the resolved windows.
    logger.info("matching design windows resolved: %s.", {k: list(v) for k, v in windows.items()})
    # Return the windows.
    return dict(windows)


# Compose one matching design.
def compose_matching_design(
    policy_name: str,
    outcome_name: str,
    axis: str,
    partition: Partition,
    panel: pd.DataFrame,
    lagged: pd.DataFrame,
    baseline_config: Mapping[str, Any],
    data: Mapping[str, Any],
    size_M: int = DEFAULT_MATCH_SIZE,
) -> MatchingDesignResult:
    """
    Compose one matching design.

    Parameters
    ----------
    policy_name : str
        The policy instrument of the pair.
    outcome_name : str
        The canonical outcome variable of the pair.
    axis : str
        The matching axis.
    partition : Partition
        The baseline partition.
    panel : pandas.DataFrame
        The analysis panel.
    lagged : pandas.DataFrame
        The lagged panel.
    baseline_config : mapping
        The baseline configuration.
    data : mapping
        The frozen inputs under the declared keys.
    size_M : int, optional
        The number of nearest donors the frequency axis selects per target.

    Returns
    -------
    MatchingDesignResult
        The matched rosters, the optional weights, the effect with its
        inference, the balance vector and its maximum.

    Raises
    ------
    IngestError
        If the axis is undeclared, if a roster falls below the minimum, or if
        any design, selection or composition check fails.
    """
    # Refuse an undeclared axis.
    if axis not in MATCHING_AXES:
        # The failure names the axis.
        raise IngestError(f"Undeclared matching axis {axis!r}.")
    # Compose the covariate vectors of the union of both rosters.
    union: List[str] = sorted(
        set(str(code) for code in partition.targets)
        | set(str(code) for code in partition.donors)
    )
    # Compose the vectors.
    covariates: pd.DataFrame = build_pretreatment_covariates(panel, union)
    # Accumulate the diagnostics.
    diagnostics: List[str] = []
    # Compose the resolved windows.
    windows: Dict[str, Any] = resolve_design_windows(baseline_config)
    # Route the hard restriction through the frozen baseline interface.
    if axis == AXIS_HIGH_INCOME:
        # Restrict both rosters to the units above the income median.
        matched_targets, matched_donors, median = restrict_above_income_median(
            covariates, partition.targets, partition.donors
        )
        # Record the threshold as a diagnostic.
        diagnostics.append(f"income median {median:.6f}.")
        # Compose the restricted partition.
        restricted = Partition(
            policy=partition.policy,
            outcome=partition.outcome,
            targets=matched_targets,
            donors=matched_donors,
            exclusions=partition.exclusions,
        )
        # Compose the restricted data mapping through the factory's builder.
        restricted_data: Dict[str, Any] = build_variant_data(
            data, policy_name, outcome_name, restricted
        )
        # Estimate the restricted design through the frozen interface.
        baseline_result = run_baseline(
            policy_name, outcome_name, baseline_config, restricted_data
        )
        # Compose the balance vector over the restricted rosters.
        std_diff: pd.Series = compute_standardized_differences(
            covariates, matched_targets, matched_donors
        )
        if bool(std_diff.isna().any()):
            raise IngestError("The high-income balance vector carries an undefined cell.")
        # Record every candidate and its realised hard-restriction disposition.
        selection_audit = pd.DataFrame(
            [
                {
                    "roster_side": side,
                    "country_code": str(code),
                    "selected": bool(str(code) in set(retained)),
                    "income_average": float(covariates.loc[str(code), INCOME_COLUMN]),
                    "income_median": float(median),
                    "selection_rule": "strictly_above_pooled_median",
                }
                for side, candidates, retained in (
                    ("target", partition.targets, matched_targets),
                    ("donor", partition.donors, matched_donors),
                )
                for code in candidates
            ]
        )
        provenance = MappingProxyType(
            {
                "balance_window": tuple(BALANCE_WINDOW),
                "candidate_targets": tuple(str(code) for code in partition.targets),
                "candidate_donors": tuple(str(code) for code in partition.donors),
                "effective_targets": tuple(str(code) for code in matched_targets),
                "effective_donors": tuple(str(code) for code in matched_donors),
                "income_median": float(median),
                "covariates_sha256": _content_digest(covariates),
                "selection_sha256": _content_digest(selection_audit),
                "panel_sha256": _content_digest(panel),
                "lagged_sha256": _content_digest(lagged),
                "windows_sha256": str(
                    windows[WINDOW_PROVENANCE_KEY]["fingerprint_sha256"]
                ),
                "variance_convention": str(std_diff.attrs["variance_convention"]),
            }
        )
        # Return the frozen design record.
        fit_objects = baseline_result.fit_provenance["objects"]
        pre_years = list(range(PRE_TREATMENT_WINDOW[0], PRE_TREATMENT_WINDOW[1] + 1))
        pre_residuals: Dict[str, np.ndarray] = {}
        for target in matched_targets:
            scoped = lagged[
                lagged["country_code"].astype(str).eq(str(target))
                & lagged["year"].astype(int).isin(pre_years)
            ].sort_values("year")
            if len(scoped) != len(pre_years):
                raise IngestError(f"{target}: matched fitted-pre panel is incomplete.")
            observed = scoped[outcome_name].to_numpy(dtype=float)
            # Reconstruct fitted residuals with the controls used by the fit.
            covariates_target = scoped[
                [f"{name}{LAG_SUFFIX}" for name in CONTROL_COLUMNS]
            ].to_numpy(dtype=float)
            fitted = (
                covariates_target @ np.asarray(fit_objects["beta"])
                + np.asarray(fit_objects["factor_matrix"])
                @ np.asarray(fit_objects["country_loadings"][str(target)])
            )
            pre_residuals[str(target)] = observed - fitted.reshape(-1)
        fitted_pre_gaps = pd.DataFrame(pre_residuals, index=pre_years).T
        fitted_pre_gaps.index.name = "country_code"
        fitted_pre_gaps.columns.name = "year"
        return MatchingDesignResult(
            policy_name=str(policy_name),
            outcome_name=str(outcome_name),
            matching_axis=axis,
            matched_targets=tuple(str(code) for code in matched_targets),
            matched_donors=tuple(str(code) for code in matched_donors),
            weights=None,
            att=float(baseline_result.att_incremental),
            se_cd=float(baseline_result.se_cd),
            stars=str(baseline_result.stars),
            std_diff=std_diff,
            max_abs_std_diff=float(std_diff.abs().max()),
            max_abs_covariate=str(std_diff.abs().idxmax()),
            std_diff_unweighted=None,
            diagnostics=tuple(diagnostics),
            candidate_targets=tuple(str(code) for code in partition.targets),
            candidate_donors=tuple(str(code) for code in partition.donors),
            selection_audit=selection_audit.copy(deep=True),
            design_provenance=provenance,
            gaps=baseline_result.gaps.copy(deep=True),
            fitted_pre_gaps=fitted_pre_gaps,
        )
    # Select the nearest donors of every target under the frequency axis.
    selections, selection_diagnostics = select_mahalanobis_neighbours(
        covariates, partition.targets, partition.donors, size_M
    )
    # Record the selection's conditioning as a diagnostic.
    diagnostics.append(
        f"covariance condition number "
        f"{selection_diagnostics['covariance_condition_number']:.6e} and effective "
        f"size {int(selection_diagnostics['effective_size'])}."
    )
    # Convert the selections into inclusion frequencies.
    weights: pd.Series = compute_inclusion_weights(
        selections, partition.donors, len(partition.targets)
    )
    # Resolve the effective fitted donor roster as exactly the positive-weight
    # support, preserving candidate order for deterministic scientific axes.
    effective_donors = tuple(
        str(code)
        for code in partition.donors
        if float(weights.loc[str(code)]) > 0.0
    )
    if len(effective_donors) < MINIMUM_GROUP_SIZE:
        raise IngestError(
            "The neighbour selection retains fewer than two positive-weight donors."
        )
    # Compose the matched partition, whose donor pool is the weighted support.
    matched = Partition(
        policy=partition.policy,
        outcome=partition.outcome,
        targets=partition.targets,
        donors=effective_donors,
        exclusions=partition.exclusions,
    )
    # Compose the weighted design.
    scientific_output: Dict[str, pd.DataFrame] = {}
    effect, se_cd, stars, composition_diagnostics = compose_weighted_design(
        policy_name,
        outcome_name,
        matched,
        panel,
        lagged,
        weights,
        windows,
        scientific_output=scientific_output,
    )
    # Record the composition's diagnostics.
    diagnostics.extend(composition_diagnostics)
    # Compose the weighted balance vector.
    std_diff = compute_standardized_differences(
        covariates,
        matched.targets,
        matched.donors,
        weights.reindex(effective_donors),
        weighting_convention="frequency",
        frequency_total=int(len(selections)),
    )
    # Compose the unweighted balance vector so the effect of the weighting is
    # visible beside it.
    std_diff_unweighted: pd.Series = compute_standardized_differences(
        covariates, matched.targets, matched.donors
    )
    if bool(std_diff.isna().any()) or bool(std_diff_unweighted.isna().any()):
        raise IngestError("The frequency balance audit carries an undefined cell.")
    provenance = MappingProxyType(
        {
            "balance_window": tuple(BALANCE_WINDOW),
            "candidate_targets": tuple(str(code) for code in partition.targets),
            "candidate_donors": tuple(str(code) for code in partition.donors),
            "effective_targets": tuple(str(code) for code in matched.targets),
            "effective_donors": tuple(str(code) for code in matched.donors),
            "positive_weights": tuple(
                (code, float(weights.loc[code])) for code in effective_donors
            ),
            "selection_count": int(len(selections)),
            "neighbours_per_target": int(size_M),
            "covariates_sha256": _content_digest(covariates),
            "selection_sha256": _content_digest(selections),
            "panel_sha256": _content_digest(panel),
            "lagged_sha256": _content_digest(lagged),
            "windows_sha256": str(
                windows[WINDOW_PROVENANCE_KEY]["fingerprint_sha256"]
            ),
            "variance_convention": str(std_diff.attrs["variance_convention"]),
        }
    )
    # Return the frozen design record.
    return MatchingDesignResult(
        policy_name=str(policy_name),
        outcome_name=str(outcome_name),
        matching_axis=axis,
        matched_targets=tuple(str(code) for code in matched.targets),
        matched_donors=tuple(str(code) for code in matched.donors),
        weights=weights,
        att=float(effect),
        se_cd=float(se_cd),
        stars=str(stars),
        std_diff=std_diff,
        max_abs_std_diff=float(std_diff.abs().max()),
        max_abs_covariate=str(std_diff.abs().idxmax()),
        std_diff_unweighted=std_diff_unweighted,
        diagnostics=tuple(diagnostics),
        candidate_targets=tuple(str(code) for code in partition.targets),
        candidate_donors=tuple(str(code) for code in partition.donors),
        selection_audit=selections.copy(deep=True),
        design_provenance=provenance,
        gaps=scientific_output["gaps"],
        fitted_pre_gaps=scientific_output["fitted_pre_gaps"],
    )


# Freeze the diagnostics behind a canonical digest.
def freeze_matching_diagnostics(
    summary: pd.DataFrame,
    balance: pd.DataFrame,
    design_audit: Optional[pd.DataFrame] = None,
) -> str:
    """
    Freeze the diagnostics behind a canonical digest.

    Parameters
    ----------
    summary : pandas.DataFrame
        The design summary.
    balance : pandas.DataFrame
        The standardized differences.
    design_audit : pandas.DataFrame, optional
        The candidate/effective rosters, selection digest and fitted lineage of
        every design. The orchestrator always supplies it; omission retains the
        historical two-frame direct-call contract.

    Returns
    -------
    str
        The hexadecimal digest of the canonical payload.

    Raises
    ------
    IngestError
        If either frame is empty.
    """
    # Refuse an empty frame.
    for label, frame in (("summary", summary), ("balance", balance)):
        # Test the frame's extent.
        if len(frame) == 0:
            # The failure names the frame.
            raise IngestError(f"Cannot freeze an empty {label} frame.")
    if design_audit is not None and len(design_audit) == 0:
        raise IngestError("A supplied matching design audit cannot be empty.")
    # Bind labelled axes, schema and complete content of all three scientific
    # artefacts rather than a lossy subset of scalar columns.
    payload = "|".join(
        (
            _content_digest(summary.sort_index()),
            _content_digest(balance.sort_index()),
            (
                _content_digest(design_audit.sort_index())
                if design_audit is not None else "legacy_no_design_audit"
            ),
        )
    )
    return hashlib.sha256(payload.encode("utf-8")).hexdigest()


# -------------------------------------------------------------------------------------------------------------------------------
# Task 23, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Execute the matching designs and their balance diagnostics.
def execute_matching_designs(
    baseline_config: Mapping[str, Any],
    data: Mapping[str, Any],
    pairs: Optional[Sequence[Tuple[str, str]]] = None,
    axes: Sequence[str] = MATCHING_AXES,
    size_M: int = DEFAULT_MATCH_SIZE,
    expected_att: Optional[Mapping[Tuple[str, str, str], float]] = None,
    tolerance: float = 1.0e-3,
) -> MatchingDiagnosticsResult:
    """
    Execute the matching designs and their balance diagnostics.

    Parameters
    ----------
    baseline_config : mapping
        The baseline configuration, which is never mutated.
    data : mapping
        The frozen inputs under the declared keys.
    pairs : sequence of tuple, optional
        The pairs the execution covers. When absent every registered pair is
        covered.
    axes : sequence of str, optional
        The matching axes to execute, applied in their declared order.
    size_M : int, optional
        The number of nearest donors the frequency axis selects per target.
    expected_att : mapping of tuple to float, optional
        The declared effects keyed on the pair and the axis, compared only
        when supplied.
    tolerance : float, optional
        The tolerance of the declared comparison.

    Returns
    -------
    MatchingDiagnosticsResult
        The design records, actual effective-count summary, scheme-labelled
        balance table, complete design audit, digest, timestamp and notes.

    Raises
    ------
    KeyError
        If the data mapping lacks a declared key.
    IngestError
        If a pair is absent from the registry, an axis is undeclared, or any
        design, selection or comparison check fails.
    """
    # Refuse a data mapping lacking the declared keys.
    for key in ("registry", "panel", "lagged"):
        # Test the key's presence.
        if key not in data:
            # The failure names the absent key.
            raise KeyError(f"The data mapping lacks the declared key {key!r}.")
    if isinstance(tolerance, (bool, np.bool_)) or not np.isfinite(float(tolerance)) or float(tolerance) < 0.0:
        raise IngestError("The matching comparison tolerance must be finite and nonnegative.")
    if not axes or len(set(str(axis) for axis in axes)) != len(tuple(axes)):
        raise IngestError("Matching axes must be a nonempty unique sequence.")
    # Refuse an undeclared axis.
    for axis in axes:
        # Test the axis's membership.
        if axis not in MATCHING_AXES:
            # The failure names the axis.
            raise IngestError(f"Undeclared matching axis {axis!r}.")
    # Compose the pair list, defaulting to every registered pair.
    selected: List[Tuple[str, str]] = (
        [tuple(pair) for pair in pairs]
        if pairs is not None
        else sorted(data["registry"].partitions.keys())
    )
    if not selected or len(set(selected)) != len(selected):
        raise IngestError("Matching pairs must be a nonempty unique sequence.")
    expected_keys = set(
        (pair[0], pair[1], axis) for pair in selected for axis in axes
    )
    if expected_att is not None and set(expected_att) != expected_keys:
        raise IngestError(
            "The declared Table 7 effect cells must exactly equal the requested "
            "policy/outcome/design keys."
        )
    # Accumulate the design records.
    designs: Dict[Tuple[str, str, str], MatchingDesignResult] = {}
    # Accumulate the summary rows.
    summary_rows: List[Dict[str, Any]] = []
    # Accumulate the balance rows.
    balance_rows: List[Dict[str, Any]] = []
    # Accumulate complete candidate/effective roster and lineage evidence.
    audit_rows: List[Dict[str, Any]] = []
    # Iterate over the pairs in sorted order.
    for pair in sorted(tuple(item) for item in selected):
        # Refuse a pair the registry does not carry.
        if pair not in data["registry"].partitions:
            # The failure names the pair.
            raise IngestError(f"The registry does not carry the pair {pair}.")
        # Read the pair's baseline partition.
        partition: Partition = data["registry"].partitions[pair]
        # Iterate over the axes in their declared order.
        for axis in axes:
            # Compose the design.
            try:
                # Compose the design record.
                record: MatchingDesignResult = compose_matching_design(
                    pair[0],
                    pair[1],
                    axis,
                    partition,
                    data["panel"],
                    data["lagged"],
                    baseline_config,
                    data,
                    size_M,
                )
            # Re-raise a failure with the pair and the axis named.
            except IngestError as error:
                # Raise the localized failure.
                raise IngestError(f"{pair[0]}/{pair[1]}/{axis}: {error}") from error
            # Record the design.
            designs[(pair[0], pair[1], axis)] = record
            if tuple(record.candidate_targets) != tuple(str(code) for code in partition.targets):
                raise IngestError("The design candidate target roster drifted from its partition.")
            if tuple(record.candidate_donors) != tuple(str(code) for code in partition.donors):
                raise IngestError("The design candidate donor roster drifted from its partition.")
            if record.weights is not None:
                positive = tuple(
                    code for code in record.candidate_donors
                    if float(record.weights.get(code, 0.0)) > 0.0
                )
                if positive != tuple(record.matched_donors):
                    raise IngestError(
                        "The reported matched donors do not equal the positive-weight support."
                    )
                selections = record.selection_audit
                if len(selections) != len(record.matched_targets) * int(size_M):
                    raise IngestError("The selection audit is incomplete for the declared M.")
                grouped = selections.groupby("target_code")["donor_code"].nunique()
                if len(grouped) != len(record.matched_targets) or bool((grouped != int(size_M)).any()):
                    raise IngestError(
                        "Every target must have exactly M distinct saved neighbours."
                    )
                if set(selections["target_code"].astype(str)) != set(record.matched_targets):
                    raise IngestError("The selection audit target roster is inconsistent.")
                if not set(selections["donor_code"].astype(str)).issubset(
                    set(record.matched_donors)
                ):
                    raise IngestError("The selection audit names a non-effective donor.")
            else:
                audit = record.selection_audit
                if len(audit) != len(record.candidate_targets) + len(record.candidate_donors):
                    raise IngestError("The hard-restriction selection audit is incomplete.")
                for side, candidates, effective in (
                    ("target", record.candidate_targets, record.matched_targets),
                    ("donor", record.candidate_donors, record.matched_donors),
                ):
                    scoped_audit = audit[audit["roster_side"].astype(str) == side]
                    if set(scoped_audit["country_code"].astype(str)) != set(candidates):
                        raise IngestError(
                            f"The hard-restriction {side} candidate audit is inconsistent."
                        )
                    retained = set(
                        scoped_audit.loc[scoped_audit["selected"].astype(bool), "country_code"]
                        .astype(str)
                    )
                    if retained != set(effective):
                        raise IngestError(
                            f"The hard-restriction {side} effective roster is inconsistent."
                        )
            if bool(record.std_diff.isna().any()) or (
                record.std_diff_unweighted is not None
                and bool(record.std_diff_unweighted.isna().any())
            ):
                raise IngestError("A matching balance vector carries a NaN cell.")
            if tuple(record.std_diff.index.astype(str)) != tuple(COVARIATE_DIMENSIONS):
                raise IngestError("The matching balance vector has the wrong covariate axis.")
            if record.std_diff_unweighted is not None and tuple(
                record.std_diff_unweighted.index.astype(str)
            ) != tuple(COVARIATE_DIMENSIONS):
                raise IngestError("The unweighted balance vector has the wrong covariate axis.")
            zero_flags = record.std_diff.attrs.get("zero_variance_flags", {})
            nonfinite_dimensions = [
                str(label)
                for label, value in record.std_diff.items()
                if not np.isfinite(float(value))
            ]
            if any(
                zero_flags.get(label) != "unequal_means_infinite"
                for label in nonfinite_dimensions
            ):
                raise IngestError(
                    "A non-finite balance statistic lacks the required E21 flag."
                )
            observed_max = float(record.std_diff.abs().max())
            observed_covariate = str(record.std_diff.abs().idxmax())
            if not (
                (np.isinf(observed_max) and np.isinf(float(record.max_abs_std_diff)))
                or abs(observed_max - float(record.max_abs_std_diff)) <= IDENTITY_TOLERANCE
            ) or observed_covariate != str(record.max_abs_covariate):
                raise IngestError("The recorded balance maximum is inconsistent with its vector.")
            # Compare the effect with the declared value when supplied.
            if expected_att is not None and (pair[0], pair[1], axis) in expected_att:
                # Read the declared effect.
                raw_declared = expected_att[(pair[0], pair[1], axis)]
                if isinstance(raw_declared, (bool, np.bool_)):
                    raise IngestError("A declared Table 7 effect cannot be Boolean.")
                try:
                    declared: float = float(raw_declared)
                except (TypeError, ValueError) as exc:
                    raise IngestError("A declared Table 7 effect is not numeric.") from exc
                if not np.isfinite(declared):
                    raise IngestError("A declared Table 7 effect must be finite.")
                # Refuse a divergence beyond the tolerance.
                if abs(float(record.att) - declared) > float(tolerance):
                    # The failure names the design and both values.
                    raise IngestError(
                        f"{pair[0]}/{pair[1]}/{axis}: effect {float(record.att)!r} "
                        f"against the declared {declared!r}."
                    )
            # Append the summary row.
            summary_rows.append(
                {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "matching_axis": axis,
                    "target_count": int(len(record.matched_targets)),
                    "donor_count": int(len(record.matched_donors)),
                    "candidate_target_count": int(len(record.candidate_targets)),
                    "candidate_donor_count": int(len(record.candidate_donors)),
                    "weighted": record.weights is not None,
                    "positive_weight_count": (
                        int((record.weights > 0).sum())
                        if record.weights is not None
                        else int(len(record.matched_donors))
                    ),
                    "att": float(record.att),
                    "se_cd": float(record.se_cd),
                    "stars": str(record.stars),
                    "max_abs_std_diff": float(record.max_abs_std_diff),
                    "max_abs_covariate": str(record.max_abs_covariate),
                }
            )
            # Append the weighted balance rows.
            for covariate, value in record.std_diff.items():
                # Record the covariate's difference.
                balance_rows.append(
                    {
                        "policy": pair[0],
                        "outcome": pair[1],
                        "matching_axis": axis,
                        "covariate": str(covariate),
                        "weighting_scheme": (
                            "frequency_weighted" if record.weights is not None else "unweighted"
                        ),
                        "std_diff": float(value),
                        "weighted": record.weights is not None,
                        "variance_convention": str(
                            record.std_diff.attrs.get("variance_convention", "unrecorded")
                        ),
                        "zero_variance_status": str(
                            record.std_diff.attrs.get("zero_variance_flags", {}).get(
                                str(covariate), "positive_variance"
                            )
                        ),
                    }
                )
            # Append the unweighted balance rows of a weighted design so both
            # vectors are auditable.
            if record.std_diff_unweighted is not None:
                # Record each unweighted difference.
                for covariate, value in record.std_diff_unweighted.items():
                    # Append the unweighted row.
                    balance_rows.append(
                        {
                            "policy": pair[0],
                            "outcome": pair[1],
                            "matching_axis": axis,
                            "covariate": str(covariate),
                            "weighting_scheme": "effective_support_unweighted",
                            "std_diff": float(value),
                            "weighted": False,
                            "variance_convention": str(
                                record.std_diff_unweighted.attrs.get(
                                    "variance_convention", "unrecorded"
                                )
                            ),
                            "zero_variance_status": str(
                                record.std_diff_unweighted.attrs.get(
                                    "zero_variance_flags", {}
                                ).get(str(covariate), "positive_variance")
                            ),
                        }
                    )
            # Retain complete roster/selection/fitted-input provenance for Table 7.
            audit_rows.append(
                {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "matching_axis": axis,
                    "candidate_targets": json.dumps(list(record.candidate_targets)),
                    "candidate_donors": json.dumps(list(record.candidate_donors)),
                    "effective_targets": json.dumps(list(record.matched_targets)),
                    "effective_donors": json.dumps(list(record.matched_donors)),
                    "weights_sha256": (
                        _content_digest(record.weights)
                        if record.weights is not None else "not_applicable"
                    ),
                    "selection_sha256": _content_digest(record.selection_audit),
                    "selection_rows": int(len(record.selection_audit)),
                    "provenance_json": json.dumps(
                        _thaw_sweep_evidence(record.design_provenance),
                        sort_keys=True,
                        separators=(",", ":"),
                        allow_nan=False,
                    ),
                }
            )
    # Assemble the summary frame keyed on the design.
    summary: pd.DataFrame = pd.DataFrame(summary_rows).set_index(
        ["policy", "outcome", "matching_axis"]
    )
    # Assemble the balance frame keyed on the design and the covariate.
    balance: pd.DataFrame = pd.DataFrame(balance_rows).set_index(
        ["policy", "outcome", "matching_axis", "weighting_scheme", "covariate"]
    )
    design_audit = pd.DataFrame(audit_rows).set_index(
        ["policy", "outcome", "matching_axis"]
    )
    # Assert the summary's coverage of the requested designs.
    # Compose the covered keys.
    covered = set(summary.index)
    # Refuse a coverage divergence.
    if covered != expected_keys:
        # The failure names the missing designs.
        missing = sorted(expected_keys - covered)
        # Raise with the missing designs attached.
        raise IngestError(f"The summary omits the design(s) {missing}.")
    # Freeze the diagnostics.
    if set(design_audit.index) != expected_keys or not bool(design_audit.index.is_unique):
        raise IngestError("The matching design audit is incomplete or non-unique.")
    fingerprint: str = freeze_matching_diagnostics(summary, balance, design_audit)
    # Accumulate the recorded notes.
    notes: List[str] = [
        "the high-income axis is a hard restriction applied to both rosters "
        "above the median of the pre-treatment income level, not a weight.",
        "the frequency axis reweights the donor cross-section of the factor and "
        "the donor rows of the pooled fit by the inclusion frequencies, so the "
        "estimated design is the design the balance vector describes.",
        "the balance vector is computed on the pre-treatment covariate window "
        "with sample variances, and the maximum absolute difference is "
        "reported beside the full vector.",
        "a weighted design also reports its unweighted balance so the effect of "
        "the weighting is visible.",
    ]
    # Record the design counts.
    notes.append(
        f"designs executed: {len(summary)} across {len(set(selected))} pair(s) "
        f"and {len(tuple(axes))} axis(es)."
    )
    # Assemble the immutable result.
    result: MatchingDiagnosticsResult = MatchingDiagnosticsResult(
        designs=designs,
        summary=summary,
        balance=balance,
        fingerprint_sha256=fingerprint,
        completed_utc=current_utc_iso(),
        notes=tuple(notes),
        design_audit=design_audit,
    )
    # Log the completed execution.
    logger.info(
        "matching diagnostics complete: %d design(s), fingerprint %s.",
        len(summary),
        fingerprint[:16],
    )
    # Return the assembled result.
    return result
````

````python
# Task 24: Execute Matching Designs

# ==============================================================================
# Task 24: Execute Matching Designs
# ==============================================================================

# Declare the label of the baseline design in the balance audit.
DESIGN_BASELINE: Final[str] = "baseline"
# Declare the label of the roster side carrying the treated units.
SIDE_TREATED: Final[str] = "treated"
# Declare the label of the roster side carrying the donors.
SIDE_DONOR: Final[str] = "donor"
# Declare the tolerance of the frequency reconstruction.
COUNT_TOLERANCE: Final[float] = 1.0e-6


# Declare the frozen report of the matching execution.
@dataclass(frozen=True)
class MatchingReportResult:
    """
    Frozen report of the matching execution.

    Attributes
    ----------
    summary : pandas.DataFrame
        One row per pair and axis carrying the rosters' sizes, the effect with
        its inference and the largest absolute standardized difference.
    dropouts : pandas.DataFrame
        One unique row per pair, axis, roster side and candidate country,
        carrying its effective/dropout status and design-specific reason.
    weight_distribution : pandas.DataFrame
        One row per pair and donor carrying the inclusion frequency, the
        reconstructed selection count and whether the donor was selected at
        all, which the specification fixes is reported in full for audit.
    balance : pandas.DataFrame
        Long full-vector audit keyed on pair, design, weighting scheme and
        covariate, including zero-dispersion and reproduction evidence.
    maxima : pandas.DataFrame
        One row per pair, design and weighting scheme carrying the reproduced
        maximum, its covariate and full-vector identity.
    fingerprint_sha256 : str
        Digest of the canonical report payload.
    completed_utc : str
        ISO-8601 UTC timestamp of the execution.
    notes : tuple of str
        Recorded decisions, including the balance audit's descriptive status.
    """
    # The design summary.
    summary: pd.DataFrame
    # The dropout record.
    dropouts: pd.DataFrame
    # The full weight distribution.
    weight_distribution: pd.DataFrame
    # The balance audit across the designs.
    balance: pd.DataFrame
    # The largest absolute difference of every design.
    maxima: pd.DataFrame
    # Digest of the canonical report payload.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the execution.
    completed_utc: str
    # Recorded decisions of the execution.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 24, Step 1: The design summary and the dropout record
# -------------------------------------------------------------------------------------------------------------------------------


# Build the dropout record of every design.
def build_dropouts(
    diagnostics: MatchingDiagnosticsResult,
    partitions: Mapping[Tuple[str, str], Partition],
) -> pd.DataFrame:
    """
    Build the dropout record of every design.

    Parameters
    ----------
    diagnostics : MatchingDiagnosticsResult
        The matching orchestrator's result.
    partitions : mapping of tuple to Partition
        The baseline partitions keyed on the pair.

    Returns
    -------
    pandas.DataFrame
        One uniquely keyed row per pair, axis, roster side and candidate country.
        Every row states whether the identity is effective, the corresponding
        design-specific reason, and the candidate/effective/dropout counts.

    Raises
    ------
    IngestError
        If a pair is absent, candidate/effective rosters disagree with the
        partition, selection evidence is incomplete, or a ledger key repeats.
    """
    # Accumulate the dropout rows.
    rows: List[Dict[str, Any]] = []
    # Iterate over the designs in sorted key order.
    for key in sorted(diagnostics.designs.keys()):
        # Read the design's pair and axis.
        policy, outcome, axis = key
        # Refuse a pair absent from the partitions.
        if (policy, outcome) not in partitions:
            # The failure names the pair.
            raise IngestError(f"The partitions do not carry the pair {(policy, outcome)}.")
        # Read the baseline partition.
        partition: Partition = partitions[(policy, outcome)]
        # Read the design record.
        record: MatchingDesignResult = diagnostics.designs[key]
        candidate_targets = tuple(str(code) for code in partition.targets)
        candidate_donors = tuple(str(code) for code in partition.donors)
        if tuple(record.candidate_targets) != candidate_targets:
            raise IngestError(
                f"{policy}/{outcome}/{axis}: candidate targets disagree with the partition."
            )
        if tuple(record.candidate_donors) != candidate_donors:
            raise IngestError(
                f"{policy}/{outcome}/{axis}: candidate donors disagree with the partition."
            )
        effective_targets = tuple(str(code) for code in record.matched_targets)
        effective_donors = tuple(str(code) for code in record.matched_donors)
        if not set(effective_targets).issubset(set(candidate_targets)) or not set(
            effective_donors
        ).issubset(set(candidate_donors)):
            raise IngestError(
                f"{policy}/{outcome}/{axis}: an effective roster is not a candidate subset."
            )
        # Bind every disposition to the executed selection evidence.
        if axis == AXIS_HIGH_INCOME:
            audit = record.selection_audit.copy(deep=True)
            required = {"roster_side", "country_code", "selected", "selection_rule"}
            if not required.issubset(audit.columns):
                raise IngestError(
                    f"{policy}/{outcome}/{axis}: income selection audit is incomplete."
                )
            audit["country_code"] = audit["country_code"].astype(str)
            if bool(audit.duplicated(["roster_side", "country_code"]).any()) or len(
                audit
            ) != len(candidate_targets) + len(candidate_donors):
                raise IngestError(
                    f"{policy}/{outcome}/{axis}: income selection audit is non-unique or incomplete."
                )
            if set(audit["selection_rule"].astype(str)) != {
                "strictly_above_pooled_median"
            }:
                raise IngestError(
                    f"{policy}/{outcome}/{axis}: income selection reason is mislabelled."
                )
            evidence = {
                (str(row.roster_side), str(row.country_code)): bool(row.selected)
                for row in audit.itertuples(index=False)
            }
        elif axis == AXIS_MAHALANOBIS:
            if record.weights is None:
                raise IngestError(
                    f"{policy}/{outcome}/{axis}: frequency design omits its weights."
                )
            if not bool(record.weights.index.is_unique) or tuple(
                str(code) for code in record.weights.index
            ) != candidate_donors:
                raise IngestError(
                    f"{policy}/{outcome}/{axis}: frequency candidate weights are incomplete."
                )
            if bool(record.weights.isna().any()) or not bool(
                np.isfinite(record.weights.to_numpy(dtype=float)).all()
            ) or bool((record.weights < 0.0).any()) or abs(
                float(record.weights.sum()) - 1.0
            ) > 1.0e-9:
                raise IngestError(
                    f"{policy}/{outcome}/{axis}: frequency candidate weights are malformed."
                )
            evidence = {
                ("target", code): True for code in candidate_targets
            }
            evidence.update(
                {
                    ("donor", code): bool(float(record.weights.get(code, 0.0)) > 0.0)
                    for code in candidate_donors
                }
            )
        else:
            raise IngestError(f"Undeclared matching axis {axis!r} in dropout ledger.")
        # Emit one row per candidate identity on each roster side.
        for side, candidates, effective in (
            (SIDE_TREATED, candidate_targets, effective_targets),
            (SIDE_DONOR, candidate_donors, effective_donors),
        ):
            audit_side = "target" if side == SIDE_TREATED else "donor"
            expected_effective = set(effective)
            for code in candidates:
                if (audit_side, code) not in evidence:
                    raise IngestError(
                        f"{policy}/{outcome}/{axis}: selection evidence omits "
                        f"{audit_side} {code}."
                    )
                retained = bool(evidence[(audit_side, code)])
                if retained != (code in expected_effective):
                    raise IngestError(
                        f"{policy}/{outcome}/{axis}: disposition evidence disagrees "
                        f"for {audit_side} {code}."
                    )
                if axis == AXIS_HIGH_INCOME:
                    reason = (
                        "retained_above_pooled_income_median"
                        if retained else "below_or_equal_pooled_income_median"
                    )
                elif side == SIDE_DONOR:
                    reason = (
                        "positive_selection_frequency"
                        if retained else "zero_selection_frequency"
                    )
                else:
                    reason = "all_targets_retained"
                rows.append(
                    {
                        "policy": policy,
                        "outcome": outcome,
                        "matching_axis": axis,
                        "roster_side": side,
                        "country_code": code,
                        "candidate": True,
                        "effective": retained,
                        "dropped": not retained,
                        "reason": reason,
                        "count_before": int(len(candidates)),
                        "count_after": int(len(effective)),
                        "dropped_count": int(len(candidates) - len(effective)),
                    }
                )
    # Assemble the country-level ledger on its complete scientific key.
    table: pd.DataFrame = pd.DataFrame(rows).set_index(
        ["policy", "outcome", "matching_axis", "roster_side", "country_code"]
    )
    if not bool(table.index.is_unique):
        raise IngestError("The matching dropout ledger carries duplicate design/role/country keys.")
    # Reconcile every repeated count against the actual candidate/effective sets.
    for key, group in table.groupby(level=[0, 1, 2, 3], sort=True):
        if int(group["candidate"].sum()) != int(group["count_before"].iloc[0]):
            raise IngestError(f"{key}: dropout candidate count is inconsistent.")
        if int(group["effective"].sum()) != int(group["count_after"].iloc[0]):
            raise IngestError(f"{key}: dropout effective count is inconsistent.")
        if int(group["dropped"].sum()) != int(group["dropped_count"].iloc[0]):
            raise IngestError(f"{key}: dropout count is inconsistent.")
    # Log the record's extent.
    logger.info("dropout record built across %d row(s).", len(table))
    # Return the frame.
    return table


# -------------------------------------------------------------------------------------------------------------------------------
# Task 24, Step 2: The full weight distribution
# -------------------------------------------------------------------------------------------------------------------------------


# Build the full weight distribution of the frequency design.
def build_weight_distribution(
    diagnostics: MatchingDiagnosticsResult,
    size_M: int = DEFAULT_MATCH_SIZE,
) -> pd.DataFrame:
    """
    Build the full weight distribution of the frequency design.

    Parameters
    ----------
    diagnostics : MatchingDiagnosticsResult
        The matching orchestrator's result.
    size_M : int, optional
        The neighbour count the designs were executed with. The reconstruction
        divides each frequency by the design's own selection total, so the
        caller's count must be supplied rather than assumed from the module's
        default, which would misreport every design executed at another count.

    Returns
    -------
    pandas.DataFrame
        One row per pair and donor carrying the inclusion frequency, the
        reconstructed selection count and whether the donor was selected at
        all. The specification fixes that the full distribution is reported,
        because the average frequency alone would hide a design whose weights
        collapse onto a few donors.

    Raises
    ------
    IngestError
        If a frequency design is absent, if its weights do not sum to one, or
        if the reconstructed selection total does not match the roster and the
        neighbour count, because either divergence would mean the frequency
        vector is inconsistent with the selection it reports.
    """
    if (
        isinstance(size_M, (bool, np.bool_))
        or not isinstance(size_M, (int, np.integer))
        or int(size_M) <= 0
    ):
        raise IngestError("The report neighbour count must be a positive integer.")
    # Accumulate the distribution rows.
    rows: List[Dict[str, Any]] = []
    # Iterate over the designs in sorted key order.
    for key in sorted(diagnostics.designs.keys()):
        # Read the design's identity and record.
        policy, outcome, axis = key
        # Select the frequency designs only, because the hard restriction
        # carries no weight vector.
        record: MatchingDesignResult = diagnostics.designs[key]
        # Skip a design without weights, recording nothing for it.
        if record.weights is None:
            # Continue to the next design.
            continue
        # Read the weight vector.
        weights: pd.Series = record.weights
        if (
            not bool(weights.index.is_unique)
            or tuple(str(code) for code in weights.index) != tuple(record.candidate_donors)
            or bool(weights.isna().any())
            or not bool(np.isfinite(weights.to_numpy(dtype=float)).all())
            or bool((weights < 0.0).any())
        ):
            raise IngestError(
                f"{policy}/{outcome}/{axis}: the candidate weight vector is malformed."
            )
        # Assert the declared normalization.
        if abs(float(weights.sum()) - 1.0) > 1.0e-9:
            # The failure names the design and the observed total.
            raise IngestError(
                f"{policy}/{outcome}/{axis}: the inclusion frequencies sum to "
                f"{float(weights.sum())!r} rather than to one."
            )
        # Recover the actual denominator from the immutable selection audit,
        # not from the number of unique positive-weight donors (which can be
        # smaller than M when targets share neighbours).
        if len(record.selection_audit) == 0:
            raise IngestError(
                f"{policy}/{outcome}/{axis}: the frequency design omits its "
                "selection audit."
            )
        target_count: int = int(len(record.matched_targets))
        required = {"target_code", "donor_code", "neighbour_rank"}
        if not required.issubset(record.selection_audit.columns):
            raise IngestError(
                f"{policy}/{outcome}/{axis}: selection audit omits required columns."
            )
        selections = record.selection_audit.copy(deep=True)
        selections["target_code"] = selections["target_code"].astype(str)
        selections["donor_code"] = selections["donor_code"].astype(str)
        ranks = pd.to_numeric(selections["neighbour_rank"], errors="coerce")
        if bool(ranks.isna().any()) or bool((ranks % 1 != 0).any()):
            raise IngestError(
                f"{policy}/{outcome}/{axis}: neighbour ranks are not integral."
            )
        selections["neighbour_rank"] = ranks.astype(int)
        if bool(selections.duplicated(["target_code", "donor_code"]).any()):
            raise IngestError(
                f"{policy}/{outcome}/{axis}: a target repeats a selected donor."
            )
        grouped = selections.groupby("target_code")["donor_code"].nunique()
        rank_sets = selections.groupby("target_code")["neighbour_rank"].apply(set)
        expected_ranks = set(range(1, int(size_M) + 1))
        if (
            len(grouped) != target_count
            or bool((grouped != int(size_M)).any())
            or len(selections) != target_count * int(size_M)
            or set(grouped.index.astype(str)) != set(record.matched_targets)
            or any(observed != expected_ranks for observed in rank_sets)
        ):
            raise IngestError(
                f"{policy}/{outcome}/{axis}: the saved selection audit does not "
                "contain exactly M distinct donors per target."
            )
        if not set(selections["donor_code"]).issubset(set(record.candidate_donors)):
            raise IngestError(
                f"{policy}/{outcome}/{axis}: selection audit names an ineligible donor."
            )
        denominator: float = float(len(selections))
        actual_counts = selections["donor_code"].value_counts().reindex(
            list(record.candidate_donors), fill_value=0
        )
        # Accumulate the reconstructed counts.
        reconstructed: List[float] = []
        # Iterate over the weight vector in its own order.
        for donor, weight in weights.items():
            # Reconstruct the donor's selection count from its frequency.
            count = float(weight) * denominator
            if abs(count - round(count)) > COUNT_TOLERANCE:
                raise IngestError(
                    f"{policy}/{outcome}/{axis}/{donor}: weight does not reconstruct "
                    "an integer selection count."
                )
            actual_count = int(actual_counts.loc[str(donor)])
            if int(round(count)) != actual_count:
                raise IngestError(
                    f"{policy}/{outcome}/{axis}/{donor}: weight-implied count "
                    f"{int(round(count))} disagrees with saved selections {actual_count}."
                )
            # Record the reconstructed count.
            reconstructed.append(count)
            # Append the donor's distribution row.
            rows.append(
                {
                    "policy": policy,
                    "outcome": outcome,
                    "matching_axis": axis,
                    "donor_code": str(donor),
                    "inclusion_frequency": float(weight),
                    "selection_count": actual_count,
                    "selected": bool(actual_count > 0),
                    "effective_donor": bool(str(donor) in set(record.matched_donors)),
                    "selection_total": int(denominator),
                    "neighbours_per_target": int(size_M),
                }
            )
        # Refuse a reconstructed total that does not match the denominator,
        # which would mean the frequencies are inconsistent with the design.
        if abs(sum(reconstructed) - denominator) > COUNT_TOLERANCE:
            # The failure names the design and both totals.
            raise IngestError(
                f"{policy}/{outcome}/{axis}: the reconstructed selection total "
                f"{sum(reconstructed)!r} does not match the design's {denominator!r}."
            )
        positive_support = tuple(
            str(code) for code in record.candidate_donors
            if int(actual_counts.loc[str(code)]) > 0
        )
        if positive_support != tuple(record.matched_donors):
            raise IngestError(
                f"{policy}/{outcome}/{axis}: effective donors disagree with positive selections."
            )
    # Assemble the distribution frame keyed on the design and the donor.
    table: pd.DataFrame = pd.DataFrame(rows).set_index(
        ["policy", "outcome", "matching_axis", "donor_code"]
    )
    if not bool(table.index.is_unique):
        raise IngestError("The matching weight distribution carries duplicate donor keys.")
    # Refuse an empty table, because a frequency design was executed.
    if len(table) == 0:
        # The failure states the condition.
        raise IngestError(
            "The weight distribution is empty although a frequency design was "
            "executed."
        )
    # Log the distribution's extent and the unselected donors.
    logger.info(
        "weight distribution built across %d donor row(s) with %d unselected "
        "donor(s).",
        len(table),
        int((~table["selected"]).sum()),
    )
    # Return the frame.
    return table


# -------------------------------------------------------------------------------------------------------------------------------
# Task 24, Step 3: The standardized-difference balance audit
# -------------------------------------------------------------------------------------------------------------------------------


# Build the balance audit across the baseline and the matched designs.
def build_balance_audit(
    diagnostics: MatchingDiagnosticsResult,
    partitions: Mapping[Tuple[str, str], Partition],
    panel: pd.DataFrame,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Build the balance audit across the baseline and the matched designs.

    Parameters
    ----------
    diagnostics : MatchingDiagnosticsResult
        The matching orchestrator's result.
    partitions : mapping of tuple to Partition
        The baseline partitions keyed on the pair.
    panel : pandas.DataFrame
        The analysis panel.

    Returns
    -------
    tuple
        A long balance table uniquely keyed by policy, outcome, design,
        weighting scheme and covariate, plus a scheme-labelled maxima frame.
        Every matched vector is independently recomputed from the effective
        roster before its maximum is accepted. The audit remains descriptive:
        reduced imbalance does not establish balance.

    Raises
    ------
    IngestError
        If a pair or design is absent; a full vector, scheme, flag or maximum
        disagrees with its independently reproduced value; or a key repeats.
    """
    expected_source_names = [
        "policy", "outcome", "matching_axis", "weighting_scheme", "covariate"
    ]
    if list(diagnostics.balance.index.names) != expected_source_names or not bool(
        diagnostics.balance.index.is_unique
    ):
        raise IngestError(
            "The Task 23 balance source lacks a unique scheme-labelled component key."
        )
    # Accumulate fully labelled component and maximum rows.
    balance_rows: List[Dict[str, Any]] = []
    maxima_rows: List[Dict[str, Any]] = []
    # Append a reproduced full vector and its maximum under one exact scheme.
    def append_vector(
        pair: Tuple[str, str],
        design: str,
        scheme: str,
        reproduced: pd.Series,
        stored: Optional[pd.Series],
    ) -> None:
        if tuple(reproduced.index.astype(str)) != tuple(COVARIATE_DIMENSIONS):
            raise IngestError(f"{pair}/{design}/{scheme}: covariate axis is incomplete.")
        if bool(reproduced.isna().any()):
            raise IngestError(f"{pair}/{design}/{scheme}: balance vector carries NaN.")
        flags = dict(reproduced.attrs.get("zero_variance_flags", {}))
        convention = str(reproduced.attrs.get("variance_convention", ""))
        if not convention:
            raise IngestError(f"{pair}/{design}/{scheme}: variance convention is absent.")
        if stored is not None:
            if tuple(stored.index.astype(str)) != tuple(COVARIATE_DIMENSIONS):
                raise IngestError(f"{pair}/{design}/{scheme}: stored vector axis is incomplete.")
            if not bool(
                np.allclose(
                    reproduced.to_numpy(dtype=float),
                    stored.to_numpy(dtype=float),
                    rtol=0.0,
                    atol=IDENTITY_TOLERANCE,
                    equal_nan=False,
                )
            ):
                raise IngestError(
                    f"{pair}/{design}/{scheme}: stored balance vector is not reproduced."
                )
            if convention != str(stored.attrs.get("variance_convention", "")) or flags != dict(
                stored.attrs.get("zero_variance_flags", {})
            ):
                raise IngestError(
                    f"{pair}/{design}/{scheme}: stored balance metadata is not reproduced."
                )
        vector_digest = _content_digest(reproduced)
        for covariate, value in reproduced.items():
            status = str(flags.get(str(covariate), "positive_variance"))
            if not np.isfinite(float(value)) and status != "unequal_means_infinite":
                raise IngestError(
                    f"{pair}/{design}/{scheme}/{covariate}: non-finite imbalance lacks E21 status."
                )
            balance_rows.append(
                {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "design": str(design),
                    "weighting_scheme": str(scheme),
                    "covariate": str(covariate),
                    "std_diff": float(value),
                    "variance_convention": convention,
                    "zero_variance_status": status,
                    "vector_sha256": vector_digest,
                    "vector_reproduced": True,
                }
            )
        absolute = reproduced.abs()
        maximum = float(absolute.max())
        maximizer = str(absolute.idxmax())
        infinite_count = int(np.isinf(reproduced.to_numpy(dtype=float)).sum())
        maxima_rows.append(
            {
                "policy": pair[0],
                "outcome": pair[1],
                "design": str(design),
                "weighting_scheme": str(scheme),
                "max_abs_std_diff": maximum,
                "max_abs_covariate": maximizer,
                "component_count": int(len(reproduced)),
                "undefined_dimensions": int(reproduced.isna().sum()),
                "infinite_dimensions": infinite_count,
                "zero_dispersion_dimensions": int(len(flags)),
                "max_status": (
                    "infinite_zero_dispersion_unequal"
                    if np.isinf(maximum) else "finite"
                ),
                "variance_convention": convention,
                "vector_sha256": vector_digest,
                "vector_reproduced": True,
            }
        )
    # Iterate over the pairs covered by the diagnostics.
    pairs = sorted({(str(policy), str(outcome)) for policy, outcome, _ in diagnostics.designs})
    # Iterate over the pairs.
    for pair in pairs:
        # Refuse a pair absent from the partitions.
        if pair not in partitions:
            # The failure names the pair.
            raise IngestError(f"The partitions do not carry the pair {pair}.")
        # Read the baseline partition.
        partition: Partition = partitions[pair]
        # Compose the covariate vectors of the union of the baseline rosters.
        union: List[str] = sorted(
            set(str(code) for code in partition.targets)
            | set(str(code) for code in partition.donors)
        )
        # Compose the vectors.
        covariates: pd.DataFrame = build_pretreatment_covariates(panel, union)
        # Compose the baseline vector through the shared formula, on the
        # untouched rosters so the comparison is not circular.
        baseline_vector: pd.Series = compute_standardized_differences(
            covariates, partition.targets, partition.donors
        )
        append_vector(pair, DESIGN_BASELINE, "unweighted", baseline_vector, None)
        # Attach each matched design's column.
        for axis in MATCHING_AXES:
            # Skip an axis the execution did not cover.
            if (pair[0], pair[1], axis) not in diagnostics.designs:
                # Continue to the next axis.
                continue
            # Read the design's record.
            record: MatchingDesignResult = diagnostics.designs[(pair[0], pair[1], axis)]
            if tuple(record.candidate_targets) != tuple(str(code) for code in partition.targets) or tuple(
                record.candidate_donors
            ) != tuple(str(code) for code in partition.donors):
                raise IngestError(f"{pair}/{axis}: candidate roster provenance drifted.")
            if axis == AXIS_HIGH_INCOME:
                reproduced = compute_standardized_differences(
                    covariates, record.matched_targets, record.matched_donors
                )
                append_vector(pair, axis, "unweighted", reproduced, record.std_diff)
            elif axis == AXIS_MAHALANOBIS:
                if record.weights is None or len(record.selection_audit) == 0:
                    raise IngestError(f"{pair}/{axis}: frequency evidence is incomplete.")
                effective_weights = record.weights.reindex(record.matched_donors)
                weighted = compute_standardized_differences(
                    covariates,
                    record.matched_targets,
                    record.matched_donors,
                    effective_weights,
                    weighting_convention="frequency",
                    frequency_total=int(len(record.selection_audit)),
                )
                append_vector(
                    pair, axis, "frequency_weighted", weighted, record.std_diff
                )
                unweighted = compute_standardized_differences(
                    covariates, record.matched_targets, record.matched_donors
                )
                append_vector(
                    pair,
                    axis,
                    "effective_support_unweighted",
                    unweighted,
                    record.std_diff_unweighted,
                )
            else:
                raise IngestError(f"Undeclared matching axis {axis!r} in balance audit.")
    # Assemble the long balance frame on the complete scheme-labelled key.
    balance: pd.DataFrame = pd.DataFrame(balance_rows).set_index(
        ["policy", "outcome", "design", "weighting_scheme", "covariate"]
    )
    # Assemble the maxima frame on the same design/scheme identity.
    maxima: pd.DataFrame = pd.DataFrame(maxima_rows).set_index(
        ["policy", "outcome", "design", "weighting_scheme"]
    )
    if not bool(balance.index.is_unique) or not bool(maxima.index.is_unique):
        raise IngestError("The matching balance audit carries duplicate labelled keys.")
    # Verify Task 23's stored component table independently, excluding baseline.
    for index, row in balance.iterrows():
        policy, outcome, design, scheme, covariate = index
        if design == DESIGN_BASELINE:
            continue
        source_key = (policy, outcome, design, scheme, covariate)
        if source_key not in diagnostics.balance.index:
            raise IngestError(f"The Task 23 balance table omits {source_key}.")
        source = diagnostics.balance.loc[source_key]
        observed = float(row["std_diff"])
        supplied = float(source["std_diff"])
        values_agree = (
            np.isinf(observed) and np.isinf(supplied) and np.sign(observed) == np.sign(supplied)
        ) or abs(observed - supplied) <= IDENTITY_TOLERANCE
        if not values_agree or str(row["variance_convention"]) != str(
            source["variance_convention"]
        ) or str(row["zero_variance_status"]) != str(source["zero_variance_status"]):
            raise IngestError(f"The Task 23 balance component {source_key} is stale.")
    # Assert the maxima's agreement with the summary for every matched design.
    for (policy, outcome, design, scheme), row in maxima.iterrows():
        # Skip the baseline, which has no summary row.
        if str(design) == DESIGN_BASELINE:
            # Continue to the next design.
            continue
        # Read the summary's maximum.
        headline_scheme = (
            str(design) == AXIS_HIGH_INCOME and str(scheme) == "unweighted"
        ) or (
            str(design) == AXIS_MAHALANOBIS and str(scheme) == "frequency_weighted"
        )
        if not headline_scheme:
            continue
        summary_value = float(
            diagnostics.summary.loc[(policy, outcome, design), "max_abs_std_diff"]
        )
        # Refuse a disagreement between the two artefacts.
        audit_value = float(row["max_abs_std_diff"])
        if not (
            (np.isinf(audit_value) and np.isinf(summary_value))
            or abs(audit_value - summary_value) <= IDENTITY_TOLERANCE
        ):
            # The failure names the design and both values.
            raise IngestError(
                f"{policy}/{outcome}/{design}: the audit's maximum "
                f"{float(row['max_abs_std_diff'])!r} disagrees with the summary's "
                f"{summary_value!r}."
            )
    # Log the audit's extent.
    logger.info(
        "balance audit built across %d component row(s) and %d design/scheme maximum row(s).",
        len(balance),
        len(maxima),
    )
    # Return the two frames.
    return balance, maxima


# Compare the audit's maxima with the declared study values.
def assert_declared_balance(
    maxima: pd.DataFrame,
    expected: Optional[Mapping[Any, Any]] = None,
    tolerance: float = 1.0e-3,
) -> None:
    """
    Compare the audit's maxima with the declared study values.

    Parameters
    ----------
    maxima : pandas.DataFrame
        The reproduced maxima keyed on policy, outcome, design and weighting
        scheme, carrying full-vector identity and zero-dispersion evidence.
    expected : mapping, optional
        Publication cells keyed by the exact four-part identity (or legacy
        design label). A value may be numeric, using the documented default of
        three displayed decimals, or a mapping with ``reported_value``,
        ``displayed_precision``, optional ``allowance`` and
        ``max_abs_covariate``.
    tolerance : float, optional
        The tolerance of the comparison.

    Returns
    -------
    None
        The function returns nothing when every supplied comparison holds.

    Raises
    ------
    IngestError
        If a vector was not independently reproduced, carries an undefined or
        mislabelled component, or a publication cell is absent or outside its
        rounding interval plus explicit allowance.
    """
    if isinstance(tolerance, (bool, np.bool_)):
        raise IngestError("Balance publication tolerance must be finite and nonnegative.")
    try:
        tolerance_value = float(tolerance)
    except (TypeError, ValueError) as exc:
        raise IngestError(
            "Balance publication tolerance must be finite and nonnegative."
        ) from exc
    if not np.isfinite(tolerance_value) or tolerance_value < 0.0:
        raise IngestError("Balance publication tolerance must be finite and nonnegative.")
    required_columns = {
        "max_abs_std_diff",
        "max_abs_covariate",
        "component_count",
        "undefined_dimensions",
        "infinite_dimensions",
        "zero_dispersion_dimensions",
        "max_status",
        "variance_convention",
        "vector_sha256",
        "vector_reproduced",
    }
    if not required_columns.issubset(maxima.columns):
        raise IngestError(
            f"The maxima audit omits column(s) {sorted(required_columns - set(maxima.columns))}."
        )
    if list(maxima.index.names) != [
        "policy", "outcome", "design", "weighting_scheme"
    ] or not bool(maxima.index.is_unique):
        raise IngestError("The maxima audit key is incomplete or non-unique.")
    allowed_schemes = {
        DESIGN_BASELINE: {"unweighted"},
        AXIS_HIGH_INCOME: {"unweighted"},
        AXIS_MAHALANOBIS: {"frequency_weighted", "effective_support_unweighted"},
    }
    # Validate the full-vector reconstruction evidence before consulting any
    # declared scalar maximum.
    for index, row in maxima.iterrows():
        design = str(index[2])
        scheme = str(index[3])
        if design not in allowed_schemes or scheme not in allowed_schemes[design]:
            raise IngestError(f"The maxima audit mislabels design/scheme {design}/{scheme}.")
        if int(row["component_count"]) != len(COVARIATE_DIMENSIONS):
            raise IngestError(f"{index}: the full covariate vector is incomplete.")
        if int(row["undefined_dimensions"]) != 0:
            raise IngestError(f"{index}: the balance vector carries undefined component(s).")
        if not bool(row["vector_reproduced"]):
            raise IngestError(f"{index}: the full balance vector was not reproduced.")
        if str(row["max_abs_covariate"]) not in set(COVARIATE_DIMENSIONS):
            raise IngestError(f"{index}: the maximizing covariate is mislabelled.")
        if len(str(row["vector_sha256"])) != 64 or not all(
            character in "0123456789abcdef" for character in str(row["vector_sha256"])
        ):
            raise IngestError(f"{index}: the balance-vector digest is malformed.")
        value = float(row["max_abs_std_diff"])
        if np.isnan(value) or value < 0.0:
            raise IngestError(f"{index}: the maximum imbalance is invalid.")
        if np.isinf(value) and str(row["max_status"]) != "infinite_zero_dispersion_unequal":
            raise IngestError(f"{index}: infinite imbalance lacks the E21 status.")
        if np.isfinite(value) and str(row["max_status"]) != "finite":
            raise IngestError(f"{index}: finite imbalance carries the wrong status.")
    # Skip scalar publication comparisons when no declared values were supplied.
    if expected is None:
        # Log the observed maxima for the evidence record.
        logger.info(
            "declared balance maxima not supplied; observed maxima: %s.",
            maxima["max_abs_std_diff"].round(6).to_dict(),
        )
        # Return without comparison.
        return
    # Compare each declared publication cell under its exact or legacy identity.
    for declared_key, declaration in expected.items():
        if isinstance(declared_key, tuple):
            if len(declared_key) not in (3, 4):
                raise IngestError(f"Malformed balance publication key {declared_key!r}.")
            policy, outcome, design = (str(value) for value in declared_key[:3])
            scheme = (
                str(declared_key[3])
                if len(declared_key) == 4
                else (
                    "frequency_weighted"
                    if design == AXIS_MAHALANOBIS else "unweighted"
                )
            )
            exact_key = (policy, outcome, design, scheme)
            scoped = maxima.loc[[exact_key]] if exact_key in maxima.index else maxima.iloc[0:0]
        else:
            design = str(declared_key)
            scheme = "frequency_weighted" if design == AXIS_MAHALANOBIS else "unweighted"
            scoped = maxima[
                (maxima.index.get_level_values("design") == design)
                & (maxima.index.get_level_values("weighting_scheme") == scheme)
            ]
        # Refuse a design the frame does not carry.
        if len(scoped) == 0:
            # The failure names the absent design.
            raise IngestError(f"The maxima frame omits declared cell {declared_key!r}.")
        if isinstance(declaration, Mapping):
            raw_reported = declaration.get(
                "reported_value", declaration.get("reported", declaration.get("value"))
            )
            raw_precision = declaration.get("displayed_precision", 3)
            raw_allowance = declaration.get("allowance", tolerance_value)
            expected_covariate = declaration.get("max_abs_covariate")
            # Reject contradictory identity metadata instead of using the key
            # for lookup and silently displaying a different labelled cell.
            identity_positions = {
                "policy": 0,
                "outcome": 1,
                "design": 2,
                "weighting_scheme": 3,
            }
            for field, position in identity_positions.items():
                if field in declaration:
                    observed_labels = set(
                        str(index[position]) for index in scoped.index.tolist()
                    )
                    if observed_labels != {str(declaration[field])}:
                        raise IngestError(
                            f"Declared balance cell {declared_key!r} contradicts "
                            f"its {field} metadata."
                        )
        else:
            raw_reported = declaration
            raw_precision = 3
            raw_allowance = tolerance_value
            expected_covariate = None
        if isinstance(raw_reported, (bool, np.bool_)) or isinstance(
            raw_precision, (bool, np.bool_)
        ) or isinstance(raw_allowance, (bool, np.bool_)):
            raise IngestError(f"Declared balance cell {declared_key!r} is malformed.")
        try:
            reported = float(raw_reported)
            precision = int(raw_precision)
            allowance = float(raw_allowance)
        except (TypeError, ValueError) as exc:
            raise IngestError(f"Declared balance cell {declared_key!r} is malformed.") from exc
        if (
            not np.isfinite(reported)
            or precision < 0
            or float(raw_precision) != float(precision)
            or not np.isfinite(allowance)
            or allowance < 0.0
        ):
            raise IngestError(f"Declared balance cell {declared_key!r} is malformed.")
        # Compose the largest observed maximum of the design, which is the
        # legacy manuscript mapping reports across panels; exact keys select one.
        observed: float = float(scoped["max_abs_std_diff"].max())
        if not np.isfinite(observed):
            raise IngestError(
                f"{declared_key!r}: an infinite E21 imbalance cannot reproduce a finite publication cell."
            )
        rounding_half_width = 0.5 * (10.0 ** (-precision))
        lower = reported - rounding_half_width - allowance
        upper = reported + rounding_half_width + allowance
        if observed < lower or observed > upper:
            # The failure names the design and both values.
            raise IngestError(
                f"{declared_key!r}: largest absolute standardized difference "
                f"{observed!r} lies outside publication interval [{lower!r}, {upper!r}]."
            )
        if expected_covariate is not None:
            maximizers = set(str(value) for value in scoped["max_abs_covariate"])
            if maximizers != {str(expected_covariate)}:
                raise IngestError(
                    f"{declared_key!r}: maximizing covariate(s) {sorted(maximizers)} "
                    f"disagree with {expected_covariate!r}."
                )
    # Log the completed comparison.
    logger.info("declared balance maxima verified across %d design(s).", len(expected))


# Freeze the report behind a canonical digest.
def freeze_matching_report(
    summary: pd.DataFrame,
    dropouts: pd.DataFrame,
    weight_distribution: pd.DataFrame,
    balance: pd.DataFrame,
) -> str:
    """
    Freeze the report behind a canonical digest.

    Parameters
    ----------
    summary : pandas.DataFrame
        The design summary.
    dropouts : pandas.DataFrame
        The unique candidate-country disposition ledger.
    weight_distribution : pandas.DataFrame
        The weight distribution.
    balance : pandas.DataFrame
        The long, scheme-labelled full-vector balance audit.

    Returns
    -------
    str
        The hexadecimal digest of the canonical payload.

    Raises
    ------
    IngestError
        If any frame is empty, carries duplicate keys, or lacks the documented
        country/scheme-labelled scientific index.
    """
    # Refuse an empty frame.
    for label, frame in (
        ("summary", summary),
        ("dropouts", dropouts),
        ("weight distribution", weight_distribution),
        ("balance", balance),
    ):
        # Test the frame's extent.
        if len(frame) == 0:
            # The failure names the frame.
            raise IngestError(f"Cannot freeze an empty {label} frame.")
    # Refuse ambiguous rows before hashing; a digest cannot repair a duplicate
    # scientific key.
    for label, frame in (
        ("summary", summary),
        ("dropouts", dropouts),
        ("weight distribution", weight_distribution),
        ("balance", balance),
    ):
        if not bool(frame.index.is_unique):
            raise IngestError(f"Cannot freeze {label} with duplicate keys.")
    expected_dropout_names = [
        "policy", "outcome", "matching_axis", "roster_side", "country_code"
    ]
    expected_balance_names = [
        "policy", "outcome", "design", "weighting_scheme", "covariate"
    ]
    if list(dropouts.index.names) != expected_dropout_names:
        raise IngestError("The dropout ledger lacks its design/role/country key.")
    if list(balance.index.names) != expected_balance_names:
        raise IngestError("The balance audit lacks its scheme-labelled key.")
    # Bind every labelled row, schema and dtype through the shared deterministic
    # content digest rather than serializing a lossy scalar subset.
    payload = "|".join(
        _content_digest(frame.sort_index())
        for frame in (summary, dropouts, weight_distribution, balance)
    )
    return hashlib.sha256(payload.encode("utf-8")).hexdigest()


# -------------------------------------------------------------------------------------------------------------------------------
# Task 24, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Execute the matching designs and their reporting artefacts.
def execute_matching_report(
    baseline_config: Mapping[str, Any],
    data: Mapping[str, Any],
    pairs: Optional[Sequence[Tuple[str, str]]] = None,
    size_M: int = DEFAULT_MATCH_SIZE,
    expected_att: Optional[Mapping[Tuple[str, str, str], float]] = None,
    expected_balance: Optional[Mapping[Any, Any]] = None,
    tolerance: float = 1.0e-3,
    diagnostics: Optional[MatchingDiagnosticsResult] = None,
) -> MatchingReportResult:
    """
    Execute the matching designs and their reporting artefacts.

    Parameters
    ----------
    baseline_config : mapping
        The baseline configuration, which is never mutated.
    data : mapping
        The frozen inputs under the declared keys.
    pairs : sequence of tuple, optional
        The pairs the execution covers. When absent every registered pair is
        covered.
    size_M : int, optional
        The number of nearest donors the frequency axis selects per target.
    expected_att : mapping of tuple to float, optional
        The declared effects keyed on the pair and the axis.
    expected_balance : mapping, optional
        Publication-aware declared maxima by exact cell or legacy design label.
    tolerance : float, optional
        The tolerance of the comparisons.

    Returns
    -------
    MatchingReportResult
        The summary, the dropout record, the weight distribution, the balance
        audit with its maxima, the digest, the timestamp and the notes.

    Raises
    ------
    KeyError
        If the data mapping lacks a declared key.
    IngestError
        If any orchestration, reconstruction, coverage or comparison check
        fails.
    """
    # Refuse a data mapping lacking the declared keys.
    for key in ("registry", "panel", "lagged"):
        # Test the key's presence.
        if key not in data:
            # The failure names the absent key.
            raise KeyError(f"The data mapping lacks the declared key {key!r}.")
    # Compose the pair list, defaulting to every registered pair.
    selected: List[Tuple[str, str]] = (
        [tuple(pair) for pair in pairs]
        if pairs is not None
        else sorted(data["registry"].partitions.keys())
    )
    if not selected or len(set(selected)) != len(selected):
        raise IngestError("Matching report pairs must be a nonempty unique sequence.")
    # Reuse an explicitly supplied, fingerprinted Task 23 result when the
    # pipeline has already executed it; direct callers retain the historical
    # fresh-execution path.
    if diagnostics is None:
        diagnostics = execute_matching_designs(
            baseline_config,
            data,
            pairs=selected,
            axes=MATCHING_AXES,
            size_M=size_M,
            expected_att=expected_att,
            tolerance=tolerance,
        )
    elif not isinstance(diagnostics, MatchingDiagnosticsResult):
        raise IngestError("Supplied matching diagnostics are malformed.")
    expected_design_keys = {
        (str(policy), str(outcome), str(axis))
        for policy, outcome in selected
        for axis in MATCHING_AXES
    }
    if set(diagnostics.designs) != expected_design_keys:
        raise IngestError("Supplied matching diagnostics do not cover the requested designs.")
    if diagnostics.fingerprint_sha256 != freeze_matching_diagnostics(
        diagnostics.summary, diagnostics.balance, diagnostics.design_audit
    ):
        raise IngestError("Supplied matching diagnostics fingerprint is stale.")
    # Compose the summary frame from the orchestrator's own summary.
    summary: pd.DataFrame = diagnostics.summary.copy(deep=True)
    # Compose the dropout record.
    dropouts: pd.DataFrame = build_dropouts(
        diagnostics, dict(data["registry"].partitions)
    )
    # Compose the weight distribution with the executed neighbour count.
    weight_distribution: pd.DataFrame = build_weight_distribution(diagnostics, size_M)
    # Compose the balance audit and its maxima.
    balance, maxima = build_balance_audit(
        diagnostics, dict(data["registry"].partitions), data["panel"]
    )
    # Require the exact four report schemes for each requested pair.
    expected_scheme_keys = {
        (pair[0], pair[1], DESIGN_BASELINE, "unweighted")
        for pair in selected
    } | {
        (pair[0], pair[1], AXIS_HIGH_INCOME, "unweighted")
        for pair in selected
    } | {
        (pair[0], pair[1], AXIS_MAHALANOBIS, scheme)
        for pair in selected
        for scheme in ("frequency_weighted", "effective_support_unweighted")
    }
    observed_scheme_keys = set(
        tuple(index[:4]) for index in balance.index.unique().tolist()
    )
    if observed_scheme_keys != expected_scheme_keys:
        raise IngestError(
            "The balance report omits or adds a design/weighting scheme: "
            f"missing {sorted(expected_scheme_keys - observed_scheme_keys)}, "
            f"extra {sorted(observed_scheme_keys - expected_scheme_keys)}."
        )
    # Reconcile summary counts to the country-level candidate/effective ledger.
    for (policy, outcome, axis), row in summary.iterrows():
        for side, candidate_column, effective_column in (
            (SIDE_TREATED, "candidate_target_count", "target_count"),
            (SIDE_DONOR, "candidate_donor_count", "donor_count"),
        ):
            scoped = dropouts.xs(
                (policy, outcome, axis, side),
                level=("policy", "outcome", "matching_axis", "roster_side"),
            )
            if int(scoped["candidate"].sum()) != int(row[candidate_column]) or int(
                scoped["effective"].sum()
            ) != int(row[effective_column]):
                raise IngestError(
                    f"{policy}/{outcome}/{axis}/{side}: report counts substitute "
                    "a candidate pool for the effective roster."
                )
    # Reconcile the frequency summary to exactly M saved selections per target.
    for pair in selected:
        key = (pair[0], pair[1], AXIS_MAHALANOBIS)
        scoped_weights = weight_distribution.xs(
            key, level=("policy", "outcome", "matching_axis")
        )
        expected_total = int(summary.loc[key, "target_count"]) * int(size_M)
        if int(scoped_weights["selection_count"].sum()) != expected_total or set(
            scoped_weights.index[scoped_weights["selected"]]
        ) != set(diagnostics.designs[key].matched_donors):
            raise IngestError(
                f"{pair}: weight distribution does not reconcile to M times targets."
            )
    # Compare the maxima with the declared values when supplied.
    assert_declared_balance(maxima, expected=expected_balance, tolerance=tolerance)
    # Freeze the report.
    fingerprint: str = freeze_matching_report(
        summary, dropouts, weight_distribution, balance
    )
    # Accumulate the recorded notes.
    notes: List[str] = [
        "the median restriction is computed over the union of the target and "
        "donor rosters from each country's pre-treatment average, and the "
        "country-level disposition ledger states every candidate retained or "
        "removed and the pooled-income reason.",
        "the frequency design's weights sum to one; the full distribution is "
        "reported because the average frequency alone would hide a design whose "
        "weights collapse onto few donors, and a donor no target selected is "
        "recorded as unselected.",
        "the balance audit is descriptive: a reduced maximum does not establish "
        "balance; each scheme-labelled nine-covariate vector is independently "
        "reproduced, and the baseline uses untouched rosters.",
    ]
    # Record the coverage of the artefacts.
    notes.append(
        f"the report covers {len(summary)} design(s), {len(dropouts)} candidate "
        f"country disposition row(s), {len(weight_distribution)} donor weight "
        f"row(s), {len(balance)} balance component row(s), and {len(maxima)} "
        "design/scheme maximum row(s)."
    )
    # Assemble the immutable result.
    result: MatchingReportResult = MatchingReportResult(
        summary=summary,
        dropouts=dropouts,
        weight_distribution=weight_distribution,
        balance=balance,
        maxima=maxima,
        fingerprint_sha256=fingerprint,
        completed_utc=current_utc_iso(),
        notes=tuple(notes),
    )
    # Log the completed execution.
    logger.info(
        "matching report complete: %d design(s), fingerprint %s.",
        len(summary),
        fingerprint[:16],
    )
    # Return the assembled result.
    return result
````

````python
# Task 25: Comtrade Cell Classification and Cell-Status Ledger

# ==============================================================================
# Task 25: Comtrade Cell Classification and Cell-Status Ledger
# ==============================================================================

# Declare the status of an observed positive record.
STATUS_POSITIVE: Final[str] = "positive"
# Declare the status of an observed zero carrying both confirmations.
STATUS_VERIFIED_ZERO: Final[str] = "verified_zero"
# Declare the status of a cell without an observed value.
STATUS_MISSING: Final[str] = "missing"
# Declare the disposition of a cell carrying an observed value or an observed
# absence.
DISPOSITION_USABLE: Final[str] = "usable"
# Declare the disposition of a missing cell the imputation layer may complete.
DISPOSITION_IMPUTABLE: Final[str] = "imputable"
# Declare the disposition of a missing canonical Russian-gas cell pending (or
# already carrying) evidence from the registered exporter-mirror channel; it is
# distinct from within-window completion eligibility.
DISPOSITION_RECOVERABLE: Final[str] = "recoverable"
# Declare the disposition of a missing cell excluded by the priority list.
DISPOSITION_CASE_EXCLUSION: Final[str] = "case_exclusion"
# Declare the disposition of a missing cell with no registered recovery or
# evidence-supported completion path.
DISPOSITION_REJECTED: Final[str] = "rejected"
# Compose the declared disposition vocabulary.
DISPOSITION_VOCABULARY: Final[Tuple[str, ...]] = (
    DISPOSITION_USABLE,
    DISPOSITION_RECOVERABLE,
    DISPOSITION_IMPUTABLE,
    DISPOSITION_CASE_EXCLUSION,
    DISPOSITION_REJECTED,
)
# Declare the primary key columns of the ledger.
LEDGER_KEY: Final[Tuple[str, ...]] = ("reporter_code", "year", "series")
# Declare the value column of the raw block.
VALUE_COLUMN: Final[str] = "value_usd"
# Declare the confirmation flag of the reporting side.
REPORT_FLAG_COLUMN: Final[str] = "hs_report_confirmed"
# Declare the confirmation flag of the individual retry.
RETRY_FLAG_COLUMN: Final[str] = "retry_succeeded"
# Declare the query-type column of the retry ledger.
QUERY_TYPE_COLUMN: Final[str] = "query_type"
# Declare the multi-country query type.
QUERY_TYPE_MULTI: Final[str] = "multi_country"
# Declare the individual query type.
QUERY_TYPE_INDIVIDUAL: Final[str] = "individual"
# Declare the canonical Russian-gas primitive of the priority exclusion.
PIPELINE_GAS_SERIES: Final[str] = "russian_gas"
# Register accepted source aliases once; every Task 25/26 handoff uses the
# canonical primitive namespace on the right-hand side.
TRADE_SERIES_ALIASES: Final[Mapping[str, str]] = MappingProxyType(
    {
        "russian_gas": "russian_gas",
        "russian gas": "russian_gas",
        "russian-gas": "russian_gas",
        "pipeline_gas": "russian_gas",
        "pipeline gas": "russian_gas",
        "world_gas": "world_gas",
        "world gas": "world_gas",
        "russian_non_gas": "russian_non_gas",
        "russian non gas": "russian_non_gas",
    }
)
# Declare the two independent exposure windows in which the coverage exclusion
# applies; no exclusion is inferred across their hard gap.
PRIORITY_EXCLUSION_WINDOWS: Final[Tuple[Tuple[str, int, int], ...]] = (
    ("w2010_2013", 2010, 2013),
    ("w2018_2021", 2018, 2021),
)
# Declare the immutable scientific reason of every priority exclusion.
PRIORITY_EXCLUSION_REASON: Final[str] = "pipeline_coverage_unrecoverable"
# Compose the countries of the priority exclusion, whose pipeline-gas imports
# cannot be recovered reliably.
PIPELINE_GAS_EXCLUSION_COUNTRIES: Final[Tuple[str, ...]] = (
    "AUT",
    "DEU",
    "POL",
    "TUR",
)


# Resolve a raw trade-series label to the single canonical primitive namespace.
def _canonical_trade_series(value: Any) -> str:
    """Return the canonical Task 25 primitive identifier for a source label."""
    if value is None or (isinstance(value, float) and np.isnan(value)):
        raise IngestError("A trade-series identifier cannot be null.")
    normalized = " ".join(str(value).strip().lower().replace("-", " ").replace("_", " ").split())
    lookup = normalized.replace(" ", "_")
    if lookup in TRADE_SERIES_ALIASES:
        return str(TRADE_SERIES_ALIASES[lookup])
    if normalized in TRADE_SERIES_ALIASES:
        return str(TRADE_SERIES_ALIASES[normalized])
    raise IngestError(f"Undeclared trade-series identifier {value!r}.")


# Resolve a year to its declared, non-overlapping exposure window.
def _priority_window_for_year(year: Any) -> Optional[Tuple[str, int, int]]:
    """Return the declared exposure window containing an integral calendar year."""
    if isinstance(year, (bool, np.bool_)):
        raise IngestError("A trade year cannot be Boolean.")
    try:
        numeric = float(year)
    except (TypeError, ValueError) as exc:
        raise IngestError(f"Invalid trade year {year!r}.") from exc
    if not np.isfinite(numeric) or not numeric.is_integer():
        raise IngestError(f"Invalid trade year {year!r}.")
    calendar = int(numeric)
    for window_id, start, end in PRIORITY_EXCLUSION_WINDOWS:
        if int(start) <= calendar <= int(end):
            return str(window_id), int(start), int(end)
    return None


# Declare the frozen ledger of the classification.
@dataclass(frozen=True)
class CellLedgerResult:
    """
    Frozen cell-status ledger of the commodity cells.

    Attributes
    ----------
    ledger : pandas.DataFrame
        One row per cell carrying the primary key, the untouched value, the two
        confirmation flags, the derived status and the derived disposition.
    counts : pandas.DataFrame
        One row per series carrying the number of cells of each status and each
        disposition.
    reclassifications : pandas.DataFrame
        One row per cell recovered by an individual retry, carrying the original
        and the final query type and the status before and after.
    exclusions : pandas.DataFrame
        One row per declared priority exclusion, uniquely keyed by country,
        canonical series and explicit exposure window.
    fingerprint_sha256 : str
        Digest of the canonical ledger payload, computed over the sorted key.
    built_utc : str
        ISO-8601 UTC timestamp of the ledger.
    notes : tuple of str
        Recorded decisions, including the applied exclusion set and the retry
        state.
    """
    # The classified and disposed ledger.
    ledger: pd.DataFrame
    # The per-series counts.
    counts: pd.DataFrame
    # The reclassification log.
    reclassifications: pd.DataFrame
    # The declared priority exclusions.
    exclusions: pd.DataFrame
    # Digest of the canonical ledger payload.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the ledger.
    built_utc: str
    # Recorded decisions of the ledger.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 25, Step 1: The verified-zero classification
# -------------------------------------------------------------------------------------------------------------------------------


# Classify one cell under the declared rule.
def classify_cell(
    value_usd: float,
    hs_report_confirmed: int,
    retry_succeeded: int,
) -> str:
    """
    Classify one cell under the declared rule.

    Parameters
    ----------
    value_usd : float
        The reported value, which is never modified by the classification.
    hs_report_confirmed : int
        The confirmation flag of the reporting side.
    retry_succeeded : int
        The confirmation flag of the individual retry.

    Returns
    -------
    str
        The cell's status. A strictly positive value is a positive record; an
        exactly zero value is a verified zero only when **both** confirming
        flags are set; and a missing value or zero with incomplete evidence is
        missing. Negative values are rejected as malformed rather than silently
        converted to missing. The
        classification is deterministic: it consumes only the declared inputs
        and no exogenous quantity.

    Raises
    ------
    IngestError
        If the value is negative/malformed/non-finite or either evidence flag
        is not binary.
    """
    # Validate both evidence flags even when the value is absent so malformed
    # provenance cannot hide behind a missing observation.
    flags: List[int] = []
    for label, raw_flag in (
        (REPORT_FLAG_COLUMN, hs_report_confirmed),
        (RETRY_FLAG_COLUMN, retry_succeeded),
    ):
        if isinstance(raw_flag, (bool, np.bool_)):
            flags.append(int(raw_flag))
            continue
        try:
            numeric_flag = float(raw_flag)
        except (TypeError, ValueError) as exc:
            raise IngestError(f"{label} must be a binary flag.") from exc
        if not np.isfinite(numeric_flag) or numeric_flag not in (0.0, 1.0):
            raise IngestError(f"{label} must be a binary flag.")
        flags.append(int(numeric_flag))
    # Treat a missing value as missing regardless of the valid flags, because a
    # confirmation cannot manufacture an observed value.
    if value_usd is None or bool(pd.isna(value_usd)):
        # Return the missing status.
        return STATUS_MISSING
    # Convert the value to a float for the comparisons.
    try:
        # Compose the numeric value.
        numeric: float = float(value_usd)
    # Refuse an unconvertible value.
    except (TypeError, ValueError) as error:
        # Re-raise as an ingest failure naming the input.
        raise IngestError(f"Cannot classify the value {value_usd!r}.") from error
    if isinstance(value_usd, (bool, np.bool_)):
        raise IngestError("A trade value cannot be Boolean.")
    # Refuse a non-finite value, which the rule does not admit.
    if not np.isfinite(numeric):
        # The failure names the value.
        raise IngestError(f"Cannot classify the non-finite value {value_usd!r}.")
    if numeric < 0.0:
        raise IngestError(f"Cannot classify the negative trade value {value_usd!r}.")
    # Classify a strictly positive value as a positive record.
    if numeric > 0.0:
        # Return the positive status.
        return STATUS_POSITIVE
    # Classify an exactly zero value as a verified zero only when both
    # confirming flags are set, because a zero without confirmation is a
    # non-record rather than an observed absence.
    if numeric == 0.0 and flags[0] == 1 and flags[1] == 1:
        # Return the verified-zero status.
        return STATUS_VERIFIED_ZERO
    # Classify the only remaining valid case, a zero whose flags are incomplete,
    # as missing.
    return STATUS_MISSING


# Classify every cell of the raw block.
def classify_cells(raw: pd.DataFrame) -> pd.DataFrame:
    """
    Classify every cell of the raw block.

    Parameters
    ----------
    raw : pandas.DataFrame
        The raw cell-status block carrying the primary key, the value and the
        two confirmation flags.

    Returns
    -------
    pandas.DataFrame
        The classified ledger, which preserves the primary key, carries the
        strictly parsed value plus its untouched source representation, and
        adds the derived status and classification evidence.

    Raises
    ------
    IngestError
        If a required column is absent, if the value column is non-numeric, or
        if the vectorized classification disagrees with the declared scalar
        rule on any row, because the two must implement one rule.
    """
    # Refuse a block lacking a required column.
    for column in (
        list(LEDGER_KEY) + [VALUE_COLUMN, REPORT_FLAG_COLUMN, RETRY_FLAG_COLUMN]
    ):
        # Test the column's presence.
        if column not in raw.columns:
            # The failure names the absent column.
            raise IngestError(f"The raw block lacks the required column {column!r}.")
    # Compose the ledger as a copy so the raw block is never modified.
    ledger: pd.DataFrame = raw.copy()
    # Canonicalize reporter identities before key construction while retaining
    # the source label as evidence of the reversible normalization.
    ledger["reporter_code_original"] = ledger["reporter_code"].copy()
    if bool(ledger["reporter_code"].isna().any()):
        raise IngestError("Reporter codes cannot be null.")
    ledger["reporter_code"] = ledger["reporter_code"].astype(str).str.strip().str.upper()
    if bool((ledger["reporter_code"].str.len() == 0).any()):
        raise IngestError("Reporter codes cannot be blank.")
    # Canonicalize the primitive exactly once while retaining its source label.
    ledger["series_original"] = ledger["series"].copy()
    ledger["series"] = ledger["series"].map(_canonical_trade_series)
    # Validate integral years before composing canonical cell keys.
    windows = [_priority_window_for_year(value) for value in ledger["year"]]
    numeric_years = pd.to_numeric(ledger["year"], errors="coerce")
    ledger["year"] = numeric_years.astype(int)
    if bool(ledger.duplicated(list(LEDGER_KEY)).any()):
        duplicated = ledger.loc[ledger.duplicated(list(LEDGER_KEY), keep=False), list(LEDGER_KEY)]
        raise IngestError(
            "Canonical series aliases create duplicate cell key(s) "
            f"{duplicated.head(8).to_dict('records')}."
        )
    # Read the value column strictly, distinguishing true missing markers from
    # malformed strings and retaining the raw source value separately.
    ledger["original_value_usd"] = ledger[VALUE_COLUMN].copy()
    boolean_values = ledger[VALUE_COLUMN].map(
        lambda value: isinstance(value, (bool, np.bool_))
    )
    if bool(boolean_values.any()):
        raise IngestError("Trade values cannot be Boolean.")
    numeric_values = pd.to_numeric(ledger[VALUE_COLUMN], errors="coerce")
    malformed_values = numeric_values.isna() & ~ledger[VALUE_COLUMN].isna()
    if bool(malformed_values.any()):
        raise IngestError(
            f"Malformed trade value(s) at row(s) {list(np.flatnonzero(malformed_values))[:8]}."
        )
    values = numeric_values.to_numpy(dtype=float)
    if bool((np.isfinite(values) & (values < 0.0)).any()) or bool(np.isinf(values).any()):
        raise IngestError("Trade values must be missing or finite and nonnegative.")
    ledger[VALUE_COLUMN] = numeric_values.astype(float)
    # Validate flags through the scalar rule's exact binary policy.
    report_flags: List[int] = []
    retry_flags: List[int] = []
    for report_flag, retry_flag in zip(
        ledger[REPORT_FLAG_COLUMN], ledger[RETRY_FLAG_COLUMN]
    ):
        classify_cell(np.nan, report_flag, retry_flag)
        report_flags.append(
            int(bool(report_flag))
            if isinstance(report_flag, (bool, np.bool_))
            else int(float(report_flag))
        )
        retry_flags.append(
            int(bool(retry_flag))
            if isinstance(retry_flag, (bool, np.bool_))
            else int(float(retry_flag))
        )
    report_array = np.asarray(report_flags, dtype=float)
    retry_array = np.asarray(retry_flags, dtype=float)
    ledger[REPORT_FLAG_COLUMN] = np.asarray(report_flags, dtype=int)
    ledger[RETRY_FLAG_COLUMN] = np.asarray(retry_flags, dtype=int)
    # Compose the positive mask, which requires a strictly positive value.
    positive_mask: np.ndarray = np.isfinite(values) & (values > 0.0)
    # Compose the verified-zero mask, which requires an exact zero and both
    # confirming flags, which is the declared both-flags requirement.
    zero_mask: np.ndarray = (
        np.isfinite(values) & (values == 0.0) & (report_array == 1.0) & (retry_array == 1.0)
    )
    # Compose the status column by elementwise selection, defaulting to the
    # missing status for every case the two masks do not claim.
    ledger["final_status"] = np.where(
        positive_mask, STATUS_POSITIVE, np.where(zero_mask, STATUS_VERIFIED_ZERO, STATUS_MISSING)
    )
    # Assert the vectorized classification agrees with the declared scalar rule
    # on every row, so the fast path cannot drift from the rule.
    for position, (_, row) in enumerate(ledger.iterrows()):
        # Compose the scalar classification of the row.
        scalar: str = classify_cell(
            float(values[position]),
            int(report_array[position]),
            int(retry_array[position]),
        )
        # Refuse a divergence between the two implementations of the rule.
        if scalar != str(ledger["final_status"].iloc[position]):
            # The failure names the position and both statuses.
            raise IngestError(
                f"Row {position}: the vectorized classification "
                f"{ledger['final_status'].iloc[position]!r} disagrees with the "
                f"declared rule {scalar!r}."
            )
    # Preserve the mutually exclusive pre-completion classification and the
    # evidence that distinguishes an observed zero from an unconfirmed empty cell.
    ledger["precompletion_status"] = ledger["final_status"].astype(str)
    ledger["original_anchor"] = ledger["final_status"].isin(
        [STATUS_POSITIVE, STATUS_VERIFIED_ZERO]
    )
    ledger["zero_verification_evidence"] = (
        (ledger[VALUE_COLUMN] == 0.0)
        & (ledger[REPORT_FLAG_COLUMN] == 1)
        & (ledger[RETRY_FLAG_COLUMN] == 1)
    )
    ledger["window_id"] = [item[0] if item is not None else None for item in windows]
    ledger["window_start"] = [item[1] if item is not None else pd.NA for item in windows]
    ledger["window_end"] = [item[2] if item is not None else pd.NA for item in windows]
    # Log the status distribution.
    logger.info(
        "cells classified across %d row(s): %s.",
        len(ledger),
        ledger["final_status"].value_counts().to_dict(),
    )
    # Return the classified ledger.
    return ledger


# -------------------------------------------------------------------------------------------------------------------------------
# Task 25, Step 2: The downstream dispositions
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the declared priority exclusions.
def build_priority_exclusions(
    countries: Sequence[str] = PIPELINE_GAS_EXCLUSION_COUNTRIES,
    series: str = PIPELINE_GAS_SERIES,
) -> pd.DataFrame:
    """
    Compose the declared priority exclusions.

    Parameters
    ----------
    countries : sequence of str, optional
        The countries whose pipeline-gas imports cannot be recovered reliably.
    series : str, optional
        The series of the priority exclusion.

    Returns
    -------
    pandas.DataFrame
        One uniquely keyed row per country, canonical series and applicable
        exposure window, carrying its inclusive bounds and documented
        pipeline-coverage reason.

    Raises
    ------
    IngestError
        If the country list or the series is empty, because an empty exclusion
        would silently disable the rule.
    """
    # Refuse an empty country list.
    if not countries:
        # The failure states the precondition.
        raise IngestError("The priority exclusion requires at least one country.")
    country_codes = tuple(str(country).strip().upper() for country in countries)
    if any(not code for code in country_codes) or len(set(country_codes)) != len(country_codes):
        raise IngestError("Priority-exclusion countries must be unique nonblank codes.")
    canonical_series = _canonical_trade_series(series)
    if canonical_series != "russian_gas":
        raise IngestError(
            "Priority pipeline coverage exclusions apply only to canonical russian_gas."
        )
    # Assemble the country/series/window rows explicitly so no downstream stage
    # can infer scope from a mismatched product alias.
    rows = [
        {
            "country_code": country,
            "series": canonical_series,
            "window_id": window_id,
            "window_start": int(start),
            "window_end": int(end),
            "reason": PRIORITY_EXCLUSION_REASON,
            "immutable": True,
        }
        for country in country_codes
        for window_id, start, end in PRIORITY_EXCLUSION_WINDOWS
    ]
    exclusions = pd.DataFrame(rows).set_index(
        ["country_code", "series", "window_id"]
    )
    if not bool(exclusions.index.is_unique):
        raise IngestError("Priority exclusions carry duplicate country/series/window keys.")
    # Log the exclusion set.
    logger.info(
        "priority exclusions declared across %d country-series pair(s) for the "
        "series %s.",
        len(exclusions),
        canonical_series,
    )
    # Return the frame.
    return exclusions


# Assign the downstream disposition of every cell.
def resolve_dispositions(
    ledger: pd.DataFrame,
    exclusions: pd.DataFrame,
) -> pd.DataFrame:
    """
    Assign the downstream disposition of every cell.

    Parameters
    ----------
    ledger : pandas.DataFrame
        The classified ledger.
    exclusions : pandas.DataFrame
        The priority exclusions keyed on country, canonical series and explicit
        exposure window.

    Returns
    -------
    pandas.DataFrame
        The ledger with retained/recoverable/imputable/case-exclusion/rejected
        dispositions, explicit reasons, original-anchor evidence, immutable
        exclusion keys and within-window completion eligibility.

    Raises
    ------
    IngestError
        If keys/statuses/exclusions are malformed or a completion/recovery state
        is not supported by actual history and immutable evidence.
    """
    # Refuse a ledger without its classification and evidence columns.
    required_ledger = set(LEDGER_KEY) | {
        "final_status", "precompletion_status", "original_anchor",
        VALUE_COLUMN, REPORT_FLAG_COLUMN, RETRY_FLAG_COLUMN,
        "window_id", "window_start", "window_end",
    }
    if not required_ledger.issubset(ledger.columns):
        raise IngestError(
            "The ledger lacks disposition input(s) "
            f"{sorted(required_ledger - set(ledger.columns))}."
        )
    if bool(ledger.duplicated(list(LEDGER_KEY)).any()):
        raise IngestError("The classified ledger carries duplicate canonical cell keys.")
    if list(exclusions.index.names) != ["country_code", "series", "window_id"]:
        raise IngestError(
            "Priority exclusions must use the country/series/window key."
        )
    required_exclusion_columns = {"window_start", "window_end", "reason", "immutable"}
    if not required_exclusion_columns.issubset(exclusions.columns) or not bool(
        exclusions.index.is_unique
    ):
        raise IngestError("Priority exclusions are incomplete or non-unique.")
    # Refuse a status outside the declared vocabulary.
    observed_statuses = set(str(value) for value in ledger["final_status"].unique())
    # Identify the undeclared statuses.
    undeclared: List[str] = sorted(observed_statuses - set(STATUS_VOCABULARY))
    # Refuse an undeclared status.
    if undeclared:
        # The failure names the offending statuses.
        raise IngestError(f"The ledger carries undeclared status(es) {undeclared}.")
    # Compose the ledger as a copy so the classified frame is never modified.
    disposed: pd.DataFrame = ledger.copy()
    disposed["_disposition_position"] = np.arange(len(disposed), dtype=int)
    disposed["series"] = disposed["series"].map(_canonical_trade_series)
    if bool(disposed.duplicated(list(LEDGER_KEY)).any()):
        raise IngestError("Canonical series aliases create duplicate disposition keys.")
    if not disposed["final_status"].astype(str).equals(
        disposed["precompletion_status"].astype(str)
    ):
        raise IngestError("Pre-completion statuses changed before disposition resolution.")
    # Recompute the classification and window evidence at this boundary so a
    # caller cannot forge eligibility fields while retaining plausible labels.
    for position, row in disposed.iterrows():
        expected_status = classify_cell(
            row[VALUE_COLUMN], row[REPORT_FLAG_COLUMN], row[RETRY_FLAG_COLUMN]
        )
        if str(row["final_status"]) != expected_status:
            raise IngestError(
                f"Disposition input row {position} disagrees with the shared "
                "cell-classification rule."
            )
        expected_anchor = expected_status in (STATUS_POSITIVE, STATUS_VERIFIED_ZERO)
        if not isinstance(row["original_anchor"], (bool, np.bool_)) or bool(
            row["original_anchor"]
        ) != expected_anchor:
            raise IngestError(
                f"Disposition input row {position} carries forged original-anchor evidence."
            )
        expected_window = _priority_window_for_year(row["year"])
        try:
            observed_window = None if pd.isna(row["window_id"]) else (
                str(row["window_id"]),
                int(row["window_start"]),
                int(row["window_end"]),
            )
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError(
                f"Disposition input row {position} carries malformed window evidence."
            ) from exc
        if observed_window != expected_window:
            raise IngestError(
                f"Disposition input row {position} carries inconsistent window evidence."
            )
    # Compose the observed-value mask, which is the usable case.
    usable_mask: np.ndarray = disposed["final_status"].isin(
        [STATUS_POSITIVE, STATUS_VERIFIED_ZERO]
    ).to_numpy()
    # Compose the missing mask.
    missing_mask: np.ndarray = (disposed["final_status"] == STATUS_MISSING).to_numpy()
    # Validate and freeze the canonical exclusion evidence.
    exclusion_records: Dict[Tuple[str, str, str], Mapping[str, Any]] = {}
    for (country, series, window_id), row in exclusions.iterrows():
        canonical_series = _canonical_trade_series(series)
        immutable_value = row["immutable"]
        if (
            canonical_series != "russian_gas"
            or str(row["reason"]) != PRIORITY_EXCLUSION_REASON
            or not isinstance(immutable_value, (bool, np.bool_))
            or not bool(immutable_value)
        ):
            raise IngestError("Priority exclusion identity/reason/immutability is malformed.")
        expected_window = next(
            (item for item in PRIORITY_EXCLUSION_WINDOWS if item[0] == str(window_id)),
            None,
        )
        if expected_window is None or (int(row["window_start"]), int(row["window_end"])) != (
            int(expected_window[1]), int(expected_window[2])
        ):
            raise IngestError("Priority exclusion window metadata is inconsistent.")
        exclusion_records[(str(country), canonical_series, str(window_id))] = row
    exclusion_mask: np.ndarray = np.array(
        [
            missing_mask[position]
            and (
                str(disposed["reporter_code"].iloc[position]),
                str(disposed["series"].iloc[position]),
                str(disposed["window_id"].iloc[position]),
            ) in exclusion_records
            for position in range(len(disposed))
        ],
        dtype=bool,
    )
    # Resolve explicit recovery evidence.  The exporter-mirror channel is a
    # registered recovery route for canonical Russian-gas cells, so an otherwise
    # valid missing cell must remain pending until that independently supplied
    # mirror is inspected by the recovery stage.  Retry success alone is never
    # evidence and no other primitive is admitted to the mirror route.
    explicit_recovery_evidence = np.zeros(len(disposed), dtype=bool)
    if "recovery_evidence" in disposed.columns:
        raw_evidence = disposed["recovery_evidence"]
        valid = raw_evidence.isin([True, False, 0, 1])
        if not bool(valid.all()):
            raise IngestError("Recovery evidence must be binary.")
        explicit_recovery_evidence |= raw_evidence.astype(bool).to_numpy()
    if "mirror_value_usd" in disposed.columns:
        mirror_raw = disposed["mirror_value_usd"]
        mirror_values = pd.to_numeric(mirror_raw, errors="coerce").to_numpy(dtype=float)
        malformed_mirror = mirror_raw.notna().to_numpy() & ~np.isfinite(mirror_values)
        if bool(malformed_mirror.any()) or bool(
            (mirror_values[np.isfinite(mirror_values)] < 0.0).any()
        ):
            raise IngestError(
                "Mirror recovery values must be finite and nonnegative when supplied."
            )
        explicit_recovery_evidence |= np.isfinite(mirror_values) & (mirror_values > 0.0)
    russian_gas_mask = (
        disposed["series"].astype(str).to_numpy() == "russian_gas"
    )
    if bool((explicit_recovery_evidence & ~russian_gas_mask).any()):
        raise IngestError(
            "Mirror recovery evidence is registered only for canonical russian_gas cells."
        )
    explicit_recovery_evidence &= missing_mask & (~exclusion_mask)
    registered_mirror_path = missing_mask & russian_gas_mask & (~exclusion_mask)
    # Determine completion eligibility from actual original anchors inside each
    # window, using at most two-cell internal gaps or a one-year edge backed by
    # at least three observed anchors.
    completion_evidence = np.zeros(len(disposed), dtype=bool)
    for (_reporter, _series, window_id), group in disposed.groupby(
        ["reporter_code", "series", "window_id"], dropna=False, sort=False
    ):
        if pd.isna(window_id):
            continue
        offsets = group["_disposition_position"].to_numpy(dtype=int)
        years = group["year"].astype(int).to_numpy()
        anchor_mask = group["original_anchor"].astype(bool).to_numpy()
        anchor_years = np.sort(years[anchor_mask])
        available_years = set(int(year) for year in years)
        if anchor_years.size == 0:
            continue
        start = int(group["window_start"].dropna().iloc[0])
        end = int(group["window_end"].dropna().iloc[0])
        for local, (offset, year, status) in enumerate(
            zip(offsets, years, group["final_status"].astype(str))
        ):
            if status != STATUS_MISSING or exclusion_mask[offset]:
                continue
            left = anchor_years[anchor_years < int(year)]
            right = anchor_years[anchor_years > int(year)]
            internal = False
            if left.size and right.size:
                lower, upper = int(left[-1]), int(right[0])
                internal = (
                    upper - lower - 1 <= MAXIMUM_INTERNAL_GAP
                    and set(range(lower, upper + 1)).issubset(available_years)
                )
            one_year_edge = (
                anchor_years.size >= MINIMUM_ANCHORS
                and (
                    (int(year) == start and int(anchor_years[0]) == start + 1)
                    or (int(year) == end and int(anchor_years[-1]) == end - 1)
                )
            )
            completion_evidence[offset] = bool(internal or one_year_edge)
    # Assign the dispositions in the declared priority order: usable cells
    # first, immutable case exclusions next, then cells pending the registered
    # mirror path, evidence-supported imputations, and finally true rejections.
    disposed["disposition"] = np.where(
        usable_mask,
        DISPOSITION_USABLE,
        np.where(
            exclusion_mask,
            DISPOSITION_CASE_EXCLUSION,
            np.where(
                registered_mirror_path,
                DISPOSITION_RECOVERABLE,
                np.where(
                    missing_mask & completion_evidence,
                    DISPOSITION_IMPUTABLE,
                    DISPOSITION_REJECTED,
                ),
            ),
        ),
    )
    disposed["recovery_eligible"] = registered_mirror_path
    disposed["recovery_evidence_state"] = np.select(
        [
            exclusion_mask,
            explicit_recovery_evidence,
            registered_mirror_path,
        ],
        [
            "barred_priority_exclusion",
            "explicit_positive_mirror",
            "pending_registered_mirror",
        ],
        default="not_registered",
    )
    disposed["completion_eligible"] = completion_evidence
    disposed["immutable_exclusion"] = exclusion_mask
    disposed["disposition_reason"] = np.select(
        [
            usable_mask & (disposed["final_status"].to_numpy() == STATUS_POSITIVE),
            usable_mask & (disposed["final_status"].to_numpy() == STATUS_VERIFIED_ZERO),
            exclusion_mask,
            explicit_recovery_evidence,
            registered_mirror_path & ~explicit_recovery_evidence,
            missing_mask & completion_evidence,
        ],
        [
            "retained_positive_observation",
            "retained_verified_zero",
            PRIORITY_EXCLUSION_REASON,
            "explicit_mirror_recovery_evidence",
            "pending_registered_russian_exporter_mirror",
            "within_window_original_anchor_rule",
        ],
        default="insufficient_recovery_or_completion_evidence",
    )
    disposed["priority_exclusion_key"] = [
        (
            f"{disposed['reporter_code'].iloc[position]}|russian_gas|"
            f"{disposed['window_id'].iloc[position]}"
        ) if exclusion_mask[position] else ""
        for position in range(len(disposed))
    ]
    disposed = disposed.drop(columns=["_disposition_position"])
    # Assert the dispositions partition the rows exactly.
    counts = disposed["disposition"].value_counts()
    # Refuse a partition that does not cover every row.
    if int(counts.sum()) != len(disposed):
        # The failure states the condition.
        raise IngestError(
            "The dispositions do not partition the ledger's rows."
        )
    # Log the disposition distribution.
    logger.info(
        "dispositions assigned across %d row(s): %s.",
        len(disposed),
        counts.to_dict(),
    )
    # Return the disposed ledger.
    return disposed


# Assert that no rejected cell can be reused as an observation.
def assert_rejected_immutability(ledger: pd.DataFrame) -> None:
    """
    Assert that no rejected cell can be reused as an observation.

    Parameters
    ----------
    ledger : pandas.DataFrame
        The disposed ledger.

    Returns
    -------
    None
        The function returns nothing when the rule holds.

    Raises
    ------
    IngestError
        If a rejected cell carries a usable status, or if a rejected cell
        carries a strictly positive value, because a positive value would be a
        positive record under the classification and the two artefacts would
        then disagree. A zero or a missing value is admissible: an unconfirmed
        zero is exactly the case the rule keeps missing, so its raw zero is an
        input rather than a reclassification, and the specification fixes that
        such a cell is never later read as zero.
    """
    # Refuse a ledger without the immutable classification/disposition evidence.
    required = set(LEDGER_KEY) | {
        "disposition", "final_status", "precompletion_status", VALUE_COLUMN,
        "original_value_usd", REPORT_FLAG_COLUMN, RETRY_FLAG_COLUMN,
        "original_anchor", "completion_eligible", "recovery_eligible",
        "immutable_exclusion",
    }
    if not required.issubset(ledger.columns):
        # The failure names the absent column.
        raise IngestError("The immutability gate requires the disposition column.")
    if bool(ledger.duplicated(list(LEDGER_KEY)).any()):
        raise IngestError("The immutability gate received duplicate canonical cell keys.")
    barred = ledger[
        ledger["disposition"].isin(
            [DISPOSITION_REJECTED, DISPOSITION_CASE_EXCLUSION]
        )
    ]
    violations: List[Tuple[str, int, str, str]] = []
    downstream_value_columns = [
        column for column in (
            "recovered_value_usd",
            "completed_value_usd",
            "imputed_value_usd",
        ) if column in barred.columns
    ]
    downstream_flag_columns = [
        column for column in (
            "recovered", "recovery_realized", "imputed", "was_completed"
        )
        if column in barred.columns
    ]
    for _, row in barred.iterrows():
        key = (str(row["reporter_code"]), int(row["year"]), str(row["series"]))
        reason = ""
        original_status = classify_cell(
            row["original_value_usd"],
            row[REPORT_FLAG_COLUMN],
            row[RETRY_FLAG_COLUMN],
        )
        if _canonical_trade_series(row["series"]) != str(row["series"]):
            reason = "noncanonical_series"
        elif (
            original_status != STATUS_MISSING
            or not isinstance(row["original_anchor"], (bool, np.bool_))
            or bool(row["original_anchor"])
        ):
            reason = "original_classification_changed"
        elif str(row["final_status"]) != STATUS_MISSING or str(
            row["precompletion_status"]
        ) != STATUS_MISSING:
            reason = "status_changed"
        elif bool(row["completion_eligible"]) or bool(row["recovery_eligible"]):
            reason = "barred_cell_marked_eligible"
        elif pd.notna(
            pd.to_numeric(pd.Series([row[VALUE_COLUMN]]), errors="coerce").iloc[0]
        ) and float(
            pd.to_numeric(pd.Series([row[VALUE_COLUMN]]), errors="coerce").iloc[0]
        ) > 0.0:
            reason = "positive_raw_value"
        elif str(row["disposition"]) == DISPOSITION_CASE_EXCLUSION and (
            not isinstance(row["immutable_exclusion"], (bool, np.bool_))
            or not bool(row["immutable_exclusion"])
        ):
            reason = "case_exclusion_not_immutable"
        else:
            expected_window = _priority_window_for_year(row["year"])
            in_priority_scope = (
                str(row["reporter_code"]) in PIPELINE_GAS_EXCLUSION_COUNTRIES
                and str(row["series"]) == "russian_gas"
                and expected_window is not None
            )
            if in_priority_scope and str(row["disposition"]) != DISPOSITION_CASE_EXCLUSION:
                reason = "priority_cell_lost_case_exclusion"
            elif str(row["disposition"]) == DISPOSITION_CASE_EXCLUSION and not in_priority_scope:
                reason = "case_exclusion_outside_declared_scope"
        if not reason:
            for column in downstream_value_columns:
                numeric = pd.to_numeric(pd.Series([row[column]]), errors="coerce").iloc[0]
                if pd.notna(numeric):
                    reason = f"downstream_value_in_{column}"
                    break
            if not reason and any(bool(row[column]) for column in downstream_flag_columns):
                reason = "downstream_recovery_or_imputation_flag"
        if reason:
            violations.append((*key, reason))
    if violations:
        raise IngestError(
            "Rejected/case-excluded cell immutability violation(s): "
            f"{violations[:8]}."
        )
    # Verify every present priority case retains the exact canonical window key
    # and documented reason through the handoff.
    priority = barred[
        barred["reporter_code"].astype(str).isin(PIPELINE_GAS_EXCLUSION_COUNTRIES)
        & (barred["series"].astype(str) == "russian_gas")
        & (barred["disposition"] == DISPOSITION_CASE_EXCLUSION)
    ]
    for _, row in priority.iterrows():
        expected_window = _priority_window_for_year(row["year"])
        if expected_window is None:
            raise IngestError(
                f"Priority exclusion lies outside a declared window at "
                f"{row['reporter_code']}/{row['year']}."
            )
        window_id, window_start, window_end = expected_window
        expected = f"{row['reporter_code']}|russian_gas|{window_id}"
        try:
            observed_window = (
                str(row.get("window_id", "")),
                int(row.get("window_start")),
                int(row.get("window_end")),
            )
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError(
                f"Priority exclusion window evidence is malformed for {expected}."
            ) from exc
        if str(row.get("priority_exclusion_key", "")) != expected or str(
            row.get("disposition_reason", "")
        ) != PRIORITY_EXCLUSION_REASON or observed_window != (
            window_id,
            window_start,
            window_end,
        ):
            raise IngestError(f"Priority exclusion provenance drifted for {expected}.")
    # Log the verified immutability.
    logger.info(
        "rejected-cell immutability verified across %d rejected cell(s).",
        len(barred),
    )


# -------------------------------------------------------------------------------------------------------------------------------
# Task 25, Step 3: The reclassification log and the freeze
# -------------------------------------------------------------------------------------------------------------------------------


# Log every cell recovered from a multi-country to an individual request.
def build_reclassification_log(
    ledger: pd.DataFrame,
    retry_records: Optional[pd.DataFrame],
) -> pd.DataFrame:
    """
    Log every cell recovered from a multi-country to an individual request.

    Parameters
    ----------
    ledger : pandas.DataFrame
        The disposed ledger.
    retry_records : pandas.DataFrame, optional
        The retry ledger carrying the primary key, the original query type, the
        final query type and the status before and after each retry.

    Returns
    -------
    pandas.DataFrame
        One row per recovered cell carrying the original and the final query
        type together with the statuses, so every reclassification is auditable
        from the cell's own record. When no retry records are supplied the
        function returns an empty frame with the declared columns and records
        that state.

    Raises
    ------
    IngestError
        If a retry record lacks a declared column, or if a logged cell is
        absent from the ledger.
    """
    # Declare the columns of the log.
    columns: List[str] = [
        *LEDGER_KEY,
        "original_query_type",
        "final_query_type",
        "status_before",
        "status_after",
    ]
    # Return an empty log when no retry records were supplied, recording the
    # state rather than implying a recovery.
    if retry_records is None:
        # Log the absent retry records.
        logger.info(
            "no retry records were supplied, so no reclassification is logged."
        )
        # Return the empty log.
        return pd.DataFrame(columns=columns).set_index(list(LEDGER_KEY))
    # Refuse a retry frame lacking a declared column.
    for column in columns:
        # Test the column's presence.
        if column not in retry_records.columns:
            # The failure names the absent column.
            raise IngestError(f"The retry records lack the declared column {column!r}.")
    # Copy the retry frame so the caller's object is never modified.
    log: pd.DataFrame = retry_records.copy()
    log["series"] = log["series"].map(_canonical_trade_series)
    numeric_years = pd.to_numeric(log["year"], errors="coerce")
    if bool(numeric_years.isna().any()) or bool((numeric_years % 1 != 0).any()):
        raise IngestError("Retry records carry a malformed calendar year.")
    log["year"] = numeric_years.astype(int)
    if bool(log.duplicated(list(LEDGER_KEY)).any()):
        raise IngestError("Retry records carry duplicate canonical cell keys.")
    query_vocabulary = {QUERY_TYPE_MULTI, QUERY_TYPE_INDIVIDUAL}
    for column in ("original_query_type", "final_query_type"):
        undeclared_query = set(log[column].astype(str)) - query_vocabulary
        if undeclared_query:
            raise IngestError(
                f"Retry records carry undeclared {column} value(s) "
                f"{sorted(undeclared_query)}."
            )
    for column in ("status_before", "status_after"):
        undeclared_status = set(log[column].astype(str)) - set(STATUS_VOCABULARY)
        if undeclared_status:
            raise IngestError(
                f"Retry records carry undeclared {column} value(s) "
                f"{sorted(undeclared_status)}."
            )
    # Retain only records whose query type moved from the multi-country request
    # to the individual one *and* whose status actually changed. An unsuccessful
    # retry is provenance, but is not a scientific reclassification.
    migrated_query = (
        (log["original_query_type"].astype(str) == QUERY_TYPE_MULTI)
        & (log["final_query_type"].astype(str) == QUERY_TYPE_INDIVIDUAL)
    )
    status_changed = (
        log["status_before"].astype(str) != log["status_after"].astype(str)
    )
    if bool((status_changed & ~migrated_query).any()):
        raise IngestError(
            "Every retry status change must be a multi_country to individual "
            "reclassification."
        )
    migrated = log[
        migrated_query
        & status_changed
    ].copy()
    if len(migrated):
        valid_transition = (
            (migrated["status_before"].astype(str) == STATUS_MISSING)
            & migrated["status_after"].astype(str).isin(
                [STATUS_POSITIVE, STATUS_VERIFIED_ZERO]
            )
        )
        if not bool(valid_transition.all()):
            raise IngestError(
                "Retry reclassifications must move missing to positive or verified_zero."
            )
    # Assert every logged cell appears in the ledger.
    ledger_keys = set(
        (str(row[LEDGER_KEY[0]]), int(row[LEDGER_KEY[1]]), str(row[LEDGER_KEY[2]]))
        for _, row in ledger.iterrows()
    )
    # Identify the logged cells absent from the ledger.
    absent: List[Any] = []
    # Iterate over the migrated records.
    for _, row in migrated.iterrows():
        # Compose the record's key.
        key = (
            str(row[LEDGER_KEY[0]]),
            int(row[LEDGER_KEY[1]]),
            str(row[LEDGER_KEY[2]]),
        )
        # Record an absent key.
        if key not in ledger_keys:
            # Append the absent key.
            absent.append(key)
    # Refuse a logged cell absent from the ledger.
    if absent:
        # The failure names the first absent keys.
        raise IngestError(
            f"The reclassification log references cell(s) absent from the ledger: "
            f"{absent[:5]}."
        )
    # Compose the log indexed on the primary key.
    result: pd.DataFrame = migrated.set_index(list(LEDGER_KEY))
    # Reconcile every logged after-status to the immutable classification.
    ledger_status = ledger.set_index(list(LEDGER_KEY))["final_status"].astype(str)
    if len(result):
        observed_after = result["status_after"].astype(str)
        if not observed_after.equals(ledger_status.reindex(result.index)):
            raise IngestError("Retry after-statuses disagree with the classified ledger.")
        retry_success = ledger.set_index(list(LEDGER_KEY))[RETRY_FLAG_COLUMN].reindex(
            result.index
        )
        if retry_success.isna().any() or not bool(
            (pd.to_numeric(retry_success, errors="coerce") == 1).all()
        ):
            raise IngestError(
                "A logged retry reclassification lacks successful individual-retry evidence."
            )
    # Log the log's extent.
    logger.info(
        "reclassification log built across %d recovered cell(s).", len(result)
    )
    # Return the log.
    return result


# Compose the per-series status and disposition counts.
def build_status_counts(ledger: pd.DataFrame) -> pd.DataFrame:
    """
    Compose the per-series status and disposition counts.

    Parameters
    ----------
    ledger : pandas.DataFrame
        The disposed ledger.

    Returns
    -------
    pandas.DataFrame
        One row per series carrying the count of every status and every
        disposition, which the specification requires so the block can be
        compared with the manuscript's panel counts without re-deriving them.

    Raises
    ------
    IngestError
        If the ledger lacks a required column or carries no rows.
    """
    # Refuse an empty ledger.
    if len(ledger) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot count an empty ledger.")
    # Refuse a ledger lacking a required column.
    for column in ("series", "final_status", "disposition"):
        # Test the column's presence.
        if column not in ledger.columns:
            # The failure names the absent column.
            raise IngestError(f"The ledger lacks the column {column!r}.")
    # Compose the status counts per series.
    status_counts: pd.DataFrame = (
        ledger.groupby(["series", "final_status"]).size().unstack(fill_value=0)
    ).reindex(columns=list(STATUS_VOCABULARY), fill_value=0)
    # Compose the disposition counts per series.
    disposition_counts: pd.DataFrame = (
        ledger.groupby(["series", "disposition"]).size().unstack(fill_value=0)
    ).reindex(columns=list(DISPOSITION_VOCABULARY), fill_value=0)
    # Compose the combined frame with prefixed columns.
    counts: pd.DataFrame = status_counts.add_prefix("status_").join(
        disposition_counts.add_prefix("disposition_"), how="outer"
    ).fillna(0)
    # Cast the counts to integers for the report.
    counts = counts.astype(int)
    counts["cells"] = ledger.groupby("series").size().astype(int)
    if "completion_eligible" in ledger.columns:
        counts["completion_eligible"] = ledger.groupby("series")[
            "completion_eligible"
        ].sum().astype(int)
    if "recovery_eligible" in ledger.columns:
        counts["recovery_eligible"] = ledger.groupby("series")[
            "recovery_eligible"
        ].sum().astype(int)
    # Prove pre-completion status counts remain a separate exhaustive partition.
    status_columns = [column for column in counts if column.startswith("status_")]
    if not bool((counts[status_columns].sum(axis=1) == counts["cells"]).all()):
        raise IngestError("Pre-completion status counts do not partition each series.")
    # Log the counts' extent.
    logger.info("per-series counts composed across %d series.", len(counts))
    # Return the counts.
    return counts


# Freeze the ledger behind a canonical digest over the sorted key.
def freeze_cell_ledger(ledger: pd.DataFrame) -> str:
    """
    Freeze the ledger behind a canonical digest over the sorted key.

    Parameters
    ----------
    ledger : pandas.DataFrame
        The disposed ledger.

    Returns
    -------
    str
        The hexadecimal digest of the canonical payload, computed over the
        primary key in sorted order so two ledgers with the same content digest
        identically regardless of their construction order.

    Raises
    ------
    IngestError
        If the ledger is empty or lacks a required column.
    """
    # Refuse an empty ledger.
    if len(ledger) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot freeze an empty ledger.")
    # Refuse a ledger lacking a required column.
    for column in ("final_status", "disposition"):
        # Test the column's presence.
        if column not in ledger.columns:
            # The failure names the absent column.
            raise IngestError(f"The ledger lacks the column {column!r}.")
    if bool(ledger.duplicated(list(LEDGER_KEY)).any()):
        raise IngestError("Cannot freeze duplicate canonical cell keys.")
    # Bind every classification, disposition, exclusion and supporting-evidence
    # column with its labelled schema after canonical key ordering.
    ordered = ledger.sort_values(list(LEDGER_KEY), kind="stable").reset_index(drop=True)
    return _content_digest(ordered)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 25, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Build the cell-status ledger from the raw block.
def build_cell_ledger(
    raw: pd.DataFrame,
    retry_records: Optional[pd.DataFrame] = None,
    exclusions: Optional[pd.DataFrame] = None,
    expected_counts: Optional[Mapping[str, Mapping[str, int]]] = None,
) -> CellLedgerResult:
    """
    Build the cell-status ledger from the raw block.

    Parameters
    ----------
    raw : pandas.DataFrame
        The raw cell-status block carrying the primary key, the value and the
        two confirmation flags.
    retry_records : pandas.DataFrame, optional
        The retry ledger, which records every query reissued as an individual
        request.
    exclusions : pandas.DataFrame, optional
        Canonical country/russian_gas/window exclusions. When absent the eight
        declared AUT/DEU/POL/TUR by exposure-window keys are used.
    expected_counts : mapping of str to mapping of str to int, optional
        The declared per-series counts, compared only when supplied.

    Returns
    -------
    CellLedgerResult
        The ledger, the counts, the reclassification log, the exclusions, the
        digest, the timestamp and the notes.

    Raises
    ------
    IngestError
        If the raw block is empty, if the classification, the dispositions, the
        immutability gate, the log or the counts fail, or if a declared count
        diverges from the observed one.
    """
    # Refuse an empty raw block.
    if len(raw) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot build a ledger from an empty raw block.")
    # Compose the exclusions, defaulting to the declared set.
    applied_exclusions: pd.DataFrame = (
        exclusions.copy(deep=True)
        if exclusions is not None else build_priority_exclusions()
    )
    if list(applied_exclusions.index.names) != [
        "country_code", "series", "window_id"
    ]:
        raise IngestError(
            "Priority exclusions must use the country/series/window key."
        )
    # Normalize a caller-supplied legacy series alias before exact identity
    # comparison; the stored exclusion audit itself is always canonical.
    exclusion_rows = applied_exclusions.reset_index()
    exclusion_rows["country_code"] = (
        exclusion_rows["country_code"].astype(str).str.strip().str.upper()
    )
    exclusion_rows["series"] = exclusion_rows["series"].map(
        _canonical_trade_series
    )
    exclusion_rows["window_id"] = exclusion_rows["window_id"].astype(str)
    applied_exclusions = exclusion_rows.set_index(
        ["country_code", "series", "window_id"]
    )
    if not bool(applied_exclusions.index.is_unique):
        raise IngestError("Priority exclusions carry duplicate canonical keys.")
    mandatory_exclusions = build_priority_exclusions()
    if set(applied_exclusions.index) != set(mandatory_exclusions.index):
        raise IngestError(
            "Priority exclusions must contain exactly AUT/DEU/POL/TUR russian_gas "
            "keys in both declared exposure windows."
        )
    # Classify every cell.
    ledger: pd.DataFrame = classify_cells(raw)
    # Assign the dispositions.
    ledger = resolve_dispositions(ledger, applied_exclusions)
    # Assert the immutability of the rejected cells.
    assert_rejected_immutability(ledger)
    # Compose the reclassification log.
    canonical_retries = retry_records.copy(deep=True) if retry_records is not None else None
    if canonical_retries is not None and "series" in canonical_retries.columns:
        canonical_retries["series"] = canonical_retries["series"].map(
            _canonical_trade_series
        )
    reclassifications: pd.DataFrame = build_reclassification_log(
        ledger, canonical_retries
    )
    # Compose the per-series counts.
    counts: pd.DataFrame = build_status_counts(ledger)
    # Compare the counts with the declared values when supplied.
    if expected_counts is not None:
        # Iterate over the declared series.
        for raw_series, declared in expected_counts.items():
            series = _canonical_trade_series(raw_series)
            # Refuse a series the ledger does not carry.
            if series not in counts.index:
                # The failure names the absent series.
                raise IngestError(f"The ledger omits the declared series {series!r}.")
            # Compare each declared count.
            for raw_column, value in declared.items():
                column = (
                    f"status_{raw_column}"
                    if str(raw_column) in set(STATUS_VOCABULARY)
                    else str(raw_column)
                )
                # Refuse a column the counts do not carry.
                if column not in counts.columns:
                    # The failure names the absent column.
                    raise IngestError(f"The counts omit the declared column {column!r}.")
                # Read the observed count.
                if isinstance(value, (bool, np.bool_)) or not isinstance(
                    value, (int, np.integer)
                ) or int(value) < 0:
                    raise IngestError(
                        f"{series}/{column}: declared count must be a nonnegative integer."
                    )
                observed: int = int(counts.loc[series, column])
                # Refuse a divergence from the declared count.
                if observed != int(value):
                    # The failure names the series, the column and both counts.
                    raise IngestError(
                        f"{series}/{column}: observed {observed} against the "
                        f"declared {int(value)}."
                    )
    # Freeze the ledger.
    ledger_fingerprint = freeze_cell_ledger(ledger)
    exclusion_fingerprint = _content_digest(applied_exclusions.sort_index())
    fingerprint: str = hashlib.sha256(
        f"{ledger_fingerprint}|{exclusion_fingerprint}".encode("utf-8")
    ).hexdigest()
    # Accumulate the recorded notes.
    notes: List[str] = [
        "the classification consumes only the value and the two confirming "
        "flags, so it carries no exogenous input and is recomputable from the "
        "frozen raw block; the original source value and series alias are retained.",
        "a zero is a verified zero only when both confirming flags are set; a "
        "zero without confirmation is missing.",
        "the priority exclusion matches canonical country/russian_gas/window "
        "keys, so another primitive or a year outside both windows is not excluded.",
        "a missing canonical russian_gas cell remains recoverable pending the "
        "registered exporter-mirror assessment; completion requires actual "
        "within-window original anchors, and retry success alone changes neither "
        "eligibility rule.",
        "rejected and case-excluded cells are immutable barriers to downstream "
        "recovery and completion.",
    ]
    # Record the applied exclusion set.
    notes.append(
        "priority exclusions applied: "
        f"{sorted(set(str(c) for c, _, _ in applied_exclusions.index))} "
        "on the series "
        f"{sorted(set(str(s) for _, s, _ in applied_exclusions.index))} "
        "and windows "
        f"{sorted(set(str(w) for _, _, w in applied_exclusions.index))}."
    )
    # Record the retry state.
    notes.append(
        "retry records were supplied, so recovered cells are logged."
        if retry_records is not None
        else "no retry records were supplied, so no cell is logged as recovered."
    )
    # Assemble the immutable result.
    result: CellLedgerResult = CellLedgerResult(
        ledger=ledger,
        counts=counts,
        reclassifications=reclassifications,
        exclusions=applied_exclusions,
        fingerprint_sha256=fingerprint,
        built_utc=current_utc_iso(),
        notes=tuple(notes),
    )
    # Log the completed ledger.
    logger.info(
        "cell ledger complete: %d cell(s), %d series, fingerprint %s.",
        len(ledger),
        len(counts),
        fingerprint[:16],
    )
    # Return the assembled result.
    return result
````

````python
# Task 26: Mirror-Flow Recovery, Denominator Construction, and Completion

# ==============================================================================
# Task 26: Mirror-Flow Recovery, Denominator Construction, and Completion
# ==============================================================================

# Declare the mechanism label of an importer-reported value.
MECHANISM_IMPORTER: Final[str] = "importer_reported"
# Declare the mechanism label of a mirror recovery.
MECHANISM_MIRROR: Final[str] = "russian_exporter_mirror"
# Declare the mechanism label of a cell no mechanism produced.
MECHANISM_ABSENT: Final[str] = "null"
# Compose the declared mechanism vocabulary.
MECHANISM_VOCABULARY: Final[Tuple[str, ...]] = (
    MECHANISM_IMPORTER,
    MECHANISM_MIRROR,
    MECHANISM_ABSENT,
)
# Declare the substitution label of a directly reported total.
SUBSTITUTE_IMPORTER: Final[str] = "importer_reported"
# Declare the substitution label of a substituted total.
SUBSTITUTE_EXPORTER_SUM: Final[str] = "exporter_sum_substitute"
# Compose the declared substitution vocabulary.
SUBSTITUTION_VOCABULARY: Final[Tuple[str, ...]] = (
    SUBSTITUTE_IMPORTER,
    SUBSTITUTE_EXPORTER_SUM,
    MECHANISM_ABSENT,
)
# Declare the label of a completion by interpolation.
METHOD_INTERPOLATION: Final[str] = "log1p_linear_interpolation"
# Declare the label of a completion by bounded growth extrapolation.
METHOD_EXTRAPOLATION: Final[str] = "bounded_median_growth_extrapolation"
# Declare the longest internal gap the interpolation fills.
MAXIMUM_INTERNAL_GAP: Final[int] = 2
# Declare the minimum number of anchors an edge extrapolation requires.
MINIMUM_ANCHORS: Final[int] = 3
# Declare the lower percentile bound of the growth rate.
GROWTH_PERCENTILE_LOWER: Final[float] = 5.0
# Declare the upper percentile bound of the growth rate.
GROWTH_PERCENTILE_UPPER: Final[float] = 95.0
# Compose the hard barrier between the two data windows, which no completion
# may bridge.
WINDOW_BARRIER: Final[Tuple[int, int]] = (2013, 2018)
# Declare the series of the three completed primitives.
PRIMITIVE_SERIES: Final[Tuple[str, ...]] = (
    "russian_gas",
    "world_gas",
    "russian_non_gas",
)
# Declare the exporter code of the mirror flow.
EXPORTER_CODE: Final[str] = "RUS"
# Declare the export flow label.
EXPORT_FLOW: Final[str] = "export"
# Declare the column carrying the recovered value.
RECOVERED_COLUMN: Final[str] = "recovered_value_usd"
# Declare the exact, non-overlapping exposure windows used by every completion.
EXPOSURE_COMPLETION_WINDOWS: Final[Mapping[str, Tuple[int, ...]]] = MappingProxyType(
    {
        "w2010_2013": (2010, 2011, 2012, 2013),
        "w2018_2021": (2018, 2019, 2020, 2021),
    }
)
# Declare the authoritative current-dollar GDP indicator consumed downstream.
NOMINAL_GDP_INDICATOR: Final[str] = "NY.GDP.MKTP.CD"
# Register the accepted source labels of the two gas subtypes. The canonical
# labels are the exact keys of the previously frozen HS registry.
GAS_SUBTYPE_ALIASES: Final[Mapping[str, str]] = MappingProxyType(
    {
        "lng": "LNG",
        "liquefied natural gas": "LNG",
        "pipeline gas": "Pipeline gas",
        "gaseous gas": "Pipeline gas",
        "natural gas in gaseous state": "Pipeline gas",
        # Retain compatibility with an explicitly pre-aggregated mirror audit;
        # production raw blocks should carry the two registered subtypes.
        "russian gas": "Pre-aggregated Russian gas",
    }
)


# Declare the frozen result of the exposure construction.
@dataclass(frozen=True)
class ExposureConstructionResult:
    """
    Frozen result of the mirror recovery, the denominator and the completion.

    Attributes
    ----------
    recovered : pandas.DataFrame
        The classified ledger with the mechanism and value columns attached,
        whose importer-reported positives are unchanged.
    denominator : pandas.DataFrame
        One row per reporter and year carrying the corrected numerator, every
        denominator trigger, the selected total, rejection status and source
        provenance.
    completed : pandas.DataFrame
        The three completed trade primitives plus authoritative nominal GDP,
        keyed on the reporter, the year and the series. GDP is joined after
        trade completion and is never interpolated or substituted.
    imputation_ledger : pandas.DataFrame
        One row per completed cell carrying the method and the anchor count.
    counts : pandas.DataFrame
        One row per series carrying the number of completed cells and the
        number of cells left missing.
    fingerprint_sha256 : str
        Digest of the canonical construction payload.
    built_utc : str
        ISO-8601 UTC timestamp of the construction.
    notes : tuple of str
        Recorded decisions, including the applied barrier span.
    """
    # The recovered ledger.
    recovered: pd.DataFrame
    # The denominator frame.
    denominator: pd.DataFrame
    # The completed primitives.
    completed: pd.DataFrame
    # The completion ledger.
    imputation_ledger: pd.DataFrame
    # The per-series completion counts.
    counts: pd.DataFrame
    # Digest of the canonical construction payload.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the construction.
    built_utc: str
    # Recorded decisions of the construction.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 26, Step 1: The mirror-flow recovery
# -------------------------------------------------------------------------------------------------------------------------------


# Resolve one exporter product row to its registered gas subtype.
def _resolve_registered_gas_subtype(
    product_description: Any,
    product_code: Any = None,
) -> str:
    """Return the canonical gas subtype after description/HS-code validation."""
    normalized = " ".join(
        str(product_description)
        .strip()
        .lower()
        .replace("-", " ")
        .replace("_", " ")
        .split()
    )
    described = GAS_SUBTYPE_ALIASES.get(normalized)
    coded: Optional[str] = None
    if product_code is not None and not bool(pd.isna(product_code)):
        digits = "".join(character for character in str(product_code) if character.isdigit())
        matches = [
            subtype
            for subtype, prefixes in HS_GAS_CODES.items()
            if any(digits.startswith(str(prefix)) for prefix in prefixes)
        ]
        if len(matches) != 1:
            raise IngestError(
                f"Unregistered or ambiguous gas HS code {product_code!r}."
            )
        coded = str(matches[0])
    if described is None and coded is None:
        raise IngestError(
            f"Unregistered exporter gas product {product_description!r}."
        )
    if (
        described is not None
        and described != "Pre-aggregated Russian gas"
        and coded is not None
        and described != coded
    ):
        raise IngestError(
            "Exporter product description and registered HS code disagree: "
            f"{product_description!r}/{product_code!r}."
        )
    if described == "Pre-aggregated Russian gas" and coded is not None:
        raise IngestError(
            "A pre-aggregated Russian-gas row cannot also claim a subtype HS code."
        )
    return str(coded or described)


# Recover importer-side missing gas cells from the exporter's mirror rows.
def recover_mirror_flow(
    ledger: pd.DataFrame,
    mirror_block: pd.DataFrame,
    exclusions: Optional[Any] = None,
) -> pd.DataFrame:
    """
    Recover importer-side missing gas cells from the exporter's mirror rows.

    Parameters
    ----------
    ledger : pandas.DataFrame
        The classified and disposed ledger of the commodity cells.
    mirror_block : pandas.DataFrame
        The raw block carrying the exporter's own rows, whose flow label,
        reporter, partner, year and product identify the mirror record.
    exclusions : pandas.DataFrame or sequence of tuple, optional
        The canonical country/series/window exclusions, or legacy country-series
        pairs/window triples, that bar recovery. When absent the declared eight
        AUT/DEU/POL/TUR window keys are used.

    Returns
    -------
    pandas.DataFrame
        The ledger with the recovered value and the two mechanism columns
        attached. Only missing cells are candidates, so an importer-reported
        positive retains precedence and is never overwritten; a case-excluded
        cell is never recovered; and a cell without a mirror row remains
        missing with an absent mechanism.

    Raises
    ------
    IngestError
        If a required column is absent, if the merge changes the ledger's key,
        or if the recovery would overwrite a positive cell.
    """
    required_ledger = set(LEDGER_KEY) | {
        VALUE_COLUMN,
        "final_status",
        "disposition",
        "recovery_eligible",
        "immutable_exclusion",
    }
    if not required_ledger.issubset(ledger.columns):
        raise IngestError(
            "The recovery ledger lacks required field(s) "
            f"{sorted(required_ledger - set(ledger.columns))}."
        )
    if bool(ledger.duplicated(list(LEDGER_KEY)).any()):
        raise IngestError("The recovery ledger carries duplicate cell keys.")
    required_mirror = {
        "reporter_code",
        "partner_code",
        "year",
        "product_description",
        VALUE_COLUMN,
        "trade_flow",
    }
    if not required_mirror.issubset(mirror_block.columns):
        raise IngestError(
            "The mirror block lacks required field(s) "
            f"{sorted(required_mirror - set(mirror_block.columns))}."
        )
    # Resolve the immutable exclusion audit in the same canonical namespace as
    # Task 25. Legacy pair/triple inputs remain accepted but are expanded to the
    # two exact exposure windows before any candidate is selected.
    source_exclusions: Any = (
        exclusions if exclusions is not None else build_priority_exclusions()
    )
    applied_exclusions: set = set()
    if isinstance(source_exclusions, pd.DataFrame):
        if list(source_exclusions.index.names) != [
            "country_code",
            "series",
            "window_id",
        ]:
            raise IngestError("Recovery exclusions lack the canonical window key.")
        required_columns = {"window_start", "window_end", "reason", "immutable"}
        if not required_columns.issubset(source_exclusions.columns):
            raise IngestError("Recovery exclusions lack immutable window evidence.")
        for (country, series, window_id), row in source_exclusions.iterrows():
            if (
                str(row["reason"]) != PRIORITY_EXCLUSION_REASON
                or not isinstance(row["immutable"], (bool, np.bool_))
                or not bool(row["immutable"])
            ):
                raise IngestError("Recovery exclusion evidence is malformed.")
            applied_exclusions.add(
                (
                    str(country).strip().upper(),
                    _canonical_trade_series(series),
                    str(window_id),
                    int(row["window_start"]),
                    int(row["window_end"]),
                )
            )
    else:
        for item in source_exclusions:
            candidate = tuple(item)
            if len(candidate) not in (2, 3):
                raise IngestError(
                    "Recovery exclusions must be pairs or window-key triples."
                )
            country, series = candidate[:2]
            selected_windows = PRIORITY_EXCLUSION_WINDOWS
            if len(candidate) == 3:
                selected_windows = tuple(
                    window
                    for window in PRIORITY_EXCLUSION_WINDOWS
                    if window[0] == str(candidate[2])
                )
                if len(selected_windows) != 1:
                    raise IngestError(
                        f"Unknown recovery exclusion window {candidate[2]!r}."
                    )
            for window_id, start, end in selected_windows:
                applied_exclusions.add(
                    (
                        str(country).strip().upper(),
                        _canonical_trade_series(series),
                        str(window_id),
                        int(start),
                        int(end),
                    )
                )
    # Canonicalize the ledger without changing its row order or input object.
    recovered = ledger.copy(deep=True)
    recovered["reporter_code"] = (
        recovered["reporter_code"].astype(str).str.strip().str.upper()
    )
    recovered["series"] = recovered["series"].map(_canonical_trade_series)
    years = pd.to_numeric(recovered["year"], errors="coerce")
    if bool(years.isna().any()) or bool((years % 1.0 != 0.0).any()):
        raise IngestError("The recovery ledger carries a malformed calendar year.")
    recovered["year"] = years.astype(int)
    if bool(recovered.duplicated(list(LEDGER_KEY)).any()):
        raise IngestError("Canonical recovery keys are not unique.")
    # Canonicalize and validate only the registered Russian exporter rows. Other
    # trade directions remain outside the recovery evidence by construction.
    raw_mirror = mirror_block.copy(deep=True)
    raw_mirror["reporter_code"] = (
        raw_mirror["reporter_code"].astype(str).str.strip().str.upper()
    )
    raw_mirror["partner_code"] = (
        raw_mirror["partner_code"].astype(str).str.strip().str.upper()
    )
    raw_mirror["trade_flow"] = raw_mirror["trade_flow"].astype(str).str.strip().str.lower()
    mirror_rows = raw_mirror[
        (raw_mirror["reporter_code"] == EXPORTER_CODE)
        & (raw_mirror["trade_flow"] == EXPORT_FLOW)
    ].copy()
    aggregate_rows: List[Dict[str, Any]] = []
    if len(mirror_rows):
        mirror_years = pd.to_numeric(mirror_rows["year"], errors="coerce")
        if bool(mirror_years.isna().any()) or bool(
            (mirror_years % 1.0 != 0.0).any()
        ):
            raise IngestError("The mirror block carries a malformed calendar year.")
        mirror_rows["year"] = mirror_years.astype(int)
        if bool((mirror_rows["partner_code"].str.len() == 0).any()):
            raise IngestError("The mirror block carries a blank importing partner.")
        mirror_rows["registered_subtype"] = [
            _resolve_registered_gas_subtype(
                description,
                mirror_rows["product_code"].iloc[position]
                if "product_code" in mirror_rows.columns
                else None,
            )
            for position, description in enumerate(
                mirror_rows["product_description"].tolist()
            )
        ]
        mirror_numeric = pd.to_numeric(mirror_rows[VALUE_COLUMN], errors="coerce")
        malformed = mirror_numeric.isna() & ~mirror_rows[VALUE_COLUMN].isna()
        if bool(malformed.any()) or bool(
            np.isinf(mirror_numeric.to_numpy(dtype=float)).any()
        ):
            raise IngestError("The mirror block carries a malformed/non-finite value.")
        if bool((mirror_numeric.dropna() < 0.0).any()):
            raise IngestError("The mirror block carries a negative trade value.")
        mirror_rows[VALUE_COLUMN] = mirror_numeric.astype(float)
        source_key = ["partner_code", "year", "registered_subtype"]
        if "product_code" in mirror_rows.columns:
            source_key.append("product_code")
        if bool(mirror_rows.duplicated(source_key).any()):
            raise IngestError("The mirror block carries duplicate registered rows.")
        for (partner, year), group in mirror_rows.groupby(
            ["partner_code", "year"], sort=True
        ):
            subtypes = tuple(sorted(set(group["registered_subtype"].astype(str))))
            if "Pre-aggregated Russian gas" in subtypes and len(subtypes) > 1:
                raise IngestError(
                    f"{partner}/{year}: pre-aggregated and subtype mirror rows overlap."
                )
            numeric = group[VALUE_COLUMN].to_numpy(dtype=float)
            positive = np.isfinite(numeric) & (numeric > 0.0)
            positive_subtypes = tuple(
                sorted(set(group.loc[positive, "registered_subtype"].astype(str)))
            )
            if set(positive_subtypes) >= {"LNG", "Pipeline gas"}:
                coverage_state = "lng_and_pipeline"
            elif positive_subtypes == ("LNG",):
                coverage_state = "lng_only"
            elif positive_subtypes == ("Pipeline gas",):
                coverage_state = "pipeline_only"
            elif positive_subtypes == ("Pre-aggregated Russian gas",):
                coverage_state = "preaggregated"
            else:
                coverage_state = "no_positive_registered_product"
            aggregate_rows.append(
                {
                    "reporter_code": str(partner),
                    "year": int(year),
                    "mirror_total_usd": float(numeric[positive].sum())
                    if bool(positive.any())
                    else float("nan"),
                    "mirror_registered_rows": int(len(group)),
                    "mirror_positive_rows": int(positive.sum()),
                    "mirror_observed_subtypes": json.dumps(subtypes),
                    "mirror_positive_subtypes": json.dumps(positive_subtypes),
                    "mirror_has_lng": bool("LNG" in positive_subtypes),
                    "mirror_has_pipeline": bool("Pipeline gas" in positive_subtypes),
                    "mirror_coverage_state": coverage_state,
                    "mirror_source_sha256": _content_digest(
                        group.sort_values(source_key, kind="stable").reset_index(drop=True)
                    ),
                }
            )
    mirror_aggregate = pd.DataFrame(
        aggregate_rows,
        columns=[
            "reporter_code",
            "year",
            "mirror_total_usd",
            "mirror_registered_rows",
            "mirror_positive_rows",
            "mirror_observed_subtypes",
            "mirror_positive_subtypes",
            "mirror_has_lng",
            "mirror_has_pipeline",
            "mirror_coverage_state",
            "mirror_source_sha256",
        ],
    )
    if bool(mirror_aggregate.duplicated(["reporter_code", "year"]).any()):
        raise IngestError("Aggregated mirror rows are not unique by partner/year.")
    recovered["_recovery_order"] = np.arange(len(recovered), dtype=int)
    recovered = recovered.merge(
        mirror_aggregate,
        on=["reporter_code", "year"],
        how="left",
        validate="many_to_one",
        sort=False,
    ).sort_values("_recovery_order", kind="stable")
    recovered = recovered.drop(columns=["_recovery_order"]).reset_index(drop=True)
    recovered["mirror_registered_rows"] = (
        recovered["mirror_registered_rows"].fillna(0).astype(int)
    )
    recovered["mirror_positive_rows"] = (
        recovered["mirror_positive_rows"].fillna(0).astype(int)
    )
    for column in ("mirror_has_lng", "mirror_has_pipeline"):
        recovered[column] = recovered[column].fillna(False).astype(bool)
    recovered["mirror_observed_subtypes"] = recovered[
        "mirror_observed_subtypes"
    ].fillna("[]")
    recovered["mirror_positive_subtypes"] = recovered[
        "mirror_positive_subtypes"
    ].fillna("[]")
    recovered["mirror_coverage_state"] = recovered[
        "mirror_coverage_state"
    ].fillna("no_registered_export_rows")
    recovered["mirror_source_sha256"] = recovered["mirror_source_sha256"].fillna("")
    russian_mask = recovered["series"].astype(str).eq("russian_gas").to_numpy()
    mapped = pd.to_numeric(
        recovered["mirror_total_usd"], errors="coerce"
    ).to_numpy(dtype=float).copy()
    mapped[~russian_mask] = np.nan
    candidate_mask = (
        russian_mask
        & recovered["final_status"].astype(str).eq(STATUS_MISSING).to_numpy()
        & recovered["disposition"].astype(str).eq(DISPOSITION_RECOVERABLE).to_numpy()
        & recovered["recovery_eligible"].astype(bool).to_numpy()
    )
    excluded = np.array(
        [
            any(
                str(recovered["reporter_code"].iloc[position]) == country
                and str(recovered["series"].iloc[position]) == series
                and start <= int(recovered["year"].iloc[position]) <= end
                for country, series, _window_id, start, end in applied_exclusions
            )
            for position in range(len(recovered))
        ],
        dtype=bool,
    )
    recovery_mask = candidate_mask & ~excluded & np.isfinite(mapped) & (mapped > 0.0)
    recovered[RECOVERED_COLUMN] = np.where(recovery_mask, mapped, np.nan)
    importer_mask = recovered["final_status"].astype(str).isin(
        [STATUS_POSITIVE, STATUS_VERIFIED_ZERO]
    ).to_numpy()
    recovered["effective_value_usd"] = pd.to_numeric(
        recovered[VALUE_COLUMN], errors="coerce"
    ).to_numpy(dtype=float)
    recovered.loc[recovery_mask, "effective_value_usd"] = mapped[recovery_mask]
    mechanism = np.where(
        importer_mask,
        MECHANISM_IMPORTER,
        np.where(recovery_mask, MECHANISM_MIRROR, MECHANISM_ABSENT),
    )
    recovered["mirror_availability"] = pd.Categorical(
        mechanism, categories=list(MECHANISM_VOCABULARY)
    )
    recovered["recovered_via"] = pd.Categorical(
        mechanism, categories=list(MECHANISM_VOCABULARY)
    )
    recovered["recovery_realized"] = recovery_mask
    recovered["recovery_attempted"] = candidate_mask & ~excluded
    if "recovery_evidence_state" in recovered.columns:
        recovered.loc[recovery_mask, "recovery_evidence_state"] = (
            "realized_positive_mirror"
        )
        recovered.loc[
            candidate_mask & ~excluded & ~recovery_mask,
            "recovery_evidence_state",
        ] = "registered_mirror_unavailable"
    # Importer observations retain exact precedence even when a positive mirror
    # is present, while excluded/rejected cells remain immutable downstream.
    observed_input = pd.to_numeric(ledger[VALUE_COLUMN], errors="coerce").to_numpy(
        dtype=float
    )
    observed_output = pd.to_numeric(
        recovered["effective_value_usd"], errors="coerce"
    ).to_numpy(dtype=float)
    if not bool(np.array_equal(observed_input[importer_mask], observed_output[importer_mask])):
        raise IngestError("Mirror recovery overwrote an importer observation.")
    if bool(recovery_mask[excluded].any()) or bool((recovery_mask & ~russian_mask).any()):
        raise IngestError("Mirror recovery crossed an immutable scientific boundary.")
    assert_rejected_immutability(recovered)
    recovered_counts = recovered.loc[recovery_mask].groupby("reporter_code").size()
    logger.info(
        "mirror recovery applied to %d cell(s) across %d countr(ies): %s; "
        "registered exporter partner-years %d.",
        int(recovery_mask.sum()),
        len(recovered_counts),
        recovered_counts.to_dict(),
        len(mirror_aggregate),
    )
    return recovered


# -------------------------------------------------------------------------------------------------------------------------------
# Task 26, Step 2: The gas-share denominator
# -------------------------------------------------------------------------------------------------------------------------------


# Construct the gas-share denominator with the declared substitution rule.
def construct_denominator(
    numerator: pd.DataFrame,
    totals: pd.DataFrame,
) -> pd.DataFrame:
    """
    Construct the gas-share denominator with the declared substitution rule.

    Parameters
    ----------
    numerator : pandas.DataFrame
        One row per reporter and year carrying the Russian flow, which is the
        mirror-recovered numerator.
    totals : pandas.DataFrame
        One row per reporter and year carrying the importer-reported total and
        the exporter-sum total.

    Returns
    -------
    pandas.DataFrame
        One row per reporter and year carrying the numerator, the retained
        total, the share and the substitution mechanism, whose categorical
        labels are the declared three.

    Raises
    ------
    IngestError
        If a required column is absent, if the two frames carry different keys,
        or if a retained cell violates the declared share bound, because a
        numerator outside the interval from zero to its total cannot be a
        share.
    """
    numerator_columns = {"reporter_code", "year", "russian_flow_usd"}
    totals_columns = {
        "reporter_code",
        "year",
        "world_total_usd",
        "exporter_sum_usd",
    }
    if not numerator_columns.issubset(numerator.columns):
        raise IngestError(
            "The numerator lacks required field(s) "
            f"{sorted(numerator_columns - set(numerator.columns))}."
        )
    if not totals_columns.issubset(totals.columns):
        raise IngestError(
            "The totals frame lacks required field(s) "
            f"{sorted(totals_columns - set(totals.columns))}."
        )
    if len(numerator) == 0 and len(totals) == 0:
        raise IngestError("The denominator requires a reporter-year universe.")
    numerator_clean = numerator.copy(deep=True)
    totals_clean = totals.copy(deep=True)
    for frame_name, frame in (
        ("numerator", numerator_clean),
        ("totals", totals_clean),
    ):
        if bool(frame["reporter_code"].isna().any()):
            raise IngestError(f"The {frame_name} carries a null reporter code.")
        frame["reporter_code"] = (
            frame["reporter_code"].astype(str).str.strip().str.upper()
        )
        years = pd.to_numeric(frame["year"], errors="coerce")
        if bool(years.isna().any()) or bool((years % 1.0 != 0.0).any()):
            raise IngestError(f"The {frame_name} carries a malformed calendar year.")
        frame["year"] = years.astype(int)
        if bool(frame.duplicated(["reporter_code", "year"]).any()):
            raise IngestError(f"The {frame_name} carries duplicate reporter-year keys.")
    # Parse all monetary fields strictly, retaining true missing values while
    # refusing strings, infinities, Booleans and negative Russian/export sums.
    for frame_name, frame, columns in (
        ("numerator", numerator_clean, ("russian_flow_usd",)),
        ("totals", totals_clean, ("world_total_usd", "exporter_sum_usd")),
    ):
        for column in columns:
            boolean = frame[column].map(
                lambda value: isinstance(value, (bool, np.bool_))
            )
            numeric = pd.to_numeric(frame[column], errors="coerce")
            malformed = numeric.isna() & ~frame[column].isna()
            if bool(boolean.any()) or bool(malformed.any()) or bool(
                np.isinf(numeric.to_numpy(dtype=float)).any()
            ):
                raise IngestError(
                    f"The {frame_name}/{column} field is malformed or non-finite."
                )
            frame[column] = numeric.astype(float)
    if bool((numerator_clean["russian_flow_usd"].dropna() < 0.0).any()):
        raise IngestError("The numerator carries a negative Russian flow.")
    if bool((totals_clean["exporter_sum_usd"].dropna() < 0.0).any()):
        raise IngestError("The exporter-sum alternative cannot be negative.")
    # Use the union as the explicit left universe so a missing numerator or
    # missing totals row remains visible rather than disappearing in a join.
    universe = pd.concat(
        [
            numerator_clean[["reporter_code", "year"]],
            totals_clean[["reporter_code", "year"]],
        ],
        ignore_index=True,
    ).drop_duplicates().sort_values(["reporter_code", "year"], kind="stable")
    merged = universe.merge(
        numerator_clean,
        on=["reporter_code", "year"],
        how="left",
        validate="one_to_one",
        indicator="numerator_merge_state",
    ).merge(
        totals_clean,
        on=["reporter_code", "year"],
        how="left",
        validate="one_to_one",
        indicator="totals_merge_state",
        suffixes=("", "_totals"),
    )
    merged["numerator_present"] = merged["numerator_merge_state"].eq("both")
    merged["totals_present"] = merged["totals_merge_state"].eq("both")
    merged = merged.drop(columns=["numerator_merge_state", "totals_merge_state"])
    russian = pd.to_numeric(
        merged["russian_flow_usd"], errors="coerce"
    ).to_numpy(dtype=float)
    reported = pd.to_numeric(
        merged["world_total_usd"], errors="coerce"
    ).to_numpy(dtype=float)
    exporter = pd.to_numeric(
        merged["exporter_sum_usd"], errors="coerce"
    ).to_numpy(dtype=float)
    trigger_absent = ~np.isfinite(reported)
    trigger_nonpositive = (
        np.isfinite(reported)
        & (reported <= 0.0)
        & np.isfinite(russian)
        & (russian > 0.0)
    )
    trigger_inconsistent = (
        np.isfinite(reported)
        & np.isfinite(russian)
        & (reported < russian)
    )
    substitute = trigger_absent | trigger_nonpositive | trigger_inconsistent
    exporter_available = np.isfinite(exporter) & (exporter >= 0.0)
    importer_available = np.isfinite(reported) & (reported >= 0.0)
    selected = np.where(
        substitute & exporter_available,
        exporter,
        np.where(~substitute & importer_available, reported, np.nan),
    )
    selected_source = np.where(
        substitute & exporter_available,
        SUBSTITUTE_EXPORTER_SUM,
        np.where(~substitute & importer_available, SUBSTITUTE_IMPORTER, MECHANISM_ABSENT),
    )
    denominator_valid = np.isfinite(selected) & (selected >= 0.0)
    relationship_rejected = (
        np.isfinite(russian)
        & denominator_valid
        & ((russian < 0.0) | (russian > selected))
    )
    substitution_unavailable = substitute & ~exporter_available
    relationship_valid = (
        ~np.isfinite(russian)
        | (denominator_valid & (russian >= 0.0) & (russian <= selected))
    )
    selected[~relationship_valid] = np.nan
    selected_source[~relationship_valid] = MECHANISM_ABSENT
    share_eligible = (
        np.isfinite(russian)
        & (russian >= 0.0)
        & np.isfinite(selected)
        & (selected > 0.0)
        & (russian <= selected)
    )
    triggers: List[str] = []
    primary_triggers: List[str] = []
    selection_reasons: List[str] = []
    for position in range(len(merged)):
        fired = [
            label
            for label, mask in (
                ("importer_total_absent", trigger_absent),
                ("importer_total_non_positive", trigger_nonpositive),
                ("importer_total_below_numerator", trigger_inconsistent),
            )
            if bool(mask[position])
        ]
        triggers.append(";".join(fired) if fired else "direct_importer_report")
        primary_triggers.append(fired[0] if fired else "direct_importer_report")
        if bool(relationship_rejected[position]):
            selection_reasons.append(
                "rejected_numerator_exceeds_selected_denominator"
            )
        elif bool(substitution_unavailable[position]):
            selection_reasons.append("required_exporter_substitute_unavailable")
        elif not bool(np.isfinite(selected[position])):
            selection_reasons.append("no_admissible_denominator")
        elif bool(share_eligible[position]):
            selection_reasons.append("ratio_eligible")
        elif not bool(np.isfinite(russian[position])):
            selection_reasons.append("numerator_missing_denominator_preserved")
        elif selected[position] == 0.0 and russian[position] == 0.0:
            selection_reasons.append("verified_zero_components_zero_denominator")
        else:
            selection_reasons.append("numerator_denominator_relationship_invalid")
    merged["trigger_importer_total_absent"] = trigger_absent
    merged["trigger_importer_total_non_positive"] = trigger_nonpositive
    merged["trigger_importer_total_below_numerator"] = trigger_inconsistent
    merged["substitution_trigger"] = triggers
    merged["primary_substitution_trigger"] = primary_triggers
    merged["retained_total_usd"] = selected
    merged["substitution_mechanism"] = pd.Categorical(
        selected_source, categories=list(SUBSTITUTION_VOCABULARY)
    )
    merged["denominator_selected"] = np.isfinite(selected)
    merged["denominator_rejected"] = relationship_rejected | substitution_unavailable
    merged["retained"] = share_eligible
    merged["denominator_selection_reason"] = selection_reasons
    merged["gas_share"] = np.where(
        share_eligible,
        russian / np.where(selected == 0.0, np.nan, selected),
        np.nan,
    )
    merged["numerator_input_sha256"] = _content_digest(
        numerator_clean.sort_values(["reporter_code", "year"], kind="stable")
        .reset_index(drop=True)
    )
    merged["totals_input_sha256"] = _content_digest(
        totals_clean.sort_values(["reporter_code", "year"], kind="stable")
        .reset_index(drop=True)
    )
    frame = merged.set_index(["reporter_code", "year"]).sort_index()
    if bool(frame.index.duplicated().any()) or len(frame) != len(universe):
        raise IngestError("The denominator did not preserve its full key universe.")
    if bool(
        (
            frame["retained"]
            & (
                (frame["russian_flow_usd"] < 0.0)
                | (frame["russian_flow_usd"] > frame["retained_total_usd"])
            )
        ).any()
    ):
        raise IngestError("A retained denominator violates 0 <= numerator <= total.")
    logger.info(
        "denominator constructed across %d reporter-year(s): %s; eligible "
        "share cells %d.",
        len(frame),
        frame["substitution_mechanism"].value_counts().to_dict(),
        int(frame["retained"].sum()),
    )
    return frame

# -------------------------------------------------------------------------------------------------------------------------------
# Task 26, Step 3: The within-window completion
# -------------------------------------------------------------------------------------------------------------------------------


# Fill one series' internal gaps by interpolation in the log space.
def interpolate_internal_gaps(
    years: np.ndarray,
    values: np.ndarray,
    maximum_gap: int = MAXIMUM_INTERNAL_GAP,
) -> Tuple[np.ndarray, np.ndarray]:
    """
    Fill one series' internal gaps by interpolation in the log space.

    Parameters
    ----------
    years : numpy.ndarray
        The ascending years of the run.
    values : numpy.ndarray
        The observed values of the run, whose missing entries are not finite.
    maximum_gap : int, optional
        The longest consecutive gap the interpolation fills.

    Returns
    -------
    tuple
        The completed values and the mask of the cells that were completed. The
        interpolation is linear in the logarithm of one plus the value, so the
        filled path is geometric rather than arithmetic, and the inverse
        transform returns the level while preserving non-negativity by
        construction.

    Raises
    ------
    IngestError
        If the years are not strictly ascending, because a gap's length would
        then be undefined.
    """
    raw_years = np.asarray(years)
    raw_values = np.asarray(values)
    if raw_years.ndim != 1 or raw_values.ndim != 1 or len(raw_years) != len(raw_values):
        raise IngestError("Interpolation years and values must be aligned vectors.")
    if raw_years.dtype.kind == "b" or isinstance(maximum_gap, (bool, np.bool_)):
        raise IngestError("Interpolation years/gap cannot be Boolean.")
    try:
        numeric_years = raw_years.astype(float)
        numeric_values = raw_values.astype(float)
        numeric_gap = float(maximum_gap)
    except (TypeError, ValueError, OverflowError) as exc:
        raise IngestError("Interpolation inputs must be numeric.") from exc
    if (
        not np.isfinite(numeric_years).all()
        or not np.equal(numeric_years, np.floor(numeric_years)).all()
        or (len(numeric_years) > 1 and not bool((np.diff(numeric_years) == 1.0).all()))
    ):
        raise IngestError(
            "Interpolation requires consecutive, strictly ascending calendar years."
        )
    if not np.isfinite(numeric_gap) or not numeric_gap.is_integer() or not 0 <= numeric_gap <= 2:
        raise IngestError("The maximum internal gap must be an integer from zero to two.")
    if bool(np.isinf(numeric_values).any()) or bool(
        (np.isfinite(numeric_values) & (numeric_values < 0.0)).any()
    ):
        raise IngestError("Observed interpolation values must be finite and nonnegative.")
    calendar_years = numeric_years.astype(int)
    # Copy the values so the caller's array is never modified.
    completed: np.ndarray = numeric_values.copy()
    # Compose the completion mask.
    filled: np.ndarray = np.zeros(len(values), dtype=bool)
    # Identify the observed positions.
    observed: np.ndarray = np.flatnonzero(np.isfinite(values))
    # Return unchanged when fewer than two observations exist, because the
    # interpolation requires a bounding pair.
    if observed.size < 2:
        # Return the copied values and the empty mask.
        return completed, filled
    # Iterate over the consecutive observed pairs.
    for left, right in zip(observed[:-1], observed[1:]):
        # Compose the number of missing cells between the pair.
        gap_length: int = int(calendar_years[right] - calendar_years[left] - 1)
        # Skip a gap longer than the declared maximum, which is left missing
        # rather than fabricated.
        if gap_length == 0 or gap_length > int(numeric_gap):
            # Continue to the next pair.
            continue
        # Compose the fractional positions of the missing years.
        fractions: np.ndarray = (
            calendar_years[left + 1 : right] - calendar_years[left]
        ) / float(
            calendar_years[right] - calendar_years[left]
        )
        # Compose the bounding values in the logarithm of one plus the value.
        left_log: float = float(np.log1p(values[left]))
        # Compose the right bound in the same space.
        right_log: float = float(np.log1p(values[right]))
        # Interpolate linearly in the log space.
        interpolated: np.ndarray = left_log + fractions * (right_log - left_log)
        # Invert the transform, which preserves non-negativity by construction.
        completed[left + 1 : right] = np.maximum(0.0, np.expm1(interpolated))
        # Record the completed cells.
        filled[left + 1 : right] = True
    # Return the completed values and the mask.
    return completed, filled


# Calculate one permitted edge completion and its full growth-bound audit.
def _edge_extrapolation_calculation(
    years: np.ndarray,
    values: np.ndarray,
    minimum_anchors: int,
    lower_percentile: float,
    upper_percentile: float,
    growth_reference: Optional[np.ndarray],
) -> Tuple[np.ndarray, np.ndarray, int, Mapping[str, Any]]:
    """Return edge-completed values, fill mask, anchor count and bound audit."""
    raw_years = np.asarray(years)
    raw_values = np.asarray(values)
    if raw_years.ndim != 1 or raw_values.ndim != 1 or len(raw_years) != len(raw_values):
        raise IngestError("Edge-extrapolation years and values must be aligned vectors.")
    if raw_years.dtype.kind == "b" or isinstance(minimum_anchors, (bool, np.bool_)):
        raise IngestError("Edge-extrapolation years/anchor count cannot be Boolean.")
    if any(isinstance(value, (bool, np.bool_)) for value in raw_values.tolist()):
        raise IngestError("Edge-extrapolation values cannot be Boolean.")
    try:
        numeric_years = raw_years.astype(float)
        numeric_values = raw_values.astype(float)
        anchors_required = float(minimum_anchors)
        lower_probability = float(lower_percentile)
        upper_probability = float(upper_percentile)
    except (TypeError, ValueError, OverflowError) as exc:
        raise IngestError("Edge-extrapolation inputs must be numeric.") from exc
    if (
        len(numeric_years) != 4
        or not np.isfinite(numeric_years).all()
        or not np.equal(numeric_years, np.floor(numeric_years)).all()
        or not bool((np.diff(numeric_years) == 1.0).all())
    ):
        raise IngestError(
            "Edge extrapolation requires one complete four-year calendar window."
        )
    if (
        not np.isfinite(anchors_required)
        or not anchors_required.is_integer()
        or anchors_required < 3
    ):
        raise IngestError("Edge extrapolation requires at least three anchors.")
    if (
        not np.isfinite(lower_probability)
        or not np.isfinite(upper_probability)
        or not 0.0 <= lower_probability < upper_probability <= 100.0
    ):
        raise IngestError(
            "Growth percentiles must be finite and ordered within [0, 100]."
        )
    if bool(np.isinf(numeric_values).any()) or bool(
        (np.isfinite(numeric_values) & (numeric_values < 0.0)).any()
    ):
        raise IngestError("Original edge anchors must be finite and nonnegative.")
    completed = numeric_values.copy()
    filled = np.zeros(len(completed), dtype=bool)
    observed = np.flatnonzero(np.isfinite(numeric_values))
    missing = np.flatnonzero(~np.isfinite(numeric_values))
    empty_audit: Dict[str, Any] = {
        "edge_side": "none",
        "anchor_years": tuple(int(numeric_years[index]) for index in observed),
        "applied_log_rate": float("nan"),
        "growth_lower_bound": float("nan"),
        "growth_upper_bound": float("nan"),
        "growth_reference_sha256": "",
        "growth_reference_scope": "not_used",
    }
    # A valid edge operation has exactly three original anchors and exactly one
    # missing first or last year. Multiple edge years and internal gaps remain
    # untouched rather than being silently converted into an extrapolation.
    if (
        len(observed) < int(anchors_required)
        or len(missing) != 1
        or int(missing[0]) not in (0, len(completed) - 1)
    ):
        return completed, filled, 0, MappingProxyType(empty_audit)
    anchor_logs = np.log1p(numeric_values[observed])
    steps = np.diff(numeric_years[observed])
    if not bool((steps == 1.0).all()):
        return completed, filled, 0, MappingProxyType(empty_audit)
    within_rates = np.diff(anchor_logs) / steps
    if growth_reference is None:
        reference = within_rates.copy()
        reference_scope = "within_run_legacy_fallback"
        logger.warning(
            "edge extrapolation used a disclosed within-run fallback because "
            "no pooled growth-reference distribution was supplied."
        )
    else:
        try:
            reference = np.asarray(growth_reference, dtype=float).reshape(-1)
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError("The growth-reference distribution is malformed.") from exc
        reference_scope = "registered_pooled_observed_growth"
    if reference.size == 0 or not bool(np.isfinite(reference).all()):
        raise IngestError(
            "The growth-reference distribution must contain finite log-growth rates."
        )
    lower = float(np.percentile(reference, lower_probability))
    upper = float(np.percentile(reference, upper_probability))
    rate = float(np.clip(np.median(within_rates), lower, upper))
    edge_position = int(missing[0])
    if edge_position == 0:
        projected_log = float(anchor_logs[0]) - rate
        edge_side = "leading"
    else:
        projected_log = float(anchor_logs[-1]) + rate
        edge_side = "trailing"
    # E19: inversion is explicitly floored at zero, so a declining leading-edge
    # projection cannot produce a negative monetary amount.
    completed[edge_position] = max(0.0, float(np.expm1(projected_log)))
    filled[edge_position] = True
    reference_payload = np.asarray(reference, dtype="<f8").tobytes()
    audit = MappingProxyType(
        {
            "edge_side": edge_side,
            "anchor_years": tuple(int(numeric_years[index]) for index in observed),
            "applied_log_rate": rate,
            "growth_lower_bound": lower,
            "growth_upper_bound": upper,
            "growth_reference_sha256": hashlib.sha256(reference_payload).hexdigest(),
            "growth_reference_scope": reference_scope,
        }
    )
    return completed, filled, int(len(observed)), audit


# Fill one run's edge gaps by the bounded median growth rate.
def extrapolate_edge_gaps(
    years: np.ndarray,
    values: np.ndarray,
    minimum_anchors: int = MINIMUM_ANCHORS,
    lower_percentile: float = GROWTH_PERCENTILE_LOWER,
    upper_percentile: float = GROWTH_PERCENTILE_UPPER,
    growth_reference: Optional[np.ndarray] = None,
) -> Tuple[np.ndarray, np.ndarray, int]:
    """
    Fill one run's edge gaps by the bounded median growth rate.

    Parameters
    ----------
    years : numpy.ndarray
        The ascending years of the run.
    values : numpy.ndarray
        The observed values of the run.
    minimum_anchors : int, optional
        The number of anchors the rate requires.
    lower_percentile : float, optional
        The lower percentile bound of the observed growth rates.
    upper_percentile : float, optional
        The upper percentile bound of the observed growth rates.
    growth_reference : numpy.ndarray, optional
        Registered pooled observed log-growth rates used to compute the two
        percentile bounds. The legacy omitted case is disclosed in the log and
        uses the run's own rates only for backward compatibility; the production
        completion orchestrator always supplies the pooled reference.

    Returns
    -------
    tuple
        The completed values, the mask of the completed cells and the number of
        anchors used. Only one missing boundary year can be completed. The rate
        is the median adjacent original-anchor log growth, clipped to the
        registered pooled reference bounds, and the inverted amount is floored
        at zero.

    Raises
    ------
    IngestError
        If the calendar, values, anchor rule, percentile bounds or registered
        reference distribution is malformed.
    """
    completed, filled, anchors, _audit = _edge_extrapolation_calculation(
        years,
        values,
        minimum_anchors,
        lower_percentile,
        upper_percentile,
        growth_reference,
    )
    return completed, filled, anchors


# Complete the primitives within their windows.
def complete_within_window(
    primitives: pd.DataFrame,
    barrier: Tuple[int, int] = WINDOW_BARRIER,
    maximum_gap: int = MAXIMUM_INTERNAL_GAP,
    minimum_anchors: int = MINIMUM_ANCHORS,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Complete the primitives within their windows.

    Parameters
    ----------
    primitives : pandas.DataFrame
        One row per reporter, year and series carrying the observed value.
    barrier : tuple of int, optional
        The hard barrier between the two data windows, which no completion may
        bridge.
    maximum_gap : int, optional
        The longest consecutive internal gap the interpolation fills.
    minimum_anchors : int, optional
        The number of anchors an edge extrapolation requires.

    Returns
    -------
    tuple
        The completed primitives and the companion ledger recording each
        completed cell's method and anchor count. Every completion is confined
        to the cell's own series, so no cross-country borrowing occurs, and the
        barrier is respected as a run delimiter, so a gap spanning it is never
        bridged.

    Raises
    ------
    IngestError
        If a required column is absent or the barrier is not an ordered pair.
    """
    required = {"reporter_code", "year", "series", VALUE_COLUMN}
    if not required.issubset(primitives.columns):
        raise IngestError(
            "The primitives lack required field(s) "
            f"{sorted(required - set(primitives.columns))}."
        )
    if tuple(int(value) for value in barrier) != WINDOW_BARRIER:
        raise IngestError(
            "Completion must use the fixed 2013/2018 boundary between the two "
            "declared four-year windows."
        )
    if isinstance(maximum_gap, (bool, np.bool_)) or isinstance(
        minimum_anchors, (bool, np.bool_)
    ):
        raise IngestError("Completion thresholds cannot be Boolean.")
    if int(maximum_gap) != float(maximum_gap) or not 0 <= int(maximum_gap) <= 2:
        raise IngestError("Completion permits at most two internal missing years.")
    if int(minimum_anchors) != float(minimum_anchors) or int(minimum_anchors) < 3:
        raise IngestError("Edge completion requires at least three original anchors.")
    source = primitives.copy(deep=True)
    if len(source) == 0:
        raise IngestError("Cannot complete an empty primitive frame.")
    if bool(source["reporter_code"].isna().any()):
        raise IngestError("Completion reporter codes cannot be null.")
    source["reporter_code"] = (
        source["reporter_code"].astype(str).str.strip().str.upper()
    )
    source["series"] = source["series"].map(_canonical_trade_series)
    years = pd.to_numeric(source["year"], errors="coerce")
    if bool(years.isna().any()) or bool((years % 1.0 != 0.0).any()):
        raise IngestError("Completion primitives carry a malformed calendar year.")
    source["year"] = years.astype(int)
    allowed_years = tuple(
        year
        for window in EXPOSURE_COMPLETION_WINDOWS.values()
        for year in window
    )
    if not set(source["year"]).issubset(set(allowed_years)):
        raise IngestError("Completion primitives contain years outside the two windows.")
    if bool(source.duplicated(list(LEDGER_KEY)).any()):
        raise IngestError("Completion primitives carry duplicate canonical cell keys.")
    boolean_values = source[VALUE_COLUMN].map(
        lambda value: isinstance(value, (bool, np.bool_))
    )
    numeric_values = pd.to_numeric(source[VALUE_COLUMN], errors="coerce")
    malformed_values = numeric_values.isna() & ~source[VALUE_COLUMN].isna()
    if bool(boolean_values.any()) or bool(malformed_values.any()) or bool(
        np.isinf(numeric_values.to_numpy(dtype=float)).any()
    ):
        raise IngestError("Primitive values must be missing or finite numeric amounts.")
    if bool((numeric_values.dropna() < 0.0).any()):
        raise IngestError("Primitive amounts cannot be negative.")
    source[VALUE_COLUMN] = numeric_values.astype(float)
    has_task25_evidence = "disposition" in source.columns
    if has_task25_evidence:
        evidence_required = {
            "final_status",
            "precompletion_status",
            "disposition",
            "completion_eligible",
            "recovery_eligible",
            "immutable_exclusion",
            "original_anchor",
        }
        if not evidence_required.issubset(source.columns):
            raise IngestError("Completion received incomplete Task 25 evidence.")
    anchor_column = (
        "completion_original_anchor"
        if "completion_original_anchor" in source.columns
        else "original_anchor"
    )
    # Reindex every observed reporter/primitive pair to all eight declared years.
    # This makes 2013 and 2018 ordinary endpoints rather than barrier casualties.
    grid_parts: List[pd.DataFrame] = []
    for (reporter, series), group in source.groupby(
        ["reporter_code", "series"], sort=True, dropna=False
    ):
        indexed = group.set_index("year").reindex(allowed_years)
        indexed["reporter_code"] = str(reporter)
        indexed["series"] = str(series)
        indexed["year"] = np.asarray(allowed_years, dtype=int)
        if has_task25_evidence and bool(indexed["disposition"].isna().any()):
            missing_years = indexed.loc[indexed["disposition"].isna(), "year"].tolist()
            raise IngestError(
                f"{reporter}/{series}: Task 25 evidence omits grid year(s) "
                f"{missing_years}."
            )
        grid_parts.append(indexed.reset_index(drop=True))
    completed = pd.concat(grid_parts, ignore_index=True, sort=False)
    if bool(completed.duplicated(list(LEDGER_KEY)).any()):
        raise IngestError("The completed primitive grid is not uniquely keyed.")
    expected_grid = {
        (str(reporter), int(year), str(series))
        for reporter, series in source[["reporter_code", "series"]]
        .drop_duplicates()
        .itertuples(index=False, name=None)
        for year in allowed_years
    }
    observed_grid = set(
        completed[list(LEDGER_KEY)].itertuples(index=False, name=None)
    )
    if observed_grid != expected_grid:
        raise IngestError("Completion failed to construct the exact declared grid.")
    if has_task25_evidence:
        for column in (
            "completion_eligible",
            "original_anchor",
            anchor_column,
            "immutable_exclusion",
        ):
            if not completed[column].map(
                lambda value: isinstance(value, (bool, np.bool_))
            ).all():
                raise IngestError(f"Completion evidence {column!r} must be Boolean.")
        barred = completed["disposition"].isin(
            [DISPOSITION_REJECTED, DISPOSITION_CASE_EXCLUSION]
        )
        if bool(
            (pd.to_numeric(completed.loc[barred, VALUE_COLUMN], errors="coerce") > 0.0).any()
        ):
            raise IngestError("A barred primitive carries a positive completion input.")
        completed.loc[barred, VALUE_COLUMN] = np.nan
        completion_allowed = completed["completion_eligible"].astype(bool).to_numpy()
        original_anchor = completed[anchor_column].astype(bool).to_numpy()
    else:
        completion_allowed = np.ones(len(completed), dtype=bool)
        original_anchor = np.isfinite(
            pd.to_numeric(completed[VALUE_COLUMN], errors="coerce").to_numpy(dtype=float)
        )
        completed["completion_eligible"] = completion_allowed
        completed["original_anchor"] = original_anchor
        completed["completion_original_anchor"] = original_anchor
        completed["immutable_exclusion"] = False
    completed["completed_value_usd"] = pd.to_numeric(
        completed[VALUE_COLUMN], errors="coerce"
    ).astype(float)
    completed["was_completed"] = False
    completed["completion_method"] = ""
    completed["completion_window"] = [
        next(
            name
            for name, window_years in EXPOSURE_COMPLETION_WINDOWS.items()
            if int(year) in window_years
        )
        for year in completed["year"]
    ]
    # Register pooled, observed adjacent log-growth distributions by primitive.
    # These distributions bind edge rates to the full data rather than to the
    # very run being extrapolated.
    pooled_growth: Dict[str, List[float]] = {
        series: [] for series in sorted(set(completed["series"].astype(str)))
    }
    for (_reporter, series), group in completed.groupby(
        ["reporter_code", "series"], sort=True
    ):
        ordered = group.sort_values("year")
        for window_years in EXPOSURE_COMPLETION_WINDOWS.values():
            scoped = ordered[ordered["year"].isin(window_years)].sort_values("year")
            scoped_values = scoped[VALUE_COLUMN].to_numpy(dtype=float)
            scoped_anchors = scoped[anchor_column].astype(bool).to_numpy()
            scoped_years = scoped["year"].to_numpy(dtype=int)
            for left in range(len(scoped_years) - 1):
                right = left + 1
                if (
                    scoped_anchors[left]
                    and scoped_anchors[right]
                    and np.isfinite(scoped_values[left])
                    and np.isfinite(scoped_values[right])
                    and scoped_years[right] - scoped_years[left] == 1
                ):
                    pooled_growth[str(series)].append(
                        float(
                            np.log1p(scoped_values[right])
                            - np.log1p(scoped_values[left])
                        )
                    )
    ledger_rows: List[Dict[str, Any]] = []
    for (reporter, series), group in completed.groupby(
        ["reporter_code", "series"], sort=True
    ):
        for window_name, window_years in EXPOSURE_COMPLETION_WINDOWS.items():
            scoped = group[group["year"].isin(window_years)].sort_values("year")
            if tuple(scoped["year"].astype(int)) != tuple(window_years):
                raise IngestError(
                    f"{reporter}/{series}/{window_name}: incomplete four-year grid."
                )
            positions = scoped.index.to_numpy(dtype=int)
            run_years = scoped["year"].to_numpy(dtype=int)
            run_values = scoped[VALUE_COLUMN].to_numpy(dtype=float)
            interpolated, filled_internal = interpolate_internal_gaps(
                run_years, run_values, int(maximum_gap)
            )
            edge_anchor_values = np.where(
                scoped[anchor_column].astype(bool).to_numpy(),
                run_values,
                np.nan,
            )
            reference_values = np.asarray(
                pooled_growth.get(str(series), []), dtype=float
            )
            edge_completed, filled_edge, edge_anchors, edge_audit = (
                _edge_extrapolation_calculation(
                    run_years,
                    edge_anchor_values,
                    int(minimum_anchors),
                    GROWTH_PERCENTILE_LOWER,
                    GROWTH_PERCENTILE_UPPER,
                    reference_values if reference_values.size else None,
                )
            )
            generated = run_values.copy()
            generated[filled_internal] = interpolated[filled_internal]
            generated[filled_edge] = edge_completed[filled_edge]
            allowed = completion_allowed[positions]
            disallowed = (filled_internal | filled_edge) & ~allowed
            generated[disallowed] = np.nan
            filled_internal[disallowed] = False
            filled_edge[disallowed] = False
            if bool(
                (np.isfinite(generated) & (generated < 0.0)).any()
            ):
                raise IngestError(
                    f"{reporter}/{series}/{window_name}: completion produced a negative amount."
                )
            completed.loc[positions, "completed_value_usd"] = generated
            completed.loc[positions[filled_internal | filled_edge], "was_completed"] = True
            completed.loc[positions[filled_internal], "completion_method"] = METHOD_INTERPOLATION
            completed.loc[positions[filled_edge], "completion_method"] = METHOD_EXTRAPOLATION
            observed_positions = np.flatnonzero(np.isfinite(run_values))
            for local_position in np.flatnonzero(filled_internal):
                left_candidates = observed_positions[observed_positions < local_position]
                right_candidates = observed_positions[observed_positions > local_position]
                anchor_years = (
                    int(run_years[left_candidates[-1]]),
                    int(run_years[right_candidates[0]]),
                )
                ledger_rows.append(
                    {
                        "reporter_code": str(reporter),
                        "series": str(series),
                        "year": int(run_years[local_position]),
                        "window_id": str(window_name),
                        "method": METHOD_INTERPOLATION,
                        "anchor_count": 2,
                        "original_anchor_years": json.dumps(anchor_years),
                        "value_before": float("nan"),
                        "value_after": float(generated[local_position]),
                        "applied_log_rate": float("nan"),
                        "growth_lower_bound": float("nan"),
                        "growth_upper_bound": float("nan"),
                        "growth_reference_sha256": "",
                        "growth_reference_scope": "not_applicable",
                    }
                )
            for local_position in np.flatnonzero(filled_edge):
                ledger_rows.append(
                    {
                        "reporter_code": str(reporter),
                        "series": str(series),
                        "year": int(run_years[local_position]),
                        "window_id": str(window_name),
                        "method": METHOD_EXTRAPOLATION,
                        "anchor_count": int(edge_anchors),
                        "original_anchor_years": json.dumps(
                            tuple(edge_audit["anchor_years"])
                        ),
                        "value_before": float("nan"),
                        "value_after": float(generated[local_position]),
                        "applied_log_rate": float(edge_audit["applied_log_rate"]),
                        "growth_lower_bound": float(edge_audit["growth_lower_bound"]),
                        "growth_upper_bound": float(edge_audit["growth_upper_bound"]),
                        "growth_reference_sha256": str(
                            edge_audit["growth_reference_sha256"]
                        ),
                        "growth_reference_scope": str(
                            edge_audit["growth_reference_scope"]
                        ),
                    }
                )
    ledger_columns = [
        "reporter_code",
        "series",
        "year",
        "window_id",
        "method",
        "anchor_count",
        "original_anchor_years",
        "value_before",
        "value_after",
        "applied_log_rate",
        "growth_lower_bound",
        "growth_upper_bound",
        "growth_reference_sha256",
        "growth_reference_scope",
    ]
    ledger = pd.DataFrame(ledger_rows, columns=ledger_columns).set_index(
        ["reporter_code", "series", "year"]
    )
    if bool(ledger.index.duplicated().any()):
        raise IngestError("A primitive cell was completed more than once.")
    if len(ledger):
        if bool((ledger["value_after"] < 0.0).any()) or bool(
            ~np.isfinite(ledger["value_after"]).all()
        ):
            raise IngestError("The completion ledger contains an invalid generated value.")
        completed_lookup = completed.set_index(
            ["reporter_code", "series", "year"]
        )["completed_value_usd"]
        if not np.allclose(
            completed_lookup.reindex(ledger.index).to_numpy(dtype=float),
            ledger["value_after"].to_numpy(dtype=float),
            rtol=0.0,
            atol=0.0,
        ):
            raise IngestError("Completion ledger values disagree with the completed grid.")
    completed["value_provenance"] = np.where(
        completed["was_completed"].astype(bool),
        completed["completion_method"].astype(str),
        np.where(
            completed.get("recovery_realized", pd.Series(False, index=completed.index))
            .fillna(False)
            .astype(bool),
            MECHANISM_MIRROR,
            np.where(
                np.isfinite(completed[VALUE_COLUMN].to_numpy(dtype=float)),
                MECHANISM_IMPORTER,
                MECHANISM_ABSENT,
            ),
        ),
    )
    if has_task25_evidence:
        assert_rejected_immutability(completed)
    logger.info(
        "within-window completion applied to %d cell(s): %s.",
        len(ledger),
        ledger["method"].value_counts().to_dict() if len(ledger) else {},
    )
    return completed.sort_values(list(LEDGER_KEY), kind="stable").reset_index(drop=True), ledger


# Compose the per-series completion counts.
def build_completion_counts(
    completed: pd.DataFrame,
    ledger: pd.DataFrame,
) -> pd.DataFrame:
    """
    Compose the per-series completion counts.

    Parameters
    ----------
    completed : pandas.DataFrame
        The completed primitives.
    ledger : pandas.DataFrame
        The completion ledger.

    Returns
    -------
    pandas.DataFrame
        One row per series carrying the number of completed cells, the number
        left missing and the number of anchors drawn on, which the
        specification requires so the counts can be compared with the
        manuscript's imputation counts.

    Raises
    ------
    IngestError
        If the completed frame is empty or lacks a required column.
    """
    # Refuse an empty frame.
    if len(completed) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot count an empty completed frame.")
    # Refuse a required column's absence.
    if "series" not in completed.columns:
        # The failure names the absent column.
        raise IngestError("The completed frame lacks the series column.")
    # Compose the per-series completion counts from the ledger.
    completed_counts = (
        ledger.reset_index().groupby("series").size()
        if len(ledger)
        else pd.Series(dtype=int)
    )
    # Compose the per-series remaining missing counts.
    remaining = completed.assign(
        still_missing=~np.isfinite(
            pd.to_numeric(completed["completed_value_usd"], errors="coerce").to_numpy(dtype=float)
        )
    )
    # Compose the remaining counts.
    remaining_counts = remaining.groupby("series")["still_missing"].sum()
    # Compose the per-series observation counts.
    observed_counts = remaining.groupby("series").size()
    # Assemble the counts frame.
    counts: pd.DataFrame = pd.DataFrame(
        {
            "cells": observed_counts,
            "completed": completed_counts,
            "still_missing": remaining_counts,
        }
    ).fillna(0)
    # Cast the counts to integers.
    counts = counts.astype(int)
    # Log the counts' extent.
    logger.info("completion counts composed across %d series.", len(counts))
    # Return the counts.
    return counts


# Freeze the construction behind a canonical digest.
def freeze_exposure_construction(
    recovered: pd.DataFrame,
    denominator: pd.DataFrame,
    completed: pd.DataFrame,
    ledger: pd.DataFrame,
) -> str:
    """
    Freeze the construction behind a canonical digest.

    Parameters
    ----------
    recovered : pandas.DataFrame
        The recovered ledger.
    denominator : pandas.DataFrame
        The denominator frame.
    completed : pandas.DataFrame
        The completed primitives.
    ledger : pandas.DataFrame
        The completion ledger.

    Returns
    -------
    str
        The hexadecimal digest of the canonical payload, computed over the
        sorted keys so the digest is invariant to construction order.

    Raises
    ------
    IngestError
        If the recovered or completed frame is empty.
    """
    for label, frame in (("recovered", recovered), ("completed", completed)):
        if len(frame) == 0:
            raise IngestError(f"Cannot freeze an empty {label} frame.")
    if bool(recovered.duplicated(list(LEDGER_KEY)).any()):
        raise IngestError("Recovered construction cells are not uniquely keyed.")
    if bool(completed.duplicated(["reporter_code", "year", "series"]).any()):
        raise IngestError("Completed construction cells are not uniquely keyed.")
    if not isinstance(denominator.index, pd.MultiIndex) or list(
        denominator.index.names
    ) != ["reporter_code", "year"] or bool(denominator.index.duplicated().any()):
        raise IngestError("Denominator construction keys are malformed or duplicated.")
    if not isinstance(ledger.index, pd.MultiIndex) or list(ledger.index.names) != [
        "reporter_code",
        "series",
        "year",
    ] or bool(ledger.index.duplicated().any()):
        raise IngestError("Completion-ledger keys are malformed or duplicated.")
    recovered_keys = set(recovered[list(LEDGER_KEY)].itertuples(index=False, name=None))
    completed_trade = completed[completed["series"].isin(PRIMITIVE_SERIES)]
    completed_keys = set(
        completed_trade[list(LEDGER_KEY)].itertuples(index=False, name=None)
    )
    if recovered_keys != completed_keys:
        raise IngestError(
            "Recovered and completed trade cells do not share one exact key universe."
        )
    russian_keys = {
        (str(row.reporter_code), int(row.year))
        for row in recovered[recovered["series"] == "russian_gas"]
        .loc[:, ["reporter_code", "year"]]
        .itertuples(index=False)
    }
    if set(denominator.index.tolist()) != russian_keys:
        raise IngestError(
            "The denominator key universe is not the recovered Russian-gas universe."
        )
    if not set(ledger.index).issubset(
        set(
            completed_trade[["reporter_code", "series", "year"]]
            .itertuples(index=False, name=None)
        )
    ):
        raise IngestError("Completion-ledger cells are absent from the completed grid.")
    payload = {
        "recovered": _content_digest(
            recovered.sort_values(list(LEDGER_KEY), kind="stable").reset_index(drop=True)
        ),
        "denominator": _content_digest(denominator.sort_index().reset_index()),
        "completed": _content_digest(
            completed.sort_values(
                ["reporter_code", "series", "year"], kind="stable"
            ).reset_index(drop=True)
        ),
        "imputation_ledger": _content_digest(ledger.sort_index().reset_index()),
    }
    return hashlib.sha256(
        json.dumps(payload, sort_keys=True, separators=(",", ":")).encode("utf-8")
    ).hexdigest()


# -------------------------------------------------------------------------------------------------------------------------------
# Task 26, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Construct the exposure primitives from the ledger and the raw blocks.
def construct_exposure_primitives(
    ledger: pd.DataFrame,
    mirror_block: pd.DataFrame,
    numerator: pd.DataFrame,
    totals: pd.DataFrame,
    primitives: pd.DataFrame,
    exclusions: Optional[Any] = None,
    barrier: Tuple[int, int] = WINDOW_BARRIER,
    maximum_gap: int = MAXIMUM_INTERNAL_GAP,
    minimum_anchors: int = MINIMUM_ANCHORS,
    expected_completions: Optional[Mapping[str, int]] = None,
    tolerance: int = 0,
    nominal_gdp: Optional[pd.DataFrame] = None,
) -> ExposureConstructionResult:
    """
    Construct the exposure primitives from the ledger and the raw blocks.

    Parameters
    ----------
    ledger : pandas.DataFrame
        The classified and disposed ledger.
    mirror_block : pandas.DataFrame
        The raw block carrying the exporter's mirror rows.
    numerator : pandas.DataFrame
        One audit row per reporter and year carrying the originally supplied
        Russian flow. The scientific numerator is re-derived from ``ledger``
        plus ``mirror_block`` and this frame must reconcile to that lineage.
    totals : pandas.DataFrame
        One row per reporter and year carrying the importer and exporter totals;
        optional Block C provenance fields are retained in the denominator.
    primitives : pandas.DataFrame
        One audit row per reporter, year and trade series carrying the raw
        observed value; its exact keys and values must reconcile to ``ledger``.
    exclusions : pandas.DataFrame or sequence of tuple, optional
        The canonical priority exclusion audit or its legacy pair/triple form.
    barrier : tuple of int, optional
        The hard barrier between the data windows.
    maximum_gap : int, optional
        The longest internal gap the interpolation fills.
    minimum_anchors : int, optional
        The number of anchors an edge extrapolation requires.
    expected_completions : mapping of str to int, optional
        The declared completion counts per series, compared only when supplied.
    tolerance : int, optional
        The admissible difference between the observed and declared counts.
    nominal_gdp : pandas.DataFrame, optional
        Authoritative ``NY.GDP.MKTP.CD`` rows keyed by country/year. Production
        callers must supply this frame; for backward compatibility, an explicit
        ``gdp`` series embedded in ``primitives`` is also recognized.

    Returns
    -------
    ExposureConstructionResult
        The recovered ledger, the denominator, the completed primitives, the
        completion ledger, the counts, the digest, the timestamp and the notes.

    Raises
    ------
    IngestError
        If any recovery, construction, completion or comparison check fails.
    """
    if isinstance(tolerance, (bool, np.bool_)) or not isinstance(
        tolerance, (int, np.integer)
    ) or int(tolerance) < 0:
        raise IngestError("Completion-count tolerance must be a nonnegative integer.")
    # Recover the registered gas products first; every later input is reconciled
    # to this Task 25/26 lineage rather than trusted as an independent shortcut.
    recovered = recover_mirror_flow(ledger, mirror_block, exclusions)
    canonical_ledger = recovered.copy(deep=True)
    canonical_ledger["series"] = canonical_ledger["series"].map(
        _canonical_trade_series
    )
    if bool(canonical_ledger.duplicated(list(LEDGER_KEY)).any()):
        raise IngestError("Task 25/26 recovered cells are not uniquely keyed.")
    recovered_keys = set(
        canonical_ledger[list(LEDGER_KEY)].itertuples(index=False, name=None)
    )
    # Separate an optional embedded GDP compatibility input before canonicalizing
    # the three trade primitives.
    supplied_primitives = primitives.copy(deep=True)
    if not {"reporter_code", "year", "series", VALUE_COLUMN}.issubset(
        supplied_primitives.columns
    ):
        raise IngestError("The supplied primitive audit lacks its declared columns.")
    normalized_series = supplied_primitives["series"].astype(str).map(
        lambda value: " ".join(
            value.strip().lower().replace("-", " ").replace("_", " ").split()
        )
    )
    embedded_gdp = supplied_primitives[
        normalized_series.isin({"gdp", "nominal gdp"})
    ].copy()
    supplied_trade = supplied_primitives[
        ~normalized_series.isin({"gdp", "nominal gdp"})
    ].copy()
    supplied_trade["reporter_code"] = (
        supplied_trade["reporter_code"].astype(str).str.strip().str.upper()
    )
    supplied_trade["series"] = supplied_trade["series"].map(
        _canonical_trade_series
    )
    trade_years = pd.to_numeric(supplied_trade["year"], errors="coerce")
    if bool(trade_years.isna().any()) or bool((trade_years % 1.0 != 0.0).any()):
        raise IngestError("The supplied primitive audit carries malformed years.")
    supplied_trade["year"] = trade_years.astype(int)
    if bool(supplied_trade.duplicated(list(LEDGER_KEY)).any()):
        raise IngestError("The supplied primitive audit carries duplicate keys.")
    supplied_values = pd.to_numeric(supplied_trade[VALUE_COLUMN], errors="coerce")
    malformed_values = supplied_values.isna() & ~supplied_trade[VALUE_COLUMN].isna()
    if bool(malformed_values.any()) or bool(
        np.isinf(supplied_values.to_numpy(dtype=float)).any()
    ) or bool((supplied_values.dropna() < 0.0).any()):
        raise IngestError("The supplied primitive audit carries invalid amounts.")
    supplied_trade[VALUE_COLUMN] = supplied_values.astype(float)
    supplied_keys = set(
        supplied_trade[list(LEDGER_KEY)].itertuples(index=False, name=None)
    )
    if supplied_keys != recovered_keys:
        raise IngestError(
            "The supplied primitive audit does not match the Task 25 cell universe."
        )
    original_lookup = canonical_ledger.set_index(list(LEDGER_KEY))[VALUE_COLUMN]
    supplied_lookup = supplied_trade.set_index(list(LEDGER_KEY))[VALUE_COLUMN]
    original_aligned = pd.to_numeric(
        original_lookup.reindex(supplied_lookup.index), errors="coerce"
    ).to_numpy(dtype=float)
    supplied_aligned = supplied_lookup.to_numpy(dtype=float)
    primitive_agreement = (
        (np.isnan(original_aligned) & np.isnan(supplied_aligned))
        | (
            np.isfinite(original_aligned)
            & np.isfinite(supplied_aligned)
            & (original_aligned == supplied_aligned)
        )
    )
    if not bool(primitive_agreement.all()):
        raise IngestError(
            "The supplied primitive audit is disconnected from Task 25 values."
        )
    # Derive the numerator from importer-priority/mirror-corrected Russian gas.
    russian_rows = canonical_ledger[
        canonical_ledger["series"] == "russian_gas"
    ].copy()
    if len(russian_rows) == 0:
        raise IngestError("The recovered ledger carries no Russian-gas cells.")
    derived_numerator = russian_rows[
        ["reporter_code", "year", "effective_value_usd", "recovered_via"]
    ].rename(columns={"effective_value_usd": "russian_flow_usd"})
    derived_numerator["numerator_source"] = derived_numerator[
        "recovered_via"
    ].astype(str)
    derived_numerator["recovered_ledger_sha256"] = _content_digest(
        canonical_ledger.sort_values(list(LEDGER_KEY), kind="stable")
        .reset_index(drop=True)
    )
    # Provenance-check the legacy numerator input against either the uncorrected
    # importer value or the just-derived corrected value, then discard it from
    # the scientific calculation.
    required_numerator = {"reporter_code", "year", "russian_flow_usd"}
    if not required_numerator.issubset(numerator.columns):
        raise IngestError("The supplied numerator audit lacks declared columns.")
    supplied_numerator = numerator.copy(deep=True)
    supplied_numerator["reporter_code"] = (
        supplied_numerator["reporter_code"].astype(str).str.strip().str.upper()
    )
    numerator_years = pd.to_numeric(supplied_numerator["year"], errors="coerce")
    if bool(numerator_years.isna().any()) or bool(
        (numerator_years % 1.0 != 0.0).any()
    ):
        raise IngestError("The supplied numerator audit carries malformed years.")
    supplied_numerator["year"] = numerator_years.astype(int)
    if bool(supplied_numerator.duplicated(["reporter_code", "year"]).any()):
        raise IngestError("The supplied numerator audit carries duplicate keys.")
    supplied_num_values = pd.to_numeric(
        supplied_numerator["russian_flow_usd"], errors="coerce"
    )
    if bool(
        (supplied_num_values.isna() & ~supplied_numerator["russian_flow_usd"].isna()).any()
    ) or bool(np.isinf(supplied_num_values.to_numpy(dtype=float)).any()) or bool(
        (supplied_num_values.dropna() < 0.0).any()
    ):
        raise IngestError("The supplied numerator audit carries invalid amounts.")
    supplied_numerator["russian_flow_usd"] = supplied_num_values.astype(float)
    numerator_keys = set(
        supplied_numerator[["reporter_code", "year"]].itertuples(
            index=False, name=None
        )
    )
    derived_numerator_keys = set(
        derived_numerator[["reporter_code", "year"]].itertuples(
            index=False, name=None
        )
    )
    if numerator_keys != derived_numerator_keys:
        raise IngestError("The supplied numerator audit has the wrong key universe.")
    supplied_num_lookup = supplied_numerator.set_index(
        ["reporter_code", "year"]
    )["russian_flow_usd"]
    raw_num_lookup = russian_rows.set_index(["reporter_code", "year"])[VALUE_COLUMN]
    corrected_num_lookup = derived_numerator.set_index(
        ["reporter_code", "year"]
    )["russian_flow_usd"]
    for key, supplied_value in supplied_num_lookup.items():
        candidates = (raw_num_lookup.loc[key], corrected_num_lookup.loc[key])
        if pd.isna(supplied_value):
            agrees = any(pd.isna(candidate) for candidate in candidates)
        else:
            agrees = any(
                pd.notna(candidate) and float(candidate) == float(supplied_value)
                for candidate in candidates
            )
        if not agrees:
            raise IngestError(
                f"The supplied numerator value is disconnected at {key}."
            )
    # Verify that totals keys/observed world totals belong to the same ledger.
    required_totals = {
        "reporter_code",
        "year",
        "world_total_usd",
        "exporter_sum_usd",
    }
    if not required_totals.issubset(totals.columns):
        raise IngestError("The supplied totals audit lacks declared columns.")
    supplied_totals = totals.copy(deep=True)
    supplied_totals["reporter_code"] = (
        supplied_totals["reporter_code"].astype(str).str.strip().str.upper()
    )
    totals_years = pd.to_numeric(supplied_totals["year"], errors="coerce")
    if bool(totals_years.isna().any()) or bool((totals_years % 1.0 != 0.0).any()):
        raise IngestError("The supplied totals audit carries malformed years.")
    supplied_totals["year"] = totals_years.astype(int)
    if bool(supplied_totals.duplicated(["reporter_code", "year"]).any()):
        raise IngestError("The supplied totals audit carries duplicate keys.")
    if not set(
        supplied_totals[["reporter_code", "year"]].itertuples(
            index=False, name=None
        )
    ).issubset(derived_numerator_keys):
        raise IngestError("The supplied totals audit contains foreign reporter-years.")
    world_rows = canonical_ledger[canonical_ledger["series"] == "world_gas"]
    world_keys = set(
        world_rows[["reporter_code", "year"]].itertuples(index=False, name=None)
    )
    nongas_keys = set(
        canonical_ledger[canonical_ledger["series"] == "russian_non_gas"]
        [["reporter_code", "year"]]
        .itertuples(index=False, name=None)
    )
    if world_keys != derived_numerator_keys or nongas_keys != derived_numerator_keys:
        raise IngestError(
            "The three primitive series do not share one reporter-year universe."
        )
    world_lookup = world_rows.set_index(["reporter_code", "year"])[VALUE_COLUMN]
    for _, row in supplied_totals.iterrows():
        key = (str(row["reporter_code"]), int(row["year"]))
        supplied_world = pd.to_numeric(
            pd.Series([row["world_total_usd"]]), errors="coerce"
        ).iloc[0]
        original_world = world_lookup.loc[key]
        if not (
            (pd.isna(supplied_world) and pd.isna(original_world))
            or (
                pd.notna(supplied_world)
                and pd.notna(original_world)
                and float(supplied_world) == float(original_world)
            )
        ):
            raise IngestError(
                f"The supplied world total is disconnected at {key}."
            )
    # Re-derive importer totals from the screened ledger over its complete key
    # universe.  The totals block contributes only the independently computed
    # exporter-sum alternative, so an omitted auxiliary row cannot erase a
    # valid importer observation already established by Task 25.
    exporter_alternatives = supplied_totals.set_index(
        ["reporter_code", "year"]
    )["exporter_sum_usd"]
    derived_totals = world_rows[["reporter_code", "year", VALUE_COLUMN]].rename(
        columns={VALUE_COLUMN: "world_total_usd"}
    )
    derived_totals["exporter_sum_usd"] = [
        exporter_alternatives.get((str(country), int(year)), np.nan)
        for country, year in derived_totals[["reporter_code", "year"]].itertuples(
            index=False, name=None
        )
    ]
    derived_totals["exporter_sum_input_present"] = [
        pd.notna(exporter_alternatives.get((str(country), int(year)), np.nan))
        for country, year in derived_totals[["reporter_code", "year"]].itertuples(
            index=False, name=None
        )
    ]
    # Retain the ten-source provenance of each independently derived exporter
    # alternative in the denominator audit, including absent Block C rows.
    for provenance_column in (
        "block_c_importer_row_present", "block_c_source_sha256",
        "block_c_recovered_via_claim",
        "exporter_sum_source_id", "exporter_sum_components_sha256",
        "exporter_sum_query_sha256",
    ):
        if provenance_column in supplied_totals.columns:
            source_lookup = supplied_totals.set_index(
                ["reporter_code", "year"]
            )[provenance_column]
            derived_totals[provenance_column] = [
                source_lookup.get((str(country), int(year)), np.nan)
                for country, year in derived_totals[["reporter_code", "year"]]
                .itertuples(index=False, name=None)
            ]
    derived_totals["task25_world_ledger_sha256"] = _content_digest(
        world_rows.sort_values(["reporter_code", "year"], kind="stable")
        .reset_index(drop=True)
    )
    denominator = construct_denominator(derived_numerator, derived_totals)
    if set(denominator.index) != derived_numerator_keys:
        raise IngestError("The denominator did not preserve the primitive universe.")
    # Construct the only completion input from the recovered ledger, then replace
    # world-gas values with the audited denominator selections.
    primitive_input = canonical_ledger.copy(deep=True)
    primitive_input["source_value_usd"] = primitive_input["effective_value_usd"]
    primitive_input[VALUE_COLUMN] = primitive_input["effective_value_usd"]
    primitive_input["construction_source"] = primitive_input[
        "recovered_via"
    ].astype(str)
    denominator_values = denominator["retained_total_usd"]
    denominator_mechanism = denominator["substitution_mechanism"].astype(str)
    denominator_reason = denominator["denominator_selection_reason"].astype(str)
    world_mask = primitive_input["series"].eq("world_gas")
    world_index = pd.MultiIndex.from_frame(
        primitive_input.loc[world_mask, ["reporter_code", "year"]]
    )
    primitive_input.loc[world_mask, VALUE_COLUMN] = denominator_values.reindex(
        world_index
    ).to_numpy(dtype=float)
    primitive_input.loc[world_mask, "construction_source"] = (
        denominator_mechanism.reindex(world_index).to_numpy(dtype=object)
    )
    primitive_input.loc[world_mask, "denominator_selection_reason"] = (
        denominator_reason.reindex(world_index).to_numpy(dtype=object)
    )
    corrected_observation = (
        primitive_input["original_anchor"].astype(bool)
        | primitive_input["recovery_realized"].astype(bool)
        | (
            world_mask
            & np.isfinite(
                pd.to_numeric(
                    primitive_input[VALUE_COLUMN], errors="coerce"
                ).to_numpy(dtype=float)
            )
        )
    )
    primitive_input["completion_original_anchor"] = corrected_observation.astype(bool)
    completed_trade, completion_ledger = complete_within_window(
        primitive_input,
        barrier=barrier,
        maximum_gap=maximum_gap,
        minimum_anchors=minimum_anchors,
    )
    assert_rejected_immutability(completed_trade)
    # D4 completes each primitive independently.  A completed world-gas
    # denominator must therefore be checked again against its paired Russian
    # gas numerator: the pre-completion denominator audit cannot see an
    # imputed value.  Missing components remain missing for the later sample
    # disposition; an observed or completed contradiction is never masked.
    completed_values = completed_trade.set_index(list(LEDGER_KEY))[
        "completed_value_usd"
    ]
    completed_flags = completed_trade.set_index(list(LEDGER_KEY))["was_completed"]
    for country, year in sorted(derived_numerator_keys):
        numerator_key = (country, year, "russian_gas")
        denominator_key = (country, year, "world_gas")
        completed_numerator = float(completed_values.loc[numerator_key])
        completed_denominator = float(completed_values.loc[denominator_key])
        if (
            np.isfinite(completed_numerator)
            and np.isfinite(completed_denominator)
            and not (0.0 <= completed_numerator <= completed_denominator)
        ):
            raise IngestError(
                f"Completed Russian-gas flow exceeds or invalidates the "
                f"world-gas total at {(country, year)}: "
                f"numerator={completed_numerator}, "
                f"denominator={completed_denominator}, "
                f"numerator_completed={bool(completed_flags.loc[numerator_key])}, "
                f"denominator_completed={bool(completed_flags.loc[denominator_key])}."
            )
    counts = build_completion_counts(completed_trade, completion_ledger)
    # Resolve and validate authoritative nominal GDP for the exact completed
    # country/year universe. GDP is appended only after trade completion and is
    # never interpolated or substituted.
    gdp_source = nominal_gdp.copy(deep=True) if nominal_gdp is not None else None
    if gdp_source is None and len(embedded_gdp):
        gdp_source = embedded_gdp.rename(columns={"reporter_code": "country_code"})
    if gdp_source is None:
        raise IngestError(
            "Authoritative nominal GDP is required for downstream exposure ratios."
        )
    if "country_code" not in gdp_source.columns and "reporter_code" in gdp_source.columns:
        gdp_source = gdp_source.rename(columns={"reporter_code": "country_code"})
    required_gdp = {"country_code", "year", "indicator_code", VALUE_COLUMN}
    if not required_gdp.issubset(gdp_source.columns):
        raise IngestError(
            "Nominal GDP lacks authoritative country/year/indicator/value fields."
        )
    if bool(gdp_source["indicator_code"].isna().any()) or not bool(
        gdp_source["indicator_code"].astype(str).eq(NOMINAL_GDP_INDICATOR).all()
    ):
        raise IngestError(
            f"Nominal GDP must use only {NOMINAL_GDP_INDICATOR}."
        )
    if bool(gdp_source["country_code"].isna().any()):
        raise IngestError("Nominal GDP carries a null country code.")
    gdp_source["country_code"] = (
        gdp_source["country_code"].astype(str).str.strip().str.upper()
    )
    if not bool(gdp_source["country_code"].str.fullmatch(r"[A-Z]{3}").all()):
        raise IngestError("Nominal GDP country codes must be nonblank ISO-3 labels.")
    if bool(
        gdp_source["year"].map(
            lambda value: isinstance(value, (bool, np.bool_))
        ).any()
    ):
        raise IngestError("Nominal GDP years cannot be Boolean.")
    gdp_years = pd.to_numeric(gdp_source["year"], errors="coerce")
    if bool(gdp_years.isna().any()) or bool((gdp_years % 1.0 != 0.0).any()):
        raise IngestError("Nominal GDP carries malformed calendar years.")
    gdp_source["year"] = gdp_years.astype(int)
    if bool(gdp_source.duplicated(["country_code", "year"]).any()):
        raise IngestError("Nominal GDP carries duplicate country-year keys.")
    boolean_gdp_values = gdp_source[VALUE_COLUMN].map(
        lambda value: isinstance(value, (bool, np.bool_))
    )
    gdp_values = pd.to_numeric(gdp_source[VALUE_COLUMN], errors="coerce")
    if bool(boolean_gdp_values.any()) or bool(gdp_values.isna().any()) or bool(
        ~np.isfinite(gdp_values.to_numpy(dtype=float)).all()
    ) or bool((gdp_values <= 0.0).any()):
        raise IngestError("Nominal GDP must be finite and strictly positive.")
    gdp_source[VALUE_COLUMN] = gdp_values.astype(float)
    required_gdp_keys = derived_numerator_keys
    gdp_lookup = gdp_source.set_index(["country_code", "year"])[VALUE_COLUMN]
    missing_gdp = sorted(required_gdp_keys - set(gdp_lookup.index))
    if missing_gdp:
        raise IngestError(
            f"Nominal GDP omits required country-year(s) {missing_gdp[:8]}."
        )
    gdp_digest = _content_digest(
        gdp_source.sort_values(["country_code", "year"], kind="stable")
        .reset_index(drop=True)
    )
    gdp_rows = pd.DataFrame(
        [
            {
                "reporter_code": country,
                "year": int(year),
                "series": "gdp",
                VALUE_COLUMN: float(gdp_lookup.loc[(country, year)]),
                "completed_value_usd": float(gdp_lookup.loc[(country, year)]),
                "original_anchor": True,
                "completion_original_anchor": True,
                "completion_eligible": False,
                "immutable_exclusion": False,
                "was_completed": False,
                "completion_method": "",
                "completion_window": next(
                    name
                    for name, window_years in EXPOSURE_COMPLETION_WINDOWS.items()
                    if int(year) in window_years
                ),
                "value_provenance": "authoritative_nominal_gdp",
                "construction_source": NOMINAL_GDP_INDICATOR,
                "gdp_source_sha256": gdp_digest,
            }
            for country, year in sorted(required_gdp_keys)
        ]
    )
    completed = pd.concat([completed_trade, gdp_rows], ignore_index=True, sort=False)
    if bool(completed.duplicated(["reporter_code", "year", "series"]).any()):
        raise IngestError("The GDP join duplicated a completed component key.")
    # Reconcile optional fixed-vintage completion counts without applying a
    # hidden default to synthetic/ad hoc fixtures.
    if expected_completions is not None:
        for raw_series, declared in expected_completions.items():
            series = _canonical_trade_series(raw_series)
            if isinstance(declared, (bool, np.bool_)) or not isinstance(
                declared, (int, np.integer)
            ) or int(declared) < 0:
                raise IngestError("Declared completion counts must be nonnegative integers.")
            if series not in counts.index:
                raise IngestError(f"The counts omit the declared series {series!r}.")
            observed = int(counts.loc[series, "completed"])
            if abs(observed - int(declared)) > int(tolerance):
                raise IngestError(
                    f"{series}: {observed} completion(s) against the declared "
                    f"{int(declared)} +/- {int(tolerance)}."
                )
    lineage_payload = {
        "task25_recovered": _content_digest(
            recovered.sort_values(list(LEDGER_KEY), kind="stable").reset_index(drop=True)
        ),
        "mirror_block": _content_digest(mirror_block.reset_index(drop=True)),
        "supplied_numerator": _content_digest(
            supplied_numerator.sort_values(["reporter_code", "year"], kind="stable")
            .reset_index(drop=True)
        ),
        "supplied_totals": _content_digest(
            supplied_totals.sort_values(["reporter_code", "year"], kind="stable")
            .reset_index(drop=True)
        ),
        "supplied_primitives": _content_digest(
            supplied_trade.sort_values(list(LEDGER_KEY), kind="stable")
            .reset_index(drop=True)
        ),
        "nominal_gdp": gdp_digest,
    }
    lineage_sha256 = hashlib.sha256(
        json.dumps(lineage_payload, sort_keys=True, separators=(",", ":")).encode(
            "utf-8"
        )
    ).hexdigest()
    recovered["construction_lineage_sha256"] = lineage_sha256
    denominator["construction_lineage_sha256"] = lineage_sha256
    completed["construction_lineage_sha256"] = lineage_sha256
    completion_ledger["construction_lineage_sha256"] = lineage_sha256
    fingerprint = freeze_exposure_construction(
        recovered, denominator, completed, completion_ledger
    )
    notes = (
        "registered positive Russian exporter LNG/pipeline rows are aggregated "
        "per partner-year; importer observations retain precedence and the "
        "no-recovery path remains explicit.",
        "the corrected Russian-gas numerator drives all three denominator "
        "substitution triggers; the selected denominator then replaces the "
        "world-gas primitive in the single completion dataflow.",
        "completion uses exact 2010-2013 and 2018-2021 grids, pooled registered "
        "growth bounds, nonnegative generated values and immutable exclusions.",
        f"authoritative nominal GDP {NOMINAL_GDP_INDICATOR} is joined without "
        f"completion; source digest {gdp_digest}.",
        "completed trade cells per series: "
        + str(counts["completed"].to_dict() if len(counts) else {})
        + ".",
        f"construction lineage: {lineage_sha256}.",
    )
    result = ExposureConstructionResult(
        recovered=recovered,
        denominator=denominator,
        completed=completed,
        imputation_ledger=completion_ledger,
        counts=counts,
        fingerprint_sha256=fingerprint,
        built_utc=current_utc_iso(),
        notes=notes,
    )
    logger.info(
        "exposure primitives complete: %d denominator row(s), %d completion(s), "
        "fingerprint %s.",
        len(denominator),
        len(completion_ledger),
        fingerprint[:16],
    )
    return result
````

````python
# Task 27: Exposure Ratio Construction and Sample Rules

# ==============================================================================
# Task 27: Exposure Ratio Construction and Sample Rules
# ==============================================================================

# Declare the ratio name of the Russian energy exposure.
RATIO_ENERGY: Final[str] = "energy_exposure"
# Declare the ratio name of the gas share.
RATIO_GAS_SHARE: Final[str] = "gas_share"
# Declare the disposition tag column of the energy ratio (remediation B-20).
RATIO_DISPOSITION_ENERGY: Final[str] = "energy_disposition"
# Declare the disposition tag column of the gas share (remediation B-20).
RATIO_DISPOSITION_GAS_SHARE: Final[str] = "gas_share_disposition"
# Compose the declared ratio names.
RATIO_NAMES: Final[Tuple[str, str]] = (RATIO_ENERGY, RATIO_GAS_SHARE)
# Declare the label of the strict sample rule.
RULE_STRICT: Final[str] = "strict"
# Declare the label of the extended sample rule.
RULE_EXTENDED: Final[str] = "extended"
# Compose the declared rule order.
SAMPLE_RULE_ORDER: Final[Tuple[str, str]] = (RULE_STRICT, RULE_EXTENDED)
# Declare the minimum usable years of the strict rule.
STRICT_MINIMUM_USABLE: Final[int] = 3
# Declare the maximum imputed years of the strict rule.
STRICT_MAXIMUM_IMPUTED: Final[int] = 1
# Declare the minimum usable years of the extended rule.
EXTENDED_MINIMUM_USABLE: Final[int] = 2
# Declare the maximum imputed years of the extended rule.
EXTENDED_MAXIMUM_IMPUTED: Final[int] = 2
# Compose the union of components used by the two independently sampled ratios.
REQUIRED_COMPONENTS: Final[Tuple[str, ...]] = (
    "russian_gas",
    "russian_non_gas",
    "world_gas",
    "gdp",
)
# Bind each exposure estimand to only the components it scientifically uses.
RATIO_COMPONENTS: Final[Mapping[str, Tuple[str, ...]]] = MappingProxyType(
    {
        RATIO_ENERGY: ("russian_gas", "russian_non_gas", "gdp"),
        RATIO_GAS_SHARE: ("russian_gas", "world_gas"),
    }
)
# Register the standard-deviation convention used by every regression sample.
EXPOSURE_SD_CONVENTION: Final[str] = "sample_sd_ddof_1"
# Compose the countries the mechanism universe omits, which are the targets and
# the implicit exclusions rather than participants.
ORDINATE_EXCLUSIONS: Final[Tuple[str, ...]] = ("RUS", "BLR", "UKR")
# Compose the declared size of the full country panel.
FULL_PANEL_SIZE: Final[int] = 129


# Declare the frozen result of the ratio construction.
@dataclass(frozen=True)
class ExposureRatioResult:
    """
    Frozen result of the exposure ratio construction.

    Attributes
    ----------
    ratios : pandas.DataFrame
        One row per country and window carrying both ratios.
    components : pandas.DataFrame
        One row per country, window, component and year carrying the yearly
        components every ratio was computed from, so the ratio is reproducible
        from the log without re-running the completion.
    sample_mask : pandas.DataFrame
        One row per country, window, rule and ratio carrying separate
        construction-coverage and mechanism-regression admission flags.
    standardized : pandas.DataFrame
        One row per country, window, rule and ratio carrying the standardized
        value, ddof=1 sample moments, convention and exact sample fingerprint.
    counts : pandas.DataFrame
        One row per window, rule and ratio carrying construction and
        mechanism-regression counts.
    fingerprint_sha256 : str
        Digest of the canonical construction payload.
    built_utc : str
        ISO-8601 UTC timestamp of the construction.
    notes : tuple of str
        Recorded decisions, including the windows and the universe size.
    """
    # The ratio panel.
    ratios: pd.DataFrame
    # The component log.
    components: pd.DataFrame
    # The sample mask.
    sample_mask: pd.DataFrame
    # The standardized values with their moments.
    standardized: pd.DataFrame
    # The per-window and rule counts.
    counts: pd.DataFrame
    # Digest of the canonical construction payload.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the construction.
    built_utc: str
    # Recorded decisions of the construction.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 27, Step 1: The ratio construction
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the window ratio of two completed series.
def compose_window_ratio(
    numerator_by_year: Mapping[int, float],
    denominator_by_year: Mapping[int, float],
    on_zero_denominator: str = "raise",
) -> float:
    """
    Compose the window ratio of two completed series.

    Parameters
    ----------
    numerator_by_year : mapping of int to float
        The numerator component of each window year.
    denominator_by_year : mapping of int to float
        The denominator component of each window year.
    on_zero_denominator : str, optional
        The zero-denominator policy: ``raise`` (the standalone default)
        refuses the cell, while ``mask`` returns a missing value so wide
        reshapes preserve their row counts without fabricating a zero. The
        stage-26 orchestrator explicitly selects ``mask`` (remediation B-20).

    Returns
    -------
    float
        The ratio of the window sums. The ratio is computed from the yearly
        components and never interpolated, which is the declared discipline:
        the sums run over the same years, so the result is a ratio of sums
        rather than a mean of yearly ratios.

    Raises
    ------
    IngestError
        If the two mappings carry different years, if a value is missing, or,
        under the strict ``raise`` policy, if the denominator sums to zero,
        because a ratio over a shorter window or over a zero denominator
        would not be the declared estimand. Under the ``mask`` policy a zero
        denominator yields a missing value recorded by the caller's
        disposition audit instead of a fabricated zero.
    """
    # Refuse an undeclared zero-denominator policy (remediation B-20).
    if on_zero_denominator not in ("raise", "mask"):
        # The failure names the received policy.
        raise IngestError(
            f"Undeclared zero-denominator policy {on_zero_denominator!r}."
        )
    # Validate year labels losslessly before comparing the two supports.
    for label, mapping in (
        ("numerator", numerator_by_year),
        ("denominator", denominator_by_year),
    ):
        for year in mapping:
            if isinstance(year, (bool, np.bool_)):
                raise IngestError(f"The {label} carries a Boolean year.")
            if not isinstance(year, (int, np.integer, float, np.floating)):
                raise IngestError(f"The {label} carries a nonnumeric year label.")
            try:
                numeric_year = float(year)
            except (TypeError, ValueError, OverflowError) as exc:
                raise IngestError(f"The {label} carries a malformed year.") from exc
            if not np.isfinite(numeric_year) or not numeric_year.is_integer():
                raise IngestError(f"The {label} carries a malformed year.")
    # Refuse mismatched year sets.
    if set(int(year) for year in numerator_by_year) != set(
        int(year) for year in denominator_by_year
    ):
        # The failure states the condition.
        raise IngestError("The ratio requires both components over the same years.")
    # Refuse an empty year set.
    if not numerator_by_year:
        # The failure states the precondition.
        raise IngestError("The ratio requires at least one year.")
    # Accumulate the numerator sum.
    numerator_sum: float = 0.0
    # Accumulate the denominator sum.
    denominator_sum: float = 0.0
    # Iterate over the years in sorted order.
    for year in sorted(int(value) for value in numerator_by_year):
        # Read the numerator of the year.
        if isinstance(numerator_by_year[year], (bool, np.bool_)) or isinstance(
            denominator_by_year[year], (bool, np.bool_)
        ):
            raise IngestError(f"Ratio components cannot be Boolean in {year}.")
        numerator = float(numerator_by_year[year])
        # Read the denominator of the year.
        denominator = float(denominator_by_year[year])
        # Refuse a missing numerator.
        if not np.isfinite(numerator):
            # The failure names the year.
            raise IngestError(f"The numerator is missing in {year}.")
        # Refuse a missing denominator.
        if not np.isfinite(denominator):
            # The failure names the year.
            raise IngestError(f"The denominator is missing in {year}.")
        if numerator < 0.0:
            raise IngestError(f"The numerator is negative in {year}.")
        if denominator < 0.0:
            raise IngestError(f"The denominator is negative in {year}.")
        # Add the numerator's contribution.
        numerator_sum += numerator
        # Add the denominator's contribution.
        denominator_sum += denominator
    # Handle a zero denominator according to the declared policy.
    if denominator_sum == 0.0:
        # Refuse the cell under the strict standalone policy.
        if on_zero_denominator == "raise":
            # The failure states the condition.
            raise IngestError(
                "The ratio's denominator sums to zero over the window."
            )
        # Return a missing value under the masked policy so wide reshapes
        # preserve their row counts without fabricating a zero (B-20).
        return float("nan")
    # Return the ratio of the sums.
    return numerator_sum / denominator_sum


# Resolve the audit disposition of one composed window ratio.
def _ratio_disposition(
    value: float,
    numerator_by_year: Mapping[int, float],
    denominator_by_year: Mapping[int, float],
) -> str:
    """
    Classify one composed ratio for the exposure audit.

    Parameters
    ----------
    value : float
        The composed ratio, which may be missing when the window is
        incomplete or masked.
    numerator_by_year : mapping of int to float
        The numerator component of each window year.
    denominator_by_year : mapping of int to float
        The denominator component of each window year.

    Returns
    -------
    str
        A diagnostic distinguishing a defined ratio (including a valid zero
        numerator), incomplete sides, a zero denominator, negative components
        and a supplied value inconsistent with its components.

    Raises
    ------
    None
        The function never raises; malformed inputs resolve to an explicit
        diagnostic rather than being mistaken for ordinary missingness.
    """
    # Validate the full labelled ratio state before assigning a disposition.
    try:
        if any(
            isinstance(year, (bool, np.bool_)) for year in numerator_by_year
        ) or any(isinstance(year, (bool, np.bool_)) for year in denominator_by_year):
            return "invalid_year_alignment"
        numerator_years = tuple(sorted(int(year) for year in numerator_by_year))
        denominator_years = tuple(sorted(int(year) for year in denominator_by_year))
        if not numerator_years or numerator_years != denominator_years:
            return "invalid_year_alignment"
        if isinstance(value, (bool, np.bool_)):
            return "invalid_component"
        numerator_values = np.asarray(
            [float(numerator_by_year[year]) for year in numerator_years], dtype=float
        )
        denominator_values = np.asarray(
            [float(denominator_by_year[year]) for year in denominator_years], dtype=float
        )
        observed_value = float(value)
    except (KeyError, TypeError, ValueError, OverflowError):
        return "invalid_component"
    if bool((numerator_values[np.isfinite(numerator_values)] < 0.0).any()):
        return "negative_numerator"
    if bool((denominator_values[np.isfinite(denominator_values)] < 0.0).any()):
        return "negative_denominator"
    if not bool(np.isfinite(numerator_values).all()):
        return "incomplete_numerator"
    if not bool(np.isfinite(denominator_values).all()):
        return "incomplete_denominator"
    denominator_sum = float(denominator_values.sum())
    if denominator_sum == 0.0:
        return "zero_denominator"
    expected_value = float(numerator_values.sum() / denominator_sum)
    if not np.isfinite(observed_value):
        return "undefined_despite_complete_components"
    if not np.isclose(observed_value, expected_value, rtol=0.0, atol=1.0e-12):
        return "inconsistent_ratio"
    # A zero numerator with a positive denominator is a valid defined ratio.
    return "defined"


# Build the exposure ratios and their component log.
def build_exposure_ratios(
    components: pd.DataFrame,
    windows: Mapping[str, Sequence[int]],
    on_zero_denominator: str = "raise",
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Build the exposure ratios and their component log.

    Parameters
    ----------
    components : pandas.DataFrame
        One row per country, series and year carrying the completed primitive
        or the reported product series, under the column ``value_usd``.
    windows : mapping of str to sequence of int
        The years of each window.
    on_zero_denominator : str, optional
        The zero-denominator policy passed to every ratio composition:
        ``raise`` (the standalone default) refuses the cell, while ``mask``
        records a missing value with an explicit disposition tag so wide
        reshapes preserve their row counts. Stage 26 selects ``mask``
        (remediation B-20).

    Returns
    -------
    tuple
        The ratio frame keyed on the country and the window, and the component
        log keyed on the country, the window, the component and the year.

    Raises
    ------
    IngestError
        If a required column is absent, if a window is empty, if a required
        component is absent for a country, or if the logged components fail to
        reproduce the ratio, which is the declared reconstructibility check.
    """
    # Refuse a components frame lacking a required column.
    for column in ("country_code", "series", "year", "value_usd"):
        # Test the column's presence.
        if column not in components.columns:
            # The failure names the absent column.
            raise IngestError(f"The components frame lacks the column {column!r}.")
    # Refuse an empty window mapping.
    if not windows:
        # The failure states the precondition.
        raise IngestError("At least one window is required.")
    # Refuse an undeclared zero-denominator policy (remediation B-20).
    if on_zero_denominator not in ("raise", "mask"):
        # The failure names the received policy.
        raise IngestError(
            f"Undeclared zero-denominator policy {on_zero_denominator!r}."
        )
    # Canonicalize and validate the component panel before any lookup can
    # collapse duplicate rows or coerce malformed scientific values.
    optional_provenance = [
        column
        for column in ("was_completed", "value_provenance", "original_anchor")
        if column in components.columns
    ]
    canonical = components[
        ["country_code", "series", "year", "value_usd"] + optional_provenance
    ].copy()
    if bool(canonical["country_code"].isna().any()):
        raise IngestError("Exposure components carry a null country code.")
    canonical["country_code"] = (
        canonical["country_code"].astype(str).str.strip().str.upper()
    )
    if bool((canonical["country_code"].str.len() == 0).any()):
        raise IngestError("Exposure components carry a blank country code.")
    canonical["series"] = canonical["series"].astype(str).str.strip()
    undeclared_series = sorted(set(canonical["series"]) - set(REQUIRED_COMPONENTS))
    if undeclared_series:
        raise IngestError(f"Exposure components carry undeclared series {undeclared_series}.")
    if bool(canonical["year"].map(lambda item: isinstance(item, (bool, np.bool_))).any()):
        raise IngestError("Exposure component years cannot be Boolean.")
    numeric_years = pd.to_numeric(canonical["year"], errors="coerce")
    if bool(numeric_years.isna().any()) or bool((numeric_years % 1.0 != 0.0).any()):
        raise IngestError("Exposure components carry malformed calendar years.")
    canonical["year"] = numeric_years.astype(int)
    boolean_values = canonical["value_usd"].map(
        lambda item: isinstance(item, (bool, np.bool_))
    )
    numeric_values = pd.to_numeric(canonical["value_usd"], errors="coerce")
    malformed_values = numeric_values.isna() & ~canonical["value_usd"].isna()
    if bool(boolean_values.any()) or bool(malformed_values.any()) or bool(
        np.isinf(numeric_values.to_numpy(dtype=float)).any()
    ) or bool((numeric_values.dropna() < 0.0).any()):
        raise IngestError(
            "Exposure component values must be missing or finite nonnegative amounts."
        )
    canonical["value_usd"] = numeric_values.astype(float)
    finite_gdp = canonical.loc[
        canonical["series"].eq("gdp") & canonical["value_usd"].notna(),
        "value_usd",
    ]
    if bool((finite_gdp <= 0.0).any()):
        raise IngestError("Nominal GDP components must be strictly positive.")
    if bool(canonical.duplicated(["country_code", "series", "year"]).any()):
        raise IngestError("Exposure components carry duplicate country/series/year keys.")
    for flag in ("was_completed", "original_anchor"):
        if flag in canonical.columns and not bool(
            canonical[flag].map(lambda item: isinstance(item, (bool, np.bool_))).all()
        ):
            raise IngestError(f"Exposure component provenance {flag!r} must be Boolean.")
    canonical_windows: Dict[str, Tuple[int, ...]] = {}
    occupied_years: set = set()
    for raw_name, raw_years in windows.items():
        name = str(raw_name).strip()
        years_list = list(raw_years)
        if not name or name in canonical_windows or any(
            isinstance(year, (bool, np.bool_)) for year in years_list
        ):
            raise IngestError("Exposure-window names/years are malformed.")
        parsed = pd.to_numeric(pd.Series(years_list), errors="coerce")
        if bool(parsed.isna().any()) or bool((parsed % 1.0 != 0.0).any()):
            raise IngestError(f"The window {name!r} contains a malformed year.")
        years_tuple = tuple(sorted(int(year) for year in parsed))
        if len(years_tuple) != 4 or len(set(years_tuple)) != 4:
            raise IngestError(f"The window {name!r} must contain four distinct years.")
        overlap = occupied_years & set(years_tuple)
        if overlap:
            raise IngestError(f"Exposure windows overlap on year(s) {sorted(overlap)}.")
        occupied_years.update(years_tuple)
        canonical_windows[name] = years_tuple
    # Accumulate the ratio rows.
    ratio_rows: List[Dict[str, Any]] = []
    # Accumulate the component log rows.
    log_rows: List[Dict[str, Any]] = []
    # Compose the component lookup keyed on the country, the series and the year.
    lookup: pd.Series = canonical.set_index(["country_code", "series", "year"])[
        "value_usd"
    ]
    provenance_lookup = canonical.set_index(["country_code", "series", "year"])
    # Compose the country roster.
    countries: List[str] = sorted(
        set(str(code) for code in canonical["country_code"].unique())
    )
    # Iterate over the countries.
    for country in countries:
        # Iterate over the windows in sorted order.
        for window_name in sorted(canonical_windows.keys()):
            # Compose the window's years.
            years: List[int] = list(canonical_windows[window_name])
            # Refuse an empty window.
            if not years:
                # The failure names the window.
                raise IngestError(f"The window {window_name!r} carries no year.")
            # Compose the yearly components of every required series.
            required: Dict[str, Dict[int, float]] = {}
            # Iterate over the required components.
            for series in REQUIRED_COMPONENTS:
                # Compose the series' yearly values.
                required[series] = {}
                # Iterate over the window's years.
                for year in years:
                    # Compose the lookup key.
                    key = (str(country), str(series), int(year))
                    # Read the value, which is missing when the key is absent.
                    value = float(lookup[key]) if key in lookup.index else float("nan")
                    # Record the value.
                    required[series][year] = value
                    # Log the component.
                    log_rows.append(
                        {
                            "country_code": str(country),
                            "window": str(window_name),
                            "component": str(series),
                            "year": int(year),
                            "value_usd": value,
                            "source_row_present": bool(key in lookup.index),
                            "finite_component": bool(np.isfinite(value)),
                            "was_completed": bool(
                                provenance_lookup.loc[key, "was_completed"]
                            ) if key in provenance_lookup.index and "was_completed" in provenance_lookup.columns else False,
                            "original_anchor": bool(
                                provenance_lookup.loc[key, "original_anchor"]
                            ) if key in provenance_lookup.index and "original_anchor" in provenance_lookup.columns else bool(np.isfinite(value)),
                            "value_provenance": str(
                                provenance_lookup.loc[key, "value_provenance"]
                            ) if key in provenance_lookup.index and "value_provenance" in provenance_lookup.columns else (
                                "observed_unspecified" if np.isfinite(value) else "missing"
                            ),
                        }
                    )
            # Preserve an incomplete year's later disposition, but never let
            # an invalid finite gas pair disappear from a partly aligned
            # window (or be offset by other years in its ratio of sums).
            for year in years:
                russian_gas = required["russian_gas"][year]
                world_gas = required["world_gas"][year]
                if (
                    np.isfinite(russian_gas)
                    and np.isfinite(world_gas)
                    and not (0.0 <= russian_gas <= world_gas)
                ):
                    raise IngestError(
                        f"Completed Russian-gas flow exceeds or invalidates the "
                        f"world-gas total at {(country, year)} in {window_name!r}."
                    )
            # Align each ratio on its own complete component-year intersection.
            # This permits the declared 3-of-4 and 2-of-4 coverage designs while
            # ensuring numerator and denominator always sum over identical years.
            energy_years = tuple(
                year
                for year in years
                if all(
                    np.isfinite(required[series][year])
                    for series in RATIO_COMPONENTS[RATIO_ENERGY]
                )
            )
            gas_years = tuple(
                year
                for year in years
                if all(
                    np.isfinite(required[series][year])
                    for series in RATIO_COMPONENTS[RATIO_GAS_SHARE]
                )
            )
            energy_numerator = {
                year: required["russian_gas"][year]
                + required["russian_non_gas"][year]
                for year in energy_years
            }
            gdp_denominator = {year: required["gdp"][year] for year in energy_years}
            gas_numerator = {year: required["russian_gas"][year] for year in gas_years}
            world_denominator = {
                year: required["world_gas"][year] for year in gas_years
            }
            if energy_years:
                energy = compose_window_ratio(
                    energy_numerator,
                    gdp_denominator,
                    on_zero_denominator=on_zero_denominator,
                )
                energy_disposition = _ratio_disposition(
                    energy, energy_numerator, gdp_denominator
                )
            else:
                energy = float("nan")
                energy_disposition = "no_aligned_component_years"
            if gas_years:
                gas_share = compose_window_ratio(
                    gas_numerator,
                    world_denominator,
                    on_zero_denominator=on_zero_denominator,
                )
                gas_share_disposition = _ratio_disposition(
                    gas_share, gas_numerator, world_denominator
                )
            else:
                gas_share = float("nan")
                gas_share_disposition = "no_aligned_component_years"
            # Append the ratio row, carrying the disposition tags so the audit
            # can distinguish a masked zero denominator from an incomplete
            # component window (remediation B-20).
            ratio_rows.append(
                {
                    "country_code": str(country),
                    "window": str(window_name),
                    RATIO_ENERGY: float(energy),
                    RATIO_GAS_SHARE: float(gas_share),
                    RATIO_DISPOSITION_ENERGY: str(energy_disposition),
                    RATIO_DISPOSITION_GAS_SHARE: str(gas_share_disposition),
                    "energy_aligned_years": int(len(energy_years)),
                    "gas_share_aligned_years": int(len(gas_years)),
                    "energy_year_roster": json.dumps(energy_years),
                    "gas_share_year_roster": json.dumps(gas_years),
                    "declared_window_years": json.dumps(tuple(years)),
                }
            )
    # Assemble the ratio frame keyed on the country and the window.
    ratios: pd.DataFrame = pd.DataFrame(ratio_rows).set_index(
        ["country_code", "window"]
    )
    # Assemble the component log.
    log: pd.DataFrame = pd.DataFrame(log_rows).set_index(
        ["country_code", "window", "component", "year"]
    )
    log["energy_aligned"] = False
    log["gas_share_aligned"] = False
    for (country, window_name), ratio_row in ratios.iterrows():
        energy_years = set(json.loads(str(ratio_row["energy_year_roster"])))
        gas_years = set(json.loads(str(ratio_row["gas_share_year_roster"])))
        for series in RATIO_COMPONENTS[RATIO_ENERGY]:
            for year in energy_years:
                log.loc[(country, window_name, series, int(year)), "energy_aligned"] = True
        for series in RATIO_COMPONENTS[RATIO_GAS_SHARE]:
            for year in gas_years:
                log.loc[(country, window_name, series, int(year)), "gas_share_aligned"] = True
    # Assert the reconstructibility: every defined ratio equals the ratio
    # recomputed from the log's own components, and every undefined ratio is
    # undefined in the log's recomputation too.
    for (country, window_name), row in ratios.iterrows():
        # Select the cell's log rows.
        scoped = log.loc[(country, window_name)]
        # Compose the logged numerators and denominators of both ratios.
        logged_inputs = {
            RATIO_ENERGY: (
                scoped.loc["russian_gas", "value_usd"].add(
                    scoped.loc["russian_non_gas", "value_usd"], fill_value=np.nan
                ),
                scoped.loc["gdp", "value_usd"],
            ),
            RATIO_GAS_SHARE: (
                scoped.loc["russian_gas", "value_usd"],
                scoped.loc["world_gas", "value_usd"],
            ),
        }
        # Recompute each ratio from the log and compare it with the frame.
        for ratio_name, (numerator_log, denominator_log) in logged_inputs.items():
            aligned_years = tuple(
                int(year)
                for year in numerator_log.index
                if np.isfinite(float(numerator_log.loc[year]))
                and np.isfinite(float(denominator_log.loc[year]))
            )
            if aligned_years:
                logged_value = compose_window_ratio(
                    {
                        year: float(numerator_log.loc[year]) for year in aligned_years
                    },
                    {
                        year: float(denominator_log.loc[year]) for year in aligned_years
                    },
                    on_zero_denominator=on_zero_denominator,
                )
            else:
                logged_value = float("nan")
            roster_column = (
                "energy_year_roster"
                if ratio_name == RATIO_ENERGY
                else "gas_share_year_roster"
            )
            if tuple(json.loads(str(row[roster_column]))) != aligned_years:
                raise IngestError(
                    f"{country}/{window_name}/{ratio_name}: saved aligned-year roster diverges."
                )
            # Read the frame's value.
            observed_value = float(row[ratio_name])
            # Refuse a divergence between the frame and the log.
            if not (
                (np.isnan(observed_value) and np.isnan(logged_value))
                or abs(observed_value - logged_value) < 1.0e-12
            ):
                # The failure names the cell and the ratio.
                raise IngestError(
                    f"{country}/{window_name}/{ratio_name}: the ratio "
                    f"{observed_value!r} is not reconstructible from the components "
                    f"({logged_value!r})."
                )
    # Tally the ratio dispositions for the exposure audit (remediation B-20).
    disposition_tally: Dict[str, int] = {}
    # Iterate over the disposition columns.
    for disposition_column in (
        RATIO_DISPOSITION_ENERGY,
        RATIO_DISPOSITION_GAS_SHARE,
    ):
        # Count the column's dispositions when the column is present.
        column_tally: Dict[str, int] = (
            ratios[disposition_column].value_counts().to_dict()
            if disposition_column in ratios.columns
            else {}
        )
        # Fold the column's tally into the audit tally.
        for key, count in column_tally.items():
            # Compose the folded key.
            folded_key: str = f"{disposition_column}={key}"
            # Record the count.
            disposition_tally[folded_key] = int(count)
    # Log the ratios' extent.
    logger.info(
        "exposure ratios computed across %d country-window(s) with %d logged "
        "component row(s); dispositions: %s.",
        len(ratios),
        len(log),
        disposition_tally,
    )
    # Return the ratios and the log.
    return ratios, log

# -------------------------------------------------------------------------------------------------------------------------------
# Task 27, Step 2: The sample rules
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the per-component usable and imputed year counts.
def build_usable_counts(
    components: pd.DataFrame,
    completion_ledger: pd.DataFrame,
    windows: Mapping[str, Sequence[int]],
) -> pd.DataFrame:
    """
    Compose the per-component usable and imputed year counts.

    Parameters
    ----------
    components : pandas.DataFrame
        One row per country, series and year carrying the value.
    completion_ledger : pandas.DataFrame
        The completion ledger of the previous task, keyed on the reporter, the
        series and the year, whose presence marks a cell as imputed.
    windows : mapping of str to sequence of int
        The years of each window.

    Returns
    -------
    pandas.DataFrame
        One row per country, window and component carrying the usable and
        imputed counts, original usable count, explicit-missing count and
        structurally absent-row count used by both rules.

    Raises
    ------
    IngestError
        If a required column is absent, because the counts would then rest on
        an undeclared quantity.
    """
    # Refuse a components frame lacking a required column.
    for column in ("country_code", "series", "year", "value_usd"):
        # Test the column's presence.
        if column not in components.columns:
            # The failure names the absent column.
            raise IngestError(f"The components frame lacks the column {column!r}.")
    # Canonicalize the component grid and refuse duplicate or malformed rows
    # before distinguishing absent rows from explicit missing values.
    usable_columns = ["country_code", "series", "year", "value_usd"]
    if "was_completed" in components.columns:
        usable_columns.append("was_completed")
    canonical = components[usable_columns].copy()
    if bool(canonical["country_code"].isna().any()) or bool(
        canonical["series"].isna().any()
    ):
        raise IngestError("Usable-count component identities cannot be null.")
    canonical["country_code"] = (
        canonical["country_code"].astype(str).str.strip().str.upper()
    )
    canonical["series"] = canonical["series"].astype(str).str.strip()
    if bool((canonical["country_code"].str.len() == 0).any()) or bool(
        (canonical["series"].str.len() == 0).any()
    ):
        raise IngestError("Usable-count component identities cannot be blank.")
    if bool(canonical["year"].map(lambda item: isinstance(item, (bool, np.bool_))).any()):
        raise IngestError("Usable-count years cannot be Boolean.")
    years_numeric = pd.to_numeric(canonical["year"], errors="coerce")
    if bool(years_numeric.isna().any()) or bool((years_numeric % 1.0 != 0.0).any()):
        raise IngestError("Usable-count components carry malformed calendar years.")
    canonical["year"] = years_numeric.astype(int)
    if bool(canonical.duplicated(["country_code", "series", "year"]).any()):
        raise IngestError("Usable-count components carry duplicate scientific keys.")
    boolean_values = canonical["value_usd"].map(
        lambda item: isinstance(item, (bool, np.bool_))
    )
    values_numeric = pd.to_numeric(canonical["value_usd"], errors="coerce")
    malformed = values_numeric.isna() & ~canonical["value_usd"].isna()
    if bool(boolean_values.any()) or bool(malformed.any()) or bool(
        np.isinf(values_numeric.to_numpy(dtype=float)).any()
    ) or bool((values_numeric.dropna() < 0.0).any()):
        raise IngestError("Usable-count components carry malformed/non-finite values.")
    canonical["value_usd"] = values_numeric.astype(float)
    if "was_completed" in canonical.columns and not bool(
        canonical["was_completed"].map(
            lambda item: isinstance(item, (bool, np.bool_))
        ).all()
    ):
        raise IngestError("The was_completed provenance flag must be Boolean.")
    if not set(canonical["series"]).issubset(set(REQUIRED_COMPONENTS)):
        raise IngestError("Usable-count components carry an undeclared series.")
    # Compose the imputed keys from the completion ledger, the sole source of
    # changed-cell identity, and bind every such key to a finite component.
    if len(completion_ledger):
        if not isinstance(completion_ledger.index, pd.MultiIndex) or list(
            completion_ledger.index.names
        ) != ["reporter_code", "series", "year"]:
            raise IngestError("The completion ledger has a malformed scientific key.")
        if bool(completion_ledger.index.duplicated().any()):
            raise IngestError("The completion ledger carries duplicate changed cells.")
        for reporter, series, year in completion_ledger.index:
            if (
                not str(reporter).strip()
                or str(series).strip() not in REQUIRED_COMPONENTS
                or isinstance(year, (bool, np.bool_))
            ):
                raise IngestError("The completion ledger carries a malformed key.")
            try:
                numeric_year = float(year)
            except (TypeError, ValueError, OverflowError) as exc:
                raise IngestError("The completion ledger carries a malformed year.") from exc
            if not np.isfinite(numeric_year) or not numeric_year.is_integer():
                raise IngestError("The completion ledger carries a malformed year.")
    imputed_keys = {
        (str(reporter).strip().upper(), str(series).strip(), int(year))
        for reporter, series, year in completion_ledger.index
    } if len(completion_ledger) else set()
    # Accumulate the count rows.
    rows: List[Dict[str, Any]] = []
    # Compose the country roster.
    countries: List[str] = sorted(
        set(str(code) for code in canonical["country_code"].unique())
    )
    # Compose the component lookup.
    lookup: pd.Series = canonical.set_index(["country_code", "series", "year"])["value_usd"]
    component_keys = set(lookup.index)
    foreign_imputations = sorted(imputed_keys - component_keys)
    if foreign_imputations:
        raise IngestError(
            f"Completion-ledger keys are absent from components: {foreign_imputations[:8]}."
        )
    for key in imputed_keys:
        if not np.isfinite(float(lookup.loc[key])):
            raise IngestError(f"Imputed component {key} is not finite.")
    if "was_completed" in canonical.columns:
        flagged_imputations = set(
            canonical.loc[
                canonical["was_completed"].astype(bool),
                ["country_code", "series", "year"],
            ].itertuples(index=False, name=None)
        )
        if flagged_imputations != imputed_keys:
            raise IngestError(
                "Completed-component flags disagree with the changed-cell ledger."
            )
    if not windows:
        raise IngestError("Usable counts require at least one four-year window.")
    canonical_windows: Dict[str, Tuple[int, ...]] = {}
    occupied_years: set = set()
    for raw_name, raw_years in windows.items():
        window_name = str(raw_name).strip()
        raw_year_list = list(raw_years)
        if not window_name or window_name in canonical_windows or any(
            isinstance(year, (bool, np.bool_)) for year in raw_year_list
        ):
            raise IngestError("Usable-count window names/years are malformed.")
        numeric_window = pd.to_numeric(pd.Series(raw_year_list), errors="coerce")
        if bool(numeric_window.isna().any()) or bool(
            (numeric_window % 1.0 != 0.0).any()
        ):
            raise IngestError(f"The window {window_name!r} is malformed.")
        years_tuple = tuple(sorted(int(year) for year in numeric_window))
        if len(years_tuple) != 4 or len(set(years_tuple)) != 4:
            raise IngestError(
                f"The window {window_name!r} must contain four distinct years."
            )
        overlap = occupied_years & set(years_tuple)
        if overlap:
            raise IngestError(f"Usable-count windows overlap on {sorted(overlap)}.")
        occupied_years.update(years_tuple)
        canonical_windows[window_name] = years_tuple
    # Iterate over the countries.
    for country in countries:
        # Iterate over the windows in sorted order.
        for window_name in sorted(canonical_windows):
            # Compose the exact four-year window.
            years: List[int] = list(canonical_windows[window_name])
            # Iterate over the required components.
            for series in REQUIRED_COMPONENTS:
                # Accumulate the usable years.
                usable: int = 0
                # Accumulate the imputed years.
                imputed_count: int = 0
                # Count original finite observations separately from repairs.
                original_usable: int = 0
                # Count structurally absent rows separately from explicit NaN.
                present_rows: int = 0
                explicit_missing: int = 0
                # Iterate over the window's years.
                for year in years:
                    # Compose the lookup key.
                    key = (str(country), str(series), int(year))
                    # Read the value.
                    present = key in lookup.index
                    present_rows += int(present)
                    value = float(lookup[key]) if present else float("nan")
                    # Count a usable year, which is a finite value.
                    if np.isfinite(value):
                        # Increment the usable count.
                        usable += 1
                        # Count an imputed year, which is a completed cell.
                        if key in imputed_keys:
                            # Increment the imputed count.
                            imputed_count += 1
                        else:
                            original_usable += 1
                    elif present:
                        explicit_missing += 1
                # Append the component's count row.
                rows.append(
                    {
                        "country_code": str(country),
                        "window": str(window_name),
                        "component": str(series),
                        "window_years": int(len(years)),
                        "present_rows": int(present_rows),
                        "missing_rows": int(len(years) - present_rows),
                        "explicit_missing_years": int(explicit_missing),
                        "usable_years": int(usable),
                        "original_usable_years": int(original_usable),
                        "imputed_years": int(imputed_count),
                    }
                )
    # Assemble the counts frame.
    counts: pd.DataFrame = pd.DataFrame(rows).set_index(
        ["country_code", "window", "component"]
    )
    if not bool(counts.index.is_unique):
        raise IngestError("Usable-count output keys are not unique.")
    if not bool(
        (
            counts["usable_years"]
            == counts["original_usable_years"] + counts["imputed_years"]
        ).all()
    ):
        raise IngestError("Usable counts double-count original and imputed cells.")
    # Log the counts' extent.
    logger.info(
        "usable-year counts composed across %d country-window-component row(s).",
        len(counts),
    )
    # Return the counts.
    return counts


# Apply the strict and extended sample rules.
def apply_sample_rules(
    counts: pd.DataFrame,
    ratios: pd.DataFrame,
    mechanism_universe: Sequence[str],
    excluded_countries: Sequence[str] = PIPELINE_GAS_EXCLUSION_COUNTRIES,
) -> pd.DataFrame:
    """
    Apply the strict and extended sample rules.

    Parameters
    ----------
    counts : pandas.DataFrame
        The per-component counts.
    ratios : pandas.DataFrame
        The ratio frame, whose defined entries bound the admissible units.
    mechanism_universe : sequence of str
        The countries the mechanism construction covers.
    excluded_countries : sequence of str, optional
        The countries excluded under both rules.

    Returns
    -------
    pandas.DataFrame
        One row per country, window, rule and ratio carrying separate
        construction and mechanism-regression flags, aligned-year evidence,
        binding component counts and failure reasons.

    Raises
    ------
    IngestError
        If the universe is empty, because admitting every unit would defeat the
        intersection, or if the counts and the ratios do not share their keys.
    """
    if not mechanism_universe:
        raise IngestError("The mechanism universe is empty.")
    if not isinstance(counts.index, pd.MultiIndex) or list(counts.index.names) != [
        "country_code", "window", "component"
    ] or not bool(counts.index.is_unique):
        raise IngestError("Component counts require unique country/window/component keys.")
    if not isinstance(ratios.index, pd.MultiIndex) or list(ratios.index.names) != [
        "country_code", "window"
    ] or not bool(ratios.index.is_unique):
        raise IngestError("Ratios require unique country/window keys.")
    count_keys = set(counts.index.droplevel("component"))
    if set(ratios.index) != count_keys:
        raise IngestError("Ratio and component-count country/window keys disagree.")
    universe_codes = tuple(str(code).strip().upper() for code in mechanism_universe)
    if any(not code for code in universe_codes) or len(set(universe_codes)) != len(
        universe_codes
    ):
        raise IngestError("Mechanism-universe codes must be unique and nonblank.")
    universe = set(universe_codes) - set(ORDINATE_EXCLUSIONS)
    excluded = {str(code).strip().upper() for code in excluded_countries}
    if excluded != set(PIPELINE_GAS_EXCLUSION_COUNTRIES):
        raise IngestError(
            "Mechanism priority exclusions must be exactly AUT/DEU/POL/TUR."
        )
    rows: List[Dict[str, Any]] = []
    for (country, window_name), ratio_row in ratios.sort_index().iterrows():
        scoped = counts.loc[(country, window_name)]
        for ratio_name in RATIO_NAMES:
            required = RATIO_COMPONENTS[ratio_name]
            ratio_value = float(ratio_row[ratio_name])
            disposition_column = (
                RATIO_DISPOSITION_ENERGY
                if ratio_name == RATIO_ENERGY
                else RATIO_DISPOSITION_GAS_SHARE
            )
            disposition = str(ratio_row[disposition_column])
            aligned_column = (
                "energy_aligned_years"
                if ratio_name == RATIO_ENERGY
                else "gas_share_aligned_years"
            )
            aligned_years = int(ratio_row[aligned_column])
            for rule, minimum_usable, maximum_imputed in (
                (RULE_STRICT, STRICT_MINIMUM_USABLE, STRICT_MAXIMUM_IMPUTED),
                (RULE_EXTENDED, EXTENDED_MINIMUM_USABLE, EXTENDED_MAXIMUM_IMPUTED),
            ):
                failures: List[str] = []
                evidence: Dict[str, Dict[str, int]] = {}
                for series in required:
                    if series not in set(scoped.index):
                        failures.append(f"{series}: counts absent")
                        continue
                    component_row = scoped.loc[series]
                    usable = int(component_row["usable_years"])
                    imputed = int(component_row["imputed_years"])
                    original = int(component_row["original_usable_years"])
                    evidence[series] = {
                        "usable": usable,
                        "original_usable": original,
                        "imputed": imputed,
                    }
                    if usable < minimum_usable or imputed > maximum_imputed:
                        failures.append(
                            f"{series}: requires >={minimum_usable} usable and "
                            f"<={maximum_imputed} imputed; observed {usable}/{imputed}"
                        )
                if aligned_years < minimum_usable:
                    failures.append(
                        f"aligned ratio years require >={minimum_usable}; observed {aligned_years}"
                    )
                if not np.isfinite(ratio_value) or disposition != "defined":
                    failures.append(f"ratio disposition={disposition}")
                construction_pass = len(failures) == 0
                priority_excluded = str(country) in excluded
                in_universe = str(country) in universe
                admitted = construction_pass and in_universe and not priority_excluded
                regression_failures: List[str] = []
                if not in_universe:
                    regression_failures.append("outside mechanism universe")
                if priority_excluded:
                    regression_failures.append("priority country excluded")
                rows.append(
                    {
                        "country_code": str(country),
                        "window": str(window_name),
                        "rule": str(rule),
                        "ratio_name": str(ratio_name),
                        "required_components": json.dumps(required),
                        "component_evidence": json.dumps(evidence, sort_keys=True),
                        "minimum_usable_years": int(minimum_usable),
                        "maximum_imputed_years": int(maximum_imputed),
                        "ratio_disposition": disposition,
                        "aligned_usable_years": int(aligned_years),
                        "construction_pass": bool(construction_pass),
                        "rule_pass": bool(construction_pass),
                        "in_universe": bool(in_universe),
                        "priority_excluded": bool(priority_excluded),
                        "regression_admitted": bool(admitted),
                        "admitted": bool(admitted),
                        "construction_reason": "admitted" if construction_pass else "; ".join(failures),
                        "regression_reason": "admitted" if admitted else "; ".join(failures + regression_failures),
                        "reason": "admitted" if admitted else "; ".join(failures + regression_failures),
                    }
                )
    mask = pd.DataFrame(rows).set_index(
        ["country_code", "window", "rule", "ratio_name"]
    ).sort_index()
    if not bool(mask.index.is_unique):
        raise IngestError("Ratio-specific sample-mask keys are not unique.")
    admitted_counts = mask[mask["admitted"]].groupby(
        ["window", "rule", "ratio_name"]
    ).size().to_dict()
    logger.info(
        "ratio-specific sample rules applied across %d row(s); admitted: %s.",
        len(mask),
        admitted_counts,
    )
    return mask


# -------------------------------------------------------------------------------------------------------------------------------
# Task 27, Step 3: The within-sample standardization
# -------------------------------------------------------------------------------------------------------------------------------


# Standardize the ratios within the admitted sample.
def standardize_within_sample(
    ratios: pd.DataFrame,
    mask: pd.DataFrame,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Standardize the ratios within the admitted sample.

    Parameters
    ----------
    ratios : pandas.DataFrame
        The ratio frame.
    mask : pandas.DataFrame
        The sample mask.

    Returns
    -------
    tuple
        The standardized frame keyed on country, window, rule and ratio, and
        the moments keyed on window, rule and ratio. Each independent sample
        uses sample standard deviation (ddof=1) and carries a roster/value
        fingerprint.

    Raises
    ------
    IngestError
        If an admitted sample has fewer than two units, is degenerate or fails
        zero-mean and sample-unit-variance post-conditions.
    """
    if not isinstance(mask.index, pd.MultiIndex) or list(mask.index.names) != [
        "country_code", "window", "rule", "ratio_name"
    ] or not bool(mask.index.is_unique):
        raise IngestError("Standardization requires the ratio-specific sample mask.")
    if not isinstance(ratios.index, pd.MultiIndex) or not bool(ratios.index.is_unique):
        raise IngestError("Standardization requires unique country/window ratios.")
    rows: List[Dict[str, Any]] = []
    moment_rows: List[Dict[str, Any]] = []
    combinations = sorted(
        set(
            zip(
                mask.index.get_level_values("window"),
                mask.index.get_level_values("rule"),
                mask.index.get_level_values("ratio_name"),
            )
        )
    )
    for window_name, rule, ratio_name in combinations:
        admitted = mask.loc[
            (mask.index.get_level_values("window") == window_name)
            & (mask.index.get_level_values("rule") == rule)
            & (mask.index.get_level_values("ratio_name") == ratio_name)
            & mask["admitted"]
        ]
        if len(admitted) < 2:
            raise IngestError(
                f"{window_name}/{rule}/{ratio_name}: sample SD requires at least two admitted units."
            )
        codes = sorted(
            str(country) for country in admitted.index.get_level_values("country_code")
        )
        values = np.asarray(
            [float(ratios.loc[(code, window_name), ratio_name]) for code in codes],
            dtype=float,
        )
        if not bool(np.isfinite(values).all()):
            raise IngestError(
                f"{window_name}/{rule}/{ratio_name}: an admitted ratio is undefined."
            )
        mean = float(np.mean(values))
        deviation = float(np.std(values, ddof=1))
        if not np.isfinite(deviation) or deviation <= 0.0:
            raise IngestError(
                f"{window_name}/{rule}/{ratio_name}: sample variation is undefined or zero."
            )
        standardized_values = (values - mean) / deviation
        if abs(float(np.mean(standardized_values))) > 1.0e-9 or not np.isclose(
            float(np.var(standardized_values, ddof=1)), 1.0, rtol=0.0, atol=1.0e-9
        ):
            raise IngestError(
                f"{window_name}/{rule}/{ratio_name}: sample-SD post-conditions failed."
            )
        sample_payload = {
            "window": str(window_name),
            "rule": str(rule),
            "ratio_name": str(ratio_name),
            "country_codes": codes,
            "raw_values": [float(value) for value in values],
            "sd_convention": EXPOSURE_SD_CONVENTION,
        }
        sample_fingerprint = hashlib.sha256(
            json.dumps(sample_payload, sort_keys=True, separators=(",", ":")).encode(
                "utf-8"
            )
        ).hexdigest()
        for position, code in enumerate(codes):
            rows.append(
                {
                    "country_code": str(code),
                    "window": str(window_name),
                    "rule": str(rule),
                    "ratio_name": str(ratio_name),
                    "standardized_value": float(standardized_values[position]),
                    "raw_value": float(values[position]),
                    "sample_mean": mean,
                    "sample_sd": deviation,
                    "sample_size": int(len(values)),
                    "sd_convention": EXPOSURE_SD_CONVENTION,
                    "sample_fingerprint_sha256": sample_fingerprint,
                }
            )
        moment_rows.append(
            {
                "window": str(window_name),
                "rule": str(rule),
                "ratio_name": str(ratio_name),
                "sample_size": int(len(values)),
                "sample_mean": mean,
                "sample_sd": deviation,
                "sd_convention": EXPOSURE_SD_CONVENTION,
                "sample_fingerprint_sha256": sample_fingerprint,
            }
        )
    standardized = pd.DataFrame(rows).set_index(
        ["country_code", "window", "rule", "ratio_name"]
    ).sort_index()
    moments = pd.DataFrame(moment_rows).set_index(
        ["window", "rule", "ratio_name"]
    ).sort_index()
    if not bool(standardized.index.is_unique) or not bool(moments.index.is_unique):
        raise IngestError("Standardization output keys are not unique.")
    logger.info(
        "sample-SD standardization applied across %d unit-row(s) over %d independent samples.",
        len(standardized),
        len(moments),
    )
    return standardized, moments


# Freeze the construction behind a canonical digest.
def freeze_exposure_ratios(
    ratios: pd.DataFrame,
    mask: pd.DataFrame,
    standardized: pd.DataFrame,
    components: Optional[pd.DataFrame] = None,
    counts: Optional[pd.DataFrame] = None,
) -> str:
    """
    Freeze the construction behind a canonical digest.

    Parameters
    ----------
    ratios : pandas.DataFrame
        The ratio frame.
    mask : pandas.DataFrame
        The sample mask.
    standardized : pandas.DataFrame
        The standardized frame.
    components : pandas.DataFrame, optional
        The full component/year reconstruction log.
    counts : pandas.DataFrame, optional
        The ratio-specific construction and regression count audit.

    Returns
    -------
    str
        The hexadecimal digest of the canonical payload over the sorted keys.

    Raises
    ------
    IngestError
        If any frame is empty.
    """
    # Refuse an empty frame.
    for label, frame in (
        ("ratios", ratios),
        ("sample mask", mask),
        ("standardized", standardized),
    ):
        # Test the frame's extent.
        if len(frame) == 0:
            # The failure names the frame.
            raise IngestError(f"Cannot freeze an empty {label} frame.")
    if not bool(ratios.index.is_unique) or not bool(mask.index.is_unique) or not bool(
        standardized.index.is_unique
    ):
        raise IngestError("Exposure-ratio artefacts must be uniquely keyed before freezing.")
    payload = {
        "ratios": _content_digest(ratios.sort_index().reset_index()),
        "sample_mask": _content_digest(mask.sort_index().reset_index()),
        "standardized": _content_digest(standardized.sort_index().reset_index()),
    }
    if components is not None:
        if len(components) == 0 or not bool(components.index.is_unique):
            raise IngestError("Cannot freeze an empty or duplicated component log.")
        payload["components"] = _content_digest(
            components.sort_index().reset_index()
        )
    if counts is not None:
        if len(counts) == 0 or not bool(counts.index.is_unique):
            raise IngestError("Cannot freeze an empty or duplicated count audit.")
        payload["counts"] = _content_digest(counts.sort_index().reset_index())
    return hashlib.sha256(
        json.dumps(payload, sort_keys=True, separators=(",", ":")).encode("utf-8")
    ).hexdigest()


# -------------------------------------------------------------------------------------------------------------------------------
# Task 27, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Construct the exposure ratios, the sample mask and the standardized values.
def construct_exposure_ratios(
    components: pd.DataFrame,
    completion_ledger: pd.DataFrame,
    windows: Mapping[str, Sequence[int]],
    mechanism_universe: Sequence[str],
    excluded_countries: Sequence[str] = PIPELINE_GAS_EXCLUSION_COUNTRIES,
    expected_counts: Optional[Mapping[Tuple[str, ...], int]] = None,
) -> ExposureRatioResult:
    """
    Construct the exposure ratios, the sample mask and the standardized values.

    Parameters
    ----------
    components : pandas.DataFrame
        One row per country, series and year carrying the completed primitive
        or the reported product series.
    completion_ledger : pandas.DataFrame
        The completion ledger of the previous task, which is the single source
        of the imputation counts.
    windows : mapping of str to sequence of int
        The years of each window.
    mechanism_universe : sequence of str
        The countries the mechanism construction covers.
    excluded_countries : sequence of str, optional
        The countries excluded under both rules.
    expected_counts : mapping of tuple to int, optional
        Declared counts keyed by window/rule/ratio and optionally a terminal
        metric (``construction_pass`` or ``regression_admitted``). Legacy
        window/rule keys are checked against both independent ratios.

    Returns
    -------
    ExposureRatioResult
        The ratios, the component log, the sample mask, the standardized frame,
        the counts, the digest, the timestamp and the notes.

    Raises
    ------
    IngestError
        If any construction, rule, standardization, coverage or comparison
        check fails.
    """
    # A verified all-zero world-gas window is an undefined share, not a zero
    # share and not a reason to abort otherwise valid countries.  Preserve the
    # ratio row and its zero_denominator disposition so the sample rules can
    # exclude it from construction, regression and standardization.  The
    # standalone ratio composer remains strict by default.
    ratios, component_log = build_exposure_ratios(
        components, windows, on_zero_denominator="mask"
    )
    # Compose the per-component counts.
    counts = build_usable_counts(components, completion_ledger, windows)
    # Apply the sample rules.
    mask = apply_sample_rules(counts, ratios, mechanism_universe, excluded_countries)
    # Standardize within the admitted sample.
    standardized, _moments = standardize_within_sample(ratios, mask)
    # Compose the per-window and rule counts.
    summary_rows: List[Dict[str, Any]] = []
    # Iterate over the windows.
    for window_name in sorted(set(mask.index.get_level_values("window"))):
        # Iterate over the rules.
        for rule in SAMPLE_RULE_ORDER:
            # Summarize each ratio's construction and regression masks
            # independently so one component set cannot determine the other.
            for ratio_name in RATIO_NAMES:
                scoped = mask.loc[
                    (mask.index.get_level_values("window") == window_name)
                    & (mask.index.get_level_values("rule") == rule)
                    & (mask.index.get_level_values("ratio_name") == ratio_name)
                ]
                summary_rows.append(
                    {
                        "window": str(window_name),
                        "rule": str(rule),
                        "ratio_name": str(ratio_name),
                        "units": int(len(scoped)),
                        "construction_pass": int(scoped["construction_pass"].sum()),
                        "rule_pass": int(scoped["rule_pass"].sum()),
                        "in_universe": int(scoped["in_universe"].sum()),
                        "priority_excluded": int(scoped["priority_excluded"].sum()),
                        "regression_admitted": int(scoped["regression_admitted"].sum()),
                        "admitted": int(scoped["admitted"].sum()),
                    }
                )
    # Assemble the summary frame.
    summary: pd.DataFrame = pd.DataFrame(summary_rows).set_index(
        ["window", "rule", "ratio_name"]
    ).sort_index()
    # Compare the counts with the declared values when supplied.
    if expected_counts is not None:
        # Iterate over the declared keys.
        for raw_key, declared in expected_counts.items():
            if isinstance(declared, (bool, np.bool_)) or not isinstance(
                declared, (int, np.integer)
            ) or int(declared) < 0:
                raise IngestError("Declared exposure counts must be nonnegative integers.")
            key = tuple(raw_key)
            if len(key) == 4:
                candidate_keys = [key[:3]]
                metric = str(key[3])
                if metric not in ("construction_pass", "regression_admitted", "admitted"):
                    raise IngestError(f"Undeclared exposure-count metric {metric!r}.")
            elif len(key) == 3:
                candidate_keys = [key]
                metric = "regression_admitted"
            elif len(key) == 2:
                # Preserve legacy declarations only by applying them to both
                # independently constructed ratio samples.
                candidate_keys = [key + (ratio_name,) for ratio_name in RATIO_NAMES]
                metric = "regression_admitted"
            else:
                raise IngestError(
                    "Declared exposure counts require window/rule[/ratio] keys."
                )
            for candidate in candidate_keys:
                if candidate not in summary.index:
                    raise IngestError(f"The summary omits the declared key {candidate}.")
                observed = int(summary.loc[candidate, metric])
                if observed != int(declared):
                    raise IngestError(
                        f"{candidate}: {observed} admitted unit(s) against the declared "
                        f"{int(declared)}."
                    )
    # Freeze the construction.
    fingerprint: str = freeze_exposure_ratios(
        ratios, mask, standardized, component_log, summary
    )
    # Accumulate the recorded notes.
    notes: List[str] = [
        "every ratio is a ratio of window sums recomputed from its components "
        "on its own common finite component-year roster and is never itself "
        "interpolated; the component log reproduces every value and roster.",
        "energy/GDP and gas share use distinct registered component sets, so "
        "missingness in an irrelevant primitive cannot remove the other ratio.",
        "the strict rule requires at least three usable years and at most one "
        "imputed year per required component and at least three aligned ratio "
        "years; the extended rule relaxes these bounds to two; construction "
        "and mechanism-regression masks remain separate.",
        f"the mechanism universe carries {len(set(str(code) for code in mechanism_universe))} "
        f"supplied countr(ies); the production design expects {FULL_PANEL_SIZE - len(ORDINATE_EXCLUSIONS)} "
        f"after excluding {list(ORDINATE_EXCLUSIONS)}.",
        f"standardization uses {EXPOSURE_SD_CONVENTION} independently within "
        "each admitted ratio/window/rule sample and binds the exact country "
        "roster and raw values to a sample fingerprint.",
    ]
    # Record the windows.
    notes.append(
        "windows: "
        + str({str(name): [int(year) for year in sorted(years)] for name, years in windows.items()})
        + "."
    )
    # Assemble the immutable result.
    result: ExposureRatioResult = ExposureRatioResult(
        ratios=ratios,
        components=component_log,
        sample_mask=mask,
        standardized=standardized,
        counts=summary,
        fingerprint_sha256=fingerprint,
        built_utc=current_utc_iso(),
        notes=tuple(notes),
    )
    # Log the completed construction.
    logger.info(
        "exposure ratios complete: %d ratio row(s), %d standardized row(s), "
        "fingerprint %s.",
        len(ratios),
        len(standardized),
        fingerprint[:16],
    )
    # Return the assembled result.
    return result
````

````python
# Task 28: Build Orchestrator 6 - Exposure Heterogeneity Orchestrator

# ==============================================================================
# Task 28: Build Orchestrator 6 - Exposure Heterogeneity Orchestrator
# ==============================================================================

# Declare the estimator of the interactive-factor specification.
ESTIMATOR_INTERACTIVE: Final[str] = "interactive_factor"
# Declare the estimator of the two-way substitution, which substitutes the
# estimator rather than a parameter of it.
ESTIMATOR_TWFE: Final[str] = "two_way_fixed_effects"
# Compose the declared estimator vocabulary.
ESTIMATOR_VOCABULARY: Final[Tuple[str, ...]] = (ESTIMATOR_INTERACTIVE, ESTIMATOR_TWFE)
# Declare the post window of the escalation indicator.
POST_WINDOW: Final[Tuple[int, int, int]] = (2022, 2023, 2024)
# Declare the name of the sanction-by-post coefficient.
NAME_BETA_1: Final[str] = "beta_1_sanction_post"
# Declare the name of the exposure-by-post coefficient.
NAME_BETA_2: Final[str] = "beta_2_exposure_post"
# Declare the name of the triple-interaction coefficient.
NAME_BETA_3: Final[str] = "beta_3_sanction_exposure_post"
# Declare the name of the composite slope.
NAME_COMPOSITE: Final[str] = "beta_2_plus_beta_3"
# Declare the scaling that converts a decimal rate to percentage points.
RATE_SCALE: Final[float] = 100.0


# Declare one specification of the heterogeneity suite.
@dataclass(frozen=True)
class SpecificationRecord:
    """
    One specification of the heterogeneity suite.

    Attributes
    ----------
    name : str
        The specification's declared name.
    exposure_variant : str
        The exposure ratio the specification uses.
    window : str
        The exposure window.
    coverage_rule : str
        The coverage rule the sample admits.
    policy_name : str
        The policy instrument whose sanction indicator enters.
    estimator : str
        The estimator the specification uses; one of the declared six
        substitutes the estimator rather than a parameter, which is why the
        estimator is an axis rather than a constant.
    """
    # The specification's declared name.
    name: str
    # The exposure ratio the specification uses.
    exposure_variant: str
    # The exposure window.
    window: str
    # The coverage rule.
    coverage_rule: str
    # The policy instrument.
    policy_name: str
    # The estimator.
    estimator: str


# Compose the declared six specifications of the study.
SPECIFICATIONS: Final[Tuple[SpecificationRecord, ...]] = (
    SpecificationRecord(
        name="baseline",
        exposure_variant=RATIO_ENERGY,
        window="w2018_2021",
        coverage_rule=RULE_STRICT,
        policy_name="strict_sanctions",
        estimator=ESTIMATOR_INTERACTIVE,
    ),
    SpecificationRecord(
        name="pre_crimea",
        exposure_variant=RATIO_ENERGY,
        window="w2010_2013",
        coverage_rule=RULE_STRICT,
        policy_name="strict_sanctions",
        estimator=ESTIMATOR_INTERACTIVE,
    ),
    SpecificationRecord(
        name="pre_crimea_extended",
        exposure_variant=RATIO_ENERGY,
        window="w2010_2013",
        coverage_rule=RULE_EXTENDED,
        policy_name="strict_sanctions",
        estimator=ESTIMATOR_INTERACTIVE,
    ),
    SpecificationRecord(
        name="twfe",
        exposure_variant=RATIO_ENERGY,
        window="w2018_2021",
        coverage_rule=RULE_STRICT,
        policy_name="strict_sanctions",
        estimator=ESTIMATOR_TWFE,
    ),
    SpecificationRecord(
        name="direct_arming_strict",
        exposure_variant=RATIO_ENERGY,
        window="w2018_2021",
        coverage_rule=RULE_STRICT,
        policy_name="direct_arming",
        estimator=ESTIMATOR_INTERACTIVE,
    ),
    SpecificationRecord(
        name="gas_share",
        exposure_variant=RATIO_GAS_SHARE,
        window="w2018_2021",
        coverage_rule=RULE_STRICT,
        policy_name="strict_sanctions",
        estimator=ESTIMATOR_INTERACTIVE,
    ),
)


# Declare the frozen result of the heterogeneity suite.
@dataclass(frozen=True)
class HeterogeneityResult:
    """
    Frozen result of the exposure heterogeneity suite.

    Attributes
    ----------
    coefficients : pandas.DataFrame
        One row per specification and reported scalar carrying the estimate,
        its cluster standard error and its two-sided probability. The reported
        scalars are the three coefficients and their composite slope.
    covariance : pandas.DataFrame
        The full coefficient covariance of every specification, so the
        composite slope's variance is recoverable from the artefact rather than
        only from the code.
    statistics : pandas.DataFrame
        One row per specification carrying actual fitted observation,
        country, sanctioner, and donor counts; estimator and absorbed/full
        ranks; residual degrees of freedom; CR1 correction; sample, donor, and
        rule-selection fingerprints.
    fingerprint_sha256 : str
        Digest of the canonical suite payload.
    built_utc : str
        ISO-8601 UTC timestamp of the suite.
    notes : tuple of str
        Recorded decisions, including the two conventions the specification
        fixes.
    """
    # The coefficient frame.
    coefficients: pd.DataFrame
    # The covariance frame.
    covariance: pd.DataFrame
    # The per-specification statistics.
    statistics: pd.DataFrame
    # Digest of the canonical suite payload.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the suite.
    built_utc: str
    # Recorded decisions of the suite.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 28, Step 1: The regression frame and the partialling
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the regression frame of one specification.
def build_regression_frame(
    panel: pd.DataFrame,
    outcome: str,
    exposures: pd.Series,
    sanction_indicator: pd.Series,
    donor_codes: Sequence[str],
    exposure_variant: str,
    rate_in_percentage_points: bool = True,
    policy_name: Optional[str] = None,
    exposure_window: Optional[str] = None,
    coverage_rule: Optional[str] = None,
) -> pd.DataFrame:
    """
    Compose the regression frame of one specification.

    Parameters
    ----------
    panel : pandas.DataFrame
        The frozen panel carrying the outcome by country and year.
    outcome : str
        The outcome variable.
    exposures : pandas.Series
        The standardized exposure keyed on country for one already selected
        ratio, window and coverage-rule regression sample.
    sanction_indicator : pandas.Series
        The binary policy-status tensor keyed on country and year.  The
        Equation 18 group indicator is resolved once per country as whether
        that country is ever in the policy group and is then held fixed over
        the complete regression history.
    donor_codes : sequence of str
        The non-policy group over which the cross-sectional mean is composed.
    exposure_variant : str
        The exposure ratio's name, recorded for the audit.
    rate_in_percentage_points : bool, optional
        Whether a rate outcome is scaled from decimals to percentage points,
        which is the study's declared convention for the rate regressions.
    policy_name, exposure_window, coverage_rule : str, optional
        Immutable specification labels retained on every emitted row.  The
        orchestrator supplies all three; the optional defaults preserve the
        historical direct-call interface.

    Returns
    -------
    pandas.DataFrame
        One row per country and year carrying the dependent variable, the post
        indicator, the sanction indicator, the exposure, the cross-sectional
        donor mean and the three interactions.

    Raises
    ------
    IngestError
        If keys are duplicated, values are non-finite, the regression sample
        is unbalanced, the policy tensor is incomplete/non-binary, or the
        panel omits a donor-year.  These failures prevent silent observation,
        exposure, or donor-roster deletion.
    """
    # Refuse a panel lacking a required column.
    if not bool(panel.columns.is_unique):
        raise IngestError("The regression panel carries duplicate column labels.")
    for column in ("country_code", "year", outcome):
        # Test the column's presence.
        if column not in panel.columns:
            # The failure names the absent column.
            raise IngestError(f"The panel lacks the column {column!r}.")
    # Canonicalize and validate the non-policy donor roster.
    donors_ordered = tuple(str(code).strip() for code in donor_codes)
    if not donors_ordered or any(not code for code in donors_ordered):
        # The failure states the precondition.
        raise IngestError("The cross-sectional mean requires a non-empty donor roster.")
    if len(set(donors_ordered)) != len(donors_ordered):
        raise IngestError("The cross-sectional donor roster contains duplicate identities.")
    if not isinstance(exposures, pd.Series) or not bool(exposures.index.is_unique):
        raise IngestError("Regression exposures must be a uniquely country-keyed Series.")
    exposure_labels = [str(code).strip() for code in exposures.index]
    if any(not code for code in exposure_labels) or len(set(exposure_labels)) != len(
        exposure_labels
    ):
        raise IngestError("Regression exposure identities are blank or collide after canonicalization.")
    exposure_values = pd.to_numeric(exposures, errors="coerce")
    if len(exposure_values) < 2 or not bool(
        np.isfinite(exposure_values.to_numpy(dtype=float)).all()
    ):
        raise IngestError("Regression exposures require at least two finite countries.")
    exposure_map = {
        str(code).strip(): float(value)
        for code, value in zip(exposures.index, exposure_values.to_numpy(dtype=float))
    }
    exposure_codes = set(exposure_map)
    # Normalize only the rows needed by the regression or its donor mean.
    relevant_codes = exposure_codes | set(donors_ordered)
    source = panel.loc[
        panel["country_code"].astype(str).str.strip().isin(relevant_codes),
        ["country_code", "year", outcome],
    ].copy()
    source["country_code"] = source["country_code"].astype(str).str.strip()
    numeric_year = pd.to_numeric(source["year"], errors="coerce")
    if len(source) == 0 or not bool(np.isfinite(numeric_year.to_numpy(dtype=float)).all()):
        raise IngestError("The regression panel carries a missing or non-numeric year.")
    if not bool(np.equal(numeric_year.to_numpy(dtype=float), np.floor(numeric_year)).all()):
        raise IngestError("The regression panel carries a non-integral calendar year.")
    source["year"] = numeric_year.astype(int)
    if bool(source.duplicated(["country_code", "year"]).any()):
        raise IngestError("The regression panel duplicates a country-year key.")
    if set(source["country_code"]) != relevant_codes:
        raise IngestError("The panel omits an exposure-sample country.")
    source["_outcome"] = pd.to_numeric(source[outcome], errors="coerce").astype(float)
    if not bool(np.isfinite(source["_outcome"].to_numpy(dtype=float)).all()):
        raise IngestError(f"The regression panel carries a missing value of {outcome!r}.")
    # Require one common regression calendar for every exposure country.
    sample_source = source[source["country_code"].isin(exposure_codes)].copy()
    sample_calendars = {
        code: tuple(sorted(group["year"].astype(int).tolist()))
        for code, group in sample_source.groupby("country_code", sort=True)
    }
    if set(sample_calendars) != exposure_codes or len(set(sample_calendars.values())) != 1:
        raise IngestError("The exposure regression sample is not a complete balanced country-year product.")
    regression_years = next(iter(sample_calendars.values()))
    if not set(POST_WINDOW).issubset(set(regression_years)):
        raise IngestError("The regression calendar omits a declared 2022--2024 post year.")
    expected_donor_keys = {
        (code, int(year)) for code in donors_ordered for year in regression_years
    }
    donor_source = source[source["country_code"].isin(set(donors_ordered))].copy()
    actual_donor_keys = set(
        zip(donor_source["country_code"], donor_source["year"].astype(int))
    )
    if actual_donor_keys != expected_donor_keys:
        missing = sorted(expected_donor_keys - actual_donor_keys)[:5]
        extra = sorted(actual_donor_keys - expected_donor_keys)[:5]
        raise IngestError(
            "The non-policy outcome-mean roster is not a complete donor-by-year "
            f"product; missing={missing!r}, extra={extra!r}."
        )
    # Copy the exact exposure-sample product so the frozen panel is untouched.
    frame: pd.DataFrame = sample_source[["country_code", "year", "_outcome"]].copy()
    dependent = frame["_outcome"].astype(float)
    # Scale a rate outcome to percentage points, which is the declared
    # convention for the rate regressions.
    if str(outcome) in RATE_OUTCOMES and bool(rate_in_percentage_points):
        # Scale the dependent variable.
        dependent = dependent * RATE_SCALE
        # Record the convention.
        logger.info(
            "%s enters in percentage points, scaled by %.1f.",
            outcome,
            RATE_SCALE,
        )
    # Attach the dependent variable.
    frame["y"] = dependent
    # Refuse a missing dependent value, which the sample rules should preclude.
    if not bool(np.isfinite(frame["y"].to_numpy(dtype=float)).all()):
        # The failure names the outcome.
        raise IngestError(
            f"The regression frame carries a missing value of {outcome!r}."
        )
    # Compose the post indicator over the declared escalation years.
    frame["post"] = frame["year"].astype(int).isin(POST_WINDOW).astype(float)
    # Validate the complete policy tensor before resolving the time-invariant
    # Equation 18 group membership S_i.
    if not isinstance(sanction_indicator, pd.Series) or not isinstance(
        sanction_indicator.index, pd.MultiIndex
    ) or sanction_indicator.index.nlevels != 2 or not bool(sanction_indicator.index.is_unique):
        raise IngestError("The policy indicator must be uniquely keyed on country and year.")
    sanction_table = sanction_indicator.rename("indicator").reset_index()
    sanction_table.columns = ["country_code", "year", "indicator"]
    sanction_table["country_code"] = sanction_table["country_code"].astype(str).str.strip()
    sanction_year = pd.to_numeric(sanction_table["year"], errors="coerce")
    sanction_value = pd.to_numeric(sanction_table["indicator"], errors="coerce")
    if not bool(np.isfinite(sanction_year.to_numpy(dtype=float)).all()) or not bool(
        np.equal(sanction_year.to_numpy(dtype=float), np.floor(sanction_year)).all()
    ):
        raise IngestError("The policy indicator carries an invalid calendar year.")
    if not bool(np.isfinite(sanction_value.to_numpy(dtype=float)).all()) or not set(
        sanction_value.astype(float).unique().tolist()
    ).issubset({0.0, 1.0}):
        raise IngestError("The policy indicator must contain only finite binary cells.")
    sanction_table["year"] = sanction_year.astype(int)
    sanction_table["indicator"] = sanction_value.astype(float)
    required_indicator_keys = {
        (code, int(year)) for code in relevant_codes for year in regression_years
    }
    observed_indicator_keys = set(
        zip(sanction_table["country_code"], sanction_table["year"])
    )
    missing_indicator = sorted(required_indicator_keys - observed_indicator_keys)[:5]
    if missing_indicator:
        raise IngestError(
            "The policy indicator is incomplete over the regression and donor "
            f"products; examples={missing_indicator!r}."
        )
    membership = (
        sanction_table[sanction_table["country_code"].isin(relevant_codes)]
        .groupby("country_code", sort=True)["indicator"]
        .max()
        .astype(float)
    )
    if any(float(membership.loc[code]) != 0.0 for code in donors_ordered):
        raise IngestError("The declared donor outcome-mean roster contains a policy-group country.")
    frame["s"] = frame["country_code"].map(membership).astype(float)
    # Attach the exposure by joining on the country.
    frame["e"] = (
        frame["country_code"].astype(str).map(
            exposure_map
        )
    )
    # Refuse a missing exposure.
    if not bool(np.isfinite(frame["e"].to_numpy(dtype=float)).all()):
        # The failure names the variant.
        raise IngestError(
            f"The exposure variant {exposure_variant!r} is missing for a country."
        )
    # Compose the cross-sectional donor mean per year, over the non-policy group.
    donor_values = donor_source["_outcome"].astype(float)
    if str(outcome) in RATE_OUTCOMES and bool(rate_in_percentage_points):
        donor_values = donor_values * RATE_SCALE
    donor_source["_dependent"] = donor_values
    # Compose the yearly donor mean over the exact fixed non-policy roster.
    yearly_mean: pd.Series = donor_source.groupby("year", sort=True)["_dependent"].mean()
    # Refuse a year whose donor mean is undefined.
    if not bool(np.isfinite(yearly_mean.to_numpy(dtype=float)).all()):
        # The failure states the condition.
        raise IngestError("The cross-sectional donor mean is undefined for a year.")
    # Attach the cross-sectional mean by year.
    frame["ybar"] = frame["year"].map(yearly_mean).astype(float)
    # Compose the three interacted regressors.
    frame["x1"] = frame["s"] * frame["post"]
    # Compose the exposure interaction.
    frame["x2"] = frame["e"] * frame["post"]
    # Compose the triple interaction.
    frame["x3"] = frame["s"] * frame["e"] * frame["post"]
    # Retain the resolved scientific design on every row so downstream audits
    # never infer it from a specification label alone.
    frame["policy"] = str(policy_name) if policy_name is not None else "unspecified"
    frame["exposure_variant"] = str(exposure_variant)
    frame["window"] = str(exposure_window) if exposure_window is not None else "unspecified"
    frame["coverage_rule"] = str(coverage_rule) if coverage_rule is not None else "unspecified"
    frame["donor_roster_sha256"] = hashlib.sha256(
        "\n".join(sorted(donors_ordered)).encode("utf-8")
    ).hexdigest()
    frame["sample_roster_sha256"] = hashlib.sha256(
        "\n".join(sorted(exposure_codes)).encode("utf-8")
    ).hexdigest()
    frame = frame.drop(columns=["_outcome"])
    # Log the frame's extent.
    logger.info(
        "%s/%s: regression frame composed across %d country-year(s) and %d "
        "countr(ies).",
        outcome,
        exposure_variant,
        len(frame),
        frame["country_code"].nunique(),
    )
    # Return the ordered frame.
    return frame.sort_values(["country_code", "year"]).reset_index(drop=True)


# Build an orthonormal basis for the registered nuisance design.
def _heterogeneity_projection_basis(
    frame: pd.DataFrame,
    estimator: str,
) -> Tuple[np.ndarray, np.ndarray, int, int]:
    """Build the nuisance design and an orthonormal basis for its column space.

    Interactive Equation 18 absorbs a country intercept and a country-specific
    loading on the non-policy-group outcome mean.  The separately registered
    TWFE alternative absorbs country and year effects.  The SVD basis exposes
    the effective absorbed dimension and supports an explicit full-design
    numerical oracle without retaining redundant nuisance columns.
    """
    if estimator not in ESTIMATOR_VOCABULARY:
        raise IngestError(f"Undeclared estimator {estimator!r}.")
    required = {"country_code", "year", "y", "ybar", "s", "e", "post", "x1", "x2", "x3"}
    missing = sorted(required - set(frame.columns))
    if missing:
        raise IngestError(f"The heterogeneity frame omits required columns {missing!r}.")
    if len(frame) == 0 or bool(frame.duplicated(["country_code", "year"]).any()):
        raise IngestError("The heterogeneity frame is empty or duplicates a country-year key.")
    numeric = frame[["year", "y", "ybar", "s", "e", "post", "x1", "x2", "x3"]].apply(
        pd.to_numeric, errors="coerce"
    )
    if not bool(np.isfinite(numeric.to_numpy(dtype=float)).all()):
        raise IngestError("The heterogeneity frame contains a non-finite scientific value.")
    if not set(numeric["s"].unique().tolist()).issubset({0.0, 1.0}) or not set(
        numeric["post"].unique().tolist()
    ).issubset({0.0, 1.0}):
        raise IngestError("The heterogeneity treatment and post indicators must be binary.")
    fixed_s = frame.assign(_s=numeric["s"]).groupby("country_code")["_s"].nunique()
    if bool(fixed_s.gt(1).any()):
        raise IngestError("Equation 18 requires the policy-group indicator S_i to be fixed by country.")
    countries = frame["country_code"].astype(str)
    unit_dummies = pd.get_dummies(countries, drop_first=False).to_numpy(dtype=float)
    if estimator == ESTIMATOR_INTERACTIVE:
        second_block = unit_dummies * numeric["ybar"].to_numpy(dtype=float)[:, None]
    else:
        second_block = pd.get_dummies(
            numeric["year"].astype(int), drop_first=True
        ).to_numpy(dtype=float)
    projection = np.column_stack([unit_dummies, second_block])
    u, singular_values, _vh = np.linalg.svd(projection, full_matrices=False)
    tolerance = (
        float(max(projection.shape) * np.finfo(float).eps * singular_values[0])
        if len(singular_values) and float(singular_values[0]) > 0.0
        else 0.0
    )
    rank = int(np.sum(singular_values > tolerance))
    if rank <= 0:
        raise IngestError("The registered nuisance design has zero rank.")
    return projection, u[:, :rank], rank, int(projection.shape[1])


# Partial out the unit effect and the loading term.
def partial_out_effects(
    frame: pd.DataFrame,
    estimator: str,
) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    """
    Partial out the unit effect and the loading term.

    Parameters
    ----------
    frame : pandas.DataFrame
        The regression frame.
    estimator : str
        The estimator. The interactive-factor estimator partials out the unit
        dummies together with the country-by-mean interactions, which is the
        declared construction; the two-way estimator partials out the unit
        dummies together with the year dummies, which is its declared
        substitution.

    Returns
    -------
    tuple
        The residualized dependent variable, the residualized three-column
        design and the projection's rank diagnostics.

    Raises
    ------
    IngestError
        If the estimator is undeclared or the projection is rank deficient
        against the declared blocks, because the residualization would then be
        incomplete.
    """
    projection, basis, rank, width = _heterogeneity_projection_basis(frame, estimator)
    # Compose the scientific outcome and three registered interactions.
    y: np.ndarray = frame["y"].to_numpy(dtype=float)
    x: np.ndarray = frame[["x1", "x2", "x3"]].to_numpy(dtype=float)
    # Apply I-QQ' without materializing an observation-by-observation matrix.
    y_residual: np.ndarray = y - basis @ (basis.T @ y)
    x_residual: np.ndarray = x - basis @ (basis.T @ x)
    # Compose the rank diagnostics.
    diagnostics: np.ndarray = np.array(
        [float(rank), float(width), float(len(y))], dtype=float
    )
    # Log the partialling's extent.
    logger.info(
        "%s: partialled on %d projection column(s) of rank %d across %d row(s).",
        estimator,
        projection.shape[1],
        int(rank),
        len(y),
    )
    # Return the residualized dependent variable, the design and the diagnostics.
    return y_residual, x_residual, diagnostics


# -------------------------------------------------------------------------------------------------------------------------------
# Task 28, Step 2: The estimation and the clustered covariance
# -------------------------------------------------------------------------------------------------------------------------------


# Estimate the interacted regression with a clustered covariance.
def estimate_interacted_regression(
    frame: pd.DataFrame,
    estimator: str,
) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    """
    Estimate the interacted regression with a clustered covariance.

    Parameters
    ----------
    frame : pandas.DataFrame
        The regression frame keyed on the country and the year.
    estimator : str
        The estimator name.

    Returns
    -------
    tuple
        The coefficient vector, its clustered covariance matrix and the
        estimation statistics in registered order: observations, clusters,
        residual degrees of freedom, residualized-design rank, absorbed rank,
        nuisance width, full effective rank, CR1 correction, fitted countries,
        and fitted sanctioners.

    Raises
    ------
    IngestError
        If the residualized design is rank deficient, because the coefficients
        would then be unidentified, or if the cluster count is below two,
        because a clustered covariance would be undefined.
    """
    # Partial out the effects and the loading term.
    y_residual, x_residual, partial_diagnostics = partial_out_effects(frame, estimator)
    # Refuse a rank-deficient design.
    design_rank = int(np.linalg.matrix_rank(x_residual))
    # Test the rank against the design's width.
    if design_rank != x_residual.shape[1]:
        # The failure names both ranks.
        raise IngestError(
            f"The residualized design carries rank {design_rank} against "
            f"{x_residual.shape[1]} column(s), so the coefficients are "
            f"unidentified."
        )
    # Compose the coefficient vector through the least-squares solve.
    beta: np.ndarray = np.linalg.lstsq(x_residual, y_residual, rcond=None)[0]
    # Compose the residuals of the partialled fit.
    residual: np.ndarray = y_residual - x_residual @ beta
    # Compose the cross-product of the design.
    cross_product: np.ndarray = x_residual.T @ x_residual
    # Compose the cluster groups.
    clusters: np.ndarray = frame["country_code"].astype(str).to_numpy()
    # Compose the cluster identities.
    unique_clusters = sorted(set(str(value) for value in clusters))
    # Refuse too few clusters.
    if len(unique_clusters) < 2:
        # The failure names the observed count.
        raise IngestError(
            f"The clustered covariance requires at least two clusters; "
            f"{len(unique_clusters)} supplied."
        )
    # Accumulate the clustered meat of the sandwich.
    meat: np.ndarray = np.zeros((x_residual.shape[1], x_residual.shape[1]), dtype=float)
    # Iterate over the clusters.
    for cluster in unique_clusters:
        # Select the cluster's rows.
        mask = clusters == cluster
        # Compose the cluster's score.
        score: np.ndarray = x_residual[mask].T @ residual[mask]
        # Accumulate the outer product of the score.
        meat += np.outer(score, score)
    # Compose the inverse of the design's cross-product.
    inverse = np.linalg.pinv(cross_product)
    # Compose the clustered covariance with the registered finite-sample
    # correction.  The absorbed dimension is part of K; using only the three
    # reported interactions would understate this CR1 adjustment.
    observations = float(x_residual.shape[0])
    absorbed_rank = int(partial_diagnostics[0])
    absorbed_width = int(partial_diagnostics[1])
    full_rank = int(absorbed_rank + design_rank)
    residual_df = int(observations) - full_rank
    if residual_df <= 0:
        raise IngestError(
            f"The Equation 18 full design leaves {residual_df} residual degree(s) of freedom."
        )
    # Compose the cluster count.
    cluster_count = float(len(unique_clusters))
    # Compose the correction factor.
    correction = (cluster_count / (cluster_count - 1.0)) * (
        (observations - 1.0) / float(residual_df)
    )
    # Compose the covariance.
    covariance: np.ndarray = inverse @ meat @ inverse * correction
    covariance = 0.5 * (covariance + covariance.T)
    if not bool(np.isfinite(beta).all()) or not bool(np.isfinite(covariance).all()):
        raise IngestError("The Equation 18 fit produced a non-finite coefficient or covariance.")
    # Reproduce the fit independently with the explicit full design [Q, X].
    # Q is the orthonormal basis of the registered nuisance column space, so
    # this is numerically full rank and its last three entries are Equation 18.
    _projection, basis, oracle_absorbed_rank, _width = _heterogeneity_projection_basis(
        frame, estimator
    )
    if oracle_absorbed_rank != absorbed_rank:
        raise IngestError("The absorbed rank changed between residualized and explicit fits.")
    original_x = frame[["x1", "x2", "x3"]].to_numpy(dtype=float)
    original_y = frame["y"].to_numpy(dtype=float)
    full_design = np.column_stack([basis, original_x])
    full_beta = np.linalg.lstsq(full_design, original_y, rcond=None)[0]
    oracle_beta = full_beta[-3:]
    full_residual = original_y - full_design @ full_beta
    full_meat = np.zeros((full_rank, full_rank), dtype=float)
    for cluster in unique_clusters:
        mask = clusters == cluster
        score = full_design[mask].T @ full_residual[mask]
        full_meat += np.outer(score, score)
    full_inverse = np.linalg.pinv(full_design.T @ full_design)
    oracle_covariance = (
        full_inverse @ full_meat @ full_inverse * correction
    )[-3:, -3:]
    oracle_covariance = 0.5 * (oracle_covariance + oracle_covariance.T)
    oracle_scale = max(
        1.0, float(np.max(np.abs(beta))), float(np.max(np.abs(covariance)))
    )
    if not np.allclose(beta, oracle_beta, rtol=1.0e-9, atol=1.0e-10 * oracle_scale):
        raise IngestError(
            "Residualized Equation 18 coefficients disagree with the explicit full design."
        )
    if not np.allclose(
        covariance, oracle_covariance, rtol=1.0e-8, atol=1.0e-10 * oracle_scale
    ):
        raise IngestError(
            "Residualized clustered covariance disagrees with the explicit full design."
        )
    sanctioners = int(
        frame[["country_code", "s"]]
        .drop_duplicates("country_code")["s"]
        .astype(float)
        .sum()
    )
    # Compose the statistics.
    statistics: np.ndarray = np.array(
        [
            observations,
            cluster_count,
            float(residual_df),
            float(design_rank),
            float(absorbed_rank),
            float(absorbed_width),
            float(full_rank),
            float(correction),
            float(len(unique_clusters)),
            float(sanctioners),
        ],
        dtype=float,
    )
    # Log the fit's extent.
    logger.info(
        "%s: fitted across %d observation(s) and %d cluster(s); coefficients %s.",
        estimator,
        int(observations),
        int(cluster_count),
        np.round(beta, 6).tolist(),
    )
    # Return the coefficients, the covariance and the statistics.
    return beta, covariance, statistics


# -------------------------------------------------------------------------------------------------------------------------------
# Task 28, Step 3: The composite slope's standard error
# -------------------------------------------------------------------------------------------------------------------------------


# Compose a linear combination's estimate and its standard error.
def compose_linear_combination(
    beta: np.ndarray,
    covariance: np.ndarray,
    weights: np.ndarray,
    naive_sum_of_standard_errors: bool = True,
) -> Tuple[float, float, float, float]:
    """
    Compose a linear combination's estimate and its standard error.

    Parameters
    ----------
    beta : numpy.ndarray
        The coefficient vector.
    covariance : numpy.ndarray
        The coefficient covariance matrix.
    weights : numpy.ndarray
        The linear combination's weights.
    naive_sum_of_standard_errors : bool, optional
        Whether to compose the naive sum of the components' standard errors for
        the comparison the specification fixes must be reported.

    Returns
    -------
    tuple
        The combination's estimate, its standard error, its two-sided
        probability and the naive sum of the components' standard errors, so
        the difference between the exact and the naive quantity is visible.

    Raises
    ------
    IngestError
        If the weights' length disagrees with the coefficient vector, or if the
        combination's variance is negative beyond round-off, because a negative
        variance would mean the covariance is not positive semi-definite.
    """
    # Refuse malformed or non-finite inputs before composing any inference.
    beta = np.asarray(beta, dtype=float)
    covariance = np.asarray(covariance, dtype=float)
    weights = np.asarray(weights, dtype=float)
    if beta.ndim != 1 or weights.ndim != 1 or weights.shape[0] != beta.shape[0] or covariance.shape != (beta.shape[0], beta.shape[0]):
        # The failure states the condition.
        raise IngestError(
            "The linear combination's weights must match the coefficient vector."
        )
    if not bool(np.isfinite(beta).all()) or not bool(np.isfinite(covariance).all()) or not bool(
        np.isfinite(weights).all()
    ):
        raise IngestError("A linear-combination input is non-finite.")
    if not np.allclose(covariance, covariance.T, rtol=0.0, atol=1.0e-12):
        raise IngestError("The coefficient covariance is not symmetric.")
    eigenvalues = np.linalg.eigvalsh(0.5 * (covariance + covariance.T))
    covariance_scale = max(1.0, float(np.max(np.abs(covariance))))
    if float(np.min(eigenvalues)) < -1.0e-10 * covariance_scale:
        raise IngestError("The coefficient covariance is not positive semi-definite.")
    # Compose the combination's estimate.
    estimate: float = float(weights @ beta)
    # Compose the combination's variance from the full covariance.
    variance: float = float(weights @ covariance @ weights)
    # Refuse a materially negative variance, which would mean the covariance
    # is not positive semi-definite. The threshold admits the round-off of a
    # variance that is genuinely zero, which is a legitimate outcome of a
    # particular design.
    if variance < -1.0e-12:
        # The failure names the variance.
        raise IngestError(f"The combination's variance is negative: {variance!r}.")
    # Compose the standard error.
    standard_error: float = float(np.sqrt(max(variance, 0.0)))
    # Compose the two-sided probability under the normal reference.
    probability: float = float(
        2.0 * (1.0 - _standard_normal_cdf(abs(estimate) / standard_error))
        if standard_error > 0.0
        else float("nan")
    )
    # Compose the naive sum of the components' standard errors when requested.
    naive: float = (
        float(sum(np.sqrt(max(float(covariance[index, index]), 0.0)) for index in np.flatnonzero(weights)))
        if naive_sum_of_standard_errors
        else float("nan")
    )
    # Return the four values.
    return estimate, standard_error, probability, naive


# Evaluate the standard normal distribution function.
def _standard_normal_cdf(value: float) -> float:
    """
    Evaluate the standard normal distribution function.

    Parameters
    ----------
    value : float
        The argument.

    Returns
    -------
    float
        The probability that a standard normal variate does not exceed the
        argument, evaluated through the complementary error function so no
        approximation table is needed.
    """
    # Evaluate the distribution function through the error function.
    return 0.5 * float(1.0 + math.erf(float(value) / float(np.sqrt(2.0))))


# Build one specification's coefficient table rows.
def build_specification_rows(
    specification: SpecificationRecord,
    beta: np.ndarray,
    covariance: np.ndarray,
    statistics: np.ndarray,
    outcome: str,
) -> Tuple[List[dict], List[dict]]:
    """
    Build one specification's coefficient table rows.

    Parameters
    ----------
    specification : SpecificationRecord
        The specification.
    beta : numpy.ndarray
        The coefficient vector.
    covariance : numpy.ndarray
        The coefficient covariance matrix.
    statistics : numpy.ndarray
        The estimation statistics.
    outcome : str
        The outcome variable.

    Returns
    -------
    tuple
        The coefficient rows and the covariance rows, both carrying the
        specification's own key.

    Raises
    ------
    IngestError
        If the coefficient/covariance contract is malformed.  Equality of the
        full-covariance standard error and a naive sum is not an error: a
        perfectly positively correlated PSD covariance legitimately produces
        that equality (the registered E20 boundary case).
    """
    beta = np.asarray(beta, dtype=float)
    covariance = np.asarray(covariance, dtype=float)
    statistics = np.asarray(statistics, dtype=float)
    if beta.shape != (3,) or covariance.shape != (3, 3):
        raise IngestError(
            "Equation 18 requires exactly three coefficients and a 3-by-3 covariance."
        )
    if statistics.ndim != 1 or len(statistics) < 5 or not bool(
        np.isfinite(statistics).all()
    ):
        raise IngestError(
            "The Equation 18 estimation statistics are malformed or non-finite."
        )
    # Compose the key of the specification. The declared five-tuple is
    # prefixed by the specification's own name, because one of the declared
    # specifications substitutes the estimator and therefore shares every
    # element of the five-tuple with another; the name is the only
    # discriminator, and carrying it keeps the declared key as the prefix.
    key = (specification.name, specification.policy_name, outcome,
           specification.exposure_variant, specification.window,
           specification.coverage_rule)
    # Compose the coefficient rows.
    rows: List[dict] = []
    # Iterate over the three coefficients in their declared order.
    for position, name in enumerate((NAME_BETA_1, NAME_BETA_2, NAME_BETA_3)):
        # Compose the coefficient's own weight vector.
        weights = np.zeros(len(beta), dtype=float)
        # Assign the unit weight.
        weights[position] = 1.0
        # Compose the coefficient's estimate and its inference.
        estimate, standard_error, probability, _naive = compose_linear_combination(
            beta, covariance, weights, naive_sum_of_standard_errors=False
        )
        # Append the coefficient's row.
        rows.append(
            {
                "specification": key[0],
                "policy": key[1],
                "outcome": key[2],
                "exposure_variant": key[3],
                "window": key[4],
                "coverage_rule": key[5],
                "scalar": str(name),
                "estimate": float(estimate),
                "se_cluster": float(standard_error),
                "p_value": float(probability),
                "observations": int(statistics[0]),
                "clusters": int(statistics[1]),
            }
        )
        rows[-1].update(
            {
                "residual_df": int(statistics[2]),
                "absorbed_rank": int(statistics[4]),
                "full_design_rank": int(statistics[6]) if len(statistics) > 6 else int(statistics[3] + statistics[4]),
                "cluster_correction": float(statistics[7]) if len(statistics) > 7 else float("nan"),
                "sanctioners": int(statistics[9]) if len(statistics) > 9 else -1,
            }
        )
    # Compose the composite slope, which is the sum of the second and third
    # coefficients, and its variance from the full covariance.
    composite_weights = np.zeros(len(beta), dtype=float)
    # Assign the two unit weights.
    composite_weights[1] = 1.0
    # Assign the third weight.
    composite_weights[2] = 1.0
    # Compose the composite's estimate, its standard error, its probability and
    # the naive sum for the comparison.
    estimate, standard_error, probability, naive = compose_linear_combination(
        beta, covariance, composite_weights, naive_sum_of_standard_errors=True
    )
    # Retain the off-diagonal covariance and the naive diagnostic.  No
    # inequality is asserted: an all-ones PSD covariance has V22=V33=V23=1,
    # for which the exact composite SE and the naive sum both equal two.
    covariance_entry = float(covariance[1, 2])
    # Append the composite's row.
    rows.append(
        {
            "specification": key[0],
            "policy": key[1],
            "outcome": key[2],
            "exposure_variant": key[3],
            "window": key[4],
            "coverage_rule": key[5],
            "scalar": str(NAME_COMPOSITE),
            "estimate": float(estimate),
            "se_cluster": float(standard_error),
            "p_value": float(probability),
            "observations": int(statistics[0]),
            "clusters": int(statistics[1]),
            "residual_df": int(statistics[2]),
            "absorbed_rank": int(statistics[4]),
            "full_design_rank": int(statistics[6]) if len(statistics) > 6 else int(statistics[3] + statistics[4]),
            "cluster_correction": float(statistics[7]) if len(statistics) > 7 else float("nan"),
            "sanctioners": int(statistics[9]) if len(statistics) > 9 else -1,
        }
    )
    # Record the covariance identity used for the composite slope.
    rows[-1]["component_covariance"] = float(covariance_entry)
    rows[-1]["naive_se_sum"] = float(naive)
    # Compose the covariance rows.
    covariance_rows: List[dict] = []
    # Iterate over the covariance's entries.
    for row_index, row_name in enumerate((NAME_BETA_1, NAME_BETA_2, NAME_BETA_3)):
        # Iterate over the columns.
        for column_index, column_name in enumerate((NAME_BETA_1, NAME_BETA_2, NAME_BETA_3)):
            # Append the entry's row.
            covariance_rows.append(
                {
                    "specification": key[0],
                    "policy": key[1],
                    "outcome": key[2],
                    "exposure_variant": key[3],
                    "window": key[4],
                    "coverage_rule": key[5],
                    "row_scalar": str(row_name),
                    "column_scalar": str(column_name),
                    "value": float(covariance[row_index, column_index]),
                }
            )
    # Return the rows.
    return rows, covariance_rows


# Freeze the suite behind a canonical digest.
def freeze_heterogeneity(
    coefficients: pd.DataFrame,
    covariance: pd.DataFrame,
    statistics: pd.DataFrame,
) -> str:
    """
    Freeze the suite behind a canonical digest.

    Parameters
    ----------
    coefficients : pandas.DataFrame
        The coefficient frame.
    covariance : pandas.DataFrame
        The covariance frame.
    statistics : pandas.DataFrame
        The per-specification statistics.

    Returns
    -------
    str
        The hexadecimal digest of the canonical payload over the sorted keys.

    Raises
    ------
    IngestError
        If any frame is empty.
    """
    # Refuse an empty frame.
    for label, frame in (
        ("coefficients", coefficients),
        ("covariance", covariance),
        ("statistics", statistics),
    ):
        # Test the frame's extent.
        if len(frame) == 0:
            # The failure names the frame.
            raise IngestError(f"Cannot freeze an empty {label} frame.")
    for label, frame in (
        ("coefficients", coefficients),
        ("covariance", covariance),
        ("statistics", statistics),
    ):
        if not bool(frame.index.is_unique):
            raise IngestError(f"The {label} heterogeneity keys are duplicated.")
    # Bind every labelled value, dtype, axis name, and shape rather than a
    # hand-selected subset of publication columns.
    payload = {
        "coefficients": _content_digest(coefficients.sort_index()),
        "covariance": _content_digest(covariance.sort_index()),
        "statistics": _content_digest(statistics.sort_index()),
    }
    return hashlib.sha256(
        json.dumps(payload, sort_keys=True, separators=(",", ":")).encode("utf-8")
    ).hexdigest()


# -------------------------------------------------------------------------------------------------------------------------------
# Task 28, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Execute the exposure heterogeneity suite.
def execute_heterogeneity_suite(
    panel: pd.DataFrame,
    outcome: str,
    exposures: pd.DataFrame,
    sanction_indicators: pd.DataFrame,
    donor_codes: Any,
    specifications: Sequence[SpecificationRecord] = SPECIFICATIONS,
    expected_slopes: Optional[Mapping[Tuple[str, str], float]] = None,
    tolerance: float = 1.0e-3,
) -> HeterogeneityResult:
    """
    Execute the exposure heterogeneity suite.

    Parameters
    ----------
    panel : pandas.DataFrame
        The frozen panel.
    outcome : str
        The outcome variable.
    exposures : pandas.DataFrame
        Standardized exposures keyed on country, window and coverage rule,
        with one column per ratio-specific sample.
    sanction_indicators : pandas.DataFrame
        The sanction indicators keyed on the country and the year, one column
        per policy.
    donor_codes : sequence of str or mapping of str to sequence of str
        The non-policy group over which the cross-sectional mean is composed.
        A mapping is resolved by policy and is required by the pipeline so the
        sanctions and direct-arming specifications retain their actual donor
        identities.  A sequence remains supported for a single-policy direct
        call.
    specifications : sequence of SpecificationRecord, optional
        The specifications to execute, defaulting to the declared six.
    expected_slopes : mapping of tuple to float, optional
        The declared slopes keyed on the specification's name and the scalar's
        name, compared only when supplied.
    tolerance : float, optional
        The tolerance of the declared comparison.

    Returns
    -------
    HeterogeneityResult
        The coefficient table, the covariance table, the statistics, the
        digest, the timestamp and the notes.

    Raises
    ------
    IngestError
        If a specification references an unavailable exposure variant or
        indicator, if a fit fails, or if a declared slope diverges.
    """
    if not isinstance(exposures, pd.DataFrame) or not isinstance(
        exposures.index, pd.MultiIndex
    ) or not {"country_code", "window", "rule"}.issubset(set(exposures.index.names)):
        raise IngestError(
            "Standardized exposures must be uniquely keyed on country, window, and rule."
        )
    if not bool(exposures.columns.is_unique):
        raise IngestError("Standardized exposure column labels are duplicated.")
    if not bool(exposures.index.is_unique):
        raise IngestError("Standardized exposure sample keys are duplicated.")
    if not isinstance(sanction_indicators, pd.DataFrame) or not isinstance(
        sanction_indicators.index, pd.MultiIndex
    ) or sanction_indicators.index.nlevels != 2 or not bool(
        sanction_indicators.index.is_unique
    ):
        raise IngestError("Policy indicators must be uniquely keyed on country and year.")
    if not bool(sanction_indicators.columns.is_unique):
        raise IngestError("Policy-indicator column labels are duplicated.")
    if not specifications:
        raise IngestError("The heterogeneity suite requires at least one specification.")
    specification_names = [str(specification.name) for specification in specifications]
    if any(not name for name in specification_names) or len(set(specification_names)) != len(
        specification_names
    ):
        raise IngestError("Heterogeneity specification names are blank or duplicated.")
    if isinstance(tolerance, (bool, np.bool_)):
        raise IngestError("The expected-slope tolerance must be finite and nonnegative.")
    try:
        resolved_tolerance = float(tolerance)
    except (TypeError, ValueError) as exc:
        raise IngestError(
            "The expected-slope tolerance must be finite and nonnegative."
        ) from exc
    if not np.isfinite(resolved_tolerance) or resolved_tolerance < 0.0:
        raise IngestError("The expected-slope tolerance must be finite and nonnegative.")
    # Accumulate the coefficient rows.
    coefficient_rows: List[dict] = []
    # Accumulate the covariance rows.
    covariance_rows: List[dict] = []
    # Accumulate the statistics rows.
    statistic_rows: List[dict] = []
    # Retain each independently selected sample for the strict/extended alias
    # guard and for lineage recorded in the statistics artefact.
    sample_evidence: Dict[Tuple[str, str, str, str, str], Tuple[Tuple[str, ...], Tuple[float, ...]]] = {}
    # Determine the number of sanctioned units from the indicator, which is the
    # count of countries ever sanctioned under the specification's policy.
    # Iterate over the specifications in their declared order.
    for specification in specifications:
        if specification.exposure_variant not in RATIO_NAMES:
            raise IngestError(
                f"{specification.name}: undeclared exposure variant "
                f"{specification.exposure_variant!r}."
            )
        if specification.coverage_rule not in SAMPLE_RULE_ORDER:
            raise IngestError(
                f"{specification.name}: undeclared coverage rule "
                f"{specification.coverage_rule!r}."
            )
        if specification.policy_name not in POLICIES:
            raise IngestError(
                f"{specification.name}: undeclared policy {specification.policy_name!r}."
            )
        if specification.window not in {"w2010_2013", "w2018_2021"}:
            raise IngestError(
                f"{specification.name}: undeclared exposure window {specification.window!r}."
            )
        # Refuse an undeclared estimator.
        if specification.estimator not in ESTIMATOR_VOCABULARY:
            # The failure names the specification and the estimator.
            raise IngestError(
                f"{specification.name}: undeclared estimator "
                f"{specification.estimator!r}."
            )
        # Refuse an undetermined exposure pair.
        if specification.exposure_variant not in set(exposures.columns):
            # The failure names the variant.
            raise IngestError(
                f"{specification.name}: the exposure variant "
                f"{specification.exposure_variant!r} is unavailable."
            )
        # Refuse an unavailable policy indicator.
        if specification.policy_name not in set(sanction_indicators.columns):
            # The failure names the policy.
            raise IngestError(
                f"{specification.name}: the policy indicator "
                f"{specification.policy_name!r} is unavailable."
            )
        # Select the exact ratio/window/coverage-rule sample constructed by
        # Task 27 rather than silently reusing the strict sample for every spec.
        if "rule" not in exposures.index.names:
            raise IngestError(
                f"{specification.name}: exposures omit the coverage-rule level."
            )
        try:
            selected_exposures = exposures.xs(
                (specification.window, specification.coverage_rule),
                level=("window", "rule"),
            )[specification.exposure_variant]
        except KeyError as exc:
            raise IngestError(
                f"{specification.name}: exposure sample "
                f"{specification.window}/{specification.coverage_rule} is unavailable."
            ) from exc
        if not bool(selected_exposures.index.is_unique):
            raise IngestError(f"{specification.name}: exposure-country keys are duplicated.")
        # Nulls mark identities outside this Task 27 ratio-specific sample;
        # selection is explicit and every retained value must be finite.
        window_exposures = selected_exposures[selected_exposures.notna()].copy()
        if window_exposures.empty:
            # The failure names the window.
            raise IngestError(
                f"{specification.name}: the window {specification.window!r} carries "
                f"no exposure."
            )
        numeric_exposure = pd.to_numeric(window_exposures, errors="coerce")
        if not bool(np.isfinite(numeric_exposure.to_numpy(dtype=float)).all()):
            raise IngestError(f"{specification.name}: the selected exposure sample is non-finite.")
        window_exposures = numeric_exposure.astype(float)
        evidence_key = (
            str(specification.exposure_variant), str(specification.window),
            str(specification.policy_name), str(specification.estimator),
            str(specification.coverage_rule),
        )
        sample_evidence[evidence_key] = (
            tuple(str(code) for code in window_exposures.index),
            tuple(float(value) for value in window_exposures.to_numpy(dtype=float)),
        )
        selection_digest = hashlib.sha256(
            json.dumps(
                {
                    "variant": str(specification.exposure_variant),
                    "window": str(specification.window),
                    "rule": str(specification.coverage_rule),
                    "countries": list(sample_evidence[evidence_key][0]),
                    "values": list(sample_evidence[evidence_key][1]),
                },
                sort_keys=True,
                separators=(",", ":"),
            ).encode("utf-8")
        ).hexdigest()
        # Compose the sanction indicator of the specification's policy.
        indicator = sanction_indicators[specification.policy_name]
        # Resolve the non-policy roster by the specification's actual policy.
        if isinstance(donor_codes, Mapping):
            pair_key = (str(specification.policy_name), str(outcome))
            if pair_key in donor_codes:
                policy_donors_raw = donor_codes[pair_key]
            elif specification.policy_name in donor_codes:
                policy_entry = donor_codes[specification.policy_name]
                if isinstance(policy_entry, Mapping):
                    if str(outcome) not in policy_entry:
                        raise IngestError(
                            f"{specification.name}: no donor roster is registered for "
                            f"{specification.policy_name}/{outcome}."
                        )
                    policy_donors_raw = policy_entry[str(outcome)]
                else:
                    policy_donors_raw = policy_entry
            else:
                raise IngestError(
                    f"{specification.name}: no donor roster is registered for policy/outcome "
                    f"{specification.policy_name!r}/{outcome!r}."
                )
        else:
            policy_donors_raw = donor_codes
        if isinstance(policy_donors_raw, (str, bytes)):
            raise IngestError(f"{specification.name}: the donor roster is not a sequence of identities.")
        policy_donors = tuple(str(code) for code in policy_donors_raw)
        # Compose the regression frame.
        frame = build_regression_frame(
            panel,
            outcome,
            window_exposures,
            indicator,
            policy_donors,
            specification.exposure_variant,
            policy_name=specification.policy_name,
            exposure_window=specification.window,
            coverage_rule=specification.coverage_rule,
        )
        # Estimate the specification.
        beta, covariance, statistics = estimate_interacted_regression(
            frame, specification.estimator
        )
        # Compose the specification's rows.
        rows, covariance_entries = build_specification_rows(
            specification, beta, covariance, statistics, outcome
        )
        # Bind every reported scalar and covariance cell to the actual fitted
        # estimator, country sample, and policy-specific non-policy roster.
        donor_digest = str(frame["donor_roster_sha256"].iloc[0])
        sample_digest = str(frame["sample_roster_sha256"].iloc[0])
        for row in rows:
            row.update(
                {
                    "estimator": str(specification.estimator),
                    "donor_count": int(len(policy_donors)),
                    "donor_roster_sha256": donor_digest,
                    "sample_roster_sha256": sample_digest,
                    "selection_fingerprint_sha256": selection_digest,
                }
            )
        for row in covariance_entries:
            row.update(
                {
                    "estimator": str(specification.estimator),
                    "donor_roster_sha256": donor_digest,
                    "sample_roster_sha256": sample_digest,
                    "selection_fingerprint_sha256": selection_digest,
                }
            )
        # Accumulate the rows.
        coefficient_rows.extend(rows)
        # Accumulate the covariance entries.
        covariance_rows.extend(covariance_entries)
        # Compose the sanctioner count from the indicator, which is the number
        # of countries ever sanctioned under the policy.
        sanctioners = int(statistics[9])
        # Append the statistics row.
        statistic_rows.append(
            {
                "policy": specification.policy_name,
                "outcome": str(outcome),
                "exposure_variant": specification.exposure_variant,
                "window": specification.window,
                "coverage_rule": specification.coverage_rule,
                "specification": specification.name,
                "estimator": specification.estimator,
                "observations": int(statistics[0]),
                "clusters": int(statistics[1]),
                "sanctioners": int(sanctioners),
                "partial_rank": int(statistics[4]),
                "residual_df": int(statistics[2]),
                "design_rank": int(statistics[3]),
                "absorbed_rank": int(statistics[4]),
                "absorbed_width": int(statistics[5]),
                "full_design_rank": int(statistics[6]),
                "cluster_correction": float(statistics[7]),
                "sample_countries": int(statistics[8]),
                "donor_count": int(len(policy_donors)),
                "donor_roster_sha256": donor_digest,
                "sample_roster_sha256": sample_digest,
                "selection_fingerprint_sha256": selection_digest,
            }
        )
    # Strict and extended variants must come from distinct rule-indexed source
    # selections.  Their realised rosters/values may legitimately coincide in
    # a fully observed fixture; the rule-labelled selection fingerprints, not
    # an artificial numerical inequality, establish separate provenance.
    evidence_groups: Dict[Tuple[str, str, str, str], Dict[str, Tuple[Tuple[str, ...], Tuple[float, ...]]]] = {}
    for key, evidence in sample_evidence.items():
        group_key = key[:4]
        evidence_groups.setdefault(group_key, {})[key[4]] = evidence
    for group_key, by_rule in evidence_groups.items():
        if RULE_STRICT in by_rule and RULE_EXTENDED in by_rule:
            strict_fingerprint = hashlib.sha256(
                json.dumps(
                    {"design": group_key, "rule": RULE_STRICT, "evidence": by_rule[RULE_STRICT]},
                    sort_keys=True,
                    separators=(",", ":"),
                ).encode("utf-8")
            ).hexdigest()
            extended_fingerprint = hashlib.sha256(
                json.dumps(
                    {"design": group_key, "rule": RULE_EXTENDED, "evidence": by_rule[RULE_EXTENDED]},
                    sort_keys=True,
                    separators=(",", ":"),
                ).encode("utf-8")
            ).hexdigest()
            if strict_fingerprint == extended_fingerprint:
                raise IngestError(f"Rule-labelled selections alias for design {group_key!r}.")
    # Assemble the coefficient frame.
    coefficients: pd.DataFrame = pd.DataFrame(coefficient_rows).set_index(
        ["specification", "policy", "outcome", "exposure_variant", "window",
         "coverage_rule", "scalar"]
    )
    # Assemble the covariance frame.
    covariance_frame: pd.DataFrame = pd.DataFrame(covariance_rows).set_index(
        [
            "specification",
            "policy",
            "outcome",
            "exposure_variant",
            "window",
            "coverage_rule",
            "row_scalar",
            "column_scalar",
        ]
    )
    # Assemble the statistics frame.
    statistics_frame: pd.DataFrame = pd.DataFrame(statistic_rows).set_index(
        ["specification", "policy", "outcome", "exposure_variant", "window",
         "coverage_rule"]
    )
    if len(coefficients) != 4 * len(specifications) or len(covariance_frame) != 9 * len(
        specifications
    ) or len(statistics_frame) != len(specifications):
        raise IngestError("The heterogeneity suite output is not complete for every specification.")
    if not bool(coefficients.index.is_unique) or not bool(covariance_frame.index.is_unique) or not bool(
        statistics_frame.index.is_unique
    ):
        raise IngestError("The heterogeneity suite output keys are not unique.")
    # Compare the slopes with the declared values when supplied.
    if expected_slopes is not None:
        # Iterate over the declared entries.
        for key, declared in expected_slopes.items():
            if not isinstance(key, tuple) or len(key) != 2:
                raise IngestError("Expected slopes must be keyed on (specification, scalar).")
            if isinstance(declared, (bool, np.bool_)):
                raise IngestError(f"{key!r}: the declared slope must be a finite number.")
            try:
                resolved_declared = float(declared)
            except (TypeError, ValueError) as exc:
                raise IngestError(
                    f"{key!r}: the declared slope must be a finite number."
                ) from exc
            if not np.isfinite(resolved_declared):
                raise IngestError(f"{key!r}: the declared slope must be a finite number.")
            # Select the specification's row by its index level, which is the
            # name, because the name is the key's leading level.
            scoped = statistics_frame[
                statistics_frame.index.get_level_values(0) == key[0]
            ]
            # Refuse an unspecified name.
            if len(scoped) == 0:
                # The failure names the specification.
                raise IngestError(f"The suite omits the specification {key[0]!r}.")
            # Compose the specification's coefficient key.
            coefficient_key = tuple(scoped.index[0]) + (str(key[1]),)
            # Read the observed slope.
            if coefficient_key not in coefficients.index:
                raise IngestError(f"The suite omits declared coefficient cell {key!r}.")
            observed = float(coefficients.loc[coefficient_key, "estimate"])
            # Refuse a divergence from the declared slope.
            if abs(observed - resolved_declared) > resolved_tolerance:
                # The failure names the specification, the scalar and both values.
                raise IngestError(
                    f"{key[0]}/{key[1]}: slope {observed!r} against the declared "
                    f"{resolved_declared!r}."
                )
    # Freeze the suite.
    fingerprint: str = freeze_heterogeneity(
        coefficients, covariance_frame, statistics_frame
    )
    # Accumulate the recorded notes.
    notes: List[str] = [
        "the same interface serves every specification; the six declared "
        "variants are configured through the exposure variant, the window, the "
        "coverage rule, the policy instrument and the estimator name.",
        "the unit effect and the country-specific loading on the "
        "cross-sectional donor mean are partialled out together, because the "
        "loading is an interactive-factor term rather than a year effect.",
        "no lagged controls enter, and a rate outcome is scaled to percentage "
        "points, which are the two conventions the specification fixes.",
        "the composite slope's variance is V22 + V33 + 2*V23; equality of "
        "its standard error with a naive component-SE sum remains admissible "
        "at the PSD all-ones E20 boundary.",
        "country-cluster CR1 uses G/(G-1)*(N-1)/(N-K), where K is the full "
        "effective rank including absorbed country/loadings (or TWFE effects).",
    ]
    # Record the sanctioner counts per specification.
    notes.append(
        "sanctioner counts: "
        + str(
            {
                str(index[0]): int(row["sanctioners"])
                for index, row in statistics_frame.iterrows()
            }
        )
        + "."
    )
    # Assemble the immutable result.
    result: HeterogeneityResult = HeterogeneityResult(
        coefficients=coefficients,
        covariance=covariance_frame,
        statistics=statistics_frame,
        fingerprint_sha256=fingerprint,
        built_utc=current_utc_iso(),
        notes=tuple(notes),
    )
    # Log the completed suite.
    logger.info(
        "heterogeneity suite complete: %d specification(s), %d coefficient "
        "row(s), fingerprint %s.",
        len(statistics_frame),
        len(coefficients),
        fingerprint[:16],
    )
    # Return the assembled result.
    return result
````

````python
# Task 29: Execute Exposure Heterogeneity Regressions

# ==============================================================================
# Task 29: Execute Exposure Heterogeneity Regressions
# ==============================================================================

# Declare the four outcomes of the panel, whose names are the analysis columns
# of the frozen panel.
OUTCOME_LOG_PRICE: Final[str] = "log_CPI"
# Declare the inflation outcome, which enters in percentage points.
OUTCOME_INFLATION: Final[str] = "inflation_att"
# Declare the level product outcome.
OUTCOME_LOG_PRODUCT: Final[str] = "log_real_GDPpc"
# Declare the growth outcome, which enters in percentage points.
OUTCOME_GROWTH: Final[str] = "growth_att"
# Compose the declared outcomes of the panel.
PANEL_OUTCOMES: Final[Tuple[str, ...]] = (
    OUTCOME_LOG_PRICE,
    OUTCOME_INFLATION,
    OUTCOME_LOG_PRODUCT,
    OUTCOME_GROWTH,
)
# Declare the window of the baseline specification.
WINDOW_BASELINE: Final[str] = "w2018_2021"
# Declare the window of the pre-Crimea specification.
WINDOW_PRE_CRIMEA: Final[str] = "w2010_2013"
# Declare the significance level the invariants are evaluated at.
SIGNIFICANCE_LEVEL: Final[float] = 0.05
# Declare the four separately reported Table 10 scalars.
TABLE10_SCALARS: Final[Tuple[str, str, str, str]] = (
    NAME_BETA_1,
    NAME_BETA_2,
    NAME_BETA_3,
    NAME_COMPOSITE,
)
# Classify narrative sign/significance statements as descriptive findings.
CHECK_CLASS_DESCRIPTIVE: Final[str] = "descriptive_empirical"
# Classify identities whose failure makes the computation invalid.
CHECK_CLASS_MATHEMATICAL: Final[str] = "mathematical_dataflow"
# Declare the name of the growth invariant.
INVARIANT_GROWTH: Final[str] = "growth_differential_negative_and_significant"
# Declare the name of the within-coalition invariant.
INVARIANT_WITHIN_COALITION: Final[str] = "log_price_within_coalition_positive_and_significant"
# Compose the declared invariants of the sweep: the outcome, the scalar and the
# required sign.
INVARIANTS: Final[Tuple[Tuple[str, str, str, int], ...]] = (
    (INVARIANT_GROWTH, OUTCOME_GROWTH, NAME_BETA_3, -1),
    (INVARIANT_WITHIN_COALITION, OUTCOME_LOG_PRICE, NAME_COMPOSITE, 1),
)


# Declare the frozen result of the heterogeneity execution.
@dataclass(frozen=True)
class HeterogeneityExecutionResult:
    """
    Frozen result of the exposure heterogeneity execution.

    Attributes
    ----------
    baseline_table : pandas.DataFrame
        One row per outcome and scalar carrying the baseline specification's
        estimates with their inference, keyed on the outcome, the exposure
        variant and the window.
    pre_crimea_table : pandas.DataFrame
        The same shape for the strict earlier-window Table 10 panel.
    table10 : pandas.DataFrame
        Both strict Table 10 panels with beta1, beta2, beta3, and the total
        coalition slope retained separately.
    sweep_table : pandas.DataFrame
        One row per specification, outcome and scalar carrying the estimates,
        the inference, the significance flags and the sample counts.
    covariance_table : pandas.DataFrame
        Every 3-by-3 coefficient covariance from every outcome/specification
        cell in the full B7 sweep, with estimator and sample lineage.
    invariant_checks : pandas.DataFrame
        One row per invariant, specification and outcome recording the sign,
        the significance and whether the declared directional requirement
        is supported as a descriptive finding.
    window_comparison : pandas.DataFrame
        The actual baseline-versus-pre-Crimea comparison, including every
        available coverage rule and all four Table 10 scalars.
    fingerprint_sha256 : str
        Digest of the canonical execution payload.
    built_utc : str
        ISO-8601 UTC timestamp of the execution.
    notes : tuple of str
        Recorded decisions, including the significance convention.
    """
    # The baseline specification's table.
    baseline_table: pd.DataFrame
    # The pre-Crimea specification's table.
    pre_crimea_table: pd.DataFrame
    # Both panels of Table 10.
    table10: pd.DataFrame
    # The six-specification sweep.
    sweep_table: pd.DataFrame
    # The complete per-specification covariance ledger.
    covariance_table: pd.DataFrame
    # The declared invariants' checks.
    invariant_checks: pd.DataFrame
    # The actual window-comparison object.
    window_comparison: pd.DataFrame
    # Digest of the canonical execution payload.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the execution.
    built_utc: str
    # Recorded decisions of the execution.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 29, Step 1: The baseline and pre-Crimea tables
# -------------------------------------------------------------------------------------------------------------------------------


# Reconcile every scalar with its retained full covariance and fitted lineage.
def _assert_heterogeneity_suite_identities(
    suite: HeterogeneityResult,
    outcome: str,
    specifications: Sequence[SpecificationRecord],
) -> None:
    """Fail closed on a coefficient/covariance/sample lineage inconsistency.

    These are mathematical/dataflow identities, unlike the narrative signs
    evaluated later.  In particular the total coalition slope must equal
    beta2+beta3 and use V22+V33+2V23, while every covariance cell must describe
    the same policy/outcome/sample/estimator fit as its coefficient rows.
    """
    scalar_order = (NAME_BETA_1, NAME_BETA_2, NAME_BETA_3)
    for specification in specifications:
        coefficient_mask = (
            (suite.coefficients.index.get_level_values("specification") == specification.name)
            & (suite.coefficients.index.get_level_values("outcome") == str(outcome))
        )
        covariance_mask = (
            (suite.covariance.index.get_level_values("specification") == specification.name)
            & (suite.covariance.index.get_level_values("outcome") == str(outcome))
        )
        coefficient_rows = suite.coefficients[coefficient_mask]
        covariance_rows = suite.covariance[covariance_mask]
        statistic_rows = suite.statistics[
            (suite.statistics.index.get_level_values("specification") == specification.name)
            & (suite.statistics.index.get_level_values("outcome") == str(outcome))
        ]
        if len(coefficient_rows) != 4 or len(covariance_rows) != 9 or len(statistic_rows) != 1:
            raise IngestError(
                f"{specification.name}/{outcome}: incomplete coefficient/covariance/statistics identity."
            )
        coefficients_by_scalar = {
            str(key[-1]): row for key, row in coefficient_rows.iterrows()
        }
        if set(coefficients_by_scalar) != set(TABLE10_SCALARS):
            raise IngestError(f"{specification.name}/{outcome}: incomplete scalar vocabulary.")
        covariance_matrix = np.empty((3, 3), dtype=float)
        for row_position, row_name in enumerate(scalar_order):
            for column_position, column_name in enumerate(scalar_order):
                cell = covariance_rows[
                    (covariance_rows.index.get_level_values("row_scalar") == row_name)
                    & (covariance_rows.index.get_level_values("column_scalar") == column_name)
                ]
                if len(cell) != 1:
                    raise IngestError(
                        f"{specification.name}/{outcome}/{row_name}/{column_name}: covariance cell missing or duplicated."
                    )
                covariance_matrix[row_position, column_position] = float(cell.iloc[0]["value"])
        beta = np.array(
            [float(coefficients_by_scalar[name]["estimate"]) for name in scalar_order],
            dtype=float,
        )
        for position, name in enumerate(scalar_order):
            expected_se = float(np.sqrt(max(float(covariance_matrix[position, position]), 0.0)))
            observed_se = float(coefficients_by_scalar[name]["se_cluster"])
            if not np.isclose(observed_se, expected_se, rtol=1.0e-10, atol=1.0e-12):
                raise IngestError(f"{specification.name}/{outcome}/{name}: SE/covariance identity failed.")
        composite = coefficients_by_scalar[NAME_COMPOSITE]
        composite_variance = float(
            covariance_matrix[1, 1]
            + covariance_matrix[2, 2]
            + 2.0 * covariance_matrix[1, 2]
        )
        if composite_variance < -1.0e-12:
            raise IngestError(f"{specification.name}/{outcome}: composite variance is negative.")
        if not np.isclose(float(composite["estimate"]), float(beta[1] + beta[2]), rtol=1.0e-10, atol=1.0e-12) or not np.isclose(
            float(composite["se_cluster"]),
            float(np.sqrt(max(composite_variance, 0.0))),
            rtol=1.0e-10,
            atol=1.0e-12,
        ):
            raise IngestError(f"{specification.name}/{outcome}: total-slope identity failed.")
        statistic = statistic_rows.iloc[0]
        lineage_columns = (
            "donor_roster_sha256", "sample_roster_sha256",
            "selection_fingerprint_sha256",
        )
        for column in lineage_columns:
            expected_lineage = str(statistic[column])
            if set(coefficient_rows[column].astype(str)) != {expected_lineage} or set(
                covariance_rows[column].astype(str)
            ) != {expected_lineage}:
                raise IngestError(
                    f"{specification.name}/{outcome}: {column} diverges across fitted artefacts."
                )


# Slice one specification's table out of a suite.
def slice_specification_table(
    suites: Mapping[str, HeterogeneityResult],
    specification_name: str,
    scalars: Sequence[str],
) -> pd.DataFrame:
    """
    Slice one specification's table out of a suite.

    Parameters
    ----------
    suites : mapping of str to HeterogeneityResult
        The per-outcome suites keyed on the outcome.
    specification_name : str
        The specification whose rows are sliced.
    scalars : sequence of str
        The reported scalars the table carries.

    Returns
    -------
    pandas.DataFrame
        One row per outcome and scalar carrying the estimate, its standard
        error and its probability, together with the sample counts and the
        significance flag.

    Raises
    ------
    IngestError
        If the specification is absent from a suite, if a scalar is absent from
        a specification, or if a specification's rows are duplicated, because a
        table would then overstate or understate its own evidence.
    """
    # Accumulate the table rows.
    rows: List[dict] = []
    # Iterate over the outcomes in sorted order.
    for outcome in sorted(suites.keys()):
        # Read the suite.
        suite: HeterogeneityResult = suites[outcome]
        # Iterate over the reported scalars.
        for scalar in scalars:
            # Select the specification's rows for the scalar.
            scoped = suite.coefficients[
                (suite.coefficients.index.get_level_values(0) == specification_name)
                & (suite.coefficients.index.get_level_values(2) == outcome)
                & (suite.coefficients.index.get_level_values(-1) == scalar)
            ]
            # Refuse a missing or duplicated row.
            if len(scoped) != 1:
                # The failure names the specification, the outcome and the scalar.
                raise IngestError(
                    f"{specification_name}/{outcome}/{scalar}: {len(scoped)} row(s) "
                    f"found where exactly one is required."
                )
            # Read the row.
            row = scoped.iloc[0]
            covariance_scope = suite.covariance[
                (suite.covariance.index.get_level_values("specification") == specification_name)
                & (suite.covariance.index.get_level_values("outcome") == outcome)
            ]
            if len(covariance_scope) != 9:
                raise IngestError(
                    f"{specification_name}/{outcome}: incomplete covariance provenance."
                )
            covariance_fingerprint = _content_digest(covariance_scope.sort_index())
            # Read the key's levels for the table's own columns.
            levels = scoped.index[0]
            # Append the table row.
            rows.append(
                {
                    "outcome": str(outcome),
                    "exposure_variant": str(levels[3]),
                    "window": str(levels[4]),
                    "coverage_rule": str(levels[5]),
                    "scalar": str(scalar),
                    "estimate": float(row["estimate"]),
                    "se_cluster": float(row["se_cluster"]),
                    "p_value": float(row["p_value"]),
                    "significant_5pct": bool(
                        np.isfinite(float(row["p_value"])) and float(row["p_value"]) < SIGNIFICANCE_LEVEL
                    ),
                    "observations": int(row["observations"]),
                    "clusters": int(row["clusters"]),
                    "estimator": str(row["estimator"]),
                    "sanctioners": int(row["sanctioners"]),
                    "donor_count": int(row["donor_count"]),
                    "donor_roster_sha256": str(row["donor_roster_sha256"]),
                    "sample_roster_sha256": str(row["sample_roster_sha256"]),
                    "selection_fingerprint_sha256": str(row["selection_fingerprint_sha256"]),
                    "covariance_fingerprint_sha256": covariance_fingerprint,
                    "residual_df": int(row["residual_df"]),
                    "absorbed_rank": int(row["absorbed_rank"]),
                    "full_design_rank": int(row["full_design_rank"]),
                    "cluster_correction": float(row["cluster_correction"]),
                }
            )
    # Assemble the table keyed on the outcome, the variant, the window, the rule
    # and the scalar.
    table: pd.DataFrame = pd.DataFrame(rows).set_index(
        ["outcome", "exposure_variant", "window", "coverage_rule", "scalar"]
    )
    # Log the table's extent.
    logger.info(
        "%s: table sliced across %d row(s).", specification_name, len(table)
    )
    # Return the table.
    return table


# -------------------------------------------------------------------------------------------------------------------------------
# Task 29, Step 3: The sweep and the declared invariants
# -------------------------------------------------------------------------------------------------------------------------------


# Evaluate the declared directional invariants over the sweep.
def evaluate_invariants(
    sweep_table: pd.DataFrame,
    invariants: Sequence[Tuple[str, str, str, int]] = INVARIANTS,
) -> pd.DataFrame:
    """
    Evaluate narrative directional claims over the validated sweep.

    Parameters
    ----------
    sweep_table : pandas.DataFrame
        The sweep table, keyed so that the specification is a level.
    invariants : sequence of tuple, optional
        The declared empirical claims, each carrying its name, outcome,
        scalar, and narrative direction.

    Returns
    -------
    pandas.DataFrame
        One uniquely keyed row per claim and specification.  Each row records
        its scalar-specific estimate, covariance-based inference, complete
        sample provenance, mathematical/dataflow validity, and whether the
        descriptive sign/significance pattern is supported.  A descriptive
        finding is never promoted to a production validity requirement.

    Raises
    ------
    IngestError
        If keys are duplicated, scientific fields are malformed, or a claim
        names a scalar/outcome absent from the requested complete sweep.
    """
    required_levels = {
        "specification", "policy", "outcome", "exposure_variant", "window",
        "coverage_rule", "scalar",
    }
    if len(sweep_table) == 0 or not isinstance(sweep_table.index, pd.MultiIndex) or not required_levels.issubset(
        set(sweep_table.index.names)
    ) or not bool(sweep_table.index.is_unique):
        raise IngestError("The heterogeneity sweep must have unique complete scientific keys.")
    required_columns = {
        "estimate", "se_cluster", "p_value", "observations", "clusters",
        "sanctioners", "estimator", "donor_roster_sha256",
        "sample_roster_sha256", "selection_fingerprint_sha256",
        "covariance_fingerprint_sha256",
    }
    missing_columns = sorted(required_columns - set(sweep_table.columns))
    if missing_columns:
        raise IngestError(f"The heterogeneity sweep omits claim provenance {missing_columns!r}.")
    claims = list(invariants)
    claim_names: List[str] = []
    for claim in claims:
        if not isinstance(claim, tuple) or len(claim) != 4:
            raise IngestError("Each empirical claim must be a four-item tuple.")
        name, outcome, scalar, required_sign = claim
        try:
            resolved_sign = int(required_sign)
        except (TypeError, ValueError) as exc:
            raise IngestError(f"Malformed empirical claim {claim!r}.") from exc
        if isinstance(required_sign, (bool, np.bool_)) or not str(name).strip() or resolved_sign not in (-1, 1):
            raise IngestError(f"Malformed empirical claim {claim!r}.")
        if scalar not in TABLE10_SCALARS:
            raise IngestError(f"{name}: undeclared claim scalar {scalar!r}.")
        claim_names.append(str(name))
    if len(set(claim_names)) != len(claim_names):
        raise IngestError("Empirical claim names are duplicated.")
    rows: List[dict] = []
    specifications = sorted(
        set(str(value) for value in sweep_table.index.get_level_values("specification"))
    )
    for name, outcome, scalar, required_sign in claims:
        for specification in specifications:
            scoped = sweep_table[
                (sweep_table.index.get_level_values("specification") == specification)
                & (sweep_table.index.get_level_values("outcome") == outcome)
                & (sweep_table.index.get_level_values("scalar") == scalar)
            ]
            if len(scoped) != 1:
                raise IngestError(
                    f"{name}/{specification}/{outcome}/{scalar}: {len(scoped)} row(s) "
                    "found where exactly one is required."
                )
            row = scoped.iloc[0]
            key = scoped.index[0]
            estimate = float(row["estimate"])
            standard_error = float(row["se_cluster"])
            probability = float(row["p_value"])
            if not np.isfinite(estimate) or not np.isfinite(standard_error) or standard_error < 0.0:
                raise IngestError(f"{name}/{specification}: coefficient inference is malformed.")
            if np.isfinite(probability):
                if probability < 0.0 or probability > 1.0:
                    raise IngestError(f"{name}/{specification}: probability is outside [0, 1].")
            elif standard_error > 0.0:
                raise IngestError(f"{name}/{specification}: positive-SE inference lacks a probability.")
            sign = int(np.sign(estimate))
            sign_matches = bool(sign == int(required_sign))
            significant = bool(np.isfinite(probability) and probability < SIGNIFICANCE_LEVEL)
            supported = bool(sign_matches and significant)
            rows.append(
                {
                    "check_class": CHECK_CLASS_DESCRIPTIVE,
                    "claim": str(name),
                    "specification": str(specification),
                    "outcome": str(outcome),
                    "scalar": str(scalar),
                    "policy": str(key[sweep_table.index.names.index("policy")]),
                    "exposure_variant": str(key[sweep_table.index.names.index("exposure_variant")]),
                    "window": str(key[sweep_table.index.names.index("window")]),
                    "coverage_rule": str(key[sweep_table.index.names.index("coverage_rule")]),
                    "estimator": str(row["estimator"]),
                    "estimate": estimate,
                    "se_cluster": standard_error,
                    "p_value": probability,
                    "sign": sign,
                    "required_sign": int(required_sign),
                    "sign_matches": sign_matches,
                    "significant_5pct": significant,
                    "claim_supported": supported,
                    "holds": supported,
                    "dataflow_valid": True,
                    "required_for_validity": False,
                    "enforcement": "report_only",
                    "observations": int(row["observations"]),
                    "clusters": int(row["clusters"]),
                    "sanctioners": int(row["sanctioners"]),
                    "donor_roster_sha256": str(row["donor_roster_sha256"]),
                    "sample_roster_sha256": str(row["sample_roster_sha256"]),
                    "selection_fingerprint_sha256": str(row["selection_fingerprint_sha256"]),
                    "covariance_fingerprint_sha256": str(row["covariance_fingerprint_sha256"]),
                    "finding": (
                        "supported"
                        if supported
                        else "direction_matches_but_not_significant"
                        if sign_matches
                        else "direction_not_supported"
                    ),
                }
            )
    checks = pd.DataFrame(rows).set_index(
        ["check_class", "claim", "specification", "outcome", "scalar"]
    ).sort_index()
    if not bool(checks.index.is_unique):
        raise IngestError("Empirical claim-check keys are duplicated.")
    logger.info(
        "descriptive claims evaluated across %d check(s); %d supported; none are validity gates.",
        len(checks),
        int(checks["claim_supported"].sum()),
    )
    return checks


# Assert the declared invariants when the caller requires them.
def assert_invariants_required(
    checks: pd.DataFrame,
    required: bool = False,
) -> None:
    """
    Enforce mathematical/dataflow checks without forcing empirical findings.

    Parameters
    ----------
    checks : pandas.DataFrame
        A labelled check frame.  Rows classified ``descriptive_empirical`` are
        report-only.  Rows classified ``mathematical_dataflow`` may be validity
        gates when ``required_for_validity`` is true.
    required : bool, optional
        Legacy request to enforce validity checks.  It never changes an
        empirical sign/significance result into a mathematical requirement.

    Returns
    -------
    None
        The function returns nothing when every required invariant holds.

    Raises
    ------
    IngestError
        If a required mathematical/dataflow identity fails.  Descriptive
        empirical rows never trigger this exception, even when ``required`` is
        true.
    """
    if not isinstance(required, (bool, np.bool_)):
        raise IngestError("The invariant-enforcement switch must be Boolean.")
    if not isinstance(checks, pd.DataFrame) or len(checks) == 0 or not bool(
        checks.index.is_unique
    ):
        raise IngestError("The invariant/claim check frame is empty or has duplicate keys.")
    classes = (
        checks["check_class"].astype(str)
        if "check_class" in checks.columns
        else pd.Series(
            checks.index.get_level_values("check_class").astype(str), index=checks.index
        )
        if isinstance(checks.index, pd.MultiIndex) and "check_class" in checks.index.names
        else pd.Series(CHECK_CLASS_DESCRIPTIVE, index=checks.index)
    )
    allowed = {CHECK_CLASS_DESCRIPTIVE, CHECK_CLASS_MATHEMATICAL}
    if not set(classes.unique()).issubset(allowed):
        raise IngestError(f"Undeclared invariant-check class(es): {sorted(set(classes) - allowed)!r}.")
    descriptive = checks[classes == CHECK_CLASS_DESCRIPTIVE]
    if len(descriptive):
        supported_column = (
            "claim_supported" if "claim_supported" in descriptive.columns else "holds"
        )
        if supported_column not in descriptive.columns:
            raise IngestError("Descriptive claim checks omit their support status.")
        if not bool(
            descriptive[supported_column].map(
                lambda value: isinstance(value, (bool, np.bool_))
            ).all()
        ):
            raise IngestError("Descriptive claim support states must be Boolean.")
        logger.info(
            "descriptive findings recorded: %d of %d supported; none are forced.",
            int(descriptive[supported_column].astype(bool).sum()),
            len(descriptive),
        )
    mathematical = checks[classes == CHECK_CLASS_MATHEMATICAL]
    if len(mathematical) == 0:
        return
    if "dataflow_valid" not in mathematical.columns:
        raise IngestError("Mathematical/dataflow checks omit dataflow_valid.")
    if not bool(
        mathematical["dataflow_valid"].map(
            lambda value: isinstance(value, (bool, np.bool_))
        ).all()
    ):
        raise IngestError("Mathematical dataflow-valid states must be Boolean.")
    if "required_for_validity" in mathematical.columns and not bool(
        mathematical["required_for_validity"].map(
            lambda value: isinstance(value, (bool, np.bool_))
        ).all()
    ):
        raise IngestError("Mathematical required-for-validity states must be Boolean.")
    required_rows = (
        mathematical["required_for_validity"].astype(bool)
        if "required_for_validity" in mathematical.columns
        else pd.Series(bool(required), index=mathematical.index)
    )
    violated = mathematical[required_rows & ~mathematical["dataflow_valid"].astype(bool)]
    if len(violated):
        raise IngestError(
            "Required mathematical/dataflow checks failed for: "
            + "; ".join("/".join(str(value) for value in index) if isinstance(index, tuple) else str(index) for index in violated.index)
            + "."
        )
    logger.info("all %d required mathematical/dataflow check(s) hold.", int(required_rows.sum()))


# -------------------------------------------------------------------------------------------------------------------------------
# Task 29, Step 2: The pre-Crimea specification under both rules
# -------------------------------------------------------------------------------------------------------------------------------


# Compare the two windows' samples side by side.
def build_window_comparison(
    baseline_table: pd.DataFrame,
    pre_crimea_table: pd.DataFrame,
) -> pd.DataFrame:
    """
    Compare the two windows' samples side by side.

    Parameters
    ----------
    baseline_table : pandas.DataFrame
        The baseline window's table.
    pre_crimea_table : pandas.DataFrame
        The earlier window's table.

    Returns
    -------
    pandas.DataFrame
        One row per outcome, Table 10 scalar, and available earlier-window
        coverage rule.  Both estimates, covariance-based inference, exact
        fitted counts, and sample/donor lineage remain visible.

    Raises
    ------
    IngestError
        If either table omits an outcome the other carries, because the
        comparison would then be partial.
    """
    if not bool(baseline_table.index.is_unique) or not bool(pre_crimea_table.index.is_unique):
        raise IngestError("A window-comparison input has duplicated scientific keys.")
    # Compose the baseline outcomes.
    baseline_outcomes = set(baseline_table.index.get_level_values("outcome"))
    # Compose the earlier window's outcomes.
    earlier_outcomes = set(pre_crimea_table.index.get_level_values("outcome"))
    # Refuse a partial overlap.
    if baseline_outcomes != earlier_outcomes:
        # The failure names the divergence.
        missing = sorted(baseline_outcomes ^ earlier_outcomes)
        # Raise with the divergent outcomes attached.
        raise IngestError(f"The two windows disagree on the outcomes: {missing}.")
    earlier_rules = sorted(
        set(str(value) for value in pre_crimea_table.index.get_level_values("coverage_rule"))
    )
    if RULE_STRICT not in earlier_rules:
        raise IngestError("The earlier-window comparison omits its strict sample.")
    # Accumulate the comparison rows.
    rows: List[dict] = []
    # Iterate over the outcomes in sorted order.
    for outcome in sorted(baseline_outcomes):
        # Iterate over every separately reported Table 10 scalar.
        for scalar in TABLE10_SCALARS:
            # Select the baseline's row.
            baseline = baseline_table[
                (baseline_table.index.get_level_values("outcome") == outcome)
                & (baseline_table.index.get_level_values("scalar") == scalar)
            ]
            if len(baseline) != 1:
                raise IngestError(
                    f"baseline/{outcome}/{scalar}: {len(baseline)} row(s), expected one."
                )
            baseline_row = baseline.iloc[0]
            # Accumulate the earlier window's rows under every executed rule.
            for rule in earlier_rules:
                # Select the earlier window's row.
                earlier = pre_crimea_table[
                    (pre_crimea_table.index.get_level_values("outcome") == outcome)
                    & (pre_crimea_table.index.get_level_values("scalar") == scalar)
                    & (pre_crimea_table.index.get_level_values("coverage_rule") == rule)
                ]
                if len(earlier) != 1:
                    raise IngestError(
                        f"pre_crimea/{outcome}/{scalar}/{rule}: {len(earlier)} row(s), expected one."
                    )
                # Read the earlier row.
                earlier_row = earlier.iloc[0]
                # Append the comparison row.
                rows.append(
                    {
                        "outcome": str(outcome),
                        "scalar": str(scalar),
                        "coverage_rule": str(rule),
                        "baseline_estimate": float(baseline_row["estimate"]),
                        "baseline_se_cluster": float(baseline_row["se_cluster"]),
                        "baseline_observations": int(baseline_row["observations"]),
                        "baseline_clusters": int(baseline_row["clusters"]),
                        "baseline_sanctioners": int(baseline_row["sanctioners"]),
                        "baseline_p_value": float(baseline_row["p_value"]),
                        "baseline_donor_roster_sha256": str(baseline_row["donor_roster_sha256"]),
                        "baseline_sample_roster_sha256": str(baseline_row["sample_roster_sha256"]),
                        "baseline_selection_fingerprint_sha256": str(baseline_row["selection_fingerprint_sha256"]),
                        "earlier_estimate": float(earlier_row["estimate"]),
                        "earlier_se_cluster": float(earlier_row["se_cluster"]),
                        "earlier_observations": int(earlier_row["observations"]),
                        "earlier_clusters": int(earlier_row["clusters"]),
                        "earlier_sanctioners": int(earlier_row["sanctioners"]),
                        "earlier_p_value": float(earlier_row["p_value"]),
                        "earlier_donor_roster_sha256": str(earlier_row["donor_roster_sha256"]),
                        "earlier_sample_roster_sha256": str(earlier_row["sample_roster_sha256"]),
                        "earlier_selection_fingerprint_sha256": str(earlier_row["selection_fingerprint_sha256"]),
                        "estimate_difference": float(earlier_row["estimate"])
                        - float(baseline_row["estimate"]),
                        "sample_difference": int(earlier_row["observations"])
                        - int(baseline_row["observations"]),
                    }
                )
    # Assemble the comparison frame.
    comparison: pd.DataFrame = pd.DataFrame(rows).set_index(
        ["outcome", "scalar", "coverage_rule"]
    )
    # Log the comparison's extent.
    logger.info("window comparison built across %d row(s).", len(comparison))
    # Return the frame.
    return comparison


# Freeze the execution behind a canonical digest.
def freeze_heterogeneity_execution(
    sweep_table: pd.DataFrame,
    invariant_checks: pd.DataFrame,
    table10: Optional[pd.DataFrame] = None,
    covariance_table: Optional[pd.DataFrame] = None,
    window_comparison: Optional[pd.DataFrame] = None,
) -> str:
    """
    Freeze the execution behind a canonical digest.

    Parameters
    ----------
    sweep_table : pandas.DataFrame
        The sweep table.
    invariant_checks : pandas.DataFrame
        The invariant check frame.
    table10, covariance_table, window_comparison : pandas.DataFrame, optional
        Additional Task 29 scientific artefacts.  The orchestrator supplies all
        three; optionality preserves the direct two-frame freezing interface.

    Returns
    -------
    str
        The hexadecimal digest of the canonical payload over the sorted keys.

    Raises
    ------
    IngestError
        If either frame is empty.
    """
    # Refuse an empty frame.
    for label, frame in (("sweep", sweep_table), ("invariant checks", invariant_checks)):
        # Test the frame's extent.
        if len(frame) == 0:
            # The failure names the frame.
            raise IngestError(f"Cannot freeze an empty {label} frame.")
    if not bool(sweep_table.index.is_unique) or not bool(invariant_checks.index.is_unique):
        raise IngestError("The heterogeneity execution carries duplicated audit keys.")
    payload = {
        "sweep": _content_digest(sweep_table.sort_index()),
        "invariant_checks": _content_digest(invariant_checks.sort_index()),
    }
    for label, frame in (
        ("table10", table10),
        ("covariance_table", covariance_table),
        ("window_comparison", window_comparison),
    ):
        if frame is not None:
            if len(frame) == 0 or not bool(frame.index.is_unique):
                raise IngestError(f"Cannot freeze an empty or duplicate-keyed {label} frame.")
            payload[label] = _content_digest(frame.sort_index())
    return hashlib.sha256(
        json.dumps(payload, sort_keys=True, separators=(",", ":")).encode("utf-8")
    ).hexdigest()


# -------------------------------------------------------------------------------------------------------------------------------
# Task 29, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Execute the exposure heterogeneity regressions.
def execute_heterogeneity_regressions(
    panel: pd.DataFrame,
    exposures: pd.DataFrame,
    sanction_indicators: pd.DataFrame,
    donor_codes: Any,
    outcomes: Sequence[str] = PANEL_OUTCOMES,
    specifications: Sequence[SpecificationRecord] = SPECIFICATIONS,
    required_invariants: bool = False,
    expected_estimates: Optional[Mapping[Tuple[str, str, str], float]] = None,
    tolerance: float = 1.0e-3,
) -> HeterogeneityExecutionResult:
    """
    Execute the exposure heterogeneity regressions.

    Parameters
    ----------
    panel : pandas.DataFrame
        The frozen panel carrying every outcome.
    exposures : pandas.DataFrame
        The standardized exposures keyed on the country and the window.
    sanction_indicators : pandas.DataFrame
        The sanction indicators keyed on the country and the year.
    donor_codes : sequence of str or mapping of str to sequence of str
        The policy-specific non-policy group(s), forwarded without collapsing
        sanctions and direct-arming donor identities.
    outcomes : sequence of str, optional
        The outcomes the execution covers. Task 29 is a complete Table 10/B7
        execution and therefore requires each of the declared four exactly
        once; the parameter preserves ordering but not subset execution.
    specifications : sequence of SpecificationRecord, optional
        The specifications the sweep covers. Task 29 requires the exact six
        registered B7 designs, although caller ordering may differ.
    required_invariants : bool, optional
        Whether a violated directional invariant is refused rather than
        recorded as a finding.
    expected_estimates : mapping of tuple to float, optional
        The declared estimates keyed on the specification, the outcome and the
        scalar, compared only when supplied.
    tolerance : float, optional
        The tolerance of the declared comparison.

    Returns
    -------
    HeterogeneityExecutionResult
        The baseline table, the pre-Crimea table, the sweep table, the
        invariant checks, the digest, the timestamp and the notes.

    Raises
    ------
    IngestError
        If required outcome/specification coverage or a mathematical/dataflow
        identity fails, or if a supplied declared numeric estimate diverges.
        Narrative sign/significance findings never cause rejection.
    """
    requested_outcomes = tuple(str(value) for value in outcomes)
    if len(set(requested_outcomes)) != len(requested_outcomes) or set(
        requested_outcomes
    ) != set(PANEL_OUTCOMES):
        raise IngestError(
            "Task 29 requires each of the four declared outcomes exactly once."
        )
    requested_specifications = tuple(specifications)
    if len({str(value.name) for value in requested_specifications}) != len(
        requested_specifications
    ):
        raise IngestError("Task 29 specification names are duplicated.")
    canonical_by_name = {record.name: record for record in SPECIFICATIONS}
    if set(record.name for record in requested_specifications) != set(canonical_by_name):
        raise IngestError("Task 29 requires the complete six-specification B7 design.")
    for record in requested_specifications:
        if record != canonical_by_name[record.name]:
            raise IngestError(
                f"Task 29 specification {record.name!r} conflicts with its registered design."
            )
    resolved_expected: Dict[Tuple[str, str, str], float] = {}
    if expected_estimates is not None:
        for key, value in expected_estimates.items():
            if not isinstance(key, tuple) or len(key) != 3:
                raise IngestError(
                    "Expected heterogeneity estimates must be keyed on "
                    "(specification, outcome, scalar)."
                )
            if key[0] not in canonical_by_name or key[1] not in PANEL_OUTCOMES or key[2] not in TABLE10_SCALARS:
                raise IngestError(f"Undeclared expected heterogeneity cell {key!r}.")
            if isinstance(value, (bool, np.bool_)):
                raise IngestError(f"{key!r}: expected estimate must be finite numeric.")
            try:
                numeric_value = float(value)
            except (TypeError, ValueError) as exc:
                raise IngestError(f"{key!r}: expected estimate must be finite numeric.") from exc
            if not np.isfinite(numeric_value):
                raise IngestError(f"{key!r}: expected estimate must be finite numeric.")
            resolved_expected[(str(key[0]), str(key[1]), str(key[2]))] = numeric_value
    # Accumulate the per-outcome suites.
    suites: dict = {}
    # Accumulate the sweep rows.
    sweep_rows: List[dict] = []
    # Accumulate the complete covariance ledgers without reducing them to
    # standard errors.
    covariance_frames: List[pd.DataFrame] = []
    # Iterate over the outcomes, running the heterogeneity orchestrator once per
    # outcome so every table derives from a single execution.
    for outcome in requested_outcomes:
        # Refuse an outcome the panel lacks.
        if str(outcome) not in panel.columns:
            # The failure names the outcome.
            raise IngestError(f"The panel lacks the outcome {outcome!r}.")
        # Run the suite for the outcome.
        suite: HeterogeneityResult = execute_heterogeneity_suite(
            panel,
            str(outcome),
            exposures,
            sanction_indicators,
            donor_codes,
            specifications=specifications,
            expected_slopes=(
                {
                    (name, scalar): float(value)
                    for (name, target_outcome, scalar), value in resolved_expected.items()
                    if str(target_outcome) == str(outcome)
                }
                if expected_estimates is not None
                else None
            ),
            tolerance=tolerance,
        )
        # Enforce only mathematical/dataflow identities at production time;
        # narrative directions are evaluated later as report-only findings.
        _assert_heterogeneity_suite_identities(
            suite, str(outcome), requested_specifications
        )
        # Record the suite.
        suites[str(outcome)] = suite
        covariance_frames.append(suite.covariance.copy(deep=True))
        # Append the sweep rows of the outcome.
        for key, row in suite.coefficients.iterrows():
            covariance_scope = suite.covariance[
                (suite.covariance.index.get_level_values("specification") == str(key[0]))
                & (suite.covariance.index.get_level_values("outcome") == str(key[2]))
            ]
            if len(covariance_scope) != 9:
                raise IngestError(
                    f"{key[0]}/{key[2]}: incomplete covariance provenance in B7."
                )
            covariance_fingerprint = _content_digest(covariance_scope.sort_index())
            # Record the row with its specification level.
            sweep_rows.append(
                {
                    "specification": str(key[0]),
                    "policy": str(key[1]),
                    "outcome": str(key[2]),
                    "exposure_variant": str(key[3]),
                    "window": str(key[4]),
                    "coverage_rule": str(key[5]),
                    "scalar": str(key[6]),
                    "estimate": float(row["estimate"]),
                    "se_cluster": float(row["se_cluster"]),
                    "p_value": float(row["p_value"]),
                    "significant_5pct": bool(
                        np.isfinite(float(row["p_value"]))
                        and float(row["p_value"]) < SIGNIFICANCE_LEVEL
                    ),
                    "observations": int(row["observations"]),
                    "clusters": int(row["clusters"]),
                    "estimator": str(row["estimator"]),
                    "sanctioners": int(row["sanctioners"]),
                    "donor_count": int(row["donor_count"]),
                    "donor_roster_sha256": str(row["donor_roster_sha256"]),
                    "sample_roster_sha256": str(row["sample_roster_sha256"]),
                    "selection_fingerprint_sha256": str(row["selection_fingerprint_sha256"]),
                    "covariance_fingerprint_sha256": covariance_fingerprint,
                    "residual_df": int(row["residual_df"]),
                    "absorbed_rank": int(row["absorbed_rank"]),
                    "full_design_rank": int(row["full_design_rank"]),
                    "cluster_correction": float(row["cluster_correction"]),
                }
            )
    # Assemble the sweep table.
    sweep_table: pd.DataFrame = pd.DataFrame(sweep_rows).set_index(
        [
            "specification",
            "policy",
            "outcome",
            "exposure_variant",
            "window",
            "coverage_rule",
            "scalar",
        ]
    )
    covariance_table = pd.concat(covariance_frames, axis=0).sort_index()
    if not bool(sweep_table.index.is_unique) or not bool(covariance_table.index.is_unique):
        raise IngestError("Task 29 coefficient or covariance keys are duplicated.")
    expected_coefficient_rows = len(requested_outcomes) * len(requested_specifications) * len(
        TABLE10_SCALARS
    )
    expected_covariance_rows = len(requested_outcomes) * len(requested_specifications) * 9
    if len(sweep_table) != expected_coefficient_rows or len(covariance_table) != expected_covariance_rows:
        raise IngestError(
            "Task 29 output coverage is incomplete: "
            f"coefficients={len(sweep_table)}/{expected_coefficient_rows}, "
            f"covariances={len(covariance_table)}/{expected_covariance_rows}."
        )
    # Assert the sweep's coverage of every specification and outcome.
    for specification in specifications:
        # Select the specification's rows.
        scoped = sweep_table[
            sweep_table.index.get_level_values("specification") == specification.name
        ]
        # Refuse a specification without rows for every outcome.
        if set(scoped.index.get_level_values("outcome")) != set(requested_outcomes):
            # The failure names the specification.
            raise IngestError(
                f"The sweep omits an outcome for the specification "
                f"{specification.name!r}."
            )
    # Slice the baseline specification's table.
    baseline_table: pd.DataFrame = slice_specification_table(
        suites, "baseline", TABLE10_SCALARS
    )
    # Slice the pre-Crimea specification's table.
    pre_crimea_table: pd.DataFrame = slice_specification_table(
        suites, "pre_crimea", TABLE10_SCALARS
    )
    extended_pre_crimea = slice_specification_table(
        suites, "pre_crimea_extended", TABLE10_SCALARS
    )
    comparison_pre_crimea = pd.concat(
        [pre_crimea_table, extended_pre_crimea], axis=0
    ).sort_index()
    if not bool(comparison_pre_crimea.index.is_unique):
        raise IngestError("Strict and extended earlier-window Table 10 keys collide.")
    # Assemble the two strict manuscript panels into one separately labelled
    # Table 10 without pooling their windows or samples.
    table10_parts: List[pd.DataFrame] = []
    for panel_name, panel_table in (
        ("panel_a_baseline", baseline_table),
        ("panel_b_pre_crimea", pre_crimea_table),
    ):
        labelled = panel_table.reset_index()
        labelled.insert(0, "panel", panel_name)
        table10_parts.append(
            labelled.set_index(
                ["panel", "outcome", "exposure_variant", "window", "coverage_rule", "scalar"]
            )
        )
    table10 = pd.concat(table10_parts, axis=0).sort_index()
    if len(table10) != 2 * len(requested_outcomes) * len(TABLE10_SCALARS) or not bool(
        table10.index.is_unique
    ):
        raise IngestError("Table 10 is incomplete or has duplicate scalar keys.")
    # Evaluate the declared invariants over the sweep.
    checks: pd.DataFrame = evaluate_invariants(sweep_table)
    # Assert the invariants when the caller requires them.
    assert_invariants_required(checks, required=required_invariants)
    # Compose the window comparison.
    comparison: pd.DataFrame = build_window_comparison(
        baseline_table, comparison_pre_crimea
    )
    # Freeze the execution.
    fingerprint: str = freeze_heterogeneity_execution(
        sweep_table,
        checks,
        table10=table10,
        covariance_table=covariance_table,
        window_comparison=comparison,
    )
    # Accumulate the recorded notes.
    notes: List[str] = [
        "each outcome runs the heterogeneity orchestrator once under every "
        "declared specification, so the four outcomes share the exposure "
        "measure and the post indicator and differ only in the dependent "
        "variable.",
        "the rate outcomes enter in percentage points, which is the "
        "estimation's own declared convention.",
        "an insignificant coefficient is reported with its probability and its "
        "significance flag rather than suppressed, so the earlier window's "
        "inconclusive interaction is visible as such.",
        f"the narrative sign/significance claims are evaluated at the {SIGNIFICANCE_LEVEL} "
        f"level and recorded as descriptive findings, never validity gates.",
    ]
    # Record the invariant outcomes.
    notes.append(
        f"descriptive claims supported: {int(checks['claim_supported'].sum())} of {len(checks)}."
    )
    # Record the windows compared.
    notes.append(
        f"the baseline window {WINDOW_BASELINE} is compared with the earlier "
        f"window {WINDOW_PRE_CRIMEA} across {len(comparison)} row(s)."
    )
    # Assemble the immutable result.
    result: HeterogeneityExecutionResult = HeterogeneityExecutionResult(
        baseline_table=baseline_table,
        pre_crimea_table=pre_crimea_table,
        table10=table10,
        sweep_table=sweep_table,
        covariance_table=covariance_table,
        invariant_checks=checks,
        window_comparison=comparison,
        fingerprint_sha256=fingerprint,
        built_utc=current_utc_iso(),
        notes=tuple(notes),
    )
    # Log the completed execution.
    logger.info(
        "heterogeneity execution complete: %d outcome(s), %d specification(s), "
        "%d sweep row(s), fingerprint %s.",
        len(suites),
        len(specifications),
        len(sweep_table),
        fingerprint[:16],
    )
    # Return the assembled result.
    return result
````

````python
# Task 30: Direct-Arming Coalition Pipeline Execution

# ==============================================================================
# Task 30: Direct-Arming Coalition Pipeline Execution
# ==============================================================================

# Declare the arming policy, whose pipeline this task executes.
POLICY_DIRECT_ARMING: Final[str] = "direct_arming"
# Compose the outcomes the arming pipeline reports.
ARMING_OUTCOMES: Final[Tuple[str, ...]] = ("log_CPI", "log_real_GDPpc")
# Declare the wave label of the first wave.
WAVE_ONE: Final[str] = WAVE_ONE_LABEL
# Declare the wave label of the second wave.
WAVE_TWO: Final[str] = WAVE_TWO_LABEL
# Compose the declared wave order.
WAVE_ORDER: Final[Tuple[str, str]] = (WAVE_ONE, WAVE_TWO)
# The first wave carries no reference window: it begins before the gap matrix
# does, so its timing figure is the level gap over the active years rather than
# an incremental form.
WAVE_ONE_REFERENCE: Final[Optional[Tuple[int, int]]] = None
# Declare the default reference window of the second wave.
WAVE_TWO_REFERENCE: Final[Tuple[int, int]] = (2019, 2021)
# Declare the interpretation label of a significant placebo probability.
INTERPRETATION_SIGNIFICANT: Final[str] = "detected_deviation_in_size_matched_placebo_diagnostic"
# Declare the interpretation label of a non-significant placebo probability,
# which the specification fixes is the correct diagnostic rather than a failure.
INTERPRETATION_CONSISTENT: Final[str] = "nondetection_in_size_matched_placebo_diagnostic"
# Declare the significance level the interpretation uses.
INTERPRETATION_LEVEL: Final[float] = 0.05
# Require the manuscript's registered inference resolution in every branch.
ARMING_INFERENCE_REPLICATES: Final[int] = 500
# Require the declared direct-arming coalition size.
ARMING_TARGET_COUNT: Final[int] = 29


# Declare the frozen result of the arming pipeline.
@dataclass(frozen=True)
class DirectArmingResult:
    """
    Frozen result of the direct-arming pipeline.

    Attributes
    ----------
    baseline : pandas.DataFrame
        One row per outcome carrying the two wave effects and the incremental
        effect with their inference, keyed on the policy and the outcome.
    inference : pandas.DataFrame
        One row per outcome and branch carrying the interval bounds and the
        corrected placebo probability, keyed on the policy, the outcome and
        the branch.
    interpretation : pandas.DataFrame
        One row per outcome recording the placebo probability's significance
        and its reading, which the specification fixes is a diagnostic rather
        than a headline test.
    active_year : pandas.DataFrame
        One row per outcome and wave carrying the active roster size, the B6
        equal-country level gap and its inference, plus a separately named
        preferred-reference incremental contrast when that reference exists.
    exclusions : pandas.DataFrame
        One row per pair and wave naming every country excluded from the
        active-window average for having no active year.
    design_audit : pandas.DataFrame
        One row per outcome binding every branch to its exact target/donor
        roster, documented removals, effective registry, and lineage digest.
    fingerprint_sha256 : str
        Digest of the canonical pipeline payload.
    built_utc : str
        ISO-8601 UTC timestamp of the pipeline.
    notes : tuple of str
        Recorded decisions, including the boundary-check reading.
    """
    # The baseline table.
    baseline: pd.DataFrame
    # The inference table.
    inference: pd.DataFrame
    # The placebo interpretation.
    interpretation: pd.DataFrame
    # The active-year timing table.
    active_year: pd.DataFrame
    # The active-year exclusions.
    exclusions: pd.DataFrame
    # The canonical effective-design audit.
    design_audit: pd.DataFrame
    # Digest of the canonical pipeline payload.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the pipeline.
    built_utc: str
    # Recorded decisions of the pipeline.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 30, Step 1: The arming baseline pipeline
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the arming partition of one outcome.
def compose_arming_partition(
    partition: Partition,
    donors_removed: Sequence[str] = (),
) -> Partition:
    """
    Compose the arming partition of one outcome.

    Parameters
    ----------
    partition : Partition
        The arming policy's registered partition.
    donors_removed : sequence of str, optional
        The donors the outcome's design drops, which the specification fixes
        applies to one outcome alone.

    Returns
    -------
    Partition
        The per-outcome partition, whose exclusion is confined to the outcome
        that declares it.

    Raises
    ------
    IngestError
        If a removed donor is absent from the pool, or if the removal leaves
        too few donors, because the design would then be undefined.
    """
    if str(partition.policy) != POLICY_DIRECT_ARMING:
        raise IngestError("An arming partition must carry the direct-arming policy identity.")
    removal_list = [str(code).strip() for code in donors_removed]
    if any(not code for code in removal_list) or len(removal_list) != len(set(removal_list)):
        raise IngestError("Outcome-specific donor removals are blank or duplicated.")
    # Compose the pool after the removals.
    removed = set(removal_list)
    # Refuse a removal that names a non-member.
    absent = sorted(removed - set(str(code) for code in partition.donors))
    # Raise when a removal names a non-member.
    if absent:
        # The failure names the absent donors.
        raise IngestError(
            f"The removal names donor(s) absent from the pool: {absent}."
        )
    # Compose the retained pool.
    donors: Tuple[str, ...] = tuple(
        str(code) for code in partition.donors if str(code) not in removed
    )
    # Refuse a pool below the declared minimum.
    if len(donors) < 2:
        # The failure names the retained count.
        raise IngestError(
            f"The removal leaves {len(donors)} donor(s), below the declared minimum."
        )
    # Log the composition's extent.
    logger.info(
        "%s: arming partition composed with %d donor(s) after removing %d.",
        partition.outcome,
        len(donors),
        len(removed),
    )
    exclusions = partition.exclusions.copy(deep=True)
    if len(exclusions) == 0 and len(exclusions.columns) == 0:
        exclusions = pd.DataFrame(
            columns=["policy", "outcome", "country_code", "reason", "detail"]
        )
    required_exclusion_columns = {"policy", "outcome", "country_code", "reason", "detail"}
    if not required_exclusion_columns.issubset(exclusions.columns):
        raise IngestError("The arming partition exclusion audit schema is incomplete.")
    added_rows = [
        {
            "policy": str(partition.policy),
            "outcome": str(partition.outcome),
            "country_code": code,
            "reason": REASON_ADDITIONAL_EXCLUSION,
            "detail": "Task30 documented outcome-specific donor removal",
        }
        for code in sorted(removed)
        if not bool(
            (
                exclusions["country_code"].astype(str).eq(code)
                & exclusions["reason"].astype(str).eq(REASON_ADDITIONAL_EXCLUSION)
            ).any()
        )
    ]
    if added_rows:
        exclusions = pd.concat([exclusions, pd.DataFrame(added_rows)], ignore_index=True)
    if bool(exclusions.duplicated(["policy", "outcome", "country_code", "reason"]).any()):
        raise IngestError("The effective arming exclusion audit contains duplicate identities.")
    # Return the per-outcome partition.
    return Partition(
        policy=partition.policy,
        outcome=partition.outcome,
        targets=partition.targets,
        donors=donors,
        exclusions=exclusions,
    )


# Resolve the one canonical effective direct-arming design of each outcome.
def _resolve_direct_arming_designs(
    data: Mapping[str, Any],
    outcomes: Sequence[str],
    outcome_removals: Optional[Mapping[str, Sequence[str]]] = None,
) -> Tuple[Mapping[str, Mapping[str, Any]], pd.DataFrame]:
    """Freeze outcome-specific partitions once for reuse by every Task 30 branch."""
    if "registry" not in data or not isinstance(data["registry"], PartitionRegistry):
        raise IngestError("Direct-arming design resolution requires a PartitionRegistry.")
    selected = tuple(str(outcome) for outcome in outcomes)
    if len(set(selected)) != len(selected) or set(selected) != set(ARMING_OUTCOMES):
        raise IngestError("Task 30 requires both declared direct-arming outcomes exactly once.")
    supplied = outcome_removals or {}
    if not isinstance(supplied, Mapping) or not set(str(key) for key in supplied).issubset(
        set(ARMING_OUTCOMES)
    ):
        raise IngestError("Direct-arming outcome-removal declarations are malformed.")
    registry: PartitionRegistry = data["registry"]
    if registry.fingerprint_sha256 != freeze_partition_registry(registry.registry):
        raise IngestError("The parent partition registry fingerprint is stale.")
    designs: Dict[str, Mapping[str, Any]] = {}
    audit_rows: List[Dict[str, Any]] = []
    target_reference: Optional[Tuple[str, ...]] = None
    for outcome in selected:
        pair = (POLICY_DIRECT_ARMING, outcome)
        if pair not in registry.partitions:
            raise IngestError(f"The registry does not carry the pair {pair}.")
        source = registry.partitions[pair]
        expected_removals = tuple(
            sorted(str(code) for code in ADDITIONAL_DONOR_EXCLUSIONS.get(pair, ()))
        )
        declared_removals = tuple(
            sorted(str(code).strip() for code in supplied.get(outcome, expected_removals))
        )
        if declared_removals != expected_removals:
            raise IngestError(
                f"{pair}: removal declaration {declared_removals!r} conflicts with "
                f"registered {expected_removals!r}."
            )
        already_removed: List[str] = []
        to_remove: List[str] = []
        exclusion_table = source.exclusions
        for code in expected_removals:
            if code in set(str(value) for value in source.donors):
                to_remove.append(code)
            else:
                if not isinstance(exclusion_table, pd.DataFrame) or not {
                    "country_code", "reason"
                }.issubset(exclusion_table.columns):
                    raise IngestError(f"{pair}: exclusion audit cannot prove removal of {code}.")
                recorded = exclusion_table[
                    exclusion_table["country_code"].astype(str).eq(code)
                    & exclusion_table["reason"].astype(str).eq(REASON_ADDITIONAL_EXCLUSION)
                ]
                if len(recorded) != 1:
                    raise IngestError(
                        f"{pair}: documented removal {code} is absent or duplicated in the audit."
                    )
                already_removed.append(code)
        effective = (
            compose_arming_partition(source, tuple(to_remove))
            if to_remove
            else Partition(
                policy=source.policy,
                outcome=source.outcome,
                targets=tuple(source.targets),
                donors=tuple(source.donors),
                exclusions=source.exclusions.copy(deep=True),
            )
        )
        targets = tuple(str(code) for code in effective.targets)
        donors = tuple(str(code) for code in effective.donors)
        if len(targets) != ARMING_TARGET_COUNT or len(set(targets)) != len(targets):
            raise IngestError(
                f"{pair}: direct-arming target roster must contain exactly "
                f"{ARMING_TARGET_COUNT} unique countries; observed {len(targets)}."
            )
        if len(set(donors)) != len(donors) or set(targets) & set(donors):
            raise IngestError(f"{pair}: target/donor rosters duplicate or overlap.")
        if any(code in set(donors) for code in expected_removals):
            raise IngestError(f"{pair}: a documented donor removal remains in the pool.")
        if target_reference is None:
            target_reference = targets
        elif set(target_reference) != set(targets):
            raise IngestError("Direct-arming target identities drift across outcomes.")
        variant_data = build_variant_data(data, pair[0], pair[1], effective)
        lineage = dict(variant_data["variant_lineage"])
        design_payload = {
            "pair": list(pair),
            "targets": list(targets),
            "donors": list(donors),
            "documented_removals": list(expected_removals),
            "partition_exclusions_sha256": _content_digest(
                effective.exclusions.sort_values(
                    [column for column in ("policy", "outcome", "country_code", "reason") if column in effective.exclusions.columns]
                ).reset_index(drop=True)
            ),
            "variant_registry_sha256": variant_data["registry"].fingerprint_sha256,
            "variant_lineage_sha256": lineage["lineage_sha256"],
        }
        design_sha256 = hashlib.sha256(
            json.dumps(design_payload, sort_keys=True, separators=(",", ":")).encode("utf-8")
        ).hexdigest()
        designs[outcome] = MappingProxyType(
            {
                "partition": effective,
                "data": variant_data,
                "design_sha256": design_sha256,
                "variant_registry_sha256": variant_data["registry"].fingerprint_sha256,
                "variant_lineage_sha256": lineage["lineage_sha256"],
            }
        )
        audit_rows.append(
            {
                "policy": pair[0],
                "outcome": pair[1],
                "target_count": len(targets),
                "donor_count": len(donors),
                "targets_json": json.dumps(list(targets)),
                "donors_json": json.dumps(list(donors)),
                "documented_removals_json": json.dumps(list(expected_removals)),
                "already_registered_removals_json": json.dumps(already_removed),
                "design_sha256": design_sha256,
                "variant_registry_sha256": variant_data["registry"].fingerprint_sha256,
                "variant_lineage_sha256": lineage["lineage_sha256"],
            }
        )
    audit = pd.DataFrame(audit_rows).set_index(["policy", "outcome"]).sort_index()
    if not bool(audit.index.is_unique):
        raise IngestError("Direct-arming effective-design audit keys are duplicated.")
    return MappingProxyType(designs), audit


# Fit the canonical arming baselines and retain their scientific objects.
def _fit_arming_baselines(
    baseline_config: Mapping[str, Any],
    designs: Mapping[str, Mapping[str, Any]],
    outcomes: Sequence[str],
) -> Tuple[pd.DataFrame, Mapping[str, BaselineResult]]:
    """Return the labelled Table 9 rows and exact fitted results by outcome."""
    rows: List[Dict[str, Any]] = []
    fitted: Dict[str, BaselineResult] = {}
    for outcome in outcomes:
        design = designs[str(outcome)]
        partition: Partition = design["partition"]
        result = run_baseline(
            POLICY_DIRECT_ARMING, str(outcome), baseline_config, design["data"]
        )
        if set(result.gaps.index.astype(str)) != set(partition.targets) or len(
            result.gaps
        ) != len(partition.targets):
            raise IngestError(
                f"{POLICY_DIRECT_ARMING}/{outcome}: fitted gap roster disagrees with the canonical design."
            )
        fitted[str(outcome)] = result
        rows.append(
            {
                "policy": POLICY_DIRECT_ARMING,
                "outcome": str(outcome),
                "target_count": len(partition.targets),
                "donor_count": len(partition.donors),
                "target_roster_json": json.dumps(list(partition.targets)),
                "donor_roster_json": json.dumps(list(partition.donors)),
                "removed_donors": "; ".join(
                    ADDITIONAL_DONOR_EXCLUSIONS.get(
                        (POLICY_DIRECT_ARMING, str(outcome)), ()
                    )
                ),
                "att_wave_1": float(result.att_wave_1),
                "att_wave_2": float(result.att_wave_2),
                "att_incremental": float(result.att_incremental),
                "wave_1_estimand": "equal_country_level_gap",
                "wave_2_estimand": "equal_country_level_gap",
                "incremental_estimand": "wave_2_level_minus_reference_level",
                "se_cd": float(result.se_cd),
                "stars": str(result.stars),
                "design_sha256": str(design["design_sha256"]),
                "variant_registry_sha256": str(design["variant_registry_sha256"]),
                "variant_lineage_sha256": str(design["variant_lineage_sha256"]),
                "gap_sha256": _content_digest(result.gaps),
                "effective_window_sha256": str(
                    result.effective_windows[WINDOW_PROVENANCE_KEY]["fingerprint_sha256"]
                ),
            }
        )
    baseline = pd.DataFrame(rows).set_index(["policy", "outcome"]).sort_index()
    if not bool(baseline.index.is_unique):
        raise IngestError("Direct-arming baseline keys are duplicated.")
    return baseline, MappingProxyType(fitted)


# Run the arming baseline pipeline across the outcomes.
def run_arming_baseline(
    baseline_config: Mapping[str, Any],
    data: Mapping[str, Any],
    outcomes: Sequence[str] = ARMING_OUTCOMES,
    outcome_removals: Optional[Mapping[str, Sequence[str]]] = None,
    effective_designs: Optional[Mapping[str, Mapping[str, Any]]] = None,
) -> pd.DataFrame:
    """
    Run the arming baseline pipeline across the outcomes.

    Parameters
    ----------
    baseline_config : mapping
        The baseline configuration.
    data : mapping
        The frozen inputs under the declared keys.
    outcomes : sequence of str, optional
        The outcomes the pipeline reports.
    outcome_removals : mapping of str to sequence of str, optional
        The donors each outcome's design drops, which the specification fixes
        applies to one outcome alone.
    effective_designs : mapping, optional
        Pre-resolved canonical designs supplied by the Task 30 orchestrator so
        point estimation, inference, and timing reuse identical partitions.

    Returns
    -------
    pandas.DataFrame
        One row per outcome carrying the two wave effects, the incremental
        effect and the inference, keyed on the policy and the outcome.

    Raises
    ------
    KeyError
        If the data mapping lacks the registry.
    IngestError
        If the arming policy carries no registered partition for an outcome, or
        if the interface refuses the composed design.
    """
    selected = tuple(str(value) for value in outcomes)
    designs = effective_designs
    if designs is None:
        designs, _audit = _resolve_direct_arming_designs(
            data, selected, outcome_removals
        )
    elif set(designs) != set(selected):
        raise IngestError("Pre-resolved direct-arming designs do not cover the requested outcomes.")
    baseline, _results = _fit_arming_baselines(baseline_config, designs, selected)
    # Log the table's extent.
    logger.info("arming baseline composed across %d outcome(s).", len(baseline))
    # Return the table.
    return baseline


# -------------------------------------------------------------------------------------------------------------------------------
# Task 30, Step 2: The arming resampling and the placebo interpretation
# -------------------------------------------------------------------------------------------------------------------------------


# Run the arming resampling and record the placebo interpretation.
def run_arming_inference(
    baseline_config: Mapping[str, Any],
    data: Mapping[str, Any],
    region_map: Mapping[str, str],
    outcomes: Sequence[str] = ARMING_OUTCOMES,
    bootstrap_replicates: int = ARMING_INFERENCE_REPLICATES,
    placebo_replicates: int = ARMING_INFERENCE_REPLICATES,
    seeds: Optional[Mapping[str, Any]] = None,
    effective_designs: Optional[Mapping[str, Mapping[str, Any]]] = None,
    baseline_results: Optional[Mapping[str, BaselineResult]] = None,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Run the arming resampling and record the placebo interpretation.

    Parameters
    ----------
    baseline_config : mapping
        The baseline configuration.
    data : mapping
        The frozen inputs under the declared keys.
    region_map : AuthoritativeRegionMap
        Versioned metadata-derived regions covering every effective-design
        country; the region branch resamples from this mapping.
    outcomes : sequence of str, optional
        The outcomes the resampling covers.
    bootstrap_replicates : int, optional
        The registered number of accepted bootstrap refits, which must equal
        500 in both bootstrap branches.
    placebo_replicates : int, optional
        The registered number of accepted size-matched pseudo-assignments,
        which must equal 500.
    seeds : mapping, optional
        The populated seed-set identity, master entropy and branch spawn keys,
        each of which must match the configuration registry before drawing.
    effective_designs : mapping, optional
        The canonical per-outcome designs already resolved by the Task 30
        orchestrator; every refit uses their exact target/donor rosters.
    baseline_results : mapping, optional
        The exact baseline results fitted on those designs, whose incremental
        ATT is the saved observed placebo reference.

    Returns
    -------
    tuple
        The inference table keyed on the policy, the outcome and the branch,
        and the interpretation frame recording the placebo probability's
        reading. The specification fixes that a non-significant probability is
        the correct diagnostic rather than a failure, so the interpretation is
        recorded as an artefact beside the number.

    Raises
    ------
    IngestError
        Propagated from the engine, with the pair named.
    """
    selected = tuple(str(outcome) for outcome in outcomes)
    if (
        isinstance(bootstrap_replicates, (bool, np.bool_))
        or isinstance(placebo_replicates, (bool, np.bool_))
        or not isinstance(bootstrap_replicates, (int, np.integer))
        or not isinstance(placebo_replicates, (int, np.integer))
        or int(bootstrap_replicates) != ARMING_INFERENCE_REPLICATES
        or int(placebo_replicates) != ARMING_INFERENCE_REPLICATES
    ):
        raise IngestError(
            "Direct-arming inference requires exactly 500 accepted draws in each "
            "bootstrap branch and 500 accepted size-matched placebo assignments."
        )
    if seeds is None or not isinstance(seeds, Mapping):
        raise IngestError("Direct-arming inference requires explicit documented seeds.")
    seed_audit = _resolve_seed_registration(baseline_config, seeds)
    if not isinstance(region_map, AuthoritativeRegionMap) or any(
        not str(region_map.provenance.get(field, "")).strip()
        for field in ("source_id", "version_id")
    ):
        raise IngestError("Direct-arming regions require authoritative source/version identity.")
    validate_region_mapping(region_map)
    designs = effective_designs
    if designs is None:
        designs, _design_audit = _resolve_direct_arming_designs(data, selected)
    elif set(designs) != set(selected):
        raise IngestError("Inference designs do not cover the direct-arming outcomes.")
    required_regions = {
        str(code)
        for outcome in selected
        for code in (
            tuple(designs[outcome]["partition"].targets)
            + tuple(designs[outcome]["partition"].donors)
        )
    }
    missing_regions = sorted(required_regions - set(region_map))
    if missing_regions:
        raise IngestError(
            "Direct-arming region metadata omits effective-design countr(ies) "
            f"{missing_regions[:INFERENCE_MAX_REPORTED_OFFENDERS]}."
        )
    region_provenance = dict(region_map.provenance)
    region_map_sha256 = hashlib.sha256(
        json.dumps(
            {
                "source_id": region_provenance["source_id"],
                "version_id": region_provenance["version_id"],
                "regions": sorted((str(code), str(region)) for code, region in region_map.items()),
            },
            sort_keys=True,
            separators=(",", ":"),
        ).encode("utf-8")
    ).hexdigest()
    if baseline_results is None:
        _baseline_table, resolved_results = _fit_arming_baselines(
            baseline_config, designs, selected
        )
        baseline_results = resolved_results
    elif set(baseline_results) != set(selected):
        raise IngestError("Inference baseline results do not cover the direct-arming outcomes.")
    pairs = [(POLICY_DIRECT_ARMING, outcome) for outcome in selected]
    # Run the resampling engine once per arming pair, which is its declared
    # per-pair interface.
    results: Dict[Tuple[str, str], ResamplingResult] = {}
    # Iterate over the pairs.
    for pair in pairs:
        # Run the engine for the pair.
        results[pair] = orchestrate_resampling_inference(
            pair[0],
            pair[1],
            baseline_config,
            designs[pair[1]]["data"],
            dict(seeds),
            region_map,
            bootstrap_replicates=int(bootstrap_replicates),
            placebo_replicates=int(placebo_replicates),
        )
    # Accumulate the inference rows.
    rows: List[dict] = []
    # Accumulate the interpretation rows.
    interpretation_rows: List[dict] = []
    # Iterate over the pairs in sorted order.
    for pair in sorted(results.keys()):
        # Read the pair's result.
        result: ResamplingResult = results[pair]
        observed = float(baseline_results[pair[1]].att_incremental)
        branch_bounds = {
            BRANCH_COUNTRY_HISTORY: (result.ci_ch_lower, result.ci_ch_upper),
            BRANCH_REGION_BLOCK: (result.ci_rb_lower, result.ci_rb_upper),
            BRANCH_PLACEBO: (float("nan"), float("nan")),
        }
        for branch in BRANCH_ORDER:
            draws = np.asarray(result.full_draw_arrays[branch], dtype=float)
            if draws.shape != (ARMING_INFERENCE_REPLICATES,) or not bool(
                np.isfinite(draws).all()
            ):
                raise IngestError(
                    f"{pair}/{branch}: accepted draw array is not the required finite 500."
                )
            attempt_audit = result.attempt_audits.get(branch)
            if not isinstance(attempt_audit, pd.DataFrame) or attempt_audit.empty:
                raise IngestError(f"{pair}/{branch}: complete attempt audit is absent.")
            required_attempt_columns = {
                "attempt_index",
                "status",
                "accepted_index",
                "reason_category",
                "refit_completed",
                "statistic",
                "registry_fingerprint",
                "draw_identity_sha256",
                "draw_content_sha256",
            }
            missing_attempt_columns = sorted(
                required_attempt_columns - set(attempt_audit.columns)
            )
            if missing_attempt_columns:
                raise IngestError(
                    f"{pair}/{branch}: attempt audit omits {missing_attempt_columns}."
                )
            ordered_attempts = attempt_audit.sort_values("attempt_index")
            if ordered_attempts["attempt_index"].tolist() != list(
                range(len(ordered_attempts))
            ):
                raise IngestError(
                    f"{pair}/{branch}: attempt indices are not deterministic and contiguous."
                )
            if not set(ordered_attempts["status"].astype(str)).issubset(
                {"accepted", "rejected"}
            ):
                raise IngestError(f"{pair}/{branch}: attempt status is undeclared.")
            accepted = int((attempt_audit["status"].astype(str) == "accepted").sum())
            rejected = int((attempt_audit["status"].astype(str) == "rejected").sum())
            attempted = len(attempt_audit)
            if accepted != ARMING_INFERENCE_REPLICATES or attempted != accepted + rejected:
                raise IngestError(f"{pair}/{branch}: attempt accounting is inconsistent.")
            accepted_rows = ordered_attempts[
                ordered_attempts["status"].astype(str).eq("accepted")
            ]
            if accepted_rows["accepted_index"].astype(int).tolist() != list(
                range(ARMING_INFERENCE_REPLICATES)
            ):
                raise IngestError(
                    f"{pair}/{branch}: accepted-draw indices are not contiguous."
                )
            if not np.array_equal(
                accepted_rows["statistic"].to_numpy(dtype=float), draws
            ):
                raise IngestError(
                    f"{pair}/{branch}: saved accepted statistics disagree with draws."
                )
            if not bool(accepted_rows["refit_completed"].astype(bool).all()):
                raise IngestError(
                    f"{pair}/{branch}: an accepted assignment lacks a complete refit."
                )
            for digest_column in (
                "registry_fingerprint",
                "draw_identity_sha256",
                "draw_content_sha256",
            ):
                if not bool(
                    accepted_rows[digest_column]
                    .astype(str)
                    .str.fullmatch(r"[0-9a-f]{64}")
                    .all()
                ):
                    raise IngestError(
                        f"{pair}/{branch}: accepted assignments lack valid "
                        f"{digest_column} provenance."
                    )
            rejected_rows = ordered_attempts[
                ordered_attempts["status"].astype(str).eq("rejected")
            ]
            if len(rejected_rows) and bool(
                rejected_rows["reason_category"].astype(str).str.strip().eq("").any()
            ):
                raise IngestError(
                    f"{pair}/{branch}: a rejected assignment lacks a reason category."
                )
            lower, upper = branch_bounds[branch]
            if branch != BRANCH_PLACEBO:
                recomputed = np.percentile(draws, [PERCENTILE_LOWER, PERCENTILE_UPPER])
                if not np.allclose(
                    np.array([lower, upper], dtype=float), recomputed, rtol=0.0, atol=1.0e-12
                ):
                    raise IngestError(f"{pair}/{branch}: percentile interval disagrees with saved draws.")
            audit_json = ordered_attempts.to_json(
                orient="records", double_precision=15
            )
            rows.append(
                {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "branch": branch,
                    "ci_lower": float(lower),
                    "ci_upper": float(upper),
                    "observed_incremental_att": observed,
                    "requested": ARMING_INFERENCE_REPLICATES,
                    "accepted": accepted,
                    "rejected": rejected,
                    "attempted": attempted,
                    "discarded": rejected,
                    "quantile_method": PERCENTILE_QUANTILE_METHOD,
                    "draws_json": json.dumps(draws.tolist(), separators=(",", ":")),
                    "draws_sha256": hashlib.sha256(draws.tobytes()).hexdigest(),
                    "attempt_audit_json": audit_json,
                    "attempt_audit_sha256": hashlib.sha256(audit_json.encode("utf-8")).hexdigest(),
                    "seed_set_id": str(seeds["seed_set_id"]),
                    "master_seed": int(seeds["master"]),
                    "region_source_id": str(region_provenance["source_id"]),
                    "region_version_id": str(region_provenance["version_id"]),
                    "region_map_sha256": region_map_sha256,
                    "spawn_key_json": str(
                        seed_audit.loc[(str(seeds["seed_set_id"]), branch), "spawn_key_json"]
                    ),
                    "design_sha256": str(designs[pair[1]]["design_sha256"]),
                    "variant_registry_sha256": str(designs[pair[1]]["variant_registry_sha256"]),
                }
            )
        # Compose the placebo probability.
        probability = float(result.p_placebo_hat)
        placebo_draws = np.asarray(result.full_draw_arrays[BRANCH_PLACEBO], dtype=float)
        exceedances = int(np.count_nonzero(np.abs(placebo_draws) >= abs(observed)))
        recomputed_probability = (exceedances + 1.0) / (
            ARMING_INFERENCE_REPLICATES + 1.0
        )
        if not np.isclose(probability, recomputed_probability, rtol=0.0, atol=1.0e-15):
            raise IngestError(f"{pair}: plus-one placebo probability disagrees with saved draws.")
        # Compose the significance flag.
        significant = bool(
            np.isfinite(probability) and probability < INTERPRETATION_LEVEL
        )
        # Append the interpretation row, which carries the reading the
        # specification fixes for a non-significant probability.
        interpretation_rows.append(
            {
                "policy": pair[0],
                "outcome": pair[1],
                "p_placebo_hat": probability,
                "pseudo_draws": int(result.full_draw_arrays[BRANCH_PLACEBO].size),
                "exceedances": exceedances,
                "plus_one_numerator": exceedances + 1,
                "plus_one_denominator": ARMING_INFERENCE_REPLICATES + 1,
                "probability_resolution": 1.0 / (ARMING_INFERENCE_REPLICATES + 1.0),
                "significant_5pct": bool(significant),
                "interpretation": (
                    INTERPRETATION_SIGNIFICANT if significant else INTERPRETATION_CONSISTENT
                ),
                "interpretation_detail": (
                    "observed statistic is detectably extreme under the registered "
                    "size-matched placebo design"
                    if significant
                    else "observed statistic is not detectably extreme under the "
                    "registered size-matched placebo design; this does not establish "
                    "random assignment"
                ),
                "diagnostic_only": True,
                "design_sha256": str(designs[pair[1]]["design_sha256"]),
            }
        )
    # Assemble the inference frame.
    inference: pd.DataFrame = pd.DataFrame(rows).set_index(
        ["policy", "outcome", "branch"]
    )
    # Assemble the interpretation frame.
    interpretation: pd.DataFrame = pd.DataFrame(interpretation_rows).set_index(
        ["policy", "outcome"]
    )
    # Log the resampling's extent.
    logger.info(
        "arming inference composed across %d pair(s) and %d row(s); placebo "
        "probabilities %s.",
        len(results),
        len(inference),
        interpretation["p_placebo_hat"].round(6).to_dict(),
    )
    # Return the two frames.
    return inference, interpretation

# -------------------------------------------------------------------------------------------------------------------------------
# Task 30, Step 3: The active-year timing check
# -------------------------------------------------------------------------------------------------------------------------------


# Compute one wave's active-window effect.
def compute_wave_active_effect(
    gaps: pd.DataFrame,
    masks: pd.DataFrame,
    reference: Optional[Tuple[int, int]],
) -> Tuple[float, float, str, int, List[Dict[str, Any]]]:
    """
    Compute one wave's active-window effect.

    Parameters
    ----------
    gaps : pandas.DataFrame
        The baseline gap matrix indexed on the country and columned by the
        years.
    masks : pandas.DataFrame
        The mask audit frame of the wave's active window.
    reference : tuple of int, optional
        The wave's pre-escalation reference window. When supplied the effect is
        the incremental form — each country's active-window mean less that
        window's pre-escalation level — and when absent it is the level form,
        which is the reading the first wave's timing figure carries: the wave
        begins before the gap matrix does, so no pre-wave level exists inside
        the matrix and the figure is the level gap over the active years.

    Returns
    -------
    tuple
        The effect, its standard error, its significance marker, the roster
        size and the exclusion rows.

    Raises
    ------
    IngestError
        If no country retains an active year, because the average would then be
        undefined, or if a country carries no pre-escalation level.
    """
    # Compose the wave's reference levels, which are absent for the level form.
    pre_levels: Optional[pd.Series] = (
        compute_pre_escalation_levels(gaps, tuple(reference))
        if reference is not None
        else None
    )
    # Compute the active-window effect through the estimation module's own
    # reduction, which excludes a country without an active year and records it.
    if pre_levels is not None:
        # Compose the incremental form through the declared reduction.
        effect, exclusions, roster_size = compute_active_year_effect(
            gaps, pre_levels, masks
        )
    else:
        # Compose the level form by reducing the active-window means directly,
        # reusing the same mask semantics and the same exclusion record.
        effect, exclusions, roster_size = _reduce_active_levels(gaps, masks)
    # Compose the per-country contrasts under the restricted window, which are
    # the active-window means less the reference level when one exists.
    contrasts: List[float] = []
    # Iterate over the countries the reduction retained.
    for country in gaps.index:
        # Select the country's mask rows.
        country_mask = masks[masks["country_code"].astype(str) == str(country)]
        # Compose the country's active years.
        active_years: List[int] = [
            int(row["year"]) for _, row in country_mask.iterrows() if bool(row["active"])
        ]
        # Skip a country without an active year, which the reduction excluded.
        if not active_years:
            # Continue to the next country.
            continue
        # Append the country's restricted contrast.
        contrasts.append(
            float(gaps.loc[country, active_years].mean())
            - (float(pre_levels.loc[country]) if pre_levels is not None else 0.0)
        )
    # Compose the contrast series for the inference.
    contrast_series: pd.Series = pd.Series(contrasts, dtype=float)
    # Compute the effect and its dispersion standard error.
    contrast_effect, se_cd = compute_country_dispersion_se(contrast_series)
    # Compute the test statistic and the degrees of freedom.
    statistic, degrees_of_freedom = compute_t_statistic(
        contrast_effect, se_cd, len(contrast_series)
    )
    # Compute the two-sided probability.
    probability: float = compute_two_sided_p(statistic, degrees_of_freedom)
    # Assign the significance marker.
    stars, _critical = assign_significance_stars(statistic, degrees_of_freedom)
    # Assert the marker's agreement with the probability route.
    assert_star_consistency(
        stars,
        probability,
        statistic=statistic,
        degrees_of_freedom=degrees_of_freedom,
    )
    # Compose the exclusion rows.
    exclusion_rows: List[Dict[str, Any]] = [
        {
            "country_code": str(row["country_code"]),
            "reason": str(row["reason"]),
        }
        for _, row in exclusions.iterrows()
    ]
    # Return the wave's quantities.
    return float(effect), float(se_cd), str(stars), int(roster_size), exclusion_rows


# Reduce the active-window means of a wave without a reference level.
def _reduce_active_levels(
    gaps: pd.DataFrame,
    masks: pd.DataFrame,
) -> Tuple[float, pd.DataFrame, int]:
    """
    Reduce the active-window means of a wave without a reference level.

    Parameters
    ----------
    gaps : pandas.DataFrame
        The gap matrix.
    masks : pandas.DataFrame
        The wave's mask audit frame.

    Returns
    -------
    tuple
        The level effect, the exclusion frame and the roster size. The
        semantics mirror the estimation module's reduction exactly: a country
        without an active year is excluded and recorded, and the effect is the
        equal-weighted mean of the retained countries' active-window means. The
        level form exists because the first wave begins before the gap matrix
        does, so no pre-wave level is available inside it.

    Raises
    ------
    IngestError
        If no country retains an active year.
    """
    # Accumulate the per-country means.
    country_means: List[float] = []
    # Accumulate the exclusions.
    exclusion_rows: List[Dict[str, Any]] = []
    # Iterate over the matrix's countries in its own order.
    for country in gaps.index:
        # Select the country's mask rows.
        country_mask = masks[masks["country_code"].astype(str) == str(country)]
        # Compose the country's active years.
        active_years: List[int] = [
            int(row["year"]) for _, row in country_mask.iterrows() if bool(row["active"])
        ]
        # Exclude a country without an active year and record the exclusion.
        if not active_years:
            # Append the exclusion row.
            exclusion_rows.append(
                {
                    "country_code": str(country),
                    "reason": "no active year inside the wave",
                }
            )
            # Continue to the next country.
            continue
        # Append the country's active-window mean.
        country_means.append(float(gaps.loc[country, active_years].mean()))
    # Refuse a wave where every country was excluded.
    if not country_means:
        # The failure states the condition.
        raise IngestError("The wave excludes every country, so the effect is undefined.")
    # Assemble the exclusion frame.
    exclusions: pd.DataFrame = pd.DataFrame(
        exclusion_rows, columns=["country_code", "reason"]
    )
    # Return the effect, the exclusions and the roster size.
    return float(np.mean(np.array(country_means, dtype=float))), exclusions, len(country_means)


# Run the arming active-year timing check.
def run_arming_active_year(
    baseline_config: Mapping[str, Any],
    data: Mapping[str, Any],
    entry_years: Mapping[str, int],
    outcomes: Sequence[str] = ARMING_OUTCOMES,
    wave_references: Optional[Mapping[str, Tuple[int, int]]] = None,
    effective_designs: Optional[Mapping[str, Mapping[str, Any]]] = None,
    baseline_results: Optional[Mapping[str, BaselineResult]] = None,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Run the arming active-year timing check.

    Parameters
    ----------
    baseline_config : mapping
        The baseline configuration.
    data : mapping
        The frozen inputs, which must carry the coding frame under the key
        ``coding``.
    entry_years : mapping of str to int
        The year each country's arming indicator turns on, which the evidence
        supplies from the coding frame.
    outcomes : sequence of str, optional
        The outcomes the check covers.
    wave_references : mapping of str to tuple of int, optional
        Each wave's pre-escalation reference window.
    effective_designs : mapping, optional
        The canonical per-outcome designs shared with point estimation and
        resampling.
    baseline_results : mapping, optional
        The exact fitted baselines whose gap matrices are reduced; no timing
        branch refits or substitutes a different partition.

    Returns
    -------
    tuple
        The timing table keyed on the policy, the outcome and the wave, and the
        exclusion frame naming every country excluded from a wave's average.

    Raises
    ------
    KeyError
        If the data mapping lacks the coding frame.
    IngestError
        If the wave references are malformed, if a country carries no entry
        year, or if the second wave's identity with the baseline fails under a
        complete roster.
    """
    # Refuse a data mapping lacking the coding frame.
    if "coding" not in data:
        # The failure names the absent key.
        raise KeyError("The data mapping lacks the coding frame.")
    # Compose the wave references, defaulting to the declared windows.
    references: Dict[str, Optional[Tuple[int, int]]] = {
        WAVE_ONE: (wave_references or {}).get(WAVE_ONE, WAVE_ONE_REFERENCE),
        WAVE_TWO: (wave_references or {}).get(WAVE_TWO, WAVE_TWO_REFERENCE),
    }
    # Refuse a malformed reference.
    for label, span in references.items():
        # Skip a wave without a reference, which is the level form.
        if span is None:
            # Continue to the next wave.
            continue
        # Validate both endpoints without allowing Boolean or fractional years
        # to be silently coerced into a different scientific window.
        try:
            endpoints = tuple(span)
        except TypeError as error:
            raise IngestError(
                f"The wave {label!r} carries a malformed reference window."
            ) from error
        if len(endpoints) != 2 or any(
            isinstance(value, (bool, np.bool_))
            or not isinstance(value, (int, np.integer))
            for value in endpoints
        ):
            raise IngestError(
                f"The wave {label!r} carries a malformed reference window."
            )
        if int(endpoints[0]) > int(endpoints[1]):
            raise IngestError(
                f"The wave {label!r} carries a reversed reference window."
            )
        references[label] = (int(endpoints[0]), int(endpoints[1]))
    selected = tuple(str(outcome) for outcome in outcomes)
    designs = effective_designs
    if designs is None:
        designs, _design_audit = _resolve_direct_arming_designs(data, selected)
    elif set(designs) != set(selected):
        raise IngestError("Active-year designs do not cover the direct-arming outcomes.")
    if baseline_results is None:
        _baseline_table, resolved_results = _fit_arming_baselines(
            baseline_config, designs, selected
        )
        baseline_results = resolved_results
    elif set(baseline_results) != set(selected):
        raise IngestError("Active-year baseline results do not cover the outcomes.")
    # Accumulate the timing rows.
    rows: List[dict] = []
    # Accumulate the exclusion rows.
    exclusion_rows: List[dict] = []
    # Iterate over the outcomes in their declared order.
    for outcome in selected:
        # Compose the pair key.
        pair: Tuple[str, str] = (POLICY_DIRECT_ARMING, str(outcome))
        result = baseline_results[str(outcome)]
        design = designs[str(outcome)]
        # Select the arming targets.
        targets: Tuple[str, ...] = tuple(
            str(code) for code in design["partition"].targets
        )
        # Verify the supplied direct-arming entry years against the canonical
        # direct-arming coding tensor, rather than sanctions timing.
        full_masks = build_active_year_masks(
            data["coding"], pair[0], targets, window=(2000, 2024)
        )
        for target in targets:
            if str(target) not in entry_years:
                raise IngestError(f"{pair}: target {target} carries no arming entry year.")
            raw_entry = entry_years[str(target)]
            if isinstance(raw_entry, (bool, np.bool_)) or not isinstance(
                raw_entry, (int, np.integer)
            ):
                raise IngestError(f"{pair}: target {target} has malformed arming timing.")
            active_years = full_masks[
                full_masks["country_code"].astype(str).eq(str(target))
                & full_masks["active"].astype(bool)
            ]["year"].astype(int)
            if active_years.empty or int(active_years.min()) != int(raw_entry):
                raise IngestError(
                    f"{pair}: supplied entry year {raw_entry!r} disagrees with "
                    f"direct-arming coding for {target}."
                )
        wave_windows = {
            WAVE_ONE: tuple(result.effective_windows[WAVE_ONE_LABEL]),
            WAVE_TWO: tuple(result.effective_windows[WAVE_TWO_LABEL]),
        }
        # Iterate over the waves in their declared order.
        for wave in WAVE_ORDER:
            # Build the wave's masks from the coding frame.
            masks: pd.DataFrame = build_active_year_masks(
                data["coding"], pair[0], targets, window=wave_windows[wave]
            )
            # Appendix B6 is always the active-year level gap.  Compute the
            # preferred-reference incremental contrast separately when a
            # reference exists; never relabel it as the B6 quantity.
            level_effect, level_se, level_stars, roster_size, exclusions = (
                compute_wave_active_effect(result.gaps, masks, None)
            )
            if references[wave] is not None:
                incremental_effect, incremental_se, incremental_stars, incremental_roster, incremental_exclusions = compute_wave_active_effect(
                    result.gaps, masks, references[wave]
                )
                if incremental_roster != roster_size or incremental_exclusions != exclusions:
                    raise IngestError(f"{pair}/{wave}: level/incremental active rosters diverge.")
            else:
                incremental_effect = float("nan")
                incremental_se = float("nan")
                incremental_stars = ""
            # Append the timing row.
            rows.append(
                {
                    "policy": pair[0],
                    "outcome": pair[1],
                    "wave": str(wave),
                    "reference_window": (str(list(references[wave])) if references[wave] is not None else "level_gap"),
                    "active_roster": int(roster_size),
                    "target_count": int(len(targets)),
                    "active_year_att": float(level_effect),
                    "active_year_estimand": "equal_country_active_year_level_gap",
                    "active_year_level_att": float(level_effect),
                    "active_year_level_se_cd": float(level_se),
                    "active_year_level_stars": str(level_stars),
                    "active_year_incremental_att": float(incremental_effect),
                    "active_year_incremental_se_cd": float(incremental_se),
                    "active_year_incremental_stars": str(incremental_stars),
                    "se_cd": float(level_se),
                    "stars": str(level_stars),
                    "complete_wave_activity": bool(masks["active"].astype(bool).all()),
                    "design_sha256": str(design["design_sha256"]),
                    "gap_sha256": _content_digest(result.gaps),
                }
            )
            # Append the wave's exclusions.
            for exclusion in exclusions:
                # Record the excluded country with its pair and wave.
                exclusion_rows.append(
                    {
                        "policy": pair[0],
                        "outcome": pair[1],
                        "wave": str(wave),
                        "country_code": exclusion["country_code"],
                        "reason": exclusion["reason"],
                    }
                )
            # Only a full target-by-year active mask proves equality with the
            # corresponding complete-wave LEVEL ATT.
            if bool(masks["active"].astype(bool).all()):
                baseline_level = (
                    float(result.att_wave_1) if wave == WAVE_ONE else float(result.att_wave_2)
                )
                if abs(float(level_effect) - baseline_level) > 1.0e-9:
                    raise IngestError(
                        f"{pair[0]}/{pair[1]}/{wave}: complete-wave active level "
                        f"{float(level_effect)!r} disagrees with baseline level {baseline_level!r}."
                    )
                if wave == WAVE_TWO and references[wave] is not None and tuple(
                    references[wave]
                ) == tuple(result.effective_windows["pre_escalation"]):
                    if abs(float(incremental_effect) - float(result.att_incremental)) > 1.0e-9:
                        raise IngestError(
                            f"{pair}/{wave}: complete-wave incremental contrast disagrees with baseline incremental ATT."
                        )
    # Assemble the timing frame.
    timing: pd.DataFrame = pd.DataFrame(rows).set_index(
        ["policy", "outcome", "wave"]
    )
    # Assemble the exclusion frame.
    exclusions_frame: pd.DataFrame = pd.DataFrame(
        exclusion_rows, columns=["policy", "outcome", "wave", "country_code", "reason"]
    )
    # Set the exclusion frame's index when it carries rows.
    if len(exclusions_frame):
        # Key the frame on the pair, the wave and the country.
        exclusions_frame = exclusions_frame.set_index(
            ["policy", "outcome", "wave", "country_code"]
        )
    # Log the timing check's extent.
    logger.info(
        "arming active-year check composed across %d row(s) with %d exclusion(s).",
        len(timing),
        len(exclusions_frame),
    )
    # Return the two frames.
    return timing, exclusions_frame


# Freeze the pipeline behind a canonical digest.
def freeze_direct_arming(
    baseline: pd.DataFrame,
    inference: pd.DataFrame,
    active_year: pd.DataFrame,
    interpretation: Optional[pd.DataFrame] = None,
    exclusions: Optional[pd.DataFrame] = None,
    design_audit: Optional[pd.DataFrame] = None,
) -> str:
    """
    Freeze the pipeline behind a canonical digest.

    Parameters
    ----------
    baseline : pandas.DataFrame
        The baseline table.
    inference : pandas.DataFrame
        The inference table.
    active_year : pandas.DataFrame
        The timing table.
    interpretation : pandas.DataFrame, optional
        The placebo diagnostic, including explicit non-causal interpretation.
    exclusions : pandas.DataFrame, optional
        The country-wave exclusions from the active-year reduction.
    design_audit : pandas.DataFrame, optional
        The canonical roster and lineage evidence reused by every branch.

    Returns
    -------
    str
        The hexadecimal digest of the canonical payload over the sorted keys.

    Raises
    ------
    IngestError
        If any frame is empty.
    """
    # Refuse an empty frame.
    for label, frame in (
        ("baseline", baseline),
        ("inference", inference),
        ("active-year", active_year),
    ):
        # Test the frame's extent.
        if len(frame) == 0:
            # The failure names the frame.
            raise IngestError(f"Cannot freeze an empty {label} frame.")
    payload = {
        "baseline": _content_digest(baseline.sort_index()),
        "inference": _content_digest(inference.sort_index()),
        "active_year": _content_digest(active_year.sort_index()),
    }
    for label, frame in (
        ("interpretation", interpretation),
        ("exclusions", exclusions),
        ("design_audit", design_audit),
    ):
        if frame is not None:
            if label != "exclusions" and len(frame) == 0:
                raise IngestError(f"Cannot freeze an empty {label} frame.")
            if len(frame) and not bool(frame.index.is_unique):
                raise IngestError(f"Cannot freeze duplicate-keyed {label} frame.")
            payload[label] = _content_digest(frame.sort_index())
    return hashlib.sha256(
        json.dumps(payload, sort_keys=True, separators=(",", ":")).encode("utf-8")
    ).hexdigest()


# -------------------------------------------------------------------------------------------------------------------------------
# Task 30, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Execute the direct-arming coalition pipeline.
def execute_direct_arming_pipeline(
    baseline_config: Mapping[str, Any],
    data: Mapping[str, Any],
    entry_years: Mapping[str, int],
    region_map: Mapping[str, str],
    outcomes: Sequence[str] = ARMING_OUTCOMES,
    outcome_removals: Optional[Mapping[str, Sequence[str]]] = None,
    bootstrap_replicates: int = ARMING_INFERENCE_REPLICATES,
    placebo_replicates: int = ARMING_INFERENCE_REPLICATES,
    seeds: Optional[Mapping[str, Any]] = None,
    wave_references: Optional[Mapping[str, Tuple[int, int]]] = None,
    expected_baseline: Optional[
        Mapping[Tuple[str, str], Mapping[str, float]]
    ] = None,
    tolerance: float = 1.0e-3,
) -> DirectArmingResult:
    """
    Execute the direct-arming coalition pipeline.

    Parameters
    ----------
    baseline_config : mapping
        The baseline configuration.
    data : mapping
        The frozen inputs under the declared keys.
    entry_years : mapping of str to int
        The year each country's arming indicator turns on.
    region_map : AuthoritativeRegionMap
        Versioned metadata-derived regions covering every design country.
    outcomes : sequence of str, optional
        The outcomes the pipeline covers.
    outcome_removals : mapping of str to sequence of str, optional
        The donors each outcome's design drops.
    bootstrap_replicates : int, optional
        The number of accepted bootstrap refits, fixed at 500.
    placebo_replicates : int, optional
        The number of accepted size-matched pseudo-assignments, fixed at 500.
    seeds : mapping, optional
        The documented seed-set identity, master entropy and branch spawn
        keys. No implicit seed is permitted.
    wave_references : mapping of str to tuple of int, optional
        Each wave's pre-escalation reference window.
    expected_baseline : mapping of tuple to mapping of str to float, optional
        The declared estimates keyed on the pair and the column, compared only
        when supplied.
    tolerance : float, optional
        The tolerance of the declared comparison.

    Returns
    -------
    DirectArmingResult
        The baseline table, the inference table, the interpretation, the timing
        table, the exclusions, the canonical design audit, the digest, the
        timestamp and the notes.

    Raises
    ------
    KeyError
        If the data mapping lacks a declared key.
    IngestError
        If any step, coverage or comparison check fails.
    """
    # Refuse a data mapping lacking a declared key.
    for key in ("registry", "panel", "lagged", "coding"):
        # Test the key's presence.
        if key not in data:
            # The failure names the absent key.
            raise KeyError(f"The data mapping lacks the declared key {key!r}.")
    if isinstance(tolerance, (bool, np.bool_)):
        raise IngestError("The direct-arming comparison tolerance cannot be Boolean.")
    try:
        comparison_tolerance = float(tolerance)
    except (TypeError, ValueError) as error:
        raise IngestError("The direct-arming comparison tolerance is malformed.") from error
    if not np.isfinite(comparison_tolerance) or comparison_tolerance < 0.0:
        raise IngestError(
            "The direct-arming comparison tolerance must be finite and nonnegative."
        )
    selected = tuple(str(outcome) for outcome in outcomes)
    if (
        isinstance(bootstrap_replicates, (bool, np.bool_))
        or isinstance(placebo_replicates, (bool, np.bool_))
        or not isinstance(bootstrap_replicates, (int, np.integer))
        or not isinstance(placebo_replicates, (int, np.integer))
        or int(bootstrap_replicates) != ARMING_INFERENCE_REPLICATES
        or int(placebo_replicates) != ARMING_INFERENCE_REPLICATES
    ):
        raise IngestError("Direct-arming execution requires exactly 500 accepted draws per inference branch.")
    if seeds is None or not isinstance(seeds, Mapping):
        raise IngestError("Direct-arming execution requires explicit documented seeds.")
    _resolve_seed_registration(baseline_config, seeds)
    if not isinstance(region_map, AuthoritativeRegionMap) or any(
        not str(region_map.provenance.get(field, "")).strip()
        for field in ("source_id", "version_id")
    ):
        raise IngestError("Direct-arming execution requires versioned authoritative regions.")
    validate_region_mapping(region_map)
    # Resolve the canonical outcome partitions once, before any branch runs.
    designs, design_audit = _resolve_direct_arming_designs(
        data, selected, outcome_removals
    )
    # Fit each baseline once and retain its gap matrix for the B6 reduction.
    baseline, baseline_results = _fit_arming_baselines(
        baseline_config, designs, selected
    )
    # Run the resampling and compose the interpretation.
    inference, interpretation = run_arming_inference(
        baseline_config,
        data,
        region_map,
        selected,
        bootstrap_replicates=bootstrap_replicates,
        placebo_replicates=placebo_replicates,
        seeds=seeds,
        effective_designs=designs,
        baseline_results=baseline_results,
    )
    # Run the active-year timing check.
    active_year, exclusions = run_arming_active_year(
        baseline_config,
        data,
        entry_years,
        selected,
        wave_references,
        effective_designs=designs,
        baseline_results=baseline_results,
    )
    expected_baseline_index = pd.MultiIndex.from_tuples(
        [(POLICY_DIRECT_ARMING, outcome) for outcome in selected],
        names=["policy", "outcome"],
    )
    expected_inference_index = pd.MultiIndex.from_tuples(
        [
            (POLICY_DIRECT_ARMING, outcome, branch)
            for outcome in selected
            for branch in BRANCH_ORDER
        ],
        names=["policy", "outcome", "branch"],
    )
    expected_active_index = pd.MultiIndex.from_tuples(
        [
            (POLICY_DIRECT_ARMING, outcome, wave)
            for outcome in selected
            for wave in WAVE_ORDER
        ],
        names=["policy", "outcome", "wave"],
    )
    for label, observed_index, expected_index in (
        ("baseline", baseline.index, expected_baseline_index),
        ("interpretation", interpretation.index, expected_baseline_index),
        ("effective-design", design_audit.index, expected_baseline_index),
        ("inference", inference.index, expected_inference_index),
        ("active-year", active_year.index, expected_active_index),
    ):
        if not bool(observed_index.is_unique) or set(observed_index.tolist()) != set(
            expected_index.tolist()
        ):
            raise IngestError(
                f"The direct-arming {label} table does not have exact declared coverage."
            )
    # Reconcile every consumer to the design object resolved before any branch
    # ran.  This prevents point estimation, resampling, and B6 timing from
    # carrying same-sized but scientifically different rosters.
    for outcome in selected:
        pair = (POLICY_DIRECT_ARMING, outcome)
        design_sha256 = str(design_audit.loc[pair, "design_sha256"])
        if not re.fullmatch(r"[0-9a-f]{64}", design_sha256):
            raise IngestError(f"{pair}: effective-design digest is malformed.")
        if int(baseline.loc[pair, "target_count"]) != ARMING_TARGET_COUNT:
            raise IngestError(f"{pair}: baseline target count is not the declared 29.")
        for label, values in (
            ("baseline", [baseline.loc[pair, "design_sha256"]]),
            ("interpretation", [interpretation.loc[pair, "design_sha256"]]),
            (
                "inference",
                inference.loc[pair, "design_sha256"].astype(str).tolist(),
            ),
            (
                "active-year",
                active_year.loc[pair, "design_sha256"].astype(str).tolist(),
            ),
        ):
            if any(str(value) != design_sha256 for value in values):
                raise IngestError(
                    f"{pair}: {label} does not reuse the canonical effective design."
                )
        if not bool(
            inference.loc[pair, "accepted"]
            .astype(int)
            .eq(ARMING_INFERENCE_REPLICATES)
            .all()
        ):
            raise IngestError(f"{pair}: inference does not retain 500 accepted draws per branch.")
        saved_probability = float(interpretation.loc[pair, "p_placebo_hat"])
        if not np.isclose(
            saved_probability,
            float(
                (
                    np.count_nonzero(
                        np.abs(
                            np.asarray(
                                json.loads(
                                    inference.loc[
                                        (*pair, BRANCH_PLACEBO), "draws_json"
                                    ]
                                ),
                                dtype=float,
                            )
                        )
                        >= abs(float(baseline.loc[pair, "att_incremental"]))
                    )
                    + 1.0
                )
                / (ARMING_INFERENCE_REPLICATES + 1.0)
            ),
            rtol=0.0,
            atol=1.0e-15,
        ):
            raise IngestError(f"{pair}: interpretation probability is not the saved plus-one tail.")
    # Compare the baseline with the declared values when supplied.
    if expected_baseline is not None:
        # Iterate over the declared entries.
        for pair, declared in expected_baseline.items():
            # Refuse a pair the table does not report.
            if pair not in baseline.index:
                # The failure names the absent pair.
                raise IngestError(f"The baseline table omits the declared pair {pair}.")
            # Compare each declared column.
            for column, value in declared.items():
                # Refuse a column the table does not carry.
                if column not in baseline.columns:
                    # The failure names the absent column.
                    raise IngestError(
                        f"The baseline table omits the declared column {column!r}."
                    )
                # Read the observed value.
                observed = float(baseline.loc[pair, column])
                try:
                    declared_value = float(value)
                except (TypeError, ValueError) as error:
                    raise IngestError(
                        f"{pair}/{column}: declared baseline value is malformed."
                    ) from error
                if not np.isfinite(observed) or not np.isfinite(declared_value):
                    raise IngestError(
                        f"{pair}/{column}: baseline comparison values must be finite."
                    )
                # Refuse a divergence beyond the tolerance.
                if abs(observed - declared_value) > comparison_tolerance:
                    # The failure names the pair, the column and both values.
                    raise IngestError(
                        f"{pair}/{column}: {observed!r} against the declared "
                        f"{declared_value!r}."
                    )
    # Assert the coverage of the requested outcomes in every table.
    for label, frame in (
        ("baseline", baseline),
        ("interpretation", interpretation),
    ):
        # Compose the covered outcomes.
        covered = set(str(value) for value in frame.index.get_level_values("outcome"))
        # Refuse a coverage divergence.
        if covered != set(selected):
            # The failure names the frame and the divergence.
            missing = sorted(set(selected) - covered)
            # Raise with the missing outcomes attached.
            raise IngestError(f"The {label} table omits the outcome(s) {missing}.")
    # Freeze the pipeline.
    fingerprint: str = freeze_direct_arming(
        baseline,
        inference,
        active_year,
        interpretation=interpretation,
        exclusions=exclusions,
        design_audit=design_audit,
    )
    # Accumulate the recorded notes.
    notes: List[str] = [
        "the arming pipeline runs the same frozen interface as the headline "
        "analysis, so the only differences are the partition, the treatment "
        "indicator and the outcome-specific donor removals.",
        "an outcome-specific donor removal is confined to the design that "
        "declares it and cannot leak across outcomes.",
        "the placebo probability is diagnostic only: a probability above the "
        "declared level means the observed statistic is not detectably extreme "
        "under the registered size-matched placebo design; it is not evidence "
        "that treatment was randomly assigned.",
        "Appendix B6 active-year values are equal-country level gaps. The "
        "preferred-reference incremental contrasts are retained in separate "
        "columns and never substituted for the B6 estimand.",
        "a complete target-by-year activity mask is reconciled to the "
        "baseline wave-level ATT; only the separately named incremental "
        "contrast is reconciled to the preferred incremental ATT.",
    ]
    # Record the roster sizes.
    notes.append(
        "active rosters: "
        + str(
            {
                f"{index[0]}/{index[1]}/{index[2]}": int(row["active_roster"])
                for index, row in active_year.iterrows()
            }
        )
        + "."
    )
    # Assemble the immutable result.
    result: DirectArmingResult = DirectArmingResult(
        baseline=baseline,
        inference=inference,
        interpretation=interpretation,
        active_year=active_year,
        exclusions=exclusions,
        design_audit=design_audit,
        fingerprint_sha256=fingerprint,
        built_utc=current_utc_iso(),
        notes=tuple(notes),
    )
    # Log the completed pipeline.
    logger.info(
        "direct-arming pipeline complete: %d baseline row(s), %d inference "
        "row(s), %d timing row(s), fingerprint %s.",
        len(baseline),
        len(inference),
        len(active_year),
        fingerprint[:16],
    )
    # Return the assembled result.
    return result
````

````python
# Task 31: Figure and Table Production

# ==============================================================================
# Task 31: Figure and Table Production
# ==============================================================================

# Import the plotting library behind a guard, because the data preparation must
# run wherever the pipeline runs while the rendering needs only a plotting
# stack; its absence is recorded as a reproducibility gap rather than ignored.
try:

    # Select a non-interactive backend so the rendering needs no display.
    matplotlib.use("Agg")

    # Record that the plotting library is available.
    _PLOTTING_AVAILABLE: bool = True
# Record the library's absence.
except Exception:  # pragma: no cover - the guard records the state
    # Record the absence.
    _PLOTTING_AVAILABLE = False
    # Bind the name so the type checkers see a definition.
    plt = None  # type: ignore[assignment]

# Probe the geographic library without importing it, since the basemap source
# is unspecified and the data preparation must not depend on it; the probe's
# outcome is recorded in the gaps register rather than assumed.
_GEO_AVAILABLE: bool = importlib.util.find_spec("geopandas") is not None


# Declare the reference year the dynamic path is normalized on.
NORMALIZATION_YEAR: Final[int] = 2013
# Declare the critical value of the pointwise ribbon.
RIBBON_CRITICAL_VALUE: Final[float] = 1.96
# Declare the first treatment year, which the path's labelling and the vertical
# marker both consume.
FIRST_TREATMENT_YEAR: Final[int] = 2014
# Declare the second wave's first year, which the second marker consumes.
SECOND_WAVE_YEAR: Final[int] = 2022
# Declare the label of a fitted year, which precedes the first treatment year.
REGION_FITTED: Final[str] = "fitted_pre_treatment"
# Declare the label of a realised year, which the treatment or its aftermath
# covers.
REGION_REALIZED: Final[str] = "realized"
# Declare the colour scale of the choropleth, whose endpoints the specification
# fixes must not cap the values near them.
CHOROPLETH_SCALE: Final[Tuple[float, float]] = (-0.18, 0.18)
# Declare the countries the choropleth greys because they carry no estimate.
GREY_COUNTRIES: Final[Tuple[str, ...]] = ("UKR",)
# Declare the label of a greyed country.
GREY_LABEL: Final[str] = "grey_no_estimate"
# Declare the reference year of the overlay's normalization, which matches the
# dynamic path's.
OVERLAY_NORMALIZATION_YEAR: Final[int] = NORMALIZATION_YEAR
# Declare the reproducibility gap of the choropleth's basemap.
BASEMAP_GAP: Final[str] = "choropleth_basemap_source_unspecified"
# Declare the two independently fitted comparison designs required by Figure 4.
OVERLAY_MATCHED_DESIGNS: Final[Tuple[str, str]] = (
    "high_income",
    "mahalanobis_m5",
)
# Declare the publication tables independently of whatever sources a caller
# happens to provide.
PRODUCTION_TABLES: Final[Tuple[str, ...]] = tuple(
    [f"table_{index}" for index in range(3, 11)]
    + [f"table_A{index}" for index in range(1, 9)]
    + [f"table_B{index}" for index in range(1, 8)]
)
# Declare the cell-level inventory schema and result schema used by every
# publication table.  The inventory is passed separately from computed cells.
TABLE_INVENTORY_COLUMNS: Final[Tuple[str, ...]] = (
    "table_id",
    "row_identifier",
    "column_identifier",
    "unit",
    "display_precision",
    "procedure",
)
TABLE_CELL_COLUMNS: Final[Tuple[str, ...]] = TABLE_INVENTORY_COLUMNS + (
    "value_unrounded",
    "status",
    "source_id",
    "gap_id",
)
TABLE_STATUS_COMPUTED: Final[str] = "computed"
TABLE_STATUS_UNRESOLVED: Final[str] = "unresolved_gap"
# Pin trusted manuscript cell-binding authority documents independently of
# caller data.  The reconstructed repository contains no genuine authoritative
# registry, so production fails closed until a reviewed authority/version
# digest is registered here.  An isolated test harness may temporarily replace
# this module registry; the production adapter never reads pins from data.
TRUSTED_TABLE_BINDING_AUTHORITY_PINS: Mapping[Tuple[str, str], str] = MappingProxyType({})
# Independent structural lower bounds from the manuscript's declared panels,
# outcomes, country rosters and specification grids. They prevent a
# self-certified one-cell-per-table inventory from masquerading as complete;
# the versioned inventory may (and normally will) enumerate more cells.
TABLE_MINIMUM_REQUIRED_CELLS: Final[Mapping[str, int]] = MappingProxyType(
    {
        "table_3": 6, "table_4": 8, "table_5": 4, "table_6": 4,
        "table_7": 4, "table_8": 4, "table_9": 6, "table_10": 16,
        "table_A1": 12, "table_A2": 4, "table_A3": 12,
        "table_A4": 129, "table_A5": 4, "table_A6": 129,
        "table_A7": 4, "table_A8": 6, "table_B1": 8,
        "table_B2": 8, "table_B3": 8, "table_B4": 4,
        "table_B5": 10, "table_B6": 4, "table_B7": 24,
    }
)


# Declare the frozen bundle of the produced figures and tables.
@dataclass(frozen=True)
class FigureTableBundle:
    """
    Frozen bundle of the produced figures and tables.

    Attributes
    ----------
    figure_data : dict of str to pandas.DataFrame
        The data of every figure, keyed on the figure's name, so the rendering
        is a pure function of the bundle.
    tables : dict of str to pandas.DataFrame
        The composed tables, keyed on their role and number.
    latex : dict of str to str
        The tabular rendering of every table, recorded as text so the artefact
        is reproducible without a rendering stack.
    rendered : pandas.DataFrame
        One row per figure recording whether it was written and where, or why
        it was not.
    reproducibility_gaps : pandas.DataFrame
        One row per unresolved or explicitly resolved production dependency,
        including the versioned geometry and plotting-stack state.
    fingerprint_sha256 : str
        Digest of the canonical production payload.
    built_utc : str
        ISO-8601 UTC timestamp of the production.
    notes : tuple of str
        Recorded decisions, including the two interpretive distinctions of the
        dynamic path.
    """
    # The data of every figure.
    figure_data: Mapping[str, pd.DataFrame]
    # The composed tables.
    tables: Mapping[str, pd.DataFrame]
    # The tabular renderings.
    latex: Mapping[str, str]
    # The rendering outcomes.
    rendered: pd.DataFrame
    # The unspecified inputs.
    reproducibility_gaps: pd.DataFrame
    # Digest of the canonical production payload.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the production.
    built_utc: str
    # Recorded decisions of the production.
    notes: Tuple[str, ...]


# -------------------------------------------------------------------------------------------------------------------------------
# Task 31, Step 1: The dynamic path
# -------------------------------------------------------------------------------------------------------------------------------


# Compute the coalition's dynamic path from a gap matrix.
def compute_coalition_path(
    gaps: pd.DataFrame,
    outcome: str,
    normalization_year: int = NORMALIZATION_YEAR,
) -> pd.DataFrame:
    """
    Compute the coalition's dynamic path from a gap matrix.

    Parameters
    ----------
    gaps : pandas.DataFrame
        The gap matrix indexed on the country and columned by the year.
    outcome : str
        The outcome the path describes.
    normalization_year : int, optional
        The reference year the path is normalized on.

    Returns
    -------
    pandas.DataFrame
        One row per year carrying the coalition mean, the cross-country
        dispersion, the pointwise ribbon bounds, the normalized mean and the
        region label that marks the fitted years.

    Raises
    ------
    IngestError
        If the matrix carries fewer than two countries, because the dispersion
        would be undefined, or if the reference year is absent, because the
        normalization would then silently use another year.
    """
    # Refuse a malformed reference year rather than truncating it.
    if isinstance(normalization_year, (bool, np.bool_)) or not isinstance(
        normalization_year, (int, np.integer)
    ):
        raise IngestError("The dynamic-path normalization year must be an exact integer.")
    # Refuse a matrix with too few unique countries.
    if gaps.shape[0] < 2 or not bool(gaps.index.is_unique):
        # The failure names the observed count.
        raise IngestError(
            f"The dynamic path requires at least two countries; {gaps.shape[0]} "
            f"supplied."
        )
    # Validate and canonicalize the full labelled year axis before reduction.
    if not bool(gaps.columns.is_unique):
        raise IngestError("The dynamic gap matrix duplicates a calendar year.")
    numeric_years = pd.to_numeric(pd.Index(gaps.columns), errors="coerce")
    if (
        bool(pd.isna(numeric_years).any())
        or not bool(np.isfinite(np.asarray(numeric_years, dtype=float)).all())
        or any(not float(value).is_integer() for value in numeric_years)
    ):
        raise IngestError("The dynamic gap matrix carries malformed calendar years.")
    canonical = gaps.copy(deep=True)
    canonical.columns = pd.Index(
        [int(value) for value in numeric_years], name="year"
    )
    canonical = canonical.sort_index(axis=1)
    numeric_gaps = canonical.apply(pd.to_numeric, errors="coerce")
    if bool(numeric_gaps.isna().to_numpy().any()) or not bool(
        np.isfinite(numeric_gaps.to_numpy(dtype=float)).all()
    ):
        raise IngestError(
            "The dynamic path requires a complete finite country-by-year gap matrix."
        )
    # Refuse a matrix without the fitted reference year.
    if int(normalization_year) not in set(numeric_gaps.columns):
        # The failure names the reference year.
        raise IngestError(
            f"The gap matrix does not cover the normalization year "
            f"{normalization_year}."
        )
    # Compose the coalition mean per year.
    mean = numeric_gaps.mean(axis=0, skipna=False)
    # Compose the cross-country dispersion per year.
    dispersion = numeric_gaps.std(axis=0, skipna=False, ddof=1)
    # Compose the standard error of the mean per year.
    standard_error = dispersion / float(np.sqrt(float(numeric_gaps.shape[0])))
    # Compose the reference mean.
    reference: float = float(mean.loc[int(normalization_year)])
    # Assemble the path frame.
    path: pd.DataFrame = pd.DataFrame(
        {
            "year": [int(year) for year in gaps.columns],
            "outcome": [str(outcome)] * len(gaps.columns),
            "coalition_mean": mean.to_numpy(dtype=float),
            "cross_country_sd": dispersion.to_numpy(dtype=float),
            "standard_error": standard_error.to_numpy(dtype=float),
        }
    ).set_index("year")
    # Compose the pointwise ribbon bounds.
    path["ribbon_lower"] = path["coalition_mean"] - RIBBON_CRITICAL_VALUE * path["standard_error"]
    # Compose the upper bound.
    path["ribbon_upper"] = path["coalition_mean"] + RIBBON_CRITICAL_VALUE * path["standard_error"]
    # Compose the normalized mean, which is zero at the reference year.
    path["normalized_mean"] = path["coalition_mean"] - reference
    # Compose the normalized bounds.
    path["normalized_lower"] = path["ribbon_lower"] - reference
    # Compose the normalized upper bound.
    path["normalized_upper"] = path["ribbon_upper"] - reference
    # Compose the region label, which separates the fitted years from the
    # realised ones so the figure cannot be read as an event study.
    path["region"] = np.where(
        path.index < FIRST_TREATMENT_YEAR, REGION_FITTED, REGION_REALIZED
    )
    # Compose the shading flags the renderer consumes.
    path["in_wave_1"] = (path.index >= FIRST_TREATMENT_YEAR) & (path.index <= 2021)
    # Compose the second wave's flag.
    path["in_wave_2"] = path.index >= SECOND_WAVE_YEAR
    # Persist the actual pointwise sample size and immutable rendering
    # semantics so the figure and its fingerprint cannot drift apart.
    path["country_count"] = int(numeric_gaps.shape[0])
    path.attrs["normalization_year"] = int(normalization_year)
    path.attrs["fit_region_end"] = int(normalization_year)
    path.attrs["first_treatment_year"] = FIRST_TREATMENT_YEAR
    path.attrs["second_wave_year"] = SECOND_WAVE_YEAR
    path.attrs["ribbon_critical_value"] = RIBBON_CRITICAL_VALUE
    path.attrs["ribbon_convention"] = "country_dispersion_ddof1_div_sqrt_N_pointwise"
    path.attrs["render_spec"] = {
        "kind": "dynamic_path",
        "x": "year",
        "y": "normalized_mean",
        "band": ["normalized_lower", "normalized_upper"],
        "vertical_lines": [FIRST_TREATMENT_YEAR, SECOND_WAVE_YEAR],
        "wave_1": [FIRST_TREATMENT_YEAR, SECOND_WAVE_YEAR - 1],
        "wave_2_start": SECOND_WAVE_YEAR,
        "normalization_year": int(normalization_year),
    }
    # Assert the normalization.
    if abs(float(path.loc[int(normalization_year), "normalized_mean"])) > 1.0e-12:
        # The failure states the condition.
        raise IngestError(
            "The normalized path is not zero at the reference year, so the "
            "normalization is inconsistent."
        )
    # Log the path's extent.
    logger.info(
        "%s: dynamic path computed across %d year(s), normalized on %d.",
        outcome,
        len(path),
        normalization_year,
    )
    # Return the path.
    return path


# -------------------------------------------------------------------------------------------------------------------------------
# Task 31, Step 2: The choropleth's data and its declared gap
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the choropleth's data from the incremental contrasts.
def build_choropleth_data(
    contrasts: Mapping[str, float],
    sample_countries: Sequence[str],
    scale: Tuple[float, float] = CHOROPLETH_SCALE,
    grey_countries: Sequence[str] = GREY_COUNTRIES,
    country_geometry: Optional[pd.DataFrame] = None,
    geographic_scope: str = "Europe",
) -> pd.DataFrame:
    """
    Compose the choropleth's data from the incremental contrasts.

    Parameters
    ----------
    contrasts : mapping of str to float
        The incremental effect of every sample country.
    sample_countries : sequence of str
        The countries the design covers.
    scale : tuple of float, optional
        The declared colour scale, whose endpoints the specification fixes must
        not cap the values near them.
    grey_countries : sequence of str, optional
        The countries the figure greys because they carry no estimate.
    country_geometry : pandas.DataFrame, optional
        Authoritative geometry for the requested scope. Requires source_id,
        version_id, geographic_scope and an exact country_codes_in_scope
        inventory. A country_code column is accepted, otherwise country codes
        form the index. European geometry and worldwide geometry are separate.
    geographic_scope : str, optional
        'Europe' for Figure 3 or 'world' for Figure B1. European geometry plots
        sampled countries within its declared scope; world geometry must cover
        every sampled target. Scope mismatches and missing geometry raise.

    Returns
    -------
    pandas.DataFrame
        One row per plotted country carrying the value, the scale position,
        whether the value lies within the scale, and the grey flag.

    Raises
    ------
    IngestError
        If an in-scope sample country lacks a contrast, or geometry, source
        identity, geographic scope or the exact scope inventory is invalid.
        A missing estimate is never concealed by assigning it a grey category.
    """
    if not isinstance(country_geometry, pd.DataFrame) or country_geometry.empty:
        raise IngestError(
            "Figure 3 requires an authoritative non-empty country geometry."
        )
    source_id = str(country_geometry.attrs.get("source_id", "")).strip()
    version_id = str(country_geometry.attrs.get("version_id", "")).strip()
    if not source_id or not version_id:
        raise IngestError(
            "Figure 3 geometry lacks authoritative source/version identity."
        )
    if geographic_scope not in {"Europe", "world"}:
        raise IngestError("Unregistered choropleth geographic scope.")
    if country_geometry.attrs.get("geographic_scope") != geographic_scope:
        raise IngestError("Geometry geographic scope differs from the requested figure.")
    scope_codes = country_geometry.attrs.get("country_codes_in_scope")
    if not isinstance(scope_codes, (tuple, list)) or not scope_codes:
        raise IngestError("Geometry requires an independently sourced scope inventory.")
    if len(scope_codes) != len(set(scope_codes)):
        raise IngestError("Geometry scope inventory has duplicated country identities.")
    geometry = country_geometry.copy(deep=True)
    if "country_code" in geometry.columns:
        if bool(geometry["country_code"].isna().any()):
            raise IngestError("Figure 3 geometry carries a null country code.")
        geometry["country_code"] = geometry["country_code"].astype(str).str.strip()
        geometry = geometry.set_index("country_code", drop=True)
    geometry.index = pd.Index(
        [str(code).strip() for code in geometry.index], name="country_code"
    )
    if (
        not bool(geometry.index.is_unique)
        or any(not code for code in geometry.index)
        or "geometry" not in geometry.columns
    ):
        raise IngestError(
            "Figure 3 geometry has duplicate/blank keys or no geometry column."
        )
    if set(str(code) for code in scope_codes) != set(geometry.index):
        raise IngestError("Geometry rows do not cover the exact geographic scope inventory.")
    for code, shape in geometry["geometry"].items():
        if (
            shape is None
            or bool(getattr(shape, "is_empty", True))
            or not bool(getattr(shape, "is_valid", False))
        ):
            raise IngestError(f"Figure 3 geometry is invalid for {code}.")
    # Compose the grey and sample sets without silently deduplicating a
    # malformed scientific roster.
    grey_codes = tuple(str(code).strip() for code in grey_countries)
    sample_codes = tuple(str(code).strip() for code in sample_countries)
    if (
        not sample_codes
        or any(not code for code in sample_codes)
        or len(sample_codes) != len(set(sample_codes))
        or any(not code for code in grey_codes)
        or len(grey_codes) != len(set(grey_codes))
    ):
        raise IngestError("Figure 3 sample/grey country rosters are malformed.")
    grey = set(grey_codes)
    sample = set(sample_codes)
    contrast_codes = set(str(code).strip() for code in contrasts)
    if contrast_codes != sample:
        raise IngestError(
            "Figure 3 contrast identities must equal the declared sample roster."
        )
    required_sample = sample if geographic_scope == "world" else sample & set(scope_codes)
    missing_sample_geometry = sorted(required_sample - set(geometry.index))
    missing_excluded_geometry = sorted(grey - set(geometry.index))
    if missing_sample_geometry or missing_excluded_geometry:
        raise IngestError(
            "Figure 3 geometry omits sampled/excluded countries: "
            f"sample={missing_sample_geometry}, excluded={missing_excluded_geometry}."
        )
    # Compose the lower bound of the scale.
    lower, upper = float(scale[0]), float(scale[1])
    # Refuse a malformed scale.
    if lower >= upper:
        # The failure names the scale.
        raise IngestError(f"The colour scale {scale!r} is not an ordered interval.")
    numeric_contrasts: Dict[str, float] = {}
    for code, raw_value in contrasts.items():
        if isinstance(raw_value, (bool, np.bool_)):
            raise IngestError(f"Figure 3 contrast for {code} is Boolean.")
        try:
            value = float(raw_value)
        except (TypeError, ValueError, OverflowError) as error:
            raise IngestError(f"Figure 3 contrast for {code} is malformed.") from error
        if not np.isfinite(value):
            raise IngestError(f"Figure 3 contrast for {code} is non-finite.")
        numeric_contrasts[str(code).strip()] = value
    frame = geometry.copy(deep=True)
    frame["value"] = [
        numeric_contrasts.get(str(code), float("nan")) for code in frame.index
    ]
    frame["map_status"] = [
        (
            "excluded_ukraine"
            if str(code) in grey
            else "sample_estimate"
            if str(code) in sample
            else "non_sample"
        )
        for code in frame.index
    ]
    required_statuses = {"sample_estimate", "non_sample", "excluded_ukraine"}
    if not required_statuses.issubset(set(frame["map_status"].astype(str))):
        raise IngestError(
            "Figure 3 geometry must distinguish sampled, non-sample and Ukraine rows."
        )
    frame["grey"] = ~frame["map_status"].eq("sample_estimate")
    frame["in_scale"] = frame["value"].map(
        lambda value: bool(np.isfinite(value) and lower <= float(value) <= upper)
    )
    frame["scale_lower"] = lower
    frame["scale_upper"] = upper
    frame.attrs.update(country_geometry.attrs)
    frame.attrs["geometry_source_id"] = source_id
    frame.attrs["geometry_version_id"] = version_id
    frame.attrs["sample_not_in_geometry"] = tuple(sorted(sample - set(geometry.index)))
    frame.attrs["geographic_scope"] = geographic_scope
    frame.attrs["render_spec"] = {
        "kind": "country_choropleth",
        "geographic_scope": geographic_scope,
        "value": "value",
        "geometry": "geometry",
        "status": "map_status",
        "scale": [lower, upper],
        "scale_unit": "log_points",
        "non_sample_colour": "#d9d9d9",
        "excluded_colour": "#969696",
        "out_of_range_policy": "colourbar_extend_both_no_data_clipping",
    }
    # Log the frame's extent and the out-of-scale values.
    logger.info(
        "choropleth data composed across %d countr(ies) with %d greyed and %d "
        "outside the scale.",
        len(frame),
        int(frame["grey"].sum()),
        int((~frame["grey"] & ~frame["in_scale"]).sum()),
    )
    # Return the frame.
    return frame


# Register the unspecified inputs as reproducibility gaps.
def register_reproducibility_gaps(
    extra_gaps: Optional[Sequence[Mapping[str, str]]] = None,
    geometry_provenance: Optional[Mapping[str, str]] = None,
) -> pd.DataFrame:
    """
    Register the unspecified inputs as reproducibility gaps.

    Parameters
    ----------
    extra_gaps : sequence of mapping, optional
        Further gaps, each carrying an identifier, a description and the
        consequence of the omission.
    geometry_provenance : mapping of str to str, optional
        The authoritative source/version identity of the geometry actually
        supplied to Figure 3. When absent the map source remains unresolved.

    Returns
    -------
    pandas.DataFrame
        One row per gap keyed on its identifier, carrying the description, the
        consequence and the state of the library the gap concerns. The
        specification identifies the choropleth's basemap as unspecified, so
        the gap is a first-class record rather than a sentence in the prose.

    Raises
    ------
    IngestError
        If an extra gap lacks one of its declared fields.
    """
    # Accumulate the gap rows.
    rows: List[dict] = [
        {
            "gap_id": "plotting_library_state",
            "description": "the rendering stack's state is recorded rather than assumed",
            "consequence": (
                "the figures are written when the stack imports, and the data "
                "bundle is complete either way"
            ),
            "state": "plotting_library_present" if _PLOTTING_AVAILABLE else "plotting_library_absent",
        },
    ]
    if geometry_provenance is None:
        rows.append(
            {
                "gap_id": BASEMAP_GAP,
                "description": "the choropleth's basemap source is unspecified",
                "consequence": "Figure 3 is blocked until real versioned geometry is supplied",
                "state": "unresolved",
            }
        )
    else:
        geometry_source = str(geometry_provenance.get("source_id", "")).strip()
        geometry_version = str(geometry_provenance.get("version_id", "")).strip()
        if not geometry_source or not geometry_version:
            raise IngestError("Resolved geometry provenance lacks source/version identity.")
        rows.append(
            {
                "gap_id": "choropleth_geometry_provenance",
                "description": "authoritative geometry used by Figure 3",
                "consequence": "none; source and version are bound into the production digest",
                "state": f"resolved:{geometry_source}:{geometry_version}",
            }
        )
    # Append the caller's gaps when supplied.
    if extra_gaps is not None:
        # Iterate over the extra gaps.
        for extra in extra_gaps:
            # Refuse a gap lacking a declared field.
            for field_name in ("gap_id", "description", "consequence"):
                # Test the field's presence.
                if field_name not in extra:
                    # The failure names the absent field.
                    raise IngestError(
                        f"An extra gap lacks the declared field {field_name!r}."
                    )
            # Append the caller's gap.
            rows.append(
                {
                    "gap_id": str(extra["gap_id"]),
                    "description": str(extra["description"]),
                    "consequence": str(extra["consequence"]),
                    "state": str(extra.get("state", "unspecified")),
                }
            )
    # Assemble the frame keyed on the gap's identifier.
    gaps: pd.DataFrame = pd.DataFrame(rows).set_index("gap_id")
    if not bool(gaps.index.is_unique):
        raise IngestError("Reproducibility-gap identifiers are duplicated.")
    # Log the gaps' extent.
    logger.info("reproducibility gaps registered across %d row(s).", len(gaps))
    # Return the frame.
    return gaps

# -------------------------------------------------------------------------------------------------------------------------------
# Task 31, Step 3: The overlay, the appendix figures and the tables
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the matched-design overlay's data.
def build_overlay_data(
    baseline_path: pd.DataFrame,
    matched_paths: Mapping[str, pd.DataFrame],
    normalization_year: int = OVERLAY_NORMALIZATION_YEAR,
) -> pd.DataFrame:
    """
    Compose the matched-design overlay's data.

    Parameters
    ----------
    baseline_path : pandas.DataFrame
        The baseline's dynamic path.
    matched_paths : mapping of str to pandas.DataFrame
        Exactly the independently fitted ``high_income`` and
        ``mahalanobis_m5`` paths, carrying content, effective-roster, positive
        weight and upstream-result lineage.
    normalization_year : int, optional
        The reference year every series is normalized on.

    Returns
    -------
    pandas.DataFrame
        One row per year and series carrying the normalized value, with the
        ribbon bounds populated for the baseline alone: the specification fixes
        that the baseline retains its band while the matched designs are lines,
        because a band on a frequency-weighted design would overstate its
        precision relative to the baseline.

    Raises
    ------
    IngestError
        If a required design, reference year, baseline ribbon, independent
        fitted lineage, effective roster or positive-weight record is absent,
        stale or inconsistent.
    """
    if isinstance(normalization_year, (bool, np.bool_)) or not isinstance(
        normalization_year, (int, np.integer)
    ):
        raise IngestError("Figure 4 normalization year must be an exact integer.")
    required_designs = set(OVERLAY_MATCHED_DESIGNS)
    if set(str(name) for name in matched_paths) != required_designs:
        raise IngestError(
            "Figure 4 requires independently fitted high_income and "
            "mahalanobis_m5 paths exactly."
        )
    # Refuse a baseline path without the unnormalised scientific path and its
    # pointwise band. Normalising a previously normalised or relabelled path
    # would not establish the own-2013 reference contract.
    for column in ("coalition_mean", "ribbon_lower", "ribbon_upper"):
        if column not in baseline_path.columns:
            raise IngestError(f"The baseline path lacks the column {column!r}.")
    if not bool(baseline_path.index.is_unique) or int(normalization_year) not in set(
        int(year) for year in baseline_path.index
    ):
        raise IngestError("The baseline path omits or duplicates its 2013 reference.")
    # Verify, rather than merely trust, each path's saved content/roster/weight
    # lineage. These are the strongest checks possible at this public
    # path-level boundary; B119 records that upstream objects supplied them.
    def verify_path_lineage(path: pd.DataFrame, design_id: str) -> str:
        roster_raw = path.attrs.get("effective_roster")
        weights_raw = path.attrs.get("effective_weights")
        provenance_origin = str(path.attrs.get("provenance_origin", "")).strip()
        upstream_result_sha256 = str(
            path.attrs.get("upstream_result_sha256", "")
        ).strip()
        expected_origin = (
            "verified_fitted_and_realised_baseline_matrices"
            if design_id == "baseline"
            else "upstream_matching_design_result"
        )
        if provenance_origin != expected_origin or not re.fullmatch(
            r"[0-9a-f]{64}", upstream_result_sha256
        ):
            raise IngestError(
                f"Overlay design {design_id!r} lacks verified upstream-object provenance."
            )
        if not isinstance(roster_raw, Sequence) or isinstance(roster_raw, (str, bytes)):
            raise IngestError(f"Overlay design {design_id!r} lacks its effective roster.")
        roster = tuple(str(code).strip() for code in roster_raw)
        if not roster or any(not code for code in roster) or len(roster) != len(set(roster)):
            raise IngestError(f"Overlay design {design_id!r} has a malformed roster.")
        if not isinstance(weights_raw, Mapping) or set(str(key) for key in weights_raw) != set(roster):
            raise IngestError(f"Overlay design {design_id!r} lacks roster-aligned weights.")
        weights: Dict[str, float] = {}
        for code in roster:
            raw_weight = weights_raw[code]
            if isinstance(raw_weight, (bool, np.bool_)):
                raise IngestError(f"Overlay design {design_id!r} has a Boolean weight.")
            weight = float(raw_weight)
            if not np.isfinite(weight) or weight <= 0.0:
                raise IngestError(f"Overlay design {design_id!r} has a nonpositive weight.")
            weights[code] = weight
        if not np.isclose(sum(weights.values()), 1.0, rtol=0.0, atol=1.0e-12):
            raise IngestError(f"Overlay design {design_id!r} weights do not sum to one.")
        content_digest = _content_digest(path.loc[:, ["coalition_mean"]])
        roster_digest = _content_digest(roster)
        weights_digest = _content_digest(weights)
        if str(path.attrs.get("path_content_sha256", "")) != content_digest:
            raise IngestError(f"Overlay design {design_id!r} path-content digest is stale.")
        if str(path.attrs.get("effective_roster_sha256", "")) != roster_digest:
            raise IngestError(f"Overlay design {design_id!r} roster digest is stale.")
        if str(path.attrs.get("effective_weights_sha256", "")) != weights_digest:
            raise IngestError(f"Overlay design {design_id!r} weight digest is stale.")
        lineage_payload = {
            "design_id": design_id,
            "path_content_sha256": content_digest,
            "effective_roster_sha256": roster_digest,
            "effective_weights_sha256": weights_digest,
            "provenance_origin": provenance_origin,
            "upstream_result_sha256": upstream_result_sha256,
        }
        if design_id != "baseline":
            # Validate the donor support and weights bound by the live producer.
            donor_roster_raw = path.attrs.get("effective_donor_roster")
            donor_weights_raw = path.attrs.get("effective_donor_weights")
            if not isinstance(donor_roster_raw, (list, tuple)):
                raise IngestError(f"Overlay design {design_id!r} lacks donor support.")
            donor_roster = tuple(str(code).strip() for code in donor_roster_raw)
            if (not donor_roster or any(not code for code in donor_roster)
                    or len(set(donor_roster)) != len(donor_roster)):
                raise IngestError(f"Overlay design {design_id!r} has invalid donor support.")
            if not isinstance(donor_weights_raw, Mapping) or set(donor_weights_raw) != set(donor_roster):
                raise IngestError(f"Overlay design {design_id!r} has unaligned donor weights.")
            donor_weights = {}
            for code in donor_roster:
                raw_weight = donor_weights_raw[code]
                if isinstance(raw_weight, (bool, np.bool_)):
                    raise IngestError(f"Overlay design {design_id!r} has Boolean donor weights.")
                weight = float(raw_weight)
                if not np.isfinite(weight) or weight <= 0.0:
                    raise IngestError(f"Overlay design {design_id!r} has nonpositive donor weights.")
                donor_weights[code] = weight
            if not np.isclose(sum(donor_weights.values()), 1.0, rtol=0.0, atol=1e-12):
                raise IngestError(f"Overlay design {design_id!r} donor weights do not sum to one.")
            donor_roster_digest = _content_digest(donor_roster)
            donor_weights_digest = _content_digest(donor_weights)
            if (path.attrs.get("effective_donor_roster_sha256") != donor_roster_digest
                    or path.attrs.get("effective_donor_weights_sha256") != donor_weights_digest):
                raise IngestError(f"Overlay design {design_id!r} donor lineage is stale.")
            lineage_payload.update({
                "effective_donor_roster_sha256": donor_roster_digest,
                "effective_donor_weights_sha256": donor_weights_digest,
            })
        expected_lineage = hashlib.sha256(
            json.dumps(lineage_payload, sort_keys=True, separators=(",", ":"))
            .encode("utf-8")
        ).hexdigest()
        if str(path.attrs.get("fit_lineage_sha256", "")) != expected_lineage:
            raise IngestError(f"Overlay design {design_id!r} lineage is stale.")
        return expected_lineage

    baseline_lineage = verify_path_lineage(baseline_path, "baseline")
    if str(baseline_path.attrs.get("design_id", "")) != "baseline":
        raise IngestError("The baseline overlay path has a wrong design identity.")
    baseline_reference = float(
        baseline_path.loc[int(normalization_year), "coalition_mean"]
    )
    baseline_years = tuple(int(year) for year in baseline_path.index)
    if not np.isfinite(
        baseline_path[["coalition_mean", "ribbon_lower", "ribbon_upper"]]
        .to_numpy(dtype=float)
    ).all():
        raise IngestError("The baseline overlay path carries non-finite scientific data.")
    # Accumulate the overlay rows.
    rows: List[dict] = []
    # Record the baseline's series.
    for year, row in baseline_path.iterrows():
        # Append the baseline's row with its band.
        rows.append(
            {
                "year": int(year),
                "series": "baseline",
                "value": float(row["coalition_mean"]) - baseline_reference,
                "lower": float(row["ribbon_lower"]) - baseline_reference,
                "upper": float(row["ribbon_upper"]) - baseline_reference,
                "band": True,
                "design_id": "baseline",
                "fit_lineage_sha256": baseline_lineage,
                "normalization_value": baseline_reference,
            }
        )
    lineage_ids = {baseline_lineage}
    # Record each matched design's series as a line without a band.
    for name in sorted(matched_paths.keys()):
        # Read the matched path.
        path: pd.DataFrame = matched_paths[name]
        if path is baseline_path or id(path) == id(baseline_path):
            raise IngestError(f"The matched path {name!r} aliases the baseline path.")
        if "coalition_mean" not in path.columns:
            raise IngestError(f"The matched path {name!r} lacks coalition_mean.")
        if not bool(path.index.is_unique) or tuple(int(year) for year in path.index) != baseline_years:
            raise IngestError(
                f"The matched path {name!r} has a different or duplicate year axis."
            )
        design_id = str(path.attrs.get("design_id", ""))
        lineage = str(path.attrs.get("fit_lineage_sha256", ""))
        roster_digest = str(path.attrs.get("effective_roster_sha256", ""))
        if design_id != str(name):
            raise IngestError(f"The matched path {name!r} has wrong design identity.")
        verified_lineage = verify_path_lineage(path, str(name))
        if lineage != verified_lineage or lineage in lineage_ids:
            raise IngestError(
                f"The matched path {name!r} lacks distinct fitted-object lineage."
            )
        if not re.fullmatch(r"[0-9a-f]{64}", roster_digest):
            raise IngestError(f"The matched path {name!r} lacks effective-roster lineage.")
        if name == "mahalanobis_m5" and str(
            path.attrs.get("weighting_scheme", "")
        ) != "selection_frequency_positive_support":
            raise IngestError(
                "The Mahalanobis overlay is not the positive-weight fitted design."
            )
        if name == "high_income" and str(
            path.attrs.get("weighting_scheme", "")
        ) != "equal_weight_high_income":
            raise IngestError("The high-income overlay has wrong weighting provenance.")
        values = pd.to_numeric(path["coalition_mean"], errors="coerce")
        if bool(values.isna().any()) or not np.isfinite(values.to_numpy(dtype=float)).all():
            raise IngestError(f"The matched path {name!r} is not finite.")
        reference = float(path.loc[int(normalization_year), "coalition_mean"])
        lineage_ids.add(lineage)
        # Iterate over the path's years.
        for year, row in path.iterrows():
            # Append the design's row, whose band columns are undefined.
            rows.append(
                {
                    "year": int(year),
                    "series": str(name),
                    "value": float(row["coalition_mean"]) - reference,
                    "lower": float("nan"),
                    "upper": float("nan"),
                    "band": False,
                    "design_id": str(name),
                    "fit_lineage_sha256": lineage,
                    "normalization_value": reference,
                }
            )
    # Assemble the overlay frame keyed on the year and the series.
    overlay: pd.DataFrame = pd.DataFrame(rows).set_index(["year", "series"])
    if not bool(overlay.index.is_unique):
        raise IngestError("Figure 4 overlay duplicates a year/design key.")
    for design in ("baseline",) + OVERLAY_MATCHED_DESIGNS:
        if abs(float(overlay.loc[(int(normalization_year), design), "value"])) > 1.0e-12:
            raise IngestError(f"Figure 4 design {design!r} is not zero at 2013.")
    overlay.attrs["normalization_year"] = int(normalization_year)
    overlay.attrs["provenance_status"] = (
        "path_content_rosters_weights_verified_against_supplied_upstream_result_digests"
    )
    overlay.attrs["render_spec"] = {
        "kind": "matched_design_overlay",
        "series": ["baseline", *OVERLAY_MATCHED_DESIGNS],
        "baseline_band": ["lower", "upper"],
        "matched_bands": False,
        "normalization_year": int(normalization_year),
    }
    # Log the overlay's extent.
    logger.info(
        "overlay data composed across %d series-row(s) over %d series.",
        len(overlay),
        len(set(overlay.index.get_level_values("series"))),
    )
    # Return the frame.
    return overlay


# Compose the placebo histogram's data.
def build_placebo_histogram_data(
    draws: np.ndarray,
    observed: float,
    bins: int = 40,
) -> pd.DataFrame:
    """
    Compose the placebo histogram's data.

    Parameters
    ----------
    draws : numpy.ndarray
        The pseudo-assignment effects.
    observed : float
        The observed effect the histogram is read against.
    bins : int, optional
        The number of histogram bins.

    Returns
    -------
    pandas.DataFrame
        One row per bin carrying its bounds and count, with the observed and
        zero reference lines recorded in the frame's attributes so the renderer
        draws them from the data rather than from constants.

    Raises
    ------
    IngestError
        If the draws are empty, because the histogram would then be vacuous.
    """
    draw_array = np.asarray(draws, dtype=float)
    # Refuse empty or non-finite draws and a malformed observed marker.
    if draw_array.size == 0:
        # The failure states the precondition.
        raise IngestError("Cannot compose a histogram from an empty draw vector.")
    if not np.isfinite(draw_array).all():
        raise IngestError("The placebo histogram carries a non-finite draw.")
    if isinstance(observed, (bool, np.bool_)) or not np.isfinite(float(observed)):
        raise IngestError("The placebo histogram observed ATT is malformed.")
    if (
        isinstance(bins, (bool, np.bool_))
        or not isinstance(bins, (int, np.integer))
        or int(bins) <= 0
    ):
        raise IngestError("The placebo histogram bin count is malformed.")
    # Compose the counts and the bin edges.
    counts, edges = np.histogram(draw_array, bins=int(bins))
    # Assemble the frame.
    histogram: pd.DataFrame = pd.DataFrame(
        {
            "bin_lower": edges[:-1],
            "bin_upper": edges[1:],
            "count": counts,
        }
    )
    # Record the reference lines in the frame's attributes.
    histogram.attrs["observed_line"] = float(observed)
    # Record the zero line.
    histogram.attrs["zero_line"] = 0.0
    # Record the draw count.
    histogram.attrs["draw_count"] = int(draw_array.size)
    histogram.attrs["observed_role"] = "preferred_incremental_att"
    histogram.attrs["render_spec"] = {
        "kind": "placebo_histogram",
        "observed_line": float(observed),
        "observed_colour": "red",
        "zero_line": 0.0,
        "zero_linestyle": "dotted",
        "bins": int(bins),
    }
    # Log the histogram's extent.
    logger.info(
        "placebo histogram composed across %d bin(s) over %d draw(s).",
        len(histogram),
        int(draw_array.size),
    )
    # Return the frame.
    return histogram


# Compose the reference-window coefficient plot's data.
def build_reference_window_data(
    estimates: Mapping[str, Mapping[str, float]],
) -> pd.DataFrame:
    """
    Compose the reference-window coefficient plot's data.

    Parameters
    ----------
    estimates : mapping of str to mapping of str to float
        The estimates keyed on the window's name, each carrying the estimate,
        its standard error and its two-sided probability.

    Returns
    -------
    pandas.DataFrame
        One row per window carrying the estimate with its interval bounds and
        its significance flag, which is the data the coefficient plot reads.

    Raises
    ------
    IngestError
        If an entry lacks a declared field, or if the mapping is empty.
    """
    # Refuse an empty mapping.
    if not estimates:
        # The failure states the precondition.
        raise IngestError("Cannot compose a coefficient plot from an empty mapping.")
    # Accumulate the rows.
    rows: List[dict] = []
    # Iterate over the windows in sorted order.
    for window in sorted(estimates.keys()):
        # Read the window's entry.
        entry = estimates[window]
        # Refuse an entry lacking a declared field.
        for field_name in ("estimate", "se_cd", "p_value"):
            # Test the field's presence.
            if field_name not in entry:
                # The failure names the absent field and the window.
                raise IngestError(
                    f"The entry for {window!r} lacks the declared field "
                    f"{field_name!r}."
                )
        # Read the estimate and its standard error.
        try:
            estimate = float(entry["estimate"])
            standard_error = float(entry["se_cd"])
            probability = float(entry["p_value"])
        except (TypeError, ValueError, OverflowError) as error:
            raise IngestError(
                f"The entry for {window!r} contains malformed numeric evidence."
            ) from error
        if (
            not np.isfinite([estimate, standard_error, probability]).all()
            or standard_error < 0.0
            or not 0.0 <= probability <= 1.0
        ):
            raise IngestError(
                f"The entry for {window!r} contains invalid coefficient evidence."
            )
        # Read the standard error.
        # Append the window's row.
        rows.append(
            {
                "window": str(window),
                "estimate": estimate,
                "se_cd": standard_error,
                "lower": estimate - RIBBON_CRITICAL_VALUE * standard_error,
                "upper": estimate + RIBBON_CRITICAL_VALUE * standard_error,
                "p_value": probability,
                "significant_5pct": bool(probability < 0.05),
                "source_id": str(entry.get("source_id", "")).strip(),
            }
        )
    # Assemble the frame keyed on the window.
    frame: pd.DataFrame = pd.DataFrame(rows).set_index("window")
    if not bool(frame.index.is_unique) or bool(frame["source_id"].eq("").any()):
        raise IngestError(
            "Reference-window coefficients duplicate a window or lack source lineage."
        )
    frame.attrs["render_spec"] = {
        "kind": "reference_window_coefficients",
        "estimate": "estimate",
        "interval": ["lower", "upper"],
        "zero_line": 0.0,
    }
    # Log the frame's extent.
    logger.info("reference-window data composed across %d window(s).", len(frame))
    # Return the frame.
    return frame


# -------------------------------------------------------------------------------------------------------------------------------
# Task 31, Step 3 (continued): The table bundle
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the declared table bundle from the frozen artefacts.
def build_table_bundle(
    sources: Mapping[str, pd.DataFrame],
    render_latex: bool = True,
) -> Tuple[Dict[str, pd.DataFrame], Dict[str, str]]:
    """
    Compose the declared table bundle from the frozen artefacts.

    Parameters
    ----------
    sources : mapping of str to pandas.DataFrame
        The table sources keyed on the table's identity, which the caller
        composes from the frozen artefacts; a table whose source is absent is
        refused rather than emitted empty. The reserved
        ``__cell_inventory__`` frame is a distinct, externally verified,
        versioned manuscript-cell inventory. Its caller-supplied attributes
        are not self-authenticating: source/version, an external verification
        reference, complete-coverage certification, per-table counts and a
        current digest are all mandatory, and independent structural lower
        bounds remain enforced.
    render_latex : bool, optional
        Whether to render each table through the declared tabular writer.

    Returns
    -------
    tuple
        The tables keyed on their identity and their tabular renderings as
        text, so the artefacts are reproducible without a rendering stack.

    Raises
    ------
    IngestError
        If a declared table or independently inventoried cell is absent,
        duplicated, mislabeled or neither a finite sourced result nor a
        concrete unresolved-gap record.
    """
    if not isinstance(render_latex, (bool, np.bool_)):
        raise IngestError("The table rendering flag must be Boolean.")
    declared = PRODUCTION_TABLES
    # Identify the missing tables.
    missing = sorted(set(declared) - set(str(key) for key in sources.keys()))
    # Refuse a missing table, because the bundle must be complete against the
    # declaration rather than partially populated.
    if missing:
        # The failure names the absent tables.
        raise IngestError(f"The table bundle is missing the declared table(s) {missing}.")
    inventory = sources.get("__cell_inventory__")
    if not isinstance(inventory, pd.DataFrame) or inventory.empty:
        raise IngestError(
            "The table bundle requires an independent non-empty cell inventory."
        )
    inventory_authority = str(inventory.attrs.get("authority_id", "")).strip()
    inventory_version = str(inventory.attrs.get("version_id", "")).strip()
    external_verification_reference = str(
        inventory.attrs.get("external_verification_reference", "")
    ).strip()
    inventory_complete = inventory.attrs.get("complete_manuscript_inventory")
    expected_counts = inventory.attrs.get("expected_cell_counts")
    declared_inventory_digest = str(
        inventory.attrs.get("inventory_sha256", "")
    ).strip()
    if (
        not inventory_authority
        or not inventory_version
        or not external_verification_reference
        or inventory_complete is not True
        or not isinstance(expected_counts, Mapping)
        or set(str(key) for key in expected_counts) != set(declared)
    ):
        raise IngestError(
            "The independent table inventory lacks authoritative source/version, "
            "an external verification reference, complete-coverage certification, "
            "or per-table cell counts. Caller-set attributes are not treated as "
            "self-authenticating."
        )
    missing_inventory_columns = sorted(
        set(TABLE_INVENTORY_COLUMNS) - set(inventory.columns)
    )
    if missing_inventory_columns:
        raise IngestError(
            f"The table inventory omits {missing_inventory_columns}."
        )
    inventory = inventory.loc[:, list(TABLE_INVENTORY_COLUMNS)].copy(deep=True)
    inventory["table_id"] = inventory["table_id"].astype(str)
    for column in ("row_identifier", "column_identifier", "unit", "procedure"):
        if bool(inventory[column].isna().any()):
            raise IngestError(f"The table inventory carries null {column} values.")
        inventory[column] = inventory[column].astype(str).str.strip()
        if bool(inventory[column].eq("").any()):
            raise IngestError(f"The table inventory carries blank {column} values.")
    if set(inventory["table_id"]) != set(declared):
        raise IngestError(
            "The independent cell inventory does not cover exactly Tables 3–10, A1–A8 and B1–B7."
        )
    key_columns = ["table_id", "row_identifier", "column_identifier"]
    if bool(inventory.duplicated(key_columns).any()):
        raise IngestError("The independent table-cell inventory duplicates a key.")
    precision_values: List[int] = []
    for raw in inventory["display_precision"]:
        if (
            isinstance(raw, (bool, np.bool_))
            or not isinstance(raw, (int, np.integer))
            or not 0 <= int(raw) <= 12
        ):
            raise IngestError("The table inventory carries malformed display precision.")
        precision_values.append(int(raw))
    inventory["display_precision"] = precision_values
    observed_counts = inventory.groupby("table_id", observed=True).size().to_dict()
    normalized_expected_counts: Dict[str, int] = {}
    for table_id, raw_count in expected_counts.items():
        if (
            isinstance(raw_count, (bool, np.bool_))
            or not isinstance(raw_count, (int, np.integer))
            or int(raw_count) <= 0
        ):
            raise IngestError("The authoritative table inventory carries a malformed count.")
        normalized_expected_counts[str(table_id)] = int(raw_count)
    if observed_counts != normalized_expected_counts:
        raise IngestError(
            "The independent table inventory is partial relative to its authoritative counts."
        )
    below_structural_minimum = {
        table_id: (observed_counts.get(table_id, 0), minimum)
        for table_id, minimum in TABLE_MINIMUM_REQUIRED_CELLS.items()
        if observed_counts.get(table_id, 0) < int(minimum)
    }
    if below_structural_minimum:
        raise IngestError(
            "The purported complete inventory is below independent manuscript "
            f"cell-count bounds: {below_structural_minimum}."
        )
    observed_inventory_digest = _content_digest(inventory)
    if not re.fullmatch(r"[0-9a-f]{64}", declared_inventory_digest) or (
        declared_inventory_digest != observed_inventory_digest
    ):
        raise IngestError("The authoritative table inventory digest is absent or stale.")
    inventory_indexed = inventory.set_index(key_columns).sort_index()
    # Accumulate the tables.
    tables: Dict[str, pd.DataFrame] = {}
    # Accumulate the renderings.
    renderings: Dict[str, str] = {}
    # Iterate over the declared tables in their order.
    for name in declared:
        # Copy the source so the caller's frame is never modified.
        source = sources[name]
        if not isinstance(source, pd.DataFrame):
            raise IngestError(f"The table {name!r} source is not a DataFrame.")
        table: pd.DataFrame = source.copy(deep=True)
        # Refuse an empty source.
        if len(table) == 0:
            # The failure names the table.
            raise IngestError(f"The table {name!r} carries no row.")
        missing_cell_columns = sorted(set(TABLE_CELL_COLUMNS) - set(table.columns))
        if missing_cell_columns:
            raise IngestError(
                f"The table {name!r} omits cell fields {missing_cell_columns}."
            )
        table = table.loc[:, list(TABLE_CELL_COLUMNS)].copy(deep=True)
        table["table_id"] = table["table_id"].astype(str)
        if set(table["table_id"]) != {name}:
            raise IngestError(f"The table {name!r} carries a contradictory identity.")
        for column in (
            "row_identifier",
            "column_identifier",
            "unit",
            "procedure",
            "status",
            "source_id",
            "gap_id",
        ):
            if bool(table[column].isna().any()):
                table[column] = table[column].fillna("")
            table[column] = table[column].astype(str).str.strip()
        table_precisions: List[int] = []
        for raw_precision in table["display_precision"]:
            if (
                isinstance(raw_precision, (bool, np.bool_))
                or not isinstance(raw_precision, (int, np.integer))
                or not 0 <= int(raw_precision) <= 12
            ):
                raise IngestError(
                    f"The table {name!r} carries malformed display precision."
                )
            table_precisions.append(int(raw_precision))
        table["display_precision"] = table_precisions
        if bool(table.duplicated(key_columns).any()):
            raise IngestError(f"The table {name!r} duplicates a cell key.")
        table_indexed = table.set_index(key_columns).sort_index()
        expected_index = inventory_indexed.loc[[name]].index
        if set(table_indexed.index) != set(expected_index):
            raise IngestError(
                f"The table {name!r} cells do not equal its independent inventory."
            )
        expected_metadata = inventory_indexed.loc[table_indexed.index]
        for column in ("unit", "display_precision", "procedure"):
            if not table_indexed[column].equals(expected_metadata[column]):
                raise IngestError(
                    f"The table {name!r} mislabels inventoried {column}."
                )
        allowed_statuses = {TABLE_STATUS_COMPUTED, TABLE_STATUS_UNRESOLVED}
        if not set(table_indexed["status"]).issubset(allowed_statuses):
            raise IngestError(f"The table {name!r} carries an undeclared cell status.")
        if bool(table_indexed["source_id"].eq("").any()):
            raise IngestError(f"The table {name!r} carries a cell without source lineage.")
        computed = table_indexed["status"].eq(TABLE_STATUS_COMPUTED)
        numeric_values = pd.to_numeric(
            table_indexed.loc[computed, "value_unrounded"], errors="coerce"
        )
        if bool(numeric_values.isna().any()) or not bool(
            np.isfinite(numeric_values.to_numpy(dtype=float)).all()
        ):
            raise IngestError(f"The table {name!r} fabricates or omits a computed value.")
        if bool(table_indexed.loc[computed, "gap_id"].ne("").any()):
            raise IngestError(f"The table {name!r} labels a computed cell as a gap.")
        unresolved = ~computed
        if bool(table_indexed.loc[unresolved, "value_unrounded"].notna().any()) or bool(
            table_indexed.loc[unresolved, "gap_id"].eq("").any()
        ):
            raise IngestError(
                f"The table {name!r} unresolved cells need null values and concrete gap IDs."
            )
        table_indexed.attrs["inventory_sha256"] = _content_digest(
            expected_metadata.reset_index()
        )
        table_indexed.attrs["inventory_authority_id"] = inventory_authority
        table_indexed.attrs["inventory_version_id"] = inventory_version
        table_indexed.attrs["external_verification_reference"] = (
            external_verification_reference
        )
        table_indexed.attrs["complete_manuscript_inventory"] = True
        table_indexed.attrs["inventory_provenance_status"] = (
            "externally_supplied_structurally_verified_not_self_authenticating"
        )
        table_indexed.attrs["render_spec"] = {
            "kind": "publication_table",
            "table_id": name,
            "unrounded_column": "value_unrounded",
            "precision_column": "display_precision",
            "gap_status": TABLE_STATUS_UNRESOLVED,
        }
        # Record the table.
        tables[name] = table_indexed
        # Render the table when requested.
        if bool(render_latex):
            # Compose the tabular rendering through the declared writer.
            renderings[name] = table_indexed.to_latex(index=True, escape=False)
    # Log the bundle's extent.
    logger.info(
        "table bundle composed across %d table(s) with %d rendering(s).",
        len(tables),
        len(renderings),
    )
    # Return the tables and their renderings.
    return tables, renderings


# Digest one production frame including attributes and geometry content.
def _production_frame_digest(frame: pd.DataFrame) -> str:
    """Return an order-, schema-, text-, missingness-, geometry- and spec-sensitive digest."""
    if not isinstance(frame, pd.DataFrame):
        raise IngestError("Production content must be a pandas DataFrame.")
    canonical = pd.DataFrame(frame.copy(deep=True))
    if "geometry" in canonical.columns:
        canonical["geometry"] = canonical["geometry"].map(
            lambda shape: (
                str(getattr(shape, "wkb_hex"))
                if shape is not None and hasattr(shape, "wkb_hex")
                else repr(shape)
            )
        )
    content_digest = _content_digest(canonical)
    attributes = json.dumps(
        dict(frame.attrs), sort_keys=True, separators=(",", ":"), default=str
    )
    return hashlib.sha256(
        (content_digest + "|" + attributes).encode("utf-8")
    ).hexdigest()


# Render each figure through its registered scientific specification.
def render_figures(
    figure_data: Mapping[str, pd.DataFrame],
    output_dir: Optional[Path],
) -> pd.DataFrame:
    """
    Render the figures when the plotting library permits.

    Parameters
    ----------
    figure_data : mapping of str to pandas.DataFrame
        The figures' data keyed on the figure's name.
    output_dir : pathlib.Path, optional
        The directory the figures are written to. When absent no file is
        written and the outcome records the absence.

    Returns
    -------
    pandas.DataFrame
        One row per figure recording its registered kind, full data/spec
        digests, semantic validation, whether and where it was written, and
        the observed PNG byte size and SHA-256 digest after rendering.

    Raises
    ------
    IngestError
        If the figure data or registered scientific rendering specification
        is empty/incomplete, or if a requested output cannot be written.
    """
    # Refuse empty figure data.
    if not figure_data:
        # The failure states the precondition.
        raise IngestError("Cannot render an empty figure set.")
    # Validate each registered rendering specification before any filesystem
    # mutation so a generic fallback plot can never conceal missing science.
    allowed_kinds = {
        "dynamic_path",
        "country_choropleth",
        "matched_design_overlay",
        "matched_design_panels",
        "placebo_histogram",
        "reference_window_coefficients",
    }
    for name, data in figure_data.items():
        if not isinstance(data, pd.DataFrame) or data.empty:
            raise IngestError(f"Figure {name!r} has no scientific data.")
        spec = data.attrs.get("render_spec")
        if not isinstance(spec, Mapping) or str(spec.get("kind", "")) not in allowed_kinds:
            raise IngestError(f"Figure {name!r} lacks a registered rendering specification.")
    # Accumulate the outcome rows.
    rows: List[dict] = []
    # Compose the output directory when supplied.
    directory: Optional[Path] = Path(output_dir) if output_dir is not None else None
    # Create the directory when it does not exist.
    if directory is not None:
        # Create the directory tree.
        directory.mkdir(parents=True, exist_ok=True)
    # Iterate over the figures in sorted order.
    for name in sorted(figure_data.keys()):
        # Record the figure's outcome.
        rows.append(
            {
                "figure": str(name),
                "written": bool(_PLOTTING_AVAILABLE and directory is not None),
                "path": (
                    str(directory / f"{name}.png")
                    if _PLOTTING_AVAILABLE and directory is not None
                    else ""
                ),
                "reason": (
                    "written"
                    if _PLOTTING_AVAILABLE and directory is not None
                    else (
                        "plotting library absent"
                        if not _PLOTTING_AVAILABLE
                        else "no output directory supplied"
                    )
                ),
                "figure_kind": str(
                    figure_data[name].attrs["render_spec"]["kind"]
                ),
                "data_sha256": _production_frame_digest(figure_data[name]),
                "spec_sha256": _content_digest(
                    dict(figure_data[name].attrs["render_spec"])
                ),
                "semantic_checks": "validated_before_render",
                "file_sha256": "",
                "byte_size": 0,
            }
        )
    # Assemble the outcome frame.
    outcomes: pd.DataFrame = pd.DataFrame(rows).set_index("figure")
    # Write the figures when the library and the directory permit.
    if _PLOTTING_AVAILABLE and directory is not None:
        # Iterate over the figures.
        for name in sorted(figure_data.keys()):
            # Compose the figure.
            figure = plt.figure(figsize=(9.0, 5.5))
            # Select the axes.
            axes = figure.add_subplot(1, 1, 1)
            # Read the figure's data.
            data: pd.DataFrame = figure_data[name]
            spec = dict(data.attrs["render_spec"])
            kind = str(spec["kind"])
            if kind == "dynamic_path":
                required = {
                    "normalized_mean", "normalized_lower", "normalized_upper",
                    "region", "in_wave_1", "in_wave_2",
                }
                if not required.issubset(data.columns):
                    raise IngestError(f"{name}: dynamic-path data are incomplete.")
                years = data.index.to_numpy(dtype=int)
                axes.axvspan(FIRST_TREATMENT_YEAR, SECOND_WAVE_YEAR - 1,
                             color="#fdd49e", alpha=0.28, label="Wave 1")
                axes.axvspan(SECOND_WAVE_YEAR, int(years.max()),
                             color="#fdbb84", alpha=0.30, label="Wave 2")
                axes.fill_between(
                    years,
                    data["normalized_lower"].to_numpy(dtype=float),
                    data["normalized_upper"].to_numpy(dtype=float),
                    color="#6baed6", alpha=0.25, label="Pointwise 95% band",
                )
                fitted = data["region"].astype(str).eq(REGION_FITTED).to_numpy()
                realized = ~fitted
                axes.plot(years[fitted], data["normalized_mean"].to_numpy(dtype=float)[fitted],
                          linestyle="--", color="#08519c", linewidth=2.0,
                          label="Fitted pre-treatment gap")
                axes.plot(years[realized], data["normalized_mean"].to_numpy(dtype=float)[realized],
                          linestyle="-", color="#08519c", linewidth=2.0,
                          label="Realized gap")
                axes.axvline(FIRST_TREATMENT_YEAR, color="black", linewidth=1.0)
                axes.axvline(SECOND_WAVE_YEAR, color="black", linewidth=1.0)
                axes.axhline(0.0, color="#636363", linewidth=0.8)
                axes.set_xlabel("Calendar year")
                axes.set_ylabel("Gap normalized to 2013")
                axes.legend(loc="best", fontsize=8)
            elif kind == "country_choropleth":
                required = {"geometry", "value", "map_status", "scale_lower", "scale_upper"}
                if not required.issubset(data.columns):
                    raise IngestError(f"{name}: choropleth data are incomplete.")
                if not _GEO_AVAILABLE:
                    raise IngestError("Figure 3 requires geopandas for real geometry rendering.")
                geopandas = __import__("geopandas")
                geo = geopandas.GeoDataFrame(
                    data.copy(deep=True), geometry="geometry",
                    crs=getattr(data, "crs", None),
                )
                lower = float(data["scale_lower"].iloc[0])
                upper = float(data["scale_upper"].iloc[0])
                colour_map = matplotlib.colormaps["RdBu_r"].copy()
                colour_map.set_under("#053061")
                colour_map.set_over("#67001f")
                normalizer = matplotlib.colors.Normalize(vmin=lower, vmax=upper, clip=False)
                non_sample = geo[geo["map_status"].eq("non_sample")]
                excluded = geo[geo["map_status"].eq("excluded_ukraine")]
                sampled = geo[geo["map_status"].eq("sample_estimate")]
                if len(non_sample):
                    non_sample.plot(ax=axes, color="#d9d9d9", edgecolor="white", linewidth=0.4)
                if len(excluded):
                    excluded.plot(ax=axes, color="#969696", edgecolor="black", linewidth=0.7,
                                  hatch="///")
                sampled.plot(ax=axes, column="value", cmap=colour_map, norm=normalizer,
                             edgecolor="white", linewidth=0.4)
                scalar = matplotlib.cm.ScalarMappable(norm=normalizer, cmap=colour_map)
                scalar.set_array([])
                figure.colorbar(scalar, ax=axes, extend="both", label="Incremental effect (log points)")
                axes.set_axis_off()
                axes.set_title("Country incremental effects; grey = non-sample/excluded")
            elif kind == "matched_design_panels":
                if data.index.nlevels != 3 or set(data.index.get_level_values("outcome")) != {"log_CPI", "log_real_GDPpc"}:
                    raise IngestError(f"{name}: Figure 4 must include the two distinct level outcomes.")
                figure.delaxes(axes)
                panel_axes = figure.subplots(1, 2)
                figure.set_size_inches(13.0, 5.5)
                for panel_axes_item, outcome in zip(panel_axes, ("log_CPI", "log_real_GDPpc")):
                    draw_matching_overlay_panel(panel_axes_item, data.xs(outcome, level="outcome"))
                    panel_axes_item.set_title(outcome)
            elif kind == "matched_design_overlay":
                required = {"value", "lower", "upper", "band"}
                if not required.issubset(data.columns) or data.index.nlevels != 2:
                    raise IngestError(f"{name}: overlay data are incomplete.")
                colours = {"baseline": "#08519c", "high_income": "#31a354", "mahalanobis_m5": "#de2d26"}
                baseline = data.xs("baseline", level="series").sort_index()
                years = baseline.index.to_numpy(dtype=int)
                axes.fill_between(years, baseline["lower"].to_numpy(dtype=float),
                                  baseline["upper"].to_numpy(dtype=float),
                                  color=colours["baseline"], alpha=0.20,
                                  label="Baseline pointwise 95% band")
                for series in ("baseline",) + OVERLAY_MATCHED_DESIGNS:
                    curve = data.xs(series, level="series").sort_index()
                    axes.plot(curve.index.to_numpy(dtype=int), curve["value"].to_numpy(dtype=float),
                              linewidth=2.0, color=colours[series], label=series)
                axes.axhline(0.0, color="#636363", linewidth=0.8)
                axes.axvline(FIRST_TREATMENT_YEAR, color="black", linewidth=0.8)
                axes.axvline(SECOND_WAVE_YEAR, color="black", linewidth=0.8)
                axes.set_xlabel("Calendar year")
                axes.set_ylabel("Coalition path normalized to 2013")
                axes.legend(loc="best", fontsize=8)
            elif kind == "placebo_histogram":
                required = {"bin_lower", "bin_upper", "count"}
                if not required.issubset(data.columns):
                    raise IngestError(f"{name}: placebo histogram data are incomplete.")
                widths = data["bin_upper"].to_numpy(dtype=float) - data["bin_lower"].to_numpy(dtype=float)
                axes.bar(data["bin_lower"].to_numpy(dtype=float), data["count"].to_numpy(dtype=float),
                         width=widths, align="edge", color="#9ecae1", edgecolor="white")
                axes.axvline(float(data.attrs["observed_line"]), color="red", linewidth=2.0,
                             label="Observed preferred ATT")
                axes.axvline(0.0, color="black", linestyle=":", linewidth=1.5, label="Zero")
                axes.set_xlabel("Size-matched placebo statistic")
                axes.set_ylabel("Frequency")
                axes.legend(loc="best", fontsize=8)
            elif kind == "reference_window_coefficients":
                required = {"estimate", "lower", "upper", "se_cd"}
                if not required.issubset(data.columns):
                    raise IngestError(f"{name}: reference-window data are incomplete.")
                positions = np.arange(len(data), dtype=float)
                estimates = data["estimate"].to_numpy(dtype=float)
                errors = np.vstack([
                    estimates - data["lower"].to_numpy(dtype=float),
                    data["upper"].to_numpy(dtype=float) - estimates,
                ])
                axes.errorbar(positions, estimates, yerr=errors, fmt="o", color="#08519c",
                              ecolor="#6baed6", capsize=4)
                axes.axhline(0.0, color="black", linestyle=":", linewidth=1.0)
                axes.set_xticks(positions, [str(value) for value in data.index], rotation=30, ha="right")
                axes.set_xlabel("Reference window")
                axes.set_ylabel("Incremental coefficient")
            else:  # pragma: no cover - guarded before filesystem mutation
                raise IngestError(f"{name}: undeclared rendering kind {kind!r}.")
            axes.set_title(str(name).replace("_", " "))
            # Tighten the layout.
            figure.tight_layout()
            # Write the figure.
            figure.savefig(directory / f"{name}.png", dpi=150)
            # Release the figure's resources.
            plt.close(figure)
            rendered_path = directory / f"{name}.png"
            if not rendered_path.is_file() or rendered_path.stat().st_size <= 0:
                raise IngestError(f"Figure {name!r} was not written completely.")
            outcomes.loc[name, "file_sha256"] = _file_sha256(rendered_path)
            outcomes.loc[name, "byte_size"] = int(rendered_path.stat().st_size)
        # Log the rendering.
        logger.info("figures rendered across %d file(s).", len(figure_data))
    else:
        # Log the recorded absence.
        logger.info(
            "figures not rendered: %s.",
            "plotting library absent" if not _PLOTTING_AVAILABLE else "no output directory",
        )
    # Return the outcomes.
    return outcomes


# Freeze the production behind a canonical digest.
def freeze_production(
    figure_data: Mapping[str, pd.DataFrame],
    tables: Mapping[str, pd.DataFrame],
    gaps: pd.DataFrame,
) -> str:
    """
    Freeze the production behind a canonical digest.

    Parameters
    ----------
    figure_data : mapping of str to pandas.DataFrame
        The figures' data.
    tables : mapping of str to pandas.DataFrame
        The composed tables.
    gaps : pandas.DataFrame
        The reproducibility gaps.

    Returns
    -------
    str
        The hexadecimal digest of complete labelled content, dtypes, text,
        missingness, geometry and rendering attributes in scientific row and
        column order, with mapping keys canonicalized.

    Raises
    ------
    IngestError
        If the figure data or the tables are empty.
    """
    # Refuse an empty figure set.
    if not figure_data:
        # The failure states the precondition.
        raise IngestError("Cannot freeze an empty figure set.")
    # Refuse an empty table set.
    if not tables:
        # The failure states the precondition.
        raise IngestError("Cannot freeze an empty table set.")
    if not isinstance(gaps, pd.DataFrame) or gaps.empty or not bool(gaps.index.is_unique):
        raise IngestError("Production gaps must be a non-empty uniquely keyed frame.")
    parts: List[str] = []
    # Hash complete labelled content in its scientific row/column order; only
    # mapping-key order is canonicalized. Frame row order remains meaningful,
    # so E10's [1,2] and [2,1] cannot collide.
    for name in sorted(figure_data.keys()):
        parts.append(f"F|{name}|{_production_frame_digest(figure_data[name])}")
    for name in sorted(tables.keys()):
        parts.append(f"T|{name}|{_production_frame_digest(tables[name])}")
    parts.append(f"G|{_production_frame_digest(gaps)}")
    return hashlib.sha256("\n".join(parts).encode("utf-8")).hexdigest()


# -------------------------------------------------------------------------------------------------------------------------------
# Task 31, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Produce the figures and tables from the frozen artefacts.
def produce_figures_and_tables(
    gap_matrices: Mapping[Tuple[str, str], pd.DataFrame],
    incremental_contrasts: Mapping[Tuple[str, str], pd.Series],
    matched_paths: Optional[Mapping[str, pd.DataFrame]] = None,
    placebo_draws: Optional[Mapping[str, np.ndarray]] = None,
    reference_window_estimates: Optional[Mapping[str, Mapping[str, float]]] = None,
    table_sources: Optional[Mapping[str, pd.DataFrame]] = None,
    output_dir: Optional[Path] = None,
    extra_gaps: Optional[Sequence[Mapping[str, str]]] = None,
    fitted_pre_gaps: Optional[Mapping[Tuple[str, str], pd.DataFrame]] = None,
    observed_preferred_att: Optional[Mapping[Any, float]] = None,
    country_geometry: Optional[pd.DataFrame] = None,
    overlay_pair: Optional[Tuple[str, str]] = None,
    matched_paths_by_pair: Optional[Mapping[Tuple[str, str], Mapping[str, pd.DataFrame]]] = None,
    world_geometry: Optional[pd.DataFrame] = None,
) -> FigureTableBundle:
    """
    Produce the figures and tables from the frozen artefacts.

    Parameters
    ----------
    gap_matrices : mapping of tuple to pandas.DataFrame
        The gap matrices keyed on the pair.
    incremental_contrasts : mapping of tuple to pandas.Series
        The per-country incremental contrasts keyed on the pair.
    matched_paths : mapping of str to pandas.DataFrame, optional
        The matched designs' paths keyed on the design's name.
    placebo_draws : mapping of str to numpy.ndarray, optional
        The pseudo-assignment draws keyed on the outcome's name.
    reference_window_estimates : mapping of str to mapping of str to float, optional
        The reference-window estimates keyed on the window's name.
    table_sources : mapping of str to pandas.DataFrame, optional
        The tables' sources keyed on their identity.
    output_dir : pathlib.Path, optional
        The directory the figures are written to.
    extra_gaps : sequence of mapping, optional
        Further reproducibility gaps.
    fitted_pre_gaps : mapping of tuple to pandas.DataFrame, optional
        Distinct fitted 2001--2013 country-gap matrices. These are joined to,
        but never confused with, the realised 2014--2024 gap matrices.
    observed_preferred_att : mapping, optional
        Actual observed preferred incremental ATTs keyed exactly like the
        placebo draw mapping. They provide Figure B2's red marker.
    country_geometry : pandas.DataFrame, optional
        Authoritative versioned European geometry for Figure 3, with an exact
        country_codes_in_scope inventory and geographic_scope='Europe'.
    world_geometry : pandas.DataFrame, optional
        Separately sourced worldwide geometry for Figure B1; requires an exact
        scope inventory and geographic_scope='world'.
    matched_paths_by_pair : mapping, optional
        Both outcome-specific matching fits required by the two-panel Figure 4.
    overlay_pair : tuple of str, optional
        The explicit policy/outcome whose independently fitted baseline,
        high-income and Mahalanobis paths form Figure 4.

    Returns
    -------
    FigureTableBundle
        The figure data, the tables, the renderings, the rendering outcomes,
        the gaps, the digest, the timestamp and the notes.

    Raises
    ------
    IngestError
        If the gap matrices are empty, or if any figure's data cannot be
        composed from the artefacts supplied.
    """
    if not gap_matrices or not incremental_contrasts:
        raise IngestError("Production requires explicit realised gaps and country contrasts.")
    if not isinstance(fitted_pre_gaps, Mapping) or set(fitted_pre_gaps) != set(gap_matrices):
        raise IngestError(
            "Figure 2 requires a distinct fitted pre-period gap matrix for every realised pair."
        )
    if not isinstance(country_geometry, pd.DataFrame) or country_geometry.empty:
        raise IngestError("Figure 3 requires real authoritative country geometry.")
    if not isinstance(world_geometry, pd.DataFrame) or world_geometry.empty:
        raise IngestError("Figure B1 requires independently sourced world geometry.")
    if not matched_paths:
        raise IngestError("Figure 4 requires independently fitted matched-design paths.")
    if not placebo_draws or not isinstance(observed_preferred_att, Mapping):
        raise IngestError(
            "Figure B2 requires placebo draws and their actual observed preferred ATTs."
        )
    if set(placebo_draws) != set(observed_preferred_att):
        raise IngestError("Figure B2 draw and observed-ATT identities diverge.")
    if not reference_window_estimates:
        raise IngestError("Figure B3 requires actual reference-window coefficient results.")
    if not table_sources:
        raise IngestError("Production requires the complete publication table sources.")
    if not set(incremental_contrasts).issubset(set(gap_matrices)):
        raise IngestError("A choropleth contrast lacks its corresponding fitted design.")
    if overlay_pair is None:
        if len(gap_matrices) != 1:
            raise IngestError("Figure 4 requires an explicit overlay_pair when multiple pairs exist.")
        resolved_overlay_pair = next(iter(gap_matrices))
    else:
        resolved_overlay_pair = (str(overlay_pair[0]), str(overlay_pair[1]))
    if resolved_overlay_pair not in gap_matrices:
        raise IngestError("Figure 4 overlay pair has no baseline gap matrix.")

    figure_data: Dict[str, pd.DataFrame] = {}
    dynamic_paths: Dict[Tuple[str, str], pd.DataFrame] = {}
    for raw_pair in sorted(gap_matrices.keys()):
        pair = (str(raw_pair[0]), str(raw_pair[1]))
        realised = gap_matrices[raw_pair]
        fitted = fitted_pre_gaps[raw_pair]
        if not isinstance(realised, pd.DataFrame) or not isinstance(fitted, pd.DataFrame):
            raise IngestError(f"{pair}: fitted/realised gaps must be DataFrames.")
        if set(str(code) for code in fitted.index) != set(str(code) for code in realised.index):
            raise IngestError(f"{pair}: fitted and realised target rosters diverge.")
        fitted_years = set(int(year) for year in fitted.columns)
        realised_years = set(int(year) for year in realised.columns)
        if NORMALIZATION_YEAR not in fitted_years or fitted_years & realised_years:
            raise IngestError(
                f"{pair}: fitted gaps must include 2013 and not overlap realised years."
            )
        if any(year >= FIRST_TREATMENT_YEAR for year in fitted_years) or any(
            year < FIRST_TREATMENT_YEAR for year in realised_years
        ):
            raise IngestError(f"{pair}: fitted/realised calendar roles are mislabeled.")
        fitted_aligned = fitted.copy(deep=True)
        fitted_aligned.index = fitted_aligned.index.astype(str)
        realised_aligned = realised.copy(deep=True)
        realised_aligned.index = realised_aligned.index.astype(str)
        fitted_aligned = fitted_aligned.reindex(realised_aligned.index)
        combined = pd.concat([fitted_aligned, realised_aligned], axis=1).sort_index(axis=1)
        path = compute_coalition_path(combined, pair[1])
        roster = tuple(str(code) for code in combined.index)
        equal_weights = {code: 1.0 / len(roster) for code in roster}
        content_digest = _content_digest(path.loc[:, ["coalition_mean"]])
        roster_digest = _content_digest(roster)
        weight_digest = _content_digest(equal_weights)
        upstream_result_sha256 = hashlib.sha256(
            (
                _content_digest(fitted_aligned)
                + "|"
                + _content_digest(realised_aligned)
            ).encode("utf-8")
        ).hexdigest()
        lineage = hashlib.sha256(
            json.dumps(
                {
                    "design_id": "baseline",
                    "path_content_sha256": content_digest,
                    "effective_roster_sha256": roster_digest,
                    "effective_weights_sha256": weight_digest,
                    "provenance_origin": "verified_fitted_and_realised_baseline_matrices",
                    "upstream_result_sha256": upstream_result_sha256,
                },
                sort_keys=True,
                separators=(",", ":"),
            ).encode("utf-8")
        ).hexdigest()
        path.attrs.update(
            {
                "design_id": "baseline",
                "effective_roster": roster,
                "effective_weights": equal_weights,
                "path_content_sha256": content_digest,
                "effective_roster_sha256": roster_digest,
                "effective_weights_sha256": weight_digest,
                "fit_lineage_sha256": lineage,
                "provenance_origin": "verified_fitted_and_realised_baseline_matrices",
                "upstream_result_sha256": upstream_result_sha256,
                "fitted_gap_sha256": _content_digest(fitted_aligned),
                "realised_gap_sha256": _content_digest(realised_aligned),
                "provenance_status": "verified_from_distinct_fitted_and_realised_matrices",
            }
        )
        dynamic_paths[pair] = path
        figure_data[f"figure_2_{pair[0]}_{pair[1]}"] = path

    geometry_provenance = {
        "source_id": str(country_geometry.attrs.get("source_id", "")),
        "version_id": str(country_geometry.attrs.get("version_id", "")),
    }
    for raw_pair in sorted(incremental_contrasts.keys()):
        pair = (str(raw_pair[0]), str(raw_pair[1]))
        # National maps here implement the sanctions log-level comparisons.
        if pair[0] != "strict_sanctions" or pair[1] not in {"log_CPI", "log_real_GDPpc"}:
            continue
        contrasts = incremental_contrasts[raw_pair]
        if not isinstance(contrasts, pd.Series) or contrasts.empty or not bool(
            contrasts.index.is_unique
        ):
            raise IngestError(f"{pair}: country contrasts are empty or duplicate-keyed.")
        figure_data[f"figure_3_{pair[0]}_{pair[1]}"] = build_choropleth_data(
            {str(code): float(value) for code, value in contrasts.items()},
            [str(code) for code in contrasts.index],
            country_geometry=country_geometry, geographic_scope="Europe",
        )
        figure_data[f"figure_B1_{pair[0]}_{pair[1]}"] = build_choropleth_data(
            {str(code): float(value) for code, value in contrasts.items()},
            [str(code) for code in contrasts.index],
            country_geometry=world_geometry, geographic_scope="world",
        )

    if matched_paths_by_pair is None:
        raise IngestError("Figure 4 requires matched paths for both level outcomes.")
    figure_data["figure_4_overlay"] = build_two_panel_overlay_data(
        dynamic_paths, matched_paths_by_pair, resolved_overlay_pair[0]
    )
    for identity in sorted(placebo_draws.keys(), key=str):
        observed = observed_preferred_att[identity]
        figure_name = re.sub(r"[^A-Za-z0-9_.-]+", "_", str(identity)).strip("_")
        if not figure_name:
            raise IngestError("A Figure B2 identity cannot be rendered safely.")
        key = f"figure_B2_{figure_name}"
        if key in figure_data:
            raise IngestError("Figure B2 identities collide after canonical labelling.")
        histogram = build_placebo_histogram_data(
            np.asarray(placebo_draws[identity], dtype=float), float(observed)
        )
        histogram.attrs["source_identity"] = repr(identity)
        figure_data[key] = histogram
    figure_data["figure_B3_reference_windows"] = build_reference_window_data(
        reference_window_estimates
    )

    tables, renderings = build_table_bundle(table_sources)
    gaps = register_reproducibility_gaps(
        extra_gaps, geometry_provenance=geometry_provenance
    )
    if BASEMAP_GAP in gaps.index and str(gaps.loc[BASEMAP_GAP, "state"]) == "unresolved":
        raise IngestError("Production cannot publish Figure 3 with unresolved geometry.")
    outcomes = render_figures(figure_data, output_dir)
    fingerprint = freeze_production(figure_data, tables, gaps)
    # Accumulate the recorded notes.
    notes: List[str] = [
        "the dynamic path is normalized on the declared reference year and "
        "carries a region label separating the fitted pre-treatment years from "
        "the realised ones, so the figure cannot be read as an event study.",
        "the post-window height of the normalized path is not the estimand "
        "tables' effect, because the path is re-referenced and the tables are "
        "not.",
        "the ribbon is pointwise per year on cross-country dispersion, not "
        "event-time uncertainty.",
        "the choropleth's colour scale is not capped: a value near or beyond an "
        "endpoint uses an extended colour scale rather than silent clipping; "
        "real country geometry and its source/version are frozen with the data.",
        "the baseline path retains its band while the matched designs are lines "
        "alone, because a band on a weighted design would overstate its "
        "precision.",
        "Figure B2 marks the actual observed preferred incremental ATT in red "
        "and zero with a dotted line; the placebo mean is never substituted.",
        "every Tables 3–10, A1–A8 and B1–B7 cell is reconciled to an independent "
        "versioned inventory as either a finite sourced value or a concrete "
        "unresolved gap before formatting.",
        "the inventory's completeness certification is an externally verified "
        "input, not a self-authenticating DataFrame attribute: production rejects "
        "a missing external verification reference and records that dependency.",
    ]
    # Record the rendering outcome.
    notes.append(
        f"figures written: {int(outcomes['written'].sum())} of {len(outcomes)}."
    )
    # Assemble the immutable bundle.
    bundle: FigureTableBundle = FigureTableBundle(
        figure_data=figure_data,
        tables=tables,
        latex=renderings,
        rendered=outcomes,
        reproducibility_gaps=gaps,
        fingerprint_sha256=fingerprint,
        built_utc=current_utc_iso(),
        notes=tuple(notes),
    )
    # Log the completed production.
    logger.info(
        "production complete: %d figure(s), %d table(s), %d gap(s), fingerprint "
        "%s.",
        len(figure_data),
        len(tables),
        len(gaps),
        fingerprint[:16],
    )
    # Return the bundle.
    return bundle
````

````python
# Task 32: Reconciliation and Verification Audit

# ==============================================================================
# Task 32: Reconciliation and Verification Audit
# ==============================================================================

# Declare the tolerance of a point estimate and of a standard error.
ESTIMATE_TOLERANCE: Final[float] = 1.0e-3
# Declare the key columns of the comparison.
COMPARISON_KEY: Final[Tuple[str, ...]] = (
    "table_number",
    "row_identifier",
    "column_identifier",
)
# Declare the localization layer of the data.
LAYER_DATA: Final[str] = "data"
# Declare the localization layer of the transformation.
LAYER_TRANSFORMATION: Final[str] = "transformation"
# Declare the localization layer of the model.
LAYER_MODEL: Final[str] = "model"
# Compose the declared layer order, from the earliest to the latest.
LAYER_ORDER: Final[Tuple[str, str, str]] = (
    LAYER_DATA,
    LAYER_TRANSFORMATION,
    LAYER_MODEL,
)
# Declare the verdict of a reconciled value.
VERDICT_RECONCILED: Final[str] = "reconciled"
# Declare the verdict of a divergent value.
VERDICT_DIVERGENT: Final[str] = "divergent"
# Declare the verdict of a value registered as a gap.
VERDICT_GAP: Final[str] = "registered_gap"
# Declare the identifier of the counterfactual invariant.
INVARIANT_COUNTERFACTUAL: Final[str] = "A_no_realized_target_entered_its_counterfactual"
# Declare the identifier of the treatment-leakage invariant.
INVARIANT_LEAKAGE: Final[str] = "B_no_treatment_information_entered_eligibility_or_imputation"
# Declare the identifier of the baseline-preservation invariant.
INVARIANT_BASELINE: Final[str] = "C_no_robustness_branch_replaced_the_baseline"
# Compose the declared invariant identifiers.
INVARIANT_IDS: Final[Tuple[str, str, str]] = (
    INVARIANT_COUNTERFACTUAL,
    INVARIANT_LEAKAGE,
    INVARIANT_BASELINE,
)
# Compose the declared archive artefacts the manifest must cover.
ARCHIVE_ARTEFACTS: Final[Tuple[str, ...]] = (
    "raw_extractions_with_vintage_tags",
    "eligibility_mask",
    "completed_panel_with_method_tags",
    "treatment_coding_ledger",
    "factor_proxy_constructions",
    "pooled_coefficient_and_loadings",
    "gap_matrices",
    "inference_outputs",
    "robustness_runs",
    "figures_and_tables_machine_readable",
)
# Bind each mandatory archive identity to one exact live pipeline artefact,
# archive section and scientific role.  Caller-selected aliases cannot map ten
# manifest rows to the same convenient object.
ARCHIVE_LIVE_BINDINGS: Final[Mapping[str, Tuple[str, str, str]]] = MappingProxyType(
    {
        "raw_extractions_with_vintage_tags": ("ingest_bundle", "raw", "authoritative_raw_inputs"),
        "eligibility_mask": ("eligibility", "interim", "precompletion_eligibility"),
        "completed_panel_with_method_tags": ("analysis_panel", "processed", "completed_analysis_panel"),
        "treatment_coding_ledger": ("treatment", "interim", "policy_timing_ledger"),
        "factor_proxy_constructions": ("donor_factor", "processed", "donor_factor_objects"),
        "pooled_coefficient_and_loadings": ("pooled_fit", "estimates", "fitted_model_objects"),
        "gap_matrices": ("gap_att", "estimates", "country_year_gap_objects"),
        "inference_outputs": ("chain_inference", "estimates", "primary_inference_outputs"),
        "robustness_runs": ("robustness_sweeps", "audit", "robustness_execution_audit"),
        "figures_and_tables_machine_readable": ("production", "tables", "publication_bundle"),
    }
)
# Declare the independently controlled per-artifact vintage authority.  The
# distributed source intentionally carries no empirical registrations; a
# production deployment must install authenticated records out of band before
# any archive declaration or file can be certified.  Caller mappings, README
# text and the archive release version never populate this registry.
TRUSTED_ARCHIVE_VINTAGE_REGISTRY: Mapping[str, Mapping[str, str]] = MappingProxyType({})
# Declare the externally sourced facts that the replication archive must bind.
REQUIRED_PROVENANCE_FACTS: Final[Tuple[str, ...]] = (
    "wdi_vintage",
    "comtrade_hs_codes",
    "random_seeds",
    "wgi_release_year",
    "b10_b13_reconstruction",
    "region_block_definition",
)
# Declare the admissible comparison types.
COMPARISON_TYPES: Final[Tuple[str, ...]] = (
    "integer_count",
    "point_estimate",
    "standard_error",
)
# Declare the two report states.  A report with any scientific blocker is never
# called complete merely because its structural audit ran.
RECONCILIATION_COMPLETE: Final[str] = "complete"
RECONCILIATION_INCOMPLETE: Final[str] = "incomplete"
# Declare the attribute by which a scientific callable supplies an executable
# baseline-versus-perturbation witness to the invariant audit.
RECONCILIATION_PROBE_ATTRIBUTE: Final[str] = "__reconciliation_probe__"
# Declare the independently pinned manuscript transcription registry.  The
# distributed module deliberately carries no study-specific pin: production
# must install a separately controlled (authority, version) digest before a
# 418-cell transcription can be certified.
TRUSTED_MANUSCRIPT_UNIVERSE_PINS: Mapping[Tuple[str, str], str] = MappingProxyType({})
# Declare the production boundaries that may establish design invariants.  An
# arbitrary callable with a copied probe attribute is never evidence.
RECONCILIATION_BOUNDARY_ROLES: Final[Mapping[str, Tuple[str, ...]]] = (
    MappingProxyType(
        {
            "orchestrate_counterfactual_projection": ("target_post_y", "target_post_x"),
            "orchestrate_eligibility_screening": ("treatment",),
            "orchestrate_control_completion": ("structural_treatment_exclusion",),
            "orchestrate_outcome_repair": ("structural_treatment_exclusion",),
            "build_cell_ledger": ("structural_treatment_exclusion",),
        }
    )
)
# Snapshot exact public library class objects at module import.  Several
# pandas aliases deliberately rewrite ``__module__`` and do not expose
# recoverable class source on Python 3.14; object identity captured before any
# runtime probe is the independent provenance witness for those aliases.
TRUSTED_RECONCILIATION_LIBRARY_CLASS_OBJECTS: Final[Tuple[type, ...]] = tuple(
    value
    for module in (
        copy, hashlib, importlib, itertools, json, logging, math, os, re, sys,
        tempfile, matplotlib, plt, np, pd, scipy_stats,
    )
    for value in vars(module).values()
    if inspect.isclass(value)
)
# Snapshot native public callables for trusted top-level modules.  This covers
# legitimate C-extension re-exports such as ``hashlib.sha256`` while ensuring
# a native callable added later at runtime (including a reflection alias) is
# not silently grandfathered.
TRUSTED_RECONCILIATION_NATIVE_CALLABLE_OBJECTS: Final[Tuple[Any, ...]] = tuple(
    value
    for module in (
        copy, hashlib, importlib, itertools, json, logging, math, os, re, sys,
        tempfile, matplotlib, plt, np, pd, scipy_stats,
    )
    for value in vars(module).values()
    if inspect.isbuiltin(value) or isinstance(value, np.ufunc)
)
# Caller-frame evaluators can read names that never appear in their caller's
# static dependency graph.  Capture the exact installed objects independently
# and reject both direct aliases and module dispatch through them.
TRUSTED_RECONCILIATION_CALLER_NAMESPACE_CALLABLES: Final[Tuple[Any, ...]] = tuple(
    candidate
    for candidate in (
        getattr(pd, "eval", None),
        getattr(pd.DataFrame, "eval", None),
        getattr(pd.DataFrame, "query", None),
    )
    if callable(candidate)
)
# Pin executable invariant scenarios separately from caller data and mutable
# function attributes.  Keys are boundary, forbidden role, scenario id and
# scenario version; values digest the complete frozen stage-input witness.
TRUSTED_RECONCILIATION_SCENARIO_PINS: Mapping[Tuple[str, str, str, str], str] = MappingProxyType({})
# Declare the exact immutable runtime classes whose C-level implementations
# cannot be recursively inspected.  These are construction/time-parsing
# infrastructure used by the four registered boundaries, not scientific data
# containers or caller-extensible dispatch surfaces.
TRUSTED_RECONCILIATION_OPAQUE_CLASSES: Final[Tuple[str, ...]] = (
    "datetime.datetime",
    "datetime.timezone",
    "decimal.Decimal",
)
# Declare the scientific fields compared at each registered boundary.  Volatile
# run timestamps are intentionally absent; no caller-provided selector can
# narrow this schema.
RECONCILIATION_BOUNDARY_OUTPUT_FIELDS: Final[Mapping[str, Tuple[str, ...]]] = (
    MappingProxyType(
        {
            "orchestrate_counterfactual_projection": (
                "loadings", "covariate_paths", "outcome_paths", "fitted_pre_paths",
                "range_summary",
            ),
            "orchestrate_eligibility_screening": (
                "mask", "exclusion_ledger", "missingness", "rule_summary",
                "fingerprint_sha256", "notes",
            ),
            "orchestrate_control_completion": (
                "tensor", "ledger", "method_summary", "fingerprint_sha256", "notes",
            ),
            "orchestrate_outcome_repair": (
                "tensor", "ledger", "repair_summary", "fingerprint_sha256", "notes",
            ),
            "build_cell_ledger": (
                "ledger", "counts", "reclassifications", "exclusions",
                "fingerprint_sha256", "notes",
            ),
        }
    )
)
# Declare the complete independently transcribed manuscript-cell count.
MANUSCRIPT_CELL_COUNT: Final[int] = 418


# Declare the frozen result of the reconciliation audit.
@dataclass(frozen=True)
class ReconciliationReport:
    """
    Frozen result of the reconciliation and verification audit.

    Attributes
    ----------
    comparisons : pandas.DataFrame
        One row per reproducible manuscript value carrying the reported and
        reproduced values, units, procedures, precision, differences, earliest
        matched divergent layer and verdict; explicit unresolved values live
        in ``gaps`` and the two key sets are an exact disjoint universe split.
    gaps : pandas.DataFrame
        One row per unreproducible value carrying its blocking input.
    localization : pandas.DataFrame
        One row per divergent value carrying the layer walk from the earliest
        to the latest.
    invariants : pandas.DataFrame
        One row per design invariant carrying its verdict and the demonstration
        that establishes it.
    provenance : pandas.DataFrame
        One row per declared input carrying its value and its source.
    manifest : pandas.DataFrame
        One row per archive artefact carrying its actual file/object type,
        location or object identifier, version/vintage, role, content hash and
        observed verification state.
    fingerprint_sha256 : str
        Digest of the canonical audit payload.
    completed_utc : str
        ISO-8601 UTC timestamp of the audit.
    notes : tuple of str
        Recorded decisions, including the attribution rule and the invariant
        readings.
    status : str
        ``complete`` only when every manuscript cell, invariant, provenance
        fact and archive artefact is verified; otherwise ``incomplete``.
    blockers : pandas.DataFrame
        Explicit reasons that prevent a complete certification.
    coverage : pandas.DataFrame
        The independent-universe attribution counts used by the certificate.
    """
    # The comparison frame.
    comparisons: pd.DataFrame
    # The reconciliation gaps.
    gaps: pd.DataFrame
    # The localization walk.
    localization: pd.DataFrame
    # The invariant table.
    invariants: pd.DataFrame
    # The provenance facts.
    provenance: pd.DataFrame
    # The archive manifest.
    manifest: pd.DataFrame
    # Digest of the canonical audit payload.
    fingerprint_sha256: str
    # ISO-8601 UTC timestamp of the audit.
    completed_utc: str
    # Recorded decisions of the audit.
    notes: Tuple[str, ...]
    # Fail-closed completion state.
    status: str = RECONCILIATION_INCOMPLETE
    # Explicit completion blockers.
    blockers: pd.DataFrame = field(default_factory=pd.DataFrame)
    # Independent-universe coverage audit.
    coverage: pd.DataFrame = field(default_factory=pd.DataFrame)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 32, Step 1: The cell-by-cell reconciliation
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the comparison table from the reported and reproduced values.
def compose_comparison_table(
    reported: pd.DataFrame,
    reproduced: pd.DataFrame,
    layer_chains: Optional[Mapping[Tuple[str, str, str], Mapping[str, Any]]] = None,
    tolerance: float = ESTIMATE_TOLERANCE,
    required_universe: Optional[pd.DataFrame] = None,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Compose the comparison table from the reported and reproduced values.

    Parameters
    ----------
    reported : pandas.DataFrame
        The independently inventoried reported values.  In addition to the
        comparison key and ``value``, every row carries ``comparison_type``,
        ``unit``, ``display_precision`` and ``source_id``.
    reproduced : pandas.DataFrame
        The reproduced values under the same key.
    layer_chains : mapping, optional
        Matched expected-versus-observed checkpoints at the data,
        transformation and model layers.  Each checkpoint states its own
        quantity, unit, key, comparison rule and precision; an intermediate
        quantity is never compared with the final ATT.
    tolerance : float, optional
        Numerical allowance for point estimates and standard errors.  Integer
        counts remain exact and publication rounding is additionally honoured.
    required_universe : pandas.DataFrame, optional
        Independent cell inventory for this comparison subset.  When supplied
        its keys and scientific metadata must agree exactly with ``reported``.

    Returns
    -------
    tuple
        The comparison frame and the localization frame.

    Raises
    ------
    IngestError
        If a required column is absent, or if a reported key is absent from the
        reproduced values, because such a key is a reconciliation gap rather
        than a comparison.
    """
    # Validate the numerical allowance before any comparison.
    if isinstance(tolerance, (bool, np.bool_)) or not np.isfinite(float(tolerance)) or float(tolerance) < 0.0:
        raise IngestError("The comparison tolerance must be finite and nonnegative.")
    # Refuse a reported frame lacking scientific identity metadata.
    required_reported = list(COMPARISON_KEY) + [
        "value", "comparison_type", "unit", "display_precision", "source_id",
        "procedure_id",
    ]
    for column in required_reported:
        # Test the column's presence.
        if column not in reported.columns:
            # The failure names the absent column.
            raise IngestError(f"The reported frame lacks the column {column!r}.")
    # Refuse a reproduced frame lacking its labelled value and unit.
    for column in list(COMPARISON_KEY) + ["value", "unit", "source_id", "procedure_id"]:
        if column not in reproduced.columns:
            raise IngestError(f"The reproduced frame lacks the column {column!r}.")
    # Refuse duplicate keys before indexing can silently multiply or collapse
    # cells.
    if bool(reported.duplicated(list(COMPARISON_KEY), keep=False).any()):
        raise IngestError("The reported comparison frame carries duplicate cell keys.")
    if bool(reproduced.duplicated(list(COMPARISON_KEY), keep=False).any()):
        raise IngestError("The reproduced comparison frame carries duplicate cell keys.")
    # Normalize key labels consistently before set and indexed operations.
    reported = reported.copy()
    reproduced = reproduced.copy()
    for column in COMPARISON_KEY:
        reported[column] = reported[column].astype(str)
        reproduced[column] = reproduced[column].astype(str)
    # Cross-check this subset against its independent inventory when supplied.
    if required_universe is not None:
        _assert_comparison_subset_matches_universe(reported, required_universe)
    # Compose the reported index.
    reported_indexed = reported.set_index(list(COMPARISON_KEY))
    # Compose the reproduced index.
    reproduced_indexed = reproduced.set_index(list(COMPARISON_KEY))
    # Identify the reported keys without a reproduced counterpart.
    reported_keys = set(tuple(str(value) for value in key) for key in reported_indexed.index)
    reproduced_keys = set(tuple(str(value) for value in key) for key in reproduced_indexed.index)
    missing = sorted(reported_keys - reproduced_keys)
    extras = sorted(reproduced_keys - reported_keys)
    # Refuse a missing key, because it belongs in the gaps frame.
    if missing:
        # The failure names the missing keys.
        raise IngestError(
            f"{len(missing)} reported key(s) carry no reproduced value, for "
            f"example {missing[:3]}; register them as reconciliation gaps."
        )
    if extras:
        raise IngestError(
            f"The reproduced frame carries {len(extras)} key(s) outside the "
            f"declared comparison subset, for example {extras[:3]}."
        )
    if len(reported_indexed) == 0:
        empty_columns = list(COMPARISON_KEY) + [
            "reported_value", "reproduced_value", "absolute_difference",
            "relative_difference", "comparison_type", "unit",
            "display_precision", "allowance", "reported_source_id",
            "reproduced_source_id", "reported_procedure_id",
            "reproduced_procedure_id", "earliest_divergent_layer", "verdict",
        ]
        comparisons = pd.DataFrame(columns=empty_columns).set_index(list(COMPARISON_KEY))
        comparisons.attrs["cell_universe_count"] = 0
        comparisons.attrs["cell_universe_sha256"] = _content_digest(
            reported[list(COMPARISON_KEY)].copy()
        )
        comparisons.attrs["universe_binding"] = (
            "independent_inventory_subset_verified"
            if required_universe is not None
            else "caller_supplied_reported_subset_only"
        )
        localization = pd.DataFrame(
            columns=list(COMPARISON_KEY) + [
                "layer", "quantity_id", "unit", "checkpoint_key",
                "expected_value", "observed_value", "absolute_difference",
                "comparison_rule", "display_precision", "allowance", "matches",
                "is_earliest_divergence",
            ]
        ).set_index(list(COMPARISON_KEY) + ["layer"])
        return comparisons, localization
    # Accumulate the comparison rows.
    rows: List[dict] = []
    # Accumulate the localization rows.
    localization_rows: List[dict] = []
    # Iterate over the reported keys.
    for key, row in reported_indexed.iterrows():
        # Normalize the string key after pandas indexing.
        key = tuple(str(value) for value in key)
        # Read and validate the reported and reproduced values.
        reported_value = float(row["value"])
        # Read the reproduced value.
        reproduced_value = float(reproduced_indexed.loc[key, "value"])
        if not np.isfinite(reported_value) or not np.isfinite(reproduced_value):
            raise IngestError(f"Comparison cell {key} carries a nonfinite value.")
        comparison_type = str(row["comparison_type"]).strip()
        if comparison_type not in COMPARISON_TYPES:
            raise IngestError(
                f"Comparison cell {key} has unsupported type {comparison_type!r}."
            )
        unit = str(row["unit"]).strip()
        reproduced_unit = str(reproduced_indexed.loc[key, "unit"]).strip()
        if not unit or reproduced_unit != unit:
            raise IngestError(
                f"Comparison cell {key} has mismatched or blank units "
                f"({unit!r} versus {reproduced_unit!r})."
            )
        source_id = str(row["source_id"]).strip()
        reproduced_source_id = str(reproduced_indexed.loc[key, "source_id"]).strip()
        procedure_id = str(row["procedure_id"]).strip()
        reproduced_procedure_id = str(reproduced_indexed.loc[key, "procedure_id"]).strip()
        if not source_id or not reproduced_source_id or not procedure_id or not reproduced_procedure_id:
            raise IngestError(f"Comparison cell {key} lacks source/procedure provenance.")
        if procedure_id != reproduced_procedure_id:
            raise IngestError(
                f"Comparison cell {key} uses different scientific procedures "
                f"({procedure_id!r} versus {reproduced_procedure_id!r})."
            )
        precision_raw = row["display_precision"]
        if (
            isinstance(precision_raw, (bool, np.bool_))
            or not np.isfinite(float(precision_raw))
            or float(precision_raw) != int(float(precision_raw))
            or not 0 <= int(float(precision_raw)) <= 12
        ):
            raise IngestError(f"Comparison cell {key} has invalid displayed precision.")
        display_precision = int(float(precision_raw))
        if comparison_type == "integer_count" and (
            reported_value != int(reported_value)
            or reproduced_value != int(reproduced_value)
        ):
            raise IngestError(
                f"Integer-count cell {key} carries a non-integral operand."
            )
        if comparison_type == "standard_error" and (
            reported_value < 0.0 or reproduced_value < 0.0
        ):
            raise IngestError(f"Standard-error cell {key} carries a negative value.")
        # Compose the absolute difference.
        absolute = abs(reproduced_value - reported_value)
        # Compose the relative difference, which is undefined at a zero
        # reported magnitude.
        relative = (
            absolute / abs(reported_value) if reported_value != 0.0 else float("nan")
        )
        # Counts are exact.  Estimates and standard errors respect both the
        # declared numerical allowance and the reported rounding interval.
        allowance = (
            0.0
            if comparison_type == "integer_count"
            else max(float(tolerance), 0.5 * (10.0 ** (-display_precision)))
        )
        reconciled = bool(absolute == 0.0) if comparison_type == "integer_count" else bool(absolute <= allowance + np.finfo(float).eps)
        # Compose the localized layer of a divergence.
        layer: str = ""
        # Localize only a divergent value.
        if not reconciled:
            # Read the value's chain.
            chain = (layer_chains or {}).get(key)
            # Refuse a divergence without a chain, because it could not be
            # localized and the audit would then be incomplete.
            if chain is None:
                # The failure names the key.
                raise IngestError(
                    f"The divergent key {key} carries no layer chain, so the "
                    f"divergence cannot be localized."
                )
            missing_layers = [name for name in LAYER_ORDER if name not in chain]
            if missing_layers:
                raise IngestError(
                    f"The divergent key {key} lacks ordered checkpoint layer(s) "
                    f"{missing_layers}."
                )
            final_checkpoint = _validate_layer_checkpoint(
                LAYER_MODEL, chain[LAYER_MODEL], key, tolerance
            )
            if (
                str(final_checkpoint["unit"]) != unit
                or float(final_checkpoint["expected_value"]) != reported_value
                or float(final_checkpoint["observed_value"]) != reproduced_value
            ):
                raise IngestError(
                    f"The model checkpoint for {key} is not the same labelled "
                    "reported-versus-reproduced manuscript quantity."
                )
            # Walk the chain from the earliest layer and record the first
            # divergence, which is the layer that must be repaired.
            layer = _earliest_divergent_layer(chain, reported_value, tolerance)
            # Record every like-for-like checkpoint so the claimed earliest
            # layer is independently auditable.
            for name in LAYER_ORDER:
                if name not in chain:
                    continue
                checkpoint = _validate_layer_checkpoint(name, chain[name], key, tolerance)
                localization_rows.append(
                    {
                        "table_number": key[0],
                        "row_identifier": key[1],
                        "column_identifier": key[2],
                        "layer": name,
                        **checkpoint,
                        "is_earliest_divergence": bool(name == layer),
                    }
                )
        # Append the comparison row.
        rows.append(
            {
                "table_number": key[0],
                "row_identifier": key[1],
                "column_identifier": key[2],
                "reported_value": reported_value,
                "reproduced_value": reproduced_value,
                "absolute_difference": absolute,
                "relative_difference": relative,
                "comparison_type": comparison_type,
                "unit": unit,
                "display_precision": display_precision,
                "allowance": allowance,
                "reported_source_id": source_id,
                "reproduced_source_id": reproduced_source_id,
                "reported_procedure_id": procedure_id,
                "reproduced_procedure_id": reproduced_procedure_id,
                "earliest_divergent_layer": layer,
                "verdict": VERDICT_RECONCILED if reconciled else VERDICT_DIVERGENT,
            }
        )
    # Assemble the comparison frame.
    comparisons: pd.DataFrame = pd.DataFrame(rows).set_index(list(COMPARISON_KEY))
    comparisons.attrs["cell_universe_count"] = int(len(reported))
    comparisons.attrs["cell_universe_sha256"] = _content_digest(
        reported[list(COMPARISON_KEY)].sort_values(list(COMPARISON_KEY)).reset_index(drop=True)
    )
    comparisons.attrs["universe_binding"] = (
        "independent_inventory_subset_verified"
        if required_universe is not None
        else "caller_supplied_reported_subset_only"
    )
    # Assemble the localization frame.
    localization: pd.DataFrame = pd.DataFrame(
        localization_rows,
        columns=list(COMPARISON_KEY) + [
            "layer", "quantity_id", "unit", "checkpoint_key", "expected_value",
            "observed_value", "absolute_difference", "comparison_rule",
            "display_precision", "allowance", "matches",
            "is_earliest_divergence",
        ],
    )
    # Key the localization frame when it carries rows.
    if len(localization):
        # Set the declared key.
        localization = localization.set_index(list(COMPARISON_KEY) + ["layer"])
    # Log the comparison's extent.
    logger.info(
        "comparison composed across %d cell(s) with %d divergence(s).",
        len(comparisons),
        int((comparisons["verdict"] == VERDICT_DIVERGENT).sum()),
    )
    # Return the frames.
    return comparisons, localization


# Identify the earliest layer whose value diverges from the reported one.
def _earliest_divergent_layer(
    chain: Mapping[str, Any],
    reported_value: float,
    tolerance: float,
) -> str:
    """
    Identify the earliest layer whose value diverges from the reported one.

    Parameters
    ----------
    chain : mapping
        Matched checkpoint records.  Each declared layer carries the expected
        and observed values of one identically labelled quantity.
    reported_value : float
        The reported value.
    tolerance : float
        The tolerance of the comparison.

    Returns
    -------
    str
        The earliest layer whose observed value fails its own checkpoint rule.

    Raises
    ------
    IngestError
        If the chain carries no declared layer, because the walk would then be
        vacuous.
    """
    # Refuse an incomplete ordered chain: omission cannot be localized to the
    # first layer that happened to be supplied.
    missing_layers = [name for name in LAYER_ORDER if name not in chain]
    if missing_layers:
        raise IngestError(f"The layer chain lacks ordered layer(s) {missing_layers}.")
    # Walk the layers from the earliest.  A scalar checkpoint is explicitly
    # refused: comparing an input or intermediate scalar with a final ATT is
    # not a localization test.
    key_hint: Optional[Tuple[str, str, str]] = None
    for name in LAYER_ORDER:
        # Skip a layer the chain does not carry.
        if name not in chain:
            # Continue to the next layer.
            continue
        raw = chain[name]
        if not isinstance(raw, Mapping):
            raise IngestError(
                f"Layer {name!r} must carry a matched expected/observed checkpoint, "
                "not a scalar compared with the final reported effect."
            )
        raw_key = raw.get("comparison_key")
        if raw_key is None:
            raise IngestError(f"Layer {name!r} lacks its comparison_key.")
        normalized_key = tuple(str(value) for value in raw_key)
        if len(normalized_key) != len(COMPARISON_KEY):
            raise IngestError(f"Layer {name!r} has a malformed comparison key.")
        if key_hint is None:
            key_hint = normalized_key
        elif normalized_key != key_hint:
            raise IngestError("Layer checkpoints do not carry the same manuscript key.")
        checkpoint = _validate_layer_checkpoint(name, raw, normalized_key, tolerance)
        if name == LAYER_MODEL and float(checkpoint["expected_value"]) != float(reported_value):
            raise IngestError(
                "The model checkpoint expected value is not the reported manuscript value."
            )
        if not bool(checkpoint["matches"]):
            # Return the divergent layer.
            return str(name)
    # A divergent final cell with no divergent matched checkpoint is not
    # localized and must not be assigned to the last layer by convention.
    raise IngestError(
        "The final manuscript cell diverges, but every recorded like-for-like "
        "checkpoint agrees; the layer chain is incomplete."
    )


# Validate and normalize one like-for-like checkpoint.
def _validate_layer_checkpoint(
    layer: str,
    checkpoint: Any,
    comparison_key: Tuple[str, str, str],
    default_tolerance: float,
) -> Dict[str, Any]:
    """Validate one expected-versus-observed checkpoint of the same quantity."""
    if not isinstance(checkpoint, Mapping):
        raise IngestError(f"Layer {layer!r} does not carry a checkpoint mapping.")
    required = (
        "quantity_id", "unit", "comparison_key", "expected", "observed",
        "comparison_type", "display_precision",
    )
    missing = [name for name in required if name not in checkpoint]
    if missing:
        raise IngestError(f"Layer {layer!r} lacks checkpoint field(s) {missing}.")
    checkpoint_key = tuple(str(value) for value in checkpoint["comparison_key"])
    if checkpoint_key != tuple(str(value) for value in comparison_key):
        raise IngestError(
            f"Layer {layer!r} checkpoint key {checkpoint_key} does not match "
            f"the manuscript cell {comparison_key}."
        )
    quantity_id = str(checkpoint["quantity_id"]).strip()
    unit = str(checkpoint["unit"]).strip()
    if not quantity_id or not unit:
        raise IngestError(f"Layer {layer!r} has a blank quantity or unit.")
    expected = float(checkpoint["expected"])
    observed = float(checkpoint["observed"])
    if not np.isfinite(expected) or not np.isfinite(observed):
        raise IngestError(f"Layer {layer!r} carries a nonfinite checkpoint value.")
    comparison_type = str(checkpoint["comparison_type"])
    if comparison_type not in COMPARISON_TYPES:
        raise IngestError(f"Layer {layer!r} has an invalid comparison type.")
    precision_raw = checkpoint["display_precision"]
    if (
        isinstance(precision_raw, (bool, np.bool_))
        or not np.isfinite(float(precision_raw))
        or int(float(precision_raw)) != float(precision_raw)
        or not 0 <= int(float(precision_raw)) <= 12
    ):
        raise IngestError(f"Layer {layer!r} has invalid displayed precision.")
    precision = int(float(precision_raw))
    if comparison_type == "integer_count":
        if expected != int(expected) or observed != int(observed):
            raise IngestError(
                f"Layer {layer!r} integer-count checkpoint has a non-integral operand."
            )
        allowance = 0.0
        rule = "exact_integer"
    else:
        if comparison_type == "standard_error" and (expected < 0.0 or observed < 0.0):
            raise IngestError(
                f"Layer {layer!r} standard-error checkpoint has a negative operand."
            )
        raw_allowance = checkpoint.get("allowance", default_tolerance)
        if isinstance(raw_allowance, (bool, np.bool_)) or not np.isfinite(float(raw_allowance)) or float(raw_allowance) < 0.0:
            raise IngestError(f"Layer {layer!r} has invalid numerical allowance.")
        allowance = max(float(raw_allowance), 0.5 * 10.0 ** (-precision))
        rule = "precision_interval_plus_declared_allowance"
    difference = abs(observed - expected)
    matches = difference == 0.0 if comparison_type == "integer_count" else difference <= allowance + np.finfo(float).eps
    return {
        "quantity_id": quantity_id,
        "unit": unit,
        "checkpoint_key": json.dumps(checkpoint_key),
        "expected_value": expected,
        "observed_value": observed,
        "absolute_difference": difference,
        "comparison_rule": rule,
        "display_precision": precision,
        "allowance": allowance,
        "matches": bool(matches),
    }


# Cross-check a comparison subset with its independent manuscript inventory.
def _assert_comparison_subset_matches_universe(
    reported: pd.DataFrame,
    universe: pd.DataFrame,
) -> None:
    """Require exact cell and metadata identity against an independent subset."""
    universe_required = list(COMPARISON_KEY) + [
        "reported_value", "comparison_type", "unit", "display_precision",
        "reported_source_id", "procedure_id",
    ]
    missing = [name for name in universe_required if name not in universe.columns]
    if missing:
        raise IngestError(f"The manuscript inventory lacks field(s) {missing}.")
    if bool(universe.duplicated(list(COMPARISON_KEY), keep=False).any()):
        raise IngestError("The manuscript inventory carries duplicate cell keys.")
    u = universe.set_index(list(COMPARISON_KEY)).sort_index()
    r = reported.set_index(list(COMPARISON_KEY)).sort_index()
    if set(u.index) != set(r.index):
        raise IngestError("The comparison subset does not equal its independent inventory.")
    for key in u.index:
        if not np.isclose(float(r.loc[key, "value"]), float(u.loc[key, "reported_value"]), rtol=0.0, atol=0.0):
            raise IngestError(f"Reported value for {key} contradicts the inventory.")
        metadata_pairs = (
            ("comparison_type", "comparison_type"),
            ("unit", "unit"),
            ("display_precision", "display_precision"),
            ("source_id", "reported_source_id"),
            ("procedure_id", "procedure_id"),
        )
        for reported_name, universe_name in metadata_pairs:
            if str(r.loc[key, reported_name]) != str(u.loc[key, universe_name]):
                raise IngestError(
                    f"Reported metadata {reported_name!r} for {key} contradicts "
                    "the independent inventory."
                )


# Register the values that cannot be reproduced.
def register_reconciliation_gaps(
    unreproduced: Sequence[Mapping[str, Any]],
) -> pd.DataFrame:
    """
    Register the values that cannot be reproduced.

    Parameters
    ----------
    unreproduced : sequence of mapping
        The unreproducible values, each carrying the declared comparison key
        and the blocking input.

    Returns
    -------
    pandas.DataFrame
        One row per gap carrying the blocking input, keyed on the declared
        comparison key.

    Raises
    ------
    IngestError
        If an entry lacks the key or the blocking input, because a gap without
        its blocker cannot be acted on.
    """
    # Accumulate the gap rows.
    rows: List[dict] = []
    # Iterate over the entries.
    for entry in unreproduced:
        # Refuse an entry lacking a declared field.
        for field_name in list(COMPARISON_KEY) + [
            "blocking_input", "description", "unit", "source_id", "procedure_id"
        ]:
            # Test the field's presence.
            if field_name not in entry:
                # The failure names the absent field.
                raise IngestError(
                    f"An unreproduced value lacks the declared field "
                    f"{field_name!r}."
                )
        # Normalize and validate the scientific identity before indexing.
        key_values = tuple(str(entry[name]).strip() for name in COMPARISON_KEY)
        if any(not value for value in key_values):
            raise IngestError("A reconciliation gap carries a blank cell key.")
        blocking_input = str(entry["blocking_input"]).strip()
        description = str(entry["description"]).strip()
        unit = str(entry["unit"]).strip()
        source_id = str(entry["source_id"]).strip()
        procedure_id = str(entry["procedure_id"]).strip()
        if not blocking_input or not description or not unit or not source_id or not procedure_id:
            raise IngestError(
                f"Reconciliation gap {key_values} has blank blocker, description, "
                "unit or source provenance."
            )
        # Append the gap row.
        rows.append(
            {
                "table_number": key_values[0],
                "row_identifier": key_values[1],
                "column_identifier": key_values[2],
                "blocking_input": blocking_input,
                "description": description,
                "unit": unit,
                "source_id": source_id,
                "procedure_id": procedure_id,
                "verdict": VERDICT_GAP,
            }
        )
    # Assemble the frame keyed on the comparison key.
    gaps: pd.DataFrame = pd.DataFrame(
        rows,
        columns=[
            "table_number",
            "row_identifier",
            "column_identifier",
            "blocking_input",
            "description",
            "unit",
            "source_id",
            "procedure_id",
            "verdict",
        ],
    )
    if len(gaps) and bool(gaps.duplicated(list(COMPARISON_KEY), keep=False).any()):
        duplicated = gaps.loc[
            gaps.duplicated(list(COMPARISON_KEY), keep=False), list(COMPARISON_KEY)
        ].head(3).to_dict("records")
        raise IngestError(f"Reconciliation gaps carry duplicate keys: {duplicated}.")
    # Key the frame when it carries rows.
    if len(gaps):
        # Set the declared key.
        gaps = gaps.set_index(list(COMPARISON_KEY))
    # Log the gaps' extent.
    logger.info("reconciliation gaps registered across %d row(s).", len(gaps))
    # Return the frame.
    return gaps


# Assert that every declared row is reconciled or registered as a gap.
def assert_every_row_attributed(
    comparisons: pd.DataFrame,
    gaps: pd.DataFrame,
    declared_keys: Sequence[Tuple[str, str, str]],
) -> pd.DataFrame:
    """
    Assert that every declared row is reconciled or registered as a gap.

    Parameters
    ----------
    comparisons : pandas.DataFrame
        The comparison frame.
    gaps : pandas.DataFrame
        The gaps frame.
    declared_keys : sequence of tuple
        The declared comparison keys of the reported tables.

    Returns
    -------
    pandas.DataFrame
        One-row coverage audit carrying declared, comparison, gap, union,
        missing, extra and overlap counts.

    Raises
    ------
    IngestError
        If a declared key appears in neither frame, because the specification
        fixes that no value is left unattributed.
    """
    normalized_declared = [tuple(str(value) for value in key) for key in declared_keys]
    if any(len(key) != len(COMPARISON_KEY) or any(not value.strip() for value in key) for key in normalized_declared):
        raise IngestError("The independent manuscript universe has a malformed key.")
    if len(normalized_declared) != len(set(normalized_declared)):
        raise IngestError("The independent manuscript universe carries duplicate keys.")
    if comparisons.index.has_duplicates:
        raise IngestError("The comparison table carries duplicate keys.")
    if (
        not isinstance(comparisons.index, pd.MultiIndex)
        or comparisons.index.nlevels != len(COMPARISON_KEY)
        or tuple(comparisons.index.names) != COMPARISON_KEY
    ):
        raise IngestError("The comparison table is not keyed on the declared cell identity.")
    if len(gaps) and gaps.index.has_duplicates:
        raise IngestError("The reconciliation-gap table carries duplicate keys.")
    if len(gaps) and (
        not isinstance(gaps.index, pd.MultiIndex)
        or gaps.index.nlevels != len(COMPARISON_KEY)
        or tuple(gaps.index.names) != COMPARISON_KEY
    ):
        raise IngestError("The gap table is not keyed on the declared cell identity.")
    if "verdict" not in comparisons.columns or not set(comparisons["verdict"].astype(str)).issubset({VERDICT_RECONCILED, VERDICT_DIVERGENT}):
        raise IngestError("The comparison table carries a missing or invalid verdict.")
    if len(gaps) and ("verdict" not in gaps.columns or not gaps["verdict"].astype(str).eq(VERDICT_GAP).all()):
        raise IngestError("The gap table carries a missing or invalid gap verdict.")
    comparison_keys = {
        tuple(str(value) for value in key) for key in comparisons.index.tolist()
    }
    gap_keys = {
        tuple(str(value) for value in key) for key in gaps.index.tolist()
    } if len(gaps) else set()
    declared_set = set(normalized_declared)
    overlap = sorted(comparison_keys & gap_keys)
    attributed = comparison_keys | gap_keys
    unattributed = sorted(declared_set - attributed)
    extras = sorted(attributed - declared_set)
    if overlap or unattributed or extras:
        raise IngestError(
            "Manuscript-cell attribution is not an exact disjoint partition: "
            f"missing={unattributed[:3]}, extras={extras[:3]}, overlap={overlap[:3]}."
        )
    coverage = pd.DataFrame(
        [
            {
                "declared_count": len(declared_set),
                "comparison_count": len(comparison_keys),
                "gap_count": len(gap_keys),
                "union_count": len(attributed),
                "missing_count": 0,
                "extra_count": 0,
                "overlap_count": 0,
                "coverage_status": "exact_disjoint_partition",
                "declared_keys_sha256": _content_digest(tuple(sorted(declared_set))),
                "comparison_keys_sha256": _content_digest(tuple(sorted(comparison_keys))),
                "gap_keys_sha256": _content_digest(tuple(sorted(gap_keys))),
            }
        ]
    )
    # Log the verified attribution.
    logger.info(
        "attribution verified across %d declared key(s).", len(declared_set)
    )
    return coverage


# -------------------------------------------------------------------------------------------------------------------------------
# Task 32, Step 2: The design invariants
# -------------------------------------------------------------------------------------------------------------------------------


# Audit the three design invariants.
def audit_design_invariants(
    counterfactual_callable: Callable[..., Any],
    imputation_callables: Sequence[Callable[..., Any]],
    eligibility_callable: Callable[..., Any],
    robustness_frame: pd.DataFrame,
    probe_registry: Optional[Mapping[Tuple[str, str], Mapping[str, Any]]] = None,
) -> pd.DataFrame:
    """
    Audit the three design invariants.

    Parameters
    ----------
    counterfactual_callable : callable
        The full projection boundary carrying an executable probe under
        ``__reconciliation_probe__``.  The probe supplies independently copied
        baseline and target-post-perturbed calls plus argument roles.
    imputation_callables : sequence of callable
        Completion and classification boundaries carrying executable
        treatment-label perturbation probes.
    eligibility_callable : callable
        The full eligibility boundary carrying the same executable treatment
        perturbation witness.
    robustness_frame : pandas.DataFrame
        Executed robustness runs.  Rows bind immutable before/after baseline
        snapshots and actual variant results through recomputed content hashes.
    probe_registry : mapping, optional
        Independently pinned frozen stage-input scenarios keyed by registered
        boundary and forbidden-input role.

    Returns
    -------
    pandas.DataFrame
        One row per invariant carrying actual input/output digests, dependency
        traces and immutable robustness evidence.

    Raises
    ------
    IngestError
        If a boundary, pinned scenario, recursive dependency classification or
        robustness object is absent, because vacuous evidence cannot pass.
    """
    if not isinstance(probe_registry, Mapping):
        raise IngestError("The invariant audit requires a pinned live-stage probe registry.")
    # Execute independent outcome and covariate perturbations across the full
    # Task-12 orchestration boundary.  A single mixed sentinel cannot prove
    # that both forbidden channels are absent.
    counter_y_evidence = _execute_reconciliation_probe(
        counterfactual_callable,
        forbidden_role="target_post_y",
        probe_spec=probe_registry.get(("orchestrate_counterfactual_projection", "target_post_y")),
    )
    counter_x_evidence = _execute_reconciliation_probe(
        counterfactual_callable,
        forbidden_role="target_post_x",
        probe_spec=probe_registry.get(("orchestrate_counterfactual_projection", "target_post_x")),
    )
    # Execute treatment-label perturbations for every completion/classification
    # boundary and for eligibility itself.
    leakage_evidence: List[Dict[str, Any]] = []
    leakage_evidence.append(
        _execute_reconciliation_probe(
            eligibility_callable,
            forbidden_role="treatment",
            probe_spec=probe_registry.get(("orchestrate_eligibility_screening", "treatment")),
        )
    )
    for callable_ in tuple(imputation_callables):
        boundary_name = str(getattr(callable_, "__name__", ""))
        leakage_evidence.append(
            _audit_structural_treatment_exclusion(
                callable_,
                probe_registry.get((boundary_name, "structural_treatment_exclusion")),
            )
        )
    # Verify robustness snapshots and actual result objects rather than key
    # spelling or copied counts.
    robustness_evidence = _audit_robustness_snapshot_evidence(robustness_frame)
    # Assemble one row for each declared invariant.
    rows: List[dict] = [
        {
            "invariant": INVARIANT_COUNTERFACTUAL,
            "holds": bool(counter_y_evidence["holds"] and counter_x_evidence["holds"]),
            "demonstration": (
                "The full projection boundary was executed twice with identical "
                "permitted inputs and independently perturbed realized target-post "
                "inputs; scientific output digests were compared."
            ),
            "evidence_json": json.dumps(
                {"target_post_y": counter_y_evidence, "target_post_x": counter_x_evidence},
                sort_keys=True,
            ),
            "baseline_output_sha256": _content_digest(
                (counter_y_evidence["baseline_output_sha256"], counter_x_evidence["baseline_output_sha256"])
            ),
            "perturbed_output_sha256": _content_digest(
                (counter_y_evidence["perturbed_output_sha256"], counter_x_evidence["perturbed_output_sha256"])
            ),
            "forbidden_input_changed": bool(
                counter_y_evidence["forbidden_input_changed"]
                and counter_x_evidence["forbidden_input_changed"]
            ),
        },
        {
            "invariant": INVARIANT_LEAKAGE,
            "holds": bool(all(item["holds"] for item in leakage_evidence)),
            "demonstration": (
                "Eligibility was dynamically rerun with independently perturbed "
                "policy indicators; completion/classification boundaries, which "
                "accept no policy argument, were recursively traced for treatment "
                "non-reachability and repeat-executed on pinned identical inputs."
            ),
            "evidence_json": json.dumps(leakage_evidence, sort_keys=True),
            "baseline_output_sha256": _content_digest(
                tuple(item["baseline_output_sha256"] for item in leakage_evidence)
            ),
            "perturbed_output_sha256": _content_digest(
                tuple(item["perturbed_output_sha256"] for item in leakage_evidence)
            ),
            "forbidden_input_changed": bool(
                all(item["forbidden_input_changed"] for item in leakage_evidence)
            ),
        },
        {
            "invariant": INVARIANT_BASELINE,
            "holds": bool(robustness_evidence["holds"]),
            "demonstration": (
                "Actual baseline objects were content-digested before and after "
                "executed robustness branches, and every saved variant result was "
                "bound to its recorded run identifier."
            ),
            "evidence_json": json.dumps(robustness_evidence, sort_keys=True),
            "baseline_output_sha256": robustness_evidence["before_digest"],
            "perturbed_output_sha256": robustness_evidence["after_digest"],
            "forbidden_input_changed": bool(robustness_evidence["variant_count"] > 0),
        },
    ]
    # Assemble the invariant frame keyed on the identifier.
    invariants: pd.DataFrame = pd.DataFrame(rows).set_index("invariant")
    # Assert the coverage of the declared set.
    if set(invariants.index) != set(INVARIANT_IDS):
        # The failure states the condition.
        raise IngestError("The invariant table omits a declared invariant.")
    # Log the verdicts.
    logger.info(
        "design invariants audited: %s.",
        invariants["holds"].to_dict(),
    )
    # Return the frame.
    return invariants


# Execute one independently labelled scientific perturbation probe.
def _clone_reconciliation_input(value: Any) -> Any:
    """Clone nested scientific inputs without failing on read-only mappings."""
    if isinstance(value, type(MappingProxyType({}))):
        return MappingProxyType(
            {key: _clone_reconciliation_input(item) for key, item in value.items()}
        )
    if is_dataclass(value) and not isinstance(value, type):
        return type(value)(
            **{
                item.name: _clone_reconciliation_input(getattr(value, item.name))
                for item in fields(value)
                if item.init
            }
        )
    if isinstance(value, pd.DataFrame):
        cloned = value.copy(deep=True)
        cloned.attrs = copy.deepcopy(value.attrs)
        return cloned
    if isinstance(value, pd.Series):
        cloned = value.copy(deep=True)
        cloned.attrs = copy.deepcopy(value.attrs)
        return cloned
    if isinstance(value, np.ndarray):
        return value.copy()
    if isinstance(value, Mapping):
        return {key: _clone_reconciliation_input(item) for key, item in value.items()}
    if isinstance(value, tuple):
        return tuple(_clone_reconciliation_input(item) for item in value)
    if isinstance(value, list):
        return [_clone_reconciliation_input(item) for item in value]
    if isinstance(value, set):
        return {_clone_reconciliation_input(item) for item in value}
    return copy.deepcopy(value)


# Execute one independently labelled scientific perturbation probe.
def _execute_reconciliation_probe(
    callable_: Callable[..., Any],
    forbidden_role: str,
    probe_spec: Optional[Mapping[str, Any]] = None,
) -> Dict[str, Any]:
    """
    Execute one registered production-boundary perturbation scenario.

    The callable must be the exact module object registered in
    ``RECONCILIATION_BOUNDARY_ROLES``.  Probe payloads merely provide frozen
    stage inputs; they cannot register a lambda, substitute an internal helper
    or narrow the scientific output because the complete returned object is
    digested.

    Parameters
    ----------
    callable_ : callable
        Exact registered production orchestration boundary.
    forbidden_role : str
        Independently perturbed channel: target-post outcome, target-post
        covariate, or policy treatment.
    probe_spec : mapping, optional
        Independently pinned live-stage arguments and role bindings.

    Returns
    -------
    dict
        Input, dependency and scientific-output digest evidence.

    Raises
    ------
    IngestError
        If identity, pin, state binding, dependency classification or
        execution evidence is incomplete.
    """
    if not callable(callable_):
        raise IngestError("An invariant boundary is not callable.")
    boundary_name = str(getattr(callable_, "__name__", ""))
    registered_roles = RECONCILIATION_BOUNDARY_ROLES.get(boundary_name)
    if (
        registered_roles is None
        or forbidden_role not in registered_roles
        or globals().get(boundary_name) is not callable_
        or str(getattr(callable_, "__module__", "")) != __name__
        or boundary_name == "<lambda>"
    ):
        raise IngestError(
            "The reconciliation probe is not attached to the exact registered "
            f"production boundary for role {forbidden_role!r}."
        )
    probe = probe_spec
    if not isinstance(probe, Mapping):
        raise IngestError(
            f"Boundary {boundary_name!r} lacks a separately registered live-stage probe."
        )
    scenarios = probe.get("scenarios")
    scenario: Mapping[str, Any]
    if scenarios is not None:
        if not isinstance(scenarios, Mapping) or forbidden_role not in scenarios:
            raise IngestError(
                f"Registered probe lacks an independent {forbidden_role!r} scenario."
            )
        scenario = scenarios[forbidden_role]
        if not isinstance(scenario, Mapping):
            raise IngestError("A reconciliation perturbation scenario is malformed.")
    else:
        scenario = probe
    for name in (
        "baseline_args", "baseline_kwargs", "perturbed_args", "perturbed_kwargs",
        "argument_roles", "state_bindings",
    ):
        if name not in scenario:
            raise IngestError(f"Reconciliation probe lacks field {name!r}.")
    for name in ("probe_id", "probe_version"):
        if not str(probe.get(name, "")).strip():
            raise IngestError(f"Reconciliation probe lacks field {name!r}.")
    expected_stage_output = str(probe.get("expected_stage_output_sha256", ""))
    if not re.fullmatch(r"[0-9a-f]{64}", expected_stage_output):
        raise IngestError("Reconciliation probe lacks a valid stored stage-output digest.")
    pin_key = (
        boundary_name, forbidden_role, str(probe["probe_id"]), str(probe["probe_version"])
    )
    actual_probe_digest = _pipeline_artifact_digest(probe)
    pinned_probe_digest = TRUSTED_RECONCILIATION_SCENARIO_PINS.get(pin_key)
    if pinned_probe_digest != actual_probe_digest:
        raise IngestError(
            f"Reconciliation scenario {pin_key!r} is absent from or disagrees with the independent pin registry."
        )
    signature = inspect.signature(callable_)
    try:
        baseline_bound = signature.bind(
            *tuple(scenario["baseline_args"]), **dict(scenario["baseline_kwargs"])
        )
        perturbed_bound = signature.bind(
            *tuple(scenario["perturbed_args"]), **dict(scenario["perturbed_kwargs"])
        )
    except TypeError as exc:
        raise IngestError(f"Reconciliation probe arguments do not bind: {exc}") from exc
    baseline_bound.apply_defaults()
    perturbed_bound.apply_defaults()
    roles = {str(name): str(role) for name, role in dict(scenario["argument_roles"]).items()}
    bound_names = set(baseline_bound.arguments) | set(perturbed_bound.arguments)
    if set(roles) != bound_names:
        raise IngestError(
            "Reconciliation probe argument_roles must classify every and only "
            f"bound argument; roles={sorted(roles)}, bound={sorted(bound_names)}."
        )
    state_bindings = scenario.get("state_bindings")
    if not isinstance(state_bindings, Mapping) or set(str(key) for key in state_bindings) != bound_names:
        raise IngestError("The reconciliation scenario does not bind every argument to frozen stage state.")
    for name in sorted(bound_names):
        declared_digest = str(state_bindings[name])
        if not re.fullmatch(r"[0-9a-f]{64}", declared_digest):
            raise IngestError(f"Stage-state binding for argument {name!r} is malformed.")
        if _pipeline_artifact_digest(baseline_bound.arguments[name]) != declared_digest:
            raise IngestError(f"Baseline argument {name!r} is stale relative to frozen stage state.")
    forbidden_names = sorted(name for name, role in roles.items() if role == forbidden_role)
    if not forbidden_names:
        raise IngestError(
            f"Reconciliation probe declares no argument with role {forbidden_role!r}."
        )
    invalid_roles = sorted(
        (name, role) for name, role in roles.items()
        if role not in ("permitted", forbidden_role, "infrastructure")
    )
    if invalid_roles:
        raise IngestError(f"Reconciliation probe carries invalid roles {invalid_roles}.")
    stable_names = sorted(name for name, role in roles.items() if role != forbidden_role)
    baseline_permitted = {name: baseline_bound.arguments[name] for name in stable_names}
    perturbed_permitted = {name: perturbed_bound.arguments[name] for name in stable_names}
    baseline_forbidden = {name: baseline_bound.arguments[name] for name in forbidden_names}
    perturbed_forbidden = {name: perturbed_bound.arguments[name] for name in forbidden_names}
    permitted_before = _pipeline_artifact_digest(baseline_permitted)
    permitted_after = _pipeline_artifact_digest(perturbed_permitted)
    forbidden_before = _pipeline_artifact_digest(baseline_forbidden)
    forbidden_after = _pipeline_artifact_digest(perturbed_forbidden)
    if permitted_before != permitted_after:
        raise IngestError("A reconciliation perturbation changed a permitted scientific input.")
    if forbidden_before == forbidden_after:
        raise IngestError(
            f"A reconciliation perturbation did not change the {forbidden_role} input."
        )
    # Execute on independent copies so mutable aliasing cannot create a false
    # invariant pass.
    try:
        baseline_result = callable_(
            *_clone_reconciliation_input(tuple(scenario["baseline_args"])),
            **_clone_reconciliation_input(dict(scenario["baseline_kwargs"])),
        )
        perturbed_result = callable_(
            *_clone_reconciliation_input(tuple(scenario["perturbed_args"])),
            **_clone_reconciliation_input(dict(scenario["perturbed_kwargs"])),
        )
    except Exception as exc:
        raise IngestError(
            f"Executable reconciliation probe {probe['probe_id']!r} failed: {exc}"
        ) from exc
    if _pipeline_artifact_digest(baseline_result) != expected_stage_output:
        raise IngestError(
            f"Re-executed boundary {boundary_name!r} disagrees with its stored live stage output."
        )
    # Hash the entire returned scientific object.  A caller-supplied selector
    # could return a constant or omit a labelled output and therefore cannot
    # establish the invariant.
    output_fields = RECONCILIATION_BOUNDARY_OUTPUT_FIELDS[boundary_name]
    baseline_output_digest = _scientific_result_digest(
        _boundary_scientific_payload(boundary_name, baseline_result)
    )
    perturbed_output_digest = _scientific_result_digest(
        _boundary_scientific_payload(boundary_name, perturbed_result)
    )
    dependency_trace = _trace_callable_dependencies(
        callable_, scenario.get("dependency_roles", probe.get("dependency_roles", {}))
    )
    declared_dependencies = scenario.get(
        "dependency_bindings", probe.get("dependency_bindings")
    )
    observed_dependencies = {
        str(item["name"]): str(item["sha256"]) for item in dependency_trace
    }
    if (
        not isinstance(declared_dependencies, Mapping)
        or {str(key): str(value) for key, value in declared_dependencies.items()}
        != observed_dependencies
    ):
        raise IngestError(
            "Reconciliation callable dependencies are stale relative to the pinned scenario."
        )
    forbidden_dependency_roles = (
        {"target_post_y", "target_post_x"}
        if forbidden_role in {"target_post_y", "target_post_x"}
        else {forbidden_role}
    )
    hidden_forbidden = [
        item["name"] for item in dependency_trace
        if item["role"] in forbidden_dependency_roles
    ]
    if hidden_forbidden:
        raise IngestError(
            f"Forbidden {forbidden_role!r} data are hidden in callable closure/global "
            f"dependencies {hidden_forbidden}; expose them at the perturbable boundary."
        )
    return {
        "callable": str(getattr(callable_, "__qualname__", getattr(callable_, "__name__", "callable"))),
        "probe_id": str(probe["probe_id"]),
        "probe_version": str(probe["probe_version"]),
        "forbidden_role": forbidden_role,
        "permitted_input_sha256": permitted_before,
        "forbidden_input_baseline_sha256": forbidden_before,
        "forbidden_input_perturbed_sha256": forbidden_after,
        "forbidden_input_changed": bool(forbidden_before != forbidden_after),
        "baseline_output_sha256": baseline_output_digest,
        "perturbed_output_sha256": perturbed_output_digest,
        "output_scope": "module_registered_scientific_schema",
        "output_fields": list(output_fields),
        "excluded_audit_only_fields": (
            ["invariant_audit", "fingerprint_sha256", "projected_utc", "notes"]
            if boundary_name == "orchestrate_counterfactual_projection"
            else (["mask.attrs.policy_audit_digest"] if boundary_name == "orchestrate_eligibility_screening" else [])
        ),
        "output_type": f"{type(baseline_result).__module__}.{type(baseline_result).__qualname__}",
        "dependency_trace": dependency_trace,
        "holds": bool(baseline_output_digest == perturbed_output_digest),
    }


# Prove that a completion/classification boundary cannot receive treatment.
def _audit_structural_treatment_exclusion(
    callable_: Callable[..., Any],
    probe_spec: Optional[Mapping[str, Any]],
) -> Dict[str, Any]:
    """
    Execute frozen inputs and prove structural treatment non-reachability.

    Parameters
    ----------
    callable_ : callable
        Exact registered Task-5 or Task-6 orchestration boundary.
    probe_spec : mapping, optional
        Independently pinned live-stage scenario with complete argument and
        expected-output digests.

    Returns
    -------
    dict
        Structural dependency trace, repeat-execution digests and a Boolean
        invariant verdict. ``dynamic_perturbation_exercised`` is false because
        these boundaries intentionally expose no treatment argument.

    Raises
    ------
    IngestError
        If the boundary/scenario is unregistered, stale, treatment-reachable,
        nondeterministic or disconnected from its stored stage output.
    """
    boundary_name = str(getattr(callable_, "__name__", ""))
    if (
        globals().get(boundary_name) is not callable_
        or RECONCILIATION_BOUNDARY_ROLES.get(boundary_name)
        != ("structural_treatment_exclusion",)
    ):
        raise IngestError("Treatment exclusion was requested for an unregistered boundary.")
    if not isinstance(probe_spec, Mapping):
        raise IngestError(f"Boundary {boundary_name!r} lacks its pinned frozen-input witness.")
    for field in (
        "probe_id", "probe_version", "baseline_args", "baseline_kwargs",
        "state_bindings", "expected_stage_output_sha256", "dependency_bindings",
    ):
        if field not in probe_spec:
            raise IngestError(f"Structural exclusion witness lacks field {field!r}.")
    pin_key = (
        boundary_name, "structural_treatment_exclusion",
        str(probe_spec["probe_id"]), str(probe_spec["probe_version"]),
    )
    if TRUSTED_RECONCILIATION_SCENARIO_PINS.get(pin_key) != _pipeline_artifact_digest(probe_spec):
        raise IngestError("Structural treatment-exclusion witness is not independently pinned.")
    signature = inspect.signature(callable_)
    forbidden_parameter_tokens = ("policy_ledger", "treatment", "sanctions", "direct_arming")
    exposed = [name for name in signature.parameters if any(token in name.lower() for token in forbidden_parameter_tokens)]
    if exposed:
        raise IngestError(f"Boundary {boundary_name!r} exposes treatment parameter(s) {exposed}.")
    try:
        bound = signature.bind(
            *tuple(probe_spec["baseline_args"]), **dict(probe_spec["baseline_kwargs"])
        )
    except TypeError as exc:
        raise IngestError(f"Structural witness arguments do not bind: {exc}") from exc
    bound.apply_defaults()
    bindings = probe_spec["state_bindings"]
    if not isinstance(bindings, Mapping) or set(str(key) for key in bindings) != set(bound.arguments):
        raise IngestError("Structural witness does not bind every argument to frozen state.")
    for name, value in bound.arguments.items():
        if _pipeline_artifact_digest(value) != str(bindings[name]):
            raise IngestError(f"Structural witness argument {name!r} is stale.")
    trace = _trace_callable_dependencies(callable_, probe_spec.get("dependency_roles", {}))
    observed_dependencies = {
        str(item["name"]): str(item["sha256"]) for item in trace
    }
    declared_dependencies = probe_spec["dependency_bindings"]
    if (
        not isinstance(declared_dependencies, Mapping)
        or {str(key): str(value) for key, value in declared_dependencies.items()}
        != observed_dependencies
    ):
        raise IngestError(
            "Structural callable dependencies are stale relative to the pinned scenario."
        )
    hidden = [row["name"] for row in trace if row["role"] == "treatment"]
    if hidden:
        raise IngestError(f"Treatment is reachable through hidden dependencies {hidden}.")
    try:
        first = callable_(
            *_clone_reconciliation_input(tuple(probe_spec["baseline_args"])),
            **_clone_reconciliation_input(dict(probe_spec["baseline_kwargs"])),
        )
        second = callable_(
            *_clone_reconciliation_input(tuple(probe_spec["baseline_args"])),
            **_clone_reconciliation_input(dict(probe_spec["baseline_kwargs"])),
        )
    except Exception as exc:
        raise IngestError(f"Structural exclusion witness failed: {exc}") from exc
    expected_stage_output = str(probe_spec["expected_stage_output_sha256"])
    if (
        not re.fullmatch(r"[0-9a-f]{64}", expected_stage_output)
        or _pipeline_artifact_digest(first) != expected_stage_output
    ):
        raise IngestError(
            f"Re-executed boundary {boundary_name!r} disagrees with stored stage state."
        )
    first_digest = _scientific_result_digest(_boundary_scientific_payload(boundary_name, first))
    second_digest = _scientific_result_digest(_boundary_scientific_payload(boundary_name, second))
    return {
        "callable": boundary_name,
        "probe_id": str(probe_spec["probe_id"]),
        "probe_version": str(probe_spec["probe_version"]),
        "forbidden_role": "treatment",
        "test_mode": "structural_nonreachability_plus_repeat_execution",
        "forbidden_input_changed": False,
        "dynamic_perturbation_exercised": False,
        "structural_nonreachability_verified": True,
        "baseline_output_sha256": first_digest,
        "perturbed_output_sha256": second_digest,
        "dependency_trace": trace,
        "holds": bool(first_digest == second_digest),
    }


# Digest the complete returned object of a scientific invariant boundary.
def _scientific_result_digest(value: Any) -> str:
    """
    Digest a complete scientific result recursively.

    Parameters
    ----------
    value : Any
        Scientific frame, series, dataclass, named tuple, mapping, sequence,
        array or scalar returned by a registered boundary.

    Returns
    -------
    str
        SHA-256 digest covering labels, values, schema and pandas attributes.

    Raises
    ------
    IngestError
        If a nested object has no registered deterministic digest contract.
    """
    if isinstance(value, (pd.DataFrame, pd.Series)):
        digest = hashlib.sha256(_content_digest(value).encode("utf-8"))
        digest.update(_canonical_audit_json(dict(value.attrs)).encode("utf-8"))
        return digest.hexdigest()
    if hasattr(value, "__dataclass_fields__"):
        digest = hashlib.sha256(type(value).__qualname__.encode("utf-8"))
        for name in value.__dataclass_fields__.keys():
            digest.update(str(name).encode("utf-8"))
            digest.update(_scientific_result_digest(getattr(value, name)).encode("utf-8"))
        return digest.hexdigest()
    if isinstance(value, tuple) and hasattr(value, "_asdict"):
        value = value._asdict()
    if isinstance(value, Mapping):
        digest = hashlib.sha256(b"mapping")
        for key in sorted(value, key=str):
            digest.update(str(key).encode("utf-8"))
            digest.update(_scientific_result_digest(value[key]).encode("utf-8"))
        return digest.hexdigest()
    if isinstance(value, (tuple, list)):
        digest = hashlib.sha256(type(value).__name__.encode("utf-8"))
        for item in value:
            digest.update(_scientific_result_digest(item).encode("utf-8"))
        return digest.hexdigest()
    return _content_digest(value)


# Project one registered boundary result onto its fixed scientific schema.
def _boundary_scientific_payload(boundary_name: str, result: Any) -> Mapping[str, Any]:
    """
    Return all registered scientific fields with named audit-only data removed.

    Parameters
    ----------
    boundary_name : str
        Exact registered production boundary name.
    result : Any
        Boundary result to project onto its module-declared scientific schema.

    Returns
    -------
    mapping
        Every registered scientific field. Only explicitly named volatile or
        forbidden-input audit provenance is excluded.

    Raises
    ------
    IngestError
        If the result omits any field in the fixed module schema.
    """
    fields_ = RECONCILIATION_BOUNDARY_OUTPUT_FIELDS[boundary_name]
    if any(not hasattr(result, field) for field in fields_):
        raise IngestError(
            f"Registered boundary {boundary_name!r} did not return its complete scientific schema."
        )
    payload: Dict[str, Any] = {field: getattr(result, field) for field in fields_}
    if boundary_name == "orchestrate_eligibility_screening":
        mask = payload["mask"].copy(deep=True)
        mask.attrs = {
            key: value for key, value in mask.attrs.items()
            if str(key) != "policy_audit_digest"
        }
        payload["mask"] = mask
    return payload


# Trace closure and global dependencies actually resolved by a callable.
def _trace_callable_dependencies(
    callable_: Callable[..., Any],
    declared_roles: Mapping[str, str],
) -> List[Dict[str, str]]:
    """
    Recursively trace helpers and reject opaque or unclassified state.

    Immutable literals are content-bound infrastructure. Python helpers are
    recursively inspected across module boundaries, while approved installed
    modules are accepted only through provenance-verified, statically selected
    attributes whose live identities and implementations are bound. Mutable
    scientific objects must be assigned an explicit role. Every reachable
    method and descriptor of an actively used class is inspected so
    class-mediated dispatch cannot conceal a target-post or treatment
    dependency. Classes used only as exception types or as
    ``isinstance``/``issubclass`` predicates remain inert and therefore do not
    cause unrelated methods to enter the scientific dependency graph.

    Parameters
    ----------
    callable_ : callable
        Transparent registered Python boundary to inspect.
    declared_roles : mapping of str to str
        Explicit classifications for mutable scientific closure/global state.

    Returns
    -------
    list of dict
        Recursive dependency path, scope, role and content/source digest.

    Raises
    ------
    IngestError
        If a callable, module dispatch or descriptor is opaque; a mutable
        dependency or class attribute is unclassified; a class/native export
        lacks verified provenance; or the dependency graph exceeds its
        recursion bound.
    """
    allowed_roles = {"permitted", "target_post_y", "target_post_x", "treatment"}
    forbidden_reflection_names = {
        "__import__",
        "eval",
        "exec",
        "getattr",
        "globals",
        "locals",
        "vars",
    }
    forbidden_reflection_objects = {
        getattr(builtins, name)
        for name in forbidden_reflection_names
        if hasattr(builtins, name)
    }
    forbidden_reflection_objects.update(
        {object.__getattribute__, type.__getattribute__}
    )
    forbidden_reflection_objects.update(
        TRUSTED_RECONCILIATION_CALLER_NAMESPACE_CALLABLES
    )
    forbidden_caller_namespace_methods = {"eval", "query"}
    forbidden_reflective_attributes = {
        "__annotations__",
        "__closure__",
        "__code__",
        "__defaults__",
        "__dict__",
        "__globals__",
        "__getattribute__",
        "__kwdefaults__",
    }
    entries: List[Dict[str, str]] = []
    visited_functions: Set[int] = set()
    visited_classes: Set[int] = set()
    visited_module_attributes: Dict[int, Set[str]] = {}
    visited_trusted_module_chains: Set[Tuple[int, Tuple[str, ...]]] = set()
    trusted_module_objects: Tuple[Any, ...] = (
        copy,
        hashlib,
        importlib,
        itertools,
        json,
        logging,
        math,
        os,
        re,
        sys,
        tempfile,
        matplotlib,
        plt,
        np,
        pd,
        scipy_stats,
    )

    # Accept a proxy as infrastructure only when every nested key and value is
    # itself an immutable literal. A read-only outer mapping can otherwise
    # conceal mutable scientific frames or callables.
    def deeply_immutable(value: Any, active: Optional[Set[int]] = None) -> bool:
        """Return whether ``value`` contains only recursively immutable literals."""
        seen = set() if active is None else active
        if isinstance(value, (str, bytes, int, float, bool, type(None), np.generic)):
            return True
        if id(value) in seen:
            return False
        if isinstance(value, (tuple, frozenset)):
            seen.add(id(value))
            verdict = all(deeply_immutable(item, seen) for item in value)
            seen.remove(id(value))
            return verdict
        if isinstance(value, type(MappingProxyType({}))):
            seen.add(id(value))
            verdict = all(
                deeply_immutable(key, seen) and deeply_immutable(item, seen)
                for key, item in value.items()
            )
            seen.remove(id(value))
            return verdict
        return False

    # Hash Python bytecode recursively so dynamically defined or monkeypatched
    # methods remain content-bound even when inspect cannot recover source text.
    def code_digest(code: Any, digest: "hashlib._Hash") -> None:
        """Update ``digest`` with a deterministic recursive code-object payload."""
        digest.update(bytes(code.co_code))
        digest.update(repr(tuple(code.co_names)).encode("utf-8"))
        digest.update(repr(tuple(code.co_varnames)).encode("utf-8"))
        digest.update(repr(tuple(code.co_freevars)).encode("utf-8"))
        digest.update(repr(tuple(code.co_cellvars)).encode("utf-8"))
        for constant in code.co_consts:
            if inspect.iscode(constant):
                code_digest(constant, digest)
            else:
                digest.update(repr(constant).encode("utf-8"))

    # Bind both source and executable implementation. Source catches semantic
    # edits that compile similarly; bytecode catches runtime monkeypatches whose
    # original class source remains unchanged.
    def callable_digest(function: Callable[..., Any]) -> str:
        """Return a stable source-and-bytecode digest for one Python function."""
        digest = hashlib.sha256()
        digest.update(
            f"{getattr(function, '__module__', '')}."
            f"{getattr(function, '__qualname__', getattr(function, '__name__', ''))}".encode(
                "utf-8"
            )
        )
        try:
            digest.update(inspect.getsource(function).encode("utf-8"))
        except (OSError, TypeError):
            # Dynamically defined test/production functions may have no source
            # file; their complete recursive code object is still available.
            pass
        code = getattr(function, "__code__", None)
        if code is None:
            raise IngestError(
                f"Opaque Python callable {function!r} has no inspectable code object."
            )
        code_digest(code, digest)
        digest.update(repr(getattr(function, "__defaults__", None)).encode("utf-8"))
        digest.update(repr(getattr(function, "__kwdefaults__", None)).encode("utf-8"))
        return digest.hexdigest()

    # Dynamic namespace/reflection primitives can resolve scientific state that
    # neither ``inspect.getclosurevars`` nor a static global-name walk can see.
    # No registered reconciliation boundary needs them, so their presence is a
    # fail-closed condition rather than an unverifiable infrastructure claim.
    def reject_dynamic_namespace_dispatch(function: Callable[..., Any], path: str) -> None:
        """
        Reject executable uses of untraceable namespace/reflection built-ins.

        Parameters
        ----------
        function : callable
            Transparent Python function whose bytecode is being traced.
        path : str
            Fully qualified evidence path used in diagnostics.

        Returns
        -------
        None
            Returns only when no forbidden reflection primitive is loaded.

        Raises
        ------
        IngestError
            If the bytecode can resolve a dynamic namespace through
            ``globals``, ``locals``, ``vars``, ``eval``, ``exec`` or
            ``__import__``.
        """
        used = {
            str(instruction.argval)
            for instruction in dis.get_instructions(function)
            if instruction.opname in {"LOAD_GLOBAL", "LOAD_NAME", "LOAD_DEREF"}
            and str(instruction.argval) in forbidden_reflection_names
        }
        reflected_attributes = {
            str(instruction.argval)
            for instruction in dis.get_instructions(function)
            if instruction.opname in {"LOAD_ATTR", "LOAD_METHOD"}
            and str(instruction.argval) in forbidden_reflective_attributes
        }
        caller_namespace_methods = {
            str(instruction.argval)
            for instruction in dis.get_instructions(function)
            if instruction.opname in {"LOAD_ATTR", "LOAD_METHOD"}
            and str(instruction.argval) in forbidden_caller_namespace_methods
        }
        if used:
            raise IngestError(
                f"Callable {path!r} uses opaque namespace/reflection dispatch "
                f"{sorted(used)!r}; registered reconciliation boundaries must expose "
                "all scientific dependencies statically."
            )
        if reflected_attributes:
            raise IngestError(
                f"Callable {path!r} reflectively accesses opaque executable "
                f"metadata {sorted(reflected_attributes)!r}; registered boundaries "
                "must expose scientific dependencies through ordinary bindings."
            )
        if caller_namespace_methods:
            raise IngestError(
                f"Callable {path!r} invokes caller-namespace evaluator(s) "
                f"{sorted(caller_namespace_methods)!r}; expression engines must "
                "receive every scientific value as an explicit argument."
            )

    # Function and method defaults participate in execution without appearing
    # in closure/global bindings. Audit them explicitly so a frame captured as
    # ``def boundary(x=secret)`` cannot evade dependency classification.
    def audit_function_defaults(
        function: Callable[..., Any], path: str
    ) -> List[Dict[str, str]]:
        """
        Return role-classified evidence for positional and keyword defaults.

        Parameters
        ----------
        function : callable
            Transparent Python function or method being inspected.
        path : str
            Fully qualified callable path used to resolve declared roles.

        Returns
        -------
        list of dict
            One content-bound evidence record per positional or keyword-only
            default.

        Raises
        ------
        IngestError
            If the signature/default layout is malformed or a mutable default
            has no supported scientific role.
        """
        evidence: List[Dict[str, str]] = []
        try:
            parameters = list(inspect.signature(function).parameters.values())
        except (TypeError, ValueError) as exc:
            raise IngestError(
                f"Cannot inspect default arguments for callable {path!r}: {exc}"
            ) from exc
        defaults = tuple(getattr(function, "__defaults__", ()) or ())
        positional = [
            parameter
            for parameter in parameters
            if parameter.kind
            in {parameter.POSITIONAL_ONLY, parameter.POSITIONAL_OR_KEYWORD}
        ]
        if len(defaults) > len(positional):
            raise IngestError(f"Callable {path!r} has an invalid positional-default layout.")
        named_defaults: List[Tuple[str, Any]] = [
            (parameter.name, value)
            for parameter, value in zip(positional[-len(defaults) :], defaults)
        ] if defaults else []
        named_defaults.extend(
            (str(name), value)
            for name, value in sorted(
                dict(getattr(function, "__kwdefaults__", {}) or {}).items()
            )
        )
        for name, value in named_defaults:
            default_path = f"{path}.__default__.{name}"
            if deeply_immutable(value):
                role = "infrastructure_default"
            else:
                declared = declared_roles.get(
                    default_path,
                    declared_roles.get(f"__default__.{name}", declared_roles.get(name)),
                )
                if declared is None:
                    raise IngestError(
                        f"Callable {path!r} has unclassified mutable default {name!r}."
                    )
                role = str(declared)
                if role not in allowed_roles:
                    raise IngestError(
                        f"Callable default {default_path!r} has unsupported role {role!r}."
                    )
            evidence.append(
                {
                    "scope": "default",
                    "name": default_path,
                    "role": role,
                    "sha256": _scientific_result_digest(value),
                }
            )
        return evidence

    # Bind passive function metadata even when it is not read by bytecode.
    # Active reflective access is rejected above; this snapshot additionally
    # ensures an annotation or caller attribute mutation invalidates the trace.
    def audit_function_metadata(
        function: Callable[..., Any], path: str
    ) -> List[Dict[str, str]]:
        """
        Return deterministic evidence for passive Python-function metadata.

        Parameters
        ----------
        function : callable
            Transparent Python function or method being inspected.
        path : str
            Fully qualified callable evidence path.

        Returns
        -------
        list of dict
            Content-bound annotation, documentation and optional caller-
            attribute records.

        Raises
        ------
        IngestError
            If identity fields are malformed or annotations/user attributes
            contain scientific objects without a deterministic metadata-only
            contract.
        """
        for identity_name in ("__module__", "__qualname__", "__name__"):
            identity_value = getattr(function, identity_name, None)
            if not isinstance(identity_value, str) or not identity_value.strip():
                raise IngestError(
                    f"Callable {path!r} has malformed identity field {identity_name!r}."
                )
        records: List[Dict[str, str]] = []
        for label, value in (
            ("annotations", dict(getattr(function, "__annotations__", {}) or {})),
            ("documentation", getattr(function, "__doc__", None)),
            ("user_attributes", dict(vars(function))),
        ):
            # ``metadata_digest`` accepts only semantic metadata structures;
            # frames, arrays and opaque instances therefore fail closed.
            digest = metadata_digest(value)
            records.append(
                {
                    "scope": "callable_metadata",
                    "name": f"{path}.__metadata__.{label}",
                    "role": "infrastructure_callable_metadata",
                    "sha256": digest,
                }
            )
        return records

    # Determine whether a class is merely a predicate/exception token. The
    # bytecode segmentation is deliberately conservative: any constructor,
    # attribute access or ambiguous use activates exhaustive class traversal.
    def class_reference_mode(
        function: Callable[..., Any], dependency_name: str, class_: type
    ) -> str:
        """Classify a class dependency as ``inert`` or actively dispatched."""
        instructions = list(dis.get_instructions(function))
        occurrences = [
            index
            for index, instruction in enumerate(instructions)
            if instruction.opname in {"LOAD_GLOBAL", "LOAD_DEREF", "LOAD_NAME"}
            and instruction.argval == dependency_name
        ]
        # Closure inspection resolved the class even if a future Python opcode
        # is unknown; active traversal is the safe default.
        if not occurrences:
            return "active"
        for occurrence in occurrences:
            start = occurrence
            while start > 0 and not (
                instructions[start - 1].opname.startswith("CALL")
                or instructions[start - 1].opname
                in {"RETURN_VALUE", "RAISE_VARARGS", "RERAISE"}
            ):
                start -= 1
            end = occurrence
            while end < len(instructions) and not instructions[end].opname.startswith("CALL"):
                if instructions[end].opname == "CHECK_EXC_MATCH":
                    break
                end += 1
            segment = instructions[start : min(end + 1, len(instructions))]
            loaded = {
                str(item.argval)
                for item in segment
                if item.opname in {"LOAD_GLOBAL", "LOAD_DEREF", "LOAD_NAME"}
            }
            next_call = instructions[end] if end < len(instructions) else None
            attribute_dispatch = any(
                item.opname in {"LOAD_ATTR", "LOAD_METHOD"}
                for item in instructions[occurrence + 1 : end]
            )
            exception_match = bool(
                next_call is not None and next_call.opname == "CHECK_EXC_MATCH"
            )
            predicate_use = bool(
                loaded & {"isinstance", "issubclass"}
                and next_call is not None
                and isinstance(next_call.arg, int)
                and int(next_call.arg) >= 2
            )
            if attribute_dispatch or not (predicate_use or exception_match):
                return "active"
        return "inert"

    # Extract every executable Python function behind a class descriptor. The
    # returned labels distinguish getters/setters and wrapper types in evidence.
    def descriptor_functions(
        descriptor_name: str, descriptor: Any
    ) -> List[Tuple[str, Callable[..., Any]]]:
        """Return labelled Python callables implemented by one class descriptor."""
        functions_: List[Tuple[str, Callable[..., Any]]] = []
        if isinstance(descriptor, staticmethod):
            functions_.append(("staticmethod", descriptor.__func__))
        elif isinstance(descriptor, classmethod):
            functions_.append(("classmethod", descriptor.__func__))
        elif isinstance(descriptor, property):
            for label, function in (
                ("property_get", descriptor.fget),
                ("property_set", descriptor.fset),
                ("property_del", descriptor.fdel),
            ):
                if function is not None:
                    functions_.append((label, function))
        elif inspect.isfunction(descriptor):
            functions_.append(("instance_method", descriptor))
        else:
            # Support transparent cached/custom descriptors exposing a Python
            # implementation through ``func`` or their descriptor class.
            wrapped = inspect.getattr_static(descriptor, "func", None)
            if inspect.isfunction(wrapped):
                functions_.append(("descriptor_func", wrapped))
            descriptor_type = type(descriptor)
            for label, method_name in (
                ("descriptor_get", "__get__"),
                ("descriptor_set", "__set__"),
                ("descriptor_delete", "__delete__"),
            ):
                method = inspect.getattr_static(descriptor_type, method_name, None)
                if inspect.isfunction(method):
                    functions_.append((label, method))
        # Deduplicate wrapper paths that expose the same underlying function.
        unique: List[Tuple[str, Callable[..., Any]]] = []
        seen_functions: Set[int] = set()
        for label, function in functions_:
            if id(function) not in seen_functions:
                seen_functions.add(id(function))
                unique.append((f"{descriptor_name}:{label}", function))
        return unique

    # Extract caller-controlled state retained by a custom descriptor object.
    # Its __get__/__set__ implementation is insufficient evidence when the
    # instance closes over a mutable payload through __dict__ or slots.
    def descriptor_state_items(descriptor: Any) -> List[Tuple[str, Any]]:
        """
        Return deterministic named state held by one custom descriptor.

        Parameters
        ----------
        descriptor : Any
            Descriptor object whose instance dictionary and declared slots are
            reachable during attribute dispatch.

        Returns
        -------
        list of tuple
            Sorted ``(name, value)`` pairs from ``__dict__`` and initialized
            slots. Standard method/property wrappers return an empty list.
        """
        if isinstance(descriptor, (staticmethod, classmethod, property)) or inspect.isfunction(descriptor):
            return []
        state: Dict[str, Any] = {}
        try:
            state.update(dict(vars(descriptor)))
        except TypeError:
            pass
        for descriptor_owner in type(descriptor).__mro__:
            slots = vars(descriptor_owner).get("__slots__", ())
            if isinstance(slots, str):
                slots = (slots,)
            for slot in slots:
                slot_name = str(slot)
                if slot_name in {"__dict__", "__weakref__"} or slot_name in state:
                    continue
                try:
                    state[slot_name] = object.__getattribute__(descriptor, slot_name)
                except AttributeError:
                    continue
        return [(name, state[name]) for name in sorted(state)]

    # A Python class may rebind normally textual namespace fields such as
    # ``__module__``. Validate identity fields before using them in paths or
    # digests, otherwise a DataFrame payload can trigger ambiguous truth-value
    # errors or evade the class-member audit.
    def safe_class_identity(class_: type) -> str:
        """
        Return the validated textual module/qualname identity of a class.

        Parameters
        ----------
        class_ : type
            Active class whose diagnostic identity is required.

        Returns
        -------
        str
            ``module.qualname`` after strict textual validation.

        Raises
        ------
        IngestError
            If writable identity metadata was rebound to blank or data-bearing
            content.
        """
        module_name = type.__getattribute__(class_, "__module__")
        qualified_name = type.__getattribute__(class_, "__qualname__")
        if (
            not isinstance(module_name, str)
            or not module_name.strip()
            or not isinstance(qualified_name, str)
            or not qualified_name.strip()
        ):
            raise IngestError(
                "Active class identity fields __module__ and __qualname__ must be "
                "nonblank strings; rebound data-bearing identity metadata is opaque."
            )
        return f"{module_name}.{qualified_name}"

    # Textual ``__module__``/``__qualname__`` fields are writable. Determine
    # built-in and trusted-opaque provenance by object identity instead, so a
    # user class cannot spoof either suppression path.
    def actual_builtin_class(class_: type) -> bool:
        """
        Return whether ``class_`` is the exact class exported by ``builtins``.

        The check uses object identity rather than writable ``__module__``
        metadata, preventing a user class from spoofing the suppression rule.
        """
        try:
            name = type.__getattribute__(class_, "__name__")
        except (AttributeError, TypeError):
            return False
        return bool(
            isinstance(name, str)
            and getattr(builtins, name, None) is class_
        )

    def trusted_opaque_class(class_: type) -> bool:
        """
        Return whether ``class_`` is an exact module-approved opaque class.

        Only the imported ``datetime``, ``timezone`` and ``Decimal`` class
        objects qualify; matching text on a caller class is insufficient.
        """
        return bool(
            (class_ is datetime or class_ is timezone or class_ is Decimal)
            and safe_class_identity(class_)
            in TRUSTED_RECONCILIATION_OPAQUE_CLASSES
        )

    # Library modules are trusted only by exact imported-object identity. Their
    # installed source/version is still bound into the trace, so a same-named
    # caller module cannot inherit this exemption.
    def trusted_module_object(module: Any) -> bool:
        """Return whether ``module`` is an exact approved runtime module object."""
        return any(module is candidate for candidate in trusted_module_objects)

    def module_implementation_digest(module: Any) -> str:
        """
        Return a source/version digest for an exact trusted or inspected module.

        The digest covers the runtime module identity, advertised version and
        installed module file bytes when a readable regular file exists.
        """
        namespace = vars(module)
        module_name = namespace.get("__name__")
        if not isinstance(module_name, str) or not module_name.strip():
            raise IngestError("A module dependency lacks a genuine nonblank name.")
        digest = hashlib.sha256(f"module:{module_name}".encode("utf-8"))
        version = namespace.get("__version__")
        if version is not None:
            digest.update(f"version:{version}".encode("utf-8"))
        file_value = namespace.get("__file__")
        if file_value is not None:
            try:
                module_path = Path(file_value).expanduser().resolve()
            except (OSError, TypeError, ValueError) as exc:
                raise IngestError(
                    f"Module dependency {module_name!r} has an invalid source path."
                ) from exc
            digest.update(str(module_path).encode("utf-8"))
            if module_path.is_file():
                try:
                    digest.update(module_path.read_bytes())
                except OSError as exc:
                    raise IngestError(
                        f"Module dependency {module_name!r} source bytes are unreadable."
                    ) from exc
        return digest.hexdigest()

    # Resolve the first statically named attribute loaded from a module global.
    # Dynamic getattr is rejected separately, so every accepted module dispatch
    # must expose at least one such attribute in bytecode.
    def module_attribute_chains(
        function: Any, dependency_name: str
    ) -> Set[Tuple[str, ...]]:
        """Return statically loaded attribute chains of a module binding."""
        instructions = list(dis.get_instructions(function))
        callable_name = str(
            getattr(
                function,
                "__qualname__",
                getattr(function, "co_qualname", getattr(function, "co_name", "<code>")),
            )
        )
        chains: Set[Tuple[str, ...]] = set()
        for index, instruction in enumerate(instructions):
            if (
                instruction.opname not in {"LOAD_GLOBAL", "LOAD_DEREF", "LOAD_NAME"}
                or instruction.argval != dependency_name
            ):
                continue
            following = index + 1
            while following < len(instructions) and instructions[following].opname in {
                "CACHE", "EXTENDED_ARG", "NOP",
            }:
                following += 1
            chain: List[str] = []
            while (
                following < len(instructions)
                and instructions[following].opname in {"LOAD_ATTR", "LOAD_METHOD"}
            ):
                chain.append(str(instructions[following].argval))
                following += 1
                while following < len(instructions) and instructions[following].opname in {
                    "CACHE", "EXTENDED_ARG", "NOP",
                }:
                    following += 1
            if not chain:
                raise IngestError(
                    f"Callable {callable_name!r} passes module dependency "
                    f"{dependency_name!r} through opaque dispatch."
                )
            chains.add(tuple(chain))
        if not chains:
            raise IngestError(
                f"Callable {callable_name!r} has no auditable attribute "
                f"dispatch for module dependency {dependency_name!r}."
            )
        return chains

    def module_attribute_names(
        function: Any, dependency_name: str
    ) -> Set[str]:
        """Return first-level names from statically loaded module attribute chains."""
        return {
            chain[0]
            for chain in module_attribute_chains(function, dependency_name)
        }

    def trusted_module_namespace(module: Any) -> bool:
        """Return whether a module is a provenance-verified approved namespace."""
        if not inspect.ismodule(module):
            return False
        name = vars(module).get("__name__")
        if not isinstance(name, str) or sys.modules.get(name) is not module:
            return False
        # The explicitly imported top-level objects are accepted by identity.
        # A prefix match alone is never sufficient for a submodule: require a
        # real import spec, an origin equal to its runtime file, and containment
        # beneath an installed approved package root. This rejects in-memory
        # ``ModuleType('numpy.spoof')`` objects inserted into ``sys.modules``.
        if trusted_module_object(module):
            return True
        file_value = vars(module).get("__file__")
        spec = vars(module).get("__spec__")
        origin = getattr(spec, "origin", None)
        loader = getattr(spec, "loader", None)
        if loader is None or not isinstance(file_value, (str, os.PathLike)):
            return False
        if not isinstance(origin, (str, os.PathLike)):
            return False
        try:
            module_file = Path(file_value).expanduser().resolve(strict=True)
            origin_file = Path(origin).expanduser().resolve(strict=True)
        except (OSError, TypeError, ValueError):
            return False
        if module_file != origin_file or not module_file.is_file():
            return False
        for candidate in trusted_module_objects:
            namespace = vars(candidate)
            candidate_name = namespace.get("__name__")
            package_paths = namespace.get("__path__")
            if (
                not isinstance(candidate_name, str)
                or not name.startswith(f"{candidate_name}.")
                or package_paths is None
            ):
                continue
            for package_path in package_paths:
                try:
                    package_root = Path(package_path).expanduser().resolve(strict=True)
                    module_file.relative_to(package_root)
                except (OSError, TypeError, ValueError):
                    continue
                return True
        return False

    def exact_trusted_library_function(function: Callable[..., Any]) -> bool:
        """Return whether a Python helper is the exact export of a trusted module."""
        public_module = sys.modules.get(str(getattr(function, "__module__", "")))
        name = getattr(function, "__name__", None)
        function_globals = getattr(function, "__globals__", None)
        defining_name = (
            function_globals.get("__name__")
            if isinstance(function_globals, dict)
            else None
        )
        defining_module = sys.modules.get(str(defining_name))
        if not (
            trusted_module_namespace(public_module)
            and trusted_module_namespace(defining_module)
            and isinstance(name, str)
            and vars(public_module).get(name) is function
            and function_globals is vars(defining_module)
        ):
            return False
        # A runtime replacement can spoof writable ``__module__``/``__name__``
        # metadata. The executable code must also originate from the verified
        # defining module file before recursive inspection may be omitted.
        module_file = vars(defining_module).get("__file__")
        code = getattr(function, "__code__", None)
        if module_file is None or code is None:
            return False
        try:
            return (
                Path(module_file).expanduser().resolve(strict=True)
                == Path(code.co_filename).expanduser().resolve(strict=True)
            )
        except (OSError, TypeError, ValueError):
            return False

    def exact_trusted_library_class(class_: type) -> bool:
        """Return whether a class is an exact provenance-bound library export."""
        module = sys.modules.get(str(type.__getattribute__(class_, "__module__")))
        name = type.__getattribute__(class_, "__name__")
        if not (
            trusted_module_namespace(module)
            and isinstance(name, str)
            and vars(module).get(name) is class_
        ):
            return False
        if any(
            class_ is candidate
            for candidate in TRUSTED_RECONCILIATION_LIBRARY_CLASS_OBJECTS
        ):
            return True
        # Non-heap extension types cannot be synthesized by ordinary caller
        # code and are already bound by exact export identity. Python/heap
        # classes additionally require recoverable source in the verified
        # defining module file; writable __module__/__name__ text is not proof.
        if not bool(type.__getattribute__(class_, "__flags__") & (1 << 9)):
            return True
        module_file = vars(module).get("__file__")
        try:
            source_file = inspect.getsourcefile(class_)
            source = inspect.getsource(class_)
            return bool(
                source.strip()
                and Path(module_file).expanduser().resolve(strict=True)
                == Path(source_file).expanduser().resolve(strict=True)
            )
        except (OSError, TypeError, ValueError):
            return False

    def trusted_library_class_digest(class_: type, parent_module: Any) -> str:
        """Digest one exact installed class export and its Python source, if any."""
        if not exact_trusted_library_class(class_):
            raise IngestError("A runtime class replacement lacks trusted provenance.")
        try:
            source_sha256: Optional[str] = hashlib.sha256(
                inspect.getsource(class_).encode("utf-8")
            ).hexdigest()
        except (OSError, TypeError):
            source_sha256 = None
        return _scientific_result_digest(
            {
                "identity": safe_class_identity(class_),
                "flags": int(type.__getattribute__(class_, "__flags__")),
                "owner_sha256": module_implementation_digest(parent_module),
                "source_sha256": source_sha256,
            }
        )

    def trusted_native_attribute_digest(value: Any, parent_module: Any) -> str:
        """
        Bind an approved native library attribute to its live runtime identity.

        Parameters
        ----------
        value : Any
            Native function, method, ufunc, descriptor or other C-backed
            callable selected through an approved module namespace.
        parent_module : module
            Exact live module object that owns the selected attribute.

        Returns
        -------
        str
            SHA-256 digest binding the owning installed module, concrete type,
            exported names and stable representation of the live object.

        Raises
        ------
        IngestError
            If the parent is not an approved live module namespace or the
            callable lacks a stable exported identity.
        """
        if value in forbidden_reflection_objects:
            raise IngestError("A native library attribute aliases forbidden reflection.")
        module_name = getattr(value, "__module__", None)
        qualified_name = getattr(value, "__qualname__", getattr(value, "__name__", None))
        if not isinstance(qualified_name, str) or not qualified_name.strip():
            raise IngestError("A native library attribute lacks a stable exported name.")
        defining_module = sys.modules.get(str(module_name))
        export_name = getattr(value, "__name__", None)
        snapshotted = any(
            value is candidate
            for candidate in TRUSTED_RECONCILIATION_NATIVE_CALLABLE_OBJECTS
        )
        verified_defining_export = bool(
            trusted_module_namespace(defining_module)
            and isinstance(export_name, str)
            and vars(defining_module).get(export_name) is value
        )
        if not trusted_module_namespace(parent_module) or not (
            snapshotted or verified_defining_export
        ):
            raise IngestError(
                "A native library attribute is not an exact verified library export."
            )
        payload = {
            "owner_sha256": module_implementation_digest(parent_module),
            "defining_module_sha256": (
                module_implementation_digest(defining_module)
                if trusted_module_namespace(defining_module)
                else None
            ),
            "type": f"{type(value).__module__}.{type(value).__qualname__}",
            "module": module_name,
            "qualified_name": qualified_name,
            "representation": repr(value),
        }
        return _scientific_result_digest(payload)

    def trusted_wrapped_callable_digest(value: Any, parent_module: Any) -> Optional[str]:
        """
        Digest an exact installed-library dispatcher with a Python implementation.

        NumPy exposes several public functions as C-backed dispatcher objects
        rather than ``inspect.isfunction`` objects. They qualify only when the
        live parent module exports the exact object and ``__wrapped__`` is a
        transparent Python function whose globals and code file match its own
        provenance-verified approved module. Runtime replacements therefore
        recurse or fail rather than inheriting dispatcher trust.
        """
        name = getattr(value, "__name__", None)
        if not (
            isinstance(name, str)
            and trusted_module_namespace(parent_module)
            and vars(parent_module).get(name) is value
        ):
            return None
        wrapped = getattr(value, "__wrapped__", None)
        if not inspect.isfunction(wrapped):
            return None
        defining_module = sys.modules.get(str(getattr(wrapped, "__module__", "")))
        if not (
            trusted_module_namespace(defining_module)
            and getattr(wrapped, "__globals__", None) is vars(defining_module)
        ):
            return None
        defining_file = vars(defining_module).get("__file__")
        try:
            same_file = (
                Path(defining_file).expanduser().resolve(strict=True)
                == Path(wrapped.__code__.co_filename).expanduser().resolve(strict=True)
            )
        except (OSError, TypeError, ValueError):
            same_file = False
        if not same_file:
            return None
        return _scientific_result_digest(
            {
                "owner_sha256": module_implementation_digest(parent_module),
                "defining_module_sha256": module_implementation_digest(defining_module),
                "type": f"{type(value).__module__}.{type(value).__qualname__}",
                "module": getattr(value, "__module__", None),
                "qualified_name": getattr(value, "__qualname__", name),
                "wrapped_sha256": callable_digest(wrapped),
            }
        )

    # Approved modules are mutable Python objects. Trace the exact attribute
    # chain used by the caller rather than trusting the top-level module name:
    # runtime replacement or addition must either recurse transparently,
    # receive an explicit scientific role, or fail closed.
    def walk_trusted_module(
        module: Any,
        path: str,
        scope: str,
        function: Callable[..., Any],
        dependency_name: str,
        depth: int,
    ) -> None:
        """
        Inspect every statically selected attribute of an approved module.

        Parameters
        ----------
        module : module
            Exact approved top-level module object loaded by ``function``.
        path, scope : str
            Fully qualified evidence path and closure/global scope label.
        function : callable
            Transparent function whose bytecode selects module attributes.
        dependency_name : str
            Name used by ``function`` to load the top-level module.
        depth : int
            Current recursive dependency depth.

        Returns
        -------
        None
            Appends module, attribute and recursively reached helper evidence
            to the enclosing trace.

        Raises
        ------
        IngestError
            If dispatch is dynamic, a selected member is absent, a nested
            module is not the exact live approved namespace, or a replacement
            member is opaque/unclassified.
        """
        if depth > 64:
            raise IngestError("Module dependency graph exceeds the safe recursion depth.")
        if not trusted_module_object(module):
            raise IngestError(f"Dependency {path!r} is not an approved module object.")
        entries.append(
            {
                "scope": scope,
                "name": f"{path}.__module_implementation__",
                "role": "infrastructure_trusted_module",
                "sha256": module_implementation_digest(module),
            }
        )
        for chain in sorted(module_attribute_chains(function, dependency_name)):
            visit_key = (id(module), chain)
            if visit_key in visited_trusted_module_chains:
                continue
            visited_trusted_module_chains.add(visit_key)
            current: Any = module
            current_path = path
            parent_module = module
            for position, attribute_name in enumerate(chain):
                if not inspect.ismodule(current):
                    raise IngestError(
                        f"Trusted-module chain {current_path!r} dispatches further "
                        "through a non-module object and is not statically auditable."
                    )
                namespace = vars(current)
                resolved_dynamically = False
                if attribute_name not in namespace:
                    # A small number of installed libraries expose documented
                    # re-exports lazily (for example
                    # ``pandas.util.hash_pandas_object``). Resolve such a name
                    # only provisionally; below it must prove exact installed
                    # function/class/module provenance. Arbitrary dynamic data
                    # and caller helpers remain forbidden.
                    try:
                        value = getattr(current, attribute_name)
                    except (AttributeError, RuntimeError) as exc:
                        raise IngestError(
                            f"Trusted module {current_path!r} cannot resolve "
                            f"attribute {attribute_name!r}."
                        ) from exc
                    resolved_dynamically = True
                else:
                    value = namespace[attribute_name]
                parent_module = current
                current_path = f"{current_path}.{attribute_name}"
                final = position == len(chain) - 1
                if any(value is candidate for candidate in forbidden_reflection_objects):
                    raise IngestError(
                        f"Trusted-module attribute {current_path!r} is a "
                        "caller-namespace/reflection evaluator and cannot prove "
                        "static scientific non-reachability."
                    )
                if resolved_dynamically:
                    dynamic_provenance_holds = bool(
                        (inspect.isfunction(value) and exact_trusted_library_function(value))
                        or (inspect.isclass(value) and exact_trusted_library_class(value))
                        or (inspect.ismodule(value) and trusted_module_namespace(value))
                        or (
                            (inspect.isbuiltin(value) or isinstance(value, np.ufunc))
                            and trusted_module_namespace(
                                sys.modules.get(str(getattr(value, "__module__", "")))
                            )
                        )
                        or trusted_wrapped_callable_digest(value, parent_module) is not None
                    )
                    if not dynamic_provenance_holds:
                        raise IngestError(
                            f"Trusted module {current_path!r} resolves through an "
                            "opaque or unregistered dynamic attribute."
                        )
                if not final:
                    if not inspect.ismodule(value) or not trusted_module_namespace(value):
                        raise IngestError(
                            f"Intermediate trusted-module attribute {current_path!r} "
                            "is not an exact approved live module namespace."
                        )
                    entries.append(
                        {
                            "scope": "module_attribute",
                            "name": current_path,
                            "role": "infrastructure_trusted_module",
                            "sha256": module_implementation_digest(value),
                        }
                    )
                    current = value
                    continue

                declared = declared_roles.get(
                    current_path,
                    declared_roles.get(attribute_name),
                )
                if inspect.isfunction(value):
                    entries.append(
                        {
                            "scope": "module_attribute",
                            "name": current_path,
                            "role": "infrastructure_helper",
                            "sha256": callable_digest(value),
                        }
                    )
                    if not exact_trusted_library_function(value):
                        walk(value, current_path, depth + 1)
                    continue
                if inspect.isclass(value):
                    if exact_trusted_library_class(value):
                        entries.append(
                            {
                                "scope": "module_attribute",
                                "name": current_path,
                                "role": "infrastructure_library_class",
                                "sha256": trusted_library_class_digest(
                                    value, parent_module
                                ),
                            }
                        )
                    else:
                        try:
                            walk_class(
                                value, current_path, "module_attribute", depth + 1
                            )
                        except IngestError as exc:
                            raise IngestError(
                                f"Runtime class attribute {current_path!r} lacks "
                                f"trusted source provenance and is not safely "
                                f"traversable: {exc}"
                            ) from exc
                    continue
                if inspect.ismodule(value):
                    if not trusted_module_namespace(value):
                        raise IngestError(
                            f"Trusted-module attribute {current_path!r} is not an "
                            "exact approved live module namespace."
                        )
                    entries.append(
                        {
                            "scope": "module_attribute",
                            "name": current_path,
                            "role": "infrastructure_trusted_module",
                            "sha256": module_implementation_digest(value),
                        }
                    )
                    continue
                if inspect.isbuiltin(value) or isinstance(value, np.ufunc):
                    try:
                        native_digest = trusted_native_attribute_digest(
                            value, parent_module
                        )
                    except IngestError as exc:
                        raise IngestError(
                            f"Native runtime attribute {current_path!r} lacks exact "
                            f"trusted-library provenance: {exc}"
                        ) from exc
                    entries.append(
                        {
                            "scope": "module_attribute",
                            "name": current_path,
                            "role": "infrastructure_native_callable",
                            "sha256": native_digest,
                        }
                    )
                    continue
                if callable(value):
                    wrapped_digest = trusted_wrapped_callable_digest(
                        value, parent_module
                    )
                    if wrapped_digest is not None:
                        entries.append(
                            {
                                "scope": "module_attribute",
                                "name": current_path,
                                "role": "infrastructure_library_dispatcher",
                                "sha256": wrapped_digest,
                            }
                        )
                        continue
                    raise IngestError(
                        f"Trusted-module attribute {current_path!r} is an opaque "
                        "runtime callable replacement."
                    )
                if deeply_immutable(value):
                    role = "infrastructure_module_constant"
                else:
                    if declared is None:
                        raise IngestError(
                            f"Trusted-module attribute {current_path!r} is mutable "
                            "and unclassified."
                        )
                    role = str(declared)
                    if role not in allowed_roles:
                        raise IngestError(
                            f"Trusted-module attribute {current_path!r} has unsupported "
                            f"role {role!r}."
                        )
                entries.append(
                    {
                        "scope": "module_attribute",
                        "name": current_path,
                        "role": role,
                        "sha256": _scientific_result_digest(value),
                    }
                )

    # Dataclass decorators synthesize executable helpers outside the class
    # source (including Python 3.14 ``__replace__`` and repr wrappers). They
    # implement the already source-bound field contract and are not dispatch
    # authored by the scientific module; explicit in-class methods remain.
    def generated_dataclass_method(
        owner: type, member_name: str, function: Callable[..., Any]
    ) -> bool:
        """Return whether ``function`` matches a known dataclass-generated helper."""
        if not is_dataclass(owner):
            return False
        code = function.__code__
        closure = inspect.getclosurevars(function)
        global_names = set(closure.globals)
        nonlocal_names = set(closure.nonlocals)
        if member_name in {"__annotate__", "__annotate_func__"}:
            return bool(
                function.__module__ == owner.__module__
                and not global_names
                and nonlocal_names <= {"__classdict__"}
            )
        if member_name == "__replace__":
            return bool(
                function.__module__ == "dataclasses"
                and Path(code.co_filename).name == "dataclasses.py"
                and global_names
                <= {"MISSING", "_FIELD_CLASSVAR", "_FIELDS", "_FIELD_INITVAR"}
                and not nonlocal_names
            )
        if member_name == "__repr__":
            return bool(
                Path(code.co_filename).name == "reprlib.py"
                and global_names <= {"get_ident"}
                and nonlocal_names <= {"fillvalue", "repr_running", "user_function"}
            )
        if member_name in {"__hash__", "__eq__", "__lt__", "__le__", "__gt__", "__ge__"}:
            return bool(
                code.co_filename == "<string>"
                and function.__module__ == owner.__module__
                and not global_names
                and not nonlocal_names
            )
        if member_name == "__init__":
            return bool(
                code.co_filename == "<string>"
                and function.__module__ == owner.__module__
                and not global_names
                and nonlocal_names
                <= {"__dataclass_builtins_object__", "__dataclass_HAS_DEFAULT_FACTORY__"}
            )
        if member_name in {"__setattr__", "__delattr__"}:
            return bool(
                code.co_filename == "<string>"
                and function.__module__ == owner.__module__
                and not global_names
                and nonlocal_names == {"FrozenInstanceError", "cls"}
            )
        return False

    # Canonicalize annotation/dataclass metadata without trusting mutable
    # dictionaries merely because Python generated their attribute names.
    def metadata_digest(value: Any, active: Optional[Set[int]] = None) -> str:
        """Return a deterministic digest for class annotation/dataclass metadata."""
        seen = set() if active is None else active
        if value is None or isinstance(value, (str, bytes, int, float, bool, np.generic)):
            return _scientific_result_digest(value)
        if isinstance(value, type):
            return hashlib.sha256(
                f"type:{value.__module__}.{value.__qualname__}".encode("utf-8")
            ).hexdigest()
        if id(value) in seen:
            raise IngestError("Class metadata contains a recursive mutable structure.")
        if isinstance(value, Mapping):
            seen.add(id(value))
            digest = hashlib.sha256(b"class-metadata-mapping")
            for key in sorted(value, key=repr):
                digest.update(metadata_digest(key, seen).encode("ascii"))
                digest.update(metadata_digest(value[key], seen).encode("ascii"))
            seen.remove(id(value))
            return digest.hexdigest()
        if isinstance(value, (tuple, list, set, frozenset)):
            seen.add(id(value))
            items = list(value)
            if isinstance(value, (set, frozenset)):
                items = sorted(items, key=repr)
            digest = hashlib.sha256(type(value).__name__.encode("ascii"))
            for item in items:
                digest.update(metadata_digest(item, seen).encode("ascii"))
            seen.remove(id(value))
            return digest.hexdigest()
        if inspect.isfunction(value):
            return callable_digest(value)
        if inspect.isbuiltin(value):
            return hashlib.sha256(
                f"builtin:{value.__module__}.{value.__qualname__}".encode("utf-8")
            ).hexdigest()
        if type(value).__module__ == "dataclasses" and type(value).__name__ == "Field":
            seen.add(id(value))
            digest = metadata_digest(
                {
                    name: getattr(value, name)
                    for name in (
                        "name", "type", "default", "default_factory", "repr",
                        "hash", "init", "compare", "metadata", "kw_only",
                    )
                },
                seen,
            )
            seen.remove(id(value))
            return digest
        if type(value).__module__ == "dataclasses":
            public_scalars = {
                name: getattr(value, name)
                for name in dir(value)
                if not name.startswith("_")
                and not callable(getattr(value, name))
            }
            if public_scalars:
                return metadata_digest(public_scalars, seen)
            return hashlib.sha256(
                f"dataclasses:{type(value).__qualname__}".encode("utf-8")
            ).hexdigest()
        raise IngestError(
            f"Class metadata type {type(value).__name__!r} has no deterministic digest."
        )

    # ``__dict__`` and the default ``__weakref__`` are C-level get/set
    # descriptors, not caller data. Recognize only their exact structural form;
    # if either name is rebound to mutable content it is audited normally.
    def structural_namespace_descriptor(
        owner: type, member_name: str, member: Any
    ) -> bool:
        """Return whether a namespace member is an exact built-in structural descriptor."""
        return bool(
            member_name in {"__dict__", "__weakref__"}
            and inspect.isgetsetdescriptor(member)
            and getattr(member, "__objclass__", None) is owner
        )

    # Compute a snapshot digest of the class namespace and every current method
    # implementation. This changes when a method is monkeypatched even if the
    # original source file is unchanged.
    def class_digest(class_: type) -> str:
        """Return a deterministic digest of a class, its MRO and descriptors."""
        generated_metadata = {
            "__annotations__", "__annotations_cache__", "__dataclass_fields__", "__dataclass_params__",
            "__match_args__", "__slots__", "__static_attributes__",
            "__firstlineno__",
        }
        digest = hashlib.sha256()
        digest.update(safe_class_identity(class_).encode("utf-8"))
        digest.update(
            repr(tuple(safe_class_identity(base) for base in class_.__mro__)).encode(
                "utf-8"
            )
        )
        try:
            digest.update(inspect.getsource(class_).encode("utf-8"))
        except (OSError, TypeError):
            pass
        for owner in class_.__mro__:
            owner_identity = safe_class_identity(owner)
            if owner is object or actual_builtin_class(owner):
                continue
            digest.update(f"owner:{owner_identity}".encode("utf-8"))
            for member_name, member in sorted(vars(owner).items()):
                if structural_namespace_descriptor(owner, member_name, member):
                    digest.update(
                        f"structural:{member_name}:{owner_identity}".encode(
                            "utf-8"
                        )
                    )
                    continue
                digest.update(member_name.encode("utf-8"))
                if member_name in generated_metadata:
                    # Class source and the explicit member-method digest bind
                    # this generated annotation/dataclass metadata. Mutable
                    # mappings and Field nodes are folded by value so a runtime
                    # annotation/default mutation invalidates the trace.
                    digest.update(metadata_digest(member).encode("ascii"))
                    continue
                functions_ = descriptor_functions(member_name, member)
                if functions_:
                    for label, function in functions_:
                        digest.update(label.encode("utf-8"))
                        digest.update(callable_digest(function).encode("utf-8"))
                    for state_name, state_value in descriptor_state_items(member):
                        digest.update(f"descriptor-state:{state_name}".encode("utf-8"))
                        digest.update(_scientific_result_digest(state_value).encode("ascii"))
                elif deeply_immutable(member):
                    digest.update(_scientific_result_digest(member).encode("utf-8"))
                else:
                    # Mutable members are separately role-classified by
                    # ``walk_class``; bind their current value here as well.
                    digest.update(_scientific_result_digest(member).encode("utf-8"))
        return digest.hexdigest()

    # Traverse every current method/property/descriptor and mutable class member
    # of an active class, including inherited implementations and metaclass
    # dispatch. Dynamic getattr is therefore covered by the entire namespace;
    # an opaque descriptor is rejected rather than silently treated as harmless.
    def walk_class(class_: type, path: str, scope: str, depth: int) -> None:
        """Recursively inspect one actively dispatched class dependency."""
        generated_metadata = {
            "__annotations__", "__annotations_cache__", "__dataclass_fields__", "__dataclass_params__",
            "__match_args__", "__slots__", "__static_attributes__",
            "__firstlineno__",
        }
        if depth > 64:
            raise IngestError("Class dependency graph exceeds the safe recursion depth.")
        if id(class_) in visited_classes:
            return
        visited_classes.add(id(class_))
        entries.append(
            {
                "scope": scope,
                "name": path,
                "role": "infrastructure_class",
                "sha256": class_digest(class_),
            }
        )
        for owner in class_.__mro__:
            owner_identity = safe_class_identity(owner)
            if owner is object or actual_builtin_class(owner):
                continue
            for member_name, member in sorted(vars(owner).items()):
                if structural_namespace_descriptor(owner, member_name, member):
                    continue
                if member_name in generated_metadata:
                    continue
                member_path = f"{path}.{owner_identity}.{member_name}"
                functions_ = descriptor_functions(member_name, member)
                if functions_:
                    # A custom descriptor's ``__get__`` may delegate through
                    # another method on its own class. Traverse that class as
                    # well as the protocol method itself so such dispatch
                    # cannot conceal a data-bearing global.
                    if not isinstance(
                        member, (staticmethod, classmethod, property)
                    ) and not inspect.isfunction(member):
                        walk_class(
                            type(member),
                            f"{member_path}.__descriptor_class__",
                            "descriptor_class",
                            depth + 1,
                        )
                    for label, function in functions_:
                        function_path = f"{member_path}[{label.split(':', 1)[-1]}]"
                        entries.append(
                            {
                                "scope": "class_descriptor",
                                "name": function_path,
                                "role": "infrastructure_class_descriptor",
                                "sha256": callable_digest(function),
                            }
                        )
                        if generated_dataclass_method(owner, member_name, function):
                            continue
                        walk(function, function_path, depth + 1)
                    for state_name, state_value in descriptor_state_items(member):
                        state_path = (
                            f"{member_path}.__descriptor_state__.{state_name}"
                        )
                        if deeply_immutable(state_value):
                            state_role = "infrastructure_descriptor_state"
                        else:
                            declared = declared_roles.get(
                                state_path,
                                declared_roles.get(
                                    f"{member_name}.{state_name}",
                                    declared_roles.get(state_name),
                                ),
                            )
                            if declared is None:
                                raise IngestError(
                                    f"Descriptor dependency {member_path!r} has unclassified "
                                    f"mutable state {state_name!r}."
                                )
                            state_role = str(declared)
                            if state_role not in allowed_roles:
                                raise IngestError(
                                    f"Descriptor state {state_path!r} has unsupported role "
                                    f"{state_role!r}."
                                )
                        entries.append(
                            {
                                "scope": "descriptor_state",
                                "name": state_path,
                                "role": state_role,
                                "sha256": _scientific_result_digest(state_value),
                            }
                        )
                    continue
                if inspect.isdatadescriptor(member) or hasattr(type(member), "__get__"):
                    # Built-in member descriptors occur only on trusted opaque
                    # classes; a user-class descriptor without Python code is
                    # not auditable and must fail closed.
                    raise IngestError(
                        f"Opaque class descriptor {member_path!r} cannot be inspected."
                    )
                if not deeply_immutable(member):
                    declared = declared_roles.get(
                        member_path,
                        declared_roles.get(
                            f"{owner_identity}.{member_name}",
                            declared_roles.get(member_name),
                        ),
                    )
                    if declared is None:
                        raise IngestError(
                            f"Class dependency {path!r} has unclassified mutable member {member_name!r}."
                        )
                    role = str(declared)
                    if role not in allowed_roles:
                        raise IngestError(
                            f"Class member {member_path!r} has unsupported role {role!r}."
                        )
                    entries.append(
                        {
                            "scope": "class_member",
                            "name": member_path,
                            "role": role,
                            "sha256": _scientific_result_digest(member),
                        }
                    )
        metaclass = type(class_)
        if metaclass is not type and id(metaclass) not in visited_classes:
            walk_class(metaclass, f"{path}.__class__", "metaclass", depth + 1)

    # Inspect statically selected attributes of an untrusted module object.
    # Transparent helper functions recurse through the ordinary callable walk;
    # opaque C dispatch and data-bearing module attributes fail closed.
    def walk_module(
        module: Any,
        path: str,
        scope: str,
        function: Callable[..., Any],
        dependency_name: str,
        depth: int,
    ) -> None:
        """
        Recursively inspect attributes dispatched through a module dependency.

        Parameters
        ----------
        module : module
            Non-trusted module object resolved from a function binding.
        path, scope : str
            Evidence path and closure/global scope labels.
        function : callable
            Function whose bytecode performs the module dispatch.
        dependency_name : str
            Local/global name by which ``function`` loads the module.
        depth : int
            Current recursion depth.

        Returns
        -------
        None
            Appends complete module/attribute evidence to ``entries``.

        Raises
        ------
        IngestError
            If dispatch is dynamic, an attribute is absent or opaque, or a
            mutable module value lacks an explicit scientific role.
        """
        if depth > 64:
            raise IngestError("Module dependency graph exceeds the safe recursion depth.")
        if not inspect.ismodule(module):
            raise IngestError(f"Dependency {path!r} is not an inspectable module.")
        selected = module_attribute_names(function, dependency_name)
        already = visited_module_attributes.setdefault(id(module), set())
        pending = sorted(selected - already)
        already.update(selected)
        entries.append(
            {
                "scope": scope,
                "name": f"{path}.__module_implementation__",
                "role": "infrastructure_module",
                "sha256": module_implementation_digest(module),
            }
        )
        namespace = vars(module)
        for attribute_name in pending:
            attribute_path = f"{path}.{attribute_name}"
            if attribute_name not in namespace:
                raise IngestError(
                    f"Module dependency {path!r} resolves {attribute_name!r} "
                    "through opaque module-level dynamic dispatch."
                )
            value = namespace[attribute_name]
            declared = declared_roles.get(
                attribute_path,
                declared_roles.get(attribute_name),
            )
            if inspect.isfunction(value):
                entries.append(
                    {
                        "scope": "module_attribute",
                        "name": attribute_path,
                        "role": "infrastructure_helper",
                        "sha256": callable_digest(value),
                    }
                )
                if not exact_trusted_library_function(value):
                    walk(value, attribute_path, depth + 1)
                continue
            if inspect.isclass(value):
                walk_class(value, attribute_path, "module_attribute", depth + 1)
                continue
            if inspect.ismodule(value):
                raise IngestError(
                    f"Nested module dependency {attribute_path!r} reached through "
                    "an untrusted module is opaque."
                )
            if inspect.isbuiltin(value) or inspect.ismethoddescriptor(value):
                raise IngestError(
                    f"Untrusted module dependency {attribute_path!r} dispatches "
                    "through an opaque native callable."
                )
            if callable(value):
                raise IngestError(
                    f"Untrusted module dependency {attribute_path!r} exposes an "
                    "opaque callable object."
                )
            if deeply_immutable(value):
                role = "infrastructure_module_constant"
            else:
                if declared is None:
                    raise IngestError(
                        f"Module dependency {attribute_path!r} is mutable and unclassified."
                    )
                role = str(declared)
                if role not in allowed_roles:
                    raise IngestError(
                        f"Module dependency {attribute_path!r} has unsupported role {role!r}."
                    )
            entries.append(
                {
                    "scope": "module_attribute",
                    "name": attribute_path,
                    "role": role,
                    "sha256": _scientific_result_digest(value),
                }
            )

    # Nested lambdas, comprehensions and local functions execute code objects
    # stored in ``co_consts``. Their global loads are absent from
    # ``inspect.getclosurevars`` on the outer callable, so classify them
    # explicitly rather than treating their bytecode digest as sufficient.
    def audit_nested_code_objects(
        function: Callable[..., Any], path: str, depth: int
    ) -> None:
        """
        Trace globals and module dispatch reachable from nested code objects.

        Parameters
        ----------
        function : callable
            Transparent outer Python function owning the nested code.
        path : str
            Fully qualified dependency path of the outer function.
        depth : int
            Current recursive dependency depth.

        Returns
        -------
        None
            Appends nested implementation and dependency evidence.

        Raises
        ------
        IngestError
            If nested execution uses reflection, caller-namespace evaluators,
            opaque dispatch, or mutable scientific state without a role.
        """
        nested: List[Any] = []

        def collect(code: Any) -> None:
            """Collect every descendant code object exactly once by identity."""
            for constant in code.co_consts:
                if inspect.iscode(constant):
                    nested.append(constant)
                    collect(constant)

        collect(function.__code__)
        for ordinal, code in enumerate(nested):
            nested_path = f"{path}.<nested:{code.co_name}:{ordinal}>"
            reject_dynamic_namespace_dispatch(code, nested_path)
            digest = hashlib.sha256()
            code_digest(code, digest)
            entries.append(
                {
                    "scope": "nested_code",
                    "name": f"{nested_path}.__implementation__",
                    "role": "infrastructure_callable",
                    "sha256": digest.hexdigest(),
                }
            )
            loaded_names = sorted(
                {
                    str(instruction.argval)
                    for instruction in dis.get_instructions(code)
                    if instruction.opname in {"LOAD_GLOBAL", "LOAD_NAME"}
                }
            )
            for name in loaded_names:
                qualified = f"{nested_path}.{name}"
                if name in function.__globals__:
                    value = function.__globals__[name]
                    declared = declared_roles.get(qualified, declared_roles.get(name))
                    if any(
                        value is candidate
                        for candidate in forbidden_reflection_objects
                    ):
                        raise IngestError(
                            f"Nested callable {nested_path!r} resolves forbidden "
                            f"reflection/caller-namespace object {name!r}."
                        )
                    if inspect.isfunction(value):
                        entries.append(
                            {
                                "scope": "nested_global",
                                "name": qualified,
                                "role": "infrastructure_helper",
                                "sha256": callable_digest(value),
                            }
                        )
                        if not exact_trusted_library_function(value):
                            walk(value, qualified, depth + 1)
                        continue
                    if inspect.isclass(value):
                        if actual_builtin_class(value) or trusted_opaque_class(value):
                            entries.append(
                                {
                                    "scope": "nested_global",
                                    "name": qualified,
                                    "role": "infrastructure_class",
                                    "sha256": hashlib.sha256(
                                        safe_class_identity(value).encode("utf-8")
                                    ).hexdigest(),
                                }
                            )
                        else:
                            walk_class(value, qualified, "nested_global", depth + 1)
                        continue
                    if inspect.ismodule(value):
                        if trusted_module_object(value):
                            walk_trusted_module(
                                value,
                                qualified,
                                "nested_global",
                                code,
                                name,
                                depth + 1,
                            )
                        else:
                            walk_module(
                                value,
                                qualified,
                                "nested_global",
                                code,
                                name,
                                depth + 1,
                            )
                        continue
                    if inspect.isbuiltin(value):
                        entries.append(
                            {
                                "scope": "nested_global",
                                "name": qualified,
                                "role": "infrastructure",
                                "sha256": hashlib.sha256(
                                    f"{value.__module__}.{value.__name__}".encode("utf-8")
                                ).hexdigest(),
                            }
                        )
                        continue
                    if callable(value):
                        raise IngestError(
                            f"Nested callable {nested_path!r} resolves opaque "
                            f"callable-object dependency {name!r}."
                        )
                    if deeply_immutable(value):
                        role = "infrastructure_constant"
                    else:
                        if declared is None:
                            raise IngestError(
                                f"Nested callable {nested_path!r} has unclassified "
                                f"global dependency {name!r}."
                            )
                        role = str(declared)
                        if role not in allowed_roles:
                            raise IngestError(
                                f"Nested dependency {qualified!r} has unsupported "
                                f"role {role!r}."
                            )
                    entries.append(
                        {
                            "scope": "nested_global",
                            "name": qualified,
                            "role": role,
                            "sha256": _scientific_result_digest(value),
                        }
                    )
                    continue
                builtin_value = getattr(builtins, name, None)
                if builtin_value is None:
                    raise IngestError(
                        f"Nested callable {nested_path!r} resolves unknown global "
                        f"name {name!r}."
                    )
                if any(
                    builtin_value is candidate
                    for candidate in forbidden_reflection_objects
                ):
                    raise IngestError(
                        f"Nested callable {nested_path!r} uses forbidden built-in "
                        f"{name!r}."
                    )
                entries.append(
                    {
                        "scope": "nested_builtin",
                        "name": qualified,
                        "role": "infrastructure",
                        "sha256": hashlib.sha256(
                            f"builtins.{name}".encode("utf-8")
                        ).hexdigest(),
                    }
                )

    # Walk one production callable and all transparent helpers it invokes,
    # regardless of their defining module. Exact trusted library exports are
    # source-bound exemptions; all other Python helpers recurse.
    def walk(function: Callable[..., Any], path: str, depth: int) -> None:
        """Recursively inspect one transparent Python function dependency."""
        if depth > 64:
            raise IngestError("Callable dependency graph exceeds the safe recursion depth.")
        if id(function) in visited_functions:
            return
        if not inspect.isfunction(function):
            raise IngestError(f"Opaque callable dependency {path!r} cannot be inspected.")
        visited_functions.add(id(function))
        reject_dynamic_namespace_dispatch(function, path)
        # Bind the exact executable implementation at every recursion node,
        # including the registered root. A code edit that preserves the same
        # global names must still invalidate the pinned dependency witness.
        entries.append(
            {
                "scope": "callable",
                "name": f"{path}.__implementation__",
                "role": "infrastructure_callable",
                "sha256": callable_digest(function),
            }
        )
        entries.extend(audit_function_defaults(function, path))
        entries.extend(audit_function_metadata(function, path))
        try:
            closure = inspect.getclosurevars(function)
        except (TypeError, ValueError) as exc:
            raise IngestError(f"Cannot inspect callable dependencies: {exc}") from exc
        # ``inspect`` reports attribute names (for example DataFrame columns
        # and methods) as unbound on current Python releases.  They are bound
        # through already traced objects, not hidden globals, so only resolved
        # closure/global objects are classified below.
        for scope, values in (("nonlocal", closure.nonlocals), ("global", closure.globals)):
            for name in sorted(values):
                value = values[name]
                qualified = f"{path}.{name}"
                declared = declared_roles.get(qualified, declared_roles.get(name))
                if any(value is candidate for candidate in forbidden_reflection_objects):
                    raise IngestError(
                        f"Callable {path!r} resolves opaque namespace/reflection "
                        f"dispatch through alias {name!r}; aliases of "
                        f"{sorted(forbidden_reflection_names)!r} are forbidden."
                    )
                if inspect.isfunction(value):
                    entries.append(
                        {
                            "scope": scope,
                            "name": qualified,
                            "role": "infrastructure_helper",
                            "sha256": callable_digest(value),
                        }
                    )
                    if not exact_trusted_library_function(value):
                        walk(value, qualified, depth + 1)
                    continue
                if inspect.isclass(value):
                    mode = class_reference_mode(function, name, value)
                    identity = safe_class_identity(value)
                    if mode == "inert":
                        entries.append(
                            {
                                "scope": scope,
                                "name": qualified,
                                "role": "infrastructure_inert_class",
                                "sha256": hashlib.sha256(identity.encode("utf-8")).hexdigest(),
                            }
                        )
                    elif trusted_opaque_class(value):
                        entries.append(
                            {
                                "scope": scope,
                                "name": qualified,
                                "role": "infrastructure_trusted_opaque_class",
                                "sha256": hashlib.sha256(
                                    f"{identity}|python={sys.version_info[:3]}".encode("utf-8")
                                ).hexdigest(),
                            }
                        )
                    else:
                        walk_class(value, qualified, scope, depth + 1)
                    continue
                if inspect.ismodule(value):
                    if trusted_module_object(value):
                        walk_trusted_module(
                            value,
                            qualified,
                            scope,
                            function,
                            name,
                            depth + 1,
                        )
                    else:
                        walk_module(
                            value,
                            qualified,
                            scope,
                            function,
                            name,
                            depth + 1,
                        )
                    continue
                if (
                    inspect.isbuiltin(value)
                    or isinstance(value, logging.Logger)
                    or isinstance(value, re.Pattern)
                    or str(type(value).__module__) in {"typing", "typing_extensions"}
                ):
                    entries.append(
                        {
                            "scope": scope,
                            "name": qualified,
                            "role": "infrastructure",
                            "sha256": hashlib.sha256(
                                f"{type(value).__module__}.{getattr(value, '__name__', type(value).__qualname__)}".encode("utf-8")
                            ).hexdigest(),
                        }
                    )
                    continue
                if callable(value):
                    raise IngestError(
                        f"Callable {path!r} resolves opaque callable-object "
                        f"dependency {name!r}."
                    )
                immutable_constant = deeply_immutable(value)
                if immutable_constant:
                    role = "infrastructure_constant"
                else:
                    if declared is None:
                        raise IngestError(
                            f"Callable {path!r} has unclassified {scope} dependency {name!r}."
                        )
                    role = str(declared)
                    if role not in allowed_roles:
                        raise IngestError(
                            f"Callable dependency {qualified!r} has unsupported role {role!r}."
                        )
                entries.append(
                    {
                        "scope": scope,
                        "name": qualified,
                        "role": role,
                        "sha256": _scientific_result_digest(value),
                    }
                )
        audit_nested_code_objects(function, path, depth)

    walk(callable_, str(getattr(callable_, "__qualname__", callable_.__name__)), 0)
    return entries


# Verify actual baseline snapshots and robustness result objects.
def _audit_robustness_snapshot_evidence(robustness_frame: pd.DataFrame) -> Dict[str, Any]:
    """
    Bind robustness rows to actual Task-17 objects and Task-18 executions.

    Parameters
    ----------
    robustness_frame : pandas.DataFrame
        Exact estimated or verified-identity three-axis Task-17 subset with retained baseline,
        variant-input and result objects plus live Task-17/18 stage objects.

    Returns
    -------
    dict
        Baseline-before/after verdict, run universe and content digests for the
        retained variants and both source stages.

    Raises
    ------
    IngestError
        If identities are duplicated, retained objects are detached, Task-18
        estimates disagree, a digest is malformed/stale, or baseline state was
        mutated by any retained variant.
    """
    if not isinstance(robustness_frame, pd.DataFrame) or len(robustness_frame) == 0:
        raise IngestError("The robustness invariant requires at least one executed run.")
    required = (
        "run_id", "snapshot_id", "baseline_before_sha256",
        "baseline_after_sha256", "variant_input_sha256", "result_sha256",
        "execution_status",
    )
    missing = [name for name in required if name not in robustness_frame.columns]
    if missing:
        raise IngestError(f"The robustness evidence lacks field(s) {missing}.")
    if bool(robustness_frame["run_id"].astype(str).duplicated().any()):
        raise IngestError("Robustness evidence carries duplicate run identifiers.")
    snapshots = robustness_frame.attrs.get("baseline_snapshot_objects")
    results = robustness_frame.attrs.get("variant_result_objects")
    variant_inputs = robustness_frame.attrs.get("variant_input_objects")
    task18_object = robustness_frame.attrs.get("task18_stage_output_object")
    task18_digest = str(robustness_frame.attrs.get("task18_stage_output_sha256", ""))
    task17_object = robustness_frame.attrs.get("task17_stage_output_object")
    task17_digest = str(robustness_frame.attrs.get("task17_stage_output_sha256", ""))
    if not isinstance(snapshots, Mapping) or not isinstance(results, Mapping) or not isinstance(variant_inputs, Mapping):
        raise IngestError(
            "Robustness evidence must retain actual baseline snapshot, variant input "
            "and variant result objects."
        )
    if (
        task18_object is None
        or not re.fullmatch(r"[0-9a-f]{64}", task18_digest)
        or _pipeline_artifact_digest(task18_object) != task18_digest
    ):
        raise IngestError("Robustness evidence is not bound to the exact live Task-18 output.")
    if (
        task17_object is None
        or not re.fullmatch(r"[0-9a-f]{64}", task17_digest)
        or _pipeline_artifact_digest(task17_object) != task17_digest
        or not hasattr(task17_object, "results")
    ):
        raise IngestError("Robustness evidence is not bound to the exact live Task-17 suite.")
    for attr_name, observed in (
        ("baseline_snapshot_objects", snapshots),
        ("variant_result_objects", results),
        ("variant_input_objects", variant_inputs),
    ):
        if task17_object.results.attrs.get(attr_name) is not observed:
            raise IngestError(
                f"Robustness evidence {attr_name!r} is detached from the live Task-17 suite."
            )
    retained_run_ids = set(robustness_frame["run_id"].astype(str))
    expected_task17 = task17_object.results.reset_index()
    expected_task17 = expected_task17[
        expected_task17["status"].astype(str).isin({"success", "identity"})
        & expected_task17["run_id"].astype(str).isin(retained_run_ids)
    ].reset_index(drop=True)
    try:
        pd.testing.assert_frame_equal(
            robustness_frame.sort_values("run_id").reset_index(drop=True),
            expected_task17.sort_values("run_id").reset_index(drop=True),
            check_exact=True,
            check_like=False,
        )
    except AssertionError as exc:
        raise IngestError("Robustness rows are detached from the live Task-17 result table.") from exc
    task18_rows: List[pd.DataFrame] = []
    for axis, field_name in (
        (AXIS_REFERENCE_WINDOW, "reference_table"),
        (AXIS_DONOR_POOL, "donor_table"),
        (AXIS_FACTOR_PROXY, "factor_table"),
    ):
        table = getattr(task18_object, field_name, None)
        if not isinstance(table, pd.DataFrame) or table.empty:
            raise IngestError(f"Live Task-18 output lacks its {field_name!r} table.")
        material = table.reset_index().copy()
        # The containing frozen Task-18 object already binds frame attrs. Clear
        # them on these temporary projections so pandas never attempts an
        # elementwise equality comparison of nested DataFrames during concat.
        material.attrs = {}
        material["axis"] = axis
        task18_rows.append(material)
    task18_material = pd.concat(task18_rows, ignore_index=True, sort=False)
    if "run_id" not in task18_material.columns or task18_material["run_id"].astype(str).duplicated().any():
        raise IngestError("Live Task-18 output has missing or duplicate run identities.")
    if set(task18_material["run_id"].astype(str)) != retained_run_ids:
        raise IngestError("Retained robustness rows do not equal the live Task-18 run universe.")
    task18_by_run = task18_material.assign(
        run_id=task18_material["run_id"].astype(str)
    ).set_index("run_id", verify_integrity=True)
    for _, retained in robustness_frame.iterrows():
        run_id = str(retained["run_id"])
        executed = task18_by_run.loc[run_id]
        retained_status = str(retained["status"])
        coverage_key = (
            str(retained["policy"]), str(retained["outcome"]),
            str(retained["axis"]), str(retained["variant_name"]),
        )
        try:
            coverage_status = str(task18_object.coverage_audit.loc[
                coverage_key, "execution_status"
            ])
        except (KeyError, AttributeError, TypeError) as exc:
            raise IngestError(
                f"Robustness run {run_id!r} lacks Task-18 coverage evidence."
            ) from exc
        if (
            retained_status not in {"success", "identity"}
            or str(retained["execution_status"]) != retained_status
            or str(executed["status"]) != retained_status
            or str(executed["execution_status"]) != retained_status
            or coverage_status != retained_status
        ):
            raise IngestError(
                f"Robustness run {run_id!r} disagrees with Task-17/18 execution status."
            )
        if retained_status == "identity":
            key = coverage_key
            if key[2:] != (AXIS_DONOR_POOL, "no_repaired_main_outcome_cells"):
                raise IngestError(
                    f"Robustness run {run_id!r} has an unsupported identity variant."
                )
            try:
                design = json.loads(
                    str(task17_object.effective_designs.loc[key, "effective_design_json"])
                )
                donor_audit = task18_object.donor_audit.loc[
                    (key[0], key[1], key[3])
                ]
                audited_removed = json.loads(str(donor_audit["removed_donors_json"]))
            except (KeyError, AttributeError, TypeError, json.JSONDecodeError) as exc:
                raise IngestError(
                    f"Robustness run {run_id!r} lacks verified identity evidence."
                ) from exc
            if (
                design.get("effective_identity") is not True
                or design.get("effective_design_changed") is not False
                or design.get("removed_donors") != []
                or design.get("effective_donors") != design.get("baseline_donors")
                or not re.fullmatch(
                    r"[0-9a-f]{64}", str(design.get("baseline_lineage_sha256", ""))
                )
                or str(donor_audit["audit_status"]) != "verified_identity"
                or bool(donor_audit["restriction_verified"])
                or audited_removed != []
            ):
                raise IngestError(
                    f"Robustness run {run_id!r} has contradictory identity evidence."
                )
        for identity in ("policy", "outcome", "axis", "variant_name"):
            if str(retained[identity]) != str(executed[identity]):
                raise IngestError(
                    f"Robustness run {run_id!r} disagrees with Task-18 {identity!r}."
                )
        for scalar in ("att", "se_cd"):
            if float(retained[scalar]) != float(executed[scalar]):
                raise IngestError(
                    f"Robustness run {run_id!r} disagrees with Task-18 {scalar!r}."
                )
        if str(retained["stars"]) != str(executed["stars"]):
            raise IngestError(f"Robustness run {run_id!r} disagrees with Task-18 stars.")
    before_digests: List[str] = []
    after_digests: List[str] = []
    variant_digests: List[str] = []
    holds = True
    for _, row in robustness_frame.iterrows():
        run_id = str(row["run_id"])
        snapshot_id = str(row["snapshot_id"])
        if snapshot_id not in snapshots or run_id not in results or run_id not in variant_inputs:
            raise IngestError(f"Robustness run {run_id!r} lacks its retained actual object.")
        snapshot = snapshots[snapshot_id]
        if not isinstance(snapshot, Mapping) or "before" not in snapshot or "after" not in snapshot:
            raise IngestError(f"Baseline snapshot {snapshot_id!r} is malformed.")
        before = _content_digest(snapshot["before"])
        after = _content_digest(snapshot["after"])
        result = _content_digest(results[run_id])
        variant_input = _content_digest(variant_inputs[run_id])
        if id(snapshot["before"]) == id(snapshot["after"]):
            raise IngestError(
                f"Baseline snapshot {snapshot_id!r} aliases before and after objects."
            )
        before_digests.append(before)
        after_digests.append(after)
        variant_digests.append(str(row["variant_input_sha256"]))
        for digest in (
            str(row["baseline_before_sha256"]), str(row["baseline_after_sha256"]),
            str(row["variant_input_sha256"]), str(row["result_sha256"]),
        ):
            if not re.fullmatch(r"[0-9a-f]{64}", digest):
                raise IngestError(f"Robustness run {run_id!r} carries a malformed digest.")
        row_holds = (
            before == str(row["baseline_before_sha256"])
            and after == str(row["baseline_after_sha256"])
            and result == str(row["result_sha256"])
            and variant_input == str(row["variant_input_sha256"])
            and before == after
            and str(row["execution_status"]) in {"success", "identity"}
            and str(results[run_id].get("scientific_status", "")) == str(row["status"])
        )
        holds = holds and row_holds
    return {
        "holds": bool(holds),
        "variant_count": int(len(robustness_frame)),
        "before_digest": _content_digest(tuple(before_digests)),
        "after_digest": _content_digest(tuple(after_digests)),
        "variant_input_digest": _content_digest(tuple(variant_digests)),
        "run_ids": sorted(robustness_frame["run_id"].astype(str).tolist()),
        "task17_stage_output_sha256": task17_digest,
        "task18_stage_output_sha256": task18_digest,
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 32, Step 3: The provenance, the gaps and the archive manifest
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the provenance restatement of the declared inputs.
def build_provenance(
    study_config: Mapping[str, Any],
    declared_facts: Optional[Mapping[str, Mapping[str, Any]]] = None,
) -> pd.DataFrame:
    """
    Compose the provenance restatement of the declared inputs.

    Parameters
    ----------
    study_config : mapping
        The study configuration, from which the declared seed material is read
        rather than transcribed.
    declared_facts : mapping, optional
        Required facts bound to actual source objects or files, versions,
        relevant keys, verified content digests, transformations and parents.
        A supplied assertion without an observable source remains explicitly
        unresolved; it is never relabelled as verified.

    Returns
    -------
    pandas.DataFrame
        One row per fact carrying its value and the artefact it was read from,
        so a fact's provenance is the artefact that carries it rather than a
        sentence in the prose.

    Raises
    ------
    IngestError
        If an observed source is tampered with, its digest is malformed or its
        relevant key contradicts the claimed value.
    """
    # Accumulate the fact rows.
    rows: List[dict] = []
    # Read the seed material from the actual configuration object.
    seeds = (
        study_config.get("reproducibility_gaps", {}).get("random_seeds", {})
        if isinstance(study_config, Mapping)
        else {}
    )
    config_version = (
        str(study_config.get("version", "")).strip()
        if isinstance(study_config, Mapping)
        else ""
    )
    seed_status = (
        "observed_verified"
        if isinstance(seeds, Mapping) and bool(seeds)
        and not _contains_null_or_blank(seeds)
        and bool(config_version)
        else "unresolved"
    )
    rows.append(
        {
            "fact": "random_seeds",
            "value_json": _canonical_audit_json(dict(seeds) if isinstance(seeds, Mapping) else seeds),
            "source_artifact": "study_config",
            "source_kind": "in_memory_mapping",
            "version": config_version,
            "relevant_key": "reproducibility_gaps.random_seeds",
            "content_sha256": _content_digest(study_config if isinstance(study_config, Mapping) else {}),
            "transformation": "identity",
            "parent_artifacts_json": "[]",
            "verification_status": seed_status,
            "verification_detail": (
                "read directly from the supplied study_config object"
                if seed_status == "observed_verified"
                else "seed registry or study_config version is absent, blank or null"
            ),
        }
    )
    # Verify every non-seed required fact against an actual source artefact.
    supplied = declared_facts or {}
    for name in REQUIRED_PROVENANCE_FACTS:
        if name == "random_seeds":
            continue
        if name not in supplied:
            rows.append(
                {
                    "fact": name,
                    "value_json": "null",
                    "source_artifact": "",
                    "source_kind": "missing",
                    "version": "",
                    "relevant_key": "",
                    "content_sha256": "",
                    "transformation": "",
                    "parent_artifacts_json": "[]",
                    "verification_status": "unresolved",
                    "verification_detail": "required provenance fact was not supplied",
                }
            )
            continue
        entry = supplied[name]
        rows.append(_verify_provenance_fact(name, entry))
    # Retain additional declared facts under the same verification contract.
    for name in sorted(set(supplied) - set(REQUIRED_PROVENANCE_FACTS)):
        rows.append(_verify_provenance_fact(str(name), supplied[name]))
    # Assemble the frame keyed on the fact's name.
    provenance: pd.DataFrame = pd.DataFrame(rows).set_index("fact")
    # Log the frame's extent.
    logger.info("provenance composed across %d fact(s).", len(provenance))
    # Return the frame.
    return provenance


# Test recursively whether a configuration value contains an unresolved cell.
def _contains_null_or_blank(value: Any) -> bool:
    """Return whether nested configuration material contains null/blank data."""
    if value is None:
        return True
    if isinstance(value, str):
        return not value.strip()
    if isinstance(value, Mapping):
        return any(_contains_null_or_blank(item) for item in value.values())
    if isinstance(value, (tuple, list)):
        return any(_contains_null_or_blank(item) for item in value)
    return False


# Canonically serialize audit metadata without silently dropping type labels.
def _canonical_audit_json(value: Any) -> str:
    """Serialize nested audit metadata deterministically for evidence tables."""
    if isinstance(value, Mapping):
        normalized = {str(key): json.loads(_canonical_audit_json(item)) for key, item in sorted(value.items(), key=lambda pair: str(pair[0]))}
        return json.dumps(normalized, sort_keys=True, separators=(",", ":"))
    if isinstance(value, (tuple, list)):
        return json.dumps([json.loads(_canonical_audit_json(item)) for item in value], separators=(",", ":"))
    if isinstance(value, np.generic):
        value = value.item()
    if value is None or isinstance(value, (str, int, float, bool)):
        return json.dumps(value, sort_keys=True, separators=(",", ":"), allow_nan=False)
    return json.dumps(repr(value), separators=(",", ":"))


# Hash the actual bytes of an archive or provenance file.
def _file_sha256(path: Path) -> str:
    """Return the SHA-256 digest of a nonempty regular file."""
    resolved = path.expanduser().resolve()
    if not resolved.exists() or not resolved.is_file():
        raise IngestError(f"Source artefact does not exist as a file: {resolved}.")
    if resolved.stat().st_size <= 0:
        raise IngestError(f"Source artefact is empty: {resolved}.")
    digest = hashlib.sha256()
    with resolved.open("rb") as handle:
        for block in iter(lambda: handle.read(1024 * 1024), b""):
            digest.update(block)
    return digest.hexdigest()


# Resolve a relevant key against an actual observed source object.
def _resolve_relevant_key(artifact: Any, relevant_key: Any) -> Any:
    """Resolve a dot path, DataFrame selector or sequence index from a source."""
    if isinstance(artifact, Mapping):
        current: Any = artifact
        parts = relevant_key if isinstance(relevant_key, (tuple, list)) else str(relevant_key).split(".")
        for part in parts:
            if not isinstance(current, Mapping) or part not in current:
                raise IngestError(f"Relevant source key {relevant_key!r} was not found.")
            current = current[part]
        return current
    if isinstance(artifact, pd.DataFrame):
        if isinstance(relevant_key, Mapping):
            selector = pd.Series(True, index=artifact.index)
            value_column = relevant_key.get("value_column")
            for column, value in relevant_key.items():
                if column == "value_column":
                    continue
                if column not in artifact.columns:
                    raise IngestError(f"Relevant DataFrame key column {column!r} is absent.")
                selector &= artifact[column].astype(str).eq(str(value))
            matched = artifact.loc[selector]
            if len(matched) != 1 or value_column not in artifact.columns:
                raise IngestError("Relevant DataFrame key did not resolve exactly one value.")
            return matched.iloc[0][str(value_column)]
        if relevant_key in artifact.columns:
            return artifact[relevant_key]
        raise IngestError(f"Relevant DataFrame key {relevant_key!r} was not found.")
    if isinstance(artifact, (tuple, list)):
        if isinstance(relevant_key, (bool, np.bool_)) or int(relevant_key) != float(relevant_key):
            raise IngestError("Sequence provenance key must be an integer index.")
        return artifact[int(relevant_key)]
    raise IngestError(
        f"Relevant-key verification is unsupported for {type(artifact).__name__}."
    )


# Verify one declared provenance fact against its actual source.
def _verify_provenance_fact(name: str, entry: Mapping[str, Any]) -> Dict[str, Any]:
    """Return a provenance row after digest and key/value verification."""
    required = (
        "value", "version", "relevant_key", "content_sha256",
        "transformation", "parent_artifacts",
    )
    missing = [field_name for field_name in required if field_name not in entry]
    if missing:
        raise IngestError(f"Provenance fact {name!r} lacks field(s) {missing}.")
    version = str(entry["version"]).strip()
    if not version:
        raise IngestError(f"Provenance fact {name!r} has a blank version.")
    claimed_digest = str(entry["content_sha256"])
    if not re.fullmatch(r"[0-9a-f]{64}", claimed_digest):
        raise IngestError(f"Provenance fact {name!r} has a malformed content digest.")
    if "artifact" in entry and "path" in entry:
        raise IngestError(f"Provenance fact {name!r} supplies both object and path.")
    if "artifact" in entry:
        artifact = entry["artifact"]
        actual_digest = _content_digest(artifact)
        source_kind = "in_memory_object"
        source_artifact = str(entry.get("source_artifact", name))
        key_source = artifact
    elif "path" in entry:
        path = Path(str(entry["path"])).expanduser().resolve()
        actual_digest = _file_sha256(path)
        source_kind = "file"
        source_artifact = str(path)
        suffix = path.suffix.lower()
        if suffix == ".json":
            key_source = json.loads(path.read_text(encoding="utf-8"))
        elif suffix in (".csv", ".tsv"):
            key_source = pd.read_csv(path, sep="\t" if suffix == ".tsv" else ",")
        else:
            raise IngestError(
                f"Provenance file for {name!r} must be JSON/CSV/TSV for key verification."
            )
    else:
        # Preserve, but never verify, a bare external assertion.
        return {
            "fact": name,
            "value_json": _canonical_audit_json(entry["value"]),
            "source_artifact": str(entry.get("source_artifact", entry.get("source", ""))),
            "source_kind": "supplied_assertion",
            "version": version,
            "relevant_key": _canonical_audit_json(entry["relevant_key"]),
            "content_sha256": claimed_digest,
            "transformation": str(entry["transformation"]),
            "parent_artifacts_json": _canonical_audit_json(entry["parent_artifacts"]),
            "verification_status": "supplied_unverified",
            "verification_detail": "no observable source object or file was supplied",
        }
    if actual_digest != claimed_digest:
        raise IngestError(f"Provenance fact {name!r} source digest does not match.")
    if not source_artifact.strip():
        raise IngestError(f"Provenance fact {name!r} has a blank source identifier.")
    verified_parents, parents_verified = _verify_parent_artifacts(entry["parent_artifacts"])
    observed = _resolve_relevant_key(key_source, entry["relevant_key"])
    transformation = str(entry["transformation"]).strip()
    status = "observed_verified"
    detail = "source digest, relevant key and identity value verified"
    if transformation == "identity":
        if _content_digest(observed) != _content_digest(entry["value"]):
            raise IngestError(f"Provenance fact {name!r} contradicts its source key.")
    elif callable(entry.get("transformation_callable")):
        transformed = entry["transformation_callable"](copy.deepcopy(observed))
        if _content_digest(transformed) != _content_digest(entry["value"]):
            raise IngestError(f"Provenance fact {name!r} transformation does not reproduce.")
        detail = "source digest, relevant key and executable transformation verified"
    else:
        status = "supplied_unverified"
        detail = "source/key verified but transformation was not executable"
    if not parents_verified:
        status = "supplied_unverified"
        detail += "; one or more parent links were not authenticated"
    return {
        "fact": name,
        "value_json": _canonical_audit_json(entry["value"]),
        "source_artifact": source_artifact,
        "source_kind": source_kind,
        "version": version,
        "relevant_key": _canonical_audit_json(entry["relevant_key"]),
        "content_sha256": actual_digest,
        "transformation": transformation,
        "parent_artifacts_json": _canonical_audit_json(verified_parents),
        "verification_status": status,
        "verification_detail": detail,
    }


# Authenticate the parent links of a transformed provenance fact.
def _verify_parent_artifacts(parents: Any) -> Tuple[List[Dict[str, str]], bool]:
    """Verify parent identifiers, versions and actual file/object digests."""
    if not isinstance(parents, (tuple, list)):
        raise IngestError("Provenance parent_artifacts must be a sequence.")
    rows: List[Dict[str, str]] = []
    all_verified = True
    for parent in parents:
        if not isinstance(parent, Mapping):
            rows.append(
                {
                    "source_id": str(parent),
                    "version": "",
                    "content_sha256": "",
                    "verification_status": "supplied_unverified",
                }
            )
            all_verified = False
            continue
        for field_name in ("source_id", "version", "content_sha256"):
            if not str(parent.get(field_name, "")).strip():
                raise IngestError(f"A provenance parent lacks {field_name!r}.")
        claimed = str(parent["content_sha256"])
        if not re.fullmatch(r"[0-9a-f]{64}", claimed):
            raise IngestError("A provenance parent has a malformed digest.")
        if ("artifact" in parent) == ("path" in parent):
            rows.append(
                {
                    "source_id": str(parent["source_id"]),
                    "version": str(parent["version"]),
                    "content_sha256": claimed,
                    "verification_status": "supplied_unverified",
                }
            )
            all_verified = False
            continue
        actual = (
            _content_digest(parent["artifact"])
            if "artifact" in parent
            else _file_sha256(Path(str(parent["path"])).expanduser().resolve())
        )
        if actual != claimed:
            raise IngestError(f"Provenance parent {parent['source_id']!r} digest does not match.")
        rows.append(
            {
                "source_id": str(parent["source_id"]),
                "version": str(parent["version"]),
                "content_sha256": actual,
                "verification_status": "observed_verified",
            }
        )
    return rows, all_verified


# Compose the versioned archive's manifest.
def _archive_canonical_bytes(value: Any) -> bytes:
    """
    Serialize supported live scientific state to deterministic UTF-8 bytes.

    Parameters
    ----------
    value : Any
        DataFrame, Series, array, dataclass, mapping, collection, path or
        primitive scalar that forms a live pipeline artefact.

    Returns
    -------
    bytes
        Type-labelled canonical JSON bytes. Scientific labels, dtypes, attrs,
        missingness and field names are retained; only explicitly named UTC
        runtime fields are omitted for rerun stability.

    Raises
    ------
    IngestError
        If any nested value lacks a deterministic serialization contract.
    """
    def normalize(item: Any) -> Any:
        """Convert one supported value to a type-labelled JSON tree."""
        if isinstance(item, pd.DataFrame):
            return {
                "__type__": "DataFrame",
                "index": normalize(item.index.tolist()),
                "index_names": normalize(list(item.index.names)),
                "columns": normalize(item.columns.tolist()),
                "column_names": normalize(list(item.columns.names)),
                "dtypes": [str(dtype) for dtype in item.dtypes],
                "data": normalize(item.to_numpy(dtype=object).tolist()),
                "attrs": normalize(
                    {key: value for key, value in item.attrs.items() if not str(key).lower().endswith("_utc")}
                ),
            }
        if isinstance(item, pd.Series):
            return {
                "__type__": "Series", "name": normalize(item.name),
                "index": normalize(item.index.tolist()),
                "index_names": normalize(list(item.index.names)),
                "dtype": str(item.dtype), "data": normalize(item.tolist()),
                "attrs": normalize(
                    {key: value for key, value in item.attrs.items() if not str(key).lower().endswith("_utc")}
                ),
            }
        if isinstance(item, np.ndarray):
            return {
                "__type__": "ndarray", "dtype": str(item.dtype),
                "shape": list(item.shape), "data": normalize(item.tolist()),
            }
        if isinstance(item, np.generic):
            return normalize(item.item())
        if hasattr(item, "__dataclass_fields__"):
            return {
                "__type__": f"{type(item).__module__}.{type(item).__qualname__}",
                "fields": {
                    name: normalize(getattr(item, name))
                    for name in item.__dataclass_fields__
                    if not str(name).lower().endswith("_utc")
                },
            }
        if isinstance(item, Mapping):
            return {
                "__type__": "mapping",
                "items": [[normalize(key), normalize(item[key])] for key in sorted(item, key=repr)],
            }
        if isinstance(item, (tuple, list)):
            return {"__type__": type(item).__name__, "items": [normalize(value) for value in item]}
        if isinstance(item, (set, frozenset)):
            return {"__type__": type(item).__name__, "items": [normalize(value) for value in sorted(item, key=repr)]}
        if isinstance(item, Path):
            return {"__type__": "Path", "value": str(item)}
        if isinstance(item, pd.Timestamp):
            return {"__type__": "Timestamp", "value": item.isoformat()}
        if item is None or isinstance(item, (str, int, bool)):
            return item
        if isinstance(item, float):
            if np.isnan(item):
                return {"__type__": "float", "value": "nan"}
            if np.isposinf(item):
                return {"__type__": "float", "value": "+inf"}
            if np.isneginf(item):
                return {"__type__": "float", "value": "-inf"}
            return item
        raise IngestError(
            f"Archive state type {type(item).__module__}.{type(item).__qualname__} "
            "has no deterministic serialization contract."
        )

    return json.dumps(
        normalize(value), sort_keys=True, separators=(",", ":"), ensure_ascii=False,
        allow_nan=False,
    ).encode("utf-8")


# Digest one independently controlled artifact-vintage authority claim.
def _archive_vintage_claim_digest(
    artifact_id: str,
    authority_id: str,
    source_id: str,
    source_version: str,
    vintage_tag: str,
    evidence_reference: str,
) -> str:
    """
    Return the canonical digest of one archive-vintage authority claim.

    Parameters
    ----------
    artifact_id : str
        Exact mandatory archive identity.
    authority_id : str
        Independently controlled authority registry identifier.
    source_id : str
        Source or producing-system identity to which the vintage applies.
    source_version : str
        Authenticated source/version identifier, distinct from the archive
        release version.
    vintage_tag : str
        Authenticated vintage label of this exact artifact.
    evidence_reference : str
        Stable external or registry reference for the authority evidence.

    Returns
    -------
    str
        SHA-256 digest binding the authority claim to the artifact's fixed
        live-state key, archive section and scientific role.

    Raises
    ------
    IngestError
        If the artifact is undeclared or any claim component is blank.
    """
    if artifact_id not in ARCHIVE_LIVE_BINDINGS:
        raise IngestError(f"Archive vintage claim names undeclared artefact {artifact_id!r}.")
    values = {
        "authority_id": authority_id,
        "source_id": source_id,
        "source_version": source_version,
        "vintage_tag": vintage_tag,
        "evidence_reference": evidence_reference,
    }
    malformed = sorted(
        name
        for name, value in values.items()
        if not isinstance(value, str) or not value.strip()
    )
    if malformed:
        raise IngestError(
            f"Archive vintage claim {artifact_id!r} has non-string or blank "
            f"field(s) {malformed}."
        )
    normalized = {name: value.strip() for name, value in values.items()}
    state_key, section, role = ARCHIVE_LIVE_BINDINGS[artifact_id]
    payload = {
        "artifact_id": artifact_id,
        "state_key": state_key,
        "archive_section": section,
        "scientific_role": role,
        **normalized,
    }
    return hashlib.sha256(
        json.dumps(payload, sort_keys=True, separators=(",", ":")).encode("utf-8")
    ).hexdigest()


# Resolve and validate an archive vintage only from the module trust registry.
def _trusted_archive_vintage_record(artifact_id: str) -> Mapping[str, str]:
    """
    Return the authenticated vintage record for one mandatory artifact.

    Parameters
    ----------
    artifact_id : str
        Exact identity from ``ARCHIVE_ARTEFACTS``.

    Returns
    -------
    mapping of str to str
        Read-only authority, source/version, vintage, evidence-reference and
        claim-digest fields.

    Raises
    ------
    IngestError
        If the independently controlled registry is absent, incomplete,
        malformed or internally inconsistent. Caller declarations are never
        consulted to resolve this evidence.
    """
    if artifact_id not in ARCHIVE_ARTEFACTS:
        raise IngestError(f"Archive vintage lookup names undeclared artefact {artifact_id!r}.")
    if not isinstance(TRUSTED_ARCHIVE_VINTAGE_REGISTRY, Mapping):
        raise IngestError("The trusted archive-vintage registry is malformed.")
    record = TRUSTED_ARCHIVE_VINTAGE_REGISTRY.get(artifact_id)
    if not isinstance(record, Mapping):
        raise IngestError(
            f"Archive artefact {artifact_id!r} has no independently trusted vintage record."
        )
    expected_fields = {
        "authority_id", "source_id", "source_version", "vintage_tag",
        "evidence_reference", "claim_sha256",
    }
    observed_fields = {str(name) for name in record}
    if observed_fields != expected_fields:
        raise IngestError(
            f"Trusted vintage record {artifact_id!r} has fields {sorted(observed_fields)}; "
            f"expected {sorted(expected_fields)}."
        )
    malformed = sorted(
        name
        for name in expected_fields
        if not isinstance(record[name], str) or not record[name].strip()
    )
    if malformed:
        raise IngestError(
            f"Trusted vintage record {artifact_id!r} has non-string or blank "
            f"field(s) {malformed}."
        )
    normalized = {name: record[name].strip() for name in expected_fields}
    claimed = normalized["claim_sha256"]
    if not re.fullmatch(r"[0-9a-f]{64}", claimed):
        raise IngestError(
            f"Trusted vintage record {artifact_id!r} has a malformed claim digest."
        )
    expected = _archive_vintage_claim_digest(
        artifact_id,
        normalized["authority_id"],
        normalized["source_id"],
        normalized["source_version"],
        normalized["vintage_tag"],
        normalized["evidence_reference"],
    )
    if claimed != expected:
        raise IngestError(
            f"Trusted vintage record {artifact_id!r} disagrees with its authority claim."
        )
    return MappingProxyType(normalized)


# Compose the versioned archive's manifest.
def build_archive_manifest(
    artefacts: Mapping[str, Mapping[str, Any]],
    version_tag: str,
    live_stage_state: Optional[Mapping[str, Any]] = None,
    archive_root: Optional[Path] = None,
) -> pd.DataFrame:
    """
    Compose the versioned archive's manifest.

    Parameters
    ----------
    artefacts : mapping of str to mapping of str
        The archive's artefacts keyed on their declared name. Each entry binds
        a file under ``archive_root`` to a named live-state object and its
        parent lineage. Per-artifact vintage metadata must exactly reproduce
        the independently controlled module registry; entries cannot nominate
        or authenticate their own vintage.
    version_tag : str
        The archive's own version tag.
    live_stage_state : mapping, optional
        Frozen live stage state from which every archive file is serialized.
    archive_root : pathlib.Path, optional
        Self-contained archive root containing README/config fingerprint and
        the seven required scientific subdirectories.

    Returns
    -------
    pandas.DataFrame
        One row per declared artefact carrying the exact live-state key,
        producing-stage lineage, canonical file bytes, observed type, role,
        vintage and verified content hash.

    Raises
    ------
    IngestError
        If a declared artefact is absent, stale, outside the root, mapped to
        the wrong live state, lacks exact parent-stage lineage, or has no
        independently authenticated vintage/source record.
    """
    # Refuse an empty version tag or an archive disconnected from live state.
    if not isinstance(version_tag, str) or not version_tag.strip():
        # The failure states the precondition.
        raise IngestError("The archive requires a version tag.")
    if not isinstance(live_stage_state, Mapping) or not live_stage_state:
        raise IngestError("Archive verification requires the nonempty live stage state.")
    if archive_root is None:
        raise IngestError("Archive verification requires an explicit self-contained root.")
    root = Path(archive_root).expanduser().resolve()
    if not root.exists() or not root.is_dir():
        raise IngestError("The archive root does not exist as a directory.")
    required_directories = ("raw", "interim", "processed", "estimates", "figures", "tables", "audit")
    missing_directories = [name for name in required_directories if not (root / name).is_dir()]
    if missing_directories:
        raise IngestError(f"The archive root lacks required directories {missing_directories}.")
    for filename in ("README.md", "config_fingerprint.txt"):
        path = root / filename
        if not path.is_file() or path.stat().st_size == 0:
            raise IngestError(f"The archive root lacks nonempty {filename}.")
    if str(version_tag) not in (root / "README.md").read_text(encoding="utf-8"):
        raise IngestError("The archive README does not identify the frozen version tag.")
    if "config_validation" not in live_stage_state:
        raise IngestError("The archive state lacks the validated configuration artifact.")
    expected_config_fingerprint = hashlib.sha256(
        _archive_canonical_bytes(live_stage_state["config_validation"])
    ).hexdigest()
    if (root / "config_fingerprint.txt").read_text(encoding="utf-8").strip() != expected_config_fingerprint:
        raise IngestError("The archive config fingerprint is stale relative to live state.")
    # Identify the missing artefacts.
    missing = sorted(set(ARCHIVE_ARTEFACTS) - set(str(name) for name in artefacts.keys()))
    # Refuse a partial archive.
    if missing:
        # The failure names the absent artefacts.
        raise IngestError(
            f"The archive manifest is missing the declared artefact(s) {missing}."
        )
    extras = sorted(set(str(name) for name in artefacts.keys()) - set(ARCHIVE_ARTEFACTS))
    if extras:
        raise IngestError(f"The archive manifest carries undeclared artefact(s) {extras}.")
    # Preflight every entry before writing any byte.  This prevents a malformed
    # later lineage claim from leaving a partially frozen archive.
    plans: List[Dict[str, Any]] = []
    for name in ARCHIVE_ARTEFACTS:
        # Read the artefact's entry.
        entry = artefacts[name]
        # Refuse an entry lacking a declared field.
        for field_name in (
            "archive_version_tag", "vintage_tag", "vintage_authority_id",
            "vintage_source_id", "vintage_source_version",
            "vintage_evidence_reference", "vintage_claim_sha256",
            "role", "media_type", "content_sha256", "location",
            "state_key", "parent_state_keys", "parent_sha256",
        ):
            # Test the field's presence.
            if field_name not in entry:
                # The failure names the absent field and the artefact.
                raise IngestError(
                    f"The archive artefact {name!r} lacks the declared field "
                    f"{field_name!r}."
                )
        textual_fields = (
            "archive_version_tag", "vintage_tag", "vintage_authority_id",
            "vintage_source_id", "vintage_source_version",
            "vintage_evidence_reference", "vintage_claim_sha256", "role",
            "media_type",
        )
        malformed_text = sorted(
            field_name
            for field_name in textual_fields
            if not isinstance(entry[field_name], str)
            or not entry[field_name].strip()
        )
        if malformed_text:
            raise IngestError(
                f"Archive artefact {name!r} has non-string or blank metadata "
                f"field(s) {malformed_text}."
            )
        if entry["archive_version_tag"].strip() != version_tag.strip():
            raise IngestError(
                f"Archive artefact {name!r} names an archive release version "
                "different from the manifest version."
            )
        # Resolve vintage/source evidence exclusively from the independently
        # controlled registry and require the declaration to reproduce it.
        trusted_vintage = _trusted_archive_vintage_record(name)
        trusted_entry_fields = {
            "vintage_tag": "vintage_tag",
            "vintage_authority_id": "authority_id",
            "vintage_source_id": "source_id",
            "vintage_source_version": "source_version",
            "vintage_evidence_reference": "evidence_reference",
            "vintage_claim_sha256": "claim_sha256",
        }
        for entry_field, trusted_field in trusted_entry_fields.items():
            if entry[entry_field].strip() != trusted_vintage[trusted_field]:
                raise IngestError(
                    f"Archive artefact {name!r} {entry_field!r} disagrees with "
                    "the independently trusted vintage registry."
                )
        claimed_digest = str(entry["content_sha256"])
        if not re.fullmatch(r"[0-9a-f]{64}", claimed_digest):
            raise IngestError(f"Archive artefact {name!r} has a malformed digest.")
        if "object" in entry:
            raise IngestError("Object-only archive entries are not self-contained files.")
        path = Path(str(entry["location"])).expanduser().resolve()
        try:
            relative_path = path.relative_to(root)
        except ValueError as exc:
            raise IngestError(f"Archive artefact {name!r} escapes the archive root.") from exc
        if not relative_path.parts or relative_path.parts[0] not in required_directories:
            raise IngestError(f"Archive artefact {name!r} is outside the declared archive sections.")
        state_key = str(entry["state_key"]).strip()
        expected_state_key, expected_section, expected_role = ARCHIVE_LIVE_BINDINGS[name]
        if state_key != expected_state_key:
            raise IngestError(
                f"Archive artefact {name!r} must bind live state {expected_state_key!r}, not {state_key!r}."
            )
        if relative_path.parts[0] != expected_section:
            raise IngestError(
                f"Archive artefact {name!r} must be stored in section {expected_section!r}."
            )
        if str(entry["role"]).strip() != expected_role:
            raise IngestError(f"Archive artefact {name!r} carries the wrong scientific role.")
        if not state_key or state_key not in live_stage_state:
            raise IngestError(f"Archive artefact {name!r} lacks its live-state object.")
        expected_bytes = _archive_canonical_bytes(live_stage_state[state_key])
        if not path.parent.is_dir():
            raise IngestError(f"Archive artefact {name!r} has no validated parent directory.")
        if path.exists() and (not path.is_file() or path.read_bytes() != expected_bytes):
            raise IngestError(
                f"Archive artefact {name!r} bytes are stale or unrelated to live stage state."
            )
        actual_digest = hashlib.sha256(expected_bytes).hexdigest()
        artifact_type = "file"
        location = str(path)
        byte_size = len(expected_bytes)
        size_basis = "canonical_live_state_bytes"
        observed_type = f"{type(live_stage_state[state_key]).__module__}.{type(live_stage_state[state_key]).__qualname__}"
        parents = tuple(str(key).strip() for key in entry["parent_state_keys"])
        producing_stages = [
            stage for stage in PIPELINE_STAGES if state_key in stage.produces
        ]
        if len(producing_stages) != 1:
            raise IngestError(f"Archive artefact {name!r} has no unique producing stage.")
        expected_parents = tuple(producing_stages[0].requires)
        if parents != expected_parents:
            raise IngestError(
                f"Archive artefact {name!r} parent lineage does not equal stage "
                f"{producing_stages[0].name!r} requirements {expected_parents}."
            )
        if len(parents) != len(set(parents)) or any(key not in live_stage_state for key in parents):
            raise IngestError(f"Archive artefact {name!r} has invalid parent-state lineage.")
        claimed_parents = entry["parent_sha256"]
        if not isinstance(claimed_parents, Mapping) or set(claimed_parents) != set(parents):
            raise IngestError(f"Archive artefact {name!r} has incomplete parent digests.")
        parent_digests = {
            key: hashlib.sha256(_archive_canonical_bytes(live_stage_state[key])).hexdigest()
            for key in parents
        }
        if any(str(claimed_parents[key]) != digest for key, digest in parent_digests.items()):
            raise IngestError(f"Archive artefact {name!r} parent lineage is stale.")
        if actual_digest != claimed_digest:
            raise IngestError(f"Archive artefact {name!r} content digest does not match.")
        if not str(entry["vintage_tag"]).strip() or not str(entry["role"]).strip() or not str(entry["media_type"]).strip():
            raise IngestError(f"Archive artefact {name!r} has blank version/type/role metadata.")
        plans.append(
            {
                "name": name, "entry": entry, "path": path,
                "expected_bytes": expected_bytes, "actual_digest": actual_digest,
                "state_key": state_key, "parent_digests": parent_digests,
                "parents": parents, "observed_type": observed_type,
                "trusted_vintage": trusted_vintage,
            }
        )

    resolved_locations = [str(plan["path"]) for plan in plans]
    if len(resolved_locations) != len(set(resolved_locations)):
        raise IngestError("Mandatory archive artefacts must have distinct resolved file locations.")

    # Freeze absent files using an exclusive hard-link publication step.  A
    # concurrent creator wins without being overwritten and is then verified.
    for plan in plans:
        path = plan["path"]
        expected_bytes = plan["expected_bytes"]
        if not path.exists():
            temporary_name: Optional[str] = None
            try:
                with tempfile.NamedTemporaryFile(
                    mode="wb", dir=str(path.parent), prefix=f".{path.name}.",
                    suffix=".tmp", delete=False,
                ) as temporary:
                    temporary.write(expected_bytes)
                    temporary.flush()
                    os.fsync(temporary.fileno())
                    temporary_name = temporary.name
                try:
                    os.link(str(temporary_name), str(path))
                except FileExistsError:
                    pass
            except OSError as exc:
                raise IngestError(f"Archive artefact {plan['name']!r} could not be frozen atomically.") from exc
            finally:
                if temporary_name is not None:
                    try:
                        Path(temporary_name).unlink(missing_ok=True)
                    except OSError:
                        pass
        if not path.is_file() or path.read_bytes() != expected_bytes:
            raise IngestError(
                f"Archive artefact {plan['name']!r} failed exclusive-write verification."
            )

    # Accumulate verified manifest rows only after every freeze succeeds.
    rows: List[dict] = []
    for plan in plans:
        entry = plan["entry"]
        rows.append(
            {
                "artefact": str(plan["name"]),
                "location": str(plan["path"]),
                "archive_version_tag": str(entry["archive_version_tag"]),
                "vintage_tag": str(entry["vintage_tag"]),
                "vintage_authority_id": str(entry["vintage_authority_id"]),
                "vintage_source_id": str(entry["vintage_source_id"]),
                "vintage_source_version": str(entry["vintage_source_version"]),
                "vintage_evidence_reference": str(entry["vintage_evidence_reference"]),
                "vintage_claim_sha256": str(entry["vintage_claim_sha256"]),
                "role": str(entry["role"]),
                "media_type": str(entry["media_type"]),
                "artifact_type": "file",
                "content_sha256": plan["actual_digest"],
                "byte_size": len(plan["expected_bytes"]),
                "size_basis": "canonical_live_state_bytes",
                "observed_type": plan["observed_type"],
                "state_key": plan["state_key"],
                "live_state_sha256": plan["actual_digest"],
                "parent_state_keys_json": _canonical_audit_json(plan["parents"]),
                "parent_sha256_json": _canonical_audit_json(plan["parent_digests"]),
                "serialization_schema": "canonical_json_v1_without_named_utc_fields",
                "verification_status": "observed_verified",
            }
        )
    # Assemble the manifest frame keyed on the artefact's name.
    manifest: pd.DataFrame = pd.DataFrame(rows).set_index("artefact")
    # Record the archive's version in the frame's attributes.
    manifest.attrs["version_tag"] = str(version_tag)
    manifest.attrs["archive_root"] = str(root)
    manifest.attrs["trusted_vintage_registry_sha256"] = hashlib.sha256(
        json.dumps(
            {
                artifact_id: dict(_trusted_archive_vintage_record(artifact_id))
                for artifact_id in ARCHIVE_ARTEFACTS
            },
            sort_keys=True,
            separators=(",", ":"),
        ).encode("utf-8")
    ).hexdigest()
    manifest.attrs["manifest_sha256"] = _content_digest(manifest.sort_index())
    # Log the manifest's extent.
    logger.info(
        "archive manifest composed across %d artefact(s) at version %s.",
        len(manifest),
        version_tag,
    )
    # Return the frame.
    return manifest


# Freeze the audit behind a canonical digest.
def freeze_reconciliation_audit(
    comparisons: pd.DataFrame,
    gaps: pd.DataFrame,
    invariants: pd.DataFrame,
    manifest: pd.DataFrame,
    localization: Optional[pd.DataFrame] = None,
    provenance: Optional[pd.DataFrame] = None,
    blockers: Optional[pd.DataFrame] = None,
    coverage: Optional[pd.DataFrame] = None,
    manuscript_universe: Optional[pd.DataFrame] = None,
    status: str = RECONCILIATION_INCOMPLETE,
) -> str:
    """
    Freeze the audit behind a canonical digest.

    Parameters
    ----------
    comparisons : pandas.DataFrame
        The comparison frame.
    gaps : pandas.DataFrame
        The gaps frame.
    invariants : pandas.DataFrame
        The invariant table.
    manifest : pandas.DataFrame
        The archive manifest.

    Returns
    -------
    str
        Digest of the complete labelled audit payload: schema, values, text,
        missingness, localization, provenance, archive hashes, blockers,
        coverage, universe and completion state.

    Raises
    ------
    IngestError
        If the invariant table is empty.
    """
    # Refuse an empty invariant table.
    if len(invariants) == 0:
        # The failure states the precondition.
        raise IngestError("Cannot freeze an audit without the invariant table.")
    frames: Mapping[str, pd.DataFrame] = {
        "comparisons": comparisons,
        "gaps": gaps,
        "invariants": invariants,
        "manifest": manifest,
        "localization": localization if localization is not None else pd.DataFrame(),
        "provenance": provenance if provenance is not None else pd.DataFrame(),
        "blockers": blockers if blockers is not None else pd.DataFrame(),
        "coverage": coverage if coverage is not None else pd.DataFrame(),
        "manuscript_universe": manuscript_universe if manuscript_universe is not None else pd.DataFrame(),
    }
    digest = hashlib.sha256()
    digest.update(f"reconciliation_status|{status}".encode("utf-8"))
    for name in sorted(frames):
        canonical = frames[name].sort_index().sort_index(axis=1)
        digest.update(name.encode("utf-8"))
        digest.update(_content_digest(canonical).encode("utf-8"))
        digest.update(_canonical_audit_json(dict(canonical.attrs)).encode("utf-8"))
    return digest.hexdigest()


# Validate the independent complete manuscript-cell inventory.
def _validate_complete_manuscript_universe(universe: pd.DataFrame) -> pd.DataFrame:
    """
    Validate the independently pinned 418-cell manuscript transcription.

    Parameters
    ----------
    universe : pandas.DataFrame
        Candidate transcription carrying the 418 canonical cells and the
        authority/version identifiers whose digest is pinned by the module.

    Returns
    -------
    pandas.DataFrame
        A normalized, independently authenticated copy of the inventory.

    Raises
    ------
    IngestError
        If a key is null or blank before/after normalization, or if values,
        metadata, table counts, key uniqueness or the independent pin fail.

    Notes
    -----
    The inventory's attributes identify an authority document but cannot
    authenticate it.  Authentication comes solely from the module-controlled
    ``TRUSTED_MANUSCRIPT_UNIVERSE_PINS`` registry, whose digest is compared
    with a canonical serialization of the live cell rows. Null key components
    are rejected before string conversion so ``None``, ``numpy.nan`` and
    ``pandas.NA`` cannot become apparently valid textual identities.
    """
    if not isinstance(universe, pd.DataFrame) or len(universe) == 0:
        raise IngestError("The reconciliation requires a nonempty independent manuscript inventory.")
    required = list(COMPARISON_KEY) + [
        "reported_value", "comparison_type", "unit", "display_precision",
        "reported_source_id", "procedure_id",
    ]
    missing = [name for name in required if name not in universe.columns]
    if missing:
        raise IngestError(f"The manuscript inventory lacks field(s) {missing}.")
    normalized = universe.copy()
    for name in COMPARISON_KEY:
        # Reject semantic missingness before converting values to strings;
        # otherwise pandas would turn null sentinels into 'nan', 'None' or
        # '<NA>' and admit them as manuscript-cell identities.
        null_key = normalized[name].isna()
        if bool(null_key.any()):
            offending_rows = normalized.index[null_key].tolist()
            raise IngestError(
                f"The manuscript inventory has null {name} value(s) at row(s) "
                f"{offending_rows[:MAX_REPORTED_OFFENDERS]}."
            )
        # Normalize only non-null identities before blank and duplicate checks.
        normalized[name] = normalized[name].astype(str).str.strip()
        # Empty and whitespace-only textual identities remain invalid after
        # normalization even though they are not pandas null sentinels.
        if bool(normalized[name].eq("").any()):
            raise IngestError(f"The manuscript inventory has a blank {name}.")
    if bool(normalized.duplicated(list(COMPARISON_KEY), keep=False).any()):
        raise IngestError("The manuscript inventory carries duplicate canonical cell keys.")
    present_tables = set(normalized["table_number"].astype(str))
    expected_tables = set(PRODUCTION_TABLES)
    if present_tables != expected_tables:
        raise IngestError(
            "The manuscript inventory does not cover exactly Tables 3--10, "
            f"A1--A8 and B1--B7; missing={sorted(expected_tables-present_tables)}, "
            f"extra={sorted(present_tables-expected_tables)}."
        )
    expected_counts = universe.attrs.get("expected_cell_counts")
    if not isinstance(expected_counts, Mapping) or set(str(k) for k in expected_counts) != expected_tables:
        raise IngestError("The independent inventory lacks exact per-table cell counts.")
    for table in PRODUCTION_TABLES:
        expected = expected_counts[table]
        try:
            expected_float = float(expected)
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError(f"Inventory count for {table} is malformed.") from exc
        if (
            isinstance(expected, (bool, np.bool_))
            or not np.isfinite(expected_float)
            or int(expected_float) != expected_float
            or int(expected_float) != int(TABLE_MINIMUM_REQUIRED_CELLS[table])
        ):
            raise IngestError(f"Inventory count for {table} is invalid or structurally incomplete.")
        actual = int(normalized["table_number"].eq(table).sum())
        if actual != int(TABLE_MINIMUM_REQUIRED_CELLS[table]):
            raise IngestError(f"Inventory count for {table} is {actual}, expected {int(TABLE_MINIMUM_REQUIRED_CELLS[table])}.")
    if len(normalized) != MANUSCRIPT_CELL_COUNT:
        raise IngestError(
            f"The complete manuscript transcription must contain exactly {MANUSCRIPT_CELL_COUNT} cells."
        )
    for attr in (
        "authority_id", "version_id", "external_verification_reference",
        "inventory_sha256",
    ):
        if not str(universe.attrs.get(attr, "")).strip():
            raise IngestError(f"The manuscript inventory lacks external authority field {attr!r}.")
    if universe.attrs.get("complete_manuscript_inventory") is not True:
        raise IngestError("The external inventory is not certified complete.")
    digest_columns = sorted(required)
    actual_digest = _content_digest(
        normalized[digest_columns].sort_values(list(COMPARISON_KEY)).reset_index(drop=True)
    )
    if str(universe.attrs["inventory_sha256"]) != actual_digest:
        raise IngestError("The independent manuscript inventory digest does not match its cells.")
    authority_key = (
        str(universe.attrs["authority_id"]).strip(),
        str(universe.attrs["version_id"]).strip(),
    )
    authority_envelope = {
        "authority_id": authority_key[0],
        "version_id": authority_key[1],
        "external_verification_reference": str(
            universe.attrs["external_verification_reference"]
        ).strip(),
        "expected_cell_counts": {
            table: int(TABLE_MINIMUM_REQUIRED_CELLS[table]) for table in PRODUCTION_TABLES
        },
        "cell_rows_sha256": actual_digest,
    }
    authority_envelope_digest = hashlib.sha256(
        _canonical_audit_json(authority_envelope).encode("utf-8")
    ).hexdigest()
    pinned_digest = TRUSTED_MANUSCRIPT_UNIVERSE_PINS.get(authority_key)
    if pinned_digest is None:
        raise IngestError(
            "The manuscript transcription authority/version has no independent "
            "module-controlled digest pin."
        )
    if not re.fullmatch(r"[0-9a-f]{64}", str(pinned_digest)):
        raise IngestError("The trusted manuscript-universe pin is malformed.")
    if str(pinned_digest) != authority_envelope_digest:
        raise IngestError("The manuscript transcription authority envelope disagrees with its independent digest pin.")
    for _, row in normalized.iterrows():
        comparison_type = str(row["comparison_type"])
        if comparison_type not in COMPARISON_TYPES:
            raise IngestError("The manuscript inventory carries an invalid comparison type.")
        if isinstance(row["reported_value"], (bool, np.bool_)):
            raise IngestError("The manuscript inventory carries a Boolean reported value.")
        try:
            value = float(row["reported_value"])
            precision_float = float(row["display_precision"])
        except (TypeError, ValueError, OverflowError) as exc:
            raise IngestError("The manuscript inventory carries malformed numeric metadata.") from exc
        precision = row["display_precision"]
        if not np.isfinite(value):
            raise IngestError("The manuscript inventory carries a nonfinite reported value.")
        if (
            isinstance(precision, (bool, np.bool_))
            or not np.isfinite(precision_float)
            or int(precision_float) != precision_float
            or not 0 <= int(precision_float) <= 12
        ):
            raise IngestError("The manuscript inventory has invalid displayed precision.")
        if comparison_type == "integer_count" and value != int(value):
            raise IngestError("An integer-count manuscript cell is not integral.")
        if not str(row["unit"]).strip() or not str(row["reported_source_id"]).strip() or not str(row["procedure_id"]).strip():
            raise IngestError("The manuscript inventory has blank unit/procedure/source metadata.")
    normalized.attrs = copy.deepcopy(universe.attrs)
    normalized.attrs["structural_verification_status"] = "independent_authority_digest_pin_verified"
    normalized.attrs["trusted_pin_sha256"] = str(pinned_digest)
    normalized.attrs["authority_envelope_sha256"] = authority_envelope_digest
    return normalized


# -------------------------------------------------------------------------------------------------------------------------------
# Task 32, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Execute the reconciliation and verification audit.
def execute_reconciliation_audit(
    reported: pd.DataFrame,
    reproduced: pd.DataFrame,
    declared_keys: Sequence[Tuple[str, str, str]],
    unreproduced: Sequence[Mapping[str, Any]],
    layer_chains: Optional[Mapping[Tuple[str, str, str], Mapping[str, Any]]] = None,
    counterfactual_callable: Optional[Callable[..., Any]] = None,
    imputation_callables: Optional[Sequence[Callable[..., Any]]] = None,
    eligibility_callable: Optional[Callable[..., Any]] = None,
    robustness_frame: Optional[pd.DataFrame] = None,
    study_config: Optional[Mapping[str, Any]] = None,
    declared_facts: Optional[Mapping[str, Mapping[str, Any]]] = None,
    archive_artefacts: Optional[Mapping[str, Mapping[str, Any]]] = None,
    version_tag: str = "v1.0.0-frozen",
    tolerance: float = ESTIMATE_TOLERANCE,
    require_invariants: bool = True,
    manuscript_universe: Optional[pd.DataFrame] = None,
    probe_registry: Optional[Mapping[Tuple[str, str], Mapping[str, Any]]] = None,
    archive_state: Optional[Mapping[str, Any]] = None,
    archive_root: Optional[Path] = None,
) -> ReconciliationReport:
    """
    Execute the reconciliation and verification audit.

    Parameters
    ----------
    reported : pandas.DataFrame
        The reported values keyed on the declared comparison key.
    reproduced : pandas.DataFrame
        The reproduced values under the same key.
    declared_keys : sequence of tuple
        Keys supplied by the independent manuscript inventory.  They are
        cross-checked against, never derived from, ``reported``.
    unreproduced : sequence of mapping
        The values that cannot be reproduced with their blocking inputs.
    layer_chains : mapping of tuple to mapping of str to float, optional
        The computation chain of each value.
    counterfactual_callable : callable, optional
        The projection callable the first invariant inspects.
    imputation_callables : sequence of callable, optional
        The completion and classification callables the second invariant
        inspects.
    eligibility_callable : callable, optional
        The eligibility callable the second invariant quotes.
    robustness_frame : pandas.DataFrame, optional
        The robustness runs the third invariant inspects.
    study_config : mapping, optional
        The study configuration the provenance reads.
    declared_facts : mapping of str to mapping of str, optional
        Further provenance facts.
    archive_artefacts : mapping of str to mapping of str, optional
        The archive's artefacts. When absent the manifest is refused.
    version_tag : str, optional
        The archive's version tag.
    tolerance : float, optional
        The tolerance of the comparison.
    require_invariants : bool, optional
        Compatibility flag.  Scientific completion always requires all three
        invariants; setting this false cannot waive them.
    manuscript_universe : pandas.DataFrame, optional
        Complete externally verified inventory of Tables 3--10, A1--A8 and
        B1--B7, including units, precision, procedures and source identities.
    probe_registry : mapping, optional
        Independently pinned, live-stage argument scenarios for the registered
        scientific boundaries.
    archive_state : mapping, optional
        Live stage state whose canonical bytes must equal archive files.
    archive_root : pathlib.Path, optional
        Root of the self-contained replication archive.

    Returns
    -------
    ReconciliationReport
        The comparison, the gaps, the localization, the invariants, the
        provenance, the manifest, the digest, the timestamp and the notes.

    Raises
    ------
    IngestError
        If the independent universe, comparison keys, perturbation evidence or
        supplied archive/provenance claims are malformed.  Scientifically
        unresolved but well-formed inputs yield an explicit incomplete report.
    """
    if manuscript_universe is None:
        raise IngestError(
            "The reconciliation cannot derive its manuscript-cell universe from "
            "reported values; supply the independent authoritative inventory."
        )
    if not isinstance(require_invariants, (bool, np.bool_)):
        raise IngestError("require_invariants must be Boolean; it cannot encode a waiver string.")
    try:
        universe = _validate_complete_manuscript_universe(manuscript_universe)
    except IngestError as exc:
        if "has no independent module-controlled digest pin" not in str(exc):
            raise
        # A structurally supplied but unpinned transcription is an explicit
        # scientific blocker, not a basis for comparisons or a false complete
        # certificate.
        empty_index = pd.MultiIndex.from_tuples([], names=list(COMPARISON_KEY))
        comparisons = pd.DataFrame(index=empty_index)
        gaps = pd.DataFrame(index=empty_index)
        localization = pd.DataFrame()
        invariants = pd.DataFrame(
            [
                {
                    "invariant": identifier,
                    "holds": False,
                    "demonstration": "manuscript authority is unpinned; reconciliation was not dispatched",
                    "evidence_json": "{}",
                    "baseline_output_sha256": "",
                    "perturbed_output_sha256": "",
                    "forbidden_input_changed": False,
                }
                for identifier in INVARIANT_IDS
            ]
        ).set_index("invariant")
        try:
            provenance = build_provenance(study_config or {}, declared_facts)
        except (IngestError, TypeError, ValueError) as provenance_exc:
            provenance = pd.DataFrame(
                [{
                    "fact": "provenance_preflight", "value": "",
                    "source_artifact": "", "source_version": "",
                    "content_sha256": "", "relevant_key": "",
                    "transformation": "", "parent_artifacts_json": "[]",
                    "verification_status": "unresolved",
                    "verification_detail": str(provenance_exc),
                }]
            ).set_index("fact")
        manifest = pd.DataFrame()
        coverage = pd.DataFrame(
            [{"declared_count": MANUSCRIPT_CELL_COUNT, "comparison_count": 0,
              "gap_count": 0, "coverage_status": "blocked_unpinned_authority"}]
        )
        blockers = pd.DataFrame(
            [{"category": "manuscript_authority_unpinned", "key": "manuscript_universe", "detail": str(exc)}]
        )
        fingerprint = freeze_reconciliation_audit(
            comparisons, gaps, invariants, manifest, localization=localization,
            provenance=provenance, blockers=blockers, coverage=coverage,
            manuscript_universe=manuscript_universe, status=RECONCILIATION_INCOMPLETE,
        )
        return ReconciliationReport(
            comparisons=comparisons, gaps=gaps, localization=localization,
            invariants=invariants, provenance=provenance, manifest=manifest,
            fingerprint_sha256=fingerprint, completed_utc=current_utc_iso(),
            notes=("Reconciliation was not dispatched because the complete transcription lacks an independent trust pin.",),
            status=RECONCILIATION_INCOMPLETE, blockers=blockers, coverage=coverage,
        )
    universe_keys = [
        tuple(str(value) for value in row)
        for row in universe[list(COMPARISON_KEY)].to_numpy()
    ]
    normalized_declared = [tuple(str(value) for value in key) for key in declared_keys]
    if len(normalized_declared) != len(set(normalized_declared)):
        raise IngestError("declared_keys carry duplicate manuscript cells.")
    if len(normalized_declared) != len(universe_keys) or set(normalized_declared) != set(universe_keys):
        raise IngestError("declared_keys do not equal the independent manuscript inventory.")
    # Normalize and bind every reported cell to the independent inventory.
    reported_normalized = reported.copy()
    for column in COMPARISON_KEY:
        if column not in reported_normalized.columns:
            raise IngestError(f"The reported frame lacks key column {column!r}.")
        reported_normalized[column] = reported_normalized[column].astype(str)
    _assert_comparison_subset_matches_universe(reported_normalized, universe)
    # Register explicit unresolved cells first so they cannot silently disappear
    # from the comparison subset.
    gaps = register_reconciliation_gaps(unreproduced)
    gap_keys = {
        tuple(str(value) for value in key) for key in gaps.index.tolist()
    } if len(gaps) else set()
    universe_indexed = universe.set_index(list(COMPARISON_KEY))
    for key, row in gaps.iterrows() if len(gaps) else []:
        normalized_key = tuple(str(value) for value in key)
        if normalized_key not in universe_indexed.index:
            raise IngestError(f"Reconciliation gap {normalized_key} is outside the inventory.")
        inventory_row = universe_indexed.loc[normalized_key]
        if (
            str(row["unit"]) != str(inventory_row["unit"])
            or str(row["source_id"]) != str(inventory_row["reported_source_id"])
            or str(row["procedure_id"]) != str(inventory_row["procedure_id"])
        ):
            raise IngestError(f"Reconciliation gap {normalized_key} contradicts inventory metadata.")
    comparable_mask = [
        tuple(str(value) for value in row) not in gap_keys
        for row in reported_normalized[list(COMPARISON_KEY)].to_numpy()
    ]
    comparable = reported_normalized.loc[comparable_mask].copy()
    comparable_universe = universe.loc[
        [
            tuple(str(value) for value in row) not in gap_keys
            for row in universe[list(COMPARISON_KEY)].to_numpy()
        ]
    ].copy()
    comparisons, localization = compose_comparison_table(
        comparable,
        reproduced,
        layer_chains=layer_chains,
        tolerance=tolerance,
        required_universe=comparable_universe,
    )
    coverage = assert_every_row_attributed(comparisons, gaps, universe_keys)
    blockers_rows: List[Dict[str, Any]] = []
    divergent = comparisons.loc[comparisons["verdict"].eq(VERDICT_DIVERGENT)]
    for key in divergent.index:
        blockers_rows.append(
            {
                "category": "divergent_manuscript_cell",
                "key": _canonical_audit_json(tuple(str(value) for value in key)),
                "detail": "reported and reproduced values disagree under the cell rule",
            }
        )
    for key, row in gaps.iterrows() if len(gaps) else []:
        blockers_rows.append(
            {
                "category": "registered_reconciliation_gap",
                "key": _canonical_audit_json(tuple(str(value) for value in key)),
                "detail": f"{row['blocking_input']}: {row['description']}",
            }
        )
    invariant_inputs_present = (
        counterfactual_callable is not None
        and imputation_callables is not None
        and eligibility_callable is not None
        and robustness_frame is not None
    )
    if invariant_inputs_present:
        try:
            invariants = audit_design_invariants(
                counterfactual_callable,
                imputation_callables,
                eligibility_callable,
                robustness_frame,
                probe_registry=probe_registry,
            )
        except IngestError as invariant_exc:
            if "not attached to the exact registered production boundary" in str(invariant_exc):
                raise
            invariants = pd.DataFrame(
                [
                    {
                        "invariant": invariant,
                        "holds": False,
                        "demonstration": f"invariant evidence unresolved: {invariant_exc}",
                        "evidence_json": _canonical_audit_json(
                            {"error_type": type(invariant_exc).__name__, "detail": str(invariant_exc)}
                        ),
                        "baseline_output_sha256": "",
                        "perturbed_output_sha256": "",
                        "forbidden_input_changed": False,
                    }
                    for invariant in INVARIANT_IDS
                ]
            ).set_index("invariant")
    else:
        invariants = pd.DataFrame(
            [
                {
                    "invariant": invariant,
                    "holds": False,
                    "demonstration": "required executable perturbation evidence was not supplied",
                    "evidence_json": "{}",
                    "baseline_output_sha256": "",
                    "perturbed_output_sha256": "",
                    "forbidden_input_changed": False,
                }
                for invariant in INVARIANT_IDS
            ]
        ).set_index("invariant")
    for invariant, row in invariants.loc[~invariants["holds"]].iterrows():
        blockers_rows.append(
            {
                "category": "design_invariant_not_verified",
                "key": str(invariant),
                "detail": str(row["demonstration"]),
            }
        )
    provenance = build_provenance(
        study_config if study_config is not None else {}, declared_facts
    )
    unresolved_provenance = provenance.loc[
        ~provenance["verification_status"].eq("observed_verified")
    ]
    for fact, row in unresolved_provenance.iterrows():
        blockers_rows.append(
            {
                "category": "provenance_not_verified",
                "key": str(fact),
                "detail": str(row["verification_detail"]),
            }
        )
    if archive_artefacts is None:
        manifest = pd.DataFrame(
            columns=[
                "location", "vintage_tag", "role", "media_type", "artifact_type",
                "content_sha256", "byte_size", "size_basis", "observed_type",
                "verification_status",
            ]
        )
        manifest.index.name = "artefact"
        manifest.attrs["version_tag"] = str(version_tag)
        blockers_rows.append(
            {
                "category": "archive_incomplete",
                "key": "archive_manifest",
                "detail": "no actual archive artefacts were supplied",
            }
        )
    else:
        try:
            manifest = build_archive_manifest(
                archive_artefacts,
                version_tag,
                live_stage_state=archive_state,
                archive_root=archive_root,
            )
        except IngestError as archive_exc:
            missing_evidence = any(
                token in str(archive_exc)
                for token in (
                    "requires the nonempty live stage state",
                    "requires an explicit self-contained root",
                    "root does not exist",
                    "lacks required directories",
                    "lacks nonempty README.md",
                    "lacks nonempty config_fingerprint.txt",
                )
            )
            if not missing_evidence:
                raise
            manifest = pd.DataFrame()
            manifest.attrs["version_tag"] = str(version_tag)
            blockers_rows.append(
                {
                    "category": "archive_incomplete",
                    "key": "archive_manifest",
                    "detail": str(archive_exc),
                }
            )
    blockers = pd.DataFrame(
        blockers_rows,
        columns=["category", "key", "detail"],
    )
    status = RECONCILIATION_COMPLETE if len(blockers) == 0 else RECONCILIATION_INCOMPLETE
    # The legacy flag is recorded but cannot waive mandatory scientific checks.
    if not bool(require_invariants):
        blockers = pd.concat(
            [
                blockers,
                pd.DataFrame(
                    [
                        {
                            "category": "waiver_refused",
                            "key": "require_invariants",
                            "detail": "design invariants are mandatory and cannot be disabled",
                        }
                    ]
                ),
            ],
            ignore_index=True,
        )
        status = RECONCILIATION_INCOMPLETE
    fingerprint = freeze_reconciliation_audit(
        comparisons,
        gaps,
        invariants,
        manifest,
        localization=localization,
        provenance=provenance,
        blockers=blockers,
        coverage=coverage,
        manuscript_universe=universe,
        status=status,
    )
    # Accumulate the recorded notes.
    notes: List[str] = [
        "every reported value reconciles within the declared tolerance or is "
        "registered as a gap with its blocking input; no value is left "
        "unattributed.",
        "a divergence is localized only through matched expected-versus-observed "
        "quantities at each layer; no intermediate is compared with a final ATT.",
        "the three design invariants execute independent perturbations and bind "
        "the resulting input/output digests, closure/global dependency trace and "
        "actual robustness snapshot objects.",
        "provenance and archive claims are verified against actual source objects "
        "or files and their content hashes; supplied assertions remain unverified.",
        "the complete 418-cell manuscript inventory is accepted only when its "
        "authority/version and canonical live-cell digest match the independent "
        "module-controlled pin.",
    ]
    # Record the reconciliation's outcome.
    notes.append(
        f"cells reconciled: {int((comparisons['verdict'] == VERDICT_RECONCILED).sum())} "
        f"of {len(comparisons)}; gaps registered: {len(gaps)}."
    )
    # Assemble the immutable report.
    report: ReconciliationReport = ReconciliationReport(
        comparisons=comparisons,
        gaps=gaps,
        localization=localization,
        invariants=invariants,
        provenance=provenance,
        manifest=manifest,
        fingerprint_sha256=fingerprint,
        completed_utc=current_utc_iso(),
        notes=tuple(notes),
        status=status,
        blockers=blockers,
        coverage=coverage,
    )
    # Log the completed audit.
    logger.info(
        "reconciliation audit %s: %d cell(s), %d gap(s), %d invariant(s), "
        "fingerprint %s.",
        status,
        len(comparisons),
        len(gaps),
        len(invariants),
        fingerprint[:16],
    )
    # Return the report.
    return report
````

````python
# Task 33: End-to-End Pipeline Orchestration

# ==============================================================================
# Task 33: End-to-End Pipeline Orchestration
# ==============================================================================

# Declare the outcome of a completed stage.
OUTCOME_COMPLETED: Final[str] = "completed"
# Declare the outcome of a skipped stage.
OUTCOME_SKIPPED: Final[str] = "skipped_missing_inputs"
# Declare the outcome of a refused stage.
OUTCOME_FAILED: Final[str] = "failed"
# Declare the study-level completion states independently of stage execution.
STUDY_COMPLETE: Final[str] = "complete"
STUDY_INCOMPLETE: Final[str] = "incomplete"
# Declare the policy name of the strict-sanctions coalition, which is the name
# the partition registry and the mechanism specifications declare.
POLICY_SANCTIONS: Final[str] = "strict_sanctions"
# Declare the policy name of the direct-arming coalition, which is the name
# the partition registry and the mechanism specifications declare.
POLICY_ARMING: Final[str] = "direct_arming"
# Compose the keys of the result's own report entries.
REPORT_KEYS: Final[Tuple[str, ...]] = (
    "pipeline_summary",
    "pipeline_state",
    "pipeline_digest",
    "pipeline_notes",
    "pipeline_status",
    "pipeline_blockers",
)
# Declare the column order of the pipeline summary frame.
SUMMARY_COLUMNS: Final[Tuple[str, ...]] = (
    "number",
    "name",
    "outcome",
    "produces",
    "missing",
    "error",
    "dependency_digest",
    "artifact_digest",
)
# Declare the artefact key of the ingestion bundle.
KEY_BUNDLE: Final[str] = "ingest_bundle"
# Declare the artefact key of the configuration validation.
KEY_CONFIG: Final[str] = "config_validation"
# Declare the artefact key of the cleansed bundle. The cleansing orchestrator
# emits the cleansed container itself (CF-1), not a bare report, so every
# downstream stage consumes this key.
KEY_CLEANSE: Final[str] = "cleansed_bundle"
# Declare the artefact key of the eligibility screen.
KEY_ELIGIBILITY: Final[str] = "eligibility"
# Declare the artefact key of the control completion.
KEY_CONTROLS: Final[str] = "control_completion"
# Declare the artefact key of the outcome repair.
KEY_OUTCOMES: Final[str] = "outcome_repair"
# Declare the artefact key of the analysis panel.
KEY_PANEL: Final[str] = "analysis_panel"
# Declare the artefact key of the treatment coding.
KEY_TREATMENT: Final[str] = "treatment"
# Declare the artefact key of the partition registry.
KEY_REGISTRY: Final[str] = "partition_registry"
# Declare the artefact key of the donor factor.
KEY_FACTOR: Final[str] = "donor_factor"
# Declare the artefact key of the pooled fit.
KEY_FIT: Final[str] = "pooled_fit"
# Declare the artefact key of the counterfactual projection.
KEY_COUNTERFACTUAL: Final[str] = "counterfactual"
# Declare the artefact key of the gap and effect estimation.
KEY_GAPS: Final[str] = "gap_att"
# Declare the artefact key of the country-dispersion inference.
KEY_INFERENCE_CHAIN: Final[str] = "chain_inference"
# Declare the artefact key of the baseline pipeline.
KEY_BASELINE: Final[str] = "baseline"
# Declare the artefact key of the resampling execution.
KEY_RESAMPLING: Final[str] = "resampling"
# Declare the artefact key of the sweep factory.
KEY_SWEEPS: Final[str] = "sweeps"
# Declare the artefact key of the robustness execution.
KEY_ROBUSTNESS: Final[str] = "robustness_sweeps"
# Declare the artefact key of the secondary sweeps.
KEY_SECONDARY: Final[str] = "secondary_sweeps"
# Declare the artefact key of the influence diagnostics.
KEY_INFLUENCE: Final[str] = "influence"
# Declare the artefact key of the influence tables.
KEY_INFLUENCE_TABLES: Final[str] = "influence_tables"
# Declare the artefact key of the matching designs.
KEY_MATCHING: Final[str] = "matching"
# Declare the artefact key of the matching report.
KEY_MATCHING_REPORT: Final[str] = "matching_report"
# Declare the artefact key of the cell ledger.
KEY_CELL_LEDGER: Final[str] = "cell_ledger"
# Declare the artefact key of the exposure primitives.
KEY_PRIMITIVES: Final[str] = "exposure_primitives"
# Declare the artefact key of the exposure ratios.
KEY_RATIOS: Final[str] = "exposure_ratios"
# Declare the artefact key of the heterogeneity suite.
KEY_HETEROGENEITY: Final[str] = "heterogeneity_suite"
# Declare the artefact key of the heterogeneity regressions.
KEY_HETEROGENEITY_REG: Final[str] = "heterogeneity_regressions"
# Declare the artefact key of the direct-arming pipeline.
KEY_ARMING: Final[str] = "direct_arming"
# Declare the artefact key of the figure and table production.
KEY_PRODUCTION: Final[str] = "production"
# Declare the artefact key of the reconciliation audit.
KEY_RECONCILIATION: Final[str] = "reconciliation"
# Declare the artefact key of the composed baseline data mapping, which the
# baseline-pipeline stage emits so its consumers reuse the exact structures.
KEY_DATA_MAPPING: Final[str] = "data_mapping"


# -------------------------------------------------------------------------------------------------------------------------------
# Task 33, Step 1: The stage registry and the threading contract
# -------------------------------------------------------------------------------------------------------------------------------


# Declare one stage of the end-to-end pipeline.
@dataclass(frozen=True)
class StageRecord:
    """
    One stage of the end-to-end pipeline.

    Attributes
    ----------
    number : int
        The stage's position in the chain, which orders the run. The numbers
        are chain positions: the resampling execution stage (position 16) is
        the Task 17 orchestrator, which re-estimates the whole design through
        the Task 16 orchestrator internally, so no separate Task 16 stage is
        declared. Every declared task orchestrator is exercised by the chain.
    name : str
        The stage's name, which is also the key its artefacts are stored
        under and the key the run's report is indexed on.
    orchestrator : Callable
        The task's own orchestrator, which performs the stage's work; the
        pipeline never reimplements a step.
    requires : tuple of str
        The artefact keys the stage consumes from the state, which the
        validation checks before the stage runs.
    produces : tuple of str
        The artefact keys the stage emits into the state.
    data_keys : tuple of str
        The supplementary raw structures the stage consumes from the caller's
        data mapping, which the ingestion cannot synthesize.
    adapter : Callable
        The callable that reads the state and the configuration and invokes
        the orchestrator with its own declared arguments.
    """
    # The stage's position in the chain.
    number: int
    # The stage's name.
    name: str
    # The task's own orchestrator.
    orchestrator: Callable[..., Any]
    # The artefact keys the stage consumes.
    requires: Tuple[str, ...]
    # The artefact keys the stage emits.
    produces: Tuple[str, ...]
    # The supplementary raw structures the stage consumes.
    data_keys: Tuple[str, ...]
    # The stage's adapter.
    adapter: Callable[[Mapping[str, Any], Mapping[str, Any]], Mapping[str, Any]]
    # Whether this stage is required for a complete replication.
    mandatory: bool = True


# Carry a well-formed but scientifically incomplete stage artefact to the
# executor without misclassifying it as a successful mandatory output.
class PipelineStageIncompleteError(IngestError):
    """Raised when a stage returns useful diagnostic evidence, but not completion."""

    def __init__(self, message: str, artifacts: Mapping[str, Any]) -> None:
        super().__init__(message)
        self.artifacts: Mapping[str, Any] = MappingProxyType(dict(artifacts))


# -------------------------------------------------------------------------------------------------------------------------------
# Task 33, Step 2: The adapters, one per stage
# -------------------------------------------------------------------------------------------------------------------------------


# Compose the retained universe of the eligibility screen.
def _retained_universe(screen: Any) -> Tuple[str, ...]:
    """
    Compose the retained country universe of the eligibility screen.

    Parameters
    ----------
    screen : EligibilityScreen
        The eligibility screen whose mask carries the retained flag per
        country.

    Returns
    -------
    tuple of str
        The sorted retained country codes.

    Raises
    ------
    IngestError
        If the mask is absent or retains no country.
    """
    # Read the mask of the screen.
    mask = screen.mask
    # Refuse a screen without a retained country.
    if int(mask.sum()) == 0:
        # Without a retained universe the downstream stages are undefined.
        raise IngestError("The eligibility screen retains no country.")
    # Return the sorted retained codes.
    return tuple(sorted(str(code) for code in mask[mask].index))


# Compose the standardized exposures in the wide form the mechanism consumes.
def _standardized_exposures(ratios: Any) -> pd.DataFrame:
    """
    Compose the standardized exposures in the wide ratio form.

    Parameters
    ----------
    ratios : ExposureRatioResult
        The ratio construction carrying the standardized frame indexed on the
        country, the window, the rule and the ratio name.

    Returns
    -------
    pandas.DataFrame
        The standardized values keyed on country, window and coverage rule,
        one column per independently sampled ratio name.
    """
    standardized = ratios.standardized.copy()
    required_index = ("country_code", "window", "rule", "ratio_name")
    if tuple(standardized.index.names) != required_index:
        raise IngestError(
            "Standardized exposures must retain country/window/rule/ratio keys."
        )
    required_columns = {
        "standardized_value",
        "raw_value",
        "sample_mean",
        "sample_sd",
        "sample_size",
        "sd_convention",
        "sample_fingerprint_sha256",
    }
    missing = required_columns.difference(standardized.columns)
    if missing:
        raise IngestError(
            "Standardized exposure metadata is incomplete: "
            f"{sorted(missing)}."
        )
    if standardized.index.has_duplicates:
        raise IngestError("Standardized exposure keys must be unique.")
    values = pd.to_numeric(standardized["standardized_value"], errors="coerce")
    admitted = values.notna()
    if admitted.any() and not np.isfinite(values.loc[admitted].to_numpy(dtype=float)).all():
        raise IngestError("Admitted standardized exposure values must be finite.")
    metadata_columns = tuple(sorted(required_columns - {"standardized_value", "raw_value"}))
    metadata_rows: List[Dict[str, Any]] = []
    for sample_key, block in standardized.loc[admitted].groupby(
        level=["window", "rule", "ratio_name"], sort=True
    ):
        for column in metadata_columns:
            if block[column].nunique(dropna=False) != 1:
                raise IngestError(
                    f"Exposure sample {sample_key!r} has inconsistent {column}."
                )
        declared_n = block["sample_size"].iloc[0]
        if (
            isinstance(declared_n, (bool, np.bool_))
            or not np.isfinite(float(declared_n))
            or float(declared_n) != int(float(declared_n))
            or int(float(declared_n)) != len(block)
        ):
            raise IngestError(
                f"Exposure sample {sample_key!r} has a stale sample size."
            )
        metadata_rows.append(
            {
                "window": str(sample_key[0]),
                "rule": str(sample_key[1]),
                "ratio_name": str(sample_key[2]),
                **{column: block[column].iloc[0] for column in metadata_columns},
            }
        )
    exposures = standardized["standardized_value"].unstack("ratio_name")
    exposures = exposures.sort_index().sort_index(axis=1)
    exposures.attrs["sample_metadata"] = pd.DataFrame(metadata_rows).sort_values(
        ["window", "rule", "ratio_name"], kind="mergesort"
    ).reset_index(drop=True) if metadata_rows else pd.DataFrame(
        columns=["window", "rule", "ratio_name", *metadata_columns]
    )
    exposures.attrs["ratio_result_fingerprint_sha256"] = str(
        ratios.fingerprint_sha256
    )
    exposures.attrs["standardized_content_sha256"] = _content_digest(standardized)
    return exposures


# Compose the policy indicator frame the mechanism consumes.
def _policy_indicator_frame(treatment: Any) -> pd.DataFrame:
    """
    Compose the policy indicator frame the mechanism consumes.

    Parameters
    ----------
    treatment : TreatmentCodingResult
        The treatment coding carrying the country-year indicator tensor.

    Returns
    -------
    pandas.DataFrame
        The indicators keyed on the country and the year, one column per
        declared policy name.
    """
    # Index the indicator tensor on the country and the year.
    indexed: pd.DataFrame = treatment.indicators.set_index(
        ["country_code", "year"]
    )
    # Rename the indicator columns to the declared policy names.
    renamed: pd.DataFrame = indexed.rename(
        columns={
            "S_it": POLICY_SANCTIONS,
            "A_it": POLICY_ARMING,
        }
    )
    # Return the declared policy columns.
    return renamed[[POLICY_SANCTIONS, POLICY_ARMING]]


# Map a country's entry year to its active second-wave window.
def entry_year_active_window(
    entry_year: int,
    wave_bounds: Tuple[int, int] = WAVE_TWO_WINDOW,
    available_years: Optional[Sequence[int]] = None,
    canonical_activity: Optional[Mapping[int, Any]] = None,
    country_code: Optional[str] = None,
    policy_name: Optional[str] = None,
) -> Tuple[int, int]:
    """
    Map a country's entry year to its active second-wave window.

    The active-year axis is declared as ``per_country_entry_year``: a country's
    second wave opens at the later of its own coalition entry year and the
    declared Wave-2 start, and closes at the declared projection end. Thus an
    entrant active by 2022 reproduces the baseline window while a later
    entrant contributes only its observed active Wave-2 years.

    Parameters
    ----------
    entry_year : int
        The country's resolved entry year.

    Returns
    -------
    tuple of int
        The country's inclusive active window.
    """
    if (
        isinstance(entry_year, (bool, np.bool_))
        or not isinstance(entry_year, (int, np.integer))
    ):
        raise IngestError("Active-year entry must be an integral calendar year.")
    if len(wave_bounds) != 2:
        raise IngestError("Active-year wave bounds must contain exactly two years.")
    parsed_bounds: List[int] = []
    for value in wave_bounds:
        if (
            isinstance(value, (bool, np.bool_))
            or not isinstance(value, (int, np.integer))
        ):
            raise IngestError("Active-year wave bounds must be integral years.")
        parsed_bounds.append(int(value))
    wave_start, wave_end = parsed_bounds
    if wave_start > wave_end:
        raise IngestError("Active-year wave bounds are reversed.")
    start = max(int(entry_year), wave_start)
    window = (start, wave_end)
    if available_years is not None:
        years: List[int] = []
        for value in available_years:
            if isinstance(value, (bool, np.bool_)) or not isinstance(
                value, (int, np.integer)
            ):
                raise IngestError("Available activity years must be integral.")
            years.append(int(value))
        required = set(range(wave_start, wave_end + 1))
        if not required.issubset(set(years)):
            raise IngestError(
                "Active-year resolution lacks complete declared Wave-2 years."
            )
    if canonical_activity is not None:
        expected = {
            year: int(start <= year <= wave_end)
            for year in range(wave_start, wave_end + 1)
        }
        observed: Dict[int, int] = {}
        for raw_year, raw_value in canonical_activity.items():
            if isinstance(raw_year, (bool, np.bool_)):
                raise IngestError("Canonical activity years must be integral.")
            try:
                numeric_year = float(raw_year)
            except (TypeError, ValueError, OverflowError) as exc:
                raise IngestError("Canonical activity years must be integral.") from exc
            if not np.isfinite(numeric_year) or not numeric_year.is_integer():
                raise IngestError("Canonical activity years must be integral.")
            year = int(numeric_year)
            value = pd.to_numeric(pd.Series([raw_value]), errors="coerce").iloc[0]
            if not np.isfinite(value) or float(value) not in (0.0, 1.0):
                raise IngestError("Canonical activity cells must be finite binary values.")
            if year in expected:
                observed[year] = int(value)
        if observed != expected:
            label = f" for {country_code or '<country>'}/{policy_name or '<policy>'}"
            raise IngestError(
                "Resolved entry year disagrees with the canonical activity tensor" + label + "."
            )
    # A start after the wave end is an explicit empty intersection; callers
    # that require contributing years reject the reversed tuple themselves.
    return window


# Compose the mechanism panel of the countries that carry an exposure.
def _mechanism_panel(state: Mapping[str, Any]) -> pd.DataFrame:
    """
    Compose the mechanism panel of the exposure-carrying countries.

    The panel is duplicated under its exact ratio/window/rule sample key. It
    deliberately does not construct a union sample: every block has the
    independently admitted country roster of that exposure specification.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the panel and the exposure ratios.

    Returns
    -------
    pandas.DataFrame
        The analysis panel restricted to the exposure-carrying countries.
    """
    exposures = _standardized_exposures(state[KEY_RATIOS])
    panel = state[KEY_PANEL].panel.copy()
    if panel.duplicated(["country_code", "year"]).any():
        raise IngestError("Mechanism source panel has duplicate country-year rows.")
    blocks: List[pd.DataFrame] = []
    # Pandas 3's stack implementation requires dropna to be unspecified;
    # explicitly remove missing exposure entries afterwards to preserve the
    # independently admitted sample rosters of earlier pandas versions.
    stacked = exposures.stack().dropna().rename("standardized_exposure")
    stacked.index = stacked.index.set_names(
        ["country_code", "window", "rule", "ratio_name"]
    )
    for (window, rule, ratio_name), sample in stacked.groupby(
        level=["window", "rule", "ratio_name"], sort=True
    ):
        countries = tuple(
            sorted(str(code) for code in sample.index.get_level_values("country_code"))
        )
        block = panel[panel["country_code"].astype(str).isin(countries)].copy()
        if set(block["country_code"].astype(str)) != set(countries):
            raise IngestError(
                f"Mechanism panel cannot realize sample {(window, rule, ratio_name)!r}."
            )
        value_map = sample.droplevel(["window", "rule", "ratio_name"]).to_dict()
        block["exposure_variant"] = str(ratio_name)
        block["window"] = str(window)
        block["rule"] = str(rule)
        block["standardized_exposure"] = block["country_code"].astype(str).map(value_map)
        blocks.append(block)
    if not blocks:
        raise IngestError("No independently admitted mechanism sample is available.")
    result = pd.concat(blocks, ignore_index=True)
    result = result.set_index(
        ["exposure_variant", "window", "rule", "country_code", "year"],
        drop=False,
    ).sort_index()
    result.attrs["sample_metadata"] = exposures.attrs["sample_metadata"].copy()
    result.attrs["ratio_result_fingerprint_sha256"] = exposures.attrs[
        "ratio_result_fingerprint_sha256"
    ]
    return result


# Adapt the ingestion stage.
def _adapt_ingest(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the ingestion stage.

    Parameters
    ----------
    state : mapping
        The artefact state.
    data : mapping
        The caller's data mapping, whose ``sources`` entry carries the raw
        blocks under the declared source names.

    Returns
    -------
    mapping
        The bundle stored under the stage's declared key.
    """
    # Compose the raw sources the ingestion declares.
    sources = data["sources"]
    # Invoke the ingestion orchestrator with its declared arguments.
    bundle = orchestrate_raw_data_ingest(sources, data["study_config"])
    # Return the stage's artefacts.
    return {KEY_BUNDLE: bundle}


# Adapt the configuration-validation stage.
def _adapt_config(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the configuration-validation stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the ingested bundle under the bundle key.
    data : mapping
        The caller mapping carrying `study_config` under the `study_config` key.

    Returns
    -------
    mapping
        The validation report stored under the config key.
    """
    # Invoke the validation orchestrator with its declared arguments.
    validation = orchestrate_study_config_validation(data["study_config"])
    expected_token = hashlib.sha256(
        ("study-config-ready:" + validation.config_fingerprint).encode("utf-8")
    ).hexdigest()
    if validation.readiness_token != expected_token:
        raise IngestError("Configuration validator did not emit a digest-bound readiness token.")
    if "execution_ready" in validation.gap_matrix.columns and not bool(
        validation.gap_matrix["execution_ready"].astype(bool).all()
    ):
        raise IngestError("Configuration has unresolved production-critical gaps.")
    # Return the stage's artefacts.
    return {KEY_CONFIG: validation}


# Validate the configuration readiness token before any scientific adapter runs.
def _require_ready_config(
    state: Mapping[str, Any], data: Mapping[str, Any]
) -> ConfigValidationReport:
    """Return the validated configuration report or fail closed on stale state."""
    if KEY_CONFIG not in state or not isinstance(state[KEY_CONFIG], ConfigValidationReport):
        raise IngestError("A validated configuration readiness report is required.")
    if "study_config" not in data or not isinstance(data["study_config"], Mapping):
        raise IngestError("The validated study configuration is unavailable.")
    report: ConfigValidationReport = state[KEY_CONFIG]
    live_fingerprint = compute_config_fingerprint(data["study_config"])
    expected_token = hashlib.sha256(
        ("study-config-ready:" + live_fingerprint).encode("utf-8")
    ).hexdigest()
    if report.config_fingerprint != live_fingerprint or report.readiness_token != expected_token:
        raise IngestError("The configuration readiness token is stale or belongs to another config.")
    if "execution_ready" in report.gap_matrix.columns and not bool(
        report.gap_matrix["execution_ready"].astype(bool).all()
    ):
        raise IngestError("The configuration readiness report contains unresolved gaps.")
    return report


# Adapt the cleansing stage.
def _adapt_cleanse(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the cleansing stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the ingested bundle.
    data : mapping
        The caller mapping carrying `study_config` and optionally
        `iso3_registry` or `metadata_frame` for registry derivation.

    Returns
    -------
    mapping
        The cleansed bundle stored under the cleanse key.
    """
    ready = _require_ready_config(state, data)
    if state[KEY_BUNDLE].config_fingerprint != ready.config_fingerprint:
        raise IngestError("Ingest bundle and ready configuration fingerprints disagree.")
    # Compose the ISO-3 registry the cleansing consumes: the caller's explicit
    # registry when supplied, otherwise the declared country codes of the
    # caller's metadata frame, which is the same authoritative universe the
    # eligibility and partition stages consume.
    authoritative_metadata = state[KEY_BUNDLE].frame("country_metadata_raw")
    authoritative_registry = tuple(
        sorted(set(str(code) for code in authoritative_metadata["country_code"]))
    )
    iso3_registry = data.get("iso3_registry")
    if iso3_registry is not None:
        supplied_registry = tuple(sorted(str(code) for code in iso3_registry))
        if supplied_registry != authoritative_registry:
            raise IngestError("Caller ISO registry disagrees with authoritative country metadata.")
    iso3_registry = authoritative_registry
    # Invoke the cleansing orchestrator with its declared arguments; its
    # result is the cleansed bundle the downstream stages consume.
    cleansed = orchestrate_data_cleansing(
        state[KEY_BUNDLE], data["study_config"], iso3_registry=iso3_registry,
        universe_minimum=int(data["study_config"]["execution_parameters"]["raw_universe_minimum"]),
    )
    cleansed.verify_handoff()
    if cleansed.config_fingerprint != ready.config_fingerprint:
        raise IngestError("Cleansed bundle is not bound to the ready configuration.")
    # Return the stage's artefacts.
    return {KEY_CLEANSE: cleansed}


# Adapt the eligibility stage.
def _adapt_eligibility(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the eligibility stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the cleansed bundle.
    data : mapping
        The caller mapping optionally carrying `treatment_ledger`
        and `metadata_frame` for blindness and retained-flag checks.

    Returns
    -------
    mapping
        The eligibility screen stored under the eligibility key.
    """
    # Invoke the eligibility orchestrator on the CLEANSED bundle, supplying the
    # caller's documentary ledger and metadata when they are available so the
    # screen's blindness and retained-flag checks can run.
    _require_ready_config(state, data)
    policy_ledger = state[KEY_CLEANSE].frame("policy_ledger_raw")
    metadata_frame = state[KEY_CLEANSE].frame("country_metadata_raw")
    screen = orchestrate_eligibility_screening(
        state[KEY_CLEANSE], policy_ledger=policy_ledger, metadata_frame=metadata_frame
    )
    # Return the stage's artefacts.
    return {KEY_ELIGIBILITY: screen}


# Adapt the control-completion stage.
def _adapt_controls(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the control-completion stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the cleansed bundle and the eligibility screen with its mask.
    data : mapping
        The caller mapping, unused beyond the state for this stage.

    Returns
    -------
    mapping
        The completed control tensor stored under the controls key.
    """
    # Invoke the completion orchestrator on the CLEANSED bundle with the
    # screen's own mask.
    completion = orchestrate_control_completion(
        state[KEY_CLEANSE], mask=state[KEY_ELIGIBILITY].mask,
        expected_repaired_total=int(data["study_config"]["execution_parameters"]["completion"]["repaired_control_cells"]),
        expected_balance_derived=int(data["study_config"]["execution_parameters"]["completion"]["derived_fuel_balance_missing_cells"]),
    )
    # Return the stage's artefacts.
    return {KEY_CONTROLS: completion}


# Adapt the outcome-repair stage.
def _adapt_outcomes(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the outcome-repair stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the cleansed bundle and the eligibility screen with its mask.
    data : mapping
        The caller mapping, unused beyond the state for this stage.

    Returns
    -------
    mapping
        The repaired outcome tensor stored under the outcomes key.
    """
    # Invoke the repair orchestrator on the CLEANSED bundle with the screen's
    # own mask.
    repair = orchestrate_outcome_repair(
        state[KEY_CLEANSE], mask=state[KEY_ELIGIBILITY].mask
    )
    # Return the stage's artefacts.
    return {KEY_OUTCOMES: repair}


# Adapt the panel-assembly stage.
def _adapt_panel(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the panel-assembly stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the repaired outcomes and completed controls tensors.
    data : mapping
        The caller mapping, unused beyond the state for this stage.

    Returns
    -------
    mapping
        The assembled analysis panel stored under the panel key.
    """
    # Invoke the assembly orchestrator with its declared arguments, supplying
    # the repaired outcome tensor and the completed control tensor.
    _require_ready_config(state, data)
    control_digest = compute_frame_fingerprint(
        state[KEY_CONTROLS].tensor,
        key_columns=["country_code", "indicator_name", "year"],
        source_name="control_tensor",
    )
    outcome_digest = compute_frame_fingerprint(
        state[KEY_OUTCOMES].tensor,
        key_columns=["country_code", "indicator_name", "year"],
        source_name="outcome_tensor",
    )
    if control_digest != state[KEY_CONTROLS].fingerprint_sha256:
        raise IngestError("Control-completion tensor fingerprint is stale.")
    if outcome_digest != state[KEY_OUTCOMES].fingerprint_sha256:
        raise IngestError("Outcome-repair tensor fingerprint is stale.")
    panel = orchestrate_panel_assembly(
        state[KEY_OUTCOMES].tensor, state[KEY_CONTROLS].tensor
    )
    panel_digest = compute_frame_fingerprint(
        panel.panel, key_columns=["country_code", "year"], source_name="analysis_panel"
    )
    if panel_digest != panel.panel_fingerprint_sha256:
        raise IngestError("Analysis-panel fingerprint disagrees with its live panel.")
    if freeze_pre_treatment_matrices(panel.pre_treatment) != panel.matrices_fingerprint_sha256:
        raise IngestError("Pre-treatment matrix fingerprint disagrees with its live matrices.")
    # Return the stage's artefacts.
    return {KEY_PANEL: panel}


# Adapt the treatment-coding stage.
def _adapt_treatment(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the treatment-coding stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the eligibility screen with its retained universe.
    data : mapping
        The caller mapping carrying `treatment_ledger` with the documentary entries.

    Returns
    -------
    mapping
        The resolved treatment coding stored under the treatment key.
    """
    _require_ready_config(state, data)
    ledger = state[KEY_CLEANSE].frame("policy_ledger_raw")
    if "treatment_ledger" in data:
        external = data["treatment_ledger"]
        if not isinstance(external, pd.DataFrame) or _content_digest(external) != _content_digest(ledger):
            raise IngestError(
                "Caller treatment ledger is an inconsistent duplicate of the authoritative ingested source."
            )
    # Invoke the coding orchestrator with the authoritative ingested ledger and the
    # eligibility screen's own retained universe.
    coding = orchestrate_treatment_coding(
        ledger, _retained_universe(state[KEY_ELIGIBILITY]),
        expected_categories=configured_treatment_categories(data["study_config"]),
    )
    # Return the stage's artefacts.
    return {KEY_TREATMENT: coding}


# Adapt the partition-construction stage.
def _adapt_partitions(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the partition-construction stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying treatment coding, assembled panel, and eligibility retained universe.
    data : mapping
        The caller mapping carrying `metadata_frame` with region and geopolitical roles.

    Returns
    -------
    mapping
        The partition registry stored under the registry key.
    """
    _require_ready_config(state, data)
    metadata = state[KEY_CLEANSE].frame("country_metadata_raw")
    if "metadata_frame" in data:
        external = data["metadata_frame"]
        if not isinstance(external, pd.DataFrame) or _content_digest(external) != _content_digest(metadata):
            raise IngestError(
                "Caller metadata is an inconsistent duplicate of the authoritative ingested source."
            )
    # Invoke the partitioning orchestrator with the same authoritative
    # metadata and treatment tensor used by the preceding stages.
    registry = orchestrate_partition_construction(
        state[KEY_TREATMENT].indicators,
        state[KEY_PANEL].panel,
        metadata,
        _retained_universe(state[KEY_ELIGIBILITY]),
        expected_counts=configured_partition_counts(data["study_config"]),
        expected_roles=data["study_config"]["country_validation_registries"]["geopolitical_exclusions"],
    )
    audit_computed_partition_rosters(registry, data["study_config"])
    audit_pretreatment_group_means(state[KEY_PANEL].panel, registry, data["study_config"])
    if set(registry.partitions) != set(DECLARED_PAIRS):
        raise IngestError("Partition construction did not emit exactly the eight declared pairs.")
    if registry.fingerprint_sha256 != freeze_partition_registry(registry.registry):
        raise IngestError("Partition registry fingerprint is stale.")
    panel_codes = set(state[KEY_PANEL].panel["country_code"].astype(str))
    retained = set(_retained_universe(state[KEY_ELIGIBILITY]))
    if panel_codes != retained:
        raise IngestError("Panel and eligibility universes disagree at partition construction.")
    indicator_codes = set(state[KEY_TREATMENT].indicators["country_code"].astype(str))
    if not retained.issubset(indicator_codes):
        raise IngestError("Treatment tensor does not cover the retained partition universe.")
    # Return the stage's artefacts.
    return {KEY_REGISTRY: registry}


# Adapt the donor-factor stage.
def _adapt_factor(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the donor-factor stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the partition registry, assembled panel, and lagged controls.
    data : mapping
        The caller mapping, unused beyond the state for this stage.

    Returns
    -------
    mapping
        The donor factor objects stored under the factor key.
    """
    # Invoke the factor orchestrator with the registry, the panel and the lagged
    # controls the assembly produced.
    _require_ready_config(state, data)
    factor = orchestrate_donor_factor(
        state[KEY_REGISTRY], state[KEY_PANEL].panel, state[KEY_PANEL].lagged
    )
    if set(factor.factors) != set(DECLARED_PAIRS):
        raise IngestError("Donor-factor stage lacks a declared policy/outcome pair.")
    # Return the stage's artefacts.
    return {KEY_FACTOR: factor}


# Adapt the pretreatment-fit stage.
def _adapt_fit(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the pretreatment-fit stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the donor factor result.
    data : mapping
        The caller mapping, unused beyond the state for this stage.

    Returns
    -------
    mapping
        The pretreatment fit stored under the fit key.
    """
    # Invoke the fit orchestrator with the factor result.
    _require_ready_config(state, data)
    fit = orchestrate_pretreatment_fit(
        state[KEY_FACTOR],
        condition_max=float(data["study_config"]["execution_parameters"]["numerical"]["residual_maker_gram_condition_max"]),
    )
    expected = set(state[KEY_FACTOR].factors)
    if set(fit.betas) != expected or set(fit.loadings) != expected or set(fit.intermediates) != expected:
        raise IngestError("Pooled fit and donor-factor pair coverage disagree.")
    if not bool(fit.ledger.get("passed", pd.Series(True, index=fit.ledger.index)).astype(bool).all()):
        raise IngestError("A pooled-fit rank or orthogonality gate failed.")
    # Return the stage's artefacts.
    return {KEY_FIT: fit}


# Adapt the counterfactual-projection stage.
def _adapt_counterfactual(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the counterfactual-projection stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the pretreatment fit, donor factor, and partition registry.
    data : mapping
        The caller mapping, unused beyond the state for this stage.

    Returns
    -------
    mapping
        The counterfactual paths stored under the counterfactual key.
    """
    # Invoke the projection orchestrator with the fit, factor, registry and
    # realized panel used exclusively by the executed leakage audit.
    _require_ready_config(state, data)
    counterfactual = orchestrate_counterfactual_projection(
        state[KEY_FIT],
        state[KEY_FACTOR],
        state[KEY_REGISTRY],
        observed_panel=state[KEY_PANEL].panel,
    )
    if counterfactual.fitted_pre_paths.empty or counterfactual.outcome_paths.empty:
        raise IngestError("Counterfactual stage omitted a required fitted-pre or post path product.")
    if "invariant_holds" in counterfactual.invariant_audit.columns and not bool(
        counterfactual.invariant_audit["invariant_holds"].astype(bool).all()
    ):
        raise IngestError("Counterfactual leakage audit failed.")
    # Return the stage's artefacts.
    return {KEY_COUNTERFACTUAL: counterfactual}


# Adapt the gap-and-effect stage.
def _adapt_gaps(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the gap-and-effect stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the counterfactual, observed panel, and partition registry.
    data : mapping
        The caller mapping, unused beyond the state for this stage.

    Returns
    -------
    mapping
        The gap matrices and ATT effects stored under the gaps key.
    """
    # Invoke the gap orchestrator with the counterfactual, the observed panel
    # and the registry.
    _require_ready_config(state, data)
    windows, _ = resolve_reporting_windows(data["study_config"])
    gaps = orchestrate_gap_and_att(
        state[KEY_COUNTERFACTUAL], state[KEY_PANEL].panel, state[KEY_REGISTRY],
        wave_one=windows["wave_1_level"], wave_two=windows["wave_2_level"],
        pre_escalation=windows["pre_escalation"], post_escalation=windows["post_escalation"],
    )
    if set(gaps.gaps) != set(DECLARED_PAIRS):
        raise IngestError("Gap stage does not cover every declared pair.")
    fitted_pre = state[KEY_COUNTERFACTUAL].fitted_pre_paths
    if "year" not in fitted_pre.columns:
        raise IngestError("Distinct fitted-pre figure data lack a year column.")
    fitted_years = pd.to_numeric(fitted_pre["year"], errors="coerce")
    if (
        fitted_pre.empty
        or fitted_years.isna().any()
        or not np.isfinite(fitted_years.to_numpy(dtype=float)).all()
        or not all(float(value).is_integer() for value in fitted_years)
        or set(int(value) for value in fitted_years)
        != set(range(PRE_TREATMENT_WINDOW[0], PRE_TREATMENT_WINDOW[1] + 1))
    ):
        raise IngestError("Distinct fitted-pre figure data are incomplete or mis-scoped.")
    for pair, frame in gaps.gaps.items():
        if set(int(value) for value in frame.columns) != set(
            range(PROJECTION_START_YEAR, PROJECTION_END_YEAR + 1)
        ):
            raise IngestError(f"Gap matrix {pair!r} does not have the exact post path scope.")
    # Return the stage's artefacts.
    return {KEY_GAPS: gaps}


# Adapt the country-dispersion inference stage.
def _adapt_chain_inference(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the country-dispersion inference stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the gap-and-effect result.
    data : mapping
        The caller mapping, unused beyond the state for this stage.

    Returns
    -------
    mapping
        The country-dispersion inference stored under the inference-chain key.
    """
    # Invoke the inference orchestrator with the gap result.
    windows, _ = resolve_reporting_windows(data["study_config"])
    inference = orchestrate_country_dispersion_inference(
        state[KEY_GAPS], pre_window=windows["pre_escalation"],
        post_window=windows["post_escalation"],
    )
    # Return the stage's artefacts.
    return {KEY_INFERENCE_CHAIN: inference}


# Adapt the baseline-pipeline stage.
def _adapt_baseline(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the baseline-pipeline stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the registry, panel, lagged controls, and treatment indicators.
    data : mapping
        The caller mapping carrying `study_config` for the baseline pipeline.

    Returns
    -------
    mapping
        The baseline results and data mapping stored under the baseline and data-mapping keys.
    """
    ready = _require_ready_config(state, data)
    metadata = state[KEY_CLEANSE].frame("country_metadata_raw")
    income_groups = data.get("income_groups")
    if not isinstance(income_groups, Mapping) or not income_groups:
        raise IngestError("Authoritative income-group metadata is required for baseline variants.")
    repaired_by_indicator: Dict[str, Tuple[str, ...]] = {}
    repair_ledger = state[KEY_OUTCOMES].ledger
    for indicator_name, indicator_code in OUTCOME_CODES.items():
        if "indicator_code" in repair_ledger.columns:
            selected = repair_ledger[
                repair_ledger["indicator_code"].astype(str) == str(indicator_code)
            ]
        elif "indicator" in repair_ledger.columns:
            selected = repair_ledger[
                repair_ledger["indicator"].astype(str) == str(indicator_name)
            ]
        elif "indicator_name" in repair_ledger.columns:
            selected = repair_ledger[
                repair_ledger["indicator_name"].astype(str) == str(indicator_name)
            ]
        else:
            raise IngestError("Outcome repair ledger lacks an indicator identity column.")
        repaired_by_indicator[indicator_name] = tuple(
            sorted(set(selected["country_code"].astype(str)))
        )
    canonical_repaired = {
        CANONICAL_LOG_CPI: repaired_by_indicator["CPI"],
        CANONICAL_LOG_GDPPC: repaired_by_indicator["RealGDPpc"],
        CANONICAL_INFLATION_ATT: repaired_by_indicator["InflationRate"],
        CANONICAL_GROWTH_ATT: repaired_by_indicator["RealGDPpcGrowth"],
    }
    pair_repaired = {
        (str(policy), str(outcome)): tuple(sorted(
            set(str(code) for code in partition.donors)
            & set(canonical_repaired[str(outcome)])
        ))
        for (policy, outcome), partition in state[KEY_REGISTRY].partitions.items()
    }
    # Compose the data mapping the baseline interface consumes, which is the
    # registry together with the assembled panel, its lagged controls and the
    # country-year indicator tensor the active-year step requires.
    baseline_data: Dict[str, Any] = {
        "registry": state[KEY_REGISTRY],
        "panel": state[KEY_PANEL].panel,
        "lagged": state[KEY_PANEL].lagged,
        "coding": state[KEY_TREATMENT].indicators,
        "income_groups": dict(income_groups),
        "repaired_donor_codes": pair_repaired,
        "metadata": metadata.copy(),
        "config_readiness_token": ready.readiness_token,
        "config_fingerprint": ready.config_fingerprint,
        "panel_fingerprint_sha256": state[KEY_PANEL].panel_fingerprint_sha256,
        "lagged_fingerprint_sha256": _content_digest(state[KEY_PANEL].lagged),
        "treatment_fingerprint_sha256": state[KEY_TREATMENT].fingerprint_sha256,
    }
    # Invoke the baseline orchestrator with its declared arguments.
    baseline = orchestrate_baseline_pipeline(data["study_config"], baseline_data)
    # Return the stage's artefacts, carrying the composed data mapping so the
    # downstream stages consume the same structures the baseline consumed.
    return {KEY_BASELINE: baseline, KEY_DATA_MAPPING: baseline_data}


def _verified_region_metadata(state: Mapping[str, Any]) -> pd.DataFrame:
    """Return cleansed regions only when row and release identity match ingest."""
    cleansed_bundle = state[KEY_CLEANSE]
    ingest_bundle = state[KEY_BUNDLE]
    if not isinstance(cleansed_bundle, CleansedBundle) or not isinstance(
        ingest_bundle, IngestBundle
    ):
        raise IngestError("Region mapping requires ingested and cleansed source bundles.")
    metadata = cleansed_bundle.frame("country_metadata_raw")
    original_metadata = ingest_bundle.frame("country_metadata_raw")
    metadata_record = ingest_bundle.record("country_metadata_raw")
    metadata_schema = build_source_schemas()["country_metadata_raw"]
    if (
        not metadata.equals(original_metadata)
        or metadata_record.source_name != "country_metadata_raw"
        or tuple(metadata_record.key_columns) != tuple(metadata_schema.key_columns)
        or metadata_record.region_identity_sha256 is None
        or metadata_record.region_identity_sha256
        != _region_identity_digest(original_metadata)
        or metadata_record.region_identity_sha256 != _region_identity_digest(metadata)
        or cleansed_bundle.frame_digests.get("country_metadata_raw")
        != _source_handoff_digest("country_metadata_raw", metadata)
        or metadata_record.fingerprint_sha256
        != compute_frame_fingerprint(
            original_metadata, metadata_schema.key_columns, "country_metadata_raw"
        )
    ):
        raise IngestError(
            "Region metadata rows or source/version identity diverge from ten-source ingest."
        )
    return metadata


# Adapt the resampling stage.
def _adapt_resampling(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the resampling stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the baseline data mapping.
    data : mapping
        The caller mapping carrying ``study_config``, the registered primary
        and stability seed sets. Regions come from the cleansed metadata.

    Returns
    -------
    mapping
        The resampling inference stored under the resampling key.
    """
    _require_ready_config(state, data)
    metadata = _verified_region_metadata(state)
    # Invoke the resampling execution over the declared pairs with registered
    # seed material. Its region map is derived from the cleansed metadata and
    # must carry that source's version identity.
    resampling = execute_inference(
        data["study_config"],
        state[KEY_DATA_MAPPING],
        data["seeds"],
        metadata_frame=metadata,
        stability_seeds=data["stability_seeds"],
        bootstrap_replicates=int(data["study_config"]["algorithm_parameters"]["B1_country_history_bootstrap"]["replicates_B"]),
        placebo_replicates=int(data["study_config"]["algorithm_parameters"]["B3_size_matched_placebo"]["replicates_M"]),
    )
    # Return the stage's artefacts.
    return {KEY_RESAMPLING: resampling}


# Compose the entry year of every resolved country.
def _entry_year_map(treatment: Any) -> Dict[Tuple[str, str], int]:
    """
    Compose the entry year of every resolved country.

    Parameters
    ----------
    treatment : TreatmentCodingResult
        The treatment coding carrying the resolved country-policy entries.

    Returns
    -------
    dict of tuple to int
        The exactly one resolved entry year of each country-policy pair.
    """
    # Initialise the entry-year mapping.
    entry_years: Dict[Tuple[str, str], int] = {}
    # Iterate over the resolved country-policy entries.
    for _, row in treatment.resolved.iterrows():
        # Read the country and its entry year.
        country = str(row["country_code"])
        raw_policy = str(row["policy"])
        policy = POLICY_SANCTIONS if raw_policy in {"sanctions", POLICY_SANCTIONS} else raw_policy
        if policy not in {POLICY_SANCTIONS, POLICY_ARMING}:
            raise IngestError(f"Resolved treatment carries unknown policy {raw_policy!r}.")
        # Compose the resolved entry year.
        raw_year = row["entry_year"]
        if isinstance(raw_year, (bool, np.bool_)) or not np.isfinite(float(raw_year)) or float(raw_year) != int(float(raw_year)):
            raise IngestError(f"Resolved treatment has malformed entry year for {(country, policy)!r}.")
        year = int(float(raw_year))
        key = (country, policy)
        if key in entry_years:
            raise IngestError(f"Resolved treatment duplicates country-policy key {key!r}.")
        entry_years[key] = year
    # Return the entry-year mapping.
    return entry_years


# Select the country-keyed entry-year map for one policy without cross-policy collapse.
def _select_policy_entry_years(
    entry_years: Mapping[Any, int], policy: str
) -> Dict[str, int]:
    """Return one policy's unique country entry years from a tuple-keyed map."""
    selected: Dict[str, int] = {}
    for key, year in entry_years.items():
        if not isinstance(key, tuple) or len(key) != 2:
            raise IngestError("Policy-specific entry-year maps must use (country, policy) keys.")
        country, key_policy = str(key[0]), str(key[1])
        if key_policy == str(policy):
            if country in selected:
                raise IngestError(f"Duplicate entry year for {(country, policy)!r}.")
            selected[country] = int(year)
    if not selected:
        raise IngestError(f"No entry-year records are available for policy {policy!r}.")
    return selected


# Adapt the sweep-factory stage.
def _adapt_sweeps(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the sweep-factory stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying treatment coding and baseline data mapping.
    data : mapping
        The caller mapping carrying `study_config` for sweep construction.

    Returns
    -------
    mapping
        The sweep suite stored under the sweeps key.
    """
    _require_ready_config(state, data)
    # Compose the entry year of every resolved country, which the active-year
    # axis consumes as its per-country window argument, and the functional
    # window rule the declared variant names.
    entry_years: Dict[Tuple[str, str], int] = _entry_year_map(state[KEY_TREATMENT])
    # Invoke the factory with the configuration, the baseline data mapping,
    # the resolved entry years and the declared active-year window rule.
    sweeps = execute_sweeps(
        data["study_config"],
        state[KEY_DATA_MAPPING],
        entry_years=entry_years,
        active_year_callable=entry_year_active_window,
    )
    # Return the stage's artefacts.
    return {KEY_SWEEPS: sweeps}


# Adapt the robustness-sweeps stage.
def _adapt_robustness(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the robustness-sweeps stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the baseline data mapping.
    data : mapping
        The caller mapping carrying `study_config` for robustness execution.

    Returns
    -------
    mapping
        The robustness results stored under the robustness key.
    """
    # Invoke the robustness execution with the configuration and the mapping.
    _require_ready_config(state, data)
    robustness = execute_robustness_sweeps(
        data["study_config"], state[KEY_DATA_MAPPING],
        threshold=float(data["study_config"]["execution_parameters"]["numerical"]["factor_proxy_condition_flag"]),
    )
    if not robustness.coverage_audit.empty and "execution_status" in robustness.coverage_audit.columns:
        coverage = robustness.coverage_audit
        statuses = coverage["execution_status"].astype(str)
        verified_identity = (
            statuses.eq("identity")
            & (coverage.index.get_level_values("axis").astype(str) == AXIS_DONOR_POOL)
            & (coverage.index.get_level_values("variant_name").astype(str)
               == "no_repaired_main_outcome_cells")
            & coverage["audit_status"].astype(str).eq("verified_identity")
            & ~coverage["included_in_headline_range"].astype(bool)
        )
        unresolved = robustness.coverage_audit[
            ~(statuses.eq("success") | verified_identity)
        ]
        if not unresolved.empty:
            raise IngestError(
                f"Robustness execution retains {len(unresolved)} unverified registered run(s)."
            )
    # Return the stage's artefacts.
    return {KEY_ROBUSTNESS: robustness}


# Adapt the secondary-sweeps stage.
def _adapt_secondary(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the secondary-sweeps stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the baseline results with incremental estimates.
    data : mapping
        The caller mapping carrying `study_config` for secondary execution.

    Returns
    -------
    mapping
        The secondary sweep results stored under the secondary key.
    """
    _require_ready_config(state, data)
    # Compose the baseline effects the active-year comparison consumes from the
    # baseline result's own incremental estimates.
    baseline_effects = {
        pair: float(result.att_incremental)
        for pair, result in state[KEY_BASELINE].results.items()
    }
    # Invoke the secondary sweeps with the declared arguments.
    secondary = execute_secondary_sweeps(
        data["study_config"],
        state[KEY_DATA_MAPPING],
        baseline_effects=baseline_effects,
        tolerance=float(data["study_config"]["execution_parameters"]["numerical"]["comparison_tolerance"]),
    )
    # Return the stage's artefacts.
    return {KEY_SECONDARY: secondary}


# Adapt the influence-diagnostics stage.
def _adapt_influence(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the influence-diagnostics stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the baseline data mapping.
    data : mapping
        The caller mapping carrying `study_config` for diagnostics.

    Returns
    -------
    mapping
        The influence diagnostics stored under the influence key.
    """
    # Invoke the diagnostics with the configuration and the mapping.
    influence = execute_influence_diagnostics(
        data["study_config"], state[KEY_DATA_MAPPING],
        count=int(data["study_config"]["execution_parameters"]["influence"]["top_donors"]),
    )
    # Return the stage's artefacts.
    return {KEY_INFLUENCE: influence}


# Adapt the influence-tables stage.
def _adapt_influence_tables(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the influence-tables stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the influence result and baseline contrasts.
    data : mapping
        The caller mapping, unused beyond the state for this stage.

    Returns
    -------
    mapping
        The influence tables stored under the influence-tables key.
    """
    # Compose the per-pair contrasts the tables consume from the baseline's own
    # result objects.
    contrasts = {
        pair: result.incremental_contrasts
        for pair, result in state[KEY_BASELINE].results.items()
    }
    # Invoke the tables with the influence result and the contrasts.
    settings = data["study_config"]["execution_parameters"]["influence"]
    tables = execute_influence_tables(
        state[KEY_INFLUENCE], contrasts, count=int(settings["top_donors"]),
        muted_threshold=float(settings["muted_response_threshold"]),
    )
    # Return the stage's artefacts.
    return {KEY_INFLUENCE_TABLES: tables}


# Adapt the matching-designs stage.
def _adapt_matching(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the matching-designs stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the baseline data mapping.
    data : mapping
        The caller mapping carrying `study_config` for matching execution.

    Returns
    -------
    mapping
        The matching designs stored under the matching key.
    """
    # Invoke the matching orchestrator with the configuration and the mapping.
    _require_ready_config(state, data)
    matching = execute_matching_designs(
        data["study_config"], state[KEY_DATA_MAPPING],
        size_M=int(data["study_config"]["algorithm_parameters"]["E2_mahalanobis_matching"]["M_neighbors"]),
        tolerance=float(data["study_config"]["execution_parameters"]["numerical"]["comparison_tolerance"]),
    )
    assert_configured_matching_counts(matching, data["study_config"])
    # Return the stage's artefacts.
    return {KEY_MATCHING: matching}


# Adapt the matching-report stage.
def _adapt_matching_report(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the matching-report stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the baseline data mapping.
    data : mapping
        The caller mapping carrying `study_config` for reporting.

    Returns
    -------
    mapping
        The matching report stored under the matching-report key.
    """
    _require_ready_config(state, data)
    # Invoke the report from the exact designs already executed by Task 23.
    report = execute_matching_report(
        data["study_config"], state[KEY_DATA_MAPPING], diagnostics=state[KEY_MATCHING],
        size_M=int(data["study_config"]["algorithm_parameters"]["E2_mahalanobis_matching"]["M_neighbors"]),
        tolerance=float(data["study_config"]["execution_parameters"]["numerical"]["comparison_tolerance"]),
    )
    # Return the stage's artefacts.
    return {KEY_MATCHING_REPORT: report}


# Adapt the cell-ledger stage.
def _adapt_cell_ledger(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the cell-ledger stage.

    Parameters
    ----------
    state : mapping
        The artefact state, unused beyond the caller data for this stage.
    data : mapping
        The caller mapping. The authoritative cell-status source already
        carries its retry evidence; no disconnected retry table is consumed.

    Returns
    -------
    mapping
        The classified cell ledger stored under the cell-ledger key.
    """
    _require_ready_config(state, data)
    raw_status = state[KEY_CLEANSE].frame("comtrade_cell_status_raw")
    if "comtrade_cell_status_raw" in data and _content_digest(data["comtrade_cell_status_raw"]) != _content_digest(raw_status):
        raise IngestError("External cell-status input disagrees with the authoritative ingested source.")
    # Invoke ledger construction from the authoritative raw cell block alone.
    ledger = build_cell_ledger(
        raw_status, retry_records=None,
        expected_counts=data["study_config"]["execution_parameters"]["comtrade_status_counts"],
    )
    # Return the stage's artefacts.
    return {KEY_CELL_LEDGER: ledger}


# Adapt the exposure-primitives stage.
def _totals_from_block_c(
    block_c: pd.DataFrame,
    world_ledger: pd.DataFrame,
    source_record: IngestRecord,
) -> pd.DataFrame:
    """Reconcile Block C and derive optional exporter totals from its raw evidence."""
    source_name = "comtrade_block_c_raw"
    schema = build_source_schemas()[source_name]
    validate_columns(block_c, schema, source_name)
    if (
        not isinstance(source_record, IngestRecord)
        or source_record.source_name != source_name
        or tuple(source_record.key_columns) != ("reporter_code", "year")
        or source_record.n_rows != len(block_c)
        or source_record.n_columns != len(block_c.columns)
        or tuple(source_record.validation_stages) != (
            "canonicalize_missing_markers", "normalize_year_columns",
            "enforce_dtypes", "validate_columns", "assert_primary_key_unique",
        )
        or source_record.fingerprint_sha256 != compute_frame_fingerprint(
            block_c, schema.key_columns, source_name
        )
    ):
        raise IngestError("Block C is not bound to its validated ten-source ingest record.")
    if bool(block_c.duplicated(["reporter_code", "year"]).any()):
        raise IngestError("Block C contains duplicate reporter-year keys.")
    if (
        not block_c["partner_code"].astype(str).eq("WLD").all()
        or not block_c["trade_flow"].astype(str).str.lower().eq("import").all()
        or not block_c["product_description"].astype(str)
        .str.lower().str.replace("_", " ", regex=False).eq("total gas").all()
    ):
        raise IngestError("Block C must contain world-partner total-gas imports only.")
    allowed_years = {
        year for window in EXPOSURE_COMPLETION_WINDOWS.values() for year in window
    }
    # The fingerprint above authenticates the entire Block C source.  Only its
    # declared exposure-window rows enter denominator reconciliation; source
    # rows from other calendar years remain in the archived ten-source record.
    scoped_block_c = block_c.loc[block_c["year"].isin(allowed_years)].copy()
    world = world_ledger.loc[:, ["reporter_code", "year", VALUE_COLUMN]].copy()
    if bool(world.duplicated(["reporter_code", "year"]).any()):
        raise IngestError("The world-gas ledger has duplicate reporter-year keys.")
    if not set(world["year"].astype(int)) <= allowed_years:
        raise IngestError("World-gas ledger leaves the declared exposure windows.")
    block_keys = set(scoped_block_c[["reporter_code", "year"]].itertuples(index=False, name=None))
    world_keys = set(world[["reporter_code", "year"]].itertuples(index=False, name=None))
    if not block_keys <= world_keys:
        raise IngestError("Block C carries reporter-years outside the world-gas ledger.")
    block_values = scoped_block_c.set_index(["reporter_code", "year"])[VALUE_COLUMN]
    world_values = world.set_index(["reporter_code", "year"])[VALUE_COLUMN]
    for key, block_value in block_values.items():
        ledger_value = world_values.loc[key]
        if not (
            pd.isna(ledger_value) and pd.isna(block_value)
            or pd.notna(ledger_value) and pd.notna(block_value)
            and float(ledger_value) == float(block_value)
        ):
            raise IngestError(f"Block C disagrees with the world-gas ledger at {key}.")
    totals = world.rename(columns={VALUE_COLUMN: "world_total_usd"})
    totals["block_c_importer_row_present"] = [
        (str(country), int(year)) in block_keys
        for country, year in totals[["reporter_code", "year"]].itertuples(
            index=False, name=None
        )
    ]
    totals = totals.set_index(["reporter_code", "year"])
    totals["exporter_sum_usd"] = np.nan
    totals["exporter_sum_source_id"] = ""
    totals["exporter_sum_components_sha256"] = ""
    totals["block_c_source_sha256"] = source_record.fingerprint_sha256
    totals["exporter_sum_query_sha256"] = ""
    totals["block_c_recovered_via_claim"] = ""
    for _, source_row in block_c.iterrows():
        key = (str(source_row["reporter_code"]), int(source_row["year"]))
        raw_claim = source_row["recovered_via"]
        claim = "" if pd.isna(raw_claim) else str(raw_claim).strip().lower()
        if claim in {"none", "missing", "unresolved"}:
            claim = ""
        if claim not in {"", SUBSTITUTE_IMPORTER, SUBSTITUTE_EXPORTER_SUM}:
            raise IngestError(f"Block C recovered_via category is invalid at {key}.")
        if key in block_keys:
            totals.at[key, "block_c_recovered_via_claim"] = claim
    if schema.extension_columns[0] not in block_c.columns:
        return totals.reset_index()
    def unique_json_object(pairs: List[Tuple[str, Any]]) -> Dict[str, Any]:
        if len(pairs) != len({key for key, _ in pairs}):
            raise ValueError("duplicate JSON object key")
        return dict(pairs)
    for _, source_row in scoped_block_c.iterrows():
        key = (str(source_row["reporter_code"]), int(source_row["year"]))
        fields = [source_row[column] for column in schema.extension_columns]
        if all(pd.isna(value) or str(value).strip() == "" for value in fields):
            continue
        if any(pd.isna(value) or str(value).strip() == "" for value in fields):
            raise IngestError(f"Block C exporter evidence is partial at {key}.")
        try:
            components = json.loads(
                str(source_row["exporter_sum_components_json"]),
                object_pairs_hook=unique_json_object,
            )
            query = json.loads(
                str(source_row["exporter_sum_query_json"]),
                object_pairs_hook=unique_json_object,
            )
        except (TypeError, ValueError) as exc:
            raise IngestError(f"Block C exporter evidence is invalid JSON at {key}.") from exc
        if not isinstance(components, list) or not components:
            raise IngestError(f"Block C exporter components are empty at {key}.")
        expected_query = {
            "source": "UN_COMTRADE", "reporter_scope": "all_exporters",
            "partner_code": key[0], "trade_flow": "export",
            "product_codes": ["271111", "271121"], "year": key[1],
        }
        if query != expected_query:
            raise IngestError(f"Block C exporter query does not cover both gas products at {key}.")
        source_id = str(source_row["exporter_sum_source_id"]).strip()
        if not re.fullmatch(r"UN_COMTRADE:[A-Za-z0-9._:/?&=%+-]+", source_id):
            raise IngestError(f"Block C exporter source identity is invalid at {key}.")
        seen_components: Set[Tuple[str, str]] = set()
        component_values: List[float] = []
        for component in components:
            if not isinstance(component, dict) or set(component) != {
                "exporter_code", "product_code", "value_usd"
            }:
                raise IngestError(f"Block C exporter component schema is invalid at {key}.")
            exporter_code = component["exporter_code"]
            product_code = component["product_code"]
            value = component["value_usd"]
            if (
                not isinstance(exporter_code, str)
                or not re.fullmatch(r"[A-Z]{3}", exporter_code)
                or product_code not in {"271111", "271121"}
                or isinstance(value, (bool, np.bool_))
                or not isinstance(value, (int, float))
                or not np.isfinite(float(value))
                or float(value) < 0.0
                or (exporter_code, product_code) in seen_components
            ):
                raise IngestError(f"Block C exporter component is invalid at {key}.")
            seen_components.add((exporter_code, product_code))
            component_values.append(float(value))
        canonical_components = sorted(
            components, key=lambda item: (item["exporter_code"], item["product_code"])
        )
        digest = hashlib.sha256(json.dumps(
            canonical_components, sort_keys=True, separators=(",", ":"),
        ).encode("utf-8")).hexdigest()
        if str(source_row["exporter_sum_components_sha256"]).strip() != digest:
            raise IngestError(f"Block C exporter components digest disagrees at {key}.")
        exporter_total = float(math.fsum(component_values))
        if not np.isfinite(exporter_total):
            raise IngestError(f"Block C exporter sum overflows at {key}.")
        totals.at[key, "exporter_sum_usd"] = exporter_total
        totals.at[key, "exporter_sum_source_id"] = source_id
        totals.at[key, "exporter_sum_components_sha256"] = digest
        totals.at[key, "exporter_sum_query_sha256"] = hashlib.sha256(
            json.dumps(query, sort_keys=True, separators=(",", ":")).encode("utf-8")
        ).hexdigest()
    return totals.reset_index()


# Adapt the exposure-primitives stage.
def _adapt_primitives(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the exposure-primitives stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the classified cell ledger.
    data : mapping
        The caller mapping. All trade and nominal-GDP inputs are obtained from
        the cleansed ten-source bundle. An optional exporter-component extension
        to Block C supplies independently reported denominator alternatives.

    Returns
    -------
    mapping
        The completed exposure primitives stored under the primitives key.
    """
    ready = _require_ready_config(state, data)
    ingest_bundle = state[KEY_BUNDLE]
    cleansed_bundle = state[KEY_CLEANSE]
    if not isinstance(ingest_bundle, IngestBundle) or not isinstance(
        cleansed_bundle, CleansedBundle
    ):
        raise IngestError("Task 25 requires the ingested and cleansed ten-source bundles.")
    if (
        set(ingest_bundle.frames) != set(DECLARED_SOURCES)
        or set(ingest_bundle.records) != set(DECLARED_SOURCES)
        or set(cleansed_bundle.frames) != set(DECLARED_SOURCES)
        or set(cleansed_bundle.frame_digests) != set(DECLARED_SOURCES)
        or ingest_bundle.config_fingerprint != ready.config_fingerprint
        or cleansed_bundle.config_fingerprint != ready.config_fingerprint
    ):
        raise IngestError("Task 25 ten-source/configuration lineage is incomplete.")
    cleansed_bundle.verify_handoff()
    full_ledger = state[KEY_CELL_LEDGER].ledger
    exposure_years = {
        year for window in EXPOSURE_COMPLETION_WINDOWS.values() for year in window
    }
    ledger = full_ledger.loc[full_ledger["year"].isin(exposure_years)].copy()
    if ledger.empty:
        raise IngestError("The cell ledger has no declared exposure-window rows.")
    mirror = state[KEY_CLEANSE].frame("comtrade_block_b_raw")
    recovered_for_audit = recover_mirror_flow(ledger, mirror)
    primitives_audit = ledger.loc[:, ["reporter_code", "year", "series", VALUE_COLUMN]].copy()
    russian = recovered_for_audit[
        recovered_for_audit["series"].map(_canonical_trade_series) == "russian_gas"
    ].copy()
    numerator = russian.loc[:, ["reporter_code", "year", "effective_value_usd"]].rename(
        columns={"effective_value_usd": "russian_flow_usd"}
    )
    world = primitives_audit[
        primitives_audit["series"].map(_canonical_trade_series) == "world_gas"
    ].copy()
    block_c = cleansed_bundle.frame("comtrade_block_c_raw")
    original_block_c = ingest_bundle.frame("comtrade_block_c_raw")
    if not block_c.equals(original_block_c):
        raise IngestError(
            "Cleansed Block C differs from its authoritative ten-source ingest frame."
        )
    totals = _totals_from_block_c(
        block_c, world, ingest_bundle.record("comtrade_block_c_raw")
    )
    # Keep a raw missing importer total missing when an independent substitute
    # is unavailable; D4 may subsequently complete that cell within its own
    # window. Contradictory observed totals and false substitution claims fail.
    denominator_check = construct_denominator(numerator, totals)
    invalid: List[Tuple[str, int, str]] = []
    for (country, year), row in denominator_check.iterrows():
        key = (str(country), int(year))
        claim = str(row["block_c_recovered_via_claim"])
        mechanism = str(row["substitution_mechanism"])
        reason = str(row["denominator_selection_reason"])
        if claim == SUBSTITUTE_IMPORTER and mechanism != SUBSTITUTE_IMPORTER:
            invalid.append((*key, "false importer_reported claim"))
        if claim == SUBSTITUTE_EXPORTER_SUM and (
            mechanism != SUBSTITUTE_EXPORTER_SUM
            or not str(row["exporter_sum_components_sha256"])
            or not str(row["exporter_sum_source_id"])
        ):
            invalid.append((*key, "unverified exporter_sum_substitute claim"))
        if mechanism == SUBSTITUTE_EXPORTER_SUM and claim != SUBSTITUTE_EXPORTER_SUM:
            invalid.append((*key, "selected exporter sum lacks Block C mechanism claim"))
        if reason == "rejected_numerator_exceeds_selected_denominator":
            invalid.append((*key, reason))
        if reason == "required_exporter_substitute_unavailable" and pd.notna(
            row["world_total_usd"]
        ):
            invalid.append((*key, "contradictory observed importer total lacks substitute"))
        if bool(row["denominator_rejected"]) and reason not in {
            "required_exporter_substitute_unavailable",
            "rejected_numerator_exceeds_selected_denominator",
        }:
            invalid.append((*key, f"unexpected denominator rejection: {reason}"))
    if invalid:
        raise IngestError(
            "Block C denominator evidence is contradictory at reporter-year(s) "
            f"{invalid[:MAX_REPORTED_OFFENDERS]}."
        )
    nominal_gdp = state[KEY_CLEANSE].frame("nominal_gdp_raw")
    # Invoke primitive construction only with audits derived from the upstream
    # authoritative ledger and ingested trade/GDP blocks.
    primitives = construct_exposure_primitives(
        ledger,
        mirror,
        numerator,
        totals,
        primitives_audit,
        nominal_gdp=nominal_gdp,
        maximum_gap=int(data["study_config"]["algorithm_parameters"]["D4_within_window_completion"]["internal_gap_max_consecutive"]),
        minimum_anchors=int(data["study_config"]["algorithm_parameters"]["D4_within_window_completion"]["edge_extrapolation_min_anchors"]),
        expected_completions=data["study_config"]["algorithm_parameters"]["D4_within_window_completion"]["primitive_imputations"],
        tolerance=int(data["study_config"]["execution_parameters"]["numerical"]["exposure_completion_count_tolerance"]),
    )
    # Return the stage's artefacts.
    return {KEY_PRIMITIVES: primitives}


# Adapt the exposure-ratios stage.
def _adapt_ratios(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the exposure-ratios stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying completed primitives and eligibility retained universe.
    data : mapping
        The caller mapping. If present, `exposure_windows` must agree exactly
        with the validated configuration and canonical named windows.

    Returns
    -------
    mapping
        The exposure ratios stored under the ratios key.
    """
    _require_ready_config(state, data)
    configured_windows = read_dotted_path(
        data["study_config"],
        "series_code_registry.temporal_universe_by_layer.exposure_windows",
    )
    if not _values_match(
        configured_windows, EXPECTED_TEMPORAL_LAYERS["exposure_windows"]
    ):
        raise IngestError("Validated exposure windows diverge from the declared study periods.")
    windows = {
        name: tuple(years) for name, years in EXPOSURE_COMPLETION_WINDOWS.items()
    }
    if "exposure_windows" in data and not _values_match(
        data["exposure_windows"], windows
    ):
        raise IngestError(
            "Caller exposure windows disagree with the authoritative named periods."
        )
    # Read the completed primitives of the previous stage.
    primitives = state[KEY_PRIMITIVES]
    # Select the completed frame's own columns and rename them into the
    # component shape the ratio construction declares: the reporter becomes the
    # country and the completed value becomes the value. Selection precedes the
    # rename so the original value column cannot shadow the completed one.
    components: pd.DataFrame = primitives.completed[
        [
            "reporter_code",
            "series",
            "year",
            "completed_value_usd",
            "was_completed",
            "value_provenance",
            "original_anchor",
        ]
    ].rename(
        columns={
            "reporter_code": "country_code",
            "completed_value_usd": "value_usd",
        }
    )
    # Invoke the ratio construction with the completed components, the
    # completion ledger, the declared windows and the retained universe.
    ratios = construct_exposure_ratios(
        components,
        primitives.imputation_ledger,
        windows,
        _retained_universe(state[KEY_ELIGIBILITY]),
        expected_counts=configured_exposure_counts(data["study_config"]),
    )
    # Return the stage's artefacts.
    return {KEY_RATIOS: ratios}


# Adapt the heterogeneity-suite stage.
def _adapt_heterogeneity(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the heterogeneity-suite stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying the partition registry, mechanism panel inputs, and exposure ratios.
    data : mapping
        The caller mapping, unused beyond the state for this stage.

    Returns
    -------
    mapping
        The heterogeneity suite stored under the heterogeneity key.
    """
    _require_ready_config(state, data)
    validate_heterogeneity_specification_bindings(data["study_config"])
    # Materialize the exact independently keyed samples as a validation audit;
    # the estimator itself receives the complete panel so non-policy donors are
    # never lost merely because they lack exposure observations.
    _mechanism_panel(state)
    # Compose the policy-specific donor rosters consumed by the declared suite.
    donor_rosters = {
        (policy, outcome): tuple(
            str(code)
            for code in state[KEY_REGISTRY].partitions[(policy, outcome)].donors
        )
        for policy, outcome in DECLARED_PAIRS
    }
    # Invoke the suite for the primary outcome with the standardized exposures
    # over the mechanism panel the coverage rules admit.
    suite = execute_heterogeneity_suite(
        state[KEY_PANEL].panel,
        "log_CPI",
        _standardized_exposures(state[KEY_RATIOS]),
        _policy_indicator_frame(state[KEY_TREATMENT]),
        donor_rosters,
    )
    expected_specifications = {record.name for record in SPECIFICATIONS}
    if set(suite.statistics.index.get_level_values("specification")) != expected_specifications:
        raise IngestError("Heterogeneity suite does not cover the six registered specifications.")
    # Return the stage's artefacts.
    return {KEY_HETEROGENEITY: suite}


# Adapt the heterogeneity-regressions stage.
def _adapt_heterogeneity_regressions(
    state: Mapping[str, Any], data: Mapping[str, Any]
) -> Mapping[str, Any]:
    """
    Adapt the heterogeneity-regressions stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying mechanism panel inputs, standardized exposures, and donor roster.
    data : mapping
        The caller mapping, unused beyond the state for this stage.

    Returns
    -------
    mapping
        The heterogeneity regressions stored under the heterogeneity-regressions key.
    """
    _require_ready_config(state, data)
    validate_heterogeneity_specification_bindings(data["study_config"])
    _mechanism_panel(state)
    # Compose the exposures in the wide form the regression consumes.
    exposures = _standardized_exposures(state[KEY_RATIOS])
    # Compose exact policy-specific donor rosters rather than reusing the
    # sanctions roster for the direct-arming specification.
    donors = {
        (policy, outcome): tuple(
            str(code)
            for code in state[KEY_REGISTRY].partitions[(policy, outcome)].donors
        )
        for policy, outcome in DECLARED_PAIRS
    }
    # Invoke the regression suite across the declared outcomes over the
    # mechanism panel the coverage rules admit.
    regressions = execute_heterogeneity_regressions(
        state[KEY_PANEL].panel,
        exposures,
        _policy_indicator_frame(state[KEY_TREATMENT]),
        donors,
    )
    if set(regressions.sweep_table.index.get_level_values("outcome")) != set(PANEL_OUTCOMES):
        raise IngestError("Heterogeneity regressions do not cover all four declared outcomes.")
    # Return the stage's artefacts.
    return {KEY_HETEROGENEITY_REG: regressions}


# Adapt the direct-arming pipeline stage.
def _adapt_direct_arming(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the direct-arming pipeline stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying treatment indicators, baseline data and
        cleansed authoritative country metadata.
    data : mapping
        The caller mapping carrying `study_config` and registered seeds. An
        optional `region_map` is checked against, never substituted for, the
        versioned metadata-derived mapping.

    Returns
    -------
    mapping
        The direct-arming results stored under the arming key.
    """
    _require_ready_config(state, data)
    registry = state[KEY_DATA_MAPPING].get("registry")
    if not isinstance(registry, PartitionRegistry):
        raise IngestError("Direct-arming region mapping requires a frozen partition registry.")
    required_countries: Set[str] = set()
    for outcome in ARMING_OUTCOMES:
        pair = (POLICY_DIRECT_ARMING, outcome)
        if pair not in registry.partitions:
            raise IngestError(f"Direct-arming region mapping omits partition {pair}.")
        partition = registry.partitions[pair]
        required_countries.update(str(code) for code in partition.targets)
        required_countries.update(str(code) for code in partition.donors)
    metadata = _verified_region_metadata(state)
    region_map = build_region_map(
        metadata, required_countries=sorted(required_countries)
    )
    if "region_map" in data:
        supplied_regions = data["region_map"]
        supplied_provenance = getattr(supplied_regions, "provenance", None)
        if (
            not isinstance(supplied_regions, Mapping)
            or dict(supplied_regions) != dict(region_map)
            or (
                supplied_provenance is not None
                and (
                    not isinstance(supplied_provenance, Mapping)
                    or dict(supplied_provenance) != dict(region_map.provenance)
                )
            )
        ):
            raise IngestError(
                "Caller region_map disagrees with the authoritative metadata-derived map."
            )
    entry_years = _select_policy_entry_years(
        _entry_year_map(state[KEY_TREATMENT]), POLICY_ARMING
    )
    # Invoke the arming pipeline with the configuration, the mapping, the entry
    # years and the versioned metadata-derived region map.
    arming = execute_direct_arming_pipeline(
        data["study_config"],
        state[KEY_DATA_MAPPING],
        entry_years,
        region_map,
        seeds=data["seeds"],
        bootstrap_replicates=int(data["study_config"]["algorithm_parameters"]["B1_country_history_bootstrap"]["replicates_B"]),
        placebo_replicates=int(data["study_config"]["algorithm_parameters"]["B3_size_matched_placebo"]["replicates_M"]),
        wave_references={key: (None if value is None else tuple(value)) for key, value in data["study_config"]["execution_parameters"]["direct_arming_wrapper"]["wave_references"].items()},
        tolerance=float(data["study_config"]["execution_parameters"]["numerical"]["comparison_tolerance"]),
    )
    # Return the stage's artefacts.
    return {KEY_ARMING: arming}


# Collect finite scalar values from one computed artefact for table-cell binding.
def _artifact_numeric_values(value: Any) -> Tuple[float, ...]:
    """Return finite numeric leaves of an artefact without inventing selectors."""
    values: List[float] = []
    if isinstance(value, pd.DataFrame):
        for column in value.select_dtypes(include=[np.number]).columns:
            array = pd.to_numeric(value[column], errors="coerce").to_numpy(dtype=float)
            values.extend(float(item) for item in array[np.isfinite(array)])
    elif isinstance(value, pd.Series):
        array = pd.to_numeric(value, errors="coerce").to_numpy(dtype=float)
        values.extend(float(item) for item in array[np.isfinite(array)])
    elif is_dataclass(value) and not isinstance(value, type):
        for item in fields(value):
            values.extend(_artifact_numeric_values(getattr(value, item.name)))
    elif isinstance(value, Mapping):
        for item in value.values():
            values.extend(_artifact_numeric_values(item))
    elif isinstance(value, (tuple, list)):
        for item in value:
            values.extend(_artifact_numeric_values(item))
    elif isinstance(value, (int, float, np.number)) and not isinstance(value, (bool, np.bool_)):
        numeric = float(value)
        if np.isfinite(numeric):
            values.append(numeric)
    return tuple(values)


# Enforce the exact manuscript universe before publication output can be made.
def _validate_production_manuscript_counts(
    counts: Mapping[str, int], inventory_size: int
) -> None:
    """Reject under- and over-complete production cell inventories alike."""
    if set(counts) != set(PRODUCTION_TABLES):
        raise IngestError("Trusted table counts do not cover exactly the declared tables.")
    mismatched = {
        table_id: (counts[table_id], int(TABLE_MINIMUM_REQUIRED_CELLS[table_id]))
        for table_id in PRODUCTION_TABLES
        if counts[table_id] != int(TABLE_MINIMUM_REQUIRED_CELLS[table_id])
    }
    if mismatched or inventory_size != MANUSCRIPT_CELL_COUNT:
        raise IngestError(
            "Trusted production inventory does not match the exact manuscript "
            f"cell universe of {MANUSCRIPT_CELL_COUNT} cells: {mismatched}."
        )


# Verify publication table cells are bound to computed pipeline artefacts.
def _verified_table_sources(
    state: Mapping[str, Any],
    raw_sources: Any,
    trusted_authority: Optional[Mapping[str, Any]] = None,
) -> Mapping[str, pd.DataFrame]:
    """
    Return table sources bound to a separate trusted cell-source authority.

    ``raw_sources`` supplies layout and proposed values only.  Its DataFrame
    attributes are never a trust anchor.  ``trusted_authority`` must be an
    independently obtained mapping that pins the complete inventory digest,
    table counts, every computed-cell selector, every unresolved gap, and the
    combined registry digest.  Each computed selector is then resolved against
    the live pipeline state, whose current full-content digest and scalar value
    must agree before the value is copied into the verified table.

    Parameters
    ----------
    state : mapping of str to Any
        Live upstream pipeline artefacts addressable by authoritative selectors.
    raw_sources : Any
        Caller-supplied table frames plus ``__cell_inventory__``.
    trusted_authority : mapping of str to Any, optional
        Versioned authority document.  Absence fails closed; copying the same
        assertions into DataFrame attributes does not satisfy this input.

    Returns
    -------
    mapping of str to pandas.DataFrame
        Defensive copies whose values and provenance were reconstructed from
        the trusted selectors and live artefacts.

    Raises
    ------
    IngestError
        If the trust anchor, inventory, status universe, selector, live digest,
        scalar value, or table coverage is absent, stale, duplicated or extra.
    """
    if not isinstance(raw_sources, Mapping):
        raise IngestError("Production requires computed publication table sources.")
    if not isinstance(trusted_authority, Mapping) or not trusted_authority:
        raise IngestError(
            "Production requires a separate trusted table-binding authority; "
            "table or inventory attributes cannot authenticate themselves."
        )
    inventory_source = raw_sources.get("__cell_inventory__")
    if not isinstance(inventory_source, pd.DataFrame) or inventory_source.empty:
        raise IngestError("Production requires the independent manuscript cell inventory.")
    # Read only the independent authority mapping; similarly named frame attrs
    # are deliberately ignored and will be replaced on verified copies.
    authority_id = str(trusted_authority.get("authority_id", "")).strip()
    authority_version = str(trusted_authority.get("version_id", "")).strip()
    external_reference = str(
        trusted_authority.get("external_verification_reference", "")
    ).strip()
    inventory_sha256 = str(
        trusted_authority.get("inventory_sha256", "")
    ).strip().lower()
    binding_registry_sha256 = str(
        trusted_authority.get("binding_registry_sha256", "")
    ).strip().lower()
    expected_counts = trusted_authority.get("expected_cell_counts")
    expected_bindings = trusted_authority.get("expected_cell_bindings")
    expected_unresolved = trusted_authority.get("expected_unresolved_gaps")
    if (
        not authority_id
        or not authority_version
        or not external_reference
        or not re.fullmatch(r"[0-9a-f]{64}", inventory_sha256)
        or not re.fullmatch(r"[0-9a-f]{64}", binding_registry_sha256)
        or not isinstance(expected_counts, Mapping)
        or not isinstance(expected_bindings, Mapping)
        or not isinstance(expected_unresolved, Mapping)
    ):
        raise IngestError(
            "Trusted table authority lacks versioned inventory, count, binding, "
            "unresolved-gap, or digest evidence."
        )
    # Authenticate the authority document against a pin that is not carried
    # inside that same document.  A self-declared digest is only a checksum.
    normalized_authority_document: Mapping[str, Any] = {
        "authority_id": authority_id,
        "version_id": authority_version,
        "external_verification_reference": external_reference,
        "inventory_sha256": inventory_sha256,
        "binding_registry_sha256": binding_registry_sha256,
        "expected_cell_counts": expected_counts,
        "expected_cell_bindings": expected_bindings,
        "expected_unresolved_gaps": expected_unresolved,
    }
    authority_document_sha256 = _pipeline_artifact_digest(
        normalized_authority_document
    )
    pin_registry = TRUSTED_TABLE_BINDING_AUTHORITY_PINS
    if not isinstance(pin_registry, Mapping):
        raise IngestError("Trusted table-authority pin registry is malformed.")
    pinned_digest = str(
        pin_registry.get((authority_id, authority_version), "")
    ).strip().lower()
    if (
        not re.fullmatch(r"[0-9a-f]{64}", pinned_digest)
        or pinned_digest != authority_document_sha256
    ):
        raise IngestError(
            "Table-binding authority is not authenticated by an independent "
            "authority/version pin."
        )
    missing_inventory_columns = sorted(
        set(TABLE_INVENTORY_COLUMNS) - set(inventory_source.columns)
    )
    if missing_inventory_columns:
        raise IngestError(
            f"Independent manuscript inventory lacks {missing_inventory_columns}."
        )
    inventory = inventory_source.loc[:, list(TABLE_INVENTORY_COLUMNS)].copy(deep=True)
    key_columns = ["table_id", "row_identifier", "column_identifier"]
    for column in key_columns:
        if bool(inventory[column].isna().any()):
            raise IngestError(f"Independent manuscript inventory has null {column} values.")
        inventory[column] = inventory[column].astype(str).str.strip()
        if bool(inventory[column].eq("").any()):
            raise IngestError(f"Independent manuscript inventory has blank {column} values.")
    for column in ("unit", "procedure"):
        if bool(inventory[column].isna().any()):
            raise IngestError(f"Independent manuscript inventory has null {column} values.")
        inventory[column] = inventory[column].astype(str).str.strip()
        if bool(inventory[column].eq("").any()):
            raise IngestError(f"Independent manuscript inventory has blank {column} values.")
    normalized_precisions: List[int] = []
    for raw_precision in inventory["display_precision"]:
        if (
            isinstance(raw_precision, (bool, np.bool_))
            or not isinstance(raw_precision, (int, np.integer))
            or not 0 <= int(raw_precision) <= 12
        ):
            raise IngestError(
                "Independent manuscript inventory has malformed display precision."
            )
        normalized_precisions.append(int(raw_precision))
    inventory["display_precision"] = normalized_precisions
    if bool(inventory.duplicated(key_columns, keep=False).any()):
        raise IngestError("Independent manuscript inventory duplicates a cell key.")
    if _content_digest(inventory) != inventory_sha256:
        raise IngestError("Trusted manuscript inventory digest is stale or contradictory.")
    inventory_keys: Set[Tuple[str, str, str]] = {
        tuple(str(value) for value in row)
        for row in inventory[key_columns].to_numpy()
    }
    if set(inventory["table_id"].astype(str)) != set(PRODUCTION_TABLES):
        raise IngestError(
            "Trusted manuscript inventory does not cover exactly the declared tables."
        )
    observed_counts = inventory.groupby("table_id", observed=True).size().to_dict()
    normalized_counts: Dict[str, int] = {}
    for table_id, raw_count in expected_counts.items():
        if (
            isinstance(raw_count, (bool, np.bool_))
            or not isinstance(raw_count, (int, np.integer))
            or int(raw_count) <= 0
        ):
            raise IngestError("Trusted table authority carries a malformed cell count.")
        normalized_counts[str(table_id)] = int(raw_count)
    if len(normalized_counts) != len(expected_counts):
        raise IngestError("Trusted table counts contain aliased table identities.")
    if observed_counts != normalized_counts:
        raise IngestError("Trusted table counts disagree with the manuscript inventory.")
    # The production inventory must be the same complete 418-cell manuscript
    # universe that reconciliation later authenticates.  Counts above the
    # structural bounds are not a harmless extension: they create cells that
    # cannot have a counterpart in the independently pinned transcription.
    _validate_production_manuscript_counts(normalized_counts, len(inventory))
    # Normalize cell keys without accepting duplicate textual aliases.
    def _normalized_cell_mapping(
        raw_mapping: Mapping[Any, Any], label: str
    ) -> Dict[Tuple[str, str, str], Any]:
        """Return a uniquely keyed, stripped three-part cell mapping."""
        normalized: Dict[Tuple[str, str, str], Any] = {}
        for raw_key, value in raw_mapping.items():
            if (
                not isinstance(raw_key, tuple)
                or len(raw_key) != 3
                or any(not str(component).strip() for component in raw_key)
            ):
                raise IngestError(f"Trusted {label} registry has a malformed cell key.")
            key = tuple(str(component).strip() for component in raw_key)
            if key in normalized:
                raise IngestError(f"Trusted {label} registry duplicates cell {key!r}.")
            normalized[key] = value
        return normalized
    normalized_bindings = _normalized_cell_mapping(
        expected_bindings, "computed-binding"
    )
    normalized_unresolved = _normalized_cell_mapping(
        expected_unresolved, "unresolved-gap"
    )
    if set(normalized_bindings) & set(normalized_unresolved):
        raise IngestError("Trusted table authority overlaps computed and unresolved cells.")
    if set(normalized_bindings) | set(normalized_unresolved) != inventory_keys:
        raise IngestError(
            "Trusted table authority does not classify the exact inventory universe."
        )
    for key, gap_record in normalized_unresolved.items():
        if (
            not isinstance(gap_record, Mapping)
            or not str(gap_record.get("gap_id", "")).strip()
            or not str(gap_record.get("source_id", "")).strip()
        ):
            raise IngestError(
                f"Trusted unresolved cell {key!r} lacks gap/source provenance."
            )
    registry_payload = {
        "computed_bindings": normalized_bindings,
        "unresolved_gaps": normalized_unresolved,
    }
    if _pipeline_artifact_digest(registry_payload) != binding_registry_sha256:
        raise IngestError("Trusted cell-source registry digest is stale or contradictory.")
    expected_source_keys = set(PRODUCTION_TABLES) | {"__cell_inventory__"}
    if set(raw_sources) != expected_source_keys:
        raise IngestError(
            "Publication table sources do not equal the complete declared table set."
        )
    # Normalize and verify every authoritative computed selector up front.
    normalized_selector_digests: Dict[Tuple[str, str, str], str] = {}
    for cell_key, binding in normalized_bindings.items():
        if (
            not isinstance(binding, Mapping)
            or not str(binding.get("artifact_key", "")).strip()
            or not str(binding.get("source_id", "")).strip()
            or not re.fullmatch(
                r"[0-9a-f]{64}",
                str(binding.get("artifact_sha256", "")).strip().lower(),
            )
        ):
            raise IngestError(
                f"Trusted selector for publication cell {cell_key!r} is malformed."
            )
        path = binding.get("attribute_path", ())
        if isinstance(path, (str, bytes)) or not isinstance(path, Sequence):
            raise IngestError(
                f"Trusted selector for publication cell {cell_key!r} has a malformed path."
            )
        normalized_selector_digests[cell_key] = _pipeline_artifact_digest(binding)
    verified: Dict[str, pd.DataFrame] = {}
    observed_verified_computed_keys: Set[Tuple[str, str, str]] = set()
    for table_id, raw_frame in raw_sources.items():
        if table_id == "__cell_inventory__":
            copied_inventory = inventory.copy(deep=True)
            copied_inventory.attrs.clear()
            copied_inventory.attrs.update(
                {
                    "authority_id": authority_id,
                    "version_id": authority_version,
                    "external_verification_reference": external_reference,
                    "complete_manuscript_inventory": True,
                    "expected_cell_counts": dict(normalized_counts),
                    "inventory_sha256": inventory_sha256,
                    "binding_registry_sha256": binding_registry_sha256,
                    "binding_authority_id": authority_id,
                    "binding_registry_version": authority_version,
                    "authority_document_sha256": authority_document_sha256,
                    "binding_verification_status": (
                        "separate_trusted_authority_and_live_state_verified"
                    ),
                }
            )
            verified[str(table_id)] = copied_inventory
            continue
        if not isinstance(raw_frame, pd.DataFrame) or raw_frame.empty:
            raise IngestError(f"Publication table {table_id!r} is absent or empty.")
        missing_columns = sorted(set(TABLE_CELL_COLUMNS) - set(raw_frame.columns))
        if missing_columns:
            raise IngestError(
                f"Publication table {table_id!r} lacks {missing_columns}."
            )
        copied = raw_frame.loc[:, list(TABLE_CELL_COLUMNS)].copy(deep=True)
        for key_column in key_columns:
            if bool(copied[key_column].isna().any()):
                raise IngestError(
                    f"Publication table {table_id!r} has a null cell identity."
                )
            copied[key_column] = copied[key_column].astype(str).str.strip()
            if bool(copied[key_column].eq("").any()):
                raise IngestError(
                    f"Publication table {table_id!r} has a blank cell identity."
                )
        if bool(copied.duplicated(key_columns, keep=False).any()):
            raise IngestError(f"Publication table {table_id!r} duplicates a cell key.")
        if set(copied["table_id"].astype(str)) != {str(table_id)}:
            raise IngestError(f"Publication table {table_id!r} carries another table identity.")
        table_keys: Set[Tuple[str, str, str]] = {
            tuple(str(value).strip() for value in row)
            for row in copied[key_columns].to_numpy()
        }
        expected_table_keys = {
            key for key in inventory_keys if key[0] == str(table_id)
        }
        if table_keys != expected_table_keys:
            raise IngestError(
                f"Publication table {table_id!r} has missing or extra inventoried cells."
            )
        caller_bindings = raw_frame.attrs.get("cell_bindings")
        if caller_bindings is not None:
            if not isinstance(caller_bindings, Mapping):
                raise IngestError(f"Publication table {table_id!r} has malformed binding attrs.")
            normalized_caller = _normalized_cell_mapping(
                caller_bindings, "caller-asserted binding"
            )
            expected_for_table = {
                key: normalized_bindings[key]
                for key in expected_table_keys
                if key in normalized_bindings
            }
            if (
                set(normalized_caller) != set(expected_for_table)
                or any(
                    _pipeline_artifact_digest(normalized_caller[key])
                    != normalized_selector_digests[key]
                    for key in expected_for_table
                )
            ):
                raise IngestError(
                    f"Publication table {table_id!r} binding attrs contradict "
                    "the separate trusted authority."
                )
        verified_bindings: Dict[Tuple[str, str, str], Mapping[str, Any]] = {}
        for row_index, cell in copied.iterrows():
            cell_key = (
                str(cell["table_id"]).strip(),
                str(cell["row_identifier"]).strip(),
                str(cell["column_identifier"]).strip(),
            )
            status = str(cell["status"]).strip()
            if cell_key in normalized_unresolved:
                gap_record = normalized_unresolved[cell_key]
                if (
                    status != TABLE_STATUS_UNRESOLVED
                    or pd.notna(cell["value_unrounded"])
                    or str(cell["gap_id"]).strip()
                    != str(gap_record["gap_id"]).strip()
                    or str(cell["source_id"]).strip()
                    != str(gap_record["source_id"]).strip()
                ):
                    raise IngestError(
                        f"Publication cell {cell_key!r} contradicts its trusted unresolved status."
                    )
                continue
            if status != TABLE_STATUS_COMPUTED or str(cell["gap_id"]).strip():
                raise IngestError(
                    f"Publication cell {cell_key!r} contradicts its trusted computed status."
                )
            binding = normalized_bindings[cell_key]
            artifact_key = str(binding.get("artifact_key", "")).strip()
            if artifact_key not in state:
                raise IngestError(f"Publication cell {cell_key!r} names an unavailable artefact.")
            declared_digest = str(binding.get("artifact_sha256", "")).strip().lower()
            live_digest = _pipeline_artifact_digest(state[artifact_key])
            if declared_digest != live_digest:
                raise IngestError(f"Publication cell {cell_key!r} has stale artefact lineage.")
            resolved: Any = state[artifact_key]
            path = binding.get("attribute_path", ())
            for component in path:
                if isinstance(resolved, Mapping):
                    if component not in resolved:
                        raise IngestError(f"Publication cell {cell_key!r} binding path is absent.")
                    resolved = resolved[component]
                elif hasattr(resolved, str(component)):
                    resolved = getattr(resolved, str(component))
                else:
                    raise IngestError(f"Publication cell {cell_key!r} binding path is absent.")
            if isinstance(resolved, pd.DataFrame):
                if "index" not in binding or "column" not in binding:
                    raise IngestError(f"Publication cell {cell_key!r} DataFrame binding is incomplete.")
                try:
                    resolved = resolved.loc[binding["index"], binding["column"]]
                except (KeyError, IndexError, TypeError) as exc:
                    raise IngestError(f"Publication cell {cell_key!r} DataFrame locator is absent.") from exc
            elif isinstance(resolved, pd.Series):
                if "index" not in binding:
                    raise IngestError(f"Publication cell {cell_key!r} Series binding is incomplete.")
                try:
                    resolved = resolved.loc[binding["index"]]
                except (KeyError, IndexError, TypeError) as exc:
                    raise IngestError(f"Publication cell {cell_key!r} Series locator is absent.") from exc
            if isinstance(resolved, (bool, np.bool_)):
                raise IngestError(f"Publication cell {cell_key!r} resolves to a Boolean.")
            try:
                actual_value = float(resolved)
                reported_value = float(cell["value_unrounded"])
            except (TypeError, ValueError, OverflowError) as exc:
                raise IngestError(
                    f"Publication cell {cell_key!r} binding/value is not scalar numeric."
                ) from exc
            if not np.isfinite(actual_value) or not np.isfinite(reported_value) or not math.isclose(
                actual_value, reported_value, rel_tol=0.0, abs_tol=IDENTITY_TOLERANCE
            ):
                raise IngestError(
                    f"Publication cell {cell_key!r} disagrees with its live upstream value."
                )
            if str(cell["source_id"]).strip() != str(binding["source_id"]).strip():
                raise IngestError(
                    f"Publication cell {cell_key!r} source identity contradicts "
                    "the trusted selector."
                )
            # Reconstruct the verified value from live state rather than
            # preserving a numerically equal but caller-owned object.
            copied.at[row_index, "value_unrounded"] = actual_value
            verified_bindings[cell_key] = _deep_freeze_sweep_evidence(binding)
        copied.attrs.clear()
        copied.attrs["binding_registry_sha256"] = binding_registry_sha256
        copied.attrs["binding_authority_id"] = authority_id
        copied.attrs["binding_registry_version"] = authority_version
        copied.attrs["authority_document_sha256"] = authority_document_sha256
        copied.attrs["binding_verification_status"] = (
            "separate_trusted_authority_and_live_state_verified"
        )
        observed_verified_computed_keys.update(verified_bindings)
        copied.attrs["verified_computed_cell_keys"] = tuple(
            sorted(verified_bindings)
        )
        copied.attrs["verified_binding_digests"] = tuple(
            sorted(
                (
                    key,
                    normalized_selector_digests[key],
                )
                for key in verified_bindings
            )
        )
        verified[str(table_id)] = copied
    if observed_verified_computed_keys != set(normalized_bindings):
        raise IngestError(
            "Computed publication cells do not equal the trusted binding universe."
        )
    return MappingProxyType(verified)


# Adapt the production stage.
def _adapt_production(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the production stage.

    Parameters
    ----------
    state : mapping
        The artefact state carrying baseline gap matrices and incremental contrasts.
    data : mapping
        The caller mapping carrying table layouts, a separately authenticated
        binding-authority document, authoritative geometry, the overlay pair
        and the output directory. Scientific values are derived from state.

    Returns
    -------
    mapping
        The figure and table bundle stored under the production key.
    """
    required_state = (
        KEY_BASELINE,
        KEY_COUNTERFACTUAL,
        KEY_RESAMPLING,
        KEY_SWEEPS,
        KEY_SECONDARY,
        KEY_INFLUENCE_TABLES,
        KEY_MATCHING,
        KEY_MATCHING_REPORT,
        KEY_HETEROGENEITY_REG,
        KEY_ARMING,
        KEY_CONFIG,
    )
    missing_state = [key for key in required_state if key not in state]
    if missing_state:
        raise IngestError(
            f"Production lacks required live upstream artefact(s) {missing_state}."
        )
    required_data = (
        "table_sources",
        "table_binding_authority",
        "country_geometry",
        "world_geometry",
        "overlay_pair",
        "output_dir",
    )
    missing_data = [
        key for key in required_data if key not in data or data[key] is None
    ]
    if missing_data:
        raise IngestError(
            f"Production lacks required external prerequisite(s) {missing_data}."
        )
    _require_ready_config(state, data)
    # Compose the gap matrices of every baseline pair.
    gap_matrices = {
        pair: result.gaps for pair, result in state[KEY_BASELINE].results.items()
    }
    # Compose the incremental contrasts of every baseline pair.
    contrasts = {
        pair: result.incremental_contrasts
        for pair, result in state[KEY_BASELINE].results.items()
    }
    if KEY_COUNTERFACTUAL not in state:
        raise IngestError(
            "Production requires Task 12 fitted pre-period figure paths."
        )
    fitted_long = state[KEY_COUNTERFACTUAL].fitted_pre_paths
    fitted_pre_gaps: Dict[Tuple[str, str], pd.DataFrame] = {}
    for pair in sorted(gap_matrices):
        scoped = fitted_long[
            fitted_long["policy"].astype(str).eq(str(pair[0]))
            & fitted_long["outcome"].astype(str).eq(str(pair[1]))
        ]
        if scoped.empty or bool(scoped.duplicated(["country_code", "year"]).any()):
            raise IngestError(f"{pair}: fitted pre-period figure rows are absent or duplicated.")
        fitted_pre_gaps[pair] = scoped.pivot(
            index="country_code", columns="year", values="pre_fit_residual"
        ).sort_index().sort_index(axis=1)
    if KEY_RESAMPLING not in state:
        raise IngestError("Figure B2 requires computed resampling artefacts.")
    placebo_input = {
        pair: np.asarray(draws, dtype=float)
        for (pair, branch), draws in state[KEY_RESAMPLING].draws.items()
        if branch == BRANCH_PLACEBO
    }
    observed_input = {
        pair: float(state[KEY_BASELINE].results[pair].att_incremental)
        for pair in placebo_input
    }
    for duplicate_key, actual in (
        ("placebo_draws", placebo_input),
        ("observed_preferred_att", observed_input),
    ):
        if duplicate_key in data and _content_digest(data[duplicate_key]) != _content_digest(actual):
            raise IngestError(
                f"Caller {duplicate_key} contradicts the computed upstream artefact."
            )
    raw_overlay_pair = data["overlay_pair"]
    if (
        isinstance(raw_overlay_pair, (str, bytes))
        or not isinstance(raw_overlay_pair, Sequence)
        or len(raw_overlay_pair) != 2
    ):
        raise IngestError("Production overlay_pair must identify one policy/outcome pair.")
    overlay_pair = (str(raw_overlay_pair[0]), str(raw_overlay_pair[1]))
    if overlay_pair not in state[KEY_BASELINE].results:
        raise IngestError("Production overlay_pair is absent from live baseline results.")
    if KEY_MATCHING not in state:
        raise IngestError("Figure 4 requires upstream MatchingDiagnosticsResult.")
    matching_result = state[KEY_MATCHING]
    verified_matched_paths_by_pair: Dict[Tuple[str, str], Mapping[str, pd.DataFrame]] = {}
    for outcome in data["study_config"]["execution_parameters"]["figure_coverage"]["figure4_outcomes"]:
        effective_pair = (overlay_pair[0], str(outcome))
        verified_matched_paths: Dict[str, pd.DataFrame] = {}
        for label, axis, weighting_scheme in (
            ("high_income", AXIS_HIGH_INCOME, "equal_weight_high_income"),
            ("mahalanobis_m5", AXIS_MAHALANOBIS, "selection_frequency_positive_support"),
        ):
            key = (str(effective_pair[0]), str(effective_pair[1]), axis)
            if key not in matching_result.designs:
                raise IngestError(f"Figure 4 lacks upstream design {label!r}.")
            record = matching_result.designs[key]
            if record.gaps.empty or record.fitted_pre_gaps.empty:
                raise IngestError(f"Figure 4 upstream design {label!r} lacks fitted paths.")
            if set(record.gaps.index.astype(str)) != set(record.fitted_pre_gaps.index.astype(str)):
                raise IngestError(f"Figure 4 upstream design {label!r} has roster drift.")
            combined = pd.concat(
                [record.fitted_pre_gaps.copy(deep=True), record.gaps.copy(deep=True)],
                axis=1,
            ).sort_index(axis=1)
            path = compute_coalition_path(combined, str(effective_pair[1]))
            roster = tuple(str(code) for code in record.matched_targets)
            donor_roster = tuple(str(code) for code in record.matched_donors)
            if not roster or not donor_roster:
                raise IngestError(
                    f"Figure 4 upstream design {label!r} has no target or donor roster."
                )
            target_weights = {code: 1.0 / len(roster) for code in roster}
            if record.weights is None:
                donor_weights = {code: 1.0 / len(donor_roster) for code in donor_roster}
            else:
                raw_weights = {
                    code: float(record.weights.loc[code]) for code in donor_roster
                }
                total_weight = float(sum(raw_weights.values()))
                if total_weight <= 0.0:
                    raise IngestError(f"Figure 4 upstream design {label!r} has no weight mass.")
                donor_weights = {
                    code: float(value) / total_weight
                    for code, value in raw_weights.items()
                }
            content_digest = _content_digest(path.loc[:, ["coalition_mean"]])
            roster_digest = _content_digest(roster)
            weight_digest = _content_digest(target_weights)
            donor_roster_digest = _content_digest(donor_roster)
            donor_weight_digest = _content_digest(donor_weights)
            upstream_digest = hashlib.sha256(
                (
                    str(matching_result.fingerprint_sha256)
                    + "|"
                    + _content_digest(record.design_provenance)
                ).encode("utf-8")
            ).hexdigest()
            lineage = hashlib.sha256(
                json.dumps(
                    {
                        "design_id": label,
                        "path_content_sha256": content_digest,
                        "effective_roster_sha256": roster_digest,
                        "effective_weights_sha256": weight_digest,
                        "effective_donor_roster_sha256": donor_roster_digest,
                        "effective_donor_weights_sha256": donor_weight_digest,
                        "provenance_origin": "upstream_matching_design_result",
                        "upstream_result_sha256": upstream_digest,
                    },
                    sort_keys=True,
                    separators=(",", ":"),
                ).encode("utf-8")
            ).hexdigest()
            path.attrs.update(
                {
                    "design_id": label,
                    "effective_roster": roster,
                    "effective_weights": target_weights,
                    "effective_donor_roster": donor_roster,
                    "effective_donor_weights": donor_weights,
                    "path_content_sha256": content_digest,
                    "effective_roster_sha256": roster_digest,
                    "effective_weights_sha256": weight_digest,
                    "effective_donor_roster_sha256": donor_roster_digest,
                    "effective_donor_weights_sha256": donor_weight_digest,
                    "fit_lineage_sha256": lineage,
                    "weighting_scheme": weighting_scheme,
                    "provenance_origin": "upstream_matching_design_result",
                    "upstream_result_sha256": upstream_digest,
                }
            )
            verified_matched_paths[label] = path
        verified_matched_paths_by_pair[effective_pair] = verified_matched_paths
    reference_rows = state[KEY_SWEEPS].results.reset_index()
    reference_rows = reference_rows[
        reference_rows["axis"].astype(str).eq(AXIS_REFERENCE_WINDOW)
        & reference_rows["policy"].astype(str).eq(str(overlay_pair[0]))
        & reference_rows["outcome"].astype(str).eq(str(overlay_pair[1]))
        & reference_rows["status"].astype(str).eq("success")
    ]
    if reference_rows.empty:
        raise IngestError("Figure B3 lacks successful computed reference-window variants.")
    reference_estimates: Dict[str, Mapping[str, float]] = {}
    for _, row in reference_rows.iterrows():
        estimate, se = float(row["att"]), float(row["se_cd"])
        target_count = len(state[KEY_BASELINE].results[overlay_pair].incremental_contrasts)
        statistic, degrees_of_freedom = compute_t_statistic(
            estimate, se, target_count
        )
        p_value = compute_two_sided_p(statistic, degrees_of_freedom)
        reference_estimates[str(row["variant_name"])] = {
            "estimate": estimate,
            "se_cd": se,
            "p_value": p_value,
            "source_id": str(row["run_id"]),
        }
    # Invoke production with paths and reference coefficients derived from
    # their executed upstream artefacts.
    table_sources = _verified_table_sources(
        state,
        data["table_sources"],
        data["table_binding_authority"],
    )
    production = produce_figures_and_tables(
        gap_matrices,
        contrasts,
        matched_paths=verified_matched_paths_by_pair[overlay_pair],
        matched_paths_by_pair=verified_matched_paths_by_pair,
        placebo_draws=placebo_input,
        reference_window_estimates=reference_estimates,
        table_sources=table_sources,
        output_dir=Path(data["output_dir"]) if "output_dir" in data else None,
        extra_gaps=None,
        fitted_pre_gaps=fitted_pre_gaps,
        observed_preferred_att=observed_input,
        country_geometry=data["country_geometry"],
        world_geometry=data["world_geometry"],
        overlay_pair=overlay_pair,
    )
    # Return the stage's artefacts.
    return {KEY_PRODUCTION: production}


# Perturb a live policy ledger without changing its schema or row count.
def _perturb_reconciliation_policy_ledger(policy_ledger: pd.DataFrame) -> pd.DataFrame:
    """
    Return an independently changed treatment-identity/timing ledger.

    Parameters
    ----------
    policy_ledger : pandas.DataFrame
        Live Task-4 policy ledger. At least one country, policy or entry-year
        field must be present and perturbable.

    Returns
    -------
    pandas.DataFrame
        A deep copy whose available policy identities and timing differ from
        the live ledger while all non-treatment fields remain unchanged.

    Raises
    ------
    IngestError
        If the ledger is empty, malformed or exposes no perturbable treatment
        identity/timing field.
    """
    if not isinstance(policy_ledger, pd.DataFrame) or policy_ledger.empty:
        raise IngestError("Live eligibility evidence lacks a nonempty policy ledger.")
    perturbed = policy_ledger.copy(deep=True)
    changed = False
    if "country_code" in perturbed.columns and len(perturbed) > 1:
        original = perturbed["country_code"].copy()
        perturbed["country_code"] = original.iloc[::-1].to_numpy()
        changed = changed or not perturbed["country_code"].equals(original)
    if "policy" in perturbed.columns:
        original = perturbed["policy"].copy()
        perturbed["policy"] = original.astype(str).map(lambda value: f"reconciliation_perturbed::{value}")
        changed = changed or not perturbed["policy"].equals(original)
    if "entry_year" in perturbed.columns:
        original = perturbed["entry_year"].copy()
        numeric = pd.to_numeric(original, errors="raise")
        perturbed["entry_year"] = numeric + 37
        changed = changed or not perturbed["entry_year"].equals(original)
    if not changed:
        raise IngestError("Live policy ledger exposes no perturbable policy identity or timing field.")
    return perturbed


# Perturb only one forbidden target-post channel in the realized panel.
def _perturb_reconciliation_target_post(
    panel: pd.DataFrame,
    registry: PartitionRegistry,
    columns: Sequence[str],
) -> pd.DataFrame:
    """
    Change finite target observations after 2013 in exactly the named columns.

    Parameters
    ----------
    panel : pandas.DataFrame
        Live Task-12 analysis panel with country and year identifiers.
    registry : PartitionRegistry
        Frozen partition registry defining the union of target countries.
    columns : sequence of str
        Forbidden outcome or covariate channel to perturb.

    Returns
    -------
    pandas.DataFrame
        A deep copy with a deterministic offset applied only to finite target
        cells in 2014--2024 and only for the requested channel.

    Raises
    ------
    IngestError
        If identifiers, targets, requested columns or finite post-period cells
        are unavailable.
    """
    required = {COUNTRY_COLUMN, YEAR_COLUMN}
    if not isinstance(panel, pd.DataFrame) or not required.issubset(panel.columns):
        raise IngestError("Live Task-12 panel lacks country/year identity columns.")
    target_codes = {
        str(code) for partition in registry.partitions.values() for code in partition.targets
    }
    selected_columns = [str(name) for name in columns if str(name) in panel.columns]
    if not target_codes or not selected_columns:
        raise IngestError("Task-12 live state lacks target rows or the forbidden channel columns.")
    perturbed = panel.copy(deep=True)
    row_mask = (
        perturbed[COUNTRY_COLUMN].astype(str).isin(target_codes)
        & pd.to_numeric(perturbed[YEAR_COLUMN], errors="coerce").between(
            PROJECTION_START_YEAR, PROJECTION_END_YEAR
        )
    )
    changed = 0
    for column in selected_columns:
        numeric = pd.to_numeric(perturbed.loc[row_mask, column], errors="coerce")
        finite = numeric.notna() & np.isfinite(numeric.to_numpy(dtype=float))
        if bool(finite.any()):
            indices = numeric.index[finite]
            perturbed.loc[indices, column] = numeric.loc[indices].to_numpy(dtype=float) + 10_000.0
            changed += len(indices)
    if changed == 0:
        raise IngestError("Task-12 forbidden channel has no finite target-post cell to perturb.")
    return perturbed


# Construct a fully state-bound and independently pinnable probe scenario.
def _state_bound_reconciliation_spec(
    boundary: Callable[..., Any],
    role: str,
    probe_id: str,
    baseline_args: Sequence[Any],
    baseline_kwargs: Mapping[str, Any],
    perturbed_args: Sequence[Any],
    perturbed_kwargs: Mapping[str, Any],
    argument_roles: Mapping[str, str],
    expected_output: Any,
    dependency_roles: Optional[Mapping[str, str]] = None,
) -> Mapping[str, Any]:
    """
    Build one complete live-stage scenario with argument/output digests.

    Parameters
    ----------
    boundary : callable
        Exact registered production boundary to execute.
    role : str
        Scientific output role represented by the live stage artefact.
    probe_id : str
        Stable scenario identifier used by the independent pin registry.
    baseline_args, perturbed_args : sequence
        Positional arguments for the baseline and independently perturbed run.
    baseline_kwargs, perturbed_kwargs : mapping
        Keyword arguments for the two runs.
    argument_roles : mapping of str to str
        Complete classification of every bound argument as permitted,
        forbidden or infrastructure.
    expected_output : Any
        Live stage output whose normalized digest the baseline rerun must
        reproduce.
    dependency_roles : mapping of str to str, optional
        Module-declared roles for mutable scientific constants encountered by
        recursive dependency tracing.

    Returns
    -------
    mapping
        Read-only scenario carrying bound arguments, per-argument digests and
        the expected live-stage output digest.

    Raises
    ------
    IngestError
        If either call does not bind or the role map is incomplete.
    """
    signature = inspect.signature(boundary)
    try:
        baseline_bound = signature.bind(*tuple(baseline_args), **dict(baseline_kwargs))
        perturbed_bound = signature.bind(*tuple(perturbed_args), **dict(perturbed_kwargs))
    except TypeError as exc:
        raise IngestError(f"Live reconciliation scenario {probe_id!r} does not bind: {exc}") from exc
    baseline_bound.apply_defaults()
    perturbed_bound.apply_defaults()
    if set(argument_roles) != set(baseline_bound.arguments) or set(baseline_bound.arguments) != set(perturbed_bound.arguments):
        raise IngestError(f"Live reconciliation scenario {probe_id!r} has incomplete argument roles.")
    dependency_role_map = dict(dependency_roles or {})
    dependency_trace = _trace_callable_dependencies(boundary, dependency_role_map)
    return MappingProxyType(
        {
            "probe_id": str(probe_id),
            "probe_version": "pipeline-live-state-v1",
            "baseline_args": tuple(baseline_args),
            "baseline_kwargs": dict(baseline_kwargs),
            "perturbed_args": tuple(perturbed_args),
            "perturbed_kwargs": dict(perturbed_kwargs),
            "argument_roles": dict(argument_roles),
            "state_bindings": {
                name: _pipeline_artifact_digest(value)
                for name, value in baseline_bound.arguments.items()
            },
            "dependency_roles": dependency_role_map,
            "dependency_bindings": {
                str(item["name"]): str(item["sha256"])
                for item in dependency_trace
            },
            "expected_stage_output_sha256": _pipeline_artifact_digest(expected_output),
            "stage_output_key": str(role),
        }
    )


# Derive every invariant scenario from the exact artifacts already in state.
def _live_reconciliation_probe_registry(
    state: Mapping[str, Any],
) -> Mapping[Tuple[str, str], Mapping[str, Any]]:
    """
    Return Task-4/5/6/12 scenarios derived only from frozen pipeline state.

    Parameters
    ----------
    state : mapping
        Completed stage state containing cleansed inputs, eligibility,
        completion/repair results, panel, partitions, fit and projection.

    Returns
    -------
    mapping
        Read-only scenarios keyed by exact boundary name and forbidden role:
        Task-4 treatment, structural Task-5/6 exclusions and separate Task-12
        target-post outcome and covariate perturbations.

    Raises
    ------
    IngestError
        If a required live stage artefact is absent or cannot supply a genuine
        perturbation/bound call.
    """
    required_keys = (
        KEY_CLEANSE, KEY_ELIGIBILITY, KEY_CONTROLS, KEY_OUTCOMES, KEY_PANEL,
        KEY_REGISTRY, KEY_FACTOR, KEY_FIT, KEY_COUNTERFACTUAL,
    )
    missing = [key for key in required_keys if key not in state]
    if missing:
        raise IngestError(f"Live reconciliation probes lack stage artifact(s) {missing}.")
    bundle = state[KEY_CLEANSE]
    policy = bundle.frame("policy_ledger_raw")
    metadata = bundle.frame("country_metadata_raw")
    perturbed_policy = _perturb_reconciliation_policy_ledger(policy)
    observed_panel = state[KEY_PANEL].panel
    perturbed_y = _perturb_reconciliation_target_post(
        observed_panel, state[KEY_REGISTRY], OUTCOME_VARIABLES,
    )
    perturbed_x = _perturb_reconciliation_target_post(
        observed_panel, state[KEY_REGISTRY], CONTROL_COLUMNS,
    )
    specs: Dict[Tuple[str, str], Mapping[str, Any]] = {}
    eligibility_roles = {
        "bundle": "permitted", "policy_ledger": "treatment",
        "metadata_frame": "permitted", "expected_otherwise_eligible": "infrastructure",
        "expected_final": "infrastructure", "enforce_declared_gates": "infrastructure",
    }
    specs[("orchestrate_eligibility_screening", "treatment")] = _state_bound_reconciliation_spec(
        orchestrate_eligibility_screening, "eligibility", "stage04_policy_perturbation",
        (bundle,), {"policy_ledger": policy, "metadata_frame": metadata},
        (bundle,), {"policy_ledger": perturbed_policy, "metadata_frame": metadata},
        eligibility_roles, state[KEY_ELIGIBILITY],
        dependency_roles={
            "DOCUMENTED_EXCLUSIONS": "permitted",
            "OUTCOME_CODES": "permitted",
            "WDI_CONTROL_CODES": "permitted",
        },
    )
    structural_inputs = (
        (
            orchestrate_control_completion, "stage05_completion", (bundle,),
            {
                "mask": state[KEY_ELIGIBILITY].mask,
                "expected_repaired_total": int(bundle.study_config["execution_parameters"]["completion"]["repaired_control_cells"]),
                "expected_balance_derived": int(bundle.study_config["execution_parameters"]["completion"]["derived_fuel_balance_missing_cells"]),
            }, state[KEY_CONTROLS],
            {"WDI_CONTROL_CODES": "permitted"},
        ),
        (
            orchestrate_outcome_repair, "stage06_repair", (bundle,),
            {"mask": state[KEY_ELIGIBILITY].mask}, state[KEY_OUTCOMES],
            {
                "DECLARED_REPAIRS_BY_OUTCOME": "permitted",
                "OUTCOME_CODES": "permitted",
            },
        ),
    )
    for boundary, probe_id, args, kwargs, output, dependency_roles in structural_inputs:
        bound = inspect.signature(boundary).bind(*args, **kwargs)
        bound.apply_defaults()
        roles = {
            name: ("permitted" if name in {"bundle", "mask"} else "infrastructure")
            for name in bound.arguments
        }
        specs[(boundary.__name__, "structural_treatment_exclusion")] = _state_bound_reconciliation_spec(
            boundary, "structural_treatment_exclusion", probe_id,
            args, kwargs, args, kwargs, roles, output,
            dependency_roles=dependency_roles,
        )
    projection_roles = {
        "fit_result": "permitted", "factor_result": "permitted",
        "registry": "permitted", "start_year": "infrastructure",
        "end_year": "infrastructure", "observed_panel": "target_post_y",
    }
    projection_args = (state[KEY_FIT], state[KEY_FACTOR], state[KEY_REGISTRY])
    base_kwargs = {"observed_panel": observed_panel}
    y_kwargs = {"observed_panel": perturbed_y}
    specs[("orchestrate_counterfactual_projection", "target_post_y")] = _state_bound_reconciliation_spec(
        orchestrate_counterfactual_projection, "counterfactual", "stage12_target_post_y",
        projection_args, base_kwargs, projection_args, y_kwargs,
        projection_roles, state[KEY_COUNTERFACTUAL],
    )
    projection_roles_x = dict(projection_roles)
    projection_roles_x["observed_panel"] = "target_post_x"
    specs[("orchestrate_counterfactual_projection", "target_post_x")] = _state_bound_reconciliation_spec(
        orchestrate_counterfactual_projection, "counterfactual", "stage12_target_post_x",
        projection_args, base_kwargs, projection_args, {"observed_panel": perturbed_x},
        projection_roles_x, state[KEY_COUNTERFACTUAL],
    )
    return MappingProxyType(specs)


# Derive the ten core archive declarations from live state and stage lineage.
def _live_archive_declarations(
    state: Mapping[str, Any],
    archive_root: Path,
    version_tag: str,
) -> Mapping[str, Mapping[str, Any]]:
    """
    Return deterministic core archive declarations without caller selectors.

    Parameters
    ----------
    state : mapping
        Live pipeline artefacts and every direct parent required by their
        producing stages.
    archive_root : pathlib.Path
        Validated self-contained archive root.
    version_tag : str
        Frozen archive release version recorded separately from each artifact's
        independently authenticated source vintage.

    Returns
    -------
    mapping
        Read-only ten-entry registry with exact state key, section, role,
        canonical content digest and direct parent digests.

    Raises
    ------
    IngestError
        If a trusted vintage record, live object, unique producing stage or
        direct parent is absent.
    """
    if not isinstance(version_tag, str) or not version_tag.strip():
        raise IngestError("Core archive declarations require a nonblank release version.")
    root = Path(archive_root).expanduser().resolve()
    declarations: Dict[str, Mapping[str, Any]] = {}
    for artifact_id in ARCHIVE_ARTEFACTS:
        # Resolve source/vintage facts before deriving or writing any archive
        # payload. The empty production registry therefore fails atomically.
        trusted_vintage = _trusted_archive_vintage_record(artifact_id)
        state_key, section, role = ARCHIVE_LIVE_BINDINGS[artifact_id]
        if state_key not in state:
            raise IngestError(
                f"Core archive declaration {artifact_id!r} lacks live state {state_key!r}."
            )
        producing = [stage for stage in PIPELINE_STAGES if state_key in stage.produces]
        if len(producing) != 1:
            raise IngestError(f"Core archive declaration {artifact_id!r} has no unique producer.")
        parents = tuple(producing[0].requires)
        missing_parents = [key for key in parents if key not in state]
        if missing_parents:
            raise IngestError(
                f"Core archive declaration {artifact_id!r} lacks parent(s) {missing_parents}."
            )
        serialized = _archive_canonical_bytes(state[state_key])
        declarations[artifact_id] = MappingProxyType(
            {
                "archive_version_tag": str(version_tag),
                "vintage_tag": trusted_vintage["vintage_tag"],
                "vintage_authority_id": trusted_vintage["authority_id"],
                "vintage_source_id": trusted_vintage["source_id"],
                "vintage_source_version": trusted_vintage["source_version"],
                "vintage_evidence_reference": trusted_vintage["evidence_reference"],
                "vintage_claim_sha256": trusted_vintage["claim_sha256"],
                "role": role,
                "media_type": "application/vnd.do-sanctions-backfire.canonical+json",
                "content_sha256": hashlib.sha256(serialized).hexdigest(),
                "location": str(root / section / f"{artifact_id}.json"),
                "state_key": state_key,
                "parent_state_keys": parents,
                "parent_sha256": {
                    key: hashlib.sha256(_archive_canonical_bytes(state[key])).hexdigest()
                    for key in parents
                },
            }
        )
    return MappingProxyType(declarations)


# Resolve the archive release identity without coercing null/non-string input.
def _resolve_archive_release_version(data: Mapping[str, Any]) -> str:
    """
    Return the validated archive release version from pipeline inputs.

    Parameters
    ----------
    data : mapping
        Pipeline inputs. Omitting ``archive_version_tag`` selects the declared
        default; explicitly supplying null, non-string or blank data is an
        error rather than a request for that default.

    Returns
    -------
    str
        Stripped, nonblank archive release identity.

    Raises
    ------
    IngestError
        If the explicit release version is not a genuine nonblank string.
    """
    if not isinstance(data, Mapping):
        raise IngestError("Archive release-version inputs must be a mapping.")
    value = (
        data["archive_version_tag"]
        if "archive_version_tag" in data
        else "v1.0.0-frozen"
    )
    if not isinstance(value, str) or not value.strip():
        raise IngestError(
            "archive_version_tag must be a genuine nonblank string when supplied."
        )
    return value.strip()


# Freeze one deterministic support file without overwriting unrelated bytes.
def _freeze_archive_support_file(
    path: Path,
    payload: bytes,
    archive_root: Optional[Path] = None,
) -> None:
    """
    Exclusively publish bytes or verify an identical existing regular file.

    Parameters
    ----------
    path : pathlib.Path
        Destination below a caller-validated archive root.
    payload : bytes
        Deterministic support-file bytes.
    archive_root : pathlib.Path, optional
        Resolved archive root. When supplied, the destination must be a direct,
        non-symlink child of this root; production callers always supply it.

    Returns
    -------
    None
        Returns after an exclusive hard-link publication or byte-identical
        verification of an existing regular file.

    Raises
    ------
    IngestError
        If the parent is absent, an existing file differs, publication fails,
        or the final bytes cannot be verified. Existing unrelated bytes are
        never overwritten.
    """
    requested = Path(path).expanduser()
    if requested.is_symlink():
        raise IngestError(f"Archive support file cannot be a symbolic link: {requested}.")
    resolved = requested.resolve()
    if archive_root is not None:
        root = Path(archive_root).expanduser().resolve()
        if resolved.parent != root:
            raise IngestError(
                f"Archive support file escapes the validated archive root: {resolved}."
            )
    if not resolved.parent.is_dir():
        raise IngestError(f"Archive support-file parent does not exist: {resolved.parent}.")
    if resolved.exists():
        if not resolved.is_file() or resolved.read_bytes() != payload:
            raise IngestError(f"Archive support file has unrelated existing bytes: {resolved}.")
        return
    temporary_name: Optional[str] = None
    try:
        with tempfile.NamedTemporaryFile(
            mode="wb", dir=str(resolved.parent), prefix=f".{resolved.name}.",
            suffix=".tmp", delete=False,
        ) as temporary:
            temporary.write(payload)
            temporary.flush()
            os.fsync(temporary.fileno())
            temporary_name = temporary.name
        try:
            os.link(str(temporary_name), str(resolved))
        except FileExistsError:
            pass
    except OSError as exc:
        raise IngestError(f"Archive support file could not be frozen: {resolved}.") from exc
    finally:
        if temporary_name is not None:
            try:
                Path(temporary_name).unlink(missing_ok=True)
            except OSError:
                pass
    if not resolved.is_file() or resolved.read_bytes() != payload:
        raise IngestError(f"Archive support file failed byte verification: {resolved}.")


# Freeze README, configuration and raw-source identity from live state.
def _freeze_archive_support_files(
    state: Mapping[str, Any], archive_root: Path, version_tag: str
) -> None:
    """
    Freeze deterministic support metadata for the self-contained archive.

    Parameters
    ----------
    state : mapping
        Live state carrying the validated configuration and ingest bundle.
    archive_root : pathlib.Path
        Root containing the already-created archive directory structure.
    version_tag : str
        Version written into the deterministic README.

    Returns
    -------
    None
        Writes or verifies README, configuration fingerprint and the sorted
        ten-source fingerprint registry.

    Raises
    ------
    IngestError
        If config/ingest state is absent or any support file cannot be frozen
        without overwriting unrelated bytes.
    """
    root = Path(archive_root).expanduser().resolve()
    if KEY_CONFIG not in state or KEY_BUNDLE not in state:
        raise IngestError("Archive support metadata lacks live config or ingest state.")
    readme = (
        "# Do Sanctions Backfire replication archive\n\n"
        f"Frozen version: {version_tag}\n"
        "All scientific files are canonical serializations of live pipeline artifacts.\n"
    ).encode("utf-8")
    config_fingerprint = (
        hashlib.sha256(_archive_canonical_bytes(state[KEY_CONFIG])).hexdigest() + "\n"
    ).encode("ascii")
    source_fingerprints = {
        str(name): str(record.fingerprint_sha256)
        for name, record in sorted(state[KEY_BUNDLE].records.items())
    }
    source_payload = (
        json.dumps(source_fingerprints, sort_keys=True, separators=(",", ":")) + "\n"
    ).encode("utf-8")
    _freeze_archive_support_file(root / "README.md", readme, archive_root=root)
    _freeze_archive_support_file(
        root / "config_fingerprint.txt", config_fingerprint, archive_root=root
    )
    _freeze_archive_support_file(
        root / "source_fingerprints.json", source_payload, archive_root=root
    )


# Adapt the reconciliation stage.
def _adapt_reconciliation(state: Mapping[str, Any], data: Mapping[str, Any]) -> Mapping[str, Any]:
    """
    Adapt the reconciliation stage.

    Parameters
    ----------
    state : mapping
        The completed live pipeline state. The adapter derives reproduced
        cells, invariant scenarios, robustness evidence and the ten archive
        declarations from these exact artefacts.
    data : mapping
        The caller mapping carrying the independently pinned
        ``manuscript_universe``, declared provenance facts, layer checkpoints
        and archive root. Caller-supplied reproduced values, scientific
        callables and archive selectors are not authoritative.

    Returns
    -------
    mapping
        The reconciliation audit stored under the reconciliation key.

    Raises
    ------
    IngestError
        If the pinned universe, live production cells, Task-17/18 linkage,
        genuine probe inputs or archive declarations are missing, duplicated
        or stale. Scientifically unresolved but well-formed evidence is
        retained by the audit as an incomplete report.
    """
    _require_ready_config(state, data)
    if KEY_PRODUCTION not in state:
        raise IngestError("Reconciliation requires the completed production bundle.")
    # Authenticate the manuscript universe before deriving live evidence or
    # writing archive support files. A structurally valid but unpinned universe
    # retains the audit's explicit incomplete disposition; no archive payload
    # is created for that scientifically unresolved case.
    manuscript_universe = data["manuscript_universe"]
    try:
        _validate_complete_manuscript_universe(manuscript_universe)
    except IngestError as exc:
        if "has no independent module-controlled digest pin" not in str(exc):
            raise
        incomplete = execute_reconciliation_audit(
            pd.DataFrame(), pd.DataFrame(), (), (),
            study_config=data["study_config"],
            declared_facts=data.get("declared_facts"),
            manuscript_universe=manuscript_universe,
        )
        return {KEY_RECONCILIATION: incomplete}
    # Derive reported cells exclusively from the independently pinned
    # manuscript universe; a second caller-provided value frame could override
    # the authority and is therefore never consumed.
    required_universe_columns = list(COMPARISON_KEY) + [
        "reported_value", "comparison_type", "unit", "display_precision",
        "reported_source_id", "procedure_id",
    ]
    missing_universe_columns = [
        name for name in required_universe_columns if name not in manuscript_universe.columns
    ] if isinstance(manuscript_universe, pd.DataFrame) else required_universe_columns
    if missing_universe_columns:
        raise IngestError(
            f"Reconciliation manuscript universe lacks field(s) {missing_universe_columns}."
        )
    reported = manuscript_universe.loc[:, required_universe_columns].rename(
        columns={"reported_value": "value", "reported_source_id": "source_id"}
    ).copy(deep=True)
    # Derive reproduced cells from the actual frozen production tables. Caller
    # supplied reproduced values are deliberately refused at preflight.
    reproduced_rows: List[Dict[str, Any]] = []
    reproduced_keys: Set[Tuple[str, str, str]] = set()
    derived_gaps: List[Dict[str, str]] = []
    for table_id, table in state[KEY_PRODUCTION].tables.items():
        if not isinstance(table, pd.DataFrame) or table.empty:
            raise IngestError(f"Production table {table_id!r} is absent or empty.")
        material = table.reset_index()
        required = {
            "table_id", "row_identifier", "column_identifier", "status",
            "value_unrounded", "gap_id",
        }
        if not required.issubset(material.columns):
            raise IngestError(f"Production table {table_id!r} lacks reconciliation fields.")
        for _, row in material.iterrows():
            key = (
                str(row["table_id"]), str(row["row_identifier"]),
                str(row["column_identifier"]),
            )
            if key in reproduced_keys or any(tuple(item.get(field, "") for field in COMPARISON_KEY) == key for item in derived_gaps):
                raise IngestError(f"Production duplicates reconciliation cell {key!r}.")
            if str(row["status"]) == TABLE_STATUS_COMPUTED:
                value = float(row["value_unrounded"])
                if not np.isfinite(value):
                    raise IngestError(f"Production cell {key!r} is not finite.")
                unit = str(row.get("unit", "")).strip()
                source_id = str(row.get("source_id", "")).strip()
                procedure_id = str(row.get("procedure", "")).strip()
                if not unit or not source_id or not procedure_id:
                    raise IngestError(
                        f"Production cell {key!r} lacks unit/source/procedure provenance."
                    )
                reproduced_rows.append(
                    {
                        "table_number": key[0],
                        "row_identifier": key[1],
                        "column_identifier": key[2],
                        "value": value,
                        "unit": unit,
                        "source_id": source_id,
                        "procedure_id": procedure_id,
                    }
                )
                reproduced_keys.add(key)
            else:
                gap_id = str(row["gap_id"]).strip()
                if not gap_id:
                    raise IngestError(f"Unresolved production cell {key!r} lacks a gap ID.")
                derived_gaps.append(
                    {
                        "table_number": key[0],
                        "row_identifier": key[1],
                        "column_identifier": key[2],
                        "blocking_input": gap_id,
                        "description": "production table cell remains unresolved",
                        "unit": str(row.get("unit", "unspecified")),
                        "source_id": str(row.get("source_id", "production_bundle")),
                        "procedure_id": str(row.get("procedure", "production_table")),
                    }
                )
    reproduced = pd.DataFrame(
        reproduced_rows,
        columns=list(COMPARISON_KEY) + ["value", "unit", "source_id", "procedure_id"],
    )
    # Compose declared keys only from the independent inventory; deriving them
    # from reported values would make omission impossible to detect.
    declared_keys = [tuple(str(value) for value in row) for row in manuscript_universe[
        ["table_number", "row_identifier", "column_identifier"]
    ].to_numpy()]
    task18_coverage = state[KEY_ROBUSTNESS].coverage_audit
    if (
        not isinstance(task18_coverage, pd.DataFrame)
        or task18_coverage.empty
        or "run_id" not in task18_coverage.columns
        or "execution_status" not in task18_coverage.columns
        or not task18_coverage["execution_status"].astype(str).isin(
            {"success", "identity"}
        ).all()
        or task18_coverage["run_id"].astype(str).duplicated().any()
    ):
        raise IngestError(
            "Live Task-18 output lacks unique estimated or verified-identity run coverage."
        )
    task18_run_ids = set(task18_coverage["run_id"].astype(str))
    sweep_material = state[KEY_SWEEPS].results.reset_index()
    robustness_frame = sweep_material[
        sweep_material["status"].astype(str).isin({"success", "identity"})
        & sweep_material["run_id"].astype(str).isin(task18_run_ids)
    ].reset_index(drop=True)
    if set(robustness_frame["run_id"].astype(str)) != task18_run_ids:
        raise IngestError("Live Task-17 runs do not cover the exact Task-18 execution universe.")
    for attr_name in (
        "baseline_snapshot_objects", "variant_result_objects", "variant_input_objects"
    ):
        robustness_frame.attrs[attr_name] = state[KEY_SWEEPS].results.attrs.get(
            attr_name, {}
        )
    robustness_frame.attrs["task18_stage_output_object"] = state[KEY_ROBUSTNESS]
    robustness_frame.attrs["task18_stage_output_sha256"] = _pipeline_artifact_digest(
        state[KEY_ROBUSTNESS]
    )
    robustness_frame.attrs["task17_stage_output_object"] = state[KEY_SWEEPS]
    robustness_frame.attrs["task17_stage_output_sha256"] = _pipeline_artifact_digest(
        state[KEY_SWEEPS]
    )
    # Construct the executable evidence only from the live Task4/5/6/12 stage
    # inputs and outputs. Independent module pins still decide whether those
    # exact scenarios are trusted.
    probe_registry = _live_reconciliation_probe_registry(state)
    archive_root = Path(data["replication_archive_root"])
    version_tag = _resolve_archive_release_version(data)
    # Resolve all independently controlled vintage records before creating
    # README/config/source support files, preventing a missing authority pin
    # from leaving a partially initialized archive.
    archive_artefacts = _live_archive_declarations(state, archive_root, version_tag)
    caller_archive = data.get("archive_artefacts")
    if caller_archive is not None and _pipeline_artifact_digest(caller_archive) != _pipeline_artifact_digest(archive_artefacts):
        raise IngestError("Caller archive declarations disagree with declarations derived from live state.")
    _freeze_archive_support_files(state, archive_root, version_tag)
    # Invoke the audit with the declared arguments.
    audit = execute_reconciliation_audit(
        reported,
        reproduced,
        declared_keys,
        tuple(derived_gaps),
        layer_chains=data["layer_chains"],
        counterfactual_callable=orchestrate_counterfactual_projection,
        imputation_callables=(orchestrate_control_completion, orchestrate_outcome_repair),
        eligibility_callable=orchestrate_eligibility_screening,
        robustness_frame=robustness_frame,
        study_config=data["study_config"],
        declared_facts=data.get("declared_facts"),
        archive_artefacts=archive_artefacts,
        version_tag=version_tag,
        tolerance=float(data["study_config"]["execution_parameters"]["numerical"]["comparison_tolerance"]),
        require_invariants=True,
        manuscript_universe=manuscript_universe,
        probe_registry=probe_registry,
        archive_state=state,
        archive_root=(
            archive_root if "replication_archive_root" in data else None
        ),
    )
    # Return the stage's artefacts.
    return {KEY_RECONCILIATION: audit}


# Compose the declared pipeline registry.
PIPELINE_STAGES: Final[Tuple[StageRecord, ...]] = (
    StageRecord(1, "raw_data_ingest", orchestrate_raw_data_ingest, (), (KEY_BUNDLE,), ("sources",), _adapt_ingest),
    StageRecord(2, "study_config_validation", orchestrate_study_config_validation, (), (KEY_CONFIG,), (), _adapt_config),
    StageRecord(3, "data_cleansing", orchestrate_data_cleansing, (KEY_BUNDLE, KEY_CONFIG), (KEY_CLEANSE,), (), _adapt_cleanse),
    StageRecord(4, "eligibility_screening", orchestrate_eligibility_screening, (KEY_CLEANSE, KEY_CONFIG), (KEY_ELIGIBILITY,), (), _adapt_eligibility),
    StageRecord(5, "control_completion", orchestrate_control_completion, (KEY_CLEANSE, KEY_ELIGIBILITY, KEY_CONFIG), (KEY_CONTROLS,), (), _adapt_controls),
    StageRecord(6, "outcome_repair", orchestrate_outcome_repair, (KEY_CLEANSE, KEY_ELIGIBILITY, KEY_CONFIG), (KEY_OUTCOMES,), (), _adapt_outcomes),
    StageRecord(7, "panel_assembly", orchestrate_panel_assembly, (KEY_OUTCOMES, KEY_CONTROLS, KEY_CONFIG), (KEY_PANEL,), (), _adapt_panel),
    StageRecord(8, "treatment_coding", orchestrate_treatment_coding, (KEY_CLEANSE, KEY_ELIGIBILITY, KEY_CONFIG), (KEY_TREATMENT,), (), _adapt_treatment),
    StageRecord(9, "partition_construction", orchestrate_partition_construction, (KEY_CLEANSE, KEY_TREATMENT, KEY_PANEL, KEY_ELIGIBILITY, KEY_CONFIG), (KEY_REGISTRY,), (), _adapt_partitions),
    StageRecord(10, "donor_factor", orchestrate_donor_factor, (KEY_REGISTRY, KEY_PANEL, KEY_CONFIG), (KEY_FACTOR,), (), _adapt_factor),
    StageRecord(11, "pretreatment_fit", orchestrate_pretreatment_fit, (KEY_FACTOR, KEY_CONFIG), (KEY_FIT,), (), _adapt_fit),
    StageRecord(12, "counterfactual_projection", orchestrate_counterfactual_projection, (KEY_FIT, KEY_FACTOR, KEY_REGISTRY, KEY_PANEL, KEY_CONFIG), (KEY_COUNTERFACTUAL,), (), _adapt_counterfactual),
    StageRecord(13, "gap_and_att", orchestrate_gap_and_att, (KEY_COUNTERFACTUAL, KEY_PANEL, KEY_REGISTRY, KEY_CONFIG), (KEY_GAPS,), (), _adapt_gaps),
    StageRecord(14, "country_dispersion_inference", orchestrate_country_dispersion_inference, (KEY_GAPS, KEY_CONFIG), (KEY_INFERENCE_CHAIN,), (), _adapt_chain_inference),
    StageRecord(15, "baseline_pipeline", orchestrate_baseline_pipeline, (KEY_CLEANSE, KEY_OUTCOMES, KEY_REGISTRY, KEY_PANEL, KEY_TREATMENT, KEY_CONFIG), (KEY_BASELINE, KEY_DATA_MAPPING), ("income_groups",), _adapt_baseline),
    StageRecord(16, "resampling_inference", execute_inference, (KEY_BUNDLE, KEY_CLEANSE, KEY_BASELINE, KEY_DATA_MAPPING, KEY_CONFIG), (KEY_RESAMPLING,), ("seeds", "stability_seeds"), _adapt_resampling),
    StageRecord(17, "sweep_factory", execute_sweeps, (KEY_TREATMENT, KEY_DATA_MAPPING, KEY_CONFIG), (KEY_SWEEPS,), (), _adapt_sweeps),
    StageRecord(18, "robustness_sweeps", execute_robustness_sweeps, (KEY_DATA_MAPPING, KEY_CONFIG), (KEY_ROBUSTNESS,), (), _adapt_robustness),
    StageRecord(19, "secondary_sweeps", execute_secondary_sweeps, (KEY_BASELINE, KEY_DATA_MAPPING, KEY_CONFIG), (KEY_SECONDARY,), (), _adapt_secondary),
    StageRecord(20, "influence_diagnostics", execute_influence_diagnostics, (KEY_DATA_MAPPING, KEY_CONFIG), (KEY_INFLUENCE,), (), _adapt_influence),
    StageRecord(21, "influence_tables", execute_influence_tables, (KEY_INFLUENCE, KEY_BASELINE, KEY_CONFIG), (KEY_INFLUENCE_TABLES,), (), _adapt_influence_tables),
    StageRecord(22, "matching_designs", execute_matching_designs, (KEY_DATA_MAPPING, KEY_CONFIG), (KEY_MATCHING,), (), _adapt_matching),
    StageRecord(23, "matching_report", execute_matching_report, (KEY_MATCHING, KEY_DATA_MAPPING, KEY_CONFIG), (KEY_MATCHING_REPORT,), (), _adapt_matching_report),
    StageRecord(24, "cell_ledger", build_cell_ledger, (KEY_CLEANSE, KEY_CONFIG), (KEY_CELL_LEDGER,), (), _adapt_cell_ledger),
    StageRecord(25, "exposure_primitives", construct_exposure_primitives, (KEY_BUNDLE, KEY_CLEANSE, KEY_CELL_LEDGER, KEY_CONFIG), (KEY_PRIMITIVES,), (), _adapt_primitives),
    StageRecord(26, "exposure_ratios", construct_exposure_ratios, (KEY_PRIMITIVES, KEY_ELIGIBILITY, KEY_CONFIG), (KEY_RATIOS,), (), _adapt_ratios),
    StageRecord(27, "heterogeneity_suite", execute_heterogeneity_suite, (KEY_PANEL, KEY_REGISTRY, KEY_TREATMENT, KEY_RATIOS, KEY_CONFIG), (KEY_HETEROGENEITY,), (), _adapt_heterogeneity),
    StageRecord(28, "heterogeneity_regressions", execute_heterogeneity_regressions, (KEY_PANEL, KEY_REGISTRY, KEY_TREATMENT, KEY_RATIOS, KEY_CONFIG), (KEY_HETEROGENEITY_REG,), (), _adapt_heterogeneity_regressions),
    StageRecord(29, "direct_arming_pipeline", execute_direct_arming_pipeline, (KEY_BUNDLE, KEY_CLEANSE, KEY_DATA_MAPPING, KEY_TREATMENT, KEY_CONFIG), (KEY_ARMING,), ("seeds",), _adapt_direct_arming),
    StageRecord(
        30,
        "production",
        produce_figures_and_tables,
        (KEY_BASELINE, KEY_COUNTERFACTUAL, KEY_RESAMPLING, KEY_SWEEPS, KEY_SECONDARY, KEY_INFLUENCE_TABLES, KEY_MATCHING, KEY_MATCHING_REPORT, KEY_HETEROGENEITY_REG, KEY_ARMING, KEY_CONFIG),
        (KEY_PRODUCTION,),
        (
            "table_sources", "table_binding_authority", "country_geometry",
            "world_geometry", "output_dir", "overlay_pair",
        ),
        _adapt_production,
    ),
    StageRecord(31, "reconciliation", execute_reconciliation_audit, (KEY_PRODUCTION, KEY_SWEEPS, KEY_ROBUSTNESS, KEY_CONFIG, KEY_CLEANSE, KEY_ELIGIBILITY, KEY_CONTROLS, KEY_OUTCOMES, KEY_PANEL, KEY_REGISTRY, KEY_FACTOR, KEY_FIT, KEY_COUNTERFACTUAL), (KEY_RECONCILIATION,), ("manuscript_universe", "declared_facts", "replication_archive_root", "layer_chains"), _adapt_reconciliation),
)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 33, Step 3: The result dictionary, the summary and the freeze
# -------------------------------------------------------------------------------------------------------------------------------


# Validate the caller's data mapping before any stage runs.
def validate_pipeline_inputs(data: Mapping[str, Any]) -> Tuple[str, ...]:
    """
    Validate the caller's pipeline data mapping.

    Parameters
    ----------
    data : mapping
        The caller's data mapping, which must carry the study configuration
        and the ten raw source frames under their declared keys.

    Returns
    -------
    tuple of str
        Explicit unresolved branch prerequisites. Core authoritative source
        or configuration failures raise; optional-but-mandatory-for-study
        branch inputs are returned so the run can retain an incomplete audit.

    Raises
    ------
    IngestError
        If the mapping is not a mapping, if it lacks ``study_config`` or
        ``sources``, or if either entry is not itself a mapping.
    """
    # Refuse an input that is not a mapping.
    if not isinstance(data, Mapping):
        # The failure names the received type.
        raise IngestError(
            f"The pipeline data must be a mapping; received "
            f"{type(data).__name__}."
        )
    # Refuse a mapping without the study configuration.
    if "study_config" not in data:
        # Without the configuration no stage can run.
        raise IngestError("The data mapping lacks the declared key 'study_config'.")
    # Refuse a configuration that is not a mapping.
    if not isinstance(data["study_config"], Mapping):
        # The failure names the received type.
        raise IngestError("The study_config entry is not a mapping.")
    # Refuse a mapping without the raw sources.
    if "sources" not in data:
        # Without the sources the ingestion cannot run.
        raise IngestError("The data mapping lacks the declared key 'sources'.")
    # Refuse sources that are not a mapping.
    if not isinstance(data["sources"], Mapping):
        # The failure names the received type.
        raise IngestError("The sources entry is not a mapping.")
    source_names = tuple(str(name) for name in data["sources"])
    if len(source_names) != len(set(source_names)) or set(source_names) != set(DECLARED_SOURCES):
        raise IngestError(
            "Pipeline sources must contain exactly the ten authoritative raw sources."
        )
    for source_name in DECLARED_SOURCES:
        frame = data["sources"][source_name]
        if not isinstance(frame, pd.DataFrame) or frame.empty:
            raise IngestError(
                f"Authoritative raw source {source_name!r} must be a non-empty DataFrame."
            )
    deprecated_independent = {
        "exposure_numerator",
        "exposure_totals",
        "exposure_primitives",
        "reproduced_values",
        "matched_paths",
        "reference_window_estimates",
        "placebo_draws",
        "observed_preferred_att",
        "robustness_frame",
        "counterfactual_callable",
        "imputation_callables",
        "eligibility_callable",
        "reconciliation_probe_registry",
        "comtrade_retry_records",
        "production_gaps",
    }
    supplied_deprecated = sorted(deprecated_independent.intersection(data))
    if supplied_deprecated:
        raise IngestError(
            "Pipeline refuses disconnected independently supplied derived artefacts: "
            f"{supplied_deprecated}."
        )
    # Validate the configuration completely before any raw computation. The
    # stage still emits the immutable report, but preflight proves readiness.
    readiness = orchestrate_study_config_validation(data["study_config"])
    live_fingerprint = compute_config_fingerprint(data["study_config"])
    expected_token = hashlib.sha256(
        ("study-config-ready:" + live_fingerprint).encode("utf-8")
    ).hexdigest()
    if readiness.config_fingerprint != live_fingerprint or readiness.readiness_token != expected_token:
        raise IngestError("Pipeline configuration did not pass the readiness-token gate.")
    branch_requirements = {
        "resampling": ("seeds", "stability_seeds"),
        "sweep_income": ("income_groups",),
        "production": (
            "country_geometry", "world_geometry", "table_sources", "table_binding_authority",
            "output_dir", "overlay_pair",
        ),
        "reconciliation": (
            "manuscript_universe",
            "declared_facts",
            "replication_archive_root",
            "layer_chains",
        ),
    }
    blockers = tuple(
        f"{branch}:{key}"
        for branch, keys in branch_requirements.items()
        for key in keys
        if key not in data or data[key] is None
    )
    malformed: List[str] = []
    mapping_inputs = (
        "seeds", "stability_seeds", "region_map", "income_groups",
        "exposure_windows", "table_sources", "table_binding_authority",
        "declared_facts",
    )
    for key in mapping_inputs:
        if key in data and data[key] is not None and (
            not isinstance(data[key], Mapping) or len(data[key]) == 0
        ):
            malformed.append(f"malformed:{key}")
    frame_inputs = ("country_geometry", "world_geometry", "manuscript_universe")
    for key in frame_inputs:
        if key in data and data[key] is not None and (
            not isinstance(data[key], pd.DataFrame) or data[key].empty
        ):
            malformed.append(f"malformed:{key}")
    if "overlay_pair" in data and data["overlay_pair"] is not None:
        pair_value = data["overlay_pair"]
        if (
            isinstance(pair_value, (str, bytes))
            or not isinstance(pair_value, Sequence)
            or len(pair_value) != 2
            or any(not str(value).strip() for value in pair_value)
        ):
            malformed.append("malformed:overlay_pair")
    if "output_dir" in data and data["output_dir"] is not None and not str(
        data["output_dir"]
    ).strip():
        malformed.append("malformed:output_dir")
    if "layer_chains" in data and data["layer_chains"] is not None and not isinstance(
        data["layer_chains"], Mapping
    ):
        malformed.append("malformed:layer_chains")
    universe_input = data.get("manuscript_universe")
    if isinstance(universe_input, pd.DataFrame):
        authority_key = (
            str(universe_input.attrs.get("authority_id", "")).strip(),
            str(universe_input.attrs.get("version_id", "")).strip(),
        )
        if authority_key not in TRUSTED_MANUSCRIPT_UNIVERSE_PINS:
            malformed.append("untrusted:manuscript_universe_authority_pin")
    if not TRUSTED_RECONCILIATION_SCENARIO_PINS:
        malformed.append("untrusted:reconciliation_scenario_pins")
    blockers = tuple(dict.fromkeys((*blockers, *malformed)))
    # Log the validated input.
    logger.info(
        "pipeline inputs validated: %d source frame(s), %d branch blocker(s).",
        len(data["sources"]), len(blockers),
    )
    return blockers


# Verify the declared registry's threading contract before the run.
def assert_pipeline_registry(stages: Sequence[StageRecord]) -> None:
    """
    Verify the declared stage registry's threading contract.

    Parameters
    ----------
    stages : sequence of StageRecord
        The declared pipeline stages in their run order.

    Returns
    -------
    None
        The function returns nothing when the registry is consistent.

    Raises
    ------
    IngestError
        If the stage positions are not strictly ascending, if two stages
        produce the same artefact key, if a required state key has no
        producer, or if a producer sits at or after its consumer.
    """
    # Refuse an empty registry.
    if not stages:
        # Without stages the pipeline has nothing to run.
        raise IngestError("The pipeline registry is empty.")
    names = [str(stage.name).strip() for stage in stages]
    numbers = [int(stage.number) for stage in stages]
    if any(not name for name in names) or len(names) != len(set(names)):
        raise IngestError("Pipeline stage names must be nonblank and unique.")
    if numbers != list(range(1, len(stages) + 1)):
        raise IngestError("Pipeline stage numbers must form a complete one-based sequence.")
    # Compose the producing stage position of every artefact key.
    producers: Dict[str, int] = {}
    # Track the previous position so the order is proved ascending.
    previous: int = -1
    # Iterate over the stages in their declared order.
    for position, stage in enumerate(stages):
        if not callable(stage.orchestrator) or not callable(stage.adapter):
            raise IngestError(f"Stage {stage.name!r} lacks a callable orchestrator/adapter.")
        if not stage.produces or len(stage.produces) != len(set(stage.produces)):
            raise IngestError(f"Stage {stage.name!r} has an empty or duplicate output contract.")
        if len(stage.requires) != len(set(stage.requires)) or len(stage.data_keys) != len(set(stage.data_keys)):
            raise IngestError(f"Stage {stage.name!r} duplicates a dependency key.")
        if stage.number >= 3 and KEY_CONFIG not in stage.requires:
            raise IngestError(f"Stage {stage.name!r} is not gated by validated configuration readiness.")
        forbidden_data = {
            "exposure_numerator", "exposure_totals", "exposure_primitives",
            "reproduced_values", "matched_paths", "reference_window_estimates",
            "robustness_frame", "comtrade_retry_records", "production_gaps",
        }
        overlap = sorted(forbidden_data.intersection(stage.data_keys))
        if overlap:
            raise IngestError(
                f"Stage {stage.name!r} declares disconnected derived input(s) {overlap}."
            )
        # Refuse a stage whose position is not strictly ascending.
        if int(stage.number) <= previous:
            # The failure names the stage and both positions.
            raise IngestError(
                f"Stage {stage.name!r} declares position {stage.number} after "
                f"{previous}; the positions must ascend."
            )
        # Record the new previous position.
        previous = int(stage.number)
        # Record every artefact key the stage produces.
        for key in stage.produces:
            # Refuse a key produced twice.
            if key in producers:
                # The failure names the key and both producers.
                raise IngestError(
                    f"Artefact key {key!r} is produced by stages "
                    f"{producers[key]} and {position}; it must be unique."
                )
            # Record the producing position.
            producers[key] = position
    # Verify every required state key against its producer.
    for position, stage in enumerate(stages):
        # Iterate over the stage's declared requirements.
        for key in stage.requires:
            # Refuse a requirement no stage produces.
            if key not in producers:
                # The failure names the stage and the unproduced key.
                raise IngestError(
                    f"Stage {stage.name!r} requires artefact {key!r}, which no "
                    f"stage produces."
                )
            # Refuse a producer at or after its consumer.
            if producers[key] >= position:
                # The failure names the stage, the key and both positions.
                raise IngestError(
                    f"Stage {stage.name!r} requires artefact {key!r} produced at "
                    f"position {producers[key]}, which is not earlier than its own "
                    f"position {position}."
                )
    # Log the verified registry.
    logger.info(
        "pipeline registry verified: %d stage(s), %d artefact key(s).",
        len(stages),
        len(producers),
    )


# Resolve the missing inputs of one stage.
def resolve_stage_missing_keys(
    stage: StageRecord,
    state: Mapping[str, Any],
    data: Mapping[str, Any],
) -> Tuple[str, ...]:
    """
    Resolve the inputs one stage lacks.

    Parameters
    ----------
    stage : StageRecord
        The stage whose declared inputs are checked.
    state : mapping
        The artefact state built by the earlier stages.
    data : mapping
        The caller's data mapping.

    Returns
    -------
    tuple of str
        The missing artefact keys and data keys, in the stage's declared
        order; empty when the stage can run.
    """
    # Accumulate the missing state keys.
    missing: List[str] = [
        key for key in stage.requires if key not in state
    ]
    # Accumulate the missing data keys.
    missing.extend(
        key for key in stage.data_keys if key not in data
    )
    # Return the missing keys.
    return tuple(missing)


# Runtime wall-clock stamps are recorded for audit readability but are not
# scientific content; excluding only these named fields makes identical
# reruns reproducible without weakening source, schema or lineage hashes.
PIPELINE_VOLATILE_TIME_FIELDS: Final[Set[str]] = {
    "built_utc", "completed_utc", "created_utc", "generated_utc", "run_utc"
}


# Digest a complete pipeline artefact, including dataclass fields and pandas attrs.
def _pipeline_artifact_digest(value: Any) -> str:
    """Return a deterministic full-content digest for a pipeline artefact."""
    if isinstance(value, pd.DataFrame):
        try:
            content_digest = _content_digest(value)
        except (IngestError, TypeError, ValueError):
            fallback = {
                "index": [repr(item) for item in value.index.tolist()],
                "index_names": [repr(item) for item in value.index.names],
                "columns": [repr(item) for item in value.columns.tolist()],
                "column_names": [repr(item) for item in value.columns.names],
                "dtypes": [str(item) for item in value.dtypes.tolist()],
                "values": [
                    [repr(cell) for cell in row]
                    for row in value.astype(object).to_numpy().tolist()
                ],
            }
            content_digest = hashlib.sha256(
                json.dumps(fallback, sort_keys=True, separators=(",", ":")).encode("utf-8")
            ).hexdigest()
        payload = {
            "content": content_digest,
            "attrs": _pipeline_artifact_digest(
                {
                    key: item for key, item in value.attrs.items()
                    if str(key) not in PIPELINE_VOLATILE_TIME_FIELDS
                }
            ),
        }
        return hashlib.sha256(
            json.dumps(payload, sort_keys=True, separators=(",", ":")).encode("utf-8")
        ).hexdigest()
    if isinstance(value, pd.Series):
        payload = {
            "content": _content_digest(value),
            "attrs": _pipeline_artifact_digest(
                {
                    key: item for key, item in value.attrs.items()
                    if str(key) not in PIPELINE_VOLATILE_TIME_FIELDS
                }
            ),
        }
        return hashlib.sha256(
            json.dumps(payload, sort_keys=True, separators=(",", ":")).encode("utf-8")
        ).hexdigest()
    if is_dataclass(value) and not isinstance(value, type):
        return _pipeline_artifact_digest(
            {
                item.name: getattr(value, item.name)
                for item in fields(value)
                if item.name not in PIPELINE_VOLATILE_TIME_FIELDS
            }
        )
    if isinstance(value, Path):
        resolved = value.expanduser().resolve()
        if resolved.is_file():
            file_hash = hashlib.sha256(resolved.read_bytes()).hexdigest()
            return _pipeline_artifact_digest(
                {"path": str(resolved), "byte_size": resolved.stat().st_size, "sha256": file_hash}
            )
        # Output/archive directory declarations are infrastructure locations,
        # not file artefacts. Their scientific contents are hashed separately.
        return _pipeline_artifact_digest(
            {"directory_declaration": str(resolved), "exists": resolved.is_dir()}
        )
    if callable(value):
        module = str(getattr(value, "__module__", ""))
        qualname = str(getattr(value, "__qualname__", getattr(value, "__name__", "")))
        try:
            source = inspect.getsource(value)
        except (OSError, TypeError):
            source = ""
        if not module or not qualname or not source:
            raise IngestError("Pipeline callable dependency lacks stable source identity.")
        return _pipeline_artifact_digest(
            {"module": module, "qualname": qualname, "source": source}
        )
    if isinstance(value, Mapping):
        folded = hashlib.sha256(b"mapping")
        scientific_keys = [
            key for key in value if str(key) not in PIPELINE_VOLATILE_TIME_FIELDS
        ]
        for key in sorted(scientific_keys, key=repr):
            folded.update(repr(key).encode("utf-8"))
            folded.update(_pipeline_artifact_digest(value[key]).encode("ascii"))
        return folded.hexdigest()
    if isinstance(value, (tuple, list)):
        folded = hashlib.sha256(type(value).__name__.encode("ascii"))
        for item in value:
            folded.update(_pipeline_artifact_digest(item).encode("ascii"))
        return folded.hexdigest()
    if isinstance(value, set):
        return _pipeline_artifact_digest(tuple(sorted(value, key=repr)))
    if isinstance(value, np.ndarray):
        return _content_digest(value)
    if value is None or isinstance(value, (str, int, float, bool, np.generic)):
        return _content_digest(value)
    raise IngestError(
        f"Pipeline artefact type {type(value).__name__} has no deterministic digest."
    )


# Refuse an artefact whose own completion contract says it is incomplete.
def _assert_stage_artifact_complete(stage: StageRecord, produced: Mapping[str, Any]) -> None:
    """Validate semantic artefact completion rather than mere key presence."""
    for key in stage.produces:
        value = produced[key]
        if value is None:
            raise IngestError(f"Stage {stage.name!r} produced null artefact {key!r}.")
        if isinstance(value, (pd.DataFrame, pd.Series)) and value.empty:
            raise IngestError(f"Stage {stage.name!r} produced empty artefact {key!r}.")
        if isinstance(value, Mapping) and not value:
            raise IngestError(f"Stage {stage.name!r} produced empty artefact {key!r}.")
        if hasattr(value, "complete") and getattr(value, "complete") is not True:
            raise IngestError(
                f"Stage {stage.name!r} artefact {key!r} reports incomplete coverage."
            )
        if hasattr(value, "status"):
            status = str(getattr(value, "status")).strip().lower()
            if status not in {"complete", "completed", "success", "ready", "verified"}:
                raise IngestError(
                    f"Stage {stage.name!r} artefact {key!r} reports status {status!r}."
                )
        if hasattr(value, "reproducibility_gaps"):
            gap_frame = getattr(value, "reproducibility_gaps")
            if not isinstance(gap_frame, pd.DataFrame) or "state" not in gap_frame.columns:
                raise IngestError(
                    f"Stage {stage.name!r} artefact {key!r} has malformed gap evidence."
                )
            unresolved = gap_frame["state"].astype(str).map(
                lambda state: state == "unspecified"
                or state == "unresolved"
                or state.endswith("_absent")
            )
            if bool(unresolved.any()):
                raise IngestError(
                    f"Stage {stage.name!r} artefact {key!r} retains unresolved production gaps."
                )
        if hasattr(value, "rendered"):
            rendered = getattr(value, "rendered")
            if (
                not isinstance(rendered, pd.DataFrame)
                or rendered.empty
                or "written" not in rendered.columns
                or not bool(rendered["written"].astype(bool).all())
            ):
                raise IngestError(
                    f"Stage {stage.name!r} did not write every mandatory figure."
                )
        if key == KEY_RECONCILIATION:
            reconciliation_blockers = getattr(value, "blockers", None)
            if reconciliation_blockers is None:
                has_reconciliation_blockers = False
            elif isinstance(reconciliation_blockers, pd.DataFrame):
                has_reconciliation_blockers = not reconciliation_blockers.empty
            elif isinstance(reconciliation_blockers, (tuple, list, set, Mapping)):
                has_reconciliation_blockers = len(reconciliation_blockers) > 0
            else:
                raise IngestError(
                    "Reconciliation blockers must be a DataFrame or a finite collection."
                )
            if (
                getattr(value, "status", "") != RECONCILIATION_COMPLETE
                or has_reconciliation_blockers
            ):
                raise IngestError("Reconciliation audit is not complete and blocker-free.")
        _pipeline_artifact_digest(value)


# Run one stage through its adapter and verify the produced keys.
def run_pipeline_stage(
    stage: StageRecord,
    state: Mapping[str, Any],
    data: Mapping[str, Any],
) -> Mapping[str, Any]:
    """
    Run one stage through its adapter.

    Parameters
    ----------
    stage : StageRecord
        The stage to run.
    state : mapping
        The artefact state the adapter reads.
    data : mapping
        The caller's data mapping the adapter reads.

    Returns
    -------
    mapping
        The artefacts the stage produced, keyed as the record declares.

    Raises
    ------
    IngestError
        If the adapter does not return a mapping, or if the returned keys
        diverge from the record's declared ``produces``.
    """
    if stage.number >= 3:
        _require_ready_config(state, data)
    # Run the stage's adapter.
    produced = stage.adapter(state, data)
    # Refuse an adapter that returns no mapping.
    if not isinstance(produced, Mapping):
        # The failure names the stage and the received type.
        raise IngestError(
            f"Stage {stage.name!r} returned {type(produced).__name__}; a mapping "
            f"is required."
        )
    # Refuse a produced key set that diverges from the declaration.
    if set(produced.keys()) != set(stage.produces):
        # The failure names the stage, the declared keys and the returned keys.
        raise IngestError(
            f"Stage {stage.name!r} returned keys {sorted(produced)} against the "
            f"declared {sorted(stage.produces)}."
        )
    try:
        _assert_stage_artifact_complete(stage, produced)
    except IngestError as exc:
        reports_incomplete = any(
            (
                hasattr(value, "complete")
                and getattr(value, "complete") is not True
            )
            or (
                hasattr(value, "status")
                and str(getattr(value, "status")).strip().lower()
                not in {"complete", "completed", "success", "ready", "verified"}
            )
            or (
                hasattr(value, "reproducibility_gaps")
                and isinstance(getattr(value, "reproducibility_gaps"), pd.DataFrame)
            )
            or hasattr(value, "rendered")
            for value in produced.values()
            if value is not None
        )
        if reports_incomplete:
            raise PipelineStageIncompleteError(str(exc), produced) from exc
        raise
    # Return the verified artefacts.
    return produced


# Execute every stage in order and record each outcome.
def execute_pipeline_stages(
    data: Mapping[str, Any],
) -> Tuple[Dict[str, Any], List[Dict[str, Any]]]:
    """
    Execute every declared stage in order.

    A stage whose declared inputs are absent is recorded as skipped with the
    missing keys named and is never silently omitted. A stage that refuses is
    recorded as failed with its error, and the chain continues: every later
    stage is still evaluated against its own requirements, so independent
    branches of the pipeline complete and every completed artefact is
    retained.

    Parameters
    ----------
    data : mapping
        The caller's data mapping.

    Returns
    -------
    tuple
        The artefact state and one outcome record per stage, in run order.
    """
    # Initialise the artefact state the stages thread.
    state: Dict[str, Any] = {}
    # Accumulate the per-stage outcome records.
    records: List[Dict[str, Any]] = []
    # Iterate over the stages in their declared order.
    for stage in PIPELINE_STAGES:
        try:
            dependency_payload = {
                "state": {
                    key: _pipeline_artifact_digest(state[key])
                    for key in stage.requires
                    if key in state
                },
                "data": {
                    key: _pipeline_artifact_digest(data[key])
                    for key in stage.data_keys
                    if key in data
                },
            }
            dependency_digest = _pipeline_artifact_digest(dependency_payload)
        except Exception as exc:  # noqa: BLE001 - retain the typed dependency failure
            records.append(
                {
                    "number": int(stage.number),
                    "name": str(stage.name),
                    "outcome": OUTCOME_FAILED,
                    "produces": ", ".join(stage.produces),
                    "missing": "",
                    "error": "dependency digest: %s: %s" % (type(exc).__name__, exc),
                    "dependency_digest": "",
                    "artifact_digest": "",
                }
            )
            logger.error(
                "stage %d (%s) failed while hashing dependencies: %s: %s.",
                stage.number,
                stage.name,
                type(exc).__name__,
                exc,
            )
            continue
        # Resolve the stage's missing inputs.
        missing: Tuple[str, ...] = resolve_stage_missing_keys(stage, state, data)
        # Record an unmet stage as skipped with its missing keys named.
        if missing:
            # Append the skip record.
            records.append(
                {
                    "number": int(stage.number),
                    "name": str(stage.name),
                    "outcome": OUTCOME_SKIPPED,
                    "produces": ", ".join(stage.produces),
                    "missing": ", ".join(missing),
                    "error": "",
                    "dependency_digest": dependency_digest,
                    "artifact_digest": "",
                }
            )
            # Log the skip with its missing keys.
            logger.warning(
                "stage %d (%s) skipped: missing input(s) %s.",
                stage.number,
                stage.name,
                ", ".join(missing),
            )
            # Move to the next stage.
            continue
        # Attempt the stage and retain the state regardless of its outcome.
        try:
            # Run the stage through its adapter.
            produced = run_pipeline_stage(stage, state, data)
        except Exception as exc:  # noqa: BLE001 - the failure is recorded
            # A well-formed incomplete scientific result remains valuable
            # diagnostic evidence even though it cannot satisfy its mandatory
            # output contract or unblock descendants.
            failure_artifact_digest = ""
            if isinstance(exc, PipelineStageIncompleteError):
                for key, partial in exc.artifacts.items():
                    state[str(key) + "_diagnostic"] = partial
                failure_artifact_digest = _pipeline_artifact_digest(
                    dict(exc.artifacts)
                )
            # Append the failure record with the exception's type and message.
            records.append(
                {
                    "number": int(stage.number),
                    "name": str(stage.name),
                    "outcome": OUTCOME_FAILED,
                    "produces": ", ".join(stage.produces),
                    "missing": "",
                    "error": "%s: %s" % (type(exc).__name__, exc),
                    "dependency_digest": dependency_digest,
                    "artifact_digest": failure_artifact_digest,
                }
            )
            # Log the failure with its error.
            logger.error(
                "stage %d (%s) failed: %s: %s.",
                stage.number,
                stage.name,
                type(exc).__name__,
                exc,
            )
            # Move to the next stage, retaining every earlier artefact.
            continue
        # Store the stage's artefacts under their declared keys.
        state.update(produced)
        artifact_digest = _pipeline_artifact_digest(dict(produced))
        # Append the completion record.
        records.append(
            {
                "number": int(stage.number),
                "name": str(stage.name),
                "outcome": OUTCOME_COMPLETED,
                "produces": ", ".join(stage.produces),
                "missing": "",
                "error": "",
                "dependency_digest": dependency_digest,
                "artifact_digest": artifact_digest,
            }
        )
        # Log the completion with its keys.
        logger.info(
            "stage %d (%s) completed: %s.",
            stage.number,
            stage.name,
            ", ".join(stage.produces),
        )
    # Return the state and the records.
    return state, records


# Assemble the per-stage summary frame.
def build_pipeline_summary(records: Sequence[Mapping[str, Any]]) -> pd.DataFrame:
    """
    Assemble the per-stage summary frame.

    Parameters
    ----------
    records : sequence of mapping
        The per-stage outcome records.

    Returns
    -------
    pandas.DataFrame
        One row per stage, indexed on the stage's position, carrying the
        stage's name, outcome, produced keys, missing keys and error.
    """
    # Assemble the summary frame with the declared column order.
    frame: pd.DataFrame = pd.DataFrame(list(records), columns=list(SUMMARY_COLUMNS))
    # Index the frame on the stage's position.
    frame = frame.set_index("number")
    # Return the ordered frame.
    return frame.sort_index()


# Freeze the run's report into one digest.
def freeze_pipeline_result(
    records: Sequence[Mapping[str, Any]],
    study_status: Optional[str] = None,
    blockers: Sequence[str] = (),
    archive_evidence: Optional[Mapping[str, Any]] = None,
) -> str:
    """
    Freeze the per-stage report into one digest.

    The digest is computed over the canonical stage outcomes, their observed
    dependency and artefact digests, the terminal study state, its blockers,
    and the archive evidence digest.  It therefore certifies the executed
    dataflow rather than stage labels alone.

    Parameters
    ----------
    records : sequence of mapping
        The per-stage outcome records.
    study_status : str, optional
        The terminal complete/incomplete state.
    blockers : sequence of str
        The terminal blocker descriptions.
    archive_evidence : mapping, optional
        The archive declarations whose full content is bound into the digest.

    Returns
    -------
    str
        The SHA-256 hexadecimal digest of the canonical report.
    """
    if not records:
        raise IngestError("Cannot freeze an empty pipeline execution.")
    numbers = [int(record["number"]) for record in records]
    if len(numbers) != len(set(numbers)):
        raise IngestError("Pipeline execution records duplicate a stage number.")
    # Project every record onto the declared columns with canonical types.
    canonical: List[Dict[str, Any]] = [
        {
            "number": int(record["number"]),
            "name": str(record["name"]),
            "outcome": str(record["outcome"]),
            "produces": str(record["produces"]),
            "missing": str(record["missing"]),
            "error": str(record["error"]),
            "dependency_digest": str(record.get("dependency_digest", "")),
            "artifact_digest": str(record.get("artifact_digest", "")),
        }
        for record in records
    ]
    # Sort the canonical records by their position so the digest is order-stable.
    canonical.sort(key=lambda record: record["number"])
    # Serialise the canonical records with sorted keys and a stable separator.
    payload = {
        "stages": canonical,
        "study_status": None if study_status is None else str(study_status),
        "blockers": tuple(str(value) for value in blockers),
        "archive_evidence": (
            None
            if archive_evidence is None
            else _pipeline_artifact_digest(dict(archive_evidence))
        ),
    }
    serialized: str = json.dumps(
        payload, sort_keys=True, separators=(",", ":"), default=str
    )
    # Return the digest of the canonical payload.
    return hashlib.sha256(serialized.encode("utf-8")).hexdigest()


# Verify that the replication archive is a self-contained directory tree.
def _self_contained_archive_blockers(
    data: Mapping[str, Any],
    state: Optional[Mapping[str, Any]] = None,
) -> Tuple[str, ...]:
    """
    Return blockers for byte-identical, lineage-complete archive state.

    Parameters
    ----------
    data : mapping
        Pipeline inputs carrying the archive root and optional version tag.
        Core declarations are reconstructed from ``state`` and cannot be
        nominated by this mapping.
    state : mapping, optional
        Live completed stage state. Without it archive identity cannot be
        established and completion is refused.

    Returns
    -------
    tuple of str
        Granular self-containment, byte, lineage, source and render blockers.
    """
    root_value = data.get("replication_archive_root")
    if root_value is None:
        return ("replication_archive_root is not supplied",)
    root = Path(root_value).expanduser().resolve()
    if not root.is_dir():
        return (f"replication archive root does not exist: {root}",)
    required_dirs = (
        "raw", "interim", "processed", "estimates", "figures", "tables", "audit"
    )
    missing_dirs = tuple(name for name in required_dirs if not (root / name).is_dir())
    blockers: List[str] = [
        "replication archive lacks directory " + name for name in missing_dirs
    ]
    try:
        version_tag = _resolve_archive_release_version(data)
    except IngestError as exc:
        blockers.append(str(exc))
        return tuple(blockers)
    if not isinstance(state, Mapping):
        blockers.append("live pipeline state is unavailable for archive identity checks")
        return tuple(blockers)
    for required_file in ("README.md", "config_fingerprint.txt"):
        path = root / required_file
        if not path.is_file() or path.stat().st_size == 0:
            blockers.append(f"replication archive lacks nonempty {required_file}")
    readme_path = root / "README.md"
    if readme_path.is_file():
        expected_readme = (
            "# Do Sanctions Backfire replication archive\n\n"
            f"Frozen version: {version_tag}\n"
            "All scientific files are canonical serializations of live pipeline artifacts.\n"
        ).encode("utf-8")
        try:
            if readme_path.read_bytes() != expected_readme:
                blockers.append("replication archive README disagrees with the frozen archive version")
        except OSError as exc:
            blockers.append(f"replication archive README is unreadable: {exc}")
    fingerprint_path = root / "config_fingerprint.txt"
    if fingerprint_path.is_file():
        try:
            archived_fingerprint = fingerprint_path.read_text(encoding="utf-8").strip()
        except (OSError, UnicodeError) as exc:
            blockers.append(f"replication archive config fingerprint is unreadable: {exc}")
        else:
            if KEY_CONFIG not in state:
                blockers.append("live validated configuration is unavailable")
            else:
                expected_fingerprint = hashlib.sha256(
                    _archive_canonical_bytes(state[KEY_CONFIG])
                ).hexdigest()
                if archived_fingerprint != expected_fingerprint:
                    blockers.append("replication archive config fingerprint disagrees with live state")
    source_path = root / "source_fingerprints.json"
    if KEY_BUNDLE not in state:
        blockers.append("live ingest bundle is unavailable for source fingerprint checks")
    elif not source_path.is_file() or source_path.stat().st_size == 0:
        blockers.append("replication archive lacks nonempty source_fingerprints.json")
    else:
        try:
            archived_sources = json.loads(source_path.read_text(encoding="utf-8"))
        except (OSError, UnicodeError, json.JSONDecodeError) as exc:
            blockers.append(f"source_fingerprints.json is unreadable: {exc}")
        else:
            expected_sources = {
                str(name): str(record.fingerprint_sha256)
                for name, record in state[KEY_BUNDLE].records.items()
            }
            if archived_sources != expected_sources:
                blockers.append("archived source fingerprints disagree with the live ingest records")
    try:
        artefacts = _live_archive_declarations(
            state, root, version_tag
        )
    except IngestError as exc:
        blockers.append(f"live archive declarations are unavailable: {exc}")
        return tuple(blockers)
    declared_ids = {str(value) for value in artefacts}
    if declared_ids != set(ARCHIVE_ARTEFACTS):
        blockers.append(
            "replication archive manifest does not equal the mandatory artefact registry"
        )
    reconciliation = state.get(KEY_RECONCILIATION)
    manifest = getattr(reconciliation, "manifest", None)
    if (
        not isinstance(manifest, pd.DataFrame)
        or len(manifest) != len(ARCHIVE_ARTEFACTS)
        or not manifest.index.is_unique
        or set(manifest.index.astype(str)) != set(ARCHIVE_ARTEFACTS)
    ):
        blockers.append("verified reconciliation manifest is absent or incomplete")
        manifest = pd.DataFrame()
    resolved_locations: Set[str] = set()
    represented_dirs: Set[str] = set()
    for artifact_id in ARCHIVE_ARTEFACTS:
        declaration = artefacts.get(artifact_id)
        if not isinstance(declaration, Mapping) or "location" not in declaration:
            blockers.append(f"archive artefact {artifact_id!r} lacks a concrete location")
            continue
        path = Path(str(declaration["location"])).expanduser().resolve()
        try:
            relative = path.relative_to(root)
        except ValueError:
            blockers.append(f"archive artefact {artifact_id!r} is outside the archive root")
            continue
        if str(path) in resolved_locations:
            blockers.append(f"archive artefact {artifact_id!r} reuses another manifest location")
            continue
        resolved_locations.add(str(path))
        expected_state_key, expected_section, expected_role = ARCHIVE_LIVE_BINDINGS[artifact_id]
        if not relative.parts or relative.parts[0] != expected_section:
            blockers.append(f"archive artefact {artifact_id!r} is outside its registered section")
        if str(declaration.get("state_key", "")) != expected_state_key:
            blockers.append(f"archive artefact {artifact_id!r} names the wrong live state key")
        if str(declaration.get("role", "")) != expected_role:
            blockers.append(f"archive artefact {artifact_id!r} names the wrong scientific role")
        if expected_state_key not in state:
            blockers.append(f"archive artefact {artifact_id!r} lacks live state {expected_state_key!r}")
            continue
        if not path.is_file() or path.stat().st_size == 0:
            blockers.append(f"archive artefact {artifact_id!r} is not a nonempty regular file")
            continue
        expected_bytes = _archive_canonical_bytes(state[expected_state_key])
        if path.read_bytes() != expected_bytes:
            blockers.append(f"archive artefact {artifact_id!r} bytes disagree with live state")
        claimed_digest = str(declaration.get("content_sha256", "")).strip()
        if not re.fullmatch(r"[0-9a-f]{64}", claimed_digest):
            blockers.append(f"archive artefact {artifact_id!r} has a malformed digest")
            continue
        actual_digest = _file_sha256(path)
        if actual_digest != claimed_digest:
            blockers.append(f"archive artefact {artifact_id!r} digest does not match its file")
        producing = [stage for stage in PIPELINE_STAGES if expected_state_key in stage.produces]
        expected_parents = tuple(producing[0].requires) if len(producing) == 1 else ()
        declared_parents = tuple(str(value) for value in declaration.get("parent_state_keys", ()))
        if len(producing) != 1 or declared_parents != expected_parents:
            blockers.append(f"archive artefact {artifact_id!r} parent stage lineage is incorrect")
        parent_digests = declaration.get("parent_sha256")
        if not isinstance(parent_digests, Mapping) or set(parent_digests) != set(expected_parents):
            blockers.append(f"archive artefact {artifact_id!r} parent digests are incomplete")
        else:
            for parent in expected_parents:
                if parent not in state or str(parent_digests[parent]) != hashlib.sha256(
                    _archive_canonical_bytes(state[parent])
                ).hexdigest():
                    blockers.append(f"archive artefact {artifact_id!r} parent {parent!r} is stale")
        if not manifest.empty and artifact_id in manifest.index:
            manifest_row = manifest.loc[artifact_id]
            if (
                str(manifest_row.get("location", "")) != str(path)
                or str(manifest_row.get("content_sha256", "")) != actual_digest
                or str(manifest_row.get("live_state_sha256", "")) != actual_digest
            ):
                blockers.append(f"archive artefact {artifact_id!r} disagrees with reconciliation manifest")
            for vintage_field in (
                "archive_version_tag", "vintage_tag", "vintage_authority_id",
                "vintage_source_id", "vintage_source_version",
                "vintage_evidence_reference", "vintage_claim_sha256",
            ):
                if str(manifest_row.get(vintage_field, "")) != str(
                    declaration.get(vintage_field, "")
                ):
                    blockers.append(
                        f"archive artefact {artifact_id!r} {vintage_field!r} "
                        "disagrees with its trusted live declaration"
                    )
        if relative.parts:
            represented_dirs.add(relative.parts[0])

    # Figures are independently byte-bound to the live render ledger; they do
    # not consume an eleventh core manifest identity.
    production = state.get(KEY_PRODUCTION)
    rendered = getattr(production, "rendered", None)
    figure_data = getattr(production, "figure_data", None)
    if not isinstance(rendered, pd.DataFrame) or rendered.empty or not isinstance(figure_data, Mapping):
        blockers.append("live production figure ledger/data are unavailable")
    else:
        required_render_columns = {"written", "path", "file_sha256", "byte_size", "data_sha256", "spec_sha256"}
        if not required_render_columns.issubset(rendered.columns):
            blockers.append("live render ledger lacks byte/data/spec digest fields")
        else:
            expected_figure_paths: Set[Path] = set()
            for figure_name, row in rendered.iterrows():
                path = Path(str(row["path"])).expanduser().resolve()
                expected_figure_paths.add(path)
                try:
                    relative = path.relative_to(root)
                except ValueError:
                    blockers.append(f"rendered figure {figure_name!r} is outside archive root")
                    continue
                if not relative.parts or relative.parts[0] != "figures":
                    blockers.append(f"rendered figure {figure_name!r} is outside archive/figures")
                if not bool(row["written"]) or not path.is_file() or path.stat().st_size <= 0:
                    blockers.append(f"rendered figure {figure_name!r} is missing or empty")
                    continue
                if _file_sha256(path) != str(row["file_sha256"]):
                    blockers.append(f"rendered figure {figure_name!r} byte digest is stale")
                try:
                    recorded_size = int(row["byte_size"])
                except (TypeError, ValueError, OverflowError):
                    blockers.append(f"rendered figure {figure_name!r} byte size is malformed")
                else:
                    if recorded_size != int(path.stat().st_size):
                        blockers.append(f"rendered figure {figure_name!r} byte size is stale")
                if str(figure_name) not in figure_data:
                    blockers.append(f"rendered figure {figure_name!r} lacks live figure data")
                    continue
                frame = figure_data[str(figure_name)]
                if str(row["data_sha256"]) != _production_frame_digest(frame):
                    blockers.append(f"rendered figure {figure_name!r} data digest is stale")
                if str(row["spec_sha256"]) != _content_digest(dict(frame.attrs.get("render_spec", {}))):
                    blockers.append(f"rendered figure {figure_name!r} render-spec digest is stale")
            actual_pngs = {path.resolve() for path in (root / "figures").rglob("*.png") if path.is_file()}
            if actual_pngs != expected_figure_paths:
                blockers.append("archive/figures PNG inventory disagrees with live render ledger")
            if expected_figure_paths:
                represented_dirs.add("figures")
    for directory in required_dirs:
        if directory not in represented_dirs:
            blockers.append(f"replication archive has no verified live artefact in {directory}")
    return tuple(blockers)


# -------------------------------------------------------------------------------------------------------------------------------
# Task 33, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------


# Orchestrate the study's end-to-end pipeline from the caller's data mapping.
def orchestrate_study_pipeline(data: Mapping[str, Any]) -> Dict[str, Any]:
    """
    Run the end-to-end study pipeline over the caller's data mapping.

    Parameters
    ----------
    data : mapping
        The caller's data mapping. ``study_config`` and ``sources`` are
        mandatory; every other declared key is consumed by the stage that
        needs it. Configured runtime_inputs are bound explicitly into a copy;
        conflicting duplicates raise. A stage whose keys are absent is recorded as skipped
        with the missing keys named rather than silently omitted.

    Returns
    -------
    dict
        The result dictionary. One entry per declared stage name carries the
        artefacts that stage produced. Report entries carry the ordered stage
        summary, live state, content-sensitive digest, notes, explicit study
        status and granular blockers. A run is complete only when all 31
        stages, 418 cells, three invariants and the self-contained archive are
        independently verified.

    Raises
    ------
    IngestError
        If the input mapping fails validation or the declared registry fails
        its threading contract.
    """
    # Bind documented runtime settings into a copy without changing caller data.
    data = bind_pipeline_runtime_inputs(data)
    authority_blockers = register_independent_authorities(data["study_config"])
    # Validate the caller's data mapping before any stage runs.
    preflight_blockers = tuple(authority_blockers) + validate_pipeline_inputs(data)
    # Verify the declared registry's threading contract.
    assert_pipeline_registry(PIPELINE_STAGES)
    # Execute every stage in order, recording each outcome.
    state, records = execute_pipeline_stages(data)
    # Assemble the per-stage summary frame.
    summary: pd.DataFrame = build_pipeline_summary(records)
    # Compose the result dictionary keyed on the stage's declared name.
    result: Dict[str, Any] = {}
    # Iterate over the declared stages and store the artefacts each produced.
    for stage in PIPELINE_STAGES:
        # Record the stage's artefacts under its own name.
        result[str(stage.name)] = {
            key: state[key] for key in stage.produces if key in state
        }
        diagnostic_key = next(
            (
                key + "_diagnostic"
                for key in stage.produces
                if key + "_diagnostic" in state
            ),
            None,
        )
        if diagnostic_key is not None:
            result[str(stage.name)][diagnostic_key] = state[diagnostic_key]
    # Count the outcomes for the run's notes.
    completed: int = sum(1 for row in records if row["outcome"] == OUTCOME_COMPLETED)
    # Count the skipped stages for the run's notes.
    skipped: int = sum(1 for row in records if row["outcome"] == OUTCOME_SKIPPED)
    # Count the failed stages for the run's notes.
    failed: int = sum(1 for row in records if row["outcome"] == OUTCOME_FAILED)
    mandatory_blockers: List[str] = list(preflight_blockers)
    if len(records) != len(PIPELINE_STAGES):
        mandatory_blockers.append(
            f"stage execution returned {len(records)} records for {len(PIPELINE_STAGES)} registered stages"
        )
    elif [int(row["number"]) for row in records] != [stage.number for stage in PIPELINE_STAGES]:
        mandatory_blockers.append("stage execution records are not the exact ordered 31-stage registry")
    for stage, row in zip(PIPELINE_STAGES, records):
        if stage.mandatory and row["outcome"] != OUTCOME_COMPLETED:
            detail = str(row.get("missing") or row.get("error") or row["outcome"])
            mandatory_blockers.append(f"stage {stage.number} {stage.name}: {detail}")
        elif row["outcome"] == OUTCOME_COMPLETED:
            missing_outputs = [key for key in stage.produces if key not in state]
            if missing_outputs:
                mandatory_blockers.append(
                    f"stage {stage.number} {stage.name} lost output(s) {missing_outputs}"
                )
            else:
                live_digest = _pipeline_artifact_digest(
                    {key: state[key] for key in stage.produces}
                )
                if live_digest != str(row.get("artifact_digest", "")):
                    mandatory_blockers.append(
                        f"stage {stage.number} {stage.name} artifact digest is stale"
                    )
    mandatory_blockers.extend(_self_contained_archive_blockers(data, state))
    if KEY_RECONCILIATION not in state:
        mandatory_blockers.append("complete reconciliation artefact is absent")
        diagnostic = state.get(KEY_RECONCILIATION + "_diagnostic")
        diagnostic_blockers = getattr(diagnostic, "blockers", pd.DataFrame())
        if isinstance(diagnostic_blockers, pd.DataFrame):
            for _, item in diagnostic_blockers.iterrows():
                mandatory_blockers.append(
                    "reconciliation diagnostic: {category} | {key} | {detail}".format(
                        category=str(item.get("category", "unclassified")),
                        key=str(item.get("key", "")),
                        detail=str(item.get("detail", "")),
                    )
                )
    elif getattr(state[KEY_RECONCILIATION], "status", "") != "complete":
        # An empty blocker table cannot certify an incomplete, failed or
        # missing reconciliation status. Record the status independently.
        reconciliation_status = str(
            getattr(state[KEY_RECONCILIATION], "status", "")
        )
        mandatory_blockers.append(
            f"reconciliation status is not complete: {reconciliation_status!r}"
        )
        reconciliation_blockers = getattr(state[KEY_RECONCILIATION], "blockers", pd.DataFrame())
        if isinstance(reconciliation_blockers, pd.DataFrame):
            for _, item in reconciliation_blockers.iterrows():
                mandatory_blockers.append(
                    "reconciliation: {category} | {key} | {detail}".format(
                        category=str(item.get("category", "unclassified")),
                        key=str(item.get("key", "")),
                        detail=str(item.get("detail", "")),
                    )
                )
        else:
            mandatory_blockers.append("reconciliation blockers have malformed structure")
    else:
        reconciliation = state[KEY_RECONCILIATION]
        reconciliation_blockers = getattr(reconciliation, "blockers", None)
        if not isinstance(reconciliation_blockers, pd.DataFrame) or not reconciliation_blockers.empty:
            mandatory_blockers.append("complete reconciliation is not blocker-free")
        invariants = getattr(reconciliation, "invariants", None)
        invariant_holds_are_boolean = (
            isinstance(invariants, pd.DataFrame)
            and "holds" in invariants.columns
            and invariants["holds"].map(
                lambda value: isinstance(value, (bool, np.bool_))
            ).all()
        )
        if (
            not isinstance(invariants, pd.DataFrame)
            or len(invariants) != len(INVARIANT_IDS)
            or not invariants.index.is_unique
            or set(invariants.index.astype(str)) != set(INVARIANT_IDS)
            or "holds" not in invariants.columns
            or not invariant_holds_are_boolean
            or not bool(invariants["holds"].all())
        ):
            mandatory_blockers.append("reconciliation does not verify all three real design invariants")
        coverage = getattr(reconciliation, "coverage", None)
        coverage_valid = isinstance(coverage, pd.DataFrame) and len(coverage) == 1
        if coverage_valid:
            try:
                raw_declared_count = coverage.iloc[0].get("declared_count", -1)
                raw_union_count = coverage.iloc[0].get("union_count", -1)
                if isinstance(raw_declared_count, (bool, np.bool_)) or isinstance(
                    raw_union_count, (bool, np.bool_)
                ):
                    raise ValueError("Boolean coverage counts are invalid.")
                declared_numeric = float(raw_declared_count)
                union_numeric = float(raw_union_count)
                if (
                    not np.isfinite(declared_numeric)
                    or not np.isfinite(union_numeric)
                    or not declared_numeric.is_integer()
                    or not union_numeric.is_integer()
                ):
                    raise ValueError("Coverage counts must be finite integers.")
                declared_count = int(declared_numeric)
                union_count = int(union_numeric)
            except (TypeError, ValueError, OverflowError):
                coverage_valid = False
            else:
                coverage_valid = (
                    declared_count == MANUSCRIPT_CELL_COUNT
                    and union_count == MANUSCRIPT_CELL_COUNT
                    and str(coverage.iloc[0].get("coverage_status", ""))
                    == "exact_disjoint_partition"
                )
        if not coverage_valid:
            mandatory_blockers.append("reconciliation does not cover the exact 418-cell universe")
        comparisons = getattr(reconciliation, "comparisons", None)
        gaps = getattr(reconciliation, "gaps", None)
        if (
            not isinstance(comparisons, pd.DataFrame)
            or not isinstance(gaps, pd.DataFrame)
            or len(comparisons) + len(gaps) != MANUSCRIPT_CELL_COUNT
        ):
            mandatory_blockers.append("reconciliation cell frames do not total 418")
        elif (
            len(gaps)
            or "verdict" not in comparisons.columns
            or not comparisons["verdict"].astype(str).eq(VERDICT_RECONCILED).all()
        ):
            mandatory_blockers.append("reconciliation retains gaps or divergent manuscript cells")
        manifest = getattr(reconciliation, "manifest", None)
        if (
            not isinstance(manifest, pd.DataFrame)
            or len(manifest) != len(ARCHIVE_ARTEFACTS)
            or not manifest.index.is_unique
            or set(manifest.index.astype(str)) != set(ARCHIVE_ARTEFACTS)
            or "verification_status" not in manifest.columns
            or not bool(manifest["verification_status"].astype(str).eq("observed_verified").all())
        ):
            mandatory_blockers.append("reconciliation archive manifest is not exactly verified")
    blockers = tuple(dict.fromkeys(str(item) for item in mandatory_blockers if str(item)))
    study_status = (
        STUDY_COMPLETE
        if not blockers and completed == len(PIPELINE_STAGES) and skipped == 0 and failed == 0
        else STUDY_INCOMPLETE
    )
    # Bind terminal archive evidence to live state and actual support/render
    # bytes, never to caller declarations alone.
    archive_evidence: Dict[str, Any] = {
        "archive_blockers": _self_contained_archive_blockers(data, state),
        "reconciliation_manifest": (
            getattr(state[KEY_RECONCILIATION], "manifest", pd.DataFrame())
            if KEY_RECONCILIATION in state else pd.DataFrame()
        ),
        "rendered_figures": (
            getattr(state[KEY_PRODUCTION], "rendered", pd.DataFrame())
            if KEY_PRODUCTION in state else pd.DataFrame()
        ),
    }
    root_value = data.get("replication_archive_root")
    if root_value is not None:
        root = Path(root_value).expanduser().resolve()
        archive_evidence["support_files"] = {
            name: (_file_sha256(root / name) if (root / name).is_file() else "missing")
            for name in ("README.md", "config_fingerprint.txt", "source_fingerprints.json")
        }
    # Freeze the actual dependency/artifact graph together with the terminal
    # study state and archive evidence.
    digest: str = freeze_pipeline_result(
        records,
        study_status=study_status,
        blockers=blockers,
        archive_evidence=archive_evidence,
    )
    # Compose the run's notes.
    notes: Tuple[str, ...] = (
        f"stages run: {len(records)}; completed {completed}, skipped {skipped}, "
        f"failed {failed}.",
        f"artefact keys produced: {len(state)}.",
        f"pipeline digest: {digest}.",
        f"study status: {study_status}; blockers {len(blockers)}.",
    )
    # Record the report entries.
    result["pipeline_summary"] = summary
    result["pipeline_state"] = state
    result["pipeline_digest"] = digest
    result["pipeline_notes"] = notes
    result["pipeline_status"] = study_status
    result["pipeline_blockers"] = blockers
    # Log the completed run.
    logger.info(
        "study pipeline %s: %d completed, %d skipped, %d failed; digest %s.",
        study_status,
        completed,
        skipped,
        failed,
        digest[:16],
    )
    # Return the result dictionary.
    return result
````
